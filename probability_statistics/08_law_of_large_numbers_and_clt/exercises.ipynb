## Level 0 — Concept Check

### Problem L0.1: Weak versus Strong — What Actually Differs?

State the weak and strong laws of large numbers precisely, and describe a hypothetical sequence of averages that would satisfy the weak law but violate the strong law. Why is no such sequence produced by i.i.d. integrable summands?

**Solution**

**Weak LLN.** For every $\varepsilon \gt 0$,

$$
\lim_{n \to \infty} P\left(\left\vert \bar{X}_n - \mu \right\vert \gt \varepsilon\right) = 0 .
$$

This constrains each *individual* probability in the sequence; it says nothing about whether excursions recur.

**Strong LLN.** The set of sample paths that converge has full measure,

$$
P\left(\left\{\omega : \lim_{n\to\infty}\bar{X}_n(\omega) = \mu\right\}\right) = 1 ,
$$

equivalently, for every $\varepsilon \gt 0$ the event $\{\vert \bar{X}_n - \mu\vert \gt \varepsilon\}$ occurs only finitely often, almost surely.

**A separating example.** Let $Y_n$ be *independent* with $P(Y_n = 1) = 1/n$ and $P(Y_n = 0) = 1 - 1/n$. Then $P(\vert Y_n - 0\vert \gt \varepsilon) = 1/n \to 0$, so $Y_n \xrightarrow{P} 0$. But $\sum_n P(Y_n = 1) = \sum_n 1/n = \infty$ with independence, so the second Borel–Cantelli lemma gives $P(Y_n = 1 \text{ i.o.}) = 1$: the sequence spikes to $1$ infinitely often and does **not** converge almost surely. Convergence in probability without almost sure convergence is exactly the failure of summability.

**Why averages of i.i.d. data never behave this way.** Kolmogorov's theorem shows that for i.i.d. summands with $E\vert X_1\vert \lt \infty$, the excursion probabilities decay fast enough (after maximal-inequality blocking) to be summable along dyadic blocks, forcing almost sure convergence. Integrability is the sharp condition: if $E\vert X_1\vert = \infty$, then $\limsup_n \vert \bar{X}_n\vert = \infty$ a.s.

$$
\boxed{\text{Weak} = \text{each } P(\text{bad}_n) \to 0; \quad \text{Strong} = P(\text{bad}_n \text{ i.o.}) = 0}
$$

*Key takeaway:* The gap between the two laws is the gap between a sequence tending to zero and a series converging.

### Problem L0.2: What the CLT Does and Does Not Say

A student writes: "By the CLT, for large $n$ the sample mean $\bar{X}_n$ converges to a normal distribution." Identify the two errors and give the correct statement, including why the factor $\sqrt{n}$ is the unique correct scaling.

**Solution**

**Error 1 — the object.** $\bar{X}_n$ itself converges to the *constant* $\mu$ (LLN), so its limiting law is a point mass, not a normal law. The CLT concerns the centered and rescaled deviation.

**Error 2 — the mode.** Even for the rescaled quantity, convergence is *in distribution*, not pointwise or almost sure. Nothing is claimed about the random variables converging, only about their CDFs.

**Correct statement.**

$$
Z_n = \frac{\sqrt{n}\left(\bar{X}_n - \mu\right)}{\sigma} \xrightarrow{d} \mathcal{N}(0,1), \qquad \text{equivalently} \qquad \bar{X}_n \overset{\text{approx}}{\sim} \mathcal{N}\!\left(\mu, \frac{\sigma^2}{n}\right) .
$$

**Uniqueness of the scaling.** Consider $n^{\alpha}(\bar{X}_n - \mu)$, whose variance is $n^{2\alpha}\sigma^2/n = \sigma^2 n^{2\alpha - 1}$. This limit is:

- $0$ if $\alpha \lt 1/2$ (over-shrunk: degenerate at $0$),
- $\infty$ if $\alpha \gt 1/2$ (over-magnified: no proper limit),
- $\sigma^2$ exactly when $\alpha = 1/2$.

Only $\alpha = 1/2$ balances the shrinking fluctuation against the magnification, which is why $\sqrt{n}$ is *the* rate.

$$
\boxed{\sqrt{n}\left(\bar{X}_n - \mu\right) \xrightarrow{d} \mathcal{N}(0, \sigma^2), \text{ while } \bar{X}_n \xrightarrow{P} \mu}
$$

*Key takeaway:* LLN kills the fluctuation; the CLT resurrects it under a microscope of magnification $\sqrt{n}$.

### Problem L0.3: A Chebyshev Bound for Coin Flips

For $n = 100$ fair coin flips let $\hat{p}$ be the observed fraction of heads. Bound $P(\vert \hat{p} - 0.5\vert \ge 0.1)$ using Chebyshev, then compare with the CLT approximation. Comment on the gap.

**Solution**

**Chebyshev.** Each flip is Bernoulli($0.5$) with $\sigma^2 = p(1-p) = 0.25$, so $\operatorname{Var}(\hat{p}) = 0.25/100 = 0.0025$ and

$$
P\left(\left\vert \hat{p} - 0.5 \right\vert \ge 0.1\right) \le \frac{0.0025}{0.1^2} = 0.25 .
$$

**CLT.** The standard error is $\operatorname{sd}(\hat p) = \sqrt{0.0025} = 0.05$, so the threshold sits at $0.1/0.05 = 2$ standard errors:

$$
P\left(\left\vert \hat{p} - 0.5\right\vert \ge 0.1\right) \approx 2\left(1 - \Phi(2)\right) = 2(0.02275) = 0.0455 .
$$

(The exact binomial answer, $P(\vert B - 50\vert \ge 10)$ with $B \sim \text{Bin}(100, 0.5)$, is $\approx 0.0569$; with a continuity correction at $2.5$ standard errors minus half a count the normal approximation lands near $0.057$.)

**The gap.** Chebyshev overestimates by a factor of roughly $5$ because it uses only two moments and must remain valid for the worst distribution with that variance — e.g. a three-point law placing all excess mass exactly at $\pm\varepsilon$. The CLT exploits the actual shape (here, near-Gaussian), obtaining exponential rather than $\varepsilon^{-2}$ decay.

$$
\boxed{\text{Chebyshev} \le 0.25, \quad \text{CLT} \approx 0.046, \quad \text{exact} \approx 0.057}
$$

*Key takeaway:* Chebyshev is a guarantee (valid always, loose usually); the CLT is an approximation (sharp usually, silent about validity).

### Problem L0.4: Which Distributions Are Excluded?

For each summand law, state whether the weak LLN, strong LLN, and CLT apply: (a) $\text{Bernoulli}(0.3)$; (b) $\text{Exponential}(\lambda)$; (c) Pareto with density $f(x) = 1.5 x^{-2.5}$ on $x \gt 1$; (d) standard Cauchy.

**Solution**

Check the two thresholds: $E\vert X\vert \lt \infty$ (both LLNs) and $E[X^2] \lt \infty$ (CLT).

**(a) Bernoulli($0.3$).** Bounded, so all moments are finite: WLLN ✓, SLLN ✓, CLT ✓ with $\mu = 0.3$, $\sigma^2 = 0.21$.

**(b) Exponential($\lambda$).** All moments finite ($E[X^k] = k!/\lambda^k$): WLLN ✓, SLLN ✓, CLT ✓ with $\mu = 1/\lambda$, $\sigma^2 = 1/\lambda^2$.

**(c) Pareto with tail index $\alpha = 1.5$.** Here $f(x) = \alpha x^{-(\alpha+1)}$ with $\alpha = 1.5$. Then

$$
E[X] = \int_1^\infty x \cdot 1.5x^{-2.5}\,dx = 1.5\int_1^\infty x^{-1.5}\,dx = 3 \lt \infty ,
$$

but

$$
E[X^2] = 1.5\int_1^\infty x^{-0.5}\,dx = \infty .
$$

So WLLN ✓, SLLN ✓, **CLT ✗** — the correct limit is a $1.5$-stable law with scaling $n^{1/1.5} = n^{2/3}$, not $\sqrt{n}$.

**(d) Standard Cauchy.** $E\vert X\vert = \frac{2}{\pi}\int_0^\infty \frac{x}{1+x^2}dx = \infty$: **all three fail**. Indeed $\bar{X}_n$ is exactly standard Cauchy for every $n$.

$$
\boxed{\text{(a),(b): all three; (c): LLNs only; (d): none}}
$$

*Key takeaway:* One integral of $\vert x\vert$ and one of $x^2$ decide which limit theorems are available.

## Level 1 — Foundation

### Problem L1.1: Sample Size from Chebyshev, Hoeffding, and the CLT

A pollster wants $P(\vert \hat{p} - p\vert \ge 0.03) \le 0.05$ for an unknown proportion $p$. Determine the required $n$ from (a) Chebyshev, (b) Hoeffding, (c) the CLT, using the worst case over $p$.

**Solution**

The worst-case variance for a Bernoulli sample is $p(1-p) \le 1/4$, attained at $p = 1/2$.

**(a) Chebyshev.**

$$
P\left(\left\vert \hat p - p\right\vert \ge \varepsilon\right) \le \frac{p(1-p)}{n\varepsilon^2} \le \frac{1}{4n\varepsilon^2} \le \delta \iff n \ge \frac{1}{4\delta\varepsilon^2} = \frac{1}{4(0.05)(0.0009)} \approx 5556 .
$$

**(b) Hoeffding.** With summands in $[0,1]$,

$$
2\exp\left(-2n\varepsilon^2\right) \le \delta \iff n \ge \frac{\ln(2/\delta)}{2\varepsilon^2} = \frac{\ln 40}{2(0.0009)} \approx \frac{3.689}{0.0018} \approx 2050 .
$$

**(c) CLT.** Requiring $z_{1-\delta/2}\sqrt{p(1-p)/n} \le \varepsilon$ with $z_{0.975} = 1.96$:

$$
n \ge \frac{z^2 p(1-p)}{\varepsilon^2} \le \frac{(1.96)^2 (0.25)}{0.0009} \approx 1068 .
$$

**Interpretation.** Chebyshev's dependence on $1/\delta$ is the expensive part; Hoeffding replaces it with $\ln(1/\delta)$ and is still a rigorous guarantee; the CLT number is an approximation that happens to be closest to the truth and is the one used in practice (the classic "about 1000 respondents for $\pm 3\%$").

$$
\boxed{n_{\text{Cheb}} \approx 5556, \quad n_{\text{Hoeff}} \approx 2050, \quad n_{\text{CLT}} \approx 1068}
$$

*Key takeaway:* Guarantees cost sample size; the price of rigor is the difference between $1/\delta$, $\log(1/\delta)$, and a Gaussian quantile.

### Problem L1.2: Variance of the Sample Mean under Correlation

Let $X_1, \ldots, X_n$ be identically distributed with variance $\sigma^2$ and equicorrelation $\operatorname{Corr}(X_i, X_j) = \rho$ for $i \ne j$. Compute $\operatorname{Var}(\bar{X}_n)$ and determine the limit as $n \to \infty$. What does this say about the LLN?

**Solution**

Expand the variance of the sum with all covariance terms retained:

$$
\operatorname{Var}\left(\sum_{i=1}^n X_i\right) = \sum_{i=1}^n \operatorname{Var}(X_i) + \sum_{i \ne j}\operatorname{Cov}(X_i, X_j) = n\sigma^2 + n(n-1)\rho\sigma^2 .
$$

Dividing by $n^2$:

$$
\operatorname{Var}(\bar{X}_n) = \frac{\sigma^2}{n}\left[1 + (n-1)\rho\right] = \sigma^2\left(\rho + \frac{1-\rho}{n}\right) .
$$

**Limit.** As $n \to \infty$, $\operatorname{Var}(\bar{X}_n) \to \rho\sigma^2$. For $\rho \gt 0$ this is a *positive constant*: averaging never eliminates the common component, and the LLN fails — the sample mean converges to a random limit, not to $\mu$. Only $\rho = 0$ recovers $\sigma^2/n$. (Positive semi-definiteness also forces $\rho \ge -1/(n-1)$, so strong negative equicorrelation is impossible for large $n$.)

**Effective sample size.** Matching $\sigma^2/n_{\text{eff}}$ to the true variance gives

$$
n_{\text{eff}} = \frac{n}{1 + (n-1)\rho} \xrightarrow{n\to\infty} \frac{1}{\rho} .
$$

Ten thousand highly correlated MCMC draws with $\rho = 0.9$ carry about the information of $\sim 1.1$ independent draws in the equicorrelated idealization; in practice one uses the integrated autocorrelation time.

$$
\boxed{\operatorname{Var}(\bar{X}_n) = \sigma^2\left(\rho + \frac{1-\rho}{n}\right) \to \rho\sigma^2}
$$

*Key takeaway:* Independence is not decoration — correlated data have an information ceiling that more samples cannot break.

### Problem L1.3: Normal Approximation to the Binomial with Continuity Correction

Let $X \sim \text{Bin}(400, 0.25)$. Estimate $P(X \le 90)$ using the CLT with and without a continuity correction, and comment on the accuracy (exact value $\approx 0.1520$).

**Solution**

**Moments.** $\mu = np = 100$, $\sigma^2 = np(1-p) = 400(0.25)(0.75) = 75$, so $\sigma = \sqrt{75} \approx 8.660$.

**Without correction.**

$$
P(X \le 90) \approx \Phi\!\left(\frac{90 - 100}{8.660}\right) = \Phi(-1.1547) \approx 0.1241 .
$$

**With continuity correction.** The discrete event $\{X \le 90\}$ corresponds to the continuous region up to $90.5$:

$$
P(X \le 90) \approx \Phi\!\left(\frac{90.5 - 100}{8.660}\right) = \Phi(-1.0970) \approx 0.1363 .
$$

**Accuracy.** The exact binomial value is $0.1520$. Errors are $0.0279$ (uncorrected) and $0.0157$ (corrected): the half-unit shift roughly halves the error, because it removes the systematic bias from approximating a step function by a smooth one. The residual error is the skewness term of the Edgeworth expansion,

$$
\frac{\gamma}{6\sqrt{n}}(1-x^2)\varphi(x), \qquad \gamma = \frac{1-2p}{\sqrt{p(1-p)}} = \frac{0.5}{0.4330} \approx 1.155 ,
$$

which is nonzero precisely because $p \ne 1/2$ makes the binomial skewed.

$$
\boxed{P(X \le 90) \approx 0.136 \text{ (corrected)}, \quad 0.124 \text{ (uncorrected)}, \quad 0.152 \text{ (exact)}}
$$

*Key takeaway:* Always apply the $\pm \tfrac{1}{2}$ correction when approximating lattice distributions by a continuous one.

### Problem L1.4: CLT for a Sum of Exponentials versus the Exact Gamma Law

Let $X_1, \ldots, X_{50}$ be i.i.d. $\text{Exponential}(1)$ and $S = \sum_i X_i$. Compute the CLT approximation to $P(S \gt 60)$, compare it in structure with the exact Gamma tail, and explain the direction of the error.

**Solution**

**Moments.** For $\text{Exponential}(1)$, $\mu = 1$ and $\sigma^2 = 1$, so $E[S] = 50$ and $\operatorname{Var}(S) = 50$, $\operatorname{sd}(S) = \sqrt{50} \approx 7.071$.

**CLT approximation.**

$$
P(S \gt 60) \approx 1 - \Phi\!\left(\frac{60 - 50}{7.071}\right) = 1 - \Phi(1.4142) \approx 0.0786 .
$$

**Exact law.** $S \sim \text{Gamma}(50, 1)$, whose survival function is the Erlang tail

$$
P(S \gt s) = e^{-s}\sum_{k=0}^{49}\frac{s^k}{k!} , \qquad P(S \gt 60) \approx 0.0877 .
$$

**Direction of the error.** The exponential law has skewness $\gamma = 2 \gt 0$, so the sum's law is right-skewed: it carries more mass in the upper tail than a symmetric Gaussian with the same first two moments. The CLT therefore *underestimates* the right tail (and overestimates the left). The Edgeworth correction predicts the sign and magnitude:

$$
P(Z_n \le x) \approx \Phi(x) + \frac{\gamma}{6\sqrt{n}}\left(1 - x^2\right)\varphi(x), \qquad \frac{2}{6\sqrt{50}}(1 - 2)\varphi(1.414) \approx -0.0072 ,
$$

so the corrected tail is $\approx 0.0786 + 0.0072 = 0.0858$, much closer to $0.0877$.

$$
\boxed{P(S \gt 60): \ \text{CLT} \approx 0.079, \ \text{Edgeworth} \approx 0.086, \ \text{exact} \approx 0.088}
$$

*Key takeaway:* For skewed summands the CLT error is systematic and one-sided in the tails; the first Edgeworth term repairs most of it.

### Problem L1.5: The Delta Method in Action

Let $X_1, \ldots, X_n$ be i.i.d. $\text{Bernoulli}(p)$ with $\hat{p} = \bar{X}_n$. Find the asymptotic distribution of (a) the odds $\hat{p}/(1-\hat{p})$ and (b) the log-odds $\operatorname{logit}(\hat p) = \ln\frac{\hat p}{1-\hat p}$. Then find a transformation whose asymptotic variance does not depend on $p$.

**Solution**

**Base CLT.** $\sqrt{n}(\hat{p} - p) \xrightarrow{d} \mathcal{N}(0, p(1-p))$.

**(a) Odds.** With $g(p) = \frac{p}{1-p}$,

$$
g'(p) = \frac{(1-p) + p}{(1-p)^2} = \frac{1}{(1-p)^2} ,
$$

so by the delta method

$$
\sqrt{n}\left(\frac{\hat p}{1-\hat p} - \frac{p}{1-p}\right) \xrightarrow{d} \mathcal{N}\!\left(0, \frac{p(1-p)}{(1-p)^4}\right) = \mathcal{N}\!\left(0, \frac{p}{(1-p)^3}\right) .
$$

**(b) Log-odds.** With $h(p) = \ln p - \ln(1-p)$,

$$
h'(p) = \frac{1}{p} + \frac{1}{1-p} = \frac{1}{p(1-p)} ,
$$

hence

$$
\sqrt{n}\left(\operatorname{logit}(\hat p) - \operatorname{logit}(p)\right) \xrightarrow{d} \mathcal{N}\!\left(0, \frac{p(1-p)}{p^2(1-p)^2}\right) = \mathcal{N}\!\left(0, \frac{1}{p(1-p)}\right) .
$$

This is why logistic-regression coefficients have standard error $1/\sqrt{n\hat p(1-\hat p)}$ in the intercept-only model, and why logit intervals back-transformed to probabilities respect the $[0,1]$ range.

**(c) Variance stabilization.** Seek $g$ with $g'(p)^2 p(1-p) = 1$, i.e. $g'(p) = \left[p(1-p)\right]^{-1/2}$. Integrating,

$$
g(p) = 2\arcsin\sqrt{p} \quad \Longrightarrow \quad \sqrt{n}\left(2\arcsin\sqrt{\hat p} - 2\arcsin\sqrt{p}\right) \xrightarrow{d} \mathcal{N}(0, 1) .
$$

$$
\boxed{\operatorname{AVar}(\hat p) = \tfrac{p(1-p)}{n}, \quad \operatorname{AVar}(\operatorname{logit}\hat p) = \tfrac{1}{np(1-p)}, \quad \operatorname{AVar}(2\arcsin\sqrt{\hat p}) = \tfrac{1}{n}}
$$

*Key takeaway:* The delta method converts one CLT into a CLT for every smooth functional; choosing the transform can make the variance parameter-free.

### Problem L1.6: Convergence in Probability Implies Convergence in Distribution

Prove that $X_n \xrightarrow{P} X$ implies $X_n \xrightarrow{d} X$, and show by counterexample that the converse fails. Then prove the partial converse when the limit is a constant.

**Solution**

**Forward implication.** Fix a continuity point $x$ of $F_X$ and $\varepsilon \gt 0$. Split on whether $X_n$ and $X$ are close:

$$
\{X_n \le x\} \subseteq \{X \le x + \varepsilon\} \cup \{\vert X_n - X\vert \gt \varepsilon\} ,
$$

because if $X_n \le x$ and $\vert X_n - X\vert \le \varepsilon$ then $X \le x + \varepsilon$. Hence

$$
F_n(x) \le F_X(x+\varepsilon) + P\left(\vert X_n - X\vert \gt \varepsilon\right) .
$$

Symmetrically, $\{X \le x - \varepsilon\} \subseteq \{X_n \le x\} \cup \{\vert X_n - X\vert \gt \varepsilon\}$ gives

$$
F_X(x-\varepsilon) - P\left(\vert X_n - X\vert \gt \varepsilon\right) \le F_n(x) .
$$

Letting $n \to \infty$ kills the probability terms, so

$$
F_X(x - \varepsilon) \le \liminf_n F_n(x) \le \limsup_n F_n(x) \le F_X(x+\varepsilon) .
$$

Now let $\varepsilon \downarrow 0$; continuity of $F_X$ at $x$ squeezes both bounds to $F_X(x)$, giving $F_n(x) \to F_X(x)$. $\blacksquare$

**Counterexample to the converse.** Let $X \sim \mathcal{N}(0,1)$ and set $X_n = -X$ for every $n$. Then $X_n \sim \mathcal{N}(0,1)$ for all $n$, so trivially $X_n \xrightarrow{d} X$. But $\vert X_n - X\vert = 2\vert X\vert$, and $P(2\vert X\vert \gt \varepsilon) \to$ a positive constant, not $0$. Convergence in distribution ignores the joint coupling entirely.

**Partial converse (constant limit).** If $X_n \xrightarrow{d} c$, then $F_n \to \mathbf{1}_{[c,\infty)}$ at all points except $c$, so for $\varepsilon \gt 0$,

$$
P\left(\vert X_n - c\vert \gt \varepsilon\right) \le F_n(c - \varepsilon) + \left(1 - F_n(c + \varepsilon/2)\right) \to 0 + 0 = 0 . \qquad \blacksquare
$$

$$
\boxed{\text{a.s.} \Rightarrow P \Rightarrow d; \quad d \Rightarrow P \text{ only for constant limits}}
$$

*Key takeaway:* Distributional convergence is a statement about laws; probabilistic convergence is a statement about the random variables themselves.

### Problem L2.1: Monte Carlo Integration — Budgeting the Error

Estimate $I = \int_0^1 e^{-x^2}dx$ by Monte Carlo with $X_i \sim \text{Unif}(0,1)$ and $h(x) = e^{-x^2}$. Given $\sigma_h \approx 0.2$, how many samples are needed for a $95\%$ interval of half-width $10^{-3}$? Then show how the antithetic estimator changes the budget.

**Solution**

**Plain Monte Carlo.** The estimator $\hat I_n = \frac{1}{n}\sum_i e^{-X_i^2}$ is unbiased with standard error $\sigma_h/\sqrt{n}$. Requiring $1.96\,\sigma_h/\sqrt{n} \le 10^{-3}$:

$$
n \ge \left(\frac{1.96 \times 0.2}{10^{-3}}\right)^2 = (392)^2 \approx 1.54 \times 10^{5} .
$$

**The scaling law.** Because error $\propto n^{-1/2}$, tightening the interval by $10\times$ costs $100\times$ the samples — deterministic quadrature would be far better here in one dimension, but Monte Carlo's rate is dimension-free, so the comparison reverses beyond $d \approx 4$–$8$.

**Antithetic variates.** Use pairs $\left(h(U_i) + h(1-U_i)\right)/2$ for $n/2$ uniforms. Since $h(x) = e^{-x^2}$ is monotone decreasing on $[0,1]$, $h(U)$ and $h(1-U)$ are negatively correlated, and

$$
\operatorname{Var}\left(\frac{h(U) + h(1-U)}{2}\right) = \frac{\sigma_h^2}{2}\left(1 + \rho\right), \qquad \rho = \operatorname{Corr}\left(h(U), h(1-U)\right) \lt 0 .
$$

Numerically $\rho \approx -0.97$ for this integrand, so the per-pair variance is about $\sigma_h^2(0.03)/2$ and the effective variance reduction versus $n$ independent draws is a factor $\approx 1 + \rho \approx 0.03$: roughly $n \approx 4.6\times 10^{3}$ suffices for the same accuracy — a $30$-fold saving at no extra cost per sample.

$$
\boxed{n_{\text{plain}} \approx 1.5\times 10^5, \qquad n_{\text{antithetic}} \approx 5\times 10^3}
$$

*Key takeaway:* You cannot beat $n^{-1/2}$ by brute force, but you can shrink the constant $\sigma_h$ dramatically with structure-aware estimators.

### Problem L2.2: Minibatch Gradient Noise and the Linear Scaling Rule

In SGD the minibatch gradient is $\hat{g}_B = \frac{1}{B}\sum_{i \in \mathcal{B}}\nabla \ell_i(\theta)$ with i.i.d. sampling from a dataset whose per-example gradient covariance is $\Sigma$. Derive the distribution of the parameter update noise and explain the empirical "linear scaling rule" (double the batch, double the learning rate).

**Solution**

**Unbiasedness and covariance.** Sampling examples uniformly with replacement makes each $\nabla \ell_i$ an i.i.d. draw with mean $g = \nabla L(\theta)$ and covariance $\Sigma$. Hence

$$
E[\hat g_B] = g, \qquad \operatorname{Cov}(\hat g_B) = \frac{\Sigma}{B} .
$$

By the CLT, for moderately large $B$,

$$
\hat{g}_B \approx g + \frac{1}{\sqrt{B}}\Sigma^{1/2}\xi, \qquad \xi \sim \mathcal{N}(0, I) .
$$

**Update noise.** One SGD step with learning rate $\eta$ is $\Delta\theta = -\eta \hat g_B$, so

$$
E[\Delta\theta] = -\eta g, \qquad \operatorname{Cov}(\Delta\theta) = \frac{\eta^2}{B}\Sigma .
$$

**Noise per unit progress.** Over a fixed number of *epochs*, the number of steps is $\propto 1/B$. Accumulating $k = T/B$ independent steps, the drift is $-\eta k g \propto \eta T g/B$ and the accumulated noise covariance is $k \eta^2 \Sigma/B \propto \eta^2 T\Sigma/B^2$. Matching both the drift per epoch and the noise scale across batch sizes requires

$$
\frac{\eta}{B} = \text{const} \quad \Longrightarrow \quad \eta \propto B ,
$$

which is exactly the linear scaling rule. Equivalently, the continuous-time limit of SGD is the SDE

$$
d\theta = -\nabla L(\theta)\,dt + \sqrt{\frac{\eta}{B}}\,\Sigma^{1/2}\,dW ,
$$

whose stationary distribution depends on $\theta$ only through the *temperature* $\eta/B$. Two configurations with equal $\eta/B$ explore the loss landscape identically — the basis for the "large batch needs large learning rate (with warm-up)" practice.

$$
\boxed{\operatorname{Cov}(\hat g_B) = \Sigma/B; \quad \text{SGD temperature} = \eta/B \text{ must be held fixed}}
$$

*Key takeaway:* The CLT turns minibatching into calibrated Gaussian noise whose strength is $\eta/B$ — the single knob controlling SGD's implicit regularization.

### Problem L2.3: A/B Test Sizing for a Model Deployment

Model A has click-through rate $p_A = 0.10$. You want $80\%$ power to detect a lift to $p_B = 0.11$ at two-sided significance $\alpha = 0.05$. Derive the required per-arm sample size using the CLT.

**Solution**

**Test statistic.** With $\hat p_A, \hat p_B$ independent sample proportions from $n$ users each, the CLT gives

$$
\hat p_B - \hat p_A \approx \mathcal{N}\!\left(\delta, \ \frac{p_A(1-p_A) + p_B(1-p_B)}{n}\right), \qquad \delta = p_B - p_A = 0.01 .
$$

**Power condition.** Rejecting when $\vert \hat p_B - \hat p_A\vert \gt z_{1-\alpha/2}\,\text{se}_0$ and requiring power $1-\beta$ under the alternative gives the standard requirement

$$
\delta \ \ge\ \left(z_{1-\alpha/2} + z_{1-\beta}\right)\sqrt{\frac{p_A(1-p_A)+p_B(1-p_B)}{n}} ,
$$

so

$$
n \ \ge\ \frac{\left(z_{1-\alpha/2} + z_{1-\beta}\right)^2\left[p_A(1-p_A) + p_B(1-p_B)\right]}{\delta^2} .
$$

**Numbers.** $z_{0.975} = 1.96$, $z_{0.80} = 0.8416$, sum $= 2.8016$, squared $= 7.849$. Variances: $0.10(0.90) = 0.09$ and $0.11(0.89) = 0.0979$, total $0.1879$. Thus

$$
n \ge \frac{7.849 \times 0.1879}{10^{-4}} \approx 14{,}748 \ \text{per arm} \approx 29{,}500 \text{ total} .
$$

**Sensitivity.** Because $n \propto \delta^{-2}$, halving the detectable lift to $0.005$ quadruples the cost to $\approx 59{,}000$ per arm. This inverse-square wall is why teams switch to variance-reduced metrics (CUPED), stratification, or sequential tests rather than simply waiting longer.

$$
\boxed{n \approx 1.5\times 10^4 \text{ per arm for a } 10\% \to 11\% \text{ lift at } 80\% \text{ power}}
$$

*Key takeaway:* Experiment cost scales as (variance)/(effect size)$^2$ — shrinking metric variance is worth as much as doubling traffic.

### Problem L2.4: Random Walk to Diffusion — Deriving the Heat Kernel

A particle takes $n$ independent steps of length $\pm \ell$ with equal probability every $\tau$ seconds. Use the CLT to derive the position distribution at time $t$ and identify the diffusion coefficient.

**Solution**

**Step statistics.** Each step $S_i$ satisfies $E[S_i] = 0$ and $E[S_i^2] = \ell^2$, so $\sigma^2 = \ell^2$. After $n = t/\tau$ steps the displacement is $X_n = \sum_{i=1}^n S_i$ with

$$
E[X_n] = 0, \qquad \operatorname{Var}(X_n) = n\ell^2 = \frac{\ell^2}{\tau}\,t .
$$

**CLT limit.** Since the steps are i.i.d. with finite variance,

$$
\frac{X_n}{\ell\sqrt{n}} \xrightarrow{d} \mathcal{N}(0,1) \quad \Longrightarrow \quad X(t) \approx \mathcal{N}\!\left(0, \frac{\ell^2}{\tau}t\right) .
$$

Writing the variance as $2Dt$ identifies the diffusion coefficient

$$
D = \frac{\ell^2}{2\tau} ,
$$

and the density is the heat kernel

$$
p(x, t) = \frac{1}{\sqrt{4\pi D t}}\exp\left(-\frac{x^2}{4Dt}\right) ,
$$

which indeed solves $\partial_t p = D\,\partial_x^2 p$.

**Physical reading.** Two consequences follow immediately. First, $\langle x^2\rangle = 2Dt$: diffusive spreading grows as $\sqrt{t}$, not $t$ — the microscopic $\sqrt{n}$ of the CLT becomes the macroscopic $\sqrt{t}$ of diffusion. Second, the limit shape is Gaussian regardless of the step distribution (any zero-mean finite-variance step law gives the same kernel with $D = \sigma^2/(2\tau)$), which is why Fick's law describes systems with wildly different microphysics. Heavy-tailed step lengths (Lévy flights) break this and yield fractional diffusion with $\langle \vert x\vert^\alpha\rangle \propto t$.

$$
\boxed{X(t) \sim \mathcal{N}(0, 2Dt), \quad D = \frac{\ell^2}{2\tau}, \quad \langle x^2\rangle = 2Dt}
$$

*Key takeaway:* Diffusion is the CLT in continuous time; universality of the Gaussian is why one coefficient $D$ summarizes all microscopic detail.

### Problem L2.5: Why He Initialization Keeps Deep Networks Alive

A fully connected layer computes $z_j = \sum_{k=1}^{d} W_{jk} x_k$ with i.i.d. weights of mean $0$ and variance $\sigma_W^2$, independent of the inputs with $E[x_k^2] = q$. Determine the distribution of $z_j$ for large $d$, and derive the variance scaling that keeps signal magnitude constant through a ReLU network.

**Solution**

**Pre-activation law.** The terms $W_{jk}x_k$ are i.i.d. (conditionally on $x$), zero-mean, with variance $\sigma_W^2 q$. By the CLT, for large fan-in $d$,

$$
z_j \approx \mathcal{N}\!\left(0,\ d\,\sigma_W^2 q\right) .
$$

This is why pre-activations at initialization are empirically Gaussian even when the inputs and weights are not — the layer performs the sum that the CLT governs.

**Propagating through ReLU.** Let $q^{(l)} = E[(x^{(l)})^2]$ denote the second moment of layer-$l$ activations. With $z \sim \mathcal{N}(0, v)$ and $x = \max(z, 0)$, symmetry gives

$$
E\left[\max(z,0)^2\right] = \frac{1}{2}E[z^2] = \frac{v}{2} .
$$

Therefore

$$
q^{(l+1)} = \frac{1}{2}\,d_l\,\sigma_W^2\,q^{(l)} .
$$

**Fixed point.** Signal neither explodes nor vanishes across depth iff the multiplier is $1$:

$$
\frac{d_l \sigma_W^2}{2} = 1 \quad \Longrightarrow \quad \sigma_W^2 = \frac{2}{d_l} ,
$$

which is exactly He (Kaiming) initialization. For $\tanh$-like activations with $E[\phi(z)^2] \approx E[z^2]$ near the origin the factor $1/2$ disappears and one recovers Glorot/Xavier scaling $\sigma_W^2 = 1/d_l$ (or $2/(d_{\text{in}} + d_{\text{out}})$ symmetrized for backward passes).

**Why it matters.** With $\sigma_W^2 = c/d$, the layer multiplier is $c/2$ per layer, so after $L$ layers the signal scales as $(c/2)^L$ — exponential in depth. A $10\%$ error in $c$ compounds to a factor $\approx 2.6$ over $L = 10$ and $\approx 1.3\times 10^{4}$ over $L = 100$, which is precisely the vanishing/exploding activation problem.

$$
\boxed{z \approx \mathcal{N}(0, d\sigma_W^2 q); \quad \text{ReLU stability} \iff \sigma_W^2 = 2/d_{\text{in}}}
$$

*Key takeaway:* Initialization schemes are CLT variance bookkeeping — they set the one constant that stops depth from exponentiating.

### Problem L2.6: Control Variates for a Noisy Estimator

You estimate $\theta = E[h(X)]$ by Monte Carlo but also know $E[c(X)] = \mu_c$ exactly for a correlated function $c$. Derive the optimal control-variate estimator, its variance, and the resulting reduction factor. Apply it to reinforcement-learning baselines.

**Solution**

**Estimator family.** For any constant $\beta$, define

$$
\hat\theta_\beta = \frac{1}{n}\sum_{i=1}^n \left[h(X_i) - \beta\left(c(X_i) - \mu_c\right)\right] .
$$

Since $E[c(X) - \mu_c] = 0$, every member is unbiased: $E[\hat\theta_\beta] = \theta$.

**Variance.** Per sample,

$$
\operatorname{Var}\left(h - \beta(c-\mu_c)\right) = \sigma_h^2 - 2\beta\operatorname{Cov}(h,c) + \beta^2\sigma_c^2 .
$$

Minimize over $\beta$ by setting the derivative to zero:

$$
-2\operatorname{Cov}(h,c) + 2\beta\sigma_c^2 = 0 \quad \Longrightarrow \quad \beta^\star = \frac{\operatorname{Cov}(h,c)}{\sigma_c^2} ,
$$

which is exactly the least-squares regression slope of $h$ on $c$. Substituting back,

$$
\operatorname{Var}\left(\hat\theta_{\beta^\star}\right) = \frac{\sigma_h^2}{n}\left(1 - \rho^2\right), \qquad \rho = \operatorname{Corr}(h, c) .
$$

**Reduction factor.** The variance shrinks by $1 - \rho^2$: a control variate correlated at $\rho = 0.9$ removes $81\%$ of the variance (a $5.3\times$ effective sample-size gain), while $\rho = 0.5$ removes only $25\%$. Correlation, not cleverness, is the currency.

**RL baselines.** In policy-gradient methods the estimator $\hat g = \frac{1}{n}\sum_i \nabla_\theta \ln \pi_\theta(a_i \mid s_i)\,R_i$ has enormous variance. Subtracting a state-dependent baseline $b(s)$ gives $\left(R_i - b(s_i)\right)$; because $E\left[\nabla_\theta \ln\pi_\theta(a\mid s)\right] = 0$ for each fixed $s$, the estimator stays unbiased for any $b$. The optimal $b$ is the regression of returns on the score — approximated in practice by the value function $V(s)$, which is why actor-critic and advantage estimation $A = R - V(s)$ work.

$$
\boxed{\beta^\star = \frac{\operatorname{Cov}(h,c)}{\operatorname{Var}(c)}, \qquad \operatorname{Var}(\hat\theta_{\beta^\star}) = \frac{\sigma_h^2(1-\rho^2)}{n}}
$$

*Key takeaway:* Any zero-mean correlated quantity is free variance reduction; the achievable gain is exactly $1 - \rho^2$.

### Problem L3.1: Berry–Esseen and the Rare-Event Sample Size

For i.i.d. $\text{Bernoulli}(p)$ summands with small $p$, compute the Berry–Esseen ratio $\rho/\sigma^3$ and derive how large $n$ must be for the normal approximation to be uniformly accurate to $\epsilon$. Reconcile with the classical rule $np \ge 10$.

**Solution**

**Third absolute central moment.** For $X \sim \text{Bernoulli}(p)$ with $\mu = p$:

$$
\rho = E\left\vert X - p\right\vert^3 = p(1-p)^3 + (1-p)p^3 = p(1-p)\left[(1-p)^2 + p^2\right] .
$$

Since $\sigma^3 = \left[p(1-p)\right]^{3/2}$, the Berry–Esseen ratio is

$$
\frac{\rho}{\sigma^3} = \frac{(1-p)^2 + p^2}{\sqrt{p(1-p)}} .
$$

**Small-$p$ asymptotics.** As $p \to 0$, the numerator tends to $1$ and the denominator to $\sqrt{p}$:

$$
\frac{\rho}{\sigma^3} \sim \frac{1}{\sqrt{p}} \longrightarrow \infty .
$$

**Sample size.** Theorem 2.12 requires

$$
\frac{C\rho}{\sigma^3\sqrt{n}} \le \epsilon \quad \Longrightarrow \quad n \ \ge\ \left(\frac{C}{\epsilon}\right)^2 \frac{\left[(1-p)^2+p^2\right]^2}{p(1-p)} \ \approx\ \frac{C^2}{\epsilon^2 p} \quad (p \ll 1) .
$$

So $n$ must grow like $1/p$, i.e. **$np$ must exceed a constant** depending only on the target accuracy. With $C = 0.4748$ and $\epsilon = 0.05$ this constant is $(0.4748/0.05)^2 \approx 90$; with the sharper practical behavior (Berry–Esseen is conservative) the working threshold drops to the familiar $np \gtrsim 10$, and symmetrically $n(1-p) \gtrsim 10$ by applying the same analysis to $1 - X$.

**Interpretation.** $np$ is the *expected number of successes*: normal approximation needs enough successes, not merely enough trials. When $np$ stays bounded as $n$ grows, the correct limit is not Gaussian at all but Poisson($\lambda = np$) — the law of rare events, a different limit theorem entirely.

$$
\boxed{n \gtrsim \frac{C^2}{\epsilon^2 p} \iff np \gtrsim \text{const}; \text{ otherwise the limit is Poisson, not Gaussian}}
$$

*Key takeaway:* The CLT's validity threshold is governed by a scale-free third moment, and for rare events that quantity blows up like $p^{-1/2}$.

### Problem L3.2: The Cauchy Distribution Defeats Averaging

Let $X_1, X_2, \ldots$ be i.i.d. standard Cauchy with density $f(x) = \frac{1}{\pi(1+x^2)}$ and characteristic function $\varphi(t) = e^{-\vert t\vert}$. (a) Show $E\vert X_1\vert = \infty$. (b) Find the exact law of $\bar{X}_n$. (c) Show that no normalization $a_n$ makes $\bar{X}_n/a_n$ Gaussian, and identify what does converge.

**Solution**

**(a) Non-integrability.** By symmetry,

$$
E\vert X_1\vert = \frac{2}{\pi}\int_0^\infty \frac{x}{1+x^2}\,dx = \frac{1}{\pi}\Big[\ln(1+x^2)\Big]_0^\infty = \infty .
$$

The mean is undefined, so neither LLN has a target to converge to.

**(b) Exact law of the average.** Using independence and the scaling rule $\varphi_{aX}(t) = \varphi_X(at)$,

$$
\varphi_{\bar{X}_n}(t) = \prod_{i=1}^n \varphi_{X_i}\!\left(\frac{t}{n}\right) = \left(e^{-\vert t\vert/n}\right)^{n} = e^{-\vert t\vert} = \varphi_{X_1}(t) .
$$

By uniqueness of characteristic functions, $\bar{X}_n \overset{d}{=} X_1$ for **every** $n$: averaging a million Cauchy observations gives a value distributed exactly like a single observation. The sample mean is worthless as an estimator; the sample *median*, by contrast, is asymptotically normal with variance $\pi^2/(4n)$ and is the practical location estimator.

**(c) No Gaussian limit.** Suppose $\bar{X}_n/a_n \xrightarrow{d} \mathcal{N}(0,1)$ for some $a_n \gt 0$. Then

$$
\varphi_{\bar{X}_n/a_n}(t) = e^{-\vert t\vert/a_n} \longrightarrow e^{-t^2/2} .
$$

At $t$ and $2t$ the left side satisfies $\varphi(2t) = \varphi(t)^2$ for all $n$, whereas the Gaussian satisfies $\varphi(2t) = \varphi(t)^4$; these are incompatible for $t \ne 0$. Concretely, if $a_n \to a \in (0,\infty)$ the limit is Cauchy; if $a_n \to \infty$ the limit is the constant $0$; if $a_n \to 0$ there is no proper limit. In no case is it Gaussian.

**What does converge.** The Cauchy law is $1$-stable: $S_n/n \overset{d}{=} X_1$, matching the general stable scaling $S_n/n^{1/\alpha}$ with $\alpha = 1$. The Gaussian is the special case $\alpha = 2$, available exactly when the variance is finite.

$$
\boxed{\bar{X}_n \overset{d}{=} X_1 \ \forall n; \ \text{Cauchy is } 1\text{-stable, outside the Gaussian domain of attraction}}
$$

*Key takeaway:* Finite variance is not a technicality — without it the sample mean can carry literally no more information than a single observation.

### Problem L3.3: From Chernoff to Hoeffding

Derive Hoeffding's inequality for bounded i.i.d. summands: for $X_i \in [a,b]$ with mean $\mu$,

$$
P\left(\bar{X}_n - \mu \ge \varepsilon\right) \le \exp\left(-\frac{2n\varepsilon^2}{(b-a)^2}\right) .
$$

Compare the resulting sample complexity with Chebyshev's.

**Solution**

**Step 1 — the Chernoff device.** For any $\lambda \gt 0$, monotonicity of $x \mapsto e^{\lambda x}$ and Markov's inequality give

$$
P\left(\sum_{i=1}^n (X_i - \mu) \ge n\varepsilon\right) = P\left(e^{\lambda\sum_i (X_i-\mu)} \ge e^{\lambda n \varepsilon}\right) \le e^{-\lambda n \varepsilon}\, E\!\left[e^{\lambda \sum_i (X_i-\mu)}\right] .
$$

By independence the expectation factorizes into $\prod_i E\left[e^{\lambda(X_i-\mu)}\right] = \left(E\left[e^{\lambda(X_1-\mu)}\right]\right)^n$. Raising to a power is where the exponential gain over Chebyshev originates: Chebyshev uses one moment of the sum, Chernoff uses the whole moment generating function and then optimizes.

**Step 2 — Hoeffding's lemma.** Claim: if $Y$ has mean $0$ and $Y \in [a - \mu, b - \mu]$ with range $R = b-a$, then

$$
E\left[e^{\lambda Y}\right] \le \exp\left(\frac{\lambda^2 R^2}{8}\right) .
$$

*Proof.* Convexity of $e^{\lambda y}$ on the interval gives, writing $y$ as a convex combination of the endpoints $\alpha = a-\mu$ and $\beta = b - \mu$,

$$
e^{\lambda y} \le \frac{\beta - y}{\beta - \alpha}e^{\lambda \alpha} + \frac{y - \alpha}{\beta-\alpha}e^{\lambda\beta} .
$$

Taking expectations and using $E[Y] = 0$ yields $E[e^{\lambda Y}] \le \frac{\beta}{\beta - \alpha}e^{\lambda\alpha} - \frac{\alpha}{\beta-\alpha}e^{\lambda\beta} =: e^{\psi(\lambda)}$. Setting $p = -\alpha/(\beta-\alpha) \in [0,1]$ and $u = \lambda(\beta-\alpha)$,

$$
\psi(\lambda) = -pu + \ln\left(1 - p + pe^{u}\right) ,
$$

which satisfies $\psi(0) = \psi'(0) = 0$ and $\psi''(u) = \frac{p e^u(1-p)}{(1-p+pe^u)^2} \le \frac{1}{4}$ (the maximum of $s(1-s)$ over $s \in [0,1]$). Taylor with remainder gives $\psi(\lambda) \le u^2/8 = \lambda^2R^2/8$. $\square$

**Step 3 — optimize $\lambda$.** Combining,

$$
P\left(\bar X_n - \mu \ge \varepsilon\right) \le \exp\left(-\lambda n \varepsilon + \frac{n\lambda^2 R^2}{8}\right) .
$$

The exponent is a quadratic in $\lambda$, minimized at $\lambda^\star = 4\varepsilon/R^2$, giving exponent $-2n\varepsilon^2/R^2$:

$$
P\left(\bar X_n - \mu \ge \varepsilon\right) \le \exp\left(-\frac{2n\varepsilon^2}{(b-a)^2}\right) . \qquad \blacksquare
$$

**Sample complexity.** To achieve failure probability $\delta$ (two-sided) Hoeffding needs $n \ge \frac{R^2\ln(2/\delta)}{2\varepsilon^2}$, versus Chebyshev's $n \ge \frac{\sigma^2}{\delta\varepsilon^2}$. The dependence on $\delta$ improves from $1/\delta$ to $\ln(1/\delta)$, which is what makes union bounds over $\vert \mathcal{H}\vert$ hypotheses affordable: the cost becomes $\ln\vert\mathcal{H}\vert$ rather than $\vert\mathcal{H}\vert$.

$$
\boxed{n_{\text{Hoeffding}} = O\!\left(\varepsilon^{-2}\log(1/\delta)\right) \text{ versus } n_{\text{Chebyshev}} = O\!\left(\varepsilon^{-2}\delta^{-1}\right)}
$$

*Key takeaway:* Exponentiating before applying Markov converts polynomial tail bounds into exponential ones — the foundation of generalization bounds and bandit algorithms.

### Problem L3.4: When Does a Weighted Average Obey a CLT? (Lindeberg–Feller)

Let $Y_n = \sum_{i=1}^n w_{n,i}\,\xi_i$ where $\xi_i$ are i.i.d. with mean $0$, variance $1$, and $\sum_i w_{n,i}^2 = 1$. State the Lindeberg condition, show it reduces to $\max_i \vert w_{n,i}\vert \to 0$ for bounded $\xi$, and analyze the exponentially weighted average $w_{n,i} \propto \gamma^{n-i}$.

**Solution**

**Setup.** The triangular array $\{X_{n,i} = w_{n,i}\xi_i\}$ has row sums $Y_n$ with $E[Y_n] = 0$ and $\operatorname{Var}(Y_n) = \sum_i w_{n,i}^2 = 1$.

**Lindeberg condition.** For every $\varepsilon \gt 0$,

$$
L_n(\varepsilon) = \sum_{i=1}^n E\left[X_{n,i}^2\,\mathbf{1}\left\{\vert X_{n,i}\vert \gt \varepsilon\right\}\right] \longrightarrow 0 .
$$

**Theorem (Lindeberg–Feller).** If $L_n(\varepsilon) \to 0$ for all $\varepsilon \gt 0$, then $Y_n \xrightarrow{d} \mathcal{N}(0,1)$. The condition says no single term contributes a non-vanishing share of the variance — the sum must be a genuine aggregation of many comparable pieces.

**Reduction for bounded summands.** If $\vert \xi_i\vert \le M$, then $\vert X_{n,i}\vert \le M\vert w_{n,i}\vert \le M\,w_n^\star$ where $w_n^\star = \max_i \vert w_{n,i}\vert$. Whenever $M w_n^\star \lt \varepsilon$ the indicator is identically zero, so $L_n(\varepsilon) = 0$. Hence

$$
w_n^\star = \max_{i \le n} \vert w_{n,i}\vert \to 0 \quad \Longrightarrow \quad Y_n \xrightarrow{d}\mathcal{N}(0,1) .
$$

For the plain average $w_{n,i} = 1/\sqrt{n}$, $w_n^\star = n^{-1/2}\to 0$ — recovering the classical CLT.

**Exponentially weighted average.** Take $w_{n,i} = c_n\gamma^{\,n-i}$ with fixed $\gamma \in (0,1)$, normalized so that

$$
c_n^2\sum_{k=0}^{n-1}\gamma^{2k} = c_n^2\,\frac{1-\gamma^{2n}}{1-\gamma^2} = 1 \quad \Longrightarrow \quad c_n \to \sqrt{1-\gamma^2} .
$$

The largest weight is the most recent one, $w_{n,n} = c_n \to \sqrt{1-\gamma^2} \gt 0$, so $w_n^\star \not\to 0$ and the Lindeberg condition **fails**. Indeed the limit is $Y_\infty = \sqrt{1-\gamma^2}\sum_{k\ge 0}\gamma^k \xi_{n-k}$, an infinite weighted sum whose law is *not* Gaussian unless $\xi$ already is — it is a convolution dominated by the few most recent terms.

**Consequence for optimizers.** Momentum and Adam maintain exponentially weighted gradient averages with $\gamma \approx 0.9$–$0.999$. Their effective sample size is

$$
n_{\text{eff}} = \frac{1}{\sum_i w_{n,i}^2 / \left(\sum_i w_{n,i}\right)^2}\Bigg|_{\text{normalized}} \approx \frac{1+\gamma}{1-\gamma} ,
$$

roughly $19$ for $\gamma = 0.9$ and $1999$ for $\gamma = 0.999$. Noise reduction saturates at that finite level no matter how long training runs — momentum averages a *window*, not the whole history.

$$
\boxed{\text{CLT holds iff } \max_i \vert w_{n,i}\vert \to 0; \ \text{EMA has } n_{\text{eff}} \approx \tfrac{1+\gamma}{1-\gamma} \text{ and no CLT}}
$$

*Key takeaway:* Universality requires democratic weights; whenever a few terms dominate the variance the Gaussian limit disappears and the effective sample size stays bounded.
