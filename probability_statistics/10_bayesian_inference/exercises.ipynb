### Problem L0.1: The Base Rate Dominates — A Screening Test

A disease affects $1$ in $1000$ people. A test has sensitivity $99\%$ and specificity $99\%$. A randomly screened person tests positive. Compute the posterior probability of disease and explain why the answer surprises most people.

**Solution**

**Set up.** Let $D$ be disease and $+$ a positive test. Given: $P(D) = 0.001$, $P(+\mid D) = 0.99$, $P(+\mid D^c) = 0.01$.

**Evidence.** By the law of total probability,

$$
P(+) = P(+\mid D)P(D) + P(+\mid D^c)P(D^c) = 0.99(0.001) + 0.01(0.999) = 0.00099 + 0.00999 = 0.01098 .
$$

**Posterior.**

$$
P(D\mid +) = \frac{P(+\mid D)P(D)}{P(+)} = \frac{0.00099}{0.01098} \approx 0.0902 .
$$

**Why it surprises.** Intuition substitutes $P(+\mid D) = 0.99$ for $P(D\mid +)$ — the classic *base-rate neglect* or confusion of the conditional. The odds form makes the mechanism transparent:

$$
\underbrace{\frac{P(D\mid+)}{P(D^c\mid+)}}_{\text{posterior odds } \approx 1:10} = \underbrace{\frac{0.99}{0.01}}_{\text{likelihood ratio } = 99} \times \underbrace{\frac{0.001}{0.999}}_{\text{prior odds } \approx 1:999} .
$$

A likelihood ratio of $99$ cannot overcome prior odds of $1{:}999$; it lands at about $1{:}10$. In a population of $100{,}000$: $100$ have the disease and $99$ test positive, while $99{,}900$ are healthy and $999$ still test positive — false positives outnumber true positives $10{:}1$.

**What fixes it.** A second independent test multiplies the odds by $99$ again, giving posterior odds $\approx 9.8{:}1$, i.e. $P(D\mid ++) \approx 0.91$. This is exactly sequential updating: the first posterior is the second test's prior.

$$
\boxed{P(D\mid +) \approx 0.090, \qquad P(D\mid ++) \approx 0.91}
$$

*Key takeaway:* Posterior odds = likelihood ratio $\times$ prior odds; a strong test cannot rescue a rare hypothesis in one step.

### Problem L0.2: Credible Interval versus Confidence Interval

State precisely what a $95\%$ credible interval and a $95\%$ confidence interval each claim, and construct a scenario where they differ substantially.

**Solution**

**Credible interval.** A set $C$ with

$$
P\left(\theta\in C\mid x\right) = \int_C p(\theta\mid x)\,d\theta = 0.95 .
$$

The data $x$ are fixed and observed; $\theta$ is random. The statement "there is a $95\%$ probability that $\theta$ lies in $C$" is literally correct.

**Confidence interval.** A random set $C(X)$ with

$$
P_\theta\left(\theta\in C(X)\right) = 0.95 \quad\text{for every } \theta .
$$

Here $\theta$ is fixed and $X$ is random. The guarantee is about the *procedure* across hypothetical repetitions; once you observe $x$, the realized interval either contains $\theta$ or it does not, and no probability attaches to that particular interval.

**A scenario where they differ.** Observe one Bernoulli trial that comes up heads ($n = 1$, $k = 1$).

- A Bayesian with a $\text{Beta}(1,1)$ prior has posterior $\text{Beta}(2,1)$, density $2p$, so the equal-tailed $95\%$ interval solves $F(p) = p^2$ at $0.025$ and $0.975$, giving $(0.158,\ 0.987)$ — informative and honest.
- The Clopper–Pearson confidence interval is $\left(0.025^{1},\ 1\right) = (0.025, 1)$, far wider, because it must guarantee coverage for every $p$ simultaneously and no data-dependent probability statement is allowed.

Conversely, priors can hurt: with a strong $\text{Beta}(50,1)$ prior the credible interval would sit near $1$ regardless of a single contrary observation, and its frequentist coverage at $p = 0.5$ would be near zero.

**Reconciliation.** By Bernstein–von Mises the two agree asymptotically: both converge to $\hat\theta \pm 1.96/\sqrt{nI(\hat\theta)}$. They differ exactly where it matters — small $n$, boundaries, and hierarchical models.

$$
\boxed{P(\theta\in C\mid x) = 0.95 \ \text{(credible)} \quad\text{versus}\quad P_\theta(\theta\in C(X)) = 0.95 \ \text{(confidence)}}
$$

*Key takeaway:* Credibility conditions on the data you have; confidence averages over data you did not see.

### Problem L0.3: A Flat Prior Is Not an Absence of Information

Take $\theta\sim\text{Unif}(0,1)$ and let $\phi = \theta^2$. Derive the induced prior on $\phi$ and explain why "uninformative" is parameterization-dependent. What is the invariant alternative?

**Solution**

**Induced prior.** With $\phi = \theta^2$ on $(0,1)$, the inverse is $\theta = \sqrt{\phi}$ and $\frac{d\theta}{d\phi} = \frac{1}{2\sqrt\phi}$. By change of variables,

$$
p_\phi(\phi) = p_\theta\left(\sqrt\phi\right)\left\vert \frac{d\theta}{d\phi}\right\vert = 1\cdot\frac{1}{2\sqrt\phi} = \frac{1}{2\sqrt{\phi}}, \qquad 0\lt\phi\lt1 .
$$

This is *not* flat: it diverges at $0$ and puts $P(\phi \lt 0.25) = \sqrt{0.25} = 0.5$, i.e. half the mass in the lowest quarter of the range. A statement of ignorance about $\theta$ is a strong statement of belief about $\phi$ — and $\phi$ might be the physically meaningful quantity (an area, an intensity, a variance).

**Worse cases.** Flat priors on unbounded parameters are *improper* ($\int d\theta = \infty$). Sometimes the posterior is still proper (a flat prior on a Gaussian mean gives $\mathcal{N}(\bar x,\sigma^2/n)$), but sometimes it is not — the classic failure is a flat prior on the variance of a hierarchical group-level effect, which yields an improper posterior that MCMC will happily and silently sample from without ever converging.

**The invariant alternative.** Jeffreys' prior,

$$
p_J(\theta)\ \propto\ \sqrt{\det I(\theta)} ,
$$

transforms correctly: under $\phi = g(\theta)$, $I(\phi) = I(\theta)\left(\frac{d\theta}{d\phi}\right)^2$, so $\sqrt{I(\phi)} = \sqrt{I(\theta)}\left\vert \frac{d\theta}{d\phi}\right\vert$ — exactly the Jacobian factor a density must acquire. For a Bernoulli rate, $I(p) = \frac{1}{p(1-p)}$ gives $p_J \propto p^{-1/2}(1-p)^{-1/2} = \text{Beta}(1/2,1/2)$; for a scale parameter, $p_J(\sigma)\propto1/\sigma$, i.e. uniform in $\ln\sigma$.

**Practical advice.** Prefer *weakly informative* priors chosen on an interpretable scale (e.g. $\mathcal{N}(0,2.5^2)$ on standardized regression coefficients) and check them with prior predictive simulation, rather than reaching for flatness as a proxy for objectivity.

$$
\boxed{\theta\sim\text{Unif}(0,1) \Rightarrow p_\phi(\phi) = \tfrac{1}{2\sqrt\phi}: \text{ flatness is not invariant}}
$$

*Key takeaway:* There is no such thing as "no prior"; there is only a prior you have or have not examined on the scale that matters.

### Problem L0.4: Batch Equals Stream

You observe $8$ heads in $10$ flips, then $3$ heads in $5$ more, starting from a $\text{Beta}(2,2)$ prior. Show that updating twice gives the same posterior as updating once on all $15$ flips, and explain why order does not matter.

**Solution**

**Sequential.** First update:

$$
\text{Beta}(2,2) + (8,\ 2) \ \longrightarrow\ \text{Beta}(2+8,\ 2+2) = \text{Beta}(10, 4) .
$$

Using that posterior as the new prior for the second batch ($3$ heads, $2$ tails):

$$
\text{Beta}(10,4) + (3,\ 2)\ \longrightarrow\ \text{Beta}(13, 6) .
$$

**Batch.** All $15$ flips at once give $11$ heads and $4$ tails:

$$
\text{Beta}(2,2) + (11,\ 4)\ \longrightarrow\ \text{Beta}(2+11,\ 2+4) = \text{Beta}(13,6) . \ \checkmark
$$

**Why, in general.** With conditional independence given $\theta$,

$$
p\left(\theta\mid x_{1:n}\right)\ \propto\ p(\theta)\prod_{i=1}^n p(x_i\mid\theta) ,
$$

and a product is invariant under both grouping (associativity) and reordering (commutativity). Any partition of the data into batches, processed in any order, yields the same posterior. Structurally: the hyperparameters accumulate the sufficient statistics, here $\left(\sum_i x_i,\ n\right)$, and sums do not care about order.

**Where the property is used.** Recursive Bayesian filters (Kalman, particle filters) process one measurement at a time with no loss; online learning updates a posterior per example; distributed inference can shard data, accumulate sufficient statistics locally, and combine exactly. Note the property depends on exchangeability — with time-dependent parameters $\theta_t$ one must insert a prediction step between updates, which is precisely what turns Bayes' rule into a filter.

$$
\boxed{\text{Beta}(2,2) \xrightarrow{\ (8,2)\ }\text{Beta}(10,4)\xrightarrow{\ (3,2)\ }\text{Beta}(13,6) = \text{Beta}(2,2)\xrightarrow{\ (11,4)\ }}
$$

*Key takeaway:* Sequential and batch Bayesian updating are identical because likelihoods multiply and sufficient statistics add.

## L1 — Foundations

### Problem L1.1: A Complete Beta–Binomial Analysis

A new drug succeeds in $7$ of $20$ patients. Using a $\text{Beta}(2,2)$ prior, derive the posterior, its mean, mode and standard deviation, an approximate $95\%$ credible interval, and $P(p \gt 0.5\mid x)$.

**Solution**

**Posterior.** By conjugacy with $k = 7$, $n = 20$:

$$
p\mid x \ \sim\ \text{Beta}(\alpha+k,\ \beta+n-k) = \text{Beta}(9,\ 15) .
$$

**Summaries.** For $\text{Beta}(a,b)$ with $a = 9$, $b = 15$, $a+b = 24$:

$$
E[p\mid x] = \frac{a}{a+b} = \frac{9}{24} = 0.375, \qquad \text{mode} = \frac{a-1}{a+b-2} = \frac{8}{22} \approx 0.364 ,
$$

$$
\operatorname{Var}(p\mid x) = \frac{ab}{(a+b)^2(a+b+1)} = \frac{135}{576\cdot25} = 0.009375, \qquad \operatorname{sd} \approx 0.0968 .
$$

Compare the MLE $\hat p = 7/20 = 0.35$: the posterior mean is pulled toward the prior mean $0.5$ by the weight $\frac{m}{m+n} = \frac{4}{24} = \tfrac16$, i.e. $0.35 + \tfrac16(0.5-0.35) = 0.375$ ✓.

**Credible interval.** The posterior is close to Gaussian here ($a, b$ both comfortably above $1$), so

$$
0.375 \pm 1.96(0.0968) = (0.185,\ 0.565) .
$$

The exact equal-tailed interval from the Beta quantile function is $(0.196,\ 0.570)$ — the normal approximation is good to about $0.01$, with the small discrepancy caused by the posterior's mild right skew.

**Posterior probability of exceeding $0.5$.** Using the normal approximation,

$$
P(p \gt 0.5\mid x) \approx 1 - \Phi\!\left(\frac{0.5-0.375}{0.0968}\right) = 1 - \Phi(1.291) \approx 0.098 ,
$$

with the exact Beta value $\approx 0.106$. So the data leave roughly a $10\%$ posterior chance that the drug helps more often than not — a direct, decision-ready statement that no $p$-value provides.

$$
\boxed{p\mid x\sim\text{Beta}(9,15),\ E = 0.375,\ \text{sd} = 0.097,\ \text{CI} \approx (0.196, 0.570),\ P(p\gt0.5\mid x)\approx0.11}
$$

*Key takeaway:* Conjugacy turns inference into arithmetic on hyperparameters, and every question becomes an integral of one known density.

### Problem L1.2: Combining a Prior Measurement with New Data (Normal–Normal)

A physical constant has a published estimate $\mu_0 = 10.0$ with standard error $0.5$. You take $n = 25$ new measurements with sample mean $\bar{x} = 10.4$ and known measurement noise $\sigma = 1.0$. Compute the posterior and interpret the weights.

**Solution**

**Prior and likelihood precisions.**

$$
\lambda_0 = \frac{1}{\tau_0^2} = \frac{1}{0.25} = 4, \qquad \lambda_{\text{data}} = \frac{n}{\sigma^2} = \frac{25}{1} = 25 .
$$

**Posterior precision and variance.**

$$
\lambda_n = \lambda_0 + \lambda_{\text{data}} = 29, \qquad \tau_n^2 = \frac{1}{29} \approx 0.03448, \qquad \tau_n \approx 0.1857 .
$$

**Posterior mean.**

$$
\mu_n = \frac{\lambda_0\mu_0 + \lambda_{\text{data}}\bar x}{\lambda_n} = \frac{4(10.0) + 25(10.4)}{29} = \frac{40 + 260}{29} = \frac{300}{29} \approx 10.345 .
$$

**Interpretation of the weights.** The new data carry $25/29 \approx 86\%$ of the total precision, so the posterior sits close to $\bar x$; the prior contributes $14\%$ and pulls the estimate down by $0.055$. Equivalently, the prior is worth $\lambda_0\sigma^2/1 = 4$ observations — a published standard error of $0.5$ with per-measurement noise $1.0$ is exactly the information content of $4$ measurements.

**Credible interval.**

$$
10.345 \pm 1.96(0.1857) = (9.981,\ 10.709) .
$$

Note that the posterior standard deviation $0.186$ is smaller than *both* the prior's $0.5$ and the data's standard error $\sigma/\sqrt n = 0.2$: combining independent information always sharpens, since precisions add.

**Consistency check.** Setting $\tau_0\to\infty$ ($\lambda_0 \to 0$) gives $\mu_n \to \bar x = 10.4$ and $\tau_n^2\to\sigma^2/n = 0.04$, recovering the pure-likelihood answer.

$$
\boxed{\mu\mid x\sim\mathcal{N}(10.345,\ 0.1857^2), \qquad \text{CI} = (9.98,\ 10.71)}
$$

*Key takeaway:* Bayesian updating with Gaussians is inverse-variance weighting — the same formula physicists use to combine independent measurements, derived rather than assumed.

### Problem L1.3: Counting Rare Events (Gamma–Poisson)

A detector observes $x = 3$ events in $t = 2$ hours. With a $\text{Gamma}(a=2, b=1)$ prior on the rate $\lambda$ (events per hour), derive the posterior, a credible interval, and the predictive distribution for the next hour.

**Solution**

**Posterior.** For a Poisson process with exposure $t$, the likelihood kernel is $\lambda^{x}e^{-\lambda t}$, so

$$
p(\lambda\mid x)\ \propto\ \lambda^{a+x-1}e^{-(b+t)\lambda} \quad\Longrightarrow\quad \lambda\mid x \sim \text{Gamma}\left(a+x,\ b+t\right) = \text{Gamma}(5,\ 3) .
$$

**Summaries.** For $\text{Gamma}(a', b')$ with $a' = 5$, $b' = 3$:

$$
E[\lambda\mid x] = \frac{a'}{b'} = \frac53 \approx 1.667, \qquad \operatorname{Var} = \frac{a'}{b'^2} = \frac59 \approx 0.556, \qquad \operatorname{sd}\approx 0.745 .
$$

The MLE is $x/t = 1.5$; the prior (worth $a = 2$ events in $b = 1$ hour, mean $2$) pulls upward to $1.667$.

**Credible interval.** Since $2b'\lambda\mid x\sim\chi^2_{2a'} = \chi^2_{10}$, the exact equal-tailed $95\%$ interval is

$$
\left(\frac{\chi^2_{10,0.025}}{2b'},\ \frac{\chi^2_{10,0.975}}{2b'}\right) = \left(\frac{3.247}{6},\ \frac{20.483}{6}\right) = (0.541,\ 3.414) .
$$

Wide, as it must be after only three events.

**Posterior predictive for the next hour.** By Proof 3.4 with exposure $1$,

$$
\tilde x\mid x\ \sim\ \text{NB}\left(a' = 5,\ \frac{b'}{b'+1} = \frac34\right), \qquad P(\tilde x = j) = \binom{j+4}{j}\left(\frac34\right)^5\left(\frac14\right)^j .
$$

Its mean is $a'/b' = 1.667$ (matching the plug-in Poisson) but its variance is

$$
\frac{a'}{b'}\left(1+\frac{1}{b'}\right) = 1.667\times\frac43 = 2.222 \ \gt\ 1.667 ,
$$

overdispersed by exactly the epistemic term. For example $P(\tilde x = 0) = (3/4)^5 = 0.2373$, versus the plug-in Poisson's $e^{-1.667} = 0.189$: accounting for rate uncertainty makes a quiet hour noticeably more likely.

$$
\boxed{\lambda\mid x\sim\text{Gamma}(5,3),\ \ \text{CI} = (0.54, 3.41),\ \ \tilde x\mid x\sim\text{NB}(5, 3/4)}
$$

*Key takeaway:* Integrating out the rate converts a Poisson predictive into a negative binomial — parameter uncertainty is exactly overdispersion.

### Problem L1.4: Dirichlet–Categorical and Smoothing for Language Models

A corpus contains word counts $(n_1,\ldots,n_K)$ over a vocabulary of size $K$. With a symmetric $\text{Dirichlet}(\alpha)$ prior, derive the posterior, the posterior mean, and the predictive probability of the next word. Relate this to add-$\alpha$ smoothing.

**Solution**

**Posterior.** The Dirichlet density is $p(\pi)\propto\prod_k \pi_k^{\alpha-1}$ on the simplex and the categorical likelihood contributes $\prod_k \pi_k^{n_k}$, so

$$
p(\pi\mid n)\ \propto\ \prod_{k=1}^K \pi_k^{n_k+\alpha-1} \quad\Longrightarrow\quad \pi\mid n\ \sim\ \text{Dirichlet}\left(n_1+\alpha,\ldots,n_K+\alpha\right) .
$$

**Posterior mean.** For $\text{Dirichlet}(a_1,\ldots,a_K)$ with $A = \sum_k a_k$, $E[\pi_k] = a_k/A$. With $N = \sum_k n_k$,

$$
E\left[\pi_k\mid n\right] = \frac{n_k+\alpha}{N + K\alpha} .
$$

**Predictive.** The next word's probability is exactly this posterior mean, because

$$
P\left(w_{N+1} = k\mid n\right) = \int \pi_k\,p(\pi\mid n)\,d\pi = E[\pi_k\mid n] = \frac{n_k+\alpha}{N+K\alpha} .
$$

This *is* add-$\alpha$ (Lidstone) smoothing: $\alpha = 1$ gives Laplace smoothing $\frac{n_k+1}{N+K}$ and $\alpha = 0.5$ gives Krichevsky–Trofimov. The MLE $n_k/N$ is the $\alpha\to0$ limit.

**Why smoothing is mandatory.** An unseen word has $n_k = 0$, so the MLE assigns probability $0$ and the model's log-likelihood on any test set containing it is $-\infty$; perplexity $= \exp\left(-\frac1M\sum\ln p\right)$ becomes infinite. The Dirichlet prior guarantees $P \ge \frac{\alpha}{N+K\alpha} \gt 0$.

**Cost of the prior.** The total pseudo-count is $K\alpha$, which for a $50{,}000$-word vocabulary and $\alpha = 1$ is $50{,}000$ — potentially larger than the corpus itself for small $N$, badly flattening the distribution. This is why practical language models use small $\alpha$, or better, backoff and Kneser–Ney interpolation (which correspond to hierarchical Pitman–Yor priors — the same Bayesian machinery with a heavier-tailed base measure matching Zipf's law).

$$
\boxed{\pi\mid n\sim\text{Dirichlet}(n+\alpha), \qquad P(\text{next} = k) = \frac{n_k+\alpha}{N+K\alpha}}
$$

*Key takeaway:* Every smoothing recipe in language modeling is a Dirichlet-family prior; choosing $\alpha$ is choosing how many pseudo-observations to spread over the vocabulary.

### Problem L1.5: The Posterior Predictive Is Wider Than the Plug-In

For a $\text{Beta}(a,b)$ posterior on a coin's bias, derive the distribution of the number of heads $\tilde{k}$ in $m$ future flips, and compare its variance with the plug-in binomial.

**Solution**

**Predictive density.** Integrate the binomial likelihood against the Beta posterior:

$$
P(\tilde k = j\mid x) = \int_0^1 \binom{m}{j}p^j(1-p)^{m-j}\frac{p^{a-1}(1-p)^{b-1}}{B(a,b)}dp = \binom{m}{j}\frac{1}{B(a,b)}\int_0^1 p^{a+j-1}(1-p)^{b+m-j-1}dp .
$$

The integral is a Beta function, giving the **beta-binomial** distribution

$$
P(\tilde k = j\mid x) = \binom{m}{j}\frac{B(a+j,\ b+m-j)}{B(a,b)}, \qquad j = 0,\ldots,m .
$$

**Moments.** With $\mu = \frac{a}{a+b}$ (posterior mean of $p$),

$$
E[\tilde k\mid x] = m\mu, \qquad \operatorname{Var}(\tilde k\mid x) = m\mu(1-\mu)\,\underbrace{\frac{a+b+m}{a+b+1}}_{\ \ge\ 1} .
$$

**Derivation of the variance by total variance.** Conditioning on $p$,

$$
\operatorname{Var}(\tilde k\mid x) = \underbrace{E\left[mp(1-p)\mid x\right]}_{\text{aleatoric}} + \underbrace{\operatorname{Var}\left(mp\mid x\right)}_{\text{epistemic}} = mE[p(1-p)] + m^2\operatorname{Var}(p) ,
$$

and substituting $E[p(1-p)] = \mu(1-\mu)\frac{a+b}{a+b+1}$, $\operatorname{Var}(p) = \frac{\mu(1-\mu)}{a+b+1}$ reproduces the formula.

**Comparison with plug-in.** The plug-in predictive $\text{Binomial}(m,\hat p)$ has variance $m\mu(1-\mu)$. The inflation factor is

$$
\frac{a+b+m}{a+b+1} \ \approx\ 1 + \frac{m}{n} \quad \text{for } a+b\approx n \gg 1 .
$$

Predicting one future flip ($m = 1$) costs almost nothing, but predicting a batch as large as the training set ($m = n$) doubles the predictive variance. Plug-in prediction is therefore most dangerous exactly where it is most used: forecasting aggregates from a small calibration sample.

$$
\boxed{\tilde k\mid x\sim\text{BetaBin}(m,a,b), \quad \operatorname{Var} = m\mu(1-\mu)\frac{a+b+m}{a+b+1} \ \gt\ m\mu(1-\mu)}
$$

*Key takeaway:* The posterior predictive splits into aleatoric plus epistemic variance; the plug-in estimate silently deletes the second term.

### Problem L1.6: Choosing an Estimator by Choosing a Loss

A posterior for a demand parameter is $\text{Gamma}(5,3)$. You must order inventory $a$; each unsold (overage) unit costs $c_o = 1$ and each unmet (underage) unit of demand costs $c_u = 4$. Find the optimal $a$ and compare it with the posterior mean and median.

**Solution**

**Loss.** The asymmetric ("newsvendor" or pinball) loss, in the standard convention where $c_o$ is the overage (unsold) cost and $c_u$ the underage (unmet-demand) cost, is

$$
L(\theta, a) = c_o\left(a-\theta\right)_+ + c_u\left(\theta - a\right)_+ = 1\cdot(a-\theta)_+ + 4\cdot(\theta-a)_+ .
$$

**Minimization.** Write the posterior expected loss and differentiate as in Proof 3.5:

$$
\rho(a) = c_o\int_{-\infty}^a (a-\theta)p\,d\theta + c_u\int_a^\infty(\theta-a)p\,d\theta, \qquad \rho'(a) = c_o F(a) - c_u\left(1-F(a)\right) .
$$

Setting $\rho' = 0$:

$$
F(a^\star) = \frac{c_u}{c_u+c_o} = \frac{4}{5} = 0.8 .
$$

The optimum is the posterior **$0.8$-quantile**, not the mean. Convexity of $\rho$ (its derivative is non-decreasing) makes this the global minimum.

**Numbers.** For $\text{Gamma}(5,3)$, use $2b\theta\mid x \sim\chi^2_{2a} = \chi^2_{10}$: the $0.8$ quantile of $\chi^2_{10}$ is $13.442$, so

$$
a^\star = \frac{13.442}{6} \approx 2.240 .
$$

Compare:

$$
E[\theta\mid x] = \frac{5}{3} \approx 1.667, \qquad \text{median} = \frac{\chi^2_{10,0.5}}{6} = \frac{9.342}{6}\approx 1.557, \qquad \text{mode} = \frac{a'-1}{b'} = \frac43\approx1.333 .
$$

**Interpretation.** Because stockouts cost four times as much as overstock, the optimal order sits well above every symmetric summary — $34\%$ above the posterior mean and $68\%$ above the mode. Reporting the mean and letting a downstream process "add a safety margin" reinvents this calculation badly; the loss function belongs inside the decision.

$$
\boxed{a^\star = F^{-1}\!\left(\tfrac{c_o}{c_o+c_u}\right) = F^{-1}(0.8) \approx 2.24 \ \gg\ E[\theta\mid x] = 1.67}
$$

*Key takeaway:* The right point estimate is a property of the loss, not of the posterior; asymmetric costs give quantiles, and quantiles can be far from the center.

## L2 — Applications (AI/ML and Physics)

### Problem L2.1: Thompson Sampling for a Two-Armed Bandit

Arm A has $\text{Beta}(31, 71)$ posterior (30 successes in 100 pulls, uniform prior) and arm B has $\text{Beta}(8, 4)$ (7 successes in 10). Compute each arm's posterior mean and standard deviation, describe the Thompson-sampling decision, and explain why it explores automatically.

**Solution**

**Posterior summaries.**

$$
E[p_A] = \frac{31}{102} \approx 0.304, \quad \operatorname{sd}_A = \sqrt{\frac{31\cdot71}{102^2\cdot103}} \approx 0.0453 ,
$$

$$
E[p_B] = \frac{8}{12} \approx 0.667, \quad \operatorname{sd}_B = \sqrt{\frac{8\cdot4}{12^2\cdot13}} \approx 0.1307 .
$$

Arm B looks better but is known far less precisely — its posterior is nearly three times wider.

**Thompson sampling.** At each round draw

$$
\tilde p_A \sim \text{Beta}(31,71), \qquad \tilde p_B\sim\text{Beta}(8,4) ,
$$

and pull $\arg\max_k \tilde p_k$. The probability of choosing an arm is exactly its posterior probability of being best,

$$
P\left(\text{pull } k\right) = P\left(p_k = \max_j p_j \,\middle|\, \text{data}\right) .
$$

A normal approximation (means differ by $0.36$ against a combined sd of $\sqrt{0.0453^2+0.1307^2} = 0.138$, i.e. $2.6$ standard deviations) suggests $P(p_B \gt p_A) \approx 0.995$, but Beta$(8,4)$ is visibly skewed at these parameters, so that approximation is not reliable to three digits. The exact value, $P(p_B \gt p_A) = \int_0^1 f_{\text{Beta}(8,4)}(x)\,F_{\text{Beta}(31,71)}(x)\,dx \approx 0.993$, is close but not identical — B is pulled almost always, but not *always*.

**Why exploration is automatic.** Three regimes fall out of one rule:

1. **Wide posterior** (few pulls) $\Rightarrow$ samples scatter widely $\Rightarrow$ the arm is tried even if its mean is lower. Exploration is proportional to uncertainty.
2. **Narrow posterior** (many pulls) $\Rightarrow$ samples concentrate $\Rightarrow$ behavior becomes greedy.
3. **Evidence accumulates** on any arm that is pulled, so a genuinely inferior arm's posterior narrows around a low value and it stops being sampled.

No $\epsilon$ parameter, no annealing schedule, no confidence-bonus constant needs tuning — contrast with $\epsilon$-greedy (explores uniformly forever) and UCB (needs a bonus scale). Thompson sampling achieves $O(\log T)$ regret, matching the Lai–Robbins lower bound up to constants.

**Update.** After pulling B and observing a success, $\text{Beta}(8,4)\to\text{Beta}(9,4)$ — one line, and the loop repeats.

$$
\boxed{P(\text{pull } k) = P\left(p_k \text{ is best}\mid \text{data}\right); \text{ exploration} \propto \text{posterior width}}
$$

*Key takeaway:* Sampling from the posterior converts uncertainty directly into exploration — the cleanest example of a decision rule that needs no tuning knob.

### Problem L2.2: Bayesian Linear Regression — Posterior and Predictive Variance

For $y = Xw+\varepsilon$, $\varepsilon\sim\mathcal{N}(0,\sigma^2 I)$, with prior $w\sim\mathcal{N}(0,\tau^2 I)$, derive the exact posterior over $w$ and the predictive distribution at a new input $x_\star$. Interpret the predictive variance geometrically.

**Solution**

**Posterior.** The log-posterior is quadratic in $w$:

$$
\ln p(w\mid y) = -\frac{1}{2\sigma^2}\Vert y - Xw\Vert^2 - \frac{1}{2\tau^2}\Vert w\Vert^2 + C .
$$

Expanding and collecting terms in $w$,

$$
-\frac12\left[w^{\top}\underbrace{\left(\frac{X^{\top}X}{\sigma^2}+\frac{I}{\tau^2}\right)}_{=\,A}w - \frac{2}{\sigma^2}w^{\top}X^{\top}y\right] + C' ,
$$

so completing the square gives a Gaussian posterior:

$$
w\mid y\ \sim\ \mathcal{N}\left(m,\ A^{-1}\right), \qquad A = \frac{X^{\top}X}{\sigma^2} + \frac{I}{\tau^2}, \qquad m = \frac{1}{\sigma^2}A^{-1}X^{\top}y .
$$

The mean equals the ridge solution $\left(X^{\top}X+\lambda I\right)^{-1}X^{\top}y$ with $\lambda = \sigma^2/\tau^2$ — but now it comes with a covariance, which MAP alone discards.

**Predictive.** Since $y_\star = x_\star^{\top}w + \varepsilon_\star$ is a linear function of a Gaussian plus independent Gaussian noise,

$$
y_\star\mid y\ \sim\ \mathcal{N}\left(x_\star^{\top}m,\ \underbrace{x_\star^{\top}A^{-1}x_\star}_{\text{epistemic}} + \underbrace{\sigma^2}_{\text{aleatoric}}\right) .
$$

**Geometric reading.** $A^{-1}$ has large eigenvalues along directions poorly covered by the rows of $X$. Decomposing $x_\star$ in the eigenbasis of $X^{\top}X$ with eigenvalues $s_j^2$,

$$
x_\star^{\top}A^{-1}x_\star = \sum_j \frac{\left(x_\star^{\top}v_j\right)^2}{s_j^2/\sigma^2 + 1/\tau^2} .
$$

A test point aligned with a well-sampled direction (large $s_j$) gets a small epistemic term; a point extrapolating along an unexplored direction (small $s_j$) gets a term approaching $\tau^2\left(x_\star^{\top}v_j\right)^2$, i.e. the prior variance. The model *knows it does not know*, which is exactly the behavior missing from an ordinary least-squares prediction that reports the same $\sigma^2$ everywhere.

**Uses.** This is the linear special case of a Gaussian process ($x_\star^{\top}A^{-1}x_\star$ becomes $k_{\star\star} - k_\star^{\top}(K+\sigma^2I)^{-1}k_\star$), and the epistemic term drives acquisition functions in Bayesian optimization and active learning.

$$
\boxed{w\mid y\sim\mathcal{N}(m, A^{-1}), \qquad \operatorname{Var}(y_\star\mid y) = x_\star^{\top}A^{-1}x_\star + \sigma^2}
$$

*Key takeaway:* Bayesian regression returns ridge's point estimate plus the input-dependent uncertainty that ridge throws away.

### Problem L2.3: The Kalman Update Is a Conjugate Bayesian Step

A tracker's prior belief about position is $\mathcal{N}(\hat{x} = 100, P = 4)$ metres. A sensor reports $z = 103$ with noise variance $R = 1$. Derive the updated belief from Bayes' rule and identify the Kalman gain.

**Solution**

**Bayesian update.** Prior $x\sim\mathcal{N}(\hat x, P)$, likelihood $z\mid x\sim\mathcal{N}(x, R)$. This is Normal–Normal conjugacy (Proof 3.3) with $n = 1$:

$$
\frac{1}{P^+} = \frac1P + \frac1R = \frac14 + 1 = 1.25 \quad\Longrightarrow\quad P^+ = 0.8 ,
$$

$$
\hat x^+ = P^+\left(\frac{\hat x}{P} + \frac{z}{R}\right) = 0.8\left(\frac{100}{4} + \frac{103}{1}\right) = 0.8(128) = 102.4 .
$$

**Kalman form.** Rearranging the precision-weighted mean into an innovation form,

$$
K = \frac{P}{P+R} = \frac{4}{5} = 0.8, \qquad \hat x^+ = \hat x + K\left(z - \hat x\right) = 100 + 0.8(3) = 102.4 \ \checkmark ,
$$

$$
P^+ = (1-K)P = 0.2\times4 = 0.8 \ \checkmark .
$$

The Kalman gain $K = \frac{P}{P+R}$ is exactly the fraction of total precision contributed by the measurement. Two limits confirm the interpretation: a perfect sensor ($R\to0$) gives $K\to1$ and the estimate jumps to $z$; a useless sensor ($R\to\infty$) gives $K\to0$ and the prior is retained.

**The full filter.** A tracker alternates this update with a *prediction* step that propagates the belief through the dynamics $x_{t+1} = Fx_t + w_t$, $w_t\sim\mathcal{N}(0,Q)$:

$$
\hat x^-_{t+1} = F\hat x^+_t, \qquad P^-_{t+1} = FP^+_tF^{\top} + Q .
$$

Prediction *adds* uncertainty (via $Q$); measurement *removes* it. The Kalman filter is therefore nothing but recursive Bayesian inference in which every distribution stays Gaussian because the model is linear-Gaussian — a closure property that fails for nonlinear dynamics, where EKF (linearize), UKF (sigma points), and particle filters (Monte Carlo) are the standard approximations.

$$
\boxed{K = \frac{P}{P+R} = 0.8, \quad \hat x^+ = 102.4, \quad P^+ = 0.8}
$$

*Key takeaway:* The Kalman gain is a precision ratio, and the whole filter is Bayes' rule alternating with a prediction step.

### Problem L2.4: A Bayesian A/B Test — Probability That B Beats A

Variant A: $200$ conversions from $1000$ visitors. Variant B: $230$ from $1000$. With uniform priors, compute the posteriors, the distribution of the lift, $P(p_B \gt p_A\mid\text{data})$, and the expected loss from choosing B.

**Solution**

**Posteriors.** With $\text{Beta}(1,1)$ priors,

$$
p_A\mid x\sim\text{Beta}(201,\ 801), \qquad p_B\mid x\sim\text{Beta}(231,\ 771) .
$$

Means and standard deviations:

$$
E[p_A] = \frac{201}{1002} = 0.2006, \quad \operatorname{sd}_A = \sqrt{\frac{201\cdot801}{1002^2\cdot1003}} \approx 0.01264 ,
$$

$$
E[p_B] = \frac{231}{1002} = 0.2305, \quad \operatorname{sd}_B \approx 0.01330 .
$$

**Lift.** With $n \gg 1$ both posteriors are near-Gaussian and independent, so

$$
\delta = p_B - p_A \ \approx\ \mathcal{N}\left(0.0299,\ 0.01264^2 + 0.01330^2\right) = \mathcal{N}\left(0.0299,\ 0.01835^2\right) .
$$

**Probability of superiority.**

$$
P\left(p_B \gt p_A\mid\text{data}\right) = P(\delta \gt 0) \approx \Phi\!\left(\frac{0.0299}{0.01835}\right) = \Phi(1.629) \approx 0.948 .
$$

(Exact numerical integration of the two Beta posteriors gives $\approx 0.949$.) A $95\%$ credible interval for the lift is $0.0299 \pm 1.96(0.01835) = (-0.0060,\ 0.0659)$, which barely includes zero — consistent with a frequentist two-proportion $z$-test giving $p \approx 0.10$ two-sided.

**Expected loss (the decision-relevant number).** If you ship B, the loss when you are wrong is the conversion rate forgone:

$$
E\left[\text{loss}\mid\text{ship B}\right] = E\left[\left(p_A - p_B\right)_+\right] = \int_{-\infty}^0 (-\delta)\,\phi_\delta(\delta)\,d\delta = \sigma_\delta\varphi\!\left(\frac{\mu_\delta}{\sigma_\delta}\right) - \mu_\delta\Phi\!\left(-\frac{\mu_\delta}{\sigma_\delta}\right) ,
$$

with $\mu_\delta = 0.0299$, $\sigma_\delta = 0.01835$: $0.01835(0.1057) - 0.0299(0.0517) \approx 0.00194 - 0.00155 = 0.00039$. So shipping B risks losing about $0.04$ percentage points of conversion in expectation, against an expected *gain* of $\approx 0.0303$. The risk-to-reward ratio is roughly $1{:}78$ — ship it.

$$
\boxed{P(p_B \gt p_A\mid\text{data}) \approx 0.95, \quad E[\text{loss}\mid\text{ship B}] \approx 0.0004}
$$

*Key takeaway:* The Bayesian output is a decision-ready expected loss, not a binary significance verdict — and it can recommend shipping even when a $p$-value would not.

### Problem L2.5: Hierarchical Shrinkage — Estimating Many Rates at Once

$J$ hospitals report surgical success rates from small samples. Model $y_j\mid\theta_j\sim\mathcal{N}(\theta_j,\sigma_j^2)$ with $\theta_j\sim\mathcal{N}(\mu,\tau^2)$. Derive the posterior for $\theta_j$, interpret the shrinkage factor, and explain the empirical-Bayes estimate of $\tau^2$.

**Solution**

**Conditional posterior.** For fixed hyperparameters $(\mu,\tau^2)$, each $\theta_j$ has an independent Normal–Normal update with prior $\mathcal{N}(\mu,\tau^2)$ and one observation of precision $1/\sigma_j^2$:

$$
\theta_j\mid y_j,\mu,\tau^2\ \sim\ \mathcal{N}\left(\hat\theta_j,\ V_j\right), \qquad \frac{1}{V_j} = \frac{1}{\tau^2}+\frac{1}{\sigma_j^2} ,
$$

$$
\hat\theta_j = \frac{\frac{y_j}{\sigma_j^2} + \frac{\mu}{\tau^2}}{\frac{1}{\sigma_j^2}+\frac{1}{\tau^2}} = \left(1 - B_j\right)y_j + B_j\,\mu, \qquad B_j = \frac{\sigma_j^2}{\sigma_j^2+\tau^2} .
$$

**The shrinkage factor $B_j$.** It is the fraction of a hospital's total variance that is measurement noise:

- Small hospital (large $\sigma_j^2$) $\Rightarrow$ $B_j\to1$ $\Rightarrow$ estimate is pulled almost entirely to the population mean $\mu$.
- Large hospital (small $\sigma_j^2$) $\Rightarrow$ $B_j\to0$ $\Rightarrow$ estimate stays at its own $y_j$.
- Homogeneous population ($\tau^2\to0$) $\Rightarrow$ complete pooling, all $\hat\theta_j = \mu$.
- Heterogeneous population ($\tau^2\to\infty$) $\Rightarrow$ no pooling, $\hat\theta_j = y_j$.

This *partial pooling* is why the hospital with 3 operations and 3 successes is not ranked best in the country, and why a batter with 5 hits in 10 at-bats is not projected to hit $.500$.

**Estimating the hyperparameters.** Marginally, integrating out $\theta_j$,

$$
y_j\mid\mu,\tau^2\ \sim\ \mathcal{N}\left(\mu,\ \sigma_j^2+\tau^2\right) ,
$$

so empirical Bayes maximizes this marginal likelihood over $(\mu,\tau^2)$ — a small optimization in two parameters. With equal $\sigma_j^2 = \sigma^2$ the moment estimator is

$$
\hat\tau^2 = \max\left\{0,\ \frac{1}{J-1}\sum_j\left(y_j-\bar y\right)^2 - \sigma^2\right\} ,
$$

i.e. observed spread minus the spread noise alone would produce. A fully Bayesian treatment instead puts a prior on $\tau$ (half-normal or half-Cauchy) and samples, which is preferable because empirical Bayes understates uncertainty by treating $\hat\tau^2$ as known — and can collapse to $\hat\tau^2 = 0$, forcing total pooling.

**Connection to James–Stein.** With $\mu$ estimated as $\bar{y}$ and $\hat B$ from the formula above, $\hat\theta_j$ is essentially the James–Stein estimator, which for $J\ge4$ dominates the raw $y_j$ in total squared error *no matter what the true $\theta_j$ are*. Borrowing strength across units is not a heuristic; it is provably better.

$$
\boxed{\hat\theta_j = (1-B_j)y_j + B_j\mu, \qquad B_j = \frac{\sigma_j^2}{\sigma_j^2+\tau^2}}
$$

*Key takeaway:* Hierarchical priors implement automatic, variance-aware pooling — the Bayesian form of multi-task learning and the reason small-sample rankings must be shrunk.

### Problem L2.6: The VAE Objective from the ELBO

For a latent-variable model $p(x,z) = p_\theta(x\mid z)p(z)$ with $p(z) = \mathcal{N}(0,I)$ and an approximate posterior $q_\phi(z\mid x) = \mathcal{N}\left(\mu_\phi(x), \operatorname{diag}\sigma_\phi^2(x)\right)$, derive the training objective and the closed-form KL term.

**Solution**

**Start from the ELBO** (Proof 3.7):

$$
\ln p_\theta(x) \ \ge\ \mathcal{L}(\theta,\phi;x) = E_{q_\phi(z\mid x)}\left[\ln p_\theta(x\mid z)\right] - D_{\text{KL}}\left(q_\phi(z\mid x)\,\Vert\,p(z)\right) ,
$$

obtained by splitting $\ln p(x,z) = \ln p_\theta(x\mid z) + \ln p(z)$ inside $E_q\left[\ln\frac{p(x,z)}{q(z)}\right]$. The first term is a **reconstruction** log-likelihood; the second is a **regularizer** pulling the encoder's output toward the prior.

**Closed-form KL between diagonal Gaussians.** For $q = \mathcal{N}(\mu,\operatorname{diag}\sigma^2)$ and $p = \mathcal{N}(0,I)$ in $d$ dimensions, the divergence separates coordinatewise. For one dimension,

$$
D_{\text{KL}} = E_q\left[\ln\frac{q}{p}\right] = E_q\left[-\ln\sigma - \frac{(z-\mu)^2}{2\sigma^2} + \frac{z^2}{2}\right] = -\ln\sigma - \frac12 + \frac{\sigma^2+\mu^2}{2} ,
$$

using $E_q[(z-\mu)^2] = \sigma^2$ and $E_q[z^2] = \mu^2+\sigma^2$. Summing over coordinates:

$$
D_{\text{KL}}\left(q_\phi\,\Vert\,p\right) = \frac12\sum_{j=1}^{d}\left(\mu_j^2 + \sigma_j^2 - \ln\sigma_j^2 - 1\right) .
$$

**Reparameterization.** The reconstruction expectation must be differentiated with respect to $\phi$, but $\phi$ sits inside the sampling distribution. Write

$$
z = \mu_\phi(x) + \sigma_\phi(x)\odot\epsilon, \qquad \epsilon\sim\mathcal{N}(0,I) ,
$$

which moves the randomness to a parameter-free variable so that $\nabla_\phi E_q[f(z)] = E_\epsilon\left[\nabla_\phi f\left(\mu_\phi+\sigma_\phi\odot\epsilon\right)\right]$ — a low-variance pathwise gradient, vastly better than the score-function (REINFORCE) estimator.

**The loss.** Minimize $-\mathcal{L}$, i.e. for a Gaussian decoder with fixed variance,

$$
\mathcal{L}_{\text{VAE}} = \underbrace{\left\Vert x - \hat{x}(z)\right\Vert^2}_{\text{reconstruction}} + \underbrace{\frac12\sum_j\left(\mu_j^2+\sigma_j^2-\ln\sigma_j^2-1\right)}_{\text{KL to prior}} .
$$

**Why the bound is not tight.** The gap is $D_{\text{KL}}\left(q_\phi(z\mid x)\,\Vert\,p(z\mid x)\right)$: a diagonal-Gaussian $q$ cannot match a correlated or multimodal true posterior, so the ELBO underestimates $\ln p(x)$ and the learned posterior is too narrow. The $\beta$-VAE reweights the KL term to trade reconstruction against disentanglement; importance-weighted (IWAE) bounds tighten the gap with multiple samples.

$$
\boxed{\mathcal{L} = E_{q}\left[\ln p_\theta(x\mid z)\right] - \tfrac12\sum_j\left(\mu_j^2+\sigma_j^2-\ln\sigma_j^2-1\right)}
$$

*Key takeaway:* A VAE is amortized variational Bayes — the encoder is a learned posterior and the loss is a KL-regularized reconstruction, both read directly off the ELBO identity.

### Problem L3.1: Bayes Factors, Occam's Razor, and Lindley's Paradox

Compare $M_0 : \theta = 0$ against $M_1 : \theta\sim\mathcal{N}(0,\tau^2)$ for data $\bar{x}\sim\mathcal{N}(\theta,\sigma^2/n)$. Derive the Bayes factor, show how it penalizes flexibility, and exhibit Lindley's paradox.

**Solution**

**Marginal likelihoods.** Under $M_0$ the data density is immediate:

$$
p\left(\bar x\mid M_0\right) = \mathcal{N}\left(\bar x;\ 0,\ \frac{\sigma^2}{n}\right) .
$$

Under $M_1$, integrate the Gaussian likelihood against the Gaussian prior; since $\bar x = \theta + \text{noise}$ with independent Gaussian pieces, the variances add:

$$
p\left(\bar x\mid M_1\right) = \mathcal{N}\left(\bar x;\ 0,\ \tau^2 + \frac{\sigma^2}{n}\right) .
$$

**Bayes factor.** Writing $v = \sigma^2/n$ and $z = \bar x/\sqrt{v}$ (the usual test statistic),

$$
\text{BF}_{01} = \frac{p(\bar x\mid M_0)}{p(\bar x\mid M_1)} = \sqrt{\frac{\tau^2+v}{v}}\ \exp\left(-\frac{\bar x^2}{2v} + \frac{\bar x^2}{2(\tau^2+v)}\right) = \sqrt{1+\frac{\tau^2}{v}}\ \exp\left(-\frac{z^2}{2}\cdot\frac{\tau^2}{\tau^2+v}\right) .
$$

**Occam's razor made explicit.** The two factors pull in opposite directions:

- $\sqrt{1+\tau^2/v}$ *favors* $M_0$ and grows with the prior width $\tau$: a model that spreads its prior over a wide range of $\theta$ dilutes its predictive mass everywhere and is punished for being able to explain too much.
- $\exp(-z^2/2\cdot\frac{\tau^2}{\tau^2+v})$ *favors* $M_1$ when the data are far from $0$.

No complexity penalty was imposed; it emerged from the requirement that $\int p(\bar x\mid M)d\bar x = 1$ — flexible models must spread a fixed unit of predictive mass thinly.

**Lindley's paradox.** Fix $z = 1.96$, so a frequentist rejects $M_0$ at exactly $p = 0.05$ for every $n$. As $n\to\infty$ with $\tau$ fixed, $v = \sigma^2/n\to0$, so $\frac{\tau^2}{\tau^2+v}\to1$ and

$$
\text{BF}_{01} \approx \sqrt{\frac{n\tau^2}{\sigma^2}}\;e^{-z^2/2} \ \longrightarrow\ \infty .
$$

With $\sigma = \tau = 1$ and $n = 10^4$: $\text{BF}_{01}\approx 100\times e^{-1.92} = 14.6$, i.e. odds of about $15{:}1$ **in favor of** $M_0$ — while the $p$-value says reject at $5\%$. Both are computing correctly; they answer different questions. A fixed-$\alpha$ test holds the false-positive rate constant regardless of $n$, so at huge $n$ it rejects on effects too small to matter, whereas the Bayes factor asks whether the data are better predicted by "exactly zero" or by "somewhere in a range of width $\tau$" — and a tiny nonzero $\bar x$ is much better predicted by the former.

**Practical caution.** $\text{BF}_{01}\propto\tau$ means Bayes factors for point-null tests are *never* insensitive to the prior width, even asymptotically, and improper priors under $M_1$ make them undefined. Use proper, defensible priors (or switch to predictive comparison such as cross-validated log-likelihood or WAIC).

$$
\boxed{\text{BF}_{01} = \sqrt{1+\tfrac{n\tau^2}{\sigma^2}}\,\exp\left(-\tfrac{z^2}{2}\cdot\tfrac{\tau^2}{\tau^2+\sigma^2/n}\right)\ \xrightarrow[n\to\infty]{z \text{ fixed}}\ \infty}
$$

*Key takeaway:* Marginal likelihoods penalize flexibility automatically, and at large $n$ a fixed-$\alpha$ test and a Bayes factor can point in opposite directions — because they ask different questions.

### Problem L3.2: Jeffreys' Prior and Why It Is Invariant

Derive Jeffreys' prior for (a) a Bernoulli rate and (b) a Gaussian scale parameter, and prove the invariance property $p_J(\phi) = p_J(\theta)\left\vert d\theta/d\phi\right\vert$ under reparameterization.

**Solution**

**Definition.** $p_J(\theta)\propto\sqrt{\det I(\theta)}$ with $I$ the Fisher information.

**Proof of invariance.** Let $\phi = g(\theta)$ be a smooth bijection. The score transforms by the chain rule,

$$
\frac{\partial}{\partial\phi}\ln p\left(x\mid\theta(\phi)\right) = \frac{\partial}{\partial\theta}\ln p(x\mid\theta)\cdot\frac{d\theta}{d\phi} ,
$$

so squaring and taking expectations,

$$
I(\phi) = I(\theta)\left(\frac{d\theta}{d\phi}\right)^2 \quad\Longrightarrow\quad \sqrt{I(\phi)} = \sqrt{I(\theta)}\left\vert \frac{d\theta}{d\phi}\right\vert .
$$

The right-hand side is precisely the change-of-variables formula for a density. Hence computing Jeffreys' prior in the $\phi$ parameterization gives the same measure as transforming the $\theta$ version — the rule commutes with reparameterization, which no flat prior does. (In $d$ dimensions the same argument with the Jacobian matrix $J$ gives $I(\phi) = J^{\top}I(\theta)J$ and $\sqrt{\det I(\phi)} = \sqrt{\det I(\theta)}\,\vert\det J\vert$.) $\blacksquare$

**(a) Bernoulli.** From Topic 09, $I(p) = \frac{1}{p(1-p)}$, so

$$
p_J(p)\ \propto\ p^{-1/2}(1-p)^{-1/2} = \text{Beta}\!\left(\tfrac12,\tfrac12\right) .
$$

This arcsine prior is U-shaped, placing extra mass near $0$ and $1$ — reflecting that in the *natural* (logit or arcsine) geometry, extreme rates are far apart and deserve more prior spread. Its posterior after $k$ successes is $\text{Beta}(k+\tfrac12,\ n-k+\tfrac12)$, giving the estimate $\frac{k+1/2}{n+1}$, which has better small-sample frequentist coverage than either the MLE or the uniform-prior answer.

**(b) Gaussian scale.** For $\mathcal{N}(\mu,\sigma^2)$ with $\mu$ known, $\ln p = -\ln\sigma - \frac{(x-\mu)^2}{2\sigma^2}+C$, so

$$
\frac{\partial^2\ln p}{\partial\sigma^2} = \frac{1}{\sigma^2} - \frac{3(x-\mu)^2}{\sigma^4}, \qquad I(\sigma) = -E\left[\cdot\right] = -\frac{1}{\sigma^2}+\frac{3\sigma^2}{\sigma^4} = \frac{2}{\sigma^2} .
$$

Hence

$$
p_J(\sigma)\ \propto\ \frac{1}{\sigma} ,
$$

which is flat in $\ln\sigma$ — scale invariance, as it must be for a quantity whose units are arbitrary. It is improper ($\int_0^\infty \sigma^{-1}d\sigma = \infty$) but yields a proper posterior for $n\ge2$.

**Caveats.** In multiple dimensions Jeffreys' rule can behave badly (for $\mathcal{N}(\mu,\sigma^2)$ with both unknown it gives $1/\sigma^2$, whereas the independence-Jeffreys $1/\sigma$ has better properties), which is why reference priors and weakly informative priors are usually preferred in practice.

$$
\boxed{p_J(p) = \text{Beta}\!\left(\tfrac12,\tfrac12\right), \quad p_J(\sigma)\propto1/\sigma, \quad \sqrt{I(\phi)} = \sqrt{I(\theta)}\left\vert\tfrac{d\theta}{d\phi}\right\vert}
$$

*Key takeaway:* Jeffreys' prior is flat with respect to the Fisher-information metric, which is the only notion of "flat" that survives a change of coordinates.

### Problem L3.3: Where Bernstein–von Mises Succeeds and Where It Fails

State the Bernstein–von Mises theorem, verify it explicitly for the Beta–Binomial model, and give three models where it fails, explaining which hypothesis breaks in each.

**Solution**

**Statement.** Under regularity — identifiability, smoothness of $\theta\mapsto p(x\mid\theta)$, $\theta_0$ interior, nonsingular $I(\theta_0)$, prior positive and continuous at $\theta_0$ —

$$
\left\Vert\, p\left(\theta\mid x_{1:n}\right) - \mathcal{N}\left(\hat\theta_n,\ \left[nI(\theta_0)\right]^{-1}\right)\right\Vert_{\text{TV}}\ \xrightarrow{P}\ 0 .
$$

**Explicit verification (Beta–Binomial).** The posterior is $\text{Beta}(\alpha+k,\ \beta+n-k)$. With $k/n\to p_0$:

$$
E[p\mid x] = \frac{\alpha+k}{\alpha+\beta+n} = \frac{k}{n} + O\!\left(\frac1n\right) = \hat p + O(n^{-1}) ,
$$

$$
\operatorname{Var}(p\mid x) = \frac{(\alpha+k)(\beta+n-k)}{(\alpha+\beta+n)^2(\alpha+\beta+n+1)} = \frac{p_0(1-p_0)}{n} + O\!\left(\frac{1}{n^2}\right) .
$$

Since $I(p) = \frac{1}{p(1-p)}$, the claimed limit variance is $\left[nI(p_0)\right]^{-1} = \frac{p_0(1-p_0)}{n}$ ✓. The prior's hyperparameters appear only at order $1/n$ in the mean and $1/n^2$ in the variance: they are washed out. Skewness of the Beta is $O(n^{-1/2})$, so the shape converges to Gaussian as well.

**Three failures.**

1. **Non-regular support: $\text{Unif}(0,\theta)$.** The likelihood is $\theta^{-n}\mathbf{1}\{\theta\ge x_{(n)}\}$, so with a smooth prior the posterior concentrates on $[x_{(n)},\infty)$ with

$$
p(\theta\mid x)\ \propto\ \theta^{-n}p(\theta)\ \mathbf{1}\{\theta\ge x_{(n)}\} ,
$$

which converges after rescaling by $n$ (not $\sqrt n$) to an *exponential*-shaped, one-sided law. Broken hypothesis: differentiability / support independent of $\theta$; the Fisher information does not exist.

2. **Prior with a hole.** If $p(\theta) = 0$ on a neighborhood of $\theta_0$, then $p(\theta\mid x) = 0$ there for all $n$: Cromwell's rule. No amount of data can resurrect a hypothesis assigned zero prior probability. Broken hypothesis: prior positivity at $\theta_0$.

3. **High dimension, $d$ growing with $n$.** BvM is an asymptotic in $n$ with $d$ fixed. When $d/n\not\to0$ the quadratic expansion's remainder no longer vanishes and the posterior can be badly non-Gaussian and prior-sensitive; credible sets can have coverage far from nominal even asymptotically (a phenomenon studied extensively for nonparametric and semiparametric models). Related failures occur on parameter-space boundaries (testing $\tau^2 = 0$ in a hierarchical model) and for non-identified mixtures.

**Why it matters.** BvM is the reason Bayesian and frequentist answers coincide in routine large-sample problems — and its failure cases are exactly the situations (small $n$, boundaries, hierarchies, nonparametrics) where the choice of framework genuinely changes conclusions.

$$
\boxed{p(\theta\mid x)\to\mathcal{N}\left(\hat\theta,(nI)^{-1}\right) \text{ under regularity; fails for } \text{Unif}(0,\theta), \text{ zero priors, growing } d}
$$

*Key takeaway:* The prior washes out at rate $O(1/n)$ whenever the model is regular — and the interesting statistical problems are precisely the irregular ones.

### Problem L3.4: Why Metropolis–Hastings Samples the Right Distribution

Prove that the Metropolis–Hastings chain satisfies detailed balance with respect to the posterior, show that only unnormalized densities are needed, and derive Gibbs sampling as the special case with acceptance probability $1$.

**Solution**

**Setup.** Let $\pi(\theta) = p(\theta\mid x) = \tilde\pi(\theta)/Z$ with $\tilde\pi(\theta) = p(x\mid\theta)p(\theta)$ and unknown $Z = p(x)$. Given a proposal density $q(\theta'\mid\theta)$, the algorithm proposes $\theta'$ and accepts with

$$
a\left(\theta\to\theta'\right) = \min\left\{1,\ \frac{\tilde\pi(\theta')q(\theta\mid\theta')}{\tilde\pi(\theta)q(\theta'\mid\theta)}\right\} .
$$

Note the ratio $\tilde\pi(\theta')/\tilde\pi(\theta) = \pi(\theta')/\pi(\theta)$ — the unknown $Z$ cancels. **This cancellation is the entire reason MCMC is usable**: the intractable evidence never has to be computed.

**Detailed balance.** For $\theta'\ne\theta$ the transition kernel is $P(\theta\to\theta') = q(\theta'\mid\theta)a(\theta\to\theta')$. We must show

$$
\pi(\theta)P(\theta\to\theta') = \pi(\theta')P(\theta'\to\theta) .
$$

Write $r = \frac{\pi(\theta')q(\theta\mid\theta')}{\pi(\theta)q(\theta'\mid\theta)}$, so $a(\theta\to\theta') = \min\{1,r\}$ and $a(\theta'\to\theta) = \min\{1,1/r\}$. Consider $r \le 1$ (the case $r \gt 1$ is symmetric by swapping labels). Then $a(\theta\to\theta') = r$ and $a(\theta'\to\theta) = 1$, so

$$
\pi(\theta)q(\theta'\mid\theta)\,r = \pi(\theta)q(\theta'\mid\theta)\cdot\frac{\pi(\theta')q(\theta\mid\theta')}{\pi(\theta)q(\theta'\mid\theta)} = \pi(\theta')q(\theta\mid\theta') = \pi(\theta')q(\theta\mid\theta')\cdot 1 . \qquad \square
$$

Both sides agree, so detailed balance holds; the self-transition $\theta' = \theta$ is trivially balanced.

**From balance to correctness.** Integrating detailed balance over $\theta$ gives

$$
\int \pi(\theta)P(\theta\to\theta')\,d\theta = \pi(\theta')\int P(\theta'\to\theta)\,d\theta = \pi(\theta') ,
$$

so $\pi$ is a stationary distribution. If the chain is additionally irreducible (it can reach any positive-probability region) and aperiodic, the ergodic theorem gives, for $\pi$-integrable $f$,

$$
\frac{1}{T}\sum_{t=1}^T f\left(\theta_t\right)\ \xrightarrow{a.s.}\ E_\pi[f] ,
$$

regardless of the starting point. $\blacksquare$

**Gibbs as a special case.** Propose from the exact full conditional, $q(\theta'_j\mid\theta) = \pi\left(\theta'_j\mid\theta_{-j}\right)$, updating one block at a time with $\theta'_{-j} = \theta_{-j}$. Then

$$
r = \frac{\pi(\theta'_j,\theta_{-j})\,\pi(\theta_j\mid\theta_{-j})}{\pi(\theta_j,\theta_{-j})\,\pi(\theta'_j\mid\theta_{-j})} = \frac{\pi(\theta'_j\mid\theta_{-j})\pi(\theta_{-j})\pi(\theta_j\mid\theta_{-j})}{\pi(\theta_j\mid\theta_{-j})\pi(\theta_{-j})\pi(\theta'_j\mid\theta_{-j})} = 1 ,
$$

so every proposal is accepted. Gibbs sampling is Metropolis–Hastings with a perfect local proposal — which is why it needs conditional conjugacy, and why it mixes poorly when coordinates are strongly correlated (it can only move along axes).

**Practical notes.** Tune random-walk proposals to $\approx 23\%$ acceptance in high dimension (the optimal scaling result); prefer HMC/NUTS, which uses $\nabla\ln\tilde\pi$ to propose distant moves with high acceptance; and always monitor $\hat R$ and effective sample size, since stationarity is guaranteed only asymptotically.

$$
\boxed{\pi(\theta)q(\theta'\mid\theta)a(\theta\to\theta') = \pi(\theta')q(\theta\mid\theta')a(\theta'\to\theta); \ \ Z \text{ cancels}}
$$

*Key takeaway:* MCMC works because acceptance depends only on density *ratios*, so the intractable normalizer disappears — detailed balance then does the rest.
