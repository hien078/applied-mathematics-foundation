# Topic 03: Fixed-Point Iteration and Convergence Theory

## 1. First-Principles Intuition & Motivation

Almost every iterative algorithm in numerical mathematics — Newton's method, gradient descent, Jacobi and Gauss–Seidel, power iteration, value iteration in reinforcement learning, even the training loop of an implicit neural network layer — has the same skeleton:

$$
x_{n+1} = g(x_n) .
$$

If the sequence settles down, its limit $x^{*}$ satisfies $x^{*} = g(x^{*})$: a *fixed point* of $g$. Fixed-point theory is therefore the unifying language of iteration. Instead of analyzing each algorithm from scratch, we ask two universal questions:

1. **Existence/uniqueness** — does $g$ have a fixed point, and only one?
2. **Convergence** — does the iteration approach it, from where, and how fast?

The single most important answer is the **contraction principle**: if $g$ shrinks distances by a uniform factor $L \lt 1$, the iteration converges from *every* starting point, at a geometric rate, to the *unique* fixed point — and we even get computable error bounds along the way.

### The geometric picture: cobwebs

Plot $y = g(x)$ against $y = x$. The iteration bounces between the curve and the diagonal: from $x_n$ go vertically to the curve ($y = g(x_n)$), then horizontally to the diagonal ($x_{n+1} = y$). Near a fixed point:

- $\lvert g'(x^{*}) \rvert \lt 1$: the staircase/spiral is pulled *in* — attracting fixed point.
- $\lvert g'(x^{*}) \rvert \gt 1$: pushed *out* — repelling; the iteration flees even from excellent starts.
- $g'(x^{*}) \gt 0$: monotone staircase; $g'(x^{*}) \lt 0$: alternating spiral (errors flip sign — useful, since the root is then bracketed by consecutive iterates).

The same equation $f(x) = 0$ can be rearranged into many fixed-point forms $x = g(x)$, and *the rearrangement decides everything*: one choice of $g$ converges, another diverges. Newton's method is precisely the rearrangement engineered to make $g'(x^{*}) = 0$.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition 1 (Fixed point).** $x^{*}$ is a *fixed point* of $g: D \to \mathbb{R}$ if $g(x^{*}) = x^{*}$.

**Definition 2 (Lipschitz map, contraction).** $g$ is *Lipschitz* on $D$ with constant $L$ if

$$
\lvert g(x) - g(y) \rvert \le L\, \lvert x - y \rvert \qquad \text{for all } x, y \in D .
$$

If $L \lt 1$, $g$ is a *contraction*. If $g \in C^1$ on an interval, the mean value theorem gives $L = \sup_{x \in D} \lvert g'(x) \rvert$.

**Definition 3 (Attracting / repelling fixed point).** A fixed point $x^{*}$ of $g \in C^1$ is *attracting* if $\lvert g'(x^{*}) \rvert \lt 1$ and *repelling* if $\lvert g'(x^{*}) \rvert \gt 1$.

**Theorem 1 (Banach fixed-point theorem, interval version).** Let $g: [a, b] \to [a, b]$ be a contraction with constant $L \lt 1$. Then:

1. $g$ has exactly one fixed point $x^{*} \in [a, b]$;
2. for every $x_0 \in [a, b]$, the iteration $x_{n+1} = g(x_n)$ converges to $x^{*}$;
3. the *a priori* bound $\lvert x_n - x^{*} \rvert \le \dfrac{L^{n}}{1 - L} \lvert x_1 - x_0 \rvert$ holds;
4. the *a posteriori* bound $\lvert x_n - x^{*} \rvert \le \dfrac{L}{1 - L} \lvert x_n - x_{n-1} \rvert$ holds.

**Theorem 2 (Local convergence).** If $g \in C^1$ near a fixed point $x^{*}$ with $\lvert g'(x^{*}) \rvert \lt 1$, there is a neighborhood from which the iteration converges, and the convergence is linear with asymptotic rate $\lvert g'(x^{*}) \rvert$ when $g'(x^{*}) \neq 0$.

**Theorem 3 (Higher-order fixed-point convergence).** Let $g \in C^{p}$ near a fixed point $x^{*}$ with

$$
g'(x^{*}) = g''(x^{*}) = \cdots = g^{(p-1)}(x^{*}) = 0, \qquad g^{(p)}(x^{*}) \neq 0, \qquad p \ge 2 .
$$

Then the iteration converges locally with order exactly $p$, and

$$
\lim_{n \to \infty} \frac{\lvert x_{n+1} - x^{*} \rvert}{\lvert x_n - x^{*} \rvert^{p}} = \frac{\lvert g^{(p)}(x^{*}) \rvert}{p!} .
$$

**Theorem 4 (Aitken acceleration).** If $x_n \to x^{*}$ linearly ($e_{n+1} = (\lambda + \delta_n) e_n$ with $\lvert \lambda \rvert \lt 1$, $\delta_n \to 0$), the transformed sequence

$$
\hat{x}_n = x_n - \frac{(x_{n+1} - x_n)^2}{x_{n+2} - 2x_{n+1} + x_n}
$$

converges to $x^{*}$ *faster*: $\dfrac{\hat{x}_n - x^{*}}{x_n - x^{*}} \to 0$. Applying Aitken to every third iterate of a fixed-point map (Steffensen's method) yields quadratic convergence without derivatives.

## 3. Step-by-Step Mathematical Proofs & Derivations

### Proof 1 — The Banach fixed-point theorem

**Existence.** Since $g$ maps $[a,b]$ into itself, the iterates are well defined. Contraction gives

$$
\lvert x_{n+1} - x_n \rvert = \lvert g(x_n) - g(x_{n-1}) \rvert \le L \lvert x_n - x_{n-1} \rvert \le \cdots \le L^{n} \lvert x_1 - x_0 \rvert .
$$

For $m \gt n$, telescope and sum the geometric series:

$$
\lvert x_m - x_n \rvert \le \sum_{k=n}^{m-1} \lvert x_{k+1} - x_k \rvert \le \lvert x_1 - x_0 \rvert \sum_{k=n}^{m-1} L^{k} \le \frac{L^{n}}{1 - L}\, \lvert x_1 - x_0 \rvert .
$$

The right side tends to 0, so $(x_n)$ is Cauchy and converges to some $x^{*} \in [a,b]$ (closed set). Continuity of $g$ (Lipschitz maps are continuous) allows passing to the limit in $x_{n+1} = g(x_n)$: $x^{*} = g(x^{*})$.

**Uniqueness.** If $x^{*}$ and $y^{*}$ are both fixed, then

$$
\lvert x^{*} - y^{*} \rvert = \lvert g(x^{*}) - g(y^{*}) \rvert \le L \lvert x^{*} - y^{*} \rvert,
$$

and $L \lt 1$ forces $\lvert x^{*} - y^{*} \rvert = 0$.

**Error bounds.** Letting $m \to \infty$ in the Cauchy estimate gives the a priori bound (3). For the a posteriori bound, $\lvert x_n - x^{*} \rvert \le \lvert x_n - x_{n+1} \rvert + \lvert x_{n+1} - x^{*} \rvert \le L\lvert x_{n-1} - x_n \rvert + L \lvert x_n - x^{*} \rvert$; solving for $\lvert x_n - x^{*} \rvert$ yields (4). $\blacksquare$

### Proof 2 — Local convergence and the exact linear rate

**Claim.** (Theorem 2.) If $\lvert g'(x^{*}) \rvert = \lambda \lt 1$, the iteration converges locally and $e_{n+1}/e_n \to g'(x^{*})$ (signed errors $e_n = x_n - x^{*}$).

**Proof.** Pick $\varepsilon \gt 0$ with $\lambda + \varepsilon \lt 1$. By continuity of $g'$ there is $\delta \gt 0$ such that $\lvert g'(x) \rvert \le \lambda + \varepsilon$ on $I = [x^{*} - \delta, x^{*} + \delta]$. For $x \in I$, the mean value theorem gives

$$
\lvert g(x) - x^{*} \rvert = \lvert g(x) - g(x^{*}) \rvert = \lvert g'(\xi) \rvert \lvert x - x^{*} \rvert \le (\lambda + \varepsilon)\, \lvert x - x^{*} \rvert \lt \delta,
$$

so $g$ maps $I$ into itself and is a contraction there; Banach's theorem applies on $I$, giving convergence with $e_n \to 0$ at geometric rate $\lambda + \varepsilon$ for every $\varepsilon$.

For the exact rate: by the MVT, $e_{n+1} = g(x_n) - g(x^{*}) = g'(\xi_n) e_n$ with $\xi_n$ between $x_n$ and $x^{*}$; as $x_n \to x^{*}$, $\xi_n \to x^{*}$ and $g'(\xi_n) \to g'(x^{*})$, hence $e_{n+1}/e_n \to g'(x^{*})$. $\blacksquare$

Note the sign information: if $g'(x^{*}) \lt 0$ the errors alternate, so $x^{*}$ always lies between consecutive iterates — a free error bracket.

### Proof 3 — Order-$p$ convergence when derivatives vanish

**Claim.** (Theorem 3.) With $g'(x^{*}) = \cdots = g^{(p-1)}(x^{*}) = 0$ and $g^{(p)}(x^{*}) \neq 0$, the local order is exactly $p$.

**Proof.** Taylor-expand $g$ about $x^{*}$ with Lagrange remainder. Because all derivatives through order $p-1$ vanish at $x^{*}$,

$$
x_{n+1} = g(x_n) = g(x^{*}) + \frac{g^{(p)}(\xi_n)}{p!} (x_n - x^{*})^{p} = x^{*} + \frac{g^{(p)}(\xi_n)}{p!}\, e_n^{p}
$$

for some $\xi_n$ between $x_n$ and $x^{*}$. Hence

$$
\frac{\lvert e_{n+1} \rvert}{\lvert e_n \rvert^{p}} = \frac{\lvert g^{(p)}(\xi_n) \rvert}{p!} \longrightarrow \frac{\lvert g^{(p)}(x^{*}) \rvert}{p!} \neq 0,
$$

so the order is exactly $p$ (not more, since the limit is nonzero; not less, since it is finite). Local convergence follows as in Proof 2 because $e_{n+1} = O(e_n^{p})$ with $p \ge 2$ contracts any sufficiently small ball. $\blacksquare$

**Newton as corollary.** For $g(x) = x - f(x)/f'(x)$ at a simple root,

$$
g'(x) = 1 - \frac{f'(x)^2 - f(x) f''(x)}{f'(x)^2} = \frac{f(x) f''(x)}{f'(x)^2} \implies g'(x^{*}) = 0,
$$

and generically $g''(x^{*}) = f''(x^{*})/f'(x^{*}) \neq 0$: order exactly 2, recovering Topic 02's theorem from pure fixed-point principles.

### Proof 4 — Aitken's $\Delta^2$ acceleration

**Claim.** (Theorem 4.) For linearly convergent $x_n$ with $e_{n+1} = (\lambda + \delta_n)e_n$, $\delta_n \to 0$, $0 \lt \lvert \lambda \rvert \lt 1$, Aitken's $\hat{x}_n$ satisfies $(\hat{x}_n - x^{*})/(x_n - x^{*}) \to 0$.

**Derivation of the formula.** If convergence were *exactly* geometric, $e_{n+1} = \lambda e_n$, then

$$
\frac{x_{n+1} - x^{*}}{x_n - x^{*}} = \frac{x_{n+2} - x^{*}}{x_{n+1} - x^{*}} .
$$

Cross-multiplying and solving for $x^{*}$:

$$
x^{*} = \frac{x_n x_{n+2} - x_{n+1}^2}{x_{n+2} - 2x_{n+1} + x_n} = x_n - \frac{(x_{n+1} - x_n)^2}{x_{n+2} - 2x_{n+1} + x_n},
$$

which is exact for a true geometric sequence — Aitken extrapolates *as if* the error were geometric.

**Proof of acceleration.** Substitute $e_{n+1} = (\lambda + \delta_n)e_n$ into the formula written in error form:

$$
\hat{x}_n - x^{*} = e_n - \frac{(e_{n+1} - e_n)^2}{e_{n+2} - 2e_{n+1} + e_n} = e_n\left( 1 - \frac{(\lambda + \delta_n - 1)^2}{(\lambda + \delta_{n+1})(\lambda + \delta_n) - 2(\lambda + \delta_n) + 1} \right).
$$

As $n \to \infty$, the fraction tends to $\dfrac{(\lambda - 1)^2}{\lambda^2 - 2\lambda + 1} = 1$, so the bracket tends to 0, proving $(\hat{x}_n - x^{*})/e_n \to 0$. $\blacksquare$

Steffensen's method feeds the accelerated value back in ($x_{\mathrm{new}} = \hat{x}$ computed from $x, g(x), g(g(x))$) and achieves order 2 with no derivatives.

### Proof 5 — Sharpness: divergence from a repelling fixed point

**Claim.** If $\lvert g'(x^{*}) \rvert \gt 1$, then for every $x_0 \neq x^{*}$ close enough to $x^{*}$, the iteration *leaves* every sufficiently small neighborhood of $x^{*}$ — no matter how close $x_0$ is.

**Proof.** Choose $\varepsilon \gt 0$ with $\mu := \lvert g'(x^{*}) \rvert - \varepsilon \gt 1$ and $\delta \gt 0$ with $\lvert g'(x) \rvert \ge \mu$ on $I = [x^{*} - \delta, x^{*} + \delta]$. While $x_n \in I$ and $x_n \neq x^{*}$, the MVT gives

$$
\lvert x_{n+1} - x^{*} \rvert = \lvert g'(\xi_n) \rvert\, \lvert x_n - x^{*} \rvert \ge \mu\, \lvert x_n - x^{*} \rvert .
$$

So the error grows at least geometrically with ratio $\mu \gt 1$ and must exceed $\delta$ after at most $\lceil \log_{\mu}(\delta / e_0) \rceil$ steps: the orbit exits $I$. $\blacksquare$

Consequence: rearrangement matters. To solve $x^2 - x - 1 = 0$ ($x^{*} = \varphi \approx 1.618$), the form $x = x^2 - 1$ has $g' = 2x^{*} \approx 3.24$ (repelling — diverges), while $x = \sqrt{x + 1}$ has $g' = \frac{1}{2\sqrt{x^{*} + 1}} \approx 0.31$ (attracting — converges). Same equation, opposite fates.

## 4. Computational & Algorithmic Insights

### Choosing and engineering the map $g$

- **Rate is everything**: iterations needed $\approx \dfrac{\log(\varepsilon / e_0)}{\log \lambda}$ with $\lambda = \lvert g'(x^{*}) \rvert$. Rate 0.9 needs ~22x the iterations of rate 0.1 for each digit.
- **Relaxation/damping**: replace $g$ by $g_\omega(x) = (1 - \omega) x + \omega\, g(x)$. Then $g_\omega' = 1 - \omega + \omega g'$, and choosing $\omega = \frac{1}{1 - g'(x^{*})}$ makes $g_\omega'(x^{*}) = 0$ — turning a divergent or slow iteration into a superlinear one. This is the scalar seed of SOR and of momentum-style accelerations.
- **Stopping**: the a posteriori bound $\lvert x_n - x^{*} \rvert \le \frac{L}{1-L} \lvert x_n - x_{n-1} \rvert$ shows the step size $\lvert x_n - x_{n-1} \rvert$ *understates* the error by the factor $\frac{L}{1-L}$ — dangerously so when $L \approx 1$ (e.g. $L = 0.99$ gives factor 99). Estimate $L$ on the fly from $\hat{\lambda}_n = \frac{x_{n+1} - x_n}{x_n - x_{n-1}}$.

### Fixed points in $n$ dimensions

Everything generalizes with $\lvert \cdot \rvert$ replaced by a norm: $G: \mathbb{R}^{d} \to \mathbb{R}^{d}$ is a contraction if $\lVert G(x) - G(y) \rVert \le L \lVert x - y \rVert$, $L \lt 1$, and Banach's proof goes through verbatim on any complete space. For affine iterations $x_{n+1} = Mx_n + c$, convergence for every start holds **iff** the spectral radius satisfies $\rho(M) \lt 1$ — the spectral radius, not any particular norm, is the true asymptotic rate:

$$
e_n = M^{n} e_0, \qquad \lVert M^{n} \rVert^{1/n} \to \rho(M).
$$

This single criterion governs Jacobi ($M = -D^{-1}(L+U)$), Gauss–Seidel, gradient descent on quadratics ($M = I - \alpha A$), and the value-iteration operator of reinforcement learning ($L = \gamma$ in the sup norm). The nonlinear local version: an iterate map $G \in C^1$ converges locally iff $\rho(J_G(x^{*})) \lt 1$.

## 5. Real-World Physics & AI/ML Applications

**Value iteration and Q-learning (reinforcement learning).** The Bellman optimality operator

$$
(TV)(s) = \max_{a} \left[ r(s,a) + \gamma \sum_{s'} P(s' \mid s, a) V(s') \right]
$$

is a $\gamma$-contraction in the sup norm: $\lVert TV - TW \rVert_\infty \le \gamma \lVert V - W \rVert_\infty$. Banach's theorem *is* the convergence proof of value iteration: a unique optimal value function exists, iteration converges from any initialization, and the a priori bound $\gamma^{n}/(1-\gamma)$ tells you exactly how many sweeps a target accuracy costs.

**Gradient descent as a fixed-point map.** $x_{n+1} = x_n - \alpha \nabla f(x_n)$ is fixed-point iteration on $g = \mathrm{id} - \alpha \nabla f$. For $\mu \preceq \nabla^2 f \preceq L_s$ (strong convexity/smoothness), $g$ is a contraction with factor $\max(\lvert 1 - \alpha\mu \rvert, \lvert 1 - \alpha L_s \rvert)$, minimized at $\alpha = \frac{2}{\mu + L_s}$ giving rate $\frac{\kappa - 1}{\kappa + 1}$, $\kappa = L_s/\mu$. Divergence for $\alpha \gt 2/L_s$ is Proof 5 in action.

**Deep equilibrium models and implicit layers.** DEQ networks define their output as the fixed point $z^{*} = g_\theta(z^{*}, x)$ of a learned map, computed by accelerated fixed-point iteration (Anderson acceleration — a multidimensional cousin of Aitken); backpropagation differentiates through the fixed point via the implicit function theorem rather than through the iterations.

**Physics and engineering.**

- **Implicit time stepping (Topic 08):** each backward-Euler step solves $y_{n+1} = y_n + h f(t_{n+1}, y_{n+1})$ — a fixed-point equation in $y_{n+1}$. Direct iteration contracts iff $h \lVert \partial f / \partial y \rVert \lt 1$, which is precisely why *stiff* problems (huge Jacobians) demand Newton inner solvers instead.
- **Self-consistent field methods:** Hartree–Fock and density functional theory iterate "density $\to$ potential $\to$ new density" to a self-consistent fixed point, stabilized by damping/relaxation exactly as in Section 4.
- **Kepler's equation:** $E = M + e\sin E$ is already in fixed-point form; $g'(E) = e\cos E$ gives $L \le e \lt 1$ — guaranteed convergence for all elliptical orbits, with rate $e$ (slow for comets, fast for near-circular satellites).
- **Markov chains / PageRank:** $\pi_{n+1}^{T} = \pi_n^{T} P$ converges to the stationary distribution; with damping factor $d = 0.85$, the Google matrix is a contraction with $L = d$ in total-variation distance, so ~50 iterations give bulletin-board accuracy regardless of web size.

### Summary of key results

| Result | Statement |
| :--- | :--- |
| Banach | contraction on complete set: unique $x^{*}$, global geometric convergence |
| A priori bound | $\lvert x_n - x^{*} \rvert \le \frac{L^{n}}{1-L}\lvert x_1 - x_0 \rvert$ |
| A posteriori bound | $\lvert x_n - x^{*} \rvert \le \frac{L}{1-L}\lvert x_n - x_{n-1} \rvert$ |
| Local rate | $e_{n+1}/e_n \to g'(x^{*})$; converges iff $\lvert g'(x^{*}) \rvert \lt 1$ (generically) |
| Order $p$ | $g' = \cdots = g^{(p-1)} = 0$ at $x^{*}$ $\implies$ order exactly $p$ |
| Newton | fixed-point form with $g'(x^{*}) = 0$ — quadratic |
| Aitken/Steffensen | geometric-error extrapolation; order 2 derivative-free |
| Linear maps | $x \mapsto Mx + c$ converges globally iff $\rho(M) \lt 1$ |
