## Level 0 — Concept Check

### Problem L0.1: Roots vs fixed points

Show that root-finding and fixed-point problems are interconvertible: convert $f(x) = x^3 - x - 2 = 0$ into three different fixed-point forms, and convert the fixed-point problem $x = \cos x$ into a root-finding problem.

**Solution.**

Any rearrangement isolating an $x$ works. Three forms for $f(x) = x^3 - x - 2 = 0$:

1. $g_1(x) = x^3 - 2$ (from $x = x^3 - 2$),
2. $g_2(x) = (x + 2)^{1/3}$ (from $x^3 = x + 2$),
3. $g_3(x) = x - \dfrac{x^3 - x - 2}{3x^2 - 1}$ (Newton's rearrangement).

Conversely $x = \cos x$ becomes $f(x) = x - \cos x = 0$.

At the root $x^{*} \approx 1.5214$: $g_1'(x^{*}) = 3x^{*2} \approx 6.94$ (diverges), $g_2'(x^{*}) = \tfrac{1}{3}(x^{*}+2)^{-2/3} \approx 0.144$ (converges), $g_3'(x^{*}) = 0$ (quadratic).

$$
\boxed{\text{one equation, many maps } g \text{ — with derivatives } 6.94,\ 0.144,\ 0 \text{: divergent, linear, quadratic}}
$$

*Key takeaway:* The equation does not determine the iteration; the chosen rearrangement's slope at the solution does.

### Problem L0.2: Reading the cobweb

For an iteration $x_{n+1} = g(x_n)$ with $g'(x^{*}) = -0.8$ at the fixed point, describe (a) the qualitative behavior of the iterates, (b) the convergence rate, (c) one practical advantage this sign of $g'$ provides.

**Solution.**

(a) Since $g'(x^{*}) \lt 0$, errors flip sign each step: $e_{n+1} \approx -0.8\,e_n$, so iterates *alternate* around $x^{*}$ — an inward spiral on the cobweb diagram.

(b) $\lvert g'(x^{*}) \rvert = 0.8 \lt 1$: linear convergence with asymptotic rate 0.8 — about $\frac{\log 10}{-\log 0.8} \approx 10.3$ iterations per decimal digit. Slow but sure.

(c) Because consecutive iterates straddle $x^{*}$, the interval $[\min(x_n, x_{n+1}), \max(x_n, x_{n+1})]$ *brackets* the fixed point — a free rigorous error bound $\lvert x_n - x^{*} \rvert \le \lvert x_{n+1} - x_n \rvert$, unavailable when $g' \gt 0$.

$$
\boxed{\text{alternating spiral, linear rate } 0.8 \text{ (}\approx 10 \text{ iters/digit), consecutive iterates bracket } x^{*}}
$$

*Key takeaway:* The sign of $g'(x^{*})$ is diagnostic — oscillation means negative slope and gives self-validating error bounds.

### Problem L0.3: What contraction buys you

List the four distinct guarantees of the Banach fixed-point theorem and explain, for each, what weaker local theory ($\lvert g'(x^{*}) \rvert \lt 1$ only) fails to provide.

**Solution.**

Banach (contraction with $L \lt 1$ on a closed invariant set $D$) guarantees:

1. **Existence** of a fixed point in $D$ — local theory *assumes* a fixed point exists.
2. **Uniqueness** in all of $D$ — local theory says nothing about other fixed points elsewhere.
3. **Global convergence** from *every* $x_0 \in D$ — local theory only covers an unspecified "sufficiently small" neighborhood.
4. **Computable error bounds** (a priori $\frac{L^n}{1-L}\lvert x_1 - x_0 \rvert$, a posteriori $\frac{L}{1-L}\lvert x_n - x_{n-1} \rvert$) — local theory gives an asymptotic rate but no finite-$n$ certificate.

$$
\boxed{\text{existence + uniqueness + global convergence + explicit bounds — all four require the uniform } L \lt 1}
$$

*Key takeaway:* Verifying a uniform contraction constant on an invariant set upgrades an asymptotic hope into a finite-iteration certificate.

### Problem L0.4: Order from vanishing derivatives

An iteration function satisfies $g(x^{*}) = x^{*}$, $g'(x^{*}) = g''(x^{*}) = 0$, $g'''(x^{*}) = 12$. State the convergence order and asymptotic error constant, and how many iterations reduce an initial error of $10^{-2}$ below $10^{-15}$.

**Solution.**

By Theorem 3 the order is $p = 3$ (first nonvanishing derivative), with constant

$$
C = \frac{\lvert g'''(x^{*}) \rvert}{3!} = \frac{12}{6} = 2, \qquad e_{n+1} \approx 2\, e_n^3 .
$$

Track $t_n = \sqrt{2}\,e_n$ (so that $t_{n+1} = t_n^3$): $t_0 = \sqrt{2} \times 10^{-2} \approx 1.41 \times 10^{-2}$, and $t_n = t_0^{3^{n}}$.

- $n = 1$: $e_1 \approx 2 \times 10^{-6}$
- $n = 2$: $e_2 \approx 2(2\times10^{-6})^3 = 1.6 \times 10^{-17} \lt 10^{-15}$ ✓

$$
\boxed{p = 3,\ C = 2; \text{ two iterations suffice } (10^{-2} \to 2\times10^{-6} \to 1.6\times10^{-17})}
$$

*Key takeaway:* Cubic convergence *triples* correct digits per step — the count of vanishing derivatives is the order.

## Level 1 — Foundation

### Problem L1.1: The Dottie number

Show that $g(x) = \cos x$ has a unique fixed point in $[0, 1]$, that the iteration converges from any $x_0 \in [0, 1]$, and estimate the number of iterations to reach error $10^{-6}$ from $x_0 = 1$. (Fixed point: $x^{*} \approx 0.739085$.)

**Solution.**

**Invariance:** for $x \in [0, 1]$, $\cos x \in [\cos 1, 1] = [0.5403, 1] \subset [0, 1]$. ✓

**Contraction:** $\lvert g'(x) \rvert = \lvert \sin x \rvert \le \sin 1 = 0.8415 \lt 1$ on $[0,1]$, so $L = 0.8415$.

Banach: unique fixed point, global convergence on $[0,1]$. A priori bound with $\lvert x_1 - x_0 \rvert = \lvert \cos 1 - 1 \rvert = 0.4597$:

$$
\frac{L^{n}}{1 - L}\, (0.4597) \le 10^{-6} \iff L^{n} \le \frac{10^{-6} \times 0.1585}{0.4597} = 3.45 \times 10^{-7} .
$$

$n \ge \frac{\log(3.45 \times 10^{-7})}{\log 0.8415} = \frac{-14.88}{-0.1726} \approx 86.2$, so $n = 87$ guaranteed. (In practice the local rate $\sin x^{*} \approx 0.6736$ needs only $\approx 35$ iterations — a priori bounds are conservative.)

$$
\boxed{L = \sin 1 \approx 0.8415; \text{ guaranteed convergence, } \le 87 \text{ iterations for } 10^{-6}}
$$

*Key takeaway:* Verify invariance *and* contraction; then the geometric bound turns "it converges" into "done by iteration 87."

### Problem L1.2: Two rearrangements, two fates

To solve $x^2 - x - 1 = 0$ for the golden ratio $\varphi = \frac{1+\sqrt{5}}{2} \approx 1.6180$, compare the iterations (a) $x_{n+1} = x_n^2 - 1$ and (b) $x_{n+1} = \sqrt{x_n + 1}$, both from $x_0 = 1.5$. Compute two steps of each and explain the outcomes via $g'(\varphi)$.

**Solution.**

**(a)** $g_a(x) = x^2 - 1$, $g_a'(\varphi) = 2\varphi \approx 3.236 \gt 1$ — repelling.

- $x_1 = 1.5^2 - 1 = 1.25$, $x_2 = 1.25^2 - 1 = 0.5625$: moving *away* from 1.618 (later iterates wander chaotically / diverge).

**(b)** $g_b(x) = \sqrt{x+1}$, $g_b'(\varphi) = \frac{1}{2\sqrt{\varphi + 1}} = \frac{1}{2\varphi} \approx 0.309$ — attracting.

- $x_1 = \sqrt{2.5} = 1.5811$, $x_2 = \sqrt{2.5811} = 1.6066$: errors $0.118 \to 0.0369 \to 0.0114$, ratio $\approx 0.31 = g_b'(\varphi)$ as predicted.

$$
\boxed{g_a' (\varphi) \approx 3.24 \implies \text{divergence}; \quad g_b'(\varphi) = \tfrac{1}{2\varphi} \approx 0.31 \implies \text{linear convergence, rate } 0.31}
$$

*Key takeaway:* Before iterating, differentiate: the size of $\lvert g' \rvert$ at the solution — not the algebraic correctness of the rearrangement — decides convergence.

### Problem L1.3: A posteriori stopping

An iteration with contraction constant $L = 0.9$ reaches $\lvert x_n - x_{n-1} \rvert = 10^{-8}$. (a) Bound the true error. (b) How does the answer change for $L = 0.5$? (c) What step size must be reached to *guarantee* error $10^{-8}$ when $L = 0.9$?

**Solution.**

A posteriori bound: $\lvert x_n - x^{*} \rvert \le \dfrac{L}{1-L}\, \lvert x_n - x_{n-1} \rvert$.

(a) $L = 0.9$: factor $\frac{0.9}{0.1} = 9$, so error $\le 9 \times 10^{-8}$ — nearly an order of magnitude larger than the step.

(b) $L = 0.5$: factor $\frac{0.5}{0.5} = 1$, error $\le 10^{-8}$ — the step size is an honest error estimate.

(c) Require $\frac{0.9}{0.1} \lvert x_n - x_{n-1} \rvert \le 10^{-8} \iff \lvert x_n - x_{n-1} \rvert \le 1.11 \times 10^{-9}$.

$$
\boxed{\text{(a) } \le 9 \times 10^{-8}; \text{ (b) } \le 10^{-8}; \text{ (c) step } \le 1.1 \times 10^{-9}}
$$

*Key takeaway:* "The iterates stopped moving" understates the error by $\frac{L}{1-L}$ — fatal for slowly contracting iterations unless the factor is applied.

### Problem L1.4: Kepler's equation as a fixed point

For $E = M + e \sin E$ with $e = 0.2$, $M = 0.8$: verify the contraction hypothesis on $\mathbb{R}$, run three iterations from $E_0 = M$, and bound the error of $E_3$ a priori.

**Solution.**

$g(E) = 0.8 + 0.2\sin E$; $\lvert g'(E) \rvert = 0.2\lvert \cos E \rvert \le 0.2 = L \lt 1$ everywhere, and $g$ maps $\mathbb{R}$ into $[0.6, 1.0]$ — Banach applies globally.

- $E_1 = 0.8 + 0.2\sin(0.8) = 0.8 + 0.2(0.71736) = 0.943472$
- $E_2 = 0.8 + 0.2\sin(0.943472) = 0.8 + 0.2(0.80962) = 0.961924$
- $E_3 = 0.8 + 0.2\sin(0.961924) = 0.8 + 0.2(0.82043) = 0.964086$

A priori bound: $\lvert E_3 - E^{*} \rvert \le \frac{L^3}{1-L}\lvert E_1 - E_0 \rvert = \frac{0.008}{0.8}(0.143472) = 1.43 \times 10^{-3}$. (True $E^{*} \approx 0.964334$, actual error $2.5 \times 10^{-4}$.)

$$
\boxed{E_3 = 0.964086, \quad \lvert E_3 - E^{*} \rvert \le 1.4 \times 10^{-3} \text{ (guaranteed)}, \quad L = e = 0.2}
$$

*Key takeaway:* For orbital eccentricity $e$, the natural iteration contracts with $L = e$ — centuries of astronomers solved Kepler exactly this way.

### Problem L1.5: Aitken in action

The iteration $x_{n+1} = e^{-x_n}$ from $x_0 = 0.5$ gives $x_0 = 0.5$, $x_1 = 0.606531$, $x_2 = 0.545239$, $x_3 = 0.579703$. Apply Aitken's $\Delta^2$ to $(x_0, x_1, x_2)$ and to $(x_1, x_2, x_3)$, and compare with the true fixed point $x^{*} = 0.567143$ (where $x^{*} = e^{-x^{*}}$).

**Solution.**

Aitken: $\hat{x}_n = x_n - \dfrac{(x_{n+1} - x_n)^2}{x_{n+2} - 2x_{n+1} + x_n}$.

**From $(x_0, x_1, x_2)$:** $\Delta_1 = 0.106531$, $\Delta_2 = -0.061292$; denominator $= x_2 - 2x_1 + x_0 = -0.167823$.

$$
\hat{x}_0 = 0.5 - \frac{(0.106531)^2}{-0.167823} = 0.5 + 0.067623 = 0.567623 .
$$

**From $(x_1, x_2, x_3)$:** $\Delta = -0.061292$, denominator $= x_3 - 2x_2 + x_1 = 0.095756$.

$$
\hat{x}_1 = 0.606531 - \frac{(-0.061292)^2}{0.095756} = 0.606531 - 0.039233 = 0.567298 .
$$

Errors: raw $x_3$ has error $1.26 \times 10^{-2}$; $\hat{x}_0$ has error $4.8 \times 10^{-4}$, $\hat{x}_1$ has $1.55 \times 10^{-4}$ — two orders of magnitude gained using the *same* function evaluations.

$$
\boxed{\hat{x}_0 = 0.567623,\ \hat{x}_1 = 0.567298 \text{ vs } x^{*} = 0.567143 \text{ — errors shrink } \sim 100\times}
$$

*Key takeaway:* Aitken extrapolates the geometric error pattern to its limit — acceleration is free intelligence applied to numbers you already computed.

### Problem L1.6: Optimal relaxation

The iteration $x_{n+1} = g(x_n)$ with $g'(x^{*}) = -3$ diverges. Show that the relaxed map $g_\omega(x) = (1-\omega)x + \omega g(x)$ converges locally for $0 \lt \omega \lt \tfrac{1}{2}$, find the optimal $\omega$, and the resulting convergence order.

**Solution.**

$g_\omega'(x^{*}) = (1 - \omega) + \omega g'(x^{*}) = 1 - \omega + \omega(-3) = 1 - 4\omega$.

**Convergence condition:** $\lvert 1 - 4\omega \rvert \lt 1 \iff 0 \lt 4\omega \lt 2 \iff 0 \lt \omega \lt \tfrac{1}{2}$. ✓

**Optimum:** minimize $\lvert 1 - 4\omega \rvert$: choose $\omega^{*} = \tfrac{1}{4}$, giving $g_\omega'(x^{*}) = 0$ — the linear error term vanishes entirely, so convergence becomes (at least) *quadratic* (order given by the first nonvanishing derivative of $g_\omega$, generically $p = 2$).

Note $x^{*}$ is unchanged: $g_\omega(x^{*}) = (1-\omega)x^{*} + \omega x^{*} = x^{*}$ for every $\omega$.

$$
\boxed{\text{converges for } \omega \in (0, \tfrac{1}{2}); \quad \omega^{*} = \tfrac{1}{4} \text{ gives } g_\omega'(x^{*}) = 0 \text{ — quadratic}}
$$

*Key takeaway:* Damping doesn't change the fixed point, only the slope at it — and the right damping can transform divergence into superlinear convergence.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: Value iteration is a contraction

Prove that the Bellman optimality operator $(TV)(s) = \max_a \bigl[ r(s,a) + \gamma \sum_{s'} P(s' \mid s,a) V(s') \bigr]$ satisfies $\lVert TV - TW \rVert_\infty \le \gamma \lVert V - W \rVert_\infty$, and compute how many iterations guarantee $\lVert V_n - V^{*} \rVert_\infty \le 10^{-3}$ for $\gamma = 0.99$ with $\lVert V_1 - V_0 \rVert_\infty = 1$.

**Solution.**

**Contraction.** Fix $s$ and use $\lvert \max_a \phi(a) - \max_a \psi(a) \rvert \le \max_a \lvert \phi(a) - \psi(a) \rvert$:

$$
\lvert (TV)(s) - (TW)(s) \rvert \le \max_a \gamma \left\lvert \sum_{s'} P(s' \mid s,a)\bigl(V(s') - W(s')\bigr) \right\rvert \le \gamma \max_a \sum_{s'} P(s' \mid s,a)\, \lVert V - W \rVert_\infty = \gamma \lVert V - W \rVert_\infty,
$$

since the $P(\cdot \mid s,a)$ sum to 1. Taking the sup over $s$ proves the claim; the reward term cancels.

**Iteration count.** A priori bound: $\lVert V_n - V^{*} \rVert_\infty \le \frac{\gamma^n}{1-\gamma}\lVert V_1 - V_0 \rVert_\infty = \frac{0.99^n}{0.01} = 100 \times 0.99^n \le 10^{-3}$ requires

$$
0.99^{n} \le 10^{-5} \iff n \ge \frac{-5\ln 10}{\ln 0.99} = \frac{11.513}{0.010050} \approx 1145.6 .
$$

$$
\boxed{T \text{ is a } \gamma\text{-contraction; } n = 1146 \text{ sweeps certify } 10^{-3} \text{ accuracy at } \gamma = 0.99}
$$

*Key takeaway:* RL's discount factor *is* a Lipschitz constant — and the $\frac{1}{1-\gamma}$ blow-up explains why long-horizon ($\gamma \to 1$) RL is intrinsically slow to converge.

### Problem L2.2: Gradient descent step-size window

For $f(x) = \tfrac{1}{2} x^{T} A x - b^{T} x$ with SPD $A$ having eigenvalues in $[\mu, L_s] = [1, 100]$, gradient descent is the fixed-point map $g(x) = x - \alpha(Ax - b)$. Find the exact convergence condition on $\alpha$, the optimal $\alpha$, and the optimal rate.

**Solution.**

$g$ is affine with Jacobian $M = I - \alpha A$; the error evolves as $e_{n+1} = M e_n$. Convergence for every start iff $\rho(M) \lt 1$. Eigenvalues of $M$ are $1 - \alpha\lambda_i$, so

$$
\rho(M) = \max\bigl( \lvert 1 - \alpha\mu \rvert,\ \lvert 1 - \alpha L_s \rvert \bigr) \lt 1 \iff 0 \lt \alpha \lt \frac{2}{L_s} = 0.02 .
$$

**Optimum:** the max of the two V-shaped functions is minimized where they are equal: $\alpha\mu - 1$ rising branch meets $1 - \alpha L_s$... solve $1 - \alpha\mu = \alpha L_s - 1$:

$$
\alpha^{*} = \frac{2}{\mu + L_s} = \frac{2}{101} \approx 0.0198, \qquad \rho^{*} = \frac{L_s - \mu}{L_s + \mu} = \frac{99}{101} \approx 0.9802 .
$$

With $\kappa = L_s/\mu = 100$: $\rho^{*} = \frac{\kappa - 1}{\kappa + 1}$, needing $\approx \frac{\ln 10}{-\ln 0.9802} \approx 115$ iterations per digit.

$$
\boxed{0 \lt \alpha \lt \tfrac{2}{L_s}; \quad \alpha^{*} = \tfrac{2}{\mu + L_s}, \quad \rho^{*} = \tfrac{\kappa - 1}{\kappa + 1} = 0.9802}
$$

*Key takeaway:* The classic "learning rate must be below $2/L$" rule is spectral-radius contraction analysis; ill-conditioning ($\kappa$ large) drives the rate toward 1 — the opening for momentum and preconditioning.

### Problem L2.3: Jacobi iteration via spectral radius

For the system $Ax = b$ with $A = \begin{bmatrix} 4 & 1 \\ 2 & 3 \end{bmatrix}$, Jacobi iteration is $x_{n+1} = D^{-1}(b - Rx_n)$ with $D = \mathrm{diag}(4, 3)$, $R = A - D$. Compute the iteration matrix, its spectral radius, decide convergence, and estimate iterations per decimal digit.

**Solution.**

$$
M = -D^{-1}R = -\begin{bmatrix} 1/4 & 0 \\ 0 & 1/3 \end{bmatrix} \begin{bmatrix} 0 & 1 \\ 2 & 0 \end{bmatrix} = \begin{bmatrix} 0 & -1/4 \\ -2/3 & 0 \end{bmatrix} .
$$

Eigenvalues: $\lambda^2 = \frac{1}{4}\cdot\frac{2}{3} = \frac{1}{6}$, so $\lambda = \pm\frac{1}{\sqrt{6}}$ and

$$
\rho(M) = \frac{1}{\sqrt{6}} \approx 0.4082 \lt 1 \implies \text{convergent from any } x_0 .
$$

(Consistent with $A$ being strictly diagonally dominant by rows: $4 \gt 1$, $3 \gt 2$.) Iterations per digit: $\frac{\ln 10}{-\ln 0.4082} = \frac{2.3026}{0.8959} \approx 2.57$.

$$
\boxed{\rho(M) = \tfrac{1}{\sqrt{6}} \approx 0.408 \text{ — converges, } \approx 2.6 \text{ iterations per decimal digit}}
$$

*Key takeaway:* For linear iterations the entire convergence question collapses to one number, $\rho(M)$ — norms and starting points are irrelevant asymptotically.

### Problem L2.4: PageRank damping

PageRank iterates $\pi_{n+1}^{T} = d\,\pi_n^{T} S + (1-d)\,v^{T}$ with damping $d = 0.85$, stochastic matrix $S$, and teleport vector $v$. Show the map is a contraction in the $\ell_1$ norm with constant $d$, and bound the iterations needed for $\lVert \pi_n - \pi^{*} \rVert_1 \le 10^{-8}$ from any stochastic start.

**Solution.**

For probability (row) vectors $\pi, \sigma$: the difference $u^{T} = \pi^{T} - \sigma^{T}$ has $\sum_i u_i = 0$, and the affine parts cancel:

$$
\lVert G(\pi) - G(\sigma) \rVert_1 = d\, \lVert u^{T} S \rVert_1 \le d\, \lVert u \rVert_1 \lVert S \rVert_{1 \to 1} = d\, \lVert \pi - \sigma \rVert_1,
$$

since column-stochastic multiplication cannot increase the $\ell_1$ norm of a vector ($\lvert \sum_j u_j S_{ji} \rvert$ summed over $i$ is at most $\sum_j \lvert u_j \rvert \sum_i S_{ji} = \sum_j \lvert u_j \rvert$). So $L = d = 0.85$ and Banach gives a unique stationary $\pi^{*}$ with global convergence.

Since $\lVert \pi_0 - \pi^{*} \rVert_1 \le 2$ for any two probability vectors,

$$
\lVert \pi_n - \pi^{*} \rVert_1 \le 2\,(0.85)^n \le 10^{-8} \iff n \ge \frac{\ln(2 \times 10^{8})}{-\ln 0.85} = \frac{19.11}{0.16252} \approx 117.6 .
$$

$$
\boxed{L = d = 0.85; \quad 118 \text{ power iterations certify } \lVert \pi_n - \pi^{*} \rVert_1 \le 10^{-8} \text{ for a web of any size}}
$$

*Key takeaway:* The damping factor was chosen to *make* the web a contraction — convergence in ~100 iterations independent of graph size is engineered, not lucky.

### Problem L2.5: Implicit Euler's inner fixed-point loop

A backward-Euler step for $y' = f(y)$ solves $z = y_n + h f(z)$. For the stiff ODE $f(y) = -50 y$ with $h = 0.1$: (a) show direct fixed-point iteration on $z$ fails; (b) show Newton's method on $F(z) = z - y_n - h f(z)$ succeeds in one step; (c) state the general step-size restriction for direct iteration.

**Solution.**

(a) The map is $g(z) = y_n + h f(z) = y_n - 5z$, so $\lvert g'(z) \rvert = h \cdot 50 = 5 \gt 1$: repelling — direct iteration diverges (each sweep quintuples the error).

(b) $F(z) = z - y_n + 5z = 6z - y_n$ is *linear*, so Newton solves it exactly in one step:

$$
z = \frac{y_n}{6} \quad \left( = \frac{y_n}{1 - h\lambda} \text{ with } \lambda = -50 \right).
$$

(c) In general direct iteration contracts iff $h \bigl\lvert \frac{\partial f}{\partial y} \bigr\rvert \lt 1$, i.e. $h \lt 1/\lvert \lambda \rvert$ — exactly the tiny steps implicit methods were introduced to avoid. Hence stiff solvers pair implicit formulas with Newton (not Picard) inner iterations.

$$
\boxed{\text{direct: diverges since } h\lvert\lambda\rvert = 5 \gt 1; \text{ Newton: } z = y_n/6 \text{ exactly; direct needs } h \lt 1/\lvert\lambda\rvert}
$$

*Key takeaway:* An implicit method is only as unconditional as its inner solver — Picard iteration silently reimposes the explicit step-size limit that implicitness was meant to remove.

### Problem L2.6: Deep equilibrium layers

A deep equilibrium (DEQ) layer defines its output as the fixed point $z^{*} = \tanh(Wz^{*} + Ux)$. Suppose $\lVert W \rVert_2 = 0.9$. (a) Prove the layer is well-defined (unique $z^{*}$, any input $x$). (b) Bound the iterations to solve it to $10^{-5}$ from $z_0 = 0$ given the first step has $\lVert z_1 - z_0 \rVert_2 \le 1$. (c) Explain why backprop can avoid storing the iterates.

**Solution.**

(a) $g(z) = \tanh(Wz + Ux)$ (componentwise $\tanh$). Since $\tanh$ is 1-Lipschitz componentwise,

$$
\lVert g(z) - g(w) \rVert_2 \le \lVert W(z - w) \rVert_2 \le \lVert W \rVert_2\, \lVert z - w \rVert_2 = 0.9\, \lVert z - w \rVert_2 :
$$

a contraction on all of $\mathbb{R}^{d}$ for *every* $x$ — Banach gives a unique equilibrium $z^{*}(x)$, so the layer is a well-defined function.

(b) A priori: $\lVert z_n - z^{*} \rVert \le \frac{0.9^{n}}{0.1} \times 1 \le 10^{-5} \iff 0.9^n \le 10^{-6} \iff n \ge \frac{6\ln 10}{-\ln 0.9} \approx 131.1$, so 132 iterations (Anderson acceleration cuts this to a dozen in practice).

(c) At the solution, the implicit function theorem differentiates $z^{*} = g(z^{*}, x)$ directly:

$$
\frac{\partial z^{*}}{\partial x} = \bigl( I - \partial_z g(z^{*}) \bigr)^{-1} \partial_x g(z^{*}),
$$

where $I - \partial_z g$ is invertible because $\lVert \partial_z g \rVert \le 0.9 \lt 1$. Gradients need only the converged $z^{*}$ — memory cost is $O(1)$ in depth, the DEQ selling point.

$$
\boxed{L = \lVert W \rVert_2 = 0.9 \implies \text{unique equilibrium; } \le 132 \text{ iterations; implicit-function backprop through } (I - \partial_z g)^{-1}}
$$

*Key takeaway:* Constraining $\lVert W \rVert \lt 1$ makes an "infinitely deep" network a Banach fixed point — existence, uniqueness, convergence and differentiability all follow from one contraction estimate.

## Level 3 — Challenge

### Problem L3.1: Error of the a priori bound is sharp

Show by explicit example that the Banach a priori bound $\lvert x_n - x^{*} \rvert \le \frac{L^{n}}{1-L}\lvert x_1 - x_0 \rvert$ cannot be improved: exhibit a contraction for which the bound is attained asymptotically up to any factor $c \lt 1$.

**Solution.**

Take the linear contraction $g(x) = Lx$ on $\mathbb{R}$ ($0 \lt L \lt 1$), fixed point $x^{*} = 0$, start $x_0 = 1$.

Iterates: $x_n = L^{n}$, first step $\lvert x_1 - x_0 \rvert = 1 - L$. The bound reads

$$
\lvert x_n - x^{*} \rvert \le \frac{L^{n}}{1-L}(1 - L) = L^{n},
$$

and indeed $\lvert x_n - 0 \rvert = L^{n}$ — **equality for every $n$**. So no smaller constant can replace $\frac{1}{1-L}$ uniformly: for any candidate bound $c\,\frac{L^n}{1-L}\lvert x_1 - x_0 \rvert$ with $c \lt 1$, this example violates it at every step.

The bound's structure is forced: the tail $\sum_{k \ge n} L^k \lvert x_1 - x_0 \rvert$ of the telescoping series is exactly $\frac{L^n}{1-L}\lvert x_1 - x_0 \rvert$, and the linear map realizes the worst case where every triangle inequality is tight.

$$
\boxed{g(x) = Lx \text{ attains } \lvert x_n - x^{*} \rvert = \frac{L^{n}}{1-L}\lvert x_1 - x_0 \rvert \text{ exactly — the Banach bound is sharp}}
$$

*Key takeaway:* Banach's constants are not analyst's slack — a one-dimensional linear map saturates them, so any tighter general claim is false.

### Problem L3.2: Steffensen's method has order 2

Define Steffensen's iteration for solving $x = g(x)$: from $x_n$, set $y = g(x_n)$, $z = g(y)$, and

$$
x_{n+1} = x_n - \frac{(y - x_n)^2}{z - 2y + x_n} .
$$

Prove that at a fixed point with $g'(x^{*}) \neq 1$, convergence is (at least) quadratic, using only Taylor expansions.

**Solution.**

Let $e = x_n - x^{*}$, $\lambda = g'(x^{*})$, $\beta = \tfrac{1}{2}g''(x^{*})$. Expanding ($O(e^3)$ terms tracked):

$$
y - x^{*} = \lambda e + \beta e^2 + O(e^3), \qquad z - x^{*} = \lambda(\lambda e + \beta e^2) + \beta(\lambda e)^2 + O(e^3) = \lambda^2 e + \beta\lambda(1 + \lambda)e^2 + O(e^3).
$$

Numerator:

$$
(y - x_n)^2 = \bigl( (\lambda - 1)e + \beta e^2 + O(e^3) \bigr)^2 = (\lambda - 1)^2 e^2 + 2\beta(\lambda - 1) e^3 + O(e^4).
$$

Denominator:

$$
z - 2y + x_n = (\lambda^2 - 2\lambda + 1) e + \beta(\lambda^2 + \lambda - 2) e^2 + O(e^3) = (\lambda - 1)^2 e \left( 1 + \frac{\beta(\lambda + 2)}{\lambda - 1}\, e + O(e^2) \right),
$$

using $\lambda^2 + \lambda - 2 = (\lambda - 1)(\lambda + 2)$. Then, expanding the quotient with $(1 + \varepsilon)^{-1} = 1 - \varepsilon + O(\varepsilon^2)$:

$$
\frac{(y - x_n)^2}{z - 2y + x_n} = e\left( 1 + \frac{2\beta}{\lambda - 1} e \right)\left( 1 - \frac{\beta(\lambda+2)}{\lambda - 1} e \right) + O(e^3) = e - \frac{\beta \lambda}{\lambda - 1} e^2 + O(e^3),
$$

since $\frac{2\beta - \beta(\lambda + 2)}{\lambda - 1} = \frac{-\beta\lambda}{\lambda - 1}$. Finally

$$
x_{n+1} - x^{*} = e - \left( e - \frac{\beta\lambda}{\lambda - 1} e^2 \right) + O(e^3) = \frac{\beta \lambda}{\lambda - 1}\, e^2 + O(e^3). \qquad \blacksquare
$$

$$
\boxed{e_{n+1} = \frac{\lambda\, g''(x^{*})}{2(\lambda - 1)}\, e_n^2 + O(e_n^3) \text{ — quadratic, with no derivatives evaluated}}
$$

*Key takeaway:* Steffensen buys Newton's order with two extra $g$-evaluations instead of a derivative — Aitken extrapolation applied dynamically is a second-order method.

### Problem L3.3: Spectral radius is the exact asymptotic rate

For the affine iteration $x_{n+1} = Mx_n + c$ with $\rho(M) \lt 1$ and fixed point $x^{*}$, prove that (a) for every $\varepsilon \gt 0$ there is a norm and constant with $\lVert e_n \rVert \le C(\rho(M) + \varepsilon)^n \lVert e_0 \rVert$, and (b) $\rho(M) \lt 1$ is also *necessary* for convergence from every $x_0$. (Use the Jordan form.)

**Solution.**

Errors satisfy $e_{n} = M^{n} e_0$ (subtract $x^{*} = Mx^{*} + c$ from the iteration).

**(a) Sufficiency with rate $\rho + \varepsilon$.** Write $M = PJP^{-1}$ in Jordan form, $J = D + N$ with $D$ diagonal (eigenvalues) and $N$ nilpotent ($N^{d} = 0$). Scale each Jordan chain by $S_\delta = \mathrm{diag}(1, \delta, \delta^2, \ldots)$: then $S_\delta^{-1} J S_\delta = D + \delta N$. Define the norm $\lVert x \rVert_\delta = \lVert (PS_\delta)^{-1} x \rVert_\infty$. In it,

$$
\lVert M \rVert_\delta = \lVert D + \delta N \rVert_\infty \le \rho(M) + \delta .
$$

Choosing $\delta = \varepsilon$ gives $\lVert e_n \rVert_\delta \le (\rho + \varepsilon)^n \lVert e_0 \rVert_\delta$; norm equivalence on $\mathbb{R}^d$ converts this to any fixed norm at the cost of a constant $C$.

**(b) Necessity.** Suppose $\rho(M) \ge 1$: take an eigenpair $Mv = \lambda v$ with $\lvert \lambda \rvert \ge 1$ (if complex, take real/imaginary parts). Choose $x_0 = x^{*} + v$; then $e_n = \lambda^{n} v$ and $\lVert e_n \rVert = \lvert \lambda \rvert^{n} \lVert v \rVert$ does not tend to 0 — convergence fails from this start. Hence global convergence forces $\rho(M) \lt 1$. Moreover the rate cannot beat $\rho$: along that eigenvector $\lVert e_n \rVert^{1/n} \to \lvert \lambda \rvert$, and maximizing over eigenvalues shows $\lim \sup \lVert M^n \rVert^{1/n} = \rho(M)$ (Gelfand). $\blacksquare$

$$
\boxed{\text{convergence } \forall x_0 \iff \rho(M) \lt 1; \text{ asymptotic rate exactly } \rho(M)}
$$

*Key takeaway:* Norm estimates can lie in both directions for finitely many steps; the spectral radius — via Gelfand's formula — is the one true asymptotic contraction factor.

### Problem L3.4: A non-contraction with a unique attracting fixed point

Let $g(x) = x + e^{-x} - 1$ on $[0, \infty)$. Show that (a) $x^{*} = 0$ is the unique fixed point, (b) $g$ is *not* a contraction on $[0, \infty)$ in the Banach sense (no uniform $L \lt 1$), yet (c) the iteration converges to 0 from every $x_0 \gt 0$ — but sublinearly, with $x_n \sim 2/n$. Reconcile with the theory.

**Solution.**

(a) $g(x) = x \iff e^{-x} = 1 \iff x = 0$. ✓

(b) $g'(x) = 1 - e^{-x}$, so $\sup_{x \ge 0} g'(x) = 1$ (approached as $x \to \infty$) and, crucially, $g'(0) = 1$: no uniform $L \lt 1$ exists on any interval containing 0 — Banach and the local theorem ($\lvert g' \rvert \lt 1$ *at* the fixed point) both fail to apply.

(c) **Convergence.** For $x \gt 0$: $e^{-x} \lt 1$ gives $g(x) \lt x$, and $e^{-x} \gt 1 - x$ gives $g(x) \gt 0$. So $(x_n)$ is strictly decreasing and bounded below — it converges, and its limit is a fixed point, hence 0.

**Rate.** For small $x$, $g(x) = x - \frac{x^2}{2} + \frac{x^3}{6} - \cdots$. Set $u_n = 1/x_n$:

$$
u_{n+1} = \frac{1}{x_n - \tfrac{x_n^2}{2} + O(x_n^3)} = \frac{1}{x_n}\left( 1 + \frac{x_n}{2} + O(x_n^2) \right) = u_n + \frac{1}{2} + O(x_n).
$$

So $u_n = \frac{n}{2} + o(n)$, i.e. $x_n \sim \frac{2}{n}$: *sublinear* — a million iterations for six digits.

**Reconciliation:** the theory is not contradicted but *vindicated*: with $g'(x^{*}) = 1$ the linear term of the error recursion is neutral, and convergence (here driven by the negative quadratic term) drops below every geometric rate. Banach's uniform $L \lt 1$ is exactly what rules such degeneracy out. $\blacksquare$

$$
\boxed{g'(0) = 1: \text{ convergent but sublinear, } x_n \sim \tfrac{2}{n} \text{ — the boundary case the contraction hypothesis excludes}}
$$

*Key takeaway:* $\lvert g'(x^{*}) \rvert = 1$ is a genuine phase boundary — convergence may survive, but geometric speed does not, which is why every theorem in this topic demands strict inequality.
