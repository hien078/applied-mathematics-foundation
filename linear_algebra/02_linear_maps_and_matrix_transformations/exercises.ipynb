### Exercise 1

Linearity of Affine Maps

Determine whether the transformation $T: \mathbb{R}^2 \to \mathbb{R}^2$ given by $T(x,y) = (x+1, y)$ is a linear map.

**Intuition:** Linear maps must preserve the zero vector, mapping the origin to the origin.

**Solution & Proof:**

We test the zero vector mapping condition:

$$
T(0,0) = (0+1, 0) = (1, 0) \neq (0,0)
$$

Since $T(\mathbf{0}) \neq \mathbf{0}$, the transformation fails the necessary condition for linearity.

$$
\boxed{\text{No, } T \text{ is not linear.}}
$$

> **Note:** Affine transformations with non-zero translation vectors are not linear transformations.

### Exercise 2

Identity Map Representation

Find the matrix representation of the identity transformation $I: \mathbb{R}^n \to \mathbb{R}^n$ with respect to the standard basis.

**Intuition:** The identity map sends every standard basis vector $\mathbf{e}_i$ to itself.

**Solution & Proof:**

For each standard basis vector $\mathbf{e}_i \in \mathbb{R}^n$, we have $I(\mathbf{e}_i) = \mathbf{e}_i$.

The columns of the matrix representation are the images of the standard basis vectors:

$$
[I] = \begin{pmatrix} I(\mathbf{e}_1) & I(\mathbf{e}_2) & \dots & I(\mathbf{e}_n) \end{pmatrix} = \begin{pmatrix} \mathbf{e}_1 & \mathbf{e}_2 & \dots & \mathbf{e}_n \end{pmatrix} = \mathbf{I}_n
$$

$$
\boxed{\mathbf{I}_n}
$$

> **Note:** With respect to identical domain and codomain bases, the identity transformation is represented by the identity matrix.

### Exercise 3

Linearity of Reflection Maps

Show that the reflection map across the $x$-axis $T(x, y) = (x, -y)$ is a linear map and find its standard matrix.

**Intuition:** Reflecting vectors across a linear axis preserves vector addition and scalar scaling geometrically.

**Solution & Proof:**

1. **Additivity:** For $\mathbf{u} = (x_1, y_1)$ and $\mathbf{v} = (x_2, y_2)$:
   
$$
T(\mathbf{u} + \mathbf{v}) = T(x_1+x_2, y_1+y_2) = (x_1+x_2, -(y_1+y_2)) = (x_1, -y_1) + (x_2, -y_2) = T(\mathbf{u}) + T(\mathbf{v})
$$

2. **Homogeneity:** For scalar $c \in \mathbb{R}$:
   
$$
T(c\mathbf{u}) = T(cx_1, cy_1) = (cx_1, -cy_1) = c(x_1, -y_1) = cT(\mathbf{u})
$$

Evaluating on standard basis vectors:

$$
T(1,0) = (1, 0), \quad T(0,1) = (0, -1)
$$

$$
\boxed{\mathbf{A} = \begin{pmatrix} 1 & 0 \\ 0 & -1 \end{pmatrix}}
$$

> **Note:** Reflections across lines through the origin are linear maps with orthogonal standard matrices.

### Exercise 5

Non-Linearity of Norm Map

Is the transformation $T: \mathbb{R}^2 \to \mathbb{R}$ defined by $T(x, y) = \sqrt{x^2 + y^2}$ a linear map?

**Intuition:** The Euclidean norm involves non-linear operations (squaring and square roots) and fails homogeneity for negative scalars.

**Solution & Proof:**

We check homogeneity with scalar $c = -1$ and vector $\mathbf{v} = (1, 0)$:

$$
T(-1 \cdot (1,0)) = T(-1, 0) = \sqrt{(-1)^2 + 0^2} = 1
$$

However:

$$
-1 \cdot T(1,0) = -1 \cdot \sqrt{1^2 + 0^2} = -1
$$

Since $T(-\mathbf{v}) \neq -T(\mathbf{v})$, $T$ fails homogeneity.

$$
\boxed{\text{No, } T \text{ is not linear.}}
$$

> **Note:** Norms measure magnitude and obey triangle inequality, but fail linear homogeneity for negative scalars.

### Exercise 6

Finding the Kernel of a Linear Map

Find a basis for the kernel of $T: \mathbb{R}^3 \to \mathbb{R}^2$ defined by $T(x, y, z) = (x + y, y + z)$.

**Intuition:** The kernel consists of all vectors $(x,y,z)$ mapped to $(0,0)$.

**Solution & Proof:**

Set $T(x,y,z) = (0,0)$:

$$
\begin{cases} x + y = 0 \implies x = -y \\ y + z = 0 \implies z = -y \end{cases}
$$

Let $y = -t$ for parameter $t \in \mathbb{R}$. Then $x = t$ and $z = t$.

Vectors in $\ker(T)$ have the form:

$$
\begin{pmatrix} x \\ y \\ z \end{pmatrix} = \begin{pmatrix} t \\ -t \\ t \end{pmatrix} = t \begin{pmatrix} 1 \\ -1 \\ 1 \end{pmatrix}
$$

Since $(1, -1, 1)^T \neq \mathbf{0}$, it forms a basis for $\ker(T)$.

$$
\boxed{\text{Basis for } \ker(T) = \left\{ \begin{pmatrix} 1 \\ -1 \\ 1 \end{pmatrix} \right\}, \quad \dim(\ker(T)) = 1}
$$

> **Note:** The kernel of a linear transformation is the solution space (nullspace) of $T(\mathbf{x}) = \mathbf{0}$.

### Exercise 7

Finding the Image (Range) of a Linear Map

Determine a basis for the image $\operatorname{im}(T)$ of the map $T: \mathbb{R}^3 \to \mathbb{R}^3$ represented by $\mathbf{A} = \begin{pmatrix} 1 & 2 & 3 \\ 2 & 4 & 6 \\ 0 & 1 & 1 \end{pmatrix}$.

**Intuition:** The image of $T$ is the column space of $\mathbf{A}$.

**Solution & Proof:**

We analyze the column vectors $\mathbf{c}_1 = (1, 2, 0)^T$, $\mathbf{c}_2 = (2, 4, 1)^T$, $\mathbf{c}_3 = (3, 6, 1)^T$.

Notice that $\mathbf{c}_3 = \mathbf{c}_1 + \mathbf{c}_2$:

$$
\begin{pmatrix} 1 \\ 2 \\ 0 \end{pmatrix} + \begin{pmatrix} 2 \\ 4 \\ 1 \end{pmatrix} = \begin{pmatrix} 3 \\ 6 \\ 1 \end{pmatrix}
$$

Thus $\mathbf{c}_3$ is linearly dependent.
The first two columns $\mathbf{c}_1, \mathbf{c}_2$ are not scalar multiples of each other, hence linearly independent.

$$
\boxed{\text{Basis for } \operatorname{im}(T) = \left\{ \begin{pmatrix} 1 \\ 2 \\ 0 \end{pmatrix}, \begin{pmatrix} 2 \\ 4 \\ 1 \end{pmatrix} \right\}, \quad \dim(\operatorname{im}(T)) = 2}
$$

> **Note:** The image of a matrix transformation is spanned by its pivot columns.

### Exercise 8

Verification of Rank-Nullity Theorem

Verify the Rank-Nullity Theorem for $T: \mathbb{R}^4 \to \mathbb{R}^3$ given by matrix $\mathbf{A} = \begin{pmatrix} 1 & 0 & 2 & 1 \\ 0 & 1 & -1 & 2 \\ 1 & 1 & 1 & 3 \end{pmatrix}$.

**Intuition:** The sum of the rank (dimension of image) and nullity (dimension of kernel) must equal the domain dimension $n=4$.

**Solution & Proof:**

Row reduce $\mathbf{A}$:

$$
R_3 \leftarrow R_3 - R_1 \implies \begin{pmatrix} 1 & 0 & 2 & 1 \\ 0 & 1 & -1 & 2 \\ 0 & 1 & -1 & 2 \end{pmatrix}
$$

$$
R_3 \leftarrow R_3 - R_2 \implies \begin{pmatrix} 1 & 0 & 2 & 1 \\ 0 & 1 & -1 & 2 \\ 0 & 0 & 0 & 0 \end{pmatrix}
$$

There are 2 pivot columns, so $\operatorname{rank}(T) = 2$.

For the nullspace $\mathbf{A}\mathbf{x} = \mathbf{0}$, free variables are $x_3, x_4$:

$$
x_1 = -2x_3 - x_4, \quad x_2 = x_3 - 2x_4
$$

Nullity is 2. Verifying: $\operatorname{rank}(T) + \operatorname{nullity}(T) = 2 + 2 = 4 = \dim(\mathbb{R}^4)$.

$$
\boxed{\operatorname{rank}(T) = 2, \quad \operatorname{nullity}(T) = 2, \quad 2 + 2 = 4}
$$

> **Note:** Rank-Nullity guarantees conservation of dimension between kernel collapse and image preservation.

Matrix Representation Under Non-Standard Basis

Let $T: \mathbb{R}^2 \to \mathbb{R}^2$ be $T(x,y) = (2x+y, x-y)$. Find the matrix $[T]_B$ of $T$ with respect to the basis $B = \{ \mathbf{v}_1 = (1,1), \mathbf{v}_2 = (1,-1) \}$.

**Solution & Proof:**

T(\mathbf{v}_1) = T(1,1) = (3, 0), \quad T(\mathbf{v}_2) = T(1,-1) = (1, 2)

(3, 0) = c_1(1,1) + c_2(1,-1) \implies \begin{cases} c_1 + c_2 = 3 \\ c_1 - c_2 = 0 \end{cases} \implies c_1 = \frac{3}{2}, c_2 = \frac{3}{2}

$$
\boxed{[T]_B = \begin{pmatrix} \frac{3}{2} & \frac{3}{2} \\ \frac{3}{2} & -\frac{1}{2} \end{pmatrix}}

### Exercise 10

Composition of Linear Transformations

Let $T_1: \mathbb{R}^2 \to \mathbb{R}^2$ rotate vectors by $90^\circ$ counterclockwise, and $T_2: \mathbb{R}^2 \to \mathbb{R}^2$ reflect vectors across the $y$-axis. Find the matrix representation of $T = T_2 \circ T_1$.

**Intuition:** The matrix of a composite map is the matrix product of the individual transformation matrices in reverse application order.

**Solution & Proof:**

Standard matrix for $90^\circ$ CCW rotation:

$$
\mathbf{A}_1 = \begin{pmatrix} \cos 90^\circ & -\sin 90^\circ \\ \sin 90^\circ & \cos 90^\circ \end{pmatrix} = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}
$$

Standard matrix for reflection across $y$-axis:

$$
\mathbf{A}_2 = \begin{pmatrix} -1 & 0 \\ 0 & 1 \end{pmatrix}
$$

Matrix for composition $T = T_2 \circ T_1$:

$$
\mathbf{A} = \mathbf{A}_2 \mathbf{A}_1 = \begin{pmatrix} -1 & 0 \\ 0 & 1 \end{pmatrix} \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix} = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}
$$

$$
\boxed{\mathbf{A} = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}}
$$

> **Note:** Composing a $90^\circ$ rotation with a $y$-axis reflection yields a reflection across the line $y = x$.

### Exercise 12

Derivative Operator on Polynomials

Let $D: P_2(\mathbb{R}) \to P_2(\mathbb{R})$ be the differentiation map $D(p(x)) = p'(x)$. Find the matrix $[D]_B$ relative to the standard polynomial basis $B = \{1, x, x^2\}$.

**Intuition:** Differentiation lowers polynomial degree by 1, defining a linear operator on polynomial spaces.

**Solution & Proof:**

Apply $D$ to each basis element:

1. $D(1) = 0 = 0(1) + 0(x) + 0(x^2) \implies [D(1)]_B = (0, 0, 0)^T$

Assembling the columns:

$$
\boxed{[D]_B = \begin{pmatrix} 0 & 1 & 0 \\ 0 & 0 & 2 \\ 0 & 0 & 0 \end{pmatrix}}
$$

> **Note:** Differential operators on finite-dimensional polynomial spaces map to strictly upper-triangular nilpotent matrices.

### Exercise 13

Invertibility of Linear Maps

Determine whether $T: \mathbb{R}^3 \to \mathbb{R}^3$ given by $T(x,y,z) = (x+y, y+z, z+x)$ is invertible, and if so, find $T^{-1}(x,y,z)$.

**Intuition:** $T$ is invertible if and only if its matrix representation is non-singular ($\det \mathbf{A} \neq 0$).

**Solution & Proof:**

The standard matrix is:

$$
\mathbf{A} = \begin{pmatrix} 1 & 1 & 0 \\ 0 & 1 & 1 \\ 1 & 0 & 1 \end{pmatrix}
$$

Compute determinant: $\det(\mathbf{A}) = 1(1 - 0) - 1(0 - 1) + 0 = 2 \neq 0$. Thus $T$ is invertible.

To find $T^{-1}$, solve $\mathbf{A}(u,v,w)^T = (x,y,z)^T$:

$$
\begin{cases} u + v = x \\ v + w = y \\ w + u = z \end{cases}
$$

Adding all three equations: $2(u+v+w) = x+y+z \implies u+v+w = \frac{1}{2}(x+y+z)$.

Subtracting individual equations yields:

$$
u = \frac{1}{2}(x - y + z), \quad v = \frac{1}{2}(x + y - z), \quad w = \frac{1}{2}(-x + y + z)
$$

$$
\boxed{T^{-1}(x,y,z) = \frac{1}{2}(x - y + z, \; x + y - z, \; -x + y + z)}
$$

> **Note:** A linear transformation is invertible if and only if its kernel is trivial ($\ker T = \{\mathbf{0}\}$).

### Exercise 14

Projection onto a Line

Find the standard matrix for the orthogonal projection map $P: \mathbb{R}^2 \to \mathbb{R}^2$ onto the line $y = 2x$.

**Intuition:** The projection matrix onto a subspace spanned by unit vector $\mathbf{u}$ is $\mathbf{P} = \mathbf{u}\mathbf{u}^T$.

**Solution & Proof:**

A direction vector for the line $y = 2x$ is $\mathbf{d} = (1, 2)^T$.

Using the projection matrix formula $\mathbf{P} = \frac{\mathbf{d}\mathbf{d}^T}{\mathbf{d}^T\mathbf{d}}$ with $\mathbf{d}^T\mathbf{d} = 1^2 + 2^2 = 5$:

$$
\mathbf{P} = \frac{1}{5} \begin{pmatrix} 1 \\ 2 \end{pmatrix} \begin{pmatrix} 1 & 2 \end{pmatrix} = \frac{1}{5} \begin{pmatrix} 1 & 2 \\ 2 & 4 \end{pmatrix}
$$

$$
\boxed{\mathbf{P} = \begin{pmatrix} \frac{1}{5} & \frac{2}{5} \\ \frac{2}{5} & \frac{4}{5} \end{pmatrix}}
$$

> **Note:** Orthogonal projection matrices are symmetric and idempotent ($\mathbf{P}^2 = \mathbf{P} = \mathbf{P}^T$).

\boxed{\dim(\ker \mathbf{W}) = 1}
$$

> **Note:** Rank-deficient weight matrices destroy information by collapsing 1 or more feature dimensions to zero.

\mathbf{E}_{11} = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}, \quad \mathbf{E}_{12} = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}, \quad \mathbf{E}_{21} = \begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix}, \quad \mathbf{E}_{22} = \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix}

### Exercise 18

Trace as a Linear Functional

Let $f: \mathbb{R}^{n \times n} \to \mathbb{R}$ be defined by $f(\mathbf{A}) = \operatorname{tr}(\mathbf{A})$. Prove $f$ is a linear functional and find $\dim(\ker(f))$.

**Intuition:** Trace sums diagonal entries, which is a linear operation mapping $n^2$-dimensional matrix space to 1-dimensional real numbers.

**Solution & Proof:**

1. **Linearity:** $\operatorname{tr}(c\mathbf{A} + \mathbf{B}) = \sum_{i=1}^n (c a_{ii} + b_{ii}) = c \sum_{i=1}^n a_{ii} + \sum_{i=1}^n b_{ii} = c \operatorname{tr}(\mathbf{A}) + \operatorname{tr}(\mathbf{B})$.

2. Since $f(\mathbf{I}_n) = n \neq 0$, the range of $f$ is all of $\mathbb{R}$, so $\operatorname{rank}(f) = 1$.

3. By Rank-Nullity on $\mathbb{R}^{n \times n}$ (which has dimension $n^2$):

$$
\dim(\mathbb{R}^{n \times n}) = \operatorname{rank}(f) + \dim(\ker(f)) \implies n^2 = 1 + \dim(\ker(f))
$$

$$
\boxed{\dim(\ker(f)) = n^2 - 1}
$$

> **Note:** Traceless matrices form a maximal proper subspace (hyperplane) of dimension $n^2 - 1$ in $\mathbb{R}^{n \times n}$.

### Exercise 20

Similarity Transformation of Operators

Let $\mathbf{A} = \begin{pmatrix} 1 & 2 \\ 0 & 3 \end{pmatrix}$ represent an operator in standard basis, and let change-of-basis matrix be $\mathbf{P} = \begin{pmatrix} 1 & 1 \\ 1 & 2 \end{pmatrix}$. Compute the matrix $\mathbf{B} = \mathbf{P}^{-1} \mathbf{A} \mathbf{P}$ representing the operator in the new basis.

**Intuition:** Matrix representations of the same operator under different bases are similar ($\mathbf{B} = \mathbf{P}^{-1}\mathbf{A}\mathbf{P}$).

**Solution & Proof:**

$$
\mathbf{P}^{-1} = \begin{pmatrix} 2 & -1 \\ -1 & 1 \end{pmatrix}
$$

2. Compute product $\mathbf{A}\mathbf{P}$:

$$
\mathbf{A}\mathbf{P} = \begin{pmatrix} 1 & 2 \\ 0 & 3 \end{pmatrix} \begin{pmatrix} 1 & 1 \\ 1 & 2 \end{pmatrix} = \begin{pmatrix} 3 & 5 \\ 3 & 6 \end{pmatrix}

$$
\mathbf{B} = \begin{pmatrix} 2 & -1 \\ -1 & 1 \end{pmatrix} \begin{pmatrix} 3 & 5 \\ 3 & 6 \end{pmatrix} = \begin{pmatrix} 6-3 & 10-6 \\ -3+3 & -5+6 \end{pmatrix} = \begin{pmatrix} 3 & 4 \\ 0 & 1 \end{pmatrix}
$$

$$
\boxed{\mathbf{B} = \begin{pmatrix} 3 & 4 \\ 0 & 1 \end{pmatrix}}
$$

> **Note:** Similar matrices share trace ($\operatorname{tr}(\mathbf{A}) = \operatorname{tr}(\mathbf{B}) = 4$) and determinant ($\det(\mathbf{A}) = \det(\mathbf{B}) = 3$).

### Exercise 21

Nilpotent Linear Map

Let $T: V \to V$ be a linear map such that $T^3 = 0$ but $T^2 \neq 0$. Prove that there exists a vector $\mathbf{v} \in V$ such that $\{\mathbf{v}, T\mathbf{v}, T^2\mathbf{v}\}$ is linearly independent.

**Intuition:** If $T^2\mathbf{v} \neq \mathbf{0}$, applying $T$ repeatedly peels off lower-order terms in linear dependence equations.

**Solution & Proof:**

Since $T^2 \neq 0$, choose $\mathbf{v} \in V$ such that $T^2\mathbf{v} \neq \mathbf{0}$.

Set up linear independence relation:

$$
c_1 \mathbf{v} + c_2 T\mathbf{v} + c_3 T^2\mathbf{v} = \mathbf{0} \quad \text{(*)}
$$

Apply $T^2$ to (*):

$$
c_1 T^2\mathbf{v} + c_2 T^3\mathbf{v} + c_3 T^4\mathbf{v} = \mathbf{0} \implies c_1 T^2\mathbf{v} + \mathbf{0} + \mathbf{0} = \mathbf{0} \implies c_1 = 0
$$

Equation (*) simplifies to $c_2 T\mathbf{v} + c_3 T^2\mathbf{v} = \mathbf{0}$. Apply $T$:

$$
c_2 T^2\mathbf{v} + c_3 T^3\mathbf{v} = \mathbf{0} \implies c_2 T^2\mathbf{v} = \mathbf{0} \implies c_2 = 0
$$

Finally $c_3 T^2\mathbf{v} = \mathbf{0} \implies c_3 = 0$.

$$
\boxed{\{\mathbf{v}, T\mathbf{v}, T^2\mathbf{v}\} \text{ is linearly independent.}}
$$

> **Note:** Nilpotent operators generate cyclic subspaces with linearly independent Krylov sequences.

### Exercise 22

Reflection Across a General Hyperplane

\|\mathbf{A}\mathbf{x}\|_2^2 = (\mathbf{A}\mathbf{x})^T(\mathbf{A}\mathbf{x}) = \mathbf{x}^T (\mathbf{A}^T \mathbf{A}) \mathbf{x}
$$

Since $\mathbf{A}^T \mathbf{A}$ is symmetric positive semi-definite:

$$
\max_{\|\mathbf{x}\|_2=1} \mathbf{x}^T (\mathbf{A}^T \mathbf{A}) \mathbf{x} = \lambda_{\max}(\mathbf{A}^T \mathbf{A})
$$

Taking the square root:

$$
\|T\|_2 = \sqrt{\lambda_{\max}(\mathbf{A}^T \mathbf{A})} = \sigma_{\max}(\mathbf{A})
$$

$$
\boxed{\|T\|_2 = \sigma_{\max}(\mathbf{A}) = \sqrt{\lambda_{\max}(\mathbf{A}^T \mathbf{A})}}
$$

> **Note:** The spectral norm of a linear map measures the maximum stretch factor along its principal right singular vector.

### Exercise 40

Jacobian Matrix of Neural Network Layer Transformation

Consider a neural network hidden layer transformation $f(\mathbf{x}) = \sigma(\mathbf{W}\mathbf{x} + \mathbf{b})$, where $\mathbf{W} \in \mathbb{R}^{m \times n}$, $\mathbf{b} \in \mathbb{R}^m$, and $\sigma$ is an element-wise smooth activation function. Derive the Jacobian matrix $\mathbf{J}_f(\mathbf{x})$ as a linear transformation of tangent vectors.

**Intuition:** The local linearization of a non-linear network layer is given by scaling weight matrix rows by activation derivatives.

**Solution & Proof:**

Let $\mathbf{z} = \mathbf{W}\mathbf{x} + \mathbf{b}$, so component $i$ is $z_i = \sum_{j=1}^n W_{ij} x_j + b_i$.

The output component is $f_i(\mathbf{x}) = \sigma(z_i)$.

By the chain rule, entry $(i, j)$ of the Jacobian matrix is:

$$
(\mathbf{J}_f(\mathbf{x}))_{ij} = \frac{\partial f_i}{\partial x_j} = \sigma'(z_i) \frac{\partial z_i}{\partial x_j} = \sigma'(z_i) W_{ij}
$$

In matrix notation, this corresponds to multiplying $\mathbf{W}$ on the left by the diagonal matrix of activation derivatives:

$$
\boxed{\mathbf{J}_f(\mathbf{x}) = \operatorname{diag}(\sigma'(\mathbf{W}\mathbf{x} + \mathbf{b})) \mathbf{W}}
$$

> **Note:** The local linear transformation of a neural layer modulates global weights $\mathbf{W}$ via pointwise activation derivatives.
