## Level 0 — Concept Check

### Problem L0.1: Which sets are convex?

Decide whether each set is convex, with a one-line justification: (a) the Euclidean ball $\{\mathbf{x} : \lVert \mathbf{x} \rVert_2 \le 1\}$; (b) the unit sphere $\{\mathbf{x} : \lVert \mathbf{x} \rVert_2 = 1\}$; (c) a halfspace $\{\mathbf{x} : \mathbf{a}^\top\mathbf{x} \le b\}$; (d) the set of positive semidefinite matrices; (e) the union of two disjoint balls.

**Solution.**

**(a) Convex.** For $\lVert \mathbf{x}\rVert_2, \lVert \mathbf{y}\rVert_2 \le 1$ and $\theta\in[0,1]$, the triangle inequality and homogeneity give $\lVert \theta\mathbf{x}+(1-\theta)\mathbf{y}\rVert_2 \le \theta\lVert \mathbf{x}\rVert_2+(1-\theta)\lVert \mathbf{y}\rVert_2 \le 1$.

**(b) Not convex.** Take $\mathbf{x}$ and $\mathbf{y}=-\mathbf{x}$ on the sphere; their midpoint is $\mathbf{0}$, whose norm is $0 \neq 1$.

**(c) Convex.** $\mathbf{a}^\top\!\left(\theta\mathbf{x}+(1-\theta)\mathbf{y}\right) = \theta\,\mathbf{a}^\top\mathbf{x}+(1-\theta)\,\mathbf{a}^\top\mathbf{y} \le \theta b+(1-\theta)b = b$. (Linear functions preserve convex combinations exactly.)

**(d) Convex.** If $A,B \succeq 0$ then for any $\mathbf{u}$, $\mathbf{u}^\top\!\left(\theta A+(1-\theta)B\right)\mathbf{u} = \theta\,\mathbf{u}^\top A\mathbf{u}+(1-\theta)\,\mathbf{u}^\top B\mathbf{u} \ge 0$. The PSD cone is convex (and a cone).

**(e) Not convex.** Any segment joining a point of one ball to a point of the other must leave the union, since the balls are disjoint.

$$
\boxed{\text{(a) yes, (b) no, (c) yes, (d) yes, (e) no}}
$$

*Key takeaway:* Convexity is preserved by intersections, affine maps and linear inequalities — but never by unions or by taking boundaries. Feasible regions in optimization are built from (c) and (d) for exactly this reason.

### Problem L0.2: Classify the canonical saddle

For $f(x,y) = x^2 - y^2$, find all critical points, compute the Hessian, and classify.

**Solution.**

Gradient: $\nabla f = (2x,\, -2y)^\top$, which vanishes only at the origin, so $(0,0)$ is the unique critical point.

Hessian (constant everywhere):

$$
H = \begin{bmatrix} 2 & 0 \\ 0 & -2 \end{bmatrix}, \qquad \lambda_1 = -2, \ \lambda_2 = 2 .
$$

Mixed signs, so $H$ is indefinite and the origin is a **saddle**. Concretely, along the $x$-axis $f(t,0)=t^2 \gt 0$ and along the $y$-axis $f(0,t)=-t^2 \lt 0$: every neighborhood of the origin contains points above and below $f(0,0)=0$.

$$
\boxed{(0,0) \text{ is a saddle: } \lambda\left(H\right) = \{+2,-2\}, \text{ index } \alpha = 1}
$$

*Key takeaway:* Gradient descent started exactly on the $x$-axis converges to the saddle forever; started anywhere else it escapes, because the $y$-mode carries the expanding factor $1+2\eta \gt 1$. The trapping set is a line — measure zero in the plane.

### Problem L0.3: The value of convexity in one sentence

State precisely what convexity buys you at a stationary point, and give a non-convex function where the same stationarity test fails.

**Solution.**

**What it buys.** If $f$ is convex and differentiable, then $\nabla f(\mathbf{x}_c)=0$ implies $\mathbf{x}_c$ is a *global* minimizer. The proof is one line from the first-order characterization: for every $\mathbf{y}$,

$$
f(\mathbf{y}) \ge f(\mathbf{x}_c) + \nabla f(\mathbf{x}_c)^\top(\mathbf{y}-\mathbf{x}_c) = f(\mathbf{x}_c).
$$

So a *local, checkable* condition yields a *global* conclusion — no search over the domain is required.

**Where it fails without convexity.** Take $f(x) = x^3$. Then $f'(0)=0$, yet $0$ is neither a local minimum nor a local maximum ($f$ is strictly increasing through it). Or $f(x)=x^4-x^2$: $x=0$ is a stationary point and a local *maximum*, while the global minima are at $x=\pm 1/\sqrt{2}$.

$$
\boxed{\text{convex} + \nabla f(\mathbf{x}_c)=0 \implies \mathbf{x}_c \in \arg\min f \ \text{(globally)}}
$$

*Key takeaway:* This is the whole reason convex modeling is worth the constraint. Every guarantee of the form "the algorithm found the answer" in classical ML traces back to this implication.

### Problem L0.4: A degenerate Hessian tells you nothing

At a critical point, the Hessian is $H = \operatorname{diag}(2,0)$. What can and cannot be concluded? Support the answer with explicit examples.

**Solution.**

**What can be concluded.** $H \succeq 0$, which is *necessary* for a local minimum, so the point is not a local maximum and is not a strict saddle. That is all.

**What cannot.** The sufficient condition requires $H \succ 0$, and here $\lambda_2 = 0$. Along the second coordinate the quadratic term vanishes entirely, so the $o(\lVert \mathbf{h}\rVert_2^2)$ remainder decides. All three outcomes are realized with this same Hessian at the origin:

| Function | Behavior along $y$ | Verdict at origin |
|---|---|---|
| $f_1(x,y)=x^2+y^4$ | $+y^4 \gt 0$ | strict local (and global) minimum |
| $f_2(x,y)=x^2-y^4$ | $-y^4 \lt 0$ | saddle |
| $f_3(x,y)=x^2$ | identically $0$ | non-strict minimum: the whole $y$-axis minimizes |

$$
\boxed{H \succeq 0 \text{ is necessary, } H \succ 0 \text{ is sufficient; a zero eigenvalue leaves the test inconclusive}}
$$

*Key takeaway:* Degeneracy is not an exotic edge case in machine learning — it is the normal state of an overparameterized model, whose Hessian at a solution has a large null space (row $f_3$ generalized to many dimensions).

## Level 1 — Foundation

### Problem L1.1: Intersections preserve convexity, unions do not

Prove that the intersection of an arbitrary family of convex sets is convex, and give a counterexample for unions. What does this imply for feasible regions defined by many constraints?

**Solution.**

**Intersection.** Let $\{C_i\}_{i\in I}$ be convex and $C = \bigcap_{i\in I}C_i$. Take $\mathbf{x},\mathbf{y}\in C$ and $\theta\in[0,1]$. For each $i$, both $\mathbf{x},\mathbf{y}\in C_i$, and $C_i$ is convex, so $\theta\mathbf{x}+(1-\theta)\mathbf{y}\in C_i$. Since this holds for *every* $i$, the point lies in the intersection:

$$
\theta\mathbf{x}+(1-\theta)\mathbf{y} \in \bigcap_{i\in I}C_i = C .
$$

Note the index set $I$ may be infinite — the proof never counts.

**Union fails.** In $\mathbb{R}$, take $C_1=[0,1]$ and $C_2=[2,3]$, both convex. Their union contains $1$ and $2$ but not their midpoint $1.5$.

**Consequence.** A feasible set defined by any number of linear inequalities $\mathbf{a}_i^\top\mathbf{x}\le b_i$ (halfspaces, each convex) and equalities (hyperplanes, convex) is an intersection, hence convex — this is why *polyhedra*, and by extension linear and quadratic programs, are convex problems no matter how many constraints there are. Conversely, "or"-type constraints (disjunctions) break convexity immediately, which is exactly why integer/combinatorial constraints are hard.

$$
\boxed{\bigcap_{i\in I}C_i \text{ convex}; \quad C_1\cup C_2 \text{ generally not}}
$$

*Key takeaway:* Convexity is an "AND" property. Modeling with conjunctions of convex constraints stays tractable; introducing a disjunction leaves convex optimization for combinatorial optimization.

### Problem L1.2: Norms are convex; the squared Euclidean norm is strongly convex

Prove that every norm $\lVert \cdot \rVert$ is a convex function, and that $f(\mathbf{x}) = \lVert \mathbf{x}\rVert_2^2$ is $2$-strongly convex.

**Solution.**

**Norms are convex.** Using absolute homogeneity and the triangle inequality, for $\theta\in[0,1]$:

$$
\lVert \theta\mathbf{x}+(1-\theta)\mathbf{y} \rVert \le \lVert \theta\mathbf{x}\rVert + \lVert (1-\theta)\mathbf{y}\rVert = \theta\lVert \mathbf{x}\rVert + (1-\theta)\lVert \mathbf{y}\rVert .
$$

That is exactly Definition 2.2. Note the proof uses no differentiability — essential, since $\lVert \cdot \rVert_1$ is not differentiable at $0$, yet lasso remains a convex problem.

**Squared Euclidean norm.** $f(\mathbf{x}) = \mathbf{x}^\top\mathbf{x}$ gives $\nabla f = 2\mathbf{x}$ and $\nabla^2 f = 2I$. Since $2I \succeq 2I$, Theorem 2.4 gives $\mu$-strong convexity with $\mu = 2$. Equivalently, $f - \frac{2}{2}\lVert \cdot\rVert_2^2 = 0$ is convex.

**Why $\lVert \cdot\rVert_2$ itself is *not* strongly convex:** it is linear along any ray from the origin ($\lVert t\mathbf{u}\rVert_2 = t$ for $t\ge 0$, $\lVert \mathbf{u}\rVert_2=1$), and a function that is affine on a segment cannot be strongly convex there.

$$
\boxed{\text{every norm is convex}; \quad \nabla^2\lVert \mathbf{x}\rVert_2^2 = 2I \succ 0 \implies \mu = 2}
$$

*Key takeaway:* This pair explains the two standard regularizers. $\lambda\lVert \mathbf{w}\rVert_1$ keeps a problem convex (and adds sparsity) without adding curvature; $\lambda\lVert \mathbf{w}\rVert_2^2$ adds $2\lambda$ to every Hessian eigenvalue, which is where its conditioning benefit in Topic 03 comes from.

### Problem L1.3: Find and classify all critical points

Let $f(x,y) = x^3 - 3xy + y^3$. Find every critical point and classify it using the Hessian.

**Solution.**

**Step 1 (critical points).** $\nabla f = \left(3x^2-3y,\ -3x+3y^2\right)^\top = \mathbf{0}$ gives the system $y = x^2$ and $x = y^2$. Substituting, $x = x^4$, so $x(x^3-1)=0$, giving $x = 0$ or $x = 1$. Hence

$$
\mathbf{x}_c \in \{(0,0),\ (1,1)\}.
$$

**Step 2 (Hessian).**

$$
H(x,y) = \begin{bmatrix} 6x & -3 \\ -3 & 6y \end{bmatrix}, \qquad \det H = 36xy - 9 .
$$

**Step 3 (classify at $(0,0)$).** $H = \begin{bmatrix}0 & -3\\ -3 & 0\end{bmatrix}$ with $\det H = -9 \lt 0$. A negative determinant in 2D means eigenvalues of opposite sign (here $\pm 3$), so the origin is a **saddle**.

**Step 4 (classify at $(1,1)$).** $H = \begin{bmatrix}6 & -3\\ -3 & 6\end{bmatrix}$ with $\det H = 27 \gt 0$ and trace $12 \gt 0$, so both eigenvalues are positive (explicitly $6\mp 3 = 3$ and $9$). Therefore $(1,1)$ is a **strict local minimum**, with $f(1,1) = 1-3+1 = -1$.

**Step 5 (not global).** $f(t,0) = t^3 \to -\infty$ as $t\to-\infty$, so $f$ is unbounded below and $(1,1)$ is only a *local* minimum — confirming that $f$ is not convex (as the saddle at the origin already showed).

$$
\boxed{(0,0)\ \text{saddle } (\lambda = \pm 3); \quad (1,1)\ \text{strict local min } (\lambda = 3, 9),\ f = -1;\ \inf f = -\infty}
$$

*Key takeaway:* The 2D shortcut $\det H \lt 0 \Rightarrow$ saddle, $\det H \gt 0$ with $\operatorname{tr}H \gt 0 \Rightarrow$ minimum, is just the sign rules $\lambda_1\lambda_2 = \det H$ and $\lambda_1+\lambda_2 = \operatorname{tr}H$ in disguise; in $d$ dimensions there is no substitute for the full spectrum.

### Problem L1.4: Log-sum-exp is convex

Prove that $f(\mathbf{x}) = \log\sum_{i=1}^{d}e^{x_i}$ is convex on $\mathbb{R}^d$ by computing its Hessian and showing it is positive semidefinite.

**Solution.**

**Step 1 (gradient).** Let $S = \sum_j e^{x_j}$ and $p_i = e^{x_i}/S$, so $\mathbf{p}$ is the softmax of $\mathbf{x}$ with $p_i \gt 0$ and $\sum_i p_i = 1$. Then

$$
\frac{\partial f}{\partial x_i} = \frac{e^{x_i}}{S} = p_i \qquad \Longrightarrow \qquad \nabla f(\mathbf{x}) = \mathbf{p}.
$$

**Step 2 (Hessian).** Differentiating $p_i = e^{x_i}/S$ with respect to $x_j$ and using $\partial S/\partial x_j = e^{x_j}$:

$$
\frac{\partial p_i}{\partial x_j} = \frac{\delta_{ij}e^{x_i}S - e^{x_i}e^{x_j}}{S^2} = \delta_{ij}p_i - p_ip_j \qquad \Longrightarrow \qquad \nabla^2 f = \operatorname{diag}(\mathbf{p}) - \mathbf{p}\mathbf{p}^\top .
$$

**Step 3 (positive semidefiniteness).** For any $\mathbf{u}\in\mathbb{R}^d$,

$$
\mathbf{u}^\top\nabla^2 f\,\mathbf{u} = \sum_i p_iu_i^2 - \left(\sum_i p_iu_i\right)^2 .
$$

Interpreting $\mathbf{p}$ as a probability distribution over indices and $U$ as the random variable taking value $u_i$ with probability $p_i$, this is exactly $\mathbb{E}[U^2]-\left(\mathbb{E}[U]\right)^2 = \operatorname{Var}(U) \ge 0$. (Equivalently, apply the Cauchy–Schwarz inequality to the vectors with components $\sqrt{p_i}$ and $\sqrt{p_i}u_i$.) Hence $\nabla^2 f \succeq 0$ everywhere and $f$ is convex. $\blacksquare$

Note the null space: variance is zero iff $U$ is constant, i.e. $\mathbf{u} = c\mathbf{1}$. So $\nabla^2 f$ is singular along $\mathbf{1}$ — reflecting the shift invariance $f(\mathbf{x}+c\mathbf{1}) = f(\mathbf{x})+c$. Log-sum-exp is convex but never *strictly* convex.

$$
\boxed{\nabla^2 f = \operatorname{diag}(\mathbf{p})-\mathbf{p}\mathbf{p}^\top \succeq 0, \qquad \mathbf{u}^\top\nabla^2 f\mathbf{u} = \operatorname{Var}_{\mathbf{p}}(U) \ge 0}
$$

*Key takeaway:* This is the Hessian of the softmax cross-entropy's normalizer, so multinomial logistic regression is convex in its logits — and the singular direction $\mathbf{1}$ is the familiar fact that logits are only identified up to an additive constant.

### Problem L1.5: Composition rules, and where they stop

Prove: if $h:\mathbb{R}^d\to\mathbb{R}$ is convex and $g:\mathbb{R}\to\mathbb{R}$ is convex and non-decreasing, then $g\circ h$ is convex. Then show by counterexample that dropping "non-decreasing" breaks the result.

**Solution.**

**Proof.** Take $\mathbf{x},\mathbf{y}$ and $\theta\in[0,1]$, and write $\mathbf{z} = \theta\mathbf{x}+(1-\theta)\mathbf{y}$.

*Step 1 (inner convexity).* $h(\mathbf{z}) \le \theta h(\mathbf{x})+(1-\theta)h(\mathbf{y})$.

*Step 2 (monotonicity transports the inequality through $g$).* Because $g$ is non-decreasing, applying it to both sides preserves the direction:

$$
g\!\left(h(\mathbf{z})\right) \le g\!\left(\theta h(\mathbf{x})+(1-\theta)h(\mathbf{y})\right).
$$

*Step 3 (outer convexity).* Applying convexity of $g$ to the right side,

$$
g\!\left(\theta h(\mathbf{x})+(1-\theta)h(\mathbf{y})\right) \le \theta\, g\!\left(h(\mathbf{x})\right)+(1-\theta)\,g\!\left(h(\mathbf{y})\right).
$$

Chaining Steps 2 and 3 gives exactly the convexity of $g\circ h$. $\blacksquare$

**Counterexample without monotonicity.** Let $g(u) = e^{-u}$, which is convex but *decreasing*, and $h(x)=x^2$, convex. Then $\left(g\circ h\right)(x) = e^{-x^2}$, whose second derivative is $\left(4x^2-2\right)e^{-x^2}$, negative for $\lvert x\rvert \lt 1/\sqrt{2}$ — a concave bump near the origin. Not convex.

$$
\boxed{g \text{ convex non-decreasing},\ h \text{ convex} \implies g\circ h \text{ convex}; \ e^{-x^2} \text{ shows monotonicity is required}}
$$

*Key takeaway:* Monotonicity is what lets an outer function inherit an inner inequality. The mirrored rule — $g$ convex non-increasing with $h$ *concave* — also gives a convex composite, and together they generate most of the convex atoms used by disciplined-convex-programming libraries.

### Problem L1.6: Jensen's inequality and the general chord condition

Prove that convexity of $f$ implies $f\!\left(\sum_{i=1}^{m}\theta_i\mathbf{x}_i\right) \le \sum_{i=1}^{m}\theta_i f(\mathbf{x}_i)$ for any $\theta_i \ge 0$ with $\sum_i\theta_i = 1$. Then deduce $f(\mathbb{E}[X]) \le \mathbb{E}[f(X)]$.

**Solution.**

**Induction on $m$.** The base case $m=2$ is Definition 2.2. Assume the claim for $m$ points and take $m+1$ points with weights $\theta_1,\dots,\theta_{m+1}$ summing to $1$. If $\theta_{m+1}=1$ the statement is trivial, so assume $\theta_{m+1} \lt 1$ and set $s = 1-\theta_{m+1} = \sum_{i=1}^{m}\theta_i \gt 0$. Write

$$
\sum_{i=1}^{m+1}\theta_i\mathbf{x}_i = s\underbrace{\left(\sum_{i=1}^{m}\frac{\theta_i}{s}\mathbf{x}_i\right)}_{=:\ \mathbf{z}} + \theta_{m+1}\mathbf{x}_{m+1},
$$

where the weights $\theta_i/s$ are nonnegative and sum to $1$, so $\mathbf{z}$ is a legitimate convex combination of the first $m$ points. Apply the two-point definition, then the induction hypothesis to $f(\mathbf{z})$:

$$
f\!\left(\sum_{i=1}^{m+1}\theta_i\mathbf{x}_i\right) \le s\,f(\mathbf{z}) + \theta_{m+1}f(\mathbf{x}_{m+1}) \le s\sum_{i=1}^{m}\frac{\theta_i}{s}f(\mathbf{x}_i)+\theta_{m+1}f(\mathbf{x}_{m+1}) = \sum_{i=1}^{m+1}\theta_i f(\mathbf{x}_i).
$$

**Probabilistic form.** A discrete random variable $X$ taking value $\mathbf{x}_i$ with probability $\theta_i$ turns the statement into $f(\mathbb{E}[X]) \le \mathbb{E}[f(X)]$; the general case follows by approximating $X$ with discrete variables and passing to the limit (using the tangent-plane bound of Proof 3.1 at $\mathbf{x}=\mathbb{E}[X]$ and taking expectations gives a one-line proof directly: $\mathbb{E}[f(X)] \ge f(\mathbb{E}X)+\nabla f(\mathbb{E}X)^\top\mathbb{E}[X-\mathbb{E}X] = f(\mathbb{E}X)$).

$$
\boxed{f\!\left(\textstyle\sum_i\theta_i\mathbf{x}_i\right) \le \sum_i\theta_i f(\mathbf{x}_i); \qquad f(\mathbb{E}[X]) \le \mathbb{E}[f(X)]}
$$

*Key takeaway:* Jensen is convexity applied to a distribution instead of two points. It is the engine behind the ELBO in variational inference, the non-negativity of KL divergence, the AM–GM inequality, and the EM algorithm's monotonicity proof.

## Level 2 — Applications in AI/ML

### Problem L2.1: Logistic regression is convex — and only strongly convex with weight decay

Show that $f(\mathbf{w}) = \frac{1}{n}\sum_i \log\!\left(1+e^{-y_i\mathbf{x}_i^\top\mathbf{w}}\right)$ is convex in $\mathbf{w}$ but generally not strongly convex, and that adding $\lambda\lVert \mathbf{w}\rVert_2^2$ makes it $2\lambda$-strongly convex.

**Solution.**

**Step 1 (Hessian).** With $s_i = \sigma(-y_i\mathbf{x}_i^\top\mathbf{w}) \in (0,1)$,

$$
\nabla^2 f(\mathbf{w}) = \frac{1}{n}\sum_{i=1}^{n}s_i(1-s_i)\,\mathbf{x}_i\mathbf{x}_i^\top = \frac{1}{n}X^\top S X, \qquad S = \operatorname{diag}\!\left(s_i(1-s_i)\right) \succ 0 .
$$

**Step 2 (convexity).** For any $\mathbf{u}$,

$$
\mathbf{u}^\top\nabla^2 f\,\mathbf{u} = \frac{1}{n}\sum_i s_i(1-s_i)\left(\mathbf{x}_i^\top\mathbf{u}\right)^2 \ge 0,
$$

a sum of nonnegative terms. Hence $\nabla^2 f \succeq 0$ everywhere and $f$ is convex (Theorem 2.4).

**Step 3 (no strong convexity, two independent reasons).**
- *Rank.* If $d \gt n$ then $X^\top S X$ has rank at most $n \lt d$, so it is singular: $\lambda_{\min}=0$ along any $\mathbf{u}\in\ker X$.
- *Saturation.* Even with full rank, along a separating direction $\mathbf{u}$ with $y_i\mathbf{x}_i^\top\mathbf{u} \gt 0$ for all $i$, taking $\mathbf{w}=t\mathbf{u}$ with $t\to\infty$ drives every $s_i \to 0$, so $s_i(1-s_i)\to 0$ and $\nabla^2 f \to 0$. The loss flattens out; $\inf_{\mathbf{w}}\lambda_{\min}(\nabla^2 f) = 0$, and the infimum $f^\star=0$ is not attained.

**Step 4 (ridge restores it).** For $f_\lambda = f + \lambda\lVert \mathbf{w}\rVert_2^2$, $\nabla^2 f_\lambda = \nabla^2 f + 2\lambda I \succeq 2\lambda I$, so $f_\lambda$ is $2\lambda$-strongly convex, has a *unique* finite minimizer, and by Topic 03 converges geometrically with $\kappa \le \frac{L+2\lambda}{2\lambda}$.

$$
\boxed{\nabla^2 f = \tfrac{1}{n}X^\top SX \succeq 0 \ (\text{convex}); \quad \mu = 0; \quad \nabla^2 f_\lambda \succeq 2\lambda I \ (\mu = 2\lambda)}
$$

*Key takeaway:* On separable data, unregularized logistic regression has *no minimizer* — the weights diverge while the loss creeps to zero. Weight decay is what makes the problem well-posed, and it simultaneously converts an $O(1/k)$ rate into a geometric one.

### Problem L2.2: Why deep networks cannot be convex in their weights

Consider a network with one hidden layer of two identical-architecture units. Use the permutation symmetry to prove the loss is not a convex function of the weights.

**Solution.**

**Step 1 (the symmetry).** Let the network compute $F(\mathbf{x};\boldsymbol{\theta}) = a_1\phi(\mathbf{w}_1^\top\mathbf{x}) + a_2\phi(\mathbf{w}_2^\top\mathbf{x})$ with $\boldsymbol{\theta} = (\mathbf{w}_1,\mathbf{w}_2,a_1,a_2)$. Swapping the two hidden units,

$$
\boldsymbol{\theta}' = (\mathbf{w}_2,\mathbf{w}_1,a_2,a_1),
$$

leaves the function identical: $F(\cdot;\boldsymbol{\theta}) = F(\cdot;\boldsymbol{\theta}')$ pointwise. Hence any loss satisfies $\mathcal{L}(\boldsymbol{\theta}) = \mathcal{L}(\boldsymbol{\theta}')$ for every dataset.

**Step 2 (two global minimizers).** If $\boldsymbol{\theta}$ is a global minimizer with $\mathbf{w}_1\neq\mathbf{w}_2$ (or $a_1\neq a_2$), then so is the distinct point $\boldsymbol{\theta}'$.

**Step 3 (the midpoint test).** If $\mathcal{L}$ were convex, then by Proof 3.3(b) the *entire segment* between two global minimizers would consist of global minimizers. In particular the midpoint

$$
\bar{\boldsymbol{\theta}} = \frac{\boldsymbol{\theta}+\boldsymbol{\theta}'}{2} = \left(\frac{\mathbf{w}_1+\mathbf{w}_2}{2},\ \frac{\mathbf{w}_1+\mathbf{w}_2}{2},\ \frac{a_1+a_2}{2},\ \frac{a_1+a_2}{2}\right)
$$

would have to achieve the minimum. But $\bar{\boldsymbol{\theta}}$ has two *identical* hidden units, so it computes $\left(a_1+a_2\right)\phi\!\left(\bar{\mathbf{w}}^\top\mathbf{x}\right)$ — a network with an effective width of **one**. Whenever the task genuinely requires two distinct units (e.g. XOR with ReLU, or fitting any function outside the span of a single unit), this collapsed network has strictly larger loss, contradicting convexity.

$$
\boxed{\mathcal{L}(\boldsymbol{\theta})=\mathcal{L}(\boldsymbol{\theta}') = \min \ \text{but} \ \mathcal{L}\!\left(\tfrac{\boldsymbol{\theta}+\boldsymbol{\theta}'}{2}\right) \gt \min \implies \mathcal{L} \text{ not convex}}
$$

*Key takeaway:* Non-convexity in deep learning is *structural*, not a matter of a badly chosen loss: any architecture with interchangeable units has at least $H!$ equivalent global minimizers (times $2^H$ for ReLU sign symmetries), and a convex function cannot have separated minimizers. The useful question is never "is it convex" but "is the reachable region benign".

### Problem L2.3: Counting minima in high dimensions

At a critical point of a $d$-parameter model, model the Hessian eigenvalues as i.i.d. with $\Pr[\lambda_i \gt 0] = p$. Compute the probability that the point is a local minimum for $p = 0.5$ and $d\in\{10, 100, 10^6\}$, and the distribution of the index. Interpret for deep learning.

**Solution.**

**Probability of a minimum.** All $d$ eigenvalues must be positive, and by independence

$$
\Pr[\text{local min}] = p^{\,d} = 2^{-d} \quad (p = 0.5).
$$

| $d$ | $2^{-d}$ |
|---|---|
| $10$ | $\approx 9.8\times 10^{-4}$ |
| $100$ | $\approx 7.9\times 10^{-31}$ |
| $10^6$ | $\approx 10^{-301030}$ |

**Index distribution.** The index $\alpha = \#\{i:\lambda_i \lt 0\}$ is Binomial$(d, 1-p)$, so

$$
\mathbb{E}[\alpha] = (1-p)d = \frac{d}{2}, \qquad \operatorname{sd}(\alpha) = \sqrt{d\,p(1-p)} = \frac{\sqrt{d}}{2}.
$$

The *fraction* of negative directions concentrates: $\alpha/d = \frac{1}{2}\pm O(d^{-1/2})$. For $d=10^6$ a typical critical point has about $500{,}000$ escape directions, with fluctuation only $\pm 500$.

**Interpretation.** A randomly encountered critical point of a large model is, with overwhelming probability, a saddle with an $\Theta(d)$-dimensional descent subspace — never a trap. The real theory (Bray & Dean; Dauphin et al., 2014) refines this by making $p$ depend on the loss value: $p \to 1$ only as the loss approaches its global minimum, so the few genuine minima are all *good* ones, and high-loss critical points are high-index saddles.

$$
\boxed{\Pr[\text{min}] = p^d = 2^{-d}; \quad \alpha \sim \mathrm{Bin}(d, 1-p), \ \tfrac{\alpha}{d}\to 1-p}
$$

*Key takeaway:* Low-dimensional intuition ("the landscape is full of bad local minima") is a dimensionality artifact. In high dimensions the obstacle is *slowness near saddles and plateaus*, which is why noise and momentum matter more than restart strategies.

### Problem L2.4: Quantifying sharpness at a minimum

Two minima have Hessians $H_A = \operatorname{diag}(100, 1, 1)$ and $H_B = \operatorname{diag}(4, 4, 4)$, with equal training loss. Compute the worst-case and average loss increase under a perturbation of size $\lVert \boldsymbol{\delta}\rVert_2 = \epsilon = 0.1$, and the $\epsilon$-sharpness of each. Which is "flatter", and by which measure?

**Solution.**

Near a minimum $\nabla f = 0$, so $f(\mathbf{x}^\star+\boldsymbol{\delta}) - f(\mathbf{x}^\star) \approx \frac{1}{2}\boldsymbol{\delta}^\top H\boldsymbol{\delta}$.

**Worst case ($\epsilon$-sharpness).** By the Rayleigh quotient, the maximum over $\lVert \boldsymbol{\delta}\rVert_2 = \epsilon$ is attained along the top eigenvector:

$$
\max_{\lVert \boldsymbol{\delta}\rVert_2 \le \epsilon}\Delta f = \frac{\epsilon^2}{2}\lambda_{\max}, \qquad
\Delta f_A = \frac{0.01}{2}(100) = 0.5, \qquad \Delta f_B = \frac{0.01}{2}(4) = 0.02 .
$$

**Average over random directions.** For $\boldsymbol{\delta}$ uniform on the sphere of radius $\epsilon$ in $\mathbb{R}^3$, $\mathbb{E}\left[\boldsymbol{\delta}\boldsymbol{\delta}^\top\right] = \frac{\epsilon^2}{3}I$, so

$$
\mathbb{E}[\Delta f] = \frac{\epsilon^2}{2d}\operatorname{tr}H, \qquad \operatorname{tr}H_A = 102,\ \operatorname{tr}H_B = 12
$$

$$
\Longrightarrow \quad \mathbb{E}[\Delta f_A] = \frac{0.01}{6}(102) = 0.17, \qquad \mathbb{E}[\Delta f_B] = \frac{0.01}{6}(12) = 0.02 .
$$

**Verdict and the caveat.** $B$ is flatter by both measures — $25\times$ by worst case, $8.5\times$ on average. But the two measures disagree in *character*: $A$ is anisotropic (extremely sharp in one direction, flat in two), while $B$ is isotropic. A single scalar hides this. Note also that $A$'s $\lambda_{\max}=100$ caps the learning rate at $\eta \lt 2/100$ (Topic 03) while $B$ tolerates $\eta \lt 0.5$ — sharpness has an immediate *optimization* consequence independent of any generalization claim.

$$
\boxed{\text{sharpness}_\epsilon = \tfrac{\epsilon^2}{2}\lambda_{\max}: \ A = 0.5 \ \text{vs} \ B = 0.02; \quad \mathbb{E}[\Delta f] = \tfrac{\epsilon^2}{2d}\operatorname{tr}H: \ 0.17 \ \text{vs} \ 0.02}
$$

*Key takeaway:* $\lambda_{\max}$ measures worst-case sharpness and sets the stability ceiling; $\operatorname{tr}H$ measures average sharpness and is cheaply estimable by Hutchinson's method. Both are heavily used and, as the next problem shows, both are reparameterization-dependent.

### Problem L2.5: Sharpness is not a property of the function (Dinh et al.)

For a two-layer ReLU network $F(\mathbf{x}) = W^{(2)}\phi\!\left(W^{(1)}\mathbf{x}\right)$ with $\phi = \mathrm{ReLU}$, show that the reparameterization $\left(\alpha W^{(1)},\ \alpha^{-1}W^{(2)}\right)$ leaves $F$ unchanged, and describe its effect on the loss Hessian. What does this imply about "flat minima generalize better"?

**Solution.**

**Step 1 (function invariance).** ReLU is positively homogeneous: $\phi(\alpha u) = \alpha\phi(u)$ for $\alpha \gt 0$. Hence for any $\alpha \gt 0$,

$$
\alpha^{-1}W^{(2)}\,\phi\!\left(\alpha W^{(1)}\mathbf{x}\right) = \alpha^{-1}W^{(2)}\,\alpha\,\phi\!\left(W^{(1)}\mathbf{x}\right) = W^{(2)}\phi\!\left(W^{(1)}\mathbf{x}\right) = F(\mathbf{x}).
$$

The network computes *exactly the same function* for every $\alpha$, so its training loss, test loss, and generalization gap are all unchanged. The map $T_\alpha$ carries every minimizer to another minimizer, generating a one-parameter family of equally-good solutions.

**Step 2 (Hessian is not invariant).** Loss curvature transforms by the chain rule. With $\boldsymbol{\theta}' = T_\alpha\boldsymbol{\theta}$ and $J = \partial\boldsymbol{\theta}/\partial\boldsymbol{\theta}' = \operatorname{diag}\!\left(\alpha^{-1}I,\ \alpha I\right)$, at a minimum (where the gradient vanishes and no second-derivative-of-the-map term survives),

$$
H' = J^\top H J .
$$

Blocks scale as $H'_{11} = \alpha^{-2}H_{11}$, $H'_{22} = \alpha^{2}H_{22}$, $H'_{12} = H_{12}$. Taking $\alpha$ large makes the second-layer block — and therefore $\lambda_{\max}(H')$ and $\operatorname{tr}H'$ — arbitrarily large; taking $\alpha$ small does the same to the first-layer block. In either direction the minimum becomes arbitrarily "sharp" by these measures while remaining the same model.

**Step 3 (implication).** Since sharpness can be inflated without limit at fixed generalization, no *unmodified* Euclidean sharpness measure can be a cause of generalization. Sharpness is a property of the coordinate system, not of the learned function.

$$
\boxed{F \text{ invariant under } \left(\alpha W^{(1)}, \alpha^{-1}W^{(2)}\right), \ \text{but } \lambda_{\max}(H) \to \infty \text{ as } \alpha\to\infty}
$$

*Key takeaway:* The empirical correlation between flatness and generalization is real but its formalization is subtle. Fixes in use: normalize by parameter scale (filter normalization when *plotting*, as in Li et al.), use the Fisher/natural metric instead of the Euclidean one, or define sharpness through a scale-free adversarial objective (SAM's $\max_{\lVert \boldsymbol{\epsilon}\rVert\le\rho}$ with layerwise-normalized $\rho$).

### Problem L2.6: Overparameterization creates connected minima

Analyze $f(u,v) = (uv-1)^2$ on $\mathbb{R}^2$: find all critical points, describe the set of global minimizers, compute the Hessian there, and relate the picture to overparameterized networks.

**Solution.**

**Step 1 (critical points).** Let $r = uv-1$. Then $\nabla f = \left(2rv,\ 2ru\right)^\top$, which vanishes iff $r = 0$, or $u = v = 0$.

- $r = 0$: the hyperbola $\mathcal{M} = \{(u,v) : uv = 1\}$, a one-dimensional *manifold* of points with $f = 0$ — the global minimum, since $f \ge 0$.
- $u = v = 0$: the origin, where $f(0,0)=1$.

**Step 2 (the origin).** $\nabla^2 f = \begin{bmatrix}2v^2 & 4uv-2\\ 4uv-2 & 2u^2\end{bmatrix}$, which at the origin equals $\begin{bmatrix}0&-2\\-2&0\end{bmatrix}$ with eigenvalues $\pm 2$. It is a **saddle**, not a spurious local minimum: there are descent directions (along $u=v$, $f(t,t)=(t^2-1)^2$ decreases from $1$).

**Step 3 (Hessian on the solution manifold).** At a point with $uv=1$ we have $r=0$, so

$$
H = \begin{bmatrix}2v^2 & 2\\ 2 & 2u^2\end{bmatrix}, \qquad \det H = 4u^2v^2 - 4 = 4(uv)^2-4 = 0 .
$$

The Hessian is singular *everywhere on the manifold*: one eigenvalue is $0$ (tangent to $\mathcal{M}$ — moving along the solution set costs nothing) and the other is $\operatorname{tr}H = 2(u^2+v^2) \gt 0$ (normal to $\mathcal{M}$). The minimum is non-strict, and no amount of second-order information will single out a point.

**Step 4 (the ML reading).** The two structural features generalize: **(i)** no spurious local minima — the only non-optimal critical point is a saddle, which gradient descent escapes; **(ii)** the global minimizers form a *connected*, degenerate manifold, so the Hessian at any solution has a large null space. Real overparameterized networks show the same signature: an eigenvalue bulk concentrated at zero with a handful of large outliers, and low-loss paths connecting apparently distinct solutions (mode connectivity). Which point of $\mathcal{M}$ you reach is decided by the optimizer's *implicit bias* — from $(u_0,v_0)$ gradient flow conserves $u^2-v^2$, so the trajectory lands at a specific, initialization-determined solution.

$$
\boxed{\arg\min f = \{uv=1\} \ \text{(connected manifold)}; \ \det H = 0 \ \text{on it}; \ (0,0) \ \text{is a saddle, } \lambda=\pm 2}
$$

*Key takeaway:* This two-parameter toy is the smallest honest model of overparameterized deep learning: a benign landscape whose only obstruction is a saddle, and whose "solution" is a set rather than a point — making implicit regularization, not optimization, the thing that determines the answer.

### Problem L3.1: Prove the second-order characterization of convexity

Prove that for $f\in C^2$ on an open convex domain, $f$ is convex if and only if $\nabla^2 f(\mathbf{x})\succeq 0$ for all $\mathbf{x}$.

**Solution.**

**Step 0 (reduce to one dimension).** For fixed $\mathbf{x}$ in the domain and $\mathbf{u}\in\mathbb{R}^d$, define $g(t) = f(\mathbf{x}+t\mathbf{u})$ on the open interval where it is defined. Two chain-rule computations give

$$
g'(t) = \nabla f(\mathbf{x}+t\mathbf{u})^\top\mathbf{u}, \qquad g''(t) = \mathbf{u}^\top\nabla^2 f(\mathbf{x}+t\mathbf{u})\,\mathbf{u}.
$$

By Definition 2.2 applied along segments, $f$ is convex on its (convex) domain **iff** every such restriction $g$ is convex. So the multivariate claim is equivalent to the scalar claim "$g$ convex $\iff g''\ge0$", plus the observation that "$g''(t)\ge0$ for all $\mathbf{x},\mathbf{u},t$" is literally the statement $\nabla^2 f\succeq 0$ everywhere.

**Step 1 ($g''\ge 0 \Rightarrow g$ convex).** Fix $s,t$ in the interval. By Taylor's theorem with Lagrange remainder at order $k=1$ (Topic 02, Proof 3.1), there is $\xi$ between $s$ and $t$ with

$$
g(t) = g(s) + g'(s)(t-s) + \frac{g''(\xi)}{2}(t-s)^2 \ \ge \ g(s)+g'(s)(t-s),
$$

since $g''(\xi)\ge 0$ and $(t-s)^2 \ge 0$. That is the first-order condition in 1D, which by Proof 3.1 is equivalent to convexity of $g$.

**Step 2 ($g$ convex $\Rightarrow g''\ge 0$).** Convexity gives the first-order condition at both points:

$$
g(t)\ge g(s)+g'(s)(t-s), \qquad g(s)\ge g(t)+g'(t)(s-t).
$$

Adding them and cancelling $g(s)+g(t)$ yields $0 \ge \left(g'(s)-g'(t)\right)(t-s)$, i.e.

$$
\left(g'(t)-g'(s)\right)(t-s) \ge 0,
$$

so $g'$ is non-decreasing. For $t \gt s$, dividing by $(t-s) \gt 0$ and letting $t\downarrow s$ gives $g''(s) = \lim_{t\to s}\frac{g'(t)-g'(s)}{t-s} \ge 0$.

**Step 3 (assemble).** Steps 1–2 prove the scalar equivalence for every $\mathbf{x},\mathbf{u}$. Since $g''(0) = \mathbf{u}^\top\nabla^2 f(\mathbf{x})\mathbf{u}$, requiring it for all $\mathbf{u}$ is exactly $\nabla^2 f(\mathbf{x})\succeq 0$, and requiring it for all $\mathbf{x}$ completes the equivalence. $\blacksquare$

**Sharp remark on strictness.** $\nabla^2 f \succ 0$ implies strict convexity (Step 1 gives a strict inequality), but the converse *fails*: $f(x)=x^4$ is strictly convex while $f''(0)=0$. The correct equivalences are: convex $\iff \nabla^2 f\succeq 0$; $\mu$-strongly convex $\iff \nabla^2 f\succeq\mu I$; strictly convex sits strictly between them.

$$
\boxed{f \text{ convex} \iff \nabla^2 f \succeq 0; \quad \mu\text{-strongly convex} \iff \nabla^2 f\succeq\mu I; \quad \nabla^2 f\succ0 \Rightarrow \text{strict, not} \Leftarrow}
$$

*Key takeaway:* Every multivariate convexity statement is a one-dimensional statement about restrictions to lines. That reduction — the same one used for multivariate Taylor in Topic 02 — is the single most reusable technique in this whole curriculum.

### Problem L3.2: Gradient descent escapes a strict saddle

Let $\mathbf{x}_c$ be a critical point with $\nabla^2 f(\mathbf{x}_c) = H$ having $\lambda_{\min}(H) = -\gamma \lt 0$, and run gradient descent with $0 \lt \eta \lt 1/L$. (a) Show the linearized dynamics has an expanding mode and estimate the escape time. (b) Argue that the set of initializations converging to $\mathbf{x}_c$ has measure zero.

**Solution.**

**(a) The expanding mode.**

*Step 1 (linearize).* Put $\mathbf{e}_k = \mathbf{x}_k-\mathbf{x}_c$. Since $\nabla f(\mathbf{x}_c) = 0$, Taylor gives $\nabla f(\mathbf{x}_k) = H\mathbf{e}_k+o\!\left(\lVert \mathbf{e}_k\rVert_2\right)$, so

$$
\mathbf{e}_{k+1} = \left(I-\eta H\right)\mathbf{e}_k + o\!\left(\lVert \mathbf{e}_k\rVert_2\right).
$$

*Step 2 (diagonalize).* With $H = Q\Lambda Q^\top$ and $\tilde{\mathbf{e}} = Q^\top\mathbf{e}$, mode $i$ carries the multiplier $1-\eta\lambda_i$. For $\lambda_i \gt 0$ and $\eta \lt 1/L \le 1/\lambda_i$ we get $0 \lt 1-\eta\lambda_i \lt 1$: contraction. For the negative eigenvalue $\lambda_{\min}=-\gamma$:

$$
1-\eta\lambda_{\min} = 1+\eta\gamma \gt 1 .
$$

*Step 3 (escape time).* Ignoring higher-order terms while the iterate is close, the escape component grows as $\lvert \tilde{e}_{k,-}\rvert \approx (1+\eta\gamma)^k\lvert \tilde{e}_{0,-}\rvert$. It exceeds a fixed radius $R$ after

$$
k \approx \frac{\log\!\left(R/\lvert \tilde{e}_{0,-}\rvert\right)}{\log(1+\eta\gamma)} \approx \frac{1}{\eta\gamma}\log\frac{R}{\lvert \tilde{e}_{0,-}\rvert} \quad (\eta\gamma \ll 1).
$$

Escape is guaranteed but takes $\Theta(1/(\eta\gamma))$ steps — long when the negative curvature $\gamma$ is small, which is exactly the plateau seen in practice. A random perturbation of size $\rho$ ensures $\lvert \tilde{e}_{0,-}\rvert \gtrsim \rho$ and caps the wait at $O\!\left(\frac{1}{\eta\gamma}\log\frac{1}{\rho}\right)$ — the design of perturbed gradient descent, and a free service provided by SGD's minibatch noise.

**(b) The trapping set is measure zero.**

*Step 1 (the map is a diffeomorphism).* Let $T(\mathbf{x}) = \mathbf{x}-\eta\nabla f(\mathbf{x})$, so $DT(\mathbf{x}) = I-\eta\nabla^2 f(\mathbf{x})$. With $\eta \lt 1/L$, every eigenvalue of $DT$ lies in $(0, 2)$ and is bounded away from $0$, so $DT$ is invertible everywhere and $T$ is a local diffeomorphism.

*Step 2 (stable manifold).* At the fixed point $\mathbf{x}_c$, $DT(\mathbf{x}_c) = I-\eta H$ has $\#\{\lambda_i \lt 0\} \ge 1$ eigenvalues of modulus $\gt 1$. The Center-Stable Manifold Theorem gives a local invariant manifold $W^{cs}$ containing every point whose forward orbit stays near $\mathbf{x}_c$, of dimension

$$
\dim W^{cs} = d - \#\{i : \lambda_i \lt 0\} \le d-1 .
$$

A submanifold of dimension $\lt d$ has Lebesgue measure zero.

*Step 3 (globalize).* Any initialization converging to $\mathbf{x}_c$ eventually enters the neighborhood, so it lies in $\bigcup_{k\ge0}T^{-k}\!\left(W^{cs}\right)$. Since $T$ is a diffeomorphism, preimages of measure-zero sets are measure zero, and a countable union of measure-zero sets is measure zero. Hence almost every $\mathbf{x}_0$ escapes (Lee et al., 2016). $\blacksquare$

$$
\boxed{\text{expanding factor } 1+\eta\gamma \gt 1; \ \text{escape in } \Theta\!\left(\tfrac{1}{\eta\gamma}\log\tfrac{1}{\rho}\right) \text{ steps}; \ \Pr[\text{trapped}] = 0}
$$

*Key takeaway:* "Almost surely escapes" and "escapes quickly" are different statements. The theory rules out permanent entrapment; only the *magnitude* of the negative curvature governs how much of training is spent crawling across saddle plateaus.

### Problem L3.3: Convex sublevel sets do not imply convexity

Prove that every sublevel set of a convex function is convex, and give a function with convex sublevel sets that is not convex. Name the weaker class such functions belong to and say why optimizers still care.

**Solution.**

**Step 1 (convex $\Rightarrow$ convex sublevel sets).** Fix $c$ and let $S_c = \{\mathbf{x} : f(\mathbf{x})\le c\}$. Take $\mathbf{x},\mathbf{y}\in S_c$ and $\theta\in[0,1]$. Convexity gives

$$
f\!\left(\theta\mathbf{x}+(1-\theta)\mathbf{y}\right) \le \theta f(\mathbf{x})+(1-\theta)f(\mathbf{y}) \le \theta c + (1-\theta)c = c,
$$

so the combination is in $S_c$. Hence *every* sublevel set is convex.

**Step 2 (the converse fails).** Take $f(x) = -e^{-x^2}$ on $\mathbb{R}$. Its sublevel sets are

$$
S_c = \{x : -e^{-x^2}\le c\} = \begin{cases} \emptyset & c \lt -1,\\ \left[-\sqrt{\log\frac{1}{-c}},\ \sqrt{\log\frac{1}{-c}}\right] & -1\le c \lt 0,\\ \mathbb{R} & c \ge 0,\end{cases}
$$

all intervals, hence all convex. But $f$ is not convex: $f''(x) = \left(2-4x^2\right)e^{-x^2}$, which is negative for $\lvert x\rvert \gt 1/\sqrt{2}$ — the tails are concave. Directly: $f(0) = -1$ while $\frac{1}{2}f(-2)+\frac{1}{2}f(2) = -e^{-4}\approx -0.018$, and $-1 \le -0.018$ holds, but testing the chord between $x=1$ and $x=3$ shows the graph rising above it.

**Step 3 (the right class).** Functions all of whose sublevel sets are convex are called **quasiconvex**, characterized by

$$
f\!\left(\theta\mathbf{x}+(1-\theta)\mathbf{y}\right) \le \max\left\{f(\mathbf{x}),\ f(\mathbf{y})\right\} .
$$

Convex $\Rightarrow$ quasiconvex, never conversely. Quasiconvexity still forbids "interior bumps" that would create a separated basin, so descent methods cannot be walled off from the optimum, and bisection on the sublevel sets solves quasiconvex programs. What is lost: a stationary point need not be optimal ($f(x)=-e^{-x^2}$ has a *maximum* at $x=0$ with $f'(0)=0$), so the one-line global certificate of Proof 3.3(c) disappears.

$$
\boxed{f \text{ convex} \Rightarrow S_c \text{ convex } \forall c; \ \text{converse false } (f = -e^{-x^2}, \text{ quasiconvex})}
$$

*Key takeaway:* Convexity is a statement about the *epigraph*, which is strictly stronger than a statement about any or all sublevel sets. The hierarchy convex $\subset$ quasiconvex $\subset$ general is the right map for judging what a landscape assumption actually buys.

### Problem L3.4: A benign non-convex landscape — matrix factorization

For symmetric $M \in \mathbb{R}^{n\times n}$ with $M \succeq 0$ of rank $r$, consider $f(U) = \frac{1}{4}\lVert UU^\top - M\rVert_F^2$ over $U\in\mathbb{R}^{n\times r}$. (a) Show $f$ is non-convex. (b) Derive the critical-point equation. (c) Show the global minimizers form a connected set, and explain in what sense the landscape is nevertheless benign.

**Solution.**

**(a) Non-convexity from symmetry.** For any orthogonal $R\in\mathbb{R}^{r\times r}$, $\left(UR\right)\left(UR\right)^\top = URR^\top U^\top = UU^\top$, so $f(UR) = f(U)$: the objective is invariant under the whole orthogonal group $O(r)$. Also $f(-U)=f(U)$. Taking $U^\star$ a global minimizer with $r\ge1$, both $U^\star$ and $-U^\star$ are global minimizers while their midpoint $0$ gives $f(0)=\frac{1}{4}\lVert M\rVert_F^2 \gt 0$ (for $M\neq0$). A convex function's minimizer set is convex (Proof 3.3(b)), so $f$ is not convex.

**(b) Critical points.** Write $E = UU^\top - M$ (symmetric). A first-order perturbation $U\to U+\Delta$ changes $UU^\top$ by $\Delta U^\top + U\Delta^\top$ to first order, so

$$
\mathrm{d}f = \frac{1}{2}\left\langle E,\ \Delta U^\top + U\Delta^\top \right\rangle_F = \left\langle EU,\ \Delta \right\rangle_F \quad \Longrightarrow \quad \nabla f(U) = \left(UU^\top - M\right)U .
$$

Critical points therefore satisfy

$$
\left(UU^\top - M\right)U = 0 .
$$

Solutions include $U = 0$ and, more generally, $U = V_S\Lambda_S^{1/2}R$ where $V_S,\Lambda_S$ collect *any* subset $S$ of $r$ eigenpairs of $M$ and $R$ is orthogonal — verify: $UU^\top = V_S\Lambda_S V_S^\top$, and $\left(V_S\Lambda_SV_S^\top - M\right)V_S\Lambda_S^{1/2} = 0$ because $M V_S = V_S\Lambda_S$.

**(c) Global minimizers and benignity.** By the Eckart–Young theorem the minimum of $\lVert UU^\top-M\rVert_F$ over rank-$r$ PSD matrices is attained by the top-$r$ eigen-decomposition, so with $\operatorname{rank}M = r$ the minimum value is $0$ and

$$
\arg\min f = \left\{ U^\star R \ :\ R \in O(r) \right\}, \qquad U^\star = V_r\Lambda_r^{1/2},
$$

an orbit of the (connected, for $\det R = 1$) rotation group — a *manifold* of solutions, not a point, of dimension $\binom{r}{2}$. The Hessian at any minimizer is singular along the directions tangent to that orbit.

The landscape is benign in the precise sense established for this problem (Ge, Lee & Ma, 2016; Bhojanapalli et al., 2016): **every critical point is either a global minimizer or a strict saddle** with $\lambda_{\min}\left(\nabla^2 f\right) \lt 0$. The mechanism is visible in the family above: if the chosen subset $S$ omits a larger eigenvalue $\lambda_j$ in favor of a smaller $\lambda_i$, then rotating the corresponding column of $U$ toward $\mathbf{v}_j$ strictly decreases $f$, so that critical point has negative curvature. By Problem L3.2, gradient descent from a random start therefore converges to a *global* minimum despite the non-convexity.

$$
\boxed{\nabla f(U) = \left(UU^\top-M\right)U; \quad \arg\min f = \{U^\star R : R\in O(r)\}; \quad \text{all saddles strict}}
$$

*Key takeaway:* "Non-convex" and "hard" are not synonyms. Matrix factorization, phase retrieval, tensor decomposition, and (empirically) overparameterized neural networks all share the *strict-saddle + connected-minima* profile, which is enough for randomly initialized first-order methods to succeed — the modern replacement for convexity as a sufficient condition for trainability.
