## Level 0 — Concept Check

### Problem L0.1: The quadratic model of the exponential

Write the degree-2 Taylor polynomial of $f(x) = e^x$ at $x = 0$, and state the exact Lagrange form of the error at a point $h$.

**Solution.**

All derivatives of $e^x$ equal $e^x$, and $e^0 = 1$. Hence

$$
T_2(h) = 1 + h + \frac{h^2}{2}.
$$

By Taylor's theorem with Lagrange remainder ($k = 2$), there is a $\xi$ between $0$ and $h$ with

$$
e^h - T_2(h) = \frac{f'''(\xi)}{3!}h^3 = \frac{e^{\xi}}{6}h^3.
$$

$$
\boxed{e^h = 1 + h + \frac{h^2}{2} + \frac{e^{\xi}}{6}h^3, \quad \xi \text{ between } 0 \text{ and } h}
$$

*Key takeaway:* The remainder is an exact equality at a mystery point $\xi$ — bounding $e^{\xi}$ over the interval converts it into a usable worst-case error bar.

### Problem L0.2: What $L$-smoothness looks like

State the definition of $L$-smoothness and describe geometrically what the descent lemma says about the graph of $f$.

**Solution.**

$f$ is $L$-smooth if its gradient is $L$-Lipschitz:

$$
\lVert \nabla f(x) - \nabla f(y) \rVert_2 \le L\lVert x - y \rVert_2 \quad \text{for all } x, y.
$$

The descent lemma then states that at every anchor point $x$, the whole graph of $f$ lies *below* the quadratic $q_x(y) = f(x) + \nabla f(x)^\top (y-x) + \frac{L}{2}\lVert y-x \rVert_2^2$, which touches the graph at $x$ with matching slope. (Symmetrically, $f$ lies above the same expression with $-\frac{L}{2}$.)

$$
\boxed{f \text{ is trapped between two tangent parabolas of opening } \pm\frac{L}{2} \text{ at every point}}
$$

*Key takeaway:* $L$ is a universal curvature cap. Any algorithm that respects it can guarantee progress without ever seeing more than one gradient.

### Problem L0.3: One-step Newton

Why does Newton's method minimize a positive-definite quadratic in exactly one step, from any starting point?

**Solution.**

Newton's step minimizes the local second-order model $m(h) = f(x) + \nabla f(x)^\top h + \frac{1}{2}h^\top \nabla^2 f(x) h$. For a quadratic $f$, the second-order Taylor expansion has zero remainder (all third derivatives vanish), so the model equals the function: $m(h) = f(x + h)$ identically.

Minimizing the model therefore minimizes the function itself, and the step $h = -\left[\nabla^2 f\right]^{-1}\nabla f(x)$ lands exactly on the unique global minimizer.

$$
\boxed{\text{Model} = \text{function for quadratics} \implies \text{one Newton step is exact}}
$$

*Key takeaway:* Newton's method is only as good as its model; on quadratics the model is perfect, and near any nondegenerate minimum it is nearly perfect — hence local quadratic convergence.

### Problem L0.4: Slopes on a log-log error plot

An experiment measures the error of the first- and second-order Taylor models of a smooth function at step sizes $h = 10^{-1}, 10^{-2}, 10^{-3}$. What slopes should appear on a log-log plot of error versus $h$, and why?

**Solution.**

The first-order model has error $\frac{f''(\xi)}{2}h^2 = O(h^2)$; the second-order model has error $\frac{f'''(\xi)}{6}h^3 = O(h^3)$. If $E(h) = C h^p$, then

$$
\log E(h) = \log C + p \log h,
$$

a straight line of slope $p$ on log-log axes. So the first-order error line has slope $2$ and the second-order line slope $3$: each decade of $h$ buys two (resp. three) decades of accuracy.

$$
\boxed{\text{slope } 2 \text{ (first order)}, \qquad \text{slope } 3 \text{ (second order)}}
$$

*Key takeaway:* Convergence-order plots are executable proofs — a buggy derivative implementation betrays itself by producing the wrong slope.

## Level 1 — Foundation

### Problem L1.1: Certified logarithm approximation

Compute the degree-2 Taylor polynomial of $f(x) = \ln(1+x)$ at $0$, and bound the error for all $x \in [0, 0.1]$ using the Lagrange remainder.

**Solution.**

Derivatives: $f'(x) = (1+x)^{-1}$, $f''(x) = -(1+x)^{-2}$, $f'''(x) = 2(1+x)^{-3}$. At $0$: $f(0) = 0$, $f'(0) = 1$, $f''(0) = -1$. Hence

$$
T_2(x) = x - \frac{x^2}{2}.
$$

Lagrange remainder with $k = 2$: for some $\xi \in (0, x)$,

$$
R_2(x) = \frac{f'''(\xi)}{6}x^3 = \frac{x^3}{3(1+\xi)^3}.
$$

For $x \in [0, 0.1]$ we have $\xi \ge 0$, so $(1+\xi)^{-3} \le 1$ and

$$
0 \le R_2(x) \le \frac{(0.1)^3}{3} = \frac{1}{3000} \approx 3.34 \times 10^{-4}.
$$

$$
\boxed{\ln(1+x) = x - \frac{x^2}{2} + R_2(x), \qquad 0 \le R_2(x) \le \tfrac{1}{3000} \text{ on } [0, 0.1]}
$$

*Key takeaway:* Because the remainder here is positive, the model is a certified *underestimate* — sign information from the remainder is often as useful as its size.

### Problem L1.2: The sine inequality

Prove that $\lvert \sin x - x \rvert \le \dfrac{\lvert x \rvert^3}{6}$ for all real $x$.

**Solution.**

Apply Taylor's theorem to $f(x) = \sin x$ at $0$ with $k = 2$. Since $f(0) = 0$, $f'(0) = 1$, $f''(0) = 0$, the degree-2 polynomial is $T_2(x) = x$, and for some $\xi$ between $0$ and $x$:

$$
\sin x = x + \frac{f'''(\xi)}{6}x^3 = x - \frac{\cos\xi}{6}x^3.
$$

Taking absolute values and using $\lvert \cos\xi \rvert \le 1$:

$$
\lvert \sin x - x \rvert = \frac{\lvert \cos\xi \rvert}{6}\lvert x \rvert^3 \le \frac{\lvert x \rvert^3}{6}. \qquad \blacksquare
$$

$$
\boxed{\lvert \sin x - x \rvert \le \frac{\lvert x \rvert^3}{6}}
$$

*Key takeaway:* The small-angle approximation of physics comes with a free, explicit error bar — cubic in the angle, which is why it works so well below a few degrees.

### Problem L1.3: A multivariate expansion by hand

Compute the second-order Taylor expansion of $f(x, y) = e^{x}\cos y$ at the origin.

**Solution.**

Value and gradient: $f(0,0) = 1$; $f_x = e^x\cos y$ gives $f_x(0,0) = 1$; $f_y = -e^x \sin y$ gives $f_y(0,0) = 0$.

Second partials: $f_{xx} = e^x\cos y = 1$ at the origin; $f_{xy} = -e^x\sin y = 0$; $f_{yy} = -e^x\cos y = -1$. Hessian at origin:

$$
H = \begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}.
$$

Second-order model with $h = (x, y)$:

$$
f(x,y) \approx 1 + x + \frac{1}{2}\begin{bmatrix} x & y \end{bmatrix}\begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}\begin{bmatrix} x \\ y \end{bmatrix} = 1 + x + \frac{x^2 - y^2}{2}.
$$

$$
\boxed{e^{x}\cos y = 1 + x + \frac{x^2 - y^2}{2} + o(x^2 + y^2)}
$$

*Key takeaway:* The mixed-sign Hessian (indefinite) says the origin sits on saddle-shaped level geometry along the graph — a preview of Topic 04's classification.

### Problem L1.4: Smoothness constants of quadratics

Find the best $L$-smoothness constant of (a) $f(w) = \frac{1}{2}w^\top H w$ with $H = \operatorname{diag}(1, 8)$; (b) the least-squares loss $L(w) = \frac{1}{n}\lVert Xw - y \rVert_2^2$ in terms of the spectrum of $X^\top X$.

**Solution.**

**(a)** The gradient map is $\nabla f(w) = Hw$, which is linear, so the Lipschitz constant of the gradient equals $\lVert H \rVert_{\mathrm{op}} = \lambda_{\max}(H) = 8$ (attained along $e_2$).

**(b)** Here $\nabla L(w) = \frac{2}{n}X^\top(Xw - y)$ and

$$
\nabla L(w_1) - \nabla L(w_2) = \frac{2}{n}X^\top X (w_1 - w_2),
$$

so the best constant is the operator norm of the constant Hessian:

$$
L = \frac{2}{n}\lambda_{\max}(X^\top X) = \frac{2}{n}\sigma_{\max}(X)^2,
$$

where $\sigma_{\max}$ is the largest singular value.

$$
\boxed{\text{(a) } L = 8, \qquad \text{(b) } L = \tfrac{2}{n}\lambda_{\max}(X^\top X)}
$$

*Key takeaway:* For quadratic losses the abstract constant $L$ is a concrete spectral quantity of the data — and it directly caps the stable learning rate in Topic 03.

### Problem L1.5: Sufficient decrease from the descent lemma

Let $f$ be $L$-smooth. Show that the gradient step $y = x - \frac{1}{L}\nabla f(x)$ guarantees

$$
f(y) \le f(x) - \frac{1}{2L}\lVert \nabla f(x) \rVert_2^2.
$$

**Solution.**

Apply the descent lemma with $y - x = -\frac{1}{L}\nabla f(x)$:

$$
f(y) \le f(x) + \nabla f(x)^\top\left( -\frac{1}{L}\nabla f(x) \right) + \frac{L}{2}\left\lVert \frac{1}{L}\nabla f(x) \right\rVert_2^2.
$$

Compute the two correction terms: the inner product gives $-\frac{1}{L}\lVert \nabla f(x) \rVert_2^2$, and the quadratic term gives $\frac{L}{2}\cdot\frac{1}{L^2}\lVert \nabla f(x) \rVert_2^2 = \frac{1}{2L}\lVert \nabla f(x) \rVert_2^2$. Summing:

$$
f(y) \le f(x) - \frac{1}{L}\lVert \nabla f(x) \rVert_2^2 + \frac{1}{2L}\lVert \nabla f(x) \rVert_2^2 = f(x) - \frac{1}{2L}\lVert \nabla f(x) \rVert_2^2. \qquad \blacksquare
$$

$$
\boxed{f\!\left( x - \tfrac{1}{L}\nabla f(x) \right) \le f(x) - \tfrac{1}{2L}\lVert \nabla f(x) \rVert_2^2}
$$

*Key takeaway:* Every gradient step with $\eta = 1/L$ pays for itself: the loss drops by at least a computable amount proportional to the squared gradient norm — the engine of all convergence proofs in Topic 03.

### Problem L1.6: Digit doubling in Newton's method

Run two Newton iterations for $f(x) = x^2 - 2$ starting from $x_0 = 1.5$, and verify the quadratic (digit-doubling) convergence toward $\sqrt{2} \approx 1.41421356$.

**Solution.**

Newton's update for a root of $f$: $x_{t+1} = x_t - \dfrac{f(x_t)}{f'(x_t)} = x_t - \dfrac{x_t^2 - 2}{2x_t} = \dfrac{x_t}{2} + \dfrac{1}{x_t}$.

Iteration 1: $x_1 = \dfrac{1.5}{2} + \dfrac{1}{1.5} = 0.75 + 0.66667 = 1.41667$. Error: $\lvert x_1 - \sqrt{2} \rvert \approx 2.45 \times 10^{-3}$.

Iteration 2: $x_2 = \dfrac{1.41667}{2} + \dfrac{1}{1.41667} = 0.708333 + 0.705882 = 1.414216$. Error: $\lvert x_2 - \sqrt{2} \rvert \approx 2.1 \times 10^{-6}$.

The error went from $\sim 8 \times 10^{-2}$ (start) to $\sim 2 \times 10^{-3}$ to $\sim 2 \times 10^{-6}$: each step roughly *squares* the error, doubling the number of correct digits, consistent with $e_{t+1} \approx \frac{f''}{2f'}e_t^2 = \frac{e_t^2}{2x_t}$.

$$
\boxed{x_2 \approx 1.414216, \quad \text{errors } 8\times 10^{-2} \to 2.4\times 10^{-3} \to 2.1\times 10^{-6} \text{ (quadratic)}}
$$

*Key takeaway:* Quadratic convergence is visible after two iterations — and this square-root iteration (Heron's method) is literally how hardware computes $\sqrt{2}$.

## Level 2 — Applications in AI/ML

### Problem L2.1: Quadratic model of the logistic loss

For a single sample with $x \in \mathbb{R}^d$, $y \in \{0,1\}$, the logistic loss is $\ell(w) = -y\log\sigma(x^\top w) - (1-y)\log(1 - \sigma(x^\top w))$. Build its second-order Taylor model at $w = 0$.

**Solution.**

From Topic 01: $\nabla \ell(w) = (\sigma(x^\top w) - y)x$ and, differentiating again with $\sigma' = \sigma(1-\sigma)$,

$$
\nabla^2 \ell(w) = \sigma(x^\top w)\left( 1 - \sigma(x^\top w) \right) x x^\top.
$$

At $w = 0$: $\sigma(0) = \frac{1}{2}$, so $\ell(0) = \log 2$ (either branch), $\nabla \ell(0) = \left( \frac{1}{2} - y \right)x$, and $\nabla^2 \ell(0) = \frac{1}{4}xx^\top$. The quadratic model is

$$
\ell(w) \approx \log 2 + \left( \tfrac{1}{2} - y \right) x^\top w + \frac{1}{8}\left( x^\top w \right)^2.
$$

$$
\boxed{\ell(w) \approx \log 2 + \left( \tfrac{1}{2} - y \right)x^\top w + \tfrac{1}{8}(x^\top w)^2}
$$

*Key takeaway:* The logistic Hessian $\sigma(1-\sigma)xx^\top$ is at most $\frac{1}{4}xx^\top$: logistic regression is automatically $L$-smooth with $L = \frac{1}{4n}\lambda_{\max}(X^\top X)$ — a free, data-computable learning-rate certificate.

### Problem L2.2: Second-order pruning (Optimal Brain Damage)

A trained network sits at a minimum $w^\star$ with diagonal Hessian entries $H_{11} = 10$, $H_{22} = 0.1$. Weights are $w_1 = 0.3$, $w_2 = 0.8$. Using the local quadratic model, which weight should be pruned (set to zero), and what loss increase does the model predict?

**Solution.**

At a minimum, $\nabla L(w^\star) = 0$, so the quadratic model for a perturbation $h$ is $\Delta L \approx \frac{1}{2}h^\top H h$. Zeroing weight $i$ means $h = -w_i e_i$, giving the saliency

$$
\Delta L_i \approx \frac{1}{2}H_{ii}w_i^2.
$$

Compute both: $\Delta L_1 = \frac{1}{2}(10)(0.09) = 0.45$ and $\Delta L_2 = \frac{1}{2}(0.1)(0.64) = 0.032$.

Weight 2 is 14 times cheaper to remove despite being larger in magnitude — magnitude pruning would have chosen wrongly here.

$$
\boxed{\text{Prune } w_2: \ \Delta L \approx 0.032 \ \text{vs} \ 0.45 \text{ for } w_1}
$$

*Key takeaway:* Curvature, not magnitude, prices a parameter's importance; this is the Taylor-model logic of LeCun's Optimal Brain Damage and modern second-order pruning.

### Problem L2.3: Softplus near the origin

The softplus activation is $s(x) = \log(1 + e^x)$. Compute its second-order Taylor model at $0$ and interpret the coefficients.

**Solution.**

Derivatives: $s'(x) = \sigma(x)$ (the sigmoid) and $s''(x) = \sigma'(x) = \sigma(x)(1 - \sigma(x))$.

At $0$: $s(0) = \log 2$, $s'(0) = \frac{1}{2}$, $s''(0) = \frac{1}{4}$. Hence

$$
s(x) \approx \log 2 + \frac{x}{2} + \frac{x^2}{8}.
$$

Interpretation: near $0$, softplus behaves like a shifted half-slope line with gentle upward curvature $\frac{1}{4}$ — the smooth interpolation between the ReLU branches $0$ (slope 0) and $x$ (slope 1), whose average is exactly the $\frac{1}{2}$ linear coefficient.

$$
\boxed{\log(1 + e^x) = \log 2 + \frac{x}{2} + \frac{x^2}{8} + O(x^4)}
$$

(The cubic term vanishes because $s(x) - \frac{x}{2}$ is even, as $s(x) - s(-x) = x$.)

*Key takeaway:* Taylor coefficients of activations govern early-training dynamics: with small inputs, every smooth activation is its low-order model, so initialization theory is largely Taylor analysis.

### Problem L2.4: A learning-rate certificate from data

A design matrix has singular values $\sigma_1 = 20$, $\sigma_2 = 5$, $\sigma_3 = 1$, with $n = 100$ samples, and we minimize $L(w) = \frac{1}{n}\lVert Xw - y \rVert_2^2$. Compute the smoothness constant $L$ and the largest provably safe constant learning rate $\eta \lt 2/L$ for gradient descent.

**Solution.**

The Hessian is $\frac{2}{n}X^\top X$ with eigenvalues $\frac{2}{n}\sigma_i^2$. The smoothness constant is the largest one:

$$
L = \frac{2}{100} \cdot 20^2 = \frac{800}{100} = 8.
$$

Gradient descent on a quadratic is stable precisely for $\eta \lt 2/L$ (Topic 03), so

$$
\eta \lt \frac{2}{8} = 0.25,
$$

and the model-minimizing conservative choice is $\eta = 1/L = 0.125$.

$$
\boxed{L = 8, \qquad \eta_{\text{safe}} \lt 0.25, \qquad \eta_{1/L} = 0.125}
$$

*Key takeaway:* "Tune the learning rate" has a closed form for quadratic losses: it is a singular-value computation on the data matrix.

### Problem L2.5: Laplace approximation from the quadratic model

In 1D Bayesian inference, the negative log-posterior is $E(w)$ with a minimum at $w^\star$ where $E''(w^\star) = 25$. Use the second-order Taylor model to approximate the posterior $p(w) \propto e^{-E(w)}$ by a Gaussian. What are its mean and standard deviation?

**Solution.**

Expand around the mode: since $E'(w^\star) = 0$,

$$
E(w) \approx E(w^\star) + \frac{1}{2}E''(w^\star)(w - w^\star)^2.
$$

Exponentiating,

$$
p(w) \propto \exp\left( -\frac{E''(w^\star)}{2}(w - w^\star)^2 \right),
$$

which is a Gaussian with mean $w^\star$ and variance $1/E''(w^\star) = 1/25$. Hence the standard deviation is $1/5$.

$$
\boxed{p(w) \approx \mathcal{N}\left( w^\star,\ \sigma^2 = \tfrac{1}{25} \right), \qquad \sigma = 0.2}
$$

*Key takeaway:* Curvature at the mode *is* (inverse) uncertainty. In $d$ dimensions the covariance becomes $\left[\nabla^2 E(w^\star)\right]^{-1}$ — the Laplace approximation used for Bayesian neural networks and model evidence estimates.

### Problem L2.6: When no global $L$ exists

Show that $f(x) = x^4$ is not $L$-smooth on $\mathbb{R}$ for any finite $L$, but is $L_R$-smooth on $[-R, R]$ with $L_R = 12R^2$. What does this imply for gradient descent step sizes?

**Solution.**

Here $f'(x) = 4x^3$ and $f''(x) = 12x^2$. For a twice-differentiable function, the best Lipschitz constant of $f'$ on an interval is the supremum of $\lvert f'' \rvert$ there.

On $\mathbb{R}$: $\sup_x 12x^2 = \infty$, so no finite $L$ works — concretely, $\dfrac{\lvert f'(x) - f'(0) \rvert}{\lvert x \rvert} = 4x^2$ grows without bound.

On $[-R, R]$: $\lvert f''(x) \rvert \le 12R^2$, and by the mean value theorem $\lvert f'(x) - f'(y) \rvert = \lvert f''(\zeta) \rvert\,\lvert x - y \rvert \le 12R^2\lvert x - y \rvert$, attained near the endpoints. So $L_R = 12R^2$.

Consequence: a *constant* step size that is safe near the start (large $\lvert x \rvert$, huge curvature) is tiny; conversely a step size tuned near the minimum will diverge if the iterate ever wanders out to large $\lvert x \rvert$. Step sizes must adapt to the local curvature radius — the theoretical root of learning-rate schedules, clipping, and line searches.

$$
\boxed{f(x) = x^4: \text{ no global } L; \quad L_R = 12R^2 \text{ on } [-R, R]}
$$

*Key takeaway:* Neural losses behave like this: smooth on bounded regions, with curvature exploding elsewhere. That is why the "descent lemma with a global $L$" story must be localized in deep learning practice.

## Level 3 — Challenge

### Problem L3.1: Two-sided quadratic trap

Prove that if $f$ is $L$-smooth, then for all $x, y$:

$$
\left\lvert f(y) - f(x) - \nabla f(x)^\top (y - x) \right\rvert \le \frac{L}{2}\lVert y - x \rVert_2^2.
$$

**Solution.**

Let $h = y - x$ and $g(t) = f(x + th)$, so $g'(t) = \nabla f(x + th)^\top h$. By the fundamental theorem of calculus,

$$
f(y) - f(x) - \nabla f(x)^\top h = \int_0^1 \left( \nabla f(x + th) - \nabla f(x) \right)^\top h \, dt.
$$

Bound the absolute value of the integral by the integral of the absolute value, then apply Cauchy–Schwarz and $L$-Lipschitzness of the gradient:

$$
\left\lvert \left( \nabla f(x+th) - \nabla f(x) \right)^\top h \right\rvert \le \lVert \nabla f(x+th) - \nabla f(x) \rVert_2\, \lVert h \rVert_2 \le L t \lVert h \rVert_2^2.
$$

Integrating $\int_0^1 Lt\lVert h \rVert_2^2\, dt = \frac{L}{2}\lVert h \rVert_2^2$ yields the two-sided bound. $\blacksquare$

$$
\boxed{\left\lvert f(y) - f(x) - \nabla f(x)^\top(y-x) \right\rvert \le \frac{L}{2}\lVert y - x \rVert_2^2}
$$

*Key takeaway:* $L$-smoothness controls the linear model's error in *both* directions: the descent lemma (upper) powers convergence proofs, while the lower bound limits how much progress any single step can claim.

### Problem L3.2: The flat function that defeats Taylor series

Let $f(x) = e^{-1/x^2}$ for $x \neq 0$ and $f(0) = 0$. Prove that $f^{(n)}(0) = 0$ for every $n \ge 0$, so every Taylor polynomial of $f$ at $0$ is identically zero even though $f(x) \gt 0$ for $x \neq 0$.

**Solution.**

**Step 1 (structure of derivatives).** By induction: for $x \neq 0$, every derivative has the form $f^{(n)}(x) = P_n(1/x)\, e^{-1/x^2}$ for some polynomial $P_n$. Base case: $P_0 = 1$. Inductive step: differentiating $P_n(1/x)e^{-1/x^2}$ gives

$$
\left[ -\frac{1}{x^2}P_n'(1/x) + \frac{2}{x^3}P_n(1/x) \right] e^{-1/x^2},
$$

again a polynomial in $1/x$ times $e^{-1/x^2}$.

**Step 2 (the key limit).** For any integer $m \ge 0$, substituting $u = 1/x$ ($u \to \pm\infty$ as $x \to 0^{\pm}$):

$$
\lim_{x \to 0} \frac{e^{-1/x^2}}{x^m} = \lim_{u \to \pm\infty} u^m e^{-u^2} = 0,
$$

because $e^{u^2}$ grows faster than any power (e.g. $e^{u^2} \ge u^{2k}/k!$ for every $k$). Hence $P(1/x)e^{-1/x^2} \to 0$ as $x \to 0$ for every polynomial $P$.

**Step 3 (induction on the derivative at 0).** Suppose $f^{(n)}(0) = 0$ (true for $n = 0$). Then

$$
f^{(n+1)}(0) = \lim_{x \to 0} \frac{f^{(n)}(x) - 0}{x} = \lim_{x \to 0} \frac{1}{x}P_n(1/x)e^{-1/x^2} = 0
$$

by Step 2, since $\frac{1}{x}P_n(1/x)$ is again a polynomial in $1/x$. By induction all $f^{(n)}(0) = 0$. $\blacksquare$

$$
\boxed{f^{(n)}(0) = 0 \ \forall n \implies T_k \equiv 0 \text{ for every } k, \text{ yet } f \not\equiv 0}
$$

*Key takeaway:* Smoothness does not imply analyticity: derivative information at a single point can miss the function entirely. Finite-order Taylor *with remainder bounds* — not infinite series — is the honest foundation for optimization. (This same function builds the bump functions and smooth cutoffs used throughout analysis.)

### Problem L3.3: Newton's method can diverge

Apply Newton's root-finding iteration to $f(x) = x^{1/3}$ (root at $x = 0$). Derive the iteration map and show that it *diverges* from every starting point $x_0 \neq 0$. Why does the quadratic-convergence theorem not apply?

**Solution.**

With $f(x) = x^{1/3}$ and $f'(x) = \frac{1}{3}x^{-2/3}$:

$$
x_{t+1} = x_t - \frac{f(x_t)}{f'(x_t)} = x_t - \frac{x_t^{1/3}}{\frac{1}{3}x_t^{-2/3}} = x_t - 3x_t = -2x_t.
$$

So $\lvert x_{t+1} \rvert = 2\lvert x_t \rvert$: the iterate flips sign and *doubles* in magnitude every step, diverging geometrically from every $x_0 \neq 0$.

The convergence theorem requires $f$ to be smooth with $f'$ nonzero and bounded behavior near the root. Here $f'(x) = \frac{1}{3}x^{-2/3} \to \infty$ as $x \to 0$: the tangent lines at points near the root are nearly vertical, so each tangent line crosses zero far on the *other side* of the root — overshooting by a factor of 2 forever.

$$
\boxed{x_{t+1} = -2x_t \implies \lvert x_t \rvert = 2^t \lvert x_0 \rvert \to \infty}
$$

*Key takeaway:* Newton's power comes entirely from the local model being trustworthy; when curvature blows up (unbounded second derivative of the objective, singular Hessians in optimization), the raw Newton step is worse than useless — hence damping, trust regions, and line searches.

### Problem L3.4: Cubic error bound for the quadratic model

Suppose $\nabla^2 f$ is $M$-Lipschitz: $\lVert \nabla^2 f(u) - \nabla^2 f(v) \rVert_{\mathrm{op}} \le M\lVert u - v \rVert_2$. Prove

$$
\left\lvert f(x+h) - f(x) - \nabla f(x)^\top h - \frac{1}{2}h^\top \nabla^2 f(x)\, h \right\rvert \le \frac{M}{6}\lVert h \rVert_2^3.
$$

**Solution.**

Let $g(t) = f(x + th)$, so $g''(t) = h^\top \nabla^2 f(x + th) h$. Taylor with *integral* remainder for $g$ on $[0,1]$ (obtained by integrating by parts twice) gives

$$
g(1) = g(0) + g'(0) + \frac{1}{2}g''(0) + \int_0^1 \frac{(1-t)^2}{2}\, g'''(t)\, dt
$$

when $g \in C^3$; to avoid assuming third derivatives, use instead the first-order integral form of the *second* derivative:

$$
f(x+h) - f(x) - \nabla f(x)^\top h - \frac{1}{2}h^\top \nabla^2 f(x) h = \int_0^1 (1-t)\, h^\top\left[ \nabla^2 f(x+th) - \nabla^2 f(x) \right] h \, dt,
$$

which follows from $f(x+h) - f(x) - \nabla f(x)^\top h = \int_0^1 (1-t)\, h^\top \nabla^2 f(x+th)\, h\, dt$ (integrate $g'' \cdot (1-t)$ by parts) after subtracting $\int_0^1 (1-t)\,dt \cdot h^\top\nabla^2 f(x) h = \frac{1}{2}h^\top \nabla^2 f(x)h$.

Now bound the integrand using the operator-norm inequality and Hessian Lipschitzness:

$$
\left\lvert h^\top\left[ \nabla^2 f(x+th) - \nabla^2 f(x) \right] h \right\rvert \le \lVert \nabla^2 f(x+th) - \nabla^2 f(x) \rVert_{\mathrm{op}} \lVert h \rVert_2^2 \le Mt\lVert h \rVert_2^3.
$$

Therefore the error is at most $\int_0^1 (1-t)\, Mt \lVert h \rVert_2^3\, dt = M\lVert h \rVert_2^3 \int_0^1 t(1-t)\,dt = \frac{M}{6}\lVert h \rVert_2^3$. $\blacksquare$

$$
\boxed{\left\lvert f(x+h) - m_x^{(2)}(h) \right\rvert \le \frac{M}{6}\lVert h \rVert_2^3}
$$

*Key takeaway:* This is the multivariate engine behind Newton's quadratic convergence and behind cubic-regularized Newton (Nesterov–Polyak), which *minimizes* the model plus $\frac{M}{6}\lVert h \rVert^3$ to get global guarantees on nonconvex problems.
