## Level 0 — Concept Check

### Problem L0.1: Partials exist, differentiability fails?

True or false: if all partial derivatives of $f: \mathbb{R}^2 \to \mathbb{R}$ exist at a point, then $f$ is differentiable there. Justify with a counterexample if false.

**Solution.**

False. Consider $f(x,y) = \dfrac{xy}{x^2 + y^2}$ for $(x,y) \neq (0,0)$ and $f(0,0) = 0$.

Along the axes, $f(h, 0) = 0$ and $f(0, h) = 0$, so both partials at the origin exist and equal $0$:

$$
\frac{\partial f}{\partial x}(0,0) = \lim_{h \to 0} \frac{f(h,0) - f(0,0)}{h} = 0, \qquad \frac{\partial f}{\partial y}(0,0) = 0.
$$

But along the diagonal $y = x$ we get $f(x,x) = \dfrac{x^2}{2x^2} = \dfrac{1}{2}$ for every $x \neq 0$, so $f$ is not even continuous at the origin, hence not differentiable there.

$$
\boxed{\text{False: existence of partials does not imply differentiability (nor even continuity).}}
$$

*Key takeaway:* Partial derivatives only probe axis directions; differentiability demands a single linear model that works in *every* direction.

### Problem L0.2: Fastest decrease

At a point $x$ with $\nabla f(x) \neq 0$, in which unit direction does $f$ decrease fastest, and what is the instantaneous rate of change in that direction?

**Solution.**

By Cauchy–Schwarz, for unit $v$, $D_v f(x) = \nabla f(x)^\top v \ge -\lVert \nabla f(x) \rVert_2$, with equality exactly when $v$ points opposite to the gradient. Hence the direction of fastest decrease is

$$
v = -\frac{\nabla f(x)}{\lVert \nabla f(x) \rVert_2},
$$

and the rate there is $D_v f(x) = -\lVert \nabla f(x) \rVert_2$.

$$
\boxed{v = -\nabla f(x)/\lVert \nabla f(x) \rVert_2, \qquad D_v f(x) = -\lVert \nabla f(x) \rVert_2}
$$

*Key takeaway:* Gradient descent's update direction is not a heuristic — it is the unique first-order-optimal descent direction under the Euclidean norm.

### Problem L0.3: Jacobian shapes

A neural network layer computes $g: \mathbb{R}^3 \to \mathbb{R}^2$. What is the shape of its Jacobian $J_g$? If a scalar loss $\ell: \mathbb{R}^2 \to \mathbb{R}$ follows, what are the shapes in the backprop product $\nabla_x(\ell \circ g) = J_g^\top \nabla \ell$?

**Solution.**

The Jacobian has one row per output and one column per input: $J_g \in \mathbb{R}^{2 \times 3}$.

The loss gradient $\nabla \ell \in \mathbb{R}^{2}$ (one entry per layer output). The backprop product is

$$
\underbrace{J_g^\top}_{3 \times 2} \underbrace{\nabla \ell}_{2 \times 1} \in \mathbb{R}^{3 \times 1},
$$

one sensitivity per input — exactly the shape of $\nabla_x(\ell \circ g)$.

$$
\boxed{J_g \in \mathbb{R}^{2 \times 3}, \qquad \nabla_x(\ell \circ g) = J_g^\top \nabla \ell \in \mathbb{R}^{3}}
$$

*Key takeaway:* Shape bookkeeping ("outputs $\times$ inputs") catches most backprop implementation bugs before any math does.

### Problem L0.4: Subdifferential of the absolute value

Compute the subdifferential $\partial f(0)$ for $f(x) = \lvert x \rvert$, and explain why frameworks may set $\text{ReLU}'(0) = 0$.

**Solution.**

A scalar $s$ is a subgradient of $f$ at $0$ if $\lvert z \rvert \ge f(0) + s(z - 0) = sz$ for all $z$. For $z \gt 0$ this requires $s \le 1$; for $z \lt 0$ it requires $s \ge -1$. Both hold precisely when $s \in [-1, 1]$, so

$$
\boxed{\partial f(0) = [-1,\, 1]}
$$

Similarly $\mathrm{ReLU}(x) = \max(0, x)$ has $\partial\,\mathrm{ReLU}(0) = [0, 1]$; any element (including $0$) is a valid subgradient, so the convention $\mathrm{ReLU}'(0) = 0$ is a legitimate selection from the subdifferential.

*Key takeaway:* At a kink there is no single tangent line, but a whole interval of supporting lines — any one of them drives valid subgradient descent.

## Level 1 — Foundation

### Problem L1.1: Gradient and Hessian of a quadratic-plus-coupling

Let $f(x, y) = x^2 + 3y^2 + 2xy - 4x$. Compute $\nabla f(1, 2)$ and the Hessian $\nabla^2 f$. Is the Hessian positive definite?

**Solution.**

Partial derivatives:

$$
\frac{\partial f}{\partial x} = 2x + 2y - 4, \qquad \frac{\partial f}{\partial y} = 6y + 2x.
$$

At $(1, 2)$: $\frac{\partial f}{\partial x} = 2 + 4 - 4 = 2$ and $\frac{\partial f}{\partial y} = 12 + 2 = 14$.

The Hessian is constant (the function is quadratic):

$$
\nabla^2 f = \begin{bmatrix} 2 & 2 \\ 2 & 6 \end{bmatrix}.
$$

Positive definiteness via leading principal minors (Sylvester): $2 \gt 0$ and $\det = 12 - 4 = 8 \gt 0$, so yes.

$$
\boxed{\nabla f(1,2) = (2,\ 14)^\top, \qquad \nabla^2 f = \begin{bmatrix} 2 & 2 \\ 2 & 6 \end{bmatrix} \succ 0}
$$

*Key takeaway:* For quadratics the Hessian is constant; the cross term $2xy$ shows up as symmetric off-diagonal entries.

### Problem L1.2: Directional derivative

For $f(x,y) = x e^{y}$, compute the directional derivative at $(2, 0)$ in the direction of $u = (3, 4)$.

**Solution.**

First normalize: $\lVert u \rVert_2 = \sqrt{9 + 16} = 5$, so $v = (3/5,\ 4/5)$.

Gradient: $\nabla f = (e^{y},\ x e^{y})$, so $\nabla f(2, 0) = (1,\ 2)$.

Directional derivative:

$$
D_v f(2,0) = \nabla f(2,0)^\top v = 1 \cdot \frac{3}{5} + 2 \cdot \frac{4}{5} = \frac{11}{5}.
$$

$$
\boxed{D_v f(2,0) = \frac{11}{5}}
$$

*Key takeaway:* Always normalize the direction first — $D_v f$ is defined for unit vectors, and forgetting the normalization silently rescales the answer.

### Problem L1.3: The two workhorse identities

Derive from first principles: (a) $\nabla_x (a^\top x) = a$; (b) $\nabla_x (x^\top A x) = (A + A^\top) x$, which reduces to $2Ax$ for symmetric $A$.

**Solution.**

**(a)** Let $f(x) = a^\top x$. Then $f(x + h) = a^\top x + a^\top h$, exactly linear with zero remainder. Matching $f(x+h) = f(x) + g^\top h + o(\lVert h \rVert)$ gives $g = a$.

**(b)** Let $f(x) = x^\top A x$. Expand:

$$
f(x+h) = (x+h)^\top A (x+h) = x^\top A x + x^\top A h + h^\top A x + h^\top A h.
$$

Since $h^\top A x$ is a scalar, $h^\top A x = (h^\top A x)^\top = x^\top A^\top h$. Therefore

$$
f(x+h) = f(x) + \left( (A + A^\top) x \right)^\top h + h^\top A h,
$$

and $\lvert h^\top A h \rvert \le \lVert A \rVert_{\mathrm{op}} \lVert h \rVert_2^2 = o(\lVert h \rVert)$. Hence the gradient is $(A + A^\top)x$; symmetry $A = A^\top$ gives $2Ax$.

$$
\boxed{\nabla_x (a^\top x) = a, \qquad \nabla_x (x^\top A x) = (A + A^\top)x}
$$

*Key takeaway:* Nearly every ML gradient (least squares, ridge, Gaussians, attention scores) reduces to these two identities plus the chain rule.

### Problem L1.4: Chain rule with explicit Jacobians

Let $g(x_1, x_2) = (x_1 x_2,\ x_1 + x_2)$ and $f(u_1, u_2) = u_1^2 + u_2$. Compute $\nabla_x (f \circ g)$ at $x = (1, 2)$ both by the Jacobian chain rule and by direct substitution.

**Solution.**

**Chain rule route.** $J_g = \begin{bmatrix} x_2 & x_1 \\ 1 & 1 \end{bmatrix}$, so $J_g(1,2) = \begin{bmatrix} 2 & 1 \\ 1 & 1 \end{bmatrix}$. Also $g(1,2) = (2, 3)$ and $\nabla f(u) = (2u_1,\ 1)$, so $\nabla f(2,3) = (4,\ 1)$.

$$
\nabla_x (f \circ g)(1,2) = J_g^\top \nabla f = \begin{bmatrix} 2 & 1 \\ 1 & 1 \end{bmatrix} \begin{bmatrix} 4 \\ 1 \end{bmatrix} = \begin{bmatrix} 9 \\ 5 \end{bmatrix}.
$$

**Direct route.** $(f \circ g)(x) = x_1^2 x_2^2 + x_1 + x_2$. Partials: $2x_1 x_2^2 + 1$ and $2x_1^2 x_2 + 1$; at $(1,2)$: $8 + 1 = 9$ and $4 + 1 = 5$. The routes agree.

$$
\boxed{\nabla_x (f \circ g)(1,2) = (9,\ 5)^\top}
$$

*Key takeaway:* $J_g^\top \nabla f$ is a complete recipe — this two-line computation is a one-layer backprop by hand.

### Problem L1.5: Gradient normal to a contour

Show that for $f(x,y) = x^2 + y^2$, the gradient at any point of the circle $x^2 + y^2 = r^2$ is normal to the circle.

**Solution.**

Parametrize the level set by $r(t) = (r\cos t,\ r\sin t)$, whose tangent is $r'(t) = (-r\sin t,\ r\cos t)$.

The gradient is $\nabla f = (2x,\ 2y)$, so on the curve $\nabla f(r(t)) = (2r\cos t,\ 2r\sin t)$.

Inner product:

$$
\nabla f(r(t))^\top r'(t) = -2r^2 \cos t \sin t + 2r^2 \sin t \cos t = 0.
$$

The gradient is orthogonal to the tangent at every point, i.e. normal to the circle; indeed it points radially outward, the direction of fastest growth of $x^2 + y^2$.

$$
\boxed{\nabla f \perp \text{level set} \quad \text{(gradient is radial, tangent is angular)}}
$$

*Key takeaway:* This is the picture behind every contour plot of gradient descent: steps cross contour lines at right angles.

### Problem L1.6: Least-squares gradient, concretely

For $X = \begin{bmatrix} 1 & 0 \\ 1 & 1 \\ 1 & 2 \end{bmatrix}$, $y = (1,\ 2,\ 2)^\top$, and $L(w) = \frac{1}{3}\lVert Xw - y \rVert_2^2$, compute $\nabla L(w)$ at $w = (0, 0)^\top$ and the Hessian $\nabla^2 L$.

**Solution.**

From the general formula $\nabla L(w) = \frac{2}{3} X^\top (Xw - y)$. At $w = 0$, the residual is $Xw - y = -y$, so

$$
\nabla L(0) = -\frac{2}{3} X^\top y = -\frac{2}{3} \begin{bmatrix} 1 & 1 & 1 \\ 0 & 1 & 2 \end{bmatrix} \begin{bmatrix} 1 \\ 2 \\ 2 \end{bmatrix} = -\frac{2}{3}\begin{bmatrix} 5 \\ 6 \end{bmatrix} = \begin{bmatrix} -10/3 \\ -4 \end{bmatrix}.
$$

The Hessian is constant:

$$
\nabla^2 L = \frac{2}{3} X^\top X = \frac{2}{3}\begin{bmatrix} 3 & 3 \\ 3 & 5 \end{bmatrix} = \begin{bmatrix} 2 & 2 \\ 2 & 10/3 \end{bmatrix}.
$$

Since $X$ has full column rank, $\nabla^2 L \succ 0$: the loss is strictly convex with a unique minimizer.

$$
\boxed{\nabla L(0) = \left( -\tfrac{10}{3},\ -4 \right)^\top, \qquad \nabla^2 L = \tfrac{2}{3} X^\top X \succ 0}
$$

*Key takeaway:* The gradient at $w = 0$ is just $-\frac{2}{n} X^\top y$ — correlation between features and targets is the first descent signal a linear model sees.

## Level 2 — Applications in AI/ML

### Problem L2.1: Logistic regression gradient

For the binary cross-entropy loss with sigmoid $\sigma(z) = 1/(1 + e^{-z})$,

$$
L(w) = -\frac{1}{n}\sum_{i=1}^{n}\left[ y_i \log \sigma(x_i^\top w) + (1 - y_i)\log\left( 1 - \sigma(x_i^\top w) \right) \right],
$$

derive $\nabla L(w) = \frac{1}{n} X^\top (\sigma(Xw) - y)$, where $\sigma$ applies elementwise.

**Solution.**

Fix one sample and write $z = x^\top w$, $p = \sigma(z)$, $\ell = -\left[ y\log p + (1-y)\log(1-p) \right]$. Chain rule in three links: $\ell \to p \to z \to w$.

Step 1: $\dfrac{\partial \ell}{\partial p} = -\dfrac{y}{p} + \dfrac{1-y}{1-p} = \dfrac{p - y}{p(1-p)}$.

Step 2: $\dfrac{dp}{dz} = \sigma'(z) = p(1-p)$ (Problem L2.3).

Step 3: $\nabla_w z = x$.

Multiplying the three factors, the $p(1-p)$ terms cancel:

$$
\nabla_w \ell = \frac{p - y}{p(1-p)} \cdot p(1-p) \cdot x = (p - y)\, x.
$$

Averaging over samples and stacking rows $x_i^\top$ into $X$:

$$
\boxed{\nabla L(w) = \frac{1}{n} X^\top\left( \sigma(Xw) - y \right)}
$$

*Key takeaway:* "Prediction minus target, weighted by inputs" — the same gradient shape appears for linear regression, logistic regression, and softmax, because each pairs a matching loss with its canonical link.

### Problem L2.2: Forward vs reverse mode, counted

A network computes $L = \ell(f_3(f_2(f_1(x))))$ with layer Jacobians $J_1 \in \mathbb{R}^{1000 \times 1000}$, $J_2 \in \mathbb{R}^{1000 \times 1000}$, $J_3 \in \mathbb{R}^{10 \times 1000}$ and loss gradient $\nabla \ell \in \mathbb{R}^{10}$. Count the scalar multiplications to form $\nabla_x L$ (a) right-to-left (reverse mode) and (b) left-to-right after forming $J_3 J_2 J_1$ (naive forward accumulation of full matrices).

**Solution.**

The gradient is $\nabla_x L = J_1^\top J_2^\top J_3^\top \nabla\ell$.

**(a) Reverse mode: matrix–vector only.**

$u_1 = J_3^\top \nabla\ell$: $1000 \times 10$ times $10 \times 1$ costs $10^4$.

$u_2 = J_2^\top u_1$: $1000 \times 1000$ times vector costs $10^6$.

$u_3 = J_1^\top u_2$: another $10^6$.

Total $\approx 2.01 \times 10^6$ multiplications.

**(b) Full-matrix accumulation.** $J_3 J_2$ costs $10 \cdot 1000 \cdot 1000 = 10^7$; then $(J_3 J_2) J_1$ costs another $10^7$; final matrix–vector $10^4$. Total $\approx 2.0 \times 10^7$ — an order of magnitude worse here, and the gap grows with depth and width (for a scalar loss and $d$ inputs, matrix accumulation is worse by a factor of about $d$).

$$
\boxed{\text{Reverse mode: } \approx 2\times 10^6 \text{ mults} \quad \text{vs} \quad \text{matrix accumulation: } \approx 2\times 10^7}
$$

*Key takeaway:* Backprop's efficiency is pure associativity: keep everything as vector–Jacobian products and never materialize a full Jacobian.

### Problem L2.3: Sigmoid derivative and vanishing gradients

Show $\sigma'(z) = \sigma(z)(1 - \sigma(z))$ and that $\sigma'(z) \le \frac{1}{4}$ everywhere. What does this imply for a deep chain of sigmoid layers?

**Solution.**

With $\sigma(z) = (1 + e^{-z})^{-1}$, differentiate by the chain rule:

$$
\sigma'(z) = -(1 + e^{-z})^{-2} \cdot (-e^{-z}) = \frac{e^{-z}}{(1 + e^{-z})^2} = \frac{1}{1+e^{-z}} \cdot \frac{e^{-z}}{1+e^{-z}} = \sigma(z)\left( 1 - \sigma(z) \right).
$$

Let $p = \sigma(z) \in (0,1)$. The map $p \mapsto p(1-p)$ is a downward parabola maximized at $p = \frac{1}{2}$ with value $\frac{1}{4}$. Hence $\sigma'(z) \le \frac{1}{4}$, with equality only at $z = 0$.

In a depth-$k$ chain of sigmoid activations, backprop multiplies $k$ such factors, so the gradient magnitude carries a factor at most $4^{-k}$ from the activations alone — exponentially small in depth.

$$
\boxed{\sigma'(z) = \sigma(z)(1-\sigma(z)) \le \tfrac{1}{4}}
$$

*Key takeaway:* The vanishing-gradient problem is a chain-rule inequality, and it is why ReLU (derivative $1$ on its active half) replaced sigmoid in deep architectures.

### Problem L2.4: Softmax cross-entropy gradient

For logits $z \in \mathbb{R}^K$, softmax $p_k = e^{z_k}/\sum_{j} e^{z_j}$, and one-hot target $y$, show that the cross-entropy loss $\ell(z) = -\sum_k y_k \log p_k$ has gradient $\nabla_z \ell = p - y$.

**Solution.**

Let $S = \sum_j e^{z_j}$ and let $c$ be the true class ($y_c = 1$). Then

$$
\ell(z) = -\log p_c = -z_c + \log S.
$$

Differentiate with respect to $z_k$:

$$
\frac{\partial \ell}{\partial z_k} = -\mathbf{1}[k = c] + \frac{1}{S} \cdot e^{z_k} = p_k - y_k.
$$

Stacking over $k$ gives $\nabla_z \ell = p - y$. (The same holds for soft targets by linearity of $-\sum_k y_k \log p_k$ in $y$.)

$$
\boxed{\nabla_z \ell = p - y}
$$

*Key takeaway:* Softmax and cross-entropy are designed for each other: their composition collapses to the cleanest possible error signal, numerically stable and cheap.

### Problem L2.5: Gradient checking in action

For $f(x) = x^3$ at $x = 1$, compute the forward-difference and central-difference estimates of $f'(1) = 3$ with $\epsilon = 10^{-2}$, and confirm the predicted error orders $O(\epsilon)$ and $O(\epsilon^2)$.

**Solution.**

Exact expansions with $\epsilon = 10^{-2}$:

Forward: $\dfrac{(1+\epsilon)^3 - 1}{\epsilon} = \dfrac{3\epsilon + 3\epsilon^2 + \epsilon^3}{\epsilon} = 3 + 3\epsilon + \epsilon^2 = 3.0301$.

Central: $\dfrac{(1+\epsilon)^3 - (1-\epsilon)^3}{2\epsilon} = \dfrac{6\epsilon + 2\epsilon^3}{2\epsilon} = 3 + \epsilon^2 = 3.0001$.

Errors: forward $3.01 \times 10^{-2} \approx 3\epsilon$ (the theory predicts leading error $\frac{f''(1)}{2}\epsilon = 3\epsilon$); central $10^{-4} = \epsilon^2$ (theory: $\frac{f'''(1)}{6}\epsilon^2 = \epsilon^2$). Both match exactly.

$$
\boxed{\text{forward error} = 3\epsilon + \epsilon^2, \qquad \text{central error} = \epsilon^2}
$$

*Key takeaway:* Central differences buy an order of accuracy for one extra function evaluation — always the right choice for gradient checking.

### Problem L2.6: A one-dimensional Lasso step

Minimize $f(w) = \frac{1}{2}(w - 1)^2 + \lambda \lvert w \rvert$ for $\lambda = \frac{1}{2}$ using subdifferential calculus.

**Solution.**

$f$ is convex (sum of convex functions), so $w^\star$ is a global minimizer if and only if $0 \in \partial f(w^\star)$, where

$$
\partial f(w) = (w - 1) + \lambda\, \partial \lvert w \rvert = \begin{cases} \{w - 1 + \lambda\} & w \gt 0 \\ [\,-1 - \lambda,\ -1 + \lambda\,] & w = 0 \\ \{w - 1 - \lambda\} & w \lt 0 \end{cases}
$$

Try $w \gt 0$: $w - 1 + \frac{1}{2} = 0$ gives $w = \frac{1}{2} \gt 0$ — consistent. (Check the other branches: at $w = 0$ the interval is $[-\frac{3}{2}, -\frac{1}{2}]$, which misses $0$; the $w \lt 0$ branch gives $w = \frac{3}{2}$, inconsistent.)

$$
\boxed{w^\star = \tfrac{1}{2} = \operatorname{sign}(1)\max\left( \lvert 1 \rvert - \lambda,\ 0 \right) \quad \text{(soft-thresholding)}}
$$

*Key takeaway:* The $L^1$ penalty shrinks the unregularized answer toward zero by exactly $\lambda$ and snaps it to zero when small — the mechanism behind Lasso sparsity, derived purely from subgradients.

## Level 3 — Challenge

### Problem L3.1: Gradient of the Euclidean norm and its subdifferential at the origin

Let $f(x) = \lVert x \rVert_2$ on $\mathbb{R}^d$. (a) Show $\nabla f(x) = x/\lVert x \rVert_2$ for $x \neq 0$. (b) Show $\partial f(0) = \{ s : \lVert s \rVert_2 \le 1 \}$, the closed unit ball.

**Solution.**

**(a)** Write $f(x) = (x^\top x)^{1/2}$. By the chain rule with outer map $t \mapsto t^{1/2}$ and inner map $x \mapsto x^\top x$ (gradient $2x$ by Problem L1.3):

$$
\nabla f(x) = \frac{1}{2}(x^\top x)^{-1/2} \cdot 2x = \frac{x}{\lVert x \rVert_2}, \qquad x \neq 0.
$$

It is a unit vector: the norm grows at unit rate in its own radial direction.

**(b)** $s \in \partial f(0)$ means $\lVert z \rVert_2 \ge s^\top z$ for all $z$.

(*If* $\lVert s \rVert_2 \le 1$): Cauchy–Schwarz gives $s^\top z \le \lVert s \rVert_2 \lVert z \rVert_2 \le \lVert z \rVert_2$. So $s$ is a subgradient.

(*Only if*): suppose $\lVert s \rVert_2 \gt 1$. Choose $z = s$: then $s^\top z = \lVert s \rVert_2^2 \gt \lVert s \rVert_2 = \lVert z \rVert_2$, violating the inequality. Hence no such $s$ is a subgradient.

$$
\boxed{\nabla f(x) = \frac{x}{\lVert x \rVert_2} \ (x \neq 0), \qquad \partial f(0) = \left\{ s : \lVert s \rVert_2 \le 1 \right\}}
$$

*Key takeaway:* Group-lasso and weight-decay-style penalties inherit their geometry from this computation: away from zero the norm has a clean gradient; at zero an entire ball of supporting slopes creates the "trap" that produces exactly-zero groups.

### Problem L3.2: A function differentiable at exactly one point

Show that $f(x) = x^2$ for rational $x$ and $f(x) = 0$ for irrational $x$ is differentiable at $x = 0$ and discontinuous at every $x \neq 0$.

**Solution.**

**Differentiability at 0.** For any $h \neq 0$,

$$
\left\lvert \frac{f(h) - f(0)}{h} \right\rvert = \frac{\lvert f(h) \rvert}{\lvert h \rvert} \le \frac{h^2}{\lvert h \rvert} = \lvert h \rvert \xrightarrow[h \to 0]{} 0,
$$

since $\lvert f(h) \rvert$ is either $h^2$ (rational $h$) or $0$ (irrational $h$). Hence $f'(0)$ exists and equals $0$.

**Discontinuity elsewhere.** Fix $x \neq 0$. Rationals and irrationals are both dense, so there are sequences $q_n \to x$ (rational) and $r_n \to x$ (irrational) with $f(q_n) = q_n^2 \to x^2 \neq 0$ and $f(r_n) = 0 \to 0$. Two subsequential limits disagree, so $\lim_{t \to x} f(t)$ does not exist and $f$ is discontinuous at $x$.

$$
\boxed{f'(0) = 0 \text{ exists, yet } f \text{ is discontinuous at every } x \neq 0}
$$

*Key takeaway:* Differentiability is a strictly *local* squeeze: a quadratic envelope at one point is enough, no matter how pathological the function is elsewhere.

### Problem L3.3: The log-det gradient

For symmetric positive definite $A$ and symmetric $B$, prove

$$
\frac{d}{dt}\Big\vert_{t=0} \log\det(A + tB) = \operatorname{tr}(A^{-1} B),
$$

and conclude $\nabla_A \log\det A = A^{-1}$. Where does this appear in ML?

**Solution.**

Factor out $A$: $\det(A + tB) = \det(A)\det(I + t A^{-1} B)$, so

$$
\log\det(A + tB) = \log\det A + \log\det(I + tC), \qquad C = A^{-1}B.
$$

Let $\mu_1, \dots, \mu_d$ be the eigenvalues of $C$ (real, since $C$ is similar to the symmetric matrix $A^{-1/2} B A^{-1/2}$). Then $\det(I + tC) = \prod_i (1 + t\mu_i)$ and

$$
\log\det(I + tC) = \sum_{i=1}^{d} \log(1 + t\mu_i) = t\sum_{i=1}^{d} \mu_i + O(t^2) = t\operatorname{tr}(C) + O(t^2).
$$

Differentiating at $t = 0$ gives $\operatorname{tr}(C) = \operatorname{tr}(A^{-1}B)$. Since the directional derivative in direction $B$ is $\operatorname{tr}(A^{-1}B) = \langle A^{-1}, B \rangle$ (Frobenius inner product) for every symmetric $B$, the gradient is $A^{-1}$.

$$
\boxed{\nabla_A \log\det A = A^{-1}}
$$

*ML appearances:* the Gaussian log-likelihood term $-\frac{1}{2}\log\det\Sigma$ in covariance estimation and Gaussian processes; the log-determinant Jacobian in normalizing flows; log-det barriers in interior-point methods (Boyd & Vandenberghe, Ch. 11).

*Key takeaway:* Matrix-variable calculus works exactly like vector calculus once the right inner product ($\operatorname{tr}(X^\top Y)$) is chosen — the gradient is whatever pairs with the perturbation.

### Problem L3.4: Hessian-vector products without the Hessian

Show that $g(x) = \nabla f(x)^\top v$ (for fixed $v$) satisfies $\nabla g(x) = \nabla^2 f(x)\, v$ for $f \in C^2$, and explain why this makes Hessian-vector products cheap while full Hessians stay expensive.

**Solution.**

Write $g(x) = \sum_{j} \frac{\partial f}{\partial x_j}(x)\, v_j$. Differentiating with respect to $x_i$,

$$
\frac{\partial g}{\partial x_i}(x) = \sum_{j} \frac{\partial^2 f}{\partial x_i \partial x_j}(x)\, v_j = \left[ \nabla^2 f(x)\, v \right]_i,
$$

using symmetry of second partials (Clairaut, valid for $f \in C^2$). Hence $\nabla g = \nabla^2 f\, v$.

**Cost argument.** $g$ is a scalar-valued function computable by one forward pass plus one reverse pass (to get $\nabla f$), then one inner product. Its gradient — which *is* the Hessian-vector product — therefore costs one more reverse pass through that computation: a constant multiple of the cost of evaluating $f$, independent of $d$. A full Hessian, by contrast, needs $d$ independent products $\nabla^2 f\, e_1, \dots, \nabla^2 f\, e_d$: cost $O(d)$ passes and $O(d^2)$ memory, infeasible for modern $d$.

$$
\boxed{\nabla_x\left( \nabla f(x)^\top v \right) = \nabla^2 f(x)\, v \quad \text{at } O(1) \text{ extra passes}}
$$

*Key takeaway:* Curvature information in ML is accessed through matrix-free products (Pearlmutter's trick) — powering conjugate-gradient Newton steps, eigenvalue probes of the loss landscape, and influence-function computations.
