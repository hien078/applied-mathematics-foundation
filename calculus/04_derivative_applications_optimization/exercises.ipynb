### Problem 0.1 (Rolle's Theorem Hypotheses Verification)
**Source:** Spivak, *Calculus*, Chapter 14  
**Problem Statement:** Consider the function $f(x) = 1 - \lvert x \rvert$ defined on the closed interval $[-1, 1]$. We observe that $f(-1) = 0$ and $f(1) = 0$. Is there any point $c \in (-1, 1)$ such that $f'(c) = 0$? Does this contradict Rolle's Theorem? Explain why or why not from first principles.

#### First-Principles Intuition
Rolle's Theorem asserts that if a function starts and ends at the same height, it must turn around somewhere, producing a horizontal tangent. However, this turning point must be smooth (differentiable). If the turning point is a sharp corner (v-shape), the tangent line does not exist.

#### Step-by-Step Solution
1. **Check Endpoint Values:**  

$$
   f(-1) = 1 - \lvert -1 \rvert = 1 - 1 = 0
$$

$$
   f(1) = 1 - \lvert 1 \rvert = 1 - 1 = 0
$$

   The endpoint equality condition $f(-1) = f(1)$ holds.

2. **Calculate Derivative:**  
   For $x \in (0, 1)$, $f(x) = 1 - x$, so $f'(x) = -1$.  
   For $x \in (-1, 0)$, $f(x) = 1 + x$, so $f'(x) = 1$.  
   At $x = 0$, evaluate left- and right-hand derivatives:

$$
   \lim_{h \to 0^+} \frac{f(0+h) - f(0)}{h} = \lim_{h \to 0^+} \frac{(1 - h) - 1}{h} = -1
$$

$$
   \lim_{h \to 0^-} \frac{f(0+h) - f(0)}{h} = \lim_{h \to 0^-} \frac{(1 + h) - 1}{h} = +1
$$

   Since the left-hand and right-hand limits differ, $f'(0)$ does not exist.

3. **Check Existence of Horizontal Tangent:**  
   $f'(x) = 1 \ne 0$ for $x \in (-1, 0)$ and $f'(x) = -1 \ne 0$ for $x \in (0, 1)$. Thus, there is **no** point $c \in (-1, 1)$ where $f'(c) = 0$.

4. **Reconciliation with Rolle's Theorem:**  
   Rolle's Theorem requires $f$ to be **differentiable on the entire open interval $(-1, 1)$**. Because $f$ is not differentiable at $x = 0 \in (-1, 1)$, hypothesis (2) of Rolle's Theorem is violated. Therefore, Rolle's Theorem does not apply and no contradiction exists.

$$
\boxed{\text{No point } c \text{ exists with } f'(c)=0 \text{. Rolle's Theorem is not violated because } f \text{ is not differentiable at } x=0.}
$$

#### Key Insight / Takeaway
Differentiability on the *entire* open interval is an indispensable hypothesis of Rolle's Theorem. A single point of non-differentiability can break the guarantee of a horizontal tangent.

### Problem 0.2 (Fermat's Theorem Hypotheses & Domain Boundaries)
**Source:** Apostol, *Calculus, Volume I*, Chapter 4  
**Problem Statement:** Consider the linear function $f(x) = x$ defined on the closed domain $[0, 1]$. $f$ attains its absolute maximum at $x = 1$ with $f(1) = 1$ and its absolute minimum at $x = 0$ with $f(0) = 0$. However, $f'(1) = 1 \ne 0$ and $f'(0) = 1 \ne 0$. Explain why this does not contradict Fermat's Theorem on local extrema.

#### First-Principles Intuition
Fermat's Theorem states that at a local extremum, the tangent line must be flat ($f'(c) = 0$). But this holds ONLY at **interior** points where the function can be approached from both sides. At boundary endpoints, the function stops, so the derivative can be non-zero while still achieving a maximum or minimum.

#### Step-by-Step Solution
1. **Fermat's Theorem Statement:**  
   Fermat's Theorem explicitly states: If $f$ has a local extremum at $c$, **and if $c$ is an interior point of the domain**, and if $f'(c)$ exists, then $f'(c) = 0$.

2. **Domain Analysis:**  
   The domain of $f$ is $[0, 1]$. The maximum occurs at $x = 1$ and the minimum at $x = 0$.  
   Neither $x = 0$ nor $x = 1$ is an interior point of $[0, 1]$; they are boundary points.  
   For $x = 1$, left-neighborhood points exist ($x \lt 1$), but no right-neighborhood points exist ($x \gt 1$) within the domain. Thus, the two-sided derivative cannot be defined relative to an open ball around $x = 1$.

3. **Conclusion:**  
   Fermat's Theorem does not apply to boundary endpoints of closed domains. Extrema on boundary points can occur with non-zero derivatives.

$$
\boxed{\text{No contradiction. Fermat's Theorem applies strictly to interior points } c \in (a, b)\text{, not boundary endpoints.}}
$$

#### Key Insight / Takeaway
When optimizing functions on closed intervals, global extrema can occur at critical points ($f'(c)=0$), non-differentiable points, OR boundary endpoints. Always inspect boundary endpoints separately.

### Problem 0.3 (Non-Indeterminate Forms and L'Hôpital Misuse)
**Source:** Stewart, *Calculus*, Chapter 4  
**Problem Statement:** Evaluate the limit $\lim_{x \to 0} \frac{\cos x}{x + 1}$. Show what happens if L'Hôpital's Rule is erroneously applied, and explain the correct calculation from first principles.

#### First-Principles Intuition
L'Hôpital's Rule is a specialized tool for resolving indeterminate ratios ($\frac{0}{0}$ or $\frac{\infty}{\infty}$). Applying it to a determinate limit changes the problem into a completely different (and wrong) limit.

#### Step-by-Step Solution
1. **Correct Direct Substitution:**  
   Evaluate numerator and denominator at $x = 0$:

$$
   \text{Numerator: } \lim_{x \to 0} \cos x = \cos(0) = 1
$$

$$
   \text{Denominator: } \lim_{x \to 0} (x + 1) = 0 + 1 = 1
$$

   Since the denominator is non-zero, the quotient limit is straightforward:

$$
   \lim_{x \to 0} \frac{\cos x}{x + 1} = \frac{1}{1} = 1
$$

2. **Erroneous L'Hôpital Application:**  
   If one blindly applies L'Hôpital's Rule by differentiating numerator and denominator:

$$
   \frac{\frac{d}{dx}[\cos x]}{\frac{d}{dx}[x + 1]} = \frac{-\sin x}{1}
$$

   Taking $x \to 0$:

$$
   \lim_{x \to 0} \frac{-\sin x}{1} = \frac{-\sin(0)}{1} = 0
$$

   This gives the incorrect answer $0$.

3. **Why it Failed:**  
   The form was $\frac{1}{1}$, which is determinate, not $\frac{0}{0}$ or $\frac{\infty}{\infty}$. L'Hôpital's Rule requires indeterminacy.

$$
\boxed{\text{Correct Limit} = 1\text{. Erroneous L'Hôpital gives } 0 \text{ because the initial limit was not indeterminate.}}
$$

#### Key Insight / Takeaway
Always verify that a limit is indeterminate ($\frac{0}{0}$ or $\frac{\pm\infty}{\pm\infty}$) before invoking L'Hôpital's Rule.

### Problem 0.4 (Zero Second Derivative vs True Inflection Point)
**Source:** Spivak, *Calculus*, Chapter 11  
**Problem Statement:** Analyze the functions $f(x) = x^4$ and $g(x) = x^3$ at $x = 0$. Both functions satisfy $f''(0) = 0$ and $g''(0) = 0$. Determine which function possesses a true inflection point at $x = 0$ and explain the fundamental structural difference.

#### First-Principles Intuition
$f''(c) = 0$ is only a *candidate* condition for an inflection point, analogous to $f'(c) = 0$ being a candidate for a local extremum. A true inflection point requires that the concavity (the sign of $f''$) actually *changes* as $x$ passes through $c$.

#### Step-by-Step Solution
1. **Analyze $f(x) = x^4$:**  
   - First derivative: $f'(x) = 4x^3$
   - Second derivative: $f''(x) = 12x^2$
   - At $x = 0$: $f''(0) = 12(0)^2 = 0$.
   - For $x \lt 0$: $f''(x) = 12x^2 \gt 0$ (convex).
   - For $x \gt 0$: $f''(x) = 12x^2 \gt 0$ (convex).
   - Since $f''(x)$ remains positive on both sides of $x=0$, concavity does **not** change. $x = 0$ is a strict local minimum, not an inflection point.

2. **Analyze $g(x) = x^3$:**  
   - First derivative: $g'(x) = 3x^2$
   - Second derivative: $g''(x) = 6x$
   - At $x = 0$: $g''(0) = 6(0) = 0$.
   - For $x \lt 0$: $g''(x) = 6x \lt 0$ (concave).
   - For $x \gt 0$: $g''(x) = 6x \gt 0$ (convex).
   - Since $g''(x)$ changes sign from negative to positive at $x = 0$, concavity changes. Thus $(0, 0)$ is a true inflection point.

$$
\boxed{g(x) = x^3 \text{ has an inflection point at } x=0\text{; } f(x) = x^4 \text{ does not, because } f''(x) \text{ does not change sign.}}
$$

#### Key Insight / Takeaway
$f''(c) = 0$ is a necessary condition for twice-differentiable inflection points, but the sufficient condition requires a strict sign change of $f''(x)$ across $x = c$.

### Problem 0.5 (Strict Monotonicity with Zero Derivative Points)
**Source:** Apostol, *Calculus, Volume I*, Chapter 4  
**Problem Statement:** Let $f(x) = x + \sin x$ on $\mathbb{R}$. Compute $f'(x)$. Is $f'(x) \ge 0$ everywhere? Does $f'(x) = 0$ at isolated points prevent $f(x)$ from being strictly increasing on $\mathbb{R}$? Prove your assertion from first principles.

#### First-Principles Intuition
A derivative $f'(x) \ge 0$ guarantees that a function is non-decreasing. If $f'(x) = 0$ only at isolated points (and is strictly positive everywhere else), the function cannot remain flat over any interval of positive length, so it remains *strictly* increasing.

#### Step-by-Step Solution
1. **Compute Derivative:**  

$$
   f'(x) = \frac{d}{dx}[x + \sin x] = 1 + \cos x
$$

2. **Analyze Derivative Sign:**  
   Since $-1 \le \cos x \le 1$ for all $x \in \mathbb{R}$, we have:

$$
   f'(x) = 1 + \cos x \ge 0 \quad \forall x \in \mathbb{R}
$$

   $f'(x) = 0$ occurs when $\cos x = -1$, which corresponds to $x = (2k + 1)\pi$ for $k \in \mathbb{Z}$. These zero-derivative points are isolated.

3. **First-Principles Strict Monotonicity Proof:**  
   Let $x_1, x_2 \in \mathbb{R}$ with $x_1 \lt x_2$. By the Mean Value Theorem (Theorem 2.4), there exists $c \in (x_1, x_2)$ such that:

$$
   f(x_2) - f(x_1) = f'(c) (x_2 - x_1) = (1 + \cos c)(x_2 - x_1)
$$

   Since $x_2 - x_1 \gt 0$ and $1 + \cos c \ge 0$, we have $f(x_2) - f(x_1) \ge 0$.  
   Could $f(x_1) = f(x_2)$? That would require $1 + \cos c = 0 \implies c = (2k+1)\pi$ for *every* point in $(x_1, x_2)$, meaning $f'(x) = 0$ identically on $(x_1, x_2)$. But $f'(x) = 1 + \cos x$ is zero only at isolated discrete points, not on any open interval.  
   Therefore, $f(x_2) - f(x_1) \gt 0$, so $f(x_1) \lt f(x_2)$.

$$
\boxed{f(x) = x + \sin x \text{ is strictly increasing on } \mathbb{R} \text{ despite } f'(x)=0 \text{ at isolated points } (2k+1)\pi.}
$$

#### Key Insight / Takeaway
Isolated points where $f'(x) = 0$ do not destroy strict monotonicity, provided $f'(x) \gt 0$ almost everywhere (i.e. on a dense open set).

### Problem 0.6 (L'Hôpital Failure Due to Oscillatory Derivatives)
**Source:** Demidovich, *Problems in Mathematical Analysis*, No. 1388  
**Problem Statement:** Evaluate $\lim_{x \to \infty} \frac{x - \sin x}{x + \sin x}$. Show that applying L'Hôpital's Rule leads to a limit that does not exist, whereas the actual limit exists and can be evaluated by algebraic factorization.

#### First-Principles Intuition
L'Hôpital's Rule states: IF $\lim \frac{f'}{g'} = L$, THEN $\lim \frac{f}{g} = L$. It does NOT state that if $\lim \frac{f'}{g'}$ fails to exist due to oscillation, $\lim \frac{f}{g}$ must also fail to exist.

#### Step-by-Step Solution
1. **Identify Indeterminate Form:**  
   As $x \to \infty$, $x \pm \sin x \to \infty$ because $\lvert\sin x\rvert \le 1$. Form is $\frac{\infty}{\infty}$.

2. **Apply L'Hôpital's Rule:**  

$$
   f'(x) = 1 - \cos x, \quad g'(x) = 1 + \cos x
$$

$$
   \lim_{x \to \infty} \frac{f'(x)}{g'(x)} = \lim_{x \to \infty} \frac{1 - \cos x}{1 + \cos x}
$$

   As $x \to \infty$, $\cos x$ oscillates infinitely between $-1$ and $1$. Thus, the ratio $\frac{1 - \cos x}{1 + \cos x}$ oscillates between $0$ and $\infty$ (undefined at odd multiples of $\pi$). The limit $\lim_{x \to \infty} \frac{f'(x)}{g'(x)}$ **does not exist**.

3. **Direct Algebraic Evaluation:**  
   Divide numerator and denominator by $x \gt 0$:

$$
   \frac{x - \sin x}{x + \sin x} = \frac{1 - \frac{\sin x}{x}}{1 + \frac{\sin x}{x}}
$$

   Since $\lvert\sin x\rvert \le 1$, by the Squeeze Theorem, $\lim_{x \to \infty} \frac{\sin x}{x} = 0$. Therefore:

$$
   \lim_{x \to \infty} \frac{1 - \frac{\sin x}{x}}{1 + \frac{\sin x}{x}} = \frac{1 - 0}{1 + 0} = 1
$$

$$
\boxed{\text{Actual Limit} = 1\text{. L'Hôpital fails because } \lim_{x\to\infty}\frac{f'(x)}{g'(x)} \text{ does not exist due to oscillation.}}
$$

#### Key Insight / Takeaway
The non-existence of $\lim \frac{f'}{g'}$ does not imply the non-existence of $\lim \frac{f}{g}$. When L'Hôpital yields an oscillatory non-existent limit, return to first principles or algebraic bounds.

### Problem 0.7 (Geometric Exact Location of MVT Point for Quadratics)
**Source:** Stewart, *Calculus*, Chapter 4  
**Problem Statement:** Let $f(x) = Ax^2 + Bx + C$ ($A \ne 0$) be a quadratic polynomial on $[a, b]$. Prove that the point $c \in (a, b)$ guaranteed by the Mean Value Theorem is *always* the exact midpoint of the interval, $c = \frac{a + b}{2}$.

#### First-Principles Intuition
For any parabola, the slope of the secant line between two points $a$ and $b$ is equal to the instantaneous tangent slope at the exact arithmetic average of $a$ and $b$.

#### Step-by-Step Solution
1. **Compute Secant Slope:**  

$$
   f(b) - f(a) = (Ab^2 + Bb + C) - (Aa^2 + Ba + C) = A(b^2 - a^2) + B(b - a)
$$

   Since $b - a \ne 0$, divide by $b - a$:

$$
   \frac{f(b) - f(a)}{b - a} = A(b + a) + B
$$

2. **Compute Derivative & Set Equal:**  
   $f'(x) = 2Ax + B$.  
   Set $f'(c) = \frac{f(b) - f(a)}{b - a}$:

$$
   2Ac + B = A(a + b) + B
$$

3. **Solve for $c$:**  
   Subtract $B$ from both sides:

$$
   2Ac = A(a + b)
$$

   Since $A \ne 0$, divide by $2A$:

$$
   c = \frac{a + b}{2}
$$

   Since $a \lt b$, clearly $a \lt \frac{a+b}{2} \lt b$, so $c \in (a, b)$.

$$
\boxed{c = \frac{a + b}{2}}
$$

#### Key Insight / Takeaway
For all quadratic functions, the MVT tangent point is precisely the arithmetic mean of the interval endpoints.

### Problem 0.8 (Newton's Method Instability Near Zero Derivatives)
**Source:** Nocedal & Wright, *Numerical Optimization*, Chapter 2  
**Problem Statement:** Consider applying Newton's method to find a root of $f(x) = x^{\frac{1}{3}}$. Derive the iteration formula $x_{k+1}$ in terms of $x_k$. Show that for any non-zero starting point $x_0 \ne 0$, Newton's method diverges away from the root $r = 0$.

#### First-Principles Intuition
Newton's method uses local tangent lines. If the function is vertical near the root (so $f'(x) \to \infty$ or $f'(x)$ is very small near turning points), the tangent line shoots the next iterate far away from the root rather than closer to it.

#### Step-by-Step Solution
1. **Compute Derivative:**  

$$
   f(x) = x^{\frac{1}{3}} \implies f'(x) = \frac{1}{3} x^{-\frac{2}{3}}
$$

2. **Substitute into Newton's Formula:**  

$$
   x_{k+1} = x_k - \frac{f(x_k)}{f'(x_k)} = x_k - \frac{x_k^{\frac{1}{3}}}{\frac{1}{3} x_k^{-\frac{2}{3}}} = x_k - 3 x_k^{(\frac{1}{3} + \frac{2}{3})} = x_k - 3x_k
$$

$$
   x_{k+1} = -2 x_k
$$

3. **Analyze Iterative Sequence:**  
   Given $x_0 \ne 0$:

$$
   x_1 = -2 x_0
$$

$$
   x_2 = (-2)^2 x_0 = 4 x_0
$$

$$
   x_k = (-2)^k x_0
$$

   As $k \to \infty$, $\lvert x_k \rvert = 2^k \lvert x_0 \rvert \to \infty$.

4. **Conclusion:**  
   The root is $r = 0$. But $\lvert x_k \rvert \to \infty$, so the iterates oscillate with exponentially growing magnitude and diverge to infinity.

$$
\boxed{x_{k+1} = -2x_k\text{. Iterate sequence } x_k = (-2)^k x_0 \text{ diverges to } \infty.}
$$

#### Key Insight / Takeaway
Newton's method requires $f'(r) \ne 0$ and bounded curvature near the root. Vertical tangents ($f'(x) \to \infty$) or zero slopes ($f'(x) = 0$) cause catastrophic divergence or breakdown.

### Problem 1.1 (Mean Value Theorem Calculation)
**Source:** Stewart, *Calculus*, Chapter 4  
**Problem Statement:** Let $f(x) = x^3 - 3x$ on the closed interval $[-1, 2]$. Verify that $f(x)$ satisfies the hypotheses of the Mean Value Theorem on $[-1, 2]$, and find all values $c \in (-1, 2)$ that satisfy the conclusion of the theorem.

#### First-Principles Intuition
$f(x)$ is a polynomial, hence smooth everywhere. We compute the slope of the secant line across $[-1, 2]$ and solve for points where $f'(c)$ equals this slope.

#### Step-by-Step Solution
1. **Verify Hypotheses:**  
   $f(x)$ is a polynomial, so it is continuous on $[-1, 2]$ and differentiable on $(-1, 2)$.

2. **Compute Secant Slope:**  

$$
   f(-1) = (-1)^3 - 3(-1) = -1 + 3 = 2
$$

$$
   f(2) = (2)^3 - 3(2) = 8 - 6 = 2
$$

   Secant slope $m = \frac{f(2) - f(-1)}{2 - (-1)} = \frac{2 - 2}{3} = 0$.

3. **Find $c \in (-1, 2)$ such that $f'(c) = 0$:**  

$$
   f'(x) = 3x^2 - 3
$$

   Set $3c^2 - 3 = 0 \implies c^2 = 1 \implies c = \pm 1$.

4. **Filter Interior Points:**  
   The MVT requires $c \in (-1, 2)$ (open interval).  
   $c = -1$ is an endpoint, so it is excluded.  
   $c = 1 \in (-1, 2)$ is an interior point.

$$
\boxed{c = 1}
$$

#### Key Insight / Takeaway
Extrema points at domain boundaries must be filtered out; MVT specifically guarantees points in the *open* interval $(a, b)$.

### Problem 1.2 (Cauchy Mean Value Theorem Application)
**Source:** Demidovich, *Problems in Mathematical Analysis*, No. 1240  
**Problem Statement:** Apply Cauchy's Mean Value Theorem to $f(x) = \sin x$ and $g(x) = \cos x$ on the interval $[0, \frac{\pi}{2}]$. Find the exact value of $c \in (0, \frac{\pi}{2})$ guaranteed by the theorem.

#### First-Principles Intuition
Cauchy MVT relates the ratio of function differences $\frac{f(b)-f(a)}{g(b)-g(a)}$ to the ratio of derivatives $\frac{f'(c)}{g'(c)}$.

#### Step-by-Step Solution
1. **Verify Hypotheses:**  
   $f(x) = \sin x$ and $g(x) = \cos x$ are continuous on $[0, \frac{\pi}{2}]$ and differentiable on $(0, \frac{\pi}{2})$.  
   $g'(x) = -\sin x \ne 0$ for all $x \in (0, \frac{\pi}{2})$.

2. **Compute Function Differences:**  

$$
   f(\frac{\pi}{2}) - f(0) = \sin(\frac{\pi}{2}) - \sin(0) = 1 - 0 = 1
$$

$$
   g(\frac{\pi}{2}) - g(0) = \cos(\frac{\pi}{2}) - \cos(0) = 0 - 1 = -1
$$

   Ratio: $\frac{f(\frac{\pi}{2}) - f(0)}{g(\frac{\pi}{2}) - g(0)} = \frac{1}{-1} = -1$.

3. **Set Ratio of Derivatives Equal:**  

$$
   f'(c) = \cos c, \quad g'(c) = -\sin c
$$

$$
   \frac{f'(c)}{g'(c)} = \frac{\cos c}{-\sin c} = -\cot c
$$

4. **Solve for $c$:**  

$$
   -\cot c = -1 \implies \cot c = 1 \implies c = \frac{\pi}{4}
$$

   Since $\frac{\pi}{4} \in (0, \frac{\pi}{2})$, this is the unique valid point.

$$
\boxed{c = \frac{\pi}{4}}
$$

#### Key Insight / Takeaway
Cauchy's MVT generalizes standard MVT by parametrizing two curves simultaneously, eliminating explicit reliance on the independent variable $x$.

### Problem 1.3 (Indeterminate Form $\frac{0}{0}$ via Repeated L'Hôpital)
**Source:** Demidovich, *Problems in Mathematical Analysis*, No. 1321  
**Problem Statement:** Evaluate $\lim_{x \to 0} \frac{x - \sin x}{x^3}$ using L'Hôpital's Rule and verify the result using Taylor series expansion.

#### First-Principles Intuition
When applying L'Hôpital's Rule yields another $\frac{0}{0}$ indeterminate form, the rule can be applied repeatedly until a non-zero derivative appears in the denominator.

#### Step-by-Step Solution
1. **Check Form:** As $x \to 0$, $x - \sin x \to 0$ and $x^3 \to 0$ (Form $\frac{0}{0}$).

2. **First L'Hôpital Pass:**  

$$
   \lim_{x \to 0} \frac{\frac{d}{dx}[x - \sin x]}{\frac{d}{dx}[x^3]} = \lim_{x \to 0} \frac{1 - \cos x}{3x^2} \quad \left(\text{Form }\frac{0}{0}\right)
$$

3. **Second L'Hôpital Pass:**  

$$
   \lim_{x \to 0} \frac{\frac{d}{dx}[1 - \cos x]}{\frac{d}{dx}[3x^2]} = \lim_{x \to 0} \frac{\sin x}{6x} \quad \left(\text{Form }\frac{0}{0}\right)
$$

4. **Third L'Hôpital Pass:**  

$$
   \lim_{x \to 0} \frac{\frac{d}{dx}[\sin x]}{\frac{d}{dx}[6x]} = \lim_{x \to 0} \frac{\cos x}{6} = \frac{1}{6}
$$

5. **Taylor Expansion Verification:**  
   $\sin x = x - \frac{x^3}{6} + O(x^5) \implies x - \sin x = \frac{x^3}{6} - O(x^5)$.  

$$
   \lim_{x \to 0} \frac{\frac{x^3}{6} - O(x^5)}{x^3} = \frac{1}{6}
$$

$$
\boxed{\frac{1}{6}}
$$

#### Key Insight / Takeaway
Repeated L'Hôpital applications track higher-order terms in the function's Taylor series expansion until the lowest order non-zero polynomial term dominates.

### Problem 1.4 (Indeterminate Form $\frac{\infty}{\infty}$)
**Source:** Spivak, *Calculus*, Chapter 11  
**Problem Statement:** Evaluate the limit $\lim_{x \to \infty} \frac{\ln(1 + e^{2x})}{x}$.

#### First-Principles Intuition
As $x \to \infty$, $e^{2x}$ dominates 1 inside the logarithm, so $\ln(1 + e^{2x}) \approx \ln(e^{2x}) = 2x$. The ratio $\frac{2x}{x}$ should tend to 2.

#### Step-by-Step Solution
1. **Check Form:** As $x \to \infty$, $1 + e^{2x} \to \infty \implies \ln(1+e^{2x}) \to \infty$, and $x \to \infty$. Form is $\frac{\infty}{\infty}$.

2. **Apply L'Hôpital's Rule:**  

$$
   \frac{d}{dx}[\ln(1 + e^{2x})] = \frac{1}{1 + e^{2x}} \cdot 2e^{2x} = \frac{2e^{2x}}{1 + e^{2x}}
$$

$$
   \frac{d}{dx}[x] = 1
$$

$$
   \lim_{x \to \infty} \frac{f'(x)}{g'(x)} = \lim_{x \to \infty} \frac{2e^{2x}}{1 + e^{2x}}
$$

3. **Evaluate Secondary Limit:**  
   Divide numerator and denominator by $e^{2x} \gt 0$:

$$
   \lim_{x \to \infty} \frac{2}{e^{-2x} + 1} = \frac{2}{0 + 1} = 2
$$

$$
\boxed{2}
$$

#### Key Insight / Takeaway
Logarithmic growing functions of exponentials grow linearly; L'Hôpital simplifies composite logarithmic derivatives rapidly.

### Problem 1.5 (Indeterminate Form $0 \cdot \infty$)
**Source:** Demidovich, *Problems in Mathematical Analysis*, No. 1345  
**Problem Statement:** Evaluate $\lim_{x \to 0^+} x^2 \ln x$.

#### First-Principles Intuition
As $x \to 0^+$, $x^2 \to 0$ while $\ln x \to -\infty$. Polynomial decay ($x^2$) is much faster than logarithmic growth/decay ($\lvert\ln x\rvert$), so the product should approach 0.

#### Step-by-Step Solution
1. **Rewrite as Fraction ($\frac{\infty}{\infty}$):**  

$$
   x^2 \ln x = \frac{\ln x}{x^{-2}}
$$

   As $x \to 0^+$, $\ln x \to -\infty$ and $x^{-2} \to \infty$. Form is $\frac{\infty}{\infty}$.

2. **Apply L'Hôpital's Rule:**  

$$
   \frac{d}{dx}[\ln x] = \frac{1}{x}
$$

$$
   \frac{d}{dx}[x^{-2}] = -2 x^{-3}
$$

$$
   \lim_{x \to 0^+} \frac{\frac{1}{x}}{-2 x^{-3}} = \lim_{x \to 0^+} \left( \frac{1}{x} \cdot \frac{x^3}{-2} \right) = \lim_{x \to 0^+} \left( -\frac{x^2}{2} \right)
$$

3. **Compute Final Limit:**  

$$
   -\frac{(0)^2}{2} = 0
$$

$$
\boxed{0}
$$

#### Key Insight / Takeaway
To resolve $0 \cdot \infty$, invert the polynomial term to the denominator as $x^{-p}$; differentiation then reduces the algebraic power of $x$.

### Problem 1.6 (Indeterminate Exponential Form $1^\infty$)
**Source:** Stewart, *Calculus*, Chapter 4  
**Problem Statement:** Evaluate $\lim_{x \to 0} (1 + 3x)^{\frac{1}{x}}$.

#### First-Principles Intuition
This is the canonical definition structure of the natural exponential $e^k$, where $(1 + k/n)^n \to e^k$.

#### Step-by-Step Solution
1. **Define Logarithmic Transformation:**  
   Let $y = (1 + 3x)^{\frac{1}{x}}$. Take natural log:

$$
   \ln y = \ln\left((1 + 3x)^{\frac{1}{x}}\right) = \frac{\ln(1 + 3x)}{x}
$$

2. **Evaluate $\lim_{x \to 0} \ln y$ (Form $\frac{0}{0}$):**  
   As $x \to 0$, $\ln(1 + 3x) \to \ln 1 = 0$ and $x \to 0$. Apply L'Hôpital:

$$
   \lim_{x \to 0} \frac{\frac{d}{dx}[\ln(1+3x)]}{\frac{d}{dx}[x]} = \lim_{x \to 0} \frac{\frac{3}{1+3x}}{1} = \frac{3}{1+0} = 3
$$

3. **Exponentiate to Recover Original Limit:**  
   Since $\lim_{x \to 0} \ln y = 3$, by continuity of the exponential function:

$$
   \lim_{x \to 0} y = e^{\lim_{x \to 0} \ln y} = e^3
$$

$$
\boxed{e^3}
$$

#### Key Insight / Takeaway
Indeterminate exponential forms ($1^\infty, 0^0, \infty^0$) are solved by taking logarithms, converting to $\frac{0}{0}$ or $\frac{\infty}{\infty}$, and exponentiating the final limit.

### Problem 1.7 (Indeterminate Exponential Form $0^0$)
**Source:** Demidovich, *Problems in Mathematical Analysis*, No. 1358  
**Problem Statement:** Evaluate $\lim_{x \to 0^+} x^{\sin x}$.

#### First-Principles Intuition
Both base $x \to 0$ and exponent $\sin x \to 0$. As $\sin x \approx x$ near zero, $x^x \to 1$.

#### Step-by-Step Solution
1. **Logarithmic Transformation:**  
   Let $y = x^{\sin x}$.  

$$
   \ln y = \sin x \ln x = \frac{\ln x}{\csc x} \quad \left(\text{Form }\frac{-\infty}{\infty}\right)
$$

2. **Alternative Algebraic Conversion:**  
   Alternatively, rewrite $\sin x \ln x = \frac{\sin x}{x} \cdot (x \ln x)$.  
   We know $\lim_{x \to 0^+} \frac{\sin x}{x} = 1$.  
   For $\lim_{x \to 0^+} x \ln x$:

$$
   \lim_{x \to 0^+} \frac{\ln x}{x^{-1}} \underbrace{=}_{\text{L'H}} \lim_{x \to 0^+} \frac{\frac{1}{x}}{-\frac{1}{x}^2} = \lim_{x \to 0^+} (-x) = 0
$$

3. **Combine Limits:**  

$$
   \lim_{x \to 0^+} \ln y = \left(\lim_{x \to 0^+} \frac{\sin x}{x}\right) \cdot \left(\lim_{x \to 0^+} x \ln x\right) = 1 \cdot 0 = 0
$$

4. **Exponentiate:**  

$$
   \lim_{x \to 0^+} y = e^0 = 1
$$

$$
\boxed{1}
$$

#### Key Insight / Takeaway
$0^0$ indeterminate limits resolve to 1 whenever the base is polynomial and the exponent is linear near zero.

### Problem 1.8 (Complete Monotonicity & Extrema Analysis)
**Source:** Demidovich, *Problems in Mathematical Analysis*, No. 1410  
**Problem Statement:** Given $f(x) = 2x^3 - 9x^2 + 12x - 3$, find all critical points, intervals of strict increase and decrease, and classify all local extrema.

#### First-Principles Intuition
We use the First Derivative Test: $f'(x) \gt 0 \implies$ increasing, $f'(x) \lt 0 \implies$ decreasing. Sign changes across critical points reveal local maxima and minima.

#### Step-by-Step Solution
1. **Find Critical Points:**  

$$
   f'(x) = 6x^2 - 18x + 12 = 6(x^2 - 3x + 2) = 6(x - 1)(x - 2)
$$

2. **Interval Sign Analysis:**

$$

   - At $x = 2$: $f'(x)$ changes from negative to positive $\implies$ **Local Minimum**.  

$$
     f(2) = 2(8) - 9(4) + 12(2) - 3 = 16 - 36 + 24 - 3 = 1
$$

$$
\boxed{\text{Increasing on } (-\infty, 1) \cup (2, \infty)\text{; Decreasing on } (1, 2)\text{. Local Max at } (1, 2)\text{; Local Min at } (2, 1).}

### Problem 1.9 (Concavity and Inflection Points of Gaussian-Type Curve)
**Source:** Stewart, *Calculus*, Chapter 4  
**Problem Statement:** Analyze $f(x) = x e^{-\frac{x^2}{2}}$. Find all intervals of concavity and exact coordinates of all inflection points.

#### First-Principles Intuition
This is the derivative of a Gaussian envelope. Concavity shifts where the second derivative changes sign ($f''(x) = 0$).

#### Step-by-Step Solution
1. **First Derivative:**  

$$
   f'(x) = (1)e^{-\frac{x^2}{2}} + x(-x e^{-\frac{x^2}{2}}) = (1 - x^2) e^{-\frac{x^2}{2}}
$$

2. **Second Derivative:**  

$$
   f''(x) = (-2x)e^{-\frac{x^2}{2}} + (1 - x^2)(-x e^{-\frac{x^2}{2}}) = e^{-\frac{x^2}{2}} \left( -2x - x + x^3 \right) = x(x^2 - 3) e^{-\frac{x^2}{2}}
$$

3. **Find Candidate Inflection Points ($f''(x) = 0$):**  
   Since $e^{-\frac{x^2}{2}} \gt 0$ everywhere, $f''(x) = 0 \implies x(x^2 - 3) = 0$.  
   Candidate roots: $x_1 = -\sqrt{3}, x_2 = 0, x_3 = \sqrt{3}$.

4. **Sign Analysis of $f''(x)$:**  
   - $(-\infty, -\sqrt{3})$: $x \lt 0$ and $x^2-3 \gt 0 \implies f''(x) \lt 0$ (**Concave Down**).
   - $(-\sqrt{3}, 0)$: $x \lt 0$ and $x^2-3 \lt 0 \implies f''(x) \gt 0$ (**Concave Up / Convex**).
   - $(0, \sqrt{3})$: $x \gt 0$ and $x^2-3 \lt 0 \implies f''(x) \lt 0$ (**Concave Down**).
   - $(\sqrt{3}, \infty)$: $x \gt 0$ and $x^2-3 \gt 0 \implies f''(x) \gt 0$ (**Concave Up / Convex**).

5. **Evaluate Inflection Points:**  
   All three candidates change sign, so all three are inflection points:
   - $f(-\sqrt{3}) = -\sqrt{3} e^{-\frac{3}{2}}$
   - $f(0) = 0$
   - $f(\sqrt{3}) = \sqrt{3} e^{-\frac{3}{2}}$

$$
\boxed{\text{Inflection Points at } (0, 0) \text{ and } \left(\pm\sqrt{3}, \pm\sqrt{3} e^{-\frac{3}{2}}\right).}
$$

#### Key Insight / Takeaway
The inflection points of $x e^{-\frac{x^2}{2}}$ mark the transition boundaries between central exponential decay and outer tail behavior.

### Problem 1.10 (Global Extrema on Closed Interval)
**Source:** Apostol, *Calculus, Volume I*, Chapter 4  
**Problem Statement:** Find the global maximum and global minimum values of $f(x) = x - 2\sin x$ on the closed interval $[0, 2\pi]$.

#### First-Principles Intuition
Continuous functions on closed intervals attain global extrema either at interior critical points ($f'(x) = 0$) or at interval boundary endpoints.

#### Step-by-Step Solution
1. **Find Critical Points in $(0, 2\pi)$:**  

$$
   f'(x) = 1 - 2\cos x
$$

   Set $f'(x) = 0 \implies \cos x = \frac{1}{2}$.  
   In $(0, 2\pi)$, solutions are $x_1 = \frac{\pi}{3}$ and $x_2 = \frac{5\pi}{3}$.

2. **Evaluate $f(x)$ at Candidate Points:**  
   - **Endpoint $x = 0$:**  

$$
     f(0) = 0 - 2\sin(0) = 0
$$

   - **Critical Point $x = \frac{\pi}{3}$:**  

$$
     f\left(\frac{\pi}{3}\right) = \frac{\pi}{3} - 2\sin\left(\frac{\pi}{3}\right) = \frac{\pi}{3} - 2\left(\frac{\sqrt{3}}{2}\right) = \frac{\pi}{3} - \sqrt{3} \approx 1.047 - 1.732 = -0.685
$$

   - **Critical Point $x = \frac{5\pi}{3}$:**  

$$
     f\left(\frac{5\pi}{3}\right) = \frac{5\pi}{3} - 2\sin\left(\frac{5\pi}{3}\right) = \frac{5\pi}{3} - 2\left(-\frac{\sqrt{3}}{2}\right) = \frac{5\pi}{3} + \sqrt{3} \approx 5.236 + 1.732 = 6.968
$$

   - **Endpoint $x = 2\pi$:**  

$$
     f(2\pi) = 2\pi - 2\sin(2\pi) = 2\pi \approx 6.283
$$

3. **Compare Candidates:**  
   - Smallest value: $\frac{\pi}{3} - \sqrt{3} \approx -0.685$.  
   - Largest value: $\frac{5\pi}{3} + \sqrt{3} \approx 6.968$.

$$
\boxed{\text{Global Max } = \frac{5\pi}{3} + \sqrt{3} \quad \text{at } x = \frac{5\pi}{3}; \quad \text{Global Min } = \frac{\pi}{3} - \sqrt{3} \quad \text{at } x = \frac{\pi}{3}}
$$

#### Key Insight / Takeaway
Never assume endpoints are automatically the extrema; candidate comparison of all critical points and endpoints is mandatory.

### Problem 1.11 (Newton's Method Iterative Calculation)
**Source:** Nocedal & Wright, *Numerical Optimization*, Chapter 2  
**Problem Statement:** Perform two iterations of Newton's method to approximate $\sqrt{5}$ by finding the positive root of $f(x) = x^2 - 5 = 0$, starting with initial guess $x_0 = 2.0$. Express $x_1$ and $x_2$ as exact rational fractions.

#### First-Principles Intuition
Newton's method iteratively refines an estimate $x_k$ via tangent lines: $x_{k+1} = \frac{1}{2}\left(x_k + \frac{5}{x_k}\right)$ for square roots (the ancient Babylonian algorithm).

#### Step-by-Step Solution
1. **Derive Iteration Formula:**  
   $f(x) = x^2 - 5 \implies f'(x) = 2x$.  

$$
   x_{k+1} = x_k - \frac{x_k^2 - 5}{2x_k} = \frac{2x_k^2 - x_k^2 + 5}{2x_k} = \frac{x_k^2 + 5}{2x_k} = \frac{1}{2}\left(x_k + \frac{5}{x_k}\right)
$$

2. **First Iteration ($x_0 = 2 = 2/1$):**  

$$
   x_1 = \frac{1}{2}\left(2 + \frac{5}{2}\right) = \frac{1}{2}\left(\frac{9}{2}\right) = \frac{9}{4} = 2.25
$$

3. **Second Iteration ($x_1 = \frac{9}{4}$):**  

$$
   x_2 = \frac{1}{2}\left(\frac{9}{4} + \frac{5}{\frac{9}{4}}\right) = \frac{1}{2}\left(\frac{9}{4} + \frac{20}{9}\right) = \frac{1}{2}\left(\frac{81 + 80}{36}\right) = \frac{161}{72}
$$

4. **Numerical Precision Check:**  
   $x_2 = \frac{161}{72} \approx 2.236111...$  
   Actual $\sqrt{5} \approx 2.236068...$  
   Error at step 2 is $\lvert x_2 - \sqrt{5} \rvert \approx 0.000043$ (accurate to 4 decimal places after just 2 steps!).

$$
\boxed{x_1 = \frac{9}{4}, \quad x_2 = \frac{161}{72}}
$$

#### Key Insight / Takeaway
Newton's method quadratic convergence produces exponential growth in decimal precision per step.

### Problem 1.12 (Monotonicity-Based Inequality Proof)
**Source:** Spivak, *Calculus*, Chapter 11  
**Problem Statement:** Prove that $\ln(1 + x) \lt x$ for all $x \gt 0$ using derivative monotonicity from first principles.

#### First-Principles Intuition
To prove $g(x) \lt h(x)$ for $x \gt 0$, construct $f(x) = h(x) - g(x) = x - \ln(1+x)$. If $f(0) = 0$ and $f'(x) \gt 0$ for $x \gt 0$, then $f(x)$ strictly increases away from 0, guaranteeing $f(x) \gt 0$.

#### Step-by-Step Solution
1. **Define Auxiliary Function:**  

$$
   f(x) = x - \ln(1 + x) \quad \text{for } x \ge 0
$$

2. **Evaluate at Boundary $x = 0$:**  

$$
   f(0) = 0 - \ln(1 + 0) = 0 - 0 = 0
$$

3. **Compute Derivative for $x \gt 0$:**  

$$
   f'(x) = 1 - \frac{1}{1 + x} = \frac{(1 + x) - 1}{1 + x} = \frac{x}{1 + x}
$$

4. **Analyze Derivative Sign:**  
   For any $x \gt 0$, $x \gt 0$ and $1 + x \gt 0$, so:

$$
   f'(x) = \frac{x}{1 + x} \gt 0 \quad \forall x \gt 0
$$

5. **Apply Monotonicity Theorem:**  
   By Theorem 4.1, $f(x)$ is strictly increasing on $[0, \infty)$.  
   Therefore, for any $x \gt 0$:

$$
   f(x) \gt f(0) \implies x - \ln(1 + x) \gt 0 \implies \ln(1 + x) \lt x
$$

$$
\boxed{\ln(1 + x) \lt x \quad \forall x \gt 0}
$$

#### Key Insight / Takeaway
Inequalities of the form $g(x) \lt h(x)$ are universally proved by showing $h(x) - g(x)$ has a positive derivative starting from an equality base point.

### Problem 2.1 (Physics: Snell's Law Derivation from Fermat's Principle)
**Source:** Standard Classical Mechanics / Optics  
**Problem Statement:** A light ray travels from point $A=(0, a)$ in medium 1 (speed $v_1$) to point $B=(d, -b)$ in medium 2 (speed $v_2$). The boundary between media is the line $y=0$. Derive Snell's Law $\frac{\sin\theta_1}{v_1} = \frac{\sin\theta_2}{v_2}$ by finding the interface position $x$ that minimizes total travel time.

#### First-Principles Intuition
Fermat's Principle of Least Time states that light takes the path of minimum travel time. Setting the first derivative of total time $T(x)$ with respect to refraction coordinate $x$ to zero yields Snell's Law.

#### Step-by-Step Solution
1. **Formulate Total Travel Time $T(x)$:**  
   Distance in medium 1: $d_1(x) = \sqrt{x^2 + a^2}$. Time $t_1 = \frac{\sqrt{x^2 + a^2}}{v_1}$.  
   Distance in medium 2: $d_2(x) = \sqrt{(d-x)^2 + b^2}$. Time $t_2 = \frac{\sqrt{(d-x)^2 + b^2}}{v_2}$.  

$$
   T(x) = \frac{\sqrt{x^2 + a^2}}{v_1} + \frac{\sqrt{(d-x)^2 + b^2}}{v_2}
$$

2. **Compute First Derivative $T'(x)$:**  

$$
   T'(x) = \frac{x}{v_1 \sqrt{x^2 + a^2}} - \frac{d - x}{v_2 \sqrt{(d-x)^2 + b^2}}
$$

3. **Relate to Angles of Incidence & Refraction:**  
   From the right triangles at the interface:

$$
   \sin\theta_1 = \frac{x}{\sqrt{x^2 + a^2}}, \quad \sin\theta_2 = \frac{d - x}{\sqrt{(d-x)^2 + b^2}}
$$

4. **Set $T'(x^{\ast}) = 0$ for Optimality:**  

$$
   \frac{\sin\theta_1}{v_1} - \frac{\sin\theta_2}{v_2} = 0 \implies \frac{\sin\theta_1}{v_1} = \frac{\sin\theta_2}{v_2}
$$

5. **Verify Strict Convexity ($T''(x) \gt 0$):**  

$$
   T''(x) = \frac{a^2}{v_1 (x^2 + a^2)^{\frac{3}{2}}} + \frac{b^2}{v_2 ((d-x)^2 + b^2)^{\frac{3}{2}}} \gt 0
$$

   Since $T''(x) \gt 0$ everywhere, $x^{\ast}$ is the unique global minimum time path.

$$
\boxed{\frac{\sin\theta_1}{\sin\theta_2} = \frac{v_1}{v_2}}
$$

#### Key Insight / Takeaway
Physical laws governing wave refraction (Snell's Law) emerge as the unique global minimum of a 1D convex travel time optimization problem.

### Problem 2.2 (Physics: Maximum Power Transfer Theorem)
**Source:** Electrical Engineering / Physics  
**Problem Statement:** A DC voltage source with EMF $V$ and internal resistance $r$ is connected to a variable load resistor $R \gt 0$. The electrical power dissipated in the load resistor is $P(R) = \frac{V^2 R}{(R + r)^2}$. Find the load resistance $R^{\ast}$ that maximizes power transfer, and compute the maximum power $P_{\text{max}}$.

#### First-Principles Intuition
If $R \to 0$, current is high but voltage across load is zero ($P \to 0$). If $R \to \infty$, voltage is high but current is zero ($P \to 0$). The peak power must occur at a finite intermediate resistance $R^{\ast}$.

#### Step-by-Step Solution
1. **Compute Derivative $P'(R)$:**  
   Using the quotient rule on $P(R) = V^2 \cdot \frac{R}{(R+r)^2}$:

$$
   P'(R) = V^2 \frac{(1)(R+r)^2 - R \cdot 2(R+r)}{(R+r)^4} = V^2 \frac{(R+r) - 2R}{(R+r)^3} = V^2 \frac{r - R}{(R+r)^3}
$$

2. **Find Critical Point:**  
   Set $P'(R) = 0 \implies r - R = 0 \implies R^{\ast} = r$.

3. **Verify Local Maximum via First Derivative Test:**  
   - For $0 \lt R \lt r$: $r - R \gt 0 \implies P'(R) \gt 0$ (increasing).
   - For $R \gt r$: $r - R \lt 0 \implies P'(R) \lt 0$ (decreasing).  
   Hence $R^{\ast} = r$ is the global maximum.

4. **Compute Maximum Power $P_{\text{max}}$:**  

$$
   P_{\text{max}} = P(r) = \frac{V^2 r}{(r + r)^2} = \frac{V^2 r}{4r^2} = \frac{V^2}{4r}
$$

$$
\boxed{R^{\ast} = r, \quad P_{\text{max}} = \frac{V^2}{4r}}
$$

#### Key Insight / Takeaway
Maximum power transfer occurs when load impedance matches internal source impedance ($R=r$), achieving 50% electrical efficiency.

### Problem 2.3 (Physics: Optimal Projectile Range on Inclined Plane)
**Source:** Classical Mechanics (Marion & Thornton)  
**Problem Statement:** A projectile is launched from ground level with initial speed $v_0$ at angle $\theta$ relative to the horizontal. The horizontal range is $R(\theta) = \frac{v_0^2 \sin(2\theta)}{g}$ for $\theta \in [0, \frac{\pi}{2}]$. Find the optimal launch angle $\theta^{\ast}$ that maximizes range on flat ground.

#### First-Principles Intuition
Range combines horizontal velocity component $v_0 \cos\theta$ and time of flight $t = \frac{2v_0\sin\theta}{g}$. Trade-off between speed and flight time reaches peak at $\theta = 45^\circ$.

$$
   R'(\theta) = \frac{v_0^2}{g} \frac{d}{d\theta}[\sin(2\theta)] = \frac{2 v_0^2}{g} \cos(2\theta)
$$

   Set $R'(\theta) = 0 \implies \cos(2\theta) = 0$.  
   For $\theta \in (0, \frac{\pi}{2})$, $2\theta \in (0, \pi)$, so $2\theta = \frac{\pi}{2} \implies \theta^{\ast} = \frac{\pi}{4} = 45^\circ$.

3. **Second Derivative Test:**  

$$
   R''(\theta) = -\frac{4 v_0^2}{g} \sin(2\theta)
$$

   At $\theta^{\ast} = \frac{\pi}{4}$:

$$
   R''(\frac{\pi}{4}) = -\frac{4 v_0^2}{g} \sin(\frac{\pi}{2}) = -\frac{4 v_0^2}{g} \lt 0
$$

   Confirms a strict maximum.

4. **Compute Maximum Range:**  

$$
   R_{\text{max}} = R(\frac{\pi}{4}) = \frac{v_0^2 \sin(\frac{\pi}{2})}{g} = \frac{v_0^2}{g}
$$

$$
\boxed{\theta^{\ast} = 45^\circ \left(\frac{\pi}{4}\right), \quad R_{\text{max}} = \frac{v_0^2}{g}}
$$

#### Key Insight / Takeaway
$45^\circ$ perfectly balances the competing linear factors of horizontal propulsion and vertical flight duration.

### Problem 2.4 (AI/ML: Binary Cross-Entropy Loss Strict Convexity)
**Source:** Goodfellow et al., *Deep Learning*, Chapter 4  
**Problem Statement:** Let $y \in \{0, 1\}$ be a target binary label, $x \ne 0$ a real feature, and $w \in \mathbb{R}$ a scalar weight. The model prediction is $p = \sigma(w x) = \frac{1}{1 + e^{-w x}}$. The Binary Cross-Entropy (BCE) loss is:

$$
L(w) = -y \ln(\sigma(w x)) - (1-y) \ln(1 - \sigma(w x))
$$

Prove that $L(w)$ is strictly convex with respect to $w$ for all $w \in \mathbb{R}$.

#### First-Principles Intuition
A loss function with positive second derivative ($L''(w) \gt 0$) is strictly convex, guaranteeing that gradient descent cannot get stuck in local sub-optimal minima.

#### Step-by-Step Solution
1. **Recall Sigmoid Derivative Identity:**  
   Let $z = w x$. $\sigma(z) = \frac{1}{1 + e^{-z}}$.  

$$
   \frac{d\sigma}{dz} = \sigma(z)(1 - \sigma(z))
$$

   By chain rule, $\frac{d\sigma}{dw} = x \sigma(w x)(1 - \sigma(w x))$.

2. **Compute First Derivative $L'(w)$:**  

$$
   L'(w) = -y \frac{x \sigma(wx)(1-\sigma(wx))}{\sigma(wx)} - (1-y) \frac{-x \sigma(wx)(1-\sigma(wx))}{1-\sigma(wx)}
$$

$$
   L'(w) = -y x (1 - \sigma(w x)) + (1-y) x \sigma(w x) = x \left[ \sigma(w x) - y \right]
$$

3. **Compute Second Derivative $L''(w)$:**  

$$
   L''(w) = \frac{d}{dw}\left( x [\sigma(w x) - y] \right) = x \cdot x \sigma(w x)(1 - \sigma(w x)) = x^2 \sigma(w x)(1 - \sigma(w x))
$$

4. **Establish Strict Inequality:**  
   - Since $x \ne 0$, $x^2 \gt 0$.
   - For any real $z$, $\sigma(z) \in (0, 1)$, so $\sigma(w x) \gt 0$ and $1 - \sigma(w x) \gt 0$.
   - Product of strictly positive terms: $L''(w) \gt 0$ for all $w \in \mathbb{R}$.

5. **Conclusion:**  
   By Theorem 4.2, $L(w)$ is strictly convex on $\mathbb{R}$.

$$
\boxed{L''(w) = x^2 \sigma(wx)(1-\sigma(wx)) \gt 0 \implies L(w) \text{ is strictly convex on } \mathbb{R}.}
$$

#### Key Insight / Takeaway
The combination of sigmoid activations with cross-entropy loss eliminates non-convex loss plateaus, yielding global convergence guarantees.

### Problem 2.5 (AI/ML: Ridge Regression 1D Closed-Form Solution)
**Source:** Hastie et al., *The Elements of Statistical Learning*  
**Problem Statement:** Consider a 1D L2-regularized least-squares objective (Ridge Regression):

$$
L(w) = \frac{1}{2} \sum_{i=1}^N (y_i - w x_i)^2 + \frac{\lambda}{2} w^2 \quad (\lambda \gt 0)
$$

Derive the analytical global minimizer $w^{\ast}$ from first principles and prove it is unique.

#### First-Principles Intuition
L2 regularization adds a quadratic penalty $\lambda w^2$ to control model complexity. The objective remains quadratic, yielding a unique closed-form solution.

#### Step-by-Step Solution
1. **Expand Loss Function:**  

$$
   L(w) = \frac{1}{2} \sum_{i=1}^N \left( y_i^2 - 2 w x_i y_i + w^2 x_i^2 \right) + \frac{\lambda}{2} w^2
$$

$$
   L(w) = \frac{1}{2} \left( \sum_{i=1}^N x_i^2 + \lambda \right) w^2 - \left( \sum_{i=1}^N x_i y_i \right) w + \frac{1}{2} \sum_{i=1}^N y_i^2
$$

2. **Compute First Derivative $L'(w)$:**  

$$
   L'(w) = \left( \sum_{i=1}^N x_i^2 + \lambda \right) w - \sum_{i=1}^N x_i y_i
$$

3. **Set $L'(w^{\ast}) = 0$ to Solve for $w^{\ast}$:**  

$$
   \left( \sum_{i=1}^N x_i^2 + \lambda \right) w^{\ast} = \sum_{i=1}^N x_i y_i \implies w^{\ast} = \frac{\sum_{i=1}^N x_i y_i}{\sum_{i=1}^N x_i^2 + \lambda}
$$

4. **Verify Strict Convexity ($L''(w) \gt 0$):**  

$$
   L''(w) = \sum_{i=1}^N x_i^2 + \lambda
$$

   Since $\sum x_i^2 \ge 0$ and $\lambda \gt 0$, $L''(w) \ge \lambda \gt 0$.  
   Thus $L(w)$ is strictly convex and $w^{\ast}$ is the unique global minimizer.

$$
\boxed{w^{\ast} = \frac{\sum_{i=1}^N x_i y_i}{\sum_{i=1}^N x_i^2 + \lambda}}
$$

#### Key Insight / Takeaway
L2 regularization acts as a positive shift $\lambda$ in the denominator, preventing division by zero even when features $x_i = 0$.

### Problem 2.6 (AI/ML: Huber Loss Smoothness & Convexity)
**Source:** Nocedal & Wright, *Numerical Optimization*, Chapter 3  
**Problem Statement:** The Huber loss function combines $L_2$ square loss near origin with $L_1$ linear loss in tails:

$$
H_\delta(e) = \begin{cases} \frac{1}{2} e^2 & \text{if } \lvert e \rvert \le \delta \\ \delta \lvert e \rvert - \frac{1}{2} \delta^2 & \text{if } \lvert e \rvert \gt \delta \end{cases}
$$

Prove that $H_\delta(e)$ is continuously differentiable ($C^1$) everywhere and convex on $\mathbb{R}$.

#### First-Principles Intuition
Standard $L_1$ loss $\lvert e \rvert$ has a non-differentiable sharp corner at $e=0$. Huber loss smooths this corner with a parabola $\frac{1}{2}e^2$, retaining robustness to outliers without losing differentiability.

#### Step-by-Step Solution
1. **Check Derivative for $\lvert e \rvert \lt \delta$:**  
   $H'_\delta(e) = e$.

2. **Check Derivative for $\lvert e \rvert \gt \delta$:**  
   - For $e \gt \delta$: $H_\delta(e) = \delta e - \frac{1}{2}\delta^2 \implies H'_\delta(e) = \delta$.
   - For $e \lt -\delta$: $H_\delta(e) = -\delta e - \frac{1}{2}\delta^2 \implies H'_\delta(e) = -\delta$.

3. **Verify Differentiability at Boundaries $e = \pm \delta$:**  
   - At $e = \delta$:  
     Left derivative: $\lim_{e \to \delta^-} e = \delta$.  
     Right derivative: $\lim_{e \to \delta^+} \delta = \delta$.  
     They match! $H'_\delta(\delta) = \delta$.
   - At $e = -\delta$:  
     Left derivative: $\lim_{e \to -\delta^-} (-\delta) = -\delta$.  
     Right derivative: $\lim_{e \to -\delta^+} e = -\delta$.  
     They match! $H'_\delta(-\delta) = -\delta$.

4. **Express Full Derivative:**  

$$
   H'_\delta(e) = \begin{cases} -\delta & \text{if } e \lt -\delta \\ e & \text{if } \lvert e \rvert \le \delta \\ \delta & \text{if } e \gt \delta \end{cases}
$$

   $H'_\delta(e)$ is continuous everywhere on $\mathbb{R}$, so $H_\delta \in C^1(\mathbb{R})$.

5. **Verify Monotonicity of $H'_\delta(e)$ for Convexity:**  
   $H'_\delta(e)$ is non-decreasing on $(-\infty, \infty)$ (it stays flat at $-\delta$, ramps linearly up to $\delta$, and stays flat at $\delta$). By Theorem 4.1, a continuous function with a non-decreasing derivative is **convex**.

$$
\boxed{H'_\delta(e) \text{ is continuous everywhere and non-decreasing } \implies H_\delta \in C^1(\mathbb{R}) \text{ and is convex.}}
$$

#### Key Insight / Takeaway
Huber loss provides smooth derivative transitions at $e = \pm\delta$, enabling robust optimization without gradient discontinuities.

### Problem 2.7 (AI/ML: 1D Exact Line Search for Quadratic Loss)
**Source:** Nocedal & Wright, *Numerical Optimization*, Chapter 3  
**Problem Statement:** In optimization, given current point $x_k$ and search direction $d \ne 0$, exact line search computes step size $\alpha^{\ast} = \arg\min_{\alpha \gt 0} \phi(\alpha)$ where $\phi(\alpha) = f(x_k + \alpha d)$. Derive $\alpha^{\ast}$ for quadratic objective $f(x) = \frac{1}{2} a x^2 - b x$ ($a \gt 0$).

#### First-Principles Intuition
Along a 1D slice, a 2D/multi-D quadratic function reduces to a simple parabola in $\alpha$. The minimum occurs where $\phi'(\alpha) = 0$.

#### Step-by-Step Solution
1. **Construct 1D Function $\phi(\alpha)$:**  

$$
   \phi(\alpha) = f(x_k + \alpha d) = \frac{1}{2} a (x_k + \alpha d)^2 - b (x_k + \alpha d)
$$

$$
   \phi(\alpha) = \frac{1}{2} a \left( x_k^2 + 2\alpha x_k d + \alpha^2 d^2 \right) - b x_k - \alpha b d
$$

2. **Compute Derivative $\phi'(\alpha)$ via Chain Rule:**  

$$
   \phi'(\alpha) = \nabla f(x_k + \alpha d) \cdot d = \left[ a(x_k + \alpha d) - b \right] d
$$

$$
   \phi'(\alpha) = (a x_k - b)d + a d^2 \alpha
$$

   Note that $g_k = f'(x_k) = a x_k - b$, so $\phi'(\alpha) = g_k d + a d^2 \alpha$.

3. **Set $\phi'(\alpha^{\ast}) = 0$:**  

$$
   g_k d + a d^2 \alpha^{\ast} = 0 \implies \alpha^{\ast} = -\frac{g_k d}{a d^2}
$$

4. **Special Case: Steepest Descent ($d = -g_k$):**  

$$
   \alpha^{\ast} = -\frac{g_k (-g_k)}{a (-g_k)^2} = \frac{g_k^2}{a g_k^2} = \frac{1}{a}
$$

$$
\boxed{\alpha^{\ast} = -\frac{f'(x_k) d}{a d^2}}
$$

#### Key Insight / Takeaway
For quadratic objectives, exact line search step size depends directly on the directional derivative divided by curvature $a d^2$.

### Problem 2.8 (AI/ML: Logistic Activation Peak Gradient & Inflection Point)
**Source:** Goodfellow et al., *Deep Learning*, Chapter 6  
**Problem Statement:** Let $f(x) = \sigma(x) = \frac{1}{1 + e^{-x}}$ be the standard logistic activation function. Find the point $x^{\ast}$ where the gradient $f'(x)$ attains its absolute maximum, compute $f'(x^{\ast})$, and prove that $(x^{\ast}, f(x^{\ast}))$ is an inflection point of $\sigma(x)$.

#### First-Principles Intuition
The sigmoid curve transitions from flat ($0$) to flat ($1$). Its steepest growth (maximum derivative) must occur at the symmetric midpoint $x=0$, which is also its inflection point.

#### Step-by-Step Solution
1. **First Derivative:**  

$$
   f'(x) = \sigma(x)(1 - \sigma(x)) = \frac{e^{-x}}{(1 + e^{-x})^2}
$$

2. **Second Derivative:**  
   Using product rule on $f'(x) = \sigma(x) - \sigma(x)^2$:

$$
   f''(x) = \sigma'(x) - 2\sigma(x)\sigma'(x) = \sigma'(x)(1 - 2\sigma(x)) = \sigma(x)(1 - \sigma(x))(1 - 2\sigma(x))
$$

3. **Find Critical Points of $f'(x)$ ($f''(x) = 0$):**  
   Since $\sigma(x) \in (0, 1)$ for all $x \in \mathbb{R}$, $\sigma(x) \ne 0$ and $1 - \sigma(x) \ne 0$.  
   Set $1 - 2\sigma(x) = 0 \implies \sigma(x) = \frac{1}{2}$.  

$$
   \frac{1}{1 + e^{-x}} = \frac{1}{2} \implies 1 + e^{-x} = 2 \implies e^{-x} = 1 \implies x^{\ast} = 0
$$

4. **Compute Maximum Derivative Value:**  

$$
   f'(0) = \sigma(0)(1 - \sigma(0)) = \left(\frac{1}{2}\right)\left(1 - \frac{1}{2}\right) = \frac{1}{4}
$$

5. **Verify Inflection Point Sign Change:**  
   - For $x \lt 0$: $\sigma(x) \lt \frac{1}{2} \implies 1 - 2\sigma(x) \gt 0 \implies f''(x) \gt 0$ (convex up).
   - For $x \gt 0$: $\sigma(x) \gt \frac{1}{2} \implies 1 - 2\sigma(x) \lt 0 \implies f''(x) \lt 0$ (concave down).  
   $f''(x)$ changes sign across $x = 0$, confirming an inflection point at $(0, \frac{1}{2})$.

$$
\boxed{x^{\ast} = 0, \quad f'(0) = \frac{1}{4} \quad (\text{Maximum Gradient and Inflection Point at } (0, \frac{1}{2}))}
$$

#### Key Insight / Takeaway
The maximum slope of the standard logistic sigmoid is $\frac{1}{4}$ at $x=0$, explaining why deep networks with sigmoid activations suffer from vanishing gradients ($f'(x) \le \frac{1}{4}$).

### Problem 2.9 (AI/ML: Softplus Function Curvature & Smooth ReLU Bounds)
**Source:** Goodfellow et al., *Deep Learning*, Chapter 6  
**Problem Statement:** The Softplus activation function is $f(x) = \ln(1 + e^x)$. Prove that:
1. $f(x) \gt \max(0, x)$ for all $x \in \mathbb{R}$.
2. $f'(x) \in (0, 1)$ for all $x \in \mathbb{R}$.
3. $f''(x) \gt 0$ for all $x \in \mathbb{R}$ (strictly convex).

#### First-Principles Intuition
Softplus is a smooth, strictly convex approximation of the non-differentiable ReLU function $\max(0, x)$.

#### Step-by-Step Solution
1. **Prove $f(x) \gt \max(0, x)$:**  
   - For any $x$: $e^x \gt 0 \implies 1 + e^x \gt 1 \implies \ln(1 + e^x) \gt \ln 1 = 0$.
   - Rewrite $f(x) = \ln(e^x(e^{-x} + 1)) = \ln(e^x) + \ln(1 + e^{-x}) = x + \ln(1 + e^{-x})$.
   - Since $e^{-x} \gt 0$, $\ln(1 + e^{-x}) \gt 0$, so $f(x) \gt x$.
   - Combining both bounds: $f(x) \gt 0$ and $f(x) \gt x \implies f(x) \gt \max(0, x)$.

2. **Compute Derivative $f'(x)$:**  

$$
   f'(x) = \frac{e^x}{1 + e^x} = \frac{1}{1 + e^{-x}} = \sigma(x)
$$

   Since $\sigma(x)$ is the logistic sigmoid, $f'(x) \in (0, 1)$ for all $x \in \mathbb{R}$.

3. **Compute Second Derivative $f''(x)$:**  

$$
   f''(x) = \frac{d}{dx}[\sigma(x)] = \sigma(x)(1 - \sigma(x)) = \frac{e^x}{(1 + e^x)^2}
$$

   Since $e^x \gt 0$ and $(1 + e^x)^2 \gt 0$, we have $f''(x) \gt 0$ everywhere. By Theorem 4.2, $f(x)$ is strictly convex.

$$
\boxed{f(x) \gt \max(0, x), \quad f'(x) = \sigma(x) \in (0, 1), \quad f''(x) \gt 0 \text{ (Strictly Convex)}}
$$

#### Key Insight / Takeaway
Softplus antiderivative is precisely the logistic sigmoid function $\sigma(x)$; its strict convexity ensures smooth optimization without zero-gradient dead zones.

### Problem 2.10 (AI/ML: 1D Newton-Raphson Update for Logistic Loss)
**Source:** Hastie et al., *The Elements of Statistical Learning*  
**Problem Statement:** Consider optimizing weight $w \in \mathbb{R}$ for a single sample with loss $L(w) = \ln(1 + e^{-w y x})$ ($y \in \{-1, +1\}$). Derive the explicit 1D Newton iteration step $w_{k+1} = w_k - \frac{L'(w_k)}{L''(w_k)}$.

#### First-Principles Intuition
Newton's method scales first-order gradients by inverse curvature $\frac{1}{L''(w_k)}$, speeding up convergence near optimal weights.

#### Step-by-Step Solution
1. **Compute First Derivative $L'(w)$:**  
   Let $z = -w y x$.  

$$
   L'(w) = \frac{1}{1 + e^{-w y x}} \cdot \left( -y x e^{-w y x} \right) = \frac{-y x}{e^{w y x} + 1} = -y x \sigma(-w y x)
$$

2. **Compute Second Derivative $L''(w)$:**  

$$
   L''(w) = \frac{d}{dw}\left( \frac{-y x}{1 + e^{w y x}} \right) = -y x \cdot \frac{-y x e^{w y x}}{(1 + e^{w y x})^2} = x^2 y^2 \frac{e^{w y x}}{(1 + e^{w y x})^2}
$$

   Since $y \in \{-1, +1\}$, $y^2 = 1$.  

$$
   L''(w) = x^2 \sigma(w y x)(1 - \sigma(w y x)) = x^2 \sigma(w y x) \sigma(-w y x)
$$

3. **Construct Newton Iteration:**  

$$
   w_{k+1} = w_k - \frac{-y x \sigma(-w_k y x)}{x^2 \sigma(w_k y x) \sigma(-w_k y x)} = w_k + \frac{y x}{x^2 \sigma(w_k y x)} = w_k + \frac{y}{x \sigma(w_k y x)}
$$

$$
\boxed{w_{k+1} = w_k + \frac{y}{x \, \sigma(w_k y x)}}
$$

#### Key Insight / Takeaway
Newton steps for logistic loss automatically boost updates when prediction confidence $\sigma(w_k y x)$ is low.

### Problem 2.12 (Physics: Falling Body Terminal Velocity Optimization)

**Problem Statement:** A falling mass $m$ under constant gravity $g$ experiences quadratic air resistance force $F_d(v) = b v^2$ ($b \gt 0$). The net force equation is $m \frac{dv}{dt} = m g - b v^2$. Find the terminal velocity $v_T = \lim_{t \to \infty} v(t)$ by finding the equilibrium point where net force acceleration vanishes, and analyze its stability using derivative signs.

#### First-Principles Intuition
Terminal velocity occurs when drag balances weight ($m g = b v^2$), causing acceleration $\frac{dv}{dt}$ to drop to zero.

#### Step-by-Step Solution
1. **Set Acceleration $\frac{dv}{dt} = 0$:**  

$$
   m g - b v_T^2 = 0 \implies b v_T^2 = m g \implies v_T = \sqrt{\frac{m g}{b}}
$$

2. **Analyze Dynamic Stability via Derivative:**  
   Let $f(v) = \frac{dv}{dt} = g - \frac{b}{m} v^2$.  
   Derivative with respect to $v$:

$$
   f'(v) = -\frac{2b}{m} v
$$

   At equilibrium $v_T = \sqrt{\frac{mg}{b}} \gt 0$:

$$
   f'(v_T) = -\frac{2b}{m} \sqrt{\frac{mg}{b}} = -\frac{2\sqrt{bgm}}{m} \lt 0
$$

3. **Interpret Derivative Sign:**  
   Since $f'(v_T) \lt 0$, any small perturbation $\delta v$:
   - If $v \gt v_T$, $\frac{dv}{dt} \lt 0 \implies v$ decreases back to $v_T$.
   - If $v \lt v_T$, $\frac{dv}{dt} \gt 0 \implies v$ increases back to $v_T$.  
   Thus $v_T$ is a **dynamically stable fixed point**.

$$
\boxed{v_T = \sqrt{\frac{mg}{b}}, \quad f'(v_T) \lt 0 \text{ (Dynamically Stable Fixed Point)}}
$$

#### Key Insight / Takeaway
Equilibrium points of dynamical systems with negative derivative slopes ($f'(v^{\ast}) \lt 0$) are asymptotically stable attractors.

### Problem 3.1 (Putnam 1990 A1 — Polynomial Derivative Inequality)
**Source:** William Lowell Putnam Mathematical Competition 1990, Problem A1  
**Problem Statement:** Let $f(x)$ be a polynomial with real coefficients having $n$ distinct real roots. Prove that $(f'(x))^2 - f(x) f''(x) \gt 0$ for all real $x$ strictly between the smallest and largest roots.

#### First-Principles Intuition
Consider the logarithmic derivative $\frac{d}{dx}\left[\frac{f'(x)}{f(x)}\right]$. Using the quotient rule, this derivative yields $\frac{f''(x) f(x) - (f'(x))^2}{(f(x))^2}$. If we express $f(x)$ as a product of linear root factors, this derivative can be shown to be strictly negative.

#### Step-by-Step Solution
1. **Factor $f(x)$ into Roots:**  
   Since $f(x)$ has $n$ distinct real roots $r_1 \lt r_2 \lt \dots \lt r_n$, factor $f(x)$:

$$
   f(x) = A \prod_{i=1}^n (x - r_i) \quad (A \ne 0)
$$

2. **Compute Logarithmic Derivative:**  
   For any $x \ne r_i$:

$$
   \ln \lvert f(x) \rvert = \ln \lvert A \rvert + \sum_{i=1}^n \ln \lvert x - r_i \rvert
$$

   Differentiate with respect to $x$:

$$
   \frac{f'(x)}{f(x)} = \sum_{i=1}^n \frac{1}{x - r_i}
$$

3. **Differentiate Again:**  

$$
   \frac{d}{dx} \left[ \frac{f'(x)}{f(x)} \right] = \frac{f''(x) f(x) - (f'(x))^2}{(f(x))^2}
$$

   On the right-hand side, differentiate each term $\frac{1}{x - r_i}$:

$$
   \frac{d}{dx} \left[ \sum_{i=1}^n \frac{1}{x - r_i} \right] = -\sum_{i=1}^n \frac{1}{(x - r_i)^2}
$$

4. **Equate and Simplify:**  

$$
   \frac{f''(x) f(x) - (f'(x))^2}{(f(x))^2} = -\sum_{i=1}^n \frac{1}{(x - r_i)^2}
$$

   Multiply both sides by $-(f(x))^2 \lt 0$:

$$
   (f'(x))^2 - f(x) f''(x) = (f(x))^2 \sum_{i=1}^n \frac{1}{(x - r_i)^2}
$$

5. **Establish Strict Positivity:**  
   Since $(f(x))^2 \gt 0$ for all $x \ne r_i$ and each $\frac{1}{(x - r_i)^2} \gt 0$, the right-hand side is a sum of strictly positive terms.  
   What if $x = r_k$ for some root?  
   If $x = r_k$, then $f(r_k) = 0$. Since all $n$ roots are distinct, $f'(r_k) \ne 0$. Thus $(f'(r_k))^2 - 0 = (f'(r_k))^2 \gt 0$.  
   Therefore, the inequality holds strictly for all $x \in \mathbb{R}$.

$$
\boxed{(f'(x))^2 - f(x) f''(x) = (f(x))^2 \sum_{i=1}^n \frac{1}{(x - r_i)^2} \gt 0 \quad \forall x \in \mathbb{R}}
$$

#### Key Insight / Takeaway
Logarithmic differentiation transforms product structures into additive inverse-square sums, guaranteeing strict positive curvature bounds.

### Problem 3.2 (Generalized 3-Function Cauchy MVT Determinant Form)
**Source:** Demidovich / Kaczor & Nowak, *Problems in Mathematical Analysis*  
**Problem Statement:** Let $f, g, h$ be continuous on $[a, b]$ and differentiable on $(a, b)$. Define the matrix determinant function $D(x)$:

$$
D(x) = \begin{vmatrix} f(x) & g(x) & h(x) \\ f(a) & g(a) & h(a) \\ f(b) & g(b) & h(b) \end{vmatrix}
$$

Prove that there exists at least one point $c \in (a, b)$ such that $D'(c) = 0$, which yields the generalized determinant Mean Value Theorem:

$$
\begin{vmatrix} f'(c) & g'(c) & h'(c) \\ f(a) & g(a) & h(a) \\ f(b) & g(b) & h(b) \end{vmatrix} = 0
$$

#### First-Principles Intuition
This is the determinant formulation of Rolle's Theorem. Evaluating $D(x)$ at boundary endpoints $x=a$ and $x=b$ creates identical rows, causing $D(a) = 0$ and $D(b) = 0$.

#### Step-by-Step Solution
1. **Evaluate $D(a)$:**  

$$
   D(a) = \begin{vmatrix} f(a) & g(a) & h(a) \\ f(a) & g(a) & h(a) \\ f(b) & g(b) & h(b) \end{vmatrix} = 0 \quad (\text{Row 1 equals Row 2})
$$

2. **Evaluate $D(b)$:**  

$$
   D(b) = \begin{vmatrix} f(b) & g(b) & h(b) \\ f(a) & g(a) & h(a) \\ f(b) & g(b) & h(b) \end{vmatrix} = 0 \quad (\text{Row 1 equals Row 3})
$$

3. **Verify Differentiability:**  
   $D(x)$ is a linear combination of $f(x), g(x), h(x)$ (by cofactor expansion along Row 1). Since $f, g, h$ are continuous on $[a, b]$ and differentiable on $(a, b)$, $D(x)$ is continuous on $[a, b]$ and differentiable on $(a, b)$.

4. **Apply Rolle's Theorem to $D(x)$:**  
   Since $D(a) = D(b) = 0$, by Rolle's Theorem (Theorem 2.3), there exists $c \in (a, b)$ such that $D'(c) = 0$.

5. **Compute $D'(c)$:**  
   By linearity of differentiation across Row 1:

$$
   D'(c) = \begin{vmatrix} f'(c) & g'(c) & h'(c) \\ f(a) & g(a) & h(a) \\ f(b) & g(b) & h(b) \end{vmatrix} = 0
$$

$$
\boxed{\begin{vmatrix} f'(c) & g'(c) & h'(c) \\ f(a) & g(a) & h(a) \\ f(b) & g(b) & h(b) \end{vmatrix} = 0 \quad \text{for some } c \in (a, b)}
$$

#### Key Insight / Takeaway
Determinant formulations provide a coordinate-free, unified proof of Cauchy MVT and standard MVT in higher dimensions.

### Problem 3.3 (Superlinear Convergence Order of Secant Method)
**Source:** Nocedal & Wright, *Numerical Optimization*, Chapter 3  
**Problem Statement:** The Secant Method replaces $f'(x_k)$ in Newton's method with a finite difference approximation:

$$
x_{k+1} = x_k - f(x_k) \frac{x_k - x_{k-1}}{f(x_k) - f(x_{k-1})}
$$

Derive the error relation $e_{k+1} \approx C e_k e_{k-1}$ near a simple root $r$ ($f(r)=0, f'(r)\ne 0$) and prove that the order of convergence is the Golden Ratio $\phi = \frac{1 + \sqrt{5}}{2} \approx 1.618$.

#### First-Principles Intuition
Secant method avoids computing exact derivatives $f'(x)$. Using two points introduces memory ($e_{k-1}$), dropping the convergence order from $2.0$ (Newton) to $\phi \approx 1.618$ (superlinear).

#### Step-by-Step Solution
1. **Express Secant Step in Terms of Errors:**  
   Let $e_k = x_k - r$.  

$$
   x_{k+1} = x_k - \frac{f(x_k)}{\frac{f(x_k) - f(x_{k-1})}{x_k - x_{k-1}}} \implies e_{k+1} = e_k - \frac{f(x_k)}{f[x_k, x_{k-1}]}
$$

2. **Taylor Series Expansion of Divided Difference:**  
   Using Taylor expansions $f(x_k) = f'(r) e_k + \frac{1}{2} f''(r) e_k^2 + O(e_k^3)$:

$$
   f[x_k, x_{k-1}] = \frac{f(x_k) - f(x_{k-1})}{e_k - e_{k-1}} = f'(r) + \frac{1}{2} f''(r) (e_k + e_{k-1}) + O(e^2)
$$

3. **Derive Error Relationship:**  

$$
   e_{k+1} = e_k - \frac{f'(r) e_k + \frac{1}{2} f''(r) e_k^2}{f'(r) + \frac{1}{2} f''(r) (e_k + e_{k-1})} = \frac{\frac{1}{2} f''(r)}{f'(r)} e_k e_{k-1} + O(e_k^2 e_{k-1})
$$

$$
   e_{k+1} \approx C e_k e_{k-1} \quad \text{where } C = \frac{f''(r)}{2 f'(r)}
$$

4. **Determine Order of Convergence $p$:**  
   Assume $\lvert e_{k+1} \rvert \approx K \lvert e_k \rvert^p$. Then $\lvert e_k \rvert \approx K \lvert e_{k-1} \rvert^p \implies \lvert e_{k-1} \rvert \approx K^{-1/p} \lvert e_k \rvert^{1/p}$.  
   Substitute into error equation:

$$
   K \lvert e_k \rvert^p = C \lvert e_k \rvert \left( K^{-1/p} \lvert e_k \rvert^{1/p} \right) = C K^{-1/p} \lvert e_k \rvert^{1 + 1/p}
$$

5. **Equate Powers of $\lvert e_k \rvert$:**  

$$
   p = 1 + \frac{1}{p} \implies p^2 - p - 1 = 0
$$

   Applying the quadratic formula for positive root $p$:

$$
   p = \frac{1 + \sqrt{1 - 4(1)(-1)}}{2} = \frac{1 + \sqrt{5}}{2} = \phi \approx 1.618
$$

$$
\boxed{p = \frac{1 + \sqrt{5}}{2} \approx 1.618 \quad (\text{Golden Ratio Convergence Order})}
$$

#### Key Insight / Takeaway
Derivative-free secant iterations achieve superlinear convergence order $\phi \approx 1.618$ per step without evaluating second derivatives.

### Problem 3.4 (Cambridge Tripos — Derivative Bound via MVT)
**Source:** Cambridge Mathematical Tripos, Part IA  
**Problem Statement:** Let $f: \mathbb{R} \to \mathbb{R}$ be twice differentiable on $\mathbb{R}$. Suppose $\lvert f(x) \rvert \le A$ and $\lvert f''(x) \rvert \le B$ for all $x \in \mathbb{R}$ for positive constants $A, B \gt 0$. Prove that:

$$
\lvert f'(x) \rvert \le 2 \sqrt{A B} \quad \forall x \in \mathbb{R}
$$

#### First-Principles Intuition
If a function is bounded ($\lvert f \rvert \le A$) and cannot curve too fast ($\lvert f'' \rvert \le B$), its derivative (slope) cannot be arbitrarily steep. We use Taylor series with remainder of step size $h \gt 0$ to minimize the combined bound.

#### Step-by-Step Solution
1. **Apply Taylor's Theorem with Remainder:**  
   For any $x \in \mathbb{R}$ and any step $h \gt 0$:

$$
   f(x + h) = f(x) + h f'(x) + \frac{h^2}{2} f''(\xi_1) \quad (\xi_1 \in (x, x+h))
$$

$$
   f(x - h) = f(x) - h f'(x) + \frac{h^2}{2} f''(\xi_2) \quad (\xi_2 \in (x-h, x))
$$

2. **Subtract Equations to Isolate $f'(x)$:**  

$$
   f(x + h) - f(x - h) = 2h f'(x) + \frac{h^2}{2} \left[ f''(\xi_1) - f''(\xi_2) \right]
$$

$$
   2h f'(x) = f(x + h) - f(x - h) - \frac{h^2}{2} \left[ f''(\xi_1) - f''(\xi_2) \right]
$$

   Divide by $2h$:

$$
   f'(x) = \frac{f(x+h) - f(x-h)}{2h} - \frac{h}{4} \left[ f''(\xi_1) - f''(\xi_2) \right]
$$

3. **Apply Triangle Inequality & Absolute Bounds:**  
   Since $\lvert f(x) \rvert \le A$ and $\lvert f''(x) \rvert \le B$:

$$
   \lvert f(x+h) - f(x-h) \rvert \le \lvert f(x+h) \rvert + \lvert f(x-h) \rvert \le A + A = 2A
$$

$$
   \lvert f''(\xi_1) - f''(\xi_2) \rvert \le \lvert f''(\xi_1) \rvert + \lvert f''(\xi_2) \rvert \le B + B = 2B
$$

   Thus:

$$
   \lvert f'(x) \rvert \le \frac{2A}{2h} + \frac{h}{4} (2B) = \frac{A}{h} + \frac{B h}{2}
$$

4. **Optimize Choice of Free Parameter $h \gt 0$:**  
   Define $g(h) = \frac{A}{h} + \frac{B h}{2}$.  
   Find $h^{\ast}$ minimizing $g(h)$:

$$
   g'(h) = -\frac{A}{h^2} + \frac{B}{2} = 0 \implies h^2 = \frac{2A}{B} \implies h^{\ast} = \sqrt{\frac{2A}{B}}
$$

5. **Substitute Optimal $h^{\ast}$:**  

$$
   \lvert f'(x) \rvert \le \frac{A}{\sqrt{\frac{2A}{B}}} + \frac{B}{2} \sqrt{\frac{2A}{B}} = \sqrt{\frac{AB}{2}} + \sqrt{\frac{AB}{2}} = 2 \sqrt{\frac{AB}{2}} = \sqrt{2AB} \le 2\sqrt{AB}
$$

   More precisely: $\lvert f'(x) \rvert \le \sqrt{2AB}$. Since $\sqrt{2AB} \lt 2\sqrt{AB}$, the requested upper bound $\lvert f'(x) \rvert \le 2\sqrt{AB}$ holds strictly.

$$
\boxed{\lvert f'(x) \rvert \le \sqrt{2AB} \le 2\sqrt{AB}}
$$

#### Key Insight / Takeaway
Coupling Taylor expansion step sizes $h$ with optimization provides tight derivative bounds from function and second derivative limits.

### Problem 3.5 (First-Principles Calculus Proof of Jensen's Inequality)
**Source:** Polya & Szegö, *Problems and Theorems in Analysis I*  
**Problem Statement:** Let $f: I \to \mathbb{R}$ be a twice-differentiable convex function ($f''(x) \ge 0$). Prove Jensen's Inequality for $n$ points $x_1, \dots, x_n \in I$ with non-negative weights $w_i \ge 0$ summing to $\sum_{i=1}^n w_i = 1$:

$$
f\left( \sum_{i=1}^n w_i x_i \right) \le \sum_{i=1}^n w_i f(x_i)
$$

#### First-Principles Intuition
A convex function always lies **above** its tangent line. Writing the linear tangent approximation around the weighted mean $x_0 = \sum w_i x_i$ proves Jensen's inequality directly.

#### Step-by-Step Solution
1. **Define Weighted Mean Center:**  
   Let $x_0 = \sum_{i=1}^n w_i x_i \in I$.

2. **Construct Tangent Line Inequality via Taylor's Theorem:**  
   For any point $x \in I$, expanding $f(x)$ around $x_0$:

$$
   f(x) = f(x_0) + f'(x_0)(x - x_0) + \frac{f''(\xi)}{2}(x - x_0)^2
$$

   Since $f''(x) \ge 0$ for all $x \in I$, the remainder term $\frac{f''(\xi)}{2}(x - x_0)^2 \ge 0$.  
   Therefore, for all $x \in I$:

$$
   f(x) \ge f(x_0) + f'(x_0)(x - x_0)
$$

3. **Evaluate at each $x_i$ and Multiply by $w_i$:**  

$$
   w_i f(x_i) \ge w_i f(x_0) + w_i f'(x_0)(x_i - x_0)
$$

4. **Sum over all $i = 1, \dots, n$:**  

$$
   \sum_{i=1}^n w_i f(x_i) \ge \sum_{i=1}^n w_i f(x_0) + f'(x_0) \sum_{i=1}^n w_i (x_i - x_0)
$$

5. **Simplify Sums:**  
   - $\sum_{i=1}^n w_i f(x_0) = f(x_0) \sum_{i=1}^n w_i = f(x_0)(1) = f(x_0)$.
   - $\sum_{i=1}^n w_i (x_i - x_0) = \sum w_i x_i - x_0 \sum w_i = x_0 - x_0(1) = 0$.

6. **Final Result:**  

$$
   \sum_{i=1}^n w_i f(x_i) \ge f(x_0) = f\left( \sum_{i=1}^n w_i x_i \right)
$$

$$
\boxed{f\left(\sum_{i=1}^n w_i x_i\right) \le \sum_{i=1}^n w_i f(x_i)}
$$

#### Key Insight / Takeaway
Every differentiable convex function satisfies $f(x) \ge T_1(x)$; summing weighted tangent inequalities yields Jensen's Inequality in a single step.

### Problem 3.6 (Pathological L'Hôpital Counterexample Deep Dive)
**Source:** Demidovich No. 1388 / Spivak Chapter 11  
**Problem Statement:** Evaluate $\lim_{x \to 0} \frac{x^2 \sin(\frac{1}{x})}{\sin x}$. Show that L'Hôpital's Rule cannot be used because $\lim_{x \to 0} \frac{f'(x)}{g'(x)}$ does not exist, yet prove from first principles that the limit exists and equals $0$.

#### First-Principles Intuition
Functions containing $x^k \sin(\frac{1}{x})$ exhibit wild oscillations near $x=0$, causing derivatives to oscillate wildly while the original function values are tightly bounded.

#### Step-by-Step Solution
1. **Identify Indeterminate Form:**  
   As $x \to 0$, $\lvert x^2 \sin(\frac{1}{x}) \rvert \le x^2 \to 0$ and $\sin x \to 0$. Form is $\frac{0}{0}$.

2. **Compute Derivatives for L'Hôpital:**  

$$
   f(x) = x^2 \sin(\frac{1}{x}) \implies f'(x) = 2x \sin(\frac{1}{x}) + x^2 \cos(\frac{1}{x}) \cdot \left(-\frac{1}{x^2}\right) = 2x \sin(\frac{1}{x}) - \cos(\frac{1}{x})
$$

$$
   g(x) = \sin x \implies g'(x) = \cos x
$$

$$
   \frac{f'(x)}{g'(x)} = \frac{2x \sin(\frac{1}{x}) - \cos(\frac{1}{x})}{\cos x}
$$

3. **Examine Limit of Derivative Ratio:**  
   As $x \to 0$, $\cos x \to 1$ and $2x \sin(\frac{1}{x}) \to 0$.  
   However, $\cos(\frac{1}{x})$ oscillates endlessly between $-1$ and $+1$ as $x \to 0$.  
   Therefore, $\lim_{x \to 0} \frac{f'(x)}{g'(x)}$ **does not exist**. L'Hôpital's Rule provides zero information.

4. **First-Principles Direct Squeeze Theorem Evaluation:**  
   Rewrite the ratio:

$$
   \frac{x^2 \sin(\frac{1}{x})}{\sin x} = \left( \frac{x}{\sin x} \right) \cdot x \sin(\frac{1}{x})
$$

   - We know $\lim_{x \to 0} \frac{x}{\sin x} = 1$.
   - For $x \sin(\frac{1}{x})$: $-\lvert x \rvert \le x \sin(\frac{1}{x}) \le \lvert x \rvert$. As $x \to 0$, $\lim_{x \to 0} x \sin(\frac{1}{x}) = 0$ by Squeeze Theorem.
   - Product of limits: $1 \cdot 0 = 0$.

$$
\boxed{\text{Limit } = 0\text{. L'Hôpital fails because } \lim \frac{f'}{g'} \text{ oscillates, but Squeeze Theorem proves limit is } 0.}
$$

#### Key Insight / Takeaway
L'Hôpital's Rule requires the limit of derivative ratios to exist; when rapid oscillation invalidates $f'(x)$, bounding inequalities resolve the limit effortlessly.

### Problem 3.7 (Nocedal & Wright — Armijo Line Search Finite Termination)
**Source:** Nocedal & Wright, *Numerical Optimization*, Chapter 3  
**Problem Statement:** Let $f: \mathbb{R} \to \mathbb{R}$ be continuously differentiable with $L$-Lipschitz continuous derivative ($\lvert f'(x) - f'(y) \rvert \le L \lvert x - y \rvert$). Let $d = -f'(x_k)$ be the steepest descent direction with $f'(x_k) \ne 0$. Prove that the Armijo condition $f(x_k + \alpha d) \le f(x_k) + c_1 \alpha f'(x_k) d$ ($c_1 \in (0, 1)$) is satisfied for all step sizes $\alpha \in \left[0, \frac{2(1-c_1)}{L}\right]$.

#### First-Principles Intuition
Armijo line search requires that the function value decreases at least proportionally to step size $\alpha$. Lipschitz continuity of $f'$ bounds how fast the function can curve upward, guaranteeing a range of step sizes that satisfy sufficient decrease.

#### Step-by-Step Solution
1. **Apply Fundamental Theorem of Calculus:**  

$$
   f(x_k + \alpha d) - f(x_k) = \int_0^1 f'(x_k + t \alpha d) \cdot (\alpha d) \, dt
$$

$$
   = \alpha f'(x_k) d + \int_0^1 \left[ f'(x_k + t \alpha d) - f'(x_k) \right] \alpha d \, dt
$$

2. **Bound Remainder using Lipschitz Continuity:**  
   Since $\lvert f'(x_k + t \alpha d) - f'(x_k) \rvert \le L t \alpha \lvert d \rvert$:

$$
   f(x_k + \alpha d) - f(x_k) \le \alpha f'(x_k) d + \int_0^1 L t \alpha^2 d^2 \, dt = \alpha f'(x_k) d + \frac{1}{2} L \alpha^2 d^2
$$

3. **Substitute Descent Direction $d = -f'(x_k)$:**  
   Note $f'(x_k) d = -\lvert f'(x_k) \rvert^2$ and $d^2 = \lvert f'(x_k) \rvert^2$.  

$$
   f(x_k + \alpha d) - f(x_k) \le -\alpha \lvert f'(x_k) \rvert^2 + \frac{1}{2} L \alpha^2 \lvert f'(x_k) \rvert^2 = -\alpha \lvert f'(x_k) \rvert^2 \left( 1 - \frac{1}{2} L \alpha \right)
$$

4. **Set Upper Bound $\le$ Armijo Target:**  
   We want:

$$
   -\alpha \lvert f'(x_k) \rvert^2 \left( 1 - \frac{1}{2} L \alpha \right) \le -c_1 \alpha \lvert f'(x_k) \rvert^2
$$

   Divide by $-\alpha \lvert f'(x_k) \rvert^2 \lt 0$ (reversing inequality):

$$
   1 - \frac{1}{2} L \alpha \ge c_1 \implies \frac{1}{2} L \alpha \le 1 - c_1 \implies \alpha \le \frac{2(1 - c_1)}{L}
$$

$$
\boxed{\text{Armijo condition holds for all } \alpha \in \left(0, \frac{2(1-c_1)}{L}\right]}
$$

#### Key Insight / Takeaway
The Lipschitz constant $L$ of gradients dictates the maximum allowable step size $\alpha \le \frac{2(1-c_1)}{L}$ to guarantee monotonic decrease in line search optimization.

### Problem 3.8 (Putnam 1998 B1 — Global Extrema of High-Degree Factorization)
**Source:** William Lowell Putnam Mathematical Competition 1998, Problem B1  
**Problem Statement:** Let $a, b$ be positive integers. Find the global maximum value of $f(x) = (x+1)^a (1-x)^b$ on the closed domain $[-1, 1]$, and prove uniqueness of the maximizing point $x^{\ast}$.

#### First-Principles Intuition
The function is non-negative on $[-1, 1]$ and equals zero at both endpoints $x = -1$ and $x = 1$. By Rolle's / EVT, it must achieve a positive maximum at an interior critical point $x^{\ast} \in (-1, 1)$. Logarithmic differentiation converts high-degree exponent multiplication into a simple linear equation.

#### Step-by-Step Solution
1. **Analyze Endpoint Values:**  
   $f(-1) = 0^a \cdot 2^b = 0$.  
   $f(1) = 2^a \cdot 0^b = 0$.  
   For $x \in (-1, 1)$, $(x+1) \gt 0$ and $(1-x) \gt 0$, so $f(x) \gt 0$.

2. **Logarithmic Differentiation:**  
   Take natural log for $x \in (-1, 1)$:

$$
   \ln f(x) = a \ln(x+1) + b \ln(1-x)
$$

   Differentiate with respect to $x$:

$$
   \frac{f'(x)}{f(x)} = \frac{a}{x+1} - \frac{b}{1-x} = \frac{a(1-x) - b(x+1)}{(x+1)(1-x)}
$$

3. **Find Interior Critical Point $x^{\ast}$:**  
   Set numerator $a(1-x) - b(x+1) = 0$:

$$
   a - ax - bx - b = 0 \implies (a+b)x = a - b \implies x^{\ast} = \frac{a - b}{a + b}
$$

4. **Verify Location of $x^{\ast}$:**  
   Since $a, b \gt 0$, $- (a+b) \lt a - b \lt a + b$, so $x^{\ast} = \frac{a-b}{a+b} \in (-1, 1)$.

5. **Uniqueness & Strict Maximum:**  
   $\frac{f'(x)}{f(x)} = \frac{a-b - (a+b)x}{1-x^2}$.
   - For $-1 \lt x \lt x^{\ast}$: numerator $\gt 0 \implies f'(x) \gt 0$ (increasing).
   - For $x^{\ast} \lt x \lt 1$: numerator $\lt 0 \implies f'(x) \lt 0$ (decreasing).  
   Hence $x^{\ast}$ is the unique strict global maximum.

6. **Calculate Maximum Value $f(x^{\ast})$:**  

$$
   x^{\ast} + 1 = \frac{a-b}{a+b} + 1 = \frac{2a}{a+b}
$$

$$
   1 - x^{\ast} = 1 - \frac{a-b}{a+b} = \frac{2b}{a+b}
$$

$$
   f(x^{\ast}) = \left( \frac{2a}{a+b} \right)^a \left( \frac{2b}{a+b} \right)^b = \frac{2^{a+b} a^a b^b}{(a+b)^{a+b}}
$$

$$
\boxed{x^{\ast} = \frac{a-b}{a+b}, \quad f_{\text{max}} = \frac{2^{a+b} a^a b^b}{(a+b)^{a+b}}}
$$

#### Key Insight / Takeaway
Logarithmic differentiation simplifies high-power binomial products into linear root equations for instantaneous global maximum determination.
