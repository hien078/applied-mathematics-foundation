### Problem L0.1
**Source:** Spivak, *Calculus*, Chapter 3, Problem 1  
**Statement:** Determine whether each of the following mathematical relations defines a valid real function $y = f(x)$:  
(a) $R_1 = \{(x, y) \in \mathbb{R}^2 \mid x^2 + y^2 = 9\}$  
(b) $R_2 = \{(x, y) \in \mathbb{R}^2 \mid y = \sqrt{9 - x^2}\}$  

#### First-Principles Intuition
A binary relation $R \subseteq X \times Y$ is a function if and only if for every $x$ in the domain, there is *exactly one* corresponding output $y$. Geometrically, this corresponds to the Vertical Line Test: no vertical line can intersect the graph at more than one point.

#### Step-by-Step Solution
**(a)** Solving $x^2 + y^2 = 9$ for $y$ yields:

$$
y = \pm \sqrt{9 - x^2}
$$

For any $x \in (-3, 3)$, such as $x = 0$, we have $y = 3$ and $y = -3$. Since the single input $x = 0$ maps to two distinct outputs $y \in \{3, -3\}$, $R_1$ violates uniqueness and is **not** a function.

**(b)** In $R_2$, the principal square root symbol $\sqrt{\cdot}$ denotes the non-negative square root. For each $x \in [-3, 3]$, $y = \sqrt{9 - x^2} \ge 0$ is uniquely defined.  
Hence, $R_2$ is a valid function with domain $[-3, 3]$ and range $[0, 3]$.

$$
\boxed{\text{(a) Not a function; (b) Valid function with } \operatorname{Dom}(f) = [-3,3], \operatorname{Im}(f) = [0,3]}
$$

#### Key Insight / Takeaway
Implicit equations $F(x,y)=0$ often define multi-valued curves. A function requires selecting a single branch to ensure output uniqueness.

### Problem L0.2
**Source:** Apostol, *Calculus Vol I*, Chapter 1  
**Statement:** Explain the fundamental distinction between the **codomain** and the **range (image)** of a function $f: X \to Y$. Illustrate with $f: \mathbb{R} \to \mathbb{R}$ defined by $f(x) = x^2$.

#### First-Principles Intuition
The codomain is the target set assigned *a priori* in the function's declaration. The range (image) is the actual set of outputs produced when evaluating $f(x)$ across all $x \in X$.

#### Step-by-Step Solution
For $f: \mathbb{R} \to \mathbb{R}$ where $f(x) = x^2$:
- The **Domain** is $\operatorname{Dom}(f) = \mathbb{R}$.
- The **Codomain** is the specified target set $\operatorname{Cod}(f) = \mathbb{R}$.
- Since $x^2 \ge 0$ for all real $x$, and for every $y \ge 0$ there exists $x = \sqrt{y} \in \mathbb{R}$ such that $f(x) = y$, the **Range (Image)** is:

$$
  \operatorname{Im}(f) = [0, \infty)
$$

Since $\operatorname{Im}(f) = [0, \infty) \subsetneq \mathbb{R} = \operatorname{Cod}(f)$, $f$ is **not surjective** onto $\mathbb{R}$.

$$
\boxed{\text{Codomain } = \mathbb{R}, \quad \text{Range } = [0, \infty); \quad \text{Range } \subsetneq \text{Codomain} \implies \text{Not Surjective}}
$$

#### Key Insight / Takeaway
Surjectivity depends directly on the declared codomain. Redefining the function as $f: \mathbb{R} \to [0, \infty)$ makes it surjective.

### Problem L0.3
**Source:** Spivak, *Calculus*, Chapter 3  
**Statement:** Prove algebraically from first principles that if $f: \mathbb{R} \to \mathbb{R}$ is strictly increasing, then $f$ is injective (one-to-one).

#### First-Principles Intuition
Strict monotonicity ensures that moving right along the input axis strictly changes the output. Thus, two distinct inputs can never land on the same height.

#### Step-by-Step Solution

We are given that $f$ is strictly increasing:

$$
\forall x_1, x_2 \in \mathbb{R}, \quad x_1 \lt x_2 \implies f(x_1) \lt f(x_2)
$$

We must show $f$ is injective: $f(x_1) = f(x_2) \implies x_1 = x_2$.  
We prove the contrapositive: $x_1 \neq x_2 \implies f(x_1) \neq f(x_2)$.

Assume $x_1 \neq x_2$. By trichotomy of real numbers, either $x_1 \lt x_2$ or $x_1 \gt x_2$.
1. If $x_1 \lt x_2$, strict monotonicity implies $f(x_1) \lt f(x_2)$, so $f(x_1) \neq f(x_2)$.
2. If $x_1 \gt x_2$, strict monotonicity implies $f(x_1) \gt f(x_2)$, so $f(x_1) \neq f(x_2)$.

In both cases, $x_1 \neq x_2 \implies f(x_1) \neq f(x_2)$. Thus, $f$ is injective.

$$
\boxed{\text{Strict Monotonicity } \implies x_1 \neq x_2 \implies f(x_1) \neq f(x_2) \implies \text{Injectivity}}
$$

#### Key Insight / Takeaway
Strict monotonicity is a sufficient condition for injectivity on any domain.

### Problem L0.4
**Source:** Stewart, *Calculus ET*, Chapter 1  
**Statement:** Let $f_1, f_2$ be even functions and $g_1, g_2$ be odd functions defined on $\mathbb{R}$. Determine the parity of:  
(a) $h_1(x) = f_1(x) \cdot f_2(x)$  
(b) $h_2(x) = g_1(x) \cdot g_2(x)$  
(c) $h_3(x) = f_1(x) \cdot g_1(x)$  

#### First-Principles Intuition
Function multiplication under negation $x \to -x$ mirrors integer sign multiplication: $(-1) \cdot (-1) = 1$ and $(+1) \cdot (-1) = -1$.

#### Step-by-Step Solution
Recall definitions: $f_i(-x) = f_i(x)$ (even) and $g_i(-x) = -g_i(x)$ (odd).

**(a)** Evaluate $h_1(-x)$:

$$
h_1(-x) = f_1(-x) \cdot f_2(-x) = f_1(x) \cdot f_2(x) = h_1(x) \implies \text{Even}
$$

**(b)** Evaluate $h_2(-x)$:

$$
h_2(-x) = g_1(-x) \cdot g_2(-x) = (-g_1(x)) \cdot (-g_2(x)) = g_1(x) \cdot g_2(x) = h_2(x) \implies \text{Even}
$$

**(c)** Evaluate $h_3(-x)$:

$$
h_3(-x) = f_1(-x) \cdot g_1(-x) = f_1(x) \cdot (-g_1(x)) = - (f_1(x) \cdot g_1(x)) = -h_3(x) \implies \text{Odd}
$$

$$
\boxed{\text{(a) Even} \cdot \text{Even} = \text{Even}; \quad \text{(b) Odd} \cdot \text{Odd} = \text{Even}; \quad \text{(c) Even} \cdot \text{Odd} = \text{Odd}}
$$

#### Key Insight / Takeaway
The product of two functions of the same parity is even; the product of two functions of opposite parity is odd.

### Problem L0.5
**Source:** Demidovich, *Problems in Mathematical Analysis*, Chapter 1  
**Statement:** Prove that any constant function $f(x) = c$ ($c \in \mathbb{R}$) is periodic with period $T$ for every $T \gt 0$, but has no fundamental period $T_0$.

#### First-Principles Intuition
A fundamental period is defined as the *smallest positive* period $T_0 \gt 0$. If periods exist arbitrarily close to zero, no minimum positive period can exist.

#### Step-by-Step Solution
1. **Periodicity**: For any $T \gt 0$ and any $x \in \mathbb{R}$:

$$
   f(x + T) = c = f(x)
$$

   Therefore, every positive real number $T \in (0, \infty)$ is a valid period of $f(x) = c$.

2. **Absence of Fundamental Period**:  
   The fundamental period $T_0$ must satisfy:

$$
   T_0 = \min \{ T \in \mathbb{R} \mid T \gt 0 \text{ and } f(x+T) = f(x) \}
$$

   However, the set of periods is $(0, \infty)$. For any candidate $T_1 \gt 0$, we can choose $T_2 = \frac{T_1}{2} \gt 0$, which is also a period and strictly smaller than $T_1$.  
   Since $\inf(0, \infty) = 0 \notin (0, \infty)$, the minimum positive element does not exist.

$$
\boxed{\text{Every } T \gt 0 \text{ is a period, but } \inf(0,\infty) = 0 \notin (0,\infty) \implies \text{No fundamental period } T_0}
$$

#### Key Insight / Takeaway
A function can be periodic without possessing a fundamental period if its set of positive periods has no minimum element.

### Problem L0.6
**Source:** Stewart, *Calculus ET*, Chapter 1  
**Statement:** Let $f(x) = x^2 + 1$ and $g(x) = 2x - 3$. Compute $(f \circ g)(x)$ and $(g \circ f)(x)$, and state their domains.

#### First-Principles Intuition
Function composition is non-commutative in general: applying transformation $g$ then $f$ alters the state differently than applying $f$ then $g$.

#### Step-by-Step Solution
1. **Compute $(f \circ g)(x)$**:

$$
   \begin{aligned}
   (f \circ g)(x) = f(g(x)) &= f(2x - 3) \\
   &= (2x - 3)^2 + 1 \\
   &= 4x^2 - 12x + 9 + 1 = 4x^2 - 12x + 10
   \end{aligned}
$$

   Domain: $\operatorname{Dom}(g) = \mathbb{R}$ and $g(x) \in \operatorname{Dom}(f) = \mathbb{R} \implies \operatorname{Dom}(f \circ g) = \mathbb{R}$.

2. **Compute $(g \circ f)(x)$**:

$$
   \begin{aligned}
   (g \circ f)(x) = g(f(x)) &= g(x^2 + 1) \\
   &= 2(x^2 + 1) - 3 \\
   &= 2x^2 + 2 - 3 = 2x^2 - 1
   \end{aligned}
$$

   Domain: $\operatorname{Dom}(f) = \mathbb{R}$ and $f(x) \in \operatorname{Dom}(g) = \mathbb{R} \implies \operatorname{Dom}(g \circ f) = \mathbb{R}$.

Notice that $4x^2 - 12x + 10 \neq 2x^2 - 1$.

$$
\boxed{(f \circ g)(x) = 4x^2 - 12x + 10, \quad (g \circ f)(x) = 2x^2 - 1; \quad \operatorname{Dom} = \mathbb{R}}
$$

#### Key Insight / Takeaway
Function composition $f \circ g \neq g \circ f$ reflects operator non-commutativity.

### Problem L0.7
**Source:** Spivak, *Calculus*, Chapter 4  
**Statement:** Prove geometrically and algebraically why the graph of the inverse function $y = f^{-1}(x)$ is the reflection of $y = f(x)$ across the line $y = x$.

#### First-Principles Intuition
If a function transforms input point $a$ to output height $b$, its inverse transforms input height $b$ back to point $a$. Swapping coordinates $(a,b) \to (b,a)$ is a reflection across the line $y=x$.

#### Step-by-Step Solution
1. Let $(a, b)$ be a point on the graph of $y = f(x)$. By definition, $b = f(a)$.
2. Since $f$ is invertible, applying $f^{-1}$ to both sides yields $f^{-1}(b) = f^{-1}(f(a)) = a$.
3. Thus, the point $(b, a)$ lies on the graph of $y = f^{-1}(x)$.
4. The midpoint of segment joining $(a, b)$ and $(b, a)$ is:

$$
   M = \left( \frac{a+b}{2}, \frac{a+b}{2} \right)
$$

   Since its coordinates are equal, $M$ lies on the line $y = x$.
5. The slope of the segment joining $(a, b)$ and $(b, a)$ is:

$$
   m = \frac{a - b}{b - a} = -1 \quad (a \neq b)
$$

   Since the line $y = x$ has slope $m_0 = 1$, the product of slopes is $m \cdot m_0 = (-1)(1) = -1$, proving perpendicularity.  
   Therefore, $(b, a)$ is the orthogonal reflection of $(a, b)$ across $y = x$.

$$
\boxed{(a,b) \in \text{Graph}(f) \iff (b,a) \in \text{Graph}(f^{-1}); \quad \text{Midpoint on } y=x, \text{ Slope } = -1 \implies \text{Reflection across } y=x}
$$

#### Key Insight / Takeaway
Inverting a function is geometrically equivalent to interchanging the horizontal and vertical axes.

### Problem L0.8
**Source:** Apostol, *Calculus Vol I*, Chapter 1  
**Statement:** Define the signum function $\operatorname{sgn}(x)$. Determine its domain, range, and parity.

#### First-Principles Intuition
The signum function extracts only the algebraic sign of a real number, collapsing all positive numbers to $+1$, all negative numbers to $-1$, and zero to $0$.

#### Step-by-Step Solution
1. **Definition**:

$$
   \operatorname{sgn}(x) = \begin{cases} 1, & x \gt 0 \\ 0, & x = 0 \\ -1, & x \lt 0 \end{cases}
$$

2. **Domain**: Evaluated for any real number $x \in \mathbb{R} \implies \operatorname{Dom}(\operatorname{sgn}) = \mathbb{R}$.
3. **Range**: The output values are strictly $\{ -1, 0, 1 \} \implies \operatorname{Im}(\operatorname{sgn}) = \{-1, 0, 1\}$.
4. **Parity**: Test $\operatorname{sgn}(-x)$:
   - For $x \gt 0 \implies -x \lt 0 \implies \operatorname{sgn}(-x) = -1 = -\operatorname{sgn}(x)$.
   - For $x = 0 \implies -x = 0 \implies \operatorname{sgn}(0) = 0 = -\operatorname{sgn}(0)$.
   - For $x \lt 0 \implies -x \gt 0 \implies \operatorname{sgn}(-x) = 1 = -\operatorname{sgn}(x)$.  
   In all cases, $\operatorname{sgn}(-x) = -\operatorname{sgn}(x)$. Thus, the signum function is **odd**.

$$
\boxed{\operatorname{Dom} = \mathbb{R}, \quad \operatorname{Range} = \{-1, 0, 1\}, \quad \text{Parity: Odd}}
$$

#### Key Insight / Takeaway
The signum function is an odd step function essential for absolute value derivatives: $\frac{d}{dx}\vert x \vert = \operatorname{sgn}(x)$ for $x \neq 0$.

### Problem L1.1
**Source:** Demidovich, *Problems in Mathematical Analysis*, No. 12  
**Statement:** Find the natural domain of the function:

$$
f(x) = \sqrt{\frac{x-1}{x+2}} + \ln(4 - x^2)
$$

#### First-Principles Intuition
The natural domain is the intersection of the domains of all sub-expressions: square roots require non-negative radicands, logarithms require positive arguments, and denominators must be non-zero.

#### Step-by-Step Solution
Let $f(x) = A(x) + B(x)$ where $A(x) = \sqrt{\frac{x-1}{x+2}}$ and $B(x) = \ln(4 - x^2)$.

1. **Condition for $A(x)$**:
   Radicand must be non-negative and denominator non-zero:

$$
   \frac{x-1}{x+2} \ge 0 \quad \text{and} \quad x+2 \neq 0
$$

   Testing intervals bounded by critical points $x = -2$ and $x = 1$:
   - $x \in (-\infty, -2)$: $( - ) / ( - ) \gt 0 \quad \checkmark$
   - $x \in (-2, 1)$: $( - ) / ( + ) \lt 0 \quad \times$
   - $x \in [1, \infty)$: $( + ) / ( + ) \ge 0 \quad \checkmark$  
   Thus, $\operatorname{Dom}(A) = (-\infty, -2) \cup [1, \infty)$.

2. **Condition for $B(x)$**:
   Argument of logarithm must be strictly positive:

$$
   4 - x^2 \gt 0 \iff x^2 \lt 4 \iff x \in (-2, 2)
$$

   Thus, $\operatorname{Dom}(B) = (-2, 2)$.

3. **Intersection of Domains**:

$$
   \operatorname{Dom}(f) = \operatorname{Dom}(A) \cap \operatorname{Dom}(B) = \Big( (-\infty, -2) \cup [1, \infty) \Big) \cap (-2, 2)
$$

   - $(-\infty, -2) \cap (-2, 2) = \emptyset$
   - $[1, \infty) \cap (-2, 2) = [1, 2)$

$$
\boxed{\operatorname{Dom}(f) = [1, 2)}
$$

#### Key Insight / Takeaway
Domain intersection eliminates candidate intervals that fail any single sub-expression requirement.

### Problem L1.2
**Source:** Stewart, *Calculus ET*, Chapter 1  
**Statement:** Determine the range of the function $f(x) = \frac{x^2 - 1}{x^2 + 1}$ for $x \in \mathbb{R}$.

#### First-Principles Intuition
We can solve for $x$ in terms of $y = f(x)$ and determine which $y$ values permit real solutions $x \in \mathbb{R}$, or rewrite the fraction algebraically.

#### Step-by-Step Solution
Rewrite $f(x)$ using the Pythagorean and double-angle identities:

$$
f(x) = \frac{x^2 + 1 - 2}{x^2 + 1} = 1 - \frac{2}{x^2 + 1}
$$

Analyze bounds on $x^2 + 1$ for $x \in \mathbb{R}$:
1. Since $x^2 \ge 0$, we have $x^2 + 1 \ge 1$, with minimum $1$ occurring at $x = 0$.
2. As $x \to \pm \infty$, $x^2 + 1 \to \infty$.  
Thus, the range of $g(x) = x^2 + 1$ is $[1, \infty)$.

Now apply transformations to the inequality $1 \le x^2 + 1 \lt \infty$:

$$
0 \lt \frac{1}{x^2 + 1} \le 1
$$

Multiply by $-2$ (reversing inequalities):

$$
-2 \le -\frac{2}{x^2 + 1} \lt 0
$$

Add $1$:

$$
1 - 2 \le 1 - \frac{2}{x^2 + 1} \lt 1 + 0 \implies -1 \le f(x) \lt 1
$$

Minimum value $f(0) = -1$ is attained. Upper bound $1$ is an asymptote as $x \to \pm\infty$.

$$
\boxed{\operatorname{Range}(f) = [-1, 1)}
$$

#### Key Insight / Takeaway
Algebraic rewriting $1 - \frac{c}{g(x)}$ simplifies range determination by isolating the variable into a single monotonic denominator.

### Problem L1.3
**Source:** Spivak, *Calculus*, Chapter 3, Problem 14  
**Statement:** Decompose the function $f(x) = e^x + \sin x + x^3 + x^2 + 5$ into its unique even part $f_E(x)$ and odd part $f_O(x)$.

#### First-Principles Intuition
Apply Theorem 3.1: $f_E(x) = \frac{f(x)+f(-x)}{2}$ and $f_O(x) = \frac{f(x)-f(-x)}{2}$.

#### Step-by-Step Solution
1. Compute $f(-x)$:

$$
   f(-x) = e^{-x} + \sin(-x) + (-x)^3 + (-x)^2 + 5 = e^{-x} - \sin x - x^3 + x^2 + 5
$$

2. Compute Even Component $f_E(x)$:

$$
   \begin{aligned}
   f_E(x) &= \frac{f(x) + f(-x)}{2} \\
   &= \frac{(e^x + \sin x + x^3 + x^2 + 5) + (e^{-x} - \sin x - x^3 + x^2 + 5)}{2} \\
   &= \frac{(e^x + e^{-x}) + 2x^2 + 10}{2} = \frac{e^x + e^{-x}}{2} + x^2 + 5 \\
   &= \cosh x + x^2 + 5
   \end{aligned}
$$

3. Compute Odd Component $f_O(x)$:

$$
   \begin{aligned}
   f_O(x) &= \frac{f(x) - f(-x)}{2} \\
   &= \frac{(e^x + \sin x + x^3 + x^2 + 5) - (e^{-x} - \sin x - x^3 + x^2 + 5)}{2} \\
   &= \frac{(e^x - e^{-x}) + 2\sin x + 2x^3}{2} = \frac{e^x - e^{-x}}{2} + \sin x + x^3 \\
   &= \sinh x + \sin x + x^3
   \end{aligned}
$$

$$
\boxed{f_E(x) = \cosh x + x^2 + 5, \quad f_O(x) = \sinh x + \sin x + x^3}
$$

#### Key Insight / Takeaway
Exponential functions decompose naturally into hyperbolic functions: $e^x = \cosh x + \sinh x$.

### Problem L1.4
**Source:** Apostol, *Calculus Vol I*, Chapter 1  
**Statement:** Prove algebraically from first principles (without using derivatives) that $f(x) = x^3 + 3x - 5$ is strictly increasing on $\mathbb{R}$.

#### First-Principles Intuition
We must show that for any $x_1 \lt x_2$, the difference $f(x_2) - f(x_1) \gt 0$.

#### Step-by-Step Solution
Let $x_1, x_2 \in \mathbb{R}$ with $x_1 \lt x_2$. Compute $f(x_2) - f(x_1)$:

$$
\begin{aligned}
f(x_2) - f(x_1) &= (x_2^3 + 3x_2 - 5) - (x_1^3 + 3x_1 - 5) \\
&= (x_2^3 - x_1^3) + 3(x_2 - x_1)
\end{aligned}
$$

Factor the difference of cubes:

$$
x_2^3 - x_1^3 = (x_2 - x_1)(x_2^2 + x_1 x_2 + x_1^2)
$$

Thus:

$$
f(x_2) - f(x_1) = (x_2 - x_1) \left[ x_2^2 + x_1 x_2 + x_1^2 + 3 \right]
$$

Analyze the quadratic term $Q(x_1, x_2) = x_2^2 + x_1 x_2 + x_1^2 + 3$ by completing the square in $x_2$:

$$
Q(x_1, x_2) = \left( x_2 + \frac{x_1}{2} \right)^2 + \frac{3x_1^2}{4} + 3
$$

Since squares of real numbers are non-negative:

$$
\left( x_2 + \frac{x_1}{2} \right)^2 \ge 0, \quad \frac{3x_1^2}{4} \ge 0 \implies Q(x_1, x_2) \ge 3 \gt 0
$$

Since $x_1 \lt x_2$, we have $(x_2 - x_1) \gt 0$.  
Therefore, the product $(x_2 - x_1) Q(x_1, x_2) \gt 0 \cdot 3 = 0$.  
Thus $f(x_2) - f(x_1) \gt 0 \implies f(x_1) \lt f(x_2)$, proving strict monotonicity on $\mathbb{R}$.

$$
\boxed{f(x_2) - f(x_1) = (x_2 - x_1)\left[ \left(x_2 + \frac{x_1}{2}\right)^2 + \frac{3x_1^2}{4} + 3 \right] \gt 0 \implies \text{Strictly Increasing}}
$$

#### Key Insight / Takeaway
Completing the square proves strict positivity of quadratic forms without relying on differential calculus.

### Problem L1.6
**Source:** Demidovich, *Problems in Mathematical Analysis*, No. 45  
**Statement:** Find the inverse function $f^{-1}(x)$ for $f(x) = \frac{2x + 3}{5x - 4}$. State its domain and range.

#### First-Principles Intuition
A linear fractional (homographic) transformation is bijective away from its vertical asymptote. Its inverse is also a homographic transformation.

#### Step-by-Step Solution
1. **Domain and Range of $f$**:
   - Denominator $5x - 4 \neq 0 \implies \operatorname{Dom}(f) = \mathbb{R} \setminus \{\frac{4}{5}\}$.
   - As $x \to \pm \infty$, $f(x) \to \frac{2}{5} \implies \operatorname{Range}(f) = \mathbb{R} \setminus \{\frac{2}{5}\}$.

2. **Solve $y = f(x)$ for $x$**:

$$
   y = \frac{2x + 3}{5x - 4} \implies y(5x - 4) = 2x + 3
$$

   Expand and group terms containing $x$:

$$
   5xy - 4y = 2x + 3 \implies 5xy - 2x = 4y + 3 \implies x(5y - 2) = 4y + 3
$$

   Divide by $(5y - 2)$:

$$
   x = \frac{4y + 3}{5y - 2}
$$

3. **Formulate Inverse Function**:
   Swap variable labels $y \to x$:

$$

### Problem L1.7
**Source:** Spivak, *Calculus*, Chapter 4  
**Statement:** Determine the fundamental period $T_0$ of the function $f(x) = \sin^4 x + \cos^4 x$.

#### First-Principles Intuition
Applying trigonometric double-angle identities reduces higher powers to linear cosine terms, revealing the true fundamental period.

#### Step-by-Step Solution
Rewrite $f(x)$ using the Pythagorean and double-angle identities:

$$
\sin^4 x + \cos^4 x = (\sin^2 x + \cos^2 x)^2 - 2 \sin^2 x \cos^2 x
$$

Using the Pythagorean identity $\sin^2 x + \cos^2 x = 1$ and double-angle identity $\sin(2x) = 2\sin x \cos x$:

$$
\sin^4 x + \cos^4 x = 1^2 - 2 \left( \frac{\sin(2x)}{2} \right)^2 = 1 - \frac{1}{2} \sin^2(2x)
$$

Apply power-reduction identity $\sin^2(\theta) = \frac{1 - \cos(2\theta)}{2}$ for $\theta = 2x$:

$$
f(x) = 1 - \frac{1}{2} \left[ \frac{1 - \cos(4x)}{2} \right] = 1 - \frac{1}{4} + \frac{1}{4} \cos(4x) = \frac{3}{4} + \frac{1}{4} \cos(4x)
$$

The function $\cos(\omega x)$ has fundamental period $T_0 = \frac{2\pi}{\omega}$.  
Here $\omega = 4$, so:

$$
T_0 = \frac{2\pi}{4} = \frac{\pi}{2}
$$

$$
\boxed{f(x) = \frac{3}{4} + \frac{1}{4} \cos(4x) \implies \text{Fundamental Period } T_0 = \frac{\pi}{2}}
$$

#### Key Insight / Takeaway
Superposition of powers can hide smaller periods; identity reduction to $\cos(\omega x)$ exposes the fundamental frequency.

### Problem L1.8
**Source:** Apostol, *Calculus Vol I*, Chapter 1  
**Statement:** Derive from first principles the explicit logarithmic expression for $\operatorname{arcsinh}(x)$.

#### First-Principles Intuition
Set $y = \operatorname{arcsinh}(x)$, express $x = \sinh y$ in terms of exponentials, and solve the resulting quadratic equation in $e^y$.

#### Step-by-Step Solution
1. Let $y = \operatorname{arcsinh}(x)$. By definition:

$$
   x = \sinh y = \frac{e^y - e^{-y}}{2}

2. Multiply by $2 e^y$:

$$
   2x e^y = e^{2y} - 1 \implies e^{2y} - 2x e^y - 1 = 0
$$

3. Let $u = e^y \gt 0$. We obtain quadratic equation $u^2 - 2xu - 1 = 0$.
4. Solve for $u$ using the quadratic formula:

$$
   u = \frac{2x \pm \sqrt{4x^2 - 4(1)(-1)}}{2} = x \pm \sqrt{x^2 + 1}

6. Thus, we must select the positive root:

$$
   e^y = x + \sqrt{x^2 + 1}
$$

7. Taking the natural logarithm:

$$
   y = \ln\left( x + \sqrt{x^2 + 1} \right)
$$

$$
\boxed{\operatorname{arcsinh}(x) = \ln\left( x + \sqrt{x^2 + 1} \right) \quad \forall x \in \mathbb{R}}
$$

#### Key Insight / Takeaway
Inverse hyperbolic functions map naturally to logarithms of algebraic expressions via exponential quadratics.

### Problem L2.1
**Source:** Goodfellow et al., *Deep Learning*, Chapter 3 (Sigmoid Analysis)  
**Statement:** The standard Sigmoid activation function is defined by $\sigma(x) = \frac{1}{1 + e^{-x}}$.  
(a) Prove $\sigma(x)$ is strictly increasing on $\mathbb{R}$ and has range $(0, 1)$.  
(b) Prove the reflection identity $\sigma(-x) = 1 - \sigma(x)$.  
(c) Decompose $\sigma(x) - \frac{1}{2}$ into its odd parity representation.

#### First-Principles Intuition
The Sigmoid function squashes the real line into probability space $(0,1)$. Its centered form $\sigma(x) - \frac{1}{2}$ is an odd function (point-symmetric about $(0, \frac{1}{2})$).

#### Step-by-Step Solution
**(a) Monotonicity & Range**:
Let $x_1 \lt x_2$. Exponential is strictly increasing $\implies -x_1 \gt -x_2 \implies e^{-x_1} \gt e^{-x_2}$.  
Add $1$: $1 + e^{-x_1} \gt 1 + e^{-x_2} \gt 1$.  
Taking reciprocals (reversing inequality):

$$
\frac{1}{1 + e^{-x_1}} \lt \frac{1}{1 + e^{-x_2}} \implies \sigma(x_1) \lt \sigma(x_2)
$$

Thus $\sigma(x)$ is strictly increasing. As $x \to -\infty$, $e^{-x} \to \infty \implies \sigma(x) \to 0$. As $x \to \infty$, $e^{-x} \to 0 \implies \sigma(x) \to 1$. Range is $(0, 1)$.

**(b) Reflection Identity**:

$$
\sigma(-x) = \frac{1}{1 + e^x} = \frac{e^{-x}}{e^{-x}(1 + e^x)} = \frac{e^{-x}}{e^{-x} + 1} = \frac{(1 + e^{-x}) - 1}{1 + e^{-x}} = 1 - \frac{1}{1 + e^{-x}} = 1 - \sigma(x)
$$

**(c) Odd Function Decomposition**:
Define $g(x) = \sigma(x) - \frac{1}{2}$. Evaluate $g(-x)$:

$$
g(-x) = \sigma(-x) - \frac{1}{2} = (1 - \sigma(x)) - \frac{1}{2} = \frac{1}{2} - \sigma(x) = - \left( \sigma(x) - \frac{1}{2} \right) = -g(x)
$$

Expressing using hyperbolic tangent:

$$
\sigma(x) - \frac{1}{2} = \frac{1 - e^{-x}}{2(1 + e^{-x})} = \frac{e^{\frac{x}{2}} - e^{-\frac{x}{2}}}{2(e^{\frac{x}{2}} + e^{-\frac{x}{2}})} = \frac{1}{2} \tanh\left(\frac{x}{2}\right)
$$

$$
\boxed{\operatorname{Range} = (0,1); \quad \sigma(-x) = 1 - \sigma(x); \quad \sigma(x) - \frac{1}{2} = \frac{1}{2}\tanh\left(\frac{x}{2}\right) \quad (\text{Odd})}
$$

#### Key Insight / Takeaway
The Sigmoid function is a shifted and scaled hyperbolic tangent: $\sigma(x) = \frac{1}{2}\left[1 + \tanh(\frac{x}{2})\right]$.

### Problem L2.2
**Source:** Goodfellow et al., *Deep Learning*, Chapter 4 (Softplus Activation)  
**Statement:** The Softplus activation function is $S(x) = \ln(1 + e^x)$.  
(a) Prove $S(x) \gt \max(0, x)$ for all $x \in \mathbb{R}$.  
(b) Compute asymptotic limits of $S(x)$ as $x \to -\infty$ and $x \to +\infty$.  
(c) Derive the explicit inverse function $S^{-1}(y)$ and state its domain.

#### First-Principles Intuition
Softplus is a smooth lower-bounded approximation to the rectified linear unit ($\operatorname{ReLU}(x) = \max(0,x)$).

#### Step-by-Step Solution
**(a) Upper Bound Proof**:
- For any $x \in \mathbb{R}$, $e^x \gt 0 \implies 1 + e^x \gt 1 \implies \ln(1 + e^x) \gt \ln(1) = 0$.
- Also $1 + e^x \gt e^x \implies \ln(1 + e^x) \gt \ln(e^x) = x$.  
Combining both inequalities gives $S(x) \gt 0$ and $S(x) \gt x \implies S(x) \gt \max(0, x)$.

**(b) Asymptotic Behavior**:
- As $x \to -\infty$: $e^x \to 0 \implies S(x) \to \ln(1) = 0$.
- As $x \to +\infty$: $S(x) = \ln\left(e^x(1 + e^{-x})\right) = x + \ln(1 + e^{-x}) \to x + 0 = x$.

**(c) Compute Inverse $S^{-1}(y)$**:
Set $y = \ln(1 + e^x)$ for $y \gt 0$:

$$
e^y = 1 + e^x \implies e^x = e^y - 1
$$

Since $y \gt 0$, $e^y \gt 1 \implies e^y - 1 \gt 0$. Taking natural logarithm:

$$
x = \ln(e^y - 1)
$$

Swapping variables: $S^{-1}(y) = \ln(e^y - 1)$ with domain $(0, \infty)$.

$$
\boxed{S(x) \gt \max(0,x); \quad \lim_{x \to -\infty} S(x) = 0, \ \lim_{x \to \infty} (S(x) - x) = 0; \quad S^{-1}(y) = \ln(e^y - 1), \ y \in (0, \infty)}
$$

#### Key Insight / Takeaway
Softplus smoothly interpolates between $0$ and $x$, providing a strictly monotonic, infinitely differentiable alternative to ReLU.

### Problem L2.3
**Source:** Hendrycks & Gimpel (2016), *Gaussian Error Linear Units (GELUs)*

(a) Decompose $\operatorname{GELU}(x)$ into its even and odd components.  
(b) Evaluate $\lim_{x \to +\infty} \operatorname{GELU}(x)$ and $\lim_{x \to -\infty} \operatorname{GELU}(x)$.

#### First-Principles Intuition
GELU weights inputs by their probability under a Gaussian distribution, producing a smooth non-monotonic curve with a small negative dip.

#### Step-by-Step Solution
**(a) Even/Odd Decomposition**:
Expand GELU:

$$
\operatorname{GELU}(x) = \frac{x}{2} + \frac{x}{2} \operatorname{erf}\left(\frac{x}{\sqrt{2}}\right)
$$

Evaluate components under negation:
- Let $g_1(x) = \frac{x}{2}$. Then $g_1(-x) = -\frac{x}{2} = -g_1(x) \implies g_1$ is **odd**.
- Let $g_2(x) = \frac{x}{2} \operatorname{erf}\left(\frac{x}{\sqrt{2}}\right)$. Then:

  g_2(-x) = \frac{-x}{2} \operatorname{erf}\left(\frac{-x}{\sqrt{2}}\right) = \frac{-x}{2} \left[ -\operatorname{erf}\left(\frac{x}{\sqrt{2}}\right) \right] = \frac{x}{2} \operatorname{erf}\left(\frac{x}{\sqrt{2}}\right) = g_2(x) \implies g_2 \text{ is } \mathbf{even}.
$$

Thus, the unique decomposition is:

$$
f_E(x) = \frac{x}{2} \operatorname{erf}\left(\frac{x}{\sqrt{2}}\right), \quad f_O(x) = \frac{x}{2}
$$

**(b) Limits at Infinity**:
Recall $\lim_{z \to +\infty} \operatorname{erf}(z) = 1$ and $\lim_{z \to -\infty} \operatorname{erf}(z) = -1$.
- As $x \to +\infty$: $\operatorname{GELU}(x) \to \frac{x}{2}[1 + 1] = x$.
- As $x \to -\infty$: $\Phi(x) \sim \frac{1}{\vert x \vert\sqrt{2\pi}} e^{-\frac{x^2}{2}} \implies \operatorname{GELU}(x) \to 0$.

$$
\boxed{\operatorname{GELU}_E(x) = \frac{x}{2} \operatorname{erf}\left(\frac{x}{\sqrt{2}}\right), \quad \operatorname{GELU}_O(x) = \frac{x}{2}; \quad \lim_{x \to \infty} \operatorname{GELU}(x) = x, \ \lim_{x \to -\infty} \operatorname{GELU}(x) = 0}
$$

#### Key Insight / Takeaway
The even part of GELU contains the error function interaction, while the odd part is a simple linear scale $\frac{x}{2}$.

### Problem L2.4
**Source:** Dinh et al. (2016), *Density estimation using Real NVP* (Normalizing Flows)  
**Statement:** An affine coupling layer $f: \mathbb{R}^2 \to \mathbb{R}^2$ transforms input vector $(x_1, x_2)$ into output vector $(y_1, y_2)$ via arbitrary continuous functions $s, t: \mathbb{R} \to \mathbb{R}$:

$$
y_1 = x_1, \quad y_2 = x_2 \exp(s(x_1)) + t(x_1)
$$

(a) Prove $f$ is bijective on $\mathbb{R}^2$.  
(b) Derive the explicit inverse transformation $(x_1, x_2) = f^{-1}(y_1, y_2)$.

#### First-Principles Intuition
Coupling layers leave a subset of features unchanged ($y_1 = x_1$), enabling direct inversion of remaining features regardless of the complexity of networks $s$ and $t$.

#### Step-by-Step Solution
**(a) Bijectivity Proof**:
1. **Injectivity**: Suppose $f(x_1, x_2) = f(x_1', x_2')$.
   - From $y_1 = y_1'$, we get $x_1 = x_1'$.
   - Substitute $x_1 = x_1'$ into $y_2 = y_2'$:

$$
     x_2 \exp(s(x_1)) + t(x_1) = x_2' \exp(s(x_1)) + t(x_1)
$$

     Subtract $t(x_1)$ and divide by $\exp(s(x_1)) \gt 0$:

$$
     x_2 = x_2'
$$

     Thus $(x_1, x_2) = (x_1', x_2')$, proving injectivity.

2. **Surjectivity**: For any target output $(y_1, y_2) \in \mathbb{R}^2$:
   - Set $x_1 = y_1$.
   - Solve $y_2 = x_2 \exp(s(y_1)) + t(y_1)$ for $x_2$:

$$
     x_2 = \frac{y_2 - t(y_1)}{\exp(s(y_1))} = (y_2 - t(y_1)) \exp(-s(y_1))
$$

   Since $x_1, x_2$ are uniquely determined for any $(y_1, y_2)$, $f$ is surjective and bijective.

**(b) Explicit Inverse**:

$$
f^{-1}(y_1, y_2) = \left( y_1, \ (y_2 - t(y_1)) \exp(-s(y_1)) \right)
$$

$$
\boxed{f^{-1}(y_1, y_2) = \Big( y_1, \ [y_2 - t(y_1)] e^{-s(y_1)} \Big)}
$$

#### Key Insight / Takeaway
Triangular coupling architectures guarantee exact invertibility even when $s(x)$ and $t(x)$ are non-invertible deep neural networks.

### Problem L2.6
**Source:** Griffiths, *Introduction to Quantum Mechanics*, Chapter 2  
**Statement:** In a 1D quantum harmonic oscillator, the stationary wavefunctions are $\psi_n(x) = A_n H_n(x) e^{-\frac{x^2}{2}}$, where $H_n(x)$ is the $n$-th Hermite polynomial satisfying $H_n(-x) = (-1)^n H_n(x)$.  
(a) Determine the parity of $\psi_n(x)$.  
(b) Prove that the probability density function $P_n(x) = \vert\psi_n(x)\vert^2$ is an even function for all $n \in \mathbb{N}$.

#### First-Principles Intuition
While quantum state vectors can have odd parity, physical observables like spatial probability densities must remain symmetric under spatial inversion.

#### Step-by-Step Solution
**(a) Parity of $\psi_n(x)$**:
Evaluate $\psi_n(-x)$:

$$
\begin{aligned}
\psi_n(-x) &= A_n H_n(-x) e^{-(-x)^2/2} \\
&= A_n \left( (-1)^n H_n(x) \right) e^{-\frac{x^2}{2}} \\
&= (-1)^n \left( A_n H_n(x) e^{-\frac{x^2}{2}} \right) = (-1)^n \psi_n(x)
\end{aligned}
$$

Thus, $\psi_n(x)$ is **even** if $n$ is even, and **odd** if $n$ is odd. Parity is $(-1)^n$.

**(b) Parity of Probability Density $P_n(x)$**:
Evaluate $P_n(-x)$:

$$
\begin{aligned}
P_n(-x) &= \vert\psi_n(-x)\vert^2 = \vert(-1)^n \psi_n(x)\vert^2 \\
&= \vert(-1)^n\vert^2 \cdot \vert\psi_n(x)\vert^2 = 1 \cdot P_n(x) = P_n(x)
\end{aligned}
$$

Since $P_n(-x) = P_n(x)$ for all $n$, the probability density is always an even function.

$$
\boxed{\text{Parity of } \psi_n(x) = (-1)^n; \quad P_n(-x) = \vert\psi_n(-x)\vert^2 = \vert\psi_n(x)\vert^2 = P_n(x) \implies \text{Always Even}}
$$

#### Key Insight / Takeaway
Squaring any odd or even real wavefunction yields a strictly even probability distribution.

### Problem L2.7

**Statement:** Consider the regularized linear regression loss function $L(w) = \frac{1}{N} \sum_{i=1}^N (w^T x_i - y_i)^2 + \lambda \|w\|_2^2$ for weight vector $w \in \mathbb{R}^d$.

#### Step-by-Step Solution
**(a) Case $y_i = 0$ for all $i$**:
The loss simplifies to:

$$
L(w) = \frac{1}{N} \sum_{i=1}^N (w^T x_i)^2 + \lambda \|w\|_2^2

$$
\begin{aligned}
L(-w) &= \frac{1}{N} \sum_{i=1}^N ((-w)^T x_i)^2 + \lambda \|-w\|_2^2 \\
&= \frac{1}{N} \sum_{i=1}^N (- (w^T x_i))^2 + \lambda \|w\|_2^2 \\
&= \frac{1}{N} \sum_{i=1}^N (w^T x_i)^2 + \lambda \|w\|_2^2 = L(w)
\end{aligned}
$$

Thus, $L(w)$ is an **even function** of weight vector $w$.

**(b) Case $y_i \neq 0$**:
Evaluate $L(-w)$:

$$
L(-w) = \frac{1}{N} \sum_{i=1}^N (-w^T x_i - y_i)^2 + \lambda \|w\|_2^2 = \frac{1}{N} \sum_{i=1}^N (w^T x_i + y_i)^2 + \lambda \|w\|_2^2
$$

In general $(w^T x_i + y_i)^2 \neq (w^T x_i - y_i)^2$ unless $y_i = 0$ or $w^T x_i = 0$.  
Furthermore $L(-w) \neq -L(w)$ since $L(w) \ge 0$.  
Thus $L(w)$ is **neither even nor odd** when $y_i \neq 0$.

$$
\boxed{\text{(a) Even function of } w \text{ when } y_i = 0; \quad \text{(b) Neither even nor odd when } y_i \neq 0}
$$

#### Key Insight / Takeaway
Zero-target loss landscapes possess reflection symmetry $L(w) = L(-w)$, creating dual global minima $\pm w^\ast$.

### Problem L2.8
**Source:** Taylor, *Classical Mechanics* (Relativistic Kinematics)  
**Statement:** The relativistic mass of a particle moving at speed $v$ is $m(v) = \frac{m_0}{\sqrt{1 - \frac{v^2}{c^2}}}$, where $m_0 \gt 0$ is rest mass and $c$ is the speed of light.  
(a) Determine natural domain, range, and monotonicity of $m(v)$ for $v \in [0, c)$.  
(b) Derive the inverse function $v(m)$ expressing velocity as a function of mass.

#### First-Principles Intuition
As velocity approaches the speed of light $c$, relativistic mass grows without bound, making light speed an unreachable asymptotic domain boundary for massive particles.

#### Step-by-Step Solution
**(a) Domain, Range & Monotonicity**:
- Domain: Radicand $1 - \frac{v^2}{c^2} \gt 0$ and $v \ge 0 \implies v \in [0, c)$.
- Monotonicity: For $0 \le v_1 \lt v_2 \lt c$:

$$
  v_1^2 \lt v_2^2 \implies 1 - \frac{v_1^2}{c^2} \gt 1 - \frac{v_2^2}{c^2} \implies \sqrt{1 - \frac{v_1^2}{c^2}} \gt \sqrt{1 - \frac{v_2^2}{c^2}}
$$

  Taking reciprocals: $m(v_1) \lt m(v_2)$. Thus $m(v)$ is **strictly increasing**.
- Range: At $v = 0$, $m(0) = m_0$. As $v \to c^-$, $m(v) \to \infty$. Range is $[m_0, \infty)$.

**(b) Derive Inverse $v(m)$**:
Set $m = \frac{m_0}{\sqrt{1 - \frac{v^2}{c^2}}}$ for $m \ge m_0$:

$$
\sqrt{1 - \frac{v^2}{c^2}} = \frac{m_0}{m} \implies 1 - \frac{v^2}{c^2} = \left(\frac{m_0}{m}\right)^2
$$

Solve for $v^2$:

$$
\frac{v^2}{c^2} = 1 - \left(\frac{m_0}{m}\right)^2 \implies v(m) = c \sqrt{1 - \left(\frac{m_0}{m}\right)^2}
$$

$$
\boxed{\operatorname{Dom} = [0, c), \ \operatorname{Range} = [m_0, \infty), \ \text{Strictly Increasing}; \quad v(m) = c \sqrt{1 - \left(\frac{m_0}{m}\right)^2}}
$$

#### Key Insight / Takeaway
The inverse function $v(m)$ maps required mass inflation back to physical particle velocity.

### Problem L2.9

**Statement:** The binary temperature Softmax function for logit $x$ and temperature $T \gt 0$ is $p_1(x; T) = \frac{e^{\frac{x}{T}}}{e^{\frac{x}{T}} + e^{-\frac{x}{T}}}$.

(b) Analyze the asymptotic behavior as $T \to 0^+$ (cold temperature) and $T \to \infty$ (high temperature).

#### First-Principles Intuition
Temperature scales logit differences: low temperatures turn probabilities into hard step decisions, while high temperatures flatten probabilities into uniform distributions.

$$
p_1(x; T) = \frac{1}{1 + e^{-2\frac{x}{T}}}
$$

$$
p_1(x; T) = \frac{1}{2} \left[ 1 + \tanh\left(\frac{x}{T}\right) \right]

2. **High Temperature Limit ($T \to \infty$)**:

$$
\boxed{p_1(x; T) = \frac{1}{2}\left[1 + \tanh\left(\frac{x}{T}\right)\right]; \quad \lim_{T \to 0^+} p_1 = \text{Step Function}, \ \lim_{T \to \infty} p_1 = \frac{1}{2} \ (\text{Uniform})}

Temperature parameterization smoothly interpolates between deterministic max selection ($T \to 0^+$) and maximum entropy noise ($T \to \infty$).

### Problem L3.1
**Source:** Cauchy (1821) / Spivak, *Calculus*, Chapter 4 (Cauchy Functional Equation)  
**Statement:** Let $f: \mathbb{R} \to \mathbb{R}$ be a continuous function satisfying Cauchy's functional equation:

$$
f(x + y) = f(x) + f(y) \quad \forall x, y \in \mathbb{R}
$$

Prove that $f(x) = cx$ for some constant $c \in \mathbb{R}$.

#### First-Principles Intuition
Additivity $f(x+y)=f(x)+f(y)$ propagates linearity from integers to rational numbers. Continuity then extends linearity across all real numbers.

#### Step-by-Step Solution
Let $c = f(1)$.

1. **Base Case $x = 0$ and Negatives**:
   - $f(0 + 0) = f(0) + f(0) \implies f(0) = 0$.
   - $f(x + (-x)) = f(x) + f(-x) = f(0) = 0 \implies f(-x) = -f(x)$ (Odd function).

2. **Induction for Positive Integers $\mathbb{Z}^+$**:
   - For $n = 1$, $f(1) = c \cdot 1$.
   - Assume $f(n) = cn$. Then $f(n+1) = f(n) + f(1) = cn + c = c(n+1)$.  
   By induction, $f(n) = cn$ for all $n \in \mathbb{N}$. Since $f$ is odd, $f(n) = cn$ for all $n \in \mathbb{Z}$.

3. **Extension to Rational Numbers $\mathbb{Q}$**:
   Let $q = \frac{m}{n} \in \mathbb{Q}$ with $m \in \mathbb{Z}, n \in \mathbb{Z}^+$.  
   Then:

$$
   f(m) = f\left( n \cdot \frac{m}{n} \right) = f\left( \underbrace{\frac{m}{n} + \dots + \frac{m}{n}}_{n \text{ times}} \right) = n f\left( \frac{m}{n} \right)
$$

   Since $f(m) = cm$, we have $cm = n f(\frac{m}{n}) \implies f\left( \frac{m}{n} \right) = c \frac{m}{n}$.  
   Thus $f(q) = cq$ for all $q \in \mathbb{Q}$.

4. **Extension to Real Numbers $\mathbb{R}$ via Continuity**:
   Let $x \in \mathbb{R}$. Since $\mathbb{Q}$ is dense in $\mathbb{R}$, there exists a sequence of rational numbers $\{q_k\} \subset \mathbb{Q}$ such that $\lim_{k \to \infty} q_k = x$.  
   By continuity of $f$:

$$
   f(x) = f\left( \lim_{k \to \infty} q_k \right) = \lim_{k \to \infty} f(q_k) = \lim_{k \to \infty} (c q_k) = c \lim_{k \to \infty} q_k = cx
$$

$$
\boxed{f(x) = cx \quad \text{where } c = f(1)}
$$

#### Key Insight / Takeaway
Density of rationals combined with continuity uniquely fixes functional behavior across all real numbers.

### Problem L3.2
**Source:** Functional-equation folklore (polynomial fixed-point argument); Spivak, *Calculus*, Chapter 4 style  
**Statement:** Find all polynomials $P(x)$ with real coefficients satisfying $P(0) = 0$ and:

$$
P(x^2 + 1) = (P(x))^2 + 1 \quad \forall x \in \mathbb{R}
$$

#### First-Principles Intuition
Degrees of polynomials must balance across identities: $\text{deg}(P(x^2+1)) = 2 \text{deg}(P)$, while $\text{deg}((P(x))^2+1) = 2 \text{deg}(P)$. Testing leading coefficients and evaluating small values pinpoints the polynomial.

#### Step-by-Step Solution
1. **Degree Analysis**:
   Let $n = \text{deg}(P)$.
   - $\text{deg}(P(x^2 + 1)) = 2n$.
   - $\text{deg}((P(x))^2 + 1) = 2n$.  
   Degrees match for any $n \ge 0$.

2. **Evaluate at $x = 0$**:
   Given $P(0) = 0$:

$$
   P(0^2 + 1) = (P(0))^2 + 1 \implies P(1) = 0^2 + 1 = 1
$$

3. **Evaluate at $x = 1$**:

$$
   P(1^2 + 1) = (P(1))^2 + 1 \implies P(2) = 1^2 + 1 = 2
$$

4. **Construct Sequence of Fixed Points**:
   Define sequence $a_0 = 0$, $a_{k+1} = a_k^2 + 1$.  
   Then $a_0 = 0, a_1 = 1, a_2 = 2, a_3 = 5, a_4 = 26, \dots$.  
   By induction, $P(a_k) = a_k$ for all $k \ge 0$.

5. **Uniqueness Argument**:
   Consider polynomial $Q(x) = P(x) - x$.  
   The roots of $Q(x)$ include all elements of the infinite sequence $\{a_k\}_{k=0}^\infty$.  
   Since a non-zero polynomial of finite degree cannot have infinitely many distinct roots, $Q(x)$ must be the zero polynomial:

$$
   Q(x) \equiv 0 \implies P(x) = x
$$

   Check: $P(x^2 + 1) = x^2 + 1$, and $(P(x))^2 + 1 = x^2 + 1 \quad \checkmark$.

$$
\boxed{P(x) = x}
$$

#### Key Insight / Takeaway
If a polynomial equation generates an infinite sequence of fixed points, the polynomial must equal the identity function.

### Problem L3.3
**Source:** Demidovich, *Problems in Mathematical Analysis*, No. 14 / Spivak Ch 4  
**Statement:** Prove that $f(x) = \sin(x) + \cos(\sqrt{2}x)$ is non-periodic on $\mathbb{R}$.

#### First-Principles Intuition
Apply Theorem 3.4: the ratio of fundamental periods $\frac{2\pi}{2\pi/\sqrt{2}} = \sqrt{2} \notin \mathbb{Q}$ is irrational, preventing synchronized periodic alignment.

#### Step-by-Step Solution
Assume by contradiction that $f(x)$ is periodic with period $T \gt 0$.  
Then $f(x + T) = f(x)$ for all $x \in \mathbb{R}$:

$$
\sin(x + T) + \cos(\sqrt{2}(x + T)) = \sin(x) + \cos(\sqrt{2}x)
$$

Evaluate at $x = 0$:

$$
\sin(T) + \cos(\sqrt{2}T) = 1 \quad \text{--- (Eq. 1)}
$$

Differentiate twice (or evaluate second derivative structure for smooth continuous periodic functions):
For any periodic smooth function $f(x+T) = f(x)$, its derivatives must also be periodic with period $T$.  
Taking second derivative $f''(x) = -\sin(x) - 2\cos(\sqrt{2}x)$:

$$
f''(0 + T) = f''(0) \implies -\sin(T) - 2\cos(\sqrt{2}T) = -2 \implies \sin(T) + 2\cos(\sqrt{2}T) = 2 \quad \text{--- (Eq. 2)}
$$

Subtract (Eq. 1) from (Eq. 2):

$$
\cos(\sqrt{2}T) = 1
$$

Substitute $\cos(\sqrt{2}T) = 1$ back into (Eq. 1):

$$
\sin(T) + 1 = 1 \implies \sin(T) = 0
$$

Now solve both conditions:
- $\sin(T) = 0 \implies T = k \pi$ for some integer $k \in \mathbb{Z}^+$.
- $\cos(\sqrt{2}T) = 1 \implies \sqrt{2}T = 2m \pi$ for some integer $m \in \mathbb{Z}^+ \implies T = \frac{2m\pi}{\sqrt{2}} = m \pi \sqrt{2}$.

Equating expressions for $T$:

$$
k \pi = m \pi \sqrt{2} \implies \sqrt{2} = \frac{k}{m} \quad (k, m \in \mathbb{Z}^+)
$$

This asserts that $\sqrt{2}$ is rational, contradicting the established irrationality of $\sqrt{2}$.  
Therefore, no such period $T \gt 0$ exists, and $f(x)$ is non-periodic.

$$
\boxed{\text{Periodicity } \implies \sqrt{2} = \frac{k}{m} \in \mathbb{Q} \quad (\text{Contradiction}) \implies \text{Non-Periodic}}
$$

#### Key Insight / Takeaway
Irrational frequency ratios create non-repeating continuous trajectories across the domain.

### Problem L3.4
**Source:** IMO Shortlist / Cambridge Tripos Part IA  
**Statement:** Find all continuous functions $f: (0, \infty) \to \mathbb{R} \setminus \{0\}$ satisfying both functional equations:  
(1) $f(f(x)) = x$ for all $x \gt 0$ (Involution) whenever $f(x) \gt 0$, and more generally $f$ is its own inverse on its domain.  
(2) $f\left(\frac{1}{x}\right) = \frac{1}{f(x)}$ for all $x \gt 0$.  
We restrict to $(0, \infty)$, an interval, because Theorem 3.2b (continuity + injectivity $\implies$ strict monotonicity) requires an interval domain; $\mathbb{R} \setminus \{0\}$ is not an interval and the same four solutions restricted to $x \lt 0$ must be checked separately.

Condition (1) means $f$ is its own inverse. Condition (2) means $f$ commutes with reciprocal inversion. Testing simple power forms $f(x) = x^a$ isolates valid continuous solutions.

#### Step-by-Step Solution
1. Evaluate $f(1)$:
   From (2) with $x = 1$: $f(1) = \frac{1}{f(1)} \implies (f(1))^2 = 1 \implies f(1) \in \{1, -1\}$.
   From (2) with $x = -1$: $f(-1) = \frac{1}{f(-1)} \implies f(-1) \in \{1, -1\}$.

2. Evaluate $f(x)$ for $x \gt 0$:
   Since $f$ is continuous on $(0, \infty)$ and $f(x) \neq 0$, by IVT $f(x)$ cannot change sign.
   - Case A: $f(x) \gt 0$ for all $x \gt 0$. Then $f(1) = 1$.
   - Case B: $f(x) \lt 0$ for all $x \gt 0$. Then $f(1) = -1$.

3. Candidate Functions:
   - Try $f_1(x) = x$. Check: $f(f(x)) = x \checkmark$, $f(\frac{1}{x}) = \frac{1}{x} = \frac{1}{f(x)} \checkmark$.
   - Try $f_2(x) = -x$. Check: $f(f(x)) = -(-x) = x \checkmark$, $f(\frac{1}{x}) = -\frac{1}{x} = \frac{1}{-x} = \frac{1}{f(x)} \checkmark$.
   - Try $f_3(x) = \frac{1}{x}$. Check: $f(f(x)) = 1/(\frac{1}{x}) = x \checkmark$, $f(\frac{1}{x}) = x = 1/(\frac{1}{x}) = \frac{1}{f(x)} \checkmark$.
   - Try $f_4(x) = -\frac{1}{x}$. Check: $f(f(x)) = -1/(-\frac{1}{x}) = x \checkmark$, $f(\frac{1}{x}) = -x = 1/(-\frac{1}{x}) = \frac{1}{f(x)} \checkmark$.

4. Prove these are the only continuous solutions:
   $f$ is injective (it is its own inverse), and $(0,\infty)$ is an interval, so by **Theorem 3.2b**
   (continuity + injectivity $\implies$ strict monotonicity, proved in `first_principles.ipynb`)
   $f$ is strictly monotonic on $(0,\infty)$. Checking the sign of each candidate's derivative:
   $\frac{d}{dx}(x) = 1 \gt 0$, $\frac{d}{dx}\!\left(-\frac{1}{x}\right) = \frac{1}{x^2} \gt 0$ (both increasing), while
   $\frac{d}{dx}(-x) = -1 \lt 0$, $\frac{d}{dx}\!\left(\frac{1}{x}\right) = -\frac{1}{x^2} \lt 0$ (both decreasing).
   - If strictly increasing: $f(x) = x$ or $f(x) = -\frac{1}{x}$.
   - If strictly decreasing: $f(x) = -x$ or $f(x) = \frac{1}{x}$.

$$
\boxed{f(x) \in \left\{ x, \ -x, \ \frac{1}{x}, \ -\frac{1}{x} \right\}}
$$

#### Key Insight / Takeaway
Continuous self-inverse functions commuting with inversion form a finite group of four basic rational transformations under composition.

### Problem L3.5
**Source:** Apostol, *Calculus Vol I* / Kaczor & Nowak, *Problems in Mathematical Analysis I*  
**Statement:** Let the Dirichlet function $D(x)$ be defined as:

$$
D(x) = \begin{cases} 1, & x \in \mathbb{Q} \\ 0, & x \in \mathbb{R} \setminus \mathbb{Q} \end{cases}
$$

(a) Prove every rational number $q \gt 0$ is a period of $D(x)$.  
(b) Prove $D(x)$ has no fundamental period $T_0$.  
(c) Prove $D(x)$ is nowhere monotonic on any open interval $(a, b)$.

#### First-Principles Intuition
The rational numbers $\mathbb{Q}$ and irrational numbers $\mathbb{R} \setminus \mathbb{Q}$ are both dense in $\mathbb{R}$. Any interval contains infinitely many of both.

#### Step-by-Step Solution
**(a) Rational Periodicity**:
Let $q \in \mathbb{Q}$ with $q \gt 0$.
- If $x \in \mathbb{Q}$, then $x + q \in \mathbb{Q} \implies D(x + q) = 1 = D(x)$.
- If $x \in \mathbb{R} \setminus \mathbb{Q}$, then $x + q \in \mathbb{R} \setminus \mathbb{Q}$ (since irrational + rational = irrational) $\implies D(x + q) = 0 = D(x)$.  
Thus $D(x + q) = D(x)$ for all $x \in \mathbb{R}$, proving $q$ is a period.

**(b) Absence of Fundamental Period**:
The set of positive periods is $P = \mathbb{Q} \cap (0, \infty)$.  
For any candidate period $q_1 \gt 0$, choose $q_2 = q_1 / 2 \in P$. Since $0 \lt q_2 \lt q_1$, no smallest positive rational exists. Thus $\inf(P) = 0 \notin P$, so no fundamental period $T_0$ exists.

**(c) Nowhere Monotonicity**:
Let $(a, b) \subset \mathbb{R}$ be any open interval. By density of $\mathbb{Q}$ and $\mathbb{R} \setminus \mathbb{Q}$:
- Select $q_1 \in \mathbb{Q} \cap (a, b) \implies D(q_1) = 1$.
- Select $r_1 \in (\mathbb{R} \setminus \mathbb{Q}) \cap (q_1, b) \implies D(r_1) = 0$.
- Select $q_2 \in \mathbb{Q} \cap (r_1, b) \implies D(q_2) = 1$.  
Here $q_1 \lt r_1 \lt q_2$, but $D(q_1) = 1 \gt D(r_1) = 0 \lt D(q_2) = 1$.  
This violates increasing ($1 \le 0$) and decreasing ($0 \ge 1$) conditions. Thus $D(x)$ is nowhere monotonic.

$$
\boxed{\forall q \in \mathbb{Q}^+, D(x+q)=D(x); \quad \inf(\mathbb{Q}^+) = 0 \implies \text{No } T_0; \quad \text{Dense oscillations } \implies \text{Nowhere Monotonic}}
$$

#### Key Insight / Takeaway
Density of sets creates functions with extreme local oscillations that fail standard calculus concepts like fundamental period or monotonicity.

### Problem L3.6
**Source:** Kaczor & Nowak, *Problems in Mathematical Analysis I*, Section 1.5  
**Statement:** Solve Jensen's functional equation for a continuous function $f: \mathbb{R} \to \mathbb{R}$:

$$
f\left(\frac{x + y}{2}\right) = \frac{f(x) + f(y)}{2} \quad \forall x, y \in \mathbb{R}
$$

#### First-Principles Intuition
Midpoint convexity paired with midpoint concavity forces the graph of $f(x)$ to be a straight line $f(x) = ax + b$.

#### Step-by-Step Solution
1. **Define Shifted Function**:
   Let $b = f(0)$. Define $g(x) = f(x) - b$.  
   Then $g(0) = f(0) - b = 0$.  
   Substitute $f(x) = g(x) + b$ into Jensen's equation:

$$
   g\left(\frac{x+y}{2}\right) + b = \frac{g(x) + b + g(y) + b}{2} = \frac{g(x) + g(y)}{2} + b
$$

   Subtract $b$:

$$
   g\left(\frac{x+y}{2}\right) = \frac{g(x) + g(y)}{2} \quad \text{--- (Eq. 1)}
$$

2. **Convert to Cauchy Equation**:
   Set $y = 0$ in (Eq. 1) using $g(0) = 0$:

$$
   g\left(\frac{x}{2}\right) = \frac{g(x) + 0}{2} = \frac{g(x)}{2} \implies g(x) = 2 g\left(\frac{x}{2}\right)
$$

   Substitute this identity into (Eq. 1):

$$
   g\left(\frac{x+y}{2}\right) = \frac{g(x+y)}{2}
$$

   Equating with (Eq. 1):

$$
   \frac{g(x+y)}{2} = \frac{g(x) + g(y)}{2} \implies g(x + y) = g(x) + g(y)
$$

3. **Apply Cauchy's Theorem (Problem L3.1)**:
   Since $f$ is continuous, $g(x) = f(x) - b$ is continuous and satisfies Cauchy's equation.  
   Therefore, $g(x) = ax$ for some constant $a = g(1) \in \mathbb{R}$.  
   Hence $f(x) = ax + b$.

$$
\boxed{f(x) = ax + b \quad (a, b \in \mathbb{R})}
$$

#### Key Insight / Takeaway
Midpoint linearity plus continuity implies global affine linearity.

### Problem L3.7
**Source:** Functional-equation folklore (d'Alembert-type cosine equation); Kaczor & Nowak, *Problems in Mathematical Analysis I*  
**Statement:** Let $f: \mathbb{R} \to \mathbb{R}$ be a continuous function such that:

$$
f(x + y) = f(x)f(y) - f(x - y) \quad \forall x, y \in \mathbb{R}
$$

If $f(0) \neq 0$, determine $f(x)$ for all $x$.

#### First-Principles Intuition
Product-to-sum identities like $\cos(x+y) + \cos(x-y) = 2\cos x \cos y$ mirror this functional structure.

#### Step-by-Step Solution
1. **Determine $f(0)$**:
   Set $x = 0, y = 0$:

$$
   f(0) = f(0)f(0) - f(0) = (f(0))^2 - f(0)
$$

   Since $f(0) \neq 0$, divide by $f(0)$:

$$
   1 = f(0) - 1 \implies f(0) = 2
$$

2. **Evaluate Parity**:
   Set $x = 0$ in the original equation:

$$
   f(y) = f(0)f(y) - f(-y) = 2f(y) - f(-y) \implies f(-y) = f(y)
$$

   Thus, $f(x)$ is an **even function**.

3. **Set $y = x$**:

$$
   f(2x) = (f(x))^2 - f(0) = (f(x))^2 - 2
$$

4. **Identify Candidate Function**:
   Recall double-angle identity for hyperbolic cosine: $\cosh(2\theta) = 2\cosh^2\theta - 1$, and for cosine: $2\cos(2\theta) = (2\cos\theta)^2 - 2$.  
   Let $f(x) = 2\cos(cx)$ or $f(x) = 2\cosh(cx)$.
   - Test $f(x) = 2\cos(cx)$:

$$
     f(x)f(y) - f(x-y) = 4\cos(cx)\cos(cy) - 2\cos(c(x-y))
$$

     Using product formula $2\cos A \cos B = \cos(A+B) + \cos(A-B)$:

$$
     = 2[\cos(c(x+y)) + \cos(c(x-y))] - 2\cos(c(x-y)) = 2\cos(c(x+y)) = f(x+y) \quad \checkmark
$$

   - Test $f(x) = 2\cosh(cx)$, using $2\cosh A \cosh B = \cosh(A+B) + \cosh(A-B)$:

$$
     f(x)f(y) - f(x-y) = 4\cosh(cx)\cosh(cy) - 2\cosh(c(x-y)) = 2[\cosh(c(x+y)) + \cosh(c(x-y))] - 2\cosh(c(x-y)) = 2\cosh(c(x+y)) = f(x+y) \quad \checkmark
$$

5. **Uniqueness (sketch)**: Setting $y=x$ gave $f(2x) = f(x)^2 - 2$, which together with $f(0)=2$ and
   continuity pins $f$ uniquely on the dyadic-rational multiples of any fixed $x_1$ once $f(x_1)$ is
   chosen, and continuity extends this to all of $\mathbb{R}$. Writing $f(x_1) = 2\cos(c x_1)$ when
   $\lvert f(x_1) \rvert \le 2$ (bounded case) or $f(x_1) = 2\cosh(c x_1)$ when $\lvert f(x_1) \rvert \gt 2$
   (unbounded case) reproduces exactly the doubling recursion $2\cos(2\theta) = (2\cos\theta)^2 - 2$ /
   $2\cosh(2\theta) = (2\cosh\theta)^2 - 2$, so no other continuous family satisfies the recursion.

$$
\boxed{f(x) = 2\cos(cx) \quad \text{or} \quad f(x) = 2\cosh(cx) \quad (c \in \mathbb{R})}
$$

#### Key Insight / Takeaway
Functional equations matching cosine product identities select trigonometric or hyperbolic bounds depending on growth constraints; boundedness of $f$ ($\lvert f\rvert \le 2$ vs. $\gt 2$) is exactly what decides trigonometric vs. hyperbolic.

### Problem L3.8
**Source:** Rudin, *Principles of Mathematical Analysis*, Chapter 4 / Polya & Szego  
**Statement:** Let $\{q_n\}_{n=1}^\infty$ be an enumeration of all rational numbers in $(0, 1)$. Define $f: [0, 1] \to \mathbb{R}$ by:

$$
f(x) = \sum_{n: q_n \lt x} \frac{1}{2^n}
$$

(a) Prove $f(x)$ is strictly increasing on $[0, 1]$.  
(b) Prove $f(x)$ is discontinuous at every rational point $q_k \in (0, 1)$.

#### First-Principles Intuition
We construct a strictly monotonic function that accumulates discrete jump steps of size $2^{-k}$ at each rational location $q_k$.

#### Step-by-Step Solution
**(a) Proof of Strict Monotonicity**:
Let $0 \le x_1 \lt x_2 \le 1$.  
By density of rationals, there exists at least one rational $q_m \in \mathbb{Q}$ such that $x_1 \lt q_m \lt x_2$.  
Compute difference $f(x_2) - f(x_1)$:

$$
f(x_2) - f(x_1) = \sum_{n: q_n \lt x_2} \frac{1}{2^n} - \sum_{n: q_n \lt x_1} \frac{1}{2^n} = \sum_{n: x_1 \le q_n \lt x_2} \frac{1}{2^n}
$$

Since $q_m$ satisfies $x_1 \le q_m \lt x_2$, the index $m$ is present in the sum:

$$
f(x_2) - f(x_1) \ge \frac{1}{2^m} \gt 0
$$

Thus $f(x_1) \lt f(x_2)$, proving $f$ is strictly increasing on $[0, 1]$.

**(b) Discontinuity at Rational Points**:
Let $q_k$ be any specific rational in $(0, 1)$.  
Calculate left-hand limit $f(q_k^-)$ and right-hand limit $f(q_k^+)$:
- As $x \to q_k^-$, only rationals strictly smaller than $q_k$ contribute:

$$
  f(q_k^-) = \sum_{n: q_n \lt q_k} \frac{1}{2^n}
$$

- For $x \ge q_k$, the term $q_k$ itself enters the sum:

$$
  f(q_k^+) = \sum_{n: q_n \le q_k} \frac{1}{2^n} = f(q_k^-) + \frac{1}{2^k}
$$

Compute jump discontinuity magnitude:

$$
\Delta f(q_k) = f(q_k^+) - f(q_k^-) = \frac{1}{2^k} \gt 0
$$

Since the left and right limits differ by $\frac{1}{2^k} \neq 0$, $f(x)$ has a jump discontinuity at every rational point $q_k \in (0, 1)$.

$$
\boxed{x_1 \lt x_2 \implies f(x_2) - f(x_1) \ge 2^{-m} \gt 0; \quad \text{Jump at } q_k \text{ equals } 2^{-k} \gt 0 \implies \text{Dense Discontinuities}}
$$

#### Key Insight / Takeaway
A monotonic function can have countably infinite jump discontinuities dense across an interval while maintaining strict monotonicity.

### Problem L3.9
**Source:** Cambridge Mathematical Tripos Part IA, Analysis I  
**Statement:** Find all continuous functions $f: (0, \infty) \to \mathbb{R}$ satisfying:

$$
f(xy) = x f(y) + y f(x) \quad \forall x, y \gt 0
$$

#### First-Principles Intuition
Divide by $xy$ to convert the product structure into an additive logarithmic functional equation.

#### Step-by-Step Solution
1. Divide both sides of $f(xy) = x f(y) + y f(x)$ by $xy \gt 0$:

$$
   \frac{f(xy)}{xy} = \frac{f(y)}{y} + \frac{f(x)}{x}
$$

2. Define auxiliary function $g(x) = \frac{f(x)}{x}$ for $x \gt 0$.  
   The equation becomes:

$$
   g(xy) = g(x) + g(y)
$$

3. Perform change of variables $x = e^u$ and $y = e^v$ for $u, v \in \mathbb{R}$.  
   Define $h(u) = g(e^u)$. Then:

$$
   h(u + v) = g(e^{u+v}) = g(e^u \cdot e^v) = g(e^u) + g(e^v) = h(u) + h(v)
$$

4. Apply Cauchy's Functional Equation (Problem L3.1):  
   Since $f$ is continuous, $h(u)$ is continuous on $\mathbb{R}$ and satisfies $h(u+v) = h(u)+h(v)$.  
   Therefore, $h(u) = c u$ for some constant $c \in \mathbb{R}$.
5. Revert variable substitutions:

$$
   g(x) = h(\ln x) = c \ln x
$$

   Substitute back $g(x) = \frac{f(x)}{x}$:

$$
   \frac{f(x)}{x} = c \ln x \implies f(x) = c x \ln x
$$

$$
\boxed{f(x) = c x \ln x \quad (c \in \mathbb{R})}
$$

#### Key Insight / Takeaway
Dividing by $xy$ transforms product-derivative structures into logarithmic Cauchy equations.

### Problem L3.10
**Source:** Functional-equation folklore (continuous involutions on $\mathbb{R}$); Rudin, *Principles of Mathematical Analysis*, Chapter 4 style  
**Statement:** Let $f: \mathbb{R} \to \mathbb{R}$ be a continuous function such that $f(f(x)) = x$ for all $x \in \mathbb{R}$. Prove that there exists at least one fixed point $x_0 \in \mathbb{R}$ such that $f(x_0) = x_0$, and show that if $f$ is not the identity function, it must be strictly decreasing.

#### First-Principles Intuition
Self-inverse continuous functions on $\mathbb{R}$ are homeomorphisms. They must be strictly monotonic. If strictly increasing, $f(x)=x$; if strictly decreasing, the Intermediate Value Theorem forces a unique intersection with $y=x$.

#### Step-by-Step Solution
1. **Injectivity & Monotonicity**:
   - $f(x_1) = f(x_2) \implies f(f(x_1)) = f(f(x_2)) \implies x_1 = x_2$, so $f$ is injective.
   - By **Theorem 3.2b** (the converse of Theorem 3.2), any continuous injective function on $\mathbb{R}$ is strictly monotonic.

2. **Case 1: $f$ is Strictly Increasing**:
   Suppose by contradiction that $\exists a \in \mathbb{R}$ such that $f(a) \neq a$.
   - If $f(a) \gt a$, strict monotonicity implies $f(f(a)) \gt f(a) \implies a \gt f(a)$, a contradiction.
   - If $f(a) \lt a$, strict monotonicity implies $f(f(a)) \lt f(a) \implies a \lt f(a)$, a contradiction.  
   Thus, if $f$ is strictly increasing, $f(x) = x$ for all $x \in \mathbb{R}$ (the identity function).

3. **Case 2: $f$ is Strictly Decreasing**:
   Assume $f$ is not the identity function, so $f$ must be strictly decreasing.
   - Define $g(x) = f(x) - x$.  
     Since $f$ is strictly decreasing, $g(x)$ is strictly decreasing on $\mathbb{R}$.
   - Pick any $a \in \mathbb{R}$. If $f(a) = a$, we are done.  
     If $f(a) \neq a$, let $b = f(a)$. Then $f(b) = f(f(a)) = a$.
   - Evaluate $g(a)$ and $g(b)$:

$$
     g(a) = f(a) - a = b - a
$$

$$
     g(b) = f(b) - b = a - b = -(b - a) = -g(a)
$$

   - Since $g(a)$ and $g(b)$ have opposite signs ($g(a) \cdot g(b) = -(b-a)^2 \lt 0$), by the **Intermediate Value Theorem (IVT)**, there exists $x_0$ between $a$ and $b$ such that $g(x_0) = 0 \implies f(x_0) = x_0$.
   - Uniqueness of $x_0$: Since $g(x) = f(x) - x$ is strictly decreasing, it can cross zero at most once. Thus $x_0$ is the unique fixed point.

$$
\boxed{f(f(x))=x \implies f \text{ is strictly monotonic}; \ f \text{ decreasing } \implies \exists! x_0 \text{ such that } f(x_0) = x_0 \text{ via IVT}}
$$

#### Key Insight / Takeaway
Topological continuity restricts continuous involutions on $\mathbb{R}$ to either the identity mapping or strictly decreasing functions with a single fixed point.
