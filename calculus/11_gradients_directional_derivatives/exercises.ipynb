### Problem L0.1 — Gradient Normal to a Level Curve
**Source:** Marsden & Tromba, *Vector Calculus* (Chapter 2)  
**Statement:** Explain why the gradient vector $\nabla f(\mathbf{x}_0)$ of a smooth scalar field $f: \mathbb{R}^2 \to \mathbb{R}$ is always perpendicular (orthogonal) to the level curve $f(x, y) = c$ passing through point $\mathbf{x}_0 = (x_0, y_0)$.

#### First-Principles Intuition
A level curve represents points of constant elevation $c$. If you move along a level curve, the function value does not change at all, meaning the rate of change in that direction is zero. Since the directional derivative in direction $\mathbf{u}$ is $D_{\mathbf{u}}f = \nabla f \cdot \mathbf{u}$, for the rate of change to be zero along the tangent direction of the level curve, the gradient $\nabla f$ must be perpendicular to that tangent vector.

#### Step-by-Step Solution
1. Let $\mathbf{r}(t) = (x(t), y(t))^T$ be any smooth parameterized curve lying entirely on the level curve $f(x, y) = c$, such that $\mathbf{r}(0) = \mathbf{x}_0 = (x_0, y_0)^T$.
2. Since $\mathbf{r}(t)$ lies on the level curve, $f(\mathbf{r}(t)) = c$ for all $t$ in a neighborhood of $0$.
3. Differentiating both sides with respect to $t$ at $t = 0$:

$$
   \frac{d}{dt} \Big\vert_{t=0} f(\mathbf{r}(t)) = \frac{d}{dt} \Big\vert_{t=0} c = 0
$$

4. Applying the multivariable chain rule to the left-hand side:

$$
   \frac{d}{dt} f(\mathbf{r}(t)) = \frac{\partial f}{\partial x}(\mathbf{r}(t)) \frac{dx}{dt}(t) + \frac{\partial f}{\partial y}(\mathbf{r}(t)) \frac{dy}{dt}(t) = \nabla f(\mathbf{r}(t)) \cdot \mathbf{r}'(t)
$$

5. Evaluating at $t = 0$:

$$
   \nabla f(\mathbf{x}_0) \cdot \mathbf{r}'(0) = 0
$$

6. Here $\mathbf{r}'(0)$ is a non-zero vector tangent to the level curve at $\mathbf{x}_0$. Since their inner product is zero, $\nabla f(\mathbf{x}_0)$ is orthogonal to $\mathbf{r}'(0)$ and thus perpendicular to the level curve.

$$
\text{Result: } \boxed{\nabla f(\mathbf{x}_0) \perp \mathbf{r}'(0) \quad \text{because } \nabla f(\mathbf{x}_0) \cdot \mathbf{r}'(0) = \frac{d}{dt}\Big\vert_{t=0} f(\mathbf{r}(t)) = 0}
$$

#### Key Insight / Takeaway
The gradient points directly across level sets (in the direction of maximum change), never along level sets (where change is zero).

### Problem L0.2 — Cauchy-Schwarz Bounds on the Directional Derivative
**Source:** Apostol, *Calculus, Volume II* (Chapter 8)  
**Statement:** Using the Cauchy-Schwarz inequality, prove that the directional derivative $D_{\mathbf{u}} f(\mathbf{x})$ of a differentiable scalar field is bounded by $-\|\nabla f(\mathbf{x})\| \le D_{\mathbf{u}} f(\mathbf{x}) \le \|\nabla f(\mathbf{x})\|$ for any unit vector $\|\mathbf{u}\| = 1$. Under what condition are the upper and lower bounds attained?

#### First-Principles Intuition
The directional derivative is the scalar projection of the gradient vector onto the direction vector $\mathbf{u}$. The maximum projection occurs when $\mathbf{u}$ aligns perfectly with the gradient, and the minimum (most negative) projection occurs when $\mathbf{u}$ points in the exact opposite direction.

#### Step-by-Step Solution
1. By Theorem 4.1, the master formula for directional derivatives of differentiable functions:

$$
   D_{\mathbf{u}} f(\mathbf{x}) = \nabla f(\mathbf{x}) \cdot \mathbf{u} = \langle \nabla f(\mathbf{x}), \mathbf{u} \rangle
$$

2. Apply the Cauchy-Schwarz inequality for the Euclidean inner product:

$$
   \lvert \langle \nabla f(\mathbf{x}), \mathbf{u} \rangle \rvert \le \|\nabla f(\mathbf{x})\| \|\mathbf{u}\|
$$

3. Since $\mathbf{u}$ is a unit vector, $\|\mathbf{u}\| = 1$. Therefore:

$$
   \lvert D_{\mathbf{u}} f(\mathbf{x}) \rvert \le \|\nabla f(\mathbf{x})\|
$$

4. Expanding the absolute value inequality yields:

$$
   -\|\nabla f(\mathbf{x})\| \le D_{\mathbf{u}} f(\mathbf{x}) \le \|\nabla f(\mathbf{x})\|
$$

5. Equality holds in Cauchy-Schwarz if and only if vectors $\nabla f(\mathbf{x})$ and $\mathbf{u}$ are linearly dependent (collinear).
   - Upper bound $+\|\nabla f(\mathbf{x})\|$ is attained when $\mathbf{u} = \frac{\nabla f(\mathbf{x})}{\|\nabla f(\mathbf{x})\|}$ (steepest ascent).
   - Lower bound $-\|\nabla f(\mathbf{x})\|$ is attained when $\mathbf{u} = -\frac{\nabla f(\mathbf{x})}{\|\nabla f(\mathbf{x})\|}$ (steepest descent).

$$
\text{Result: } \boxed{-\|\nabla f(\mathbf{x})\| \le D_{\mathbf{u}} f(\mathbf{x}) \le \|\nabla f(\mathbf{x})\|, \quad \text{attained when } \mathbf{u} = \pm \frac{\nabla f(\mathbf{x})}{\|\nabla f(\mathbf{x})\|}}
$$

#### Key Insight / Takeaway
The norm of the gradient vector $\|\nabla f(\mathbf{x})\|$ is the physical speed of maximum function increase, and $-\nabla f(\mathbf{x})$ gives the direction of steepest descent.

### Problem L0.3 — Partials Without Differentiability
**Source:** Spivak, *Calculus on Manifolds* (Chapter 2)  
**Statement:** Consider the function $f: \mathbb{R}^2 \to \mathbb{R}$ defined by:

$$
f(x, y) = \begin{cases} \frac{xy}{x^2 + y^2}, & (x, y) \neq (0, 0) \\ 0, & (x, y) = (0, 0) \end{cases}
$$

Determine whether partial derivatives $\frac{\partial f}{\partial x}(0, 0)$ and $\frac{\partial f}{\partial y}(0, 0)$ exist. Is $f$ differentiable at $(0,0)$? Does the formula $D_{\mathbf{u}} f(0,0) = \nabla f(0,0) \cdot \mathbf{u}$ hold?

#### First-Principles Intuition
Partial derivatives evaluate change strictly along the single line of a coordinate axis. Along the axes ($x=0$ or $y=0$), the function value is identically zero. However, approaching $(0,0)$ along diagonal lines $y = mx$ yields non-zero constant values, making the function discontinuous at the origin. Continuity is a necessary condition for differentiability.

#### Step-by-Step Solution
1. Compute $\frac{\partial f}{\partial x}(0,0)$ by definition:

$$
   \frac{\partial f}{\partial x}(0, 0) = \lim_{h \to 0} \frac{f(h, 0) - f(0, 0)}{h} = \lim_{h \to 0} \frac{0 - 0}{h} = 0
$$

2. Compute $\frac{\partial f}{\partial y}(0,0)$ by definition:

$$
   \frac{\partial f}{\partial y}(0, 0) = \lim_{k \to 0} \frac{f(0, k) - f(0, 0)}{k} = \lim_{k \to 0} \frac{0 - 0}{k} = 0
$$

   Both partial derivatives exist and equal $0$, so candidate gradient is $\nabla f(0,0) = (0,0)^T$.
3. Check continuity of $f(x,y)$ at $(0,0)$. Approach along path $y = mx$:

$$
   \lim_{x \to 0} f(x, mx) = \lim_{x \to 0} \frac{x(mx)}{x^2 + (mx)^2} = \lim_{x \to 0} \frac{m x^2}{x^2(1 + m^2)} = \frac{m}{1 + m^2}
$$

   The limit depends on slope $m$, so $\lim_{(x,y)\to(0,0)} f(x,y)$ does not exist. $f$ is discontinuous at $(0,0)$.
4. Since differentiability implies continuity, $f$ is **not differentiable** at $(0,0)$.
5. Compute directional derivative along $\mathbf{u} = (u_1, u_2)^T$ with $u_1, u_2 \neq 0$:

$$
   D_{\mathbf{u}} f(0,0) = \lim_{h \to 0} \frac{f(hu_1, hu_2) - 0}{h} = \lim_{h \to 0} \frac{\frac{h^2 u_1 u_2}{h^2(u_1^2 + u_2^2)}}{h} = \lim_{h \to 0} \frac{u_1 u_2}{h(u_1^2 + u_2^2)} = \pm \infty \ (\text{undefined})
$$

   Thus $D_{\mathbf{u}}f(0,0) \neq \nabla f(0,0) \cdot \mathbf{u} = 0$.

$$
\text{Result: } \boxed{\frac{\partial f}{\partial x}(0,0)=\frac{\partial f}{\partial y}(0,0)=0, \text{ but } f \text{ is NOT differentiable and master formula fails.}}
$$

#### Key Insight / Takeaway
The existence of partial derivatives is a weak property; Fréchet differentiability requires joint linear approximation in all directions simultaneously.

---

```python
# L0.3 - partials exist, f is discontinuous, D_u f blows up off the axes
def f_L03(x, y):
    return 0.0 if (x == 0.0 and y == 0.0) else x*y/(x*x + y*y)

h = 1e-6
fx = (f_L03(h, 0.0) - f_L03(0.0, 0.0))/h
fy = (f_L03(0.0, h) - f_L03(0.0, 0.0))/h
print("partial f/partial x (0,0) =", fx, " partial f/partial y (0,0) =", fy)
assert fx == 0.0 and fy == 0.0

for m in [0.5, 1.0, 2.0]:
    print(f"limit along y = {m}x  ->  {f_L03(1e-9, m*1e-9):.6f}   (theory m/(1+m^2) = {m/(1+m*m):.6f})")

u = np.array([1.0, 1.0])/np.sqrt(2)
for hh in [1e-2, 1e-4, 1e-6]:
    print(f"difference quotient along u=(1,1)/sqrt(2), h={hh:g}: {f_L03(hh*u[0], hh*u[1])/hh:.4g}")
print("-> the quotient diverges like 1/(2h), so D_u f(0,0) does not exist off the axes")
```

### Problem L0.4 — Domain Gradient Versus Surface Normal
**Source:** Nocedal & Wright, *Numerical Optimization* (Chapter 2)  
**Statement:** Clarify the distinction between the 2D gradient vector $\nabla f(x_0, y_0) \in \mathbb{R}^2$ of a scalar field $z = f(x, y)$ and the 3D surface normal vector $\mathbf{n} \in \mathbb{R}^3$ to the graph surface $\mathcal{G} = \{(x, y, z) \mid z = f(x, y)\}$.

#### First-Principles Intuition
The domain of $f$ is the 2D plane $\mathbb{R}^2$, whereas the graph of $f$ is a 2D surface embedded in 3D space $\mathbb{R}^3$. The gradient lives in the domain $\mathbb{R}^2$, pointing in the direction of steepest ascent across the map. The normal vector lives in 3D space $\mathbb{R}^3$, standing perpendicular to the physical surface.

#### Step-by-Step Solution
1. The gradient vector:

$$
   \nabla f(x_0, y_0) = \begin{bmatrix} \frac{\partial f}{\partial x}(x_0, y_0) \\ \frac{\partial f}{\partial y}(x_0, y_0) \end{bmatrix}
$$

   is a **2D vector** in the domain space $\mathbb{R}^2$. It points perpendicular to the level curve $f(x, y) = c$.
2. Express the graph surface $\mathcal{G}$ as a 3D level surface of the scalar field $F(x, y, z) = f(x, y) - z = 0$.
3. The 3D normal vector to this level surface is given by the 3D gradient $\nabla F(x_0, y_0, z_0)$:

$$
   \mathbf{n} = \nabla F(x_0, y_0, z_0) = \begin{bmatrix} \frac{\partial F}{\partial x} \\ \frac{\partial F}{\partial y} \\ \frac{\partial F}{\partial z} \end{bmatrix} = \begin{bmatrix} \frac{\partial f}{\partial x}(x_0, y_0) \\ \frac{\partial f}{\partial y}(x_0, y_0) \\ -1 \end{bmatrix} \in \mathbb{R}^3
$$

4. Notice that the first two components of $\mathbf{n}$ match $\nabla f(x_0, y_0)$, while the third component $-1$ accounts for elevation change $z$.

$$
\text{Result: } \boxed{\nabla f \in \mathbb{R}^2 \text{ (domain vector } \perp \text{ level curve)}, \quad \mathbf{n} = \begin{bmatrix} \nabla f \\ -1 \end{bmatrix} \in \mathbb{R}^3 \text{ (surface normal } \perp \text{ graph surface)}}
$$

#### Key Insight / Takeaway
Gradient vectors live in input/domain space, while surface normal vectors live in ambient output/graph space.

### Problem L0.5 — Directional Derivatives at a Stationary Point
**Source:** Stewart, *Multivariable Calculus* (Chapter 14)  
**Statement:** Suppose $\mathbf{x}_0$ is a critical/stationary point of a smooth scalar field $f: \mathbb{R}^n \to \mathbb{R}$, meaning $\nabla f(\mathbf{x}_0) = \mathbf{0}$. What is the value of the directional derivative $D_{\mathbf{u}} f(\mathbf{x}_0)$ along any arbitrary unit vector $\mathbf{u}$? What does this mean geometrically for local tangent approximations?

#### First-Principles Intuition
At a peak, trough, or saddle point, the surface is locally flat (horizontal). Moving a tiny distance in any direction causes zero first-order change in elevation.

#### Step-by-Step Solution
1. Apply the master formula for directional derivatives:

$$
   D_{\mathbf{u}} f(\mathbf{x}_0) = \nabla f(\mathbf{x}_0) \cdot \mathbf{u}
$$

2. Since $\mathbf{x}_0$ is a stationary point, $\nabla f(\mathbf{x}_0) = \mathbf{0}$.
3. Substituting $\mathbf{0}$ into the inner product:

$$
   D_{\mathbf{u}} f(\mathbf{x}_0) = \mathbf{0} \cdot \mathbf{u} = 0 \quad \forall \|\mathbf{u}\| = 1
$$

4. Geometrically, the tangent hyperplane at $\mathbf{x}_0$ is completely horizontal ($\nabla f(\mathbf{x}_0) \cdot (\mathbf{x} - \mathbf{x}_0) = 0 \implies z = z_0$). First-order Taylor variation $\delta f = 0$ in all directions.

$$
\text{Result: } \boxed{D_{\mathbf{u}} f(\mathbf{x}_0) = 0 \quad \text{for all unit vectors } \mathbf{u}}
$$

#### Key Insight / Takeaway
At stationary points, first derivatives provide no directional preference; higher-order curvature (Hessian matrix $\nabla^2 f$) is required to classify the landscape.

### Problem L0.6 — Gradient Fields Are Irrotational
**Source:** Marsden & Tromba, *Vector Calculus* (Chapter 4)  
**Statement:** Prove that any smooth gradient vector field $\mathbf{F}(\mathbf{x}) = \nabla f(\mathbf{x})$ in $\mathbb{R}^3$ is irrotational, meaning its curl is identically zero ($\nabla \times (\nabla f) = \mathbf{0}$).

#### First-Principles Intuition
Clairaut's theorem guarantees that mixed second partial derivatives are equal ($f_{xy} = f_{yx}$). The curl measures microscopic rotational circulation. Since the components of a gradient vector field are pure partial derivatives, their cross-differences cancel out exactly.

#### Step-by-Step Solution
1. Let $f(x, y, z)$ be a $C^2$ scalar field. The gradient vector field is:

$$
   \mathbf{F} = \nabla f = \begin{bmatrix} \frac{\partial f}{\partial x} \\ \frac{\partial f}{\partial y} \\ \frac{\partial f}{\partial z} \end{bmatrix} = \begin{bmatrix} F_1 \\ F_2 \\ F_3 \end{bmatrix}
$$

2. By definition of the curl operator $\nabla \times \mathbf{F}$:

$$
   \nabla \times (\nabla f) = \begin{vmatrix} \mathbf{i} & \mathbf{j} & \mathbf{k} \\ \frac{\partial}{\partial x} & \frac{\partial}{\partial y} & \frac{\partial}{\partial z} \\ \frac{\partial f}{\partial x} & \frac{\partial f}{\partial y} & \frac{\partial f}{\partial z} \end{vmatrix} = \begin{bmatrix} \frac{\partial^2 f}{\partial y \partial z} - \frac{\partial^2 f}{\partial z \partial y} \\ \frac{\partial^2 f}{\partial z \partial x} - \frac{\partial^2 f}{\partial x \partial z} \\ \frac{\partial^2 f}{\partial x \partial y} - \frac{\partial^2 f}{\partial y \partial x} \end{bmatrix}
$$

3. By Clairaut's theorem on equality of mixed partials for $C^2$ functions, $\frac{\partial^2 f}{\partial y \partial z} = \frac{\partial^2 f}{\partial z \partial y}$, $\frac{\partial^2 f}{\partial z \partial x} = \frac{\partial^2 f}{\partial x \partial z}$, and $\frac{\partial^2 f}{\partial x \partial y} = \frac{\partial^2 f}{\partial y \partial x}$.
4. Therefore, every component of the curl vector evaluates to $0$.

$$
\text{Result: } \boxed{\nabla \times (\nabla f) = \mathbf{0} \quad (\text{Gradient fields are conservative and irrotational})}
$$

#### Key Insight / Takeaway
Gradient fields store conservative potential energy; work done along any closed loop in a gradient field is zero ($\oint \nabla f \cdot d\mathbf{r} = 0$).

### Problem L0.7 — Directional Derivative Along a Straight Line
**Source:** Apostol, *Calculus, Volume II* (Chapter 8)  
**Statement:** Let $f: \mathbb{R}^n \to \mathbb{R}$ be differentiable, and let $\mathbf{r}(t) = \mathbf{x}_0 + t \mathbf{u}$ be a straight line path parametrized by arc length $t \in \mathbb{R}$, where $\|\mathbf{u}\| = 1$. Express the single-variable derivative $\frac{d}{dt} f(\mathbf{r}(t))$ in terms of the directional derivative $D_{\mathbf{u}} f$.

#### First-Principles Intuition
Walking along a straight line at unit speed $t$ converts a multivariable field exploration into a 1D scalar function $g(t) = f(\mathbf{r}(t))$. The rate of change of $g(t)$ with respect to time $t$ is precisely the directional derivative along $\mathbf{u}$.

#### Step-by-Step Solution
1. Define 1D single-variable function $g(t) = f(\mathbf{x}_0 + t \mathbf{u})$.
2. Compute $g'(0)$ by limit definition of ordinary derivative:

$$
   g'(0) = \lim_{t \to 0} \frac{g(t) - g(0)}{t} = \lim_{t \to 0} \frac{f(\mathbf{x}_0 + t\mathbf{u}) - f(\mathbf{x}_0)}{t}
$$

3. By Definition 3.3 of `first_principles.ipynb`, this limit is precisely $D_{\mathbf{u}} f(\mathbf{x}_0)$.
4. Alternatively, using the chain rule:

$$
   \frac{d}{dt} f(\mathbf{r}(t)) = \nabla f(\mathbf{r}(t)) \cdot \mathbf{r}'(t)
$$

   Since $\mathbf{r}(t) = \mathbf{x}_0 + t\mathbf{u}$, the velocity vector is $\mathbf{r}'(t) = \mathbf{u}$.
5. Evaluating at $t=0$:

$$
   \frac{d}{dt}\Big\vert_{t=0} f(\mathbf{x}_0 + t\mathbf{u}) = \nabla f(\mathbf{x}_0) \cdot \mathbf{u} = D_{\mathbf{u}} f(\mathbf{x}_0)
$$


$$
\text{Result: } \boxed{\frac{d}{dt} f(\mathbf{x}_0 + t\mathbf{u}) = D_{\mathbf{u}} f(\mathbf{x}_0 + t\mathbf{u})}
$$

#### Key Insight / Takeaway
The directional derivative is simply the ordinary 1D derivative of a scalar field evaluated along a straight-line trajectory.

### Problem L0.8 — Gradient Descent as Forward Euler
**Source:** Goodfellow et al., *Deep Learning* (Chapter 4)  
**Statement:** Show that discrete gradient descent $\mathbf{x}_{k+1} = \mathbf{x}_k - \eta \nabla f(\mathbf{x}_k)$ is the Forward Euler numerical approximation of the continuous differential equation $\dot{\mathbf{x}}(t) = -\nabla f(\mathbf{x}(t))$ with step size $\Delta t = \eta$.

#### First-Principles Intuition
Continuous gradient flow describes a particle sliding down a smooth hill. Discrete gradient descent takes finite jump steps of size $\eta$ along the local steepest slope. As step size approaches zero ($\eta \to 0$), discrete steps converge to the smooth continuous trajectory.

#### Step-by-Step Solution
1. Consider the continuous autonomous initial value problem:

$$
   \frac{d\mathbf{x}}{dt}(t) = -\nabla f(\mathbf{x}(t)), \quad \mathbf{x}(0) = \mathbf{x}_0
$$

2. Approximate the continuous time derivative $\frac{d\mathbf{x}}{dt}$ at time $t_k = k \Delta t$ using forward finite difference:

$$
   \frac{d\mathbf{x}}{dt}(t_k) \approx \frac{\mathbf{x}(t_k + \Delta t) - \mathbf{x}(t_k)}{\Delta t} = \frac{\mathbf{x}_{k+1} - \mathbf{x}_k}{\Delta t}
$$

3. Substitute this approximation into the ODE:

$$
   \frac{\mathbf{x}_{k+1} - \mathbf{x}_k}{\Delta t} = -\nabla f(\mathbf{x}_k)
$$

4. Rearrange to isolate next state $\mathbf{x}_{k+1}$:

$$
   \mathbf{x}_{k+1} = \mathbf{x}_k - \Delta t \nabla f(\mathbf{x}_k)
$$

5. Identifying time step size $\Delta t$ as learning rate $\eta$ yields standard discrete gradient descent.

$$
\text{Result: } \boxed{\mathbf{x}_{k+1} = \mathbf{x}_k - \eta \nabla f(\mathbf{x}_k) \iff \text{Forward Euler discretization of } \dot{\mathbf{x}} = -\nabla f(\mathbf{x})}
$$

#### Key Insight / Takeaway
Gradient descent in ML is a numerical integration step for continuous gradient flow; if learning rate $\eta$ is too large, Euler's method becomes unstable and diverges.

### Problem L1.1 — Gradient of a Mixed Exponential Field
**Source:** Stewart, *Multivariable Calculus* (Chapter 14)  
**Statement:** Calculate the gradient vector field $\nabla f(x, y)$ of the scalar field $f(x, y) = x^3 y^2 - e^{2x} \sin(y)$, and evaluate $\nabla f(0, \frac{\pi}{2})$.

#### First-Principles Intuition
The gradient combines partial derivative rates of change along each coordinate axis into a single vector. We compute $\frac{\partial f}{\partial x}$ treating $y$ as constant, and $\frac{\partial f}{\partial y}$ treating $x$ as constant.

#### Step-by-Step Solution
1. Compute partial derivative with respect to $x$:

$$
   \frac{\partial f}{\partial x} = \frac{\partial}{\partial x}(x^3 y^2) - \frac{\partial}{\partial x}(e^{2x} \sin y) = 3x^2 y^2 - 2e^{2x} \sin y
$$

2. Compute partial derivative with respect to $y$:

$$
   \frac{\partial f}{\partial y} = \frac{\partial}{\partial y}(x^3 y^2) - \frac{\partial}{\partial y}(e^{2x} \sin y) = 2x^3 y - e^{2x} \cos y
$$

3. Form the 2D gradient vector field:

$$
   \nabla f(x, y) = \begin{bmatrix} 3x^2 y^2 - 2e^{2x} \sin y \\ 2x^3 y - e^{2x} \cos y \end{bmatrix}
$$

4. Evaluate at point $(x, y) = (0, \frac{\pi}{2})$:
   - $\frac{\partial f}{\partial x}(0, \frac{\pi}{2}) = 3(0)^2(\frac{\pi}{2})^2 - 2e^0 \sin(\frac{\pi}{2}) = 0 - 2(1)(1) = -2$
   - $\frac{\partial f}{\partial y}(0, \frac{\pi}{2}) = 2(0)^3(\frac{\pi}{2}) - e^0 \cos(\frac{\pi}{2}) = 0 - 1(0) = 0$
5. Thus:

$$
   \nabla f\left(0, \frac{\pi}{2}\right) = \begin{bmatrix} -2 \\ 0 \end{bmatrix}
$$

$$
\text{Result: } \boxed{\nabla f(x, y) = \begin{bmatrix} 3x^2 y^2 - 2e^{2x} \sin y \\ 2x^3 y - e^{2x} \cos y \end{bmatrix}, \quad \nabla f\left(0, \frac{\pi}{2}\right) = \begin{bmatrix} -2 \\ 0 \end{bmatrix}}
$$

#### Key Insight / Takeaway
At $(0, \frac{\pi}{2})$, moving in the positive $x$-direction decreases the function value at rate $-2$, while moving in the $y$-direction produces zero instantaneous change.

---

```python
# L1.1 - symbolic gradient of x^3 y^2 - e^{2x} sin y, evaluated at (0, pi/2)
x, y = sp.symbols("x y", real=True)
f = x**3*y**2 - sp.exp(2*x)*sp.sin(y)
grad = sp.Matrix([sp.diff(f, x), sp.diff(f, y)])
print("grad f(x,y) =", sp.simplify(grad.T))
val = sp.simplify(grad.subs({x: 0, y: sp.pi/2}))
print("grad f(0, pi/2) =", val.T)
assert val == sp.Matrix([-2, 0])
```

### Problem L1.2 — Directional Derivative After Normalising
**Source:** Marsden & Tromba, *Vector Calculus* (Chapter 2)  
**Statement:** Find the directional derivative of $f(x, y, z) = x^2 y z + 4xz^2$ at the point $P(1, -2, 1)$ in the direction of the vector $\mathbf{v} = 2\mathbf{i} - \mathbf{j} - 2\mathbf{k}$.

#### First-Principles Intuition
The directional derivative measures rate of change per unit distance. Since $\mathbf{v}$ has magnitude $\|\mathbf{v}\| \neq 1$, we must first normalize $\mathbf{v}$ into a unit vector $\mathbf{u} = \frac{\mathbf{v}}{\|\mathbf{v}\|}$ before taking the dot product with $\nabla f(P)$.

#### Step-by-Step Solution
1. Compute partial derivatives of $f(x, y, z) = x^2 y z + 4xz^2$:
   - $\frac{\partial f}{\partial x} = 2xyz + 4z^2$
   - $\frac{\partial f}{\partial y} = x^2 z$
   - $\frac{\partial f}{\partial z} = x^2 y + 8xz$
2. Evaluate gradient vector at $P(1, -2, 1)$:
   - $\frac{\partial f}{\partial x}(1, -2, 1) = 2(1)(-2)(1) + 4(1)^2 = -4 + 4 = 0$
   - $\frac{\partial f}{\partial y}(1, -2, 1) = (1)^2(1) = 1$
   - $\frac{\partial f}{\partial z}(1, -2, 1) = (1)^2(-2) + 8(1)(1) = -2 + 8 = 6$

$$
   \nabla f(1, -2, 1) = \begin{bmatrix} 0 \\ 1 \\ 6 \end{bmatrix}
$$

3. Normalize direction vector:

$$
   \mathbf{v} = \begin{bmatrix} 2 \\ -1 \\ -2 \end{bmatrix}
$$

$$
   \|\mathbf{v}\| = \sqrt{2^2 + (-1)^2 + (-2)^2} = \sqrt{4 + 1 + 4} = \sqrt{9} = 3
$$

$$
   \mathbf{u} = \frac{\mathbf{v}}{\|\mathbf{v}\|} = \begin{bmatrix} \frac{2}{3} \\ -\frac{1}{3} \\ -\frac{2}{3} \end{bmatrix}
$$

4. Calculate directional derivative via Master Formula:

$$
   D_{\mathbf{u}} f(P) = \nabla f(P) \cdot \mathbf{u} = \begin{bmatrix} 0 \\ 1 \\ 6 \end{bmatrix} \cdot \begin{bmatrix} \frac{2}{3} \\ -\frac{1}{3} \\ -\frac{2}{3} \end{bmatrix} = 0\left(\frac{2}{3}\right) + 1\left(-\frac{1}{3}\right) + 6\left(-\frac{2}{3}\right) = -\frac{1}{3} - 4 = -\frac{13}{3}
$$

$$
\text{Result: } \boxed{D_{\mathbf{u}} f(1, -2, 1) = -\frac{13}{3}}
$$

#### Key Insight / Takeaway
Failing to normalize $\mathbf{v}$ yields a scaled value ($3 \times$ larger); always normalize vectors to unit length when computing directional derivatives.

---

```python
# L1.2 - directional derivative of x^2 y z + 4 x z^2 at (1,-2,1) along (2,-1,-2)
x, y, z = sp.symbols("x y z", real=True)
f = x**2*y*z + 4*x*z**2
g = sp.Matrix([sp.diff(f, v) for v in (x, y, z)]).subs({x: 1, y: -2, z: 1})
v = sp.Matrix([2, -1, -2])
u = v/v.norm()
D = (g.T*u)[0]
print("grad f(1,-2,1) =", g.T, "  ||v|| =", v.norm(), "  u =", u.T)
print("D_u f =", D, "=", float(D))
assert sp.simplify(D - sp.Rational(-13, 3)) == 0
```

### Problem L1.3 — Steepest Ascent and the Rate Toward a Point
**Source:** Apostol, *Calculus, Volume II* (Chapter 8)  
**Statement:** For $f(x, y) = x e^{y} + y e^x$ at point $P(0, 0)$:
1. Find the direction of maximum rate of increase.
2. Find the maximum rate of increase.
3. Find the directional derivative in the direction pointing toward point $Q(3, 4)$.

#### First-Principles Intuition
Steepest ascent is governed by the direction of $\nabla f(0,0)$, and its magnitude $\|\nabla f(0,0)\|$ is the maximum rate. For any other target point $Q$, the unit direction vector is $\mathbf{u} = \frac{\vec{PQ}}{\|\vec{PQ}\|}$.

#### Step-by-Step Solution
1. Compute gradient of $f(x, y) = x e^y + y e^x$:

$$
   \nabla f(x, y) = \begin{bmatrix} e^y + y e^x \\ x e^y + e^x \end{bmatrix}
$$

2. Evaluate at $P(0, 0)$:

$$
   \nabla f(0, 0) = \begin{bmatrix} e^0 + 0 \\ 0 + e^0 \end{bmatrix} = \begin{bmatrix} 1 \\ 1 \end{bmatrix}
$$

3. **Direction of maximum rate of increase:**

$$
   \mathbf{u}_{\text{max}} = \frac{\nabla f(0,0)}{\|\nabla f(0,0)\|} = \frac{1}{\sqrt{1^2 + 1^2}} \begin{bmatrix} 1 \\ 1 \end{bmatrix} = \begin{bmatrix} \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{bmatrix}
$$

4. **Maximum rate of increase:**

$$
   \max D_{\mathbf{u}} f(0,0) = \|\nabla f(0,0)\| = \sqrt{1^2 + 1^2} = \sqrt{2}
$$

5. **Directional derivative toward $Q(3, 4)$:**
   Displacement vector $\vec{PQ} = Q - P = (3-0, 4-0)^T = (3, 4)^T$.
   Magnitude $\|\vec{PQ}\| = \sqrt{3^2 + 4^2} = 5$.
   Unit vector:

$$
   \mathbf{u}_{PQ} = \begin{bmatrix} \frac{3}{5} \\ \frac{4}{5} \end{bmatrix}
$$

$$
   D_{\mathbf{u}_{PQ}} f(0,0) = \nabla f(0,0) \cdot \mathbf{u}_{PQ} = \begin{bmatrix} 1 \\ 1 \end{bmatrix} \cdot \begin{bmatrix} \frac{3}{5} \\ \frac{4}{5} \end{bmatrix} = \frac{3}{5} + \frac{4}{5} = \frac{7}{5}
$$

$$
\text{Result: } \boxed{\text{(1) Direction: } \begin{bmatrix} \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{bmatrix}, \quad \text{(2) Max Rate: } \sqrt{2}, \quad \text{(3) Rate toward Q: } \frac{7}{5}}
$$

#### Key Insight / Takeaway
The rate toward $Q$ ($1.4$) is slightly less than the maximum possible rate $\sqrt{2} \approx 1.414$ because direction $(\frac{3}{5}, \frac{4}{5})^T$ is angled near $(\frac{1}{\sqrt{2}}, \frac{1}{\sqrt{2}})^T$.

---

```python
# L1.3 - steepest ascent of x e^y + y e^x at the origin, and the rate toward Q(3,4)
x, y = sp.symbols("x y", real=True)
f = x*sp.exp(y) + y*sp.exp(x)
g = sp.Matrix([sp.diff(f, x), sp.diff(f, y)]).subs({x: 0, y: 0})
rate_max = g.norm()
u_max = g/rate_max
u_pq = sp.Matrix([3, 4])/5
rate_pq = (g.T*u_pq)[0]
print("grad f(0,0) =", g.T)
print("max rate =", rate_max, "=", float(rate_max), " direction =", u_max.T)
print("rate toward Q =", rate_pq, "=", float(rate_pq))
assert g == sp.Matrix([1, 1]) and rate_max == sp.sqrt(2) and rate_pq == sp.Rational(7, 5)
assert float(rate_pq) < float(rate_max)   # Cauchy-Schwarz
```

### Problem L1.4 — Tangent Plane to an Ellipsoid
**Source:** Marsden & Tromba, *Vector Calculus* (Chapter 2)  
**Statement:** Find the equation of the tangent hyperplane to the level surface $x^2 + 2y^2 + 3z^2 = 12$ at the point $P_0(1, 2, 1)$.

#### First-Principles Intuition
The level surface is defined by scalar field $F(x, y, z) = x^2 + 2y^2 + 3z^2 = 12$. The 3D gradient vector $\nabla F(P_0)$ is perpendicular to the tangent plane. Any point $(x,y,z)$ on the tangent plane satisfies $(\mathbf{x} - \mathbf{x}_0) \cdot \nabla F(P_0) = 0$.

#### Step-by-Step Solution
1. Define implicit function $F(x, y, z) = x^2 + 2y^2 + 3z^2$.
2. Compute gradient field $\nabla F$:

$$
   \nabla F(x, y, z) = \begin{bmatrix} 2x \\ 4y \\ 6z \end{bmatrix}
$$

3. Evaluate gradient at point $P_0(1, 2, 1)$:

$$
   \mathbf{n} = \nabla F(1, 2, 1) = \begin{bmatrix} 2(1) \\ 4(2) \\ 6(1) \end{bmatrix} = \begin{bmatrix} 2 \\ 8 \\ 6 \end{bmatrix}
$$

4. Tangent plane equation $\mathbf{n} \cdot (\mathbf{x} - \mathbf{x}_0) = 0$:

$$
   2(x - 1) + 8(y - 2) + 6(z - 1) = 0
$$

5. Expand and simplify:

$$
   2x - 2 + 8y - 16 + 6z - 6 = 0 \implies 2x + 8y + 6z = 24
$$

6. Divide by 2:

$$
   x + 4y + 3z = 12
$$

$$
\text{Result: } \boxed{x + 4y + 3z = 12}
$$

#### Key Insight / Takeaway
The normal vector components $(1, 4, 3)^T$ match the partial derivatives of the ellipsoid scaled by half.

---

```python
# L1.4 - tangent plane to x^2 + 2y^2 + 3z^2 = 12 at (1,2,1)
x, y, z = sp.symbols("x y z", real=True)
F = x**2 + 2*y**2 + 3*z**2
p0 = {x: 1, y: 2, z: 1}
n = sp.Matrix([sp.diff(F, v) for v in (x, y, z)]).subs(p0)
plane = sp.expand(n.dot(sp.Matrix([x - 1, y - 2, z - 1])))
print("point on the surface? F(1,2,1) =", F.subs(p0), "(should be 12)")
print("normal n =", n.T, " plane equation:", plane, "= 0")
print("simplified:", sp.simplify(plane/2 + 12), "= 12")
assert F.subs(p0) == 12 and sp.simplify(plane - (2*x + 8*y + 6*z - 24)) == 0
```

### Problem L1.5 — Tangent Plane to a Graph
**Source:** Stewart, *Multivariable Calculus* (Chapter 14)  
**Statement:** Find the equation of the tangent plane to the graph surface $z = f(x, y) = x^2 - 3xy + y^2$ at the point where $(x_0, y_0) = (2, 1)$.

#### First-Principles Intuition
A graph $z = f(x, y)$ can be converted into a level surface $F(x, y, z) = f(x, y) - z = 0$. The tangent plane equation is given by local linearization $z - z_0 = f_x(x_0, y_0)(x - x_0) + f_y(x_0, y_0)(y - y_0)$.

#### Step-by-Step Solution
1. Evaluate $z_0 = f(2, 1)$:

$$
   z_0 = 2^2 - 3(2)(1) + 1^2 = 4 - 6 + 1 = -1
$$

2. Compute partial derivatives of $f(x, y) = x^2 - 3xy + y^2$:
   - $f_x(x, y) = 2x - 3y$
   - $f_y(x, y) = -3x + 2y$
3. Evaluate partial derivatives at $(2, 1)$:
   - $f_x(2, 1) = 2(2) - 3(1) = 4 - 3 = 1$
   - $f_y(2, 1) = -3(2) + 2(1) = -6 + 2 = -4$
4. Write tangent plane equation:

$$
   z - z_0 = f_x(2, 1)(x - 2) + f_y(2, 1)(y - 1)
$$

$$
   z - (-1) = 1(x - 2) - 4(y - 1)
$$

$$
   z + 1 = x - 2 - 4y + 4 \implies z + 1 = x - 4y + 2
$$

5. Rearrange into standard form:

$$
   x - 4y - z = -1
$$

$$
\text{Result: } \boxed{x - 4y - z = -1}
$$

#### Key Insight / Takeaway
The normal vector to the graph surface is $\mathbf{n} = (f_x, f_y, -1)^T = (1, -4, -1)^T$.

---

```python
# L1.5 - tangent plane to z = x^2 - 3xy + y^2 at (2,1)
x, y = sp.symbols("x y", real=True)
f = x**2 - 3*x*y + y**2
z0 = f.subs({x: 2, y: 1})
fx = sp.diff(f, x).subs({x: 2, y: 1})
fy = sp.diff(f, y).subs({x: 2, y: 1})
print("z0 =", z0, " f_x(2,1) =", fx, " f_y(2,1) =", fy)
print("tangent plane: z =", sp.expand(z0 + fx*(x - 2) + fy*(y - 1)))
assert (z0, fx, fy) == (-1, 1, -4)
# the boxed form x - 4y - z = -1 must hold at the point of tangency
assert 2 - 4*1 - (-1) == -1
```

### Problem L1.6 — Temperature Rate Along a Helix
**Source:** Apostol, *Calculus, Volume II* (Chapter 8)  
**Statement:** A particle moves along a circular helix trajectory $\mathbf{r}(t) = (\cos t, \sin t, t)^T$ in a temperature field $T(x, y, z) = x^2 + y^2 - z$. Compute the rate of temperature change $\frac{dT}{dt}$ experienced by the particle at time $t = \frac{\pi}{4}$ using the multivariable chain rule.

#### First-Principles Intuition
As the particle moves along the helix, its spatial position changes continuously. The total rate of temperature change is the dot product of the spatial temperature gradient $\nabla T$ and the particle's velocity vector $\mathbf{r}'(t)$.

#### Step-by-Step Solution
1. Compute spatial temperature gradient $\nabla T(x, y, z)$:

$$
   \nabla T(x, y, z) = \begin{bmatrix} 2x \\ 2y \\ -1 \end{bmatrix}
$$

2. Evaluate particle position $\mathbf{r}(t)$ at $t = \frac{\pi}{4}$:

$$
   \mathbf{r}(\frac{\pi}{4}) = \begin{bmatrix} \cos(\frac{\pi}{4}) \\ \sin(\frac{\pi}{4}) \\ \frac{\pi}{4} \end{bmatrix} = \begin{bmatrix} \frac{\sqrt{2}}{2} \\ \frac{\sqrt{2}}{2} \\ \frac{\pi}{4} \end{bmatrix}
$$

3. Evaluate gradient at particle position $\mathbf{r}(\frac{\pi}{4})$:

$$
   \nabla T(\mathbf{r}(\frac{\pi}{4})) = \begin{bmatrix} 2(\frac{\sqrt{2}}{2}) \\ 2(\frac{\sqrt{2}}{2}) \\ -1 \end{bmatrix} = \begin{bmatrix} \sqrt{2} \\ \sqrt{2} \\ -1 \end{bmatrix}
$$

4. Compute velocity vector $\mathbf{r}'(t)$:

$$
   \mathbf{r}'(t) = \begin{bmatrix} -\sin t \\ \cos t \\ 1 \end{bmatrix} \implies \mathbf{r}'(\frac{\pi}{4}) = \begin{bmatrix} -\frac{\sqrt{2}}{2} \\ \frac{\sqrt{2}}{2} \\ 1 \end{bmatrix}
$$

5. Compute chain rule dot product $\frac{dT}{dt} = \nabla T(\mathbf{r}(t)) \cdot \mathbf{r}'(t)$:

$$
   \frac{dT}{dt}\Big\vert_{t=\frac{\pi}{4}} = \begin{bmatrix} \sqrt{2} \\ \sqrt{2} \\ -1 \end{bmatrix} \cdot \begin{bmatrix} -\frac{\sqrt{2}}{2} \\ \frac{\sqrt{2}}{2} \\ 1 \end{bmatrix} = (\sqrt{2})\left(-\frac{\sqrt{2}}{2}\right) + (\sqrt{2})\left(\frac{\sqrt{2}}{2}\right) + (-1)(1) = -1 + 1 - 1 = -1
$$

$$
\text{Result: } \boxed{\frac{dT}{dt}\Big\vert_{t=\frac{\pi}{4}} = -1}
$$

#### Key Insight / Takeaway
Even though horizontal circular motion keeps $x^2 + y^2 = 1$ constant, vertical climb along $z$ at speed $dz/dt = 1$ causes temperature to drop at rate $-1$.

---

```python
# L1.6 - dT/dt along the helix r(t) = (cos t, sin t, t) in T = x^2 + y^2 - z
t = sp.symbols("t", real=True)
r = sp.Matrix([sp.cos(t), sp.sin(t), t])
T = r[0]**2 + r[1]**2 - r[2]
chain = sp.simplify(sp.Matrix([2*r[0], 2*r[1], -1]).dot(sp.diff(r, t)))
direct = sp.simplify(sp.diff(T, t))
print("chain rule dT/dt =", chain, "   direct d/dt T(r(t)) =", direct)
print("value at t = pi/4:", chain.subs(t, sp.pi/4))
assert sp.simplify(chain - direct) == 0 and chain.subs(t, sp.pi/4) == -1
```

### Problem L1.7 — Gradient of a Radial Field
**Source:** Demidovich, *Problems in Mathematical Analysis* (No. 3201)  
**Statement:** Let $\mathbf{r} = (x, y, z)^T$ and $r = \|\mathbf{r}\| = \sqrt{x^2 + y^2 + z^2}$. Show that for any differentiable single-variable function $g(r)$, the gradient of the radial scalar field $f(\mathbf{r}) = g(r)$ is given by $\nabla f(\mathbf{r}) = g'(r) \frac{\mathbf{r}}{r} = g'(r) \hat{\mathbf{r}}$.

#### First-Principles Intuition
A radial function depends solely on the distance $r$ from the origin. Level surfaces are concentric spheres. The gradient vector must point purely in the radial direction $\hat{\mathbf{r}} = \frac{\mathbf{r}}{r}$ with magnitude equal to $g'(r)$.

#### Step-by-Step Solution
1. Apply the chain rule to partial derivative $\frac{\partial f}{\partial x}$:

$$
   \frac{\partial f}{\partial x} = \frac{d g}{d r} \frac{\partial r}{\partial x}
$$

2. Compute $\frac{\partial r}{\partial x}$ where $r = (x^2 + y^2 + z^2)^{\frac{1}{2}}$:

$$
   \frac{\partial r}{\partial x} = \frac{1}{2}(x^2 + y^2 + z^2)^{-\frac{1}{2}} (2x) = \frac{x}{r}
$$

3. Similarly, for $y$ and $z$:

$$
   \frac{\partial r}{\partial y} = \frac{y}{r}, \quad \frac{\partial r}{\partial z} = \frac{z}{r}
$$

4. Assemble the gradient vector field:

$$
   \nabla f(\mathbf{r}) = \begin{bmatrix} g'(r) \frac{x}{r} \\ g'(r) \frac{y}{r} \\ g'(r) \frac{z}{r} \end{bmatrix} = g'(r) \frac{1}{r} \begin{bmatrix} x \\ y \\ z \end{bmatrix} = g'(r) \frac{\mathbf{r}}{r}
$$

5. Since $\hat{\mathbf{r}} = \frac{\mathbf{r}}{r}$ is the unit radial vector, $\nabla g(r) = g'(r) \hat{\mathbf{r}}$.

$$
\text{Result: } \boxed{\nabla g(r) = g'(r) \hat{\mathbf{r}} = g'(r) \frac{\mathbf{r}}{\|\mathbf{r}\|}}
$$

#### Key Insight / Takeaway
For spherically symmetric fields, gradient computation simplifies to 1D differentiation along unit radial direction $\hat{\mathbf{r}}$.

### Problem L1.8 — Normal and Tangent Lines to a Hyperbola
**Source:** Stewart, *Multivariable Calculus* (Chapter 14)  
**Statement:** Find vector equations for both the normal line and the tangent line to the level curve $x^2 - y^2 = 3$ at the point $P_0(2, 1)$.

#### First-Principles Intuition
The level curve is $f(x, y) = x^2 - y^2 = 3$. The gradient $\nabla f(P_0)$ provides the normal direction $\mathbf{n}$. The tangent direction $\mathbf{T}$ is orthogonal to $\mathbf{n}$ ($\mathbf{n} \cdot \mathbf{T} = 0$).

#### Step-by-Step Solution
1. Compute gradient of $f(x, y) = x^2 - y^2$:

$$
   \nabla f(x, y) = \begin{bmatrix} 2x \\ -2y \end{bmatrix}
$$

2. Evaluate at $P_0(2, 1)$:

$$
   \mathbf{n} = \nabla f(2, 1) = \begin{bmatrix} 4 \\ -2 \end{bmatrix}
$$

3. Simplify normal vector direction:

$$
   \mathbf{n}_{\text{simple}} = \begin{bmatrix} 2 \\ -1 \end{bmatrix}
$$

4. **Normal Line Equation:** Passes through $P_0(2, 1)^T$ in direction $\mathbf{n}_{\text{simple}}$:

$$
   \mathbf{r}_{\text{norm}}(t) = \begin{bmatrix} 2 \\ 1 \end{bmatrix} + t \begin{bmatrix} 2 \\ -1 \end{bmatrix}, \quad t \in \mathbb{R}
$$

5. **Tangent Line Direction:** Vector orthogonal to $\mathbf{n}_{\text{simple}}$ is:

$$
   \mathbf{T} = \begin{bmatrix} 1 \\ 2 \end{bmatrix}
$$

   (since $2(1) + (-1)(2) = 0$).
6. **Tangent Line Equation:** Passes through $P_0(2, 1)^T$ in direction $\mathbf{T}$:

$$
   \mathbf{r}_{\text{tang}}(s) = \begin{bmatrix} 2 \\ 1 \end{bmatrix} + s \begin{bmatrix} 1 \\ 2 \end{bmatrix}, \quad s \in \mathbb{R}
$$

$$
\text{Result: } \boxed{\text{Normal: } \begin{bmatrix} 2 \\ 1 \end{bmatrix} + t \begin{bmatrix} 2 \\ -1 \end{bmatrix}, \quad \text{Tangent: } \begin{bmatrix} 2 \\ 1 \end{bmatrix} + s \begin{bmatrix} 1 \\ 2 \end{bmatrix}}
$$

#### Key Insight / Takeaway
In 2D, if normal vector is $(a, b)^T$, tangent vector is $(-b, a)^T$.

---

```python
# L1.8 - normal and tangent lines to x^2 - y^2 = 3 at (2,1)
x, y = sp.symbols("x y", real=True)
f = x**2 - y**2
g = sp.Matrix([sp.diff(f, x), sp.diff(f, y)]).subs({x: 2, y: 1})
n_simple = sp.Matrix([2, -1])
T = sp.Matrix([1, 2])
print("point on the curve? f(2,1) =", f.subs({x: 2, y: 1}), "(should be 3)")
print("grad f(2,1) =", g.T, " simplified normal =", n_simple.T, " tangent =", T.T)
print("n . T =", n_simple.dot(T))
assert f.subs({x: 2, y: 1}) == 3 and g == sp.Matrix([4, -2]) and n_simple.dot(T) == 0
```

### Problem L1.9 — Gradient of a Quadratic Form
**Source:** Nocedal & Wright, *Numerical Optimization* (Chapter 2)  
**Statement:** Let $\mathbf{A} \in \mathbb{R}^{n \times n}$ be a symmetric matrix ($\mathbf{A} = \mathbf{A}^T$) and $\mathbf{b} \in \mathbb{R}^n$. Prove that the gradient of the quadratic scalar field $f(\mathbf{x}) = \frac{1}{2} \mathbf{x}^T \mathbf{A} \mathbf{x} - \mathbf{b}^T \mathbf{x}$ is $\nabla f(\mathbf{x}) = \mathbf{A}\mathbf{x} - \mathbf{b}$.

#### First-Principles Intuition
In 1D, the derivative of $\frac{1}{2} a x^2 - b x$ is $a x - b$. Matrix quadratic forms generalize scalar squares to $n$-dimensional vector spaces.

#### Step-by-Step Solution
1. Expand $f(\mathbf{x})$ in component summation notation:

$$
   f(\mathbf{x}) = \frac{1}{2} \sum_{i=1}^n \sum_{j=1}^n A_{ij} x_i x_j - \sum_{i=1}^n b_i x_i
$$

2. Differentiate $f(\mathbf{x})$ with respect to component $x_k$:

$$
   \frac{\partial f}{\partial x_k} = \frac{\partial}{\partial x_k} \left( \frac{1}{2} \sum_{i=1}^n \sum_{j=1}^n A_{ij} x_i x_j \right) - \frac{\partial}{\partial x_k} \left( \sum_{i=1}^n b_i x_i \right)
$$

3. The linear term derivative is simply $b_k$.
4. For the quadratic term, isolate terms involving $x_k$:
   - When $i=k, j=k$: term is $\frac{1}{2} A_{kk} x_k^2 \implies$ derivative is $A_{kk} x_k$.
   - When $i=k, j \neq k$: terms are $\frac{1}{2} \sum_{j \neq k} A_{kj} x_k x_j \implies$ derivative is $\frac{1}{2} \sum_{j \neq k} A_{kj} x_j$.
   - When $j=k, i \neq k$: terms are $\frac{1}{2} \sum_{i \neq k} A_{ik} x_i x_k \implies$ derivative is $\frac{1}{2} \sum_{i \neq k} A_{ik} x_i$.
5. Combine all quadratic derivative terms:

$$
   \frac{\partial}{\partial x_k} \left( \frac{1}{2} \mathbf{x}^T \mathbf{A} \mathbf{x} \right) = \frac{1}{2} \sum_{j=1}^n A_{kj} x_j + \frac{1}{2} \sum_{i=1}^n A_{ik} x_i
$$

6. Since $\mathbf{A}$ is symmetric ($A_{ik} = A_{ki}$):

$$
   \frac{\partial}{\partial x_k} \left( \frac{1}{2} \mathbf{x}^T \mathbf{A} \mathbf{x} \right) = \sum_{j=1}^n A_{kj} x_j = (\mathbf{A}\mathbf{x})_k
$$

7. Assemble all partial derivatives into vector format:

$$
   \nabla f(\mathbf{x}) = \mathbf{A}\mathbf{x} - \mathbf{b}
$$

$$
\text{Result: } \boxed{\nabla \left( \frac{1}{2} \mathbf{x}^T \mathbf{A} \mathbf{x} - \mathbf{b}^T \mathbf{x} \right) = \mathbf{A}\mathbf{x} - \mathbf{b}}
$$

#### Key Insight / Takeaway
Setting $\nabla f(\mathbf{x}^{\ast}) = \mathbf{0} \implies \mathbf{A}\mathbf{x}^{\ast} = \mathbf{b}$ connects optimization of quadratic forms directly to solving linear systems of equations.

---

```python
# L1.9 - grad of (1/2) x^T A x - b^T x for a random symmetric A, against finite differences
n = 5
M = rng.standard_normal((n, n))
A = M + M.T
b = rng.standard_normal(n)
xq = rng.standard_normal(n)
f = lambda p: 0.5 * p @ A @ p - b @ p
analytic = A @ xq - b
numeric = approx_fprime(xq, f, 1e-7)
print("analytic A x - b =", analytic)
print("finite difference =", numeric)
print("max abs error     =", np.max(np.abs(analytic - numeric)))
assert np.max(np.abs(analytic - numeric)) < 1e-5
```

### Problem L1.10 — Zero Rate Along a Level Curve
**Source:** Apostol, *Calculus, Volume II* (Chapter 8)  
**Statement:** Let $\mathbf{T}$ be a unit vector tangent to the level curve $f(x, y) = c$ at point $\mathbf{x}_0$. Show by direct calculation that $D_{\mathbf{T}} f(\mathbf{x}_0) = 0$.

#### First-Principles Intuition
A tangent vector points locally along the level curve. Since the function value remains constant along the level curve, the rate of change along its tangent vector must be zero.

#### Step-by-Step Solution
1. Let $\nabla f(\mathbf{x}_0)$ be the gradient vector of $f$ at $\mathbf{x}_0$.
2. By Theorem 4.3 (gradient normal to the level set), $\nabla f(\mathbf{x}_0)$ is orthogonal to any vector $\mathbf{T}$ tangent to the level set at $\mathbf{x}_0$:

$$
   \nabla f(\mathbf{x}_0) \cdot \mathbf{T} = 0
$$

3. By Theorem 4.1 (the master formula):

$$
   D_{\mathbf{T}} f(\mathbf{x}_0) = \nabla f(\mathbf{x}_0) \cdot \mathbf{T}
$$

4. Substituting $\nabla f(\mathbf{x}_0) \cdot \mathbf{T} = 0$:

$$
   D_{\mathbf{T}} f(\mathbf{x}_0) = 0
$$

$$
\text{Result: } \boxed{D_{\mathbf{T}} f(\mathbf{x}_0) = 0}
$$

#### Key Insight / Takeaway
Moving along tangent directions to level sets produces zero first-order variation in function value.

### Problem L2.1 — Electric Field from a Potential
**Source:** Physics / Marsden & Tromba, *Vector Calculus*  
**Statement:** The electric potential field in a region of space is given by $V(x, y, z) = \frac{100}{x^2 + y^2 + z^2 + 1}$ Volts.
1. Compute the electric field vector field $\mathbf{E}(x, y, z) = -\nabla V(x, y, z)$.
2. Calculate the magnitude and direction of the electric force $\mathbf{F} = q \mathbf{E}$ acting on a point charge $q = 2 \mu\text{C}$ located at position $P(1, 2, 2)$.

#### First-Principles Intuition
Electric field is the negative gradient of electrostatic potential, driving positive charges toward regions of lower potential.

#### Step-by-Step Solution
1. Let $r^2 = x^2 + y^2 + z^2$. Potential is $V = 100(r^2 + 1)^{-1}$.
2. Compute partial derivative $\frac{\partial V}{\partial x}$:

$$
   \frac{\partial V}{\partial x} = -100(r^2 + 1)^{-2} (2x) = -\frac{200x}{(x^2 + y^2 + z^2 + 1)^2}
$$

3. By radial symmetry:

$$
   \nabla V(x, y, z) = -\frac{200}{(x^2 + y^2 + z^2 + 1)^2} \begin{bmatrix} x \\ y \\ z \end{bmatrix}
$$

4. Thus electric field $\mathbf{E} = -\nabla V$:

$$
   \mathbf{E}(x, y, z) = \frac{200}{(x^2 + y^2 + z^2 + 1)^2} \begin{bmatrix} x \\ y \\ z \end{bmatrix}
$$

5. Evaluate $\mathbf{E}$ at $P(1, 2, 2)$:
   $r^2 = 1^2 + 2^2 + 2^2 = 9 \implies r^2 + 1 = 10$.

$$
   \mathbf{E}(1, 2, 2) = \frac{200}{10^2} \begin{bmatrix} 1 \\ 2 \\ 2 \end{bmatrix} = \frac{200}{100} \begin{bmatrix} 1 \\ 2 \\ 2 \end{bmatrix} = \begin{bmatrix} 2 \\ 4 \\ 4 \end{bmatrix} \text{ V/m}
$$

6. Compute electric force $\mathbf{F} = q \mathbf{E}$ for $q = 2 \times 10^{-6} \text{ C}$:

$$
   \mathbf{F} = 2 \times 10^{-6} \begin{bmatrix} 2 \\ 4 \\ 4 \end{bmatrix} = \begin{bmatrix} 4 \\ 8 \\ 8 \end{bmatrix} \mu\text{N}
$$

7. Magnitude of force:

$$
   \|\mathbf{F}\| = \sqrt{4^2 + 8^2 + 8^2} = \sqrt{16 + 64 + 64} = \sqrt{144} = 12 \mu\text{N}
$$

$$
\text{Result: } \boxed{\mathbf{E} = \begin{bmatrix} 2 \\ 4 \\ 4 \end{bmatrix} \text{ V/m}, \quad \mathbf{F} = \begin{bmatrix} 4 \\ 8 \\ 8 \end{bmatrix} \mu\text{N}, \quad \|\mathbf{F}\| = 12 \mu\text{N}}
$$

#### Key Insight / Takeaway
The electric field vector points radially outward away from origin, accelerating positive charges toward lower potential.

---

```python
# L2.1 - E = -grad V for V = 100/(x^2+y^2+z^2+1), and the force on q = 2 microcoulomb
x, y, z = sp.symbols("x y z", real=True)
V = 100/(x**2 + y**2 + z**2 + 1)
E = -sp.Matrix([sp.diff(V, v) for v in (x, y, z)])
print("E(x,y,z) =", sp.simplify(E.T))
E_P = sp.simplify(E.subs({x: 1, y: 2, z: 2}))
print("E(1,2,2) =", E_P.T, "V/m")
q = 2e-6
F = np.array([float(c) for c in E_P]) * q
print("F = qE =", F*1e6, "microN   ||F|| =", np.linalg.norm(F)*1e6, "microN")
assert E_P == sp.Matrix([2, 4, 4])
assert abs(np.linalg.norm(F)*1e6 - 12.0) < 1e-9
```

### Problem L2.2 — Fourier Heat Flux on a Plate
**Source:** Physics / Apostol, *Calculus, Volume II*  
**Statement:** A metallic plate has a temperature distribution $T(x, y) = 50 + 20x^2 - 10y^2$ (in ${}^\circ\text{C}$). By Fourier's Law, heat flux is $\mathbf{q} = -k \nabla T$ with thermal conductivity $k = 3 \text{ W/(m}\cdot{}^\circ\text{C)}$.
1. Find heat flux vector $\mathbf{q}$ at point $(1, 2)$.
2. In what direction is heat flowing fastest at $(1, 2)$?

#### First-Principles Intuition
Heat naturally diffuses from hotter to colder regions, driven by the negative temperature gradient.

#### Step-by-Step Solution
1. Compute temperature gradient $\nabla T(x, y)$:

$$
   \nabla T(x, y) = \begin{bmatrix} 40x \\ -20y \end{bmatrix}
$$

2. Evaluate gradient at $(1, 2)$:

$$
   \nabla T(1, 2) = \begin{bmatrix} 40(1) \\ -20(2) \end{bmatrix} = \begin{bmatrix} 40 \\ -40 \end{bmatrix} {}^\circ\text{C/m}
$$

3. Compute heat flux vector $\mathbf{q} = -k \nabla T$ for $k = 3$:

$$
   \mathbf{q}(1, 2) = -3 \begin{bmatrix} 40 \\ -40 \end{bmatrix} = \begin{bmatrix} -120 \\ 120 \end{bmatrix} \text{ W/m}^2
$$

4. Direction of heat flow is unit vector of $\mathbf{q}$:

$$
   \|\mathbf{q}\| = \sqrt{(-120)^2 + 120^2} = 120\sqrt{2}
$$

$$
   \mathbf{u}_{\text{heat}} = \frac{1}{120\sqrt{2}} \begin{bmatrix} -120 \\ 120 \end{bmatrix} = \begin{bmatrix} -\frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{bmatrix}
$$

$$
\text{Result: } \boxed{\mathbf{q} = \begin{bmatrix} -120 \\ 120 \end{bmatrix} \text{ W/m}^2, \quad \text{Direction: } \begin{bmatrix} -\frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{bmatrix}}
$$

#### Key Insight / Takeaway
Heat flows along $-\nabla T$, moving toward the top-left quadrant where temperature drops.

---

```python
# L2.2 - Fourier heat flux q = -k grad T on T = 50 + 20x^2 - 10y^2 at (1,2), k = 3
x, y = sp.symbols("x y", real=True)
T = 50 + 20*x**2 - 10*y**2
gT = sp.Matrix([sp.diff(T, x), sp.diff(T, y)]).subs({x: 1, y: 2})
k = 3
q = -k*gT
print("grad T(1,2) =", gT.T, "deg C / m")
print("q = -k grad T =", q.T, "W/m^2   ||q|| =", q.norm(), "=", float(q.norm()))
print("unit direction =", (q/q.norm()).T)
assert gT == sp.Matrix([40, -40]) and q == sp.Matrix([-120, 120])
assert sp.simplify(q.norm() - 120*sp.sqrt(2)) == 0
```

### Problem L2.3 — Gradient of the Least-Squares Loss
**Source:** Goodfellow et al., *Deep Learning* (Chapter 4)  
**Statement:** Consider the mean squared error (MSE) loss for linear regression without bias:

$$
\mathcal{L}(\mathbf{w}) = \frac{1}{2m} \|\mathbf{X}\mathbf{w} - \mathbf{y}\|^2 = \frac{1}{2m} (\mathbf{X}\mathbf{w} - \mathbf{y})^T (\mathbf{X}\mathbf{w} - \mathbf{y})
$$

where $\mathbf{X} \in \mathbb{R}^{m \times d}$, $\mathbf{y} \in \mathbb{R}^m$, and $\mathbf{w} \in \mathbb{R}^d$. Derive the gradient $\nabla_{\mathbf{w}} \mathcal{L}(\mathbf{w})$ from first principles.

#### First-Principles Intuition
Expanding the quadratic loss into matrix products allows us to apply linear and quadratic gradient formulas.

#### Step-by-Step Solution
1. Expand the loss function $\mathcal{L}(\mathbf{w})$:

$$
   \mathcal{L}(\mathbf{w}) = \frac{1}{2m} (\mathbf{w}^T \mathbf{X}^T - \mathbf{y}^T)(\mathbf{X}\mathbf{w} - \mathbf{y}) = \frac{1}{2m} \left( \mathbf{w}^T \mathbf{X}^T \mathbf{X} \mathbf{w} - \mathbf{w}^T \mathbf{X}^T \mathbf{y} - \mathbf{y}^T \mathbf{X} \mathbf{w} + \mathbf{y}^T \mathbf{y} \right)
$$

2. Since scalar quantities equal their transpose, $\mathbf{y}^T \mathbf{X} \mathbf{w} = (\mathbf{w}^T \mathbf{X}^T \mathbf{y})^T = \mathbf{w}^T \mathbf{X}^T \mathbf{y}$:

$$
   \mathcal{L}(\mathbf{w}) = \frac{1}{2m} \mathbf{w}^T (\mathbf{X}^T \mathbf{X}) \mathbf{w} - \frac{1}{m} (\mathbf{X}^T \mathbf{y})^T \mathbf{w} + \frac{1}{2m} \|\mathbf{y}\|^2
$$

3. Apply matrix gradient rules (Problem L1.9):
   - $\nabla_{\mathbf{w}} \left( \frac{1}{2} \mathbf{w}^T \mathbf{A} \mathbf{w} \right) = \mathbf{A}\mathbf{w}$ for symmetric matrix $\mathbf{A} = \mathbf{X}^T \mathbf{X}$.
   - $\nabla_{\mathbf{w}} (\mathbf{c}^T \mathbf{w}) = \mathbf{c}$ for vector $\mathbf{c} = \mathbf{X}^T \mathbf{y}$.
4. Therefore:

$$
   \nabla_{\mathbf{w}} \mathcal{L}(\mathbf{w}) = \frac{1}{m} \mathbf{X}^T \mathbf{X} \mathbf{w} - \frac{1}{m} \mathbf{X}^T \mathbf{y} = \frac{1}{m} \mathbf{X}^T (\mathbf{X}\mathbf{w} - \mathbf{y})
$$

$$
\text{Result: } \boxed{\nabla_{\mathbf{w}} \mathcal{L}(\mathbf{w}) = \frac{1}{m} \mathbf{X}^T (\mathbf{X}\mathbf{w} - \mathbf{y})}
$$

#### Key Insight / Takeaway
The gradient is the cross-product of feature inputs $\mathbf{X}^T$ with residual prediction errors $\mathbf{e} = \mathbf{X}\mathbf{w} - \mathbf{y}$.

---

```python
# L2.3 - grad of the least-squares loss against finite differences
m_, d_ = 12, 4
X = rng.standard_normal((m_, d_))
yv = rng.standard_normal(m_)
loss = lambda w: (0.5/m_) * np.sum((X @ w - yv)**2)
w0 = rng.standard_normal(d_)
analytic = (X.T @ (X @ w0 - yv))/m_
numeric = approx_fprime(w0, loss, 1e-7)
print("analytic (1/m) X^T (Xw - y) =", analytic)
print("finite difference           =", numeric)
print("max abs error               =", np.max(np.abs(analytic - numeric)))
assert np.max(np.abs(analytic - numeric)) < 1e-5
```

### Problem L2.4 — Softmax Cross-Entropy Gradient
**Source:** Goodfellow et al., *Deep Learning* (Chapter 4)  
**Statement:** For a multi-class softmax classifier with logit vector $\mathbf{z} \in \mathbb{R}^K$, predicted probability vector $\mathbf{p} = \operatorname{softmax}(\mathbf{z})$ has components $p_k = \frac{e^{z_k}}{\sum_{j=1}^K e^{z_j}}$.
Given target class $y \in \{1, \dots, K\}$, the cross-entropy loss is $\mathcal{L}(\mathbf{z}) = -\ln p_y$.
Derive the gradient vector $\nabla_{\mathbf{z}} \mathcal{L}(\mathbf{z})$.

#### First-Principles Intuition
Softmax cross-entropy loss simplifies remarkably upon differentiation: the gradient is the error vector between predicted probability distribution $\mathbf{p}$ and one-hot target distribution $\mathbf{e}_y$.

#### Step-by-Step Solution
1. Write loss function explicitly in terms of logits $\mathbf{z}$:

$$
   \mathcal{L}(\mathbf{z}) = -\ln \left( \frac{e^{z_y}}{\sum_{j=1}^K e^{z_j}} \right) = -z_y + \ln \left( \sum_{j=1}^K e^{z_j} \right)
$$

2. Compute partial derivative $\frac{\partial \mathcal{L}}{\partial z_k}$ for arbitrary index $k$:

$$
   \frac{\partial \mathcal{L}}{\partial z_k} = -\frac{\partial z_y}{\partial z_k} + \frac{\partial}{\partial z_k} \ln \left( \sum_{j=1}^K e^{z_j} \right)
$$

3. The derivative of log-sum-exp is:

$$
   \frac{\partial}{\partial z_k} \ln \left( \sum_{j=1}^K e^{z_j} \right) = \frac{e^{z_k}}{\sum_{j=1}^K e^{z_j}} = p_k
$$

4. For the first term, $\frac{\partial z_y}{\partial z_k} = 1$ if $k = y$, and $0$ if $k \neq y$ (Kronecker delta $\delta_{ky}$).
5. Combining terms:

$$
   \frac{\partial \mathcal{L}}{\partial z_k} = p_k - \delta_{ky}
$$

6. Assemble into vector format:

$$
   \nabla_{\mathbf{z}} \mathcal{L}(\mathbf{z}) = \mathbf{p} - \mathbf{e}_y
$$

$$
\text{Result: } \boxed{\nabla_{\mathbf{z}} \mathcal{L}(\mathbf{z}) = \mathbf{p} - \mathbf{e}_y}
$$

#### Key Insight / Takeaway
Gradient descent updates logits proportional to the probability residual $\mathbf{p} - \mathbf{e}_y$.

---

```python
# L2.4 - softmax cross-entropy gradient p - e_y against finite differences
K, y_true = 5, 2
zv = rng.standard_normal(K)
def ce(zz):
    zz = zz - zz.max()
    return -zz[y_true] + np.log(np.exp(zz).sum())
p = np.exp(zv - zv.max()); p /= p.sum()
analytic = p.copy(); analytic[y_true] -= 1.0
numeric = approx_fprime(zv, ce, 1e-7)
print("p            =", p)
print("p - e_y      =", analytic)
print("finite diff  =", numeric)
print("max abs error=", np.max(np.abs(analytic - numeric)))
assert np.max(np.abs(analytic - numeric)) < 1e-5
```

### Problem L2.5 — Steepest-Descent Streamline on a Terrain
**Source:** Marsden & Tromba, *Vector Calculus* (Chapter 2)  
**Statement:** A terrain is modeled by elevation function $z = f(x, y) = 100 - x^2 - 2y^2$. A water drop placed at point $(1, 1, 97)$ flows along the trajectory of steepest descent $\mathbf{r}(t) = (x(t), y(t))^T$.
1. Set up the system of differential equations for the streamline trajectory.
2. Solve for the curve $y = g(x)$ followed by the water drop.

#### First-Principles Intuition
Water drops follow the path of steepest descent, meaning their velocity vector $\mathbf{r}'(t)$ is parallel to the negative gradient $-\nabla f(x, y)$.

#### Step-by-Step Solution
1. Compute negative gradient of $f(x, y) = 100 - x^2 - 2y^2$:

$$
   -\nabla f(x, y) = -\begin{bmatrix} -2x \\ -4y \end{bmatrix} = \begin{bmatrix} 2x \\ 4y \end{bmatrix}
$$

2. The steepest descent trajectory satisfies ODE system:

$$
   \frac{dx}{dt} = 2x, \quad \frac{dy}{dt} = 4y
$$

3. Divide the $y$-equation by the $x$-equation to eliminate time parameter $t$:

$$
   \frac{dy}{dx} = \frac{\frac{dy}{dt}}{\frac{dx}{dt}} = \frac{4y}{2x} = \frac{2y}{x}
$$

4. Separate variables:

$$
   \frac{dy}{y} = 2 \frac{dx}{x}
$$

5. Integrate both sides:

$$
   \ln \lvert y \rvert = 2 \ln \lvert x \rvert + C \implies y = C x^2
$$

6. Apply initial condition $(x(0), y(0)) = (1, 1)$:

$$
   1 = C(1)^2 \implies C = 1
$$

7. Thus the water drop path across the $xy$-plane is parabolic curve $y = x^2$.

$$
\text{Result: } \boxed{y = x^2 \quad (\text{Path of Steepest Descent Streamline})}
$$

#### Key Insight / Takeaway
Streamlines (orthogonal trajectories to level curves $x^2 + 2y^2 = c$) form parabolic paths $y = k x^2$.

---

```python
# L2.5 - the steepest-descent streamline of f = 100 - x^2 - 2y^2 from (1,1) is y = x^2
sol = solve_ivp(lambda t, p: np.array([2*p[0], 4*p[1]]), (0.0, 1.0), np.array([1.0, 1.0]),
                rtol=1e-11, atol=1e-13, dense_output=True)
ts = np.linspace(0.0, 1.0, 11)
P = sol.sol(ts)
resid = np.max(np.abs(P[1] - P[0]**2))
print(" t      x(t)        y(t)       x(t)^2")
for tt, xx, yy in zip(ts[:5], P[0][:5], P[1][:5]):
    print(f"{tt:4.2f} {xx:10.5f} {yy:10.5f} {xx**2:10.5f}")
print("max |y(t) - x(t)^2| along the trajectory =", resid)
assert resid < 1e-8
```

### Problem L2.6 — Logistic Regression Gradient
**Source:** Goodfellow et al., *Deep Learning* (Chapter 4)  
**Statement:** Derive the gradient of the binary cross-entropy loss for logistic regression with weight vector $\mathbf{w} \in \mathbb{R}^d$ and single data pair $(\mathbf{x}, y)$ where $y \in \{0, 1\}$:

$$
\mathcal{L}(\mathbf{w}) = - y \ln \sigma(\mathbf{w}^T \mathbf{x}) - (1-y) \ln (1 - \sigma(\mathbf{w}^T \mathbf{x}))
$$

where $\sigma(z) = \frac{1}{1 + e^{-z}}$ is the sigmoid activation function.

#### First-Principles Intuition
Using the identity $\sigma'(z) = \sigma(z)(1 - \sigma(z))$, the multivariable chain rule simplifies the loss derivative.

#### Step-by-Step Solution
1. Let $a = \sigma(z)$ where $z = \mathbf{w}^T \mathbf{x}$. Note $\nabla_{\mathbf{w}} z = \mathbf{x}$.
2. Derivative of sigmoid $\sigma(z)$:

$$
   \frac{da}{dz} = \frac{d}{dz}(1 + e^{-z})^{-1} = -(1 + e^{-z})^{-2}(-e^{-z}) = \frac{e^{-z}}{(1 + e^{-z})^2} = a(1-a)
$$

3. Differentiate loss $\mathcal{L}(a) = -y \ln a - (1-y) \ln(1-a)$ with respect to $a$:

$$
   \frac{d\mathcal{L}}{da} = -\frac{y}{a} + \frac{1-y}{1-a} = \frac{-y(1-a) + (1-y)a}{a(1-a)} = \frac{a - y}{a(1-a)}
$$

4. Apply chain rule $\nabla_{\mathbf{w}} \mathcal{L} = \frac{d\mathcal{L}}{da} \frac{da}{dz} \nabla_{\mathbf{w}} z$:

$$
   \nabla_{\mathbf{w}} \mathcal{L} = \left( \frac{a - y}{a(1-a)} \right) \cdot \left( a(1-a) \right) \cdot \mathbf{x} = (a - y) \mathbf{x}
$$

5. Substitute $a = \sigma(\mathbf{w}^T \mathbf{x})$:

$$
   \nabla_{\mathbf{w}} \mathcal{L}(\mathbf{w}) = (\sigma(\mathbf{w}^T \mathbf{x}) - y) \mathbf{x}
$$

$$
\text{Result: } \boxed{\nabla_{\mathbf{w}} \mathcal{L}(\mathbf{w}) = (\sigma(\mathbf{w}^T \mathbf{x}) - y) \mathbf{x}}
$$

#### Key Insight / Takeaway
The derivative of sigmoid cancels the denominator of cross-entropy, eliminating vanishing gradient problems near errors.

---

```python
# L2.6 - logistic-regression gradient (sigma(w.x) - y) x against finite differences
d_ = 4
xv = rng.standard_normal(d_)
w0 = rng.standard_normal(d_)
y_lab = 1.0
sig = lambda t: 1.0/(1.0 + np.exp(-t))
def bce(w):
    a = sig(w @ xv)
    return -y_lab*np.log(a) - (1 - y_lab)*np.log(1 - a)
analytic = (sig(w0 @ xv) - y_lab)*xv
numeric = approx_fprime(w0, bce, 1e-7)
print("analytic (sigma(w.x) - y) x =", analytic)
print("finite difference           =", numeric)
print("max abs error               =", np.max(np.abs(analytic - numeric)))
assert np.max(np.abs(analytic - numeric)) < 1e-5
```

### Problem L2.7 — Gradient Flow on an Anisotropic Bowl
**Source:** Nocedal & Wright, *Numerical Optimization* (Chapter 3)  
**Statement:** Consider quadratic bowl $f(x, y) = \frac{1}{2} (a x^2 + b y^2)$ with $a, b \gt 0$. Solve the continuous gradient flow ODE system $\dot{\mathbf{x}}(t) = -\nabla f(\mathbf{x}(t))$ with initial point $(x_0, y_0)$. Show that trajectory point $(x(t), y(t))$ converges exponentially to origin $(0,0)$.

#### First-Principles Intuition
Gradient flow decouples into independent 1D exponential decay differential equations along each principal axis of the Hessian.

#### Step-by-Step Solution
1. Compute gradient vector:

$$
   \nabla f(x, y) = \begin{bmatrix} ax \\ by \end{bmatrix}
$$

2. Set up gradient flow system $\dot{\mathbf{x}} = -\nabla f(\mathbf{x})$:

$$
   \frac{dx}{dt} = -a x, \quad \frac{dy}{dt} = -b y
$$

3. Solve independent ODEs by separation of variables:

$$
   x(t) = x_0 e^{-at}, \quad y(t) = y_0 e^{-bt}
$$

4. Take limit as $t \to \infty$:

$$
   \lim_{t \to \infty} x(t) = \lim_{t \to \infty} x_0 e^{-at} = 0, \quad \lim_{t \to \infty} y(t) = \lim_{t \to \infty} y_0 e^{-bt} = 0
$$

5. Thus $\lim_{t \to \infty} \mathbf{x}(t) = (0, 0)^T$ exponentially fast.

$$
\text{Result: } \boxed{(x(t), y(t)) = (x_0 e^{-at}, y_0 e^{-bt}) \xrightarrow{t \to \infty} (0, 0)}
$$

#### Key Insight / Takeaway
If $a \gg b$, $x(t)$ decays much faster than $y(t)$, causing the trajectory to quickly collapse onto the $y$-axis before slowly sliding to the origin.

---

```python
# L2.7 - gradient flow on f = (1/2)(a x^2 + b y^2) decays as (x0 e^{-at}, y0 e^{-bt})
a, b = 1.0, 9.0
p0 = np.array([2.0, -1.0])
sol = solve_ivp(lambda t, p: -np.array([a*p[0], b*p[1]]), (0.0, 3.0), p0,
                rtol=1e-11, atol=1e-13, dense_output=True)
ts = np.array([0.0, 0.5, 1.0, 2.0, 3.0])
num = sol.sol(ts)
exact = np.stack([p0[0]*np.exp(-a*ts), p0[1]*np.exp(-b*ts)])
print("   t     x numeric    x exact     y numeric    y exact")
for i, tt in enumerate(ts):
    print(f"{tt:5.2f} {num[0,i]:11.6f} {exact[0,i]:11.6f} {num[1,i]:12.6f} {exact[1,i]:11.6f}")
print("max abs error =", np.max(np.abs(num - exact)))
print("||x(3)|| =", np.linalg.norm(num[:, -1]), "-> 0 exponentially")
assert np.max(np.abs(num - exact)) < 1e-8
```

### Problem L2.8 — Sign of a Loss Directional Derivative
**Source:** Goodfellow et al., *Deep Learning* (Chapter 4)  
**Statement:** In neural network training, a modeler perturbs weight vector $\mathbf{w}_0$ along direction $\mathbf{u} = (\frac{1}{\sqrt{2}}, \frac{1}{\sqrt{2}})^T$. Given gradient $\nabla_{\mathbf{w}} \mathcal{L}(\mathbf{w}_0) = (3, -1)^T$, compute directional derivative $D_{\mathbf{u}} \mathcal{L}(\mathbf{w}_0)$ and determine whether loss increases or decreases along $\mathbf{u}$.

#### First-Principles Intuition
The sign of $D_{\mathbf{u}}\mathcal{L}$ dictates whether moving along $\mathbf{u}$ raises ($\gt 0$) or lowers ($\lt 0$) the loss.

#### Step-by-Step Solution
1. Apply master formula:

$$
   D_{\mathbf{u}} \mathcal{L}(\mathbf{w}_0) = \nabla_{\mathbf{w}} \mathcal{L}(\mathbf{w}_0) \cdot \mathbf{u}
$$

2. Compute dot product:

$$
   D_{\mathbf{u}} \mathcal{L}(\mathbf{w}_0) = \begin{bmatrix} 3 \\ -1 \end{bmatrix} \cdot \begin{bmatrix} \frac{1}{\sqrt{2}} \\ \frac{1}{\sqrt{2}} \end{bmatrix} = \frac{3 - 1}{\sqrt{2}} = \frac{2}{\sqrt{2}} = \sqrt{2}
$$

3. Since $D_{\mathbf{u}}\mathcal{L}(\mathbf{w}_0) = \sqrt{2} \gt 0$, moving in direction $\mathbf{u}$ **increases** the loss at instantaneous rate $\sqrt{2}$.

$$
\text{Result: } \boxed{D_{\mathbf{u}} \mathcal{L}(\mathbf{w}_0) = \sqrt{2} \gt 0 \quad (\text{Loss Increases})}
$$

#### Key Insight / Takeaway
To decrease loss, one must move in a direction $\mathbf{v}$ where $\nabla \mathcal{L} \cdot \mathbf{v} \lt 0$.

---

```python
# L2.8 - directional derivative of a loss with gradient (3,-1) along (1,1)/sqrt(2)
gL = np.array([3.0, -1.0])
u = np.array([1.0, 1.0])/np.sqrt(2)
D = gL @ u
print("D_u L =", D, " sqrt(2) =", np.sqrt(2), " sign:", "increase" if D > 0 else "decrease")
print("bound ||grad L|| =", np.linalg.norm(gL), " so D_u L <= ||grad L||:", D <= np.linalg.norm(gL))
assert abs(D - np.sqrt(2)) < 1e-14 and D <= np.linalg.norm(gL)
```

### Problem L2.9 — Gravitational Field from a Potential
**Source:** Physics / Marsden & Tromba, *Vector Calculus*  
**Statement:** Gravitational potential of a point mass $M$ at origin is $V(\mathbf{r}) = -\frac{G M}{\|\mathbf{r}\|}$. Derive gravitational acceleration field $\mathbf{g}(\mathbf{r}) = -\nabla V(\mathbf{r})$.

#### First-Principles Intuition
Gravitational force is attractive, pulling objects toward origin down potential gradients.

#### Step-by-Step Solution
1. Write potential function in terms of radial distance $r = \|\mathbf{r}\|$:

$$
   V(r) = -G M r^{-1}
$$

2. Apply radial gradient formula $\nabla V(r) = V'(r) \hat{\mathbf{r}}$ (Problem L1.7):

$$
   V'(r) = \frac{d}{dr}(-G M r^{-1}) = G M r^{-2} = \frac{G M}{r^2}
$$

3. Thus gradient is:

$$
   \nabla V(\mathbf{r}) = \frac{G M}{r^2} \hat{\mathbf{r}} = \frac{G M}{r^3} \mathbf{r}
$$

4. Gravitational field is negative gradient $\mathbf{g} = -\nabla V$:

$$
   \mathbf{g}(\mathbf{r}) = -\frac{G M}{r^2} \hat{\mathbf{r}} = -\frac{G M}{r^3} \mathbf{r}
$$

$$
\text{Result: } \boxed{\mathbf{g}(\mathbf{r}) = -\frac{G M}{\|\mathbf{r}\|^2} \hat{\mathbf{r}} = -\frac{G M}{\|\mathbf{r}\|^3} \mathbf{r}}
$$

#### Key Insight / Takeaway
The negative sign ensures gravity is attractive (pointing along $-\hat{\mathbf{r}}$ toward origin).

### Problem L2.10 — Snell's Law from Fermat's Principle
**Source:** Physics / Apostol, *Calculus, Volume II*  
**Statement:** Light travels from point $A(0, a)$ in medium 1 (speed $v_1$) to point $B(d, -b)$ in medium 2 (speed $v_2$), crossing boundary $y=0$ at point $P(x, 0)$. Total travel time is:

$$
T(x) = \frac{\sqrt{x^2 + a^2}}{v_1} + \frac{\sqrt{(d-x)^2 + b^2}}{v_2}
$$

Using Fermat's principle of least time ($T'(x) = 0$), derive Snell's Law of Refraction: $\frac{\sin \theta_1}{v_1} = \frac{\sin \theta_2}{v_2}$.

#### First-Principles Intuition
Fermat's principle states light follows paths that minimize travel time. Setting time gradient to zero enforces angle balance across media boundary.

#### Step-by-Step Solution
1. Compute derivative $T'(x)$:

$$
   T'(x) = \frac{d}{dx} \left( \frac{\sqrt{x^2 + a^2}}{v_1} \right) + \frac{d}{dx} \left( \frac{\sqrt{(d-x)^2 + b^2}}{v_2} \right)
$$

$$
   T'(x) = \frac{x}{v_1 \sqrt{x^2 + a^2}} - \frac{d-x}{v_2 \sqrt{(d-x)^2 + b^2}}
$$

2. Set $T'(x) = 0$:

$$
   \frac{x}{v_1 \sqrt{x^2 + a^2}} = \frac{d-x}{v_2 \sqrt{(d-x)^2 + b^2}}
$$

3. Identify geometric angles from normal line:
   - In triangle 1: $\sin \theta_1 = \frac{\text{opposite}}{\text{hypotenuse}} = \frac{x}{\sqrt{x^2 + a^2}}$.
   - In triangle 2: $\sin \theta_2 = \frac{\text{opposite}}{\text{hypotenuse}} = \frac{d-x}{\sqrt{(d-x)^2 + b^2}}$.
4. Substituting trigonometric ratios into optimization equation:

$$
   \frac{\sin \theta_1}{v_1} = \frac{\sin \theta_2}{v_2}
$$

$$
\text{Result: } \boxed{\frac{\sin \theta_1}{v_1} = \frac{\sin \theta_2}{v_2} \quad (\text{Snell's Law of Refraction})}
$$

#### Key Insight / Takeaway
Refraction of light at boundaries is a physical realization of gradient-based path optimization.

### Problem L2.11 — One Exact-Line-Search Step and Orthogonality
**Source:** Nocedal & Wright, *Numerical Optimization* (Chapter 3)  
**Statement:** Consider ill-conditioned quadratic function $f(x, y) = \frac{1}{2}(x^2 + 10y^2)$.
Starting from point $\mathbf{x}_0 = (10, 1)^T$, compute the first discrete gradient descent step $\mathbf{x}_1 = \mathbf{x}_0 - \eta \nabla f(\mathbf{x}_0)$ using exact line search step size $\eta^{\ast} = \frac{\|\nabla f\|^2}{\nabla f^T \mathbf{H} \nabla f}$. Show that consecutive search directions $-\nabla f(\mathbf{x}_0)$ and $-\nabla f(\mathbf{x}_1)$ are orthogonal.

#### First-Principles Intuition
Exact line search minimizes $f(\mathbf{x}_k - \eta \nabla f(\mathbf{x}_k))$ along the line of steepest descent. At the minimum point along a line, the new gradient must be orthogonal to the search line.

#### Step-by-Step Solution
1. Compute gradient:

$$
   \nabla f(x, y) = \begin{bmatrix} x \\ 10y \end{bmatrix}
$$

   and Hessian:

$$
   \mathbf{H} = \begin{bmatrix} 1 & 0 \\ 0 & 10 \end{bmatrix}
$$

2. Evaluate at $\mathbf{x}_0 = (10, 1)^T$:

$$
   \mathbf{g}_0 = \nabla f(10, 1) = \begin{bmatrix} 10 \\ 10 \end{bmatrix}
$$

3. Compute numerator of $\eta^{\ast}$:

$$
   \|\mathbf{g}_0\|^2 = 10^2 + 10^2 = 200
$$

4. Compute denominator of $\eta^{\ast}$:

$$
   \mathbf{g}_0^T \mathbf{H} \mathbf{g}_0 = \begin{bmatrix} 10 & 10 \end{bmatrix} \begin{bmatrix} 1 & 0 \\ 0 & 10 \end{bmatrix} \begin{bmatrix} 10 \\ 10 \end{bmatrix} = \begin{bmatrix} 10 & 10 \end{bmatrix} \begin{bmatrix} 10 \\ 100 \end{bmatrix} = 100 + 1000 = 1100
$$

5. Optimal step size:

$$
   \eta^{\ast} = \frac{200}{1100} = \frac{2}{11}
$$

6. Compute next state $\mathbf{x}_1$:

$$
   \mathbf{x}_1 = \begin{bmatrix} 10 \\ 1 \end{bmatrix} - \frac{2}{11} \begin{bmatrix} 10 \\ 10 \end{bmatrix} = \begin{bmatrix} 10 - 20/11 \\ 1 - 20/11 \end{bmatrix} = \begin{bmatrix} 90/11 \\ -9/11 \end{bmatrix}
$$

7. Compute new gradient $\mathbf{g}_1 = \nabla f(\mathbf{x}_1)$:

$$
   \mathbf{g}_1 = \begin{bmatrix} 90/11 \\ 10(-9/11) \end{bmatrix} = \begin{bmatrix} 90/11 \\ -90/11 \end{bmatrix}
$$

8. Verify orthogonality $\mathbf{g}_0 \cdot \mathbf{g}_1$:

$$
   \mathbf{g}_0 \cdot \mathbf{g}_1 = \begin{bmatrix} 10 \\ 10 \end{bmatrix} \cdot \begin{bmatrix} 90/11 \\ -90/11 \end{bmatrix} = 10\left(\frac{90}{11}\right) + 10\left(-\frac{90}{11}\right) = 0
$$

$$
\text{Result: } \boxed{\mathbf{x}_1 = \begin{bmatrix} 90/11 \\ -9/11 \end{bmatrix}, \quad \mathbf{g}_0 \cdot \mathbf{g}_1 = 0 \quad (\text{Orthogonal Zigzagging})}
$$

#### Key Insight / Takeaway
Exact line search causes steepest descent directions to turn at $90^\circ$ angles at every step, creating inefficient zigzag trajectories in narrow valleys.

---

```python
# L2.11 - one exact-line-search step on f = (1/2)(x^2 + 10 y^2) from (10,1)
H = np.diag([1.0, 10.0])
xk = np.array([10.0, 1.0])
g0 = H @ xk
eta = (g0 @ g0)/(g0 @ H @ g0)
x1 = xk - eta*g0
g1 = H @ x1
print("g0 =", g0, "  eta* =", eta, " = 2/11 =", 2/11)
print("x1 =", x1, " exact (90/11, -9/11) =", (90/11, -9/11))
print("g1 =", g1, "  g0 . g1 =", g0 @ g1)
assert abs(eta - 2/11) < 1e-15
assert np.allclose(x1, [90/11, -9/11])
assert abs(g0 @ g1) < 1e-12
```

### Problem L2.12 — Reverse-Mode Differentiation of a Composite
**Source:** Goodfellow et al., *Deep Learning* (Chapter 6)  
**Statement:** Consider composite scalar function $f(x, y) = \sin(x y + y^2)$.
Construct the computational graph and compute derivative $\frac{\partial f}{\partial x}$ at $(x, y) = (1, \frac{\sqrt{\pi}}{2})$ using reverse-mode automatic differentiation (backpropagation).

#### First-Principles Intuition
Reverse-mode AD computes gradients by traversing the computational graph backwards, propagating adjoint values $\bar{v}_i = \frac{\partial f}{\partial v_i}$.

#### Step-by-Step Solution
1. **Forward Pass:** Define intermediate node variables:
   - $v_1 = x = 1$
   - $v_2 = y = \frac{\sqrt{\pi}}{2}$
   - $v_3 = v_1 v_2 = \frac{\sqrt{\pi}}{2}$
   - $v_4 = v_2^2 = \frac{\pi}{4}$
   - $v_5 = v_3 + v_4 = \frac{\sqrt{\pi}}{2} + \frac{\pi}{4}$
   - $f = v_6 = \sin(v_5) = \sin(\frac{\sqrt{\pi}}{2} + \frac{\pi}{4})$
2. **Reverse Pass (Adjoints):** Initialize output adjoint $\bar{v}_6 = \frac{\partial f}{\partial v_6} = 1$.
   - $\bar{v}_5 = \bar{v}_6 \frac{\partial v_6}{\partial v_5} = 1 \cdot \cos(v_5) = \cos(\frac{\sqrt{\pi}}{2} + \frac{\pi}{4})$
   - $\bar{v}_4 = \bar{v}_5 \frac{\partial v_5}{\partial v_4} = \bar{v}_5 \cdot 1 = \cos(\frac{\sqrt{\pi}}{2} + \frac{\pi}{4})$
   - $\bar{v}_3 = \bar{v}_5 \frac{\partial v_5}{\partial v_3} = \bar{v}_5 \cdot 1 = \cos(\frac{\sqrt{\pi}}{2} + \frac{\pi}{4})$
   - $\bar{v}_1 = \bar{v}_3 \frac{\partial v_3}{\partial v_1} = \bar{v}_3 \cdot v_2 = v_2 \cos(v_5)$
3. Substitute numerical values at $x = 1, y = \frac{\sqrt{\pi}}{2}$:

$$
   \frac{\partial f}{\partial x} = \bar{v}_1 = \left(\frac{\sqrt{\pi}}{2}\right) \cos\left(\frac{\sqrt{\pi}}{2} + \frac{\pi}{4}\right)
$$

$$
\text{Result: } \boxed{\frac{\partial f}{\partial x} = \frac{\sqrt{\pi}}{2} \cos\left(\frac{\sqrt{\pi}}{2} + \frac{\pi}{4}\right)}
$$

#### Key Insight / Takeaway
Reverse-mode AD accumulates gradients backward in a single pass regardless of input space dimension $n$.

---

```python
# L2.12 - reverse-mode adjoint for f = sin(xy + y^2) at (1, sqrt(pi)/2)
x, y = sp.symbols("x y", real=True)
f = sp.sin(x*y + y**2)
pt = {x: 1, y: sp.sqrt(sp.pi)/2}
sym = sp.simplify(sp.diff(f, x).subs(pt))
print("symbolic df/dx =", sym, " =", float(sym))
xv, yv = 1.0, np.sqrt(np.pi)/2
v3, v4 = xv*yv, yv**2
v5 = v3 + v4
adjoint = np.cos(v5)*yv          # reverse pass: vbar5 = cos(v5), vbar1 = vbar5 * y
numeric = (np.sin((xv + 1e-6)*yv + yv**2) - np.sin((xv - 1e-6)*yv + yv**2))/2e-6
print("reverse-mode adjoint =", adjoint, "  central difference =", numeric)
assert abs(adjoint - float(sym)) < 1e-12 and abs(adjoint - numeric) < 1e-8
```

### Problem L3.1 — Second Derivative Along a Curve
**Source:** Cambridge Mathematical Tripos  
**Statement:** Let $f: \mathbb{R}^2 \to \mathbb{R}$ be a smooth $C^2$ scalar field. Let $\mathbf{r}(s)$ be a curve parameterized by arc length $s$ with unit tangent vector $\mathbf{T}(s) = \mathbf{r}'(s)$ and principal normal vector $\mathbf{N}(s)$ with curvature $\kappa(s) = \|\mathbf{r}''(s)\|$.
Derive the second derivative of $f$ along the curve, $\frac{d^2}{ds^2} f(\mathbf{r}(s))$, in terms of the gradient $\nabla f$, Hessian matrix $\mathbf{H}_f = \nabla^2 f$, curvature $\kappa$, and vectors $\mathbf{T}, \mathbf{N}$.

#### First-Principles Intuition
Differentiating the directional rate of change along a curve requires accounting for both function landscape curvature (Hessian matrix) and curve trajectory bending (Frenet-Serret curvature).

#### Step-by-Step Solution
1. Express first derivative via chain rule:

$$
   \frac{d}{ds} f(\mathbf{r}(s)) = \nabla f(\mathbf{r}(s)) \cdot \mathbf{r}'(s) = \nabla f(\mathbf{r}(s)) \cdot \mathbf{T}(s)
$$

2. Differentiate both sides with respect to arc length $s$ using product rule:

$$
   \frac{d^2}{ds^2} f(\mathbf{r}(s)) = \frac{d}{ds} \left( \nabla f(\mathbf{r}(s)) \right) \cdot \mathbf{T}(s) + \nabla f(\mathbf{r}(s)) \cdot \mathbf{T}'(s)
$$

3. By the chain rule applied to vector field $\nabla f$:

$$
   \frac{d}{ds} \left( \nabla f(\mathbf{r}(s)) \right) = \mathbf{H}_f(\mathbf{r}(s)) \mathbf{r}'(s) = \mathbf{H}_f(\mathbf{r}(s)) \mathbf{T}(s)
$$

   where $\mathbf{H}_f$ is the $2 \times 2$ Hessian matrix of second partial derivatives.
4. By the Frenet-Serret formulas for arc-length parameterized curves:

$$
   \mathbf{T}'(s) = \kappa(s) \mathbf{N}(s)
$$

5. Substitute both expressions back into the second derivative formula:

$$
   \frac{d^2}{ds^2} f(\mathbf{r}(s)) = \left( \mathbf{H}_f \mathbf{T} \right) \cdot \mathbf{T} + \nabla f \cdot (\kappa \mathbf{N})
$$

6. Using quadratic form matrix notation $\left( \mathbf{H}_f \mathbf{T} \right) \cdot \mathbf{T} = \mathbf{T}^T \mathbf{H}_f \mathbf{T}$:

$$
   \frac{d^2}{ds^2} f(\mathbf{r}(s)) = \mathbf{T}(s)^T \mathbf{H}_f(\mathbf{r}(s)) \mathbf{T}(s) + \kappa(s) \left( \nabla f(\mathbf{r}(s)) \cdot \mathbf{N}(s) \right)
$$

$$
\text{Result: } \boxed{\frac{d^2}{ds^2} f(\mathbf{r}(s)) = \mathbf{T}^T \mathbf{H}_f \mathbf{T} + \kappa (\nabla f \cdot \mathbf{N})}
$$

#### Key Insight / Takeaway
The second variation along a path combines internal landscape curvature ($\mathbf{T}^T \mathbf{H}_f \mathbf{T}$) with path steering acceleration ($\kappa \nabla f \cdot \mathbf{N}$).

---

```python
# L3.1 - d^2/ds^2 f(r(s)) = T^T H T + kappa (grad f . N) on the unit circle
s = sp.symbols("s", real=True)
x, y = sp.symbols("x y", real=True)
f = sp.exp(x)*sp.cos(y) + x*y**2
r = sp.Matrix([sp.cos(s), sp.sin(s)])          # arc-length parameterised, kappa = 1
direct = sp.diff(f.subs({x: r[0], y: r[1]}), s, 2)
gradf = sp.Matrix([sp.diff(f, x), sp.diff(f, y)])
Hf = sp.hessian(f, (x, y))
T = sp.diff(r, s)
N = sp.diff(r, s, 2)                           # = kappa * N with kappa = 1
sub = {x: r[0], y: r[1]}
formula = (T.T*Hf.subs(sub)*T)[0] + gradf.subs(sub).dot(N)
print("residual (direct - formula) =", sp.simplify(direct - formula))
print("check at s = 0.7:", float(direct.subs(s, 0.7)), "vs", float(formula.subs(s, 0.7)))
assert sp.simplify(direct - formula) == 0
```

### Problem L3.2 — Rotation Invariance of the Gradient Norm
**Source:** Apostol, *Calculus, Volume II* / Spivak  
**Statement:** Prove coordinate invariance of the gradient operator: if $\mathbf{x} = \mathbf{Q} \mathbf{y}$ is an orthogonal change of basis ($\mathbf{Q}^T \mathbf{Q} = \mathbf{I}$), and $g(\mathbf{y}) = f(\mathbf{Q}\mathbf{y})$, show that $\|\nabla_{\mathbf{y}} g(\mathbf{y})\| = \|\nabla_{\mathbf{x}} f(\mathbf{x})\|$.

#### First-Principles Intuition
Rotating or reflecting the coordinate grid does not alter physical slope or maximum rate of elevation change.

#### Step-by-Step Solution
1. Apply the vector chain rule (Theorem 4.4(b)) to $g(\mathbf{y}) = f(\mathbf{Q}\mathbf{y})$:

$$
   \nabla_{\mathbf{y}} g(\mathbf{y}) = \mathbf{J}_{\mathbf{Q}\mathbf{y}}^T \nabla_{\mathbf{x}} f(\mathbf{Q}\mathbf{y})
$$

2. Since $\mathbf{x} = \mathbf{Q}\mathbf{y}$ is linear, its Jacobian matrix is simply $\mathbf{J} = \mathbf{Q}$.
3. Therefore:

$$
   \nabla_{\mathbf{y}} g(\mathbf{y}) = \mathbf{Q}^T \nabla_{\mathbf{x}} f(\mathbf{x})
$$

4. Compute the squared Euclidean norm of $\nabla_{\mathbf{y}} g(\mathbf{y})$:

$$
   \|\nabla_{\mathbf{y}} g(\mathbf{y})\|^2 = (\nabla_{\mathbf{y}} g)^T (\nabla_{\mathbf{y}} g) = (\mathbf{Q}^T \nabla_{\mathbf{x}} f)^T (\mathbf{Q}^T \nabla_{\mathbf{x}} f)
$$

$$
   = (\nabla_{\mathbf{x}} f)^T \mathbf{Q} \mathbf{Q}^T (\nabla_{\mathbf{x}} f)
$$

5. Since $\mathbf{Q}$ is an orthogonal matrix, $\mathbf{Q} \mathbf{Q}^T = \mathbf{I}$:

$$
   \|\nabla_{\mathbf{y}} g(\mathbf{y})\|^2 = (\nabla_{\mathbf{x}} f)^T \mathbf{I} (\nabla_{\mathbf{x}} f) = \|\nabla_{\mathbf{x}} f(\mathbf{x})\|^2
$$

6. Taking square roots yields $\|\nabla_{\mathbf{y}} g(\mathbf{y})\| = \|\nabla_{\mathbf{x}} f(\mathbf{x})\|$.

$$
\text{Result: } \boxed{\|\nabla_{\mathbf{y}} g(\mathbf{y})\| = \|\nabla_{\mathbf{x}} f(\mathbf{x})\| \quad (\text{Gradient Norm is Rotation Invariant})}
$$

#### Key Insight / Takeaway
The length of the gradient vector represents a physical invariant independent of spatial coordinate orientation.

---

```python
# L3.2 - orthogonal change of basis preserves the gradient norm
n = 4
Q, _ = np.linalg.qr(rng.standard_normal((n, n)))
c = rng.standard_normal(n)
f = lambda p: np.exp(-0.3*p @ p) + c @ np.sin(p)
gradf = lambda p: -0.6*np.exp(-0.3*p @ p)*p + c*np.cos(p)
yv = rng.standard_normal(n)
xv = Q @ yv
g_y = approx_fprime(yv, lambda w: f(Q @ w), 1e-7)
g_x = gradf(xv)
print("Q^T Q = I ?", np.allclose(Q.T @ Q, np.eye(n)))
print("grad_y g =", g_y)
print("Q^T grad_x f =", Q.T @ g_x)
print("||grad_y g|| =", np.linalg.norm(g_y), "  ||grad_x f|| =", np.linalg.norm(g_x))
assert abs(np.linalg.norm(g_y) - np.linalg.norm(g_x)) < 1e-6
```

### Problem L3.3 — Euler's Homogeneous Function Theorem
**Source:** Demidovich (No. 3225) / Apostol, *Calculus, Volume II*  
**Statement:** A scalar field $f: \mathbb{R}^n \to \mathbb{R}$ is homogeneous of degree $k \in \mathbb{R}$ if $f(t\mathbf{x}) = t^k f(\mathbf{x})$ for all $t \gt 0$ and $\mathbf{x} \in \mathbb{R}^n$. Prove Euler's Theorem for Homogeneous Functions:

$$
\mathbf{x} \cdot \nabla f(\mathbf{x}) = k f(\mathbf{x})
$$

#### First-Principles Intuition
Scaling input vectors by $t$ scales the output by $t^k$. Differentiating this scaling relationship at $t=1$ equates the directional derivative along radial direction $\mathbf{x}$ to $k f(\mathbf{x})$.

#### Step-by-Step Solution
1. Start with the homogeneity identity:

$$
   f(t x_1, t x_2, \dots, t x_n) = t^k f(x_1, x_2, \dots, x_n)
$$

2. Differentiate both sides with respect to $t$:

$$
   \frac{d}{dt} \left[ f(t \mathbf{x}) \right] = \frac{d}{dt} \left[ t^k f(\mathbf{x}) \right]
$$

3. Apply multivariable chain rule to the left-hand side, letting $u_i(t) = t x_i$:

$$
   \frac{d}{dt} \left[ f(t \mathbf{x}) \right] = \sum_{i=1}^n \frac{\partial f}{\partial u_i}(t\mathbf{x}) \frac{d u_i}{d t} = \sum_{i=1}^n \frac{\partial f}{\partial u_i}(t\mathbf{x}) x_i = \mathbf{x} \cdot \nabla f(t\mathbf{x})
$$

4. Differentiate the right-hand side with respect to $t$:

$$
   \frac{d}{dt} \left[ t^k f(\mathbf{x}) \right] = k t^{k-1} f(\mathbf{x})
$$

5. Equate both derivative expressions:

$$
   \mathbf{x} \cdot \nabla f(t\mathbf{x}) = k t^{k-1} f(\mathbf{x})
$$

6. Evaluate at $t = 1$:

$$
   \mathbf{x} \cdot \nabla f(\mathbf{x}) = k f(\mathbf{x})
$$

$$
\text{Result: } \boxed{\mathbf{x} \cdot \nabla f(\mathbf{x}) = k f(\mathbf{x}) \quad (\text{Euler's Homogeneity Theorem})}
$$

#### Key Insight / Takeaway
For degree-$k$ homogeneous functions, the radial directional derivative $D_{\mathbf{x}}f$ equals $k$ times the function value.

---

```python
# L3.3 - Euler's theorem on a degree-3 homogeneous field
x, y, z = sp.symbols("x y z", real=True)
k = 3
f = x**3 + 2*x*y*z + y**2*z + z**3
t = sp.symbols("t", positive=True)
print("homogeneity check f(t x) - t^k f(x) =",
      sp.simplify(f.subs({x: t*x, y: t*y, z: t*z}) - t**k*f))
lhs = sp.simplify(sp.Matrix([x, y, z]).dot(sp.Matrix([sp.diff(f, v) for v in (x, y, z)])))
print("x . grad f =", lhs, "   k f =", sp.expand(k*f))
assert sp.simplify(lhs - k*f) == 0
```

### Problem L3.4 — Exponential Convergence of Strongly Convex Flow
**Source:** Nocedal & Wright / Advanced Analysis  
**Statement:** Let $f: \mathbb{R}^n \to \mathbb{R}$ be $\mu$-strongly convex ($\mu \gt 0$), meaning $\langle \nabla f(\mathbf{x}) - \nabla f(\mathbf{y}), \mathbf{x} - \mathbf{y} \rangle \ge \mu \|\mathbf{x} - \mathbf{y}\|^2$ for all $\mathbf{x}, \mathbf{y}$.
Prove that the solution $\mathbf{x}(t)$ to gradient flow $\dot{\mathbf{x}}(t) = -\nabla f(\mathbf{x}(t))$ converges exponentially to the unique minimizer $\mathbf{x}^{\ast}$:

$$
\|\mathbf{x}(t) - \mathbf{x}^{\ast}\| \le \|\mathbf{x}(0) - \mathbf{x}^{\ast}\| e^{-\mu t}
$$

#### First-Principles Intuition
Strong convexity implies the landscape is everywhere curved at least like a parabola of width $1/\mu$, creating a strong restorative gradient force pulling toward the minimizer.

#### Step-by-Step Solution
1. Let $\mathbf{x}^{\ast}$ be the unique minimizer where $\nabla f(\mathbf{x}^{\ast}) = \mathbf{0}$.
2. Define Lyapunov distance function $V(t) = \frac{1}{2} \|\mathbf{x}(t) - \mathbf{x}^{\ast}\|^2$.
3. Differentiate $V(t)$ with respect to time $t$:

$$
   \frac{dV}{dt} = \frac{1}{2} \frac{d}{dt} \langle \mathbf{x}(t) - \mathbf{x}^{\ast}, \mathbf{x}(t) - \mathbf{x}^{\ast} \rangle = \langle \mathbf{x}(t) - \mathbf{x}^{\ast}, \dot{\mathbf{x}}(t) \rangle
$$

4. Substitute gradient flow equation $\dot{\mathbf{x}}(t) = -\nabla f(\mathbf{x}(t))$:

$$
   \frac{dV}{dt} = \langle \mathbf{x}(t) - \mathbf{x}^{\ast}, -\nabla f(\mathbf{x}(t)) \rangle = -\langle \nabla f(\mathbf{x}(t)) - \mathbf{0}, \mathbf{x}(t) - \mathbf{x}^{\ast} \rangle
$$

5. Since $\mathbf{0} = \nabla f(\mathbf{x}^{\ast})$, apply strong convexity inequality with $\mathbf{y} = \mathbf{x}^{\ast}$:

$$
   \langle \nabla f(\mathbf{x}(t)) - \nabla f(\mathbf{x}^{\ast}), \mathbf{x}(t) - \mathbf{x}^{\ast} \rangle \ge \mu \|\mathbf{x}(t) - \mathbf{x}^{\ast}\|^2 = 2 \mu V(t)
$$

6. Substituting into the derivative yields differential inequality:

$$
   \frac{dV}{dt} \le -2 \mu V(t)
$$

7. Integrate Grönwall's differential inequality:

$$
   V(t) \le V(0) e^{-2\mu t} \implies \frac{1}{2} \|\mathbf{x}(t) - \mathbf{x}^{\ast}\|^2 \le \frac{1}{2} \|\mathbf{x}(0) - \mathbf{x}^{\ast}\|^2 e^{-2\mu t}
$$

8. Taking square roots gives exponential convergence:

$$
   \|\mathbf{x}(t) - \mathbf{x}^{\ast}\| \le \|\mathbf{x}(0) - \mathbf{x}^{\ast}\| e^{-\mu t}
$$

$$
\text{Result: } \boxed{\|\mathbf{x}(t) - \mathbf{x}^{\ast}\| \le \|\mathbf{x}(0) - \mathbf{x}^{\ast}\| e^{-\mu t} \quad (\text{Exponential Global Convergence})}
$$

#### Key Insight / Takeaway
Strong convexity guarantees continuous gradient flow converges to the global optimum at an exponential rate $\mu$.

### Problem L3.5 — Orthogonal Trajectories to Hyperbolas
**Source:** Demidovich (No. 3210)  
**Statement:** Find the family of orthogonal trajectories to the family of level curves (hyperbolas) $x^2 - y^2 = c$.

#### First-Principles Intuition
Orthogonal trajectories intersect a family of curves at $90^\circ$ everywhere. Since level curve tangents satisfy slope $y'$, orthogonal trajectory tangents must satisfy slope $-\frac{1}{y'}$.

#### Step-by-Step Solution
1. Implicitly differentiate level curves $x^2 - y^2 = c$ with respect to $x$:

$$
   2x - 2y y' = 0 \implies y' = \frac{x}{y}
$$

2. The slope of orthogonal trajectories $y_{\text{orth}}'$ must be the negative reciprocal of level curve slope:

$$
   y_{\text{orth}}' = -\frac{1}{y'} = -\frac{y}{x}
$$

3. Set up differential equation for orthogonal trajectories:

$$
   \frac{dy}{dx} = -\frac{y}{x}
$$

4. Separate variables:

$$
   \frac{dy}{y} = -\frac{dx}{x}
$$

5. Integrate both sides:

$$
   \ln \lvert y \rvert = -\ln \lvert x \rvert + C_1 \implies \ln \lvert y \rvert + \ln \lvert x \rvert = C_1 \implies \ln \lvert xy \rvert = C_1
$$

6. Exponentiate:

$$
   x y = C \quad (C \in \mathbb{R} \setminus \{0\})
$$

$$
\text{Result: } \boxed{x y = C \quad (\text{Orthogonal Trajectories are Rectangular Hyperbolas})}
$$

#### Key Insight / Takeaway
The orthogonal trajectories to hyperbolas $x^2 - y^2 = c$ are rotated hyperbolas $x y = C$.

---

```python
# L3.5 - the curves xy = C meet x^2 - y^2 = c orthogonally
x, y = sp.symbols("x y", real=True)
g1 = sp.Matrix([sp.diff(x**2 - y**2, v) for v in (x, y)])   # normal to x^2 - y^2 = c
g2 = sp.Matrix([sp.diff(x*y, v) for v in (x, y)])           # normal to xy = C
print("grad(x^2-y^2) . grad(xy) =", sp.expand(g1.dot(g2)))
for px, py in [(2.0, 1.0), (3.0, -0.5), (0.7, 2.2)]:
    n1 = np.array([2*px, -2*py]); n2 = np.array([py, px])
    print(f"at ({px}, {py}): n1.n2 = {n1 @ n2:.3e}")
    assert abs(n1 @ n2) < 1e-12
print("the two normals are orthogonal everywhere, so the curve families are orthogonal")
```

### Problem L3.6 — All Directional Derivatives Without Differentiability
**Source:** Marsden & Tromba, *Vector Calculus*, Ch. 2.3, Exercise on non-differentiable functions with all directional derivatives
**Statement:** Let $f(x,y) = \dfrac{x^3}{x^2+y^2}$ for $(x,y) \neq (0,0)$ and $f(0,0) = 0$. Show that the (two-sided) directional derivative $D_{\mathbf{u}} f(0,0)$ exists for every unit vector $\mathbf{u}$, but $f$ is not Fréchet differentiable at $(0,0)$.

#### First-Principles Intuition
Restricted to any line through the origin, $f$ is a smooth (in fact linear-in-$h$) function of the parameter $h$, so every directional derivative exists as a genuine two-sided limit — matching Definition 3.3 of `first_principles.ipynb`. But the resulting map $\mathbf{u} \mapsto D_{\mathbf{u}} f(0,0)$ is cubic in $u_1$, not linear, so it cannot be written as $\nabla f(0,0) \cdot \mathbf{u}$: this is exactly the situation the Fréchet definition rules out.

#### Step-by-Step Solution
1. Compute the two-sided directional derivative along a unit vector $\mathbf{u} = (u_1, u_2)^T$, $u_1^2+u_2^2=1$:

$$
   D_{\mathbf{u}} f(0,0) = \lim_{h \to 0} \frac{f(hu_1, hu_2) - f(0,0)}{h} = \lim_{h \to 0} \frac{1}{h}\cdot\frac{h^3 u_1^3}{h^2(u_1^2+u_2^2)} = \lim_{h \to 0} \frac{h^3 u_1^3}{h^3} = u_1^3.
$$

   The ratio is $u_1^3$ for every $h \neq 0$ (no dependence on the sign of $h$), so the limit exists from both sides and $D_{\mathbf{u}} f(0,0) = u_1^3$.
2. Test linearity in $\mathbf{u}$: if $f$ were Fréchet differentiable at $(0,0)$, Theorem 4.1 of `first_principles.ipynb` would force $D_{\mathbf{u}} f(0,0) = \nabla f(0,0)\cdot\mathbf{u}$, a **linear** function of $\mathbf{u}$. But $u_1^3$ is not linear: e.g. for $\mathbf{u}=(1,0)$, $D_{\mathbf{u}}f = 1$, while for $-\mathbf{u}=(-1,0)$, $D_{-\mathbf{u}}f = -1 = -D_{\mathbf{u}}f$ (consistent with linearity), but for $\mathbf{u} = (1/\sqrt2, 1/\sqrt2)$, $D_{\mathbf{u}}f = (1/\sqrt2)^3 = 1/(2\sqrt2) \approx 0.354$, whereas a linear map matching the axis values $D_{(1,0)}f=1,\ D_{(0,1)}f=0$ would predict $\nabla f(0,0)\cdot\mathbf{u} = 1/\sqrt2 \approx 0.707 \neq 0.354$.
3. Therefore no linear map reproduces all directional derivatives, so $f$ is not Fréchet differentiable at $(0,0)$, even though every directional derivative exists.

$$
\text{Result: } \boxed{D_{\mathbf{u}} f(0,0) = u_1^3 \text{ exists } \forall \|\mathbf{u}\|=1, \text{ but } u_1^3 \text{ is non-linear in } \mathbf{u} \implies f \text{ NOT Fréchet differentiable at } (0,0).}
$$

#### Key Insight / Takeaway
All directional derivatives existing is strictly weaker than Fréchet differentiability: the master formula $D_{\mathbf{u}}f = \nabla f \cdot \mathbf{u}$ (Theorem 4.1) requires the map $\mathbf{u} \mapsto D_{\mathbf{u}}f$ to be linear, and this example shows that requirement can fail even when every individual directional derivative is a genuine two-sided limit.

---

```python
# L3.6 - D_u f(0,0) = u1^3 exists in every direction, but is not linear in u
def f_L36(p):
    x, y = p
    d = x*x + y*y
    return 0.0 if d == 0.0 else x**3/d

print(f"{'theta':>8}{'D_u f (h>0)':>14}{'D_u f (h<0)':>14}{'u1^3':>10}{'<grad,u>':>11}")
grad0 = np.array([1.0, 0.0])       # the two partials at the origin
for th in [0.0, np.pi/6, np.pi/4, np.pi/3, np.pi/2]:
    u = np.array([np.cos(th), np.sin(th)])
    hp, hm = 1e-7, -1e-7
    Dp = f_L36(hp*u)/hp
    Dm = f_L36(hm*u)/hm
    print(f"{th:8.4f}{Dp:14.8f}{Dm:14.8f}{u[0]**3:10.6f}{grad0 @ u:11.6f}")
    assert abs(Dp - Dm) < 1e-12            # genuine two-sided limit
    assert abs(Dp - u[0]**3) < 1e-9        # equals u1^3
u45 = np.array([1.0, 1.0])/np.sqrt(2)
print("gap at 45 deg: <grad,u> - u1^3 =", grad0 @ u45 - u45[0]**3)
assert abs(grad0 @ u45 - u45[0]**3) > 0.3  # not linear, so not Frechet differentiable
```

### Problem L3.7 — Projected Gradient on the Unit Sphere
**Source:** Nocedal & Wright / Advanced Optimization  
**Statement:** Derive the projected gradient operator $\nabla_{\mathcal{S}} f(\mathbf{x})$ for optimizing scalar field $f(\mathbf{x})$ subject to spherical constraint $\mathcal{S} = \{ \mathbf{x} \in \mathbb{R}^n \mid \|\mathbf{x}\|^2 = 1 \}$.

#### First-Principles Intuition
Constrained optimization on a manifold restricts updates to the tangent space. The projected gradient is obtained by subtracting the normal component of the unconstrained gradient from $\nabla f(\mathbf{x})$.

#### Step-by-Step Solution
1. The constraint manifold is sphere $c(\mathbf{x}) = \|\mathbf{x}\|^2 - 1 = 0$.
2. The unit normal vector to sphere at point $\mathbf{x} \in \mathcal{S}$ is $\mathbf{n} = \frac{\nabla c(\mathbf{x})}{\|\nabla c(\mathbf{x})\|} = \frac{2\mathbf{x}}{2\|\mathbf{x}\|} = \mathbf{x}$ (since $\|\mathbf{x}\| = 1$).
3. The orthogonal projection matrix onto the normal space is $\mathbf{P}_{\mathbf{n}} = \mathbf{x} \mathbf{x}^T$.
4. The orthogonal projection matrix onto the tangent plane $T_{\mathbf{x}}\mathcal{S}$ is:

$$
   \mathbf{P}_{T} = \mathbf{I} - \mathbf{x} \mathbf{x}^T
$$

5. Project the unconstrained gradient $\nabla f(\mathbf{x})$ onto $T_{\mathbf{x}}\mathcal{S}$:

$$
   \nabla_{\mathcal{S}} f(\mathbf{x}) = \mathbf{P}_{T} \nabla f(\mathbf{x}) = (\mathbf{I} - \mathbf{x} \mathbf{x}^T) \nabla f(\mathbf{x}) = \nabla f(\mathbf{x}) - (\mathbf{x} \cdot \nabla f(\mathbf{x})) \mathbf{x}
$$

$$
\text{Result: } \boxed{\nabla_{\mathcal{S}} f(\mathbf{x}) = \nabla f(\mathbf{x}) - (\mathbf{x}^T \nabla f(\mathbf{x})) \mathbf{x}}
$$

#### Key Insight / Takeaway
Riemannian gradient descent on spheres updates parameters along $\mathbf{x}_{k+1} = \operatorname{proj}\left(\mathbf{x}_k - \eta \nabla_{\mathcal{S}} f(\mathbf{x}_k)\right)$.

---

```python
# L3.7 - the projected gradient (I - x x^T) grad f is tangent to the unit sphere
n = 5
A = rng.standard_normal((n, n)); A = A + A.T
xs = rng.standard_normal(n); xs /= np.linalg.norm(xs)
gradf = A @ xs                       # gradient of (1/2) x^T A x
proj = gradf - (xs @ gradf)*xs
P = np.eye(n) - np.outer(xs, xs)
print("||x|| =", np.linalg.norm(xs))
print("projected gradient =", proj)
print("P grad f           =", P @ gradf)
print("x . projected grad =", xs @ proj, " (tangency)")
assert np.allclose(proj, P @ gradf)
assert abs(xs @ proj) < 1e-12
assert np.allclose(P @ P, P) and np.allclose(P, P.T)   # P is an orthogonal projector
```

### Problem L3.8 — Matrix Gradient of a Trace Quadratic
**Source:** Cambridge Mathematical Tripos / Matrix Analysis  
**Statement:** Let $\mathbf{A} \in \mathbb{R}^{n \times n}$ be a symmetric matrix ($\mathbf{A} = \mathbf{A}^T$). Consider matrix objective function $f(\mathbf{X}) = \frac{1}{2} \operatorname{tr}(\mathbf{X}^T \mathbf{A} \mathbf{X})$ defined over real matrices $\mathbf{X} \in \mathbb{R}^{n \times n}$. Compute matrix gradient $\nabla_{\mathbf{X}} f(\mathbf{X})$.

#### First-Principles Intuition
Using trace differential properties $\text{d}\operatorname{tr}(\mathbf{Z}) = \operatorname{tr}(\text{d}\mathbf{Z})$, matrix derivatives follow inner product matching $\text{d}f = \langle \nabla_{\mathbf{X}} f, \text{d}\mathbf{X} \rangle_F = \operatorname{tr}((\nabla_{\mathbf{X}} f)^T \text{d}\mathbf{X})$.

#### Step-by-Step Solution
1. Take differential of $f(\mathbf{X}) = \frac{1}{2} \operatorname{tr}(\mathbf{X}^T \mathbf{A} \mathbf{X})$:

$$
   \text{d} f = \frac{1}{2} \operatorname{tr}\left( (\text{d}\mathbf{X})^T \mathbf{A} \mathbf{X} + \mathbf{X}^T \mathbf{A} (\text{d}\mathbf{X}) \right)
$$

2. Use trace linearity and cyclic property $\operatorname{tr}(\mathbf{U}\mathbf{V}) = \operatorname{tr}(\mathbf{V}\mathbf{U})$:

$$
   \operatorname{tr}\left( (\text{d}\mathbf{X})^T \mathbf{A} \mathbf{X} \right) = \operatorname{tr}\left( \mathbf{A} \mathbf{X} (\text{d}\mathbf{X})^T \right) = \operatorname{tr}\left( (\mathbf{X}^T \mathbf{A}^T \text{d}\mathbf{X})^T \right) = \operatorname{tr}\left( \mathbf{X}^T \mathbf{A}^T \text{d}\mathbf{X} \right)
$$

3. Combine trace terms:

$$
   \text{d} f = \frac{1}{2} \operatorname{tr}\left( \mathbf{X}^T \mathbf{A}^T \text{d}\mathbf{X} + \mathbf{X}^T \mathbf{A} \text{d}\mathbf{X} \right) = \operatorname{tr}\left( \mathbf{X}^T \left( \frac{\mathbf{A} + \mathbf{A}^T}{2} \right) \text{d}\mathbf{X} \right)
$$

4. Since $\mathbf{A}$ is symmetric ($\mathbf{A} = \mathbf{A}^T$):

$$
   \text{d} f = \operatorname{tr}\left( \mathbf{X}^T \mathbf{A} \text{d}\mathbf{X} \right) = \operatorname{tr}\left( (\mathbf{A}\mathbf{X})^T \text{d}\mathbf{X} \right)
$$

5. Comparing with standard Frobenius differential definition $\text{d} f = \operatorname{tr}\left( (\nabla_{\mathbf{X}} f)^T \text{d}\mathbf{X} \right)$:

$$
   \nabla_{\mathbf{X}} f(\mathbf{X}) = \mathbf{A}\mathbf{X}
$$

$$
\text{Result: } \boxed{\nabla_{\mathbf{X}} \left( \frac{1}{2} \operatorname{tr}(\mathbf{X}^T \mathbf{A} \mathbf{X}) \right) = \mathbf{A}\mathbf{X}}
$$

#### Key Insight / Takeaway
Matrix gradient formulas directly generalize vector matrix products via trace inner products.

---

```python
# L3.8 - grad_X (1/2) tr(X^T A X) = A X for symmetric A, checked entrywise
n = 4
A = rng.standard_normal((n, n)); A = A + A.T
X = rng.standard_normal((n, n))
fX = lambda M: 0.5*np.trace(M.T @ A @ M)
analytic = A @ X
numeric = np.zeros((n, n))
eps = 1e-6
for i in range(n):
    for j in range(n):
        E = np.zeros((n, n)); E[i, j] = eps
        numeric[i, j] = (fX(X + E) - fX(X - E))/(2*eps)
print("max |A X - finite difference| =", np.max(np.abs(analytic - numeric)))
assert np.max(np.abs(analytic - numeric)) < 1e-6
```

### Problem L3.9 — Jacobi's Formula for the Determinant
**Source:** Matrix Calculus / Goodfellow et al.  
**Statement:** Compute directional derivative of matrix function $f(\mathbf{X}) = \text{det}(\mathbf{X})$ for invertible matrix $\mathbf{X} \in \mathbb{R}^{n \times n}$ along direction matrix $\mathbf{H} \in \mathbb{R}^{n \times n}$ (Jacobi's Formula).

#### First-Principles Intuition
Perturbing matrix $\mathbf{X}$ to $\mathbf{X} + t\mathbf{H} = \mathbf{X}(\mathbf{I} + t \mathbf{X}^{-1}\mathbf{H})$ isolates determinant change into characteristic polynomial eigenvalues.

#### Step-by-Step Solution
1. Write directional derivative limit definition:

$$
   D_{\mathbf{H}} f(\mathbf{X}) = \lim_{t \to 0} \frac{\text{det}(\mathbf{X} + t\mathbf{H}) - \text{det}(\mathbf{X})}{t}
$$

2. Factor out $\mathbf{X}$ using determinant multiplicative property:

$$
   \text{det}(\mathbf{X} + t\mathbf{H}) = \text{det}\left( \mathbf{X} (\mathbf{I} + t \mathbf{X}^{-1}\mathbf{H}) \right) = \text{det}(\mathbf{X}) \text{det}(\mathbf{I} + t \mathbf{X}^{-1}\mathbf{H})
$$

3. Expand $\text{det}(\mathbf{I} + t \mathbf{M})$ for matrix $\mathbf{M} = \mathbf{X}^{-1}\mathbf{H}$ using eigenvalues $\lambda_i$:

$$
   \text{det}(\mathbf{I} + t \mathbf{M}) = \prod_{i=1}^n (1 + t \lambda_i) = 1 + t \sum_{i=1}^n \lambda_i + O(t^2) = 1 + t \operatorname{tr}(\mathbf{M}) + O(t^2)
$$

4. Substitute expansion into determinant formula:

$$
   \text{det}(\mathbf{X} + t\mathbf{H}) = \text{det}(\mathbf{X}) \left( 1 + t \operatorname{tr}(\mathbf{X}^{-1}\mathbf{H}) + O(t^2) \right)
$$

5. Subtract $\text{det}(\mathbf{X})$ and divide by $t$:

$$
   \frac{\text{det}(\mathbf{X} + t\mathbf{H}) - \text{det}(\mathbf{X})}{t} = \text{det}(\mathbf{X}) \operatorname{tr}(\mathbf{X}^{-1}\mathbf{H}) + O(t)
$$

6. Take limit as $t \to 0$:

$$
   D_{\mathbf{H}} \text{det}(\mathbf{X}) = \text{det}(\mathbf{X}) \operatorname{tr}(\mathbf{X}^{-1}\mathbf{H})
$$

7. Note that matrix gradient is $\nabla_{\mathbf{X}} \text{det}(\mathbf{X}) = \text{det}(\mathbf{X}) (\mathbf{X}^{-1})^T = \operatorname{adj}(\mathbf{X})^T$.

$$
\text{Result: } \boxed{D_{\mathbf{H}} \text{det}(\mathbf{X}) = \text{det}(\mathbf{X}) \operatorname{tr}(\mathbf{X}^{-1}\mathbf{H}) \quad (\text{Jacobi's Formula})}
$$

#### Key Insight / Takeaway
The derivative of determinant is trace of matrix inverse multiplied by determinant.

---

```python
# L3.9 - Jacobi's formula: D_H det(X) = det(X) tr(X^{-1} H)
n = 4
X = rng.standard_normal((n, n)) + 2*np.eye(n)
H = rng.standard_normal((n, n))
analytic = np.linalg.det(X)*np.trace(np.linalg.solve(X, H))
eps = 1e-6
numeric = (np.linalg.det(X + eps*H) - np.linalg.det(X - eps*H))/(2*eps)
print("det(X) =", np.linalg.det(X))
print("det(X) tr(X^{-1} H) =", analytic)
print("central difference  =", numeric)
print("relative error      =", abs(analytic - numeric)/abs(analytic))
assert abs(analytic - numeric)/abs(analytic) < 1e-7
# the matrix gradient is det(X) (X^{-1})^T = adj(X)^T
G = np.linalg.det(X)*np.linalg.inv(X).T
assert abs(np.sum(G*H) - analytic) < 1e-8
```

### Problem L3.10 — The Gradient as a Riesz Representative
**Source:** Functional Analysis / Spivak  
**Statement:** Let $\mathcal{H}$ be a real Hilbert space with inner product $\langle \cdot, \cdot \rangle$. Let $f: \mathcal{H} \to \mathbb{R}$ be Fréchet differentiable at $\mathbf{x} \in \mathcal{H}$. Use the Riesz Representation Theorem to prove the existence and uniqueness of the gradient vector $\nabla f(\mathbf{x}) \in \mathcal{H}$ satisfying $D f(\mathbf{x})[\mathbf{h}] = \langle \nabla f(\mathbf{x}), \mathbf{h} \rangle$ for all $\mathbf{h} \in \mathcal{H}$.

#### First-Principles Intuition
The Fréchet differential $Df(\mathbf{x})$ is a bounded linear functional on Hilbert space. The Riesz Representation Theorem guarantees that every bounded linear functional corresponds uniquely to inner product with a fixed vector (the gradient).

#### Step-by-Step Solution
1. By definition of Fréchet differentiability, $L = Df(\mathbf{x}): \mathcal{H} \to \mathbb{R}$ is a linear mapping:

$$
   L(a \mathbf{h}_1 + b \mathbf{h}_2) = a L(\mathbf{h}_1) + b L(\mathbf{h}_2)
$$

2. Furthermore, $L$ is a bounded (continuous) linear functional on $\mathcal{H}$, meaning there exists $M \gt 0$ such that $\lvert L(\mathbf{h}) \rvert \le M \|\mathbf{h}\|$ for all $\mathbf{h} \in \mathcal{H}$.
3. State the Riesz Representation Theorem:
   *For every bounded linear functional $L$ on a Hilbert space $\mathcal{H}$, there exists a UNIQUE element $\mathbf{g} \in \mathcal{H}$ such that $L(\mathbf{h}) = \langle \mathbf{g}, \mathbf{h} \rangle$ for all $\mathbf{h} \in \mathcal{H}$, with $\|\mathbf{g}\| = \|L\|_{op}$.*
4. Define the gradient vector $\nabla f(\mathbf{x})$ as this unique element $\mathbf{g} \in \mathcal{H}$.
5. Therefore:

$$
   Df(\mathbf{x})[\mathbf{h}] = \langle \nabla f(\mathbf{x}), \mathbf{h} \rangle \quad \forall \mathbf{h} \in \mathcal{H}
$$

6. Uniqueness: Suppose $\mathbf{g}_1, \mathbf{g}_2 \in \mathcal{H}$ both satisfy the inner product representation. Then $\langle \mathbf{g}_1 - \mathbf{g}_2, \mathbf{h} \rangle = 0$ for all $\mathbf{h} \in \mathcal{H}$. Setting $\mathbf{h} = \mathbf{g}_1 - \mathbf{g}_2$ yields $\|\mathbf{g}_1 - \mathbf{g}_2\|^2 = 0 \implies \mathbf{g}_1 = \mathbf{g}_2$.

$$
\text{Result: } \boxed{\exists ! \ \nabla f(\mathbf{x}) \in \mathcal{H} \text{ such that } Df(\mathbf{x})[\mathbf{h}] = \langle \nabla f(\mathbf{x}), \mathbf{h} \rangle \quad (\text{Riesz Representation})}
$$

#### Key Insight / Takeaway
The gradient is fundamentally the Riesz dual representation of the Fréchet total differential operator under a choice of inner product.
