### Problem L0.1 — Classifying Type I, Type II, and Hybrid
**Source**: Stewart, *Calculus: Early Transcendentals*, Ch. 7.8  

#### Statement
Classify each of the following integrals as a Type I improper integral, a Type II improper integral, or a hybrid (combination of Type I and Type II). Identify all points of singular behavior or unbounded domains:
1. $I_1 = \int_0^\infty \frac{dx}{x^2 + 1}$
2. $I_2 = \int_0^1 \frac{dx}{\sqrt{x}}$
3. $I_3 = \int_0^\infty \frac{dx}{x^{\frac{1}{2}} (x + 1)}$

#### Intuition
Improper integrals are classified by the nature of the obstacle preventing standard Riemann integration on a compact interval:
- Type I integrals have an infinite limit of integration ($[a, \infty)$, $(-\infty, b]$, or $(-\infty, \infty)$).
- Type II integrals have an integrand that approaches $\pm \infty$ at one or more points in the interval of integration.
- Hybrid integrals exhibit both infinite limits of integration and interior/endpoint singularities.

#### Solution
1. **Analysis of $I_1$**:
   - The integrand $f(x) = \frac{1}{x^2 + 1}$ is continuous and bounded on $[0, \infty)$ since $0 \lt f(x) \le 1$ for all $x \ge 0$.
   - The domain $[0, \infty)$ is infinite.
   - Therefore, $I_1$ is a **Type I improper integral**.

2. **Analysis of $I_2$**:
   - The domain of integration is the finite interval $[0, 1]$.
   - As $x \to 0^+$, the integrand $g(x) = \frac{1}{\sqrt{x}} \to \infty$.
   - Therefore, $I_2$ has a singularity at the lower limit $x = 0$, making it a **Type II improper integral**.

3. **Analysis of $I_3$**:
   - As $x \to 0^+$, $h(x) = \frac{1}{\sqrt{x}(x+1)} \sim \frac{1}{\sqrt{x}} \to \infty$ (Type II singularity at $x = 0$).
   - The upper limit of integration is $\infty$ (Type I unbounded domain).
   - Thus, $I_3$ must be split into two integrals: $\int_0^1 h(x) \, dx + \int_1^\infty h(x) \, dx$.
   - Therefore, $I_3$ is a **Hybrid (Type I + Type II) improper integral**.

$$
\boxed{I_1: \text{Type I}, \quad I_2: \text{Type II}, \quad I_3: \text{Hybrid (Type I + Type II)}}
$$

#### Key takeaway
Always check both the domain boundaries *and* the continuity of the integrand across the entire closed domain before setting up integral evaluation. Hybrid integrals must be split at an arbitrary interior point $c$ into separate Type I and Type II components.

### Problem L0.2 — The p-Test on the Half-Line
**Source**: Apostol, *Calculus Vol. 1*, Ch. 10.22  

#### Statement
Using the definition of a Type I improper integral, determine whether $I = \int_1^\infty \frac{dx}{x^p}$ converges or diverges for $p = 1$, $p = 2$, and $p = \frac{1}{2}$. Calculate its value when it converges.

#### Intuition
The $p$-test for Type I integrals asks whether the integrand decay rate $x^{-p}$ as $x \to \infty$ is fast enough for the accumulated area under the curve to remain finite as the upper bound approaches infinity.

#### Solution
By definition:

$$
\int_1^\infty \frac{dx}{x^p} = \lim_{R \to \infty} \int_1^R x^{-p} \, dx
$$

1. **For $p = 1$**:

$$
\lim_{R \to \infty} \int_1^R \frac{dx}{x} = \lim_{R \to \infty} \left[ \ln x \right]_1^R = \lim_{R \to \infty} (\ln R - 0) = \infty
$$

Thus, for $p = 1$, the integral **diverges**.

2. **For $p = 2$**:

$$
\lim_{R \to \infty} \int_1^R x^{-2} \, dx = \lim_{R \to \infty} \left[ -\frac{1}{x} \right]_1^R = \lim_{R \to \infty} \left( -\frac{1}{R} + 1 \right) = 1
$$

Thus, for $p = 2$, the integral **converges to $1$**.

3. **For $p = \frac{1}{2}$**:

$$
\lim_{R \to \infty} \int_1^R x^{-\frac{1}{2}} \, dx = \lim_{R \to \infty} \left[ 2\sqrt{x} \right]_1^R = \lim_{R \to \infty} (2\sqrt{R} - 2) = \infty
$$

Thus, for $p = \frac{1}{2}$, the integral **diverges**.

$$
\boxed{p=1: \text{Diverges}, \quad p=2: \text{Converges to } 1, \quad p=\frac{1}{2}: \text{Diverges}}
$$

#### Key takeaway
For unbounded domains $[1, \infty)$, $x^{-p}$ decays fast enough if and only if $p \gt 1$. The threshold case $p = 1$ gives logarithmic divergence.

---

```python
# p-test on [1, inf): converge iff p > 1, with value 1/(p-1).
for p in (1.0, 2.0, 0.5):
    partial = [integrate.quad(lambda x: x**(-p), 1, R)[0] for R in (1e2, 1e4, 1e6)]
    print(f"p={p}:  partial integrals at R=1e2,1e4,1e6 -> "
          + ", ".join(f"{v:.6f}" for v in partial))
assert abs(integrate.quad(lambda x: x**(-2.0), 1, np.inf)[0] - 1.0) < 1e-10
print("\nchecked: p=2 converges to 1; p=1 and p=0.5 grow without bound")
```

### Problem L0.3 — The p-Test near a Singular Endpoint
**Source**: Apostol, *Calculus Vol. 1*, Ch. 10.23  

#### Statement
Using the definition of a Type II improper integral, evaluate $J = \int_0^1 \frac{dx}{x^p}$ for $p = \frac{1}{2}$, $p = 1$, and $p = \frac{3}{2}$.

#### Intuition
For Type II integrals on $(0, 1]$, the singularity is at $x = 0$. Here, we test whether the integrand grows slowly enough near $0$ so that the region under the singularity has finite area.

#### Solution
By definition:

$$
\int_0^1 \frac{dx}{x^p} = \lim_{\epsilon \to 0^+} \int_\epsilon^1 x^{-p} \, dx
$$

1. **For $p = \frac{1}{2}$**:

$$
\lim_{\epsilon \to 0^+} \int_\epsilon^1 x^{-\frac{1}{2}} \, dx = \lim_{\epsilon \to 0^+} \left[ 2\sqrt{x} \right]_\epsilon^1 = \lim_{\epsilon \to 0^+} (2 - 2\sqrt{\epsilon}) = 2
$$

Thus, for $p = \frac{1}{2}$, the integral **converges to $2$**.

2. **For $p = 1$**:

$$
\lim_{\epsilon \to 0^+} \int_\epsilon^1 \frac{dx}{x} = \lim_{\epsilon \to 0^+} \left[ \ln x \right]_\epsilon^1 = \lim_{\epsilon \to 0^+} (0 - \ln\epsilon) = +\infty
$$

Thus, for $p = 1$, the integral **diverges**.

3. **For $p = \frac{3}{2}$**:

$$
\lim_{\epsilon \to 0^+} \int_\epsilon^1 x^{-\frac{3}{2}} \, dx = \lim_{\epsilon \to 0^+} \left[ -2 x^{-\frac{1}{2}} \right]_\epsilon^1 = \lim_{\epsilon \to 0^+} \left( -2 + \frac{2}{\sqrt{\epsilon}} \right) = +\infty
$$

Thus, for $p = \frac{3}{2}$, the integral **diverges**.

$$
\boxed{p=\frac{1}{2}: \text{Converges to } 2, \quad p=1: \text{Diverges}, \quad p=\frac{3}{2}: \text{Diverges}}
$$

#### Key takeaway
The $p$-test behavior for Type II integrals $(0, 1]$ is exact opposite of Type I integrals $[1, \infty)$: $\int_0^1 x^{-p} \, dx$ converges if and only if $p \lt 1$.

---

```python
# p-test on (0, 1]: converge iff p < 1, with value 1/(1-p).
for p in (0.5, 1.0, 1.5):
    partial = [integrate.quad(lambda x, p=p: x**(-p), eps, 1)[0] for eps in (1e-2, 1e-4, 1e-6)]
    print(f"p={p}:  partial integrals from eps=1e-2,1e-4,1e-6 -> "
          + ", ".join(f"{v:.6f}" for v in partial))
assert abs(integrate.quad(lambda x: x**(-0.5), 0, 1)[0] - 2.0) < 1e-9
print("\nchecked: p=1/2 converges to 2; p=1 and p=3/2 blow up as eps -> 0")
```

### Problem L0.4 — Principal Value versus Standard Convergence
**Source**: Spivak, *Calculus* (4th ed.), Ch. 19  

#### Statement
Evaluate the standard improper integral $I = \int_{-\infty}^\infty x \, dx$ and its Cauchy Principal Value $\operatorname{P.V.} \int_{-\infty}^\infty x \, dx$. Explain why they differ.

#### Intuition
Standard improper integration over $(-\infty, \infty)$ requires that limits to $-\infty$ and $+\infty$ approach independently. Cauchy Principal Value enforces symmetric limits $(-R, R)$, which can cancel out odd symmetric divergences.

#### Solution
1. **Standard Improper Integral**:
By definition, for any chosen point $c = 0$:

$$
\int_{-\infty}^\infty x \, dx = \lim_{L \to -\infty} \int_L^0 x \, dx + \lim_{R \to \infty} \int_0^R x \, dx
$$

Evaluating the limits:

$$
\lim_{R \to \infty} \int_0^R x \, dx = \lim_{R \to \infty} \frac{R^2}{2} = +\infty
$$

Since one of the independent limits does not exist (diverges to $+\infty$), the standard improper integral **diverges** (undefined).

2. **Cauchy Principal Value**:
By definition:

$$
\operatorname{P.V.} \int_{-\infty}^\infty x \, dx = \lim_{R \to \infty} \int_{-R}^R x \, dx
$$

Since $f(x) = x$ is an odd function, for every $R \gt 0$:

$$
\int_{-R}^R x \, dx = \left[ \frac{x^2}{2} \right]_{-R}^R = \frac{R^2}{2} - \frac{(-R)^2}{2} = 0
$$

Taking the limit as $R \to \infty$:

$$
\operatorname{P.V.} \int_{-\infty}^\infty x \, dx = \lim_{R \to \infty} 0 = 0
$$

$$
\boxed{\text{Standard Integral: Diverges (Undefined)}, \quad \text{Cauchy P.V.}: 0}
$$

#### Key takeaway
Cauchy Principal Value is a weaker conditional limit concept. A convergent Cauchy Principal Value does *not* imply standard improper integral convergence unless the integrand is non-negative or absolutely integrable.

---

```python
# Symmetric (principal-value) truncation cancels; the one-sided limit does not.
for R in (10.0, 100.0, 1000.0):
    pv = integrate.quad(lambda x: x, -R, R)[0]
    one_sided = integrate.quad(lambda x: x, 0, R)[0]
    print(f"R={R:>7.0f}:  int_-R^R x dx = {pv:.3e},   int_0^R x dx = {one_sided:.3e}")
print("\nPV stays at 0 while the one-sided piece diverges like R^2/2")
```

### Problem L0.5 — Gamma at Small Integers
**Source**: Demidovich, *Problems in Mathematical Analysis*, No. 2351  

#### Statement
Compute the value of the Gamma function $\Gamma(n)$ for $n = 1, 2, 5$.

#### Intuition
The Gamma function generalizes the factorial function to real and complex numbers, satisfying $\Gamma(n) = (n-1)!$ for positive integers.

#### Solution
Using the fundamental identity $\Gamma(n) = (n-1)!$:

1. **For $n = 1$**:

$$
\Gamma(1) = (1-1)! = 0! = 1
$$

Alternatively, from the definition: $\Gamma(1) = \int_0^\infty e^{-t} \, dt = \left[-e^{-t}\right]_0^\infty = 1$.

2. **For $n = 2$**:

$$
\Gamma(2) = (2-1)! = 1! = 1
$$

3. **For $n = 5$**:

$$
\Gamma(5) = (5-1)! = 4! = 4 \times 3 \times 2 \times 1 = 24
$$

$$
\boxed{\Gamma(1) = 1, \quad \Gamma(2) = 1, \quad \Gamma(5) = 24}
$$

#### Key takeaway
The shift in indices $\Gamma(n) = (n-1)!$ is a historical artifact of Euler's definition. Always remember to subtract $1$ when evaluating $\Gamma(n)$ for integer $n$.

---

```python
# Gamma(n) = (n-1)! at n = 1, 2, 5, both from the integral and from scipy.
from math import factorial
for n in (1, 2, 5):
    quad_val = integrate.quad(lambda t, n=n: t**(n - 1) * np.exp(-t), 0, np.inf)[0]
    exact = float(factorial(n - 1))
    print(f"n={n}: integral={quad_val:.10f}  scipy.gamma={special.gamma(n):.10f}  (n-1)!={exact:.0f}")
    assert abs(quad_val - exact) < 1e-8
print("\nall three agree to 1e-8")
```

### Problem L0.6 — Direct Comparison against a Cubic Tail
**Source**: Stewart, *Calculus*, Ch. 7.8  

#### Statement
Suppose $0 \le f(x) \le \frac{1}{x^3 + \sqrt{x}}$ for all $x \ge 1$. Determine whether $\int_1^\infty f(x) \, dx$ converges using the Direct Comparison Test.

#### Intuition
To show convergence via DCT, we must bound $f(x)$ above by a known convergent benchmark function $g(x)$. For large $x$, $x^3 + \sqrt{x} \gt x^3$.

#### Solution
For all $x \ge 1$:

$$
x^3 + \sqrt{x} \gt x^3 \gt 0 \implies \frac{1}{x^3 + \sqrt{x}} \lt \frac{1}{x^3}
$$

Thus, for all $x \ge 1$:

$$
0 \le f(x) \le \frac{1}{x^3 + \sqrt{x}} \lt \frac{1}{x^3}
$$

We know by the $p$-test that $\int_1^\infty \frac{dx}{x^3}$ converges (since $p = 3 \gt 1$).

By the Direct Comparison Test (DCT), since $0 \le f(x) \lt x^{-3}$ and $\int_1^\infty x^{-3} \, dx$ converges, the integral $\int_1^\infty f(x) \, dx$ must also **converge**.

$$
\boxed{\text{Converges by DCT with } g(x) = \frac{1}{x^3}}
$$

#### Key takeaway
Dropping positive lower-order terms in the denominator creates a larger fraction, providing a clean upper bound for DCT.

---

```python
# The comparison bound and the benchmark integral, both evaluated.
g = integrate.quad(lambda x: x**(-3.0), 1, np.inf)[0]
h = integrate.quad(lambda x: 1.0 / (x**3 + np.sqrt(x)), 1, np.inf)[0]
print(f"int_1^inf x^-3 dx            = {g:.10f}   (= 1/2, p = 3 > 1)")
print(f"int_1^inf dx/(x^3 + sqrt(x)) = {h:.10f}")
assert h < g and abs(g - 0.5) < 1e-10
print("the dominated integral is finite and strictly below the benchmark, as DCT predicts")
```

### Problem L1.1 — Convergence without a Vanishing Integrand
**Source**: Spivak, *Calculus*, Ch. 19  

#### Statement
True or False: "If $f(x)$ is continuous on $[0, \infty)$ and $\int_0^\infty f(x) \, dx$ converges, then $\lim_{x \to \infty} f(x) = 0$." Prove your statement or provide a explicit counterexample.

#### Intuition
For infinite series $\sum a_n$, convergence forces $a_n \to 0$. However, for continuous integrals, area is width times height; if a function forms infinitely many tall but extremely narrow triangular spikes whose areas shrink rapidly, the integral can converge while the function height does not vanish.

#### Solution
The statement is **FALSE**.

**Counterexample Construction**:
Construct a continuous non-negative function $f(x)$ on $[0, \infty)$ consisting of triangular spikes centered at integers $n = 2, 3, 4, \dots$ with height $1$ and base width $\frac{2}{n^2}$.

Specifically, for each integer $n \ge 2$, define a triangle with vertices at $\left(n - \frac{1}{n^2}, 0\right)$, $(n, 1)$, and $\left(n + \frac{1}{n^2}, 0\right)$. Outside these triangular regions, set $f(x) = 0$.

1. **Area under $f(x)$**:
The area of the $n$-th triangular spike is:

$$
A_n = \frac{1}{2} \times \text{base} \times \text{height} = \frac{1}{2} \left( \frac{2}{n^2} \right) (1) = \frac{1}{n^2}
$$

The total integral over $[0, \infty)$ is:

$$
\int_0^\infty f(x) \, dx = \sum_{n=2}^\infty A_n = \sum_{n=2}^\infty \frac{1}{n^2} = \frac{\pi^2}{6} - 1 \lt \infty
$$

Thus, $\int_0^\infty f(x) \, dx$ converges.

2. **Limit of $f(x)$ as $x \to \infty$**:
At every integer $n \ge 2$, $f(n) = 1$. Between spikes, $f(x) = 0$.
Therefore, $\lim_{x \to \infty} f(x)$ does not exist (and certainly does not equal $0$).

$$
\boxed{\text{FALSE. Counterexample: Triangular spikes of height } 1 \text{ and base width } \frac{2}{n^2} \text{ at } x=n.}
$$

#### Key takeaway
Unlike infinite series, convergence of an improper integral does not require $\lim_{x \to \infty} f(x) = 0$ unless $f(x)$ is uniformly continuous or monotonic.

---

```python
# Total spike area equals sum_{n>=2} 1/n^2 = pi^2/6 - 1 while f(n) = 1 for every n.
n = np.arange(2, 200_001)
area = np.sum(1.0 / n**2)
print(f"sum_(n=2)^200000 1/n^2 = {area:.10f}")
print(f"pi^2/6 - 1             = {np.pi**2 / 6 - 1:.10f}")
assert abs(area - (np.pi**2 / 6 - 1)) < 1e-4
print("finite total area, yet f(n) = 1 at every integer n >= 2, so f(x) does not tend to 0")
```

### Problem L1.2 — Symmetry of the Beta Function
**Source**: Apostol, *Calculus Vol. 2*, Ch. 11  

#### Statement
Prove the symmetry property of the Beta function: $B(x, y) = B(y, x)$ for all $x \gt 0, y \gt 0$.

#### Intuition
The Beta integral $\int_0^1 t^{x-1} (1-t)^{y-1} dt$ integrates a product of powers of $t$ and $(1-t)$. The reflection substitution $u = 1-t$ swaps the roles of $t$ and $1-t$.

#### Solution
Start with the definition:

$$
B(x, y) = \int_0^1 t^{x-1} (1-t)^{y-1} \, dt
$$

Perform the substitution $u = 1 - t$, which implies $t = 1 - u$ and $dt = -du$.
Change limits of integration:
- When $t = 0 \implies u = 1$.
- When $t = 1 \implies u = 0$.

Substituting into the integral:

$$
\begin{aligned}
B(x, y) &= \int_1^0 (1-u)^{x-1} u^{y-1} (-du) \\
&= \int_0^1 u^{y-1} (1-u)^{x-1} \, du \\
&= B(y, x) \quad \blacksquare
\end{aligned}
$$

$$
\boxed{B(x, y) = B(y, x) \quad \text{for all } x \gt 0,\ y \gt 0}
$$

#### Key takeaway
Symmetry in integral representations often emerges naturally from a simple linear reflection substitution such as $u = 1-t$ or $u = -x$.

---

```python
# B(x, y) = B(y, x) on a spread of arguments.
pairs = [(2.3, 1.7), (0.5, 4.0), (3.0, 3.0), (0.25, 0.75)]
for x, y in pairs:
    lhs = integrate.quad(lambda t: t**(x - 1) * (1 - t)**(y - 1), 0, 1)[0]
    rhs = integrate.quad(lambda t: t**(y - 1) * (1 - t)**(x - 1), 0, 1)[0]
    print(f"B({x},{y}) = {lhs:.12f}   B({y},{x}) = {rhs:.12f}   |diff| = {abs(lhs - rhs):.2e}")
    assert abs(lhs - rhs) < 1e-9
```

### Problem L1.3 — The Logarithmic p-Test
**Source**: Demidovich, *Problems in Mathematical Analysis*, No. 2362  

#### Statement
Determine the values of $p \in \mathbb{R}$ for which the integral $I = \int_2^\infty \frac{dx}{x (\ln x)^p}$ converges, and evaluate the integral when it converges.

#### Intuition
The factor $\frac{1}{x}$ is precisely the derivative of $\ln x$. Substituting $u = \ln x$ converts this logarithmic Type I integral into a standard $p$-power integral.

#### Solution
Substitute $u = \ln x \implies du = \frac{dx}{x}$.
Change limits:
- When $x = 2 \implies u = \ln 2$.
- As $x \to \infty \implies u \to \infty$.

The integral transforms to:

$$
I = \int_{\ln 2}^\infty \frac{du}{u^p}
$$

By Theorem 4.1 ($p$-test on $[\ln 2, \infty)$), this integral converges if and only if $p \gt 1$.

For $p \gt 1$:

$$
I = \lim_{R \to \infty} \left[ \frac{u^{1-p}}{1-p} \right]_{\ln 2}^R = \lim_{R \to \infty} \frac{R^{1-p} - (\ln 2)^{1-p}}{1-p} = \frac{(\ln 2)^{1-p}}{p - 1}
$$

$$
\boxed{\text{Converges for } p \gt 1 \text{ with value } \frac{1}{(p-1)(\ln 2)^{p-1}} ; \text{ Diverges for } p \le 1}
$$

#### Key takeaway
Logarithmic decay $x^{-1}(\ln x)^{-p}$ is slower than any power decay $x^{-(1+\epsilon)}$, but converges if and only if the logarithmic power satisfies $p \gt 1$.

---

```python
# Finite-window check of the antiderivative, then the R -> infinity limit.
for p in (1.5, 2.0, 3.0):
    R = 1e6
    num = integrate.quad(lambda x, p=p: 1.0 / (x * np.log(x)**p), 2, R, limit=400)[0]
    partial = ((np.log(2))**(1 - p) - (np.log(R))**(1 - p)) / (p - 1)
    closed = 1.0 / ((p - 1) * np.log(2)**(p - 1))
    print(f"p={p}: quad_2^R={num:.8f}  exact_2^R={partial:.8f}  limit={closed:.8f}")
    assert abs(num - partial) < 1e-6
print("\nthe finite windows match exactly; the limit as R -> inf is 1/((p-1)(ln 2)^(p-1))")
```

### Problem L1.4 — An Arcsine Endpoint Singularity
**Source**: Spivak, *Calculus*, Ch. 19  

#### Statement
Evaluate the improper integral $I = \int_0^1 \frac{dx}{\sqrt{1 - x^2}}$.

#### Intuition
The integrand has a Type II singularity at the upper limit $x = 1$. The antiderivative is the inverse sine function $\arcsin x$.

#### Solution
1. **Identify Singularity**: As $x \to 1^-$, $\sqrt{1-x^2} \to 0 \implies \frac{1}{\sqrt{1-x^2}} \to \infty$.
2. **Apply Limit Definition**:

$$
I = \lim_{\epsilon \to 0^+} \int_0^{1-\epsilon} \frac{dx}{\sqrt{1 - x^2}}
$$

3. **Evaluate Primitive**:

$$
\int \frac{dx}{\sqrt{1 - x^2}} = \arcsin x
$$

$$
I = \lim_{\epsilon \to 0^+} \left[ \arcsin(1-\epsilon) - \arcsin(0) \right] = \arcsin(1) - 0 = \frac{\pi}{2}
$$

$$
\boxed{\frac{\pi}{2}}
$$

#### Key takeaway
Standard inverse trigonometric antiderivatives remain fully valid for Type II improper integrals when taking appropriate limits at singular endpoints.

---

```python
val = integrate.quad(lambda x: 1.0 / np.sqrt(1 - x * x), 0, 1)[0]
print(f"int_0^1 dx/sqrt(1-x^2) = {val:.12f}")
print(f"pi/2                   = {np.pi / 2:.12f}")
assert abs(val - np.pi / 2) < 1e-8
```

### Problem L1.5 — Limit Comparison with an Oscillating Denominator
**Source**: Apostol, *Calculus Vol. 1*, Ch. 10  

#### Statement
Use the Limit Comparison Test (LCT) to determine the convergence or divergence of:

$$
I = \int_1^\infty \frac{x + 2}{x^3 + \sin^2 x} \, dx
$$

#### Intuition
For large $x$, $\sin^2 x \in [0, 1]$ becomes completely negligible compared to $x^3$, and $x + 2 \sim x$. Thus, the integrand behaves asymptotically like $\frac{x}{x^3} = \frac{1}{x^2}$.

#### Solution
1. **Choose Comparison Function**: Let $f(x) = \frac{x+2}{x^3+\sin^2 x}$ and $g(x) = \frac{1}{x^2}$. Both functions are strictly positive on $[1, \infty)$.
2. **Compute Limit of Ratio**:

$$
L = \lim_{x \to \infty} \frac{f(x)}{g(x)} = \lim_{x \to \infty} \frac{\frac{x+2}{x^3+\sin^2 x}}{\frac{1}{x^2}} = \lim_{x \to \infty} \frac{x^3 + 2x^2}{x^3 + \sin^2 x}
$$

Divide numerator and denominator by $x^3$:

$$
L = \lim_{x \to \infty} \frac{1 + \frac{2}{x}}{1 + \frac{\sin^2 x}{x^3}} = \frac{1 + 0}{1 + 0} = 1
$$

3. **Apply LCT**: Since $L = 1 \in (0, \infty)$, $\int_1^\infty f(x) \, dx$ shares the same convergence behavior as $\int_1^\infty \frac{dx}{x^2}$.
Since $p = 2 \gt 1$, $\int_1^\infty \frac{dx}{x^2}$ converges.

$$
\boxed{\text{Converges by Limit Comparison Test with } g(x) = \frac{1}{x^2}}
$$

#### Key takeaway
LCT eliminates bounded oscillating terms like $\sin^2 x$ by establishing asymptotic dominance as $x \to \infty$.

---

```python
# The limit-comparison ratio tends to 1, and the integral is finite.
x = np.array([1e1, 1e2, 1e3, 1e4])
ratio = ((x + 2) / (x**3 + np.sin(x)**2)) / (1.0 / x**2)
print("f/g at x = 10, 100, 1000, 10000:", np.round(ratio, 6))
val = integrate.quad(lambda t: (t + 2) / (t**3 + np.sin(t)**2), 1, np.inf)[0]
print(f"int_1^inf f = {val:.8f}  (finite, as LCT against 1/x^2 predicts)")
assert abs(ratio[-1] - 1.0) < 1e-3 and np.isfinite(val)
```

### Problem L1.6 — Conditional Convergence of sin x over x
**Source**: Demidovich, *Problems in Mathematical Analysis*, No. 2381  

#### Statement
Prove that $I = \int_1^\infty \frac{\sin x}{x} \, dx$ is conditionally convergent (i.e., it converges, but $\int_1^\infty \left\vert\frac{\sin x}{x}\right\vert dx$ diverges).

#### Intuition
The sign oscillations of $\sin x$ cause destructive cancellation, allowing the integral to converge via Dirichlet's test. However, taking absolute values destroys cancellation, leaving an integral that diverges like $\frac{1}{x}$.

#### Solution
1. **Proof of Convergence (Dirichlet Test)**:
Set $f(x) = \sin x$ and $g(x) = \frac{1}{x}$.
- Primitive of $f(x)$: $F(R) = \int_1^R \sin x \, dx = \cos(1) - \cos(R)$.
  Clearly, $\lvert F(R) \rvert \le \lvert\cos(1)\rvert + \lvert\cos(R)\rvert \le 2$ (bounded for all $R \ge 1$).
- Function $g(x) = \frac{1}{x}$ is monotonically decreasing to $0$ as $x \to \infty$.
By Dirichlet's Test (Theorem 4.9), $\int_1^\infty \frac{\sin x}{x} \, dx$ **converges**.

2. **Proof of Divergence of Absolute Value Integral**:
Notice $\lvert\sin x\rvert \ge \sin^2 x = \frac{1 - \cos(2x)}{2}$.
For any $R \gt 1$:

$$
\int_1^R \frac{\lvert\sin x\rvert}{x} \, dx \ge \int_1^R \frac{\sin^2 x}{x} \, dx = \frac{1}{2} \int_1^R \frac{dx}{x} - \frac{1}{2} \int_1^R \frac{\cos(2x)}{x} \, dx
$$

- $\int_1^R \frac{dx}{x} = \ln R \to \infty$ as $R \to \infty$.
- $\int_1^\infty \frac{\cos(2x)}{x} \, dx$ converges by Dirichlet's Test (bounded primitive $\frac{\sin(2R)}{2}$).
Thus, $\int_1^\infty \frac{\lvert\sin x\rvert}{x} \, dx = \infty$.

$$
\boxed{\int_1^\infty \frac{\sin x}{x} dx \text{ is Conditionally Convergent}}
$$

#### Key takeaway
Conditional convergence relies entirely on alternating signs. Squaring or taking absolute values eliminates cancellation and exposes underlying divergence.

---

```python
# sin(x)/x converges; |sin(x)|/x grows like (2/pi) ln R.
for R in (1e2, 1e3, 1e4):
    signed = integrate.quad(lambda x: np.sin(x) / x, 1, R, limit=2000)[0]
    absol = integrate.quad(lambda x: abs(np.sin(x)) / x, 1, R, limit=2000)[0]
    print(f"R={R:>8.0f}:  int sin(x)/x = {signed:.6f}   int |sin(x)|/x = {absol:.6f}   "
          f"(2/pi) ln R = {2 / np.pi * np.log(R):.6f}")
print("\nthe signed partial integrals settle; the absolute ones track (2/pi) ln R upward")
```

### Problem L1.7 — An Exponential Moment via Gamma

Evaluate $I = \int_0^\infty x^3 e^{-2x} \, dx$ using the Gamma function.

Integrals of the form $\int_0^\infty x^n e^{-k x} \, dx$ can be directly mapped to the Gamma function $\Gamma(z) = \int_0^\infty t^{z-1} e^{-t} \, dt$ via linear substitution $t = k x$.

$$

$$
\boxed{\frac{3}{8}}
$$

I = \frac{24 \times 6}{40320} = \frac{144}{40320} = \frac{1}{280}

\boxed{\frac{1}{280}}

\boxed{\ln\left(\frac{5}{2}\right)}

### Problem L1.11 — Feynman's Trick on the Damped Dirichlet Integral
**Source**: MIT 18.01 / Demidovich No. 3811  

#### Statement
Using Feynman's parameter trick (differentiation under the integral sign), evaluate for $a \gt 0$:

$$
I(a) = \int_0^\infty \frac{e^{-ax} \sin x}{x} \, dx
$$

#### Intuition
The factor $\frac{1}{x}$ makes direct integration difficult. Differentiating with respect to the parameter $a$ eliminates the denominator $x$.

#### Solution
1. **Differentiate with respect to $a$**:

$$
I'(a) = \frac{d}{da} \int_0^\infty \frac{e^{-ax} \sin x}{x} \, dx = \int_0^\infty \frac{\partial}{\partial a} \left( \frac{e^{-ax} \sin x}{x} \right) \, dx = -\int_0^\infty e^{-ax} \sin x \, dx
$$

2. **Evaluate Primitive Integral**:
Using standard integration by parts twice:

$$
\int_0^\infty e^{-ax} \sin x \, dx = \frac{1}{a^2 + 1} \implies I'(a) = -\frac{1}{a^2 + 1}
$$

3. **Integrate back to find $I(a)$**:

$$
I(a) = \int -\frac{da}{a^2 + 1} = -\arctan(a) + C
$$

4. **Determine Constant $C$**:
As $a \to \infty$, $e^{-ax} \to 0 \implies \lim_{a \to \infty} I(a) = 0$.

$$
0 = -\arctan(\infty) + C \implies 0 = -\frac{\pi}{2} + C \implies C = \frac{\pi}{2}
$$

Thus:

$$
I(a) = \frac{\pi}{2} - \arctan(a) = \operatorname{arccot}(a) = \arctan\left(\frac{1}{a}\right)
$$

$$
\boxed{\arctan\left(\frac{1}{a}\right)}
$$

#### Key takeaway
Differentiating under the integral sign converts difficult fractional integrands into standard elementary integrals.

---

```python
for a in (0.5, 1.3, 4.0):
    num = integrate.quad(lambda x, a=a: np.exp(-a * x) * np.sin(x) / x, 0, np.inf, limit=400)[0]
    print(f"a={a}: quad={num:.10f}   arctan(1/a)={np.arctan(1/a):.10f}   "
          f"|diff|={abs(num - np.arctan(1/a)):.1e}")
    assert abs(num - np.arctan(1 / a)) < 1e-7
```

### Problem L1.12 — The Shifted Gaussian Integral
**Source**: Apostol, *Calculus Vol. 2*, Ch. 11  

#### Statement
Evaluate the shifted Gaussian integral:

$$
I = \int_{-\infty}^\infty e^{-a x^2 + b x} \, dx
$$

for $a \gt 0, b \in \mathbb{R}$.

#### Intuition
Completing the square in the exponent reduces the integrand to a standard Gaussian integral $\int_{-\infty}^\infty e^{-u^2} du = \sqrt{\pi}$.

#### Solution
1. **Complete the Square in Exponent**:

$$
-a x^2 + b x = -a \left( x^2 - \frac{b}{a} x \right) = -a \left( \left( x - \frac{b}{2a} \right)^2 - \frac{b^2}{4a^2} \right) = -a \left( x - \frac{b}{2a} \right)^2 + \frac{b^2}{4a}
$$

2. **Substitute into Integral**:

$$
I = \int_{-\infty}^\infty \exp\left( -a \left( x - \frac{b}{2a} \right)^2 + \frac{b^2}{4a} \right) \, dx = e^{\frac{b^2}{4a}} \int_{-\infty}^\infty e^{-a \left( x - \frac{b}{2a} \right)^2} \, dx
$$

3. **Change of Variables**: Let $u = \sqrt{a} \left( x - \frac{b}{2a} \right) \implies dx = \frac{du}{\sqrt{a}}$.

$$
I = e^{\frac{b^2}{4a}} \int_{-\infty}^\infty e^{-u^2} \frac{du}{\sqrt{a}} = \frac{e^{\frac{b^2}{4a}}}{\sqrt{a}} \int_{-\infty}^\infty e^{-u^2} \, du
$$

4. **Use Known Standard Gaussian Value**: $\int_{-\infty}^\infty e^{-u^2} du = \sqrt{\pi}$.

$$
I = \sqrt{\frac{\pi}{a}} e^{\frac{b^2}{4a}}
$$

$$
\boxed{\sqrt{\frac{\pi}{a}} \exp\left(\frac{b^2}{4a}\right)}
$$

#### Key takeaway
Completing the square is the foundational algebraic step for evaluating all Gaussian integrals in probability theory, physics, and machine learning.

---

```python
for a, b in ((1.4, 0.7), (0.5, -2.0), (3.0, 0.0)):
    num = integrate.quad(lambda x, a=a, b=b: np.exp(-a * x * x + b * x), -np.inf, np.inf)[0]
    closed = np.sqrt(np.pi / a) * np.exp(b * b / (4 * a))
    print(f"a={a}, b={b}: quad={num:.10f}  closed={closed:.10f}  |diff|={abs(num-closed):.1e}")
    assert abs(num - closed) < 1e-8
```

### Problem L1.13 — An Interior Singularity at the Origin
**Source**: Stewart, *Calculus*, Ch. 7.8  

#### Statement
Evaluate $I = \int_{-1}^1 \frac{dx}{x^{\frac{2}{3}}}$.

#### Intuition
The integrand has an interior Type II singularity at $x = 0$. The integral must be split at $0$ into two separate improper integrals.

#### Solution
1. **Split Integral at Singularity $x = 0$**:

$$
I = \int_{-1}^0 x^{-\frac{2}{3}} \, dx + \int_0^1 x^{-\frac{2}{3}} \, dx
$$

2. **Evaluate Right-Side Integral**:

$$
\int_0^1 x^{-\frac{2}{3}} \, dx = \lim_{\epsilon_2 \to 0^+} \int_{\epsilon_2}^1 x^{-\frac{2}{3}} \, dx = \lim_{\epsilon_2 \to 0^+} \left[ 3 x^{\frac{1}{3}} \right]_{\epsilon_2}^1 = 3 - 0 = 3
$$

3. **Evaluate Left-Side Integral**:

$$
\int_{-1}^0 x^{-\frac{2}{3}} \, dx = \lim_{\epsilon_1 \to 0^+} \int_{-1}^{-\epsilon_1} x^{-\frac{2}{3}} \, dx = \lim_{\epsilon_1 \to 0^+} \left[ 3 x^{\frac{1}{3}} \right]_{-1}^{-\epsilon_1} = 0 - 3(-1) = 3
$$

4. **Combine Results**:

$$
I = 3 + 3 = 6
$$

$$
\boxed{6}
$$

#### Key takeaway
Failure to split interior singularities often leads to false claims of divergence or sign errors. Always isolate interior singular points.

---

```python
left = integrate.quad(lambda x: abs(x)**(-2 / 3), -1, 0)[0]
right = integrate.quad(lambda x: x**(-2 / 3), 0, 1)[0]
print(f"left piece = {left:.10f},  right piece = {right:.10f},  total = {left + right:.10f}")
assert abs(left + right - 6.0) < 1e-8
```

### Problem L2.1 — Gaussian Normalizer of a VAE Prior
**Source**: Machine Learning Theory / Variational Autoencoders (VAEs)  

#### Statement
In VAEs, the prior over latent variables $\mathbf{z} \in \mathbb{R}^d$ is a isotropic Gaussian $\mathcal{N}(\mathbf{0}, \mathbf{I}_d)$. Compute the normalization constant $Z = \int_{\mathbb{R}^d} \exp\left(-\frac{1}{2} \lVert \mathbf{z} \rVert^2\right) d\mathbf{z}$ from first principles.

#### Intuition
A $d$-dimensional uncoupled Gaussian integral factors into the product of $d$ independent 1D Gaussian integrals.

#### Solution
1. **Factorize Multidimensional Integral**:
Since $\lVert \mathbf{z} \rVert^2 = z_1^2 + z_2^2 + \dots + z_d^2$:

$$
Z = \int_{-\infty}^\infty \dots \int_{-\infty}^\infty \exp\left(-\frac{1}{2} \sum_{i=1}^d z_i^2\right) dz_1 dz_2 \dots dz_d = \prod_{i=1}^d \left( \int_{-\infty}^\infty e^{-\frac{1}{2} z_i^2} dz_i \right)
$$

2. **Evaluate 1D Component Integral**:
Let $I_1 = \int_{-\infty}^\infty e^{-\frac{1}{2} u^2} du$. Substitute $v = \frac{u}{\sqrt{2}} \implies du = \sqrt{2} dv$:

$$
I_1 = \int_{-\infty}^\infty e^{-v^2} (\sqrt{2} dv) = \sqrt{2} \int_{-\infty}^\infty e^{-v^2} dv = \sqrt{2} \sqrt{\pi} = \sqrt{2\pi}
$$

3. **Multiply Across $d$ Dimensions**:

$$
Z = (I_1)^d = \left(\sqrt{2\pi}\right)^d = (2\pi)^{\frac{d}{2}}
$$

$$
\boxed{(2\pi)^{\frac{d}{2}}}
$$

#### Key takeaway
The factor $(2\pi)^{-\frac{d}{2}}$ in standard Gaussian probability density functions directly balances this multidimensional improper integral.

---

```python
# (2 pi)^(d/2) by factorising the d-dimensional Gaussian into d identical 1-D integrals.
one_d = integrate.quad(lambda z: np.exp(-0.5 * z * z), -np.inf, np.inf)[0]
print(f"1-D integral = {one_d:.12f},  sqrt(2 pi) = {np.sqrt(2*np.pi):.12f}")
for d in (1, 2, 5, 10):
    print(f"d={d:>3}:  (1-D)^d = {one_d**d:.6e}   (2 pi)^(d/2) = {(2*np.pi)**(d/2):.6e}")
assert abs(one_d - np.sqrt(2 * np.pi)) < 1e-10
```

### Problem L2.2 — Closed-Form KL Divergence in the ELBO
**Source**: Kingma & Welling (2013), *Auto-Encoding Variational Bayes*  

#### Statement
Compute the analytical KL-divergence integral between a 1D Gaussian posterior $q(z) = \mathcal{N}(\mu, \sigma^2)$ and standard prior $p(z) = \mathcal{N}(0, 1)$:

$$
D_{\text{KL}}(q \parallel p) = \int_{-\infty}^\infty q(z) \ln\left(\frac{q(z)}{p(z)}\right) dz
$$

#### Intuition
Substitute the explicit Gaussian densities into the log-ratio. The improper integral then reduces to calculating expectations of $1$, $z$, and $z^2$ under $q(z)$.

#### Solution
1. **Write Log Density Ratio**:

$$
q(z) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(z-\mu)^2}{2\sigma^2}\right), \quad p(z) = \frac{1}{\sqrt{2\pi}} \exp\left(-\frac{z^2}{2}\right)
$$

$$
\ln\left(\frac{q(z)}{p(z)}\right) = \ln\left(\frac{1}{\sigma}\right) - \frac{(z-\mu)^2}{2\sigma^2} + \frac{z^2}{2} = -\ln\sigma - \frac{(z-\mu)^2}{2\sigma^2} + \frac{z^2}{2}
$$

2. **Expand Integral into Expectations under $q(z)$**:

$$
D_{\text{KL}}(q \parallel p) = \mathbb{E}_{q}\left[ -\ln\sigma - \frac{(z-\mu)^2}{2\sigma^2} + \frac{z^2}{2} \right]
$$

Using linearity of expectation:
- $\mathbb{E}_q[-\ln\sigma] = -\ln\sigma$
- $\mathbb{E}_q\left[\frac{(z-\mu)^2}{2\sigma^2}\right] = \frac{1}{2\sigma^2} \operatorname{Var}(z) = \frac{\sigma^2}{2\sigma^2} = \frac{1}{2}$
- $\mathbb{E}_q\left[\frac{z^2}{2}\right] = \frac{1}{2} \mathbb{E}_q[z^2] = \frac{1}{2} (\mu^2 + \sigma^2)$
3. **Sum Expectations**:

$$
D_{\text{KL}}(q \parallel p) = -\ln\sigma - \frac{1}{2} + \frac{1}{2}(\mu^2 + \sigma^2) = \frac{1}{2} \left( \sigma^2 + \mu^2 - 1 - \ln(\sigma^2) \right)
$$

$$
\boxed{\frac{1}{2} \left( \sigma^2 + \mu^2 - 1 - \ln(\sigma^2) \right)}
$$

#### Key takeaway
The analytic VAE loss formula is derived directly by evaluating standard Gaussian moments via improper integration.

---

```python
# KL(q || p) by direct quadrature against the closed form.
# The log-ratio is expanded analytically so the integrand stays finite in the tails.
for mu, sig in ((0.0, 1.0), (1.5, 0.5), (-0.8, 2.0)):
    def integrand(z, mu=mu, sig=sig):
        q = np.exp(-(z - mu)**2 / (2 * sig**2)) / np.sqrt(2 * np.pi * sig**2)
        log_ratio = -np.log(sig) - (z - mu)**2 / (2 * sig**2) + z**2 / 2
        return q * log_ratio
    num = integrate.quad(integrand, mu - 12 * sig, mu + 12 * sig, limit=400)[0]
    closed = 0.5 * (sig**2 + mu**2 - 1 - np.log(sig**2))
    print(f"mu={mu}, sigma={sig}: quad={num:.10f}  closed={closed:.10f}  |diff|={abs(num-closed):.1e}")
    assert abs(num - closed) < 1e-8
```

#### Statement
In Bayesian estimation, the prior over success probability $\theta$ is $\operatorname{Beta}(a, b)$ with density $p(\theta) = \frac{1}{B(a, b)} \theta^{a-1} (1-\theta)^{b-1}$. Given $k$ successes in $n$ Bernoulli trials, prove that the posterior normalization integral equals $B(a+k, b+n-k)$.

Comparing with Definition 3.11:

$$
Z = B(a+k, \, b+n-k)
$$

$$
\boxed{B(a+k, \, b+n-k)}
$$

The Beta function is the exact marginal likelihood (evidence) normalizing constant for Binomial observations under a Beta prior.

### Problem L2.4 — Mean of a Gamma Conjugate Prior
**Source**: Bayesian Probability / Poisson-Gamma Model  

#### Statement
Let $\lambda \sim \operatorname{Gamma}(\alpha, \beta)$ with probability density $p(\lambda) = \frac{\beta^\alpha}{\Gamma(\alpha)} \lambda^{\alpha-1} e^{-\beta \lambda}$ for $\lambda \gt 0$. Compute the expected value $\mathbb{E}[\lambda] = \int_0^\infty \lambda p(\lambda) \, d\lambda$.

#### Intuition
The expectation integral increases the power of $\lambda$ by $1$, shifting the effective Gamma parameter from $\alpha$ to $\alpha + 1$.

#### Solution
1. **Set up Expectation Integral**:

$$
\mathbb{E}[\lambda] = \int_0^\infty \lambda \left( \frac{\beta^\alpha}{\Gamma(\alpha)} \lambda^{\alpha-1} e^{-\beta \lambda} \right) \, d\lambda = \frac{\beta^\alpha}{\Gamma(\alpha)} \int_0^\infty \lambda^{(\alpha+1)-1} e^{-\beta \lambda} \, d\lambda
$$

2. **Evaluate Integral via Gamma Formula**:
Using $\int_0^\infty x^{n-1} e^{-k x} dx = \frac{\Gamma(n)}{k^n}$:

$$
\int_0^\infty \lambda^{(\alpha+1)-1} e^{-\beta \lambda} \, d\lambda = \frac{\Gamma(\alpha+1)}{\beta^{\alpha+1}}
$$

3. **Substitute back and Simplify**:

$$
\mathbb{E}[\lambda] = \frac{\beta^\alpha}{\Gamma(\alpha)} \cdot \frac{\Gamma(\alpha+1)}{\beta^{\alpha+1}} = \frac{\beta^\alpha}{\beta^{\alpha+1}} \cdot \frac{\alpha \Gamma(\alpha)}{\Gamma(\alpha)} = \frac{\alpha}{\beta}
$$

$$
\boxed{\frac{\alpha}{\beta}}
$$

#### Key takeaway
Moments of Gamma-distributed variables map directly to ratio of shifted Gamma functions.

---

```python
# E[lambda] = alpha / beta for the Gamma density.
for alpha, beta in ((2.0, 3.0), (5.5, 0.75)):
    dens = lambda lam: beta**alpha / special.gamma(alpha) * lam**(alpha - 1) * np.exp(-beta * lam)
    mass = integrate.quad(dens, 0, np.inf)[0]
    mean = integrate.quad(lambda lam: lam * dens(lam), 0, np.inf)[0]
    print(f"alpha={alpha}, beta={beta}: mass={mass:.10f}  E[lam]={mean:.10f}  alpha/beta={alpha/beta:.10f}")
    assert abs(mass - 1) < 1e-9 and abs(mean - alpha / beta) < 1e-9
```

### Problem L2.5 — Combining Two Propagator Denominators
**Source**: Cambridge Tripos IB / Quantum Field Theory Loop Integrals  

#### Statement
In Feynman diagram evaluation, evaluate the basic propagator integral $I = \int_0^1 \frac{dx}{[x A + (1-x) B]^2}$ for constants $A \gt 0, B \gt 0, A \neq B$.

#### Intuition
The denominator is linear in $x$. Direct substitution $u = x A + (1-x) B$ simplifies the integral into a standard power integral.

#### Solution
1. **Substitution**: Let $u = x A + (1-x) B = (A - B) x + B$.
   Then $du = (A - B) dx \implies dx = \frac{du}{A - B}$.
2. **Change Limits**:
   - When $x = 0 \implies u = B$.
   - When $x = 1 \implies u = A$.
3. **Evaluate Integral**:

$$
I = \int_B^A \frac{1}{u^2} \frac{du}{A - B} = \frac{1}{A - B} \left[ -\frac{1}{u} \right]_B^A = \frac{1}{A - B} \left( -\frac{1}{A} + \frac{1}{B} \right)
$$

4. **Simplify Fraction**:

$$
-\frac{1}{A} + \frac{1}{B} = \frac{A - B}{A B}
$$

$$
I = \frac{1}{A - B} \cdot \frac{A - B}{A B} = \frac{1}{A B}
$$

$$
\boxed{\frac{1}{A B}}
$$

#### Key takeaway
This verifies Feynman's parameter identity $\frac{1}{A B} = \int_0^1 \frac{dx}{[x A + (1-x) B]^2}$, which combines two separate propagator denominators into a single square.

---

```python
# Feynman's two-denominator identity.
for A, B in ((2.0, 7.0), (0.3, 1.9), (5.0, 5.0 + 1e-6)):
    num = integrate.quad(lambda x: 1.0 / (x * A + (1 - x) * B)**2, 0, 1)[0]
    print(f"A={A}, B={B}: integral={num:.12f}   1/(AB)={1/(A*B):.12f}")
    assert abs(num - 1 / (A * B)) < 1e-10
```

### Problem L2.6 — Maxwell-Boltzmann Normalization
**Source**: Statistical Mechanics / Maxwell-Boltzmann Distribution  

#### Statement
The Maxwell-Boltzmann speed distribution of particles of mass $m$ at temperature $T$ is $f(v) = C v^2 \exp\left(-\frac{m v^2}{2 k_B T}\right)$ for $v \ge 0$. Find the normalization constant $C$ such that $\int_0^\infty f(v) \, dv = 1$.

#### Intuition
Setting total probability to 1 requires evaluating a standard Gaussian-type integral $\int_0^\infty v^2 e^{-\alpha v^2} dv$.

#### Solution
1. **Set up Normalization Equation**:

$$
C \int_0^\infty v^2 \exp\left(-\frac{m v^2}{2 k_B T}\right) \, dv = 1
$$

Let $\alpha = \frac{m}{2 k_B T}$. We need to compute $J = \int_0^\infty v^2 e^{-\alpha v^2} \, dv$.
2. **Substitute $t = \alpha v^2$**:

$$
v = \frac{t^{\frac{1}{2}}}{\alpha^{\frac{1}{2}}} \implies dv = \frac{dt}{2 \alpha^{\frac{1}{2}} t^{\frac{1}{2}}}
$$

$$
J = \int_0^\infty \left(\frac{t}{\alpha}\right) e^{-t} \left(\frac{dt}{2 \alpha^{\frac{1}{2}} t^{\frac{1}{2}}}\right) = \frac{1}{2 \alpha^{\frac{3}{2}}} \int_0^\infty t^{\frac{1}{2}} e^{-t} \, dt
$$

3. **Evaluate via Gamma Function**:

$$
\int_0^\infty t^{\frac{1}{2}} e^{-t} \, dt = \Gamma\left(\frac{3}{2}\right) = \frac{1}{2} \Gamma\left(\frac{1}{2}\right) = \frac{\sqrt{\pi}}{2}
$$

Thus:

$$
J = \frac{1}{2 \alpha^{\frac{3}{2}}} \cdot \frac{\sqrt{\pi}}{2} = \frac{\sqrt{\pi}}{4 \alpha^{\frac{3}{2}}}
$$

4. **Substitute $\alpha = \frac{m}{2 k_B T}$ and solve for $C$**:

$$
C \cdot \frac{\sqrt{\pi}}{4 \left(\frac{m}{2 k_B T}\right)^{\frac{3}{2}}} = 1 \implies C = \frac{4}{\sqrt{\pi}} \left(\frac{m}{2 k_B T}\right)^{\frac{3}{2}} = 4\pi \left(\frac{m}{2\pi k_B T}\right)^{\frac{3}{2}}
$$

$$
\boxed{4\pi \left(\frac{m}{2\pi k_B T}\right)^{\frac{3}{2}}}
$$

#### Key takeaway
Physical kinetic distribution normalization factors are computed directly by expressing $v^2 e^{-\alpha v^2}$ integrals in terms of $\Gamma(\frac{3}{2})$.

---

```python
# Maxwell-Boltzmann normalisation, in units where m = k_B = T = 1.
m = k_B = T = 1.0
C_mb = 4 * np.pi * (m / (2 * np.pi * k_B * T))**1.5
mass = integrate.quad(lambda v: C_mb * v**2 * np.exp(-m * v**2 / (2 * k_B * T)), 0, np.inf)[0]
print(f"C = {C_mb:.12f}")
print(f"total probability = {mass:.12f}")
assert abs(mass - 1.0) < 1e-10
```

### Problem L2.7 — The Planck Blackbody Integral
**Source**: Quantum Physics / Planck Blackbody Radiation Law  

#### Statement
Evaluating total blackbody energy density requires computing the integral $I = \int_0^\infty \frac{x^3}{e^x - 1} \, dx$. Prove that $I = 6 \zeta(4) = \frac{\pi^4}{15}$.

#### Intuition
Expand the denominator $(e^x - 1)^{-1} = e^{-x} (1 - e^{-x})^{-1}$ as a geometric series $\sum_{k=1}^\infty e^{-k x}$, turning the integral into a sum of Gamma functions.

#### Solution
1. **Geometric Series Expansion**:
For $x \gt 0$, $e^{-x} \lt 1$. Thus:

$$
\frac{1}{e^x - 1} = \frac{e^{-x}}{1 - e^{-x}} = \sum_{k=1}^\infty e^{-k x}
$$

2. **Swap Integration and Summation**:

$$
I = \int_0^\infty x^3 \left( \sum_{k=1}^\infty e^{-k x} \right) \, dx = \sum_{k=1}^\infty \int_0^\infty x^3 e^{-k x} \, dx
$$

3. **Evaluate Each Term using Gamma Function**:
Let $u = k x \implies dx = \frac{du}{k}$.

$$
\int_0^\infty x^3 e^{-k x} \, dx = \int_0^\infty \left(\frac{u}{k}\right)^3 e^{-u} \frac{du}{k} = \frac{1}{k^4} \int_0^\infty u^3 e^{-u} \, du = \frac{\Gamma(4)}{k^4} = \frac{6}{k^4}
$$

4. **Sum Infinite Series**:

$$
I = \sum_{k=1}^\infty \frac{6}{k^4} = 6 \sum_{k=1}^\infty \frac{1}{k^4} = 6 \zeta(4)
$$

Using the known Euler value $\zeta(4) = \frac{\pi^4}{90}$:

$$
I = 6 \times \frac{\pi^4}{90} = \frac{\pi^4}{15}
$$

$$
\boxed{\frac{\pi^4}{15}}
$$

#### Key takeaway
Bose-Einstein integrals evaluate by converting denominators into geometric series and summing Riemann zeta function values.

---

```python
# x^3 e^-x / (1 - e^-x) is the overflow-free form of x^3 / (e^x - 1).
val = integrate.quad(lambda x: x**3 * np.exp(-x) / (1 - np.exp(-x)), 0, np.inf)[0]
print(f"int_0^inf x^3/(e^x - 1) dx = {val:.12f}")
print(f"6 zeta(4) = pi^4/15        = {np.pi**4 / 15:.12f}")
assert abs(val - np.pi**4 / 15) < 1e-9
```

### Problem L2.8 — Characteristic Function of the Laplace Law
**Source**: Probability Theory / Characteristic Functions  

#### Statement
Compute the characteristic function $\phi(t) = \int_{-\infty}^\infty e^{i t x} \frac{1}{2} e^{-\lvert x \rvert} \, dx$ of the standard Laplace distribution.

#### Intuition
Split the real line at $x = 0$ to eliminate $\lvert x \rvert$ and compute two convergent Type I exponential integrals.

#### Solution
1. **Split Integral at $x = 0$**:

$$
\phi(t) = \frac{1}{2} \int_{-\infty}^0 e^{i t x} e^x \, dx + \frac{1}{2} \int_0^\infty e^{i t x} e^{-x} \, dx
$$

2. **Evaluate Upper Integral**:

$$
\int_0^\infty e^{-(1 - i t) x} \, dx = \left[ -\frac{e^{-(1 - i t) x}}{1 - i t} \right]_0^\infty = \frac{1}{1 - i t}
$$

3. **Evaluate Lower Integral**:
Let $y = -x$:

$$
\int_{-\infty}^0 e^{(1 + i t) x} \, dx = \int_0^\infty e^{-(1 + i t) y} \, dy = \frac{1}{1 + i t}
$$

4. **Combine and Simplify**:

$$
\phi(t) = \frac{1}{2} \left( \frac{1}{1 + i t} + \frac{1}{1 - i t} \right) = \frac{1}{2} \left( \frac{(1 - i t) + (1 + i t)}{(1 + i t)(1 - i t)} \right) = \frac{1}{2} \left( \frac{2}{1 + t^2} \right) = \frac{1}{1 + t^2}
$$

$$
\boxed{\frac{1}{1 + t^2}}
$$

#### Key takeaway
The Fourier transform / characteristic function of the Laplace distribution is the Cauchy distribution density (up to scale).

---

```python
# The Laplace characteristic function is real by symmetry, so the cosine transform suffices.
for t in (0.0, 0.7, 3.0):
    num = integrate.quad(lambda x, t=t: 0.5 * np.exp(-abs(x)) * np.cos(t * x), -np.inf, np.inf)[0]
    print(f"t={t}: phi(t) numeric = {num:.12f}   1/(1+t^2) = {1/(1+t*t):.12f}")
    assert abs(num - 1 / (1 + t * t)) < 1e-8
```

### Problem L2.9 — Integrated Softplus and the Eta Function
**Source**: Machine Learning Optimization / Loss Landscapes  

#### Statement
In neural networks, the Softplus function $\operatorname{softplus}(x) = \ln(1 + e^x)$ smooths the ReLU activation. Evaluate the integral $I = \int_{-\infty}^0 \operatorname{softplus}(x) \, dx$.

#### Intuition
For $x \lt 0$, $e^x \in (0, 1)$, so we can use Taylor expansion $\ln(1+u) = \sum_{k=1}^\infty \frac{(-1)^{k-1} u^k}{k}$ for $u = e^x$.

#### Solution
1. **Express Logarithm as Infinite Series**:
For $x \in (-\infty, 0]$, $0 \lt e^x \le 1$:

$$
\ln(1 + e^x) = \sum_{k=1}^\infty \frac{(-1)^{k-1}}{k} e^{k x}
$$

2. **Integrate Term-by-Term**:

$$
I = \int_{-\infty}^0 \left( \sum_{k=1}^\infty \frac{(-1)^{k-1}}{k} e^{k x} \right) dx = \sum_{k=1}^\infty \frac{(-1)^{k-1}}{k} \int_{-\infty}^0 e^{k x} \, dx
$$

3. **Evaluate Primitive Integral**:

$$
\int_{-\infty}^0 e^{k x} \, dx = \left[ \frac{e^{k x}}{k} \right]_{-\infty}^0 = \frac{1}{k}
$$

4. **Sum Series**:

$$
I = \sum_{k=1}^\infty \frac{(-1)^{k-1}}{k^2} = 1 - \frac{1}{4} + \frac{1}{9} - \frac{1}{16} + \dots = \frac{\pi^2}{12}
$$

$$
\boxed{\frac{\pi^2}{12}}
$$

#### Key takeaway
The integrated softplus loss over negative logits yields Dirichlet's alternating eta function $\eta(2) = \frac{\pi^2}{12}$.

---

```python
val = integrate.quad(lambda x: np.log1p(np.exp(x)), -np.inf, 0)[0]
print(f"int_-inf^0 softplus(x) dx = {val:.12f}")
print(f"eta(2) = pi^2/12          = {np.pi**2 / 12:.12f}")
assert abs(val - np.pi**2 / 12) < 1e-9
```

### Problem L2.10 — Volume of the d-Dimensional Ball
**Source**: High-Dimensional Geometry & ML  

#### Statement
Compute the volume $V_d(R)$ of a $d$-dimensional sphere of radius $R$ by evaluating the total Gaussian integral $I = \int_{\mathbb{R}^d} e^{-\lVert \mathbf{x} \rVert^2} d\mathbf{x}$ in spherical coordinates.

#### Intuition
Evaluating the same Gaussian integral in Cartesian coordinates gives $\pi^{\frac{d}{2}}$ and in $d$-dimensional spherical coordinates gives $S_{d-1} \int_0^\infty r^{d-1} e^{-r^2} dr$, where $S_{d-1}$ is the surface area of the unit sphere.

#### Solution
1. **Cartesian Evaluation**:

$$
I = \left(\int_{-\infty}^\infty e^{-x^2} dx\right)^d = \pi^{\frac{d}{2}}
$$

2. **Spherical Coordinates Evaluation**:

$$
I = S_{d-1} \int_0^\infty r^{d-1} e^{-r^2} \, dr
$$

Substitute $u = r^2 \implies dr = \frac{du}{2\sqrt{u}}$:

$$
\int_0^\infty r^{d-1} e^{-r^2} \, dr = \int_0^\infty u^{\frac{d-1}{2}} e^{-u} \frac{du}{2 u^{\frac{1}{2}}} = \frac{1}{2} \int_0^\infty u^{\frac{d}{2}-1} e^{-u} \, du = \frac{1}{2} \Gamma\left(\frac{d}{2}\right)
$$

3. **Equate and Solve for Surface Area $S_{d-1}$**:

$$
\pi^{\frac{d}{2}} = S_{d-1} \cdot \frac{1}{2} \Gamma\left(\frac{d}{2}\right) \implies S_{d-1} = \frac{2 \pi^{\frac{d}{2}}}{\Gamma\left(\frac{d}{2}\right)}
$$

4. **Compute Volume $V_d(R)$**:

$$
V_d(R) = \int_0^R S_{d-1} r^{d-1} \, dr = \frac{S_{d-1} R^d}{d} = \frac{2 \pi^{\frac{d}{2}} R^d}{d \, \Gamma\left(\frac{d}{2}\right)} = \frac{\pi^{\frac{d}{2}} R^d}{\Gamma\left(\frac{d}{2} + 1\right)}
$$

$$
\boxed{V_d(R) = \frac{\pi^{\frac{d}{2}} R^d}{\Gamma\left(\frac{d}{2} + 1\right)}}
$$

#### Key takeaway
The Gamma function provides the exact analytical formula for volumes of hyperspheres across all integer dimensions $d$.

---

```python
# Gamma-based ball volume against the elementary formulas in d = 1, 2, 3.
R = 2.0
formula = lambda d: np.pi**(d / 2) * R**d / special.gamma(d / 2 + 1)
elementary = {1: 2 * R, 2: np.pi * R**2, 3: 4 / 3 * np.pi * R**3}
for d in (1, 2, 3, 4, 10):
    line = f"d={d:>3}: V = {formula(d):.10f}"
    if d in elementary:
        line += f"   elementary = {elementary[d]:.10f}"
        assert abs(formula(d) - elementary[d]) < 1e-10
    print(line)
```

### Problem L2.11 — The Student-t Normalizing Constant
**Source**: Machine Learning & Robust Statistics / Student-$t$ Distribution  

#### Statement
The standard Student-$t$ distribution with $\nu \gt 0$ degrees of freedom has density $p(x) = C_\nu \left(1 + \frac{x^2}{\nu}\right)^{-\frac{\nu+1}{2}}$. Determine the normalization constant $C_\nu$ in terms of the Beta function.

#### Intuition
Substitute $x = \sqrt{\nu} \tan\theta$ to transform the algebraic power integral into a trigonometric Beta integral over $[-\frac{\pi}{2}, \frac{\pi}{2}]$.

#### Solution
1. **Set up Normalization Condition**:

$$
C_\nu \int_{-\infty}^\infty \left(1 + \frac{x^2}{\nu}\right)^{-\frac{\nu+1}{2}} \, dx = 1
$$

By symmetry:

$$
2 C_\nu \int_0^\infty \left(1 + \frac{x^2}{\nu}\right)^{-\frac{\nu+1}{2}} \, dx = 1
$$

2. **Substitute $x = \sqrt{\nu} \tan\theta \implies dx = \sqrt{\nu} \sec^2\theta d\theta$**:
Limits: $x=0 \implies \theta=0$, and $x \to \infty \implies \theta \to \frac{\pi}{2}$.

$$
1 + \frac{x^2}{\nu} = 1 + \tan^2\theta = \sec^2\theta
$$

$$
\int_0^\infty \left(1 + \frac{x^2}{\nu}\right)^{-\frac{\nu+1}{2}} \, dx = \int_0^{\frac{\pi}{2}} (\sec^2\theta)^{-\frac{\nu+1}{2}} \sqrt{\nu} \sec^2\theta \, d\theta = \sqrt{\nu} \int_0^{\frac{\pi}{2}} \cos^{\nu-1}\theta \, d\theta
$$

3. **Relate to Beta Function**:

$$
\int_0^{\frac{\pi}{2}} \cos^{\nu-1}\theta \, d\theta = \frac{1}{2} B\left(\frac{1}{2}, \frac{\nu}{2}\right)
$$

4. **Solve for $C_\nu$**:

$$
2 C_\nu \left[ \sqrt{\nu} \cdot \frac{1}{2} B\left(\frac{1}{2}, \frac{\nu}{2}\right) \right] = 1 \implies C_\nu = \frac{1}{\sqrt{\nu} B\left(\frac{1}{2}, \frac{\nu}{2}\right)} = \frac{\Gamma\left(\frac{\nu+1}{2}\right)}{\sqrt{\pi \nu} \, \Gamma\left(\frac{\nu}{2}\right)}
$$

$$
\boxed{C_\nu = \frac{1}{\sqrt{\nu} \, B\left(\frac{1}{2}, \frac{\nu}{2}\right)} = \frac{\Gamma\left(\frac{\nu+1}{2}\right)}{\sqrt{\pi \nu} \, \Gamma\left(\frac{\nu}{2}\right)}}
$$

#### Key takeaway
Heavy-tailed Student-$t$ densities are normalized by Beta/Gamma functions obtained via trigonometric substitutions.

---

```python
# The Student-t constant makes the density integrate to 1.
for nu in (1.0, 3.0, 30.0):
    C_nu = 1.0 / (np.sqrt(nu) * special.beta(0.5, nu / 2))
    gamma_form = special.gamma((nu + 1) / 2) / (np.sqrt(np.pi * nu) * special.gamma(nu / 2))
    mass = integrate.quad(lambda x, nu=nu, C=C_nu: C * (1 + x * x / nu)**(-(nu + 1) / 2),
                          -np.inf, np.inf)[0]
    print(f"nu={nu:>5}: C={C_nu:.10f}  Gamma form={gamma_form:.10f}  total mass={mass:.10f}")
    assert abs(C_nu - gamma_form) < 1e-12 and abs(mass - 1) < 1e-8
```

### Problem L3.1 — The Fresnel Integral
**Source**: Cambridge Tripos IB / Advanced Physics  

#### Statement
Evaluate the Fresnel Integral $I = \int_0^\infty \sin(x^2) \, dx$.

#### Intuition
Substitute $u = x^2$ to transform into an improper gamma-type integral $\int_0^\infty \frac{\sin u}{2\sqrt{u}} du$, then introduce a parameter $e^{-\alpha u}$ and apply Feynman's parameter trick / Laplace transform.

#### Solution
1. **Substitute $u = x^2 \implies dx = \frac{du}{2\sqrt{u}}$**:

$$
I = \frac{1}{2} \int_0^\infty u^{-\frac{1}{2}} \sin u \, du
$$

2. **Introduce Parametric Helper Function**:
For $\alpha \gt 0$, define $J(\alpha) = \frac{1}{2} \int_0^\infty u^{-\frac{1}{2}} e^{-\alpha u} \sin u \, du$.
Recall Gamma integral identity for complex exponents:

$$
\int_0^\infty u^{-\frac{1}{2}} e^{-z u} \, du = \frac{\Gamma(\frac{1}{2})}{z^{\frac{1}{2}}} = \frac{\sqrt{\pi}}{\sqrt{z}} \quad \text{for } \operatorname{Re}(z) \gt 0
$$

3. **Set $z = \alpha - i$**:

$$
\int_0^\infty u^{-\frac{1}{2}} e^{-(\alpha - i) u} \, du = \frac{\sqrt{\pi}}{\sqrt{\alpha - i}}
$$

Taking the imaginary part of both sides:

$$
\int_0^\infty u^{-\frac{1}{2}} e^{-\alpha u} \sin u \, du = \operatorname{Im}\left( \frac{\sqrt{\pi}}{\sqrt{\alpha - i}} \right)
$$

4. **Take Limit as $\alpha \to 0^+$**:

$$
\lim_{\alpha \to 0^+} \frac{1}{\sqrt{\alpha - i}} = \frac{1}{\sqrt{-i}} = \frac{1}{e^{-i \frac{\pi}{4}}} = e^{i \frac{\pi}{4}} = \cos\left(\frac{\pi}{4}\right) + i \sin\left(\frac{\pi}{4}\right) = \frac{\sqrt{2}}{2} + i \frac{\sqrt{2}}{2}
$$

Taking the imaginary part gives $\frac{\sqrt{2}}{2} = \frac{1}{\sqrt{2}}$.
Thus:

$$
I = \frac{1}{2} \left( \sqrt{\pi} \cdot \frac{1}{\sqrt{2}} \right) = \frac{\sqrt{2\pi}}{4} = \sqrt{\frac{\pi}{8}}
$$

$$
\boxed{\sqrt{\frac{\pi}{8}}}
$$

#### Key takeaway
Fresnel integrals oscillate endlessly without vanishing, yet converge to $\sqrt{\frac{\pi}{8}}$ via complex parameter analytic continuation.

---

```python
# Fresnel: scipy's S(z) -> 1/2, and int_0^inf sin(x^2) dx = sqrt(pi/2) * S(inf).
S_inf, _ = special.fresnel(1e6)
val = np.sqrt(np.pi / 2) * S_inf
print(f"S(1e6)                  = {S_inf:.10f}   (limit 1/2)")
print(f"int_0^inf sin(x^2) dx   = {val:.10f}")
print(f"sqrt(pi/8)              = {np.sqrt(np.pi / 8):.10f}")
assert abs(val - np.sqrt(np.pi / 8)) < 1e-5
```

### Problem L3.2 — The Squared Dirichlet Integral
**Source**: Demidovich, *Problems in Mathematical Analysis*, No. 3816  

#### Statement
Evaluate the squared Dirichlet integral $I = \int_0^\infty \left(\frac{\sin x}{x}\right)^2 dx$.

#### Intuition
Apply integration by parts to drop the square power, converting it into the standard Dirichlet integral $\int_0^\infty \frac{\sin(2x)}{x} dx = \frac{\pi}{2}$.

#### Solution
1. **Integration by Parts**:
Set $u = \sin^2 x \implies du = 2\sin x \cos x \, dx = \sin(2x) \, dx$.
Set $dv = \frac{dx}{x^2} \implies v = -\frac{1}{x}$.

$$
I = \left[ -\frac{\sin^2 x}{x} \right]_0^\infty - \int_0^\infty \left(-\frac{1}{x}\right) \sin(2x) \, dx
$$

2. **Evaluate Boundary Term**:
- As $x \to \infty$, $\left\vert\frac{\sin^2 x}{x}\right\vert \le \frac{1}{x} \to 0$.
- As $x \to 0^+$, $\frac{\sin^2 x}{x} \sim \frac{x^2}{x} = x \to 0$.
Thus, boundary term vanishes: $\left[ -\frac{\sin^2 x}{x} \right]_0^\infty = 0$.
3. **Evaluate Remaining Integral**:

$$
I = \int_0^\infty \frac{\sin(2x)}{x} \, dx
$$

Substitute $y = 2x \implies dy = 2dx, \frac{dy}{y} = \frac{dx}{x}$:

$$
I = \int_0^\infty \frac{\sin y}{y} \, dy = \frac{\pi}{2}
$$

$$
\boxed{\frac{\pi}{2}}
$$

#### Key takeaway
Integration by parts removes higher power denominators in trigonometric improper integrals, mapping them to Dirichlet's integral.

---

```python
# sin^2(x)/x^2 = (1 - cos 2x) / (2 x^2); the tail uses quad's oscillatory weight.
head = integrate.quad(lambda x: (np.sin(x) / x)**2, 0, 1)[0]
tail = (integrate.quad(lambda x: 0.5 / x**2, 1, np.inf)[0]
        - integrate.quad(lambda x: 0.5 / x**2, 1, np.inf, weight="cos", wvar=2.0)[0])
print(f"head + tail = {head + tail:.12f}")
print(f"pi/2        = {np.pi / 2:.12f}")
assert abs(head + tail - np.pi / 2) < 1e-9
```

### Problem L3.3 — Putnam 2005 A5: a Log-Tangent Integral
**Source**: William Lowell Putnam Mathematical Competition 2005, Problem A5  

#### Statement
Evaluate $I = \int_0^1 \frac{\ln(1+x)}{1+x^2} \, dx$.

#### Intuition
Substitute $x = \tan\theta$ to transform the rational denominator $1+x^2$ into $1$, converting the integral into a symmetric trigonometric logarithm integral over $[0, \frac{\pi}{4}]$.

#### Solution
1. **Trigonometric Substitution**:
Let $x = \tan\theta \implies dx = \sec^2\theta d\theta$.
Limits: $x = 0 \implies \theta = 0$, and $x = 1 \implies \theta = \frac{\pi}{4}$.

$$
I = \int_0^{\frac{\pi}{4}} \frac{\ln(1 + \tan\theta)}{1 + \tan^2\theta} \sec^2\theta \, d\theta = \int_0^{\frac{\pi}{4}} \ln(1 + \tan\theta) \, d\theta
$$

2. **Apply Reflection Identity $\theta \to \frac{\pi}{4} - \theta$**:
Let $\theta = \frac{\pi}{4} - \phi \implies d\theta = -d\phi$.

$$
1 + \tan\left(\frac{\pi}{4} - \phi\right) = 1 + \frac{1 - \tan\phi}{1 + \tan\phi} = \frac{1 + \tan\phi + 1 - \tan\phi}{1 + \tan\phi} = \frac{2}{1 + \tan\phi}
$$

Thus:

$$
I = \int_0^{\frac{\pi}{4}} \ln\left(\frac{2}{1 + \tan\phi}\right) \, d\phi = \int_0^{\frac{\pi}{4}} \left( \ln 2 - \ln(1 + \tan\phi) \right) d\phi
$$

3. **Solve for $I$**:

$$
I = \ln 2 \int_0^{\frac{\pi}{4}} d\phi - I = \frac{\pi}{4} \ln 2 - I \implies 2I = \frac{\pi}{4} \ln 2 \implies I = \frac{\pi}{8} \ln 2
$$

$$
\boxed{\frac{\pi}{8} \ln 2}
$$

#### Key takeaway
The symmetry substitution $\phi = \frac{\pi}{4} - \theta$ combined with the tangent addition formula simplifies log-trigonometric integrals cleanly.

---

```python
val = integrate.quad(lambda x: np.log1p(x) / (1 + x * x), 0, 1)[0]
print(f"int_0^1 ln(1+x)/(1+x^2) dx = {val:.12f}")
print(f"(pi/8) ln 2                = {np.pi / 8 * np.log(2):.12f}")
assert abs(val - np.pi / 8 * np.log(2)) < 1e-10
```

### Problem L3.4 — Frullani for a Cosine with No Limit
**Source**: Polya & Szego, *Problems and Theorems in Analysis I*, Part I  

#### Statement
Evaluate the Frullani cosine integral $I = \int_0^\infty \frac{\cos(ax) - \cos(bx)}{x} \, dx$ for $a \gt 0, b \gt 0$.

#### Intuition
Although $f(t) = \cos t$ does not have a limit at $\infty$, integration by parts or parametric integration shows that the oscillation averages out to $0$ as $R \to \infty$.

#### Solution
1. **Parametric Integration Method**:
Observe that $\frac{\cos(ax) - \cos(bx)}{x} = \int_a^b \sin(y x) \, dy$.
2. **Swap Order of Integration**:

$$
I = \int_a^b \left( \int_0^\infty \sin(y x) \, dx \right) \, dy
$$

Using convergence factor $e^{-\epsilon x}$ and taking $\epsilon \to 0^+$:

$$
\int_0^\infty e^{-\epsilon x} \sin(y x) \, dx = \frac{y}{\epsilon^2 + y^2} \xrightarrow{\epsilon \to 0^+} \frac{1}{y}
$$

3. **Integrate over $y$**:

$$
I = \int_a^b \frac{dy}{y} = \ln b - \ln a = \ln\left(\frac{b}{a}\right)
$$

$$
\boxed{\ln\left(\frac{b}{a}\right)}
$$

#### Key takeaway
For oscillating Frullani functions where $f(\infty)$ is undefined, continuous averaging over oscillations sets the effective boundary value $f(\infty) = 0$.

---

```python
# int_eps^inf cos(ax)/x dx = -Ci(a eps); the difference tends to ln(b/a).
Ci = lambda z: special.sici(z)[1]
a, b = 1.5, 4.0
for eps in (1e-4, 1e-7, 1e-10):
    val = Ci(b * eps) - Ci(a * eps)
    print(f"eps={eps:.0e}:  Ci(b eps) - Ci(a eps) = {val:.12f}")
print(f"ln(b/a) = {np.log(b / a):.12f}")
assert abs(Ci(b * 1e-10) - Ci(a * 1e-10) - np.log(b / a)) < 1e-8
```

### Problem L3.5 — Stirling's Formula by Laplace's Method
**Source**: Polya & Szego, *Problems in Analysis I* / Cambridge Tripos  

#### Statement
Derive the leading asymptotic term of $\Gamma(n+1) = \int_0^\infty x^n e^{-x} \, dx$ as $n \to \infty$ using Laplace's Method (Stirling's approximation setup).

#### Intuition
Rewrite the integrand $x^n e^{-x} = e^{n \ln x - x}$. The exponent $g(x) = n \ln x - x$ has a sharp maximum at $x = n$. Expanding around $x = n$ converts the integral into a Gaussian integral.

#### Solution
1. **Find Peak Location**:
Set $g(x) = n \ln x - x$. $g'(x) = \frac{n}{x} - 1 = 0 \implies x = n$.
2. **Change Variables to Center at $x = n$**:
Let $x = n(1 + t) \implies dx = n \, dt$.

$$
g(x) = n \ln(n(1+t)) - n(1+t) = n \ln n - n + n[\ln(1+t) - t]
$$

Thus:

$$
\Gamma(n+1) = \int_{-1}^\infty e^{n \ln n - n} e^{n[\ln(1+t) - t]} n \, dt = n^{n+1} e^{-n} \int_{-1}^\infty e^{n[\ln(1+t) - t]} \, dt
$$

3. **Taylor Expand Exponent around $t = 0$**:
For small $t$, $\ln(1+t) = t - \frac{t^2}{2} + O(t^3) \implies \ln(1+t) - t \approx -\frac{t^2}{2}$.
4. **Evaluate Gaussian Approximation**:
As $n \to \infty$, the integral concentrates tightly near $t = 0$:

$$
\int_{-1}^\infty e^{-n t^2 / 2} \, dt \approx \int_{-\infty}^\infty e^{-n t^2 / 2} \, dt = \sqrt{\frac{2\pi}{n}}
$$

5. **Combine Terms**:

$$
\Gamma(n+1) \approx n^{n+1} e^{-n} \sqrt{\frac{2\pi}{n}} = n^n e^{-n} \sqrt{2\pi n}
$$

$$
\boxed{\Gamma(n+1) \sim \sqrt{2\pi n} \left(\frac{n}{e}\right)^n \quad \text{as } n \to \infty}
$$

#### Key takeaway
Laplace's Method approximates improper integrals of exponentiated large parameters by local Gaussian expansion near the global maximum.

---

```python
# Relative error of Stirling's leading term, which must fall like 1/(12n).
print(f"{'n':>6} {'ln Gamma(n+1)':>16} {'Stirling':>16} {'rel err':>12} {'1/(12n)':>12}")
for n in (5, 20, 100, 1000):
    exact = special.gammaln(n + 1)
    stir = 0.5 * np.log(2 * np.pi * n) + n * (np.log(n) - 1)
    rel = abs(np.expm1(exact - stir))
    print(f"{n:>6} {exact:>16.8f} {stir:>16.8f} {rel:>12.3e} {1/(12*n):>12.3e}")
    assert rel < 1.2 / (12 * n)
```

### Problem L3.6 — Euler's Log-Gamma Integral
**Source**: Cambridge Tripos Part IA / Special Functions  

#### Statement
Prove Euler's Log-Gamma integral identity: $I = \int_0^1 \ln(\Gamma(x)) \, dx = \frac{1}{2} \ln(2\pi)$.

#### Intuition
Use the reflection formula $\Gamma(x)\Gamma(1-x) = \frac{\pi}{\sin(\pi x)}$ and symmetry over $[0, 1]$.

#### Solution
1. **Apply Reflection Substitution $u = 1-x$**:

$$
I = \int_0^1 \ln(\Gamma(x)) \, dx = \int_0^1 \ln(\Gamma(1-x)) \, dx
$$

$$
2I = \int_0^1 \left( \ln(\Gamma(x)) + \ln(\Gamma(1-x)) \right) dx = \int_0^1 \ln(\Gamma(x) \Gamma(1-x)) \, dx
$$

2. **Substitute Euler's Reflection Formula**:

$$
\Gamma(x) \Gamma(1-x) = \frac{\pi}{\sin(\pi x)}
$$

$$
2I = \int_0^1 \ln\left( \frac{\pi}{\sin(\pi x)} \right) dx = \int_0^1 \ln\pi \, dx - \int_0^1 \ln(\sin(\pi x)) \, dx = \ln\pi - \int_0^1 \ln(\sin(\pi x)) \, dx
$$

3. **Evaluate Sine Log Integral**:
Let $y = \pi x \implies dx = \frac{dy}{\pi}$:

$$
\int_0^1 \ln(\sin(\pi x)) \, dx = \frac{1}{\pi} \int_0^\pi \ln(\sin y) \, dy = \frac{2}{\pi} \int_0^{\frac{\pi}{2}} \ln(\sin y) \, dy
$$

Using the result $\int_0^{\frac{\pi}{2}} \ln(\sin y) \, dy = -\frac{\pi}{2} \ln 2$ (proved in Problem L3.10 below by the double-angle self-referential argument, which uses neither the reflection formula nor this identity — so the chain is not circular):

$$
\int_0^1 \ln(\sin(\pi x)) \, dx = \frac{2}{\pi} \left( -\frac{\pi}{2} \ln 2 \right) = -\ln 2
$$

4. **Solve for $I$**:

$$
2I = \ln\pi - (-\ln 2) = \ln\pi + \ln 2 = \ln(2\pi) \implies I = \frac{1}{2} \ln(2\pi)
$$

$$
\boxed{\frac{1}{2} \ln(2\pi)}
$$

#### Key takeaway
Euler's reflection formula converts log-Gamma integrals into standard log-sine integrals.

---

```python
val = integrate.quad(lambda x: special.gammaln(x), 0, 1)[0]
print(f"int_0^1 ln Gamma(x) dx = {val:.12f}")
print(f"(1/2) ln(2 pi)         = {0.5 * np.log(2 * np.pi):.12f}")
assert abs(val - 0.5 * np.log(2 * np.pi)) < 1e-9
```

### Problem L3.7 — A Log Integral Killed by Inversion
**Source**: Demidovich, *Problems in Mathematical Analysis*, No. 3825  

#### Statement
Evaluate $I = \int_0^\infty \frac{\ln x}{1 + x^2} \, dx$.

#### Intuition
Split the integral at $x = 1$ and apply inversion substitution $u = \frac{1}{x}$ on the interval $[1, \infty)$.

#### Solution
1. **Split Integral**:

$$
I = \int_0^1 \frac{\ln x}{1 + x^2} \, dx + \int_1^\infty \frac{\ln x}{1 + x^2} \, dx
$$

2. **Transform Second Integral**:
Let $u = \frac{1}{x} \implies x = \frac{1}{u}, dx = -\frac{du}{u^2}$.
Limits: $x = 1 \implies u = 1$, and $x \to \infty \implies u \to 0$.

$$
\int_1^\infty \frac{\ln x}{1 + x^2} \, dx = \int_1^0 \frac{\ln(\frac{1}{u})}{1 + (\frac{1}{u})^2} \left(-\frac{du}{u^2}\right) = \int_0^1 \frac{-\ln u}{\frac{u^2 + 1}{u^2}} \frac{du}{u^2} = -\int_0^1 \frac{\ln u}{1 + u^2} \, du
$$

3. **Combine Terms**:

$$
I = \int_0^1 \frac{\ln x}{1 + x^2} \, dx - \int_0^1 \frac{\ln u}{1 + u^2} \, du = 0
$$

$$
\boxed{0}
$$

#### Key takeaway
Inversion substitutions $u = \frac{1}{x}$ demonstrate exact destructive anti-symmetry for logarithmic integrands over $[0, 1]$ and $[1, \infty)$.

---

```python
whole = integrate.quad(lambda x: np.log(x) / (1 + x * x), 0, np.inf)[0]
lower = integrate.quad(lambda x: np.log(x) / (1 + x * x), 0, 1)[0]
upper = integrate.quad(lambda x: np.log(x) / (1 + x * x), 1, np.inf)[0]
print(f"int_0^1   = {lower:.12f}")
print(f"int_1^inf = {upper:.12f}   (the exact negative of the first piece)")
print(f"total     = {whole:.3e}")
assert abs(lower + upper) < 1e-10 and abs(whole) < 1e-10
```

### Problem L3.8 — The Reflection-Formula Power Integral
**Source**: Kaczor & Nowak / Demidovich No. 3848  

#### Statement
Prove that for $0 \lt a \lt b$, $I = \int_0^\infty \frac{x^{a-1}}{1 + x^b} \, dx = \frac{\pi}{b \sin\left(\frac{\pi a}{b}\right)}$.

#### Intuition
Substitute $u = x^b$ to transform into the infinite domain representation of the Beta function $B(x, y) = \int_0^\infty \frac{u^{x-1}}{(1+u)^{x+y}} du$, then apply Euler's reflection formula.

#### Solution
1. **Substitute $u = x^b$**:

$$
x = u^{\frac{1}{b}} \implies dx = \frac{1}{b} u^{\frac{1}{b} - 1} du
$$

$$
I = \int_0^\infty \frac{(u^{\frac{1}{b}})^{a-1}}{1 + u} \left( \frac{1}{b} u^{\frac{1}{b}-1} du \right) = \frac{1}{b} \int_0^\infty \frac{u^{\frac{a}{b}-1}}{1 + u} \, du
$$

2. **Identify Beta Function**:
Recall Theorem 4.9: $\int_0^\infty \frac{u^{x-1}}{(1+u)^{x+y}} du = B(x, y)$.
Here:

$$
x - 1 = \frac{a}{b} - 1 \implies x = \frac{a}{b}
$$

$$
x + y = 1 \implies y = 1 - \frac{a}{b}
$$

Thus:

$$
I = \frac{1}{b} B\left(\frac{a}{b}, 1 - \frac{a}{b}\right)
$$

3. **Apply Gamma-Beta and Reflection Formulas**:

$$
B\left(\frac{a}{b}, 1 - \frac{a}{b}\right) = \frac{\Gamma\left(\frac{a}{b}\right) \Gamma\left(1 - \frac{a}{b}\right)}{\Gamma(1)} = \Gamma\left(\frac{a}{b}\right) \Gamma\left(1 - \frac{a}{b}\right)
$$

By Euler's Reflection Formula $\Gamma(z)\Gamma(1-z) = \frac{\pi}{\sin(\pi z)}$:

$$
\Gamma\left(\frac{a}{b}\right) \Gamma\left(1 - \frac{a}{b}\right) = \frac{\pi}{\sin\left(\frac{\pi a}{b}\right)}
$$

$$
I = \frac{\pi}{b \sin\left(\frac{\pi a}{b}\right)}
$$

$$
\boxed{\frac{\pi}{b \sin\left(\frac{\pi a}{b}\right)}}
$$

#### Key takeaway
Euler's reflection formula provides a closed form for algebraic power integrals over $[0, \infty)$.

---

```python
for a, b in ((1.2, 3.0), (0.5, 1.0), (2.0, 7.0)):
    num = integrate.quad(lambda x, a=a, b=b: x**(a - 1) / (1 + x**b), 0, np.inf)[0]
    closed = np.pi / (b * np.sin(np.pi * a / b))
    print(f"a={a}, b={b}: quad={num:.10f}  pi/(b sin(pi a/b))={closed:.10f}  "
          f"|diff|={abs(num-closed):.1e}")
    assert abs(num - closed) < 1e-8
```

### Problem L3.9 — N-Denominator Feynman Parameterization
**Source**: Advanced Quantum Field Theory / Tripos Part III  

#### Statement
Prove the General Feynman Parameter identity for $N$ denominators $A_1, A_2, \dots, A_N \gt 0$:

$$
\frac{1}{A_1 A_2 \dots A_N} = (N-1)! \int_0^1 \dots \int_0^1 d x_1 \dots d x_N \, \frac{\delta\left(1 - \sum_{i=1}^N x_i\right)}{\left(\sum_{i=1}^N x_i A_i\right)^N}
$$

#### Intuition
Use the Schwinger representation of denominators $\frac{1}{A} = \int_0^\infty e^{-t A} dt$ and perform a scale transformation of integration variables.

#### Solution
1. **Schwinger Parameterization**:

$$
\frac{1}{\prod_{i=1}^N A_i} = \prod_{i=1}^N \left( \int_0^\infty e^{-t_i A_i} dt_i \right) = \int_0^\infty \dots \int_0^\infty dt_1 \dots dt_N \, \exp\left(-\sum_{i=1}^N t_i A_i\right)
$$

2. **Introduce Unity via Delta Function**:
Insert $1 = \int_0^\infty d\lambda \, \delta\left(\lambda - \sum_{i=1}^N t_i\right)$:

$$
\frac{1}{\prod_{i=1}^N A_i} = \int_0^\infty d\lambda \int_0^\infty \dots \int_0^\infty dt_1 \dots dt_N \, \delta\left(\lambda - \sum_{i=1}^N t_i\right) \exp\left(-\sum_{i=1}^N t_i A_i\right)
$$

3. **Rescale Variables $t_i = \lambda x_i$**:
Then $dt_i = \lambda dx_i$, and $\delta\left(\lambda - \lambda \sum x_i\right) = \frac{1}{\lambda} \delta\left(1 - \sum x_i\right)$.

$$
\begin{aligned}
\frac{1}{\prod_{i=1}^N A_i} &= \int_0^\infty d\lambda \int_0^1 \dots \int_0^1 (\lambda^N dx_1 \dots dx_N) \, \frac{1}{\lambda} \delta\left(1 - \sum_{i=1}^N x_i\right) \exp\left(-\lambda \sum_{i=1}^N x_i A_i\right) \\
&= \int_0^1 \dots \int_0^1 dx_1 \dots dx_N \, \delta\left(1 - \sum_{i=1}^N x_i\right) \left[ \int_0^\infty \lambda^{N-1} e^{-\lambda \left(\sum x_i A_i\right)} d\lambda \right]
\end{aligned}
$$

4. **Evaluate Internal $\lambda$ Integral**:

$$
\int_0^\infty \lambda^{N-1} e^{-\lambda \left(\sum x_i A_i\right)} d\lambda = \frac{\Gamma(N)}{\left(\sum_{i=1}^N x_i A_i\right)^N} = \frac{(N-1)!}{\left(\sum_{i=1}^N x_i A_i\right)^N}
$$

Substitution yields the final identity. $\blacksquare$

$$
\boxed{\frac{1}{\prod_{i=1}^N A_i} = (N-1)! \int_0^1 \dots \int_0^1 d x_1 \dots d x_N \, \frac{\delta\left(1 - \sum_{i=1}^N x_i\right)}{\left(\sum_{i=1}^N x_i A_i\right)^N}}
$$

#### Key takeaway
Schwinger parameterization converts arbitrary product denominators into a single simplex integral over Feynman parameters $x_i$.

---

```python
# N = 3: integrate over the simplex x1 + x2 + x3 = 1 by eliminating x3.
A = np.array([1.7, 3.1, 0.6])
inner = lambda x2, x1: 1.0 / (x1 * A[0] + x2 * A[1] + (1 - x1 - x2) * A[2])**3
val = 2.0 * integrate.dblquad(inner, 0, 1, lambda x1: 0, lambda x1: 1 - x1)[0]
print(f"(N-1)! * simplex integral = {val:.12f}")
print(f"1/(A1 A2 A3)              = {1 / np.prod(A):.12f}")
assert abs(val - 1 / np.prod(A)) < 1e-9
```

### Problem L3.10 — The Log-Sine Integral
**Source**: Putnam Competition / Euler / Spivak Ch. 19  

#### Statement
Evaluate $I = \int_0^{\frac{\pi}{2}} \ln(\sin x) \, dx$.

#### Intuition
Use trigonometric symmetry and double-angle formulas to set up a self-referential equation for $I$.

#### Solution
0. **Convergence first.** The self-referential step below divides by a finite number, so finiteness must be established before it is used. On $(0, \frac{\pi}{2}]$ the integrand is negative with $\lvert \ln(\sin x) \rvert = -\ln(\sin x)$, and $\sin x \ge \frac{2x}{\pi}$ there (concavity of $\sin$ on $[0, \frac{\pi}{2}]$), so $0 \le -\ln(\sin x) \le -\ln\left(\frac{2x}{\pi}\right)$. Since $\int_0^{\frac{\pi}{2}} -\ln\left(\frac{2x}{\pi}\right) dx = \frac{\pi}{2}$ is finite, Theorem 4.3 (Direct Comparison) gives absolute convergence, so $I \in \mathbb{R}$.
1. **Apply Reflection Symmetry**:
By $x \to \frac{\pi}{2} - x$:

$$
I = \int_0^{\frac{\pi}{2}} \ln(\cos x) \, dx
$$

2. **Add the Two Expressions**:

$$
2I = \int_0^{\frac{\pi}{2}} \left( \ln(\sin x) + \ln(\cos x) \right) dx = \int_0^{\frac{\pi}{2}} \ln(\sin x \cos x) \, dx
$$

3. **Use Double-Angle Formula $\sin x \cos x = \frac{\sin(2x)}{2}$**:

$$
2I = \int_0^{\frac{\pi}{2}} \ln\left(\frac{\sin(2x)}{2}\right) dx = \int_0^{\frac{\pi}{2}} \ln(\sin(2x)) \, dx - \int_0^{\frac{\pi}{2}} \ln 2 \, dx
$$

$$
2I = \int_0^{\frac{\pi}{2}} \ln(\sin(2x)) \, dx - \frac{\pi}{2} \ln 2
$$

4. **Evaluate Integral of $\ln(\sin(2x))$**:
Substitute $u = 2x \implies dx = \frac{du}{2}$:

$$
\int_0^{\frac{\pi}{2}} \ln(\sin(2x)) \, dx = \frac{1}{2} \int_0^\pi \ln(\sin u) \, du
$$

By symmetry of $\sin u$ on $[0, \pi]$:

$$
\int_0^\pi \ln(\sin u) \, du = 2 \int_0^{\frac{\pi}{2}} \ln(\sin u) \, du = 2 I
$$

Thus:

$$
\int_0^{\frac{\pi}{2}} \ln(\sin(2x)) \, dx = \frac{1}{2} (2I) = I
$$

5. **Solve for $I$**:

$$
2I = I - \frac{\pi}{2} \ln 2 \implies I = -\frac{\pi}{2} \ln 2
$$

$$
\boxed{-\frac{\pi}{2} \ln 2}
$$

#### Key takeaway
Combining double-angle identities with domain splitting generates self-referential integral equations that yield exact values without primitive integration.

```python
val = integrate.quad(lambda x: np.log(np.sin(x)), 0, np.pi / 2)[0]
print(f"int_0^(pi/2) ln(sin x) dx = {val:.12f}")
print(f"-(pi/2) ln 2              = {-np.pi / 2 * np.log(2):.12f}")
assert abs(val + np.pi / 2 * np.log(2)) < 1e-9
```
