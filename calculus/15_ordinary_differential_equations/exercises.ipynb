### Problem L0.1 — Verification of Solution & Linearity Classification
**Source**: Boyce & DiPrima, *Elementary Differential Equations*, Ch. 1  
**Statement**  
Consider the ordinary differential equation:

$$
t^2 y'' + 2t y' - 6y = 0 \quad (t \gt 0)
$$

1. Classify the order and linearity of the equation.
2. Verify by direct substitution whether $y_1(t) = t^2$ and $y_2(t) = t^{-3}$ are solutions.

#### Intuition
A differential equation is linear if the state function $y$ and its derivatives appear only to the first power without product interactions (e.g., no $y \cdot y'$ or $\sin(y)$ terms). Verification requires computing the necessary derivatives $\frac{dy}{dt}$ and $\frac{d^2 y}{dt^2}$ and demonstrating that the linear operator action equals zero.

#### Solution
1. **Classification**:
   - Highest derivative present is $y''$, so the order is **2 (Second-Order)**.
   - The coefficients $t^2, 2t, -6$ depend solely on the independent variable $t$, and $y, y', y''$ appear linearly. Thus, the ODE is **Linear and Homogeneous**.

2. **Verification of $y_1(t) = t^2$**:
   Compute derivatives:

$$
   y_1'(t) = 2t, \quad y_1''(t) = 2
$$

   Substitute into the differential operator $L[y] = t^2 y'' + 2t y' - 6y$:

$$
   \begin{aligned}
   L[y_1] &= t^2 (2) + 2t (2t) - 6 (t^2) \\
   &= 2t^2 + 4t^2 - 6t^2 = 0
   \end{aligned}
$$

   Hence, $y_1(t) = t^2$ is a solution.

3. **Verification of $y_2(t) = t^{-3}$**:
   Compute derivatives:

$$
   y_2'(t) = -3t^{-4}, \quad y_2''(t) = 12t^{-5}
$$

   Substitute into $L[y]$:

$$
   \begin{aligned}
   L[y_2] &= t^2 (12t^{-5}) + 2t (-3t^{-4}) - 6 (t^{-3}) \\
   &= 12t^{-3} - 6t^{-3} - 6t^{-3} = 0
   \end{aligned}
$$

   Hence, $y_2(t) = t^{-3}$ is also a solution.

$$
\boxed{\text{Order: 2, Linear Homogeneous. Both } y_1(t) = t^2 \text{ and } y_2(t) = t^{-3} \text{ are valid solutions.}}
$$

#### Key takeaway
Power functions $y(t) = t^r$ are canonical trial solutions for Euler-Cauchy linear ODEs $a t^2 y'' + b t y' + c y = 0$.

---

```python
import sympy as sp
t = sp.symbols("t", positive=True)
L = lambda y: t**2 * sp.diff(y, t, 2) + 2 * t * sp.diff(y, t) - 6 * y
for y in (t**2, t**-3):
    r = sp.simplify(L(y))
    print(f"L[{y}] = {r}")
    assert r == 0
print("both are solutions; the operator is linear in y, y', y''")
```

### Problem L0.2 — Fixed Points and Dynamic Direction of 1D Systems
**Source**: S. Strogatz, *Nonlinear Dynamics and Chaos*, Ch. 2  
**Statement**  
Analyze the autonomous first-order ODE:

$$
\frac{dx}{dt} = x^2 - 4
$$

1. Find all fixed points (equilibria).
2. Determine the stability of each fixed point by analyzing the sign of $\frac{dx}{dt}$.

#### Intuition
Fixed points occur where velocity $\frac{dx}{dt} = 0$. If $\frac{dx}{dt} \gt 0$ to the right and $\frac{dx}{dt} \lt 0$ to the left of a fixed point, trajectories diverge from it (unstable). If $\frac{dx}{dt} \lt 0$ to the right and $\frac{dx}{dt} \gt 0$ to the left, trajectories converge to it (stable).

#### Solution
1. **Fixed Points**:
   Set $\frac{dx}{dt} = 0$:

$$
   x^2 - 4 = 0 \implies (x - 2)(x + 2) = 0 \implies x_1^\ast = -2, \quad x_2^\ast = 2
$$

2. **Linear Stability Analysis**:
   Let $f(x) = x^2 - 4$. Differentiate $f(x)$:

$$
   f'(x) = 2x
$$

   - At $x_1^\ast = -2$: $f'(-2) = -4 \lt 0$. Since the linearized derivative is negative, $x^\ast = -2$ is **Asymptotically Stable** (Attractor).
   - At $x_2^\ast = 2$: $f'(2) = +4 \gt 0$. Since the linearized derivative is positive, $x^\ast = 2$ is **Unstable** (Repeller).

$$
\boxed{x^\ast = -2 \text{ is Asymptotically Stable}, \quad x^\ast = 2 \text{ is Unstable.}}
$$

#### Key takeaway
For 1D autonomous flows $\dot{x} = f(x)$, $f'(x^\ast) \lt 0$ guarantees local exponential stability, while $f'(x^\ast) \gt 0$ guarantees instability.

---

```python
import numpy as np
f = lambda x: x**2 - 4
fp = lambda x: 2 * x
for xs in (-2.0, 2.0):
    print(f"x* = {xs:+.1f}: f(x*) = {f(xs):.1e}, f'(x*) = {fp(xs):+.1f} "
          f"-> {'stable' if fp(xs) < 0 else 'unstable'}")
assert fp(-2.0) < 0 < fp(2.0)
```

### Problem L0.3 — Geometric Interpretation of Slope Fields & Integral Curves
**Source**: V.I. Arnold, *Ordinary Differential Equations*, Ch. 1  
**Statement**  
Consider the first-order differential equation:

$$
\frac{dy}{dx} = -\frac{x}{y} \quad (y \ne 0)
$$

Identify the family of integral curves geometrically and algebraically.

#### Intuition
The slope of the integral curve at any point $(x,y)$ is $m_1 = -\frac{x}{y}$. The position vector from the origin to $(x,y)$ has slope $m_2 = \frac{y}{x}$. Notice that $m_1 \cdot m_2 = \left(-\frac{x}{y}\right) \left(\frac{y}{x}\right) = -1$. The velocity vector of the trajectory is perpendicular to the radial vector.

#### Solution
Separate variables:

$$
y \, dy = -x \, dx
$$

Integrate both sides:

$$
\int y \, dy = -\int x \, dx \implies \frac{y^2}{2} = -\frac{x^2}{2} + C'
$$

Multiply by 2 and define $R^2 = 2C'$:

$$
x^2 + y^2 = R^2 \quad (R \gt 0)
$$

$$
\boxed{\text{The integral curves form a family of concentric circles centered at the origin: } x^2 + y^2 = R^2.}
$$

#### Key takeaway
Perpendicularity between state vectors and slope fields generates circular rotational orbits.

### Problem L0.4 — Superposition Principle & Vector Space of Solutions
**Source**: MIT 18.03 Differential Equations  
**Statement**  
Prove that the set of solutions $V = \{y \in C^2(\mathbb{R}) : y'' + p(t) y' + q(t) y = 0\}$ forms a vector space over $\mathbb{R}$.

#### Intuition
Linearity of the differential operator $L = \frac{d^2}{dt^2} + p(t) \frac{d}{dt} + q(t)$ implies $L[c_1 y_1 + c_2 y_2] = c_1 L[y_1] + c_2 L[y_2]$.

#### Solution
Let $y_1, y_2 \in V$ and $\alpha, \beta \in \mathbb{R}$. Compute $L[\alpha y_1 + \beta y_2]$:

$$
\begin{aligned}
L[\alpha y_1 + \beta y_2] &= (\alpha y_1 + \beta y_2)'' + p(t)(\alpha y_1 + \beta y_2)' + q(t)(\alpha y_1 + \beta y_2) \\
&= \alpha y_1'' + \beta y_2'' + p(t)\alpha y_1' + p(t)\beta y_2' + q(t)\alpha y_1 + q(t)\beta y_2 \\
&= \alpha \left( y_1'' + p(t)y_1' + q(t)y_1 \right) + \beta \left( y_2'' + p(t)y_2' + q(t)y_2 \right) \\
&= \alpha L[y_1] + \beta L[y_2]
\end{aligned}
$$

Since $y_1, y_2 \in V$, $L[y_1] = 0$ and $L[y_2] = 0$.
Thus:

$$
L[\alpha y_1 + \beta y_2] = \alpha(0) + \beta(0) = 0 \implies \alpha y_1 + \beta y_2 \in V
$$

$$
\boxed{V \text{ is closed under addition and scalar multiplication, hence it is a vector subspace of } C^2(\mathbb{R}).}
$$

#### Key takeaway
Homogeneous linear ODE solutions form vector spaces, enabling linear algebra techniques (bases, dimension, span).

### Problem L0.5 — Wronskian Non-Vanishing Condition
**Source**: Boyce & DiPrima, *Elementary Differential Equations*, Ch. 3  
**Statement**  
Compute the Wronskian of $y_1(t) = e^{2t}$ and $y_2(t) = e^{-3t}$, and determine whether they form a fundamental set of solutions.

#### Intuition
The Wronskian determinant $W(y_1, y_2)(t)$ measures the linear independence of solution curves.

#### Solution
Compute derivatives:

$$
y_1'(t) = 2e^{2t}, \quad y_2'(t) = -3e^{-3t}
$$

Evaluate the Wronskian determinant:

$$
\begin{aligned}
W(y_1, y_2)(t) &= \det \begin{pmatrix} e^{2t} & e^{-3t} \\ 2e^{2t} & -3e^{-3t} \end{pmatrix} \\
&= e^{2t}(-3e^{-3t}) - e^{-3t}(2e^{2t}) \\
&= -3e^{-t} - 2e^{-t} = -5e^{-t}
\end{aligned}
$$

Since $-5e^{-t} \ne 0$ for all $t \in \mathbb{R}$, $y_1(t)$ and $y_2(t)$ are linearly independent and form a fundamental set of solutions.

$$
\boxed{W(y_1, y_2)(t) = -5e^{-t} \ne 0; \text{ the solutions form a fundamental basis.}}
$$

#### Key takeaway
A non-vanishing Wronskian guarantees that linear combinations can satisfy any arbitrary initial state $(y(t_0), y'(t_0))$.

---

```python
import sympy as sp
t = sp.symbols("t")
y1, y2 = sp.exp(2 * t), sp.exp(-3 * t)
W = sp.simplify(y1 * sp.diff(y2, t) - sp.diff(y1, t) * y2)
print("W(t) =", W)
assert sp.simplify(W + 5 * sp.exp(-t)) == 0
```

### Problem L0.6 — Trace-Determinant Criterion for 2D Linear Systems
**Source**: S. Strogatz, *Nonlinear Dynamics and Chaos*, Ch. 5  
**Statement**  
Classify the equilibrium at $(0,0)$ for the linear system $\dot{\mathbf{x}} = A \mathbf{x}$ with:

$$
A = \begin{pmatrix} -2 & 1 \\ 1 & -4 \end{pmatrix}
$$

#### Intuition
System behavior is dictated by the trace $\tau = \operatorname{tr}(A)$ and determinant $\Delta = \det(A)$ of the system matrix.

#### Solution
Compute trace and determinant:

$$
\tau = \operatorname{tr}(A) = -2 + (-4) = -6
$$

$$
\Delta = \det(A) = (-2)(-4) - (1)(1) = 8 - 1 = 7
$$

Compute the discriminant:

$$
\Delta_{disc} = \tau^2 - 4\Delta = (-6)^2 - 4(7) = 36 - 28 = 8 \gt 0
$$

Since $\Delta = 7 \gt 0$, $\tau = -6 \lt 0$, and $\tau^2 - 4\Delta \gt 0$, the eigenvalues are real, distinct, and both negative:

$$
\lambda_{1,2} = \frac{-6 \pm \sqrt{8}}{2} = -3 \pm \sqrt{2} \quad (\lambda_1 \approx -1.586, \, \lambda_2 \approx -4.414)
$$

$$
\boxed{\text{The equilibrium at } (0,0) \text{ is a Stable Node (Sink).}}
$$

#### Key takeaway
Negative trace and positive determinant with positive discriminant guarantee pure exponential decay along distinct eigen-directions.

---

```python
import numpy as np
A = np.array([[-2.0, 1.0], [1.0, -4.0]])
tau, det = np.trace(A), np.linalg.det(A)
ev = np.sort(np.linalg.eigvals(A).real)
print(f"tau = {tau:.0f}, det = {det:.0f}, disc = {tau**2 - 4*det:.0f}")
print(f"eigenvalues = {ev}   (-3 -/+ sqrt 2 = {-3-np.sqrt(2):.3f}, {-3+np.sqrt(2):.3f})")
assert tau < 0 < det and tau**2 - 4 * det > 0
assert np.allclose(ev, sorted([-3 - np.sqrt(2), -3 + np.sqrt(2)]))
print("stable node (sink)")
```

### Problem L0.7 — Matrix Exponential Commutativity Condition
**Source**: T.M. Apostol, *Calculus Volume II*, Ch. 7  
**Statement**  
Prove that $e^{At} e^{Bt} = e^{(A+B)t}$ for all $t$ if and only if $A$ and $B$ commute ($AB = BA$).

#### Intuition
The exponential series relies on the binomial theorem $(A+B)^k = \sum \binom{k}{j} A^j B^{k-j}$, which is valid if and only if $A$ and $B$ commute.

#### Solution
Expand both sides in Taylor series up to $t^2$:
LHS:

$$
\begin{aligned}
e^{At} e^{Bt} &= \left( I + A t + \frac{A^2 t^2}{2} + O(t^3) \right) \left( I + B t + \frac{B^2 t^2}{2} + O(t^3) \right) \\
&= I + (A+B)t + \left( \frac{A^2}{2} + AB + \frac{B^2}{2} \right) t^2 + O(t^3)
\end{aligned}
$$

RHS:

$$
e^{(A+B)t} = I + (A+B)t + \frac{(A+B)^2}{2} t^2 + O(t^3) = I + (A+B)t + \left( \frac{A^2 + AB + BA + B^2}{2} \right) t^2 + O(t^3)
$$

Equating $t^2$ coefficients:

$$
\frac{A^2}{2} + AB + \frac{B^2}{2} = \frac{A^2 + AB + BA + B^2}{2}
$$

Multiply by 2 and cancel $A^2 + B^2$:

$$
2AB = AB + BA \implies AB = BA
$$

$$
\boxed{e^{At} e^{Bt} = e^{(A+B)t} \ \text{ for all } t \in \mathbb{R} \iff AB = BA}
$$

#### Key takeaway
The quantifier is what makes this an equivalence: for a **single** fixed $t$ the identity can hold with $AB \ne BA$, so commuting is sufficient but not necessary at one instant — it is necessary only if the identity is demanded for every $t$.

---

```python
import numpy as np
from scipy.linalg import expm
A = np.array([[1.0, 0.0], [0.0, 2.0]])
B_comm = np.array([[3.0, 0.0], [0.0, -1.0]])          # diagonal: commutes with A
B_noncomm = np.array([[0.0, 1.0], [1.0, 0.0]])        # does not commute
for name, B in (("commuting", B_comm), ("non-commuting", B_noncomm)):
    c = np.linalg.norm(A @ B - B @ A)
    g = np.linalg.norm(expm(A) @ expm(B) - expm(A + B))
    print(f"{name:>14}: ||AB-BA|| = {c:7.4f}   ||e^A e^B - e^(A+B)|| = {g:.4e}")
assert np.linalg.norm(expm(A) @ expm(B_comm) - expm(A + B_comm)) < 1e-12
assert np.linalg.norm(expm(A) @ expm(B_noncomm) - expm(A + B_noncomm)) > 1e-2

# at a single t the identity can hold without commuting: take t = 1 and
# A, B equal to 2*pi*i times diagonalisable matrices with integer eigenvalues
A2 = 2j * np.pi * np.array([[1, 0], [0, -1]], dtype=complex)
B2 = 2j * np.pi * np.array([[1, 1], [0, -1]], dtype=complex)
print(f"\nsingle t = 1 witness: ||AB-BA|| = {np.linalg.norm(A2@B2 - B2@A2):.4f}, "
      f"||e^A e^B - e^(A+B)|| = {np.linalg.norm(expm(A2)@expm(B2) - expm(A2+B2)):.2e}")
# but the same pair breaks the identity at other t, so 'for all t' fails
print("   t      ||e^(At) e^(Bt) - e^((A+B)t)||")
gaps = {}
for tv in (0.1, 0.25, 0.5, 1.0):
    gaps[tv] = np.linalg.norm(expm(A2 * tv) @ expm(B2 * tv) - expm((A2 + B2) * tv))
    print(f"{tv:6.2f}   {gaps[tv]:.4e}")
assert np.linalg.norm(A2 @ B2 - B2 @ A2) > 1.0
assert gaps[1.0] < 1e-12 and gaps[0.25] > 0.5
```

### Problem L0.8 — ResNet as Forward Euler Discretization
**Source**: Chen et al. (2018), *Neural Ordinary Differential Equations*  
**Statement**  
Show that a Residual Network layer $\mathbf{h}_{l+1} = \mathbf{h}_l + f(\mathbf{h}_l, \theta_l)$ is equivalent to a Forward Euler step for an ODE $\frac{d\mathbf{h}}{dt} = f(\mathbf{h}(t), \theta(t))$.

#### Intuition
Forward Euler discretizes continuous derivatives using finite differences: $\frac{\mathbf{h}(t + \Delta t) - \mathbf{h}(t)}{\Delta t} = f(\mathbf{h}(t), \theta(t))$.

#### Solution
Consider the initial value problem:

$$
\frac{d\mathbf{h}}{dt} = f(\mathbf{h}(t), t, \theta)
$$

Apply Forward Euler with uniform step size $\Delta t = 1$:

$$
\mathbf{h}(t + 1) = \mathbf{h}(t) + 1 \cdot f(\mathbf{h}(t), t, \theta)
$$

Discretizing time $t \to l$ (layer index):

$$
\mathbf{h}_{l+1} = \mathbf{h}_l + f(\mathbf{h}_l, \theta_l)
$$

$$
\boxed{\text{Residual layers are Forward Euler step approximations of continuous feature flows with step size } \Delta t = 1.}
$$

#### Key takeaway
Neural ODEs generalize deep discrete networks to continuous-depth architectures with adaptive numerical integration.

## L1 — Foundations

### Problem L1.1 — First-Order Separable ODE Initial Value Problem
**Source**: Demidovich, *Problems in Mathematical Analysis*, No. 2785  
**Statement**  
Solve the initial value problem:

$$
\frac{dy}{dx} = \frac{x^2}{y (1 + x^3)}, \quad y(0) = 2
$$

#### Intuition
Separate variables to isolate $y$ terms on the left and $x$ terms on the right, then integrate using logarithmic substitution for $x$.

#### Solution
Separate variables:

$$
y \, dy = \frac{x^2}{1 + x^3} \, dx
$$

Integrate both sides:

$$
\int y \, dy = \int \frac{x^2}{1 + x^3} \, dx
$$

For the right integral, substitute $u = 1 + x^3 \implies du = 3x^2 dx \implies x^2 dx = \frac{du}{3}$:

$$
\frac{y^2}{2} = \frac{1}{3} \ln\lvert 1 + x^3 \rvert + C
$$

Apply the initial condition $y(0) = 2$:

$$
\frac{2^2}{2} = \frac{1}{3} \ln\lvert 1 + 0 \rvert + C \implies 2 = 0 + C \implies C = 2
$$

Substitute $C = 2$:

$$
\frac{y^2}{2} = \frac{1}{3} \ln(1 + x^3) + 2 \implies y^2 = \frac{2}{3} \ln(1 + x^3) + 4
$$

Since $y(0) = +2 \gt 0$, select the positive square root:

$$
\boxed{y(x) = \sqrt{\frac{2}{3} \ln(1 + x^3) + 4}}
$$

#### Key takeaway
Initial conditions uniquely determine integral constants and select appropriate function branches (e.g., positive vs. negative root).

---

```python
import sympy as sp
x = sp.symbols("x")
y = sp.sqrt(sp.Rational(2, 3) * sp.log(1 + x**3) + 4)
res = sp.simplify(sp.diff(y, x) - x**2 / (y * (1 + x**3)))
print("ODE residual:", res, "   y(0) =", y.subs(x, 0))
assert res == 0 and y.subs(x, 0) == 2
```

### Problem L1.2 — First-Order Linear ODE via Integrating Factor
**Source**: Boyce & DiPrima, *Elementary Differential Equations*, Ch. 2.1  
**Statement**  
Solve the initial value problem:

$$
t y' + 2y = t^2 - t + 1, \quad y(1) = 0 \quad (t \gt 0)
$$

#### Intuition
Convert the ODE to standard linear form $y' + p(t) y = q(t)$ by dividing by $t$, then construct the integrating factor $\mu(t) = e^{\int p(t) dt}$.

#### Solution
Divide by $t \gt 0$:

$$
y' + \frac{2}{t} y = t - 1 + \frac{1}{t}
$$

Identify $p(t) = \frac{2}{t}$ and $q(t) = t - 1 + \frac{1}{t}$.
Compute the integrating factor $\mu(t)$:

$$
\mu(t) = \exp\left( \int \frac{2}{t} dt \right) = \exp(2 \ln t) = e^{\ln(t^2)} = t^2
$$

Multiply the ODE by $\mu(t) = t^2$:

$$
t^2 y' + 2t y = \frac{d}{dt} [t^2 y] = t^2 \left( t - 1 + \frac{1}{t} \right) = t^3 - t^2 + t
$$

Integrate both sides with respect to $t$:

$$
t^2 y(t) = \int (t^3 - t^2 + t) dt = \frac{t^4}{4} - \frac{t^3}{3} + \frac{t^2}{2} + C
$$

Divide by $t^2$:

$$
y(t) = \frac{t^2}{4} - \frac{t}{3} + \frac{1}{2} + \frac{C}{t^2}
$$

Apply the initial condition $y(1) = 0$:

$$
0 = \frac{1}{4} - \frac{1}{3} + \frac{1}{2} + C = \frac{3 - 4 + 6}{12} + C = \frac{5}{12} + C \implies C = -\frac{5}{12}
$$

$$
\boxed{y(t) = \frac{t^2}{4} - \frac{t}{3} + \frac{1}{2} - \frac{5}{12 t^2}}
$$

#### Key takeaway
Integrating factors transform a linear combination of derivatives into the exact derivative of a product.

---

```python
import sympy as sp
t = sp.symbols("t", positive=True)
y = t**2 / 4 - t / 3 + sp.Rational(1, 2) - sp.Rational(5, 12) / t**2
res = sp.simplify(t * sp.diff(y, t) + 2 * y - (t**2 - t + 1))
print("ODE residual:", res, "   y(1) =", sp.simplify(y.subs(t, 1)))
assert res == 0 and sp.simplify(y.subs(t, 1)) == 0
```

### Problem L1.3 — Exact Differential Equation
**Source**: M. Spivak / Demidovich No. 2810  
**Statement**  
Solve the initial value problem:

$$
(2x y + 3) dx + (x^2 - 1) dy = 0, \quad y(0) = 1
$$

#### Intuition
Check exactness $\frac{\partial M}{\partial y} = \frac{\partial N}{\partial x}$. If exact, find potential function $\Phi(x,y)$ such that $\frac{\partial \Phi}{\partial x} = M$ and $\frac{\partial \Phi}{\partial y} = N$.

#### Solution
Identify $M(x,y) = 2xy + 3$ and $N(x,y) = x^2 - 1$.
Compute partial derivatives:

$$
\frac{\partial M}{\partial y} = 2x, \quad \frac{\partial N}{\partial x} = 2x
$$

Since $\frac{\partial M}{\partial y} = \frac{\partial N}{\partial x}$, the ODE is **exact**.

Find $\Phi(x,y)$:

$$
\Phi(x,y) = \int M(x,y) dx = \int (2xy + 3) dx = x^2 y + 3x + h(y)
$$

Differentiate with respect to $y$ and equate to $N(x,y)$:

$$
\frac{\partial \Phi}{\partial y} = x^2 + h'(y) = x^2 - 1 \implies h'(y) = -1 \implies h(y) = -y
$$

Thus, the potential function is $\Phi(x,y) = x^2 y + 3x - y$.
The general solution is implicit level curve $\Phi(x,y) = C$:

$$
x^2 y + 3x - y = C
$$

Apply $y(0) = 1$:

$$
0^2(1) + 3(0) - 1 = C \implies C = -1
$$

Thus, $x^2 y + 3x - y = -1 \implies y(x^2 - 1) = -3x - 1 \implies y(1 - x^2) = 3x + 1$.

$$
\boxed{y(x) = \frac{3x + 1}{1 - x^2}}
$$

#### Key takeaway
Exact ODEs represent total differentials $d\Phi = 0$ of potential conservative surfaces.

---

```python
import sympy as sp
x = sp.symbols("x")
y = (3 * x + 1) / (1 - x**2)
res = sp.simplify((2 * x * y + 3) + (x**2 - 1) * sp.diff(y, x))
print("M + N y' =", res, "   y(0) =", y.subs(x, 0))
assert res == 0 and y.subs(x, 0) == 1
```

### Problem L1.4 — Non-Exact ODE with Integrating Factor $\mu(x)$
**Source**: Boyce & DiPrima, *Elementary Differential Equations*, Ch. 2.6  
**Statement**  
Find an integrating factor $\mu(x)$ and solve the ODE:

$$
(3x^2 y + 2xy + y^3) dx + (x^2 + y^2) dy = 0
$$

#### Intuition
If $\frac{\partial M}{\partial y} \ne \frac{\partial N}{\partial x}$, check whether $\frac{1}{N}\left(\frac{\partial M}{\partial y} - \frac{\partial N}{\partial x}\right)$ depends only on $x$.

#### Solution
Identify $M = 3x^2 y + 2xy + y^3$ and $N = x^2 + y^2$.
Compute partial derivatives:

$$
\frac{\partial M}{\partial y} = 3x^2 + 2x + 3y^2, \quad \frac{\partial N}{\partial x} = 2x
$$

Difference:

$$
\frac{\partial M}{\partial y} - \frac{\partial N}{\partial x} = 3x^2 + 3y^2 = 3(x^2 + y^2)
$$

Compute the ratio:

$$
\frac{1}{N} \left( \frac{\partial M}{\partial y} - \frac{\partial N}{\partial x} \right) = \frac{3(x^2 + y^2)}{x^2 + y^2} = 3 \quad (\text{depends only on } x)
$$

Compute $\mu(x)$:

$$
\mu(x) = \exp\left( \int 3 \, dx \right) = e^{3x}
$$

Multiply the original ODE by $e^{3x}$:

$$
e^{3x} (3x^2 y + 2xy + y^3) dx + e^{3x} (x^2 + y^2) dy = 0
$$

Find potential $\Phi(x,y)$:

$$
\Phi(x,y) = \int e^{3x}(x^2 + y^2) dy = e^{3x}(x^2 y + \frac{y^3}{3}) + k(x)
$$

Differentiating $\Phi$ with respect to $x$:

$$
\frac{\partial \Phi}{\partial x} = 3e^{3x}(x^2 y + \frac{y^3}{3}) + e^{3x}(2xy) + k'(x) = e^{3x}(3x^2 y + y^3 + 2xy) + k'(x)
$$

Equating to $\mu M = e^{3x}(3x^2 y + 2xy + y^3)$ yields $k'(x) = 0 \implies k(x) = 0$.

$$
\boxed{e^{3x} \left( x^2 y + \frac{y^3}{3} \right) = C}
$$

#### Key takeaway
Integrating factors convert non-conservative vector fields into integrable total differentials.

---

```python
import sympy as sp
x, y = sp.symbols("x y")
Phi = sp.exp(3 * x) * (x**2 * y + y**3 / 3)
M = sp.exp(3 * x) * (3 * x**2 * y + 2 * x * y + y**3)
N = sp.exp(3 * x) * (x**2 + y**2)
print("Phi_x - mu*M =", sp.simplify(sp.diff(Phi, x) - M))
print("Phi_y - mu*N =", sp.simplify(sp.diff(Phi, y) - N))
assert sp.simplify(sp.diff(Phi, x) - M) == 0 and sp.simplify(sp.diff(Phi, y) - N) == 0
```

### Problem L1.5 — 2nd-Order Homogeneous ODE with Complex Conjugate Roots
**Source**: Boyce & DiPrima, *Elementary Differential Equations*, Ch. 3.1  
**Statement**  
Solve the initial value problem:

$$
y'' - 4y' + 13y = 0, \quad y(0) = 2, \quad y'(0) = 1
$$

#### Intuition
Find characteristic polynomial roots $r^2 - 4r + 13 = 0$. Complex conjugate roots $\alpha \pm i \beta$ produce oscillatory solutions damped or amplified by $e^{\alpha t}$.

#### Solution
Characteristic equation:

$$
r^2 - 4r + 13 = 0 \implies r = \frac{4 \pm \sqrt{16 - 52}}{2} = \frac{4 \pm \sqrt{-36}}{2} = 2 \pm 3i
$$

Here $\alpha = 2, \beta = 3$. The general solution is:

$$
y(t) = e^{2t} \left( C_1 \cos(3t) + C_2 \sin(3t) \right)
$$

Apply $y(0) = 2$:

$$
y(0) = e^0 (C_1 \cos 0 + C_2 \sin 0) = C_1 = 2
$$

Differentiate $y(t)$:

$$
y'(t) = 2e^{2t}(C_1 \cos 3t + C_2 \sin 3t) + e^{2t}(-3C_1 \sin 3t + 3C_2 \cos 3t)
$$

Apply $y'(0) = 1$ with $C_1 = 2$:

$$
y'(0) = 2(2) + 3C_2 = 4 + 3C_2 = 1 \implies 3C_2 = -3 \implies C_2 = -1
$$

$$
\boxed{y(t) = e^{2t} \left( 2 \cos(3t) - \sin(3t) \right)}
$$

#### Key takeaway
Real parts of complex characteristic roots govern exponential growth/decay, while imaginary parts dictate oscillation frequencies.

---

```python
import sympy as sp
t = sp.symbols("t")
y = sp.exp(2 * t) * (2 * sp.cos(3 * t) - sp.sin(3 * t))
res = sp.simplify(sp.diff(y, t, 2) - 4 * sp.diff(y, t) + 13 * y)
print("residual:", res, "  y(0) =", y.subs(t, 0), "  y'(0) =", sp.diff(y, t).subs(t, 0))
assert res == 0 and y.subs(t, 0) == 2 and sp.diff(y, t).subs(t, 0) == 1
```

### Problem L1.6 — 2nd-Order Homogeneous ODE with Repeated Roots
**Source**: Demidovich, *Problems in Mathematical Analysis*, No. 2950  
**Statement**  
Solve the initial value problem:

$$
y'' - 6y' + 9y = 0, \quad y(0) = 1, \quad y'(0) = 4
$$

#### Intuition
Repeated roots $r_1 = r_2 = r$ require multiplying by $t$ to generate a second linearly independent solution $t e^{r t}$.

#### Solution
Characteristic equation:

$$
r^2 - 6r + 9 = 0 \implies (r - 3)^2 = 0 \implies r_1 = r_2 = 3
$$

The general solution is:

$$
y(t) = (C_1 + C_2 t) e^{3t}
$$

Apply $y(0) = 1$:

$$
y(0) = C_1 = 1
$$

Differentiate $y(t)$:

$$
y'(t) = C_2 e^{3t} + 3(C_1 + C_2 t) e^{3t}
$$

Apply $y'(0) = 4$ with $C_1 = 1$:

$$
y'(0) = C_2 + 3(1) = C_2 + 3 = 4 \implies C_2 = 1
$$

$$
\boxed{y(t) = (1 + t) e^{3t}}
$$

#### Key takeaway
Degenerate characteristic roots produce polynomial-exponential secular growth terms $t^k e^{r t}$.

---

```python
import sympy as sp
t = sp.symbols("t")
y = (1 + t) * sp.exp(3 * t)
res = sp.simplify(sp.diff(y, t, 2) - 6 * sp.diff(y, t) + 9 * y)
print("residual:", res, "  y(0) =", y.subs(t, 0), "  y'(0) =", sp.diff(y, t).subs(t, 0))
assert res == 0 and y.subs(t, 0) == 1 and sp.diff(y, t).subs(t, 0) == 4
```

### Problem L1.7 — Method of Undetermined Coefficients
**Source**: Boyce & DiPrima, *Elementary Differential Equations*, Ch. 3.5  
**Statement**  
Find the general solution to:

$$
y'' - 3y' - 4y = 3 e^{2t}
$$

#### Intuition
Find $y_h(t)$ for the homogeneous system, then propose a trial particular solution $y_p(t) = A e^{2t}$ matching the forcing term structure.

#### Solution
1. **Homogeneous Solution**:

$$
   r^2 - 3r - 4 = 0 \implies (r - 4)(r + 1) = 0 \implies r_1 = 4, \, r_2 = -1
$$

$$
   y_h(t) = C_1 e^{4t} + C_2 e^{-t}
$$

2. **Particular Solution**:
   Since $r = 2$ is not a root of the characteristic polynomial, propose $y_p(t) = A e^{2t}$.
   Compute derivatives:

$$
   y_p' = 2A e^{2t}, \quad y_p'' = 4A e^{2t}
$$

   Substitute into the non-homogeneous ODE:

$$
   4A e^{2t} - 3(2A e^{2t}) - 4(A e^{2t}) = 3 e^{2t}
$$

$$
   (4A - 6A - 4A) e^{2t} = -6A e^{2t} = 3 e^{2t} \implies -6A = 3 \implies A = -\frac{1}{2}
$$

   Thus $y_p(t) = -\frac{1}{2} e^{2t}$.

$$
\boxed{y(t) = C_1 e^{4t} + C_2 e^{-t} - \frac{1}{2} e^{2t}}
$$

#### Key takeaway
Undetermined coefficients works efficiently when forcing terms are eigenfunctions of linear differential operators.

---

```python
import sympy as sp
t, C1, C2 = sp.symbols("t C1 C2")
y = C1 * sp.exp(4 * t) + C2 * sp.exp(-t) - sp.exp(2 * t) / 2
res = sp.simplify(sp.diff(y, t, 2) - 3 * sp.diff(y, t) - 4 * y - 3 * sp.exp(2 * t))
print("residual for arbitrary C1, C2:", res)
assert res == 0
```

### Problem L1.8 — Variation of Parameters for 2nd-Order Non-Homogeneous ODE
**Source**: Boyce & DiPrima, *Elementary Differential Equations*, Ch. 3.6  
**Statement**  
Find the general solution to:

$$
y'' + y = \tan t \quad \left( -\frac{\pi}{2} \lt t \lt \frac{\pi}{2} \right)
$$

#### Intuition
When forcing terms are non-exponential/non-polynomial (such as $\tan t$), apply Variation of Parameters: $y_p = -y_1 \int \frac{y_2 g}{W} dt + y_2 \int \frac{y_1 g}{W} dt$.

#### Solution
1. **Homogeneous Solution**:
   $r^2 + 1 = 0 \implies r = \pm i$.

$$
   y_1(t) = \cos t, \quad y_2(t) = \sin t
$$

2. **Wronskian**:

$$
   W(t) = \det \begin{pmatrix} \cos t & \sin t \\ -\sin t & \cos t \end{pmatrix} = \cos^2 t + \sin^2 t = 1
$$

3. **Particular Solution Integrals**:

$$
   u_1(t) = -\int \frac{\sin t \cdot \tan t}{1} dt = -\int \frac{\sin^2 t}{\cos t} dt = -\int \frac{1 - \cos^2 t}{\cos t} dt = -\int (\sec t - \cos t) dt
$$

$$
   u_1(t) = -\ln\lvert \sec t + \tan t \rvert + \sin t
$$

$$
   u_2(t) = \int \frac{\cos t \cdot \tan t}{1} dt = \int \sin t \, dt = -\cos t
$$

4. **Assemble $y_p(t)$**:

$$
   \begin{aligned}
   y_p(t) &= u_1(t) \cos t + u_2(t) \sin t \\
   &= (-\ln\lvert \sec t + \tan t \rvert + \sin t)\cos t + (-\cos t)\sin t \\
   &= -\cos t \ln\lvert \sec t + \tan t \rvert + \sin t \cos t - \sin t \cos t \\
   &= -\cos t \ln\lvert \sec t + \tan t \rvert
   \end{aligned}
$$

$$
\boxed{y(t) = C_1 \cos t + C_2 \sin t - \cos t \ln\lvert \sec t + \tan t \rvert}
$$

#### Key takeaway
Variation of Parameters provides a universal constructive integral formula for any arbitrary continuous forcing function $g(t)$.

---

```python
import sympy as sp
t, C1, C2 = sp.symbols("t C1 C2")
y = C1 * sp.cos(t) + C2 * sp.sin(t) - sp.cos(t) * sp.log(sp.sec(t) + sp.tan(t))
res = sp.simplify(sp.diff(y, t, 2) + y - sp.tan(t))
print("residual:", res)
assert res == 0
```

### Problem L1.9 — Matrix Exponential Computation via Diagonalization
**Source**: T.M. Apostol, *Calculus Volume II*, Ch. 7  
**Statement**  
Compute $e^{At}$ for the matrix:

$$
A = \begin{pmatrix} 1 & 2 \\ 0 & 3 \end{pmatrix}
$$

#### Intuition
Since $A$ is upper triangular, its eigenvalues are on the main diagonal. Diagonalize $A = P D P^{-1}$ to evaluate $e^{At} = P e^{Dt} P^{-1}$.

#### Solution
1. **Eigenvalues**: $\lambda_1 = 1, \lambda_2 = 3$.
2. **Eigenvectors**:
   - For $\lambda_1 = 1$:

$$
     (A - I)\mathbf{v}_1 = \mathbf{0} \implies \begin{pmatrix} 0 & 2 \\ 0 & 2 \end{pmatrix}\begin{pmatrix} v_1 \\ v_2 \end{pmatrix} = \mathbf{0} \implies v_2 = 0 \implies \mathbf{v}_1 = \begin{pmatrix} 1 \\ 0 \end{pmatrix}
$$

   - For $\lambda_2 = 3$:

$$
     (A - 3I)\mathbf{v}_2 = \mathbf{0} \implies \begin{pmatrix} -2 & 2 \\ 0 & 0 \end{pmatrix}\begin{pmatrix} v_1 \\ v_2 \end{pmatrix} = \mathbf{0} \implies v_1 = v_2 \implies \mathbf{v}_2 = \begin{pmatrix} 1 \\ 1 \end{pmatrix}
$$

3. **Construct $P, D, P^{-1}$**:

$$
   P = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}, \quad D = \begin{pmatrix} 1 & 0 \\ 0 & 3 \end{pmatrix}, \quad P^{-1} = \begin{pmatrix} 1 & -1 \\ 0 & 1 \end{pmatrix}
$$

4. **Compute $e^{At}$**:

$$
   e^{Dt} = \begin{pmatrix} e^t & 0 \\ 0 & e^{3t} \end{pmatrix}
$$

$$
   \begin{aligned}
   e^{At} = P e^{Dt} P^{-1} &= \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix} \begin{pmatrix} e^t & 0 \\ 0 & e^{3t} \end{pmatrix} \begin{pmatrix} 1 & -1 \\ 0 & 1 \end{pmatrix} \\
   &= \begin{pmatrix} e^t & e^{3t} \\ 0 & e^{3t} \end{pmatrix} \begin{pmatrix} 1 & -1 \\ 0 & 1 \end{pmatrix} \\
   &= \begin{pmatrix} e^t & e^{3t} - e^t \\ 0 & e^{3t} \end{pmatrix}
   \end{aligned}
$$

$$
\boxed{e^{At} = \begin{pmatrix} e^t & e^{3t} - e^t \\ 0 & e^{3t} \end{pmatrix}}
$$

#### Key takeaway
Diagonalization decouples system modes, converting matrix exponentiation into independent scalar exponentiations.

---

```python
import sympy as sp
t = sp.symbols("t")
A = sp.Matrix([[1, 2], [0, 3]])
hand = sp.Matrix([[sp.exp(t), sp.exp(3 * t) - sp.exp(t)], [0, sp.exp(3 * t)]])
print(sp.simplify((A * t).exp() - hand))
assert sp.simplify((A * t).exp() - hand) == sp.zeros(2, 2)
```

### Problem L1.10 — Matrix Exponential for Non-Diagonalizable / Repeated Matrices
**Source**: T.M. Apostol, *Calculus Volume II*, Ch. 7  
**Statement**  
Compute $e^{At}$ for the non-diagonalizable matrix:

$$
A = \begin{pmatrix} 2 & 1 \\ 0 & 2 \end{pmatrix}
$$

#### Intuition
Decompose $A = 2I + N$, where

$$
N = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}
$$

is a nilpotent matrix ($N^2 = 0$). Since $2I$ and $N$ commute, $e^{At} = e^{2It} e^{Nt}$.

#### Solution
1. **Check Nilpotency**:

$$
   N = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix} \implies N^2 = \begin{pmatrix} 0 & 0 \\ 0 & 0 \end{pmatrix}
$$

2. **Compute Exponents**:

$$
   e^{2It} = e^{2t} I = \begin{pmatrix} e^{2t} & 0 \\ 0 & e^{2t} \end{pmatrix}
$$

$$
   e^{Nt} = I + N t + \frac{N^2 t^2}{2!} + \dots = \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} + \begin{pmatrix} 0 & t \\ 0 & 0 \end{pmatrix} = \begin{pmatrix} 1 & t \\ 0 & 1 \end{pmatrix}
$$

3. **Multiply ($2It$ commutes with $Nt$)**:

$$
   e^{At} = e^{2It} e^{Nt} = e^{2t} \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} \begin{pmatrix} 1 & t \\ 0 & 1 \end{pmatrix} = e^{2t} \begin{pmatrix} 1 & t \\ 0 & 1 \end{pmatrix}
$$

$$
\boxed{e^{At} = \begin{pmatrix} e^{2t} & t e^{2t} \\ 0 & e^{2t} \end{pmatrix}}
$$

#### Key takeaway
Nilpotent components in Jordan blocks create linear/polynomial secular multipliers $t^k e^{\lambda t}$.

---

```python
import sympy as sp
t = sp.symbols("t")
A = sp.Matrix([[2, 1], [0, 2]])
hand = sp.Matrix([[sp.exp(2 * t), t * sp.exp(2 * t)], [0, sp.exp(2 * t)]])
print(sp.simplify((A * t).exp() - hand))
assert sp.simplify((A * t).exp() - hand) == sp.zeros(2, 2)
```

### Problem L2.1 — Damped Harmonic Oscillator Dynamics
**Source**: Classical Physics / Boyce & DiPrima  
**Statement**  
A spring-mass-damper system obeys $m \ddot{x} + c \dot{x} + k x = 0$ with $m = 1$, $k = 25$, $x(0) = 1$, $\dot{x}(0) = 0$.
Find the exact motion $x(t)$ for damping values:
1. $c = 10$ (Critical Damping)
2. $c = 6$ (Underdamping)

#### Intuition
Critical damping ($\zeta = 1$) returns the system to equilibrium as quickly as possible without oscillation. Underdamping ($\zeta \lt 1$) generates decaying harmonic oscillations.

#### Solution
1. **Case 1: $c = 10$**:
   ODE: $\ddot{x} + 10 \dot{x} + 25 x = 0$.
   Characteristic equation: $r^2 + 10r + 25 = 0 \implies (r + 5)^2 = 0 \implies r = -5$ (Repeated).
   General solution: $x(t) = (C_1 + C_2 t) e^{-5t}$.
   Apply $x(0) = 1 \implies C_1 = 1$.
   Derivative: $\dot{x}(t) = C_2 e^{-5t} - 5(1 + C_2 t) e^{-5t}$.
   Apply $\dot{x}(0) = 0 \implies C_2 - 5 = 0 \implies C_2 = 5$.

$$
   x_1(t) = (1 + 5t) e^{-5t}
$$

2. **Case 2: $c = 6$**:
   ODE: $\ddot{x} + 6 \dot{x} + 25 x = 0$.
   Characteristic equation: $r^2 + 6r + 25 = 0 \implies r = \frac{-6 \pm \sqrt{36 - 100}}{2} = -3 \pm 4i$.
   General solution: $x(t) = e^{-3t}(C_1 \cos 4t + C_2 \sin 4t)$.
   Apply $x(0) = 1 \implies C_1 = 1$.
   Derivative: $\dot{x}(t) = -3e^{-3t}(\cos 4t + C_2 \sin 4t) + e^{-3t}(-4\sin 4t + 4C_2 \cos 4t)$.
   Apply $\dot{x}(0) = 0 \implies -3(1) + 4C_2 = 0 \implies C_2 = \frac{3}{4}$.

$$
   x_2(t) = e^{-3t} \left( \cos(4t) + \frac{3}{4} \sin(4t) \right)
$$

$$
\boxed{c=10: x(t) = (1+5t)e^{-5t}; \quad c=6: x(t) = e^{-3t}\left(\cos 4t + \frac{3}{4}\sin 4t\right)}
$$

#### Key takeaway
Transition from real to complex characteristic roots marks the boundary between non-oscillatory decay and dynamic oscillation.

---

```python
import sympy as sp
t = sp.symbols("t")
cases = {10: (1 + 5 * t) * sp.exp(-5 * t),
         6: sp.exp(-3 * t) * (sp.cos(4 * t) + sp.Rational(3, 4) * sp.sin(4 * t))}
for c, x in cases.items():
    res = sp.simplify(sp.diff(x, t, 2) + c * sp.diff(x, t) + 25 * x)
    print(f"c = {c:2d}: residual {res}, x(0) = {x.subs(t,0)}, x'(0) = {sp.diff(x,t).subs(t,0)}")
    assert res == 0 and x.subs(t, 0) == 1 and sp.diff(x, t).subs(t, 0) == 0
```

### Problem L2.2 — Driven RLC Circuit & Steady-State Resonance
**Source**: Circuit Theory / Physics  
**Statement**  
A driven series RLC circuit satisfies $L \frac{d^2 q}{dt^2} + R \frac{dq}{dt} + \frac{1}{C} q = V_0 \cos(\omega t)$.
Find the resonant frequency $\omega_r$ that maximizes the steady-state current amplitude $I_0(\omega) = \omega q_0(\omega)$.

#### Intuition
Current $I(t) = \frac{dq}{dt}$. In steady state, the system responds at driving frequency $\omega$. Electrical impedance $Z(\omega)$ reaches a minimum at resonance, maximizing current flow.

#### Solution
Differentiate the ODE with respect to $t$, setting $I = \frac{dq}{dt}$:

$$
L I'' + R I' + \frac{1}{C} I = -V_0 \omega \sin(\omega t)
$$

Propose steady-state solution $I_p(t) = I_0 \cos(\omega t - \phi)$. Using complex impedance $Z(\omega) = R + i \left( \omega L - \frac{1}{\omega C} \right)$:

$$
I_0(\omega) = \frac{V_0}{\lvert Z(\omega) \rvert} = \frac{V_0}{\sqrt{R^2 + \left( \omega L - \frac{1}{\omega C} \right)^2}}
$$

To maximize $I_0(\omega)$, we minimize the denominator $\lvert Z(\omega) \rvert$. The minimum occurs when the reactive impedance term vanishes:

$$
\omega L - \frac{1}{\omega C} = 0 \implies \omega^2 = \frac{1}{LC} \implies \omega_r = \frac{1}{\sqrt{LC}}
$$

$$
\boxed{\omega_r = \frac{1}{\sqrt{LC}}, \quad I_{0,\max} = \frac{V_0}{R}}
$$

#### Key takeaway
Resonance occurs when the natural frequency of an undamped system matches the external driving frequency.

---

```python
import numpy as np
L, R, C, V0 = 0.5, 3.0, 2e-3, 12.0
w = np.linspace(1.0, 120.0, 400000)
I0 = V0 / np.sqrt(R**2 + (w * L - 1 / (w * C))**2)
w_num = w[np.argmax(I0)]
w_pred = 1 / np.sqrt(L * C)
print(f"argmax numerically = {w_num:.4f} rad/s,  1/sqrt(LC) = {w_pred:.4f} rad/s")
print(f"max current = {I0.max():.6f} A,  V0/R = {V0/R:.6f} A")
assert abs(w_num - w_pred) < 1e-2 and abs(I0.max() - V0 / R) < 1e-6
```

### Problem L2.3 — Phase Plane Analysis of the Simple Pendulum
**Source**: S. Strogatz, *Nonlinear Dynamics and Chaos*, Ch. 6  
**Statement**  
Analyze the non-linear pendulum ODE:

$$
\ddot{\theta} + \frac{g}{L} \sin \theta = 0
$$

Convert to a 2D first-order system with $\omega = \dot{\theta}$, find all fixed points in $[-\pi, \pi]$, and linearize around each to determine stability.

#### Intuition
Bottom hanging position $\theta = 0$ is intuitively stable, while top inverted position $\theta = \pi$ is unstable.

#### Solution
1. **System Formulation**:

$$
   \begin{cases}
   \dot{\theta} = \omega \\[4pt]
   \dot{\omega} = -\frac{g}{L} \sin \theta
   \end{cases}
$$

2. **Fixed Points**:
   $\omega = 0$ and $\sin \theta = 0 \implies (\theta^\ast, \omega^\ast) = (0, 0)$ and $(\pi, 0)$.

3. **Jacobian Matrix**:

$$
   J(\theta, \omega) = \begin{pmatrix} 0 & 1 \\ -\frac{g}{L} \cos \theta & 0 \end{pmatrix}
$$

4. **Linearization at $(0,0)$**:

$$
   J(0,0) = \begin{pmatrix} 0 & 1 \\ -\frac{g}{L} & 0 \end{pmatrix} \implies \tau = 0, \, \Delta = \frac{g}{L} \gt 0
$$

   Eigenvalues $\lambda = \pm i \sqrt{\frac{g}{L}}$. The linearized system predicts a **Center**.

5. **Linearization at $(\pi,0)$**:

$$
   J(\pi,0) = \begin{pmatrix} 0 & 1 \\ \frac{g}{L} & 0 \end{pmatrix} \implies \tau = 0, \, \Delta = -\frac{g}{L} \lt 0
$$

   Eigenvalues $\lambda = \pm \sqrt{\frac{g}{L}}$. This is a **Saddle Point**.

$$
\boxed{(0,0) \text{ is a Linear Center (Stable Orbits)}, \quad (\pi,0) \text{ is an Unstable Saddle Point.}}
$$

#### Key takeaway
Non-linear phase planes combine centers (bounded oscillations) and saddles (homoclinic separatrix trajectories dividing oscillation from rotation).

---

```python
import numpy as np
g_over_L = 9.81 / 1.0
for name, th in (("(0, 0)", 0.0), ("(pi, 0)", np.pi)):
    J = np.array([[0.0, 1.0], [-g_over_L * np.cos(th), 0.0]])
    ev = np.linalg.eigvals(J)
    print(f"{name:>8}: tau = {np.trace(J):+.2f}, det = {np.linalg.det(J):+.3f}, "
          f"eigenvalues = {np.round(ev, 4)}")
J0 = np.array([[0.0, 1.0], [-g_over_L, 0.0]])
Jp = np.array([[0.0, 1.0], [g_over_L, 0.0]])
assert np.allclose(np.linalg.eigvals(J0).real, 0) and np.linalg.det(Jp) < 0
print("(0,0) linear centre;  (pi,0) saddle")
```

### Problem L2.4 — Lotka-Volterra Conserved Quantity Proof
**Source**: S. Strogatz, *Nonlinear Dynamics and Chaos*, Ch. 6  
**Statement**  
For the Lotka-Volterra system:

$$
\dot{x} = x(1 - y), \quad \dot{y} = y(x - 1)
$$

Prove that $V(x,y) = x - \ln x + y - \ln y$ is a conserved quantity along all trajectories in the positive quadrant $x, y \gt 0$.

#### Intuition
A function $V(x,y)$ is conserved if its total time derivative $\dot{V} = \frac{\partial V}{\partial x} \dot{x} + \frac{\partial V}{\partial y} \dot{y}$ vanishes identically along system solutions.

#### Solution
Compute partial derivatives of $V(x,y)$:

$$
\frac{\partial V}{\partial x} = 1 - \frac{1}{x} = \frac{x - 1}{x}
$$

$$
\frac{\partial V}{\partial y} = 1 - \frac{1}{y} = \frac{y - 1}{y}
$$

Substitute into the orbital derivative $\dot{V}$:

$$
\begin{aligned}
\dot{V}(x,y) &= \frac{\partial V}{\partial x} \dot{x} + \frac{\partial V}{\partial y} \dot{y} \\
&= \left( \frac{x - 1}{x} \right) [x(1 - y)] + \left( \frac{y - 1}{y} \right) [y(x - 1)] \\
&= (x - 1)(1 - y) + (y - 1)(x - 1) \\
&= (x - 1)(1 - y) - (x - 1)(1 - y) = 0
\end{aligned}
$$

Since $\dot{V}(x,y) = 0$ for all $x,y \gt 0$, $V(x,y)$ is constant along trajectories.

$$
\boxed{\frac{dV}{dt} = 0 \implies V(x,y) = x - \ln x + y - \ln y = C \text{ is strictly conserved.}}
$$

#### Key takeaway
Existence of a single non-trivial conserved quantity prevents chaos in 2D autonomous continuous systems.

---

```python
import numpy as np
from scipy.integrate import solve_ivp
V = lambda x, y: x - np.log(x) + y - np.log(y)
f = lambda t, z: [z[0] * (1 - z[1]), z[1] * (z[0] - 1)]
sol = solve_ivp(f, [0, 25], [2.5, 0.8], rtol=1e-12, atol=1e-14,
                t_eval=np.linspace(0, 25, 2000))
drift = np.max(np.abs(V(*sol.y) - V(2.5, 0.8)))
print(f"V(0) = {V(2.5, 0.8):.10f}   max drift over t in [0,25]: {drift:.2e}")
assert drift < 1e-9
```

### Problem L2.5 — Imperfect Pitchfork Bifurcation Analysis
**Source**: S. Strogatz, *Nonlinear Dynamics and Chaos*, Ch. 3  
**Statement**  
Analyze the 1D system with imperfection parameter $h$:

$$
\dot{x} = h + r x - x^3
$$

For $h = 0$, find the critical parameter $r_c$ and classify the bifurcation. Sketch the fixed points as a function of $r$.

#### Intuition
When $h = 0$, the system possesses reflection symmetry $x \to -x$, which dictates a pitchfork structure.

#### Solution
Set $h = 0 \implies \dot{x} = x(r - x^2)$.
Fixed points:
1. $x_1^\ast = 0$ for all $r$.
2. $x_{2,3}^\ast = \pm \sqrt{r}$ for $r \gt 0$.

Linear stability derivative $f'(x) = r - 3x^2$:
- At $x_1^\ast = 0$: $f'(0) = r$.
  - $r \lt 0$: $f'(0) \lt 0 \implies x^\ast = 0$ is **Stable**.
  - $r \gt 0$: $f'(0) \gt 0 \implies x^\ast = 0$ is **Unstable**.
- At $x_{2,3}^\ast = \pm \sqrt{r}$ ($r \gt 0$): $f'(\pm \sqrt{r}) = r - 3(r) = -2r \lt 0 \implies x_{2,3}^\ast$ are **Stable**.

At $r_c = 0$, the single stable origin branches into two symmetric stable branches.

$$
\boxed{r_c = 0 \text{ is a Supercritical Pitchfork Bifurcation.}}
$$

#### Key takeaway
Pitchfork bifurcations model symmetry breaking in physical systems (e.g., buckling beams, magnetization below Curie temperature).

---

```python
import numpy as np
for r in (-1.0, 1.0):
    fps = [0.0] + ([np.sqrt(r), -np.sqrt(r)] if r > 0 else [])
    print(f"r = {r:+.1f}: fixed points {np.round(fps, 4)}, "
          f"f'(x*) = {np.round([r - 3 * x**2 for x in fps], 4)}")
for r in (-1.0, 1.0):
    assert np.sign(r - 3 * 0.0**2) == np.sign(r)          # f'(0) = r
    if r > 0:
        assert r - 3 * (np.sqrt(r))**2 == -2 * r < 0      # outer branches stable
print("origin loses stability at r_c = 0 and two stable branches appear: supercritical pitchfork")
```

### Problem L2.6 — Supercritical Hopf Bifurcation in Polar Coordinates
**Source**: S. Strogatz, *Nonlinear Dynamics and Chaos*, Ch. 8  
**Statement**  
Consider the 2D system in polar coordinates:

$$
\begin{cases}
\dot{r} = \mu r - r^3 \\[4pt]
\dot{\theta} = 1
\end{cases}
$$

1. Analyze fixed points of $r$ as $\mu$ varies.
2. Identify the bifurcation at $\mu_c = 0$ and find the radius and period of the resulting limit cycle for $\mu \gt 0$.

#### Intuition
Radial dynamics $\dot{r} = f(r)$ decouple from angular dynamics $\dot{\theta} = 1$. The radial pitchfork bifurcation manifests as the birth of a circular limit cycle in 2D phase space.

#### Solution
1. **Radial Fixed Points**:
   $\dot{r} = r(\mu - r^2) = 0 \implies r_1^\ast = 0$, and $r_2^\ast = \sqrt{\mu}$ for $\mu \gt 0$.

2. **Stability**:
   - For $\mu \le 0$: $r_1^\ast = 0$ is the only fixed point. Since $\dot{r} \lt 0$, trajectories decay to origin (stable focus/node).
   - For $\mu \gt 0$: $r_1^\ast = 0$ becomes unstable ($\frac{d\dot{r}}{dr}\Big\vert_0 = \mu \gt 0$).
   - At $r^\ast = \sqrt{\mu}$: $\frac{d\dot{r}}{dr}\Big\vert_{\sqrt{\mu}} = \mu - 3(\sqrt{\mu})^2 = -2\mu \lt 0 \implies r^\ast = \sqrt{\mu}$ is a **Stable Limit Cycle**.

3. **Period**:
   $\dot{\theta} = 1 \implies \theta(t) = t + \theta_0 \implies T = \frac{2\pi}{\omega} = 2\pi$.

$$
\boxed{\mu_c = 0 \text{ is a Supercritical Hopf Bifurcation; Limit cycle radius } r^\ast = \sqrt{\mu}, \text{ Period } T = 2\pi.}
$$

#### Key takeaway
Hopf bifurcations represent the fundamental mechanism for the onset of self-sustained oscillations across physical and biological systems.

---

```python
import numpy as np
from scipy.integrate import solve_ivp
mu = 0.75
f = lambda t, z: [mu * z[0] - z[0]**3, 1.0]     # (r, theta)
sol = solve_ivp(f, [0, 60], [0.2, 0.0], rtol=1e-12, atol=1e-14)
r_end = sol.y[0, -1]
print(f"limit radius numerically = {r_end:.10f},  sqrt(mu) = {np.sqrt(mu):.10f}")
print(f"angular period = 2*pi/1 = {2*np.pi:.6f}")
assert abs(r_end - np.sqrt(mu)) < 1e-8
```

### Problem L2.7 — Continuous Gradient Flow Convergence Rate
**Source**: Optimization Theory / Machine Learning  
**Statement**  
Let $L(\mathbf{x}) = \frac{1}{2} \mathbf{x}^T A \mathbf{x}$ where $A \in \mathbb{R}^{n \times n}$ is symmetric positive definite ($A \succ 0$).
1. Solve the continuous gradient flow ODE $\dot{\mathbf{x}}(t) = -\nabla L(\mathbf{x}(t))$ with $\mathbf{x}(0) = \mathbf{x}_0$.
2. Prove exponential convergence to the minimum $\mathbf{x}^\ast = \mathbf{0}$.

#### Intuition
Gradient flow continuously follows the direction of steepest loss decrease. Positive definiteness of $A$ guarantees a unique global minimum at the origin.

#### Solution
1. **Gradient Computation**:

$$
   \nabla L(\mathbf{x}) = A \mathbf{x} \implies \frac{d\mathbf{x}}{dt} = -A \mathbf{x}(t)
$$

2. **Exact Solution via Matrix Exponential**:

$$
   \mathbf{x}(t) = e^{-At} \mathbf{x}_0
$$

3. **Convergence Analysis**:
   Let $\lambda_{\min} \gt 0$ be the smallest eigenvalue of $A$. By Rayleigh quotient bounds:

$$
   \lVert \mathbf{x}(t) \rVert = \lVert e^{-At} \mathbf{x}_0 \rVert \le \lVert e^{-At} \rVert_2 \lVert \mathbf{x}_0 \rVert = e^{-\lambda_{\min} t} \lVert \mathbf{x}_0 \rVert
$$

   Furthermore, evaluate loss decay:

$$
   L(\mathbf{x}(t)) \le \frac{1}{2} \lambda_{\max} \lVert \mathbf{x}(t) \rVert^2 \le \frac{1}{2} \lambda_{\max} \lVert \mathbf{x}_0 \rVert^2 e^{-2\lambda_{\min} t}
$$

$$
\boxed{\mathbf{x}(t) = e^{-At} \mathbf{x}_0, \quad \lVert \mathbf{x}(t) \rVert \le e^{-\lambda_{\min} t} \lVert \mathbf{x}_0 \rVert \implies \lim_{t \to \infty} \mathbf{x}(t) = \mathbf{0}}
$$

#### Key takeaway
Continuous gradient flow on strongly convex losses exhibits global exponential convergence governed by the smallest eigenvalue $\lambda_{\min}(A)$.

---

```python
import numpy as np
from scipy.linalg import expm
rng = np.random.default_rng(0)
Q = np.linalg.qr(rng.standard_normal((4, 4)))[0]
A = Q @ np.diag([0.5, 1.0, 2.0, 6.0]) @ Q.T
x0 = rng.standard_normal(4)
lam_min = 0.5
for t in (0.0, 1.0, 3.0, 8.0):
    x = expm(-A * t) @ x0
    print(f"t = {t:4.1f}: ||x(t)|| = {np.linalg.norm(x):.6e}   "
          f"bound e^(-lam_min t)||x0|| = {np.exp(-lam_min*t)*np.linalg.norm(x0):.6e}")
    assert np.linalg.norm(x) <= np.exp(-lam_min * t) * np.linalg.norm(x0) + 1e-12
```

### Problem L2.8 — Adjoint Sensitivity Equation Derivation for 1D Neural ODE
**Source**: Chen et al. (2018), *Neural ODEs*  
**Statement**  
For a scalar state $h(t)$ governed by $\frac{dh}{dt} = f(h(t), \theta)$ and scalar loss $L(h(T))$, derive the adjoint ODE for $a(t) = \frac{\partial L}{\partial h(t)}$.

#### Intuition
The adjoint state acts as a continuous-time backpropagation signal, propagating output sensitivity backward from time $T$ to time $0$.

#### Solution
Consider an infinitesimal time step from $t$ to $t + \epsilon$:

$$
h(t + \epsilon) = h(t) + \epsilon f(h(t), \theta)
$$

By the chain rule of calculus:

$$
a(t) = \frac{\partial L}{\partial h(t)} = \frac{\partial L}{\partial h(t + \epsilon)} \frac{\partial h(t + \epsilon)}{\partial h(t)} = a(t + \epsilon) \left( 1 + \epsilon \frac{\partial f(h(t), \theta)}{\partial h} \right)
$$

Rearranging:

$$
a(t) = a(t + \epsilon) + \epsilon a(t + \epsilon) \frac{\partial f}{\partial h} \implies \frac{a(t + \epsilon) - a(t)}{\epsilon} = -a(t + \epsilon) \frac{\partial f}{\partial h}
$$

Taking the limit as $\epsilon \to 0$:

$$
\frac{da(t)}{dt} = -a(t) \frac{\partial f(h(t), \theta)}{\partial h}
$$

Terminal condition at time $T$: $a(T) = \frac{\partial L}{\partial h(T)}$.

$$
\boxed{\frac{da(t)}{dt} = -a(t) \frac{\partial f}{\partial h}, \quad a(T) = \frac{\partial L}{\partial h(T)}}
$$

#### Key takeaway
Adjoint ODEs compute exact gradients across continuous time without storing forward activations, reducing memory consumption from $O(N_{steps})$ to $O(1)$.

### Problem L2.9 — Continuous-Time Recurrent Neural Network (CTRNN) Stability
**Source**: Neural Network Theory / Dynamical Systems  
**Statement**  
Analyze stability of the continuous-time recurrent neural network state $\mathbf{h}(t) \in \mathbb{R}^n$:

$$
\tau \frac{d\mathbf{h}}{dt} = -\mathbf{h}(t) + W \mathbf{h}(t) + \mathbf{x}_0 \quad (\tau \gt 0)
$$

Find the equilibrium $\mathbf{h}^\ast$ and state the necessary and sufficient condition on weight matrix $W$ for $\mathbf{h}^\ast$ to be globally asymptotically stable.

#### Intuition
Rewriting $\tau \dot{\mathbf{h}} = (W - I)\mathbf{h} + \mathbf{x}_0$, global stability requires all eigenvalues of $(W - I)$ to have strictly negative real parts.

#### Solution
1. **Equilibrium Point**:
   Set $\frac{d\mathbf{h}}{dt} = \mathbf{0}$:

$$
   -\mathbf{h}^\ast + W \mathbf{h}^\ast + \mathbf{x}_0 = \mathbf{0} \implies (I - W) \mathbf{h}^\ast = \mathbf{x}_0 \implies \mathbf{h}^\ast = (I - W)^{-1} \mathbf{x}_0
$$

   (Assuming $(I - W)$ is invertible).

2. **Linear Stability Analysis**:
   Define error deviation $\boldsymbol{\eta}(t) = \mathbf{h}(t) - \mathbf{h}^\ast$. Substitute:

$$
   \tau \frac{d\boldsymbol{\eta}}{dt} = (W - I) \boldsymbol{\eta}(t) \implies \frac{d\boldsymbol{\eta}}{dt} = \frac{1}{\tau} (W - I) \boldsymbol{\eta}(t)
$$

   The solution is $\boldsymbol{\eta}(t) = e^{\frac{1}{\tau}(W - I)t} \boldsymbol{\eta}(0)$.
   For $\lim_{t \to \infty} \boldsymbol{\eta}(t) = \mathbf{0}$, all eigenvalues $\lambda_i(W - I)$ must satisfy:

$$
   \operatorname{Re}\left( \frac{1}{\tau} (\lambda_i(W) - 1) \right) \lt 0 \implies \operatorname{Re}(\lambda_i(W)) \lt 1 \quad \forall i=1,\dots,n
$$

$$
\boxed{\mathbf{h}^\ast = (I - W)^{-1} \mathbf{x}_0; \text{ Globally Asymptotically Stable } \iff \operatorname{Re}(\lambda_i(W)) \lt 1 \, \forall i.}
$$

#### Key takeaway
Continuous RNN stability requires the real part of every eigenvalue of the recurrent weight matrix to stay strictly below unity.

---

```python
import numpy as np
from scipy.integrate import solve_ivp
rng = np.random.default_rng(0)
W = 0.4 * rng.standard_normal((3, 3))
x0 = rng.standard_normal(3)
tau = 2.0
h_star = np.linalg.solve(np.eye(3) - W, x0)
print("Re(eig(W)) =", np.round(np.linalg.eigvals(W).real, 4))
sol = solve_ivp(lambda t, h: (-h + W @ h + x0) / tau, [0, 400], rng.standard_normal(3),
                rtol=1e-10, atol=1e-12)
print("h* (formula)  =", np.round(h_star, 6))
print("h(400) (ODE)  =", np.round(sol.y[:, -1], 6))
assert np.max(np.linalg.eigvals(W).real) < 1
assert np.allclose(sol.y[:, -1], h_star, atol=1e-6)
```

### Problem L2.10 — RK4 Amplification Factor Derivation
**Source**: Numerical Analysis / Solvers  
**Statement**  
Derive the amplification polynomial $R(z)$ for 4th-Order Runge-Kutta (RK4) applied to $\frac{dy}{dt} = \lambda y$ ($z = \lambda h$).

#### Intuition
Apply RK4 formulas directly to linear $f(t,y) = \lambda y$ to express $y_{n+1} = R(z) y_n$.

#### Solution
1. Compute slope estimates:

$$
   k_1 = \lambda y_n
$$

$$
   k_2 = \lambda \left( y_n + \frac{h}{2} k_1 \right) = \lambda \left( y_n + \frac{h \lambda}{2} y_n \right) = \lambda \left( 1 + \frac{z}{2} \right) y_n
$$

$$
   k_3 = \lambda \left( y_n + \frac{h}{2} k_2 \right) = \lambda \left( y_n + \frac{z}{2} \left( 1 + \frac{z}{2} \right) y_n \right) = \lambda \left( 1 + \frac{z}{2} + \frac{z^2}{4} \right) y_n
$$

$$
   \begin{aligned}
   k_4 &= \lambda ( y_n + h k_3 ) = \lambda \left( y_n + z \left( 1 + \frac{z}{2} + \frac{z^2}{4} \right) y_n \right) \\
   &= \lambda \left( 1 + z + \frac{z^2}{2} + \frac{z^3}{4} \right) y_n
   \end{aligned}
$$

2. Combine into $y_{n+1}$:

$$
   y_{n+1} = y_n + \frac{h}{6} (k_1 + 2k_2 + 2k_3 + k_4)
$$

$$
   \begin{aligned}
   \frac{h}{6}(k_1 + 2k_2 + 2k_3 + k_4) &= \frac{z}{6} \left( 1 + 2(1 + \frac{z}{2}) + 2(1 + \frac{z}{2} + \frac{z^2}{4}) + (1 + z + \frac{z^2}{2} + \frac{z^3}{4}) \right) y_n \\
   &= \frac{z}{6} \left( 6 + 4z + z^2 + \frac{z^3}{4} \right) y_n = \left( z + \frac{z^2}{2} + \frac{z^3}{6} + \frac{z^4}{24} \right) y_n
   \end{aligned}
$$

   Thus:

$$
   y_{n+1} = \left( 1 + z + \frac{z^2}{2} + \frac{z^3}{6} + \frac{z^4}{24} \right) y_n
$$

$$
\boxed{R_{RK4}(z) = 1 + z + \frac{z^2}{2} + \frac{z^3}{6} + \frac{z^4}{24}}
$$

#### Key takeaway
RK4 matches the first five terms of the Taylor series of $e^z$, yielding 4th-order global numerical accuracy.

---

```python
import sympy as sp
z, y = sp.symbols("z y")
k1 = y
k2 = (1 + z / 2) * y
k3 = (1 + z / 2 * (1 + z / 2)) * y
k4 = (1 + z * (1 + z / 2 + z**2 / 4)) * y
step = sp.expand(y + z / 6 * (k1 + 2 * k2 + 2 * k3 + k4))
R = sp.simplify(step / y)
print("R(z) =", sp.expand(R))
assert sp.expand(R - (1 + z + z**2 / 2 + z**3 / 6 + z**4 / 24)) == 0
print("matches the degree-4 Taylor polynomial of e^z:",
      sp.series(sp.exp(z), z, 0, 5).removeO())
```

### Problem L2.11 — Vertical Motion under Quadratic Fluid Resistance
**Source**: Classical Dynamics / Fluid Mechanics  
**Statement**  
A mass $m$ falls from rest under gravity $g$ subject to quadratic drag force $F_d = -k v^2$.
1. Derive the differential equation for velocity $v(t)$.
2. Determine terminal velocity $v_T$ and solve for $v(t)$.

#### Intuition
As speed increases, quadratic drag balances gravity, driving acceleration $\frac{dv}{dt} \to 0$ and approaching a constant terminal velocity $v_T$.

#### Solution
1. **Newton's Second Law**:

$$
   m \frac{dv}{dt} = m g - k v^2 \implies \frac{dv}{dt} = g \left( 1 - \frac{k}{m g} v^2 \right)
$$

   Define terminal velocity $v_T = \sqrt{\frac{mg}{k}}$:

$$
   \frac{dv}{dt} = g \left( 1 - \frac{v^2}{v_T^2} \right)
$$

2. **Separate Variables & Integrate**:

$$
   \int \frac{dv}{1 - \frac{v^2}{v_T^2}} = g \int dt
$$

   Using partial fractions $\frac{1}{1 - (\frac{v}{v_T})^2} = \frac{1}{2}\left( \frac{1}{1 - \frac{v}{v_T}} + \frac{1}{1 + \frac{v}{v_T}} \right)$:

$$
   v_T \operatorname{artanh}\left( \frac{v}{v_T} \right) = g t + C
$$

   Since $v(0) = 0$, $C = 0$.
   Solving for $v(t)$:

$$
   \operatorname{artanh}\left( \frac{v}{v_T} \right) = \frac{g t}{v_T} \implies v(t) = v_T \tanh\left( \frac{g t}{v_T} \right)
$$

$$
\boxed{v_T = \sqrt{\frac{mg}{k}}, \quad v(t) = v_T \tanh\left( \frac{g t}{v_T} \right)}
$$

#### Key takeaway
Hyperbolic tangent solutions model non-linear saturation kinetics in physics and neural activation functions.

---

```python
import sympy as sp
t, m, g, k = sp.symbols("t m g k", positive=True)
vT = sp.sqrt(m * g / k)
v = vT * sp.tanh(g * t / vT)
res = sp.simplify(m * sp.diff(v, t) - (m * g - k * v**2))
print("residual of m v' = mg - k v^2 :", res, "   v(0) =", v.subs(t, 0))
print("limit as t -> oo :", sp.limit(v, t, sp.oo), " = v_T")
assert res == 0 and v.subs(t, 0) == 0
```

### Problem L2.12 — Score-Based Diffusion Probability Flow ODE
**Source**: AI Generative Modeling / Song et al. (2020)  
**Statement**  
Given an SDE $d\mathbf{x} = -\frac{1}{2} \beta(t) \mathbf{x} \, dt + \sqrt{\beta(t)} \, d\mathbf{w}$, show that its deterministic **Probability Flow ODE** sharing identical marginal probability densities $p_t(\mathbf{x})$ is:

$$
\frac{d\mathbf{x}}{dt} = -\frac{1}{2} \beta(t) \left[ \mathbf{x} + \nabla_{\mathbf{x}} \ln p_t(\mathbf{x}) \right]
$$

#### Intuition
The Fokker-Planck equation translates stochastic SDE trajectories into continuous deterministic probability density evolution equations.

#### Solution
1. Write the Fokker-Planck equation for SDE $d\mathbf{x} = \mathbf{f}(\mathbf{x},t)dt + g(t)d\mathbf{w}$ with $\mathbf{f} = -\frac{1}{2}\beta(t)\mathbf{x}$ and $g(t) = \sqrt{\beta(t)}$:

$$
   \frac{\partial p_t}{\partial t} = -\nabla \cdot \left( \mathbf{f} p_t \right) + \frac{1}{2} g(t)^2 \nabla^2 p_t
$$

2. Rewrite the diffusion term using $\nabla^2 p_t = \nabla \cdot (p_t \nabla \ln p_t)$:

$$
   \begin{aligned}
   \frac{\partial p_t}{\partial t} &= -\nabla \cdot \left( \mathbf{f} p_t \right) + \frac{1}{2} g(t)^2 \nabla \cdot \left( p_t \nabla \ln p_t \right) \\
   &= -\nabla \cdot \left[ \left( \mathbf{f} - \frac{1}{2} g(t)^2 \nabla \ln p_t \right) p_t \right]
   \end{aligned}
$$

3. This continuity equation matches the deterministic ODE flow $\frac{d\mathbf{x}}{dt} = \mathbf{u}(\mathbf{x},t)$ where:

$$
   \mathbf{u}(\mathbf{x},t) = \mathbf{f}(\mathbf{x},t) - \frac{1}{2} g(t)^2 \nabla_{\mathbf{x}} \ln p_t(\mathbf{x})
$$

   Substitute $\mathbf{f} = -\frac{1}{2}\beta(t)\mathbf{x}$ and $g(t)^2 = \beta(t)$:

$$
   \mathbf{u}(\mathbf{x},t) = -\frac{1}{2} \beta(t) \mathbf{x} - \frac{1}{2} \beta(t) \nabla_{\mathbf{x}} \ln p_t(\mathbf{x}) = -\frac{1}{2} \beta(t) \left[ \mathbf{x} + \nabla_{\mathbf{x}} \ln p_t(\mathbf{x}) \right]
$$

$$
\boxed{\frac{d\mathbf{x}}{dt} = -\frac{1}{2} \beta(t) \left[ \mathbf{x} + \nabla_{\mathbf{x}} \ln p_t(\mathbf{x}) \right]}
$$

#### Key takeaway
Score-based diffusion models construct continuous deterministic ODE trajectories that invert stochastic noise degradation processes.

### Problem L3.1 — Nonlinear Differential Inequality Bounds
**Source**: Classical differential inequality problem, cf. Boyce & DiPrima, *Elementary Differential Equations*, Ch. 2 (comparison methods)  
**Statement**  
Let $f: \mathbb{R} \to \mathbb{R}$ be a continuously differentiable function satisfying $f(x) + f'(x) \le 1$ for all $x$.
If $f(0) = 0$, prove that $f(x) \le 1 - e^{-x}$ for all $x \ge 0$.

#### Intuition
Multiply the differential inequality by an integrating factor $e^x$ to construct a non-decreasing product function $e^x f(x)$.

#### Solution
Multiply the given inequality $f(x) + f'(x) \le 1$ by $e^x \gt 0$:

$$
e^x f'(x) + e^x f(x) \le e^x \implies \frac{d}{dx} \left[ e^x f(x) \right] \le e^x
$$

Integrate both sides from $0$ to $x$ (where $x \ge 0$):

$$
\int_0^x \frac{d}{dt} \left[ e^t f(t) \right] dt \le \int_0^x e^t \, dt
$$

$$
e^x f(x) - e^0 f(0) \le e^x - e^0
$$

Substitute $f(0) = 0$ and $e^0 = 1$:

$$
e^x f(x) \le e^x - 1
$$

Divide by $e^x \gt 0$:

$$
f(x) \le \frac{e^x - 1}{e^x} = 1 - e^{-x}
$$

$$
\boxed{f(x) \le 1 - e^{-x} \quad \forall x \ge 0}
$$

#### Key takeaway
Integrating factors transform differential inequalities into monotonicity properties for integral comparisons.

---

```python
import numpy as np
from scipy.integrate import solve_ivp
rng = np.random.default_rng(0)
xs = np.linspace(0, 5, 400)
worst = -np.inf
for _ in range(200):
    # any admissible f: f' = 1 - f - s(x) with s >= 0 keeps f + f' <= 1
    amp = 3.0 * rng.random()
    freq = 3.0 * rng.random()
    s = lambda x: amp * (1 + np.sin(freq * x))**2
    sol = solve_ivp(lambda x, f: 1 - f - s(x), [0, 5], [0.0], t_eval=xs,
                    rtol=1e-10, atol=1e-12)
    worst = max(worst, np.max(sol.y[0] - (1 - np.exp(-xs))))
print(f"max over 200 random admissible f of  f(x) - (1 - e^-x)  =  {worst:.3e}")
print("(equality is attained only by the extremal case s == 0, i.e. f' = 1 - f)")
assert worst <= 1e-9
```

### Problem L3.2 — Periodicity of Non-Linear Oscillators
**Source**: Classical Hamiltonian oscillator argument, cf. Arnold, *Ordinary Differential Equations*, Ch. 4 (energy-conservation periodicity)  
**Statement**  
Let $f(x)$ solve $f'' + f^3 = 0$ with $f(0) = 0$ and $f'(0) = 1$.
1. Derive a first integral for $\left(f'\right)^2$ in terms of $f$.
2. Prove that $f(x)$ is periodic and find an integral expression for its period $T$.

#### Intuition
The ODE represents a conservative non-linear oscillator with potential $V(f) = \frac{1}{4} f^4$. Conservation of total energy $E = \frac{1}{2} (f')^2 + \frac{1}{4} f^4$ restricts trajectories to closed energy level curves.

#### Solution
1. **First Integral (Conservation of Energy)**:
   Multiply the ODE by $f'$:

$$
   f' f'' + f^3 f' = 0 \implies \frac{d}{dx} \left[ \frac{1}{2} (f')^2 + \frac{1}{4} f^4 \right] = 0
$$

   Thus $\frac{1}{2} (f')^2 + \frac{1}{4} f^4 = C$.
   Apply initial conditions $f(0) = 0, f'(0) = 1$:

$$
   C = \frac{1}{2}(1)^2 + \frac{1}{4}(0)^4 = \frac{1}{2}
$$

$$
   (f')^2 + \frac{1}{2} f^4 = 1 \implies f' = \sqrt{1 - \frac{1}{2} f^4}
$$

2. **Period Integral Calculation**:
   The maximum amplitude $A$ occurs when $f' = 0$:

$$
   1 - \frac{1}{2} A^4 = 0 \implies A = 2^{\frac{1}{4}}
$$

   The trajectory moves from $f = 0$ to maximum amplitude $f = A$ in one quarter period $\frac{T}{4}$:

$$
   \frac{dx}{df} = \frac{1}{\sqrt{1 - \frac{1}{2} f^4}} \implies \frac{T}{4} = \int_0^{2^{\frac{1}{4}}} \frac{df}{\sqrt{1 - \frac{1}{2} f^4}}
$$

   Substitute $u = 2^{-\frac{1}{4}} f \implies df = 2^{\frac{1}{4}} du$:

$$
   \frac{T}{4} = \int_0^1 \frac{2^{\frac{1}{4}} du}{\sqrt{1 - u^4}} \implies T = 4 \cdot 2^{\frac{1}{4}} \int_0^1 \frac{du}{\sqrt{1 - u^4}}
$$

$$
\boxed{T = 2^{\frac{9}{4}} \int_0^1 \frac{du}{\sqrt{1 - u^4}} = \frac{2^{\frac{1}{4}} \sqrt{\pi} \, \Gamma(\frac{1}{4})}{\Gamma(\frac{3}{4})}}
$$

#### Key takeaway
First integrals reduce second-order non-linear autonomous conservative systems to exact quadrature integrals over phase trajectories.

---

```python
import numpy as np
from scipy.integrate import quad, solve_ivp
from scipy.special import gamma
from scipy.optimize import brentq
I, _ = quad(lambda u: 1 / np.sqrt(1 - u**4), 0, 1)
T_quad = 2**(9 / 4) * I
T_closed = 2**(1 / 4) * np.sqrt(np.pi) * gamma(1 / 4) / gamma(3 / 4)
sol = solve_ivp(lambda x, z: [z[1], -z[0]**3], [0, 20], [0.0, 1.0],
                dense_output=True, rtol=1e-12, atol=1e-14)
gfun = lambda x: sol.sol(x)[0]
grid = np.linspace(1e-6, 20, 40000)
v = gfun(grid)
ups = [brentq(gfun, grid[i], grid[i + 1]) for i in range(len(grid) - 1)
       if v[i] < 0 <= v[i + 1]]
print(f"quadrature   T = {T_quad:.10f}")
print(f"closed form  T = {T_closed:.10f}")
print(f"ODE period   T = {ups[0]:.10f}")
assert abs(T_quad - T_closed) < 1e-9 and abs(ups[0] - T_closed) < 1e-6
```

### Problem L3.3 — Abel's Identity & Reduction of Order
**Source**: Cambridge Mathematical Tripos, Part IA  
**Statement**  
Given that $y_1(x) = x$ is a solution to the differential equation:

$$
x^2(1 - x) y'' + x y' - y = 0 \quad (0 \lt x \lt 1)
$$

Find the second linearly independent solution $y_2(x)$ and write the general solution.

#### Intuition
Substitute $y_2(x) = v(x) y_1(x) = v(x) x$ to reduce the second-order ODE to a first-order ODE for $v'(x)$.

#### Solution
1. Compute derivatives of $y_2 = v x$:

$$
   y_2' = v' x + v, \quad y_2'' = v'' x + 2v'
$$

2. Substitute into ODE:

$$
   x^2(1 - x)(v'' x + 2v') + x(v' x + v) - v x = 0
$$

$$
   x^3(1 - x) v'' + \left[ 2x^2(1 - x) + x^2 \right] v' + (x v - x v) = 0
$$

$$
   x^3(1 - x) v'' + x^2 (3 - 2x) v' = 0
$$

3. Separate variables for $u(x) = v'(x)$:

$$
   \frac{v''}{v'} = -\frac{3 - 2x}{x(1 - x)} = -\left( \frac{3}{x} + \frac{1}{1 - x} \right)
$$

   Integrate:

$$
   \ln\lvert v' \rvert = -3 \ln x + \ln\lvert 1 - x \rvert \implies v'(x) = \frac{1 - x}{x^3} = x^{-3} - x^{-2}
$$

4. Integrate $v'(x)$ to find $v(x)$:

$$
   v(x) = \int (x^{-3} - x^{-2}) dx = -\frac{1}{2x^2} + \frac{1}{x} = \frac{2x - 1}{2x^2}
$$

5. Compute $y_2(x) = v(x) \cdot x$:

$$
   y_2(x) = x \left( \frac{2x - 1}{2x^2} \right) = \frac{2x - 1}{2x} = 1 - \frac{1}{2x}
$$

   Scaling by $2$ to clear the fraction gives the equivalent second solution $y_2(x) = 2 - \dfrac{1}{x}$, linearly independent from $y_1(x) = x$.

$$
\boxed{y(x) = C_1 x + C_2 \left(2 - \frac{1}{x}\right)}
$$

#### Key takeaway
Knowing a single non-trivial solution allows Reduction of Order to construct the complete fundamental solution space.

---

```python
import sympy as sp

x = sp.symbols("x", positive=True)
y1, y2 = x, 2 - 1 / x
residual = sp.simplify(x**2 * (1 - x) * sp.diff(y2, x, 2) + x * sp.diff(y2, x) - y2)
wronskian = sp.simplify(y1 * sp.diff(y2, x) - y2 * sp.diff(y1, x))
print(f"ODE residual for y2 = 2 - 1/x: {residual}")
print(f"Wronskian(y1, y2) = {wronskian}  (nonzero => linearly independent)")
assert residual == 0 and wronskian != 0
```

### Problem L3.4 — Global Existence vs. Finite-Time Blow-Up
**Source**: V.I. Arnold, *Ordinary Differential Equations* / Tripos Part IB  
**Statement**  
Consider the non-linear initial value problem:

$$
\frac{dx}{dt} = x^\alpha \quad (\alpha \gt 1), \quad x(0) = x_0 \gt 0
$$

1. Solve the ODE explicitly.
2. Find the exact finite blow-up time $T^\ast$ where $\lim_{t \to T^{\ast-}} x(t) = +\infty$.

#### Intuition
Non-linear terms with power $\alpha \gt 1$ accelerate growth faster than exponential models, causing trajectories to escape to infinity in finite time.

#### Solution
Separate variables:

$$
x^{-\alpha} \, dx = dt
$$

Integrate from $0$ to $t$:

$$
\int_{x_0}^{x(t)} s^{-\alpha} \, ds = \int_0^t d\tau
$$

$$
\left[ \frac{s^{1-\alpha}}{1 - \alpha} \right]_{x_0}^{x(t)} = t \implies \frac{x(t)^{1-\alpha} - x_0^{1-\alpha}}{1 - \alpha} = t
$$

Multiply by $(1 - \alpha)$:

$$
x(t)^{1-\alpha} - x_0^{1-\alpha} = (1 - \alpha) t = -(\alpha - 1) t
$$

$$
x(t)^{1-\alpha} = x_0^{1-\alpha} - (\alpha - 1) t
$$

Take the exponent $\frac{1}{1-\alpha} = -\frac{1}{\alpha - 1}$:

$$
x(t) = \frac{1}{\left( x_0^{-(\alpha-1)} - (\alpha - 1) t \right)^{\frac{1}{\alpha - 1}}}
$$

The denominator vanishes when $x_0^{-(\alpha-1)} - (\alpha - 1) t = 0$:

$$
T^\ast = \frac{x_0^{-(\alpha-1)}}{\alpha - 1} = \frac{1}{(\alpha - 1) x_0^{\alpha - 1}}
$$

$$
\boxed{T^\ast = \frac{1}{(\alpha - 1) x_0^{\alpha - 1}}, \quad \lim_{t \to T^{\ast-}} x(t) = +\infty}
$$

#### Key takeaway
Lipschitz continuity guarantees local existence, but super-linear growth ($\alpha \gt 1$) can break global existence via finite-time blow-up.

---

```python
import numpy as np
from scipy.integrate import solve_ivp
alpha, x0 = 2.0, 0.5
T_star = 1 / ((alpha - 1) * x0**(alpha - 1))
x = lambda t: (x0**(-(alpha - 1)) - (alpha - 1) * t)**(-1 / (alpha - 1))
print(f"T* (formula) = {T_star:.6f}")
for t in (0.5, 1.5, 1.9, 1.99, 1.999):
    print(f"  x({t:5.3f}) = {x(t):12.3f}")
sol = solve_ivp(lambda t, z: z**alpha, [0, T_star - 1e-6], [x0], rtol=1e-11, atol=1e-13)
print(f"numerical x(T*-1e-6) = {sol.y[0,-1]:.3e}  (formula {x(T_star-1e-6):.3e})")
assert abs(T_star - 2.0) < 1e-12
assert abs(sol.y[0, -1] / x(T_star - 1e-6) - 1) < 1e-4
```

### Problem L3.5 — Infinitude of Zeros of Bessel Function $J_0(x)$
**Source**: Polya & Szego, *Problems and Theorems in Analysis*  
**Statement**  
Bessel's function of order zero $y(x) = J_0(x)$ satisfies:

$$
x y'' + y' + x y = 0 \quad (x \gt 0)
$$

1. Transform the ODE into normal form $u'' + q(x) u = 0$ via substitution $u(x) = \sqrt{x} y(x)$.
2. Use Sturm's Comparison Theorem to prove that $J_0(x)$ has infinitely many positive zeros on $(0, \infty)$.

#### Intuition
Transforming to normal form compares the oscillation frequency of $u'' + q(x)u = 0$ to the constant-frequency harmonic oscillator $v'' + v = 0$.

#### Solution
1. **Transform to Normal Form**:
   Given $y'' + \frac{1}{x} y' + y = 0$.
   Substitute $y(x) = x^{-\frac{1}{2}} u(x)$:

$$
   y' = -\frac{1}{2} x^{-\frac{3}{2}} u + x^{-\frac{1}{2}} u'
$$

$$
   y'' = \frac{3}{4} x^{-\frac{5}{2}} u - x^{-\frac{3}{2}} u' + x^{-\frac{1}{2}} u''
$$

   Substitute $y, y', y''$ into $y'' + \frac{1}{x} y' + y = 0$:

$$
   \left( \frac{3}{4} x^{-\frac{5}{2}} u - x^{-\frac{3}{2}} u' + x^{-\frac{1}{2}} u'' \right) + \frac{1}{x}\left( -\frac{1}{2} x^{-\frac{3}{2}} u + x^{-\frac{1}{2}} u' \right) + x^{-\frac{1}{2}} u = 0
$$

   Multiply by $x^{\frac{1}{2}}$:

$$
   u'' + \left( 1 + \frac{1}{4x^2} \right) u = 0
$$

   Here $q(x) = 1 + \frac{1}{4x^2}$.

2. **Sturm Comparison Application**:
   Notice that $q(x) = 1 + \frac{1}{4x^2} \gt 1$ for all $x \gt 0$.
   Compare $u'' + q(x) u = 0$ with the standard oscillator equation $v'' + v = 0$, whose solutions are $v(x) = \sin(x - x_0)$ with zeros separated by $\pi$.
   By **Sturm's Comparison Theorem**, since $q(x) \gt 1$, $u(x)$ must oscillate *faster* than $v(x)$, meaning $u(x)$ has at least one zero between any two consecutive zeros of $\sin x$.
   Since $\sin x$ has infinitely many zeros on $(0, \infty)$, $u(x)$ (and therefore $J_0(x) = x^{-\frac{1}{2}} u(x)$) must also possess infinitely many zeros on $(0, \infty)$.

$$
\boxed{u'' + \left(1 + \frac{1}{4x^2}\right)u = 0 \implies J_0(x) \text{ has infinitely many zeros on } (0, \infty).}
$$

#### Key takeaway
Sturm Comparison theorems bound non-linear and variable-coefficient zero distributions using simple harmonic reference systems.

---

```python
import numpy as np
import sympy as sp
from scipy.special import jn_zeros
x, u = sp.symbols("x u", positive=True)
uf = sp.Function("u")
y = uf(x) / sp.sqrt(x)
lhs = sp.simplify(sp.expand(sp.sqrt(x) * (sp.diff(y, x, 2) + sp.diff(y, x) / x + y)))
print("normal form:", sp.simplify(lhs))
zeros = jn_zeros(0, 8)
print("first eight zeros of J0:", np.round(zeros, 5))
print("consecutive gaps        :", np.round(np.diff(zeros), 5), " (all < pi =", round(np.pi, 5), ")")
assert np.all(np.diff(zeros) < np.pi)
```

### Problem L3.6 — Non-Linear Matrix Riccati Differential Equation
**Source**: Kaczor & Nowak / Tripos Analysis  
**Statement**  
Solve the matrix differential equation for $X(t) \in \mathbb{R}^{n \times n}$:

$$
\frac{dX}{dt} = -X(t) A X(t), \quad X(0) = X_0
$$

where $A$ and $X_0$ are invertible $n \times n$ matrices.

#### Intuition
For scalar $\dot{x} = -a x^2$, we integrate $\frac{dx}{x^2} = -a dt \implies -\frac{1}{x} = -at + C \implies x(t) = (x_0^{-1} + a t)^{-1}$. For matrix equations, evaluate derivative of inverse matrix $Y(t) = X(t)^{-1}$.

#### Solution
Let $Y(t) = X(t)^{-1}$, so $X(t) Y(t) = I$.
Differentiate $X(t) Y(t) = I$ with respect to $t$:

$$
\frac{dX}{dt} Y(t) + X(t) \frac{dY}{dt} = 0 \implies \frac{dY}{dt} = -X(t)^{-1} \frac{dX}{dt} Y(t)
$$

Substitute $\frac{dX}{dt} = -X(t) A X(t)$:

$$
\frac{dY}{dt} = -X(t)^{-1} \left( -X(t) A X(t) \right) X(t)^{-1} = A
$$

Integrate the linear matrix ODE $\frac{dY}{dt} = A$:

$$
Y(t) = Y(0) + A t = X_0^{-1} + A t
$$

Invert $Y(t)$ to recover $X(t)$:

$$
X(t) = Y(t)^{-1} = (X_0^{-1} + A t)^{-1}
$$

$$
\boxed{X(t) = \left( X_0^{-1} + A t \right)^{-1}}
$$

#### Key takeaway
Matrix identity derivatives $\frac{d}{dt}(X^{-1}) = -X^{-1} \dot{X} X^{-1}$ transform non-linear matrix Riccati equations into integrable linear forms.

---

```python
import numpy as np
from scipy.integrate import solve_ivp
rng = np.random.default_rng(0)
n = 3
Q = np.linalg.qr(rng.standard_normal((n, n)))[0]
A = Q @ np.diag([0.5, 1.0, 2.0]) @ Q.T     # SPD, so X0^-1 + A t stays invertible
X0 = np.eye(n)
X = lambda t: np.linalg.inv(np.linalg.inv(X0) + A * t)
rhs = lambda t, v: (-(v.reshape(n, n)) @ A @ (v.reshape(n, n))).ravel()
sol = solve_ivp(rhs, [0, 2.0], X0.ravel(), rtol=1e-11, atol=1e-13, t_eval=[2.0])
num = sol.y[:, -1].reshape(n, n)
print("closed form X(2):\n", np.round(X(2.0), 6))
print("max |closed form - numerical| =", np.max(np.abs(num - X(2.0))))
assert np.max(np.abs(num - X(2.0))) < 1e-8
```

### Problem L3.7 — Constructing Lyapunov Functions for Non-Linear Systems
**Source**: S. Strogatz / Tripos Part II  
**Statement**  
Consider the non-linear 2D system:

$$
\begin{cases}
\dot{x} = -x + y^2 \\[4pt]
\dot{y} = -y - x y
\end{cases}
$$

Prove that the origin $(0,0)$ is globally asymptotically stable by constructing a quadratic Lyapunov function $V(x,y) = \frac{1}{2} x^2 + \frac{1}{2} y^2$.

#### Intuition
A strict decrease in distance from the origin $V = \frac{1}{2}(x^2+y^2)$ along trajectories guarantees trajectory convergence to $(0,0)$.

#### Solution
1. Fixed point check: $(0,0)$ gives $\dot{x} = 0, \dot{y} = 0$.
2. Define candidate Lyapunov function $V(x,y) = \frac{1}{2} x^2 + \frac{1}{2} y^2$.
   - $V(0,0) = 0$.
   - $V(x,y) \gt 0$ for all $(x,y) \ne (0,0)$ (Positive definite).

3. Compute orbital derivative $\dot{V}$:

$$
   \begin{aligned}
   \dot{V}(x,y) &= x \dot{x} + y \dot{y} \\
   &= x(-x + y^2) + y(-y - x y) \\
   &= -x^2 + x y^2 - y^2 - x y^2 \\
   &= -x^2 - y^2 = -(x^2 + y^2)
   \end{aligned}
$$

   Notice that the cross terms $x y^2$ cancel identically!
   Thus $\dot{V}(x,y) = -2 V(x,y) \lt 0$ for all $(x,y) \ne (0,0)$ (Strictly negative definite).

4. By Lyapunov's Direct Theorem, $(0,0)$ is **Globally Asymptotically Stable**.

$$
\boxed{\dot{V}(x,y) = -(x^2 + y^2) \lt 0 \implies (0,0) \text{ is Globally Asymptotically Stable.}}
$$

#### Key takeaway
Lyapunov function candidate choices that cancel non-linear interaction cross-terms provide global stability proofs.

---

```python
import sympy as sp
x, y = sp.symbols("x y")
V = (x**2 + y**2) / 2
F = sp.Matrix([-x + y**2, -y - x * y])
Vdot = sp.simplify(sp.diff(V, x) * F[0] + sp.diff(V, y) * F[1])
print("Vdot =", sp.factor(Vdot))
assert sp.simplify(Vdot + (x**2 + y**2)) == 0
```

### Problem L3.8 — Homogeneous Riccati ODE Substitution
**Source**: Homogeneous ODE via substitution, cf. Boyce & DiPrima, *Elementary Differential Equations*, Ch. 2  
**Statement**  
Solve the homogeneous first-order ODE:

$$
x \frac{dy}{dx} = y + \sqrt{x^2 + y^2} \quad (x \gt 0)
$$

#### Intuition
The equation is homogeneous of degree 1 since $f(tx, ty) = f(x, y)$. Substitute $y = v x \implies \frac{dy}{dx} = v + x \frac{dv}{dx}$.

#### Solution
Divide by $x \gt 0$:

$$
\frac{dy}{dx} = \frac{y}{x} + \sqrt{1 + \left(\frac{y}{x}\right)^2}
$$

Substitute $y = v x \implies y' = v + x v'$:

$$
v + x \frac{dv}{dx} = v + \sqrt{1 + v^2} \implies x \frac{dv}{dx} = \sqrt{1 + v^2}
$$

Separate variables:

$$
\int \frac{dv}{\sqrt{1 + v^2}} = \int \frac{dx}{x}
$$

Using the hyperbolic integration identity $\int \frac{dv}{\sqrt{1+v^2}} = \operatorname{arsinh}(v)$:

$$
\operatorname{arsinh}(v) = \ln x + C \implies v = \sinh(\ln x + C)
$$

Using $\sinh(u) = \frac{e^u - e^{-u}}{2}$, let $k = e^C \gt 0$:

$$
v = \frac{k x - \frac{1}{k x}}{2}
$$

Substitute back $v = \frac{y}{x}$:

$$
\frac{y}{x} = \frac{k^2 x^2 - 1}{2 k x} \implies y(x) = \frac{k^2 x^2 - 1}{2 k}
$$

Defining $C_1 = \frac{k}{2}$:

$$
\boxed{y(x) = C_1 x^2 - \frac{1}{4 C_1}}
$$

#### Key takeaway
Homogeneous rational ODEs simplify to separable forms under state scale ratio substitutions $v = \frac{y}{x}$.

---

```python
import numpy as np
import sympy as sp
x, C1 = sp.symbols("x C1", positive=True)
y = C1 * x**2 - 1 / (4 * C1)
# the radicand is a perfect square; sympy needs to be shown that explicitly
rad = sp.simplify(sp.expand(x**2 + y**2) - sp.expand((C1 * x**2 + 1 / (4 * C1))**2))
print("x^2 + y^2 - (C1 x^2 + 1/(4 C1))^2 =", rad, " (so the root is C1 x^2 + 1/(4 C1) > 0)")
assert rad == 0
res = sp.simplify(x * sp.diff(y, x) - (y + (C1 * x**2 + 1 / (4 * C1))))
print("residual of x y' = y + sqrt(x^2 + y^2):", res)
assert res == 0
# numeric spot check straight from the original equation
c, xv = 0.7, 1.9
yv, ypv = c * xv**2 - 1 / (4 * c), 2 * c * xv
print("numeric: x y' =", round(xv * ypv, 12),
      " y + sqrt(x^2+y^2) =", round(yv + np.hypot(xv, yv), 12))
assert abs(xv * ypv - (yv + np.hypot(xv, yv))) < 1e-12
```

### Problem L3.9 — Sturm-Liouville Boundary Value Problem Eigenvalues
**Source**: Cambridge Mathematical Tripos, Part IA  
**Statement**  
Find the transcendental eigenvalue equation for $\lambda \gt 0$ for the Sturm-Liouville BVP:

$$
y'' + \lambda y = 0, \quad y(0) = 0, \quad y'(L) + h y(L) = 0 \quad (h \gt 0, L \gt 0)
$$

#### Intuition
Solve the general spatial wave equation $y(x) = A \cos(\sqrt{\lambda} x) + B \sin(\sqrt{\lambda} x)$ and apply mixed Robin boundary conditions.

#### Solution
Let $\omega = \sqrt{\lambda} \gt 0$. The general solution to $y'' + \omega^2 y = 0$ is:

$$
y(x) = A \cos(\omega x) + B \sin(\omega x)
$$

Apply $y(0) = 0$:

$$
y(0) = A(1) + B(0) = 0 \implies A = 0 \implies y(x) = B \sin(\omega x)
$$

Compute derivative:

$$
y'(x) = B \omega \cos(\omega x)
$$

Apply Robin boundary condition $y'(L) + h y(L) = 0$:

$$
B \omega \cos(\omega L) + h B \sin(\omega L) = 0
$$

For non-trivial solutions ($B \ne 0$):

$$
\omega \cos(\omega L) + h \sin(\omega L) = 0 \implies h \sin(\omega L) = -\omega \cos(\omega L)
$$

Divide by $h \cos(\omega L)$:

$$
\tan(\omega L) = -\frac{\omega}{h} \implies \tan(\sqrt{\lambda} L) = -\frac{\sqrt{\lambda}}{h}
$$

$$
\boxed{\tan(\sqrt{\lambda} L) = -\frac{\sqrt{\lambda}}{h}}
$$

#### Key takeaway
Mixed spatial boundary conditions generate transcendental characteristic equations whose roots form discrete orthogonal eigen-frequencies.

---

```python
import numpy as np
from scipy.optimize import brentq
L, h = 1.0, 2.0
F = lambda w: w * np.cos(w * L) + h * np.sin(w * L)
roots = []
grid = np.linspace(1e-6, 20, 200000)
v = F(grid)
for i in range(len(grid) - 1):
    if v[i] * v[i + 1] < 0:
        roots.append(brentq(F, grid[i], grid[i + 1]))
print("first five omega =", np.round(roots[:5], 6))
print("lambda = omega^2 =", np.round(np.array(roots[:5])**2, 6))
for w in roots[:5]:
    assert abs(np.tan(w * L) + w / h) < 1e-6
print("each root satisfies tan(omega L) = -omega/h")
```

### Problem L3.10 — Hamiltonian Phase Flow & Liouville's Theorem
**Source**: V.I. Arnold, *Mathematical Methods of Classical Mechanics*  
**Statement**  
Consider a 1D mechanical system with Hamiltonian $H(q, p) = \frac{p^2}{2m} + V(q)$ governing phase trajectories $(q(t), p(t))$:

$$
\dot{q} = \frac{\partial H}{\partial p} = \frac{p}{m}, \quad \dot{p} = -\frac{\partial H}{\partial q} = -V'(q)
$$

Prove Liouville's Theorem: the phase space velocity vector field $\mathbf{f}(q, p) = (\dot{q}, \dot{p})$ is incompressible, i.e., $\nabla \cdot \mathbf{f} = 0$.

#### Intuition
Incompressibility $\nabla \cdot \mathbf{f} = 0$ means that any region of initial conditions in phase space preserves its total area/volume under time evolution.

#### Solution
Compute the divergence of vector field $\mathbf{f}(q, p) = \left( \frac{\partial H}{\partial p}, -\frac{\partial H}{\partial q} \right)$ in $(q, p)$ space:

$$
\nabla \cdot \mathbf{f} = \frac{\partial}{\partial q} (\dot{q}) + \frac{\partial}{\partial p} (\dot{p})
$$

Substitute Hamilton's equations:

$$
\nabla \cdot \mathbf{f} = \frac{\partial}{\partial q} \left( \frac{\partial H}{\partial p} \right) + \frac{\partial}{\partial p} \left( -\frac{\partial H}{\partial q} \right)
$$

$$
\nabla \cdot \mathbf{f} = \frac{\partial^2 H}{\partial q \partial p} - \frac{\partial^2 H}{\partial p \partial q}
$$

By Clairaut's theorem on equality of mixed partial derivatives (assuming $H \in C^2$):

$$
\frac{\partial^2 H}{\partial q \partial p} = \frac{\partial^2 H}{\partial p \partial q}
$$

Thus:

$$
\nabla \cdot \mathbf{f} = 0
$$

$$
\boxed{\nabla \cdot (\dot{q}, \dot{p}) = \frac{\partial^2 H}{\partial q \partial p} - \frac{\partial^2 H}{\partial p \partial q} = 0 \implies \text{Phase space area is strictly preserved (Liouville's Theorem).}}
$$

#### Key takeaway
Hamiltonian dynamics preserve phase space volume, establishing the foundational link between classical mechanics, symplectic geometry, and statistical physics.

```python
import sympy as sp
q, p = sp.symbols("q p")
H = p**2 / 2 + q**4 / 4 + sp.sin(q) * p          # an arbitrary smooth Hamiltonian
qdot, pdot = sp.diff(H, p), -sp.diff(H, q)
div = sp.simplify(sp.diff(qdot, q) + sp.diff(pdot, p))
print("div(qdot, pdot) =", div)
assert div == 0
```
