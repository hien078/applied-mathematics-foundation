### Problem 0.1: Choice of Slicing Orientation for Planar Area
**Attribution:** [MIT 18.01 Concept Check]

#### Problem Statement
Consider the region $\mathcal{R}$ bounded by the curves $y = x^2$ and $x = y^2$. Compare the differential area element $dA$ obtained by slicing vertically versus horizontally. Derive the integral setup for both orientations and confirm that both yield the same area value.

#### First-Principles Intuition
Area is independent of coordinate orientation. Vertical slicing integrates the top function minus bottom function with respect to $x$, while horizontal slicing integrates the right function minus left function with respect to $y$.

#### Step-by-Step Solution
1. **Find Intersection Points:**
   Set $y = x^2$ and $x = y^2 \implies x = (x^2)^2 = x^4 \implies x(x^3 - 1) = 0$.
   The real intersection points are $(0, 0)$ and $(1, 1)$.

2. **Vertical Slicing (Integration w.r.t. $x$):**
   For $x \in [0, 1]$, the upper curve is $y_{\text{top}} = \sqrt{x}$ and the lower curve is $y_{\text{bottom}} = x^2$.
   The differential element is $dA = (\sqrt{x} - x^2)\,dx$.

$$
   A = \int_0^1 \left( x^{\frac{1}{2}} - x^2 \right) dx = \left[ \frac{2}{3} x^{\frac{3}{2}} - \frac{1}{3} x^3 \right]_0^1 = \frac{2}{3} - \frac{1}{3} = \frac{1}{3}
$$

3. **Horizontal Slicing (Integration w.r.t. $y$):**
   For $y \in [0, 1]$, the right curve is $x_{\text{right}} = \sqrt{y}$ and the left curve is $x_{\text{left}} = y^2$.
   The differential element is $dA = (\sqrt{y} - y^2)\,dy$.

$$
   A = \int_0^1 \left( y^{\frac{1}{2}} - y^2 \right) dy = \left[ \frac{2}{3} y^{\frac{3}{2}} - \frac{1}{3} y^3 \right]_0^1 = \frac{1}{3}
$$

Both orientations yield identical area.

$$
\boxed{A = \frac{1}{3}}
$$

#### Key Insight / Takeaway
Slicing direction can be chosen based on computational convenience; horizontal slicing avoids solving for inverse functions when vertical expressions are algebraically complicated.

### Problem 0.2: Disk/Washer vs. Shell Method Geometry
**Attribution:** [Stewart Ch 6 Concept Check]

#### Problem Statement
A region $\mathcal{R}$ in the first quadrant is bounded by $y = f(x)$, $x = a$, $x = b$, and $y = 0$. Explain geometrically why revolving $\mathcal{R}$ about the $y$-axis using vertical strips leads to the Cylindrical Shell method, whereas using horizontal strips leads to the Washer method.

#### First-Principles Intuition
The geometry of the differential volume element $dV$ depends on whether the slice is parallel or perpendicular to the axis of revolution.

#### Step-by-Step Solution
1. **Vertical Strip (Parallel to Revolution Axis):**
   A vertical strip at $x$ of height $f(x)$ and thickness $dx$ is parallel to the $y$-axis (the rotation axis).
   Revolving a line segment parallel to an axis at radius $r = x$ sweeps out a cylinder of radius $x$, height $f(x)$, and wall thickness $dx$.
   Unrolling this shell gives volume $dV = 2\pi x f(x)\,dx$.

2. **Horizontal Strip (Perpendicular to Revolution Axis):**
   A horizontal strip at $y$ of width $[b - g(y)]$ and thickness $dy$ is perpendicular to the $y$-axis.
   Revolving a line segment perpendicular to the axis sweeps out a circular annulus (washer) of outer radius $R(y) = b$ and inner radius $r(y) = g(y)$.
   The differential volume element is $dV = \pi [b^2 - (g(y))^2]\,dy$.

$$
\boxed{\text{Parallel slices} \implies \text{Shells } (dV = 2\pi r h\,dx); \quad \text{Perpendicular slices} \implies \text{Washers } (dV = \pi (R^2 - r^2)\,dy)}
$$

#### Key Insight / Takeaway
Always inspect strip orientation relative to the axis of rotation: parallel = Shell method, perpendicular = Disk/Washer method.

### Problem 0.3: Arc Length Differential and Hypotenuse Slicing
**Attribution:** [Spivak Ch 14 Concept Check]

#### Problem Statement
Why is the arc length integral of $y = f(x)$ written as $\int_a^b \sqrt{1 + (f'(x))^2}\,dx$ instead of simply $\int_a^b (1 + f'(x))\,dx$ or $\int_a^b f'(x)\,dx$?

#### First-Principles Intuition
Arc length measures spatial distance along a curve, which relies on the Euclidean metric (Pythagorean theorem) $\Delta s^2 = \Delta x^2 + \Delta y^2$, not on the sum of axial projections.

#### Step-by-Step Solution
1. **Euclidean Metric Decomposition:**
   A small displacement along a curve involves horizontal change $\Delta x$ and vertical change $\Delta y$.
   The straight-line distance $\Delta s$ between $(x, y)$ and $(x + \Delta x, y + \Delta y)$ is:

$$
   \Delta s = \sqrt{(\Delta x)^2 + (\Delta y)^2}
$$

2. **Factoring out $\Delta x$:**

$$
   \Delta s = \sqrt{1 + \left(\frac{\Delta y}{\Delta x}\right)^2}\,\Delta x
$$

3. **Limit Transition:**
   As $\Delta x \to 0$, $\frac{\Delta y}{\Delta x} \to f'(x)$, so $ds = \sqrt{1 + (f'(x))^2}\,dx$.
   Integrating $\int f'(x)\,dx = f(b) - f(a)$ measures net vertical displacement, not total distance traversed along the curve path.

$$
\boxed{ds = \sqrt{dx^2 + dy^2} = \sqrt{1 + (f'(x))^2}\,dx}
$$

#### Key Insight / Takeaway
Curve length requires compounding orthogonal differentials quadratically via the Euclidean distance formula.

### Problem 0.4: Surface Area Ribbon Slant Length
**Attribution:** [Apostol Vol 1 Concept Check]

#### Problem Statement
When computing the surface area of revolution of $y = f(x)$ about the $x$-axis, why must we use $d\Sigma = 2\pi f(x) \, ds$ rather than $d\Sigma = 2\pi f(x) \, dx$?

#### First-Principles Intuition
Revolving a curve segment creates a cylindrical frustum whose lateral slant height is the arc length element $ds$, not the flat domain width $dx$.

#### Step-by-Step Solution
1. **Conical Frustum Lateral Area:**
   A conical frustum with slant height $L_{\text{slant}}$ and average radius $r_{\text{avg}}$ has lateral surface area $A = 2\pi r_{\text{avg}} L_{\text{slant}}$.

2. **Infinitesimal Limit:**
   For a differential ribbon swept out by revolving $y = f(x)$, $r_{\text{avg}} = f(x)$ and $L_{\text{slant}} = ds = \sqrt{1 + (f'(x))^2}\,dx$.

3. **Error of $dx$ Substitution:**
   Using $dx$ instead of $ds$ assumes the curve is horizontal ($f'(x) = 0$). For steep curves ($\lvert f'(x) \rvert \gg 0$), $dx$ severely underestimates the physical surface area because $ds = \sqrt{1 + (f'(x))^2}\,dx \gt dx$.

$$
\boxed{d\Sigma = 2\pi f(x) \, ds = 2\pi f(x) \sqrt{1 + (f'(x))^2}\,dx}
$$

#### Key Insight / Takeaway
Surface area elements depend on the actual boundary arc length element $ds$, accounting for surface inclination.

### Problem 0.5: Centroid Location vs. Midpoint
**Attribution:** [Stewart Ch 8 Concept Check]

#### Problem Statement
For the region under $y = x^2$ on $[0, 1]$, determine if the $x$-coordinate of the centroid $\bar{x}$ is equal to, less than, or greater than the interval midpoint $x = 0.5$. Explain physically.

#### First-Principles Intuition
Since $y = x^2$ increases monotonically on $[0, 1]$, more area (mass) is concentrated toward the right end $x = 1$. Therefore, the center of mass $\bar{x}$ must be pulled to the right of $x = 0.5$.

#### Step-by-Step Solution
1. **Calculate Area $A$:**

$$
   A = \int_0^1 x^2\,dx = \left[ \frac{x^3}{3} \right]_0^1 = \frac{1}{3}
$$

2. **Calculate Moment $M_y$:**

$$
   M_y = \int_0^1 x (x^2)\,dx = \int_0^1 x^3\,dx = \left[ \frac{x^4}{4} \right]_0^1 = \frac{1}{4}
$$

3. **Calculate Centroid $\bar{x}$:**

$$
   \bar{x} = \frac{M_y}{A} = \frac{\frac{1}{4}}{\frac{1}{3}} = \frac{3}{4} = 0.75
$$

Since $0.75 \gt 0.5$, $\bar{x}$ lies to the right of the geometric midpoint.

$$
\boxed{\bar{x} = \frac{3}{4} \gt 0.5}
$$

#### Key Insight / Takeaway
Centroids reflect mass-weighted spatial distributions; region asymmetry shifts the centroid toward heavier density/area concentrations.

### Problem 0.6: Hydrostatic Pressure Depth Dependence
**Attribution:** [MIT 18.01 Physics Concept]

#### Problem Statement
Explain why the total hydrostatic force on a vertical rectangular submerged gate of height $H$ and width $W$ (with top edge at fluid surface) is $F = \frac{1}{2} \rho g W H^2$, rather than $P_{\text{bottom}} \times (W H) = \rho g H \cdot (W H) = \rho g W H^2$.

#### First-Principles Intuition
Hydrostatic pressure increases linearly with depth $y$ from $P(0) = 0$ at the surface to $P(H) = \rho g H$ at the bottom. The average pressure over the depth profile is $\frac{1}{2} P(H)$.

#### Step-by-Step Solution
1. **Pressure Profile:**
   At depth $y \in [0, H]$, pressure is $P(y) = \rho g y$.

2. **Differential Force Element:**

$$
   dF = P(y) \, dA = (\rho g y) (W \, dy)
$$

3. **Depth Integration:**

$$
   F = \rho g W \int_0^H y \, dy = \rho g W \left[ \frac{y^2}{2} \right]_0^H = \frac{1}{2} \rho g W H^2
$$

Multiplying bottom pressure by total area double-counts the force because pressure at shallower depths is strictly less than at the bottom.

$$
\boxed{F = \frac{1}{2} \rho g W H^2}
$$

#### Key Insight / Takeaway
Hydrostatic force requires integrating continuously varying pressure across depth layers, equivalent to multiplying average pressure by area.

### Problem 0.7: Probability Density Value vs Probability
**Attribution:** [Goodfellow Deep Learning Ch 3]

#### Problem Statement
Let $X$ be a continuous random variable with PDF $f_X(x) = 2x$ for $x \in [0, 1]$ and $0$ elsewhere.
1. Evaluate $f_X(0.9)$. Does a density value $\gt 1$ violate probability axioms?
2. What is $P(X = 0.5)$?

#### First-Principles Intuition
For continuous random variables, $f_X(x)$ represents probability per unit length (density), not a discrete probability. Probabilities are areas under $f_X(x)$, so single point probabilities are zero.

#### Step-by-Step Solution
1. **Density Evaluation:**

$$
   f_X(0.9) = 2(0.9) = 1.8
$$

   This does **not** violate probability axioms because $f_X(x)$ is a density. The total area is $\int_0^1 2x\,dx = [x^2]_0^1 = 1$, satisfying probability normalization.

2. **Single Point Probability:**

$$
   P(X = 0.5) = \int_{0.5}^{0.5} 2x\,dx = 0
$$

$$
\boxed{f_X(0.9) = 1.8 \quad (\text{valid density}); \quad P(X = 0.5) = 0}
$$

#### Key Insight / Takeaway
Continuous densities can exceed 1; only integrated probabilities $\int_a^b f_X(x)\,dx$ must lie in $[0, 1]$.

### Problem 0.8: Applicability of Pappus's Centroid Theorems
**Attribution:** [Apostol Vol 1 Ch 14]

#### Problem Statement
Can Pappus's Second Theorem ($V = 2\pi \bar{r} A$) be used to calculate the volume of a sphere of radius $R$ generated by revolving a semicircle of radius $R$ about its diameter? Verify.

#### First-Principles Intuition
Yes. A semicircular region of area $A = \frac{\pi R^2}{2}$ revolved about its bounding diameter generates a solid sphere of radius $R$. The centroid of the semicircle lies at distance $\bar{r} = \frac{4R}{3\pi}$ from the diameter axis.

#### Step-by-Step Solution
1. **Area of Semicircle:**

$$
   A = \frac{1}{2} \pi R^2
$$

2. **Centroid Distance $\bar{r}$:**
   For semicircle $y = \sqrt{R^2 - x^2}$ over $x \in [-R, R]$:

$$
   M_x = \frac{1}{2} \int_{-R}^R (R^2 - x^2)\,dx = \frac{1}{2} \left[ R^2 x - \frac{x^3}{3} \right]_{-R}^R = \frac{2}{3} R^3
$$

$$
   \bar{r} = \bar{y} = \frac{M_x}{A} = \frac{\frac{2}{3} R^3}{\frac{1}{2} \pi R^2} = \frac{4R}{3\pi}
$$

3. **Apply Pappus's Theorem:**

$$
   V = 2\pi \bar{r} A = 2\pi \left( \frac{4R}{3\pi} \right) \left( \frac{\pi R^2}{2} \right) = \frac{4}{3} \pi R^3
$$

This matches the standard volume formula of a sphere.

$$
\boxed{V = \frac{4}{3} \pi R^3}
$$

#### Key Insight / Takeaway
Pappus's theorem converts geometric volumes into centroid-area products when the rotation axis does not cross the region interior.

### Problem 1.1: Area Between Parabola and Line
**Attribution:** [Stewart Ch 6.1]

#### Problem Statement
Find the area of the region $\mathcal{R}$ bounded by the parabola $y = 2 - x^2$ and the line $y = -x$.

#### First-Principles Intuition
Determine intersection limits, identify the upper function $f(x) = 2 - x^2$ and lower function $g(x) = -x$, and integrate their difference $dA = (f(x) - g(x))\,dx$.

#### Step-by-Step Solution
1. **Intersection Points:**

$$
   2 - x^2 = -x \implies x^2 - x - 2 = 0 \implies (x - 2)(x + 1) = 0 \implies x = -1, \, x = 2
$$

2. **Setup Area Integral:**
   On $[-1, 2]$, $2 - x^2 \ge -x$.

$$
   \begin{aligned}
   A &= \int_{-1}^2 \left[ (2 - x^2) - (-x) \right] dx = \int_{-1}^2 (2 + x - x^2)\,dx \\
   &= \left[ 2x + \frac{x^2}{2} - \frac{x^3}{3} \right]_{-1}^2 \\
   &= \left( 4 + 2 - \frac{8}{3} \right) - \left( -2 + \frac{1}{2} + \frac{1}{3} \right) \\
   &= \frac{10}{3} - \left( -\frac{7}{6} \right) = \frac{20}{6} + \frac{7}{6} = \frac{27}{6} = \frac{9}{2}
   \end{aligned}
$$

$$
\boxed{A = \frac{9}{2}}
$$

#### Key Insight / Takeaway
Correctly solving boundary intersections establishes integration limits for planar region mensuration.

### Problem 1.2: Area Enclosed by Polar Cardioid
**Attribution:** [Apostol Vol 1 Ch 14]

#### Problem Statement
Calculate the total area enclosed by the cardioid $r = a(1 + \cos\theta)$ for $a \gt 0$.

#### First-Principles Intuition
A polar sector of angle $d\theta$ has area $dA = \frac{1}{2} r^2 \, d\theta$. Integrating over $\theta \in [0, 2\pi]$ sweeps out the full cardioid surface.

#### Step-by-Step Solution
1. **Polar Area Formula:**

$$
   A = \frac{1}{2} \int_0^{2\pi} r^2 \, d\theta = \frac{1}{2} a^2 \int_0^{2\pi} (1 + \cos\theta)^2 \, d\theta
$$

2. **Expand Integrand:**

$$
   (1 + \cos\theta)^2 = 1 + 2\cos\theta + \cos^2\theta = 1 + 2\cos\theta + \frac{1 + \cos(2\theta)}{2} = \frac{3}{2} + 2\cos\theta + \frac{1}{2}\cos(2\theta)
$$

3. **Integrate:**

$$
   \begin{aligned}
   A &= \frac{1}{2} a^2 \int_0^{2\pi} \left( \frac{3}{2} + 2\cos\theta + \frac{1}{2}\cos(2\theta) \right) d\theta \\
   &= \frac{1}{2} a^2 \left[ \frac{3}{2} \theta + 2\sin\theta + \frac{1}{4}\sin(2\theta) \right]_0^{2\pi} \\
   &= \frac{1}{2} a^2 \left( 3\pi + 0 + 0 \right) = \frac{3}{2} \pi a^2
   \end{aligned}
$$

$$
\boxed{A = \frac{3}{2} \pi a^2}
$$

#### Key Insight / Takeaway
Using double-angle trigonometric identities simplifies squared polar radius integrands into easily integrable linear terms.

### Problem 1.3: Volume of Sphere via Disk Method
**Attribution:** [Stewart Ch 6.2]

#### Problem Statement
Derive the volume of a sphere of radius $R$ by revolving the circular region bounded by $x^2 + y^2 = R^2$ about the $x$-axis using the Disk Method.

#### First-Principles Intuition
Perpendicular slicing at position $x \in [-R, R]$ yields a solid circular disk of radius $y = \sqrt{R^2 - x^2}$ and thickness $dx$.

#### Step-by-Step Solution
1. **Differential Volume Element:**

$$
   dV = \pi y^2 \, dx = \pi (R^2 - x^2)\,dx
$$

2. **Integrate over $[-R, R]$:**
   By symmetry:

$$
   V = 2\pi \int_0^R (R^2 - x^2)\,dx = 2\pi \left[ R^2 x - \frac{x^3}{3} \right]_0^R = 2\pi \left( R^3 - \frac{R^3}{3} \right) = \frac{4}{3} \pi R^3
$$

$$
\boxed{V = \frac{4}{3} \pi R^3}
$$

#### Key Insight / Takeaway
Symmetry simplifies integration domains from $[-R, R]$ to $2 \times [0, R]$.

### Problem 1.4: Volume of Solid via Cylindrical Shell Method
**Attribution:** [Stewart Ch 6.3]

#### Problem Statement
Find the volume of the solid generated by revolving the region bounded by $y = 3x - x^2$ and $y = 0$ about the $y$-axis using the Cylindrical Shell method.

#### First-Principles Intuition
Slicing parallel to the $y$-axis at radius $x$ yields thin cylindrical shells of height $h(x) = 3x - x^2$ and thickness $dx$.

#### Step-by-Step Solution
1. **Find Region Endpoints:**

$$
   3x - x^2 = 0 \implies x(3 - x) = 0 \implies x = 0, \, x = 3
$$

2. **Differential Shell Volume Element:**

$$
   dV = 2\pi \, r(x) \, h(x) \, dx = 2\pi x (3x - x^2)\,dx = 2\pi (3x^2 - x^3)\,dx
$$

3. **Integrate over $[0, 3]$:**

$$
   \begin{aligned}
   V &= 2\pi \int_0^3 (3x^2 - x^3)\,dx = 2\pi \left[ x^3 - \frac{x^4}{4} \right]_0^3 \\
   &= 2\pi \left( 27 - \frac{81}{4} \right) = 2\pi \left( \frac{108 - 81}{4} \right) = 2\pi \left( \frac{27}{4} \right) = \frac{27\pi}{2}
   \end{aligned}
$$

$$
\boxed{V = \frac{27\pi}{2}}
$$

#### Key Insight / Takeaway
The shell method avoids taking inverse functions when region boundaries are given naturally as $y = f(x)$.

### Problem 1.5: Arc Length of Catenary Curve
**Attribution:** [Demidovich No. 1640]

#### Problem Statement
Calculate the exact arc length of the catenary $y = a \cosh(\frac{x}{a})$ from $x = 0$ to $x = b$ ($a \gt 0, b \gt 0$).

#### First-Principles Intuition
Hyperbolic functions satisfy $\cosh^2 u - \sinh^2 u = 1$, making $1 + (y')^2$ a perfect square under the radical.

#### Step-by-Step Solution
1. **Compute Derivative:**

$$
   y' = \frac{d}{dx} \left( a \cosh\left(\frac{x}{a}\right) \right) = \sinh\left(\frac{x}{a}\right)
$$

2. **Arc Length Element $ds$:**

$$
   \sqrt{1 + (y')^2} = \sqrt{1 + \sinh^2\left(\frac{x}{a}\right)} = \sqrt{\cosh^2\left(\frac{x}{a}\right)} = \cosh\left(\frac{x}{a}\right)
$$

$$
   ds = \cosh\left(\frac{x}{a}\right) dx
$$

3. **Integrate over $[0, b]$:**

$$
   L = \int_0^b \cosh\left(\frac{x}{a}\right) dx = \left[ a \sinh\left(\frac{x}{a}\right) \right]_0^b = a \sinh\left(\frac{b}{a}\right)
$$

$$
\boxed{L = a \sinh\left(\frac{b}{a}\right)}
$$

#### Key Insight / Takeaway
Hyperbolic identities naturally eliminate square roots in arc length calculations for hanging chain profiles.

### Problem 1.6: Surface Area of Sphere of Radius $R$
**Attribution:** [Spivak Ch 14]

#### Problem Statement
Derive the surface area of a sphere of radius $R$ by revolving the upper semicircle $y = \sqrt{R^2 - x^2}$ ($x \in [-R, R]$) about the $x$-axis.

#### First-Principles Intuition
Revolving a differential arc length element $ds$ at radius $y$ sweeps out a circular ribbon of area $d\Sigma = 2\pi y \, ds$.

#### Step-by-Step Solution
1. **Compute Derivative & $ds$:**

$$
   y = \sqrt{R^2 - x^2} \implies y' = \frac{-x}{\sqrt{R^2 - x^2}} = -\frac{x}{y}
$$

$$
   ds = \sqrt{1 + (y')^2}\,dx = \sqrt{1 + \frac{x^2}{y^2}}\,dx = \sqrt{\frac{y^2 + x^2}{y^2}}\,dx = \sqrt{\frac{R^2}{y^2}}\,dx = \frac{R}{y}\,dx
$$

2. **Differential Surface Element:**

$$
   d\Sigma = 2\pi y \, ds = 2\pi y \left( \frac{R}{y} \, dx \right) = 2\pi R \, dx
$$

Remarkably, $y$ cancels completely out of the expression!

3. **Integrate over $[-R, R]$:**

$$
   S = \int_{-R}^R 2\pi R \, dx = 2\pi R [x]_{-R}^R = 2\pi R (2R) = 4\pi R^2
$$

$$
\boxed{S = 4\pi R^2}
$$

#### Key Insight / Takeaway
Archimedes' Hat-Box Theorem: The surface area of a spherical zone between two parallel planes depends only on height $\Delta x$, because $y$ and $\frac{ds}{dx}$ cancel out.

### Problem 1.7: Centroid of a Planar Semicircular Region
**Attribution:** [Stewart Ch 8.3]

#### Problem Statement
Find the centroid coordinates $(\bar{x}, \bar{y})$ of the planar semicircular region bounded by $y = \sqrt{R^2 - x^2}$ and $y = 0$.

#### First-Principles Intuition
By vertical axis symmetry, $\bar{x} = 0$. The height centroid $\bar{y}$ is found by integrating horizontal slice moments $y \, dA$.

#### Step-by-Step Solution
1. **Symmetry:**
   The region is symmetric across the $y$-axis, so $\bar{x} = 0$.

2. **Area $A$:**

$$
   A = \frac{1}{2} \pi R^2
$$

3. **Moment $M_x$:**
   Using vertical strips of width $dx$ and height $y = \sqrt{R^2 - x^2}$, the strip centroid is at $\tilde{y} = \frac{y}{2}$:

$$
   M_x = \int_{-R}^R \frac{y}{2} (y \, dx) = \frac{1}{2} \int_{-R}^R (R^2 - x^2)\,dx = \frac{1}{2} \left[ R^2 x - \frac{x^3}{3} \right]_{-R}^R = \frac{2}{3} R^3
$$

4. **Compute $\bar{y}$:**

$$
   \bar{y} = \frac{M_x}{A} = \frac{\frac{2}{3} R^3}{\frac{1}{2} \pi R^2} = \frac{4R}{3\pi}
$$

$$
\boxed{(\bar{x}, \bar{y}) = \left(0, \frac{4R}{3\pi}\right)}
$$

#### Key Insight / Takeaway
Symmetry determines zero-moment axes directly, leaving only one non-trivial centroid coordinate to compute.

### Problem 1.8: Work Done Stretching a Non-Linear Spring
**Attribution:** [MIT 18.01 Physics]

#### Problem Statement
A non-linear spring exerts a restoring force $F(x) = k_1 x + k_2 x^3$, where $x$ is displacement from equilibrium. Calculate the work required to stretch the spring from $x = 0$ to displacement $x = L$.

#### First-Principles Intuition
Work is the line integral of force against displacement: $W = \int_0^L F(x)\,dx$.

#### Step-by-Step Solution
1. **Setup Work Integral:**

$$
   W = \int_0^L (k_1 x + k_2 x^3)\,dx
$$

2. **Evaluate Anti-derivative:**

$$
   W = \left[ \frac{1}{2} k_1 x^2 + \frac{1}{4} k_2 x^4 \right]_0^L = \frac{1}{2} k_1 L^2 + \frac{1}{4} k_2 L^4
$$

$$
\boxed{W = \frac{1}{2} k_1 L^2 + \frac{1}{4} k_2 L^4}
$$

#### Key Insight / Takeaway
Superposition of linear and cubic spring terms integrates term-by-term into quadratic energy storage and quartic hardening components.

### Problem 1.9: Work to Pump Water Out of a Conical Tank
**Attribution:** [Stewart Ch 8.1]

#### Problem Statement
An inverted right circular conical tank has height $H = 10\text{ m}$ and top radius $R = 4\text{ m}$. It is filled with water of mass density $\rho = 1000\text{ kg/m}^3$ to a depth of $8\text{ m}$. Calculate the work required to pump all water out over the top rim of the tank ($g = 9.8\text{ m/s}^2$).

#### First-Principles Intuition

### Problem 1.11: Mean and Variance of Exponential Distribution
**Attribution:** [Apostol Vol 1 Ch 14]

#### Problem Statement
For a continuous random variable $X \sim \text{Exp}(\lambda)$ with PDF $f(x) = \lambda e^{-\lambda x}$ for $x \ge 0$ ($\lambda \gt 0$), calculate the mean $\mathbb{E}[X]$ and variance $\operatorname{Var}(X)$ using integration by parts.

#### First-Principles Intuition
Integration by parts $\int u \, dv = uv - \int v \, du$ lowers polynomial powers in expectation integrals $\int_0^\infty x^k e^{-\lambda x}\,dx$.

#### Step-by-Step Solution
1. **Compute Mean $\mathbb{E}[X]$:**

$$
   \mathbb{E}[X] = \int_0^\infty x (\lambda e^{-\lambda x})\,dx
$$

   Let $u = x \implies du = dx$, $dv = \lambda e^{-\lambda x}\,dx \implies v = -e^{-\lambda x}$.

$$
   \mathbb{E}[X] = \left[ -x e^{-\lambda x} \right]_0^\infty + \int_0^\infty e^{-\lambda x}\,dx = 0 + \left[ -\frac{1}{\lambda} e^{-\lambda x} \right]_0^\infty = \frac{1}{\lambda}
$$

2. **Compute Second Moment $\mathbb{E}[X^2]$:**

$$
   \mathbb{E}[X^2] = \int_0^\infty x^2 (\lambda e^{-\lambda x})\,dx
$$

   Let $u = x^2 \implies du = 2x\,dx$, $v = -e^{-\lambda x}$.

$$
   \mathbb{E}[X^2] = \left[ -x^2 e^{-\lambda x} \right]_0^\infty + 2 \int_0^\infty x e^{-\lambda x}\,dx = 0 + \frac{2}{\lambda} \mathbb{E}[X] = \frac{2}{\lambda^2}
$$

3. **Compute Variance $\operatorname{Var}(X)$:**

$$
   \operatorname{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 = \frac{2}{\lambda^2} - \left(\frac{1}{\lambda}\right)^2 = \frac{1}{\lambda^2}
$$

$$
\boxed{\mathbb{E}[X] = \frac{1}{\lambda}, \quad \operatorname{Var}(X) = \frac{1}{\lambda^2}}
$$

#### Key Insight / Takeaway
The exponential distribution rate parameter $\lambda$ determines inverse scale for both mean and standard deviation.

### Problem 1.12: Simpson's Rule and Error Bound Analysis
**Attribution:** [Demidovich No. 1705]

#### Problem Statement
Estimate $I = \int_0^1 e^{-x^2}\,dx$ using Simpson's $\frac{1}{3}$ Rule with $n = 4$ subintervals. Calculate the theoretical error bound using $\lvert E_S \rvert \le \frac{M_4 (b - a)^5}{180 n^4}$.

#### First-Principles Intuition
Simpson's Rule fits parabolas across pairs of subintervals, yielding $\mathcal{O}(h^4)$ accuracy for smooth functions.

#### Step-by-Step Solution
1. **Subinterval Setup ($n = 4, a = 0, b = 1, h = \frac{1}{4} = 0.25$):**

$$
   x_0 = 0, \quad x_1 = 0.25, \quad x_2 = 0.5, \quad x_3 = 0.75, \quad x_4 = 1.0
$$

$$
   f(x) = e^{-x^2}
$$

   - $f(0) = e^0 = 1.000000$
   - $f(0.25) = e^{-0.0625} \approx 0.939413$
   - $f(0.5) = e^{-0.25} \approx 0.778801$
   - $f(0.75) = e^{-0.5625} \approx 0.569783$
   - $f(1.0) = e^{-1.0} \approx 0.367879$

2. **Apply Simpson's Formula:**

$$
   \begin{aligned}
   S_4 &= \frac{h}{3} \left[ f(x_0) + 4 f(x_1) + 2 f(x_2) + 4 f(x_3) + f(x_4) \right] \\
   &= \frac{0.25}{3} \left[ 1.0 + 4(0.939413) + 2(0.778801) + 4(0.569783) + 0.367879 \right] \\
   &= \frac{0.25}{3} \left[ 1.0 + 3.757652 + 1.557602 + 2.279132 + 0.367879 \right] \\
   &= \frac{0.25}{3} [8.962265] \approx 0.746855
   \end{aligned}
$$

3. **Compute Derivative & Error Bound $M_4$:**

$$
   f^{(4)}(x) = (12 - 48x^2 + 16x^4) e^{-x^2}
$$

   On $[0, 1]$, $\lvert f^{(4)}(x) \rvert \le \lvert f^{(4)}(0) \rvert = 12 \implies M_4 = 12$.

$$
   \lvert E_S \rvert \le \frac{12 (1 - 0)^5}{180 (4^4)} = \frac{12}{180 \times 256} = \frac{1}{15 \times 256} = \frac{1}{3840} \approx 0.000260
$$

$$
\boxed{S_4 \approx 0.746855, \quad \lvert E_S \rvert \le \frac{1}{3840} \approx 2.6 \times 10^{-4}}
$$

#### Key Insight / Takeaway
Simpson's Rule provides exceptional accuracy with minimal grid points ($n=4$) due to fourth-order error decay.

### Problem 2.2: Hydrostatic Force on a Vertical Circular Dam Gate
**Attribution:** [MIT 18.01 Physics]

#### Problem Statement
A vertical circular inspection gate of radius $R = 2\text{ m}$ is built into the wall of a dam. The center of the circle is located at depth $H = 5\text{ m}$ below the water surface ($H \gt R$). Calculate the total hydrostatic force acting on the circular gate.

#### First-Principles Intuition
By the Hydrostatic Centroid Theorem, total force equals fluid pressure at the gate's geometric centroid multiplied by total gate area: $F = \rho g \bar{h} A$. We verify this via first-principles depth integration.

#### Step-by-Step Solution
1. **Coordinate Setup:**
   Let vertical coordinate $y \in [-R, R]$ be measured from the circle's center ($y = 0$ at depth $H$).
   Depth below surface at coordinate $y$ is $h(y) = H - y$.
   Horizontal width at position $y$ is $w(y) = 2\sqrt{R^2 - y^2}$.

2. **Differential Force Element:**

$$
   dF = \rho g \, h(y) \, w(y) \, dy = \rho g (H - y) \left( 2\sqrt{R^2 - y^2} \right) dy
$$

3. **Integrate over $[-R, R]$:**

$$
   F = 2\rho g H \int_{-R}^R \sqrt{R^2 - y^2}\,dy - 2\rho g \int_{-R}^R y \sqrt{R^2 - y^2}\,dy
$$

   - First Integral (Semicircle Area): $\int_{-R}^R \sqrt{R^2 - y^2}\,dy = \frac{\pi R^2}{2}$.
   - Second Integral (Odd Function over Symmetric Interval): $\int_{-R}^R y \sqrt{R^2 - y^2}\,dy = 0$.

4. **Combine Terms:**

$$
   F = 2\rho g H \left( \frac{\pi R^2}{2} \right) - 0 = \rho g H (\pi R^2) = \rho g \bar{h} A
$$

5. **Substitute Values ($R = 2, H = 5, \rho g = 9800$):**

$$
   F = 9800 \times 5 \times (\pi \times 2^2) = 49000 \times 4\pi = 196,000 \pi \approx 615,752\text{ N}
$$

$$
\boxed{F = \rho g H (\pi R^2) = 196{,}000 \pi\,\text{N} \approx 6.16 \times 10^5\,\text{N}}
$$

#### Key Insight / Takeaway
Linear depth variation cancels odd moment terms around symmetric centroids, confirming $F = P(\text{centroid}) \times A$.

In machine learning optimization, exponential decay weight penalties define unnormalized region profiles $y = e^{-\alpha x}$ for $x \in [0, \infty)$ ($\alpha \gt 0$). Find the centroid coordinates $(\bar{x}, \bar{y})$ of the infinite planar region bounded by $y = e^{-\alpha x}$, $y = 0$, and $x = 0$.

#### First-Principles Intuition
Improper integrals over $[0, \infty)$ determine finite total area and spatial moments due to rapid exponential decay.

1. **Total Area $A$:**

$$
   A = \int_0^\infty e^{-\alpha x}\,dx = \left[ -\frac{1}{\alpha} e^{-\alpha x} \right]_0^\infty = \frac{1}{\alpha}
$$

$$
   M_y = \int_0^\infty x e^{-\alpha x}\,dx = \frac{1}{\alpha^2}
$$

$$
   \bar{x} = \frac{M_y}{A} = \frac{\frac{1}{\alpha^2}}{\frac{1}{\alpha}} = \frac{1}{\alpha}
$$

$$

### Problem 2.5: Expectation and Variance of Uniform Random Variable
**Attribution:** [ML Foundation]

#### Problem Statement
Let feature $X$ be uniformly distributed over $[a, b]$. Derive $\mathbb{E}[X]$, $\mathbb{E}[X^2]$, and $\operatorname{Var}(X)$ from first principles.

#### First-Principles Intuition
Uniform density $f_X(x) = \frac{1}{b-a}$ places equal probability mass across $[a, b]$, making moments straightforward power integrals.

#### Step-by-Step Solution
1. **Mean $\mathbb{E}[X]$:**

$$
   \mathbb{E}[X] = \int_a^b x \cdot \frac{1}{b-a}\,dx = \frac{1}{b-a} \left[ \frac{x^2}{2} \right]_a^b = \frac{b^2 - a^2}{2(b-a)} = \frac{a+b}{2}
$$

$$
   \mathbb{E}[X^2] = \int_a^b x^2 \cdot \frac{1}{b-a}\,dx = \frac{1}{b-a} \left[ \frac{x^3}{3} \right]_a^b = \frac{b^3 - a^3}{3(b-a)} = \frac{a^2 + ab + b^2}{3}

$$

   \end{aligned}
$$

$$
\boxed{\mathbb{E}[X] = \frac{a+b}{2}, \quad \operatorname{Var}(X) = \frac{(b-a)^2}{12}}

### Problem 2.6: KL Divergence Between Two Exponential Distributions
**Attribution:** [AI/ML Information Theory]

#### Problem Statement
Given two continuous probability distributions $p(x) = \lambda_1 e^{-\lambda_1 x}$ and $q(x) = \lambda_2 e^{-\lambda_2 x}$ for $x \ge 0$ ($\lambda_1, \lambda_2 \gt 0$), calculate the Kullback-Leibler divergence $D_{\text{KL}}(p \parallel q) = \int_0^\infty p(x) \ln\left(\frac{p(x)}{q(x)}\right) dx$.

#### First-Principles Intuition
KL divergence measures information lost when approximating distribution $p$ by $q$. Logarithmic properties decompose the integrand into constant entropy and expectation terms.

#### Step-by-Step Solution
1. **Logarithmic Ratio Simplification:**

$$
   \ln\left(\frac{p(x)}{q(x)}\right) = \ln\left(\frac{\lambda_1 e^{-\lambda_1 x}}{\lambda_2 e^{-\lambda_2 x}}\right) = \ln\left(\frac{\lambda_1}{\lambda_2}\right) + (\lambda_2 - \lambda_1) x
$$

2. **Expand Integral:**

$$
   D_{\text{KL}}(p \parallel q) = \int_0^\infty p(x) \left[ \ln\left(\frac{\lambda_1}{\lambda_2}\right) + (\lambda_2 - \lambda_1) x \right] dx
$$

3. **Substitute Expectations under $p(x)$:**
   Since $\int_0^\infty p(x)\,dx = 1$ and $\int_0^\infty x p(x)\,dx = \mathbb{E}_p[X] = \frac{1}{\lambda_1}$:

$$
   \begin{aligned}
   D_{\text{KL}}(p \parallel q) &= \ln\left(\frac{\lambda_1}{\lambda_2}\right) \int_0^\infty p(x)\,dx + (\lambda_2 - \lambda_1) \int_0^\infty x p(x)\,dx \\
   &= \ln\left(\frac{\lambda_1}{\lambda_2}\right) + (\lambda_2 - \lambda_1) \left(\frac{1}{\lambda_1}\right) \\
   &= \ln\left(\frac{\lambda_1}{\lambda_2}\right) + \frac{\lambda_2}{\lambda_1} - 1
   \end{aligned}
$$

$$
\boxed{D_{\text{KL}}(p \parallel q) = \ln\left(\frac{\lambda_1}{\lambda_2}\right) + \frac{\lambda_2}{\lambda_1} - 1}
$$

#### Key Insight / Takeaway
Using $e^u \ge 1 + u \implies u - 1 \ge -\ln(1-u)$, $D_{\text{KL}}(p \parallel q) \ge 0$ with equality if and only if $\lambda_1 = \lambda_2$.

### Problem 2.7: Expected Loss Minimization under Gaussian Noise

$$
   y = x^2 \implies y' = 2x \implies ds = \sqrt{1 + 4x^2}\,dx
$$

2. **Differential Surface Element:**

$$
   d\Sigma = 2\pi x \, ds = 2\pi x \sqrt{1 + 4x^2}\,dx
$$

3. **Integrate via $u$-substitution ($u = 1 + 4x^2 \implies du = 8x\,dx \implies x\,dx = \frac{du}{8}$):**
   When $x = 0 \implies u = 1$; when $x = 2 \implies u = 17$.

$$
   \begin{aligned}
   S &= 2\pi \int_1^{17} \sqrt{u} \left(\frac{du}{8}\right) = \frac{\pi}{4} \int_1^{17} u^{\frac{1}{2}}\,du \\
   &= \frac{\pi}{4} \left[ \frac{2}{3} u^{\frac{3}{2}} \right]_1^{17} = \frac{\pi}{6} \left( 17^{\frac{3}{2}} - 1 \right) = \frac{\pi}{6} (17\sqrt{17} - 1)
   \end{aligned}
$$

4. **Numerical Approximation:**

$$
   17\sqrt{17} \approx 17(4.1231) \approx 70.093 \implies S \approx \frac{\pi}{6} (69.093) \approx 36.18
$$

$$
\boxed{S = \frac{\pi}{6} (17\sqrt{17} - 1) \approx 36.18}
$$

#### Key Insight / Takeaway
The linear factor $x$ in the $y$-axis revolution formula $2\pi x \, ds$ provides the derivative required for $u$-substitution under the radical.

### Problem 2.10: Volume of Revolution About Off-Center Axis $x = 3$
**Attribution:** [Stewart Ch 6.3]

#### Problem Statement
Find the volume of the solid generated by revolving the region bounded by $y = x^2$ and $y = 4$ about the vertical line $x = 3$.

#### First-Principles Intuition
Using the Cylindrical Shell method, vertical strips at position $x \in [-2, 2]$ are at radius $r(x) = 3 - x$ from the axis of rotation $x = 3$.

#### Step-by-Step Solution
1. **Region Boundaries:**
   Parabola $y = x^2$ intersects line $y = 4$ at $x = -2$ and $x = 2$.
   Height of vertical strip at $x$ is $h(x) = 4 - x^2$.

2. **Differential Shell Volume Element:**

$$
   dV = 2\pi \, r(x) \, h(x) \, dx = 2\pi (3 - x)(4 - x^2)\,dx = 2\pi (12 - 3x^2 - 4x + x^3)\,dx
$$

3. **Integrate over $[-2, 2]$:**
   By linear decomposition into even and odd components:

$$
   \begin{aligned}
   V &= 2\pi \int_{-2}^2 (12 - 3x^2 - 4x + x^3)\,dx \\
   &= 2\pi \left[ \int_{-2}^2 (12 - 3x^2)\,dx + \int_{-2}^2 (-4x + x^3)\,dx \right]
   \end{aligned}
$$

   The second integral vanishes because $(-4x + x^3)$ is an odd function over $[-2, 2]$.

$$
   \begin{aligned}
   V &= 2\pi \cdot 2 \int_0^2 (12 - 3x^2)\,dx = 4\pi \left[ 12x - x^3 \right]_0^2 \\
   &= 4\pi (24 - 8) = 4\pi (16) = 64\pi
   \end{aligned}
$$

$$
\boxed{V = 64\pi}
$$

#### Key Insight / Takeaway
Odd terms in radius-height products integrate to zero over symmetric intervals $[-a, a]$.

### Problem 2.11: Pappus Theorem Applied to Solid Torus
**Attribution:** [Apostol Vol 1]

#### Problem Statement
A circular disk of radius $r = 2\text{ cm}$ is revolved about an axis in its plane located at distance $R = 5\text{ cm}$ from the disk's center ($R \gt r$). Calculate the volume $V$ and total surface area $S$ of the resulting solid torus using Pappus's Centroid Theorems.

### Problem 2.12: Expectation of Half-Normal Distribution in ReLU Nets
**Attribution:** [AI/ML Neural Nets]

#### Problem Statement
In deep neural networks initialized with standard Gaussian weights $W \sim \mathcal{N}(0, \sigma^2)$, let $X = \lvert W \rvert$ be the magnitude of a pre-activation. $X$ is Half-Normal with PDF $f_X(x) = \sqrt{\frac{2}{\pi \sigma^2}} \exp\left(-\frac{x^2}{2\sigma^2}\right)$ for $x \ge 0$. Compute $\mathbb{E}[X]$ and $\operatorname{Var}(X)$.

#### First-Principles Intuition
Folding the negative half of the Gaussian onto the positive axis doubles the density there, producing a non-zero mean. This is *not* the distribution of $\operatorname{ReLU}(W) = \max(0, W)$: ReLU keeps an atom of probability mass $1/2$ at exactly $0$ rather than folding the negative half over, so $\mathbb{E}[\operatorname{ReLU}(W)] = \frac{1}{2}\mathbb{E}[X] = \sigma/\sqrt{2\pi}$, half of $\mathbb{E}[X]$ below.

#### Step-by-Step Solution
1. **Compute Mean $\mathbb{E}[X]$:**

$$
   \mathbb{E}[X] = \sqrt{\frac{2}{\pi \sigma^2}} \int_0^\infty x \exp\left(-\frac{x^2}{2\sigma^2}\right) dx
$$

   Let $u = \frac{x^2}{2\sigma^2} \implies du = \frac{x}{\sigma^2}\,dx \implies x\,dx = \sigma^2\,du$:

$$
   \mathbb{E}[X] = \sqrt{\frac{2}{\pi \sigma^2}} (\sigma^2) \int_0^\infty e^{-u}\,du = \sigma \sqrt{\frac{2}{\pi}} [ -e^{-u} ]_0^\infty = \sigma \sqrt{\frac{2}{\pi}}
$$

2. **Compute Second Moment $\mathbb{E}[X^2]$:**

   Since $x^2 \exp(-x^2/2\sigma^2)$ is even, the integral over $[0,\infty)$ is exactly half the integral over $(-\infty,\infty)$, and the prefactor $\sqrt{2/(\pi\sigma^2)}$ is exactly twice the normal density's normalizing constant $1/\sqrt{2\pi\sigma^2}$; the factor of $2$ and the factor of $\frac{1}{2}$ cancel, leaving

$$
   \mathbb{E}[X^2] = \sqrt{\frac{2}{\pi \sigma^2}} \int_0^\infty x^2 \exp\left(-\frac{x^2}{2\sigma^2}\right) dx = \mathbb{E}_{W \sim \mathcal{N}(0, \sigma^2)}[W^2] = \sigma^2
$$

3. **Compute Variance $\operatorname{Var}(X)$:**

$$
   \operatorname{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 = \sigma^2 - \sigma^2 \left(\frac{2}{\pi}\right) = \sigma^2 \left(1 - \frac{2}{\pi}\right)
$$

$$
\boxed{\mathbb{E}[X] = \sigma \sqrt{\frac{2}{\pi}}, \quad \operatorname{Var}(X) = \sigma^2 \left(1 - \frac{2}{\pi}\right)}
$$

#### Key Insight / Takeaway
Both $\lvert W \rvert$ and $\operatorname{ReLU}(W)$ have strictly positive mean ($\sigma\sqrt{2/\pi}$ and $\sigma/\sqrt{2\pi}$ respectively), which explains why batch normalization or zero-centered activations (like GELU/SiLU) are introduced to counteract the resulting activation-mean drift.

### Problem 3.1: Gabriel's Horn Paradox
**Attribution:** [Spivak Ch 14 / Demidovich No. 1665]

#### Problem Statement
Gabriel's Horn (Torricelli's Trumpet) is the surface generated by revolving $y = \frac{1}{x}$ for $x \ge 1$ about the $x$-axis.
1. Prove that the volume $V$ of Gabriel's Horn is finite and evaluate it.
2. Prove that the surface area $S$ of Gabriel's Horn is infinite ($\lim_{b \to \infty} S = \infty$).
3. Resolve the physical paradox: How can a solid horn hold a finite volume of paint yet require an infinite amount of paint to cover its inner surface?

#### First-Principles Intuition
Volume integrals weigh radius squared $\frac{1}{x^2}$, which converges rapidly ($p \gt 1$), whereas surface area integrals weigh radius linearly $\frac{1}{x}$, which diverges logarithmically ($p = 1$).

#### Step-by-Step Solution
1. **Volume Evaluation:**

$$
   V = \pi \int_1^\infty [f(x)]^2 \, dx = \pi \int_1^\infty \frac{1}{x^2} \, dx = \pi \left[ -\frac{1}{x} \right]_1^\infty = \pi (0 - (-1)) = \pi
$$

$$
   \boxed{V = \pi \lt \infty}
$$

2. **Surface Area Evaluation:**

$$
   S = 2\pi \int_1^\infty f(x) \sqrt{1 + [f'(x)]^2} \, dx = 2\pi \int_1^\infty \frac{1}{x} \sqrt{1 + \frac{1}{x^4}} \, dx
$$

   Since $\sqrt{1 + \frac{1}{x^4}} \gt 1$ for all $x \ge 1$:

$$
   S \gt 2\pi \int_1^\infty \frac{1}{x} \, dx = 2\pi \lim_{b \to \infty} [\ln x]_1^b = 2\pi \lim_{b \to \infty} \ln b = \infty
$$

$$
   \boxed{S = \infty}
$$

3. **Paradox Resolution:**
   There is no contradiction, because "filling" and "painting" measure different things. Filling the horn's interior with paint is a *volume*: the paint occupies $\{(x,y,z): x \ge 1,\ y^2+z^2 \le 1/x^2\}$, whose volume is $\pi < \infty$, so a finite amount of paint suffices to fill it. Painting the inner *surface* with a coat of some fixed positive thickness $\varepsilon$, however, would require volume $\approx \varepsilon \cdot S = \infty$ for any $\varepsilon > 0$, since $S = \infty$.

   The resolution is that filling delivers a layer of paint whose thickness at station $x$ is on the order of the horn's radius $r(x) = 1/x \to 0$ — a thickness that vanishes as $x \to \infty$ rather than staying fixed. A vanishing-thickness coat can cover an infinite-area surface with finite volume ($\int_1^\infty (\text{thickness}) \cdot 2\pi r(x)\sqrt{1+r'(x)^2}\,dx$ converges even though $\int (\text{thickness}) \, ds$ would diverge if thickness were bounded below). "Filling has finite volume but infinite surface area" is not a paradox about paint; it is a statement about how fast $r(x)^2$ (volume weight) can converge while $r(x)$ (area weight) still diverges.

#### Key Insight / Takeaway
Mathematical infinity paradoxes often disappear when continuum assumptions are bounded by real-world physical length scales.

### Problem 3.2: Volume of Steinmetz Solid (Intersection of Two Cylinders)
**Attribution:** [Putnam Archive / Demidovich No. 1660]

#### Problem Statement
Find the volume of the 3D solid $\Omega$ formed by the intersection of two right circular cylinders of radius $R$ whose axes intersect at right angles:

$$
x^2 + y^2 \le R^2 \quad \text{and} \quad x^2 + z^2 \le R^2
$$

#### First-Principles Intuition
Slice the solid horizontally with planes perpendicular to the $x$-axis at position $x \in [-R, R]$. At each $x$, $y^2 \le R^2 - x^2$ and $z^2 \le R^2 - x^2$, making every cross-section a square of side length $2\sqrt{R^2 - x^2}$.

#### Step-by-Step Solution
1. **Cross-Sectional Geometry:**
   For a fixed $x \in [-R, R]$:

$$
   -\sqrt{R^2 - x^2} \le y \le \sqrt{R^2 - x^2} \implies \text{Width } = 2\sqrt{R^2 - x^2}
$$

$$
   -\sqrt{R^2 - x^2} \le z \le \sqrt{R^2 - x^2} \implies \text{Height } = 2\sqrt{R^2 - x^2}
$$

   The cross-section perpendicular to the $x$-axis is a square!

2. **Cross-Sectional Area:**

$$
   A(x) = \left( 2\sqrt{R^2 - x^2} \right)^2 = 4(R^2 - x^2)
$$

3. **Volume Integration:**

$$
   V = \int_{-R}^R A(x)\,dx = \int_{-R}^R 4(R^2 - x^2)\,dx = 8 \int_0^R (R^2 - x^2)\,dx
$$

$$
   V = 8 \left[ R^2 x - \frac{x^3}{3} \right]_0^R = 8 \left( R^3 - \frac{R^3}{3} \right) = \frac{16}{3} R^3
$$

$$
\boxed{V = \frac{16}{3} R^3}
$$

#### Key Insight / Takeaway
Perpendicular cross-sections of non-spherical spatial intersections can yield simple polynomial squares, avoiding square-root integration completely.

### Problem 3.3: Exact Arc Length of an Ellipse & Elliptic Integrals
**Attribution:** [Cambridge Mathematical Tripos Part IA]

#### Problem Statement
Consider the ellipse parameterized by $x(t) = a \sin t, y(t) = b \cos t$ for $t \in [0, 2\pi]$ ($a \gt b \gt 0$).
1. Express the total perimeter $L$ of the ellipse in terms of the Complete Elliptic Integral of the Second Kind $E(k) = \int_0^{\frac{\pi}{2}} \sqrt{1 - k^2 \sin^2 t}\,dt$, where $k = \frac{\sqrt{a^2 - b^2}}{a}$ is the eccentricity.
2. Calculate the first two non-zero terms in the asymptotic expansion of $L$ for small eccentricity $k \ll 1$.

#### First-Principles Intuition
The circumference of an ellipse cannot be expressed in terms of elementary functions, requiring elliptic integrals.

#### Step-by-Step Solution
1. **Derive Parametric Arc Length Element:**

$$
   x'(t) = a \cos t, \quad y'(t) = -b \sin t
$$

$$
   ds = \sqrt{a^2 \cos^2 t + b^2 \sin^2 t}\,dt
$$

   Using $\cos^2 t = 1 - \sin^2 t$:

$$
   ds = \sqrt{a^2 (1 - \sin^2 t) + b^2 \sin^2 t}\,dt = \sqrt{a^2 - (a^2 - b^2) \sin^2 t}\,dt = a \sqrt{1 - k^2 \sin^2 t}\,dt
$$

   where $k^2 = \frac{a^2 - b^2}{a^2} = e^2$.

2. **Total Perimeter via Symmetry:**

$$
   L = 4 \int_0^{\frac{\pi}{2}} a \sqrt{1 - k^2 \sin^2 t}\,dt = 4a E(k)
$$

3. **Asymptotic Expansion for Small $k \ll 1$:**
   Use the binomial expansion $\sqrt{1 - u} = 1 - \frac{1}{2} u - \frac{1}{8} u^2 - \dots$:

$$
   \sqrt{1 - k^2 \sin^2 t} \approx 1 - \frac{k^2}{2} \sin^2 t - \frac{k^4}{8} \sin^4 t
$$

   Integrating term-by-term using $\int_0^{\frac{\pi}{2}} \sin^2 t \, dt = \frac{\pi}{4}$ and $\int_0^{\frac{\pi}{2}} \sin^4 t \, dt = \frac{3\pi}{16}$:

$$
   E(k) \approx \frac{\pi}{2} \left[ 1 - \frac{k^2}{4} - \frac{3k^4}{64} \right]
$$

$$
   L \approx 2\pi a \left( 1 - \frac{k^2}{4} - \frac{3k^4}{64} \right)
$$

$$
\boxed{L = 4a E(k) \approx 2\pi a \left( 1 - \frac{k^2}{4} \right)}
$$

#### Key Insight / Takeaway
Elliptic integrals arise naturally when integrating non-uniform curvature along non-circular conic paths.

### Problem 3.4: Variational Derivation of Catenary Curve
**Attribution:** [Cambridge Mathematical Tripos Part IA]

#### Problem Statement
A uniform heavy cable of length $L$ hangs under gravity between two fixed points $(-a, h)$ and $(a, h)$. Minimize the total potential energy $\mathcal{U}[y] = \int_{-a}^a \rho g y \sqrt{1 + (y')^2}\,dx$ subject to fixed length $\int_{-a}^a \sqrt{1 + (y')^2}\,dx = L$. Derive the differential equation of the catenary using the Euler-Lagrange functional derivative with a Lagrange multiplier $\lambda$.

#### First-Principles Intuition
Physical systems in static equilibrium minimize total potential energy. Applying the Beltrami Identity to the augmented Lagrangian yields the hyperbolic cosine shape.

#### Step-by-Step Solution
1. **Formulate Augmented Lagrangian:**

$$
   \mathcal{L}(y, y') = (\rho g y - \lambda) \sqrt{1 + (y')^2}
$$

   Let $\tilde{y} = y - \frac{\lambda}{\rho g}$. Then:

$$
   \mathcal{L} = \rho g \tilde{y} \sqrt{1 + (\tilde{y}')^2}
$$

2. **Apply Beltrami Identity:**
   Since $\mathcal{L}$ does not depend explicitly on $x$, the Hamiltonian energy invariant is constant:

$$
   \mathcal{L} - \tilde{y}' \frac{\partial \mathcal{L}}{\partial \tilde{y}'} = C
$$

$$
   \rho g \tilde{y} \sqrt{1 + (\tilde{y}')^2} - \tilde{y}' \cdot \frac{\rho g \tilde{y} \tilde{y}'}{\sqrt{1 + (\tilde{y}')^2}} = C
$$

$$
   \frac{\rho g \tilde{y}}{\sqrt{1 + (\tilde{y}')^2}} \left[ 1 + (\tilde{y}')^2 - (\tilde{y}')^2 \right] = C \implies \frac{\rho g \tilde{y}}{\sqrt{1 + (\tilde{y}')^2}} = C
$$

3. **Solve Differential Equation:**

$$
   \sqrt{1 + (\tilde{y}')^2} = \frac{\rho g \tilde{y}}{C} = \frac{\tilde{y}}{K} \quad \text{where } K = \frac{C}{\rho g}
$$

$$
   (\tilde{y}')^2 = \frac{\tilde{y}^2}{K^2} - 1 \implies \frac{d\tilde{y}}{dx} = \sqrt{\frac{\tilde{y}^2}{K^2} - 1}
$$

$$
   \int \frac{d\tilde{y}}{\sqrt{\tilde{y}^2 - K^2}} = \frac{1}{K} \int dx \implies \operatorname{arccosh}\left(\frac{\tilde{y}}{K}\right) = \frac{x}{K} + C_1
$$

   Setting $C_1 = 0$ by horizontal symmetry at $x = 0$:

$$
   \tilde{y}(x) = K \cosh\left(\frac{x}{K}\right) \implies y(x) = K \cosh\left(\frac{x}{K}\right) + \frac{\lambda}{\rho g}
$$

$$
\boxed{y(x) = C_0 + K \cosh\left(\frac{x}{K}\right)}
$$

#### Key Insight / Takeaway
The catenary $y = K \cosh(\frac{x}{K})$ is the unique variational curve minimizing gravitational energy for a hanging continuous line mass.

### Problem 3.5: Surface Area of Oblate Ellipsoid of Revolution
**Attribution:** [Demidovich No. 1655]

#### Problem Statement
Find the surface area of an oblate ellipsoid generated by revolving the ellipse $\frac{x^2}{a^2} + \frac{y^2}{b^2} = 1$ ($a \gt b$) about its minor axis ($y$-axis). Express the result in terms of eccentricity $e = \frac{\sqrt{a^2 - b^2}}{a}$.

#### First-Principles Intuition
Revolving about the minor $y$-axis yields $d\Sigma = 2\pi x \, ds$. Expressing $x$ in terms of $y$ leads to an integrand requiring inverse hyperbolic or inverse trigonometric substitution.

#### Step-by-Step Solution
1. **Ellipse Equation & Parametrization:**
   $x = a \sqrt{1 - \frac{y^2}{b^2}}$ for $y \in [-b, b]$.

$$
   \frac{dx}{dy} = -\frac{a^2 y}{b^2 x}
$$

2. **Arc Length Element $ds$:**

$$
   ds = \sqrt{1 + \left(\frac{dx}{dy}\right)^2}\,dy = \sqrt{1 + \frac{a^4 y^2}{b^4 x^2}}\,dy = \sqrt{\frac{b^4 x^2 + a^4 y^2}{b^4 x^2}}\,dy
$$

3. **Surface Area Integral Formulation:**

$$
   S = 2\pi \int_{-b}^b x \, ds = 2\pi \int_{-b}^b x \cdot \frac{\sqrt{b^4 x^2 + a^4 y^2}}{b^2 x}\,dy = \frac{2\pi}{b^2} \int_{-b}^b \sqrt{b^4 x^2 + a^4 y^2}\,dy
$$

   Substituting $x^2 = a^2\left(1 - \frac{y^2}{b^2}\right)$:

$$
   b^4 x^2 + a^4 y^2 = a^2 b^4 - a^2 b^2 y^2 + a^4 y^2 = a^2\left[ b^4 + (a^2 - b^2) y^2 \right] = a^2 b^4 \left(1 + \frac{a^2 e^2}{b^4} y^2\right)
$$

   using $a^2 - b^2 = a^2 e^2$. Hence $\sqrt{b^4x^2+a^4y^2} = a b^2 \sqrt{1 + \left(\frac{a^2 e^2}{b^4}\right) y^2}$, and the $b^2$ cancels against the prefactor:

$$
   S = \frac{2\pi}{b^2} \int_{-b}^b a b^2 \sqrt{1 + \left(\frac{a^2 e^2}{b^4}\right) y^2}\,dy = 4\pi a \int_0^b \sqrt{1 + \left(\frac{a^2 e^2}{b^4}\right) y^2}\,dy
$$

4. **Evaluate Integral ($\int \sqrt{1 + c^2 y^2}\,dy$):**
   Using $\int_0^b \sqrt{1 + c^2 y^2}\,dy = \frac{b}{2} \sqrt{1 + c^2 b^2} + \frac{1}{2c} \ln\left( cb + \sqrt{1 + c^2 b^2} \right)$ with $c = \frac{ae}{b^2}$ (note: $c$ must carry units of inverse length, which $ae/b^2$ does and $ae/b$ does not):

$$
   S = 2\pi a^2 + \frac{\pi b^2}{e} \ln\left( \frac{1 + e}{1 - e} \right)
$$

$$
\boxed{S = 2\pi a^2 + \frac{\pi b^2}{e} \ln\left( \frac{1 + e}{1 - e} \right)}
$$

#### Key Insight / Takeaway
Oblate ellipsoid surface area transitions smoothly to $4\pi a^2$ as eccentricity $e \to 0$ (sphere limit).

### Problem 3.6: Moments of Maxwell-Boltzmann Speed Distribution
**Attribution:** [Statistical Physics]

#### Problem Statement
In statistical mechanics, gas molecular speeds $v \ge 0$ follow the Maxwell-Boltzmann PDF:

$$
f(v) = 4\pi \left(\frac{m}{2\pi k_B T}\right)^{\frac{3}{2}} v^2 \exp\left(-\frac{m v^2}{2 k_B T}\right)
$$

Let $a = \frac{m}{2 k_B T}$.
1. Verify total probability normalization $\int_0^\infty f(v)\,dv = 1$.
2. Compute the mean speed $\langle v \rangle = \mathbb{E}[v]$ and root-mean-square speed $v_{\text{rms}} = \sqrt{\mathbb{E}[v^2]}$.

#### First-Principles Intuition
Standard Gaussian integrals $I_n = \int_0^\infty v^n e^{-a v^2}\,dv$ evaluate via gamma function recursion.

#### Step-by-Step Solution
1. **Verification of Normalization:**

$$
   I_2 = \int_0^\infty v^2 e^{-a v^2}\,dv = \frac{\sqrt{\pi}}{4 a^{\frac{3}{2}}}
$$

$$
   \int_0^\infty f(v)\,dv = 4\pi \left(\frac{a}{\pi}\right)^{\frac{3}{2}} \left( \frac{\sqrt{\pi}}{4 a^{\frac{3}{2}}} \right) = 4\pi \cdot \frac{a^{\frac{3}{2}}}{\pi^{\frac{3}{2}}} \cdot \frac{\pi^{\frac{1}{2}}}{4 a^{\frac{3}{2}}} = 1
$$

2. **Compute Mean Speed $\langle v \rangle$:**

$$
   \langle v \rangle = 4\pi \left(\frac{a}{\pi}\right)^{\frac{3}{2}} \int_0^\infty v^3 e^{-a v^2}\,dv
$$

   Let $u = a v^2 \implies du = 2a v \, dv \implies \int_0^\infty v^3 e^{-a v^2}\,dv = \frac{1}{2a^2} \int_0^\infty u e^{-u}\,du = \frac{1}{2a^2}$:

$$
   \langle v \rangle = 4\pi \left(\frac{a}{\pi}\right)^{\frac{3}{2}} \left( \frac{1}{2a^2} \right) = \frac{2}{\sqrt{\pi a}} = \sqrt{\frac{8 k_B T}{\pi m}}
$$

3. **Compute Mean Square Speed $\langle v^2 \rangle$:**

$$
   I_4 = \int_0^\infty v^4 e^{-a v^2}\,dv = \frac{3\sqrt{\pi}}{8 a^{\frac{5}{2}}}
$$

$$
   \langle v^2 \rangle = 4\pi \left(\frac{a}{\pi}\right)^{\frac{3}{2}} \left( \frac{3\sqrt{\pi}}{8 a^{\frac{5}{2}}} \right) = \frac{3}{2a} = \frac{3 k_B T}{m}
$$

$$
   v_{\text{rms}} = \sqrt{\langle v^2 \rangle} = \sqrt{\frac{3 k_B T}{m}}
$$

$$
\boxed{\langle v \rangle = \sqrt{\frac{8 k_B T}{\pi m}}, \quad v_{\text{rms}} = \sqrt{\frac{3 k_B T}{m}}}
$$

#### Key Insight / Takeaway
$v_{\text{rms}} \gt \langle v \rangle$ due to positive skewness in the velocity magnitude distribution.

### Problem 3.7: Generalized Pappus Theorem for Inclined Axis
**Attribution:** [Polya & Szego / Putnam]

#### Problem Statement
A planar region $\mathcal{R}$ of area $A$ with centroid $(\bar{x}, \bar{y})$ is revolved around an arbitrary line $Ax + By + C = 0$ in the plane that does not intersect $\mathcal{R}$. Prove that the volume $V$ of the solid of revolution is $V = 2\pi d_{\text{centroid}} A$, where $d_{\text{centroid}} = \frac{\lvert A\bar{x} + B\bar{y} + C \rvert}{\sqrt{A^2 + B^2}}$ is the perpendicular distance from the centroid to the line.

#### First-Principles Intuition
A rigid rotation and translation of coordinates leaves volume and area invariant, projecting distance along the normal vector.

#### Step-by-Step Solution
1. **Perpendicular Distance Function:**
   The distance from any point $(x, y) \in \mathcal{R}$ to the line $\mathcal{L}: Ax + By + C = 0$ is:

$$
   d(x, y) = \frac{\lvert Ax + By + C \rvert}{\sqrt{A^2 + B^2}}
$$

2. **Derive $dV = 2\pi\,d(x,y)\,dA$ from the shell formula (§2.3):**
   Rotate and translate coordinates rigidly so that $\mathcal{L}$ becomes the new $v$-axis, $(x,y) \mapsto (u,v)$, with $u$ the signed perpendicular coordinate; since $\mathcal{L}$ does not meet $\mathcal{R}$, $u$ has one sign on all of $\mathcal{R}$ and $\lvert u \rvert = d(x,y)$. A rigid motion preserves area, so $dA_{uv} = dA_{xy}$.

   Slice $\mathcal{R}$ into thin strips parallel to the $v$-axis at position $u$, of infinitesimal area $dA$. Revolving one such strip about the $v$-axis ($u=0$) is exactly the cylindrical-shell configuration of §2.3 with shell radius $r = \lvert u \rvert$: the strip sweeps out a shell of circumference $2\pi\lvert u\rvert$ and cross-sectional area $dA$, giving volume element $dV = 2\pi \lvert u \rvert \, dA = 2\pi\,d(x,y)\,dA$.

3. **Integrate over Region $\mathcal{R}$:**

$$
   V = \iint_{\mathcal{R}} 2\pi \, d(x, y) \, dA = \frac{2\pi}{\sqrt{A^2 + B^2}} \iint_{\mathcal{R}} \lvert Ax + By + C \rvert\,dA
$$

4. **Centroid Property:**
   Since the line does not intersect $\mathcal{R}$, $(Ax + By + C)$ maintains constant sign over $\mathcal{R}$.

$$
   \iint_{\mathcal{R}} (Ax + By + C)\,dA = A \iint_{\mathcal{R}} x\,dA + B \iint_{\mathcal{R}} y\,dA + C \iint_{\mathcal{R}} dA = A(A_{\text{area}} \bar{x}) + B(A_{\text{area}} \bar{y}) + C A_{\text{area}}
$$

$$
   V = \frac{2\pi}{\sqrt{A^2 + B^2}} \lvert A\bar{x} + B\bar{y} + C \rvert A_{\text{area}} = 2\pi d_{\text{centroid}} A_{\text{area}}
$$

$$
\boxed{V = 2\pi d_{\text{centroid}} A}
$$

#### Key Insight / Takeaway
Pappus's theorem holds universally for revolution around any non-intersecting line in the plane due to linearity of integration.

### Problem 3.8: Asymptotic Expansion of Gaussian Tail Expectation (Q-Function)
**Attribution:** [Putnam / Advanced Analysis]

#### Problem Statement
In AI risk bounds and signal processing, the Gaussian Q-function measures tail probability $Q(x) = \frac{1}{\sqrt{2\pi}} \int_x^\infty e^{-\frac{t^2}{2}}\,dt$ for $x \gt 0$.
1. Using repeated integration by parts, derive the asymptotic expansion:

$$
   Q(x) = \frac{e^{-\frac{x^2}{2}}}{x\sqrt{2\pi}} \left( 1 - \frac{1}{x^2} + \frac{3}{x^4} - \dots + (-1)^k \frac{(2k-1)!!}{x^{2k}} \right) + R_k(x)
$$

2. Show that the error $\lvert R_k(x) \rvert$ is bounded by the magnitude of the first neglected term.

#### First-Principles Intuition
For large $x \gg 1$, the integral is dominated by contributions near the lower boundary $t = x$. Repeated integration by parts extracts descending powers of $x$.

#### Step-by-Step Solution
1. **First Integration by Parts:**
   Rewrite integrand: $e^{-\frac{t^2}{2}}\,dt = \frac{1}{t} (t e^{-\frac{t^2}{2}}\,dt)$.
   Let $u = \frac{1}{t} \implies du = -\frac{1}{t^2}\,dt$, $dv = t e^{-\frac{t^2}{2}}\,dt \implies v = -e^{-\frac{t^2}{2}}$.

$$
   \int_x^\infty e^{-\frac{t^2}{2}}\,dt = \left[ -\frac{1}{t} e^{-\frac{t^2}{2}} \right]_x^\infty - \int_x^\infty \frac{1}{t^2} e^{-\frac{t^2}{2}}\,dt = \frac{e^{-\frac{x^2}{2}}}{x} - \int_x^\infty \frac{1}{t^2} e^{-\frac{t^2}{2}}\,dt
$$

2. **Second Integration by Parts:**
   Apply same trick to $\int_x^\infty \frac{1}{t^2} e^{-\frac{t^2}{2}}\,dt = \int_x^\infty \frac{1}{t^3} (t e^{-\frac{t^2}{2}})\,dt$:

$$
   \int_x^\infty \frac{1}{t^2} e^{-\frac{t^2}{2}}\,dt = \frac{e^{-\frac{x^2}{2}}}{x^3} - 3 \int_x^\infty \frac{1}{t^4} e^{-\frac{t^2}{2}}\,dt
$$

   Substituting back:

$$
   Q(x) = \frac{e^{-\frac{x^2}{2}}}{\sqrt{2\pi}} \left( \frac{1}{x} - \frac{1}{x^3} \right) + \frac{3}{\sqrt{2\pi}} \int_x^\infty \frac{1}{t^4} e^{-\frac{t^2}{2}}\,dt
$$

3. **General Term & Remainder Bound:**
   Induction yields:

$$
   Q(x) = \frac{e^{-\frac{x^2}{2}}}{x\sqrt{2\pi}} \sum_{m=0}^{k-1} (-1)^m \frac{(2m-1)!!}{x^{2m}} + R_k(x)
$$

   where $R_k(x) = \frac{(-1)^k (2k-1)!!}{\sqrt{2\pi}} \int_x^\infty \frac{e^{-\frac{t^2}{2}}}{t^{2k}}\,dt$.
   Since for $t \ge x$, $\frac{1}{t^{2k+1}} \le \frac{t}{x^{2k+2}}$:

$$
   \lvert R_k(x) \rvert \le \frac{(2k-1)!!}{\sqrt{2\pi}} \int_x^\infty \frac{t e^{-\frac{t^2}{2}}}{x^{2k+2}}\,dt = \frac{(2k-1)!! e^{-\frac{x^2}{2}}}{\sqrt{2\pi} x^{2k+1}}
$$

   This upper bound equals the absolute magnitude of the next term in the asymptotic series!

$$
\boxed{Q(x) \sim \frac{e^{-\frac{x^2}{2}}}{x\sqrt{2\pi}} \left( 1 - \frac{1}{x^2} + \frac{3}{x^4} - \dots \right)}
$$

#### Key Insight / Takeaway
Asymptotic series for tail integrals yield high-precision numerical approximations for large $x$ despite being formally divergent.
