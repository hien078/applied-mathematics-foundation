# Topic 02: Sets, Relations, and Functions

## 1. First-Principles Intuition & Motivation

Every mathematical object you will ever meet — a number system, a vector space, a probability distribution, a dataset — is at bottom a *set* equipped with some structure. Set theory earns its role as the foundation of mathematics through radical austerity: the only primitive notion is membership, $x \in A$, and everything else is defined from it.

The governing principle is **extensionality**: a set is completely determined by which elements it contains. There is no order ($\{1,2\} = \{2,1\}$), no multiplicity ($\{1,1,2\} = \{1,2\}$), no "container material" — only membership. This is why proofs about sets reduce to logic: to reason about $A \cup B$ is to reason about the disjunction $x \in A \vee x \in B$, and every equivalence law from Topic 01 instantly becomes a set identity.

The second act of the theory is to build *associations* out of sets themselves. An ordered pair $(a,b)$ can be encoded as a set; the Cartesian product $A \times B$ collects all such pairs; a **relation** is any subset of $A \times B$; and a **function** is a relation with the discipline that each input gets exactly one output. Nothing new is postulated — orderings, groupings, and input-output machines are all just sets of pairs.

This austerity has practical payoff in modeling. A dataset is a finite set of (input, label) pairs — literally a relation. Asking whether it is a *function* asks whether any input appears with two conflicting labels (label noise). Asking whether the parameters-to-predictions map is *injective* asks whether the model is identifiable. The vocabulary of this topic is the vocabulary in which such questions become precise.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition 2.1 (Set, membership, extensionality).** A *set* is an unordered collection of distinct objects. We write $x \in A$ for membership and $x \notin A$ for non-membership. Two sets are equal iff they have the same elements. Standard descriptions: roster notation $A = \{1, 2, 3\}$ and set-builder notation $A = \{x \in \mathbb{R} : x \gt 0\}$.

**Standard sets.** $\mathbb{N} = \{0, 1, 2, \ldots\}$, $\mathbb{Z}$, $\mathbb{Q}$, $\mathbb{R}$, $\mathbb{C}$, and the empty set $\emptyset = \{\}$.

**Definition 2.2 (Subset and equality).** $A \subseteq B$ means $\forall x\,(x \in A \to x \in B)$. Set equality $A = B$ holds iff $A \subseteq B$ and $B \subseteq A$ — the *double inclusion* principle, which is also the standard proof template.

**Definition 2.3 (Set operations).** Relative to a universe $U$:

| Operation | Notation | Definition |
|---|---|---|
| Union | $A \cup B$ | $\{x : x \in A \vee x \in B\}$ |
| Intersection | $A \cap B$ | $\{x : x \in A \wedge x \in B\}$ |
| Difference | $A \setminus B$ | $\{x : x \in A \wedge x \notin B\}$ |
| Complement | $A^{c}$ | $\{x \in U : x \notin A\}$ |
| Symmetric difference | $A \triangle B$ | $(A \setminus B) \cup (B \setminus A)$ |
| Cartesian product | $A \times B$ | $\{(a,b) : a \in A, b \in B\}$ |
| Power set | $\mathcal{P}(A)$ | $\{S : S \subseteq A\}$ |

**Theorem 2.4 (Algebra of sets).** For all sets $A, B, C$ in a universe $U$: De Morgan's laws $(A \cup B)^{c} = A^{c} \cap B^{c}$ and $(A \cap B)^{c} = A^{c} \cup B^{c}$; distributivity $A \cap (B \cup C) = (A \cap B) \cup (A \cap C)$ and $A \cup (B \cap C) = (A \cup B) \cap (A \cup C)$; and for finite sets the inclusion-exclusion identity $\vert A \cup B \vert = \vert A \vert + \vert B \vert - \vert A \cap B \vert$.

**Definition 2.5 (Relation and its properties).** A *(binary) relation* from $A$ to $B$ is any subset $R \subseteq A \times B$; we write $a\,R\,b$ for $(a,b) \in R$. For a relation $R \subseteq A \times A$:

| Property | Definition |
|---|---|
| Reflexive | $\forall a \in A,\ a\,R\,a$ |
| Symmetric | $\forall a,b,\ a\,R\,b \to b\,R\,a$ |
| Antisymmetric | $\forall a,b,\ (a\,R\,b \wedge b\,R\,a) \to a = b$ |
| Transitive | $\forall a,b,c,\ (a\,R\,b \wedge b\,R\,c) \to a\,R\,c$ |

An *equivalence relation* is reflexive, symmetric, and transitive; its *equivalence classes* are $[a] = \{x \in A : x\,R\,a\}$. A *partial order* is reflexive, antisymmetric, and transitive (e.g. $\leq$ on $\mathbb{R}$, $\subseteq$ on $\mathcal{P}(U)$).

**Definition 2.6 (Partition).** A *partition* of $A$ is a family of nonempty, pairwise disjoint subsets whose union is $A$.

**Definition 2.7 (Function).** A *function* $f: A \to B$ is a relation $f \subseteq A \times B$ such that for every $a \in A$ there is exactly one $b \in B$ with $(a,b) \in f$; we write $f(a) = b$. Here $A$ is the *domain*, $B$ the *codomain*, and $f(A) = \{f(a) : a \in A\}$ the *image* (or range).

**Definition 2.8 (Injective, surjective, bijective).**

- $f$ is *injective* (one-to-one) iff $f(a_1) = f(a_2) \to a_1 = a_2$.
- $f$ is *surjective* (onto) iff $\forall b \in B\ \exists a \in A,\ f(a) = b$.
- $f$ is *bijective* iff it is both; equivalently, iff it has a (unique) inverse function $f^{-1}: B \to A$.

**Definition 2.9 (Cardinality).** Sets $A$ and $B$ are *equinumerous*, $\vert A \vert = \vert B \vert$, iff a bijection $A \to B$ exists. A set is *countable* iff it is finite or equinumerous with $\mathbb{N}$.

**Theorem 2.10 (Cantor).** For every set $A$ there is no surjection $A \to \mathcal{P}(A)$; hence $\vert A \vert \lt \vert \mathcal{P}(A) \vert$.

## 3. Step-by-Step Mathematical Proofs & Derivations

The universal template is **element chasing** under **double inclusion**: to prove $X = Y$, take an arbitrary $x \in X$ and deduce $x \in Y$, then conversely. Each chase is an exercise in the predicate logic of Topic 01.

### Proof 3.1 — De Morgan's Law for Sets

**Claim.** $(A \cup B)^{c} = A^{c} \cap B^{c}$.

**Proof.** ($\subseteq$) Let $x \in (A \cup B)^{c}$. Then $x \notin A \cup B$, i.e. $\neg(x \in A \vee x \in B)$. By the propositional De Morgan law this is $\neg(x \in A) \wedge \neg(x \in B)$, i.e. $x \in A^{c}$ and $x \in B^{c}$, so $x \in A^{c} \cap B^{c}$.

($\supseteq$) Let $x \in A^{c} \cap B^{c}$. Then $x \notin A$ and $x \notin B$, so $\neg(x \in A \vee x \in B)$ by De Morgan again, i.e. $x \notin A \cup B$, hence $x \in (A \cup B)^{c}$.

Both inclusions hold, so the sets are equal. The dual law $(A \cap B)^{c} = A^{c} \cup B^{c}$ follows by the same chase with the other propositional De Morgan law. $\blacksquare$

**Remark.** The proof is *nothing but* Topic 01's De Morgan law wrapped in membership predicates — the promised logic-to-sets dictionary in action.

### Proof 3.2 — Distributivity of Intersection over Union

**Claim.** $A \cap (B \cup C) = (A \cap B) \cup (A \cap C)$.

**Proof.** For any $x$:

$$
x \in A \cap (B \cup C) \iff x \in A \wedge (x \in B \vee x \in C)
$$

By propositional distributivity (Topic 01, Proof 3.6) the right side is equivalent to:

$$
(x \in A \wedge x \in B) \vee (x \in A \wedge x \in C) \iff x \in (A \cap B) \cup (A \cap C)
$$

Since the membership conditions are equivalent for every $x$, extensionality gives equality of the sets. $\blacksquare$

**Remark.** Chaining "iff" statements proves both inclusions simultaneously — a compressed double inclusion, legitimate only when every step is a genuine equivalence.

### Proof 3.3 — Inclusion-Exclusion for Two Finite Sets

**Claim.** For finite sets, $\vert A \cup B \vert = \vert A \vert + \vert B \vert - \vert A \cap B \vert$.

**Proof.** Decompose $A \cup B$ into three pairwise disjoint pieces:

$$
A \cup B = (A \setminus B) \cup (A \cap B) \cup (B \setminus A)
$$

Disjointness is clear (an element cannot be both in and out of $B$), and every element of the union lies in exactly one piece. For disjoint finite sets cardinalities add, so with $a = \vert A \setminus B \vert$, $m = \vert A \cap B \vert$, $b = \vert B \setminus A \vert$:

$$
\vert A \cup B \vert = a + m + b
$$

Also $\vert A \vert = a + m$ and $\vert B \vert = b + m$ (same disjoint decomposition applied to $A$ and $B$). Therefore:

$$
\vert A \vert + \vert B \vert - \vert A \cap B \vert = (a+m) + (b+m) - m = a + b + m = \vert A \cup B \vert \qquad \blacksquare
$$

**Remark.** Adding $\vert A \vert$ and $\vert B \vert$ counts the overlap twice; subtracting $\vert A \cap B \vert$ corrects it. Topic 05 generalizes this to $n$ sets.

### Proof 3.4 — Equivalence Classes Form a Partition

**Claim.** If $R$ is an equivalence relation on $A$, the equivalence classes $[a] = \{x : x\,R\,a\}$ form a partition of $A$: they are nonempty, cover $A$, and any two classes are either identical or disjoint.

**Proof.**

*Nonempty and covering.* By reflexivity $a\,R\,a$, so $a \in [a]$. Hence every class is nonempty and every element of $A$ lies in some class.

*Identical or disjoint.* Suppose $[a] \cap [b] \neq \emptyset$; pick $c$ in the intersection, so $c\,R\,a$ and $c\,R\,b$. We show $[a] \subseteq [b]$: let $x \in [a]$, i.e. $x\,R\,a$. By symmetry $a\,R\,c$; by transitivity ($x\,R\,a$, $a\,R\,c$) we get $x\,R\,c$; by transitivity again ($x\,R\,c$, $c\,R\,b$) we get $x\,R\,b$, so $x \in [b]$. The symmetric argument gives $[b] \subseteq [a]$, hence $[a] = [b]$.

So two classes that share even one element coincide, and classes are otherwise disjoint. $\blacksquare$

**Converse.** Any partition induces an equivalence relation ($x\,R\,y$ iff $x, y$ share a block), so equivalence relations and partitions are two descriptions of the same structure.

### Proof 3.5 — Composition Preserves Injectivity and Surjectivity

**Claim.** Let $f: A \to B$ and $g: B \to C$. If $f$ and $g$ are injective, so is $g \circ f$; if both are surjective, so is $g \circ f$; hence the composition of bijections is a bijection.

**Proof.**

*Injectivity.* Suppose $(g \circ f)(a_1) = (g \circ f)(a_2)$, i.e. $g(f(a_1)) = g(f(a_2))$. Since $g$ is injective, $f(a_1) = f(a_2)$. Since $f$ is injective, $a_1 = a_2$. Hence $g \circ f$ is injective.

*Surjectivity.* Let $c \in C$. Since $g$ is surjective, there is $b \in B$ with $g(b) = c$. Since $f$ is surjective, there is $a \in A$ with $f(a) = b$. Then $(g \circ f)(a) = g(f(a)) = g(b) = c$. Hence $g \circ f$ is surjective.

Combining both parts proves the bijection statement. $\blacksquare$

**Remark.** The two arguments run in opposite directions: injectivity peels $g$ then $f$ (outside in), surjectivity lifts witnesses through $g$ then $f$ (backwards from $C$). Deep learning note: a composition of non-injective layers (e.g. ReLU networks) is non-injective — information once collapsed is never recovered downstream.

### Proof 3.6 — Cantor's Theorem

**Claim.** For every set $A$, there is no surjection $f: A \to \mathcal{P}(A)$.

**Proof (diagonal argument).** Suppose for contradiction that $f: A \to \mathcal{P}(A)$ is surjective. Define the *diagonal set*:

$$
D = \{a \in A : a \notin f(a)\}
$$

$D$ is a subset of $A$, so $D \in \mathcal{P}(A)$. By surjectivity there exists $d \in A$ with $f(d) = D$. Now ask whether $d \in D$:

- If $d \in D$, then by definition of $D$, $d \notin f(d) = D$ — contradiction.
- If $d \notin D$, then $d \notin f(d)$, so by definition of $D$, $d \in D$ — contradiction.

Both cases are absurd, so no such surjection exists. Since injections $A \to \mathcal{P}(A)$ do exist (e.g. $a \mapsto \{a\}$), we conclude $\vert A \vert \lt \vert \mathcal{P}(A) \vert$. $\blacksquare$

**Consequence.** Applied to $A = \mathbb{N}$: the set of all sets of naturals is uncountable, and iterating the theorem yields an infinite hierarchy of ever-larger infinities. The same diagonal scheme proves $\mathbb{R}$ uncountable and underlies the halting problem and Gödel's incompleteness constructions.

### Proof-Template Summary for Set Arguments

| Goal | Template |
|---|---|
| $A \subseteq B$ | "Let $x \in A$ ... therefore $x \in B$." |
| $A = B$ | Double inclusion: prove $A \subseteq B$ and $B \subseteq A$ (or chain genuine iffs). |
| $A \neq B$ | Exhibit one element in exactly one of them. |
| $f$ injective | "Assume $f(a_1) = f(a_2)$ ... conclude $a_1 = a_2$." |
| $f$ surjective | "Let $b \in B$ ... construct $a$ with $f(a) = b$." |
| $R$ is an equivalence | Verify reflexive, symmetric, transitive — three separate universally quantified checks. |
| Impossibility of a map | Contradiction, often via a diagonal or counting obstruction. |

Every template is a specialization of the quantifier rules of Topic 01: inclusions are universally quantified implications, surjectivity is a $\forall\exists$ statement, and the diagonal argument is engineered self-referential negation.

## 4. Computational & Algorithmic Insights

**Representations of sets.** A set over a small universe of size $n$ can be a **bitset** (membership, union, intersection in $O(n/w)$ machine words, with bitwise AND/OR literally implementing $\cap$/$\cup$); over large or dynamic universes, **hash sets** give expected $O(1)$ membership and **balanced trees / sorted arrays** give $O(\log n)$ with ordered iteration. Python's `set` type implements the full algebra: `A | B`, `A & B`, `A - B`, `A ^ B` are $\cup, \cap, \setminus, \triangle$.

**Relations as data.** A finite relation $R \subseteq A \times B$ *is* a table — exactly the relational database model. Checking reflexivity costs $O(\vert A \vert)$ lookups; symmetry $O(\vert R \vert)$; transitivity is join-like, naively $O(\vert R \vert^2)$ lookups or via the adjacency-matrix product. Computing the transitive closure is the Floyd–Warshall algorithm in $O(n^3)$.

**Functions and their properties, algorithmically.** For a finite function given as a table of pairs:

| Check | Algorithm | Cost |
|---|---|---|
| Is the relation a function? | No input key appears twice with different values | $O(n)$ with a hash map |
| Injectivity | No output value appears twice | $O(n)$ with a hash set of seen outputs |
| Surjectivity | Image set equals codomain | $O(n + \vert B \vert)$ |
| Bijectivity | Both checks pass | $O(n + \vert B \vert)$ |

**Equivalence classes at scale.** The union-find (disjoint-set) data structure maintains a partition under "merge classes" and "same class?" queries in near-constant amortized time $O(\alpha(n))$ — the algorithmic realization of Proof 3.4, used in Kruskal's MST algorithm, image segmentation, and type unification.

**Power sets are exponential.** $\vert \mathcal{P}(A) \vert = 2^{\vert A \vert}$, so any algorithm enumerating all subsets is $\Omega(2^n)$ — the combinatorial explosion behind feature-subset selection being intractable exactly and approached greedily or with regularization instead.

## 5. Real-World Physics & AI/ML Applications

**Datasets, splits, and labels.** A supervised dataset is a finite relation $S \subseteq \mathcal{X} \times \mathcal{Y}$. It is a *function* precisely when no input carries two different labels; duplicated inputs with conflicting labels (annotator disagreement) make the Bayes error nonzero. A train/validation/test split is a *partition* of $S$, and cross-validation is a family of partitions. Stratified sampling partitions by the equivalence relation "same class label" and samples within classes.

**Identifiability is injectivity.** A parametric model defines a map $\theta \mapsto p_\theta$ from parameters to predictive distributions. The model is *identifiable* iff this map is injective. Neural networks are famously non-identifiable: permuting hidden units and rescaling ReLU weights ($W \mapsto cW$, next layer $\mapsto \frac{1}{c} \cdot$) leave the function unchanged, so the parameter-to-function map collapses entire equivalence classes. The loss landscape is best understood on the *quotient* by this symmetry relation.

**Encodings, embeddings, and physics.**

- **One-hot encoding** is an injection from a vocabulary $V$ into $\mathbb{R}^{\vert V \vert}$; an **embedding layer** composes it with a linear map into $\mathbb{R}^{d}$ with $d \ll \vert V \vert$, which for $d \lt \vert V \vert$ cannot be injective on all of $\mathbb{R}^{\vert V \vert}$ but is injective on the one-hot vectors themselves — the useful requirement.
- **Softmax** maps $\mathbb{R}^{n}$ onto the open probability simplex; it is surjective onto the simplex interior but not injective (adding a constant to all logits preserves the output) — its "classes" are lines, another quotient structure.
- **Physics: state spaces and symmetry.** Classical mechanics models a system as a set of states (phase space); observables are functions on it; and symmetry groups act by bijections. Identifying states related by a symmetry — taking the quotient by an equivalence relation — is how one passes from microstates to macrostates in statistical mechanics, with entropy $S = k_B \ln \Omega$ counting the size of an equivalence class.

**Databases.** SQL SELECT is set-builder notation executed by a machine; JOIN is relation composition; UNION/INTERSECT/EXCEPT are the Boolean algebra of Theorem 2.4, and query optimizers exploit exactly these identities to reorder operations.
