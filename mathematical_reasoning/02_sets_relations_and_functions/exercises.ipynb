## Level 0 — Concept Check

### Problem L0.1: Membership versus Inclusion

Let $A = \{1, 2, \{3\}\}$. Decide (true/false, with reasons): (a) $1 \in A$; (b) $3 \in A$; (c) $\{3\} \in A$; (d) $\{3\} \subseteq A$; (e) $\{1,2\} \subseteq A$.

**Solution:**

(a) **True.** $1$ is listed as an element.

(b) **False.** The elements of $A$ are $1$, $2$, and the *set* $\{3\}$; the number $3$ itself is not among them.

(c) **True.** The set $\{3\}$ is literally one of the three listed elements.

(d) **False.** $\{3\} \subseteq A$ would require every element of $\{3\}$ — namely $3$ — to be an element of $A$, and by (b) it is not.

(e) **True.** Both $1 \in A$ and $2 \in A$, so every element of $\{1,2\}$ belongs to $A$.

$$
\boxed{\text{(a) T, (b) F, (c) T, (d) F, (e) T}}
$$

*Key takeaway:* $\in$ looks one level deep only; $\subseteq$ tests all elements — a set can be an element without being a subset.

### Problem L0.2: The Empty Set and Its Wrapper

What are the cardinalities of $\emptyset$, $\{\emptyset\}$, and $\mathcal{P}(\{\emptyset\})$? Is $\emptyset \subseteq A$ for every set $A$?

**Solution:**

$\vert \emptyset \vert = 0$: the empty set has no elements.

$\vert \{\emptyset\} \vert = 1$: it contains exactly one element, the empty set.

$\mathcal{P}(\{\emptyset\}) = \{\emptyset, \{\emptyset\}\}$, so its cardinality is $2$ (consistent with $2^1 = 2$).

$\emptyset \subseteq A$ always holds: the defining condition $\forall x\,(x \in \emptyset \to x \in A)$ has a false hypothesis for every $x$, so it is **vacuously true** (Topic 01).

$$
\boxed{\vert \emptyset \vert = 0, \quad \vert \{\emptyset\} \vert = 1, \quad \vert \mathcal{P}(\{\emptyset\}) \vert = 2, \quad \emptyset \subseteq A \text{ always}}
$$

*Key takeaway:* The empty set is an element-less set, not "nothing"; vacuous truth makes it a subset of everything.

### Problem L0.3: Reading Set-Builder Notation

List the elements of $A = \{x \in \mathbb{Z} : x^2 \lt 10\}$ and $B = \{x^2 : x \in \mathbb{Z},\ x^2 \lt 10\}$. Are they equal?

**Solution:**

For $A$: the integers with $x^2 \lt 10$ are those with $\lvert x \rvert \leq 3$, so $A = \{-3, -2, -1, 0, 1, 2, 3\}$.

For $B$: we collect the *squares* of those integers: $9, 4, 1, 0, 1, 4, 9$. As a set, duplicates collapse: $B = \{0, 1, 4, 9\}$.

They are not equal: $A$ has 7 elements, $B$ has 4, and e.g. $-3 \in A$ but $-3 \notin B$.

$$
\boxed{A = \{-3,\ldots,3\},\ \vert A \vert = 7; \quad B = \{0,1,4,9\},\ \vert B \vert = 4; \quad A \neq B}
$$

*Key takeaway:* The expression left of the colon says what is collected; sets silently discard duplicates and order.

### Problem L0.4: Function or Not?

Which of the following relations on $A = \{1,2,3\}$, $B = \{a,b\}$ are functions $A \to B$? (i) $R_1 = \{(1,a),(2,a),(3,b)\}$; (ii) $R_2 = \{(1,a),(1,b),(2,a),(3,a)\}$; (iii) $R_3 = \{(1,a),(3,b)\}$.

**Solution:**

A function $A \to B$ must assign to **each** element of $A$ **exactly one** element of $B$.

(i) $R_1$: inputs 1, 2, 3 each appear exactly once. **Function** (not injective: 1 and 2 collide at $a$; surjective: both $a,b$ are hit).

(ii) $R_2$: input 1 appears with two different outputs $a$ and $b$ — uniqueness fails. **Not a function.**

(iii) $R_3$: input 2 never appears — totality fails. **Not a function** from $A$ (it is a function on the smaller domain $\{1,3\}$).

$$
\boxed{\text{only } R_1 \text{ is a function } A \to B}
$$

*Key takeaway:* "Every input, exactly one output" has two failure modes — a repeated input (ambiguity) or a missing input (partiality).

## Level 1 — Foundation

### Problem L1.1: A Difference Identity by Element Chasing

Prove that $A \setminus (B \cup C) = (A \setminus B) \cap (A \setminus C)$.

**Solution:**

For an arbitrary $x$, chain equivalences:

$$
x \in A \setminus (B \cup C) \iff x \in A \wedge \neg(x \in B \vee x \in C)
$$

By De Morgan's propositional law:

$$
\iff x \in A \wedge (x \notin B) \wedge (x \notin C)
$$

Duplicate the conjunct $x \in A$ (idempotence) and regroup:

$$
\iff (x \in A \wedge x \notin B) \wedge (x \in A \wedge x \notin C) \iff x \in (A \setminus B) \cap (A \setminus C)
$$

Every step is an equivalence, so both inclusions hold and the sets are equal by extensionality.

$$
\boxed{A \setminus (B \cup C) = (A \setminus B) \cap (A \setminus C)}
$$

*Key takeaway:* Set difference obeys a De Morgan law of its own; element chasing converts it to pure propositional logic.

### Problem L1.2: Symmetric Difference

Prove that $A \triangle B = (A \cup B) \setminus (A \cap B)$, where $A \triangle B = (A \setminus B) \cup (B \setminus A)$.

**Solution:**

Membership in the left side: $x \in A \triangle B$ iff ($x \in A \wedge x \notin B$) $\vee$ ($x \in B \wedge x \notin A$) — i.e. $x$ is in exactly one of the two sets. This is the *exclusive or* of $p: x \in A$ and $q: x \in B$.

Membership in the right side: $x \in (A \cup B) \setminus (A \cap B)$ iff $(p \vee q) \wedge \neg(p \wedge q)$ — in at least one, but not both.

Truth-table check of $(p \wedge \neg q) \vee (q \wedge \neg p) \equiv (p \vee q) \wedge \neg(p \wedge q)$:

| $p$ | $q$ | exactly one | at-least-one and not-both |
|---|---|---|---|
| T | T | F | F |
| T | F | T | T |
| F | T | T | T |
| F | F | F | F |

The columns agree, so the membership conditions are equivalent for every $x$, and the sets are equal.

$$
\boxed{A \triangle B = (A \cup B) \setminus (A \cap B)}
$$

*Key takeaway:* Symmetric difference is XOR at the level of sets; a 4-row table settles the identity once and for all.

### Problem L1.3: Power Set and Product Cardinalities

Let $A = \{1,2,3\}$ and $B = \{x,y\}$. List $\mathcal{P}(B)$, compute $\vert \mathcal{P}(A) \vert$ and $\vert A \times B \vert$, and state the general formulas.

**Solution:**

$\mathcal{P}(B) = \{\emptyset, \{x\}, \{y\}, \{x,y\}\}$ — the $2^2 = 4$ subsets of $B$.

For the power set of $A$: each subset is built by an independent in/out choice for each of the 3 elements, so $\vert \mathcal{P}(A) \vert = 2^3 = 8$.

For the product: each pair combines one of 3 first coordinates with one of 2 second coordinates, so $\vert A \times B \vert = 3 \cdot 2 = 6$; explicitly $\{(1,x),(1,y),(2,x),(2,y),(3,x),(3,y)\}$.

General formulas for finite sets:

$$
\boxed{\vert \mathcal{P}(A) \vert = 2^{\vert A \vert}, \qquad \vert A \times B \vert = \vert A \vert \cdot \vert B \vert}
$$

*Key takeaway:* Independent binary choices give the power set its exponential size; ordered pairing multiplies cardinalities — the seeds of Topic 05's counting rules.

### Problem L1.4: Classifying Maps

For each function, decide injective/surjective/bijective with proof: (a) $f: \mathbb{Z} \to \mathbb{Z}$, $f(x) = 2x$; (b) $g: \mathbb{R} \to [0,\infty)$, $g(x) = x^2$; (c) $h: \mathbb{R} \to \mathbb{R}$, $h(x) = 2x+1$.

**Solution:**

(a) *Injective:* $2x_1 = 2x_2 \Rightarrow x_1 = x_2$ (cancel 2). *Not surjective:* $1$ has no preimage since $2x = 1$ has no integer solution. So injective only.

(b) *Not injective:* $g(-1) = g(1) = 1$ with $-1 \neq 1$. *Surjective:* for $y \geq 0$, $x = \sqrt{y}$ satisfies $g(x) = y$. So surjective only (the codomain choice matters — as a map into $\mathbb{R}$ it would not be surjective).

(c) *Injective:* $2x_1 + 1 = 2x_2 + 1 \Rightarrow x_1 = x_2$. *Surjective:* given $y$, $x = (y-1)/2$ is a real preimage. Hence bijective, with inverse $h^{-1}(y) = (y-1)/2$.

$$
\boxed{\text{(a) injective only, (b) surjective only, (c) bijective with } h^{-1}(y) = \tfrac{y-1}{2}}
$$

*Key takeaway:* Injectivity and surjectivity are independent, and surjectivity is a property of the map *together with its declared codomain*.

### Problem L1.5: Congruence mod n Is an Equivalence Relation

Fix $n \geq 1$ and define $a \sim b$ iff $n \mid (a - b)$ on $\mathbb{Z}$. Prove $\sim$ is an equivalence relation and describe its classes for $n = 3$.

**Solution:**

*Reflexive:* $a - a = 0 = n \cdot 0$, so $n \mid (a-a)$.

*Symmetric:* if $n \mid (a-b)$, say $a - b = nk$, then $b - a = n(-k)$, so $n \mid (b-a)$.

*Transitive:* if $a - b = nk$ and $b - c = nm$, then $a - c = (a-b) + (b-c) = n(k+m)$, so $n \mid (a-c)$.

All three properties hold, so $\sim$ is an equivalence relation. For $n = 3$ the classes are the possible remainders on division by 3:

$$
[0] = \{\ldots,-3,0,3,6,\ldots\}, \quad [1] = \{\ldots,-2,1,4,7,\ldots\}, \quad [2] = \{\ldots,-1,2,5,8,\ldots\}
$$

These three classes are disjoint and cover $\mathbb{Z}$ — a partition, as guaranteed by the partition theorem.

$$
\boxed{\sim \text{ is an equivalence relation; } \mathbb{Z}/{\sim} = \{[0],[1],[2]\} \text{ for } n=3}
$$

*Key takeaway:* Modular arithmetic is the archetypal quotient construction: an equivalence relation collapsing $\mathbb{Z}$ into finitely many classes.

### Problem L1.6: Symmetric + Transitive Does Not Imply Reflexive

Exhibit a relation on a nonempty set that is symmetric and transitive but not reflexive, and locate the flaw in the classic "proof" that symmetry and transitivity imply reflexivity ($a\,R\,b \Rightarrow b\,R\,a \Rightarrow a\,R\,a$).

**Solution:**

**Counterexample.** On $A = \{1, 2\}$ take $R = \{(1,1)\}$ — or even simpler, the empty relation $R = \emptyset$.

- *Symmetric:* every pair in $R$ (there are none, or only $(1,1)$) has its reverse in $R$ — vacuously or trivially true.
- *Transitive:* the hypothesis $a\,R\,b \wedge b\,R\,c$ is satisfied only by $a=b=c=1$ (or never), and then $a\,R\,c$ holds.
- *Not reflexive:* $(2,2) \notin R$, and reflexivity demands $a\,R\,a$ for **every** $a \in A$.

**The flaw.** The argument "$a\,R\,b \Rightarrow b\,R\,a \Rightarrow a\,R\,a$" begins by *assuming* some $b$ with $a\,R\,b$ exists. For an element $a$ related to nothing (like $2$ above), the argument never starts. It proves only: reflexivity holds at elements that are related to at least one element.

$$
\boxed{R = \emptyset \text{ on } \{1,2\} \text{ is symmetric and transitive but not reflexive}}
$$

*Key takeaway:* Universally quantified properties must be checked at every element; hidden existence assumptions are a classic source of false "proofs."

## Level 2 — Applications in AI/ML & CS

### Problem L2.1: Label Noise and the Function Property

A dataset is a finite relation $S \subseteq \mathcal{X} \times \mathcal{Y}$. Explain formally when $S$ is a function, and prove: if $S$ contains pairs $(x_0, y_1)$ and $(x_0, y_2)$ with $y_1 \neq y_2$, then no classifier $f: \mathcal{X} \to \mathcal{Y}$ can achieve zero error on $S$ (0-1 loss over pairs).

**Solution:**

$S$ is a function iff for every $x$ there is at most one $y$ with $(x,y) \in S$ (and every $x$ in the input set appears): formally $\bigl((x,y_1) \in S \wedge (x,y_2) \in S\bigr) \to y_1 = y_2$.

**Proof of the claim.** Let $f$ be any classifier and suppose $(x_0,y_1), (x_0,y_2) \in S$ with $y_1 \neq y_2$. The value $f(x_0)$ is a single element of $\mathcal{Y}$ (because $f$ is a function). It cannot equal both $y_1$ and $y_2$, since $y_1 \neq y_2$. Hence $f$ misclassifies at least one of the two pairs, and its error on $S$ is at least $\frac{1}{\vert S \vert} \gt 0$. $\blacksquare$

$$
\boxed{\text{Conflicting labels } \Rightarrow \text{ minimum achievable error} \geq \tfrac{1}{\vert S \vert}}
$$

*Key takeaway:* "Bayes error is nonzero under label noise" is, at bottom, the statement that a function cannot represent a non-functional relation.

### Problem L2.2: One-Hot Encoding Is an Injection

Define the one-hot map $e: V \to \mathbb{R}^{n}$ on a vocabulary $V = \{w_1, \ldots, w_n\}$ by $e(w_i) = e_i$ (the $i$-th standard basis vector). Prove $e$ is injective, and show that a subsequent embedding $E \in \mathbb{R}^{d \times n}$ (mapping $w_i \mapsto E e_i$) is injective on the vocabulary iff the columns of $E$ are pairwise distinct.

**Solution:**

**Injectivity of $e$.** If $e(w_i) = e(w_j)$ then $e_i = e_j$. Standard basis vectors are pairwise distinct ($e_i$ has its 1 in position $i$), so $i = j$ and $w_i = w_j$. $\blacksquare$

**Embedding condition.** The composite map is $w_i \mapsto E e_i = E_{:,i}$, the $i$-th column of $E$.

- If the columns are pairwise distinct, then $E_{:,i} = E_{:,j} \Rightarrow i = j$, so the composite is injective on $V$.
- Conversely, if two columns coincide, $E_{:,i} = E_{:,j}$ with $i \neq j$, then $w_i \neq w_j$ map to the same vector — injectivity fails.

Note the linear map $x \mapsto Ex$ on all of $\mathbb{R}^{n}$ cannot be injective when $d \lt n$ (rank at most $d$), yet its restriction to the $n$ one-hot points can be — restriction can rescue injectivity.

$$
\boxed{e \text{ is injective; } E \circ e \text{ injective} \iff \text{columns of } E \text{ pairwise distinct}}
$$

*Key takeaway:* Embedding layers only need injectivity on the finite vocabulary, not on the whole space — a much weaker and easily satisfied condition.

### Problem L2.3: Hash Collisions and Non-Injectivity

A hash function maps $h: U \to \{0, 1, \ldots, m-1\}$ with $\vert U \vert \gt m$. Prove $h$ cannot be injective, and compute the minimum possible number of colliding pairs when $\vert U \vert = m + 1$.

**Solution:**

**Non-injectivity.** Suppose $h$ were injective. An injection $U \to \{0,\ldots,m-1\}$ forces $\vert U \vert \leq m$ (distinct inputs need distinct outputs, of which only $m$ exist), contradicting $\vert U \vert \gt m$. Hence some two inputs collide. (This is the pigeonhole principle, developed fully in Topic 05.)

**Minimum collisions for $\vert U \vert = m+1$.** Distribute $m+1$ inputs over $m$ buckets. To minimize colliding pairs, spread as evenly as possible: $m-1$ buckets with one input, one bucket with two. The number of unordered colliding pairs is then $\binom{2}{2} = 1$. Any other distribution puts $k \geq 3$ in some bucket, contributing $\binom{k}{2} \geq 3$ pairs.

$$
\boxed{h \text{ is never injective when } \vert U \vert \gt m; \text{ minimum } = 1 \text{ colliding pair at } \vert U \vert = m+1}
$$

*Key takeaway:* Hash tables, Bloom filters, and feature hashing all live with unavoidable non-injectivity; design manages collisions rather than pretending to avoid them.

### Problem L2.4: Cross-Validation as a Family of Partitions

Formalize $k$-fold cross-validation on a dataset $S$ with $\vert S \vert = n$ (assume $k \mid n$) in the language of partitions, and prove that each data point is used for validation exactly once and for training exactly $k-1$ times.

**Solution:**

**Formalization.** Choose a partition $\{F_1, \ldots, F_k\}$ of $S$ into $k$ blocks ("folds") with $\vert F_i \vert = n/k$: the blocks are nonempty, pairwise disjoint, and their union is $S$. Round $i$ uses validation set $F_i$ and training set $S \setminus F_i$.

**Proof.** Let $x \in S$. Since the folds cover $S$, $x \in F_j$ for some $j$; since they are pairwise disjoint, this $j$ is unique.

- *Validation:* $x$ is in the validation set of round $i$ iff $x \in F_i$ iff $i = j$ — exactly one round.
- *Training:* $x$ is in the training set of round $i$ iff $x \in S \setminus F_i$ iff $i \neq j$ — exactly $k - 1$ rounds. $\blacksquare$

$$
\boxed{\text{Each point validates once and trains } k-1 \text{ times — a direct consequence of partition disjointness and coverage}}
$$

*Key takeaway:* The statistical guarantees of cross-validation rest on the two defining properties of a partition; leakage bugs are precisely violations of disjointness.

### Problem L2.5: Neural Network Non-Identifiability as an Equivalence Relation

For a one-hidden-layer ReLU network $f_\theta(x) = \sum_{i=1}^{h} v_i\,\mathrm{ReLU}(w_i^{\top} x)$ with $\theta = (w_i, v_i)_{i=1}^{h}$, define $\theta \sim \theta'$ iff $f_\theta = f_{\theta'}$ as functions. Show $\sim$ is an equivalence relation, and exhibit two distinct parameter vectors in the same class.

**Solution:**

**Equivalence.** The relation is defined through equality of functions, which is reflexive ($f_\theta = f_\theta$), symmetric ($f_\theta = f_{\theta'} \Rightarrow f_{\theta'} = f_\theta$), and transitive (equalities chain). Any relation defined by "same image under a map $\theta \mapsto f_\theta$" is automatically an equivalence relation.

**Two symmetries producing distinct equivalent parameters.**

1. *Permutation:* swap units 1 and 2: $\theta' = ((w_2,v_2),(w_1,v_1),\ldots)$. The sum is unchanged, so $f_{\theta'} = f_\theta$, yet $\theta' \neq \theta$ whenever the units differ.
2. *Positive rescaling:* for $c \gt 0$, ReLU is positively homogeneous, $\mathrm{ReLU}(c\,t) = c\,\mathrm{ReLU}(t)$, so replacing $(w_1, v_1)$ by $(c\,w_1, v_1/c)$ leaves each term $v_1 \mathrm{ReLU}(w_1^{\top}x) = (v_1/c)\mathrm{ReLU}(c\,w_1^{\top}x)$ unchanged.

Thus the parameters-to-function map is far from injective: each function corresponds to a class of size at least $h! \cdot (0,\infty)^{h}$.

$$
\boxed{\sim \text{ is an equivalence relation; permutation and rescaling give } \theta' \neq \theta \text{ with } f_{\theta'} = f_\theta}
$$

*Key takeaway:* Loss landscapes, mode connectivity, and Bayesian posteriors over networks are properly objects on the quotient set of parameters modulo these symmetries.

### Problem L2.6: SQL Joins as Relation Composition

Given relations $R \subseteq A \times B$ ("customer placed order") and $Q \subseteq B \times C$ ("order contains product"), define the composition $Q \circ R \subseteq A \times C$. Compute it for $R = \{(\text{ann}, o_1), (\text{bob}, o_2)\}$ and $Q = \{(o_1, \text{tea}), (o_1, \text{cup}), (o_3, \text{pen})\}$, and connect to the SQL JOIN.

**Solution:**

**Definition.** $Q \circ R = \{(a,c) \in A \times C : \exists b \in B,\ (a,b) \in R \wedge (b,c) \in Q\}$ — an existential quantifier over the shared middle column.

**Computation.** Check each pair in $R$:

- $(\text{ann}, o_1)$: $o_1$ matches $(o_1,\text{tea})$ and $(o_1,\text{cup})$, contributing $(\text{ann},\text{tea})$ and $(\text{ann},\text{cup})$.
- $(\text{bob}, o_2)$: no pair in $Q$ has first coordinate $o_2$ — contributes nothing.

$(o_3, \text{pen})$ finds no partner in $R$. Hence $Q \circ R = \{(\text{ann},\text{tea}), (\text{ann},\text{cup})\}$.

**SQL.** `SELECT R.customer, Q.product FROM R JOIN Q ON R.order_id = Q.order_id` computes exactly this set (after projecting away the middle column). The existential $\exists b$ is the join condition; unmatched rows (bob, $o_3$) vanish, exactly as inner-join semantics dictate.

$$
\boxed{Q \circ R = \{(\text{ann},\text{tea}),\ (\text{ann},\text{cup})\}}
$$

*Key takeaway:* An inner join is relation composition: project the pairs joined through an existentially quantified shared key.

## Level 3 — Challenge

### Problem L3.1: The Rationals Are Countable

Prove that $\mathbb{Q}$ is countable by constructing an explicit injection or enumeration.

**Solution:**

**Step 1: $\mathbb{Z} \times \mathbb{Z}_{\gt 0}$ is countable.** Arrange pairs $(p,q)$ in a grid and enumerate along finite diagonals $\vert p \vert + q = 1, 2, 3, \ldots$; each diagonal is finite, so every pair receives a finite index. (Equivalently, $(p,q) \mapsto 2^{\sigma(p)} 3^{q}$ with $\sigma$ any injection $\mathbb{Z} \to \mathbb{N}$ is an injection into $\mathbb{N}$.)

**Step 2: surject the grid onto $\mathbb{Q}$.** The map $g(p,q) = p/q$ from $\mathbb{Z} \times \mathbb{Z}_{\gt 0}$ onto $\mathbb{Q}$ is surjective: every rational has a representation with positive denominator.

**Step 3: from a surjection to an enumeration.** Enumerate the grid as $x_1, x_2, \ldots$ and list $g(x_1), g(x_2), \ldots$, skipping values already listed. Every rational appears (by surjectivity) exactly once (by the skipping rule), giving a bijection between $\mathbb{N}$ (or an initial segment) and $\mathbb{Q}$. Since $\mathbb{Q}$ is infinite, we get a bijection with $\mathbb{N}$. $\blacksquare$

$$
\boxed{\mathbb{Q} \text{ is countably infinite}}
$$

*Key takeaway:* Countability survives products and quotients-by-representation: a countable grid plus a surjection onto $\mathbb{Q}$ yields an enumeration, despite $\mathbb{Q}$ being dense.

### Problem L3.2: The Reals Are Uncountable

Prove that the interval $(0,1)$ is uncountable using Cantor's diagonal argument.

**Solution:**

Suppose for contradiction that $(0,1)$ is countable, with enumeration $x_1, x_2, x_3, \ldots$ Write each in decimal:

$$
x_i = 0.d_{i1} d_{i2} d_{i3} \ldots
$$

choosing for definiteness the expansion not ending in repeating 9s. Define a new number $y = 0.e_1 e_2 e_3 \ldots$ by the diagonal rule:

$$
e_i = \begin{cases} 5 & \text{if } d_{ii} \neq 5 \\ 6 & \text{if } d_{ii} = 5 \end{cases}
$$

Then $y \in (0,1)$, and its expansion contains no 0s or 9s, so it is the unique decimal expansion of $y$. For every $i$, $y$ differs from $x_i$ in the $i$-th digit ($e_i \neq d_{ii}$), and since neither expansion is a forbidden repeating-9 form, $y \neq x_i$. Thus $y$ is a real in $(0,1)$ missing from the enumeration — contradicting completeness of the list. Hence no enumeration exists. $\blacksquare$

$$
\boxed{(0,1) \text{ — and hence } \mathbb{R} \text{ — is uncountable}}
$$

*Key takeaway:* Diagonalization defeats every proposed enumeration at once by engineering disagreement at position $i$ with item $i$ — the same self-reference that powers Cantor's theorem and the halting problem.

### Problem L3.3: Counting Functions and Injections

For finite sets with $\vert A \vert = m$ and $\vert B \vert = n$, count (a) all functions $A \to B$, (b) all injections $A \to B$ (for $m \leq n$), and (c) prove there are no injections when $m \gt n$.

**Solution:**

(a) Each of the $m$ elements of $A$ independently chooses one of $n$ images:

$$
n \cdot n \cdots n = n^{m}
$$

(b) Choose images one element at a time without reuse: the first element has $n$ choices, the second $n-1$ (its image must differ from the first), ..., the $m$-th has $n-m+1$:

$$
n(n-1)\cdots(n-m+1) = \frac{n!}{(n-m)!}
$$

(c) Suppose $f: A \to B$ is injective with $m \gt n$. The images $f(a)$ of the $m$ distinct elements are pairwise distinct elements of $B$, so $B$ contains at least $m \gt n$ elements — contradicting $\vert B \vert = n$. Hence no injection exists (pigeonhole principle).

$$
\boxed{n^{m} \text{ functions}; \quad \frac{n!}{(n-m)!} \text{ injections for } m \leq n; \quad 0 \text{ for } m \gt n}
$$

*Key takeaway:* Function counting is the bridge from set theory to combinatorics: products count free choices, falling factorials count choices without repetition.

### Problem L3.4: Schröder–Bernstein in Action

The Schröder–Bernstein theorem states: if there exist injections $f: A \to B$ and $g: B \to A$, then there exists a bijection $A \to B$. Use it (plus explicit injections) to prove $\vert \mathbb{N} \times \mathbb{N} \vert = \vert \mathbb{N} \vert$, and sketch why the theorem is nontrivial.

**Solution:**

**Injection $\mathbb{N} \to \mathbb{N} \times \mathbb{N}$:** $n \mapsto (n, 0)$. Distinct $n$ give distinct first coordinates — injective.

**Injection $\mathbb{N} \times \mathbb{N} \to \mathbb{N}$:** $(a,b) \mapsto 2^{a} 3^{b}$. If $2^{a}3^{b} = 2^{c}3^{d}$, unique prime factorization forces $a = c$ and $b = d$ — injective.

By Schröder–Bernstein, a bijection $\mathbb{N} \times \mathbb{N} \to \mathbb{N}$ exists, so $\vert \mathbb{N} \times \mathbb{N} \vert = \vert \mathbb{N} \vert$. (Explicitly, Cantor's pairing function $\pi(a,b) = \frac{(a+b)(a+b+1)}{2} + b$ is such a bijection, numbering the grid along diagonals.)

**Why nontrivial.** The two injections may be badly incompatible — neither surjective, with overlapping images. The proof must weave them into one bijection by tracking the back-and-forth orbits of elements under $g^{-1}, f^{-1}$ and deciding, orbit by orbit, whether to use $f$ or $g^{-1}$. No choice principle is needed, but the orbit analysis is genuinely clever — equality of cardinals is *not* just "two inequalities" by definition, it is a theorem.

$$
\boxed{\vert \mathbb{N} \times \mathbb{N} \vert = \vert \mathbb{N} \vert}
$$

*Key takeaway:* Schröder–Bernstein converts two one-way size comparisons into an exact size equality — the antisymmetry law of cardinal arithmetic.
