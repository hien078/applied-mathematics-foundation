## Level 0 — Concept Check

### Problem L0.1: Which Relations Hold?

For each pair, state every relation among $O$, $\Omega$, $\Theta$, $o$, $\omega$ that holds between $f$ and $g$. (a) $f = 2n^{2}$, $g = n^{2} + n$. (b) $f = n$, $g = n\log n$. (c) $f = 2^{n}$, $g = 3^{n}$. (d) $f = n$, $g = n^{1 + \sin n}$.

**Solution:**

(a) $f/g \to 2$, a positive finite constant, so $f = \Theta(g)$ — and therefore also $f = O(g)$ and $f = \Omega(g)$. Neither $o$ nor $\omega$ holds.

(b) $f/g = 1/\log n \to 0$, so $f = o(g)$, hence $f = O(g)$ but *not* $\Omega(g)$; dually $g = \omega(f)$.

(c) $f/g = (2/3)^{n} \to 0$, so $f = o(g)$ and $f = O(g)$, and $g = \omega(f)$. Note that unlike logarithms, the base of an *exponential* is not absorbed by constants.

(d) None of them. The exponent $1 + \sin n$ oscillates in $[0, 2]$, so $f/g = n^{-\sin n}$ has $\limsup = \infty$ and $\liminf = 0$; no constant bounds the ratio in either direction for all large $n$. Asymptotic notation gives a *partial* order on functions, not a total one.

$$
\boxed{\text{(a) } \Theta \quad \text{(b) } f = o(g) \quad \text{(c) } f = o(g) \quad \text{(d) incomparable}}
$$

*Key takeaway:* Compute the ratio and take limits — a finite positive limit means $\Theta$, a limit of $0$ means $o$, a limit of $\infty$ means $\omega$, and no limit at all can mean incomparable.

### Problem L0.2: Order the Growth Hierarchy

Sort into nondecreasing growth order and group any $\Theta$-equivalent functions: $n\log n$, $2^{n}$, $n^{2}$, $\log_{2} n$, $n!$, $n^{0.5}$, $\log_{10} n$, $n^{\log_2 3}$, $2^{2n}$, $(\log n)^{5}$.

**Solution:**

$$
\log_2 n \equiv \log_{10} n \prec (\log n)^{5} \prec n^{0.5} \prec n\log n \prec n^{\log_2 3} \prec n^{2} \prec 2^{n} \prec 2^{2n} \prec n!
$$

*Justifications for the non-obvious steps.*

- $\log_2 n$ and $\log_{10} n$ differ by the constant factor $1/\log_{10}2$, so they are $\Theta$-equivalent — the base of a logarithm never matters.
- $(\log n)^{5} = o(n^{0.5})$ because every polylogarithm is $o$ of every positive power.
- $n^{0.5} \prec n \prec n\log n \prec n^{\log_2 3} \approx n^{1.585} \prec n^{2}$: powers are ordered by exponent, and a log factor is smaller than any power increment, since $n\log n = o(n^{1.585})$.
- $2^{2n} = 4^{n}$ and $4^{n}/2^{n} = 2^{n} \to \infty$, so $2^{n} = o(2^{2n})$ — a common trap, since $2^{2n}$ is *not* $\Theta(2^{n})$ even though $2n = \Theta(n)$.
- $n! \succ 2^{2n}$ since $\log(n!) = \Theta(n\log n)$ dominates $\log(4^{n}) = \Theta(n)$.

$$
\boxed{\log n \prec \log^{5} n \prec \sqrt{n} \prec n\log n \prec n^{1.585} \prec n^{2} \prec 2^{n} \prec 4^{n} \prec n!}
$$

*Key takeaway:* Compare logs when the functions are exponential or factorial; compare exponents when they are polynomial. Constants inside an exponent are *not* free.

### Problem L0.3: Is "$O$" the Same as "Worst Case"?

A student writes: "Quicksort is $O(n\log n)$ on average and $O(n^{2})$ in the worst case, so $O$ describes the worst case." Identify the two independent axes being conflated, and restate the facts precisely.

**Solution:**

*Axis 1 — which input.* Best case, average (expected) case, worst case. This selects *which* function of $n$ we are describing: $T_{\text{best}}(n)$, $\mathbb{E}[T(n)]$, or $T_{\text{worst}}(n)$.

*Axis 2 — which bound.* $O$, $\Omega$, $\Theta$. This describes *how* we bound the chosen function.

The two are orthogonal: one can give an $\Omega$ bound on the best case or a $\Theta$ description of the average case. Precise statements for quicksort with a random pivot:

$$
T_{\text{worst}}(n) = \Theta(n^{2}), \qquad \mathbb{E}[T(n)] = \Theta(n\log n), \qquad T_{\text{best}}(n) = \Theta(n\log n)
$$

Both of the student's statements are *true* but weak: writing $O$ where $\Theta$ holds discards the matching lower bound. And "$O$ describes the worst case" is false — the $O(n\log n)$ average-case bound is a statement about the expectation, not about any single input.

$$
\boxed{\text{Name the case and the bound: "worst-case } \Theta(n^{2})\text{", "expected } \Theta(n\log n)\text{"}}
$$

*Key takeaway:* Every complexity claim needs two labels. "$O(n^{2})$" alone is ambiguous about which input distribution and wasteful if $\Theta$ is available.

### Problem L0.4: True or False, with Reasons

Decide each and justify. (a) $n^{2} + 100n = O(n^{2})$. (b) $n^{2} = O(n^{3})$. (c) $n^{3} = O(n^{2})$. (d) $\frac{1}{2}n^{2} = \Theta(n^{2})$. (e) $2^{n+1} = O(2^{n})$. (f) $(n+1)! = O(n!)$.

**Solution:**

(a) **True.** For $n \ge 100$: $100n \le n^{2}$, so $n^{2}+100n \le 2n^{2}$; witnesses $c = 2$, $n_0 = 100$.

(b) **True.** $n^{2} \le n^{3}$ for $n \ge 1$ ($c = 1$). $O$ is only an upper bound, so a loose bound is still valid — though $\Theta(n^{2})$ is the informative statement.

(c) **False.** If $n^{3} \le cn^{2}$ for all $n \ge n_0$, dividing by $n^{2}$ gives $n \le c$, contradicting the unboundedness of $n$.

(d) **True.** $\frac{1}{2}n^{2} \le 1 \cdot n^{2}$ and $\frac{1}{2}n^{2} \ge \frac{1}{2}n^{2}$; take $c_1 = c_2 = \frac12, 1$ respectively. Constant factors never affect $\Theta$-class.

(e) **True.** $2^{n+1} = 2\cdot 2^{n}$, so $c = 2$ works. A constant *added to the exponent* is a constant factor and is absorbed.

(f) **False.** $(n+1)!/n! = n+1 \to \infty$, so no constant $c$ bounds the ratio. A *variable* factor cannot be absorbed — this is exactly why (e) and (f) differ.

$$
\boxed{\text{(a) T \quad (b) T \quad (c) F \quad (d) T \quad (e) T \quad (f) F}}
$$

*Key takeaway:* $O$-classes absorb constant factors, including constants in exponents, but never factors that grow with $n$. Test by taking the ratio and asking whether it stays bounded.

## Level 1 — Foundation

### Problem L1.1: A Full Definition-Based $\Theta$ Proof

Prove from Definition 2.3 that $f(n) = 5n^{3} - 3n^{2} + 2n - 7 = \Theta(n^{3})$, giving explicit $c_1$, $c_2$, $n_0$.

**Solution:**

*Upper bound.* For $n \ge 1$, drop the negative terms and bound each positive term by a multiple of $n^{3}$:

$$
f(n) \le 5n^{3} + 2n \le 5n^{3} + 2n^{3} = 7n^{3}
$$

so $c_2 = 7$ works for $n \ge 1$.

*Lower bound.* We need $f(n) \ge c_1 n^{3}$. Since $3n^{2} + 7 \le 3n^{3} + 7n^{3} = 10n^{3}$ is far too crude, bound the subtracted terms at moderate $n$ instead: for $n \ge 4$,

$$
3n^{2} + 7 \le 3n^{2} + 7n^{2} = 10n^{2} \le \frac{10}{4}n^{3} = 2.5n^{3}
$$

Hence $f(n) \ge 5n^{3} - 2.5n^{3} = 2.5n^{3}$, and $c_1 = 2.5$ works for $n \ge 4$.

*Combining.* With $c_1 = 2.5$, $c_2 = 7$, $n_0 = 4$:

$$
0 \lt 2.5\,n^{3} \le 5n^{3} - 3n^{2} + 2n - 7 \le 7\,n^{3} \quad \text{for all } n \ge 4
$$

*Check at $n = 4$:* $f(4) = 320 - 48 + 8 - 7 = 273$, and $2.5 \cdot 64 = 160 \le 273 \le 448 = 7\cdot 64$. Confirmed.

$$
\boxed{c_1 = 2.5,\quad c_2 = 7,\quad n_0 = 4}
$$

*Key takeaway:* The lower bound is the harder half for polynomials with negative coefficients: push $n_0$ up until the leading term dominates the subtracted ones by a fixed fraction.

### Problem L1.2: Exponentials and Their Bases

Prove or disprove: (a) $2^{n+10} = O(2^{n})$; (b) $2^{2n} = O(2^{n})$; (c) $3^{n} = O(2^{n})$; (d) $n^{100} = O(2^{n})$.

**Solution:**

(a) **True.** $2^{n+10} = 2^{10}\cdot 2^{n} = 1024\cdot 2^{n}$, so $c = 1024$, $n_0 = 0$. An additive constant in the exponent is a multiplicative constant.

(b) **False.** $2^{2n}/2^{n} = 2^{n} \to \infty$. Suppose $2^{2n} \le c\,2^{n}$ for $n \ge n_0$; then $2^{n} \le c$ for all such $n$, impossible.

(c) **False.** $3^{n}/2^{n} = (3/2)^{n} \to \infty$, same contradiction. Bases of exponentials are *not* interchangeable, in sharp contrast to bases of logarithms.

(d) **True.** By the ratio test, with $r_n = n^{100}/2^{n}$,

$$
\frac{r_{n+1}}{r_n} = \Bigl(1+\frac{1}{n}\Bigr)^{100}\cdot\frac{1}{2} \longrightarrow \frac{1}{2} \lt 1
$$

so $r_n \to 0$ and in fact $n^{100} = o(2^{n})$. The crossover is late — $n^{100} \gt 2^{n}$ until roughly $n \approx 996$ — but asymptotically the exponential always wins.

$$
\boxed{\text{(a) T \quad (b) F \quad (c) F \quad (d) T}}
$$

*Key takeaway:* Constants in the exponent's *offset* are free; constants in the exponent's *slope* or in the base are not. Any polynomial is eventually beaten by any exponential, though "eventually" can be far beyond practical $n$.

### Problem L1.3: Logarithms, Bases, and $\log(n!)$

Prove: (a) $\log_a n = \Theta(\log_b n)$ for any constants $a, b \gt 1$; (b) $\log(n!) = \Theta(n\log n)$; (c) $\log(n^{n}) = \Theta(n\log n)$, so $n!$ and $n^{n}$ have the same log-order yet are not $\Theta$-equivalent.

**Solution:**

(a) By the change-of-base formula $\log_a n = \dfrac{\log_b n}{\log_b a}$, and $1/\log_b a$ is a positive constant. Taking $c_1 = c_2 = 1/\log_b a$ satisfies Definition 2.3 for all $n \ge 2$. This is why $O(\log n)$ is written without a base.

(b) *Upper:* $n! = 1\cdot 2\cdots n \le n^{n}$, so $\log(n!) \le n\log n$.
*Lower:* keep the largest $\lceil n/2 \rceil$ factors, each at least $n/2$:

$$
n! \ge \Bigl(\frac{n}{2}\Bigr)^{n/2} \implies \log(n!) \ge \frac{n}{2}\log\frac{n}{2} = \frac{n}{2}\log n - \frac{n}{2}
$$

For $n \ge 4$ the subtracted term is at most half the first, giving $\log(n!) \ge \frac{n\log n}{4}$. So $c_1 = 1/4$, $c_2 = 1$, $n_0 = 4$.

(c) $\log(n^{n}) = n\log n$ exactly, so both $\log(n!)$ and $\log(n^{n})$ are $\Theta(n\log n)$. But the *functions themselves* are far apart: by Stirling, $n!\sim\sqrt{2\pi n}(n/e)^{n}$, so

$$
\frac{n^{n}}{n!} \sim \frac{e^{n}}{\sqrt{2\pi n}} \to \infty
$$

Equality of logarithms up to $\Theta$ does not imply equality of the functions up to $\Theta$ — the log hides exponential-sized factors.

$$
\boxed{\log_a n = \Theta(\log_b n), \quad \log(n!) = \Theta(n\log n), \quad n! = o(n^{n})}
$$

*Key takeaway:* Logs let you compare enormous quantities, but a $\Theta$ statement about logs is much weaker than one about the originals — never exponentiate a $\Theta$ relation.

### Problem L1.4: Harmonic Sums and Loop Counting

(a) Prove $H_n = \sum_{i=1}^{n}\frac{1}{i} = \Theta(\log n)$. (b) Give the cost of the double loop "for $i$ from $1$ to $n$: for $j$ from $1$ to $i$: constant work". (c) Give the cost of "for $i$ from $1$ to $n$: for $j$ from $1$ to $n$ step $i$: constant work".

**Solution:**

(a) Compare with the integral of $1/x$, which is decreasing:

$$
\int_{1}^{n+1}\frac{dx}{x} \le \sum_{i=1}^{n}\frac{1}{i} \le 1 + \int_{1}^{n}\frac{dx}{x}
$$

giving $\ln(n+1) \le H_n \le 1 + \ln n$. Both bounds are $\Theta(\log n)$, so $H_n = \Theta(\log n)$; more precisely $H_n = \ln n + \gamma + O(1/n)$ with $\gamma \approx 0.5772$.

(b) The inner loop runs $i$ times, so the total is

$$
\sum_{i=1}^{n} i = \frac{n(n+1)}{2} = \Theta(n^{2})
$$

(c) The inner loop runs $\lfloor n/i \rfloor$ times, so the total is

$$
\sum_{i=1}^{n}\Bigl\lfloor \frac{n}{i}\Bigr\rfloor \le n\sum_{i=1}^{n}\frac{1}{i} = nH_n = \Theta(n\log n)
$$

and the matching lower bound follows from $\lfloor n/i \rfloor \gt n/i - 1$, whose correction sums to $\Theta(n)$. So the cost is $\Theta(n\log n)$ — the same sum that appears in the sieve of Eratosthenes ($\Theta(n\log\log n)$ when restricted to primes) and in the analysis of harmonic-decay learning-rate schedules.

$$
\boxed{H_n = \Theta(\log n), \quad \text{(b) } \Theta(n^{2}), \quad \text{(c) } \Theta(n\log n)}
$$

*Key takeaway:* Nested loops are summed, not multiplied, when the inner bound depends on the outer index; the two archetypes are $\sum i = \Theta(n^{2})$ and $\sum n/i = \Theta(n\log n)$.

### Problem L1.5: Master Theorem Drills

Solve with the Master theorem, identifying $a$, $b$, $\alpha = \log_b a$, and the case: (a) $T(n)=9T(n/3)+n$; (b) $T(n)=T(2n/3)+1$; (c) $T(n)=3T(n/4)+n\log n$; (d) $T(n)=2T(n/2)+n\log n$.

**Solution:**

| | $a$ | $b$ | $\alpha = \log_b a$ | $f(n)$ vs $n^{\alpha}$ | Case | Result |
|---|---|---|---|---|---|---|
| (a) | $9$ | $3$ | $2$ | $n = O(n^{2-\epsilon})$, $\epsilon = 1$ | 1 | $\Theta(n^{2})$ |
| (b) | $1$ | $3/2$ | $0$ | $1 = \Theta(n^{0})$ | 2 with $k=0$ | $\Theta(\log n)$ |
| (c) | $3$ | $4$ | $\log_4 3 \approx 0.793$ | $n\log n = \Omega(n^{0.793+\epsilon})$ | 3 | $\Theta(n\log n)$ |
| (d) | $2$ | $2$ | $1$ | $n\log n = \Theta(n^{1}\log^{1}n)$ | 2 with $k=1$ | $\Theta(n\log^{2} n)$ |

*Notes.* In (c) the regularity condition must be checked: $af(n/b) = 3\cdot\frac{n}{4}\log\frac{n}{4} \le \frac{3}{4}n\log n = cf(n)$ with $c = 3/4 \lt 1$, so Case 3 applies.

In (d) the *three-case* version of the theorem fails — $f(n)/n^{\alpha} = \log n$ grows, but not polynomially, so it is neither $O(n^{\alpha-\epsilon})$ nor $\Omega(n^{\alpha+\epsilon})$. The extended Case 2 with $\log^{k}$ covers it; a recursion tree confirms it, since each of the $\log n$ levels contributes $\Theta(n\log n)$ falling to $\Theta(n\log(n/2^{j}))$, and

$$
\sum_{j=0}^{\log_2 n - 1} n\bigl(\log_2 n - j\bigr) = n\sum_{t=1}^{\log_2 n} t = \Theta(n\log^{2}n)
$$

$$
\boxed{\text{(a) } \Theta(n^{2}) \quad \text{(b) } \Theta(\log n) \quad \text{(c) } \Theta(n\log n) \quad \text{(d) } \Theta(n\log^{2} n)}
$$

*Key takeaway:* Compute $n^{\log_b a}$ first, then compare $f$ against it *polynomially*. When the ratio is a logarithm rather than a power, you are in the gap and must use the extended case or draw the tree.

### Problem L1.6: A Recursion Tree with an Unbalanced Split

Solve $T(n) = T(n/3) + T(2n/3) + n$ with $T(1) = \Theta(1)$ by drawing the recursion tree, then verify the upper bound by substitution.

**Solution:**

*Recursion tree.* At the root the work is $n$. Its children have sizes $n/3$ and $2n/3$, whose work sums to $n/3 + 2n/3 = n$ — the same. By induction every complete level contributes exactly $n$ until subproblems start hitting the base case.

*Depth.* The shallowest root-to-leaf path divides by $3$ each time and has length $\log_3 n$; the deepest path divides by $3/2$ each time and has length $\log_{3/2} n$. Hence

$$
n\log_3 n \le T(n) \le n\log_{3/2} n + \Theta(n)
$$

Both bounds are $\Theta(n\log n)$, since $\log_3 n$ and $\log_{3/2}n$ are constant multiples of $\log n$. So $T(n) = \Theta(n\log n)$.

*Verification by substitution (upper bound).* Guess $T(n) \le cn\log n$ (natural log for convenience). Assuming it for smaller arguments,

$$
T(n) \le c\frac{n}{3}\log\frac{n}{3} + c\frac{2n}{3}\log\frac{2n}{3} + n = cn\log n - cn\Bigl(\frac{1}{3}\log 3 + \frac{2}{3}\log\frac{3}{2}\Bigr) + n
$$

The bracket equals $\log 3 - \frac{2}{3}\log 2 \approx 1.0986 - 0.4621 = 0.6365$, a positive constant $\beta$. So $T(n) \le cn\log n - c\beta n + n \le cn\log n$ whenever $c \ge 1/\beta \approx 1.571$. The induction closes.

$$
\boxed{T(n) = \Theta(n\log n) \text{ for any constant-fraction split}}
$$

*Key takeaway:* Divide-and-conquer is robust: any split into constant fractions leaves the per-level work unchanged and alters only the base of the logarithm — that is, a constant factor. Only splits peeling off $O(1)$ elements degrade to $\Theta(n^{2})$.

## Level 2 — Applications in AI/ML & CS

### Problem L2.1: The Cost of a Transformer Layer

For a single transformer layer with sequence length $n$, model width $d$, and MLP ratio $4$: (a) give the asymptotic time and memory of self-attention and of the feed-forward block; (b) find the crossover $n$ at which attention starts to dominate; (c) evaluate for $d = 4096$ at $n = 2048$ and $n = 32768$.

**Solution:**

(a) *Attention.* Projections $QKV$ and the output projection are $4$ matmuls of shape $(n\times d)(d\times d)$: $\Theta(nd^{2})$. Scores $QK^{\top}$ are $(n\times d)(d\times n)$: $\Theta(n^{2}d)$; the softmax is $\Theta(n^{2})$; applying to $V$ is again $\Theta(n^{2}d)$. Total

$$
\Theta(nd^{2} + n^{2}d) \text{ time}, \qquad \Theta(nd + n^{2}) \text{ memory}
$$

*Feed-forward.* Two matmuls $(n\times d)(d\times 4d)$ and $(n\times 4d)(4d\times d)$: $\Theta(nd^{2})$ time, $\Theta(nd)$ memory.

(b) Attention's quadratic term dominates the linear-in-$n$ terms when

$$
n^{2}d \gtrsim nd^{2} \iff n \gtrsim d
$$

So the crossover is at sequence length approximately equal to model width.

(c) With $d = 4096$: at $n = 2048$ we have $n \lt d$, so the model is FFN/projection-dominated — the score matmuls account for roughly $n/d = 1/2$ of the projection cost. At $n = 32768$ we have $n/d = 8$, so score computation costs about $8\times$ the projections and dominates decisively; the score matrix alone needs $n^{2} = 1.07\times10^{9}$ entries per head per layer, which is why it is never materialized (FlashAttention tiles it, keeping memory $\Theta(n)$ while leaving the $\Theta(n^{2}d)$ FLOPs unchanged).

$$
\boxed{\text{time } \Theta(nd^{2} + n^{2}d); \text{ attention dominates when } n \gtrsim d}
$$

*Key takeaway:* Two terms, one linear and one quadratic in $n$, with a crossover at $n \approx d$ — this single inequality organizes the entire long-context literature.

### Problem L2.2: Why Backpropagation Is Cheap

(a) What would gradient computation cost if done by finite differences on $N$ parameters? (b) State the Baur–Strassen result and the actual cost of reverse-mode autodiff. (c) What is the memory cost, and why is it the binding constraint?

**Solution:**

(a) A finite-difference gradient perturbs each parameter separately, requiring $N+1$ forward passes. If a forward pass costs $\Theta(F)$, the gradient costs $\Theta(NF)$. For $N = 10^{9}$ this is a billion forward passes per training step — utterly infeasible, quite apart from the truncation and cancellation errors of finite differences.

(b) **Baur–Strassen / cheap gradient principle.** If a scalar function is computed by an arithmetic circuit of size $s$, then *all* $N$ partial derivatives can be computed by a circuit of size $O(s)$ — with a small constant, typically between $2$ and $4$. Reverse-mode autodiff realizes this: one forward pass records the values, one backward pass propagates adjoints, and each forward operation contributes a constant number of backward operations. So

$$
\text{cost(gradient)} = \Theta(\text{cost(forward)}), \text{ not } \Theta(N \cdot \text{cost(forward)})
$$

In practice a training step is about $3\times$ a forward pass, giving the familiar $6N$ FLOPs per token against $2N$ for inference.

(c) The catch is memory: the backward pass needs the intermediate activations, so storage is $\Theta(L\,b\,n\,d)$ for $L$ layers, batch $b$, sequence $n$, width $d$ — linear in depth rather than constant. Since accelerator memory grows more slowly than useful model size, activations, not FLOPs, are usually what caps batch size and sequence length. Gradient checkpointing trades time for memory (Problem L2.6).

$$
\boxed{\text{Gradient time } = \Theta(\text{forward}); \text{ gradient memory } = \Theta(L\,b\,n\,d)}
$$

*Key takeaway:* Reverse mode makes the gradient asymptotically free in time but expensive in memory — the reason deep learning is possible at all, and the reason memory-saving tricks are ubiquitous.

### Problem L2.3: Training Compute from Parameter Count

Using the $C \approx 6ND$ rule (FLOPs $\approx 6 \times$ parameters $\times$ training tokens): (a) justify the factor $6$; (b) estimate the compute to train a $7$B model on $1.4$T tokens; (c) at $40\%$ utilization of $10^{15}$ FLOP/s accelerators, how many device-days is that?

**Solution:**

(a) Each parameter participates in one multiply and one add per token in the forward pass: $2N$ FLOPs per token. The backward pass computes gradients with respect to both the inputs and the weights of each layer, costing about twice the forward pass: $4N$. Total $\approx 6N$ FLOPs per token, hence $C \approx 6ND$ over $D$ tokens. (This counts the matmul terms only; attention's $\Theta(n^{2}d)$ score work is extra and becomes non-negligible when $n \gtrsim d$.)

(b) With $N = 7\times10^{9}$ and $D = 1.4\times10^{12}$:

$$
C = 6 \cdot 7\times10^{9}\cdot 1.4\times10^{12} = 5.88\times10^{22}\ \text{FLOPs}
$$

(c) Effective throughput is $0.4 \times 10^{15} = 4\times10^{14}$ FLOP/s, so

$$
t = \frac{5.88\times10^{22}}{4\times10^{14}} = 1.47\times10^{8}\ \text{s} \approx 1700\ \text{device-days}
$$

i.e. about $7$ days on $256$ devices. Note how the estimate scales: doubling both $N$ and $D$ — the compute-optimal direction under Chinchilla scaling — quadruples the cost.

$$
\boxed{C \approx 6ND = 5.9\times10^{22}\ \text{FLOPs} \approx 1700 \text{ device-days at } 40\% \text{ utilization}}
$$

*Key takeaway:* A single product-rule count plus one asymptotic constant lets you price a training run to within a factor of two before writing any code — the calculation behind every scaling-law study.

### Problem L2.4: Nearest Neighbours and the Curse of Dimensionality

(a) Give the cost of exact $k$-NN over $m$ points in $d$ dimensions. (b) Why do KD-trees stop helping as $d$ grows? (c) What do LSH and HNSW achieve asymptotically, and what is given up?

**Solution:**

(a) Brute force computes $m$ distances of $\Theta(d)$ each, then selects the $k$ smallest with a heap: $\Theta(md + m\log k)$ time and $\Theta(md)$ memory for the dataset. Batched as a matrix product against $q$ queries it is $\Theta(qmd)$ — one big matmul, which is why GPU brute force is competitive up to surprisingly large $m$.

(b) A KD-tree partitions space by axis-aligned splits and prunes branches whose bounding region is farther than the current best. The pruning succeeds only when the query ball intersects few cells. In $d$ dimensions a ball of radius $r$ intersects a number of cells growing like $2^{d}$ in the worst case, so once $d \gtrsim \log_2 m$ essentially every cell must be examined and the query degrades to $\Theta(md)$ — with worse constants than brute force, because of pointer chasing and poor locality. Concentration of distances compounds the problem: for many distributions the ratio of farthest to nearest distance tends to $1$ as $d$ grows, so "nearest" becomes less meaningful and pruning has nothing to exploit.

(c) *LSH* hashes points so that near points collide with higher probability, giving query time $O(m^{\rho}d)$ with $\rho \lt 1$ depending on the approximation factor. *HNSW* builds a navigable small-world graph with empirical query time $O(\log m)$ times a degree factor, at $\Theta(m \cdot \text{degree})$ index memory and $\Theta(m\log m)$ build time. Both give up exactness: they return the true nearest neighbour only with high probability, measured as recall@k. In retrieval-augmented systems that trade is entirely acceptable, and it is the reason vector databases can serve billions of embeddings.

$$
\boxed{\text{exact } \Theta(md);\quad \text{KD-tree} \to \Theta(md) \text{ once } d \gtrsim \log m;\quad \text{ANN } O(m^{\rho}d) \text{ or } O(\log m)}
$$

*Key takeaway:* An index that helps in low dimensions can be asymptotically useless in high ones; when exact sub-linear search is impossible, the escape is to weaken the problem — approximate answers with probabilistic guarantees.

### Problem L2.5: Amortized Cost of a Replay Buffer

A reinforcement-learning replay buffer is implemented as a dynamic array that doubles when full. (a) Prove that $m$ appends cost $\Theta(m)$. (b) What happens if it grows by a fixed $1000$ slots instead? (c) Why might a latency-sensitive system reject amortized guarantees?

**Solution:**

(a) Use the potential method with $\Phi = 2\,\mathrm{size} - \mathrm{capacity} \ge 0$ (valid because the array is at least half full after every resize).

- *Non-resizing append:* actual cost $1$, $\Delta\Phi = 2$, so $\hat{c} = 3$.
- *Resizing append* at size $s$ (capacity $s$ before, $2s$ after): actual cost $s+1$, and $\Delta\Phi = [2(s+1) - 2s] - [2s - s] = 2 - s$, so $\hat{c} = (s+1) + (2-s) = 3$.

Every operation has amortized cost $3$; since $\Phi(D_0) = 0$ and $\Phi \ge 0$, the total actual cost of $m$ appends is at most $3m = \Theta(m)$, and it is $\Omega(m)$ trivially.

(b) With fixed growth $g = 1000$, resizes occur at sizes $g, 2g, 3g, \ldots$ and copy that many elements. Over $m$ appends there are $m/g$ resizes costing

$$
\sum_{j=1}^{m/g} jg = g\cdot\frac{(m/g)(m/g+1)}{2} = \Theta\Bigl(\frac{m^{2}}{g}\Bigr)
$$

Quadratic total, i.e. $\Theta(m/g)$ amortized per append — the constant $g$ delays the pain but does not change the asymptotics. Geometric growth is what makes the copy costs a convergent geometric series.

(c) An amortized bound guarantees the *total*, not any individual operation: one unlucky append still stalls for $\Theta(m)$ while the whole buffer is copied. For a real-time inference or control loop with a p99 latency budget, that tail is fatal. The remedy is *de-amortization*: incremental resizing that copies a few elements per operation, converting the $O(1)$ amortized bound into an $O(1)$ worst-case one at the cost of extra memory and bookkeeping.

$$
\boxed{\text{doubling: } \Theta(m) \text{ total, } O(1) \text{ amortized}; \quad \text{fixed growth: } \Theta(m^{2}/g)}
$$

*Key takeaway:* Amortized analysis justifies multiplicative growth everywhere in systems code, but it is a throughput guarantee, not a latency guarantee — know which one your application needs.

### Problem L2.6: Gradient Checkpointing and the $\sqrt{L}$ Optimum

Training an $L$-layer network stores all activations for the backward pass. Checkpointing stores activations only at $k$ evenly spaced boundaries and recomputes within each segment. (a) Write memory and extra-time as functions of $k$. (b) Minimize memory over $k$. (c) State the resulting trade-off.

**Solution:**

(a) With $k$ checkpoints, each segment has $L/k$ layers. Memory must hold the $k$ checkpoints plus, transiently, one segment's activations during its recomputation:

$$
M(k) = \Theta\Bigl(k + \frac{L}{k}\Bigr) \times (\text{per-layer activation size})
$$

Extra time is one additional forward pass over each segment during the backward sweep, i.e. $\Theta(L)$ extra layer-evaluations — about one extra forward pass in total, independent of $k$.

(b) Minimize $g(k) = k + L/k$ over real $k \gt 0$: $g'(k) = 1 - L/k^{2} = 0$ gives

$$
k^{*} = \sqrt{L}, \qquad g(k^{*}) = 2\sqrt{L}
$$

(By AM–GM, $k + L/k \ge 2\sqrt{L}$ with equality iff $k = \sqrt{L}$ — the same inequality proved in Topic 03.) So memory drops from $\Theta(L)$ to $\Theta(\sqrt{L})$.

(c) For $L = 100$: memory falls by a factor of about $100/(2\cdot 10) = 5$, while training time rises by roughly $33\%$ (one extra forward within a $\approx 3\times$ forward-cost step). The trade is worth taking whenever memory, not compute, caps the batch size or sequence length — which is the usual situation for large models.

$$
\boxed{M = \Theta(\sqrt{L}) \text{ at } k^{*} = \sqrt{L}, \text{ for } \approx 33\% \text{ extra time}}
$$

*Key takeaway:* Time–memory trade-offs are optimization problems in disguise; minimizing $k + L/k$ by AM–GM gives the $\sqrt{L}$ rule that every checkpointing implementation uses.

### Problem L3.1: The Comparison-Sorting Lower Bound

Prove that any deterministic comparison-based sorting algorithm makes $\Omega(n\log n)$ comparisons in the worst case, and explain precisely why counting sort does not contradict it.

**Solution:**

*Model.* Fix $n$ and represent the algorithm's execution as a binary **decision tree**: each internal node is a comparison "$a_i \le a_j$?", the two edges are the answers, and each leaf emits the permutation to apply. The worst-case comparison count is the tree's height $h$.

*Step 1: at least $n!$ leaves.* Suppose two distinct input orderings $\pi \ne \sigma$ reached the same leaf. The algorithm applies one fixed output permutation at that leaf, so it cannot sort both — the outputs would differ. Hence the map from orderings to leaves is injective and there are at least $n!$ leaves. (This is a counting argument of exactly the Topic 05 kind.)

*Step 2: at most $2^{h}$ leaves.* By induction on $h$: a tree of height $0$ has one leaf; a tree of height $h+1$ has at most two subtrees of height $\le h$, hence at most $2^{h+1}$ leaves.

*Step 3: combine and estimate.*

$$
2^{h} \ge n! \implies h \ge \log_{2}(n!) \ge \frac{n}{2}\log_{2}\frac{n}{2} = \frac{n}{2}\log_2 n - \frac{n}{2} = \Omega(n\log n)
$$

using the lower bound on $\log(n!)$ from Problem L1.3. $\blacksquare$

*Tightness.* Merge sort and heapsort make $\Theta(n\log n)$ comparisons, so the bound is attained; the information-theoretic constant $\log_2(n!) \approx n\log_2 n - 1.44n$ is even matched closely by merge insertion sort.

*Why counting sort is not a counterexample.* Counting sort never compares two elements; it uses each key's *value* as an array index, an operation outside the model. Its $\Theta(n+k)$ running time (for keys in $\{0,\ldots,k\}$) buys sub-$n\log n$ behaviour by assuming more about the input than "the keys are comparable". Radix sort likewise decomposes keys into digits. A lower bound constrains a *model of computation*, never a problem in the abstract — changing the allowed operations changes the bound.

$$
\boxed{h \ge \log_2 (n!) = \Omega(n\log n); \text{ counting sort escapes by leaving the comparison model}}
$$

*Key takeaway:* Lower bounds have the universal shape "count the distinguishable answers, bound the information one operation reveals, divide" — and they are always statements about a model, so the first question to ask of any "impossible" result is which operations it forbids.

### Problem L3.2: The Master Theorem's Gaps

Give two recurrences that no case of the three-case Master theorem resolves, solve each by another method, and state the general lesson.

**Solution:**

**Gap 1: $T(n) = 2T(n/2) + n\log n$.** Here $a = 2$, $b = 2$, $\alpha = 1$. The ratio $f(n)/n^{\alpha} = \log n$ grows, so Case 1 fails; it is not $\Theta(n)$, so the basic Case 2 fails; and it is not $\Omega(n^{1+\epsilon})$ for any $\epsilon \gt 0$, since $\log n = o(n^{\epsilon})$, so Case 3 fails. The gap is *polynomially* narrow.

*Recursion tree.* Level $j$ has $2^{j}$ nodes of size $n/2^{j}$, contributing

$$
2^{j}\cdot\frac{n}{2^{j}}\log\frac{n}{2^{j}} = n(\log_2 n - j)
$$

Summing over $j = 0, \ldots, \log_2 n - 1$:

$$
T(n) = \Theta\Bigl(n\sum_{t=1}^{\log_2 n} t\Bigr) = \Theta\bigl(n\log^{2}n\bigr)
$$

**Gap 2: $T(n) = 2T(n/2) + n/\log n$.** Now $f(n)/n^{\alpha} = 1/\log n$ *shrinks*, but not polynomially, so Case 1 also fails.

*Recursion tree.* Level $j$ contributes $2^{j}\cdot\frac{n/2^{j}}{\log(n/2^{j})} = \frac{n}{\log_2 n - j}$. Summing over $j = 0,\ldots,\log_2 n - 1$ gives $n\sum_{t=1}^{\log_2 n}\frac{1}{t} = nH_{\log_2 n} = \Theta(n\log\log n)$. Adding the leaf work $\Theta(n)$:

$$
T(n) = \Theta(n\log\log n)
$$

**Two further failure modes.** (i) *Regularity violation:* Case 3 needs $af(n/b) \le cf(n)$ with $c \lt 1$; an oscillating $f$ can satisfy $f = \Omega(n^{\alpha+\epsilon})$ yet fail regularity, and then no case applies. (ii) *Unequal splits:* $T(n) = T(n/3) + T(2n/3) + n$ is not of the form $aT(n/b) + f(n)$ at all; the **Akra–Bazzi** theorem covers it, giving $\Theta(n\log n)$ (see Problem L1.6).

*General lesson.* The Master theorem is a lookup table generated by one recursion-tree computation. When $f$ sits within a polylogarithmic factor of $n^{\log_b a}$, the table has no entry and you must redo the computation by hand — which is always possible, since the tree argument itself never fails.

$$
\boxed{2T(n/2) + n\log n = \Theta(n\log^{2}n); \qquad 2T(n/2) + n/\log n = \Theta(n\log\log n)}
$$

*Key takeaway:* Memorize the theorem for speed, but know the recursion tree for correctness — the gaps are exactly where the geometric series stops being geometric.

### Problem L3.3: Substitution Traps — Two Failed Proofs and Their Repairs

(a) "Prove" $T(n) = 2T(n/2) + n$ is $O(n)$ and locate the error. (b) Show that the correct bound $T(n) = 2T(n/2) + 1$ is $O(n)$ resists a naive substitution, and repair it.

**Solution:**

**(a) The false proof.** Guess $T(n) \le cn$. Then

$$
T(n) \le 2\Bigl(c\frac{n}{2}\Bigr) + n = cn + n = (c+1)n = O(n) \quad \text{(claimed)}
$$

*The error.* The inductive step must re-establish the hypothesis **in exactly the assumed form, with the same constant $c$**. What was derived is $T(n) \le (c+1)n$, not $T(n) \le cn$; the constant grows by $1$ at every level, so after $\log_2 n$ levels it has grown to $c + \log_2 n$ — precisely the missing logarithmic factor. Concealing this by writing "$= O(n)$" mid-induction is the classic error, and here it would "prove" a statement contradicted by the true answer $\Theta(n\log n)$ from the recursion tree.

*The moral:* never let asymptotic notation appear inside an inductive step. Carry explicit constants, and demand a literal match.

**(b) A true bound that resists naive substitution.** For $T(n) = 2T(n/2) + 1$ with $T(1) = 1$ the answer really is $\Theta(n)$ (the tree has $n$ leaves and $n-1$ internal nodes each doing $O(1)$ work). Yet guessing $T(n) \le cn$ gives

$$
T(n) \le 2\Bigl(c\frac{n}{2}\Bigr) + 1 = cn + 1 \not\le cn
$$

The step fails by an *additive constant* — a shortfall that does not grow, unlike case (a).

*The repair: strengthen the hypothesis.* Guess the stronger $T(n) \le cn - d$ with $d \gt 0$. Then

$$
T(n) \le 2\Bigl(c\frac{n}{2} - d\Bigr) + 1 = cn - 2d + 1 \le cn - d \iff d \ge 1
$$

The induction now closes. Base case: $T(1) = 1 \le c - d$ holds for $c \ge 2$, $d = 1$. Hence $T(n) \le 2n - 1 = O(n)$ — and indeed $T(n) = 2n-1$ exactly.

*Why strengthening works.* A stronger claim gives a stronger inductive hypothesis, and the extra $-d$ is exactly the reserve needed to absorb the additive $+1$. The diagnostic is the *kind* of failure: a shortfall by a constant means strengthen; a shortfall by a factor that compounds per level means the guess itself is wrong.

$$
\boxed{\text{(a) } T(n) = \Theta(n\log n), \text{ the } O(n) \text{ "proof" leaks a constant per level;} \quad \text{(b) use } T(n) \le cn - d}
$$

*Key takeaway:* Two failure modes, two responses. Growing shortfall means the guess is wrong; constant shortfall means the guess is right but too weak — subtract a lower-order term and try again.

### Problem L3.4: A Potential Function for Shrinking Dynamic Arrays

A dynamic array doubles on overflow and halves when it drops below one *quarter* full. (a) Show that halving at one *half* full admits a sequence with $\Theta(m^{2})$ total cost. (b) Design a potential function proving $O(1)$ amortized cost for the quarter rule.

**Solution:**

**(a) The half-full rule fails.** Suppose the array halves as soon as size falls below capacity$/2$. Fill it to exactly capacity $s = 2^{j}$, then alternate append, delete, append, delete. The append overflows and triggers a doubling costing $\Theta(s)$; the immediately following delete drops the size below half and triggers a halving costing $\Theta(s)$; and the state returns to the start. Each pair of operations costs $\Theta(s)$, so $m$ operations cost

$$
\Theta\bigl(m \cdot s\bigr) = \Theta(m^{2}) \quad \text{when } s = \Theta(m)
$$

The failure is *thrashing*: the resize thresholds have no slack, so the structure sits exactly on a boundary.

**(b) The quarter rule with a potential.** After a resize the array is exactly half full, so the load factor $\lambda = \mathrm{size}/\mathrm{capacity}$ has slack on both sides — $\lambda$ must travel from $1/2$ to $1$ to trigger a doubling, or from $1/2$ down to $1/4$ to trigger a halving. Either way $\Theta(\mathrm{capacity})$ operations must occur first, which pays for the next resize. Formalize with

$$
\Phi = \begin{cases} 2\,\mathrm{size} - \mathrm{capacity}, & \lambda \ge \tfrac{1}{2} \\[4pt] \tfrac{\mathrm{capacity}}{2} - \mathrm{size}, & \lambda \lt \tfrac{1}{2} \end{cases}
$$

$\Phi \ge 0$ always, and $\Phi = 0$ exactly at $\lambda = 1/2$, immediately after any resize.

*Append, $\lambda \ge 1/2$, no resize:* $\hat{c} = 1 + 2 = 3$.
*Append triggering a doubling* at size $s$ = capacity $s$: actual $s+1$, $\Delta\Phi = [2(s+1) - 2s] - [2s - s] = 2 - s$, so $\hat{c} = 3$.
*Delete, $\lambda \lt 1/2$, no resize:* size falls by $1$, so $\Phi$ rises by $1$: $\hat{c} = 1 + 1 = 2$.
*Delete triggering a halving* at size $s = c/4$ (capacity $c$ before, $c/2$ after): actual cost $s$ (copy the survivors, $s - 1$ elements, plus the delete). Before, $\Phi = c/2 - s = 2s - s = s$; after, size $s-1$ and capacity $2s$, so $\lambda \lt 1/2$ and $\Phi = s - (s-1) = 1$. Hence $\Delta\Phi = 1 - s$ and

$$
\hat{c} = s + (1 - s) = 1
$$

Every operation has $\hat{c} \le 3$, and with $\Phi(D_0) = 0$, $\Phi \ge 0$, Definition 2.9 gives total actual cost $O(m)$ for any sequence of $m$ appends and deletes. $\blacksquare$

$$
\boxed{\text{Halve at } \tfrac{1}{4} \text{ full: } O(1) \text{ amortized. Halve at } \tfrac{1}{2}: \Theta(m^{2}) \text{ worst case}}
$$

*Key takeaway:* Amortized guarantees depend on *hysteresis* — a gap between the grow and shrink thresholds — and the potential function is precisely the ledger recording how much slack the structure currently holds.
