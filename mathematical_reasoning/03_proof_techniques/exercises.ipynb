## Level 0 — Concept Check

### Problem L0.1: Name That Technique

For each claim, name the most natural proof technique and justify the choice in one sentence: (a) "There is no smallest positive rational number." (b) "If $n$ is odd, then $n^2 + 1$ is even." (c) "Every real number is either $\geq 0$ or $\lt 0$, and in both cases $x^2 \geq 0$." (d) "Not every bounded sequence converges."

**Solution:**

(a) **Contradiction.** Assuming a smallest positive rational $r$ exists hands us a concrete object to destroy: $r/2$ is a smaller positive rational.

(b) **Direct proof.** The hypothesis unpacks into the equation $n = 2k+1$, from which $n^2 + 1 = 4k^2 + 4k + 2 = 2(2k^2+2k+1)$ follows by computation.

(c) **Cases.** The claim itself announces an exhaustive split on the sign of $x$.

(d) **Counterexample.** The claim is the negation of a universal statement; one witness — e.g. $a_n = (-1)^n$, bounded but divergent — refutes "every bounded sequence converges."

$$
\boxed{\text{(a) contradiction, (b) direct, (c) cases, (d) counterexample}}
$$

*Key takeaway:* The grammatical form of a claim (impossibility, implication with usable hypothesis, built-in split, negated universal) usually dictates the technique.

### Problem L0.2: Why Examples Never Prove Universals

The polynomial $f(n) = n^2 + n + 41$ yields primes for $n = 0, 1, 2, \ldots, 39$. Does this prove "$f(n)$ is prime for all natural $n$"? Find the failure and state the logical principle.

**Solution:**

No. Forty verified instances do not establish a universal claim over an infinite domain. In fact:

$$
f(40) = 40^2 + 40 + 41 = 1600 + 81 = 1681 = 41^2
$$

which is composite. (Structurally, $f(40) = 40 \cdot 41 + 41 = 41 \cdot 41$ — the construction makes the failure inevitable.)

The principle: $P(0), \ldots, P(39)$ do not entail $\forall n\,P(n)$; only a proof about an arbitrary $n$ does. Conversely, the single witness $n = 40$ *completely* disproves the universal claim, by $\neg\forall \equiv \exists\neg$.

$$
\boxed{f(40) = 41^2 \text{ is composite — the universal claim is false}}
$$

*Key takeaway:* Evidence and proof are asymmetric: no number of confirming examples proves $\forall$, while one counterexample refutes it.

### Problem L0.3: The Anatomy of a Contradiction Proof

In a proof by contradiction of the claim "$p$", what exactly is assumed, what must be derived, and why does this establish $p$?

**Solution:**

**Assumed:** $\neg p$, the negation of the claim (for an implication $p \to q$, the negation is $p \wedge \neg q$ — hypothesis true *and* conclusion false).

**Derived:** any absurdity — a statement of the form $r \wedge \neg r$, or a violation of a known theorem or definition (e.g. "$\gcd(p,q) = 1$ and $\gcd(p,q) \geq 2$").

**Why it works:** the derivation shows $\neg p \to \text{F}$. An implication with false conclusion and true premise cannot be true, so if the derivation is valid, $\neg p$ must be false; by double negation $p$ is true. The tautology $(\neg p \to \text{F}) \to p$ is verified by a two-row truth table.

$$
\boxed{\text{Assume } \neg p, \text{ derive } r \wedge \neg r; \text{ soundness rests on } (\neg p \to \text{F}) \to p}
$$

*Key takeaway:* A contradiction proof is a controlled demolition: the negated claim is the load-bearing assumption, and any absurdity brings it down.

### Problem L0.4: Existence versus Uniqueness

The equation $x^2 = 4$ over $\mathbb{R}$: does a solution exist? Is it unique? What two separate obligations does a proof of "$\exists! x\,P(x)$" carry?

**Solution:**

*Existence:* yes — $x = 2$ satisfies $2^2 = 4$.

*Uniqueness:* no — $x = -2$ also works, and $2 \neq -2$. So $\exists x\,(x^2 = 4)$ is true while $\exists!\,x\,(x^2=4)$ is false.

A proof of $\exists! x\,P(x)$ must deliver:

1. **Existence:** exhibit (or otherwise establish) some $a$ with $P(a)$.
2. **Uniqueness:** assume $P(a)$ and $P(b)$ for arbitrary $a, b$ and derive $a = b$.

Neither part implies the other: $x^2 = 4$ has existence without uniqueness; $x^2 = -4$ over $\mathbb{R}$ has (vacuous) uniqueness without existence.

$$
\boxed{x^2 = 4 \text{ has two real solutions: existence holds, uniqueness fails}}
$$

*Key takeaway:* "Exactly one" is two theorems in one; prove each with its own template.

## Level 1 — Foundation

### Problem L1.1: Direct Proof with Divisibility

Prove: if $a \mid b$ and $b \mid c$, then $a \mid c$ (transitivity of divisibility).

**Solution:**

Assume $a \mid b$ and $b \mid c$. By definition there exist integers $m, n$ with:

$$
b = am, \qquad c = bn
$$

Substitute the first equation into the second:

$$
c = (am)n = a(mn)
$$

Since $mn \in \mathbb{Z}$, the definition of divisibility gives $a \mid c$. $\blacksquare$

$$
\boxed{a \mid b \wedge b \mid c \implies a \mid c}
$$

*Key takeaway:* Divisibility hypotheses convert to equations with fresh witness variables ($m$ and $n$ must be distinct letters); substitution plus re-verifying the definition finishes the proof.

### Problem L1.2: Contrapositive with Divisibility by 3

Prove: for every integer $n$, if $3 \mid n^2$ then $3 \mid n$.

**Solution:**

We prove the contrapositive: if $3 \nmid n$, then $3 \nmid n^2$.

Assume $3 \nmid n$. By the division algorithm, $n = 3k + r$ with $r \in \{1, 2\}$ — two cases.

*Case $r = 1$:* $n^2 = 9k^2 + 6k + 1 = 3(3k^2 + 2k) + 1$, remainder 1, so $3 \nmid n^2$.

*Case $r = 2$:* $n^2 = 9k^2 + 12k + 4 = 3(3k^2 + 4k + 1) + 1$, remainder 1, so $3 \nmid n^2$.

In both cases $3 \nmid n^2$, proving the contrapositive, hence the claim. $\blacksquare$

$$
\boxed{3 \mid n^2 \implies 3 \mid n}
$$

*Key takeaway:* Contrapositive and cases compose naturally: the negated conclusion "$3 \nmid n$" splits into finitely many concrete remainders, each yielding to computation. This lemma is the engine of the $\sqrt{3}$ irrationality proof.

### Problem L1.3: Irrationality of $\sqrt{3}$

Prove that $\sqrt{3}$ is irrational.

**Solution:**

Suppose for contradiction $\sqrt{3} = p/q$ with $p, q \in \mathbb{Z}$, $q \neq 0$, $\gcd(p,q) = 1$. Squaring:

$$
3q^2 = p^2
$$

So $3 \mid p^2$, and by Problem L1.2, $3 \mid p$: write $p = 3k$. Substituting:

$$
3q^2 = 9k^2 \implies q^2 = 3k^2
$$

So $3 \mid q^2$, hence $3 \mid q$ by the same lemma. Then $3$ divides both $p$ and $q$, so $\gcd(p,q) \geq 3$, contradicting $\gcd(p,q) = 1$. Therefore $\sqrt{3}$ is irrational. $\blacksquare$

$$
\boxed{\sqrt{3} \notin \mathbb{Q}}
$$

*Key takeaway:* The $\sqrt{2}$ proof is a machine: swap in the lemma "$3 \mid n^2 \Rightarrow 3 \mid n$" and it runs verbatim for 3 — and for any prime $p$ (Problem L3.4).

### Problem L1.4: Proof by Cases with Absolute Values

Prove that for all real $x$: $\lvert x - 1 \rvert + \lvert x + 1 \rvert \geq 2$.

**Solution:**

Split $\mathbb{R}$ into three exhaustive cases by the breakpoints $-1$ and $1$.

*Case 1: $x \geq 1$.* Both arguments are nonnegative: the sum is $(x-1) + (x+1) = 2x \geq 2$.

*Case 2: $-1 \leq x \lt 1$.* Then $x - 1 \lt 0 \leq x + 1$: the sum is $(1-x) + (x+1) = 2$.

*Case 3: $x \lt -1$.* Both arguments are negative: the sum is $(1-x) + (-x-1) = -2x \gt 2$.

In every case the sum is $\geq 2$, with equality exactly on $[-1, 1]$. $\blacksquare$

(Alternative one-liner via the triangle inequality: $2 = \lvert (x+1) - (x-1) \rvert \leq \lvert x+1 \rvert + \lvert x-1 \rvert$.)

$$
\boxed{\lvert x-1 \rvert + \lvert x+1 \rvert \geq 2 \text{ for all } x \in \mathbb{R}}
$$

*Key takeaway:* Absolute-value inequalities split at the points where each argument changes sign; always confirm the cases cover the whole line — and look back for a slicker proof afterward (Pólya's fourth step).

### Problem L1.5: Uniqueness of the Multiplicative Inverse

Prove: every real $x \neq 0$ has exactly one multiplicative inverse.

**Solution:**

*Existence.* For $x \neq 0$, the real number $y = 1/x$ satisfies $xy = 1$ (this is the defining property of division in $\mathbb{R}$, guaranteed by the field axioms).

*Uniqueness.* Suppose $xy_1 = 1$ and $xy_2 = 1$. Then:

$$
y_1 = y_1 \cdot 1 = y_1 (x y_2) = (y_1 x) y_2 = (x y_1) y_2 = 1 \cdot y_2 = y_2
$$

using associativity and commutativity of multiplication. Hence any two inverses coincide. $\blacksquare$

$$
\boxed{\forall x \neq 0\ \exists!\,y:\ xy = 1}
$$

*Key takeaway:* The two-solutions-collide computation uses only the axioms — the same argument proves uniqueness of inverses in any group, of the identity element, and of limits of sequences.

### Problem L1.6: Disproof by Counterexample

Disprove each claim with an explicit counterexample: (a) "For all reals $a, b$: $\sqrt{a^2 + b^2} = a + b$." (b) "Every continuous function is differentiable." (c) "For all positive integers $n$, $2^n \gt n^2$."

**Solution:**

(a) Take $a = b = 1$: $\sqrt{1+1} = \sqrt{2} \approx 1.414$, but $a + b = 2$. Since $\sqrt{2} \neq 2$, the claim fails. (It holds only when $ab = 0$ with $a+b \geq 0$.)

(b) $f(x) = \lvert x \rvert$ is continuous on all of $\mathbb{R}$, but at $0$ the difference quotients from the right tend to $+1$ and from the left to $-1$, so $f'(0)$ does not exist.

(c) $n = 2$: $2^2 = 4 = 2^2$, and the claimed strict inequality $4 \gt 4$ fails. ($n = 3$ also fails: $8 \lt 9$; and $n=4$ gives equality $16 = 16$. The inequality is true only for $n = 1$ and $n \geq 5$ — proved by induction in Topic 04.)

$$
\boxed{\text{(a) } a=b=1; \quad \text{(b) } f(x) = \lvert x \rvert \text{ at } 0; \quad \text{(c) } n = 2}
$$

*Key takeaway:* Hunt for counterexamples at small values, boundary values, and symmetric configurations first — most false claims die there.

## Level 2 — Applications in AI/ML & CS

### Problem L2.1: Loop Invariant for Array Summation

Consider the loop: `s = 0; for i in range(n): s = s + a[i]`. State a loop invariant and prove by the initialization–maintenance–termination pattern that the final value of `s` is $\sum_{j=0}^{n-1} a_j$.

**Solution:**

**Invariant $I(i)$:** at the start of the iteration with index $i$, the accumulator satisfies $s = \sum_{j=0}^{i-1} a_j$ (the empty sum for $i = 0$).

*Initialization:* before the first iteration $i = 0$ and $s = 0 = \sum_{j=0}^{-1} a_j$, the empty sum. Invariant holds.

*Maintenance:* assume $s = \sum_{j=0}^{i-1} a_j$ at the start of iteration $i$. The body sets $s \leftarrow s + a_i = \sum_{j=0}^{i-1} a_j + a_i = \sum_{j=0}^{i} a_j$, which is exactly $I(i+1)$.

*Termination:* the loop ends after the iteration with $i = n-1$, i.e. when the "next" index is $n$; the invariant $I(n)$ gives $s = \sum_{j=0}^{n-1} a_j$, the specification. $\blacksquare$

$$
\boxed{s_{\text{final}} = \sum_{j=0}^{n-1} a_j}
$$

*Key takeaway:* A loop invariant argument is a direct proof stitched around a loop: initialization is the base fact, maintenance is one implication, termination converts the invariant into the postcondition.

### Problem L2.2: Termination of Euclid's Algorithm

Euclid's algorithm maps $(a, b) \mapsto (b, a \bmod b)$ for $b \gt 0$, stopping when $b = 0$. Prove it terminates for all inputs $a \geq 0$, $b \geq 0$.

**Solution:**

Consider the second component along the run: $b_0 = b$, $b_1 = a \bmod b_0$, $b_2 = b_0 \bmod b_1, \ldots$ (while nonzero).

**The measure decreases strictly.** By definition of remainder, $0 \leq a \bmod b \lt b$ whenever $b \gt 0$. Hence $b_{k+1} = b_{k-1} \bmod b_k \lt b_k$: the sequence of second components is a strictly decreasing sequence of nonnegative integers as long as the algorithm runs.

**No infinite descent.** Suppose the algorithm ran forever. Then $b_0 \gt b_1 \gt b_2 \gt \cdots$ would be an infinite strictly decreasing sequence of nonnegative integers, so $b_k \leq b_0 - k$ for all $k$, and taking $k = b_0 + 1$ gives $b_k \lt 0$ — contradicting nonnegativity. Hence the algorithm reaches $b = 0$ in at most $b_0 + 1$ steps and terminates. $\blacksquare$

(The true bound is much better — $O(\log b)$ steps, by Lamé's theorem — but termination needs only the descent.)

$$
\boxed{\text{Euclid's algorithm terminates: its measure } b_k \text{ strictly descends in } \mathbb{N}}
$$

*Key takeaway:* Termination proofs pick a nonnegative integer measure that strictly decreases; contradiction against infinite descent does the rest. This is the well-ordering principle in action (Topic 04).

### Problem L2.3: Floating-Point Associativity — Refuted

Disprove: "IEEE-754 double-precision addition is associative," and explain the consequence for summing losses over a large dataset.

**Solution:**

**Counterexample.** In double precision the unit roundoff makes $1$ negligible next to $10^{16}$: evaluating with rounding after each operation,

$$
(10^{16} + 1) - 10^{16} = 10^{16} - 10^{16} = 0
$$

because $10^{16} + 1$ rounds back to $10^{16}$ (the next representable double after $10^{16}$ is $10^{16} + 2$), while:

$$
10^{16} + (1 - 10^{16}) = 10^{16} - (10^{16} - 1) = 1
$$

since $10^{16} - 1$ is exactly representable. The two orderings give $0 \neq 1$, so associativity fails. $\blacksquare$

**Consequence.** Summing $10^{9}$ per-example losses in different orders (different batchings, parallel reductions) gives different results; small terms added to a large accumulator vanish. Remedies: compensated (Kahan) summation, pairwise/tree reduction, or higher precision accumulators — all standard in deep-learning frameworks' reduction kernels.

$$
\boxed{(10^{16} + 1) - 10^{16} = 0 \neq 1 = 10^{16} + (1 - 10^{16})}
$$

*Key takeaway:* A single well-chosen triple refutes a "law" — and explains real nondeterminism in ML training runs.

### Problem L2.4: Gradient Descent Needs Hypotheses

(a) Give a counterexample showing gradient descent with a fixed step can *increase* the loss. (b) Prove: if $f$ is $L$-smooth ($\Vert \nabla f(x) - \nabla f(y) \Vert \leq L \Vert x - y \Vert$) and $\eta \leq 1/L$, one step $x^{+} = x - \eta \nabla f(x)$ satisfies $f(x^{+}) \leq f(x) - \frac{\eta}{2} \Vert \nabla f(x) \Vert^2$.

**Solution:**

**(a) Counterexample.** $f(x) = \frac{1}{2}Lx^2$ with $L = 2$, step $\eta = 1.1$: the update is $x^{+} = x - 1.1 \cdot 2x = -1.2x$, so $\lvert x^{+} \rvert = 1.2 \lvert x \rvert$ and $f(x^{+}) = 1.44\,f(x) \gt f(x)$ for $x \neq 0$. The iterates diverge geometrically.

**(b) Proof.** $L$-smoothness implies the descent lemma (integrate the gradient along the segment):

$$
f(y) \leq f(x) + \nabla f(x)^{\top}(y - x) + \frac{L}{2}\Vert y - x \Vert^2
$$

Substitute $y = x^{+} = x - \eta \nabla f(x)$, so $y - x = -\eta \nabla f(x)$:

$$
f(x^{+}) \leq f(x) - \eta \Vert \nabla f(x) \Vert^2 + \frac{L \eta^2}{2} \Vert \nabla f(x) \Vert^2 = f(x) - \eta\Bigl(1 - \frac{L\eta}{2}\Bigr)\Vert \nabla f(x) \Vert^2
$$

For $\eta \leq 1/L$ we have $1 - \frac{L\eta}{2} \geq \frac{1}{2}$, hence $f(x^{+}) \leq f(x) - \frac{\eta}{2}\Vert \nabla f(x) \Vert^2$. $\blacksquare$

$$
\boxed{\eta \leq \tfrac{1}{L} \implies f(x^{+}) \leq f(x) - \tfrac{\eta}{2}\Vert \nabla f(x) \Vert^2}
$$

*Key takeaway:* The counterexample shows the hypothesis is not decorative; the direct proof shows exactly where $\eta \leq 1/L$ enters. Rigorous ML lives in such (counterexample, conditional theorem) pairs.

### Problem L2.5: Argmax Invariance under Monotone Transforms

Prove: if $g: \mathbb{R} \to \mathbb{R}$ is strictly increasing, then for any function $f: \mathcal{X} \to \mathbb{R}$, the sets of maximizers of $f$ and of $g \circ f$ coincide. Deduce why maximizing log-likelihood is equivalent to maximizing likelihood.

**Solution:**

**Proof.** Let $x^{*}$ maximize $f$: $f(x^{*}) \geq f(x)$ for all $x$. Applying the strictly increasing $g$ preserves order (weakly): $g(f(x^{*})) \geq g(f(x))$ for all $x$, so $x^{*}$ maximizes $g \circ f$.

Conversely, let $x^{*}$ maximize $g \circ f$ and suppose $x^{*}$ did *not* maximize $f$: some $x_0$ has $f(x_0) \gt f(x^{*})$. Strict monotonicity gives $g(f(x_0)) \gt g(f(x^{*}))$, contradicting maximality of $x^{*}$ for $g \circ f$. Hence the maximizer sets are equal. $\blacksquare$

**Deduction.** The likelihood $\mathcal{L}(\theta) \gt 0$, and $g = \ln$ is strictly increasing on $(0, \infty)$; hence:

$$
\arg\max_\theta \mathcal{L}(\theta) = \arg\max_\theta \ln \mathcal{L}(\theta)
$$

so MLE may be computed on the numerically friendlier log-likelihood (sums instead of products). The same argument shows temperature scaling of logits never changes a classifier's top-1 prediction.

$$
\boxed{\arg\max f = \arg\max (g \circ f) \text{ for strictly increasing } g}
$$

*Key takeaway:* A two-direction direct/contradiction argument justifies one of the most-used silent steps in all of statistics and ML.

### Problem L2.6: Uniqueness of the Ridge Regression Solution

Prove that for $\lambda \gt 0$ the ridge objective $J(w) = \Vert Xw - y \Vert^2 + \lambda \Vert w \Vert^2$ has a unique minimizer, while ordinary least squares ($\lambda = 0$) can have infinitely many.

**Solution:**

**Strict convexity.** For any $w_1 \neq w_2$ and $t \in (0,1)$, the map $w \mapsto \Vert Xw - y \Vert^2$ is convex (composition of the affine map $w \mapsto Xw - y$ with the convex squared norm), and $w \mapsto \lambda \Vert w \Vert^2$ is *strictly* convex for $\lambda \gt 0$ because the squared norm is strictly convex. A convex plus a strictly convex function is strictly convex, so $J$ is strictly convex.

**Uniqueness by collision.** Suppose $w_1 \neq w_2$ both minimize $J$ with value $J^{*}$. Strict convexity at $t = \frac{1}{2}$ gives:

$$
J\Bigl(\frac{w_1 + w_2}{2}\Bigr) \lt \frac{1}{2}J(w_1) + \frac{1}{2}J(w_2) = J^{*}
$$

contradicting minimality of $J^{*}$. Hence the minimizer is unique. (Existence holds since $J$ is continuous and coercive: $J(w) \geq \lambda \Vert w \Vert^2 \to \infty$.)

**OLS non-uniqueness.** If $X$ has linearly dependent columns, pick $v \neq 0$ with $Xv = 0$; then $\Vert X(w + tv) - y \Vert^2$ is constant in $t$, so any minimizer generates an affine line of minimizers.

$$
\boxed{\lambda \gt 0 \implies \text{unique ridge solution; collinear } X \text{ with } \lambda = 0 \implies \text{infinitely many OLS solutions}}
$$

*Key takeaway:* Regularization buys uniqueness through strict convexity, proved by the classic two-minimizers-collide contradiction.

## Level 3 — Challenge

### Problem L3.1: Irrational Powers, Rational Result

Prove there exist irrational numbers $a$ and $b$ such that $a^{b}$ is rational, and explain what makes the proof nonconstructive.

**Solution:**

Let $c = \sqrt{2}^{\sqrt{2}}$. Exactly one of two cases holds.

*Case 1: $c \in \mathbb{Q}$.* Choose $a = b = \sqrt{2}$, irrational by the classical proof; then $a^{b} = c$ is rational.

*Case 2: $c \notin \mathbb{Q}$.* Choose $a = c$ (irrational by the case assumption) and $b = \sqrt{2}$; then:

$$
a^{b} = \bigl(\sqrt{2}^{\sqrt{2}}\bigr)^{\sqrt{2}} = \sqrt{2}^{\sqrt{2}\cdot\sqrt{2}} = \sqrt{2}^{2} = 2 \in \mathbb{Q}
$$

Either way the required pair exists. $\blacksquare$

**Nonconstructivity.** The proof splits on a disjunction ($c$ rational or not) without deciding it, so it certifies existence while naming two candidate pairs and refusing to say which works. (Gelfond–Schneider resolves the split — $c$ is transcendental, so Case 2 is the real one — but the elementary proof stands without it. A fully constructive alternative: $a = \sqrt{2}$, $b = \log_2 9$, giving $a^{b} = 3$.)

$$
\boxed{\exists\,a, b \notin \mathbb{Q}: a^{b} \in \mathbb{Q}}
$$

*Key takeaway:* The law of excluded middle lets existence proofs branch on undecided facts — powerful, and philosophically the dividing line between classical and constructive mathematics.

### Problem L3.2: AM–GM for Two Variables, with Equality Cases

Prove that for all nonnegative reals $a, b$: $\frac{a+b}{2} \geq \sqrt{ab}$, with equality iff $a = b$.

**Solution:**

Since squares are nonnegative:

$$
0 \leq (\sqrt{a} - \sqrt{b})^2 = a - 2\sqrt{ab} + b
$$

Rearranging gives $a + b \geq 2\sqrt{ab}$, i.e. $\frac{a+b}{2} \geq \sqrt{ab}$.

**Equality analysis.** Equality in the displayed line holds iff $(\sqrt{a}-\sqrt{b})^2 = 0$ iff $\sqrt{a} = \sqrt{b}$ iff $a = b$ (the square-root function is injective on $[0,\infty)$).

Both directions of the "iff" are explicit: if $a = b$ then both sides equal $a$; if equality holds, the square vanishes forcing $a = b$. $\blacksquare$

$$
\boxed{\frac{a+b}{2} \geq \sqrt{ab}, \text{ with equality iff } a = b}
$$

*Key takeaway:* "Some square is nonnegative" is the most productive one-line lemma in inequality proving; tracking exactly when it vanishes yields the equality case for free.

### Problem L3.3: No Integer Solutions — Cases via Parity

Prove that the equation $a^2 - 4b = 2$ has no integer solutions.

**Solution:**

Suppose for contradiction that integers $a, b$ satisfy $a^2 = 4b + 2 = 2(2b+1)$. Then $a^2$ is even, so $a$ is even (Topic 03, contrapositive lemma): $a = 2k$. Substituting:

$$
4k^2 = 4b + 2 \implies 2k^2 = 2b + 1
$$

The left side is even; the right side is odd. An integer cannot be both even and odd (parity is exclusive by the division algorithm with divisor 2). Contradiction. Hence no integer solutions exist. $\blacksquare$

Equivalently in modular language: squares satisfy $a^2 \equiv 0$ or $1 \pmod 4$, but $4b + 2 \equiv 2 \pmod 4$ — the residues never meet.

$$
\boxed{a^2 - 4b = 2 \text{ has no solutions in } \mathbb{Z}}
$$

*Key takeaway:* Impossibility over the integers is often a residue argument: reduce both sides mod a small number and watch the possible values disagree.

### Problem L3.4: Irrationality of $\sqrt{p}$ for Every Prime $p$

Let $p$ be prime. Prove that $\sqrt{p}$ is irrational, using Euclid's lemma (if a prime divides a product, it divides one of the factors).

**Solution:**

**Lemma (from Euclid's lemma).** If $p \mid n^2$ then $p \mid n$: indeed $p \mid n \cdot n$ forces $p \mid n$ or $p \mid n$ — i.e. $p \mid n$.

**Main proof.** Suppose $\sqrt{p} = a/b$ with $a, b \in \mathbb{Z}$, $b \neq 0$, $\gcd(a,b) = 1$. Then:

$$
p b^2 = a^2
$$

So $p \mid a^2$, hence $p \mid a$ by the lemma: $a = pk$. Substituting:

$$
p b^2 = p^2 k^2 \implies b^2 = p k^2
$$

So $p \mid b^2$, hence $p \mid b$. Then $p$ divides both $a$ and $b$, giving $\gcd(a,b) \geq p \geq 2$, contradicting coprimality. Therefore $\sqrt{p}$ is irrational. $\blacksquare$

**Generality check.** The argument used primality only through Euclid's lemma. For composite non-squares like 12 a refined argument (or unique factorization) is needed, and for perfect squares the statement is false — $\sqrt{4} = 2$ — so the hypothesis is sharp.

$$
\boxed{\sqrt{p} \notin \mathbb{Q} \text{ for every prime } p}
$$

*Key takeaway:* Generalizing a proof means locating the one property it truly used — here, Euclid's lemma — and promoting it to a hypothesis.
