# Topic 03: Proof Techniques

## 1. First-Principles Intuition & Motivation

How can a finite argument settle a claim about infinitely many cases? "For every integer $n$, if $n^2$ is even then $n$ is even" concerns all of $\mathbb{Z}$, yet mathematicians prove it in three lines. The trick is never to look at the integers one at a time: we reason about an **arbitrary** element, using only properties shared by all of them. Whatever we deduce about an arbitrary $n$ holds for each particular one.

A proof technique is a pre-fabricated argument skeleton, and each skeleton is guaranteed sound by a propositional tautology from Topic 01. Direct proof runs on modus ponens. Proof by contrapositive runs on $(p \to q) \equiv (\neg q \to \neg p)$. Proof by contradiction runs on the fact that if $\neg p$ leads to an absurdity, then $p$ must hold. Proof by cases runs on the constructive dilemma. Learning proofs is therefore not learning new logic — it is learning to *deploy* old logic fluently.

The choice of technique is a strategic decision, and there are reliable signals:

- **Direct proof** when the hypothesis unpacks into something you can compute with ($n$ even means $n = 2k$ — an equation!).
- **Contrapositive** when the *negations* are more concrete than the originals ("$n$ odd" gives $n = 2k+1$ to compute with, while "$n^2$ even" gives little grip on $n$).
- **Contradiction** for impossibility and infinitude statements: assuming the negation hands you a concrete object — a fraction equal to $\sqrt{2}$, a finite list of all primes — that you can then destroy.
- **Cases** when the domain splits naturally (even/odd, positive/negative/zero) and each case yields to a short direct argument.
- **Counterexample** when the claim is false: one witness ends the matter.

This module treats the classical proofs as *études* — each chosen because it displays one technique in its purest form.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition 2.1 (Proof).** A *proof* of a statement $S$ from a set of premises is a finite sequence of statements, each of which is a premise, an axiom, a definition, a previously proved theorem, or follows from earlier statements by a valid inference rule, and whose last statement is $S$.

**Definition 2.2 (Even/odd, divisibility, rational).** For integers: $n$ is *even* iff $n = 2k$ for some $k \in \mathbb{Z}$, *odd* iff $n = 2k+1$ for some $k \in \mathbb{Z}$ (every integer is exactly one of the two). For $a \neq 0$: $a \mid b$ ("$a$ divides $b$") iff $b = ac$ for some $c \in \mathbb{Z}$. A real $x$ is *rational* iff $x = p/q$ with $p, q \in \mathbb{Z}$, $q \neq 0$; otherwise *irrational*. A prime is an integer $p \geq 2$ whose only positive divisors are 1 and $p$.

**The six templates**, with their logical justifications:

| Technique | To prove | Template | Justifying tautology |
|---|---|---|---|
| Direct | $p \to q$ | Assume $p$; derive $q$ | Modus ponens chain |
| Contrapositive | $p \to q$ | Assume $\neg q$; derive $\neg p$ | $(p \to q) \equiv (\neg q \to \neg p)$ |
| Contradiction | $p$ | Assume $\neg p$; derive $r \wedge \neg r$ | $(\neg p \to \text{F}) \to p$ |
| Cases | $q$ | Show $p_1 \vee \cdots \vee p_k$ exhaustive; prove each $p_i \to q$ | Constructive dilemma |
| Existence/uniqueness | $\exists! x\,P(x)$ | Exhibit or deduce a witness; assume $P(a) \wedge P(b)$, derive $a = b$ | Definition of $\exists!$ |
| Counterexample | $\neg\forall x\,P(x)$ | Exhibit $x_0$ with $\neg P(x_0)$ | $\neg\forall \equiv \exists\neg$ |

**Theorem statements proved in Section 3.**

1. The sum of two even integers is even. (Direct)
2. If $n^2$ is even, then $n$ is even. (Contrapositive)
3. $\sqrt{2}$ is irrational. (Contradiction)
4. There are infinitely many primes. (Contradiction, Euclid)
5. For every integer $n$, the number $n^2 + n$ is even; and the triangle inequality $\lvert x + y \rvert \leq \lvert x \rvert + \lvert y \rvert$. (Cases)
6. There exist irrational numbers $a, b$ with $a^{b}$ rational. (Nonconstructive existence)

## 3. Step-by-Step Mathematical Proofs & Derivations

Each proof below is annotated with its strategic reading: *why this technique*, and where the crucial move happens.

### Proof 3.1 — Direct Proof: The Sum of Two Even Integers Is Even

**Claim.** If $a$ and $b$ are even integers, then $a + b$ is even.

**Proof.** Assume $a$ and $b$ are even. By definition there exist integers $m, n$ with $a = 2m$ and $b = 2n$. Then:

$$
a + b = 2m + 2n = 2(m + n)
$$

Since $m + n \in \mathbb{Z}$, the number $a + b$ equals twice an integer, hence is even. $\blacksquare$

**Strategic reading.** The hypothesis converts directly into equations — the signature situation for direct proof. Note the essential final step: *re-verify the definition* ("twice an integer") rather than stopping at an algebraic expression. Note also that $m$ and $n$ must be different letters: writing $a = 2m, b = 2m$ would silently assume $a = b$.

### Proof 3.2 — Contrapositive: If $n^2$ Is Even, Then $n$ Is Even

**Claim.** For every integer $n$: if $n^2$ is even, then $n$ is even.

**Proof.** We prove the contrapositive: if $n$ is odd, then $n^2$ is odd. Assume $n$ is odd, so $n = 2k + 1$ for some $k \in \mathbb{Z}$. Then:

$$
n^2 = (2k+1)^2 = 4k^2 + 4k + 1 = 2(2k^2 + 2k) + 1
$$

Since $2k^2 + 2k \in \mathbb{Z}$, $n^2$ is odd. By the contrapositive equivalence, the original claim holds. $\blacksquare$

**Strategic reading.** A direct attempt would begin "assume $n^2 = 2k$" — an equation about $n^2$ that gives no algebraic handle on $n$ itself (one would need square roots and case analysis). The negated conclusion "$n$ odd" instead yields $n = 2k+1$, which powers the whole computation. *Choose the side with the usable algebra.*

### Proof 3.3 — Contradiction: $\sqrt{2}$ Is Irrational

**Claim.** There is no rational number whose square is 2.

**Proof.** Suppose for contradiction that $\sqrt{2} = p/q$ with $p, q \in \mathbb{Z}$, $q \neq 0$, and the fraction in lowest terms: $\gcd(p, q) = 1$. Squaring and clearing denominators:

$$
2q^2 = p^2
$$

Thus $p^2$ is even, so by Proof 3.2, $p$ is even: $p = 2k$. Substituting:

$$
2q^2 = 4k^2 \implies q^2 = 2k^2
$$

Thus $q^2$ is even, so $q$ is even. Now both $p$ and $q$ are even, so $\gcd(p,q) \geq 2$, contradicting $\gcd(p,q) = 1$. The assumption fails, so $\sqrt{2}$ is irrational. $\blacksquare$

**Strategic reading.** Irrationality is a *negative* claim — "no fraction equals $\sqrt{2}$" — with nothing to compute from directly. Assuming the negation delivers a concrete fraction $p/q$, and the lowest-terms normalization plants the seed of the contradiction. Observe how Proof 3.2 is *reused* as a lemma: proofs compose.

### Proof 3.4 — Contradiction: Euclid's Infinitude of Primes

**Claim.** There are infinitely many prime numbers.

**Proof.** Suppose for contradiction that there are only finitely many primes, and list them all: $p_1, p_2, \ldots, p_k$. Define:

$$
N = p_1 p_2 \cdots p_k + 1
$$

Since $N \geq 2$, it has at least one prime divisor $p$ (every integer $\geq 2$ does — provable by strong induction, Topic 04). By assumption $p = p_i$ for some $i$. Then $p_i \mid N$ and $p_i \mid p_1 \cdots p_k$, so $p_i$ divides their difference:

$$
p_i \mid \bigl(N - p_1 p_2 \cdots p_k\bigr) = 1
$$

But no prime divides 1 (primes are $\geq 2$). Contradiction. Hence no finite list contains all primes. $\blacksquare$

**Strategic reading.** The finite list obtained from the negation is the raw material; the constructed number $N$ is engineered to defeat it. Note the theorem does *not* say "$N$ is prime" — $N$ may be composite (e.g. $2 \cdot 3 \cdot 5 \cdot 7 \cdot 11 \cdot 13 + 1 = 30031 = 59 \cdot 509$); it says $N$ has a prime factor outside the list. Misremembering this is a famous student error.

### Proof 3.5 — Cases: Parity and the Triangle Inequality

**Claim A.** For every integer $n$, the number $n^2 + n$ is even.

**Proof.** Every integer is even or odd — the cases are exhaustive.

*Case 1 ($n = 2k$):* $n^2 + n = 4k^2 + 2k = 2(2k^2 + k)$, even.

*Case 2 ($n = 2k+1$):* $n^2 + n = (4k^2 + 4k + 1) + (2k + 1) = 4k^2 + 6k + 2 = 2(2k^2 + 3k + 1)$, even.

In both cases $n^2 + n$ is even. $\blacksquare$ (Slicker: $n^2 + n = n(n+1)$ is a product of consecutive integers, one of which is even — still a case split, hidden.)

### Proof 3.5b — Cases: The Triangle Inequality

**Claim B (triangle inequality).** For all real $x, y$: $\lvert x + y \rvert \leq \lvert x \rvert + \lvert y \rvert$.

**Proof.** For any real $t$ we have $-\lvert t \rvert \leq t \leq \lvert t \rvert$ (immediate from the two cases $t \geq 0$, $t \lt 0$ in the definition of absolute value). Adding the inequalities for $x$ and $y$:

$$
-(\lvert x \rvert + \lvert y \rvert) \leq x + y \leq \lvert x \rvert + \lvert y \rvert
$$

Now split on the sign of $x + y$. If $x + y \geq 0$, then $\lvert x+y \rvert = x+y \leq \lvert x \rvert + \lvert y \rvert$ by the right inequality. If $x + y \lt 0$, then $\lvert x+y \rvert = -(x+y) \leq \lvert x \rvert + \lvert y \rvert$ by the left inequality. Both cases give the claim. $\blacksquare$

**Strategic reading.** Absolute values and parities are *defined* by cases, so case analysis is the native technique. The obligation is always exhaustiveness: even/odd covers $\mathbb{Z}$; $\geq 0$ / $\lt 0$ covers $\mathbb{R}$.

### Proof 3.6 — Existence and Uniqueness; Nonconstructive Existence

**Claim A.** There is exactly one real $x$ with $2x + 3 = 7$.

**Proof.** *Existence:* $x = 2$ gives $2 \cdot 2 + 3 = 7$. *Uniqueness:* suppose $2a + 3 = 7$ and $2b + 3 = 7$. Subtracting, $2a = 2b$, so $a = b$. Hence the solution is unique. $\blacksquare$

### Proof 3.6b — Nonconstructive Existence

**Claim B (nonconstructive existence).** There exist irrational numbers $a, b$ such that $a^{b}$ is rational.

**Proof.** Consider $c = \sqrt{2}^{\sqrt{2}}$. Either $c$ is rational or irrational — exhaustive cases.

*Case 1: $c$ rational.* Take $a = b = \sqrt{2}$ (irrational by Proof 3.3); then $a^b = c$ is rational, done.

*Case 2: $c$ irrational.* Take $a = c$ and $b = \sqrt{2}$; then:

$$
a^{b} = \bigl(\sqrt{2}^{\sqrt{2}}\bigr)^{\sqrt{2}} = \sqrt{2}^{\sqrt{2} \cdot \sqrt{2}} = \sqrt{2}^{2} = 2
$$

which is rational, done. In both cases suitable $a, b$ exist. $\blacksquare$

**Strategic reading.** The proof establishes existence *without revealing which case holds* — we never learn whether $\sqrt{2}^{\sqrt{2}}$ is rational (it is in fact irrational, by the deep Gelfond–Schneider theorem, but the elementary proof does not need to know). Uniqueness proofs always follow the two-solutions-collide template: assume $P(a)$ and $P(b)$, derive $a = b$.

### Choosing a Technique: Decision Guide

| Signal in the problem | Recommended technique |
|---|---|
| Hypothesis unpacks into equations or usable structure | Direct proof |
| Negations are more concrete than the originals | Contrapositive |
| Claim asserts impossibility, irrationality, or infinitude | Contradiction |
| Domain splits naturally (parity, sign, rational/irrational) | Cases |
| Claim is existential | Exhibit a witness (or argue nonconstructively) |
| Claim says "exactly one" | Existence + two-solutions-collide |
| You suspect the claim is false | Hunt for a counterexample, starting with small and extreme inputs |

Two meta-rules from Pólya: *first understand what would count as a proof* (what is arbitrary, what is given), and *after proving, look back* — ask whether the argument used every hypothesis (unused hypotheses signal either a stronger theorem or an error).

## 4. Computational & Algorithmic Insights

**Loop invariants: direct proof running inside a program.** To prove a loop correct, exhibit an *invariant* $I$ and establish three implications — initialization (before the loop, $I$ holds), maintenance (if $I$ holds before an iteration, it holds after), and termination-with-payoff ($I$ plus the exit condition implies the postcondition). Example: summing an array with accumulator $s$, the invariant "$s$ equals the sum of the first $i$ elements" is trivially true at $i = 0$, is preserved by `s += a[i]`, and at exit ($i = n$) yields exactly the specification. This is the direct-proof template with maintenance playing the role of the inductive step (formalized in Topic 04).

**Termination by descent.** Euclid's gcd algorithm replaces $(a, b)$ by $(b, a \bmod b)$; the second component is a strictly decreasing nonnegative integer, and an infinite strictly decreasing sequence of nonnegative integers is impossible (well-ordering). Termination proofs are contradiction proofs against infinite descent.

**Counterexample search as an algorithm.** Property-based testing frameworks (QuickCheck, Hypothesis) treat a claimed property $\forall x\,P(x)$ as a *falsification target*: generate random structured inputs, hunt for $x_0$ with $\neg P(x_0)$, then *shrink* it to a minimal failing case. Fuzzing does the same for "this program never crashes." These tools automate the counterexample half of mathematical method; the proof half is automated by proof assistants (Lean, Coq, Isabelle), in which the templates of this module are literal tactics: `intro` (direct), `contrapose` (contrapositive), `by_contra` (contradiction), `rcases`/`cases` (case split), `use` (existence witness).

**Floating-point folklore, refuted by counterexample.** The claim "addition of floats is associative" is false: in IEEE double precision, $(10^{16} + 1) - 10^{16} = 0$ while $10^{16} + (1 - 10^{16}) \neq$ the same order of evaluation result in general — one well-chosen triple refutes the universal claim and explains why numerical summation algorithms (Kahan, pairwise) exist at all.

| Verification activity | Underlying technique |
|---|---|
| Loop invariant argument | Direct proof + induction |
| Termination measure | Contradiction (no infinite descent) |
| Unit test | Example (never a proof of $\forall$) |
| Property-based test / fuzzing | Automated counterexample search |
| Proof assistant tactic script | Machine-checked template deployment |

## 5. Real-World Physics & AI/ML Applications

**Model debugging by contrapositive.** A mathematical model asserts "assumptions $A_1 \wedge \cdots \wedge A_k \to$ prediction $C$." When an experiment falsifies $C$, the contrapositive yields $\neg A_1 \vee \cdots \vee \neg A_k$: at least one assumption fails, and debugging is a case analysis over which. This is the logical skeleton of the scientific method itself (and of Lakatos's *Proofs and Refutations*): a falsified prediction indicts the assumption set, never "mathematics."

**Physics impossibility results are contradiction proofs.** Classic pattern: assume a perpetual motion machine exists, derive a violation of energy conservation; assume superluminal signaling, derive a causality paradox. The proof of the no-cloning theorem in quantum mechanics assumes a unitary cloner and contradicts linearity on superpositions — structurally identical to the $\sqrt{2}$ proof: assume the object exists, extract its properties, collide them.

**AI/ML applications.**

- **Gradient descent claims need hypotheses.** "GD decreases the loss each step" is *false* in general — counterexample: $f(x) = x^2$ with learning rate $\eta = 1.1$ oscillates and diverges ($x_{t+1} = (1 - 2\eta)x_t = -1.2\,x_t$). The correct theorem — for $L$-smooth $f$ and $\eta \leq 1/L$, $f(x_{t+1}) \leq f(x_t) - \frac{\eta}{2}\Vert \nabla f(x_t) \Vert^2$ — has a three-line direct proof from the smoothness inequality. The pair (counterexample, conditional theorem) is how rigorous ML papers are structured.
- **Argmax invariance under monotone transforms.** Direct proof: if $g$ is strictly increasing and $f(x^{*}) \geq f(x)$ for all $x$, then $g(f(x^{*})) \geq g(f(x))$ for all $x$, so $\arg\max$ is preserved. This justifies maximizing log-likelihood instead of likelihood, and explains why temperature scaling of logits never changes the top-1 class.
- **Adversarial examples are counterexamples.** An adversarial attack is a constructive disproof of the robustness claim $\forall \delta\,(\Vert \delta \Vert \leq \epsilon \to f(x+\delta) = f(x))$; a certification is a proof of it. The empirical-attack literature and the verification literature are the two halves of mathematical method applied to networks.
- **Uniqueness arguments in optimization.** Strictly convex functions have at most one minimizer — proved by the two-solutions-collide template: if $x \neq y$ were both minimizers, the midpoint would have strictly smaller value, contradiction. This is why ridge regression ($\lambda \gt 0$) has a unique solution while ordinary least squares with collinear features does not.
