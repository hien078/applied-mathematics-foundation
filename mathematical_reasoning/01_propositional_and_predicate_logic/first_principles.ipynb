# Topic 01: Propositional and Predicate Logic

## 1. First-Principles Intuition & Motivation

Mathematics is built out of sentences, but not every sentence can carry mathematical weight. "Solve the equation" is a command; "what is the growth rate?" is a question; "$x \gt 5$" depends on an unspecified $x$. None of these is true or false on its own. The first move of mathematical reasoning is to isolate the sentences that *do* have a definite truth value — **propositions** — and to study how their truth values combine.

The deep insight of propositional logic is **truth-functionality**: once we fix the truth values of atomic statements $p, q, r, \ldots$, the truth value of any compound built from them with NOT, AND, OR, IF-THEN, IFF is *computed*, not debated. Reasoning becomes calculation. This is why a circuit of transistors can evaluate logic, why a SAT solver can search over truth assignments, and why two mathematicians can always settle a disagreement about a propositional formula by writing out its truth table.

Propositional logic alone cannot express "every positive real number has a square root" — statements whose truth depends on ranging over a domain. **Predicate logic** adds variables, predicates $P(x)$, and the quantifiers $\forall$ and $\exists$. This is the language in which limits, continuity, convergence, optimality, and robustness are actually defined.

A guiding example from mathematical modeling: the claim "if the growth rate $r$ is positive, then the population increases" is an implication. It says *nothing* about what happens when $r \leq 0$ — the implication is vacuously true there. Recognizing what a logical statement does and does not claim is the difference between a model assumption and a model fantasy.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition 2.1 (Proposition).** A *proposition* is a declarative sentence that is either true (T) or false (F), but not both.

**Definition 2.2 (Connectives).** Given propositions $p$ and $q$, we define the compound propositions:

| Connective | Symbol | Name | Reading |
|---|---|---|---|
| Negation | $\neg p$ | NOT | "not $p$" |
| Conjunction | $p \wedge q$ | AND | "$p$ and $q$" |
| Disjunction | $p \vee q$ | OR | "$p$ or $q$" (inclusive) |
| Implication | $p \to q$ | IF-THEN | "if $p$ then $q$" |
| Biconditional | $p \leftrightarrow q$ | IFF | "$p$ if and only if $q$" |

**Definition 2.3 (Truth-table semantics).** The truth value of a compound proposition is determined by the truth values of its components:

| $p$ | $q$ | $\neg p$ | $p \wedge q$ | $p \vee q$ | $p \to q$ | $p \leftrightarrow q$ |
|---|---|---|---|---|---|---|
| T | T | F | T | T | T | T |
| T | F | F | F | T | F | F |
| F | T | T | F | T | T | F |
| F | F | T | F | F | T | T |

The implication column is the one to memorize: $p \to q$ is false **only** in the row where $p$ is true and $q$ is false. When $p$ is false the implication is *vacuously true* — it makes no claim about that situation.

**Definition 2.4 (Logical equivalence, tautology, contradiction).** Formulas $\varphi$ and $\psi$ are *logically equivalent*, written $\varphi \equiv \psi$, if they have the same truth value under every assignment of truth values to their atomic propositions. A *tautology* is true under every assignment (e.g. $p \vee \neg p$); a *contradiction* is false under every assignment (e.g. $p \wedge \neg p$); a *contingency* is neither.

**Theorem 2.5 (Fundamental equivalences).** For all propositions $p, q, r$:

| Name | Equivalence |
|---|---|
| Double negation | $\neg(\neg p) \equiv p$ |
| De Morgan's laws | $\neg(p \wedge q) \equiv \neg p \vee \neg q$ and $\neg(p \vee q) \equiv \neg p \wedge \neg q$ |
| Contrapositive | $(p \to q) \equiv (\neg q \to \neg p)$ |
| Implication as disjunction | $(p \to q) \equiv (\neg p \vee q)$ |
| Negation of implication | $\neg(p \to q) \equiv p \wedge \neg q$ |
| Distributivity | $p \wedge (q \vee r) \equiv (p \wedge q) \vee (p \wedge r)$ and $p \vee (q \wedge r) \equiv (p \vee q) \wedge (p \vee r)$ |

**Definition 2.6 (Predicate).** A *predicate* $P(x_1, \ldots, x_k)$ is a statement containing variables that becomes a proposition when each variable is assigned a value from its *domain of discourse* $D$.

**Definition 2.7 (Quantifiers).** Let $P(x)$ be a predicate over domain $D$.

- $\forall x\,P(x)$ ("for all $x$, $P(x)$") is true iff $P(a)$ is true for **every** $a \in D$.
- $\exists x\,P(x)$ ("there exists $x$ with $P(x)$") is true iff $P(a)$ is true for **at least one** $a \in D$.

**Example (continuity of $f$ at $a$).** The full quantified form is:

$$
\forall \epsilon \gt 0,\ \exists \delta \gt 0,\ \forall x,\ \bigl(\lvert x - a \rvert \lt \delta \to \lvert f(x) - f(a) \rvert \lt \epsilon\bigr)
$$

**Theorem 2.8 (Quantifier negation).** For any predicate $P$ over any domain:

$$
\neg\bigl(\forall x\,P(x)\bigr) \equiv \exists x\,\neg P(x), \qquad \neg\bigl(\exists x\,P(x)\bigr) \equiv \forall x\,\neg P(x)
$$

**Theorem 2.9 (Quantifier order).** For any predicate $P(x,y)$:

$$
\exists y\,\forall x\,P(x,y) \implies \forall x\,\exists y\,P(x,y)
$$

and the converse implication is **false** in general.

## 3. Step-by-Step Mathematical Proofs & Derivations

We now prove the central equivalences rigorously. For propositional laws, a truth-table verification *is* a complete proof: logical equivalence is defined by agreement in all rows, and the table enumerates all rows.

### Proof 3.1 — De Morgan's Laws

**Claim.** $\neg(p \wedge q) \equiv \neg p \vee \neg q$ and $\neg(p \vee q) \equiv \neg p \wedge \neg q$.

**Proof.** Enumerate all four truth assignments:

| $p$ | $q$ | $p \wedge q$ | $\neg(p \wedge q)$ | $\neg p \vee \neg q$ | $p \vee q$ | $\neg(p \vee q)$ | $\neg p \wedge \neg q$ |
|---|---|---|---|---|---|---|---|
| T | T | T | F | F | T | F | F |
| T | F | F | T | T | T | F | F |
| F | T | F | T | T | T | F | F |
| F | F | F | T | T | F | T | T |

Columns 4 and 5 agree in every row, proving the first law; columns 7 and 8 agree in every row, proving the second. $\blacksquare$

**Interpretation.** Negation converts AND into OR and vice versa. "Not (rich and famous)" means "not rich, or not famous" — the failure of a conjunction requires only one conjunct to fail.

### Proof 3.2 — The Contrapositive Law

**Claim.** $(p \to q) \equiv (\neg q \to \neg p)$.

**Proof.** Again enumerate all assignments:

| $p$ | $q$ | $p \to q$ | $\neg q$ | $\neg p$ | $\neg q \to \neg p$ |
|---|---|---|---|---|---|
| T | T | T | F | F | T |
| T | F | F | T | F | F |
| F | T | T | F | T | T |
| F | F | T | T | T | T |

Columns 3 and 6 agree in every row. $\blacksquare$

**Warning.** The converse $q \to p$ has truth values (T, T, F, T) down the rows — it differs from $p \to q$ in rows 2 and 3, so an implication is **not** equivalent to its converse. This asymmetry is the single most important fact to internalize before writing proofs by contrapositive (Topic 03).

### Proof 3.3 — Implication as Disjunction and Its Negation

**Claim.** $(p \to q) \equiv (\neg p \vee q)$, and consequently $\neg(p \to q) \equiv p \wedge \neg q$.

**Proof.** First the truth table:

| $p$ | $q$ | $p \to q$ | $\neg p \vee q$ |
|---|---|---|---|
| T | T | T | T |
| T | F | F | F |
| F | T | T | T |
| F | F | T | T |

The columns agree, so $(p \to q) \equiv (\neg p \vee q)$. Now apply De Morgan (Proof 3.1) and double negation:

$$
\neg(p \to q) \equiv \neg(\neg p \vee q) \equiv \neg\neg p \wedge \neg q \equiv p \wedge \neg q \qquad \blacksquare
$$

**Interpretation.** To refute "if $p$ then $q$" you must exhibit a situation where $p$ holds **and** $q$ fails. The negation of an implication is *not* another implication — a pervasive beginner error.

### Proof 3.4 — Quantifier Negation Laws

**Claim.** $\neg(\forall x\,P(x)) \equiv \exists x\,\neg P(x)$ over any domain $D$.

**Proof.** We argue semantically, in both directions.

($\Rightarrow$) Suppose $\neg(\forall x\,P(x))$ is true, i.e. $\forall x\,P(x)$ is false. By Definition 2.7, it is not the case that $P(a)$ holds for every $a \in D$. Hence there is at least one $a_0 \in D$ for which $P(a_0)$ fails, i.e. $\neg P(a_0)$ holds. Therefore $\exists x\,\neg P(x)$ is true.

($\Leftarrow$) Suppose $\exists x\,\neg P(x)$ is true, so some $a_0 \in D$ satisfies $\neg P(a_0)$. Then $P(a)$ cannot hold for every $a \in D$ (it fails at $a_0$), so $\forall x\,P(x)$ is false, i.e. $\neg(\forall x\,P(x))$ is true.

Both directions hold under every interpretation, so the formulas are logically equivalent. The dual law $\neg(\exists x\,P(x)) \equiv \forall x\,\neg P(x)$ follows by applying the first law to $\neg P$ and negating both sides. $\blacksquare$

**Worked negation (discontinuity).** Negating the continuity definition by pushing $\neg$ through each quantifier in turn and using Proof 3.3:

$$
\exists \epsilon \gt 0,\ \forall \delta \gt 0,\ \exists x,\ \bigl(\lvert x - a \rvert \lt \delta \wedge \lvert f(x) - f(a) \rvert \geq \epsilon\bigr)
$$

This is precisely the formal statement "$f$ is not continuous at $a$."

### Proof 3.5 — Quantifier Order: One Direction Holds, the Other Fails

**Claim.** $\exists y\,\forall x\,P(x,y) \implies \forall x\,\exists y\,P(x,y)$, but the converse fails in general.

**Proof of the implication.** Assume $\exists y\,\forall x\,P(x,y)$. Fix a witness $y_0$ such that $P(x, y_0)$ holds for all $x$. Now let $x$ be arbitrary. Then $y_0$ itself witnesses $\exists y\,P(x,y)$. Since $x$ was arbitrary, $\forall x\,\exists y\,P(x,y)$ holds. $\blacksquare$

**Counterexample to the converse.** Take domain $\mathbb{Z}$ and $P(x,y): x + y = 0$.

- $\forall x\,\exists y\,(x+y=0)$ is **true**: given $x$, choose $y = -x$.
- $\exists y\,\forall x\,(x+y=0)$ is **false**: a single $y_0$ would need $x = -y_0$ for every $x$, impossible since $\mathbb{Z}$ has more than one element. $\blacksquare$

**Interpretation.** In $\forall x\,\exists y$ the witness $y$ may *depend* on $x$; in $\exists y\,\forall x$ it must be *uniform*. This is exactly the distinction between pointwise and uniform continuity, and between "every training example has a correct label" and "one label is correct for every example."

### Proof 3.6 — Distributivity of Conjunction over Disjunction

**Claim.** $p \wedge (q \vee r) \equiv (p \wedge q) \vee (p \wedge r)$.

**Proof.** Enumerate the eight assignments:

| $p$ | $q$ | $r$ | $q \vee r$ | $p \wedge (q \vee r)$ | $p \wedge q$ | $p \wedge r$ | $(p \wedge q) \vee (p \wedge r)$ |
|---|---|---|---|---|---|---|---|
| T | T | T | T | T | T | T | T |
| T | T | F | T | T | T | F | T |
| T | F | T | T | T | F | T | T |
| T | F | F | F | F | F | F | F |
| F | T | T | T | F | F | F | F |
| F | T | F | T | F | F | F | F |
| F | F | T | T | F | F | F | F |
| F | F | F | F | F | F | F | F |

Columns 5 and 8 agree in every row. The dual law $p \vee (q \wedge r) \equiv (p \vee q) \wedge (p \vee r)$ follows by the same enumeration with the roles of $\wedge$ and $\vee$ exchanged. $\blacksquare$

**Remark.** Distributivity plus De Morgan is exactly what is needed to convert any formula to *conjunctive normal form* (CNF) — the input format of SAT solvers.

## 4. Computational & Algorithmic Insights

**Truth-table enumeration is exponential.** A formula with $n$ atomic propositions has $2^n$ truth assignments, so verifying a tautology by brute force costs $\Theta(2^n)$ evaluations. For $n = 30$ that is already over $10^9$ rows. This exponential wall motivates:

1. **Equivalence-based simplification** — rewrite formulas using the laws of Theorem 2.5 rather than enumerating rows.
2. **SAT solving** — deciding whether a CNF formula has *some* satisfying assignment is the canonical NP-complete problem (Cook–Levin, 1971). Modern CDCL solvers routinely handle industrial instances with millions of clauses despite the worst case.
3. **Binary decision diagrams (BDDs)** — canonical compressed representations of Boolean functions that make equivalence checking a pointer comparison, widely used in hardware verification.

**Normal forms and mechanization.** Every propositional formula can be algorithmically converted to CNF (a conjunction of clauses) or DNF (a disjunction of conjunctions) using only double negation, De Morgan, and distributivity. In Python, SymPy exposes this pipeline directly: `sympy.logic.boolalg.to_cnf`, `satisfiable`, and `truth_table` implement the definitions of Section 2 verbatim.

**Short-circuit evaluation.** Programming languages evaluate `A and B` lazily: if `A` is false, `B` is never evaluated. This is sound precisely because of truth-functionality — the truth value of the conjunction is already determined. De Morgan's laws also justify compiler transformations such as rewriting `not (a and b)` into `(not a) or (not b)`, which can enable branch elimination.

**Cost summary.**

| Task | Complexity |
|---|---|
| Evaluate a formula on one assignment | $O(\text{formula size})$ |
| Tautology check by truth table | $\Theta(2^n)$ assignments |
| SAT (CNF satisfiability) | NP-complete |
| Equivalence of two BDDs | $O(1)$ after canonical construction |

## 5. Real-World Physics & AI/ML Applications

**Program verification (Hoare logic).** A verified program is an implication factory: each Hoare triple $\{P\}\ C\ \{Q\}$ asserts "if precondition $P$ holds and command $C$ runs to completion, postcondition $Q$ holds." Verification-condition generators reduce program correctness to a conjunction of quantified implications discharged by SMT solvers (Z3, CVC5). The vacuous-truth convention is essential: code guarded by an unsatisfiable path condition is trivially correct because the hypothesis of its triple is never true.

**Digital circuits and the physics of computation.** Transistor networks physically realize the connectives: CMOS naturally implements NAND and NOR, and De Morgan's laws are used daily by hardware engineers to convert between AND/OR forms and the NAND-only forms that silicon prefers. That any Boolean function is computable from NAND alone (functional completeness) is a logic theorem with direct physical consequences for chip design.

**AI/ML applications.**

- **Adversarial robustness is a quantified formula.** "Model $f$ is $\epsilon$-robust at input $x$" means $\forall \delta$ with $\Vert \delta \Vert \leq \epsilon$, $f(x + \delta) = f(x)$. By Theorem 2.8 its negation is $\exists \delta$ with $\Vert \delta \Vert \leq \epsilon$ and $f(x+\delta) \neq f(x)$ — an adversarial example. Robustness certification tools (interval bound propagation, SMT-based verifiers such as Reluplex/Marabou) literally decide these formulas for ReLU networks.
- **Model debugging by contrapositive.** A model is a chain "assumptions $A_1 \wedge \cdots \wedge A_k \to$ prediction $C$." When the prediction fails, the contrapositive yields $\neg A_1 \vee \cdots \vee \neg A_k$: at least one assumption is violated. Structured debugging is applied De Morgan.
- **Symbolic AI and knowledge bases.** Logic programming (Prolog, Datalog) and modern neuro-symbolic systems execute resolution over clauses; database query planners apply De Morgan and distributivity to normalize WHERE-clause predicates before optimization.
