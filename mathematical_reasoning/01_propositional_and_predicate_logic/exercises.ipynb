## Level 0 — Concept Check

### Problem L0.1: Recognizing Propositions

Classify each sentence as a proposition, a predicate, or neither, with justification:

1. "The population at time $t=0$ is 100."
2. "$x + 3 = 7$"
3. "Minimize the loss function."
4. "Every even integer greater than 2 is a sum of two primes."

**Solution:**

1. **Proposition.** It is a declarative sentence about a specific quantity; it is either true or false (even if we do not know which).
2. **Predicate.** Its truth depends on the free variable $x$: true when $x = 4$, false otherwise. It becomes a proposition only after $x$ is assigned or quantified.
3. **Neither.** It is a command (imperative), so it has no truth value.
4. **Proposition.** This is the Goldbach conjecture. Its truth value is unknown to humanity, but it is definitely either true or false — having an *unknown* truth value is different from having *no* truth value.

$$
\boxed{\text{1: proposition, 2: predicate, 3: neither, 4: proposition (truth value unknown)}}
$$

*Key takeaway:* A proposition needs a definite truth value, not a known one; free variables produce predicates, not propositions.

### Problem L0.2: The One False Row of Implication

For the implication $p \to q$, list all truth-value combinations of $(p, q)$ and identify exactly when the implication is false. Then evaluate: "If $2 + 2 = 5$, then the Moon is made of cheese."

**Solution:**

The truth table of implication:

| $p$ | $q$ | $p \to q$ |
|---|---|---|
| T | T | T |
| T | F | F |
| F | T | T |
| F | F | T |

The implication is false **only** when $p$ is true and $q$ is false.

For the example: the hypothesis "$2+2=5$" is false, so the implication is vacuously true regardless of the (false) conclusion.

$$
\boxed{p \to q \text{ is false only for } (p,q) = (\text{T},\text{F}); \text{ the example is vacuously true}}
$$

*Key takeaway:* An implication with a false hypothesis makes no claim and therefore cannot be broken — it is vacuously true.

### Problem L0.3: Inclusive or Exclusive?

Is the statement "$x \geq 0$ or $x \leq 0$" true for $x = 0$? What about "$x \gt 0$ or $x \lt 0$"?

**Solution:**

Mathematical "or" ($\vee$) is **inclusive**: $p \vee q$ is true when either or both disjuncts hold.

- At $x = 0$: both "$x \geq 0$" and "$x \leq 0$" are true, so their disjunction is true (both-true is allowed).
- At $x = 0$: "$x \gt 0$" is false and "$x \lt 0$" is false, so the disjunction "$x \gt 0 \vee x \lt 0$" is false.

$$
\boxed{\text{"} x \geq 0 \vee x \leq 0 \text{" is true at } x=0; \quad \text{"} x \gt 0 \vee x \lt 0 \text{" is false at } x=0}
$$

*Key takeaway:* $p \vee q$ never excludes the both-true case; exclusive or is a different connective ($\oplus$).

### Problem L0.4: Negating a Universal Claim

Write the negation of "All swans are white" as a quantified formula, and state what evidence refutes the original claim.

**Solution:**

Formalize with domain "swans" and predicate $W(x)$: "$x$ is white". The claim is $\forall x\,W(x)$. By the quantifier negation law:

$$
\neg\bigl(\forall x\,W(x)\bigr) \equiv \exists x\,\neg W(x)
$$

The negation reads: "There exists a swan that is not white." A **single** non-white swan refutes the universal claim. The negation is *not* "no swan is white" ($\forall x\,\neg W(x)$), which is a much stronger statement.

$$
\boxed{\neg\bigl(\forall x\,W(x)\bigr) \equiv \exists x\,\neg W(x)}
$$

*Key takeaway:* Negation flips the quantifier and negates the body; refuting "for all" requires only one counterexample.

## Level 1 — Foundation

### Problem L1.1: Verifying De Morgan by Truth Table

Prove by truth table that $\neg(p \wedge q) \equiv \neg p \vee \neg q$.

**Solution:**

Enumerate all four assignments:

| $p$ | $q$ | $p \wedge q$ | $\neg(p \wedge q)$ | $\neg p$ | $\neg q$ | $\neg p \vee \neg q$ |
|---|---|---|---|---|---|---|
| T | T | T | F | F | F | F |
| T | F | F | T | F | T | T |
| F | T | F | T | T | F | T |
| F | F | F | T | T | T | T |

The columns for $\neg(p \wedge q)$ and $\neg p \vee \neg q$ agree in every row. Since logical equivalence means agreement under every assignment, and the table lists every assignment, the equivalence is proved.

$$
\boxed{\neg(p \wedge q) \equiv \neg p \vee \neg q}
$$

*Key takeaway:* For propositional formulas, a complete truth table is a complete proof.

### Problem L1.2: Implication as Disjunction

Show that $(p \to q) \equiv (\neg p \vee q)$, and use it to rewrite "if the model overfits, then validation loss rises" without an if-then.

**Solution:**

Truth table:

| $p$ | $q$ | $p \to q$ | $\neg p$ | $\neg p \vee q$ |
|---|---|---|---|---|
| T | T | T | F | T |
| T | F | F | F | F |
| F | T | T | T | T |
| F | F | T | T | T |

Columns 3 and 5 agree in all rows, proving the equivalence.

Application: with $p$ = "the model overfits", $q$ = "validation loss rises", the statement becomes "the model does not overfit, **or** validation loss rises."

$$
\boxed{(p \to q) \equiv (\neg p \vee q)}
$$

*Key takeaway:* Every implication is secretly a disjunction; this rewriting is the first step of CNF conversion.

### Problem L1.3: Converse, Inverse, Contrapositive

For the statement "If $n$ is divisible by 4, then $n$ is even," write the converse, inverse, and contrapositive. Which of the four statements are true for all integers $n$, and which are logically equivalent to each other?

**Solution:**

Let $p$: "$4 \mid n$" and $q$: "$n$ is even."

- **Original** $p \to q$: true (if $n = 4k$ then $n = 2(2k)$).
- **Converse** $q \to p$: "if $n$ is even then $4 \mid n$" — false ($n = 6$ is even but $4 \nmid 6$).
- **Inverse** $\neg p \to \neg q$: "if $4 \nmid n$ then $n$ is odd" — false ($n = 6$ again).
- **Contrapositive** $\neg q \to \neg p$: "if $n$ is odd then $4 \nmid n$" — true.

Equivalences: original $\equiv$ contrapositive, and converse $\equiv$ inverse (the inverse is the contrapositive of the converse). The counterexample $n=6$ falsifies both members of the second pair, consistent with their equivalence.

$$
\boxed{p \to q \equiv \neg q \to \neg p, \qquad q \to p \equiv \neg p \to \neg q, \qquad \text{only the first pair is true here}}
$$

*Key takeaway:* Of the four related implications there are exactly two equivalence classes; truth of a statement never transfers to its converse.

### Problem L1.4: Negating an Implication

Prove $\neg(p \to q) \equiv p \wedge \neg q$ algebraically (without a truth table), then negate: "If a function is differentiable, then it is continuous."

**Solution:**

Using implication-as-disjunction, De Morgan, and double negation:

$$
\neg(p \to q) \equiv \neg(\neg p \vee q) \equiv \neg\neg p \wedge \neg q \equiv p \wedge \neg q
$$

Each step invokes a previously proved equivalence, so the chain is a valid derivation.

Negation of the example: "There is a function that **is** differentiable **and** is **not** continuous." (This negation happens to be false — differentiability implies continuity — but the point is its logical form.)

$$
\boxed{\neg(p \to q) \equiv p \wedge \neg q}
$$

*Key takeaway:* The negation of an implication is a conjunction (hypothesis true, conclusion false), never another implication.

### Problem L1.5: Modus Ponens Is a Tautology

Show that $\bigl((p \to q) \wedge p\bigr) \to q$ is a tautology.

**Solution:**

| $p$ | $q$ | $p \to q$ | $(p \to q) \wedge p$ | $\bigl((p \to q) \wedge p\bigr) \to q$ |
|---|---|---|---|---|
| T | T | T | T | T |
| T | F | F | F | T |
| F | T | T | F | T |
| F | F | T | F | T |

The final column is T in every row, so the formula is a tautology. This validates the inference rule *modus ponens*: from $p \to q$ and $p$, conclude $q$.

$$
\boxed{\bigl((p \to q) \wedge p\bigr) \to q \text{ is a tautology}}
$$

*Key takeaway:* Valid inference rules are exactly the implications that are tautologies; modus ponens is the engine of every direct proof.

### Problem L1.6: Nested Quantifiers over the Integers

Determine the truth value over domain $\mathbb{Z}$ of each statement, with justification:

1. $\forall x\,\exists y\,(x + y = 0)$
2. $\exists y\,\forall x\,(x + y = 0)$
3. $\exists y\,\forall x\,(xy = 0)$
4. $\forall x\,\exists y\,(xy = 1)$

**Solution:**

1. **True.** Given any $x$, choose $y = -x$; then $x + y = 0$. The witness depends on $x$, which $\forall x\,\exists y$ permits.
2. **False.** A single $y_0$ would need $x = -y_0$ for **every** $x \in \mathbb{Z}$; taking $x = -y_0 + 1$ gives $x + y_0 = 1 \neq 0$.
3. **True.** Choose $y = 0$; then $xy = 0$ for every $x$. Here a uniform witness exists.
4. **False.** Take $x = 0$: then $xy = 0 \neq 1$ for every $y$, so $\exists y\,(0 \cdot y = 1)$ fails, falsifying the universal statement. (Even $x = 2$ fails within $\mathbb{Z}$ since $1/2 \notin \mathbb{Z}$.)

$$
\boxed{\text{1: T}, \quad \text{2: F}, \quad \text{3: T}, \quad \text{4: F}}
$$

*Key takeaway:* $\forall\exists$ allows the witness to adapt; $\exists\forall$ demands one witness for all — and the domain of discourse can change the answer.

## Level 2 — Applications in AI/ML & CS

### Problem L2.1: Negating the Definition of Continuity

A function $f$ is continuous at $a$ iff $\forall \epsilon \gt 0\,\exists \delta \gt 0\,\forall x\,(\lvert x-a \rvert \lt \delta \to \lvert f(x)-f(a) \rvert \lt \epsilon)$. Derive the formal statement "$f$ is discontinuous at $a$" step by step.

**Solution:**

Push the negation through one quantifier at a time using $\neg\forall \equiv \exists\neg$ and $\neg\exists \equiv \forall\neg$:

$$
\neg\forall \epsilon\,\exists \delta\,\forall x\,(\cdots) \equiv \exists \epsilon\,\neg\exists \delta\,\forall x\,(\cdots) \equiv \exists \epsilon\,\forall \delta\,\neg\forall x\,(\cdots) \equiv \exists \epsilon\,\forall \delta\,\exists x\,\neg(\cdots)
$$

Finally negate the implication using $\neg(p \to q) \equiv p \wedge \neg q$:

$$
\exists \epsilon \gt 0,\ \forall \delta \gt 0,\ \exists x,\ \bigl(\lvert x-a \rvert \lt \delta \wedge \lvert f(x)-f(a) \rvert \geq \epsilon\bigr)
$$

Reading: some error tolerance $\epsilon$ can never be guaranteed — no matter how small a neighborhood $\delta$ we take, some point inside it maps at least $\epsilon$ away.

$$
\boxed{\exists \epsilon \gt 0\ \forall \delta \gt 0\ \exists x\ \bigl(\lvert x-a \rvert \lt \delta \wedge \lvert f(x)-f(a) \rvert \geq \epsilon\bigr)}
$$

*Key takeaway:* Mechanical quantifier negation converts definitions into their exact failure conditions — the blueprint for every counterexample construction in analysis.

### Problem L2.2: NAND Gates and De Morgan in Hardware

CMOS technology most naturally implements NAND: $p \uparrow q \equiv \neg(p \wedge q)$. Express NOT, AND, and OR using only NAND, proving each identity.

**Solution:**

**NOT.** $p \uparrow p \equiv \neg(p \wedge p) \equiv \neg p$, since $p \wedge p \equiv p$ (idempotence).

**AND.** $\neg(p \uparrow q) \equiv \neg\neg(p \wedge q) \equiv p \wedge q$ by double negation; and negation is itself a NAND by the first identity, so $p \wedge q \equiv (p \uparrow q) \uparrow (p \uparrow q)$.

**OR.** By De Morgan, $p \vee q \equiv \neg(\neg p \wedge \neg q) \equiv \neg p \uparrow \neg q$. Replacing each negation with a self-NAND:

$$
p \vee q \equiv (p \uparrow p) \uparrow (q \uparrow q)
$$

Since NOT, AND, OR generate all Boolean functions, NAND alone is functionally complete — one physical gate type suffices to build any digital circuit.

$$
\boxed{\neg p \equiv p \uparrow p, \quad p \wedge q \equiv (p \uparrow q) \uparrow (p \uparrow q), \quad p \vee q \equiv (p \uparrow p) \uparrow (q \uparrow q)}
$$

*Key takeaway:* De Morgan's laws are not classroom trivia — they are the daily currency of hardware design, where entire CPUs are compiled to NAND.

### Problem L2.3: Hoare Triples and Vacuous Truth

A verifier must prove the postcondition of the branch `if (x > 0) { y = 1/x }` under precondition "x is an integer". The verification condition for the branch body is $(x \gt 0) \to (x \neq 0)$. Prove it, and explain why the path where the guard fails generates no division-safety obligation.

**Solution:**

**Proving the verification condition.** Assume $x \gt 0$. Since $0 \gt 0$ is false, $x \neq 0$. Hence $(x \gt 0) \to (x \neq 0)$ holds for every integer $x$ — a direct proof of an implication.

**The untaken path.** For the implicit else-branch, the division never executes; its safety obligation has the form $\bigl(\neg(x \gt 0) \wedge \text{body executes division}\bigr) \to \text{safe}$, and the hypothesis is false because no division occurs on that path. By vacuous truth the obligation holds automatically. Verification tools rely on this: unreachable or unexercised paths contribute vacuously true conditions rather than false alarms.

$$
\boxed{(x \gt 0) \to (x \neq 0) \text{ is valid; the else-path obligation is vacuously true}}
$$

*Key takeaway:* Program verification is the industrial-scale discharge of implications, and vacuous truth is what keeps guarded code from generating spurious proof obligations.

### Problem L2.4: SAT Encoding of Graph Coloring

Encode "graph $G$ with vertices $\{1,2\}$ and edge $(1,2)$ is 2-colorable" as a CNF formula, using variables $x_{v,c}$ meaning "vertex $v$ gets color $c$". How many truth assignments would brute force examine, and how does this scale with $n$ vertices and $k$ colors?

**Solution:**

Constraints per vertex $v \in \{1,2\}$ with colors $c \in \{1,2\}$:

1. **At least one color:** $(x_{v,1} \vee x_{v,2})$.
2. **At most one color:** $(\neg x_{v,1} \vee \neg x_{v,2})$.
3. **Edge constraint** for edge $(1,2)$, each color: $(\neg x_{1,c} \vee \neg x_{2,c})$.

The full CNF is the conjunction:

$$
(x_{1,1} \vee x_{1,2}) \wedge (x_{2,1} \vee x_{2,2}) \wedge (\neg x_{1,1} \vee \neg x_{1,2}) \wedge (\neg x_{2,1} \vee \neg x_{2,2}) \wedge (\neg x_{1,1} \vee \neg x_{2,1}) \wedge (\neg x_{1,2} \vee \neg x_{2,2})
$$

It is satisfied by $x_{1,1} = x_{2,2} = \text{T}$, others F (vertex 1 gets color 1, vertex 2 gets color 2), so the graph is 2-colorable.

Brute force examines $2^{nk}$ assignments over $nk$ variables — here $2^4 = 16$; for $n = 100, k = 3$ it is $2^{300}$, hopeless, which is why CDCL SAT solvers with clause learning are used instead.

$$
\boxed{2^{nk} \text{ assignments; here } 2^4 = 16, \text{ satisfiable}}
$$

*Key takeaway:* Combinatorial problems compile into CNF via at-least-one / at-most-one / conflict clauses — the universal input language of SAT solvers.

### Problem L2.5: Adversarial Robustness as a Quantified Formula

Let $f$ be a classifier. Write "$f$ is $\epsilon$-robust at input $x_0$" as a quantified formula, negate it formally, and interpret the negation.

**Solution:**

Robustness at $x_0$:

$$
\forall \delta\ \bigl(\Vert \delta \Vert \leq \epsilon \to f(x_0 + \delta) = f(x_0)\bigr)
$$

Negating with $\neg\forall \equiv \exists\neg$ and $\neg(p \to q) \equiv p \wedge \neg q$:

$$
\exists \delta\ \bigl(\Vert \delta \Vert \leq \epsilon \wedge f(x_0 + \delta) \neq f(x_0)\bigr)
$$

Interpretation: there exists a perturbation within the $\epsilon$-ball that flips the prediction — precisely an **adversarial example**. Attack algorithms (FGSM, PGD) are search procedures for the existential witness $\delta$; certification tools (interval bound propagation, SMT-based verifiers) attempt to prove the universal statement instead. The attack/defense duality in adversarial ML is exactly the $\exists$/$\forall$ duality of quantifier negation.

$$
\boxed{\neg\text{robust} \equiv \exists \delta\ \bigl(\Vert \delta \Vert \leq \epsilon \wedge f(x_0+\delta) \neq f(x_0)\bigr)}
$$

*Key takeaway:* Attacks find witnesses of an existential; certificates prove a universal — the two sides of one negation law.

### Problem L2.6: De Morgan in Database Queries

A data pipeline filters records with `NOT (age >= 18 AND consent = true)`. Rewrite this predicate without NOT applied to a compound expression, and explain why a query optimizer performs this rewriting.

**Solution:**

Let $p$: "age $\geq 18$" and $q$: "consent = true". By De Morgan:

$$
\neg(p \wedge q) \equiv \neg p \vee \neg q
$$

so the filter becomes `age < 18 OR consent = false` (assuming a two-valued consent field; SQL's three-valued logic with NULL requires `consent IS NOT TRUE` instead — an important practical caveat).

Optimizers normalize predicates by pushing negations to the leaves (negation normal form) because: (1) atomic comparisons like `age < 18` can use indexes, while a negated conjunction cannot; (2) disjunctions can be split into unions of index scans; (3) normal forms allow the planner to recognize equivalent predicates and reuse cached plans.

$$
\boxed{\neg(p \wedge q) \equiv \neg p \vee \neg q \ \Rightarrow\ \texttt{age} \lt 18 \ \text{OR} \ \texttt{consent} = \text{false}}
$$

*Key takeaway:* Negation normal form via De Morgan turns logically opaque filters into index-friendly atomic predicates — logic laws with direct performance consequences.

## Level 3 — Challenge

### Problem L3.1: Functional Completeness of NAND

Prove that every Boolean function $f: \{\text{T},\text{F}\}^n \to \{\text{T},\text{F}\}$ can be expressed using only the NAND connective.

**Solution:**

**Step 1: every Boolean function is expressible in $\{\neg, \wedge, \vee\}$.** Given $f$, build its full DNF: for each input row $(a_1, \ldots, a_n)$ where $f = \text{T}$, form the minterm $\ell_1 \wedge \cdots \wedge \ell_n$ with $\ell_i = p_i$ if $a_i = \text{T}$ and $\ell_i = \neg p_i$ otherwise; then take the disjunction of all minterms. This formula is true exactly on the rows where $f$ is true, hence equals $f$. (If $f$ is identically F, use $p_1 \wedge \neg p_1$.)

**Step 2: eliminate $\{\neg, \wedge, \vee\}$ in favor of NAND.** Using $p \uparrow q \equiv \neg(p \wedge q)$:

- $\neg p \equiv p \uparrow p$
- $p \wedge q \equiv \neg(p \uparrow q) \equiv (p \uparrow q) \uparrow (p \uparrow q)$
- $p \vee q \equiv \neg(\neg p \wedge \neg q) \equiv (p \uparrow p) \uparrow (q \uparrow q)$

Each identity is verified by a 4-row truth table (e.g. $p \uparrow p$ has values F, T on rows where $p$ is T, F — exactly $\neg p$). Rewriting the DNF bottom-up replaces every connective by NANDs, expressing $f$ in NAND alone. $\blacksquare$

$$
\boxed{\{\uparrow\} \text{ is functionally complete}}
$$

*Key takeaway:* Full DNF proves $\{\neg,\wedge,\vee\}$ complete; three local identities then collapse everything to a single gate type.

### Problem L3.2: Counting Boolean Functions

How many logically inequivalent propositional formulas are there in $n$ atomic variables? Compute the number for $n = 1, 2, 3$, and conclude how many binary connectives exist.

**Solution:**

A formula in $n$ variables is determined up to equivalence by its truth table: a function from the $2^n$ assignments to $\{\text{T},\text{F}\}$. The number of such functions is:

$$
2^{2^n}
$$

Two formulas are equivalent iff they induce the same table, and by full DNF (Problem L3.1) every table is realized, so the count is exact.

- $n = 1$: $2^2 = 4$ (constants T, F, identity $p$, negation $\neg p$).
- $n = 2$: $2^4 = 16$ — so there are exactly **16 binary connectives**, including the 5 standard ones plus XOR, NAND, NOR, and 8 degenerate/less-used ones.
- $n = 3$: $2^8 = 256$.

$$
\boxed{2^{2^n} \text{ inequivalent formulas; } 16 \text{ binary connectives}}
$$

*Key takeaway:* Truth tables biject equivalence classes of formulas with Boolean functions, giving the double-exponential count $2^{2^n}$.

### Problem L3.3: Absorption and Simplification to DNF

Prove the absorption law $p \vee (p \wedge q) \equiv p$, then simplify the formula $\varphi = (p \wedge q) \vee (p \wedge \neg q) \vee (\neg p \wedge q)$ to a minimal DNF.

**Solution:**

**Absorption.** Distribute: $p \vee (p \wedge q) \equiv (p \vee p) \wedge (p \vee q) \equiv p \wedge (p \vee q)$. If $p$ is T both conjuncts are T; if $p$ is F the first conjunct is F. So the result equals $p$ in every case. (A 4-row table confirms.)

**Simplification.** Merge the first two minterms by distributivity and the excluded middle:

$$
(p \wedge q) \vee (p \wedge \neg q) \equiv p \wedge (q \vee \neg q) \equiv p \wedge \text{T} \equiv p
$$

So $\varphi \equiv p \vee (\neg p \wedge q)$. Distribute again:

$$
p \vee (\neg p \wedge q) \equiv (p \vee \neg p) \wedge (p \vee q) \equiv \text{T} \wedge (p \vee q) \equiv p \vee q
$$

Check: $\varphi$ is true exactly when at least one of $p, q$ is true (its three minterms are the three non-(F,F) rows), matching $p \vee q$.

$$
\boxed{\varphi \equiv p \vee q}
$$

*Key takeaway:* Distributivity, excluded middle, and absorption are the algebraic engine behind logic minimization (Karnaugh maps, Quine–McCluskey).

### Problem L3.4: Knights and Knaves

On an island, knights always tell the truth and knaves always lie. Inhabitant A says: "We are both knaves." What are A and B? Formalize and solve.

**Solution:**

Let $a$: "A is a knight" and $b$: "B is a knight". A's statement is $S \equiv (\neg a \wedge \neg b)$. The island rule says a statement is true iff its speaker is a knight:

$$
a \leftrightarrow (\neg a \wedge \neg b)
$$

Case $a = \text{T}$: the right side becomes $\text{F} \wedge \neg b = \text{F}$, so the biconditional reads $\text{T} \leftrightarrow \text{F}$ — false. Contradiction.

Case $a = \text{F}$: the biconditional requires the right side to be false: $\neg(\neg a \wedge \neg b) \equiv a \vee b$ must hold. With $a = \text{F}$ this forces $b = \text{T}$.

Check: A is a knave claiming "both knaves" — the claim is false (B is a knight), consistent with A lying. B says nothing, so no further constraint.

$$
\boxed{\text{A is a knave and B is a knight}}
$$

*Key takeaway:* Encoding "speaker tells truth iff knight" as a biconditional turns logic puzzles into satisfiability problems with a unique model.
