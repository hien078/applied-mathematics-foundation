## Level 0 — Concept Check

### Problem L0.1: Is It a Flow?

The network on $\lbrace s,a,b,t \rbrace$ has capacities $c(s,a) = 3$, $c(s,b) = 2$, $c(a,b) = 2$, $c(a,t) = 2$, $c(b,t) = 3$. Decide whether each assignment is a valid flow and, if so, give its value.

(a) $f(s,a) = 3$, $f(s,b) = 1$, $f(a,b) = 2$, $f(a,t) = 2$, $f(b,t) = 3$.
(b) $f(s,a) = 3$, $f(s,b) = 1$, $f(a,b) = 1$, $f(a,t) = 2$, $f(b,t) = 2$.

**Solution**

**(a)** Capacities are all respected, but conservation fails at $a$:

$$
\text{inflow}(a) = 3, \qquad \text{outflow}(a) = f(a,b) + f(a,t) = 2 + 2 = 4
$$

A vertex cannot emit more than it receives, so **(a) is not a flow**.

**(b)** Capacities: $3 \le 3$, $1 \le 2$, $1 \le 2$, $2 \le 2$, $2 \le 3$ ✓. Conservation at $a$: in $3$, out $1 + 2 = 3$ ✓. At $b$: in $1 + 1 = 2$, out $2$ ✓. So **(b) is a flow** with value

$$
\vert f \vert = f(s,a) + f(s,b) = 3 + 1 = 4
$$

Cross-check with the flow-value lemma on the cut $S = \lbrace s,a \rbrace$: crossing arcs are $s \to b$, $a \to b$, $a \to t$ carrying $1 + 1 + 2 = 4$, with nothing returning ✓. (It is not maximum: pushing $f(s,b) = 2$, $f(a,b) = 1$, $f(b,t) = 3$ gives $\vert f \vert = 5$.)

$$
\boxed{\text{(a) invalid (conservation at } a\text{)}; \quad \text{(b) valid, } \vert f \vert = 4}
$$

*Key takeaway:* Always test both constraint families; capacity violations are obvious in code, conservation violations are the silent ones.

### Problem L0.2: Which Arcs Count in a Cut?

Take the network of Problem L0.1 and add an arc $b \to a$ of capacity $7$. For the cut $S = \lbrace s, a \rbrace$, $T = \lbrace b, t \rbrace$, compute $c(S,T)$ and explain which arcs are excluded and why.

**Solution**

Only arcs directed **from $S$ to $T$** contribute:

$$
c(S,T) = c(s,b) + c(a,b) + c(a,t) = 2 + 2 + 2 = 6
$$

Excluded:

- $b \to a$ (capacity $7$) runs from $T$ to $S$ — a *backward* arc. It contributes $-f(b,a)$ to the flow value, never $+c$ to the capacity, so it cannot help block flow.
- $s \to a$ lies entirely inside $S$ and cancels in the flow-value lemma.

This asymmetry is exactly why $\vert f \vert = f(S,T) - f(T,S) \le c(S,T)$: backward flow only *reduces* the net crossing, so ignoring backward capacities still gives a valid upper bound.

$$
\boxed{c(S,T) = 6; \ \text{backward arcs (here } b \to a \text{ of capacity } 7 \text{) never count}}
$$

*Key takeaway:* A cut's capacity is a one-directional quantity. Summing capacities of all arcs "with an endpoint on each side" is the single most common modelling bug.

### Problem L0.3: Maximal, Maximum, Perfect

In the path $P_4$ with vertices $v_1 - v_2 - v_3 - v_4$, give a maximal matching that is not maximum, a maximum matching, and state whether a perfect matching exists.

**Solution**

- **Maximal but not maximum**: $M_1 = \lbrace v_2 v_3 \rbrace$. No edge can be added ($v_1 v_2$ and $v_3 v_4$ both touch matched vertices), so $M_1$ is maximal — yet $\vert M_1 \vert = 1$.
- **Maximum**: $M_2 = \lbrace v_1 v_2, \ v_3 v_4 \rbrace$ with $\nu(P_4) = 2$.
- **Perfect**: $M_2$ covers all four vertices, so yes — $P_4$ has a perfect matching.

Berge's theorem explains the gap: $M_1$ admits the $M_1$-augmenting path $v_1 - v_2 - v_3 - v_4$ (unmatched endpoints, alternating edges), and flipping it along the path yields $M_2$.

$$
\boxed{\text{maximal } \not\Rightarrow \text{maximum}; \ \nu(P_4) = 2, \text{ and a perfect matching exists}}
$$

*Key takeaway:* "Maximal" is a local greedy property, "maximum" a global optimum; the certificate separating them is an augmenting path, which is why greedy matching can be a factor of $2$ off.

### Problem L0.4: Why König Needs Bipartiteness

Compute $\nu$ and $\tau$ for the triangle $K_3$ and for the 5-cycle $C_5$, and explain the structural obstruction.

**Solution**

**Triangle $K_3$.** Any two edges share a vertex, so $\nu(K_3) = 1$. A single vertex misses the opposite edge, so $\tau(K_3) = 2$. Hence $\nu \lt \tau$.

**Five-cycle $C_5$.** Matchings of size $3$ would need $6$ distinct vertices, so $\nu(C_5) = 2$ (e.g. two non-adjacent edges). Two vertices cover at most $4$ of the $5$ edges, so $\tau(C_5) = 3$. Again $\nu \lt \tau$.

**Obstruction.** Both graphs contain an odd cycle, and a graph is bipartite exactly when it has none. On an odd cycle $C_{2k+1}$ we always get $\nu = k$ and $\tau = k+1$. Algebraically, the matching LP $\max \sum_e x_e$ subject to $\sum_{e \ni v} x_e \le 1$ has the fractional solution $x_e = 1/2$ on an odd cycle with value $(2k+1)/2 \gt k$, so the LP is not integral — which is precisely what the bipartite case guarantees.

$$
\boxed{\nu(K_3) = 1 \lt 2 = \tau(K_3); \ \nu(C_5) = 2 \lt 3 = \tau(C_5); \ \text{odd cycles break König}}
$$

*Key takeaway:* Odd cycles are the single obstruction — they defeat König, break LP integrality, and are exactly what Edmonds' blossom algorithm contracts to restore the theory in general graphs.

## Level 1 — Foundation

### Problem L1.1: Ford–Fulkerson on the Diamond

All five arcs of the network $s \to a$, $s \to b$, $a \to b$, $a \to t$, $b \to t$ have capacity $1$. Start by augmenting along $s \to a \to b \to t$ and finish the algorithm. Give the maximum flow, the minimum cut, and explain the role of the reverse arc.

**Solution**

**Augmentation 1.** Path $s \to a \to b \to t$, bottleneck $\min(1,1,1) = 1$. Now $f(s,a) = f(a,b) = f(b,t) = 1$ and $\vert f \vert = 1$. In the *original* graph no unsaturated $s$–$t$ route remains, which is where naive greed stops.

**Residual network.** Saturated forward arcs $s \to a$, $a \to b$, $b \to t$ disappear and are replaced by reverse arcs $a \to s$, $b \to a$, $t \to b$ of residual capacity $1$; the untouched arcs $s \to b$ and $a \to t$ survive with residual capacity $1$.

**Augmentation 2.** $G_f$ contains $s \to b \to a \to t$ (the middle step is the **reverse** arc). Bottleneck $1$. Pushing it *adds* one unit on $s \to b$ and $a \to t$ and *cancels* one unit on $a \to b$:

$$
f(s,a) = f(s,b) = f(a,t) = f(b,t) = 1, \qquad f(a,b) = 0, \qquad \vert f \vert = 2
$$

**Termination and certificate.** Both source arcs are now saturated, so in the new residual network $s$ reaches nothing: $S = \lbrace s \rbrace$, $T = \lbrace a,b,t \rbrace$ and

$$
c(S,T) = c(s,a) + c(s,b) = 1 + 1 = 2 = \vert f \vert
$$

$$
\boxed{\vert f^{\ast} \vert = 2 = c(\lbrace s \rbrace, \lbrace a,b,t \rbrace); \ \text{the reverse arc } b \to a \text{ reroutes the first unit}}
$$

*Key takeaway:* The residual reverse arc is not a physical pipe; it is the ability to retract an earlier commitment, and it is exactly what upgrades "no path left" into a proof of optimality.

### Problem L1.2: Edmonds–Karp on a Six-Vertex Network

Compute the maximum flow from $s$ to $t$ with BFS augmenting paths for capacities $c(s,v_1) = 16$, $c(s,v_2) = 13$, $c(v_1,v_3) = 12$, $c(v_2,v_1) = 4$, $c(v_2,v_4) = 14$, $c(v_3,v_2) = 9$, $c(v_3,t) = 20$, $c(v_4,v_3) = 7$, $c(v_4,t) = 4$. Give every augmentation and the minimum cut.

**Solution**

| # | Shortest augmenting path (BFS) | Bottleneck | $\vert f \vert$ after |
|---|---|---|---|
| 1 | $s \to v_1 \to v_3 \to t$ | $\min(16,12,20) = 12$ | $12$ |
| 2 | $s \to v_2 \to v_4 \to t$ | $\min(13,14,4) = 4$ | $16$ |
| 3 | $s \to v_2 \to v_4 \to v_3 \to t$ | $\min(9,10,7,8) = 7$ | $23$ |

After the third augmentation the residual network offers, from $s$: $v_1$ (residual $4$) — dead end, since $v_1 \to v_3$ is saturated; and $v_2$ (residual $2$) — from which $v_2 \to v_1$ is a dead end and $v_2 \to v_4$ leads to $v_4$, whose two outgoing arcs $v_4 \to v_3$ and $v_4 \to t$ are both saturated. So $t$ is unreachable and the flow is maximum.

**Minimum cut.** The residual-reachable set is $S = \lbrace s, v_1, v_2, v_4 \rbrace$, $T = \lbrace v_3, t \rbrace$, with crossing arcs $v_1 \to v_3$, $v_4 \to v_3$, $v_4 \to t$:

$$
c(S,T) = 12 + 7 + 4 = 23 = \vert f^{\ast} \vert
$$

$$
\boxed{\vert f^{\ast} \vert = 23, \quad \text{min cut } S = \lbrace s, v_1, v_2, v_4 \rbrace \text{ of capacity } 23}
$$

Note the arc $v_3 \to v_2$ crosses from $T$ to $S$ and correctly contributes nothing.

*Key takeaway:* One BFS in the final residual network hands you the minimum cut for free — the bottleneck diagnosis is a by-product of the flow computation.

### Problem L1.3: Why "Any Augmenting Path" Is Not Enough

Capacities: $c(s,a) = c(s,b) = c(a,t) = c(b,t) = M$ and $c(a,b) = 1$, with $M = 1000$. Count the augmentations if the path choice alternates $s \to a \to b \to t$ and $s \to b \to a \to t$, and compare with BFS selection.

**Solution**

**Adversarial choice.** The path $s \to a \to b \to t$ has bottleneck $\min(M, 1, M) = 1$ and pushes one unit, saturating the middle arc. The residual then contains $b \to a$ of capacity $1$, and the path $s \to b \to a \to t$ again has bottleneck $1$, restoring $a \to b$. The pair of moves increases the flow by $2$ and returns the middle arc to its original state, so the process repeats:

$$
\text{augmentations} = 2M = 2000, \qquad \text{each pushing exactly } 1 \text{ unit}
$$

The final value is $\vert f^{\ast} \vert = 2M = 2000$, matching the min cut $c(\lbrace s \rbrace, \cdot) = 2M$.

**BFS choice.** Both $s \to a \to t$ and $s \to b \to t$ have $2$ arcs and are found first; each has bottleneck $M$:

$$
\text{augmentations} = 2
$$

**General bound.** Plain Ford–Fulkerson is only *pseudo-polynomial*, $O(m \vert f^{\ast} \vert)$ — the running time depends on the numerical magnitude of the capacities, so doubling $M$ doubles the work even though the graph is unchanged. Edmonds–Karp removes this dependence entirely with $O(nm^2)$.

$$
\boxed{2M = 2000 \text{ augmentations (adversarial) vs. } 2 \text{ (BFS)}; \ \vert f^{\ast} \vert = 2000 \text{ either way}}
$$

*Key takeaway:* Correctness of Ford–Fulkerson never depended on the path rule; *efficiency* depends on it entirely, and "shortest augmenting path" is the rule that makes the bound capacity-free.

### Problem L1.4: Maximum Matching and König's Cover

Bipartite graph with $X = \lbrace x_1,x_2,x_3,x_4 \rbrace$, $Y = \lbrace y_1,y_2,y_3,y_4 \rbrace$ and edges $x_1 y_1$, $x_1 y_2$, $x_2 y_1$, $x_3 y_2$, $x_3 y_3$, $x_4 y_3$. Find a maximum matching, a minimum vertex cover, and verify König.

**Solution**

**A matching of size 3.** Take $M = \lbrace x_1 y_2, \ x_2 y_1, \ x_4 y_3 \rbrace$.

**No matching of size 4.** $X$ has four vertices but $N(X) = \lbrace y_1, y_2, y_3 \rbrace$ has only three, so at most three vertices of $X$ can be matched (each needs its own neighbour). Hence $\nu = 3$.

**A vertex cover of size 3.** $C = \lbrace y_1, y_2, y_3 \rbrace$ meets every listed edge, since $y_4$ is isolated. So $\tau \le 3$; combined with the always-valid $\nu \le \tau$ we get $\tau = 3$.

$$
\boxed{\nu = \tau = 3, \quad M = \lbrace x_1 y_2, x_2 y_1, x_4 y_3 \rbrace, \quad C = \lbrace y_1, y_2, y_3 \rbrace}
$$

**As a flow.** With unit arcs $s \to x_i$ and $y_j \to t$ and infinite middle arcs, the maximum flow is $3$ and the residual-reachable set gives a cut of capacity $3$ — König is just max-flow min-cut in this network.

*Key takeaway:* The cover is a *proof* that no larger matching exists, checkable in $O(m)$ — always report it alongside the matching.

### Problem L1.5: Testing Hall's Condition

$X = \lbrace x_1,x_2,x_3 \rbrace$, $Y = \lbrace y_1,y_2,y_3,y_4 \rbrace$, and every $x_i$ is adjacent to exactly $y_1$ and $y_2$. Does a matching saturating $X$ exist? Find the maximum matching size from the deficiency formula.

**Solution**

Test Hall's condition $\vert N(S) \vert \ge \vert S \vert$. Take $S = X$:

$$
N(S) = \lbrace y_1, y_2 \rbrace, \qquad \vert N(S) \vert = 2 \lt 3 = \vert S \vert
$$

Hall's condition **fails**, so no matching saturates $X$. The witness set $S = X$ is the certificate: three vertices competing for two partners.

**Deficiency.** The maximum deficiency is

$$
\mathrm{def}(G) = \max_{S \subseteq X} \big( \vert S \vert - \vert N(S) \vert \big) = 3 - 2 = 1
$$

(smaller sets give $1 - 2 \lt 0$ and $2 - 2 = 0$). By the deficiency form of Hall's theorem,

$$
\nu(G) = \vert X \vert - \mathrm{def}(G) = 3 - 1 = 2
$$

realized by, say, $\lbrace x_1 y_1, \ x_2 y_2 \rbrace$, leaving $x_3$ and the isolated $y_3, y_4$ unmatched.

$$
\boxed{\text{Hall fails at } S = X; \ \mathrm{def} = 1, \ \nu = 2}
$$

*Key takeaway:* When a matching solver fails, it should hand back a Hall violator — a small set whose neighbourhood is too small — which is far more actionable than "infeasible".

### Problem L1.6: Constructing König's Cover from a Maximum Matching

Let $M$ be a maximum matching in a bipartite graph with parts $X, Y$. Let $U \subseteq X$ be the unmatched vertices of $X$ and let $Z$ be the set of vertices reachable from $U$ by $M$-alternating paths. Prove that $C = (X \setminus Z) \cup (Y \cap Z)$ is a vertex cover with $\vert C \vert = \vert M \vert$, and apply the construction to Problem L1.4.

**Solution**

**$C$ is a cover.** Suppose an edge $xy$ ($x \in X$, $y \in Y$) is uncovered, i.e. $x \in Z$ and $y \notin Z$.

- If $xy \notin M$: since $x \in Z$ is reached by an alternating path ending with a matched edge (or is itself unmatched), that path extends along the unmatched edge $xy$, so $y \in Z$ — contradiction.
- If $xy \in M$: then $x$ is matched, so $x \notin U$, so $x$ was reached via its own matching edge from $y$, giving $y \in Z$ — contradiction.

**$\vert C \vert \le \vert M \vert$.** Every vertex of $C$ is matched and no matching edge contributes two vertices of $C$:

- $x \in X \setminus Z$ is matched (unmatched $X$-vertices lie in $U \subseteq Z$).
- $y \in Y \cap Z$ is matched: it is reached by an alternating path of odd length starting at an unmatched $X$-vertex, and if $y$ were unmatched that path would be $M$-augmenting, contradicting maximality (Berge).
- No matching edge $xy$ has both $x \in X \setminus Z$ and $y \in Y \cap Z$: from $y \in Z$ and $xy \in M$ the alternating path continues to $x$, forcing $x \in Z$.

So $\vert C \vert \le \vert M \vert$. With the trivial $\vert M \vert \le \tau \le \vert C \vert$ we obtain

$$
\boxed{\nu(G) = \vert M \vert = \vert C \vert = \tau(G)}
$$

**Application to Problem L1.4** with $M = \lbrace x_1 y_2, x_2 y_1, x_4 y_3 \rbrace$: the unmatched $X$-vertex is $U = \lbrace x_3 \rbrace$. Alternating exploration: $x_3 \to y_2$ (unmatched edge) $\to x_1$ (matched) $\to y_1$ (unmatched edge) $\to x_2$ (matched); also $x_3 \to y_3 \to x_4$. Hence

$$
Z = \lbrace x_1,x_2,x_3,x_4, \ y_1,y_2,y_3 \rbrace, \qquad C = (X \setminus Z) \cup (Y \cap Z) = \emptyset \cup \lbrace y_1,y_2,y_3 \rbrace
$$

of size $3 = \vert M \vert$ — exactly the cover found by inspection earlier.

*Key takeaway:* König's theorem is *constructive*: one alternating search converts any maximum matching into the matching optimality proof, in $O(m)$ time.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: Image Segmentation as a Minimum Cut

Three pixels in a row must each be labelled foreground or background. The data terms are $D_1(\text{bg}) = 7$, $D_2(\text{bg}) = 2$, $D_3(\text{bg}) = 1$ and $D_1(\text{fg}) = 1$, $D_2(\text{fg}) = 4$, $D_3(\text{fg}) = 6$; neighbouring pixels pay a smoothness penalty $\lambda = 2$ when their labels differ. Build the flow network, find the MAP labelling, and state why min-cut is exact here.

**Solution**

**Construction (Boykov–Jolly).** Put the source on the foreground side and the sink on the background side:

- arc $s \to i$ with capacity $D_i(\text{bg})$ — cut when pixel $i$ ends up on the sink (background) side;
- arc $i \to t$ with capacity $D_i(\text{fg})$ — cut when pixel $i$ ends up on the source (foreground) side;
- arcs $i \leftrightarrow j$ of capacity $\lambda$ between neighbours — cut exactly when the labels differ.

By construction the capacity of any $s$–$t$ cut equals the energy of the corresponding labelling:

$$
E(\ell) = \sum_i D_i(\ell_i) + \lambda \sum_{i \sim j} \mathbf{1}[\ell_i \neq \ell_j]
$$

**Evaluating the eight labellings** (fg set listed first):

| Foreground set | Data terms | Boundary | Energy |
|---|---|---|---|
| $\emptyset$ | $7 + 2 + 1$ | $0$ | $10$ |
| $\lbrace 1 \rbrace$ | $1 + 2 + 1$ | $2$ | $6$ |
| $\lbrace 2 \rbrace$ | $7 + 4 + 1$ | $4$ | $16$ |
| $\lbrace 3 \rbrace$ | $7 + 2 + 6$ | $2$ | $17$ |
| $\lbrace 1,2 \rbrace$ | $1 + 4 + 1$ | $2$ | $8$ |
| $\lbrace 1,3 \rbrace$ | $1 + 2 + 6$ | $4$ | $13$ |
| $\lbrace 2,3 \rbrace$ | $7 + 4 + 6$ | $2$ | $19$ |
| $\lbrace 1,2,3 \rbrace$ | $1 + 4 + 6$ | $0$ | $11$ |

$$
\boxed{\text{min cut} = 6, \quad \text{MAP labelling: pixel 1 foreground, pixels 2 and 3 background}}
$$

The unary-only optimum picks the same labels ($1 \lt 7$, $2 \lt 4$, $1 \lt 6$) at cost $4$, plus one label change at the $1$–$2$ boundary for $+2$: total $6$ ✓.

**Why exact.** The pairwise term is **submodular**: $\theta(0,0) + \theta(1,1) = 0 \le 2\lambda = \theta(0,1) + \theta(1,0)$. Kolmogorov–Zabih show that a binary energy is graph-representable exactly under this condition, so a single min-cut computation gives the *global* MAP — not an approximation. Enumeration is $2^N$; min cut is polynomial.

*Key takeaway:* Submodularity is the discrete analogue of convexity; it is what converts an exponential MAP inference problem into one max-flow call.

### Problem L2.2: The Hungarian Algorithm on a $3 \times 3$ Assignment

Assign three workers to three jobs with cost matrix

$$
c = \begin{pmatrix} 4 & 1 & 3 \\ 2 & 0 & 5 \\ 3 & 2 & 2 \end{pmatrix}
$$

Run the Hungarian method, report the optimal assignment, and verify optimality with the dual potentials.

**Solution**

**Step 1 — row reduction.** Row minima $(1, 0, 2)$; subtract:

$$
\begin{pmatrix} 3 & 0 & 2 \\ 2 & 0 & 5 \\ 1 & 0 & 0 \end{pmatrix}
$$

**Step 2 — column reduction.** Column minima $(1, 0, 0)$; subtract from column 1:

$$
\begin{pmatrix} 2 & 0 & 2 \\ 1 & 0 & 5 \\ 0 & 0 & 0 \end{pmatrix}, \qquad u = (1, 0, 2), \ v = (1, 0, 0)
$$

**Step 3 — matching on zeros.** Rows 1 and 2 have their only zero in column 2, so the equality subgraph admits a matching of size $2$ at best (Hall violator $S = \lbrace \text{row }1, \text{row }2 \rbrace$ with $\vert N(S) \vert = 1$). A minimum cover of the zeros is $\lbrace \text{column } 2, \ \text{row } 3 \rbrace$, of size $2 \lt 3$ — König says no perfect matching yet.

**Step 4 — dual update.** The minimum uncovered slack is $\theta = \min \lbrace 2, 2, 1, 5 \rbrace = 1$. Subtract $\theta$ from uncovered entries and add it to doubly covered ones:

$$
\begin{pmatrix} 1 & 0 & 1 \\ 0 & 0 & 4 \\ 0 & 1 & 0 \end{pmatrix}, \qquad u = (2, 1, 2), \ v = (1, -1, 0)
$$

**Step 5 — perfect matching on tight edges.** Tight entries ($u_i + v_j = c_{ij}$) are $(1,2), (2,1), (2,2), (3,1), (3,3)$, which contain the perfect matching $1 \to 2$, $2 \to 1$, $3 \to 3$:

$$
\text{cost} = c_{12} + c_{21} + c_{33} = 1 + 2 + 2 = 5
$$

**Optimality certificate (LP duality).** The potentials satisfy $u_i + v_j \le c_{ij}$ for all nine pairs, and

$$
\sum_i u_i + \sum_j v_j = (2 + 1 + 2) + (1 - 1 + 0) = 5
$$

Primal value equals dual value, so both are optimal — complementary slackness holds because every assigned cell is tight.

$$
\boxed{\sigma = (1 \to 2, \ 2 \to 1, \ 3 \to 3), \quad \text{cost} = 5 = \sum_i u_i + \sum_j v_j}
$$

*Key takeaway:* The Hungarian method is primal–dual: it never enumerates the $n! = 6$ permutations, it grows a matching inside the tight subgraph and repairs the duals when König blocks progress.

### Problem L2.3: Entropic Optimal Transport and the Zero-Temperature Limit

With cost $C = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}$ and uniform marginals $a = b = (1/2, 1/2)$, compute the entropic optimal coupling $X_\varepsilon$ for regularization $\varepsilon$, evaluate its transport cost at $\varepsilon = 1$ and $\varepsilon = 0.1$, and take $\varepsilon \to 0$.

**Solution**

The entropic problem $\min_X \langle C, X \rangle - \varepsilon H(X)$ over couplings has the closed form $X_\varepsilon = \mathrm{diag}(\alpha) K \mathrm{diag}(\beta)$ with the Gibbs kernel

$$
K = e^{-C/\varepsilon} = \begin{pmatrix} 1 & k \\ k & 1 \end{pmatrix}, \qquad k = e^{-1/\varepsilon}
$$

By symmetry the Sinkhorn scalings are equal, $\alpha = \beta = \gamma \mathbf{1}$, and the row-sum constraint $\gamma^2 (1 + k) = 1/2$ gives

$$
X_\varepsilon = \frac{1}{2(1+k)} \begin{pmatrix} 1 & k \\ k & 1 \end{pmatrix}, \qquad \langle C, X_\varepsilon \rangle = \frac{2k}{2(1+k)} = \frac{k}{1+k}
$$

| $\varepsilon$ | $k = e^{-1/\varepsilon}$ | Transport cost $k/(1+k)$ |
|---|---|---|
| $1$ | $0.3679$ | $0.2689$ |
| $0.1$ | $4.54 \times 10^{-5}$ | $4.54 \times 10^{-5}$ |
| $\to 0$ | $\to 0$ | $\to 0$ |

The unregularized optimum is the permutation coupling $X_0 = \tfrac{1}{2} I$ with cost $0$ — the identity assignment, exactly what the Hungarian algorithm returns.

$$
\boxed{X_\varepsilon = \frac{1}{2(1+e^{-1/\varepsilon})} \begin{pmatrix} 1 & e^{-1/\varepsilon} \\ e^{-1/\varepsilon} & 1 \end{pmatrix} \ \xrightarrow[\varepsilon \to 0]{} \ \tfrac{1}{2} I, \quad \text{cost} \to 0}
$$

**Why practitioners pay the $\varepsilon$ tax.** $X_\varepsilon$ is smooth and strictly positive, so it is differentiable with respect to $C$ — you can back-propagate through the coupling. Sinkhorn converges linearly at rate depending on $\varepsilon$, at $O(n^2)$ per iteration, versus the $O(n^3)$ non-differentiable Hungarian solve. The price is a bias of order $\varepsilon$ (visible above as $0.27$ at $\varepsilon = 1$), which Sinkhorn divergences debias.

*Key takeaway:* Hard matching sits at zero temperature of a one-parameter family; raising the temperature buys differentiability and speed and costs exactness.

### Problem L2.4: Capacity-Constrained Expert Routing

Six tokens must be routed to three experts, each expert able to accept at most $2$ tokens. Allowed pairs: $t_1, t_2, t_3 \to \lbrace e_1, e_2 \rbrace$; $t_4, t_6 \to \lbrace e_1, e_3 \rbrace$; $t_5 \to \lbrace e_2, e_3 \rbrace$. (a) Model as a flow and decide feasibility. (b) Repeat when $t_4, t_5, t_6$ are all restricted to $\lbrace e_1, e_2 \rbrace$.

**Solution**

**(a) Flow model.** Source $s$ with arcs $s \to t_i$ of capacity $1$; arcs $t_i \to e_j$ of capacity $1$ for each allowed pair; arcs $e_j \to \text{sink}$ of capacity $2$ (the expert capacity). Full routing is feasible iff the maximum flow equals $6$.

Generalized Hall condition (capacities on the right): for every token set $S$, the total capacity of $N(S)$ must be at least $\vert S \vert$. Checking the tight cases: $S = \lbrace t_1,t_2,t_3 \rbrace$ has $N(S) = \lbrace e_1,e_2 \rbrace$ with capacity $4 \ge 3$ ✓; $S = \lbrace t_4,t_6 \rbrace$ has capacity $4 \ge 2$ ✓; $S$ = all tokens has capacity $6 \ge 6$ ✓. An explicit saturating assignment:

$$
e_1 \leftarrow \lbrace t_1, t_2 \rbrace, \qquad e_2 \leftarrow \lbrace t_3, t_5 \rbrace, \qquad e_3 \leftarrow \lbrace t_4, t_6 \rbrace
$$

so $\vert f^{\ast} \vert = 6$: **feasible, with zero dropped tokens**.

**(b) Infeasible variant.** Now $t_1, \dots, t_5$ all point only at $\lbrace e_1, e_2 \rbrace$. Take $S = \lbrace t_1, \dots, t_5 \rbrace$:

$$
\text{capacity of } N(S) = 2 + 2 = 4 \lt 5 = \vert S \vert
$$

The matching cut takes the source arc of the one remaining token $t_6$ plus the two expert-to-sink arcs, of capacity $1 + 2 + 2 = 5$, so the maximum flow is $5$: **exactly one token must be dropped or overflow to a slower path**.

$$
\boxed{\text{(a) } \vert f^{\ast} \vert = 6 \text{ feasible}; \quad \text{(b) } \vert f^{\ast} \vert = 5, \text{ deficiency } 1 \text{ certified by } S = \lbrace t_1,\dots,t_5 \rbrace}
$$

*Key takeaway:* Expert-capacity routing in sparse mixture-of-experts layers is a $b$-matching; token dropping is not a bug of the router but a *certified* consequence of a Hall deficiency in the routing graph.

### Problem L2.5: Attention Is Row-Stochastic, Matching Is Doubly Stochastic

Two queries attend over two keys with logits $L = \begin{pmatrix} 3 & 1 \\ 2.9 & 1 \end{pmatrix}$. (a) Compute the softmax attention map and its column sums. (b) Give the maximum-weight matching. (c) Perform one Sinkhorn column normalization and comment.

**Solution**

**(a) Softmax rows.** With $e^{3} = 20.09$, $e^{2.9} = 18.17$, $e^{1} = 2.72$:

$$
A = \begin{pmatrix} 0.881 & 0.119 \\ 0.870 & 0.130 \end{pmatrix}, \qquad \text{column sums } (1.751, \ 0.249)
$$

Rows sum to $1$ by construction, but the columns do **not**: key $1$ is claimed almost twice over while key $2$ is nearly ignored. Softmax attention imposes no exclusivity.

**(b) Maximum-weight matching.** Only two permutations exist:

$$
\text{score}(q_1 \to k_1, \ q_2 \to k_2) = 3 + 1 = 4, \qquad \text{score}(q_1 \to k_2, \ q_2 \to k_1) = 1 + 2.9 = 3.9
$$

so the optimal matching is $q_1 \to k_1$, $q_2 \to k_2$ — it *forces* the weaker query onto the unclaimed key, which attention never does.

**(c) One Sinkhorn step.** Divide each column by its sum:

$$
A' = \begin{pmatrix} 0.881/1.751 & 0.119/0.249 \\ 0.870/1.751 & 0.130/0.249 \end{pmatrix} = \begin{pmatrix} 0.503 & 0.478 \\ 0.497 & 0.522 \end{pmatrix}
$$

Columns now sum to $1$ exactly, rows to $0.981$ and $1.019$. Alternating row and column normalizations converges (Sinkhorn) to a doubly stochastic matrix; sharpening the logits (equivalently $\varepsilon \to 0$ in Problem L2.3) drives that limit to the permutation of part (b).

$$
\boxed{\text{softmax: rows sum to } 1 \text{ only; Sinkhorn: rows and columns; matching: a permutation}}
$$

*Key takeaway:* "Attention as matching" is a metaphor that becomes literal only after column normalization and temperature annealing — which is exactly what Sinkhorn attention, slot attention and Hungarian set-prediction losses add.

### Problem L2.6: Menger and Fault Tolerance in a Two-Layer Fabric

A fabric has $s \to a_1, a_2, a_3$, then $a_1 \to b_1$, $a_2 \to b_1$, $a_2 \to b_2$, $a_3 \to b_2$, then $b_1 \to t$, $b_2 \to t$, all links of capacity $1$. How many arc-disjoint $s$–$t$ paths exist, what is the minimum cut, and how many simultaneous link failures does the fabric survive?

**Solution**

**Upper bound.** Every $s$–$t$ path must use $b_1 \to t$ or $b_2 \to t$, and each has capacity $1$, so $\vert f^{\ast} \vert \le 2$.

**Achieving it.** $s \to a_1 \to b_1 \to t$ and $s \to a_3 \to b_2 \to t$ are arc-disjoint, giving $\vert f^{\ast} \vert = 2$. (The middle switch $a_2$ carries nothing — capacity is limited by the core layer, not the edge layer: textbook oversubscription.)

**Minimum cut.** $S = \lbrace s, a_1, a_2, a_3, b_1, b_2 \rbrace$, $T = \lbrace t \rbrace$ with crossing arcs $b_1 \to t$, $b_2 \to t$:

$$
c(S,T) = 1 + 1 = 2 = \vert f^{\ast} \vert
$$

**Menger's reading.** With unit capacities, max-flow min-cut says: the maximum number of arc-disjoint $s$–$t$ paths equals the minimum number of arcs whose removal disconnects $s$ from $t$. Here both numbers are $2$, so the fabric survives **any single link failure** but is disconnected by the simultaneous loss of $\lbrace b_1 \to t, \ b_2 \to t \rbrace$.

$$
\boxed{2 \text{ arc-disjoint paths} = \text{min cut } 2 \Rightarrow \text{tolerates } 1 \text{ arbitrary link failure}}
$$

*Key takeaway:* Redundancy is not a count of links but a min-cut: adding a fourth edge switch would change nothing, whereas adding a third core switch with its own uplink raises tolerance to two failures.

## Level 3 — Challenge

### Problem L3.1: Menger's Theorem from Max-Flow Min-Cut

Prove the arc version of Menger's theorem — for $s \neq t$, the maximum number of arc-disjoint $s$–$t$ paths equals the minimum size of an arc set whose removal leaves no $s$–$t$ path — and derive the vertex version.

**Solution**

**Setup.** Give every arc capacity $1$ and let $f^{\ast}$ be an integral maximum flow (Theorem D).

**Step 1: flow value $\ge$ number of disjoint paths.** Given $k$ arc-disjoint $s$–$t$ paths, sending one unit along each yields a feasible flow (no arc is reused, so no capacity is exceeded) of value $k$. Hence $\vert f^{\ast} \vert \ge k_{\max}$.

**Step 2: flow decomposition.** Conversely, let $f^{\ast}$ be integral with value $k$. Consider the arcs with $f^{\ast}(e) = 1$. Since $s$ has net outflow $k$ and interior vertices conserve, we can repeatedly trace a route from $s$ forward along unused unit arcs until reaching $t$ (a route cannot get stuck at an interior vertex, which always has an unused outgoing unit arc when it has an unused incoming one). Remove that path's arcs and repeat $k$ times; any leftover unit arcs form cycles carrying no $s$–$t$ value. The result is $k$ **arc-disjoint** $s$–$t$ paths, so $k_{\max} \ge \vert f^{\ast} \vert$. Together, $k_{\max} = \vert f^{\ast} \vert$.

**Step 3: cuts and disconnecting sets.** If $(S,T)$ is an $s$–$t$ cut, deleting its forward-crossing arcs destroys every $s$–$t$ path (a path must cross from $S$ to $T$ at some point), and with unit capacities its size is $c(S,T)$. Conversely, if $F$ is any arc set whose removal disconnects $s$ from $t$, let $S$ be the set reachable from $s$ in $G - F$; then $t \notin S$ and every forward-crossing arc of $(S, V \setminus S)$ lies in $F$, so $c(S, V \setminus S) \le \vert F \vert$. Hence

$$
\min_{F \text{ disconnecting}} \vert F \vert = \min_{(S,T)} c(S,T)
$$

**Conclusion.** Applying max-flow min-cut (Theorem C):

$$
\boxed{\max \lbrace \text{arc-disjoint } s\text{–}t \text{ paths} \rbrace = \min \lbrace \vert F \vert : G - F \text{ has no } s\text{–}t \text{ path} \rbrace}
$$

**Vertex version.** Split every vertex $v \notin \lbrace s,t \rbrace$ into $v_{\text{in}} \to v_{\text{out}}$ with capacity $1$, redirect incoming arcs to $v_{\text{in}}$ and outgoing arcs from $v_{\text{out}}$, and give all original arcs capacity $+\infty$. A finite cut can only use splitting arcs, so minimum cuts correspond exactly to **vertex** separators, and unit vertex capacities force the paths to be internally vertex-disjoint. Max-flow min-cut then gives: the maximum number of internally vertex-disjoint $s$–$t$ paths equals the minimum size of an $s$–$t$ vertex separator.

*Key takeaway:* Menger, König, Hall and max-flow min-cut are one theorem wearing four costumes; the vertex-splitting gadget is the universal translator between arc statements and vertex statements.

### Problem L3.2: Deficiency Form of Hall, and Regular Bipartite Graphs

(a) Prove $\nu(G) = \vert X \vert - \max_{S \subseteq X} \big( \vert S \vert - \vert N(S) \vert \big)$ for bipartite $G$ with parts $X, Y$. (b) Prove that every $k$-regular bipartite graph ($k \ge 1$) has a perfect matching, and that its edges decompose into $k$ perfect matchings.

**Solution**

**(a) Deficiency formula.** Write $d = \max_{S \subseteq X} ( \vert S \vert - \vert N(S) \vert ) \ge 0$ (the value $0$ is attained at $S = \emptyset$).

*Upper bound $\nu \le \vert X \vert - d$.* Let $S$ attain the maximum. Any matching can match at most $\vert N(S) \vert$ vertices of $S$, so it leaves at least $\vert S \vert - \vert N(S) \vert = d$ vertices of $X$ unmatched, giving $\nu \le \vert X \vert - d$.

*Lower bound $\nu \ge \vert X \vert - d$.* Build $G'$ by adding $d$ new vertices to $Y$, each joined to **every** vertex of $X$. For any $S \subseteq X$ with $S \neq \emptyset$,

$$
\vert N_{G'}(S) \vert = \vert N_G(S) \vert + d \ge \vert N_G(S) \vert + \big( \vert S \vert - \vert N_G(S) \vert \big) = \vert S \vert
$$

so Hall's condition holds in $G'$ and there is a matching $M'$ saturating $X$. Deleting the at most $d$ edges of $M'$ that touch new vertices leaves a matching of $G$ of size at least $\vert X \vert - d$.

$$
\boxed{\nu(G) = \vert X \vert - \max_{S \subseteq X} \big( \vert S \vert - \vert N(S) \vert \big)}
$$

**(b) Regular bipartite graphs.** First, counting edges from each side gives $k\vert X \vert = \vert E \vert = k \vert Y \vert$, so $\vert X \vert = \vert Y \vert$. Now verify Hall for any $S \subseteq X$: the $k\vert S \vert$ edges leaving $S$ all land in $N(S)$, and each vertex of $N(S)$ absorbs at most $k$ of them, so

$$
k \vert S \vert \le k \vert N(S) \vert \quad \Longrightarrow \quad \vert N(S) \vert \ge \vert S \vert
$$

Hall gives a matching saturating $X$, which is **perfect** because $\vert X \vert = \vert Y \vert$. Removing it leaves a $(k-1)$-regular bipartite graph; induction on $k$ (base $k = 0$: no edges) decomposes $E$ into exactly $k$ perfect matchings.

$$
\boxed{k\text{-regular bipartite} \Rightarrow \text{a perfect matching exists and } E = M_1 \sqcup \dots \sqcup M_k}
$$

**Corollaries.** The decomposition is a proper edge colouring with $k$ colours, so $\chi'(G) = \Delta(G)$ for bipartite graphs (König's edge-colouring theorem) — the mathematics behind conflict-free round-robin scheduling, crossbar switch timetabling, and the Birkhoff decomposition of a doubly stochastic matrix into permutations.

*Key takeaway:* Regularity makes Hall's condition automatic by a two-line double count — the cleanest example of a structural hypothesis discharging a combinatorial obligation.

### Problem L3.3: Ford–Fulkerson with Irrational Capacities

Suppose a network contains three "special" arcs whose residual capacities, after a block of four augmentations, return to the same pattern scaled by $r = \tfrac{\sqrt5 - 1}{2}$, with the block pushing $r^{n}$ units at stage $n$. (a) Show the total flow pushed is finite. (b) Conclude that Ford–Fulkerson need not terminate. (c) Explain what Edmonds–Karp changes.

**Solution**

**Key identity.** $r$ is the positive root of $r^2 + r - 1 = 0$, so

$$
r^2 = 1 - r \qquad \text{and} \qquad r^{n+2} = r^n - r^{n+1}
$$

This is precisely what makes the residual pattern self-similar: after each block of four augmentations the three special residual capacities are $(r^{n}, r^{n+1}, 0)$ up to relabelling, i.e. the *same shape* rescaled by $r$. Since $0 \lt r \lt 1$, the pattern never becomes degenerate and augmenting paths always exist — the algorithm never stops.

**(a) The total pushed is finite.** Suppose the initial augmentation pushes $1$ unit and each subsequent block pushes $2r^{n}$ across the two special routes. Then

$$
\sum_{\text{all augmentations}} \beta = 1 + 2\sum_{n \ge 1} r^{n} = 1 + \frac{2r}{1-r} = 1 + \frac{2r}{r^2} = 1 + \frac{2}{r} = 1 + 2\varphi \approx 4.236
$$

using $1 - r = r^2$ and $1/r = \varphi = \tfrac{1+\sqrt5}{2}$. The flow value therefore converges to a finite limit.

**(b) Non-termination and non-convergence to the optimum.** Give the network's non-special arcs capacity $M$; the true maximum flow is $\Theta(M)$, which can be made arbitrarily larger than the limit $1 + 2\varphi$. So the infinite sequence of augmentations converges to a flow value **strictly below** the maximum:

$$
\boxed{\lim_{k \to \infty} \vert f_k \vert = 1 + 2\varphi \approx 4.236 \ \lt \ \vert f^{\ast} \vert = \Theta(M)}
$$

Ford–Fulkerson thus fails both to terminate and to converge to the right answer (Ford & Fulkerson noted this; Zwick, 1995, gave the smallest such network, on $6$ vertices and $8$ arcs).

**(c) Why Edmonds–Karp is immune.** Theorem E bounds the number of augmentations by $O(nm)$ using only the *combinatorial* facts that residual BFS distances never decrease and that each arc becomes critical $O(n)$ times. No capacity value enters the argument, so irrational capacities are irrelevant: the algorithm halts after at most $O(nm)$ augmentations no matter what the numbers are. The same immunity holds for Dinic and push–relabel.

*Key takeaway:* "The algorithm makes progress each step" is not the same as "the algorithm terminates" — with real-valued data, progress can be summable. Strongly polynomial bounds exist precisely to eliminate that possibility.

### Problem L3.4: Birkhoff–von Neumann and Why Relaxation Is Exact

(a) Prove that the extreme points of the set of $n \times n$ doubly stochastic matrices are exactly the permutation matrices. (b) Deduce that the assignment LP has an integral optimum. (c) Explain the consequence for differentiable matching in machine learning.

**Solution**

Let $\mathcal{B}_n = \lbrace X \ge 0 : X \mathbf{1} = \mathbf{1}, \ X^{\top} \mathbf{1} = \mathbf{1} \rbrace$, a compact convex polytope.

**(a) Permutations are extreme.** Let $P$ be a permutation matrix and suppose $P = \tfrac12 (Y + Z)$ with $Y, Z \in \mathcal{B}_n$. Wherever $P_{ij} = 0$, nonnegativity forces $Y_{ij} = Z_{ij} = 0$; the remaining entries are then determined by the row sums, giving $Y = Z = P$. So $P$ is an extreme point.

**Non-permutations are not extreme.** Let $X \in \mathcal{B}_n$ have some entry $X_{i_1 j_1} \in (0,1)$. Because row $i_1$ sums to $1$, it contains another fractional entry $X_{i_1 j_2}$; because column $j_2$ sums to $1$, it contains another fractional entry $X_{i_2 j_2}$; continuing alternately between rows and columns among finitely many cells, some cell repeats, producing a **cycle** $(i_1 j_1), (i_1 j_2), (i_2 j_2), \dots$ of even length in the bipartite support graph, all of whose entries are strictly between $0$ and $1$. Let $\Delta$ be the matrix with $+1$ and $-1$ alternating around this cycle and $0$ elsewhere. Every row and every column of $\Delta$ sums to $0$, so for small $\varepsilon \gt 0$ both $X + \varepsilon \Delta$ and $X - \varepsilon \Delta$ lie in $\mathcal{B}_n$, and

$$
X = \tfrac12 \big( (X + \varepsilon\Delta) + (X - \varepsilon\Delta) \big)
$$

exhibits $X$ as a proper convex combination — not extreme. $\blacksquare$

**(b) Exact relaxation.** A linear objective over a compact polytope attains its minimum at an extreme point, so

$$
\boxed{\min_{X \in \mathcal{B}_n} \langle C, X \rangle = \min_{\sigma \in S_n} \sum_{i} c_{i\sigma(i)}}
$$

The continuous relaxation of the assignment problem has **no integrality gap**: solving the LP (or running the Hungarian algorithm, which builds primal and dual solutions in lockstep) already returns a permutation. Equivalently, the constraint matrix is totally unimodular. Corollary (Birkhoff): every doubly stochastic matrix is a convex combination of at most $(n-1)^2 + 1$ permutation matrices.

**(c) Consequence in machine learning.** Because $\mathcal{B}_n$ is the exact convex hull of permutations, it is the right relaxation for *learning* permutations: Sinkhorn/Gumbel-Sinkhorn networks output points inside $\mathcal{B}_n$, back-propagate through them, and anneal the temperature so the iterate approaches a vertex — which Theorem I guarantees is a genuine permutation. Contrast the *quadratic* assignment problem (graph alignment, Gromov–Wasserstein): the objective is no longer linear, minima need not sit at vertices, and the relaxation becomes genuinely lossy, which is why those problems are NP-hard while linear assignment is $O(n^3)$.

*Key takeaway:* Linearity plus the Birkhoff polytope equals exactness; the moment the objective becomes quadratic, the vertex argument collapses and hardness returns.
