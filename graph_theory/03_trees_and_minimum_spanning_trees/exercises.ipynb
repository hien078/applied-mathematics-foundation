## Level 0 — Concept Check

### Problem L0.1: Tree or Not?

Classify each graph as tree / forest (not tree) / neither: (a) $V = \{1,2,3,4\}$, $E = \{\{1,2\},\{2,3\},\{3,4\}\}$; (b) $V = \{1,2,3,4\}$, $E = \{\{1,2\},\{3,4\}\}$; (c) $V = \{1,2,3\}$, $E = \{\{1,2\},\{2,3\},\{3,1\}\}$.

**Solution**

(a) Connected (a path $1 - 2 - 3 - 4$) and acyclic, with $m = 3 = n - 1$: a **tree**.

(b) Acyclic but two components ($\{1,2\}$ and $\{3,4\}$): a **forest that is not a tree** (note $m = 2 = n - c = 4 - 2$).

(c) The triangle: connected but contains a cycle, and $m = 3 \gt n - 1 = 2$: **neither** tree nor forest.

$$
\boxed{\text{(a) tree, (b) forest with 2 components, (c) neither}}
$$

*Key takeaway:* Check any two of {connected, acyclic, $m = n-1$}; the third follows automatically.

### Problem L0.2: Edge Counting for Spanning Trees

A connected graph has $n = 12$ vertices and $m = 20$ edges. How many edges does each of its spanning trees contain, and how many edges must be deleted to obtain one?

**Solution**

Every spanning tree on $n = 12$ vertices has exactly $n - 1 = 11$ edges (fundamental tree count). Reaching a spanning tree from $G$ requires deleting

$$
m - (n-1) = 20 - 11 = 9
$$

edges — one per independent cycle. The number $m - n + 1$ is the **cycle rank** (circuit rank) of the graph.

$$
\boxed{11 \text{ tree edges}; \quad 9 = m - n + 1 \text{ deletions (the cycle rank)}}
$$

*Key takeaway:* $m - n + 1$ measures a connected graph's redundancy — it counts independent cycles and equals the number of non-tree "chords."

### Problem L0.3: Leaves of Standard Trees

How many leaves do the following trees on $n \ge 3$ vertices have: (a) the path $P_n$, (b) the star $K_{1, n-1}$? Which trees attain the minimum possible number of leaves?

**Solution**

(a) The path $P_n$ has exactly **2** leaves (its endpoints); every interior vertex has degree 2.

(b) The star $K_{1,n-1}$ has $n - 1$ leaves — every vertex except the center. This is the maximum possible.

By the leaf lemma, every tree with $n \ge 2$ vertices has **at least 2** leaves, so the path attains the minimum. (Sketch: the endpoints of a longest path must be leaves.)

$$
\boxed{P_n: 2 \text{ leaves (minimum possible)}; \quad K_{1,n-1}: n-1 \text{ leaves (maximum)}}
$$

*Key takeaway:* Paths and stars are the two extremes of tree shape: minimum leaves / maximum diameter versus maximum leaves / minimum diameter.

### Problem L0.4: When Is the MST Unique?

State a sufficient condition for the MST of a connected weighted graph to be unique, and outline why it works.

**Solution**

**Condition: all edge weights are distinct.**

Why it suffices: with distinct weights, every edge is decisively classified by the two exchange properties —

- an edge that is the (unique) lightest across *some* cut lies in **every** MST (cut property);
- an edge that is the (unique) heaviest on *some* cycle lies in **no** MST (cycle property).

For a tree $T$ produced by Kruskal/Prim, each tree edge is the unique lightest across the cut it spanned when added, and each non-tree edge is the unique heaviest on the cycle it would close. Any two MSTs would have to contain exactly the same "forced" edges, hence coincide.

$$
\boxed{\text{Distinct edge weights} \implies \text{the MST is unique}}
$$

*Key takeaway:* Ties are the only source of MST ambiguity; in practice, perturbing weights by tiny distinct amounts (or lexicographic tie-breaking) forces uniqueness.

## Level 1 — Foundation

### Problem L1.1: Forests and Components

Prove that a forest with $n$ vertices and $c$ connected components has exactly $n - c$ edges.

**Solution**

Each component of a forest is connected and acyclic — a tree. Let the components have $n_1, \dots, n_c$ vertices with $\sum_i n_i = n$. Component $i$, being a tree, has $n_i - 1$ edges. Summing:

$$
m = \sum_{i=1}^{c} (n_i - 1) = \left( \sum_{i=1}^{c} n_i \right) - c = n - c
$$

$$
\boxed{m_{\text{forest}} = n - c}
$$

*Key takeaway:* The formula interpolates between the empty graph ($c = n$, $m = 0$) and a tree ($c = 1$, $m = n-1$); each added forest edge decreases $c$ by exactly one — the invariant driving Kruskal's algorithm.

### Problem L1.2: Unique Paths Characterize Trees

Prove: a graph $T$ is a tree $\iff$ every pair of vertices is joined by exactly one path.

**Solution**

($\Rightarrow$) Connectivity gives at least one $u$–$v$ path. Suppose two distinct $u$–$v$ paths $P \neq Q$ existed. Walking along $P$ from $u$, let $x$ be the last vertex before $P$ and $Q$ first diverge, and let $y$ be the first vertex after $x$ where they meet again (exists, since both end at $v$). The segments of $P$ and $Q$ between $x$ and $y$ are internally disjoint and not identical, so their union is a cycle — contradicting acyclicity.

($\Leftarrow$) Unique paths give connectivity (at least one path). If $T$ contained a cycle $C$, any two distinct vertices on $C$ would be joined by the two arcs of $C$ — two distinct paths, contradiction. Hence acyclic, so $T$ is a tree. $\blacksquare$

$$
\boxed{\text{tree} \iff \text{every vertex pair has exactly one connecting path}}
$$

*Key takeaway:* "Unique routing" is the functional definition of a tree — it is why tree topologies need no routing tables and why tree-structured models admit exact dynamic programming.

### Problem L1.3: Kruskal Step by Step

Run Kruskal's algorithm on $V = \{a,b,c,d,e\}$ with weighted edges: $\{a,b\}{:}1$, $\{b,c\}{:}2$, $\{a,c\}{:}3$, $\{c,d\}{:}4$, $\{b,d\}{:}5$, $\{d,e\}{:}6$, $\{c,e\}{:}7$. Show accept/reject decisions and the final MST weight.

**Solution**

Scan edges in increasing weight, accepting iff endpoints lie in different components:

| Edge | $w$ | Components before | Decision |
|---|---|---|---|
| $\{a,b\}$ | 1 | $\{a\},\{b\},\{c\},\{d\},\{e\}$ | accept |
| $\{b,c\}$ | 2 | $\{a,b\},\{c\},\{d\},\{e\}$ | accept |
| $\{a,c\}$ | 3 | $\{a,b,c\},\{d\},\{e\}$ | reject (cycle $a,b,c$) |
| $\{c,d\}$ | 4 | $\{a,b,c\},\{d\},\{e\}$ | accept |
| $\{b,d\}$ | 5 | $\{a,b,c,d\},\{e\}$ | reject (cycle) |
| $\{d,e\}$ | 6 | $\{a,b,c,d\},\{e\}$ | accept — 4 edges, stop |

MST edges: $\{a,b\}, \{b,c\}, \{c,d\}, \{d,e\}$ with total weight

$$
w(T^{\ast}) = 1 + 2 + 4 + 6 = 13
$$

$$
\boxed{T^{\ast} = \{ab, bc, cd, de\}, \quad w(T^{\ast}) = 13}
$$

*Key takeaway:* Kruskal touches every edge once in sorted order; rejections are exactly the cycle-property exclusions, acceptances the cut-property inclusions.

### Problem L1.4: Prim Step by Step

Run Prim's algorithm from vertex $a$ on the same graph as Problem L1.3, showing the frontier edge chosen at each step. Confirm the same tree emerges.

**Solution**

Maintain the tree set $S$ and pick the lightest edge leaving $S$:

1. $S = \{a\}$: candidate edges $\{a,b\}{:}1$, $\{a,c\}{:}3$ → take $\{a,b\}$ (1).
2. $S = \{a,b\}$: candidates $\{b,c\}{:}2$, $\{a,c\}{:}3$, $\{b,d\}{:}5$ → take $\{b,c\}$ (2).
3. $S = \{a,b,c\}$: candidates $\{c,d\}{:}4$, $\{b,d\}{:}5$, $\{c,e\}{:}7$ → take $\{c,d\}$ (4).
4. $S = \{a,b,c,d\}$: candidates $\{d,e\}{:}6$, $\{c,e\}{:}7$ → take $\{d,e\}$ (6).

$S = V$: stop. Tree $\{ab, bc, cd, de\}$, weight $13$ — identical to Kruskal's output, as guaranteed by MST uniqueness under distinct weights.

$$
\boxed{\text{Prim's order: } ab, bc, cd, de; \quad w = 13 \text{ (same unique MST)}}
$$

*Key takeaway:* Each Prim step is one application of the cut property to the cut $(S, V \setminus S)$; different greedy schedules, same optimum.

### Problem L1.5: Cayley for $n = 4$

Use Cayley's formula to count labeled trees on 4 vertices, and verify by classifying them by shape (path vs. star).

**Solution**

Cayley: $\tau(K_4) = 4^{4-2} = 16$.

Shapes of trees on 4 vertices: only the path $P_4$ (degrees $1,2,2,1$) and the star $K_{1,3}$ (degrees $3,1,1,1$).

- **Stars**: choosing the center determines the star: $4$ labeled stars.
- **Paths**: a labeled path corresponds to an ordering of the 4 labels, with each path counted twice (once per direction): $\frac{4!}{2} = 12$.

Total: $4 + 12 = 16$ ✓.

$$
\boxed{\tau(K_4) = 16 = 12 \text{ paths} + 4 \text{ stars}}
$$

*Key takeaway:* Cayley's formula is shape-blind; decomposing by degree sequence (equivalently, by Prüfer content) recovers the finer classification.

### Problem L1.6: One New Edge, Exactly One Cycle

Prove: adding any edge $e = \{u,v\} \notin E(T)$ to a tree $T$ creates exactly one cycle.

**Solution**

**At least one cycle**: $T$ contains a unique $u$–$v$ path $P$ (Problem L1.2); $P + e$ is a cycle.

**At most one cycle**: any cycle $C$ in $T + e$ must use $e$ (else $C \subseteq T$, impossible in an acyclic graph). Then $C - e$ is a $u$–$v$ path in $T$, which is unique, forcing $C = P + e$. Hence the cycle is unique. $\blacksquare$

$$
\boxed{T + e \text{ contains the single cycle } P_{uv} + e \text{ (the fundamental cycle of } e)}
$$

*Key takeaway:* Fundamental cycles — one per non-tree edge — form a basis of the cycle space; they are the objects the cycle property tests during MST verification.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: Single-Linkage Clustering via the MST

Five data points have MST edges (from Problem L1.3's tree): $ab{:}1$, $bc{:}2$, $cd{:}4$, $de{:}6$. Produce the single-linkage clustering into $k = 3$ clusters, and state the general MST–clustering correspondence.

**Solution**

**Correspondence**: single-linkage clustering with $k$ clusters is obtained by deleting the $k-1$ **heaviest** edges of the MST; the resulting components are the clusters (merging order in the dendrogram = Kruskal's acceptance order).

Here the two heaviest MST edges are $de$ (6) and $cd$ (4). Deleting them leaves components:

$$
\{a, b, c\}, \qquad \{d\}, \qquad \{e\}
$$

The cluster $\{a,b,c\}$ holds together because its internal merge distances (1, 2) are below the deleted thresholds.

$$
\boxed{k = 3 \text{ clusters: } \{a,b,c\}, \ \{d\}, \ \{e\} \text{ (cut MST edges of weight } 6, 4)}
$$

*Key takeaway:* The MST *is* the single-linkage dendrogram in compressed form — clustering, a statistics problem, reduces exactly to a graph optimization problem.

### Problem L2.2: Chow–Liu Tree

Four random variables have pairwise mutual information (in bits): $I(X_1,X_2) = 0.9$, $I(X_1,X_3) = 0.3$, $I(X_1,X_4) = 0.2$, $I(X_2,X_3) = 0.7$, $I(X_2,X_4) = 0.4$, $I(X_3,X_4) = 0.6$. Find the Chow–Liu tree (the best tree-structured approximation of the joint distribution) and its total mutual information.

**Solution**

Chow–Liu (1968): the KL-optimal tree-structured distribution corresponds to the **maximum-weight spanning tree** of the mutual-information graph. Run Kruskal on *decreasing* weights:

1. $I_{12} = 0.9$: accept.
2. $I_{23} = 0.7$: accept.
3. $I_{34} = 0.6$: accept — 3 edges on 4 vertices: done.

(Ranked next were $0.4, 0.3, 0.2$, all cycle-closing.) The tree is the path $X_1 - X_2 - X_3 - X_4$ with total weight

$$
\sum_{e \in T} I_e = 0.9 + 0.7 + 0.6 = 2.2 \text{ bits}
$$

The fitted model factorizes as $p(x_1) p(x_2 \mid x_1) p(x_3 \mid x_2) p(x_4 \mid x_3)$.

$$
\boxed{T_{\text{CL}} = \{12, 23, 34\}, \quad \textstyle\sum I = 2.2 \text{ bits}}
$$

*Key takeaway:* Maximizing $\sum_e I_e$ over trees minimizes the KL divergence to the true joint — structure learning for tree graphical models is literally an MST computation.

### Problem L2.3: Network Build-Out vs. Star Topology

Five sites have link costs given by Problem L1.3's graph (MST weight 13). A consultant proposes instead a star centered at $c$ using links $\{a,c\}{:}3$, $\{b,c\}{:}2$, $\{c,d\}{:}4$, $\{c,e\}{:}7$. Compare total costs and name the trade-off the star buys.

**Solution**

Star cost:

$$
w(\text{star}) = 3 + 2 + 4 + 7 = 16
$$

MST cost is $13$ (computed in L1.3), so the star overpays by $16 - 13 = 3$, about $23\%$.

What the star buys: **hop count**. In the star, every pair of sites communicates in at most 2 hops through the hub $c$; in the MST (a path), $a$ and $e$ are 4 hops apart. The star also concentrates failure risk at $c$ (single point of failure, degree 4), while the path spreads it.

$$
\boxed{w_{\text{MST}} = 13 \lt w_{\text{star}} = 16; \text{ the extra } 3 \text{ buys diameter } 2 \text{ at the price of hub fragility}}
$$

*Key takeaway:* MST optimizes construction cost only; latency (diameter), resilience, and load call for different — often costlier — spanning structures. Real network design navigates this Pareto frontier.

### Problem L2.4: Bottleneck Paths and Percolation

Using the MST of Problem L1.3 ($ab{:}1, bc{:}2, cd{:}4, de{:}6$), find the minimax (bottleneck) path value between $a$ and $e$ — the smallest possible value of the maximum edge weight over all $a$–$e$ paths in the full graph. Interpret in percolation terms.

**Solution**

**Fact**: for any two vertices, the MST path minimizes the maximum edge weight among all connecting paths (proved in Problem L3.2). The MST path from $a$ to $e$ is $a - b - c - d - e$ with edge weights $1, 2, 4, 6$, so the bottleneck value is

$$
\beta(a, e) = \max\{1, 2, 4, 6\} = 6
$$

No alternative path does better: every $a$–$e$ path must leave the component $\{a,b,c,d\}$, and the two available crossing edges have weights 6 ($de$) and 7 ($ce$), so a maximum of at least 6 is unavoidable — confirming optimality.

**Percolation reading**: if edges "open" in increasing weight (bond percolation with thresholds), $a$ and $e$ first become connected exactly when the threshold reaches $\beta(a,e) = 6$ — Kruskal's process *is* the percolation process, and MST bottlenecks are first-connection thresholds.

$$
\boxed{\beta(a,e) = 6, \text{ attained by the MST path}}
$$

*Key takeaway:* MSTs answer minimax questions exactly: widest paths, percolation thresholds, and "weakest-link" routing all read answers off the MST.

### Problem L2.5: Matrix-Tree Count for the Cycle $C_4$

The Laplacian eigenvalues of the cycle $C_4$ are $0, 2, 2, 4$. Use the matrix-tree theorem to count its spanning trees, and verify combinatorially.

**Solution**

Kirchhoff's spectral form:

$$
\tau(C_4) = \frac{1}{n} \prod_{k=2}^{n} \lambda_k = \frac{1}{4} (2 \cdot 2 \cdot 4) = 4
$$

**Combinatorial check**: a spanning tree of a cycle is obtained by deleting exactly one of its edges (removing one edge keeps connectivity and yields $n-1$ edges; removing more disconnects). $C_4$ has 4 edges, hence 4 spanning trees. ✓

In general $\tau(C_n) = n$, consistent with the Laplacian eigenvalues $2 - 2\cos(2\pi k/n)$.

$$
\boxed{\tau(C_4) = \frac{2 \cdot 2 \cdot 4}{4} = 4}
$$

*Key takeaway:* The matrix-tree theorem turns exponential enumeration into one determinant/spectrum computation — a first taste of the spectral methods of Topic 06.

### Problem L2.6: Broadcast Trees in Distributed Training

A parameter server must broadcast a gradient to $n = 8$ workers. Sending one message takes 1 time unit; a node can send to one other node per unit, and nodes that have received may retransmit. Compare broadcast completion time for (a) the star (server sends to each worker in turn) and (b) a balanced binary spanning tree (binomial-tree schedule). Which spanning structure is optimal?

**Solution**

(a) **Star schedule**: the server transmits sequentially: worker $k$ finishes at time $k$. Completion time

$$
T_{\text{star}} = 8
$$

(b) **Binomial-tree schedule**: every node that has the message retransmits each round, doubling the informed set: $1 \to 2 \to 4 \to 8 \to 16$. Reaching all $9$ nodes (server + 8 workers) takes $\lceil \log_2 9 \rceil = 4$ rounds:

$$
T_{\text{tree}} = 4
$$

The doubling schedule is optimal: in one round the informed set can at most double, so any schedule needs at least $\lceil \log_2(n+1) \rceil$ rounds. The underlying communication pattern is a spanning tree (the binomial tree), not the star.

$$
\boxed{T_{\text{star}} = 8 \text{ vs. } T_{\text{tree}} = \lceil \log_2 9 \rceil = 4 \text{ (optimal)}}
$$

*Key takeaway:* Collective-communication primitives (broadcast, all-reduce) in distributed deep learning are spanning-tree schedules; tree shape — not just tree weight — governs performance.

## Level 3 — Challenge

### Problem L3.1: The Path-Maximum Criterion for MST Optimality

Let $T$ be a spanning tree of a weighted graph $G$ with distinct weights. Prove: $T$ is the MST $\iff$ for every non-tree edge $e = \{u,v\}$, $w(e) \gt w(f)$ for every edge $f$ on the tree path $P_{uv} \subseteq T$.

**Solution**

($\Rightarrow$) Let $T$ be the MST and $e$ a non-tree edge. Adding $e$ creates the fundamental cycle $C = P_{uv} + e$. If some $f \in P_{uv}$ had $w(f) \gt w(e)$, the exchange $T' = T - f + e$ would be a spanning tree ($f$ lies on the cycle $C$, so removing it after adding $e$ preserves connectivity and edge count) with $w(T') \lt w(T)$ — contradiction. Distinctness rules out equality, so $w(e) \gt w(f)$ for all $f \in P_{uv}$.

($\Leftarrow$) Suppose $T$ satisfies the criterion; we show $T$ equals the MST $T^{\ast}$ (unique by distinct weights). Assume not, and pick the lightest edge $f \in T^{\ast} \setminus T$. Adding $f$ to $T$ closes a cycle $P + f$ in which, by the criterion, $f$ is the strict maximum. Hence $f$ is the unique heaviest edge of that cycle — by the **cycle property**, $f$ belongs to no MST, contradicting $f \in T^{\ast}$. Therefore $T^{\ast} \setminus T = \emptyset$, and since both have $n-1$ edges, $T = T^{\ast}$. $\blacksquare$

$$
\boxed{T \text{ is the MST} \iff \text{every non-tree edge is the strict maximum of its fundamental cycle}}
$$

*Key takeaway:* This criterion is checkable with $m$ path-maximum queries — the basis of linear-time MST *verification* (Komlós), which is faster than any known deterministic construction.

### Problem L3.2: The MST Is a Bottleneck Optimizer

Prove: for every pair $u, v$, the $u$–$v$ path in the MST $T^{\ast}$ minimizes $\max_{f \in P} w(f)$ over all $u$–$v$ paths $P$ in $G$ (distinct weights assumed).

**Solution**

Let $P^{\ast}$ be the MST path from $u$ to $v$ and let $f^{\ast} = \arg\max_{f \in P^{\ast}} w(f)$ be its bottleneck edge, with value $\beta = w(f^{\ast})$. Suppose some other $u$–$v$ path $Q$ in $G$ had $\max_{f \in Q} w(f) \lt \beta$.

Delete $f^{\ast}$ from $T^{\ast}$: the tree splits into components $S \ni u$ and $V \setminus S \ni v$ (as $f^{\ast}$ lies on the unique $u$–$v$ tree path). The path $Q$ runs from $u \in S$ to $v \notin S$, so it contains an edge $g$ crossing the cut $(S, V \setminus S)$ with

$$
w(g) \le \max_{f \in Q} w(f) \lt \beta = w(f^{\ast})
$$

Then $T' = T^{\ast} - f^{\ast} + g$ is a spanning tree (reconnecting the two components) with $w(T') \lt w(T^{\ast})$ — contradicting MST minimality. Hence no path beats $\beta$:

$$
\boxed{\min_{P : u \to v} \ \max_{f \in P} w(f) = \max_{f \in P^{\ast}_{uv}} w(f)}
$$

$\blacksquare$

*Key takeaway:* The exchange argument transfers *minimum-sum* optimality into *minimax* optimality for free — the reason MSTs solve widest-path, percolation, and clustering problems that never mention total weight.

### Problem L3.3: Prüfer Encoding and Decoding

(a) Compute the Prüfer sequence of the tree with edges $\{1,5\}, \{2,5\}, \{3,4\}, \{4,5\}$ on labels $\{1,\dots,5\}$. (b) Decode the sequence $(2, 2, 4)$ on labels $\{1,\dots,5\}$ back into a tree.

**Solution**

**(a) Encoding.** Degrees: $\deg(5) = 3$, $\deg(4) = 2$, others 1. Repeat "record the neighbor of the smallest leaf, delete the leaf" ($n - 2 = 3$ times):

1. Leaves $\{1, 2, 3\}$; smallest is 1, neighbor 5 → record 5; delete 1.
2. Leaves $\{2, 3\}$; smallest is 2, neighbor 5 → record 5; delete 2.
3. Leaves $\{3, 5\}$; smallest is 3, neighbor 4 → record 4; delete 3. (Remaining edge $\{4,5\}$.)

$$
\text{Prüfer} = (5, 5, 4)
$$

Check: label 5 appears $\deg(5) - 1 = 2$ times, label 4 appears $1 = \deg(4) - 1$. ✓

**(b) Decoding $(2,2,4)$.** Degrees: $\deg(v) = (\text{count of } v) + 1$: $\deg(2) = 3$, $\deg(4) = 2$, $\deg(1) = \deg(3) = \deg(5) = 1$. Eligible leaves = labels absent from the remaining sequence:

1. Sequence $(2,2,4)$; smallest absent label: 1 → edge $\{1, 2\}$; drop first 2 → $(2,4)$; 1 used.
2. Sequence $(2,4)$; absent unused labels $\{3, 5\}$; smallest: 3 → edge $\{3, 2\}$; drop 2 → $(4)$; now 2 is absent and unused.
3. Sequence $(4)$; absent unused labels $\{2, 5\}$; smallest: 2 → edge $\{2, 4\}$; drop 4 → $()$.
4. Two unused labels remain: $\{4, 5\}$ → final edge $\{4, 5\}$.

$$
\boxed{\text{(a) } (5,5,4); \quad \text{(b) edges } \{1,2\}, \{2,3\}, \{2,4\}, \{4,5\}}
$$

*Key takeaway:* The bijection is fully mechanical in both directions; the degree identity $\deg(v) = (\text{occurrences of } v) + 1$ is the invariant that makes decoding forced and unique.

### Problem L3.4: Trees with Prescribed Degrees

Using the Prüfer correspondence, prove that the number of labeled trees on $\{1,\dots,n\}$ in which vertex $i$ has degree $d_i$ (with $\sum_i d_i = 2(n-1)$, $d_i \ge 1$) equals the multinomial coefficient $\binom{n-2}{d_1 - 1, \, d_2 - 1, \, \dots, \, d_n - 1}$. Evaluate for $n = 5$ with degrees $(3, 2, 1, 1, 1)$.

**Solution**

**Proof.** The Prüfer bijection maps labeled trees to sequences in $\{1,\dots,n\}^{n-2}$, and under it, vertex $v$ appears in the sequence exactly $\deg(v) - 1$ times (each recorded deletion of a neighbor leaf reduces $v$'s degree by 1 until $v$ becomes a leaf itself). Therefore trees with degree vector $(d_1, \dots, d_n)$ correspond bijectively to sequences of length $n-2$ containing symbol $i$ exactly $d_i - 1$ times. The number of arrangements of such a multiset is the multinomial coefficient

$$
\binom{n-2}{d_1 - 1, \dots, d_n - 1} = \frac{(n-2)!}{\prod_{i=1}^{n} (d_i - 1)!}
$$

(Consistency: summing over all valid degree vectors recovers $n^{n-2}$ by the multinomial theorem.) $\blacksquare$

**Evaluation** for $n = 5$, degrees $(3,2,1,1,1)$: multiplicities $(2,1,0,0,0)$ with $2 + 1 = 3 = n - 2$:

$$
\frac{3!}{2! \, 1! \, 0! \, 0! \, 0!} = \frac{6}{2} = 3
$$

The three trees attach the three leaves $\{3,4,5\}$ to the centers $1$ (twice) and $2$ (once) in $3$ distinguishable ways.

$$
\boxed{\#\text{trees} = \binom{n-2}{d_1-1, \dots, d_n-1}; \quad (3,2,1,1,1) \text{ on } n{=}5: \ 3 \text{ trees}}
$$

*Key takeaway:* Prüfer converts degree-constrained tree enumeration into multiset counting — a bijective proof strategy that also yields uniform random tree samplers.
