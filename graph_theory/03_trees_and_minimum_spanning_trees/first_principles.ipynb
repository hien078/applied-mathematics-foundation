# Topic 03: Trees and Minimum Spanning Trees

## 1. First-Principles Intuition & Motivation

Suppose you must connect $n$ villages with roads so that everyone can reach everyone. What is the *least* you can build?

Two forces shape the answer:

- **Connectivity** demands enough edges: below $n-1$ edges the graph must fall apart (Topic 02 proved $m \ge n-1$ for connected graphs).
- **Economy** forbids waste: if the network contains a cycle, some edge can be deleted while preserving connectivity — a cycle is redundancy.

The structures that satisfy both forces simultaneously — connected, and so lean that removing *any* edge disconnects them — are the **trees**. When edges have costs, the cheapest such structure is the **minimum spanning tree (MST)**, one of the oldest problems of combinatorial optimization (Borůvka, 1926, electrifying Moravia).

The deep surprise of this module: the MST can be found *greedily*, and provably so. Local decisions ("always take the cheapest safe edge") assemble a globally optimal structure — a rarity in optimization, explained by an exchange argument that this notebook develops in full.

### Trees are everywhere

- **Traversal trees**: BFS and DFS organize any connected graph around a spanning tree.
- **Hierarchies**: file systems, org charts, taxonomy, parse trees of programming languages.
- **Inference structures**: decision trees, phylogenetic trees, junction trees in probabilistic graphical models.
- **Recursion**: every divide-and-conquer computation unfolds as a tree.

A tree is the unique graph structure with *no ambiguity*: exactly one path between any two vertices, hence unique routes, unique ancestors, unique decompositions.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition (Forest, tree, leaf).** A **forest** is an acyclic graph; a **tree** is a connected acyclic graph. A **leaf** is a vertex of degree 1.

**Definition (Spanning tree).** A **spanning tree** of a connected graph $G = (V,E)$ is a subgraph $T = (V, E_T)$ that is a tree containing all vertices of $G$.

**Definition (MST).** Given connected $G$ with weights $w : E \to \mathbb{R}$, a **minimum spanning tree** is any $T^{\ast} \in \arg\min_T w(T)$ where $w(T) = \sum_{e \in T} w(e)$ over spanning trees $T$.

**Definition (Cut).** A **cut** $(S, V \setminus S)$ with $\emptyset \neq S \subsetneq V$ is a bipartition of the vertices; an edge **crosses** the cut if it has one endpoint in each side.

**Theorem (Tree characterization).** For a graph $G$ with $n$ vertices, the following are equivalent:

1. $G$ is a tree (connected and acyclic).
2. $G$ is connected and has exactly $n-1$ edges.
3. $G$ is acyclic and has exactly $n-1$ edges.
4. Every pair of vertices is joined by exactly one path.
5. $G$ is minimally connected: connected, but $G - e$ is disconnected for every edge $e$.
6. $G$ is maximally acyclic: acyclic, but $G + e$ contains a cycle for every non-edge $e$.

**Theorem (Leaf lemma).** Every tree with $n \ge 2$ vertices has at least two leaves.

**Theorem (Cut property).** If $e$ is the *unique* minimum-weight edge crossing some cut, then $e$ belongs to **every** MST.

**Theorem (Cycle property).** If $e$ is the *unique* maximum-weight edge of some cycle, then $e$ belongs to **no** MST.

**Theorem (Cayley's formula).** The number of labeled trees on $n$ vertices — equivalently spanning trees of $K_n$ — is $n^{n-2}$.

**Theorem (Kirchhoff's matrix-tree theorem).** For any connected graph, $\tau(G) = \det(L^{(i)})$ for any $i$, where $L^{(i)}$ deletes row and column $i$ of the Laplacian $L = D - A$; equivalently $\tau(G) = \frac{1}{n} \lambda_2 \lambda_3 \cdots \lambda_n$ over the nonzero Laplacian eigenvalues.

## 3. Step-by-Step Mathematical Proofs & Derivations

### Proof 1: The leaf lemma

**Theorem.** Every tree $T$ with $n \ge 2$ vertices has at least two leaves.

**Proof.** Let $P = v_0, v_1, \dots, v_k$ be a **longest path** in $T$ (finitely many paths exist, so a maximum length is attained; $k \ge 1$ since $T$ is connected with $n \ge 2$).

Consider the endpoint $v_0$. Every neighbor of $v_0$ other than $v_1$ is either on $P$ — which would close a cycle, impossible in a tree — or off $P$ — which would extend $P$ to a longer path, contradicting maximality. Hence $v_1$ is the only neighbor: $\deg(v_0) = 1$. The same argument applies to $v_k$. Two distinct leaves. $\blacksquare$

The leaf lemma powers induction on trees: remove a leaf, apply the hypothesis to the smaller tree, reattach.

### Proof 2: A tree on $n$ vertices has exactly $n-1$ edges

**Theorem.** If $T$ is a tree with $n$ vertices, then $m = n - 1$.

**Proof (induction on $n$).**

*Base*: $n = 1$ — the single vertex has $0 = 1 - 1$ edges. ✓

*Step*: let $T$ be a tree with $n \ge 2$ vertices. By the leaf lemma, pick a leaf $v$ with unique neighbor $u$. The graph $T - v$ is connected (any path between remaining vertices avoids the degree-1 vertex $v$ as an interior point) and acyclic (subgraphs of acyclic graphs are acyclic), hence a tree on $n-1$ vertices. By induction it has $n-2$ edges; restoring $v$ adds exactly the edge $\{u,v\}$, giving $m = n-1$. $\blacksquare$

$$
\boxed{\text{tree on } n \text{ vertices} \implies m = n - 1}
$$

**Converse directions** (completing the characterization): a connected graph with $n-1$ edges must be acyclic (deleting an edge of a cycle keeps it connected with $n-2$ edges, contradicting $m \ge n-1$ for connected graphs); an acyclic graph with $n-1$ edges must be connected (a forest with $c$ components has $n - c$ edges, so $c = 1$). Unique paths follow since two distinct $u$–$v$ paths contain a cycle, and conversely.

### Proof 3: The cut property

**Theorem.** Let $(S, V \setminus S)$ be a cut and let $e = \{u,v\}$ be the unique minimum-weight edge crossing it. Then every MST contains $e$.

**Proof (exchange argument).** Suppose some MST $T$ omits $e$. Adding $e$ to $T$ creates a unique cycle $C$ (maximal acyclicity), which contains $e$ and hence crosses the cut. A cycle crosses any cut an **even** number of times (following the cycle, sides alternate only at crossing edges, returning to the start), so $C$ contains a second crossing edge $f \neq e$.

Exchange: $T' = T - f + e$.

1. $T'$ is a spanning tree: $T + e$ has one cycle $C$; removing the cycle edge $f$ restores acyclicity, and connectivity survives because endpoints of $f$ remain joined via $C - f$. Edge count is again $n-1$.
2. Weight strictly drops: $f$ crosses the cut and $f \neq e$, so $w(f) \gt w(e)$ by unique minimality, giving $w(T') = w(T) - w(f) + w(e) \lt w(T)$.

This contradicts the minimality of $T$. Hence every MST contains $e$. $\blacksquare$

$$
\boxed{\text{unique lightest crossing edge } \in \text{ every MST}}
$$

### Proof 4: The cycle property

**Theorem.** Let $C$ be a cycle and $e$ its unique maximum-weight edge. Then no MST contains $e$.

**Proof.** Suppose an MST $T$ contains $e = \{u,v\}$. Deleting $e$ splits $T$ into two components, inducing a cut $(S, V \setminus S)$ with $u \in S$, $v \notin S$. The cycle $C$ connects $u$ to $v$ around the other side $C - e$, so some edge $f \in C - e$ crosses the cut. Then $T' = T - e + f$ is a spanning tree (reconnects the two components; no cycle since exactly one crossing edge is added to the two-component forest), and

$$
w(T') = w(T) - w(e) + w(f) \lt w(T)
$$

because $e$ is the strict maximum on $C$. Contradiction. $\blacksquare$

**Completeness.** With distinct edge weights, the cut and cycle properties classify *every* edge — each edge is either the lightest across some cut (in the MST) or the heaviest on some cycle (out of it) — and the MST is **unique**.

### Proof 5: Correctness of Kruskal's algorithm

**Algorithm (Kruskal).** Sort edges by weight; scan in increasing order, adding an edge iff it joins two different components of the forest built so far; stop after $n-1$ additions.

**Theorem.** With distinct weights, Kruskal's algorithm outputs the MST.

**Proof.** We show every accepted edge is MST-safe and every rejected edge is MST-forbidden.

*Accepted edges.* When edge $e = \{u,v\}$ is accepted, $u$ and $v$ lie in different components; let $S$ be the vertex set of $u$'s component. Every previously scanned crossing edge of the cut $(S, V \setminus S)$ was rejected — but rejected edges join vertices in the *same* component and thus never cross the current cut; hence no crossing edge lighter than $e$ exists, and unscanned crossing edges are heavier (sorted order). So $e$ is the unique lightest edge across $(S, V \setminus S)$: by the **cut property**, $e$ is in every MST.

*Rejected edges.* When $e$ is rejected, its endpoints are already connected by a path of strictly lighter (earlier) edges; that path plus $e$ forms a cycle on which $e$ is the unique heaviest edge: by the **cycle property**, $e$ is in no MST.

The algorithm therefore outputs exactly the set of MST edges — the MST itself. Termination with $n-1$ edges follows because the graph is connected. $\blacksquare$

**Prim's algorithm** (grow a single tree $S$ from a root, always adding the lightest edge leaving $S$) is correct by an immediate application of the cut property to the cut $(S, V \setminus S)$ at every step.

### Proof 6: Cayley's formula via the Prüfer bijection

**Theorem.** There are exactly $n^{n-2}$ labeled trees on $\{1, 2, \dots, n\}$.

**Proof.** We construct a bijection between labeled trees and sequences in $\{1, \dots, n\}^{n-2}$.

**Encoding (tree → sequence).** Repeat $n-2$ times: find the leaf with the *smallest label*, write down its unique neighbor, and delete the leaf. The recorded neighbors form the Prüfer sequence $(a_1, \dots, a_{n-2})$.

**Key invariant.** Vertex $v$ appears in the sequence exactly $\deg(v) - 1$ times: each deletion of a neighbor leaf records $v$ once and reduces its degree by one, until $v$ itself becomes a leaf and is never recorded again (it is either deleted later or survives as one of the final two vertices).

**Decoding (sequence → tree).** Maintain the list $L$ of labels *not* in the remaining sequence and not yet used. Repeat: take the smallest $\ell \in L$, join it to the first symbol $a$ of the remaining sequence, delete $\ell$ from consideration and $a$'s first occurrence from the sequence (adding $a$ to $L$ if it no longer appears). After $n-2$ steps join the last two unused labels. Each step is forced, and reversing the encoding step-by-step shows decode(encode$(T)$) $= T$ and encode(decode$(s)$) $= s$: a bijection.

Since there are $n^{n-2}$ sequences,

$$
\boxed{\tau(K_n) = n^{n-2}}
$$

$\blacksquare$

**Check** ($n = 3$): $3^1 = 3$ labeled trees — the three paths $1\text{–}2\text{–}3$, $2\text{–}1\text{–}3$, $1\text{–}3\text{–}2$. ✓

**Degree refinement.** Trees with prescribed degrees $d_1, \dots, d_n$ correspond to sequences containing each $i$ exactly $d_i - 1$ times, giving the multinomial count $\binom{n-2}{d_1 - 1, \dots, d_n - 1}$.

### The matrix-tree theorem (statement and sanity check)

**Theorem (Kirchhoff, 1847).** For a connected graph $G$ with Laplacian $L = D - A$ and Laplacian eigenvalues $0 = \lambda_1 \lt \lambda_2 \le \dots \le \lambda_n$:

$$
\tau(G) = \det\big(L^{(i)}\big) = \frac{1}{n} \prod_{k=2}^{n} \lambda_k
$$

where $L^{(i)}$ is $L$ with row and column $i$ deleted (any $i$).

**Sanity check on $K_3$**: $L = 3I - J$ restricted appropriately gives eigenvalues $0, 3, 3$, so $\tau = \frac{1}{3}(3 \cdot 3) = 3$ — matching Cayley's $3^{3-2} = 3$. ✓ In general the $K_n$ Laplacian has eigenvalue $n$ with multiplicity $n-1$, whence $\tau(K_n) = \frac{n^{n-1}}{n} = n^{n-2}$: Cayley's formula is a corollary.

The proof (via the Cauchy–Binet formula applied to $L = BB^{\top}$) is developed in Topic 06 alongside the full spectral theory of $L$.

## 4. Computational & Algorithmic Insights

### Algorithm comparison

| Algorithm | Strategy | Data structure | Time | Best regime |
|---|---|---|---|---|
| Kruskal (1956) | globally lightest safe edge | sorted edges + union–find | $O(m \log m)$ | sparse graphs, edges pre-sorted |
| Prim (1930/1957) | lightest edge leaving the tree | binary heap | $O(m \log n)$ | dense graphs, adjacency lists |
| Prim + Fibonacci heap | same | Fibonacci heap | $O(m + n \log n)$ | theory / very dense |
| Borůvka (1926) | lightest edge per component, in rounds | component labels | $O(m \log n)$ | parallel / distributed MST |

**Union–find**: with union by rank and path compression, $m$ operations cost $O(m \, \alpha(n))$, where $\alpha$ is the inverse Ackermann function — $\alpha(n) \le 4$ for every conceivable input, so Kruskal's bottleneck is purely the sort.

### Structural facts with algorithmic bite

- **Uniqueness**: distinct weights ⟹ unique MST (cut + cycle properties decide every edge). Ties can create exponentially many MSTs; all share the same *sorted weight sequence*.
- **Minimax path optimality**: the MST path between $u$ and $v$ minimizes the *maximum* edge weight over all $u$–$v$ paths (the "bottleneck shortest path") — the reason MSTs answer widest-path and percolation-threshold questions.
- **MST vs. shortest paths**: the MST need not contain any shortest path; conversely Dijkstra's shortest-path tree (Topic 04) can be arbitrarily heavier than the MST. Same greedy skeleton, different key function ($w(e)$ vs. $d(s,u) + w(e)$).
- **Verification faster than construction**: whether a given tree is the MST can be checked in $O(m)$ time (König/Komlós), while construction is $O(m \, \alpha(n))$ randomized (Karger–Klein–Tarjan).

### Verification strategy

- **Edge count check**: any output must contain exactly $n-1$ edges and span all $n$ vertices.
- **Cut sampling**: for random cuts, confirm the output contains a minimum crossing edge.
- **Cycle audit**: for each non-tree edge $e$, verify $w(e) \ge$ every weight on the tree path between its endpoints (this single condition is equivalent to MST optimality).
- **Cross-algorithm agreement**: with distinct weights, Kruskal, Prim, and Borůvka must return the identical edge set.
- **Perturbation stability**: adding $\epsilon$ noise to break ties should not change total weight by more than $(n-1)\epsilon$.

## 5. Real-World Physics & AI/ML Applications

### Physics and infrastructure

- **Network design**: electrification (Borůvka's original 1926 problem), fiber layout, pipeline networks — minimum-cost connection under budget is the literal MST problem.
- **Percolation and disordered media**: as bonds are added in increasing energy order (exactly Kruskal's process!), the emergence of a spanning cluster marks the percolation transition; the *minimax* MST path gives the critical threshold along optimal channels.
- **Polymer physics & invasion percolation**: fluid invading a porous medium follows the lightest available pore — Prim's algorithm run by nature.
- **Phylogenetics**: minimum-evolution trees over species-distance matrices approximate ancestral branching structure.

### AI and machine learning

- **Single-linkage hierarchical clustering ≡ MST**: cutting the MST's $k-1$ heaviest edges yields exactly the $k$ clusters of single-linkage clustering; the whole dendrogram is encoded by the MST (Gower & Ross, 1969).
- **Manifold learning**: neighborhood graphs pruned to spanning structures underlie Isomap-style embeddings; MST edges guarantee connectivity at minimal scale.
- **Chow–Liu trees**: the best tree-structured approximation to a joint distribution is the *maximum* spanning tree of the pairwise mutual-information graph — MST computation with flipped sign, foundational for tractable probabilistic graphical models.
- **Feature/correlation networks in finance**: MSTs of correlation distances (Mantegna, 1999) expose hierarchical market structure and drive portfolio diversification analysis.
- **Data-loading and communication trees**: broadcast/all-reduce topologies in distributed deep learning are spanning trees chosen to minimize communication cost.
