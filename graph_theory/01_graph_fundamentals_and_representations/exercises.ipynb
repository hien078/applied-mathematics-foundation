## Level 0 — Concept Check

### Problem L0.1: Order, Size, and Degrees

Let $G = (V, E)$ with $V = \{a, b, c, d\}$ and $E = \{\{a,b\}, \{a,c\}, \{a,d\}, \{b,c\}\}$.

State the order $n$, the size $m$, and the degree of every vertex. Verify the handshaking lemma numerically.

**Solution**

The order is the number of vertices and the size is the number of edges:

$$
n = \lvert V \rvert = 4, \qquad m = \lvert E \rvert = 4
$$

Counting incident edges per vertex: $\deg(a) = 3$ (edges to $b, c, d$), $\deg(b) = 2$ (edges to $a, c$), $\deg(c) = 2$ (edges to $a, b$), $\deg(d) = 1$ (edge to $a$).

Handshaking check:

$$
\sum_{v} \deg(v) = 3 + 2 + 2 + 1 = 8 = 2 \cdot 4 = 2m \checkmark
$$

$$
\boxed{n = 4, \quad m = 4, \quad \deg = (3, 2, 2, 1), \quad \textstyle\sum_v \deg(v) = 2m = 8}
$$

*Key takeaway:* Order counts vertices, size counts edges, and the degree sum is always exactly twice the edge count.

### Problem L0.2: An Impossible Degree Sum

A student claims to have drawn a simple graph on 5 vertices whose degrees are $3, 3, 3, 3, 3$. Is this possible?

**Solution**

By the handshaking lemma any graph satisfies $\sum_v \deg(v) = 2m$, an even number. Here

$$
\sum_v \deg(v) = 5 \cdot 3 = 15
$$

which is odd. No graph — simple or not — can have an odd degree sum. Equivalently, the number of odd-degree vertices (here five) must be even, and $5$ is not.

$$
\boxed{\text{Impossible: the degree sum } 15 \text{ is odd, violating } \textstyle\sum_v \deg(v) = 2m}
$$

*Key takeaway:* The handshaking lemma is a parity obstruction: any proposed degree sequence with odd sum can be rejected instantly.

### Problem L0.3: Reading the Adjacency Matrix

For a finite simple undirected graph, which two structural properties must its adjacency matrix $A$ satisfy? What does the entry $(A^2)_{ii}$ equal?

**Solution**

1. **Symmetry**: an undirected edge $\{i,j\}$ makes both $A_{ij} = 1$ and $A_{ji} = 1$, so $A = A^{\top}$.
2. **Zero diagonal**: simplicity forbids self-loops, so $A_{ii} = 0$ for all $i$.

For the diagonal of $A^2$:

$$
(A^2)_{ii} = \sum_{j=1}^{n} A_{ij} A_{ji} = \sum_{j=1}^{n} A_{ij}^2 = \sum_{j=1}^{n} A_{ij} = \deg(i)
$$

using symmetry and $A_{ij} \in \{0,1\}$. Each closed walk of length 2 from $i$ is an out-and-back over one incident edge.

$$
\boxed{A = A^{\top}, \quad \operatorname{diag}(A) = 0, \quad (A^2)_{ii} = \deg(i)}
$$

*Key takeaway:* The matrix encodes the graph faithfully: symmetry mirrors undirectedness, the zero diagonal mirrors simplicity, and matrix powers count walks.

### Problem L0.4: Choosing a Representation

You must run a full breadth-first traversal on a road network with $n = 10^7$ intersections and average degree $\approx 3$. Should you store it as an adjacency matrix or adjacency list? Justify with space and time bounds.

**Solution**

The graph is extremely sparse: $m \approx \tfrac{3n}{2} = 1.5 \times 10^7$ edges.

- **Adjacency matrix**: space $O(n^2) = 10^{14}$ entries — about 100 terabytes even at 1 byte per entry. A traversal touches every row: $O(n^2)$ time. Infeasible.
- **Adjacency list**: space $O(n + m) \approx 2.5 \times 10^7$ entries — a few hundred megabytes. BFS runs in $O(n + m)$ time, examining each edge at most twice.

$$
\boxed{\text{Adjacency list: } O(n+m) \text{ space and } O(n+m) \text{ traversal vs. } O(n^2) \text{ for the matrix}}
$$

*Key takeaway:* For sparse graphs the adjacency list is asymptotically mandatory; the matrix representation is reserved for dense graphs or algebraic (spectral) computations.

## Level 1 — Foundation

### Problem L1.1: Regular Graphs and Edge Counts

Prove that a $k$-regular graph on $n$ vertices has exactly $\frac{nk}{2}$ edges, and deduce that no 3-regular graph on 7 vertices exists.

**Solution**

In a $k$-regular graph every vertex has degree $k$, so the handshaking lemma gives

$$
2m = \sum_{v \in V} \deg(v) = nk \implies m = \frac{nk}{2}
$$

For $m$ to be an integer, $nk$ must be even. With $n = 7$ and $k = 3$ we get $nk = 21$, which is odd, so $m = 21/2$ is not an integer — no such graph exists.

$$
\boxed{m = \frac{nk}{2}; \text{ a } k\text{-regular graph on } n \text{ vertices exists only if } nk \text{ is even}}
$$

*Key takeaway:* Regularity plus handshaking forces the divisibility condition $2 \mid nk$ — a one-line feasibility test for regular graph constructions.

### Problem L1.2: Walks in the Triangle

Let $A$ be the adjacency matrix of the triangle $C_3 = K_3$. Compute $A^2$ and $A^3$, and interpret the entries $(A^3)_{11}$ and $(A^2)_{12}$ as walk counts.

**Solution**

With vertices $1, 2, 3$ pairwise adjacent:

$$
A = \begin{bmatrix} 0 & 1 & 1 \\ 1 & 0 & 1 \\ 1 & 1 & 0 \end{bmatrix}, \qquad
A^2 = \begin{bmatrix} 2 & 1 & 1 \\ 1 & 2 & 1 \\ 1 & 1 & 2 \end{bmatrix}, \qquad
A^3 = \begin{bmatrix} 2 & 3 & 3 \\ 3 & 2 & 3 \\ 3 & 3 & 2 \end{bmatrix}
$$

Interpretations:

- $(A^2)_{12} = 1$: the only walk of length 2 from vertex 1 to vertex 2 is $1 \to 3 \to 2$.
- $(A^3)_{11} = 2$: the closed walks of length 3 at vertex 1 are $1 \to 2 \to 3 \to 1$ and $1 \to 3 \to 2 \to 1$ — the two orientations of the triangle.

Consistency check: $\operatorname{tr}(A^3) = 6$, and the graph has exactly $T = 1$ triangle, matching $\operatorname{tr}(A^3) = 6T$.

$$
\boxed{(A^2)_{12} = 1, \qquad (A^3)_{11} = 2, \qquad \operatorname{tr}(A^3) = 6T = 6}
$$

*Key takeaway:* Small matrix powers already reveal structure: diagonal entries of $A^2$ give degrees, and the trace of $A^3$ counts triangles with multiplicity 6.

### Problem L1.3: The Complete Bipartite Graph

For $K_{p,q}$ with parts $X$ ($\lvert X \rvert = p$) and $Y$ ($\lvert Y \rvert = q$): find the number of edges, the degree of each vertex, and verify the handshaking lemma.

**Solution**

Every vertex of $X$ is joined to every vertex of $Y$ and there are no edges inside a part. Each of the $p$ vertices in $X$ contributes $q$ edges, and every edge arises exactly once this way:

$$
m = pq
$$

Degrees: $\deg(x) = q$ for $x \in X$ and $\deg(y) = p$ for $y \in Y$.

Handshaking verification:

$$
\sum_v \deg(v) = p \cdot q + q \cdot p = 2pq = 2m \checkmark
$$

$$
\boxed{m = pq, \quad \deg(x) = q \ (x \in X), \quad \deg(y) = p \ (y \in Y)}
$$

*Key takeaway:* Bipartite edge counting is a product, not a binomial coefficient — each edge is a choice of one vertex from each side.

### Problem L1.4: Degree 0 and Degree $n-1$ Cannot Coexist

Prove that a simple graph on $n \ge 2$ vertices cannot simultaneously contain a vertex of degree $0$ and a vertex of degree $n-1$. Deduce that some two vertices must share a degree.

**Solution**

Suppose $\deg(u) = 0$ and $\deg(w) = n - 1$ with $u \neq w$. Degree $n-1$ in a simple graph means $w$ is adjacent to every other vertex — in particular to $u$. But then $\deg(u) \ge 1$, a contradiction.

Consequently the $n$ vertex degrees all lie in $\{0, 1, \dots, n-1\}$, yet the two extreme values $0$ and $n-1$ cannot both occur, leaving at most $n - 1$ distinct available values. By the pigeonhole principle, two of the $n$ vertices have equal degree.

$$
\boxed{\text{In every simple graph with } n \ge 2 \text{ vertices, two vertices have the same degree}}
$$

*Key takeaway:* A pigeonhole argument on the shrunken value set $\{0,\dots,n-1\} \setminus \{0 \text{ or } n-1\}$ makes degree collisions unavoidable.

### Problem L1.5: Incidence Matrix of the Path $P_3$

Take the path $P_3$: vertices $1 - 2 - 3$, edges $e_1 = \{1,2\}$, $e_2 = \{2,3\}$, oriented $1 \to 2$ and $2 \to 3$. Write the signed incidence matrix $B$ and verify $BB^{\top} = D - A$ by direct computation.

**Solution**

With tail $= +1$ and head $= -1$:

$$
B = \begin{bmatrix} +1 & 0 \\ -1 & +1 \\ 0 & -1 \end{bmatrix}
$$

Compute the Gram matrix:

$$
BB^{\top} = \begin{bmatrix} 1 & -1 & 0 \\ -1 & 2 & -1 \\ 0 & -1 & 1 \end{bmatrix}
$$

Independently, the degrees are $(1, 2, 1)$ and the adjacency matrix has $A_{12} = A_{21} = A_{23} = A_{32} = 1$:

$$
D - A = \begin{bmatrix} 1 & 0 & 0 \\ 0 & 2 & 0 \\ 0 & 0 & 1 \end{bmatrix} - \begin{bmatrix} 0 & 1 & 0 \\ 1 & 0 & 1 \\ 0 & 1 & 0 \end{bmatrix} = \begin{bmatrix} 1 & -1 & 0 \\ -1 & 2 & -1 \\ 0 & -1 & 1 \end{bmatrix}
$$

The two computations agree.

$$
\boxed{BB^{\top} = D - A = L \text{ (the graph Laplacian of } P_3\text{)}}
$$

*Key takeaway:* The Laplacian factorization $L = BB^{\top}$ holds for any edge orientation — the signs cancel in the Gram product — and immediately proves $L \succeq 0$.

### Problem L1.6: Counting Labeled Graphs

How many distinct simple labeled graphs exist on $n$ vertices? Evaluate the count for $n = 5$ and explain why the growth is doubly exponential in appearance.

**Solution**

A simple labeled graph on vertex set $\{1, \dots, n\}$ is determined by deciding, independently for each of the $\binom{n}{2}$ unordered vertex pairs, whether it is an edge. Each decision is binary, so

$$
N(n) = 2^{\binom{n}{2}} = 2^{n(n-1)/2}
$$

For $n = 5$: $\binom{5}{2} = 10$, hence $N(5) = 2^{10} = 1024$.

The exponent itself grows quadratically in $n$, so $N(n) = 2^{\Theta(n^2)}$: already at $n = 40$ there are $2^{780} \approx 10^{234}$ labeled graphs — far exceeding the number of atoms in the observable universe ($\approx 10^{80}$).

$$
\boxed{N(n) = 2^{\binom{n}{2}}, \qquad N(5) = 1024}
$$

*Key takeaway:* The graph universe explodes combinatorially; brute-force enumeration is hopeless beyond tiny $n$, which is why structural theorems and invariants matter.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: Average Degree of a Social Network

A social network has $n = 2 \times 10^9$ users and $m = 3 \times 10^{11}$ friendship edges. Compute the average degree $\bar{d}$, and the memory required to store the network as (a) an adjacency list with 8-byte vertex identifiers and (b) a dense adjacency matrix at 1 bit per entry.

**Solution**

By the handshaking lemma,

$$
\bar{d} = \frac{1}{n} \sum_v \deg(v) = \frac{2m}{n} = \frac{2 \cdot 3 \times 10^{11}}{2 \times 10^9} = 300
$$

(a) **Adjacency list**: each undirected edge is stored in two endpoint lists, so $2m = 6 \times 10^{11}$ entries at 8 bytes each:

$$
6 \times 10^{11} \times 8 \text{ B} = 4.8 \times 10^{12} \text{ B} \approx 4.8 \text{ TB}
$$

(b) **Dense matrix**: $n^2 = 4 \times 10^{18}$ bits $= 5 \times 10^{17}$ bytes $\approx 500{,}000$ TB.

$$
\boxed{\bar{d} = \frac{2m}{n} = 300; \quad \text{list} \approx 4.8 \text{ TB} \ll \text{matrix} \approx 5 \times 10^{5} \text{ TB}}
$$

*Key takeaway:* $\bar{d} = 2m/n$ is the handshaking lemma in disguise, and the $10^5$-fold storage gap explains why every industrial graph system is adjacency-list (or CSR) based.

### Problem L2.2: Triangle Counting and the Clustering Coefficient

For the graph with edges $\{1,2\}, \{1,3\}, \{2,3\}, \{3,4\}$: count triangles via $\operatorname{tr}(A^3)$ and compute the local clustering coefficient of vertex 3, defined as $C_3 = \frac{2T_3}{\deg(3)(\deg(3)-1)}$ where $T_3$ is the number of triangles through vertex 3.

**Solution**

The graph is a triangle $\{1,2,3\}$ with a pendant edge to vertex 4. Closed walks of length 3 exist only around the triangle: each of its three vertices supports 2 such walks (both orientations), while vertex 4 supports none. Hence

$$
\operatorname{tr}(A^3) = 3 \cdot 2 = 6 \implies T = \frac{\operatorname{tr}(A^3)}{6} = 1
$$

Vertex 3 has $\deg(3) = 3$ (neighbors $1, 2, 4$), giving $\binom{3}{2} = 3$ neighbor pairs, of which exactly one pair $\{1,2\}$ is connected. With $T_3 = 1$:

$$
C_3 = \frac{2 T_3}{\deg(3)(\deg(3)-1)} = \frac{2 \cdot 1}{3 \cdot 2} = \frac{1}{3}
$$

$$
\boxed{T = 1, \qquad C_3 = \frac{1}{3}}
$$

*Key takeaway:* Triangle statistics — computable from $A^3$ — quantify "friend-of-my-friend" closure, a core feature separating social networks from random graphs.

### Problem L2.3: One Step of GNN Neighborhood Aggregation

On the path $1 - 2 - 3$ with node features $x = (x_1, x_2, x_3) = (1, 0, 2)$, compute one round of sum aggregation $h = Ax$ and mean aggregation $\bar{h} = D^{-1} A x$. Interpret the result as message passing.

**Solution**

The aggregation $(Ax)_v = \sum_{u \in N(v)} x_u$ sums each vertex's neighbor values:

$$
h_1 = x_2 = 0, \qquad h_2 = x_1 + x_3 = 3, \qquad h_3 = x_2 = 0
$$

Mean aggregation divides by degrees $(1, 2, 1)$:

$$
\bar{h}_1 = \frac{0}{1} = 0, \qquad \bar{h}_2 = \frac{3}{2}, \qquad \bar{h}_3 = \frac{0}{1} = 0
$$

Interpretation: each vertex receives "messages" (the raw features) from its neighbors and combines them with a permutation-invariant operator (sum or mean). A GCN layer is exactly $\bar{h}$ composed with a learned linear map and nonlinearity, $H' = \sigma(\hat{A} H W)$.

$$
\boxed{Ax = (0, 3, 0), \qquad D^{-1}Ax = \left(0, \tfrac{3}{2}, 0\right)}
$$

*Key takeaway:* The humble matrix-vector product $Ax$ *is* message passing — every GNN builds on this single sparse primitive.

### Problem L2.4: Ising Energy on the Cycle $C_4$

The Ising energy on a graph is $H(\sigma) = -J \sum_{\{u,v\} \in E} \sigma_u \sigma_v$ with spins $\sigma_v \in \{-1, +1\}$ and coupling $J \gt 0$. On the cycle $C_4$, compute the energy of (a) the aligned state $\sigma = (+1,+1,+1,+1)$ and (b) the alternating state $\sigma = (+1,-1,+1,-1)$. Which is the ground state?

**Solution**

$C_4$ has 4 edges: $\{1,2\}, \{2,3\}, \{3,4\}, \{4,1\}$.

(a) **Aligned**: every edge has $\sigma_u \sigma_v = (+1)(+1) = 1$:

$$
H = -J(1 + 1 + 1 + 1) = -4J
$$

(b) **Alternating**: every edge joins opposite spins, $\sigma_u \sigma_v = -1$:

$$
H = -J(-1 - 1 - 1 - 1) = +4J
$$

For ferromagnetic coupling $J \gt 0$, lower energy is preferred, so the aligned state is the ground state (energy $-4J$, achieved also by the all-$-1$ state). The alternating state is the ground state of the *antiferromagnet* ($J \lt 0$) — and it exists as a perfect alternation precisely because $C_4$ is bipartite (even cycle).

$$
\boxed{H_{\text{aligned}} = -4J, \qquad H_{\text{alternating}} = +4J}
$$

*Key takeaway:* Physical energies on networks are sums over the edge set; graph structure (here bipartiteness) determines which spin configurations are geometrically frustrated.

### Problem L2.5: Common-Neighbor Recommendation on a Bipartite Graph

Users $\{u_1, u_2\}$ and items $\{i_1, i_2, i_3\}$ form a bipartite interaction graph with biadjacency matrix $R = \begin{bmatrix} 1 & 1 & 0 \\ 0 & 1 & 1 \end{bmatrix}$ ($R_{ab} = 1$ iff user $a$ consumed item $b$). Compute the user-user co-consumption matrix $RR^{\top}$ and the item-item matrix $R^{\top}R$, and use them to recommend a new item to user $u_1$.

**Solution**

$$
RR^{\top} = \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix}, \qquad
R^{\top}R = \begin{bmatrix} 1 & 1 & 0 \\ 1 & 2 & 1 \\ 0 & 1 & 1 \end{bmatrix}
$$

$(RR^{\top})_{12} = 1$: users $u_1, u_2$ share one item ($i_2$), so they are similar. User $u_1$ has consumed $\{i_1, i_2\}$; the similar user $u_2$ has additionally consumed $i_3$. Equivalently, $(R^{\top}R)_{23} = 1$ says item $i_3$ co-occurs with $i_2$, which $u_1$ likes. Both walk-counting views (length-2 walks in the bipartite graph) point to the same recommendation.

$$
\boxed{\text{Recommend } i_3 \text{ to } u_1 \text{ (score from length-2 walks via } RR^{\top} \text{ and } R^{\top}R)}
$$

*Key takeaway:* Collaborative filtering is walk counting on a bipartite graph — the Gram matrices $RR^{\top}$ and $R^{\top}R$ are its user-based and item-based variants.

### Problem L2.6: The Friendship Paradox

Let $G$ have degrees $d_1, \dots, d_n$ with average $\bar{d} = \frac{2m}{n}$. Define the *average degree of a random friend* as the expected degree of a random endpoint of a uniformly random edge, $\mathbb{E}[D_{\text{friend}}] = \frac{\sum_v d_v^2}{\sum_v d_v}$. Prove that

$$
\mathbb{E}[D_{\text{friend}}] \ge \bar{d}
$$

with equality iff the graph is regular. ("Your friends have more friends than you do.")

**Solution**

Picking a uniform random edge and then a uniform endpoint selects vertex $v$ with probability proportional to $d_v$, so the friend's expected degree is $\frac{\sum_v d_v^2}{\sum_v d_v}$ as stated.

By the Cauchy–Schwarz inequality applied to the vectors $(d_1, \dots, d_n)$ and $(1, \dots, 1)$:

$$
\left( \sum_v d_v \right)^2 \le n \sum_v d_v^2
$$

Dividing both sides by $n \sum_v d_v$ (positive for any graph with an edge):

$$
\frac{\sum_v d_v^2}{\sum_v d_v} \ge \frac{\sum_v d_v}{n} = \bar{d}
$$

Equality in Cauchy–Schwarz holds iff $(d_1, \dots, d_n)$ is proportional to $(1, \dots, 1)$, i.e., all degrees equal — a regular graph. The gap equals $\operatorname{Var}(D)/\mathbb{E}[D]$, since $\mathbb{E}[D_{\text{friend}}] = \mathbb{E}[D] + \frac{\operatorname{Var}(D)}{\mathbb{E}[D]}$.

$$
\boxed{\mathbb{E}[D_{\text{friend}}] = \bar{d} + \frac{\operatorname{Var}(D)}{\bar{d}} \ge \bar{d}}
$$

*Key takeaway:* Sampling by edges biases toward high-degree vertices; this size-bias underlies the friendship paradox, epidemic sentinel surveillance, and degree-proportional negative sampling in graph ML.

## Level 3 — Challenge

### Problem L3.1: Trace of $A^3$ Counts Triangles

Prove rigorously that for any simple undirected graph, $\operatorname{tr}(A^3) = 6T$ where $T$ is the number of triangles.

**Solution**

By the walk-counting theorem, $(A^3)_{ii}$ is the number of closed walks of length 3 starting and ending at $i$: sequences $i \to j \to k \to i$ with all three steps edges.

**Step 1: closed 3-walks are exactly triangle traversals.** In such a walk, $j \neq i$ and $k \neq i$ (both adjacent to $i$, and the graph has no loops), and $j \neq k$ (they are adjacent). So $\{i, j, k\}$ are three distinct, pairwise adjacent vertices — a triangle containing $i$. Conversely every triangle $\{i,j,k\}$ yields closed 3-walks at $i$.

**Step 2: count multiplicity.** A fixed triangle $\{i,j,k\}$ produces exactly 2 closed walks based at $i$ ($i \to j \to k \to i$ and $i \to k \to j \to i$), hence

$$
(A^3)_{ii} = 2 \, T_i
$$

where $T_i$ is the number of triangles containing $i$.

**Step 3: sum over vertices.** Each triangle contains 3 vertices, so $\sum_i T_i = 3T$. Therefore

$$
\operatorname{tr}(A^3) = \sum_{i} (A^3)_{ii} = 2 \sum_i T_i = 6T
$$

$$
\boxed{\operatorname{tr}(A^3) = 6T \iff T = \frac{1}{6}\operatorname{tr}(A^3) = \frac{1}{6} \sum_{i} \lambda_i^3}
$$

where $\lambda_i$ are the adjacency eigenvalues (trace equals the sum of eigenvalue cubes).

*Key takeaway:* Careful bookkeeping of multiplicity (3 base vertices × 2 orientations) converts a walk count into a subgraph count — and the eigenvalue form connects combinatorics to the spectrum.

### Problem L3.2: Havel–Hakimi in Action

The sequence $(4, 4, 3, 2, 1)$ has even sum. Decide whether it is graphical (realizable by a simple graph) using the Havel–Hakimi reduction, and justify why the reduction is valid.

**Solution**

**Havel–Hakimi theorem.** A nonincreasing sequence $d_1 \ge d_2 \ge \dots \ge d_n$ is graphical iff the sequence obtained by deleting $d_1$ and subtracting 1 from the next $d_1$ terms is graphical (after re-sorting). Validity: if a realization exists, one can always rewire it (by edge exchanges) so the maximum-degree vertex is adjacent to the next $d_1$ highest-degree vertices; deleting it yields a realization of the reduced sequence, and conversely adding a new vertex joined to those terms lifts any realization back.

**Reduction of $(4,4,3,2,1)$:**

1. Delete $d_1 = 4$; subtract 1 from the next four terms: $(4,3,2,1) \mapsto (3,2,1,0)$. Sequence: $(3,2,1,0)$ — already sorted.
2. Delete $d_1 = 3$; subtract 1 from the next three terms: $(2,1,0) \mapsto (1,0,-1)$.

A degree of $-1$ is impossible, so $(1,0,-1)$ is not graphical, hence neither is $(3,2,1,0)$, hence neither is $(4,4,3,2,1)$.

Direct sanity check: on $n = 5$ vertices a degree-4 vertex is adjacent to *all* others, and there are two such vertices; then the vertex listed with degree 1 is adjacent to both of them, forcing degree $\ge 2$ — contradiction.

$$
\boxed{(4,4,3,2,1) \text{ is not graphical, although its sum } 14 \text{ is even}}
$$

*Key takeaway:* An even degree sum is necessary but far from sufficient; Havel–Hakimi gives a complete, constructive polynomial-time test.

### Problem L3.3: Self-Complementary Graphs

A graph $G$ is *self-complementary* if $G \cong \overline{G}$. Prove that a self-complementary graph on $n$ vertices satisfies $n \equiv 0$ or $n \equiv 1 \pmod 4$, and exhibit a self-complementary graph on 4 and on 5 vertices.

**Solution**

**Necessity.** $G$ and $\overline{G}$ partition the edge set of $K_n$: $m(G) + m(\overline{G}) = \binom{n}{2}$. Isomorphic graphs have equally many edges, so $m(G) = m(\overline{G})$, hence

$$
m(G) = \frac{1}{2}\binom{n}{2} = \frac{n(n-1)}{4}
$$

This must be an integer, so $4 \mid n(n-1)$. Since $n$ and $n-1$ are coprime and consecutive, exactly one of them is even; that even factor must itself be divisible by 4. Thus $4 \mid n$ or $4 \mid (n-1)$, i.e.,

$$
n \equiv 0 \pmod 4 \quad \text{or} \quad n \equiv 1 \pmod 4
$$

**Constructions.**

- $n = 4$: the path $P_4 = a - b - c - d$. Its complement has edges $\{a,c\}, \{a,d\}, \{b,d\}$, which is again a path $c - a - d - b$. So $P_4 \cong \overline{P_4}$.
- $n = 5$: the cycle $C_5$. Its complement is the pentagram — the cycle on the same vertices stepping by 2 — which is again a 5-cycle. So $C_5 \cong \overline{C_5}$.

$$
\boxed{n \equiv 0, 1 \pmod 4; \quad P_4 \text{ and } C_5 \text{ are self-complementary}}
$$

*Key takeaway:* Counting arguments on $\binom{n}{2}$ yield sharp arithmetic obstructions — a recurring pattern where divisibility gates the existence of symmetric structures.

### Problem L3.4: A Graph or Its Complement Is Connected

Prove that for every graph $G$, at least one of $G$ and its complement $\overline{G}$ is connected.

**Solution**

If $G$ is connected we are done. So assume $G$ is disconnected, with connected components $V_1, V_2, \dots, V_k$, $k \ge 2$. We show $\overline{G}$ is connected by exhibiting a path in $\overline{G}$ between any two vertices $u, v$.

**Case 1: $u, v$ lie in different components of $G$.** Then $\{u,v\} \notin E(G)$, so by definition of the complement $\{u,v\} \in E(\overline{G})$: they are adjacent in $\overline{G}$.

**Case 2: $u, v$ lie in the same component $V_i$.** Since $k \ge 2$, pick any vertex $w \in V_j$ with $j \neq i$. Both $u$ and $v$ lie in a different component from $w$, so by Case 1 both edges $\{u, w\}$ and $\{w, v\}$ exist in $\overline{G}$. The path $u - w - v$ connects them in $\overline{G}$.

In both cases $u$ and $v$ are joined by a path of length at most 2 in $\overline{G}$, so $\overline{G}$ is connected (indeed with diameter $\le 2$). $\blacksquare$

$$
\boxed{G \text{ disconnected} \implies \overline{G} \text{ connected with } \operatorname{diam}(\overline{G}) \le 2}
$$

*Key takeaway:* Disconnection is a strong structural constraint: the "missing" edges between components are so plentiful that the complement collapses to diameter at most 2.
