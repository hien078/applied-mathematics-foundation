# Topic 01: Graph Fundamentals and Representations

## 1. First-Principles Intuition & Motivation

Many systems consist of discrete objects and pairwise relationships between them:

- Cities connected by roads
- People connected by friendships
- Neurons connected by synapses
- Web pages connected by hyperlinks
- Atoms connected by chemical bonds

A **graph** captures this structure by abstracting away everything except the objects (**vertices**) and their pairwise connections (**edges**). The physical layout, the labels, the geometry of any drawing — all of it is discarded. What remains is pure *incidence structure*: which pairs are related.

This abstraction is powerful precisely because it is so aggressive. A theorem proved about abstract graphs applies simultaneously to road networks, molecules, social networks, and neural architectures.

### The Königsberg origin

Graph theory was born in 1736 when Euler asked whether one could walk through Königsberg crossing each of its seven bridges exactly once. His decisive insight was that only the *pattern of connections* matters: land masses become vertices, bridges become edges, and the geometric map becomes irrelevant.

The answer (no) followed from a parity argument on vertex degrees — the very first theorem of graph theory, and a direct corollary of the handshaking lemma proved below.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition (Graph).** A **graph** is an ordered pair $G = (V, E)$ where $V$ is a finite nonempty set of **vertices** and $E$ is a set of **edges**, each joining a pair of vertices.

- **Undirected graph**: each edge is an unordered pair $\{u, v\}$ with $u, v \in V$.
- **Directed graph (digraph)**: each edge is an ordered pair $(u, v) \in V \times V$, a connection *from* $u$ *to* $v$.

**Notation.** We write $n = \lvert V \rvert$ for the **order** and $m = \lvert E \rvert$ for the **size** of the graph.

**Definition (Simple graph).** A graph with no self-loops ($\{v,v\}$) and no repeated edges. Unless stated otherwise, "graph" means finite simple undirected graph.

### Taxonomy of graph types

| Type | Definition |
|---|---|
| **Simple graph** | No self-loops, no multiple edges between the same pair |
| **Multigraph** | Multiple parallel edges between the same pair allowed |
| **Directed graph** | Edges are ordered pairs: $(u,v) \neq (v,u)$ |
| **Weighted graph** | A triple $G=(V,E,w)$ with weight function $w : E \to \mathbb{R}$ |
| **Labeled / attributed graph** | Vertices and/or edges carry labels or feature vectors |

Weights model distances (km), capacities (liters/second), costs (dollars), or interaction strengths — the semantics come from the application, the mathematics is identical.

### Degree, neighborhoods, and subgraphs

**Definition (Neighborhood).** The **neighborhood** of $v$ is $N(v) = \{u \in V : \{v,u\} \in E\}$.

**Definition (Degree).** The **degree** of a vertex $v$ in an undirected graph is the number of edge-endpoint incidences at $v$; for a simple graph, $\deg(v) = \lvert N(v) \rvert$. A self-loop contributes $2$ to the degree.

For directed graphs, the **in-degree** $\deg^{-}(v)$ counts arcs entering $v$, and the **out-degree** $\deg^{+}(v)$ counts arcs leaving $v$.

**Definition (Subgraph).** $H = (V_H, E_H)$ is a **subgraph** of $G=(V,E)$ if $V_H \subseteq V$ and $E_H \subseteq E$. Given $S \subseteq V$, the **induced subgraph** $G[S]$ keeps exactly the edges of $G$ with both endpoints in $S$.

**Definition (Isomorphism).** Graphs $G_1, G_2$ are **isomorphic** if there is a bijection $\varphi : V_1 \to V_2$ with $\{u,v\} \in E_1 \iff \{\varphi(u), \varphi(v)\} \in E_2$. Isomorphism-invariant quantities (order, size, degree sequence, spectrum) are called **graph invariants**.

### The three canonical representations

**Definition (Adjacency matrix).** For vertices labeled $1, \dots, n$, the matrix $A \in \{0,1\}^{n \times n}$ with

$$
A_{ij} =
\begin{cases}
1 & \text{if } \{i,j\} \in E \\
0 & \text{otherwise}
\end{cases}
$$

For undirected graphs $A = A^{\top}$; for simple graphs $\operatorname{diag}(A) = 0$; for weighted graphs the entry $1$ is replaced by $w(i,j)$.

**Definition (Adjacency list).** For each $v \in V$ store the list $\mathrm{Adj}(v) = N(v)$. Total storage $O(n + m)$.

**Definition (Incidence matrix).** $B \in \mathbb{R}^{n \times m}$, rows indexed by vertices, columns by edges. Unsigned version: $B_{ve} = 1$ iff $v$ is an endpoint of $e$. For a chosen orientation, the **signed** incidence matrix uses $+1$ at the tail and $-1$ at the head of each edge.

### Canonical graph families

| Family | Symbol | Vertices | Edges | Degree pattern |
|---|---|---|---|---|
| Complete graph | $K_n$ | $n$ | $n(n-1)/2$ | every vertex has degree $n-1$ |
| Complete bipartite | $K_{p,q}$ | $p+q$ | $pq$ | degrees $q$ and $p$ on the two sides |
| Path | $P_n$ | $n$ | $n-1$ | two endpoints of degree 1 |
| Cycle | $C_n$ | $n$ | $n$ | 2-regular |
| Hypercube | $Q_d$ | $2^d$ | $d \cdot 2^{d-1}$ | $d$-regular |
| Empty graph | $\overline{K_n}$ | $n$ | $0$ | 0-regular |

A graph is **$k$-regular** if every vertex has degree exactly $k$.

## 3. Step-by-Step Mathematical Proofs & Derivations

### Proof 1: The Handshaking Lemma

**Theorem.** For any undirected graph $G = (V, E)$:

$$
\sum_{v \in V} \deg(v) = 2 \lvert E \rvert
$$

**Proof.** Count the set of incidences $I = \{(v, e) : v \in V,\ e \in E,\ v \text{ is an endpoint of } e\}$ in two ways.

1. **Grouping by vertex**: each vertex $v$ appears in exactly $\deg(v)$ incidences, so $\lvert I \rvert = \sum_{v \in V} \deg(v)$.
2. **Grouping by edge**: each edge $e = \{u, v\}$ has exactly two endpoints, contributing the two incidences $(u,e)$ and $(v,e)$, so $\lvert I \rvert = 2 \lvert E \rvert$. (A self-loop also contributes 2, consistent with the degree convention.)

Equating the two counts:

$$
\boxed{\sum_{v \in V} \deg(v) = 2 \lvert E \rvert}
$$

$\blacksquare$

### Proof 2: Parity corollary — the number of odd-degree vertices is even

**Corollary.** Every graph has an even number of vertices of odd degree.

**Proof.** Split $V$ into $V_{\mathrm{odd}} = \{v : \deg(v) \text{ odd}\}$ and $V_{\mathrm{even}} = \{v : \deg(v) \text{ even}\}$. By the handshaking lemma,

$$
\sum_{v \in V_{\mathrm{odd}}} \deg(v) = 2 \lvert E \rvert - \sum_{v \in V_{\mathrm{even}}} \deg(v)
$$

The right-hand side is a difference of even numbers, hence even. The left-hand side is a sum of odd numbers; a sum of $k$ odd numbers has the parity of $k$. For the sum to be even, $k = \lvert V_{\mathrm{odd}} \rvert$ must be even. $\blacksquare$

This parity argument is exactly what resolves the Königsberg bridge problem: the bridge multigraph has four vertices of odd degree, so no closed walk can use every bridge once (see Topic 02 for Euler's full theorem).

### Proof 3: The walk-counting theorem

**Definition (Walk).** A **walk** of length $k$ from $i$ to $j$ is a sequence $v_0 = i, v_1, \dots, v_k = j$ with $\{v_{t-1}, v_t\} \in E$ for every $t$. Vertices and edges may repeat.

**Theorem.** For any graph with adjacency matrix $A$ and any $k \ge 1$, the entry $(A^k)_{ij}$ equals the number of walks of length $k$ from $i$ to $j$.

**Proof (induction on $k$).**

*Base case* $k = 1$: $(A^1)_{ij} = A_{ij}$ is $1$ precisely when $\{i,j\} \in E$, i.e., when there is a single-edge walk. ✓

*Inductive step*: assume the claim for $k$. By the definition of matrix multiplication,

$$
(A^{k+1})_{ij} = \sum_{\ell=1}^{n} (A^k)_{i\ell} \, A_{\ell j}
$$

The term for a fixed $\ell$ counts (by the inductive hypothesis) the walks of length $k$ from $i$ to $\ell$, multiplied by $A_{\ell j} \in \{0,1\}$, which is $1$ exactly when the walk can be extended by the edge $\{\ell, j\}$. Every walk of length $k+1$ from $i$ to $j$ decomposes uniquely into its first $k$ steps (ending at some $\ell$) followed by a final edge $\{\ell, j\}$, so the sum counts each such walk exactly once. $\blacksquare$

**Consequences.** $(A^2)_{ii} = \deg(i)$; $\operatorname{tr}(A^2) = 2m$; and $\operatorname{tr}(A^3) = 6 \times (\text{number of triangles})$, since each triangle yields 6 closed walks of length 3 (3 starting vertices × 2 orientations).

### Proof 4: Edge count of the complete graph

**Theorem.** The complete graph $K_n$ has exactly $\binom{n}{2} = \frac{n(n-1)}{2}$ edges.

**Proof 1 (combinatorial).** An edge of $K_n$ is an unordered pair of distinct vertices, and every such pair is an edge. The number of unordered pairs from $n$ elements is $\binom{n}{2}$.

**Proof 2 (via handshaking).** Every vertex of $K_n$ is adjacent to the other $n-1$ vertices, so $\deg(v) = n-1$ for all $v$. By the handshaking lemma,

$$
2m = \sum_{v} \deg(v) = n(n-1) \implies m = \frac{n(n-1)}{2}
$$

$\blacksquare$

**Corollary.** Any simple graph on $n$ vertices satisfies $0 \le m \le \binom{n}{2}$, and the number of labeled simple graphs on $n$ vertices is $2^{\binom{n}{2}}$ — one independent binary choice per potential edge.

### Proof 5: Two vertices always share a degree

**Theorem.** Every simple graph with $n \ge 2$ vertices has two vertices of equal degree.

**Proof (pigeonhole).** Each degree lies in $\{0, 1, \dots, n-1\}$ — a set of $n$ possible values for $n$ vertices. If all degrees were distinct, every value would be used exactly once; in particular some vertex $u$ has degree $0$ and some vertex $w$ has degree $n-1$. But $\deg(w) = n - 1$ means $w$ is adjacent to *every* other vertex, including $u$, contradicting $\deg(u) = 0$. Hence the values $0$ and $n-1$ cannot coexist, leaving at most $n-1$ usable values for $n$ vertices; by the pigeonhole principle two vertices share a degree. $\blacksquare$

### Proof 6: The incidence matrix builds the Laplacian

**Theorem.** Let $B \in \mathbb{R}^{n \times m}$ be the signed incidence matrix of any orientation of a graph $G$, let $D = \operatorname{diag}(\deg v_1, \dots, \deg v_n)$, and let $A$ be the adjacency matrix. Then

$$
B B^{\top} = D - A =: L
$$

**Proof.** Compute the $(u,v)$ entry of $BB^{\top}$:

$$
(B B^{\top})_{uv} = \sum_{e \in E} B_{ue} B_{ve}
$$

*Diagonal case* $u = v$: the term $B_{ue}^2$ equals $1$ exactly when $u$ is an endpoint of $e$ (sign squared away), so the sum is $\deg(u) = D_{uu}$.

*Off-diagonal case* $u \neq v$: the product $B_{ue} B_{ve}$ is nonzero only when $e = \{u,v\}$, in which case one factor is $+1$ and the other $-1$ (whatever the orientation), giving $-1$. Hence the entry is $-A_{uv}$.

Combining, $BB^{\top} = D - A$. $\blacksquare$

This identity is the gateway to spectral graph theory: it instantly shows $L$ is symmetric positive semidefinite, since $x^{\top} L x = \lVert B^{\top} x \rVert^2 \ge 0$. Topic 06 develops this in full.

## 4. Computational & Algorithmic Insights

### Representation complexity

| Operation | Adjacency Matrix | Adjacency List |
|---|---|---|
| Check if edge $\{u,v\}$ exists | $O(1)$ | $O(\deg u)$ |
| Enumerate all neighbors of $v$ | $O(n)$ | $O(\deg v)$ |
| Space | $O(n^2)$ | $O(n + m)$ |
| Add edge | $O(1)$ | $O(1)$ |
| Full traversal (BFS/DFS) | $O(n^2)$ | $O(n + m)$ |
| Matrix algebra / spectral methods | natural | requires conversion |

**Rule of thumb**: use lists for sparse graphs ($m = O(n)$, typical of road networks and molecules) and matrices for dense graphs or whenever eigenvalues, powers, or matrix products are needed.

### Sparsity in practice

Real networks are almost always sparse. A social network with $n = 10^9$ users and average degree $200$ has $m = 10^{11}$ edges; the adjacency list needs roughly $10^{11}$ entries, while the dense matrix would need $10^{18}$ — a factor of $10^7$ waste. Sparse matrix formats (CSR/COO) combine the best of both worlds: $O(n+m)$ storage with fast matrix-vector products $Ax$ in $O(n+m)$ time.

The product $Ax$ has a beautiful graph meaning: $(Ax)_v = \sum_{u \in N(v)} x_u$ — each vertex *sums the values of its neighbors*. Iterating this "neighborhood aggregation" is the computational primitive underlying PageRank, label propagation, and every message-passing graph neural network.

### Counting structures with matrix powers

From the walk-counting theorem:

- **Degrees**: $(A^2)_{ii} = \deg(i)$, since a closed walk of length 2 is an out-and-back over one edge.
- **Edges**: $\operatorname{tr}(A^2) = \sum_i \deg(i) = 2m$.
- **Triangles**: $\operatorname{tr}(A^3) = 6 T$, where $T$ is the number of triangles.
- **Common neighbors**: $(A^2)_{ij}$ for $i \neq j$ counts common neighbors of $i$ and $j$ — the basis of similarity scores in link prediction.

Computing $A^2$ costs $O(n^{\omega})$ with fast matrix multiplication ($\omega \lt 2.372$), often beating combinatorial triangle enumeration on dense graphs.

## 5. Real-World Physics & AI/ML Applications

### Physics and engineering

- **Electrical circuits**: Kirchhoff's laws are graph-theoretic. Vertices are circuit nodes, edges are components; the signed incidence matrix $B$ expresses current conservation as $B f = 0$ (flow vector $f$), and $L = BB^{\top}$ appears directly in nodal analysis of resistor networks.
- **Statistical mechanics**: the Ising model lives on a graph; its energy $H(\sigma) = -\sum_{\{u,v\} \in E} J \sigma_u \sigma_v$ sums over edges, and phase-transition behavior depends on the graph's structure (lattice vs. random vs. complete).
- **Molecular chemistry**: molecules are labeled graphs (atoms = vertices, bonds = edges); graph invariants such as the Wiener index correlate with boiling points and chemical reactivity.

### AI and machine learning

- **Graph representation learning**: a graph neural network consumes exactly the data defined in this module — the adjacency structure $A$ plus node feature matrix $X \in \mathbb{R}^{n \times d}$. The GCN layer $H' = \sigma(\hat{A} H W)$ is neighborhood aggregation $Ax$ dressed with learnable weights (see Topic 07 and the legacy notebook [`../computation.ipynb`](../computation.ipynb)).
- **Knowledge graphs**: entities and typed relations form directed labeled multigraphs; link prediction models (TransE, ComplEx) embed vertices so that edge existence becomes a geometric test.
- **Recommender systems**: user–item interactions form a bipartite graph (Topic 05); collaborative filtering is walk counting — $(A^2)_{ij}$ over the bipartite adjacency counts shared interactions.
- **Molecule property prediction**: message-passing networks on molecular graphs (MPNN, SchNet) drive modern drug discovery and materials screening.

Degree distributions matter too: real networks are typically heavy-tailed ("scale-free"), which shapes sampling strategies (GraphSAGE neighbor sampling) and mini-batch design for large-scale GNN training.
