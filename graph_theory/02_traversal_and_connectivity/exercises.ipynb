## Level 0 — Concept Check

### Problem L0.1: Walk, Path, or Cycle?

In the graph with edges $\{1,2\}, \{2,3\}, \{3,1\}, \{3,4\}$, classify each sequence as a walk, a path, a cycle, or none: (a) $1,2,3,1,3,4$; (b) $4,3,2$; (c) $1,2,3,1$; (d) $1,2,4$.

**Solution**

(a) $1,2,3,1,3,4$: every consecutive pair is an edge, so it is a **walk** (length 5), but vertices 1 and 3 repeat, so it is neither a path nor a cycle.

(b) $4,3,2$: edges $\{4,3\}$ and $\{3,2\}$ exist and no vertex repeats — a **path** of length 2 (and also a walk).

(c) $1,2,3,1$: consecutive edges exist, first and last vertices coincide, interior vertices distinct, length $3 \ge 3$ — a **cycle**.

(d) $1,2,4$: the pair $\{2,4\}$ is not an edge — **none** (not even a walk).

$$
\boxed{\text{(a) walk only, (b) path, (c) cycle, (d) none}}
$$

*Key takeaway:* Walks allow repetition, paths forbid vertex repetition, cycles are closed paths — the hierarchy walk ⊇ path matters because shortest walks are automatically paths.

### Problem L0.2: BFS Visit Order

Run BFS from vertex $a$ on the graph with edges $\{a,b\}, \{a,c\}, \{b,d\}, \{c,d\}, \{d,e\}$, scanning neighbors in alphabetical order. Give the visit (dequeue) order and the final distance labels.

**Solution**

Initialization: $\mathrm{dist}[a] = 0$, queue $= (a)$.

1. Dequeue $a$: neighbors $b, c$ are new → $\mathrm{dist}[b] = \mathrm{dist}[c] = 1$; queue $=(b,c)$.
2. Dequeue $b$: neighbor $a$ seen, $d$ new → $\mathrm{dist}[d] = 2$; queue $=(c,d)$.
3. Dequeue $c$: neighbors $a, d$ already seen; queue $=(d)$.
4. Dequeue $d$: neighbors $b, c$ seen, $e$ new → $\mathrm{dist}[e] = 3$; queue $=(e)$.
5. Dequeue $e$: neighbor $d$ seen; queue empty.

$$
\boxed{\text{Visit order } a, b, c, d, e; \quad \mathrm{dist} = (a{:}0,\ b{:}1,\ c{:}1,\ d{:}2,\ e{:}3)}
$$

*Key takeaway:* BFS dequeues vertices in nondecreasing distance order — the queue is a faithful implementation of "expand the closest frontier first."

### Problem L0.3: Counting Components

The graph on $V = \{1, \dots, 7\}$ has edges $\{1,2\}, \{2,3\}, \{4,5\}, \{6,7\}, \{5,6\}$. How many connected components are there, and what are they?

**Solution**

Follow reachability: from 1 we reach $\{1,2,3\}$ (via $1 - 2 - 3$). From 4 we reach $\{4,5,6,7\}$ (via $4 - 5 - 6 - 7$ using $\{5,6\}$ and $\{6,7\}$). These two sets exhaust $V$ and no edge joins them.

$$
\boxed{2 \text{ components: } \{1,2,3\} \text{ and } \{4,5,6,7\}}
$$

*Key takeaway:* Components are the equivalence classes of reachability; one BFS/DFS per unvisited vertex enumerates them in $O(n+m)$ total time.

### Problem L0.4: Queue vs. Stack

BFS and DFS differ by exactly one implementation detail. What is it, and which structural guarantee does each choice buy?

**Solution**

The frontier container: BFS stores discovered-but-unprocessed vertices in a **FIFO queue**, DFS in a **LIFO stack** (or the call stack via recursion).

- FIFO processes vertices in the order discovered, hence in nondecreasing distance: BFS layers equal exact shortest distances in unweighted graphs.
- LIFO always extends the most recent discovery, producing deeply nested exploration intervals: the parenthesis structure that yields cycle detection, topological order, and SCC algorithms.

$$
\boxed{\text{FIFO queue} \Rightarrow \text{distance layers (metric)}; \quad \text{LIFO stack} \Rightarrow \text{nested intervals (order)}}
$$

*Key takeaway:* A single-word change in the data structure exchanges a metric certificate for an order certificate — same $O(n+m)$ cost.

## Level 1 — Foundation

### Problem L1.1: The Triangle Inequality for Graph Distance

Prove that in a connected graph, $d(u,w) \le d(u,v) + d(v,w)$ for all vertices $u, v, w$, and give an example where the inequality is strict.

**Solution**

Let $P_1$ be a shortest $u$–$v$ path (length $d(u,v)$) and $P_2$ a shortest $v$–$w$ path (length $d(v,w)$). Their concatenation is a $u$–$w$ **walk** of length $d(u,v) + d(v,w)$. Every walk contains a path between the same endpoints of no greater length (excise the portion between two visits of any repeated vertex and induct). Since $d(u,w)$ minimizes over all $u$–$w$ paths,

$$
d(u,w) \le d(u,v) + d(v,w)
$$

**Strictness**: in the triangle $K_3$ with vertices $u, v, w$: $d(u,w) = 1$ while $d(u,v) + d(v,w) = 2$.

$$
\boxed{d(u,w) \le d(u,v) + d(v,w); \text{ strict on } K_3}
$$

*Key takeaway:* Concatenate-then-shortcut is the standard proof pattern; the graph distance is a genuine metric, enabling all metric-space reasoning (balls, diameters, heuristics).

### Problem L1.2: BFS Layers of the Hypercube $Q_3$

Run BFS on the 3-dimensional hypercube $Q_3$ (vertices = binary strings of length 3, edges = pairs differing in one bit) from source $000$. How many vertices lie in each BFS layer $L_k$? Generalize to $Q_d$.

**Solution**

The distance between two binary strings in the hypercube is their Hamming distance (each edge flips one bit, and $k$ well-chosen flips realize any $k$-bit difference). Hence the layer $L_k$ from $000$ consists of strings with exactly $k$ ones:

$$
L_0 = \{000\}, \quad L_1 = \{100, 010, 001\}, \quad L_2 = \{110, 101, 011\}, \quad L_3 = \{111\}
$$

Sizes $1, 3, 3, 1$ — the binomial coefficients $\binom{3}{k}$. In general for $Q_d$:

$$
\lvert L_k \rvert = \binom{d}{k}, \qquad \sum_{k=0}^{d} \binom{d}{k} = 2^d \checkmark
$$

$$
\boxed{\lvert L_k(Q_d) \rvert = \binom{d}{k}; \text{ for } Q_3: 1, 3, 3, 1}
$$

*Key takeaway:* BFS layer structure often exposes hidden combinatorics — here graph distance equals Hamming distance and layers reproduce Pascal's triangle.

### Problem L1.3: Topological Orders of a Small DAG

For the DAG with edges $a \to b$, $a \to c$, $b \to d$, $c \to d$: list all valid topological orders and verify the count.

**Solution**

Constraints: $a$ before $b, c$; both $b$ and $c$ before $d$. Vertex $a$ must be first ($a$ is the only source) and $d$ last (the only sink). The middle positions hold $b$ and $c$ in either order:

$$
a, b, c, d \qquad \text{and} \qquad a, c, b, d
$$

Count check: the partial order leaves exactly one free choice (the relative order of the incomparable pair $b, c$), giving $2! = 2$ linear extensions.

$$
\boxed{\text{2 topological orders: } (a,b,c,d) \text{ and } (a,c,b,d)}
$$

*Key takeaway:* Topological order is rarely unique — the number of linear extensions measures how much parallelism the dependency structure permits (here $b$ and $c$ can run concurrently).

### Problem L1.4: Connectivity via Matrix Powers

Prove: a graph $G$ on $n$ vertices with adjacency matrix $A$ is connected $\iff$ every entry of $M = (I + A)^{n-1}$ is strictly positive.

**Solution**

Expand by the binomial theorem (valid since $I$ and $A$ commute):

$$
(I + A)^{n-1} = \sum_{k=0}^{n-1} \binom{n-1}{k} A^k
$$

All terms are entrywise nonnegative, so $M_{ij} \gt 0$ iff $(A^k)_{ij} \gt 0$ for **some** $0 \le k \le n-1$, i.e., iff there is a walk of some length $\le n-1$ from $i$ to $j$.

($\Leftarrow$) If all entries are positive, every pair is joined by a walk, hence by a path: $G$ is connected.

($\Rightarrow$) If $G$ is connected, any two vertices are joined by a path; a path has at most $n - 1$ edges (no repeated vertices among $n$), so a walk of length $\le n-1$ exists and the corresponding entry is positive. (Diagonal entries are positive via $k = 0$.)

$$
\boxed{G \text{ connected} \iff (I+A)^{n-1} \gt 0 \text{ entrywise}}
$$

*Key takeaway:* The $n-1$ bound comes from the pigeonhole fact that simple paths cannot exceed $n-1$ edges; adding $I$ absorbs the "walk of any shorter length" quantifier neatly.

### Problem L1.5: Eulerian Trails and Königsberg

State the condition for a connected multigraph to admit an open Eulerian trail, prove it from Euler's circuit theorem, and apply it to the Königsberg multigraph whose degrees are $(5, 3, 3, 3)$.

**Solution**

**Claim.** A connected multigraph has an open Eulerian trail iff exactly two vertices have odd degree.

**Proof.** ($\Leftarrow$) Let $u, w$ be the odd vertices. Add a phantom edge $\{u, w\}$: now all degrees are even, so Euler's theorem yields an Eulerian circuit. Delete the phantom edge from the circuit; what remains is an open trail from $u$ to $w$ using every original edge once.

($\Rightarrow$) Given a trail from $u$ to $w$ ($u \neq w$), add the edge $\{u,w\}$ to close it into a circuit; by Euler's theorem all degrees of the augmented graph are even, so in the original graph exactly $u$ and $w$ are odd. $\blacksquare$

**Königsberg**: degrees $(5,3,3,3)$ contain **four** odd vertices. Four $\neq 0$ (no circuit) and four $\neq 2$ (no open trail).

$$
\boxed{\text{Trail} \iff \text{exactly 2 odd vertices; Königsberg has 4, hence no Eulerian walk of any kind}}
$$

*Key takeaway:* The phantom-edge trick converts the trail problem into the circuit problem — a model example of reducing a new problem to a solved one.

### Problem L1.6: Connected Graphs Need $n-1$ Edges

Prove that every connected graph on $n$ vertices has at least $n - 1$ edges.

**Solution**

**Proof (component counting).** Start from the empty graph on $V$: it has $n$ components. Add the edges of $G$ one at a time. Each added edge joins two vertices; if they lie in different components it merges them (components decrease by exactly 1), otherwise the count is unchanged. After all $m$ edges, the number of components is therefore at least $n - m$. Connectivity means one component:

$$
1 \ge n - m \implies m \ge n - 1
$$

**Sharpness**: trees achieve equality (Topic 03 shows connected + exactly $n-1$ edges $\iff$ tree).

$$
\boxed{G \text{ connected} \implies m \ge n - 1, \text{ with equality iff } G \text{ is a tree}}
$$

*Key takeaway:* "Each edge kills at most one component" is a potential-function argument — the same bookkeeping that powers Kruskal's algorithm and union–find analysis.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: Contact Tracing Depth

An infectious disease spreads along a contact graph where each person has on average $b = 4$ distinct new contacts per generation. Starting from one infected person, estimate the number of people within BFS depth $k = 3$ (generations), and explain why health authorities trace only 1–2 layers.

**Solution**

Modeling the traced contact structure as a tree with branching factor $b$ (each new case exposes $b$ fresh contacts), layer $k$ contains $b^k$ people and the ball of radius 3 contains

$$
\lvert B_3 \rvert = \sum_{k=0}^{3} 4^k = 1 + 4 + 16 + 64 = 85
$$

The workload grows geometrically: depth 5 would already require tracing $\sum_{k=0}^5 4^k = 1365$ contacts per index case. Since tracing capacity is linear in staff but the ball grows exponentially in depth, authorities cut off at $k \in \{1, 2\}$ (direct contacts and contacts-of-contacts), where most transmission risk concentrates.

$$
\boxed{\lvert B_3 \rvert = 85 \text{ people}; \text{ exponential growth } b^k \text{ forces shallow tracing}}
$$

*Key takeaway:* Epidemic reach is BFS ball growth; the branching factor $b$ plays the role of $R_0$, and exponential layer growth is why early, shallow interventions dominate.

### Problem L2.2: GNN Receptive Fields and Over-Squashing

A message-passing GNN runs $k$ layers on a binary tree of depth $k$ rooted at $r$. (a) Show the receptive field of $r$ is the entire tree, of size $2^{k+1} - 1$. (b) If node embeddings have fixed dimension $h$, argue why information from the leaves is compressed by a factor exponential in $k$ (over-squashing).

**Solution**

(a) One message-passing layer extends each node's dependence by exactly one hop, so after $k$ layers the root depends on its BFS ball $B_k(r)$. In a binary tree of depth $k$, that ball is the whole tree:

$$
\lvert B_k(r) \rvert = \sum_{j=0}^{k} 2^j = 2^{k+1} - 1
$$

(b) All information from the $2^k$ leaves must pass through the root's final embedding, a vector of fixed dimension $h$ with (in practice) bounded precision. The available capacity is $O(h)$ real numbers, while the input grows as $\Theta(2^k)$ leaf features. The compression ratio

$$
\frac{2^k \cdot h}{h} = 2^k
$$

grows exponentially in depth, so distant information is squashed — gradients through long tree paths vanish and long-range dependencies fail. This motivates graph rewiring, global attention, and expander-based propagation in modern GNN design.

$$
\boxed{\lvert B_k(r) \rvert = 2^{k+1} - 1; \text{ bottleneck compresses leaf information by } \sim 2^k}
$$

*Key takeaway:* Receptive fields are BFS balls; whenever ball volume grows exponentially but embedding width stays fixed, message passing hits an information-theoretic wall.

### Problem L2.3: Autograd Ordering

A computation graph has operations with dependencies $x \to u$, $x \to v$, $u \to w$, $v \to w$, $w \to \ell$ (loss). Give a valid forward execution order and the corresponding reverse (backpropagation) order, and state the property of the graph that guarantees these orders exist.

**Solution**

The dependency digraph is a DAG (it has an obvious topological order and no cycles). A valid forward order is any topological sort:

$$
x, \ u, \ v, \ w, \ \ell \qquad (\text{or } x, v, u, w, \ell)
$$

Backpropagation processes nodes in **reverse topological order**, so that when a node's adjoint is needed, all of its consumers have already contributed their gradient terms:

$$
\ell, \ w, \ v, \ u, \ x
$$

with $\bar{w} = \bar{\ell} \frac{\partial \ell}{\partial w}$, then $\bar{u} = \bar{w} \frac{\partial w}{\partial u}$, $\bar{v} = \bar{w} \frac{\partial w}{\partial v}$, and finally $\bar{x} = \bar{u} \frac{\partial u}{\partial x} + \bar{v} \frac{\partial v}{\partial x}$ (sum over both paths).

Existence of these orders is guaranteed exactly by **acyclicity**: a digraph admits a topological order iff it is a DAG.

$$
\boxed{\text{Forward: } x, u, v, w, \ell; \quad \text{Backward: } \ell, w, v, u, x \text{ — valid because the graph is a DAG}}
$$

*Key takeaway:* Backpropagation is reverse topological traversal; the multi-path sum at $x$ is the chain rule aggregating over all directed paths.

### Problem L2.4: Six Degrees of Separation

Assume a social graph with $n = 8 \times 10^9$ people where BFS balls grow by an effective factor $\bar{d} = 100$ per layer. Estimate the depth $k$ at which a BFS ball covers the population, and relate the answer to the "six degrees" folklore.

**Solution**

Under the growth model $\lvert B_k \rvert \approx \bar{d}^{\,k}$, coverage requires

$$
\bar{d}^{\,k} \ge n \iff k \ge \frac{\ln n}{\ln \bar{d}} = \frac{\ln(8 \times 10^9)}{\ln 100} = \frac{22.8}{4.61} \approx 4.95
$$

So $k \approx 5$ layers suffice: the expected distance between two random people is about 5 hops, matching the empirical "six degrees of separation" (Milgram's letters experiment; Facebook measured $\approx 4.7$ in 2016).

The model's message: diameter scales as $O(\log n / \log \bar{d})$ — *logarithmic* in population — because BFS layer growth is geometric until saturation.

$$
\boxed{k \approx \frac{\ln n}{\ln \bar{d}} \approx 5 \text{ hops for } n = 8 \times 10^9, \ \bar{d} = 100}
$$

*Key takeaway:* Small-world diameters are a direct consequence of exponential BFS ball growth; the log-ratio formula is the quantitative core of the phenomenon.

### Problem L2.5: Kahn's Algorithm on a Build Graph

Packages with dependencies: $\text{core} \to \text{utils}$, $\text{core} \to \text{net}$, $\text{utils} \to \text{app}$, $\text{net} \to \text{app}$, $\text{utils} \to \text{cli}$ (an edge $x \to y$ means $y$ depends on $x$, so $x$ builds first). Execute Kahn's algorithm (alphabetical tie-breaking) showing the in-degree table and queue at each step.

**Solution**

Initial in-degrees: core 0, net 1, utils 1, app 2, cli 1. Queue $=$ (core).

| Step | Dequeued | In-degree updates | Queue after |
|---|---|---|---|
| 1 | core | utils $\to 0$, net $\to 0$ | (net, utils) |
| 2 | net | app $\to 1$ | (utils) |
| 3 | utils | app $\to 0$, cli $\to 0$ | (app, cli) |
| 4 | app | — | (cli) |
| 5 | cli | — | ( ) |

All 5 vertices emitted → the graph is a DAG and the build order is valid:

$$
\boxed{\text{core}, \ \text{net}, \ \text{utils}, \ \text{app}, \ \text{cli}}
$$

Had the queue emptied with vertices remaining, the leftover subgraph would have contained a cycle (every vertex with positive in-degree) — a circular dependency error.

*Key takeaway:* Kahn's algorithm is the constructive proof that DAGs have topological orders, and its failure mode doubles as a cycle detector for dependency managers.

### Problem L2.6: Grid Islanding After Line Failures

A power grid is the $3 \times 3$ grid graph (vertices $(i,j)$, $1 \le i, j \le 3$, edges between orthogonal neighbors). Storm damage removes the four edges $\{(1,2),(2,2)\}, \{(2,1),(2,2)\}, \{(2,3),(2,2)\}, \{(3,2),(2,2)\}$. Determine the resulting connected components and the physical consequence.

**Solution**

The removed edges are precisely the four edges incident to the center vertex $(2,2)$, which becomes isolated. Every other vertex lies on the boundary ring of the grid, which is the cycle

$$
(1,1) - (1,2) - (1,3) - (2,3) - (3,3) - (3,2) - (3,1) - (2,1) - (1,1)
$$

and remains connected (its 8 edges are intact).

Components: $\{(2,2)\}$ and the 8 boundary vertices — an **islanding** event. Physically, the substation at $(2,2)$ is cut off: it can neither draw nor supply power, and if it hosted generation, that supply is lost to the ring, potentially triggering load shedding.

$$
\boxed{2 \text{ components: the isolated center } (2,2) \text{ and the 8-vertex boundary ring}}
$$

*Key takeaway:* Post-failure component analysis (one BFS/DFS pass) is the first computation in any cascading-failure or resilience study; cut vertices and small cuts flag structural vulnerability.

## Level 3 — Challenge

### Problem L3.1: Full Proof of Euler's Theorem

Prove: a connected multigraph $G$ has an Eulerian circuit if and only if every vertex has even degree.

**Solution**

($\Rightarrow$) Fix an Eulerian circuit $W$. For any vertex $v$, group the edge-slots of $W$ at $v$ into (arrival, departure) pairs: each intermediate visit uses one edge in and one edge out; for the start vertex, the initial departure pairs with the terminal arrival. Since $W$ uses every edge exactly once, $\deg(v)$ equals twice the number of visits, hence even.

($\Leftarrow$) Induction on the number of edges $m$; the base $m = 0$ is trivial (single vertex).

**Step 1 (a closed subwalk exists).** Start at any vertex $s$ and greedily walk along unused edges. Upon arriving at any $v \neq s$, an odd number of $v$'s edge-slots are used (arrivals exceed departures by one), and $\deg(v)$ is even, so an unused edge remains to leave by. The walk can only terminate at $s$, producing a closed walk $C$ with distinct edges.

**Step 2 (recurse on the remainder).** Delete $E(C)$. Each vertex loses an even number of incident edges (paired in/out along $C$), so all degrees remain even. Each nonempty component $H_1, \dots, H_t$ of the remainder satisfies the induction hypothesis (connected, even degrees, fewer edges), yielding Eulerian circuits $C_1, \dots, C_t$.

**Step 3 (splice).** Because $G$ is connected, each $H_i$ intersects $C$ in at least one vertex $x_i$ (otherwise no edge or vertex of $H_i$ ever meets $C$, contradicting connectivity of $G$). Walk around $C$; at the first arrival at each $x_i$, insert the full circuit $C_i$ before continuing. The spliced walk is closed, and uses each edge of $C$ once and each edge of each $H_i$ once — every edge of $G$ exactly once. $\blacksquare$

$$
\boxed{\text{Eulerian circuit} \iff G \text{ connected and } \deg(v) \equiv 0 \pmod 2 \ \forall v}
$$

*Key takeaway:* The proof is an algorithm (Hierholzer's): greedy closed walks plus splicing run in $O(n+m)$ — a rare case where the classical existence proof is also the optimal algorithm.

### Problem L3.2: Back Edges Characterize Cycles

Using the white-path theorem, prove that a directed graph is acyclic if and only if a DFS produces no back edge.

**Solution**

**Back edge ⟹ cycle.** A back edge $(u, v)$ is scanned while $v$ is gray, meaning $v$ is an ancestor of $u$ in the DFS forest (gray vertices form a root-to-current chain). The tree edges give a directed path $v \rightsquigarrow u$; appending $(u,v)$ closes a directed cycle.

**Cycle ⟹ back edge.** Let $C = c_0 \to c_1 \to \dots \to c_{k-1} \to c_0$ be a directed cycle, and let $v = c_j$ be the vertex of $C$ with minimum discovery time $d(v)$. At the instant $v$ is discovered, all other vertices of $C$ are white (their discovery times are larger). The cycle provides a directed path from $v$ to its predecessor $u = c_{j-1}$ consisting solely of white vertices. By the **white-path theorem**, $u$ becomes a descendant of $v$ in the DFS forest, so by the parenthesis theorem $d(v) \lt d(u) \lt f(u) \lt f(v)$: when DFS scans the edge $(u, v)$ (which happens before $f(u)$), the interval of $v$ is still open — $v$ is gray. An edge to a gray vertex is classified as a back edge. $\blacksquare$

$$
\boxed{\text{digraph acyclic} \iff \text{DFS yields no back edge}}
$$

*Key takeaway:* "Look at the first-discovered vertex of the structure" is the canonical way to invoke the white-path theorem; it converts a global object (a cycle) into a local classification event (one gray edge).

### Problem L3.3: The Condensation Is a DAG

Let $G$ be a digraph and $G^{\mathrm{SCC}}$ its condensation: one vertex per strongly connected component, with an edge $C_1 \to C_2$ ($C_1 \neq C_2$) whenever some edge of $G$ goes from a vertex of $C_1$ to a vertex of $C_2$. Prove $G^{\mathrm{SCC}}$ is acyclic.

**Solution**

Suppose for contradiction the condensation contains a directed cycle

$$
C_1 \to C_2 \to \dots \to C_k \to C_1, \qquad k \ge 2
$$

Each condensation edge $C_i \to C_{i+1}$ lifts to an edge $x_i \to y_{i+1}$ in $G$ with $x_i \in C_i$, $y_{i+1} \in C_{i+1}$.

Take any $u \in C_i$ and $v \in C_j$ (components on the cycle). Inside each component, strong connectivity provides directed paths between any two of its vertices — in particular from each $y_i$ to the corresponding exit vertex $x_i$. Concatenating

$$
u \rightsquigarrow x_i \to y_{i+1} \rightsquigarrow x_{i+1} \to \dots \rightsquigarrow v
$$

and continuing around the cycle in the other direction from $v$ back to $u$, we obtain directed paths both from $u$ to $v$ and from $v$ to $u$. Hence all vertices of $C_1 \cup \dots \cup C_k$ are mutually reachable, so this union is a strongly connected subgraph strictly containing $C_1$ — contradicting maximality of $C_1$ as an SCC. Therefore no cycle exists. $\blacksquare$

$$
\boxed{G^{\mathrm{SCC}} \text{ is always a DAG}}
$$

*Key takeaway:* Quotienting by strong connectivity always produces a DAG — every digraph is "a DAG of tangles," which is why SCC condensation + topological sort is the universal preprocessing step for reachability reasoning.

### Problem L3.4: A Removable Vertex Always Exists

Prove that every connected graph $G$ with $n \ge 2$ vertices contains a vertex $v$ such that $G - v$ is still connected.

**Solution**

**Proof via spanning tree.** Since $G$ is connected, it contains a spanning tree $T$ (grow a tree by BFS/DFS from any root; it reaches every vertex). $T$ is a tree with $n \ge 2$ vertices, so it has a leaf $v$ (a vertex of degree 1 in $T$): the endpoint of a longest path in $T$ must be a leaf, since a neighbor other than the path's second vertex would either extend the path or close a cycle.

Consider $G - v$. The subgraph $T - v$ is a tree on the remaining $n - 1$ vertices (removing a leaf keeps a tree: any two remaining vertices are joined by their unique $T$-path, which never passed through the degree-1 vertex $v$ as an interior vertex). Hence $T - v$ is a connected spanning subgraph of $G - v$, so $G - v$ is connected. $\blacksquare$

**Remark.** The same argument gives *two* removable vertices (a tree with $n \ge 2$ has at least two leaves) — sharp on the path graph $P_n$, whose interior vertices are all cut vertices.

$$
\boxed{\text{Any leaf of a spanning tree of } G \text{ can be removed without disconnecting } G}
$$

*Key takeaway:* Spanning trees transfer tree facts to all connected graphs; "take a leaf of a spanning tree" is a standard tool for induction on connected graphs.
