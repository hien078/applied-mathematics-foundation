## Level 0 — Concept Check

### Problem L0.1: Why $\nabla f$ Does Not Vanish at a Constrained Optimum

**Problem Statement:** Minimize $f(x,y) = x + y$ on the circle $h(x,y) = x^2 + y^2 - 1 = 0$. Show that
$\nabla f$ is nonzero at the solution, and state precisely which projection of $\nabla f$ *does* vanish.

*Intuition:* On a curved feasible set you cannot move in every direction, so only the component of the gradient you are *allowed* to exploit must be zero.

**Solution:**

**Step 1 (the gradient is constant and nonzero).** $\nabla f(x,y) = (1,1)$ everywhere; it never vanishes, so
the unconstrained condition $\nabla f = \mathbf{0}$ has no solution at all. Yet the circle is compact and
$f$ is continuous, so a constrained minimizer certainly exists.

**Step 2 (parametrize and minimize).** On the circle $(x,y) = (\cos\theta, \sin\theta)$,

$$
f = \cos\theta + \sin\theta = \sqrt{2}\,\sin\left(\theta + \tfrac{\pi}{4}\right)
$$

minimized at $\theta = -3\pi/4$, i.e.

$$
(x^*, y^*) = \left(-\tfrac{1}{\sqrt2}, -\tfrac{1}{\sqrt2}\right), \qquad f^* = -\sqrt{2}
$$

**Step 3 (what actually vanishes).** The tangent space at $(x^*,y^*)$ is

$$
T = \left\{\mathbf{d} : \nabla h(x^*,y^*)^T\mathbf{d} = 0\right\}, \qquad \nabla h = 2(x^*, y^*) = -\sqrt{2}\,(1,1)
$$

so $T = \operatorname{span}\{(1,-1)\}$. The directional derivative along the only feasible direction is

$$
\nabla f^T(1,-1) = 1 - 1 = 0
$$

Thus $\nabla f \perp T$: the **projection of $\nabla f$ onto the tangent space vanishes**, while its normal
component (all of it, here) survives. Equivalently $\nabla f = \lambda^{\dagger}\nabla h$ with
$\lambda^{\dagger} = -1/\sqrt{2}$, i.e. $\nabla f + \lambda\nabla h = \mathbf{0}$ with
$\lambda = 1/\sqrt{2}$.

$$
\boxed{P_T\nabla f(\mathbf{x}^*) = \mathbf{0} \iff \nabla f(\mathbf{x}^*) \in \operatorname{span}\{\nabla h_j(\mathbf{x}^*)\}}
$$

> **Key takeaway:** Constrained optimality is *tangential* stationarity: the level curves of $f$ must be tangent to the feasible set, which is exactly the statement that $\nabla f$ is a combination of the constraint normals.

### Problem L0.2: The Multiplier Is a Price

**Problem Statement:** State the sensitivity theorem for the parametrized problem
$\min f(\mathbf{x})$ subject to $h(\mathbf{x}) = b$, and verify it explicitly on
$\min\ x^2 + y^2$ subject to $x + y = b$.

*Intuition:* Relaxing a constraint by one unit changes the achievable optimum at a rate that the multiplier measures exactly.

**Solution:**

**The theorem.** With $\mathcal{L} = f + \lambda(h - b)$ and a regular (LICQ) solution $\mathbf{x}^*(b)$
depending smoothly on $b$,

$$
\frac{d f^*}{d b} = -\lambda^*
$$

**Step 1 (solve the example).** $\mathcal{L} = x^2 + y^2 + \lambda(x + y - b)$:

$$
2x + \lambda = 0, \qquad 2y + \lambda = 0 \ \Longrightarrow\ x = y = -\frac{\lambda}{2}
$$

Feasibility $x + y = b$ gives $-\lambda = b$, i.e.

$$
\lambda^* = -b, \qquad x^* = y^* = \frac{b}{2}, \qquad f^*(b) = 2\left(\frac{b}{2}\right)^2 = \frac{b^2}{2}
$$

**Step 2 (verify the sensitivity identity).**

$$
\frac{df^*}{db} = b, \qquad -\lambda^* = -(-b) = b \quad \checkmark
$$

**Step 3 (why it is true in general).** Differentiate $f^*(b) = f(\mathbf{x}^*(b))$ using the chain rule and
then the stationarity condition $\nabla f = -\lambda\nabla h$:

$$
\frac{df^*}{db} = \nabla f(\mathbf{x}^*)^T\frac{d\mathbf{x}^*}{db} = -\lambda^*\,\nabla h(\mathbf{x}^*)^T\frac{d\mathbf{x}^*}{db}
$$

and differentiating the identity $h(\mathbf{x}^*(b)) = b$ gives
$\nabla h^T \frac{d\mathbf{x}^*}{db} = 1$, so $\frac{df^*}{db} = -\lambda^*$. (The full proof, including
smooth dependence of $\mathbf{x}^*$ on $b$ via the implicit function theorem, is Problem L1.5.)

$$
\boxed{\frac{\partial f^*}{\partial b_j} = -\lambda_j^*: \ \text{the multiplier is the exchange rate between objective and constraint level}}
$$

> **Key takeaway:** A multiplier is not an artefact of the algebra — it is a measured quantity with units of *objective per unit of constraint*, which is why it is a shadow price in economics and an inverse temperature in statistical mechanics.

### Problem L0.3: The Lagrangian Has a Saddle, Not a Minimum

**Problem Statement:** For $\min x^2$ subject to $x = 1$, write the Lagrangian and show that
$(x^*, \lambda^*)$ is a stationary point that is a minimum in $x$ and a maximum (indeed a flat extremum)
in $\lambda$ — never a joint minimum.

*Intuition:* The Lagrangian is linear in $\lambda$, and a linear function has no interior minimum.

**Solution:**

**Step 1 (Lagrangian and stationarity).**

$$
\mathcal{L}(x,\lambda) = x^2 + \lambda(x - 1)
$$

$$
\partial_x\mathcal{L} = 2x + \lambda = 0, \qquad \partial_\lambda\mathcal{L} = x - 1 = 0
$$

so $x^* = 1$, $\lambda^* = -2$, with $\mathcal{L}(1,-2) = 1$.

**Step 2 (Hessian of $\mathcal{L}$ in $(x,\lambda)$).**

$$
\nabla^2\mathcal{L} = \begin{bmatrix} 2 & 1 \\ 1 & 0\end{bmatrix}, \qquad \det = -1 \lt 0
$$

Indefinite: the eigenvalues are $1 \pm \sqrt{2}$, one positive and one negative. So $(x^*,\lambda^*)$ is a
**saddle point** of $\mathcal{L}$, never a joint minimum.

**Step 3 (why, structurally).** $\mathcal{L}$ is affine in $\lambda$ for fixed $x$:

$$
\mathcal{L}(x, \cdot) = x^2 + \lambda(x-1)
$$

If $x \neq 1$ the map $\lambda\mapsto \mathcal{L}$ is a nonconstant line, unbounded in both directions; if
$x = 1$ it is constant. Hence

$$
\sup_{\lambda}\mathcal{L}(x,\lambda) = \begin{cases} x^2, & x = 1 \\ +\infty, & x \neq 1\end{cases}
$$

which is exactly the original constrained problem: $\min_x \sup_\lambda \mathcal{L} = 1 = f^*$. The
multiplier acts as an infinite penalty enforcer, not as a variable to be minimized.

**Step 4 (the correct picture).** Descend in $\mathbf{x}$, ascend in $\boldsymbol{\lambda}$ — the
*min-max* structure that primal-dual algorithms (dual ascent, augmented Lagrangian, ADMM) implement
literally.

$$
\boxed{\min_{\mathbf{x}}\max_{\boldsymbol{\lambda}}\mathcal{L} = f^*, \quad \nabla^2\mathcal{L} \text{ is indefinite at } (\mathbf{x}^*,\boldsymbol{\lambda}^*)}
$$

> **Key takeaway:** "Set all partial derivatives of $\mathcal{L}$ to zero" is the right *equation* but the wrong *picture*: you are locating a mountain pass in $(\mathbf{x},\boldsymbol{\lambda})$-space, not a valley floor.

### Problem L0.4: LICQ Is a Genuine Hypothesis

**Problem Statement:** Consider $\min f(x) = x$ subject to $h(x) = x^2 = 0$ on $\mathbb{R}$. Find the
solution, show that no multiplier exists, and identify which hypothesis of the Lagrange theorem fails.
Then show that reparametrizing the same feasible set repairs the situation.

*Intuition:* The multiplier theorem describes the geometry of the feasible set through the *gradients* of the constraint functions — a degenerate parametrization hides that geometry.

**Solution:**

**Step 1 (the solution).** The feasible set is $\{x : x^2 = 0\} = \{0\}$, a single point, so $x^* = 0$
trivially, with $f^* = 0$.

**Step 2 (no multiplier exists).** $\nabla f(0) = 1$ and $\nabla h(0) = 2x\big\vert_{x=0} = 0$. The
stationarity condition would require

$$
1 + \lambda\cdot 0 = 0
$$

which has no solution $\lambda \in \mathbb{R}$. The Lagrange conditions therefore **fail at the optimizer**.

**Step 3 (which hypothesis).** LICQ requires the active constraint gradients
$\{\nabla h_j(\mathbf{x}^*)\}$ to be **linearly independent**. Here the single gradient is the zero vector,
and $\{\mathbf{0}\}$ is linearly dependent: LICQ fails. Geometrically, the linearized feasible set
$\{d : \nabla h(0)d = 0\} = \mathbb{R}$ is one-dimensional, while the true feasible set $\{0\}$ is
zero-dimensional — the linearization is a lie, and the tangent-space argument (Problem L1.3) collapses.

**Step 4 (reparametrize).** The *same* feasible set $\{0\}$ is also described by $\tilde{h}(x) = x = 0$,
for which $\nabla\tilde{h}(0) = 1 \neq 0$: LICQ holds, and stationarity $1 + \lambda\cdot 1 = 0$ gives
$\lambda^* = -1$ with no difficulty.

**Step 5 (the weaker rescue).** Even without LICQ, the **Fritz John** conditions always hold: there exist
$(\mu_0, \lambda) \neq (0,0)$, $\mu_0 \ge 0$, with $\mu_0\nabla f + \lambda\nabla h = 0$. Here
$\mu_0 = 0$, $\lambda = 1$ works — an *abnormal* multiplier, in which the objective has been annihilated
and the condition carries no information about $f$ (see Problem L3.3).

$$
\boxed{\text{LICQ fails} \Rightarrow \text{no } \lambda \text{ with } \nabla f + \lambda\nabla h = 0; \ \text{same set via } \tilde{h} = x \text{ gives } \lambda^* = -1}
$$

> **Key takeaway:** Constraint qualifications are statements about the *description* of the feasible set, not about the set itself — which is why modelers are advised to avoid squared or otherwise degenerate constraint forms.

## Level 1 — Foundation

### Problem L1.1: A Complete Two-Variable Lagrange Computation

**Problem Statement:** Find all extrema of $f(x,y) = xy$ on the circle $x^2 + y^2 = 2$ using Lagrange
multipliers. Identify which are maxima and which minima, and verify with a direct parametrization.

*Intuition:* Four stationary points on a circle, alternating between the two diagonals — symmetry does most of the work.

**Solution:**

**Step 1 (Lagrangian and stationarity).** With $h(x,y) = x^2 + y^2 - 2$ and
$\mathcal{L} = xy + \lambda(x^2+y^2-2)$:

$$
\partial_x\mathcal{L} = y + 2\lambda x = 0, \qquad \partial_y\mathcal{L} = x + 2\lambda y = 0, \qquad x^2 + y^2 = 2
$$

**Step 2 (solve).** From the first two, $y = -2\lambda x$ and $x = -2\lambda y = 4\lambda^2 x$. Since
$x = 0$ would force $y = 0$, violating feasibility, we may divide: $4\lambda^2 = 1$, so
$\lambda = \pm\tfrac{1}{2}$.

- $\lambda = -\tfrac12$: $y = x$; feasibility gives $2x^2 = 2$, so $(1,1)$ and $(-1,-1)$, with $f = 1$.
- $\lambda = +\tfrac12$: $y = -x$; feasibility gives $(1,-1)$ and $(-1,1)$, with $f = -1$.

**Step 3 (classify with the restricted Hessian).** $\nabla^2_{\mathbf{xx}}\mathcal{L} = \begin{bmatrix} 2\lambda & 1 \\ 1 & 2\lambda\end{bmatrix}$.

- At $(1,1)$, $\lambda = -\tfrac12$: $\nabla h = (2,2)$, tangent direction $\mathbf{d} = (1,-1)$, and

$$
\mathbf{d}^T\begin{bmatrix} -1 & 1 \\ 1 & -1\end{bmatrix}\mathbf{d} = -1 - 1 - 1 - 1 = -4 \lt 0 \ \Rightarrow \ \text{local maximum}
$$

- At $(1,-1)$, $\lambda = +\tfrac12$: $\nabla h = (2,-2)$, tangent $\mathbf{d} = (1,1)$, and

$$
\mathbf{d}^T\begin{bmatrix} 1 & 1 \\ 1 & 1\end{bmatrix}\mathbf{d} = 4 \gt 0 \ \Rightarrow \ \text{local minimum}
$$

**Step 4 (independent check).** Parametrize: $x = \sqrt2\cos\theta$, $y = \sqrt2\sin\theta$, so
$f = 2\cos\theta\sin\theta = \sin 2\theta$, with maximum $1$ at $\theta = \pi/4, 5\pi/4$ (the points
$(1,1), (-1,-1)$) and minimum $-1$ at $\theta = 3\pi/4, 7\pi/4$ (the points $(-1,1),(1,-1)$) — exactly as
found.

**Step 5 (note on the unconstrained Hessian).** $\nabla^2 f = \begin{bmatrix} 0 & 1 \\ 1 & 0\end{bmatrix}$
is indefinite at every point: the origin is a saddle of $f$. The constraint is what creates genuine maxima
and minima, and it is $\nabla^2_{\mathbf{xx}}\mathcal{L}$ *restricted to $T$* that detects them.

$$
\boxed{f_{\max} = 1 \ \text{at} \ \pm(1,1)\ (\lambda = -\tfrac12); \qquad f_{\min} = -1 \ \text{at} \ \pm(1,-1)\ (\lambda = +\tfrac12)}
$$

> **Key takeaway:** The multiplier is determined along with the point, and its sign is informative: here $\lambda$ flips sign between the maximizing and minimizing branches, matching the sensitivity reading $df^*/db = -\lambda$ with $b$ the squared radius.

### Problem L1.2: The Tangent Space Is the Kernel of the Constraint Jacobian

**Problem Statement:** Let $\mathbf{h}:\mathbb{R}^n\to\mathbb{R}^p$ be $\mathcal{C}^1$ and let
$\mathcal{M} = \{\mathbf{x} : \mathbf{h}(\mathbf{x}) = \mathbf{0}\}$. Prove (a) every feasible curve
through $\mathbf{x}^* \in \mathcal{M}$ has velocity in $\ker D\mathbf{h}(\mathbf{x}^*)$; and (b) under
LICQ, every $\mathbf{d} \in \ker D\mathbf{h}(\mathbf{x}^*)$ is the velocity of some feasible curve.

*Intuition:* Differentiating the constraint along a curve kills the normal directions; the implicit function theorem builds the curves that realize the survivors.

**Solution:**

**(a) Velocities lie in the kernel.** Let $\mathbf{x}(t)$ be $\mathcal{C}^1$ with
$\mathbf{x}(0) = \mathbf{x}^*$ and $\mathbf{h}(\mathbf{x}(t)) = \mathbf{0}$ for all $t$ near $0$.
Differentiating at $t = 0$ by the chain rule,

$$
\frac{d}{dt}\mathbf{h}(\mathbf{x}(t))\bigg\vert_{t=0} = D\mathbf{h}(\mathbf{x}^*)\,\dot{\mathbf{x}}(0) = \mathbf{0}
$$

so $\dot{\mathbf{x}}(0) \in \ker D\mathbf{h}(\mathbf{x}^*) =: T(\mathbf{x}^*)$. This direction needs no
constraint qualification.

**(b) Every kernel vector is realized (under LICQ).** Assume the rows of
$A := D\mathbf{h}(\mathbf{x}^*) \in \mathbb{R}^{p\times n}$ are linearly independent, so
$\operatorname{rank}A = p$ and $AA^T$ is invertible. Fix $\mathbf{d}$ with $A\mathbf{d} = \mathbf{0}$ and
look for a curve of the form

$$
\mathbf{x}(t) = \mathbf{x}^* + t\mathbf{d} + A^T\mathbf{u}(t)
$$

with correction $\mathbf{u}(t) \in \mathbb{R}^p$ to be determined. Define
$F(t, \mathbf{u}) = \mathbf{h}\left(\mathbf{x}^* + t\mathbf{d} + A^T\mathbf{u}\right)$. Then
$F(0,\mathbf{0}) = \mathbf{0}$ and

$$
D_{\mathbf{u}}F(0,\mathbf{0}) = A A^T \ \text{is invertible}
$$

By the implicit function theorem there is a $\mathcal{C}^1$ function $\mathbf{u}(t)$ with
$\mathbf{u}(0)=\mathbf{0}$ and $F(t,\mathbf{u}(t)) = \mathbf{0}$ for $t$ near $0$: the curve is feasible.
Differentiating that identity at $t=0$,

$$
A\mathbf{d} + AA^T\dot{\mathbf{u}}(0) = \mathbf{0} \ \Longrightarrow\ \dot{\mathbf{u}}(0) = \mathbf{0}
$$

using $A\mathbf{d} = \mathbf{0}$. Hence $\dot{\mathbf{x}}(0) = \mathbf{d}$, as required.

**Consequence.** Under LICQ the *linearized* feasible cone and the *true* set of feasible velocities
coincide:

$$
\boxed{T(\mathbf{x}^*) = \ker D\mathbf{h}(\mathbf{x}^*) = \left\{\mathbf{d} : \nabla h_j(\mathbf{x}^*)^T\mathbf{d} = 0, \ j = 1,\dots,p\right\}, \quad \dim T = n - p}
$$

> **Key takeaway:** LICQ is exactly the hypothesis that makes the linearization *faithful*; Problem L0.4 is the cautionary case where the kernel is one-dimensional while the feasible set is a point.

### Problem L1.3: Proof of the Lagrange Multiplier Theorem

**Problem Statement:** Let $\mathbf{x}^*$ be a local minimizer of $f$ on
$\mathcal{M} = \{\mathbf{h} = \mathbf{0}\}$, with $f, \mathbf{h} \in \mathcal{C}^1$ and LICQ holding at
$\mathbf{x}^*$. Prove there exists a unique $\boldsymbol{\lambda}^* \in \mathbb{R}^p$ with
$\nabla f(\mathbf{x}^*) + \sum_j \lambda_j^*\nabla h_j(\mathbf{x}^*) = \mathbf{0}$.

*Intuition:* Optimality kills the directional derivative along every tangent direction, so $\nabla f$ lies in the orthogonal complement of the tangent space — which is the span of the constraint gradients.

**Solution:**

**Step 1 (tangential stationarity).** Let $\mathbf{d} \in T(\mathbf{x}^*) = \ker D\mathbf{h}(\mathbf{x}^*)$.
By Problem L1.2(b) there is a feasible $\mathcal{C}^1$ curve $\mathbf{x}(t)$ with $\mathbf{x}(0)=\mathbf{x}^*$
and $\dot{\mathbf{x}}(0) = \mathbf{d}$. The scalar function $\phi(t) = f(\mathbf{x}(t))$ has a local
minimum at $t = 0$ (since $\mathbf{x}(t)$ is feasible and $\mathbf{x}^*$ is a local constrained minimizer),
so by Fermat's rule in one dimension,

$$
0 = \phi'(0) = \nabla f(\mathbf{x}^*)^T\mathbf{d}
$$

Because $T$ is a subspace, this holds for $\pm\mathbf{d}$ alike; the conclusion is the *equality*
$\nabla f(\mathbf{x}^*) \perp T(\mathbf{x}^*)$.

**Step 2 (orthogonal complement of a kernel).** For $A = D\mathbf{h}(\mathbf{x}^*)$, linear algebra gives

$$
\left(\ker A\right)^{\perp} = \operatorname{range}\left(A^T\right) = \operatorname{span}\left\{\nabla h_1(\mathbf{x}^*), \dots, \nabla h_p(\mathbf{x}^*)\right\}
$$

**Step 3 (existence).** By Steps 1–2, $\nabla f(\mathbf{x}^*) \in \operatorname{range}(A^T)$, so there is
$\boldsymbol{\mu}$ with $\nabla f(\mathbf{x}^*) = A^T\boldsymbol{\mu} = \sum_j \mu_j\nabla h_j(\mathbf{x}^*)$.
Setting $\boldsymbol{\lambda}^* = -\boldsymbol{\mu}$ gives the stated form

$$
\nabla f(\mathbf{x}^*) + \sum_{j=1}^p \lambda_j^*\,\nabla h_j(\mathbf{x}^*) = \mathbf{0} \quad \text{i.e.} \quad \nabla_{\mathbf{x}}\mathcal{L}(\mathbf{x}^*,\boldsymbol{\lambda}^*) = \mathbf{0}
$$

**Step 4 (uniqueness).** If $A^T\boldsymbol{\lambda} = A^T\boldsymbol{\lambda}'$ then
$A^T(\boldsymbol{\lambda}-\boldsymbol{\lambda}') = \mathbf{0}$; LICQ says the columns of $A^T$ (the
constraint gradients) are linearly independent, so $\boldsymbol{\lambda} = \boldsymbol{\lambda}'$.

**Step 5 (counting).** The system $\nabla_{\mathbf{x}}\mathcal{L} = \mathbf{0}$, $\mathbf{h} = \mathbf{0}$
comprises $n + p$ equations in the $n + p$ unknowns $(\mathbf{x},\boldsymbol{\lambda})$ — square, as it
must be for a well-posed problem.

$$
\boxed{\nabla f(\mathbf{x}^*) = -\sum_j \lambda_j^*\nabla h_j(\mathbf{x}^*), \quad \boldsymbol{\lambda}^* \text{ unique under LICQ}}
$$

> **Key takeaway:** The proof is two lines of geometry plus one line of linear algebra: minimality gives $\nabla f \perp \ker A$, and $(\ker A)^{\perp} = \operatorname{range}(A^T)$ turns "perpendicular to the tangent space" into "a combination of constraint gradients".

### Problem L1.4: The Bordered Hessian in Action

**Problem Statement:** For the problem of Problem L1.1 ($f = xy$ on $x^2+y^2 = 2$), form the bordered
Hessian at each stationary point and confirm the classification using the determinant sign rule: for
$n = 2$, $p = 1$, $\det \bar{H} \gt 0$ indicates a constrained local **maximum**, $\det \bar{H} \lt 0$ a
constrained local **minimum**.

*Intuition:* Bordering with the constraint gradient is a determinant-level way of restricting the quadratic form to the tangent space.

**Solution:**

**Step 1 (the bordered Hessian).** With $\mathcal{L} = f + \lambda h$,

$$
\bar{H} = \begin{bmatrix} 0 & h_x & h_y \\ h_x & \mathcal{L}_{xx} & \mathcal{L}_{xy} \\ h_y & \mathcal{L}_{yx} & \mathcal{L}_{yy}\end{bmatrix}, \qquad \nabla^2_{\mathbf{xx}}\mathcal{L} = \begin{bmatrix} 2\lambda & 1 \\ 1 & 2\lambda\end{bmatrix}, \qquad \nabla h = (2x, 2y)
$$

**Step 2 (at $(1,1)$ with $\lambda = -\tfrac12$).** Here $\nabla h = (2,2)$ and
$\nabla^2_{\mathbf{xx}}\mathcal{L} = \begin{bmatrix} -1 & 1 \\ 1 & -1\end{bmatrix}$, so

$$
\bar{H} = \begin{bmatrix} 0 & 2 & 2 \\ 2 & -1 & 1 \\ 2 & 1 & -1\end{bmatrix}
$$

Expanding along the first row:

$$
\det\bar{H} = -2\det\begin{bmatrix} 2 & 1 \\ 2 & -1\end{bmatrix} + 2\det\begin{bmatrix} 2 & -1 \\ 2 & 1\end{bmatrix} = -2(-4) + 2(4) = 16 \gt 0
$$

$\Rightarrow$ constrained local **maximum**, matching $f = 1$ from Problem L1.1.

**Step 3 (at $(1,-1)$ with $\lambda = +\tfrac12$).** Here $\nabla h = (2,-2)$ and
$\nabla^2_{\mathbf{xx}}\mathcal{L} = \begin{bmatrix} 1 & 1 \\ 1 & 1\end{bmatrix}$, so

$$
\bar{H} = \begin{bmatrix} 0 & 2 & -2 \\ 2 & 1 & 1 \\ -2 & 1 & 1\end{bmatrix}, \qquad \det\bar{H} = -2\det\begin{bmatrix} 2 & 1 \\ -2 & 1\end{bmatrix} + (-2)\det\begin{bmatrix} 2 & 1 \\ -2 & 1\end{bmatrix} = -2(4) - 2(4) = -16 \lt 0
$$

$\Rightarrow$ constrained local **minimum**, matching $f = -1$.

**Step 4 (why the rule works).** Restricting the quadratic form to the one-dimensional tangent space, the
identity for $n=2$, $p=1$ is

$$
\det\bar{H} = -\lVert \nabla h\rVert^2 \left(\mathbf{t}^T\nabla^2_{\mathbf{xx}}\mathcal{L}\,\mathbf{t}\right), \qquad \mathbf{t} = \frac{(-h_y, h_x)}{\lVert \nabla h\rVert}
$$

so $\det\bar{H}$ carries the *opposite* sign of the tangential curvature: positive determinant means
negative curvature (maximum), negative determinant means positive curvature (minimum). Check at $(1,1)$:
$\lVert \nabla h\rVert^2 = 8$ and the tangential curvature was $-4/2 = -2$ for the unit tangent
$\mathbf{t} = (1,-1)/\sqrt2$, giving $-8(-2) = 16$ $\checkmark$.

$$
\boxed{\det\bar{H}(1,1) = 16 \gt 0 \Rightarrow \text{max}; \qquad \det\bar{H}(1,-1) = -16 \lt 0 \Rightarrow \text{min}}
$$

> **Key takeaway:** The bordered Hessian is a bookkeeping device that automatically confines the second-order test to feasible directions — a determinant computation replacing the manual construction of a tangent basis.

### Problem L1.5: Proof of the Sensitivity Theorem

**Problem Statement:** Consider the perturbed family $\min f(\mathbf{x})$ subject to
$\mathbf{h}(\mathbf{x}) = \mathbf{b}$. Assuming a regular solution $\mathbf{x}^*(\mathbf{b})$ with
multipliers $\boldsymbol{\lambda}^*(\mathbf{b})$ depending differentiably on $\mathbf{b}$, prove
$\nabla_{\mathbf{b}} f^*(\mathbf{b}) = -\boldsymbol{\lambda}^*(\mathbf{b})$.

*Intuition:* Along the optimal path the objective can only change through the constraint level, and stationarity converts that change into the multiplier.

**Solution:**

**Step 1 (the two identities we may differentiate).** For each $\mathbf{b}$ near a reference value:

$$
\mathbf{h}\left(\mathbf{x}^*(\mathbf{b})\right) = \mathbf{b}, \qquad \nabla f\left(\mathbf{x}^*(\mathbf{b})\right) + \sum_{j}\lambda_j^*(\mathbf{b})\,\nabla h_j\left(\mathbf{x}^*(\mathbf{b})\right) = \mathbf{0}
$$

(The differentiable dependence follows from applying the implicit function theorem to the square system
$\nabla_{\mathbf{x}}\mathcal{L} = \mathbf{0}$, $\mathbf{h} - \mathbf{b} = \mathbf{0}$, whose Jacobian in
$(\mathbf{x},\boldsymbol{\lambda})$ is the KKT matrix — invertible precisely under LICQ plus the
second-order sufficient condition.)

**Step 2 (differentiate the constraint identity).** Writing $J = D_{\mathbf{b}}\mathbf{x}^*(\mathbf{b}) \in \mathbb{R}^{n\times p}$
and $A = D\mathbf{h}(\mathbf{x}^*)$,

$$
A\,J = I_p
$$

**Step 3 (differentiate the optimal value).** $f^*(\mathbf{b}) = f(\mathbf{x}^*(\mathbf{b}))$, so by the
chain rule

$$
\nabla_{\mathbf{b}}f^*(\mathbf{b})^T = \nabla f(\mathbf{x}^*)^T J
$$

**Step 4 (substitute stationarity).** Stationarity says
$\nabla f(\mathbf{x}^*)^T = -\boldsymbol{\lambda}^{*T}A$, hence

$$
\nabla_{\mathbf{b}}f^*(\mathbf{b})^T = -\boldsymbol{\lambda}^{*T}A J = -\boldsymbol{\lambda}^{*T} I_p = -\boldsymbol{\lambda}^{*T}
$$

**Step 5 (an envelope-theorem restatement).** Equivalently, define
$\mathcal{L}(\mathbf{x},\boldsymbol{\lambda},\mathbf{b}) = f(\mathbf{x}) + \boldsymbol{\lambda}^T(\mathbf{h}(\mathbf{x})-\mathbf{b})$.
At the optimum $\mathcal{L} = f^*$, and the envelope theorem says that only the *explicit* dependence on
$\mathbf{b}$ survives differentiation:

$$
\frac{\partial f^*}{\partial b_j} = \frac{\partial \mathcal{L}}{\partial b_j}\bigg\vert_{(\mathbf{x}^*,\boldsymbol{\lambda}^*)} = -\lambda_j^*
$$

$$
\boxed{\nabla_{\mathbf{b}} f^*(\mathbf{b}) = -\boldsymbol{\lambda}^*(\mathbf{b})}
$$

> **Key takeaway:** The sensitivity theorem is the envelope theorem in disguise: because $\mathbf{x}^*$ is stationary, first-order changes in the *decision* contribute nothing, and the entire derivative is the explicit price term.

### Problem L1.6: Distance from a Point to a Hyperplane

**Problem Statement:** Solve $\min \frac12\lVert \mathbf{x} - \mathbf{x}_0\rVert^2$ subject to
$\mathbf{a}^T\mathbf{x} = b$ (with $\mathbf{a} \neq \mathbf{0}$) using multipliers, obtain the classical
distance formula, and interpret $\lambda^*$ via sensitivity.

*Intuition:* The shortest path to a plane is along its normal, and the multiplier measures how much closer the plane must come to buy one unit of squared distance.

**Solution:**

**Step 1 (Lagrangian and stationarity).**

$$
\mathcal{L} = \tfrac12\lVert \mathbf{x}-\mathbf{x}_0\rVert^2 + \lambda\left(\mathbf{a}^T\mathbf{x} - b\right), \qquad \nabla_{\mathbf{x}}\mathcal{L} = \mathbf{x}-\mathbf{x}_0 + \lambda\mathbf{a} = \mathbf{0}
$$

so $\mathbf{x} = \mathbf{x}_0 - \lambda\mathbf{a}$: the solution moves from $\mathbf{x}_0$ **along the
normal** $\mathbf{a}$, and nowhere else.

**Step 2 (impose feasibility).**

$$
\mathbf{a}^T\mathbf{x}_0 - \lambda\lVert \mathbf{a}\rVert^2 = b \ \Longrightarrow\ \lambda^* = \frac{\mathbf{a}^T\mathbf{x}_0 - b}{\lVert \mathbf{a}\rVert^2}
$$

**Step 3 (the projection and the distance).**

$$
\mathbf{x}^* = \mathbf{x}_0 - \frac{\mathbf{a}^T\mathbf{x}_0 - b}{\lVert \mathbf{a}\rVert^2}\,\mathbf{a}, \qquad \lVert \mathbf{x}^*-\mathbf{x}_0\rVert = \frac{\lvert \mathbf{a}^T\mathbf{x}_0 - b\rvert}{\lVert \mathbf{a}\rVert}
$$

the familiar point-to-hyperplane distance. The optimal objective value is

$$
f^*(b) = \frac12\lVert \mathbf{x}^*-\mathbf{x}_0\rVert^2 = \frac{\left(\mathbf{a}^T\mathbf{x}_0 - b\right)^2}{2\lVert \mathbf{a}\rVert^2}
$$

**Step 4 (sensitivity check).**

$$
\frac{df^*}{db} = \frac{-\left(\mathbf{a}^T\mathbf{x}_0-b\right)}{\lVert \mathbf{a}\rVert^2} = -\lambda^* \quad \checkmark
$$

**Step 5 (SOSC).** $\nabla^2_{\mathbf{xx}}\mathcal{L} = I \succ 0$ on all of $\mathbb{R}^n$, hence in
particular on the tangent space $T = \{\mathbf{d} : \mathbf{a}^T\mathbf{d} = 0\}$: the point is a strict
(and, by convexity, global) constrained minimizer.

$$
\boxed{\mathbf{x}^* = \mathbf{x}_0 - \frac{\mathbf{a}^T\mathbf{x}_0-b}{\lVert \mathbf{a}\rVert^2}\mathbf{a}, \qquad \operatorname{dist} = \frac{\lvert \mathbf{a}^T\mathbf{x}_0-b\rvert}{\lVert \mathbf{a}\rVert}}
$$

> **Key takeaway:** The stationarity equation itself tells you the *geometry* before any algebra: $\nabla f \parallel \nabla h$ says "move along the normal", and the constraint then fixes how far. This is the computation behind the SVM margin $2/\lVert \mathbf{w}\rVert$.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: Maximum Entropy Gives the Boltzmann Distribution

**Problem Statement:** Maximize the Shannon entropy $H(\mathbf{p}) = -\sum_{i=1}^n p_i\ln p_i$ subject to
$\sum_i p_i = 1$ and $\sum_i p_i E_i = U$. Derive the optimal distribution, identify the multipliers, and
give the thermodynamic interpretation.

*Intuition:* Entropy maximization with a linear constraint always produces an exponential family; the multiplier of the energy constraint is the inverse temperature.

**Solution:**

**Step 1 (Lagrangian).**

$$
\mathcal{L} = -\sum_i p_i\ln p_i + \alpha\left(\sum_i p_i - 1\right) + \beta\left(\sum_i p_i E_i - U\right)
$$

(the sign convention for $\alpha,\beta$ is chosen to make the final formulas standard; we are maximizing,
so stationarity of $\mathcal{L}$ is still the condition).

**Step 2 (stationarity).**

$$
\frac{\partial\mathcal{L}}{\partial p_i} = -\ln p_i - 1 + \alpha + \beta E_i = 0 \ \Longrightarrow\ p_i = e^{\alpha - 1}\,e^{\beta E_i}
$$

**Step 3 (normalize).** Writing $\beta \to -\beta$ for the physics convention
($p_i \propto e^{-\beta E_i}$) and enforcing $\sum_i p_i = 1$:

$$
p_i = \frac{e^{-\beta E_i}}{Z(\beta)}, \qquad Z(\beta) = \sum_{j=1}^n e^{-\beta E_j}
$$

the **Boltzmann–Gibbs distribution**, with $e^{\alpha-1} = 1/Z$ so that $\alpha = 1 - \ln Z$. Note that
this is exactly the **softmax** of the vector $(-\beta E_1, \dots, -\beta E_n)$.

**Step 4 (the second multiplier is determined by $U$).** $\beta$ solves

$$
U = \sum_i p_i E_i = -\frac{d}{d\beta}\ln Z(\beta)
$$

which is monotone in $\beta$ because $\frac{d^2}{d\beta^2}\ln Z = \operatorname{Var}(E) \gt 0$; so $\beta$
exists and is unique for any attainable $U$.

**Step 5 (thermodynamics and sensitivity).** At the optimum,

$$
H^* = -\sum_i p_i\left(-\beta E_i - \ln Z\right) = \beta U + \ln Z
$$

Differentiating with respect to $U$ and using $\frac{d\ln Z}{d\beta} = -U$:

$$
\frac{dH^*}{dU} = \beta + U\frac{d\beta}{dU} + \frac{d\ln Z}{d\beta}\frac{d\beta}{dU} = \beta
$$

exactly the sensitivity theorem with the multiplier $\beta$, and the thermodynamic identity
$\frac{\partial S}{\partial U} = \frac{1}{k_BT}$ identifies

$$
\beta = \frac{1}{k_B T}
$$

$$
\boxed{p_i = \frac{e^{-\beta E_i}}{Z(\beta)}, \qquad \beta = \frac{\partial H^*}{\partial U} = \frac{1}{k_BT}}
$$

> **Key takeaway:** Temperature is a Lagrange multiplier. The same computation with different constraint functions generates the entire exponential family — Gaussian (fixed variance), exponential (fixed mean on $[0,\infty)$), softmax classifiers (fixed feature expectations).

### Problem L2.2: Eigenvalues as Constrained Optima — The Rayleigh Quotient

**Problem Statement:** Solve $\max\ \mathbf{x}^TA\mathbf{x}$ subject to $\mathbf{x}^T\mathbf{x} = 1$ for
symmetric $A$ using multipliers. Show that the stationary points are the eigenvectors, the multipliers are
the eigenvalues, and the maximum value is $\lambda_{\max}(A)$.

*Intuition:* Stationarity of a quadratic form on the sphere *is* the eigenvalue equation.

**Solution:**

**Step 1 (Lagrangian and stationarity).** With $h(\mathbf{x}) = \mathbf{x}^T\mathbf{x} - 1$,

$$
\mathcal{L} = \mathbf{x}^TA\mathbf{x} - \lambda\left(\mathbf{x}^T\mathbf{x}-1\right), \qquad \nabla_{\mathbf{x}}\mathcal{L} = 2A\mathbf{x} - 2\lambda\mathbf{x} = \mathbf{0}
$$

so

$$
A\mathbf{x} = \lambda\mathbf{x}
$$

**The optimality condition is literally the eigenvalue equation**, and $\lambda$ is the multiplier.

**Step 2 (LICQ).** $\nabla h = 2\mathbf{x}$, which is nonzero on the feasible sphere, so LICQ holds at
every feasible point and no stationary point is missed.

**Step 3 (objective value at a stationary point).** If $A\mathbf{x} = \lambda\mathbf{x}$ with
$\lVert \mathbf{x}\rVert = 1$,

$$
\mathbf{x}^TA\mathbf{x} = \lambda\,\mathbf{x}^T\mathbf{x} = \lambda
$$

So the objective *equals* the multiplier at every stationary point. Comparing all of them, the maximum is
$\lambda_{\max}(A)$, attained at a unit top eigenvector, and the minimum is $\lambda_{\min}(A)$.

**Step 4 (existence).** The sphere is compact and the objective continuous, so both extrema are attained —
this argument is a complete, multiplier-based proof that a real symmetric matrix has a real eigenvalue,
the first step of the spectral theorem.

**Step 5 (second-order check).** $\nabla^2_{\mathbf{xx}}\mathcal{L} = 2(A - \lambda I)$. At the top
eigenvector, the tangent space is $\mathbf{x}^{\perp} = \operatorname{span}\{\mathbf{v}_2,\dots,\mathbf{v}_n\}$,
and for $\mathbf{d} = \sum_{i\ge2}c_i\mathbf{v}_i$,

$$
\mathbf{d}^T\,2(A-\lambda_{\max}I)\,\mathbf{d} = 2\sum_{i\ge2}c_i^2\left(\lambda_i - \lambda_{\max}\right) \le 0
$$

negative semidefinite on $T$ (negative definite if $\lambda_{\max}$ is simple) — the second-order
certificate of a constrained maximum.

**Step 6 (sensitivity).** Perturbing the constraint to $\lVert \mathbf{x}\rVert^2 = b$ scales the optimum
to $f^*(b) = \lambda_{\max}b$, so $df^*/db = \lambda_{\max}$, consistent with the multiplier convention
used above.

$$
\boxed{\max_{\lVert \mathbf{x}\rVert = 1}\mathbf{x}^TA\mathbf{x} = \lambda_{\max}(A), \ \text{attained at } A\mathbf{x} = \lambda_{\max}\mathbf{x}}
$$

> **Key takeaway:** Spectral theory is constrained optimization: eigenvectors are the stationary points of a quadratic form on the sphere and eigenvalues are the multipliers, which is why PCA, spectral clustering and the power method all fall out of this one Lagrangian.

### Problem L2.3: Equality-Constrained Least Squares and the KKT Matrix

**Problem Statement:** Solve $\min \frac12\lVert X\mathbf{w}-\mathbf{y}\rVert^2$ subject to
$C\mathbf{w} = \mathbf{d}$ (with $C \in \mathbb{R}^{p\times n}$ of full row rank). Write the first-order
system as a single block linear system, solve it explicitly when $X^TX \succ 0$, and interpret the
correction term.

*Intuition:* The constrained solution is the unconstrained one, pulled back onto the feasible affine subspace by a multiple of the constraint normals.

**Solution:**

**Step 1 (stationarity).** $\mathcal{L} = \frac12\lVert X\mathbf{w}-\mathbf{y}\rVert^2 + \boldsymbol{\lambda}^T(C\mathbf{w}-\mathbf{d})$:

$$
X^TX\mathbf{w} - X^T\mathbf{y} + C^T\boldsymbol{\lambda} = \mathbf{0}, \qquad C\mathbf{w} = \mathbf{d}
$$

**Step 2 (block form — the KKT matrix).**

$$
\begin{bmatrix} X^TX & C^T \\ C & 0 \end{bmatrix}\begin{pmatrix} \mathbf{w} \\ \boldsymbol{\lambda}\end{pmatrix} = \begin{pmatrix} X^T\mathbf{y} \\ \mathbf{d}\end{pmatrix}
$$

a symmetric **indefinite** saddle-point system of size $(n+p)$ — the numerical fingerprint of Problem L0.3,
and the system every equality-constrained QP solver factors.

**Step 3 (solve by elimination).** Let $G = X^TX \succ 0$ and $\mathbf{w}_{\text{ls}} = G^{-1}X^T\mathbf{y}$
be the unconstrained least squares solution. From the first block,
$\mathbf{w} = \mathbf{w}_{\text{ls}} - G^{-1}C^T\boldsymbol{\lambda}$. Substituting into $C\mathbf{w}=\mathbf{d}$:

$$
C\mathbf{w}_{\text{ls}} - CG^{-1}C^T\boldsymbol{\lambda} = \mathbf{d} \ \Longrightarrow\ \boldsymbol{\lambda}^* = \left(CG^{-1}C^T\right)^{-1}\left(C\mathbf{w}_{\text{ls}} - \mathbf{d}\right)
$$

(the *Schur complement* $CG^{-1}C^T$ is invertible because $C$ has full row rank and $G \succ 0$). Hence

$$
\mathbf{w}^* = \mathbf{w}_{\text{ls}} - G^{-1}C^T\left(CG^{-1}C^T\right)^{-1}\left(C\mathbf{w}_{\text{ls}}-\mathbf{d}\right)
$$

**Step 4 (interpretation).** The correction is proportional to the **constraint residual**
$C\mathbf{w}_{\text{ls}} - \mathbf{d}$ of the unconstrained solution: if the unconstrained fit already
satisfies the constraint, then $\boldsymbol{\lambda}^* = \mathbf{0}$ and nothing changes — a zero shadow
price for a non-binding constraint. Otherwise the fit is projected onto $\{C\mathbf{w}=\mathbf{d}\}$ in the
$G$-norm, not the Euclidean norm: the geometry of the data decides the direction of the pullback.

**Step 5 (second-order).** $\nabla^2_{\mathbf{ww}}\mathcal{L} = X^TX$, which is positive definite on the
tangent space $\ker C$ whenever $X$ has full column rank (indeed it suffices that
$\ker X \cap \ker C = \{\mathbf{0}\}$), so SOSC holds and the solution is a strict global minimum by
convexity.

$$
\boxed{\mathbf{w}^* = \mathbf{w}_{\text{ls}} - G^{-1}C^T\left(CG^{-1}C^T\right)^{-1}\left(C\mathbf{w}_{\text{ls}}-\mathbf{d}\right), \qquad G = X^TX}
$$

> **Key takeaway:** Adding equality constraints to a quadratic problem costs one Schur-complement solve and produces a saddle-point matrix — the same structure that appears in Stokes flow, network flow, and interior-point Newton systems.

### Problem L2.4: Minimum-Variance Portfolio and the Price of the Budget

**Problem Statement:** Solve $\min \frac12\mathbf{w}^T\Sigma\mathbf{w}$ subject to
$\mathbf{1}^T\mathbf{w} = 1$ with $\Sigma \succ 0$. Give $\mathbf{w}^*$, the optimal variance, the
multiplier, and verify the sensitivity theorem against the budget level.

*Intuition:* Inverse covariance is a measure of "informativeness"; the optimal weights are proportional to it, normalized to a unit budget.

**Solution:**

**Step 1 (stationarity).** $\mathcal{L} = \frac12\mathbf{w}^T\Sigma\mathbf{w} + \lambda(\mathbf{1}^T\mathbf{w}-1)$:

$$
\Sigma\mathbf{w} + \lambda\mathbf{1} = \mathbf{0} \ \Longrightarrow\ \mathbf{w} = -\lambda\,\Sigma^{-1}\mathbf{1}
$$

**Step 2 (impose the budget).**

$$
-\lambda\,\mathbf{1}^T\Sigma^{-1}\mathbf{1} = 1 \ \Longrightarrow\ \lambda^* = -\frac{1}{\mathbf{1}^T\Sigma^{-1}\mathbf{1}}
$$

and therefore

$$
\mathbf{w}^* = \frac{\Sigma^{-1}\mathbf{1}}{\mathbf{1}^T\Sigma^{-1}\mathbf{1}}
$$

the classic **global minimum-variance portfolio**.

**Step 3 (optimal value).**

$$
f^* = \frac12\mathbf{w}^{*T}\Sigma\mathbf{w}^* = \frac{1}{2}\cdot\frac{\mathbf{1}^T\Sigma^{-1}\Sigma\Sigma^{-1}\mathbf{1}}{\left(\mathbf{1}^T\Sigma^{-1}\mathbf{1}\right)^2} = \frac{1}{2\,\mathbf{1}^T\Sigma^{-1}\mathbf{1}}
$$

so the minimum achievable variance is $\sigma_{\min}^2 = 1/(\mathbf{1}^T\Sigma^{-1}\mathbf{1})$, and
$\lambda^* = -2f^* = -\sigma^2_{\min}$.

**Step 4 (sensitivity check).** With budget level $b$ ($\mathbf{1}^T\mathbf{w}=b$) the same algebra gives
$\mathbf{w}^*(b) = b\,\mathbf{w}^*(1)$, hence

$$
f^*(b) = \frac{b^2}{2\,\mathbf{1}^T\Sigma^{-1}\mathbf{1}}, \qquad \frac{df^*}{db} = \frac{b}{\mathbf{1}^T\Sigma^{-1}\mathbf{1}} = -\lambda^*(b) \quad \checkmark
$$

**Step 5 (a special case for intuition).** If $\Sigma = \operatorname{diag}(\sigma_1^2,\dots,\sigma_n^2)$
then

$$
w_i^* = \frac{1/\sigma_i^2}{\sum_j 1/\sigma_j^2}
$$

*inverse-variance weighting* — the same formula as optimal sensor fusion, the Kalman gain in the scalar
case, and the precision-weighted average of Bayesian inference.

$$
\boxed{\mathbf{w}^* = \frac{\Sigma^{-1}\mathbf{1}}{\mathbf{1}^T\Sigma^{-1}\mathbf{1}}, \qquad \sigma^2_{\min} = \frac{1}{\mathbf{1}^T\Sigma^{-1}\mathbf{1}}, \qquad \lambda^* = -\sigma^2_{\min}}
$$

> **Key takeaway:** The multiplier of a normalization constraint is (minus) the optimal value itself here — the price of "one unit of portfolio" is exactly the variance you must accept to hold it.

### Problem L2.5: The Multiplier as a Physical Constraint Force

**Problem Statement:** A bead of mass $m$ slides frictionlessly on the sphere $x^2+y^2+z^2 = R^2$ in
uniform gravity, with potential energy $U = mgz$. Find the equilibrium positions by minimizing $U$ subject
to the constraint, and show that $-\lambda\nabla h$ is exactly the normal force exerted by the sphere.

*Intuition:* The multiplier converts "the bead cannot leave the surface" into the force that enforces it.

**Solution:**

**Step 1 (Lagrangian and stationarity).** With $h = x^2+y^2+z^2-R^2$ and $\mathcal{L} = mgz + \lambda h$:

$$
\nabla_{\mathbf{x}}\mathcal{L} = \begin{pmatrix} 2\lambda x \\ 2\lambda y \\ mg + 2\lambda z\end{pmatrix} = \mathbf{0}
$$

**Step 2 (solve).** If $\lambda = 0$ the third equation reads $mg = 0$, impossible; so $\lambda \neq 0$ and
the first two equations force $x = y = 0$. Feasibility then gives $z = \pm R$, and the third equation gives

$$
\lambda = -\frac{mg}{2z} = \mp\frac{mg}{2R}
$$

- **Bottom** $\mathbf{x}^* = (0,0,-R)$: $\lambda^* = +\frac{mg}{2R}$, $U = -mgR$ — the **minimum**
  (stable equilibrium).
- **Top** $(0,0,R)$: $\lambda = -\frac{mg}{2R}$, $U = +mgR$ — the maximum (unstable).

**Step 3 (the constraint force).** Newton's second law for a particle constrained to the surface reads
$m\ddot{\mathbf{x}} = -\nabla U + \mathbf{F}_c$ with the constraint (normal) force $\mathbf{F}_c$. At
equilibrium $\mathbf{F}_c = \nabla U$, while stationarity of $\mathcal{L}$ says
$\nabla U = -\lambda\nabla h$. At the bottom point,

$$
\mathbf{F}_c = -\lambda^*\nabla h(0,0,-R) = -\frac{mg}{2R}\cdot 2(0,0,-R) = (0,0,mg)
$$

exactly the upward normal force of magnitude $mg$ that balances gravity $(0,0,-mg)$. So

$$
\mathbf{F}_c = -\lambda\nabla h
$$

identifies the multiplier (up to the constraint's normalization) with the **magnitude of the constraint
force**.

**Step 4 (sensitivity reading).** Writing the constraint as $x^2+y^2+z^2 = b$ with $b = R^2$,

$$
U^*(b) = -mg\sqrt{b}, \qquad \frac{dU^*}{db} = -\frac{mg}{2\sqrt{b}} = -\frac{mg}{2R} = -\lambda^*
$$

$\checkmark$: enlarging the sphere lowers the achievable potential energy at exactly the rate $\lambda^*$
predicts.

**Step 5 (second order).** $\nabla^2_{\mathbf{xx}}\mathcal{L} = 2\lambda I$. At the bottom, $\lambda \gt 0$
so this is positive definite — in particular on the tangent plane $\{d_z = 0\}$ — confirming a strict
constrained minimum; at the top $\lambda \lt 0$ and the same computation certifies a maximum.

$$
\boxed{\mathbf{x}^* = (0,0,-R), \ \lambda^* = \frac{mg}{2R}, \ \mathbf{F}_c = -\lambda^*\nabla h = (0,0,mg)}
$$

> **Key takeaway:** In mechanics the multiplier is not an analogy for a force — it *is* the force. This is the content of d'Alembert's principle and the reason Lagrange multipliers appear in every constrained rigid-body simulator.

### Problem L2.6: PCA — The Second Principal Component Needs Two Multipliers

**Problem Statement:** Let $\Sigma \succ 0$ be symmetric with eigenpairs $(\lambda_1 \ge \lambda_2 \ge \dots, \mathbf{v}_1,\mathbf{v}_2,\dots)$.
Solve $\max\ \mathbf{w}^T\Sigma\mathbf{w}$ subject to $\mathbf{w}^T\mathbf{w} = 1$ and
$\mathbf{v}_1^T\mathbf{w} = 0$. Show the solution is $\mathbf{v}_2$ with value $\lambda_2$, and that the
multiplier of the orthogonality constraint is zero.

*Intuition:* Deflation is a second constraint; because the objective is already blind to the removed direction, that constraint is free.

**Solution:**

**Step 1 (Lagrangian with two multipliers).**

$$
\mathcal{L} = \mathbf{w}^T\Sigma\mathbf{w} - \alpha\left(\mathbf{w}^T\mathbf{w}-1\right) - \beta\left(\mathbf{v}_1^T\mathbf{w}\right)
$$

$$
\nabla_{\mathbf{w}}\mathcal{L} = 2\Sigma\mathbf{w} - 2\alpha\mathbf{w} - \beta\mathbf{v}_1 = \mathbf{0}
$$

**Step 2 (LICQ).** The two constraint gradients are $2\mathbf{w}$ and $\mathbf{v}_1$. Feasibility forces
$\mathbf{v}_1^T\mathbf{w} = 0$ with $\lVert \mathbf{w}\rVert = 1$, so they are orthogonal and nonzero,
hence linearly independent: LICQ holds.

**Step 3 (eliminate $\beta$).** Multiply the stationarity equation on the left by $\mathbf{v}_1^T$ and use
$\Sigma\mathbf{v}_1 = \lambda_1\mathbf{v}_1$ (so $\mathbf{v}_1^T\Sigma = \lambda_1\mathbf{v}_1^T$) together
with $\mathbf{v}_1^T\mathbf{w} = 0$:

$$
2\lambda_1\underbrace{\mathbf{v}_1^T\mathbf{w}}_{=0} - 2\alpha\underbrace{\mathbf{v}_1^T\mathbf{w}}_{=0} - \beta\underbrace{\mathbf{v}_1^T\mathbf{v}_1}_{=1} = 0 \ \Longrightarrow\ \beta = 0
$$

**Step 4 (the reduced eigenproblem).** With $\beta = 0$ the condition becomes
$\Sigma\mathbf{w} = \alpha\mathbf{w}$: $\mathbf{w}$ must be a unit eigenvector of $\Sigma$ orthogonal to
$\mathbf{v}_1$, and as in Problem L2.2 the objective equals the multiplier:
$\mathbf{w}^T\Sigma\mathbf{w} = \alpha$. Maximizing over the admissible eigenvalues
$\{\lambda_2,\dots,\lambda_n\}$ gives

$$
\mathbf{w}^* = \mathbf{v}_2, \qquad \alpha^* = \lambda_2
$$

**Step 5 (why $\beta^* = 0$ is meaningful).** By the sensitivity theorem, $\beta^*$ is the rate at which the
optimum would improve if the orthogonality requirement were relaxed to
$\mathbf{v}_1^T\mathbf{w} = \epsilon$. Zero price means: *to first order, tilting slightly back toward
$\mathbf{v}_1$ buys nothing* — because the objective is stationary at $\mathbf{v}_2$ in every direction
that respects the norm. Induction extends this to all $k$: the $k$-th principal component is $\mathbf{v}_k$
with value $\lambda_k$ and zero prices on all $k-1$ orthogonality constraints.

$$
\boxed{\mathbf{w}^* = \mathbf{v}_2, \quad \max = \lambda_2, \quad \alpha^* = \lambda_2, \quad \beta^* = 0}
$$

> **Key takeaway:** PCA is a nested sequence of Lagrange problems, and the vanishing multipliers on the orthogonality constraints are the algebraic statement that principal directions decouple — the reason PCA can be computed greedily, one component at a time.

## Level 3 — Challenge

### Problem L3.1: Proof of the Second-Order Sufficient Condition

**Problem Statement:** Let $f, \mathbf{h}$ be $\mathcal{C}^2$, let $(\mathbf{x}^*,\boldsymbol{\lambda}^*)$
satisfy the first-order conditions with LICQ, and suppose

$$
\mathbf{d}^T\nabla^2_{\mathbf{xx}}\mathcal{L}(\mathbf{x}^*,\boldsymbol{\lambda}^*)\,\mathbf{d} \gt 0 \quad \text{for all } \mathbf{d} \in T(\mathbf{x}^*)\setminus\{\mathbf{0}\}
$$

Prove that $\mathbf{x}^*$ is a strict local minimizer of $f$ on $\mathcal{M}$.

*Intuition:* On the feasible manifold the Lagrangian and the objective agree, so a positive definite restricted Hessian of $\mathcal{L}$ certifies quadratic growth of $f$ along the manifold.

**Solution:**

**Step 1 ($\mathcal{L}$ and $f$ agree on $\mathcal{M}$).** For feasible $\mathbf{x}$,
$\mathbf{h}(\mathbf{x}) = \mathbf{0}$, so

$$
\mathcal{L}(\mathbf{x},\boldsymbol{\lambda}^*) = f(\mathbf{x}) + \boldsymbol{\lambda}^{*T}\mathbf{h}(\mathbf{x}) = f(\mathbf{x})
$$

Hence it suffices to show $\mathcal{L}(\cdot,\boldsymbol{\lambda}^*)$ grows near $\mathbf{x}^*$ *along
feasible points*.

**Step 2 (Taylor expand $\mathcal{L}$).** Since $\nabla_{\mathbf{x}}\mathcal{L}(\mathbf{x}^*,\boldsymbol{\lambda}^*) = \mathbf{0}$,
for feasible $\mathbf{x} = \mathbf{x}^* + \mathbf{e}$,

$$
f(\mathbf{x}) = \mathcal{L}(\mathbf{x},\boldsymbol{\lambda}^*) = f(\mathbf{x}^*) + \frac12\mathbf{e}^T\nabla^2_{\mathbf{xx}}\mathcal{L}(\mathbf{x}^*,\boldsymbol{\lambda}^*)\,\mathbf{e} + o\!\left(\lVert \mathbf{e}\rVert^2\right)
$$

**Step 3 (feasible displacements are nearly tangent).** Write $\mathbf{e} = \mathbf{e}_T + \mathbf{e}_N$
with $\mathbf{e}_T \in T$ and $\mathbf{e}_N \in T^{\perp} = \operatorname{range}(A^T)$,
$A = D\mathbf{h}(\mathbf{x}^*)$. Feasibility and a Taylor expansion of $\mathbf{h}$ give

$$
\mathbf{0} = \mathbf{h}(\mathbf{x}^*+\mathbf{e}) = A\mathbf{e} + O\!\left(\lVert \mathbf{e}\rVert^2\right) = A\mathbf{e}_N + O\!\left(\lVert \mathbf{e}\rVert^2\right)
$$

Under LICQ, $A$ restricted to $T^{\perp}$ is invertible with bounded inverse, so
$\lVert \mathbf{e}_N\rVert = O(\lVert \mathbf{e}\rVert^2)$: the feasible manifold hugs its tangent plane to
second order.

**Step 4 (the quadratic form survives).** Let $\mu \gt 0$ be the smallest eigenvalue of
$\nabla^2_{\mathbf{xx}}\mathcal{L}$ restricted to $T$ (positive by hypothesis, attained on the compact unit
sphere of $T$), and let $M$ bound $\lVert \nabla^2_{\mathbf{xx}}\mathcal{L}\rVert$. Then

$$
\mathbf{e}^T\nabla^2_{\mathbf{xx}}\mathcal{L}\,\mathbf{e} \ge \mu\lVert \mathbf{e}_T\rVert^2 - 2M\lVert \mathbf{e}_T\rVert\lVert \mathbf{e}_N\rVert - M\lVert \mathbf{e}_N\rVert^2 = \mu\lVert \mathbf{e}_T\rVert^2 + O\!\left(\lVert \mathbf{e}\rVert^3\right)
$$

and $\lVert \mathbf{e}_T\rVert^2 = \lVert \mathbf{e}\rVert^2 - \lVert \mathbf{e}_N\rVert^2 = \lVert \mathbf{e}\rVert^2 + O(\lVert \mathbf{e}\rVert^4)$.

**Step 5 (conclude).** Combining Steps 2 and 4, for feasible $\mathbf{x}$ near $\mathbf{x}^*$,

$$
f(\mathbf{x}) \ge f(\mathbf{x}^*) + \frac{\mu}{2}\lVert \mathbf{x}-\mathbf{x}^*\rVert^2 + o\!\left(\lVert \mathbf{x}-\mathbf{x}^*\rVert^2\right) \ge f(\mathbf{x}^*) + \frac{\mu}{4}\lVert \mathbf{x}-\mathbf{x}^*\rVert^2
$$

on a small enough feasible neighbourhood: $\mathbf{x}^*$ is a strict local constrained minimizer with
quadratic growth.

$$
\boxed{\nabla_{\mathbf{x}}\mathcal{L} = \mathbf{0} \ \text{and}\ \nabla^2_{\mathbf{xx}}\mathcal{L}\big\vert_{T} \succ 0 \implies \text{strict local constrained minimum with modulus } \tfrac{\mu}{4}}
$$

> **Key takeaway:** Only the Hessian of the *Lagrangian* works — using $\nabla^2 f$ alone is wrong because the constraint manifold curves, and the term $\sum_j\lambda_j\nabla^2h_j$ is precisely the correction accounting for that curvature.

### Problem L3.2: Why $\nabla^2 f$ Alone Is the Wrong Test

**Problem Statement:** Exhibit a problem where $\nabla^2 f(\mathbf{x}^*)$ is *indefinite* yet $\mathbf{x}^*$
is a strict constrained minimum, and another where $\nabla^2 f(\mathbf{x}^*) \succ 0$ on the tangent space
yet $\mathbf{x}^*$ is *not* a constrained minimum. Explain the role of $\sum_j\lambda_j\nabla^2 h_j$.

*Intuition:* Walking on a curved surface, you feel the curvature of the surface as well as the curvature of the landscape.

**Solution:**

**Example A (indefinite $\nabla^2 f$, genuine constrained minimum).** Take $f(x,y)=xy$ on
$x^2+y^2 = 2$ from Problem L1.1. Then

$$
\nabla^2 f = \begin{bmatrix} 0 & 1 \\ 1 & 0\end{bmatrix}
$$

is indefinite (eigenvalues $\pm 1$) at *every* point, yet $(1,-1)$ is a strict constrained minimum. The
correct object is

$$
\nabla^2_{\mathbf{xx}}\mathcal{L} = \nabla^2 f + \lambda\nabla^2 h = \begin{bmatrix} 2\lambda & 1 \\ 1 & 2\lambda\end{bmatrix} \xrightarrow{\lambda = 1/2} \begin{bmatrix} 1 & 1 \\ 1 & 1\end{bmatrix}
$$

which is positive *definite* on the tangent line $\operatorname{span}\{(1,1)\}$ (value $4 \gt 0$). The
constraint term $\lambda\nabla^2 h = 2\lambda I$ supplied all of the missing curvature.

**Example B (tangentially positive $\nabla^2 f$, not a minimum).** Take

$$
f(x,y) = y - x^2, \qquad h(x,y) = y - 2x^2 = 0
$$

At the origin: $\nabla f = (0,1)$, $\nabla h = (0,1)$, so stationarity $\nabla f + \lambda\nabla h = 0$
gives $\lambda^* = -1$, and the tangent space is $T = \operatorname{span}\{(1,0)\}$. Now

$$
\nabla^2 f = \begin{bmatrix} -2 & 0 \\ 0 & 0\end{bmatrix}, \qquad \nabla^2 h = \begin{bmatrix} -4 & 0 \\ 0 & 0\end{bmatrix}, \qquad \nabla^2_{\mathbf{xx}}\mathcal{L} = \begin{bmatrix} -2 + 4 & 0 \\ 0 & 0\end{bmatrix} = \begin{bmatrix} 2 & 0\\ 0 & 0\end{bmatrix}
$$

The *Lagrangian* Hessian is positive on $T$ ($\mathbf{d}=(1,0)$ gives $2 \gt 0$), correctly certifying a
minimum; and indeed along the feasible parabola $y = 2x^2$,

$$
f(x, 2x^2) = 2x^2 - x^2 = x^2 \ge 0
$$

a strict minimum at the origin. Meanwhile $\nabla^2 f$ restricted to $T$ gives $-2 \lt 0$, which would
have (wrongly) predicted a maximum. So using $\nabla^2 f$ alone gives the *opposite* verdict.

**The mechanism.** Along a feasible curve $\mathbf{x}(t)$ with $\mathbf{x}(0)=\mathbf{x}^*$,
$\dot{\mathbf{x}}(0)=\mathbf{d}$,

$$
\frac{d^2}{dt^2}f(\mathbf{x}(t))\bigg\vert_{0} = \mathbf{d}^T\nabla^2f\,\mathbf{d} + \nabla f^T\ddot{\mathbf{x}}(0)
$$

The acceleration $\ddot{\mathbf{x}}(0)$ is not free: differentiating $\mathbf{h}(\mathbf{x}(t))=\mathbf{0}$
twice gives $\nabla h_j^T\ddot{\mathbf{x}}(0) = -\mathbf{d}^T\nabla^2h_j\mathbf{d}$. Substituting
$\nabla f = -\sum_j\lambda_j\nabla h_j$ eliminates $\ddot{\mathbf{x}}$ entirely and leaves

$$
\frac{d^2}{dt^2}f\bigg\vert_0 = \mathbf{d}^T\left(\nabla^2 f + \sum_j\lambda_j\nabla^2h_j\right)\mathbf{d} = \mathbf{d}^T\nabla^2_{\mathbf{xx}}\mathcal{L}\,\mathbf{d}
$$

$$
\boxed{\frac{d^2}{dt^2}f(\mathbf{x}(t))\Big\vert_{0} = \mathbf{d}^T\nabla^2_{\mathbf{xx}}\mathcal{L}\,\mathbf{d} \ne \mathbf{d}^T\nabla^2 f\,\mathbf{d} \ \text{in general}}
$$

> **Key takeaway:** $\sum_j\lambda_j\nabla^2h_j$ is the *centripetal* correction: it accounts for the fact that staying feasible requires accelerating, and the objective's gradient does work against that acceleration.

### Problem L3.3: Fritz John Conditions and Abnormal Multipliers

**Problem Statement:** Consider $\min f(x,y) = x$ subject to $h(x,y) = y^2 - x^3 = 0$. Show that the
minimizer is the origin, that no Lagrange multiplier exists, and that the Fritz John conditions
$\mu_0\nabla f + \lambda\nabla h = \mathbf{0}$, $(\mu_0,\lambda)\neq(0,0)$, $\mu_0 \ge 0$ *do* hold — with
$\mu_0 = 0$. Explain what an abnormal multiplier means.

*Intuition:* A cusp has no well-defined tangent line, so the linearized picture cannot see the constraint at all.

**Solution:**

**Step 1 (the feasible set and the minimizer).** $y^2 = x^3$ requires $x \ge 0$, and every $x \ge 0$ is
attained (take $y = \pm x^{3/2}$). So the feasible set is a cusp curve with tip at the origin, and

$$
\min_{\text{feasible}} x = 0 \quad \text{at} \quad (x^*,y^*) = (0,0)
$$

**Step 2 (no multiplier).**

$$
\nabla f = \begin{pmatrix} 1 \\ 0\end{pmatrix}, \qquad \nabla h(x,y) = \begin{pmatrix} -3x^2 \\ 2y\end{pmatrix}, \qquad \nabla h(0,0) = \begin{pmatrix} 0\\0\end{pmatrix}
$$

The Lagrange condition $\nabla f + \lambda\nabla h = \mathbf{0}$ reads $(1,0)^T = \mathbf{0}$: impossible
for every $\lambda$. LICQ fails because the single constraint gradient is the zero vector.

**Step 3 (the linearization is a lie).** The linearized tangent set is
$\{\mathbf{d} : \nabla h(0,0)^T\mathbf{d} = 0\} = \mathbb{R}^2$ — two-dimensional. The true set of feasible
velocities at the cusp tip is only $\{(d_1, 0) : d_1 \ge 0\}$, computed from the parametrization
$(t^2, t^3)$ whose velocity at $t=0$ is $\mathbf{0}$. The linearization has invented directions that do not
exist, so the argument of Problem L1.3 breaks at Step 1.

**Step 4 (Fritz John holds).** The Fritz John (John, 1948) conditions weaken stationarity to

$$
\mu_0\nabla f(\mathbf{x}^*) + \lambda\nabla h(\mathbf{x}^*) = \mathbf{0}, \qquad (\mu_0,\lambda)\neq(0,0), \quad \mu_0\ge 0
$$

Here $\mu_0 = 0$, $\lambda = 1$ satisfies it, since $\nabla h(0,0) = \mathbf{0}$:

$$
0\cdot\begin{pmatrix}1\\0\end{pmatrix} + 1\cdot\begin{pmatrix}0\\0\end{pmatrix} = \mathbf{0} \quad\checkmark
$$

**Step 5 (what "abnormal" means).** A multiplier set with $\mu_0 = 0$ is called **abnormal**: the objective
has been given zero weight, so the condition says something about the *constraints alone* (they are
degenerate at $\mathbf{x}^*$) and nothing about $f$. Any objective whatsoever would satisfy the same
condition, which is why abnormal multipliers carry no optimization information. Constraint qualifications
are exactly the hypotheses that force $\mu_0 \gt 0$; dividing through by $\mu_0$ then recovers ordinary
Lagrange multipliers $\lambda/\mu_0$.

$$
\boxed{\text{no } \lambda \text{ with } \nabla f + \lambda\nabla h = \mathbf{0}; \ \text{Fritz John holds with } (\mu_0,\lambda) = (0,1) — \text{an abnormal multiplier}}
$$

> **Key takeaway:** Fritz John conditions are always true and sometimes vacuous; a constraint qualification is precisely the extra hypothesis that upgrades them from "always true" to "informative" by guaranteeing $\mu_0 \neq 0$.

### Problem L3.4: The Maximum-Entropy Distribution with Fixed Mean and Variance Is Gaussian

**Problem Statement:** Among all probability densities $p$ on $\mathbb{R}$ with mean $\mu$ and variance
$\sigma^2$, maximize the differential entropy $H(p) = -\int p\ln p\,dx$. Solve with three multipliers and
identify the maximizer. Then verify optimality with a Gibbs-inequality argument that requires no calculus
of variations.

*Intuition:* Three linear constraints (normalization, mean, second moment) produce an exponential of a quadratic — which is a Gaussian.

**Solution:**

**Step 1 (variational Lagrangian).**

$$
\mathcal{L}[p] = -\int p\ln p\,dx - \alpha\left(\int p\,dx - 1\right) - \beta\left(\int x\,p\,dx - \mu\right) - \gamma\left(\int x^2 p\,dx - (\mu^2+\sigma^2)\right)
$$

Setting the functional derivative to zero (the pointwise Euler-Lagrange condition):

$$
\frac{\delta\mathcal{L}}{\delta p(x)} = -\ln p(x) - 1 - \alpha - \beta x - \gamma x^2 = 0
$$

**Step 2 (read off the form).**

$$
p(x) = \exp\left(-1-\alpha - \beta x - \gamma x^2\right)
$$

an exponential of a quadratic. For $p$ to be integrable we need $\gamma \gt 0$; completing the square,

$$
p(x) = C\exp\left(-\gamma\left(x + \frac{\beta}{2\gamma}\right)^2\right)
$$

which is a Gaussian density with mean $-\beta/(2\gamma)$ and variance $1/(2\gamma)$.

**Step 3 (fit the multipliers to the constraints).** Matching moments,

$$
-\frac{\beta}{2\gamma} = \mu, \qquad \frac{1}{2\gamma} = \sigma^2 \ \Longrightarrow\ \gamma = \frac{1}{2\sigma^2}, \quad \beta = -\frac{\mu}{\sigma^2}
$$

and normalization fixes $C = 1/\sqrt{2\pi\sigma^2}$, i.e. $1 + \alpha = \frac{\mu^2}{2\sigma^2} + \ln\sqrt{2\pi\sigma^2}$.
So

$$
p^*(x) = \frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right) = \mathcal{N}(\mu,\sigma^2)
$$

with maximum entropy

$$
H^* = \frac12\ln\left(2\pi e\,\sigma^2\right)
$$

**Step 4 (rigorous verification via Gibbs' inequality).** The variational argument produces a *candidate*;
here is a proof it is optimal, with no functional analysis. For any admissible $p$ (same mean and
variance), the Kullback-Leibler divergence is nonnegative:

$$
0 \le D_{\mathrm{KL}}(p \Vert p^*) = \int p\ln\frac{p}{p^*}\,dx = -H(p) - \int p\ln p^*\,dx
$$

Now $\ln p^*(x) = -\frac{(x-\mu)^2}{2\sigma^2} - \frac12\ln(2\pi\sigma^2)$ is a quadratic polynomial, and
its expectation depends on $p$ *only through the constrained moments*:

$$
-\int p\ln p^*\,dx = \frac{\mathbb{E}_p\left[(x-\mu)^2\right]}{2\sigma^2} + \frac12\ln(2\pi\sigma^2) = \frac12 + \frac12\ln(2\pi\sigma^2) = H^*
$$

Therefore $0 \le -H(p) + H^*$, i.e. $H(p) \le H^*$ for every admissible $p$, with equality iff $p = p^*$
almost everywhere.

$$
\boxed{p^* = \mathcal{N}(\mu,\sigma^2), \qquad H^* = \tfrac12\ln\left(2\pi e\sigma^2\right), \qquad \gamma^* = \tfrac{1}{2\sigma^2}, \ \beta^* = -\tfrac{\mu}{\sigma^2}}
$$

> **Key takeaway:** Multipliers on moment constraints are the *natural parameters* of an exponential family: normalization gives the log-partition function, the mean constraint gives $\mu/\sigma^2$, and the second-moment constraint gives $1/(2\sigma^2)$ — precisely the canonical parameters of the Gaussian.
