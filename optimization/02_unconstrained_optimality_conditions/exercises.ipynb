### Problem L0.1 — Stationarity Is Necessary, Never Sufficient

**Problem Statement:** State and prove Fermat's first-order necessary condition: if $f$ is differentiable
and $\mathbf{x}^*$ is a local minimizer, then $\nabla f(\mathbf{x}^*) = \mathbf{0}$. Then exhibit two
stationary points that are *not* minimizers, one in one dimension and one in two.

*Intuition:* A nonzero gradient always offers an escape route downhill; a zero gradient merely closes that route to first order.

**Solution:**

**Step 1 (proof of necessity).** Suppose $\nabla f(\mathbf{x}^*) \neq \mathbf{0}$ and set
$\mathbf{d} = -\nabla f(\mathbf{x}^*)$. Define the one-dimensional restriction
$\phi(t) = f(\mathbf{x}^* + t\mathbf{d})$. By the chain rule,

$$
\phi'(0) = \nabla f(\mathbf{x}^*)^\top \mathbf{d} = -\lVert \nabla f(\mathbf{x}^*)\rVert^2 \lt 0
$$

By the definition of the derivative there is $\bar{t} \gt 0$ with
$\frac{\phi(t) - \phi(0)}{t} \lt \frac{\phi'(0)}{2} \lt 0$ for all $t \in (0, \bar{t})$, hence
$f(\mathbf{x}^* + t\mathbf{d}) \lt f(\mathbf{x}^*)$ for arbitrarily small $t \gt 0$. Every neighbourhood
of $\mathbf{x}^*$ therefore contains a strictly better point, contradicting local minimality. So
$\nabla f(\mathbf{x}^*) = \mathbf{0}$.

**Step 2 (a stationary non-extremum in $\mathbb{R}$).** $f(x) = x^3$ has $f'(0) = 0$, yet
$f(-\epsilon) = -\epsilon^3 \lt 0 \lt \epsilon^3 = f(\epsilon)$: the origin is an inflection, neither a
minimum nor a maximum.

**Step 3 (a saddle in $\mathbb{R}^2$).** $f(x,y) = x^2 - y^2$ has $\nabla f = (2x, -2y)$, so the origin is
stationary. Along the $x$-axis $f(x,0) = x^2 \gt 0$ (a minimum), along the $y$-axis
$f(0,y) = -y^2 \lt 0$ (a maximum). Both behaviours occur in every neighbourhood of the origin, so the
origin is a saddle.

$$
\boxed{\nabla f(\mathbf{x}^*) = \mathbf{0} \text{ is necessary for a local minimum, and strictly weaker than minimality}}
$$

> **Key takeaway:** FONC is a *filter*, not a verdict: it produces a shortlist of candidates — minima, maxima, saddles and degenerate points alike — that only second- (or higher-) order information can separate.

```python
xs = np.array([-1e-3, 1e-3])
print("f(x) = x^3:   f'(0) =", 3 * 0.0**2,
      "   f(-1e-3) =", f"{xs[0] ** 3:+.1e}", "  f(+1e-3) =", f"{xs[1] ** 3:+.1e}")
assert xs[0] ** 3 < 0 < xs[1] ** 3

f_saddle = lambda x, y: x**2 - y**2
print("f(x,y) = x^2 - y^2:   grad(0,0) =", np.array([0.0, 0.0]),
      "   f(1e-3,0) =", f"{f_saddle(1e-3, 0.0):+.1e}", "  f(0,1e-3) =", f"{f_saddle(0.0, 1e-3):+.1e}")
assert f_saddle(1e-3, 0.0) > 0 > f_saddle(0.0, 1e-3)
print("both points are stationary and neither is an extremum: FONC is necessary only")
```

### Problem L0.2 — The Gap Between SONC and SOSC

**Problem Statement:** State the second-order necessary condition (SONC) and the second-order sufficient
condition (SOSC) at a stationary point. Then show that the gap between them is real by analysing
$f(x, y) = x^2 + y^3$ at the origin.

*Intuition:* A positive semidefinite Hessian leaves flat directions in which the cubic term, invisible to the second-order test, decides the outcome.

**Solution:**

**The two conditions.** Let $f \in \mathcal{C}^2$ and $\nabla f(\mathbf{x}^*) = \mathbf{0}$.

- **SONC:** if $\mathbf{x}^*$ is a local minimizer then $\nabla^2 f(\mathbf{x}^*) \succeq 0$.
- **SOSC:** if $\nabla^2 f(\mathbf{x}^*) \succ 0$ then $\mathbf{x}^*$ is a *strict* local minimizer.

The gap is the semidefinite-but-singular case $\nabla^2 f(\mathbf{x}^*) \succeq 0$ with
$\lambda_{\min} = 0$, where neither statement applies.

**Step 1 (the example is stationary).** For $f(x,y) = x^2 + y^3$,

$$
\nabla f(x,y) = \begin{pmatrix} 2x \\ 3y^2 \end{pmatrix}, \qquad \nabla f(0,0) = \mathbf{0}
$$

**Step 2 (the Hessian is PSD but singular).**

$$
\nabla^2 f(x,y) = \begin{bmatrix} 2 & 0 \\ 0 & 6y \end{bmatrix}, \qquad \nabla^2 f(0,0) = \begin{bmatrix} 2 & 0 \\ 0 & 0 \end{bmatrix} \succeq 0
$$

so SONC is satisfied and SOSC is not applicable ($\lambda_{\min} = 0$).

**Step 3 (the origin is not a minimizer).** Along the flat direction $x = 0$ we get $f(0,y) = y^3$, which
is strictly negative for every $y \lt 0$. Points $(0, -\epsilon)$ lie arbitrarily close to the origin with
$f = -\epsilon^3 \lt 0 = f(0,0)$.

$$
\boxed{\nabla^2 f(\mathbf{0}) \succeq 0 \text{ yet } \mathbf{0} \text{ is not a local minimum: SONC is necessary, not sufficient}}
$$

> **Key takeaway:** A zero eigenvalue makes the second-order test *silent*, not affirmative; along the corresponding eigenvector the fate of the point is decided by the first non-vanishing higher-order term.

```python
H = np.array([[2.0, 0.0], [0.0, 0.0]])
ys = -np.array([1e-1, 1e-2, 1e-3])
vals = ys**3
print("eig(Hessian at the origin) =", np.linalg.eigvalsh(H), " -> PSD and singular: SONC holds, SOSC silent")
print("f(0,y) at y =", "  ".join(f"{y:+.0e}" for y in ys), " -> ", "  ".join(f"{v:+.1e}" for v in vals))
assert np.all(np.linalg.eigvalsh(H) >= 0) and np.all(vals < 0.0)
print("points with f < f(0,0) = 0 sit arbitrarily close to the origin: not a local minimum")
```

### Problem L0.3 — The Determinant Is Not the Signature

**Problem Statement:** A student proposes to classify a stationary point in $\mathbb{R}^n$ by the sign of
$\det \nabla^2 f(\mathbf{x}^*)$ alone. Explain why this fails for $n \ge 2$, and state the correct
classification rule together with the finite test (Sylvester's criterion) that implements it.

*Intuition:* The determinant is the product of the eigenvalues — an even number of sign flips is invisible to it.

**Solution:**

**Step 1 (why the determinant is blind).** For a symmetric $H$ with eigenvalues $\lambda_1, \dots, \lambda_n$,

$$
\det H = \prod_{i=1}^n \lambda_i
$$

Any *even* number of negative eigenvalues leaves the product positive. Concretely, at the maximum of
$f(x,y) = -x^2 - y^2$ we have $H = \operatorname{diag}(-2,-2)$ with $\det H = 4 \gt 0$ — the same sign as
at the minimum of $x^2 + y^2$. In $\mathbb{R}^4$, $H = \operatorname{diag}(1, 1, -1, -1)$ is a saddle with
$\det H = 1 \gt 0$. And $\det H = 0$ says only that *some* eigenvalue vanishes, giving no information at
all about the rest of the spectrum.

**Step 2 (the correct rule).** Classification at a stationary point uses the whole eigenvalue signature
$(n_+, n_-, n_0)$ — the numbers of positive, negative and zero eigenvalues:

- $n_- = n_0 = 0$ (all positive): strict local **minimum** (SOSC).
- $n_+ = n_0 = 0$ (all negative): strict local **maximum**.
- $n_+ \gt 0$ and $n_- \gt 0$: **saddle point**.
- $n_0 \gt 0$ with no negative eigenvalue: **degenerate**, second-order test inconclusive.

**Step 3 (a finite test).** Sylvester's criterion: $H \succ 0$ if and only if all $n$ leading principal
minors $D_1, \dots, D_n$ are strictly positive; $H \prec 0$ if and only if the leading minors alternate,
$(-1)^k D_k \gt 0$. For $n = 2$ this reduces to the familiar rule with $D_1 = f_{xx}$ and
$D_2 = f_{xx}f_{yy} - f_{xy}^2$:

$$
\boxed{D_2 \gt 0,\ f_{xx} \gt 0 \Rightarrow \text{min}; \quad D_2 \gt 0,\ f_{xx} \lt 0 \Rightarrow \text{max}; \quad D_2 \lt 0 \Rightarrow \text{saddle}; \quad D_2 = 0 \Rightarrow \text{inconclusive}}
$$

> **Key takeaway:** Classification is a statement about the *spectrum*; any single scalar summary (determinant, trace) throws away exactly the information needed to distinguish a bowl from a mountain pass.

```python
cases = {
    "diag(2, 2)  minimum": np.diag([2.0, 2.0]),
    "diag(-2, -2)  maximum": np.diag([-2.0, -2.0]),
    "diag(1, 1, -1, -1)  saddle": np.diag([1.0, 1.0, -1.0, -1.0]),
}
for name, H in cases.items():
    print(f"{name:28s} det = {np.linalg.det(H):+7.1f}   eigenvalues = {np.linalg.eigvalsh(H)}")
assert np.linalg.det(cases["diag(-2, -2)  maximum"]) > 0
assert np.linalg.det(cases["diag(1, 1, -1, -1)  saddle"]) > 0
print("a positive determinant occurs at a minimum, a maximum and a saddle: det classifies nothing")
```

### Problem L0.4 — Bounded Below Does Not Mean Attained

**Problem Statement:** Give a continuous function on $\mathbb{R}$ that is bounded below but has no
minimizer. Then state the coercivity–Weierstrass existence theorem and explain which hypothesis your
example violates.

*Intuition:* An infimum is a number; a minimizer is a point. Compactness is what turns one into the other.

**Solution:**

**Step 1 (the counterexample).** $f(x) = e^x$ is continuous, and $f(x) \gt 0$ for all $x$, so
$\inf_{\mathbb{R}} f = 0$. But $e^x = 0$ has no solution, so the infimum is never attained: there is no
minimizer. (Equally good: $f(x) = 1/(1+x^2)$, or $f(x) = \arctan x$.)

**Step 2 (the existence theorem).** Call $f$ **coercive** if $f(\mathbf{x}) \to +\infty$ whenever
$\lVert \mathbf{x}\rVert \to \infty$. Then:

> If $f: \mathbb{R}^n \to \mathbb{R}$ is continuous and coercive, $f$ attains a global minimum.

*Proof sketch.* Fix any $\mathbf{x}_0$ and let $c = f(\mathbf{x}_0)$. Coercivity gives $R \gt 0$ with
$f(\mathbf{x}) \gt c$ whenever $\lVert \mathbf{x}\rVert \gt R$, so the sublevel set
$S_c = \{\mathbf{x} : f(\mathbf{x}) \le c\}$ is contained in the closed ball of radius $R$; it is closed
by continuity, hence compact and nonempty. Weierstrass gives a minimizer of $f$ on $S_c$, and since every
point outside $S_c$ has value $\gt c$, that point minimizes $f$ on all of $\mathbb{R}^n$. $\blacksquare$

**Step 3 (which hypothesis fails).** $e^x$ is continuous but **not coercive**: as $x \to -\infty$ the
value tends to $0$, not to $+\infty$. Its sublevel sets $\{x : e^x \le c\} = (-\infty, \ln c]$ are closed
but unbounded, so no compactness is available and the minimizing sequence $x_k = -k$ escapes to infinity.

$$
\boxed{\text{continuous} + \text{coercive} \implies \text{global minimizer exists}; \quad e^x \text{ shows coercivity cannot be dropped}}
$$

> **Key takeaway:** Existence is a topological statement (compact sublevel sets), entirely separate from the calculus statements FONC/SONC/SOSC, which only *recognize* a minimizer once one is known to exist.

```python
xs = -np.array([10.0, 50.0, 150.0, 300.0])
print("f(x) = e^x at x =", xs, "->", "  ".join(f"{np.exp(x):.2e}" for x in xs))
assert np.all(np.exp(xs) > 0.0) and np.exp(xs[-1]) < 1e-130
print("inf f = 0, yet e^x > 0 for every real x: the infimum is approached and never attained")
```

## L1 — Foundations

### Problem L1.1 — Full Classification of $f(x,y) = x^3 + y^3 - 3xy$

**Problem Statement:** Find all stationary points of $f(x,y) = x^3 + y^3 - 3xy$ and classify each one
using the Hessian. Is $f$ bounded below?

*Intuition:* Cubic surfaces typically produce one genuine bowl and one pass; unboundedness is settled by looking along a single ray.

**Solution:**

**Step 1 (stationary points).**

$$
\nabla f(x,y) = \begin{pmatrix} 3x^2 - 3y \\ 3y^2 - 3x \end{pmatrix} = \mathbf{0} \iff y = x^2 \ \text{and}\ x = y^2
$$

Substituting, $x = (x^2)^2 = x^4$, i.e. $x(x^3 - 1) = 0$, so $x = 0$ or $x = 1$. The stationary points are

$$
(0,0) \quad \text{and} \quad (1,1)
$$

**Step 2 (Hessian).**

$$
\nabla^2 f(x,y) = \begin{bmatrix} 6x & -3 \\ -3 & 6y \end{bmatrix}
$$

**Step 3 (classify $(0,0)$).** $\nabla^2 f(0,0) = \begin{bmatrix} 0 & -3 \\ -3 & 0 \end{bmatrix}$ has
$\det = -9 \lt 0$, so the eigenvalues $\pm 3$ have opposite signs: a **saddle point**. Directly:
$f(t,t) = 2t^3 - 3t^2 \lt 0$ for small $t \gt 0$, while $f(t,-t) = t^3 - t^3 + 3t^2 = 3t^2 \gt 0$.

**Step 4 (classify $(1,1)$).** $\nabla^2 f(1,1) = \begin{bmatrix} 6 & -3 \\ -3 & 6 \end{bmatrix}$ has
leading minors $D_1 = 6 \gt 0$ and $D_2 = 36 - 9 = 27 \gt 0$, so by Sylvester it is positive definite
(eigenvalues $3$ and $9$): a **strict local minimum** with $f(1,1) = 1 + 1 - 3 = -1$.

**Step 5 (boundedness).** Along the ray $y = 0$, $f(x,0) = x^3 \to -\infty$ as $x \to -\infty$. So $f$ is
**not** bounded below, $f$ is not coercive, and the local minimum at $(1,1)$ is *not* global.

$$
\boxed{(0,0)\ \text{saddle},\quad (1,1)\ \text{strict local min with } f = -1, \quad \inf_{\mathbb{R}^2} f = -\infty}
$$

> **Key takeaway:** Local optimality conditions are strictly local — a certified strict local minimum coexists happily with an objective that is unbounded below, which is exactly why existence needs a separate (coercivity) argument.

```python
f = lambda x, y: x**3 + y**3 - 3.0 * x * y
grad = lambda x, y: np.array([3.0 * x**2 - 3.0 * y, 3.0 * y**2 - 3.0 * x])
hess = lambda x, y: np.array([[6.0 * x, -3.0], [-3.0, 6.0 * y]])

for pt in [(0.0, 0.0), (1.0, 1.0)]:
    ev = np.linalg.eigvalsh(hess(*pt))
    print(f"{str(pt):10s} grad = {grad(*pt)}   eig = {ev}   f = {f(*pt):+.1f}")
    assert np.allclose(grad(*pt), 0.0, atol=1e-14)
assert np.allclose(np.linalg.eigvalsh(hess(0.0, 0.0)), [-3.0, 3.0])
assert np.allclose(np.linalg.eigvalsh(hess(1.0, 1.0)), [3.0, 9.0])
assert abs(f(1.0, 1.0) + 1.0) < 1e-14
print("f(x,0) = x^3 at x = -10, -100:", f(-10.0, 0.0), f(-100.0, 0.0), "-> unbounded below")
```

### Problem L1.2 — Proving the Second-Order Necessary Condition

**Problem Statement:** Let $f \in \mathcal{C}^2$ and let $\mathbf{x}^*$ be a local minimizer. Prove
$\nabla^2 f(\mathbf{x}^*) \succeq 0$ directly from Taylor's theorem.

*Intuition:* If some direction had negative curvature, the quadratic term would eventually beat the remainder and pull the function below its value at $\mathbf{x}^*$.

**Solution:**

**Step 1 (setup).** By FONC (Problem L0.1), $\nabla f(\mathbf{x}^*) = \mathbf{0}$. Taylor's theorem with
Peano remainder gives, for any fixed direction $\mathbf{d}$ and $t \to 0$,

$$
f(\mathbf{x}^* + t\mathbf{d}) = f(\mathbf{x}^*) + \frac{t^2}{2}\,\mathbf{d}^\top \nabla^2 f(\mathbf{x}^*)\,\mathbf{d} + o(t^2)
$$

**Step 2 (contradiction hypothesis).** Suppose $\nabla^2 f(\mathbf{x}^*) \not\succeq 0$: there exists
$\mathbf{d}$ with

$$
q := \mathbf{d}^\top \nabla^2 f(\mathbf{x}^*)\,\mathbf{d} \lt 0
$$

**Step 3 (the negative direction wins).** Rearranging the expansion,

$$
\frac{f(\mathbf{x}^* + t\mathbf{d}) - f(\mathbf{x}^*)}{t^2} = \frac{q}{2} + \frac{o(t^2)}{t^2} \xrightarrow[t \to 0]{} \frac{q}{2} \lt 0
$$

So there is $\bar{t} \gt 0$ such that for all $t \in (0, \bar{t})$ the left side is negative, i.e.

$$
f(\mathbf{x}^* + t\mathbf{d}) \lt f(\mathbf{x}^*)
$$

**Step 4 (conclusion).** These points lie in every neighbourhood of $\mathbf{x}^*$ once $t$ is small
enough, contradicting local minimality. Hence no such $\mathbf{d}$ exists and

$$
\boxed{\mathbf{d}^\top \nabla^2 f(\mathbf{x}^*)\,\mathbf{d} \ge 0 \ \text{for all } \mathbf{d}, \quad \text{i.e.} \quad \nabla^2 f(\mathbf{x}^*) \succeq 0}
$$

> **Key takeaway:** The proof is a competition of orders: on the scale $t^2$ the quadratic form is $O(1)$ while the remainder is $o(1)$, so any strictly negative curvature is guaranteed to dominate near enough to $\mathbf{x}^*$.

### Problem L1.3 — Proving the Second-Order Sufficient Condition

**Problem Statement:** Let $f \in \mathcal{C}^2$, $\nabla f(\mathbf{x}^*) = \mathbf{0}$ and
$\nabla^2 f(\mathbf{x}^*) \succ 0$ with smallest eigenvalue $\mu \gt 0$. Prove that $\mathbf{x}^*$ is a
strict local minimizer, and quantify the local quadratic growth.

*Intuition:* Positive definiteness gives curvature $\mu$ to spend; continuity of the Hessian guarantees a ball on which the remainder can only eat half of it.

**Solution:**

**Step 1 (uniform curvature nearby).** Since $\nabla^2 f$ is continuous and
$\lambda_{\min}(\nabla^2 f(\mathbf{x}^*)) = \mu \gt 0$, and since $\lambda_{\min}$ is a continuous
(indeed $1$-Lipschitz in the spectral norm) function of a symmetric matrix, there is $r \gt 0$ with

$$
\lambda_{\min}\left(\nabla^2 f(\mathbf{z})\right) \ge \frac{\mu}{2} \qquad \text{for all } \mathbf{z} \in B(\mathbf{x}^*, r)
$$

**Step 2 (Taylor with Lagrange remainder).** For $\mathbf{x} \in B(\mathbf{x}^*, r)$ there is a point
$\mathbf{z}$ on the segment $[\mathbf{x}^*, \mathbf{x}] \subset B(\mathbf{x}^*, r)$ with

$$
f(\mathbf{x}) = f(\mathbf{x}^*) + \underbrace{\nabla f(\mathbf{x}^*)^\top(\mathbf{x} - \mathbf{x}^*)}_{= 0} + \frac{1}{2}(\mathbf{x} - \mathbf{x}^*)^\top \nabla^2 f(\mathbf{z}) (\mathbf{x} - \mathbf{x}^*)
$$

**Step 3 (bound the quadratic form).** Using Step 1, for every $\mathbf{v}$ we have
$\mathbf{v}^\top \nabla^2 f(\mathbf{z})\mathbf{v} \ge \frac{\mu}{2}\lVert \mathbf{v}\rVert^2$, so

$$
f(\mathbf{x}) \ge f(\mathbf{x}^*) + \frac{\mu}{4}\lVert \mathbf{x} - \mathbf{x}^*\rVert^2
$$

**Step 4 (conclusion).** The right side exceeds $f(\mathbf{x}^*)$ for every
$\mathbf{x} \neq \mathbf{x}^*$ in the ball, so $\mathbf{x}^*$ is a *strict* local minimizer, and the
inequality is a **quadratic growth** (local strong convexity) estimate.

$$
\boxed{f(\mathbf{x}) \ge f(\mathbf{x}^*) + \frac{\mu}{4}\lVert \mathbf{x} - \mathbf{x}^*\rVert^2 \quad \text{on } B(\mathbf{x}^*, r)}
$$

> **Key takeaway:** SOSC delivers more than a label: it certifies quadratic growth with an explicit modulus, which is precisely the hypothesis that gives Newton's method its quadratic local rate and gradient descent its local linear rate.

### Problem L1.4 — Definiteness of a $3 \times 3$ Hessian by Sylvester

**Problem Statement:** Decide whether

$$
H = \begin{bmatrix} 2 & -1 & 0 \\ -1 & 2 & -1 \\ 0 & -1 & 2 \end{bmatrix}
$$

is positive definite, and classify the stationary point of a $\mathcal{C}^2$ function whose Hessian there
equals $H$. Then repeat for $H' = H - 3I$.

*Intuition:* Leading principal minors give a determinant-only certificate of definiteness — no eigenvalues required.

**Solution:**

**Step 1 (leading principal minors of $H$).**

$$
D_1 = 2, \qquad D_2 = \det\begin{bmatrix} 2 & -1 \\ -1 & 2\end{bmatrix} = 3, \qquad D_3 = \det H = 2(4 - 1) - (-1)(-2 - 0) = 6 - 2 = 4
$$

All three are strictly positive, so by Sylvester's criterion $H \succ 0$. Hence the stationary point is a
**strict local minimum** by SOSC.

**Step 2 (cross-check with eigenvalues).** $H$ is the standard second-difference (discrete Laplacian)
matrix, whose eigenvalues are known in closed form:

$$
\lambda_k = 2 - 2\cos\frac{k\pi}{4} = 2 - \sqrt{2},\; 2,\; 2 + \sqrt{2}, \qquad k = 1,2,3
$$

all positive, with product $(2-\sqrt{2})\cdot 2 \cdot (2+\sqrt{2}) = 2(4-2) = 4 = D_3$: consistent.

**Step 3 (the shifted matrix).** $H' = H - 3I$ has eigenvalues $\lambda_k - 3$, namely

$$
-1 - \sqrt{2} \approx -2.414, \qquad -1, \qquad \sqrt{2} - 1 \approx 0.414
$$

Two negative and one positive: $H'$ is **indefinite**, so the corresponding stationary point is a
**saddle** with index $2$ (two descent directions). Note $\det H' = (-1-\sqrt{2})(-1)(\sqrt{2}-1) = (1+\sqrt{2})(\sqrt{2}-1) = 1 \gt 0$: a positive
determinant at a saddle, exactly the trap of Problem L0.3.

$$
\boxed{H \succ 0 \Rightarrow \text{strict local min}; \quad H - 3I \text{ indefinite} \Rightarrow \text{saddle of index } 2}
$$

> **Key takeaway:** Sylvester turns definiteness into $n$ determinant evaluations, but it certifies only the *positive* definite (or, with alternating signs, negative definite) case — indefiniteness still requires looking at the actual signature.

```python
H = np.array([[2.0, -1.0, 0.0], [-1.0, 2.0, -1.0], [0.0, -1.0, 2.0]])
minors = np.array([np.linalg.det(H[:k, :k]) for k in (1, 2, 3)])
ev = np.linalg.eigvalsh(H)
print("leading principal minors:", np.round(minors, 10), "  (hand: 2, 3, 4)")
print("eigenvalues:", ev, "  hand: 2 - sqrt2, 2, 2 + sqrt2 =",
      np.array([2 - np.sqrt(2), 2.0, 2 + np.sqrt(2)]))
assert np.all(minors > 0) and np.all(ev > 0)
assert np.allclose(ev, [2 - np.sqrt(2), 2.0, 2 + np.sqrt(2)])
assert abs(minors[2] - 4.0) < 1e-12

Hp = H - 3.0 * np.eye(3)
evp = np.linalg.eigvalsh(Hp)
print("eig(H - 3I) =", evp, "   det(H - 3I) =", np.linalg.det(Hp), " (hand: 1)")
assert int((evp < 0).sum()) == 2 and int((evp > 0).sum()) == 1
assert abs(np.linalg.det(Hp) - 1.0) < 1e-12
print("two negative eigenvalues with determinant +1: an index-2 saddle, positive determinant")
```

### Problem L1.5 — Coercivity Guarantees a Global Minimizer

**Problem Statement:** Show that $f(x,y) = x^4 + y^4 - 4xy$ is coercive, conclude that a global minimizer
exists, then find all stationary points and identify the global minimizers.

*Intuition:* Quartic growth eventually swamps any quadratic cross term, so the surface is a bowl at large radius no matter how it behaves near the origin.

**Solution:**

**Step 1 (coercivity).** Use $2\lvert xy\rvert \le x^2 + y^2$, so $-4xy \ge -2(x^2+y^2)$ and

$$
f(x,y) \ge x^4 + y^4 - 2(x^2 + y^2)
$$

Each scalar term satisfies $t^4 - 2t^2 = (t^2-1)^2 - 1 \ge \frac{t^4}{2} - 2$ (since
$\frac{t^4}{2} - 2t^2 + 2 = \frac{1}{2}(t^2-2)^2 \ge 0$). Hence

$$
f(x,y) \ge \frac{x^4 + y^4}{2} - 4 \to +\infty \quad \text{as } \lVert (x,y)\rVert \to \infty
$$

so $f$ is coercive and continuous; by the theorem of Problem L0.4 a **global minimizer exists**.

**Step 2 (stationary points).**

$$
\nabla f = \begin{pmatrix} 4x^3 - 4y \\ 4y^3 - 4x \end{pmatrix} = \mathbf{0} \iff y = x^3 \ \text{and}\ x = y^3
$$

Then $x = x^9$, i.e. $x(x^8 - 1) = 0$, giving real roots $x \in \{0, 1, -1\}$ and stationary points
$(0,0)$, $(1,1)$, $(-1,-1)$.

**Step 3 (classification).**

$$
\nabla^2 f = \begin{bmatrix} 12x^2 & -4 \\ -4 & 12y^2 \end{bmatrix}
$$

- At $(0,0)$: $\begin{bmatrix} 0 & -4 \\ -4 & 0\end{bmatrix}$, $\det = -16 \lt 0$ — **saddle**,
  $f(0,0) = 0$.
- At $(\pm 1, \pm 1)$: $\begin{bmatrix} 12 & -4 \\ -4 & 12\end{bmatrix}$, $D_1 = 12 \gt 0$,
  $D_2 = 144 - 16 = 128 \gt 0$ — **strict local minima**, with $f = 1 + 1 - 4 = -2$.

**Step 4 (global minimizers).** A global minimizer exists and must be stationary; among the three
candidates the smallest value is $-2$, attained twice.

$$
\boxed{f^* = -2 \text{ attained at } (1,1) \text{ and } (-1,-1); \ (0,0) \text{ is a saddle}}
$$

> **Key takeaway:** Coercivity plus stationarity is a *complete* algorithm for global minimization of smooth functions with finitely many critical points: existence narrows the search to the critical set, and then you simply compare values.

```python
f = lambda x, y: x**4 + y**4 - 4.0 * x * y
grad = lambda x, y: np.array([4.0 * x**3 - 4.0 * y, 4.0 * y**3 - 4.0 * x])
hess = lambda x, y: np.array([[12.0 * x**2, -4.0], [-4.0, 12.0 * y**2]])

for pt in [(0.0, 0.0), (1.0, 1.0), (-1.0, -1.0)]:
    ev = np.linalg.eigvalsh(hess(*pt))
    print(f"{str(pt):14s} grad = {grad(*pt)}   eig = {ev}   f = {f(*pt):+.1f}")
    assert np.allclose(grad(*pt), 0.0, atol=1e-14)
assert abs(f(1.0, 1.0) + 2.0) < 1e-14 and abs(f(-1.0, -1.0) + 2.0) < 1e-14

g = np.linspace(-4.0, 4.0, 801)
X, Y = np.meshgrid(g, g)
print("minimum of f over a coarse grid on [-4,4]^2:", f(X, Y).min(), " (hand value -2)")
assert f(X, Y).min() >= -2.0 - 1e-12
```

### Problem L1.6 — The Quadratic Model Problem in the Nondegenerate Case

**Problem Statement:** For $f(\mathbf{x}) = \frac{1}{2}\mathbf{x}^\top A\mathbf{x} - \mathbf{b}^\top\mathbf{x}$
with $A$ symmetric, compute $\nabla f$ and $\nabla^2 f$, and prove: if $A \succ 0$ then $f$ has a unique
global minimizer $\mathbf{x}^* = A^{-1}\mathbf{b}$ with optimal value
$-\frac{1}{2}\mathbf{b}^\top A^{-1}\mathbf{b}$. What happens if $A$ has a negative eigenvalue?

*Intuition:* A quadratic is its own Taylor expansion, so the local conditions become global statements with no remainder to control.

**Solution:**

**Step 1 (derivatives).** Writing $f(\mathbf{x}) = \frac{1}{2}\sum_{i,j} A_{ij}x_i x_j - \sum_i b_i x_i$
and using $A = A^\top$:

$$
\nabla f(\mathbf{x}) = A\mathbf{x} - \mathbf{b}, \qquad \nabla^2 f(\mathbf{x}) = A \ \ \text{(constant)}
$$

**Step 2 (stationarity).** FONC reads $A\mathbf{x} = \mathbf{b}$. If $A \succ 0$ then $A$ is invertible,
so there is exactly one stationary point $\mathbf{x}^* = A^{-1}\mathbf{b}$, and $\nabla^2 f = A \succ 0$
satisfies SOSC: $\mathbf{x}^*$ is a strict local minimizer.

**Step 3 (globality by exact completion of the square).** For any $\mathbf{x}$, put
$\mathbf{e} = \mathbf{x} - \mathbf{x}^*$. Then

$$
f(\mathbf{x}) = \frac{1}{2}(\mathbf{x}^* + \mathbf{e})^\top A(\mathbf{x}^* + \mathbf{e}) - \mathbf{b}^\top(\mathbf{x}^* + \mathbf{e}) = f(\mathbf{x}^*) + \mathbf{e}^\top\left(A\mathbf{x}^* - \mathbf{b}\right) + \frac{1}{2}\mathbf{e}^\top A \mathbf{e}
$$

The middle term vanishes by stationarity, so

$$
f(\mathbf{x}) = f(\mathbf{x}^*) + \frac{1}{2}\mathbf{e}^\top A\mathbf{e} \ge f(\mathbf{x}^*) + \frac{\lambda_{\min}(A)}{2}\lVert \mathbf{e}\rVert^2
$$

with equality only at $\mathbf{e} = \mathbf{0}$: the minimizer is **unique and global**.

**Step 4 (optimal value).**

$$
f(\mathbf{x}^*) = \frac{1}{2}(A^{-1}\mathbf{b})^\top A (A^{-1}\mathbf{b}) - \mathbf{b}^\top A^{-1}\mathbf{b} = \frac{1}{2}\mathbf{b}^\top A^{-1}\mathbf{b} - \mathbf{b}^\top A^{-1}\mathbf{b} = -\frac{1}{2}\mathbf{b}^\top A^{-1}\mathbf{b}
$$

**Step 5 (negative eigenvalue).** If $A\mathbf{v} = \lambda\mathbf{v}$ with $\lambda \lt 0$ and
$\lVert \mathbf{v}\rVert = 1$, then along the ray $\mathbf{x}(t) = \mathbf{x}_0 + t\mathbf{v}$,

$$
f(\mathbf{x}(t)) = f(\mathbf{x}_0) + t\,\mathbf{v}^\top(A\mathbf{x}_0 - \mathbf{b}) + \frac{\lambda}{2}t^2 \to -\infty
$$

so $f$ is unbounded below and no minimizer exists (SONC already fails at any stationary point).

$$
\boxed{A \succ 0: \ \mathbf{x}^* = A^{-1}\mathbf{b}, \ f^* = -\tfrac{1}{2}\mathbf{b}^\top A^{-1}\mathbf{b}; \qquad \lambda_{\min}(A) \lt 0: \ \inf f = -\infty}
$$

> **Key takeaway:** The quadratic model is the Rosetta stone of the module — "solve $A\mathbf{x} = \mathbf{b}$" *is* the optimality condition, and the eigenvalue signature of $A$ decides existence, uniqueness and unboundedness with no approximation anywhere.

```python
n = 5
M = rng.standard_normal((n, n))
A = M @ M.T + n * np.eye(n)          # symmetric positive definite by construction
b = rng.standard_normal(n)
f = lambda x: 0.5 * x @ A @ x - b @ x

x_star = np.linalg.solve(A, b)
print("lambda_min(A)          =", np.linalg.eigvalsh(A).min(), "> 0, so SOSC holds")
print("||A x* - b||           =", np.linalg.norm(A @ x_star - b))
print("f(x*)                  =", f(x_star))
print("-b^T A^-1 b / 2        =", -0.5 * b @ x_star)
assert np.linalg.norm(A @ x_star - b) < 1e-10
assert abs(f(x_star) + 0.5 * b @ x_star) < 1e-12

gaps = [abs(f(x_star + e) - f(x_star) - 0.5 * e @ A @ e) for e in rng.standard_normal((6, n))]
print("max |f(x*+e) - f(x*) - e^T A e / 2| =", f"{max(gaps):.2e}", "(the expansion is exact)")
assert max(gaps) < 1e-10
```

### Problem L2.1 — Least Squares — Normal Equations from FONC and SONC

**Problem Statement:** For $f(\mathbf{w}) = \frac{1}{2}\lVert X\mathbf{w} - \mathbf{y}\rVert_2^2$ with
$X \in \mathbb{R}^{m \times n}$, derive the normal equations from FONC, verify SONC/SOSC, and state
exactly when the minimizer is unique. Describe the solution set when it is not.

*Intuition:* Training a linear model is the quadratic model problem of Problem L1.6 with $A = X^\top X$ and $\mathbf{b} = X^\top\mathbf{y}$.

**Solution:**

**Step 1 (expand into the quadratic model).**

$$
f(\mathbf{w}) = \frac{1}{2}\mathbf{w}^\top X^\top X \mathbf{w} - \mathbf{y}^\top X \mathbf{w} + \frac{1}{2}\lVert \mathbf{y}\rVert^2
$$

so with $A = X^\top X$ and $\mathbf{b} = X^\top\mathbf{y}$ this is Problem L1.6 plus a constant.

**Step 2 (FONC = normal equations).**

$$
\nabla f(\mathbf{w}) = X^\top(X\mathbf{w} - \mathbf{y}) = \mathbf{0} \iff X^\top X\mathbf{w} = X^\top\mathbf{y}
$$

**Step 3 (SONC always holds).** $\nabla^2 f = X^\top X$, and for every $\mathbf{v}$,

$$
\mathbf{v}^\top X^\top X \mathbf{v} = \lVert X\mathbf{v}\rVert^2 \ge 0
$$

so $X^\top X \succeq 0$: $f$ is convex, every stationary point is a global minimizer (Problem L3.2), and the
normal equations are *always* consistent because $X^\top\mathbf{y} \in \operatorname{range}(X^\top) = \operatorname{range}(X^\top X)$.

**Step 4 (uniqueness).** $X^\top X \succ 0$ if and only if $X\mathbf{v} = \mathbf{0} \Rightarrow \mathbf{v} = \mathbf{0}$,
i.e. $\operatorname{rank}(X) = n$ (full column rank, which requires $m \ge n$). Then

$$
\mathbf{w}^* = (X^\top X)^{-1}X^\top\mathbf{y}
$$

is the unique minimizer and SOSC holds.

**Step 5 (rank-deficient case).** If $\operatorname{rank}(X) = r \lt n$, the solution set is the affine
subspace $\mathbf{w}_0 + \ker(X)$ of dimension $n - r$; $f$ is constant along $\ker(X)$ (flat directions),
so minimizers are non-unique and SOSC never holds. The canonical selection is the minimum-norm solution
$\mathbf{w}^\dagger = X^{+}\mathbf{y}$ given by the pseudoinverse — and it is also the point gradient
descent from $\mathbf{w}_0 = \mathbf{0}$ converges to.

$$
\boxed{X^\top X\mathbf{w}^* = X^\top\mathbf{y}; \quad \text{unique} \iff \operatorname{rank}(X) = n; \quad \text{else minimizers} = \mathbf{w}_0 + \ker(X)}
$$

> **Key takeaway:** Every fact about least squares — consistency, uniqueness, the pseudoinverse, the implicit bias of gradient descent — is read off the eigen-structure of $X^\top X$ through the optimality conditions.

```python
m, n = 12, 4
X = rng.standard_normal((m, n))
y = rng.standard_normal(m)

w_star = np.linalg.solve(X.T @ X, X.T @ y)
print("full column rank:")
print("  ||X^T X w* - X^T y|| =", np.linalg.norm(X.T @ X @ w_star - X.T @ y))
print("  agrees with lstsq   :", np.allclose(w_star, np.linalg.lstsq(X, y, rcond=None)[0]))
print("  eig(X^T X)          =", np.linalg.eigvalsh(X.T @ X))
assert np.linalg.eigvalsh(X.T @ X).min() > 0
assert np.allclose(w_star, np.linalg.lstsq(X, y, rcond=None)[0])

Xd = X.copy()
Xd[:, 3] = Xd[:, 0] + 2.0 * Xd[:, 1]            # column 4 is a combination of columns 1 and 2
v = np.array([1.0, 2.0, 0.0, -1.0])             # spans ker(X_d)
w_pinv = np.linalg.pinv(Xd) @ y
print("rank deficient:")
print("  rank(X) =", np.linalg.matrix_rank(Xd),
      "  lambda_min(X^T X) =", np.linalg.eigvalsh(Xd.T @ Xd).min(), "(zero up to rounding)")
print("  ||X v|| =", np.linalg.norm(Xd @ v), " so f is flat along v")
print("  normal equations still consistent, residual =",
      np.linalg.norm(Xd.T @ Xd @ w_pinv - Xd.T @ y))
assert np.linalg.norm(Xd @ v) < 1e-12
assert np.linalg.norm(Xd.T @ Xd @ w_pinv - Xd.T @ y) < 1e-10
assert abs(0.5 * np.linalg.norm(Xd @ (w_pinv + 3.0 * v) - y) ** 2
           - 0.5 * np.linalg.norm(Xd @ w_pinv - y) ** 2) < 1e-10
```

### Problem L2.2 — Separable Logistic Regression Has No Minimizer

**Problem Statement:** For unregularized logistic regression
$f(\mathbf{w}) = \sum_{i=1}^m \log\left(1 + e^{-y_i \mathbf{w}^\top\mathbf{x}_i}\right)$ with
$y_i \in \{-1, +1\}$, show that if the data are linearly separable then $f$ has **no** minimizer, and that
adding a ridge term $\frac{\lambda}{2}\lVert \mathbf{w}\rVert^2$ restores existence and uniqueness.

*Intuition:* Perfect separation lets the model keep improving forever by scaling up the weights — the infimum sits at infinity.

**Solution:**

**Step 1 (separability).** Assume there is $\mathbf{u}$ with $y_i\mathbf{u}^\top\mathbf{x}_i \ge \gamma \gt 0$
for all $i$ (a separating direction with margin $\gamma$).

**Step 2 (the infimum is $0$ and is not attained).** Along the ray $\mathbf{w} = t\mathbf{u}$, $t \gt 0$:

$$
f(t\mathbf{u}) = \sum_{i=1}^m \log\left(1 + e^{-t\, y_i\mathbf{u}^\top\mathbf{x}_i}\right) \le m \log\left(1 + e^{-t\gamma}\right) \xrightarrow[t\to\infty]{} 0
$$

Since $\log(1+e^{-s}) \gt 0$ for every real $s$, we have $f(\mathbf{w}) \gt 0$ for every finite
$\mathbf{w}$. Therefore $\inf f = 0$ is **never attained**: no minimizer exists.

**Step 3 (why FONC cannot save you).** The gradient is

$$
\nabla f(\mathbf{w}) = -\sum_{i=1}^m \sigma\left(-y_i\mathbf{w}^\top\mathbf{x}_i\right) y_i\mathbf{x}_i, \qquad \sigma(s) = \frac{1}{1+e^{-s}}
$$

Each coefficient $\sigma(-y_i\mathbf{w}^\top\mathbf{x}_i)$ is strictly positive, so under separability the
gradient is a strictly positive combination of the vectors $-y_i\mathbf{x}_i$, which all have negative
inner product with $\mathbf{u}$; hence $\nabla f(\mathbf{w})^\top\mathbf{u} \lt 0$ for every $\mathbf{w}$ and
the gradient never vanishes. The optimality *system is empty* — a failure of existence, not of the
conditions.

**Step 4 (ridge restores existence).** Let
$f_\lambda(\mathbf{w}) = f(\mathbf{w}) + \frac{\lambda}{2}\lVert \mathbf{w}\rVert^2$ with $\lambda \gt 0$.
Since $f \ge 0$,

$$
f_\lambda(\mathbf{w}) \ge \frac{\lambda}{2}\lVert \mathbf{w}\rVert^2 \to \infty
$$

so $f_\lambda$ is coercive and continuous: a global minimizer exists. Moreover

$$
\nabla^2 f_\lambda(\mathbf{w}) = \sum_i \sigma_i(1-\sigma_i)\mathbf{x}_i\mathbf{x}_i^\top + \lambda I \succeq \lambda I \succ 0
$$

so $f_\lambda$ is $\lambda$-strongly convex and the minimizer is **unique**.

$$
\boxed{\text{separable data} \Rightarrow \inf f = 0 \text{ unattained}; \quad \lambda \gt 0 \Rightarrow \text{unique minimizer, } \nabla^2 f_\lambda \succeq \lambda I}
$$

> **Key takeaway:** The weight blow-up practitioners observe on separable data is not a numerical bug but a theorem about non-existence; regularization is the mathematically minimal fix, and it simultaneously buys strong convexity and a linear convergence rate.

```python
X = np.array([[1.0, 1.0], [2.0, 1.0], [-1.0, -1.0], [-2.0, -1.5]])
lab = np.array([1.0, 1.0, -1.0, -1.0])
u = np.array([1.0, 0.0])
margins = lab * (X @ u)
print("margins y_i u^T x_i =", margins, " -> separable with gamma =", margins.min())
assert margins.min() > 0

logistic = lambda w: float(np.sum(np.log1p(np.exp(-lab * (X @ w)))))
for t in [1.0, 5.0, 20.0, 100.0]:
    print(f"  f({t:5.0f} u) = {logistic(t * u):.3e}")
assert logistic(100.0 * u) > 0.0
assert logistic(100.0 * u) < logistic(1.0 * u)

lam = 0.5
sigma = lambda s: 1.0 / (1.0 + np.exp(-s))
grad_ridge = lambda w: -X.T @ (lab * sigma(-lab * (X @ w))) + lam * w
w = np.zeros(2)
for _ in range(50):
    p = sigma(lab * (X @ w))
    Hess = X.T @ (X * (p * (1.0 - p))[:, None]) + lam * np.eye(2)
    w = w - np.linalg.solve(Hess, grad_ridge(w))
p = sigma(lab * (X @ w))
Hess = X.T @ (X * (p * (1.0 - p))[:, None]) + lam * np.eye(2)
print("ridge minimizer w* =", w, "  ||grad f_lam(w*)|| =", np.linalg.norm(grad_ridge(w)))
print("lambda_min(Hessian of f_lam) =", np.linalg.eigvalsh(Hess).min(), ">= lambda =", lam)
assert np.linalg.norm(grad_ridge(w)) < 1e-10
assert np.linalg.eigvalsh(Hess).min() >= lam - 1e-12
```

### Problem L2.3 — Rosenbrock — A Certified Unique Global Minimum

**Problem Statement:** For the Rosenbrock function $f(x,y) = (1-x)^2 + 100(y-x^2)^2$, find all stationary
points, verify SOSC at the minimizer, compute the condition number of the Hessian there, and explain what
that number predicts about first-order methods.

*Intuition:* A banana-shaped valley: the objective is a sum of two squares, so its zero set pins the minimizer exactly.

**Solution:**

**Step 1 (stationary points).**

$$
f_x = -2(1-x) - 400x(y-x^2), \qquad f_y = 200(y - x^2)
$$

$f_y = 0$ forces $y = x^2$; substituting into $f_x = 0$ leaves $-2(1-x) = 0$, so $x = 1$, $y = 1$. There
is exactly **one** stationary point, $(1,1)$, with $f(1,1) = 0$.

**Step 2 (it is the global minimum).** $f$ is a sum of two squares, so $f \ge 0$ everywhere, and
$f(1,1) = 0$. Hence $(1,1)$ is a global minimizer; it is unique because $f(x,y) = 0$ forces $x = 1$ and
$y = x^2 = 1$.

**Step 3 (SOSC).**

$$
\nabla^2 f(x,y) = \begin{bmatrix} 2 - 400y + 1200x^2 & -400x \\ -400x & 200\end{bmatrix}, \qquad \nabla^2 f(1,1) = \begin{bmatrix} 802 & -400 \\ -400 & 200 \end{bmatrix}
$$

Leading minors: $D_1 = 802 \gt 0$ and $D_2 = 802\cdot 200 - 400^2 = 160400 - 160000 = 400 \gt 0$, so
$\nabla^2 f(1,1) \succ 0$ and SOSC certifies a strict local (here global) minimum.

**Step 4 (conditioning).** Trace $= 1002$, determinant $= 400$, so

$$
\lambda_{\pm} = \frac{1002 \pm \sqrt{1002^2 - 1600}}{2} \approx 1001.6 \ \text{and}\ 0.3994, \qquad \kappa = \frac{\lambda_{\max}}{\lambda_{\min}} \approx 2508
$$

**Step 5 (what $\kappa$ predicts).** Optimally tuned gradient descent contracts by
$\frac{\kappa-1}{\kappa+1} \approx 0.99920$ per step, i.e. about
$\ln 10 / (2/\kappa) \approx 2.9 \times 10^3$ iterations per decimal digit. Newton's method, which uses
$\nabla^2 f$ itself, is affine invariant and unaffected by $\kappa$.

$$
\boxed{\text{unique stationary point } (1,1),\ \nabla^2 f(1,1) \succ 0,\ \kappa \approx 2.5\times 10^3}
$$

> **Key takeaway:** Rosenbrock is not hard because of spurious minima — it has exactly one critical point — but because the certified minimum is extremely ill-conditioned, which is an optimality-*geometry* statement, not a topological one.

```python
f = lambda x, y: (1.0 - x) ** 2 + 100.0 * (y - x**2) ** 2
grad = lambda x, y: np.array([-2.0 * (1.0 - x) - 400.0 * x * (y - x**2), 200.0 * (y - x**2)])
H = np.array([[2.0 - 400.0 * 1.0 + 1200.0 * 1.0, -400.0], [-400.0, 200.0]])

print("||grad(1,1)|| =", np.linalg.norm(grad(1.0, 1.0)), "   f(1,1) =", f(1.0, 1.0))
print("Hessian(1,1) =", H.tolist())
minors = np.array([H[0, 0], np.linalg.det(H)])
ev = np.linalg.eigvalsh(H)
kappa = ev[-1] / ev[0]
print("leading minors =", np.round(minors, 6), " (hand: 802, 400)")
print("eigenvalues    =", ev, "   kappa =", f"{kappa:.1f}")
assert np.allclose(grad(1.0, 1.0), 0.0) and abs(f(1.0, 1.0)) < 1e-14
assert minors[0] > 0 and minors[1] > 0 and abs(minors[1] - 400.0) < 1e-8
assert abs(kappa - 2508.0) < 1.0

rho = (kappa - 1.0) / (kappa + 1.0)
print(f"optimal-step contraction rho = {rho:.6f}"
      f"   iterations per decimal digit = {np.log(10.0) / -np.log(rho):.0f}")
```

### Problem L2.4 — The Double-Well Potential and Spontaneous Symmetry Breaking

**Problem Statement:** Analyse the Landau free energy $V(x) = \frac{a}{2}x^2 + \frac{b}{4}x^4$ with
$b \gt 0$, treating $a \in \mathbb{R}$ as a temperature-like control parameter. Find all stationary points
and their classification as a function of $a$, and interpret the transition physically.

*Intuition:* At the critical value $a = 0$ the second-order test goes silent — and exactly there the physics changes character.

**Solution:**

**Step 1 (stationary points).**

$$
V'(x) = ax + bx^3 = x(a + bx^2) = 0 \iff x = 0 \quad \text{or} \quad x^2 = -\frac{a}{b}
$$

The nonzero roots $x = \pm\sqrt{-a/b}$ exist only when $a \lt 0$.

**Step 2 (second derivative).** $V''(x) = a + 3bx^2$.

- **Case $a \gt 0$:** the only stationary point is $x = 0$ with $V''(0) = a \gt 0$ — a **unique global
  minimum** (and $V$ is coercive since $b \gt 0$, so it is attained). Single-well regime.
- **Case $a = 0$:** $x = 0$ is the only stationary point and $V''(0) = 0$ — the second-order test is
  **silent**. The quartic term decides: $V(x) = \frac{b}{4}x^4 \gt 0$ for $x \neq 0$, so $x = 0$ is still
  a strict global minimum, but only "quartically flat" rather than quadratically curved.
- **Case $a \lt 0$:** $V''(0) = a \lt 0$, so the origin becomes a **local maximum**, while at
  $x_\pm = \pm\sqrt{-a/b}$,

$$
V''(x_\pm) = a + 3b\left(-\frac{a}{b}\right) = -2a \gt 0
$$

  two symmetric **strict global minima** with

$$
V(x_\pm) = \frac{a}{2}\left(-\frac{a}{b}\right) + \frac{b}{4}\frac{a^2}{b^2} = -\frac{a^2}{4b} \lt 0
$$

**Step 3 (physical reading).** $V$ is even, so the *model* is symmetric under $x \mapsto -x$ at every $a$.
For $a \lt 0$, however, each equilibrium state is *not* symmetric: the system must pick one of the two
wells. This is **spontaneous symmetry breaking**, with $a \propto T - T_c$ giving the ferromagnetic
transition (the order parameter $\lvert x_\pm\rvert = \sqrt{-a/b}$ grows like $(T_c - T)^{1/2}$, the mean-field
exponent $\beta = 1/2$).

$$
\boxed{a \gt 0: \ x^*=0 \text{ unique min}; \quad a = 0: \ \text{degenerate quartic min}; \quad a \lt 0: \ x^* = \pm\sqrt{-a/b}, \ V^* = -\frac{a^2}{4b}}
$$

> **Key takeaway:** The degenerate case that the second-order test cannot resolve is not a mathematical curiosity — it is exactly the critical point of a phase transition, where the leading behaviour is quartic and universal exponents appear.

```python
b = 1.0
for a in [1.0, 0.0, -2.0]:
    V = lambda x: 0.5 * a * x**2 + 0.25 * b * x**4
    d2V = lambda x: a + 3.0 * b * x**2
    roots = [0.0] + ([np.sqrt(-a / b), -np.sqrt(-a / b)] if a < 0 else [])
    print(f"a = {a:+.1f}:")
    for r in roots:
        print(f"    x = {r:+.4f}   V''(x) = {d2V(r):+.4f}   V(x) = {V(r):+.6f}")
    if a < 0:
        assert abs(V(np.sqrt(-a / b)) - (-(a**2) / (4.0 * b))) < 1e-14
        assert abs(d2V(np.sqrt(-a / b)) - (-2.0 * a)) < 1e-14
        assert d2V(0.0) < 0
    else:
        assert d2V(0.0) >= 0
print("at a = -2, b = 1:  V(x_pm) = -a^2/(4b) =", -(-2.0) ** 2 / (4.0 * 1.0),
      "  order parameter sqrt(-a/b) =", np.sqrt(2.0))
```

### Problem L2.5 — Normal Modes, Imaginary Frequencies, and Why Deep Networks Meet Saddles

**Problem Statement:** *(a) Physics.* Two unit masses at displacements $x_1, x_2$ are held by three
identical springs of stiffness $k = 1$: one from each mass to a fixed wall, and one coupling the masses,
so that

$$
U(x_1,x_2) = \tfrac{1}{2}x_1^2 + \tfrac{1}{2}(x_2 - x_1)^2 + \tfrac{1}{2}x_2^2 .
$$

Verify that the origin is an equilibrium, compute the stiffness matrix $K = \nabla^2 U(\mathbf{0})$, and
obtain the two normal-mode frequencies from Newton's equation $\ddot{\mathbf{x}} = -K\mathbf{x}$. Then add a
destabilizing potential $-(x_1^2 + x_2^2)$ — each mass balanced on an inverted pendulum of stiffness $2$ —
and show that the origin becomes a saddle of the total potential, with one imaginary frequency; give the
exponential growth rate and the direction in which the system falls.

*(b) Machine learning.* At a stationary point of a high-dimensional loss, model the $n$ Hessian
eigenvalues as independent symmetric random signs. Compute the probability that the point is a local
minimum for $n = 10$ and $n = 10^6$, say what the correct random-matrix answer is, and explain the
mechanism by which gradient descent is slow near — but not trapped by — a strict saddle.

*Intuition:* A negative curvature is an imaginary frequency, and an imaginary frequency is exponential escape; a minimum demands unanimity among $n$ eigenvalues, and unanimity is astronomically unlikely when $n$ is large.

**Solution:**

**Step 1 (equilibrium and stiffness matrix).** Differentiating,

$$
\nabla U(x_1,x_2) = \begin{pmatrix} x_1 - (x_2-x_1) \\ (x_2-x_1) + x_2\end{pmatrix} = \begin{pmatrix} 2x_1 - x_2 \\ -x_1 + 2x_2\end{pmatrix},
$$

so $\nabla U(\mathbf{0}) = \mathbf{0}$: the origin is an equilibrium by FONC. The stiffness matrix is the
Hessian,

$$
K = \nabla^2 U = \begin{bmatrix} 2 & -1 \\ -1 & 2 \end{bmatrix} .
$$

**Step 2 (normal modes).** $K$ has $\det(K - \lambda I) = (2-\lambda)^2 - 1 = 0$, so $\lambda = 1, 3$ with
eigenvectors $\mathbf{v}_1 = \tfrac{1}{\sqrt2}(1,1)^\top$ (both masses move together) and
$\mathbf{v}_2 = \tfrac{1}{\sqrt2}(1,-1)^\top$ (they move oppositely). Writing
$\mathbf{x}(t) = c\, \mathbf{v}_i \cos(\omega t + \phi)$ in $\ddot{\mathbf{x}} = -K\mathbf{x}$ gives
$-\omega^2 c\,\mathbf{v}_i = -\lambda_i c\, \mathbf{v}_i$, hence $\omega_i = \sqrt{\lambda_i / m}$ with
$m = 1$:

$$
\omega_1 = 1 \ \text{(symmetric mode)}, \qquad \omega_2 = \sqrt{3} \approx 1.7321 \ \text{(antisymmetric mode)} .
$$

Both frequencies are real because $K \succ 0$ — which is exactly SOSC, so the equilibrium is a strict local
minimum of $U$ and is **stable**.

**Step 3 (destabilization: a saddle with an imaginary frequency).** With
$U'(\mathbf{x}) = U(\mathbf{x}) - (x_1^2 + x_2^2)$ the gradient at the origin still vanishes, while

$$
K' = \nabla^2 U' = K - 2I = \begin{bmatrix} 0 & -1 \\ -1 & 0\end{bmatrix},
$$

whose eigenvalues are $\lambda = -1$ on $\mathbf{v}_1 = \tfrac{1}{\sqrt2}(1,1)^\top$ and $\lambda = +1$ on
$\mathbf{v}_2 = \tfrac{1}{\sqrt2}(1,-1)^\top$. SONC fails ($\lambda_{\min} = -1 \lt 0$), so by the
second-order necessary condition the origin is no longer a minimum: it is a saddle. Directly,
$U'(t,t) = -t^2 \lt 0$ while $U'(t,-t) = +t^2 \gt 0$.

The mode frequency along $\mathbf{v}_1$ is $\omega = \sqrt{\lambda/m} = \sqrt{-1} = i$: **imaginary**, so
$\cos(\omega t)$ becomes $\cosh(t)$ and the amplitude grows like $e^{t}$ with growth rate
$\lvert \lambda_{\min}\rvert^{1/2} = 1$. The system falls along $\mathbf{v}_1 = (1,1)/\sqrt2$, both masses
sliding the same way, while the antisymmetric mode keeps oscillating at $\omega = 1$.

**Step 4 (probability of unanimity in dimension $n$).** Under the independent-sign toy model each
eigenvalue is positive with probability $1/2$, so

$$
\mathbb{P}[\text{local min}] = \mathbb{P}[\text{all } n \text{ eigenvalues} \gt 0] = 2^{-n},
$$

and by symmetry the same for a local maximum; everything else — probability $1 - 2^{1-n}$ — is a saddle.
Numerically,

$$
n = 10: \ 2^{-10} \approx 9.77 \times 10^{-4}, \qquad n = 10^6: \ \log_{10} 2^{-10^6} = -10^6 \log_{10} 2 \approx -3.01 \times 10^{5} .
$$

Independence is a *toy model*, not a theorem. Eigenvalues of a symmetric random matrix repel, and the
random-matrix calculation (Bray & Dean 2007; Fyodorov & Williams 2007) gives the far smaller probability
$\exp(-\Theta(n^2))$. Both models agree on the point Dauphin et al. (2014) exploit: minima are
overwhelmingly rare relative to saddles, and low-index critical points are the ones with low loss.

**Step 5 (dynamics near a strict saddle).** Near a stationary point with
$\nabla^2 f(\mathbf{x}^{\ast}) = \operatorname{diag}(\lambda_i)$, gradient descent with step $\alpha$ acts
componentwise in the eigenbasis as

$$
e_i^{(k+1)} = (1 - \alpha\lambda_i)\, e_i^{(k)} .
$$

Along a direction with $\lambda_i \lt 0$ the factor $1 - \alpha\lambda_i \gt 1$, so that component **grows
geometrically** — the discrete-time image of the imaginary frequency of Step 3. Growth starts from whatever
tiny component $\delta_0$ the iterate happens to carry, so escape takes about

$$
k_{\text{escape}} \approx \frac{\log(1/\delta_0)}{\log(1 + \alpha\lvert \lambda_{\min}\rvert)} \approx \frac{\log(1/\delta_0)}{\alpha\lvert \lambda_{\min}\rvert}
$$

steps, which is the observed **plateau**: $\lVert \nabla f\rVert$ is tiny, the loss barely moves, and then
training suddenly resumes.

$$
\boxed{K = \begin{bmatrix} 2 & -1 \\ -1 & 2\end{bmatrix}: \ \omega = 1, \sqrt{3}; \quad K - 2I: \ \omega = i, 1 \ \text{(escape along } (1,1)\text{)}; \quad \mathbb{P}[\text{min}] = 2^{-n} \text{ (toy)}, \ e^{-\Theta(n^2)} \text{ (GOE)}}
$$

> **Key takeaway:** Negative curvature is one phenomenon wearing two costumes — an imaginary normal-mode frequency in mechanics and an expanding gradient-descent direction in training — so in both settings a strict saddle repels rather than traps, and the cost it imposes is the plateau, measured by $\log(1/\delta_0) / (\alpha\lvert \lambda_{\min}\rvert)$.

```python
K = np.array([[2.0, -1.0], [-1.0, 2.0]])
U = lambda x1, x2: 0.5 * x1**2 + 0.5 * (x2 - x1) ** 2 + 0.5 * x2**2
grad_U = lambda x1, x2: np.array([2.0 * x1 - x2, -x1 + 2.0 * x2])

lam, vec = np.linalg.eigh(K)
print("stable chain:  grad U(0,0) =", grad_U(0.0, 0.0), "   eig(K) =", lam, " (hand: 1, 3)")
print("  normal-mode frequencies omega = sqrt(lambda) =", np.sqrt(lam), " (hand: 1, sqrt3 = 1.7321)")
print("  eigenvectors (columns) =", np.round(vec, 4).tolist())
assert np.allclose(grad_U(0.0, 0.0), 0.0) and np.allclose(lam, [1.0, 3.0])
assert np.all(lam > 0)

Kp = K - 2.0 * np.eye(2)
lam_p, vec_p = np.linalg.eigh(Kp)
print("destabilized:  eig(K - 2I) =", lam_p, " (hand: -1, +1) -> SONC fails, saddle")
print("  omega along the unstable mode = sqrt(-1) = i, growth rate", np.sqrt(-lam_p[0]))
print("  U'(t,t), U'(t,-t) at t = 0.1:",
      U(0.1, 0.1) - (0.1**2 + 0.1**2), U(0.1, -0.1) - (0.1**2 + 0.1**2), " (hand: -t^2, +t^2)")
assert np.allclose(lam_p, [-1.0, 1.0])
assert abs((U(0.1, 0.1) - 2 * 0.1**2) + 0.1**2) < 1e-15
assert abs((U(0.1, -0.1) - 2 * 0.1**2) - 0.1**2) < 1e-15

for n in [10, 100]:
    print(f"n = {n:4d}:  P[all n signs positive] = 2^-n = {2.0 ** -n:.4e}")
print("n = 10^6:  log10 P = -10^6 log10 2 =", -1e6 * np.log10(2.0))
assert abs(2.0 ** -10 - 9.765625e-4) < 1e-12
assert abs(-1e6 * np.log10(2.0) + 301029.9957) < 1e-2

alpha, lam_min, delta0 = 1e-2, -0.5, 1e-6
e, k = delta0, 0
while e < 1.0:
    e *= 1.0 - alpha * lam_min
    k += 1
estimate = np.log(1.0 / delta0) / (alpha * abs(lam_min))
print(f"simulated escape from a strict saddle: {k} steps;  estimate log(1/delta0)/(alpha |lam_min|) = {estimate:.0f}")
assert abs(k - estimate) / k < 0.02
```

### Problem L2.6 — Ridge Regularization Buys Existence, Uniqueness and Conditioning

**Problem Statement:** For $f_\lambda(\mathbf{w}) = \frac{1}{2}\lVert X\mathbf{w}-\mathbf{y}\rVert^2 + \frac{\lambda}{2}\lVert \mathbf{w}\rVert^2$
with $\lambda \gt 0$, derive the minimizer, prove it is unique even when $X$ is rank deficient, and give
the condition number of the Hessian in terms of the singular values of $X$.

*Intuition:* Adding $\lambda I$ lifts the whole spectrum off zero, converting a flat valley into a genuine bowl.

**Solution:**

**Step 1 (derivatives).**

$$
\nabla f_\lambda(\mathbf{w}) = X^\top(X\mathbf{w}-\mathbf{y}) + \lambda\mathbf{w} = (X^\top X + \lambda I)\mathbf{w} - X^\top\mathbf{y}, \qquad \nabla^2 f_\lambda = X^\top X + \lambda I
$$

**Step 2 (positive definiteness).** For any $\mathbf{v} \neq \mathbf{0}$,

$$
\mathbf{v}^\top(X^\top X + \lambda I)\mathbf{v} = \lVert X\mathbf{v}\rVert^2 + \lambda\lVert \mathbf{v}\rVert^2 \ge \lambda\lVert \mathbf{v}\rVert^2 \gt 0
$$

so $\nabla^2 f_\lambda \succ 0$ regardless of $\operatorname{rank}(X)$: $f_\lambda$ is
$\lambda$-strongly convex, hence coercive, so a minimizer exists, is unique, and SOSC holds.

**Step 3 (closed form).** FONC gives the invertible system

$$
\mathbf{w}^*_\lambda = (X^\top X + \lambda I)^{-1}X^\top\mathbf{y}
$$

**Step 4 (conditioning).** If $X$ has singular values $\sigma_1 \ge \dots \ge \sigma_n \ge 0$, then
$X^\top X + \lambda I$ has eigenvalues $\sigma_i^2 + \lambda$, so

$$
\kappa_\lambda = \frac{\sigma_1^2 + \lambda}{\sigma_n^2 + \lambda}
$$

This is decreasing in $\lambda$: it equals $\kappa_0 = \sigma_1^2/\sigma_n^2$ (possibly $+\infty$) at
$\lambda = 0$ and tends to $1$ as $\lambda \to \infty$.

**Step 5 (SVD view of the shrinkage).** With $X = U\Sigma V^\top$,

$$
\mathbf{w}^*_\lambda = \sum_{i} \frac{\sigma_i}{\sigma_i^2 + \lambda}\left(\mathbf{u}_i^\top\mathbf{y}\right)\mathbf{v}_i
$$

so directions with $\sigma_i^2 \ll \lambda$ are damped by the factor $\sigma_i^2/(\sigma_i^2+\lambda)$
relative to ordinary least squares — the noisy, ill-determined directions are exactly the ones suppressed.

$$
\boxed{\mathbf{w}^*_\lambda = (X^TX+\lambda I)^{-1}X^\top\mathbf{y}, \quad \nabla^2 f_\lambda \succeq \lambda I, \quad \kappa_\lambda = \frac{\sigma_1^2+\lambda}{\sigma_n^2+\lambda}}
$$

> **Key takeaway:** One regularization parameter simultaneously fixes three separate pathologies — non-existence, non-uniqueness and ill-conditioning — because all three are statements about $\lambda_{\min}(\nabla^2 f)$ being zero.

```python
m, n, lam = 20, 5, 0.7
X = rng.standard_normal((m, n))
X[:, 4] = X[:, 0]                                # deliberately rank deficient
y = rng.standard_normal(m)

A_lam = X.T @ X + lam * np.eye(n)
w_lam = np.linalg.solve(A_lam, X.T @ y)
s = np.linalg.svd(X, compute_uv=False)
print("singular values of X          :", s)
print("eig(X^T X + lam I)            :", np.linalg.eigvalsh(A_lam))
print("hand value sigma^2 + lam      :", np.sort(s**2 + lam))
assert np.allclose(np.linalg.eigvalsh(A_lam), np.sort(s**2 + lam))
assert np.linalg.eigvalsh(A_lam).min() >= lam - 1e-12

kappa_lam = (s[0] ** 2 + lam) / (s[-1] ** 2 + lam)
print(f"kappa_lam = {kappa_lam:.4f}   while kappa_0 = sigma_1^2 / sigma_n^2 = {s[0] ** 2 / s[-1] ** 2:.3e}")

U, S, Vt = np.linalg.svd(X, full_matrices=False)
w_svd = Vt.T @ ((S / (S**2 + lam)) * (U.T @ y))
print("||w_lam - SVD shrinkage formula|| =", np.linalg.norm(w_lam - w_svd))
assert np.linalg.norm(w_lam - w_svd) < 1e-10
```

## L3 — Challenge Proofs

### Problem L3.1 — Peano's Function — Minimal Along Every Line, Yet No Local Minimum

**Problem Statement:** Let $f(x,y) = (y - x^2)(y - 2x^2)$. Show that the restriction of $f$ to every
straight line through the origin has a strict local minimum at the origin, yet the origin is *not* a local
minimum of $f$. Reconcile this with the second-order conditions.

*Intuition:* Lines are too rigid to explore the region between two tangent parabolas — the curve $y = 1.5x^2$ sneaks between them.

**Solution:**

**Step 1 (expand).**

$$
f(x,y) = y^2 - 3x^2 y + 2x^4
$$

with $f(0,0) = 0$. The gradient is $\nabla f = (-6xy + 8x^3,\ 2y - 3x^2)$, so $\nabla f(0,0) = \mathbf{0}$
and the origin is stationary.

**Step 2 (restriction to the vertical line $x = 0$).** $f(0,y) = y^2 \gt 0$ for $y \neq 0$: a strict local
minimum along that line.

**Step 3 (restriction to a line $y = mx$).**

$$
g_m(x) := f(x, mx) = m^2x^2 - 3m x^3 + 2x^4 = x^2\left(m^2 - 3mx + 2x^2\right)
$$

- If $m \neq 0$: the bracket tends to $m^2 \gt 0$ as $x \to 0$, so it is positive for all small
  $\lvert x\rvert \neq 0$, hence $g_m(x) \gt 0 = g_m(0)$ near $0$ — strict local minimum.
- If $m = 0$: $g_0(x) = 2x^4 \gt 0$ for $x \neq 0$ — again a strict local minimum.

So along **every** line through the origin, $f$ has a strict local minimum at the origin.

**Step 4 (the origin is not a local minimum of $f$).** Follow the parabola $y = \frac{3}{2}x^2$:

$$
f\!\left(x, \tfrac{3}{2}x^2\right) = \left(\tfrac{3}{2}x^2 - x^2\right)\left(\tfrac{3}{2}x^2 - 2x^2\right) = \left(\tfrac{1}{2}x^2\right)\left(-\tfrac{1}{2}x^2\right) = -\frac{x^4}{4} \lt 0
$$

for every $x \neq 0$. Such points lie arbitrarily close to the origin with strictly negative value, so the
origin is not a local minimum. (Geometrically, $f \lt 0$ exactly in the open region between the parabolas
$y = x^2$ and $y = 2x^2$, a cusp-shaped region that touches the origin but contains no line segment
through it.)

**Step 5 (reconciliation with the second-order test).**

$$
\nabla^2 f(x,y) = \begin{bmatrix} -6y + 24x^2 & -6x \\ -6x & 2\end{bmatrix}, \qquad \nabla^2 f(0,0) = \begin{bmatrix} 0 & 0 \\ 0 & 2\end{bmatrix} \succeq 0
$$

The Hessian is PSD **and singular**: SONC holds, SOSC does not apply. We are exactly in the degenerate gap
of Problem L0.2, and along the null direction $(1,0)$ the fourth-order behaviour is what decides — and it
depends on *which curve* tangent to that direction you follow.

$$
\boxed{f \text{ has a strict min along every line through } \mathbf{0}, \text{ yet } f\!\left(x,\tfrac{3}{2}x^2\right) = -\tfrac{x^4}{4} \lt 0}
$$

> **Key takeaway:** Minimality is a genuinely $n$-dimensional, full-neighbourhood property; one-dimensional slices — including the directional-derivative tests that define FONC and SONC — cannot certify it once the Hessian is singular.

```python
f = lambda x, y: (y - x**2) * (y - 2.0 * x**2)
H0 = np.array([[0.0, 0.0], [0.0, 2.0]])
ts = np.array([1e-1, 1e-2, 1e-3])
print("eig(Hessian at the origin) =", np.linalg.eigvalsh(H0), " -> PSD and singular")

print("vertical line x = 0:   f =", "  ".join(f"{v:+.2e}" for v in f(0.0 * ts, ts)))
assert np.all(f(0.0 * ts, ts) > 0)
for slope in [-2.0, -0.5, 0.0, 0.5, 2.0]:
    vals = f(ts, slope * ts)
    print(f"line y = {slope:+.1f} x:        f =", "  ".join(f"{v:+.2e}" for v in vals))
    assert np.all(vals > 0)

vals = f(ts, 1.5 * ts**2)
print("parabola y = 1.5 x^2:  f =", "  ".join(f"{v:+.2e}" for v in vals), "  hand: -x^4/4")
assert np.allclose(vals, -(ts**4) / 4.0) and np.all(vals < 0)
print("minimal along every line, yet negative on a parabola through the origin")
```

### Problem L3.2 — Convexity Upgrades Local Certificates to Global Ones

**Problem Statement:** Prove that for a differentiable convex $f$, every stationary point is a **global**
minimizer, and that the set of minimizers is convex. Then prove that $\mu$-strong convexity implies
coercivity, hence existence and uniqueness of the global minimizer.

*Intuition:* Convexity means the tangent plane never overestimates the function, so a horizontal tangent plane is a global floor.

**Solution:**

**Step 1 (first-order characterization of convexity).** For differentiable convex $f$ and all
$\mathbf{x}, \mathbf{y}$:

$$
f(\mathbf{y}) \ge f(\mathbf{x}) + \nabla f(\mathbf{x})^\top(\mathbf{y}-\mathbf{x})
$$

*Proof of this characterization.* Convexity gives
$f(\mathbf{x} + t(\mathbf{y}-\mathbf{x})) \le (1-t)f(\mathbf{x}) + t f(\mathbf{y})$, hence

$$
\frac{f(\mathbf{x} + t(\mathbf{y}-\mathbf{x})) - f(\mathbf{x})}{t} \le f(\mathbf{y}) - f(\mathbf{x})
$$

and letting $t \to 0^+$ the left side tends to $\nabla f(\mathbf{x})^\top(\mathbf{y}-\mathbf{x})$.

**Step 2 (stationary implies global).** If $\nabla f(\mathbf{x}^*) = \mathbf{0}$, Step 1 with
$\mathbf{x} = \mathbf{x}^*$ gives $f(\mathbf{y}) \ge f(\mathbf{x}^*)$ for **all** $\mathbf{y}$: global
minimality. Hence for convex problems FONC is not merely necessary but sufficient, and *globally* so.

**Step 3 (the minimizer set is convex).** The set of minimizers is the sublevel set
$S = \{\mathbf{x} : f(\mathbf{x}) \le f^*\}$, and sublevel sets of convex functions are convex: if
$f(\mathbf{x}), f(\mathbf{y}) \le c$ then
$f(\theta\mathbf{x} + (1-\theta)\mathbf{y}) \le \theta c + (1-\theta)c = c$.

**Step 4 (strong convexity implies coercivity).** With $\mu \gt 0$, strong convexity at a fixed
$\mathbf{x}_0$ gives

$$
f(\mathbf{x}) \ge f(\mathbf{x}_0) + \nabla f(\mathbf{x}_0)^\top(\mathbf{x}-\mathbf{x}_0) + \frac{\mu}{2}\lVert \mathbf{x}-\mathbf{x}_0\rVert^2 \ge f(\mathbf{x}_0) - \lVert \nabla f(\mathbf{x}_0)\rVert\, r + \frac{\mu}{2}r^2
$$

where $r = \lVert \mathbf{x}-\mathbf{x}_0\rVert$ and we used Cauchy-Schwarz. The right side is a quadratic
in $r$ with positive leading coefficient, so it tends to $+\infty$ with $r$: $f$ is **coercive**. By
Problem L0.4 a global minimizer exists.

**Step 5 (uniqueness).** If $\mathbf{x}^*$ and $\mathbf{y}^*$ were two distinct minimizers, strong
convexity at $\mathbf{x}^*$ (where $\nabla f(\mathbf{x}^*) = \mathbf{0}$) gives

$$
f^* = f(\mathbf{y}^*) \ge f^* + \frac{\mu}{2}\lVert \mathbf{y}^*-\mathbf{x}^*\rVert^2 \gt f^*
$$

a contradiction. Hence the minimizer is unique.

$$
\boxed{\text{convex}: \nabla f(\mathbf{x}^*) = \mathbf{0} \iff \mathbf{x}^* \text{ globally optimal}; \quad \mu\text{-strongly convex}: \text{minimizer exists and is unique}}
$$

> **Key takeaway:** Convexity is precisely the structural hypothesis that removes the word "local" from every conclusion in this module — and strong convexity additionally hands you existence for free, with no separate compactness argument.

### Problem L3.3 — Resolving the Degenerate Case with Higher-Order Terms

**Problem Statement:** All three functions $f_1 = x^2 + y^4$, $f_2 = x^2 - y^4$, $f_3 = x^2 + y^3$ have the
same gradient and the same Hessian at the origin. Classify each, then state and justify the general
one-dimensional rule for the first non-vanishing derivative.

*Intuition:* When the second-order test is silent, the verdict is delivered by the lowest-order term that survives along the null space of the Hessian.

**Solution:**

**Step 1 (identical second-order data).** For each $f_i$,

$$
\nabla f_i(0,0) = \mathbf{0}, \qquad \nabla^2 f_i(0,0) = \begin{bmatrix} 2 & 0 \\ 0 & 0\end{bmatrix} \succeq 0
$$

so all three pass SONC, all three fail SOSC, and second-order information cannot distinguish them. The
Hessian null space is $N = \operatorname{span}\{(0,1)\}$.

**Step 2 (restrict to the null direction).** Along $x = 0$:

$$
f_1(0,y) = y^4 \gt 0, \qquad f_2(0,y) = -y^4 \lt 0, \qquad f_3(0,y) = y^3 \ \text{(both signs)}
$$

**Step 3 (verdicts).**

- $f_1$: **strict global minimum** at the origin. Indeed $f_1 \ge 0$ with equality only at $(0,0)$; growth
  is quadratic in $x$ but only quartic in $y$ (a degenerate, "flat" minimum).
- $f_2$: **saddle**. It increases along $x$ and decreases along $y$.
- $f_3$: **neither** minimum nor maximum: along $y \lt 0$ the value is negative, along $y \gt 0$ positive
  — a degenerate saddle with odd-order behaviour (the example of Problem L0.2).

**Step 4 (the general scalar rule).** For $\phi \in \mathcal{C}^m$ with
$\phi'(0) = \dots = \phi^{(m-1)}(0) = 0$ and $\phi^{(m)}(0) \neq 0$, Taylor gives

$$
\phi(t) = \phi(0) + \frac{\phi^{(m)}(0)}{m!}t^m + o(t^m)
$$

so the sign of $\phi(t) - \phi(0)$ for small $t$ is the sign of $\phi^{(m)}(0)\,t^m$. Hence:

- $m$ **even** and $\phi^{(m)}(0) \gt 0$: strict local minimum;
- $m$ **even** and $\phi^{(m)}(0) \lt 0$: strict local maximum;
- $m$ **odd**: neither ($t^m$ changes sign), an inflection.

**Step 5 (the multivariate caveat).** Applying this rule direction by direction is *not* sufficient in
$\mathbb{R}^n$: Problem L3.1 exhibits a function passing the test along every line that still fails to have
a minimum. The correct multivariate statement requires the restriction to the Hessian null space to be
positive definite as a higher-order form, uniformly over curves, not merely over lines.

$$
\boxed{f_1 = x^2+y^4:\ \text{min}; \quad f_2 = x^2-y^4:\ \text{saddle}; \quad f_3 = x^2+y^3:\ \text{neither} — \text{decided by the first surviving order}}
$$

> **Key takeaway:** The eigenvalue $0$ is the "unknown" branch of the second-order test, and resolving it is genuinely a higher-order (and, in more than one dimension, genuinely non-linear) question.

```python
funcs = {
    "f1 = x^2 + y^4": lambda x, y: x**2 + y**4,
    "f2 = x^2 - y^4": lambda x, y: x**2 - y**4,
    "f3 = x^2 + y^3": lambda x, y: x**2 + y**3,
}
H = np.array([[2.0, 0.0], [0.0, 0.0]])
ys = np.array([-1e-2, 1e-2])
print("all three share grad = 0 and eig(Hessian) =", np.linalg.eigvalsh(H), "at the origin")
for name, g in funcs.items():
    print(f"  {name:16s} f(0,y) at y = -1e-02, +1e-02 -> ",
          "  ".join(f"{g(0.0, y):+.2e}" for y in ys))
assert funcs["f1 = x^2 + y^4"](0.0, ys[0]) > 0 and funcs["f1 = x^2 + y^4"](0.0, ys[1]) > 0
assert funcs["f2 = x^2 - y^4"](0.0, ys[1]) < 0
assert funcs["f3 = x^2 + y^3"](0.0, ys[0]) < 0 < funcs["f3 = x^2 + y^3"](0.0, ys[1])
print("identical second-order data, three different verdicts: minimum, saddle, neither")
```

### Problem L3.4 — The Singular Quadratic — Existence Requires $\mathbf{b} \in \operatorname{range}(A)$

**Problem Statement:** Let $f(\mathbf{x}) = \frac{1}{2}\mathbf{x}^\top A\mathbf{x} - \mathbf{b}^\top\mathbf{x}$
with $A$ symmetric positive **semi**definite and singular. Prove: $f$ is bounded below (and attains its
minimum) if and only if $\mathbf{b} \in \operatorname{range}(A)$, in which case the minimizer set is
$A^{+}\mathbf{b} + \ker(A)$ and the optimal value is $-\frac{1}{2}\mathbf{b}^\top A^{+}\mathbf{b}$.

*Intuition:* Along a null direction the quadratic term dies and only the linear term survives — a nonzero linear part on a flat direction is a slide to $-\infty$.

**Solution:**

**Step 1 (orthogonal decomposition).** Since $A = A^\top$, we have the orthogonal splitting
$\mathbb{R}^n = \operatorname{range}(A) \oplus \ker(A)$. Write $\mathbf{b} = \mathbf{b}_R + \mathbf{b}_N$
with $\mathbf{b}_R \in \operatorname{range}(A)$ and $\mathbf{b}_N \in \ker(A)$.

**Step 2 (necessity).** Suppose $\mathbf{b}_N \neq \mathbf{0}$ and take $\mathbf{x} = t\,\mathbf{b}_N$.
Then $A\mathbf{b}_N = \mathbf{0}$, so

$$
f(t\mathbf{b}_N) = \frac{t^2}{2}\mathbf{b}_N^\top A \mathbf{b}_N - t\,\mathbf{b}^\top\mathbf{b}_N = 0 - t\lVert \mathbf{b}_N\rVert^2 \xrightarrow[t\to+\infty]{} -\infty
$$

(using $\mathbf{b}^\top\mathbf{b}_N = \lVert \mathbf{b}_N\rVert^2$ by orthogonality). So $f$ is unbounded
below. Contrapositively, boundedness forces $\mathbf{b}_N = \mathbf{0}$, i.e.
$\mathbf{b} \in \operatorname{range}(A)$.

**Step 3 (sufficiency).** Assume $\mathbf{b} = A\bar{\mathbf{x}}$ for some $\bar{\mathbf{x}}$; take
$\bar{\mathbf{x}} = A^{+}\mathbf{b}$, the minimum-norm choice. For arbitrary
$\mathbf{x} = \bar{\mathbf{x}} + \mathbf{e}$,

$$
f(\mathbf{x}) = f(\bar{\mathbf{x}}) + \mathbf{e}^\top\left(A\bar{\mathbf{x}} - \mathbf{b}\right) + \frac{1}{2}\mathbf{e}^\top A\mathbf{e} = f(\bar{\mathbf{x}}) + \frac{1}{2}\mathbf{e}^\top A\mathbf{e} \ge f(\bar{\mathbf{x}})
$$

since $A \succeq 0$. So $\bar{\mathbf{x}}$ is a global minimizer and $f$ is bounded below.

**Step 4 (the minimizer set).** Equality above holds iff $\mathbf{e}^\top A\mathbf{e} = 0$. Writing
$A = \sum_i \lambda_i \mathbf{q}_i\mathbf{q}_i^\top$ with $\lambda_i \ge 0$, this forces
$\lambda_i (\mathbf{q}_i^\top\mathbf{e})^2 = 0$ for each $i$, i.e. $\mathbf{e} \perp \mathbf{q}_i$ whenever
$\lambda_i \gt 0$: exactly $\mathbf{e} \in \ker(A)$. Hence

$$
\arg\min f = A^{+}\mathbf{b} + \ker(A)
$$

an affine subspace of dimension $\dim\ker(A)$ — non-unique whenever $A$ is singular.

**Step 5 (optimal value).** Using $A A^{+}\mathbf{b} = \mathbf{b}$ (valid because
$\mathbf{b} \in \operatorname{range} A$) and the symmetry of $A^{+}$:

$$
f(A^{+}\mathbf{b}) = \frac{1}{2}(A^{+}\mathbf{b})^\top A A^{+}\mathbf{b} - \mathbf{b}^\top A^{+}\mathbf{b} = \frac{1}{2}\mathbf{b}^\top A^{+}\mathbf{b} - \mathbf{b}^\top A^{+}\mathbf{b} = -\frac{1}{2}\mathbf{b}^\top A^{+}\mathbf{b}
$$

$$
\boxed{\min f \gt -\infty \iff \mathbf{b} \in \operatorname{range}(A); \quad \arg\min f = A^{+}\mathbf{b} + \ker(A), \quad f^* = -\tfrac{1}{2}\mathbf{b}^\top A^{+}\mathbf{b}}
$$

> **Key takeaway:** In the semidefinite case existence is a *linear-algebra compatibility condition*, not a curvature condition: the flat directions must carry no linear force. This is precisely the consistency of the normal equations in rank-deficient least squares, where $\mathbf{b} = X^\top\mathbf{y}$ automatically lies in $\operatorname{range}(X^TX)$ and existence is therefore never at risk.

```python
Q, _ = np.linalg.qr(rng.standard_normal((4, 4)))
A = Q @ np.diag([3.0, 1.0, 0.0, 0.0]) @ Q.T
A = 0.5 * (A + A.T)
A_plus = np.linalg.pinv(A)
b_ok = A @ rng.standard_normal(4)                 # lies in range(A) by construction
b_bad = b_ok + 5.0 * Q[:, 3]                      # adds a kernel component
f = lambda x, rhs: 0.5 * x @ A @ x - rhs @ x

x_ok = A_plus @ b_ok
print("b in range(A):")
print("  ||A x* - b||      =", np.linalg.norm(A @ x_ok - b_ok))
print("  f(x*)             =", f(x_ok, b_ok))
print("  -b^T A^+ b / 2    =", -0.5 * b_ok @ A_plus @ b_ok)
assert np.linalg.norm(A @ x_ok - b_ok) < 1e-10
assert abs(f(x_ok, b_ok) + 0.5 * b_ok @ A_plus @ b_ok) < 1e-10

kern = Q[:, 2]
along = [f(x_ok + t * kern, b_ok) for t in [0.0, 1.0, 10.0]]
print("  f along x* + t ker(A), t = 0, 1, 10:", np.round(along, 10))
assert max(abs(v - along[0]) for v in along) < 1e-8

slide = [f(t * Q[:, 3], b_bad) for t in [1.0, 10.0, 100.0]]
print("b not in range(A):")
print("  f(t b_N) at t = 1, 10, 100:", np.round(slide, 6), " -> slides to -infinity")
assert slide[2] < slide[1] < slide[0] < 0
```
