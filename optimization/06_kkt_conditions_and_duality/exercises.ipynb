### Problem L0.1: What Complementary Slackness Really Says

**Problem Statement:** State the four KKT conditions for
$\min f(\mathbf{x})$ s.t. $g_i(\mathbf{x}) \le 0$, $h_j(\mathbf{x}) = 0$. Then prove that
$\mu_i^* g_i(\mathbf{x}^*) = 0$ forces $\mu_i^* = 0$ whenever $g_i(\mathbf{x}^*) \lt 0$, and explain the
consequence for support vector machines.

*Intuition:* A constraint you are not pressing against exerts no force, so its price is zero.

**Solution:**

**The KKT system.** With $\mathcal{L} = f + \sum_i\mu_ig_i + \sum_j\lambda_jh_j$:

1. **Stationarity:** $\nabla f(\mathbf{x}^*) + \sum_i \mu_i^*\nabla g_i(\mathbf{x}^*) + \sum_j\lambda_j^*\nabla h_j(\mathbf{x}^*) = \mathbf{0}$
2. **Primal feasibility:** $g_i(\mathbf{x}^*) \le 0$, $h_j(\mathbf{x}^*) = 0$
3. **Dual feasibility:** $\mu_i^* \ge 0$
4. **Complementary slackness:** $\mu_i^*\,g_i(\mathbf{x}^*) = 0$ for every $i$

**Step 1 (the elementary deduction).** Suppose $g_i(\mathbf{x}^*) \lt 0$ (the constraint is *inactive*).
Condition 4 says $\mu_i^*g_i(\mathbf{x}^*) = 0$; since the second factor is nonzero, $\mu_i^* = 0$.
Contrapositively, $\mu_i^* \gt 0 \Rightarrow g_i(\mathbf{x}^*) = 0$: a positively priced constraint must be
active.

**Step 2 (why it must be so).** If $g_i(\mathbf{x}^*) \lt 0$ then by continuity $g_i \lt 0$ on a whole
neighbourhood, so locally the problem is *identical* to the one with constraint $i$ deleted. A deleted
constraint contributes nothing to stationarity, i.e. its multiplier is $0$. Complementary slackness is the
algebraic encoding of this locality.

**Step 3 (the active set).** Define
$\mathcal{A}(\mathbf{x}^*) = \{i : g_i(\mathbf{x}^*) = 0\}$. Stationarity then collapses to

$$
\nabla f(\mathbf{x}^*) + \sum_{i\in\mathcal{A}}\mu_i^*\nabla g_i(\mathbf{x}^*) + \sum_j \lambda_j^*\nabla h_j(\mathbf{x}^*) = \mathbf{0}
$$

so the first-order conditions involve only the active constraints — the basis of active-set algorithms
(Problem L1.1).

**Step 4 (SVM consequence).** In the SVM dual the multiplier $\alpha_i$ of the margin constraint
$y_i(\mathbf{w}^T\mathbf{x}_i+b) \ge 1$ satisfies $\alpha_i\left[y_i(\mathbf{w}^T\mathbf{x}_i+b)-1\right] = 0$.
Points strictly outside the margin have $\alpha_i = 0$ and drop out of
$\mathbf{w} = \sum_i\alpha_iy_i\mathbf{x}_i$ entirely; only the points *on* the margin — the **support
vectors** — can carry weight. Deleting all non-support vectors leaves the solution unchanged.

$$
\boxed{g_i(\mathbf{x}^*) \lt 0 \implies \mu_i^* = 0; \quad \mu_i^* \gt 0 \implies g_i(\mathbf{x}^*) = 0}
$$

> **Key takeaway:** Complementary slackness is a switch that turns the combinatorial question "which constraints bind?" into an algebraic condition — and in machine learning it is precisely what makes SVM solutions sparse.

### Problem L0.2: Why Inequality Multipliers Have a Sign

**Problem Statement:** Explain, from the direction of feasible motion, why the multiplier of an inequality
$g(\mathbf{x}) \le 0$ must satisfy $\mu \ge 0$, while the multiplier of an equality $h(\mathbf{x}) = 0$ is
free in sign. Illustrate with $\min x$ subject to $x \ge 1$ and with the sensitivity interpretation.

*Intuition:* A one-sided constraint can only push you in one direction, so its price is one-sided too.

**Solution:**

**Step 1 (feasible directions).** At a point with $g(\mathbf{x}^*) = 0$, moving in direction $\mathbf{d}$
keeps $g \le 0$ to first order iff $\nabla g(\mathbf{x}^*)^T\mathbf{d} \le 0$ — a **half-space** of feasible
directions. For an equality constraint, feasibility requires $\nabla h^T\mathbf{d} = 0$ — a **subspace**.

**Step 2 (optimality on a half-space).** Optimality requires $\nabla f^T\mathbf{d} \ge 0$ for all feasible
$\mathbf{d}$. So we need

$$
\nabla g^T\mathbf{d} \le 0 \implies \nabla f^T\mathbf{d} \ge 0
$$

By Farkas' lemma this holds exactly when $-\nabla f = \mu\nabla g$ for some $\mu \ge 0$, i.e.
$\nabla f + \mu\nabla g = \mathbf{0}$ with $\mu \ge 0$. For an equality the feasible set of directions is a
subspace closed under negation, so $\nabla f^T\mathbf{d} \ge 0$ and $\nabla f^T(-\mathbf{d}) \ge 0$ force
*equality*, and $\lambda$ is unrestricted (Topic 05).

**Step 3 (the example).** $\min x$ s.t. $1 - x \le 0$. Here $g(x) = 1-x$, $\nabla f = 1$, $\nabla g = -1$:

$$
1 + \mu(-1) = 0 \ \Longrightarrow\ \mu^* = 1 \ge 0 \quad\checkmark
$$

with $x^* = 1$ (active). Had the sign condition been dropped, the *maximizer* side would also qualify: for
$\max x$ on the same set no finite solution exists, and the sign condition is exactly what excludes
"solutions" that press the wrong way against the boundary.

**Step 4 (sensitivity confirms the sign).** Perturb to $g(\mathbf{x}) \le u$. Then
$p^*(u)$ is **nonincreasing** in $u$ (relaxing a constraint can only help a minimization), and

$$
\frac{\partial p^*}{\partial u_i} = -\mu_i^* \le 0
$$

So $\mu_i^* \ge 0$ is precisely the statement "loosening an inequality never hurts". For an equality,
moving the level $b_j$ in either direction can help or hurt, so $\lambda_j^*$ is free.

$$
\boxed{\mu_i^* \ge 0 \ \text{(one-sided price)}, \qquad \lambda_j^* \in \mathbb{R} \ \text{(two-sided price)}}
$$

> **Key takeaway:** The sign condition is a monotonicity statement about the value function, not a convention — it says that a feasible region which grows cannot produce a worse optimum.

### Problem L0.3: Weak Duality in Two Lines, and Concavity of the Dual

**Problem Statement:** Prove that $g(\boldsymbol{\mu},\boldsymbol{\lambda}) = \inf_{\mathbf{x}}\mathcal{L}(\mathbf{x},\boldsymbol{\mu},\boldsymbol{\lambda})$
satisfies $g(\boldsymbol{\mu},\boldsymbol{\lambda}) \le p^*$ for all $\boldsymbol{\mu}\ge\mathbf{0}$, and
that $g$ is concave — both **without any convexity assumption on the primal**.

*Intuition:* A pointwise infimum of affine functions is concave, and evaluating that infimum at a feasible primal point can only lower the objective.

**Solution:**

**Step 1 (weak duality).** Let $\tilde{\mathbf{x}}$ be any primal-feasible point ($g_i(\tilde{\mathbf{x}}) \le 0$,
$h_j(\tilde{\mathbf{x}}) = 0$) and let $\boldsymbol{\mu}\ge\mathbf{0}$. Then

$$
\mathcal{L}(\tilde{\mathbf{x}},\boldsymbol{\mu},\boldsymbol{\lambda}) = f(\tilde{\mathbf{x}}) + \underbrace{\sum_i\mu_i g_i(\tilde{\mathbf{x}})}_{\le 0} + \underbrace{\sum_j\lambda_jh_j(\tilde{\mathbf{x}})}_{=0} \le f(\tilde{\mathbf{x}})
$$

Taking the infimum over *all* $\mathbf{x}$ on the left only decreases it:

$$
g(\boldsymbol{\mu},\boldsymbol{\lambda}) = \inf_{\mathbf{x}}\mathcal{L}(\mathbf{x},\boldsymbol{\mu},\boldsymbol{\lambda}) \le \mathcal{L}(\tilde{\mathbf{x}},\boldsymbol{\mu},\boldsymbol{\lambda}) \le f(\tilde{\mathbf{x}})
$$

Minimizing the right side over feasible $\tilde{\mathbf{x}}$ gives $g \le p^*$, hence

$$
d^* = \sup_{\boldsymbol{\mu}\ge\mathbf{0},\ \boldsymbol{\lambda}}g(\boldsymbol{\mu},\boldsymbol{\lambda}) \le p^*
$$

**Step 2 (concavity).** For each *fixed* $\mathbf{x}$, the map

$$
(\boldsymbol{\mu},\boldsymbol{\lambda}) \mapsto \mathcal{L}(\mathbf{x},\boldsymbol{\mu},\boldsymbol{\lambda}) = f(\mathbf{x}) + \boldsymbol{\mu}^T\mathbf{g}(\mathbf{x}) + \boldsymbol{\lambda}^T\mathbf{h}(\mathbf{x})
$$

is **affine** — the coefficients $\mathbf{g}(\mathbf{x}), \mathbf{h}(\mathbf{x})$ are constants. The dual
function is the pointwise infimum of this family of affine functions, and a pointwise infimum of concave
(here affine) functions is concave:

$$
g\left(\theta\mathbf{z}_1 + (1-\theta)\mathbf{z}_2\right) = \inf_{\mathbf{x}}\left[\theta\mathcal{L}(\mathbf{x},\mathbf{z}_1) + (1-\theta)\mathcal{L}(\mathbf{x},\mathbf{z}_2)\right] \ge \theta g(\mathbf{z}_1) + (1-\theta)g(\mathbf{z}_2)
$$

Note $g$ may take the value $-\infty$; its effective domain $\{g \gt -\infty\}$ is convex.

**Step 3 (consequences).** The dual problem $\max_{\boldsymbol{\mu}\ge\mathbf{0}}g$ is a **convex** problem
(maximizing a concave function over a convex set) *even when the primal is combinatorial or wildly
nonconvex*, and every dual-feasible point supplies a certified lower bound $g \le p^*$ — the basis of
branch-and-bound and of SDP relaxations.

$$
\boxed{d^* \le p^* \ \text{always}; \quad g \ \text{is concave for every primal problem}}
$$

> **Key takeaway:** Duality is not a convexity phenomenon — it is a *relaxation* phenomenon. Convexity plus Slater is only needed to make the relaxation tight.

### Problem L0.4: KKT Can Fail Without a Constraint Qualification

**Problem Statement:** Consider $\min x$ subject to $x^2 \le 0$ on $\mathbb{R}$. Find the optimum, show
that no KKT multiplier exists, identify the failing constraint qualification, and check whether Slater's
condition holds.

*Intuition:* A feasible region that is a single point, described by a tangentially degenerate inequality, has no interior for Slater and no independent gradient for LICQ.

**Solution:**

**Step 1 (the optimum).** $x^2 \le 0$ forces $x = 0$, so the feasible set is $\{0\}$ and
$p^* = 0$ at $x^* = 0$.

**Step 2 (no KKT multiplier).** With $g(x) = x^2$, $\nabla f = 1$ and $\nabla g(0) = 2x\big\vert_{0} = 0$.
Stationarity would require

$$
1 + \mu\cdot 0 = 0
$$

which fails for every $\mu \ge 0$. So the KKT system is **infeasible** at the (unique) optimizer.

**Step 3 (which qualification fails).**

- **LICQ:** the single active gradient is $\nabla g(0) = 0$, and $\{\mathbf{0}\}$ is linearly dependent —
  LICQ fails.
- **Slater:** requires a strictly feasible point, $g(x_0) \lt 0$, i.e. $x_0^2 \lt 0$ — impossible. Slater
  fails.

Both fail, as they must, since KKT holds at every optimizer under either one.

**Step 4 (what the dual says).** $\mathcal{L}(x,\mu) = x + \mu x^2$, so for $\mu \gt 0$,

$$
g(\mu) = \inf_x\left(x + \mu x^2\right) = -\frac{1}{4\mu}, \qquad g(0) = \inf_x x = -\infty
$$

Hence $d^* = \sup_{\mu \gt 0}\left(-\frac{1}{4\mu}\right) = 0 = p^*$: strong duality *holds* here by
accident (the sup is approached but not attained), yet no finite optimal multiplier exists, so KKT is still
unavailable. The dual optimum is attained only "at $\mu = \infty$".

**Step 5 (the repair).** Rewriting the same feasible set as the two inequalities $x \le 0$ and $-x \le 0$
gives gradients $1$ and $-1$; stationarity $1 + \mu_1 - \mu_2 = 0$ with $\mu_1 = 0$, $\mu_2 = 1 \ge 0$
solves the KKT system immediately.

$$
\boxed{p^* = 0 \ \text{but no } \mu \ge 0 \ \text{satisfies KKT; both LICQ and Slater fail}}
$$

> **Key takeaway:** KKT is "first-order optimality *plus* regularity". When a solver reports no multipliers, suspect the constraint description, not the solution.

## Level 1 — Foundation

### Problem L1.1: Solving a KKT System by Active-Set Enumeration

**Problem Statement:** Solve $\min\ x^2 + y^2$ subject to $x + y \ge 2$ and $x \ge 0$ by enumerating active
sets. Report the optimal point, the multipliers, and verify all four KKT conditions.

*Intuition:* With $m$ inequalities there are $2^m$ candidate active sets; for small $m$, enumerate, solve the resulting equality problem, and discard candidates violating feasibility or $\mu \ge 0$.

**Solution:**

**Step 0 (standard form).** $g_1(x,y) = 2 - x - y \le 0$, $g_2(x,y) = -x \le 0$, and

$$
\mathcal{L} = x^2+y^2 + \mu_1(2-x-y) + \mu_2(-x)
$$

$$
\partial_x\mathcal{L} = 2x - \mu_1 - \mu_2 = 0, \qquad \partial_y\mathcal{L} = 2y - \mu_1 = 0
$$

**Case A: $\mathcal{A} = \emptyset$** ($\mu_1 = \mu_2 = 0$). Then $x = y = 0$, but
$g_1(0,0) = 2 \gt 0$: **infeasible**. Reject.

**Case B: $\mathcal{A} = \{2\}$** ($\mu_1 = 0$, $x = 0$). Then $2y = 0$ gives $y = 0$, and again
$g_1 = 2 \gt 0$: **infeasible**. Reject.

**Case C: $\mathcal{A} = \{1\}$** ($\mu_2 = 0$, $x+y = 2$). Then $2x = \mu_1 = 2y$, so $x = y = 1$ and
$\mu_1 = 2$. Check: $\mu_1 = 2 \ge 0$ $\checkmark$; $g_2 = -1 \le 0$ $\checkmark$. **Accept**, with
$f = 2$.

**Case D: $\mathcal{A} = \{1,2\}$** ($x = 0$, $x+y=2$). Then $y = 2$, $\mu_1 = 2y = 4$, and
$\mu_2 = 2x - \mu_1 = -4 \lt 0$: **dual infeasible**. Reject. (Value $f = 4$, worse anyway.)

**Step 1 (verify KKT at the winner).** At $(x^*,y^*) = (1,1)$, $\mu^* = (2, 0)$:

1. Stationarity: $2(1) - 2 - 0 = 0$ and $2(1) - 2 = 0$ $\checkmark$
2. Primal feasibility: $2 - 1 - 1 = 0 \le 0$, $-1 \le 0$ $\checkmark$
3. Dual feasibility: $2 \ge 0$, $0 \ge 0$ $\checkmark$
4. Complementary slackness: $2\cdot 0 = 0$ and $0\cdot(-1) = 0$ $\checkmark$

**Step 2 (sufficiency).** The objective is strictly convex and both constraints are affine, so the problem
is convex and Slater holds (e.g. $(3,3)$ is strictly feasible). KKT is therefore **sufficient**: $(1,1)$ is
the unique global minimizer.

$$
\boxed{(x^*,y^*) = (1,1), \quad \boldsymbol{\mu}^* = (2,0), \quad p^* = 2}
$$

> **Key takeaway:** Active-set enumeration turns a KKT system into a short list of equality-constrained problems; the sign test $\mu \ge 0$ and the feasibility test are exactly what prune the list.

### Problem L1.2: Building a Dual Function from Scratch

**Problem Statement:** For $\min \frac12\lVert \mathbf{x}\rVert^2$ subject to $A\mathbf{x} = \mathbf{b}$
(with $A \in \mathbb{R}^{p\times n}$ of full row rank), compute the dual function $g(\boldsymbol{\lambda})$
in closed form, solve the dual, and verify $d^* = p^*$ explicitly.

*Intuition:* Minimizing the Lagrangian in $\mathbf{x}$ is an unconstrained quadratic problem, so the dual is again a quadratic — of the opposite sign.

**Solution:**

**Step 1 (minimize the Lagrangian in $\mathbf{x}$).**

$$
\mathcal{L}(\mathbf{x},\boldsymbol{\lambda}) = \tfrac12\lVert \mathbf{x}\rVert^2 + \boldsymbol{\lambda}^T\left(A\mathbf{x}-\mathbf{b}\right)
$$

$$
\nabla_{\mathbf{x}}\mathcal{L} = \mathbf{x} + A^T\boldsymbol{\lambda} = \mathbf{0} \ \Longrightarrow\ \mathbf{x}(\boldsymbol{\lambda}) = -A^T\boldsymbol{\lambda}
$$

(The Lagrangian is strongly convex in $\mathbf{x}$, so the infimum is attained.)

**Step 2 (dual function).** Substituting,

$$
g(\boldsymbol{\lambda}) = \tfrac12\lVert A^T\boldsymbol{\lambda}\rVert^2 - \boldsymbol{\lambda}^TAA^T\boldsymbol{\lambda} - \boldsymbol{\lambda}^T\mathbf{b} = -\tfrac12\boldsymbol{\lambda}^TAA^T\boldsymbol{\lambda} - \mathbf{b}^T\boldsymbol{\lambda}
$$

a concave quadratic (since $AA^T \succ 0$ by full row rank), as Problem L0.3 predicts.

**Step 3 (solve the dual).**

$$
\nabla g = -AA^T\boldsymbol{\lambda} - \mathbf{b} = \mathbf{0} \ \Longrightarrow\ \boldsymbol{\lambda}^* = -\left(AA^T\right)^{-1}\mathbf{b}
$$

$$
d^* = g(\boldsymbol{\lambda}^*) = -\tfrac12\mathbf{b}^T\left(AA^T\right)^{-1}\mathbf{b} + \mathbf{b}^T\left(AA^T\right)^{-1}\mathbf{b} = \tfrac12\mathbf{b}^T\left(AA^T\right)^{-1}\mathbf{b}
$$

**Step 4 (solve the primal directly).** The minimum-norm solution of $A\mathbf{x}=\mathbf{b}$ is
$\mathbf{x}^* = A^T(AA^T)^{-1}\mathbf{b}$, hence

$$
p^* = \tfrac12\lVert \mathbf{x}^*\rVert^2 = \tfrac12\mathbf{b}^T\left(AA^T\right)^{-1}AA^T\left(AA^T\right)^{-1}\mathbf{b} = \tfrac12\mathbf{b}^T\left(AA^T\right)^{-1}\mathbf{b} = d^*
$$

**Step 5 (recover the primal from the dual).**
$\mathbf{x}(\boldsymbol{\lambda}^*) = -A^T\boldsymbol{\lambda}^* = A^T(AA^T)^{-1}\mathbf{b} = \mathbf{x}^*$:
the dual solution reconstructs the primal solution through the Lagrangian minimizer — the pattern behind
every dual algorithm.

$$
\boxed{g(\boldsymbol{\lambda}) = -\tfrac12\boldsymbol{\lambda}^TAA^T\boldsymbol{\lambda} - \mathbf{b}^T\boldsymbol{\lambda}, \quad d^* = p^* = \tfrac12\mathbf{b}^T\left(AA^T\right)^{-1}\mathbf{b}}
$$

> **Key takeaway:** The dual trades $n$ variables for $p$ — a decisive win when $p \ll n$, which is exactly the situation in kernel methods and in minimum-norm interpolation problems.

### Problem L1.3: Complementary Slackness Is Equivalent to a Zero Gap

**Problem Statement:** Suppose $\mathbf{x}^*$ is primal optimal, $(\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*)$
is dual optimal, and $p^* = d^*$ (zero duality gap). Prove that $\mathbf{x}^*$ minimizes
$\mathcal{L}(\cdot,\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*)$ and that
$\mu_i^*g_i(\mathbf{x}^*) = 0$ for all $i$.

*Intuition:* The chain of inequalities in the weak-duality proof must collapse to equalities, and each collapsed inequality is a KKT condition.

**Solution:**

**Step 1 (write the chain).** Using dual optimality, the definition of $g$, and primal feasibility:

$$
f(\mathbf{x}^*) = p^* = d^* = g(\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*) = \inf_{\mathbf{x}}\left[f(\mathbf{x}) + \sum_i\mu_i^*g_i(\mathbf{x}) + \sum_j\lambda_j^*h_j(\mathbf{x})\right]
$$

$$
\le f(\mathbf{x}^*) + \sum_i\mu_i^*g_i(\mathbf{x}^*) + \sum_j\lambda_j^*h_j(\mathbf{x}^*) \le f(\mathbf{x}^*)
$$

where the last inequality uses $\mu_i^* \ge 0$, $g_i(\mathbf{x}^*)\le 0$ and $h_j(\mathbf{x}^*)=0$.

**Step 2 (both ends coincide).** The chain begins and ends at $f(\mathbf{x}^*)$, so **every** inequality in
it is an equality.

**Step 3 (first equality: $\mathbf{x}^*$ minimizes the Lagrangian).** Equality in the first inequality says

$$
\inf_{\mathbf{x}}\mathcal{L}(\mathbf{x},\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*) = \mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*)
$$

i.e. $\mathbf{x}^*$ attains the infimum. If $f, g_i$ are differentiable this yields the **stationarity**
condition $\nabla_{\mathbf{x}}\mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*) = \mathbf{0}$.

**Step 4 (second equality: complementary slackness).** Equality in the second inequality says

$$
\sum_i \mu_i^*g_i(\mathbf{x}^*) = 0
$$

Each summand satisfies $\mu_i^*g_i(\mathbf{x}^*) \le 0$ (nonnegative times nonpositive). A sum of
nonpositive numbers is zero only if each is zero, hence

$$
\mu_i^*\,g_i(\mathbf{x}^*) = 0 \qquad \text{for every } i
$$

**Step 5 (converse).** Conversely, if $\mathbf{x}^*$ is feasible, $\boldsymbol{\mu}^* \ge \mathbf{0}$,
$\mathbf{x}^*$ minimizes $\mathcal{L}(\cdot,\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*)$, and complementary
slackness holds, then reading the chain backwards gives
$g(\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*) = f(\mathbf{x}^*)$, so weak duality forces both to be optimal
with zero gap. This is the standard **optimality certificate**: a primal-dual pair with matching values
proves optimality of both.

$$
\boxed{p^* = d^* \iff \mathbf{x}^* \ \text{minimizes} \ \mathcal{L}(\cdot,\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*) \ \text{and} \ \mu_i^*g_i(\mathbf{x}^*) = 0 \ \forall i}
$$

> **Key takeaway:** Complementary slackness is not an extra axiom bolted onto KKT — it is exactly what a vanishing duality gap forces, which is why solvers monitor the gap $\sum_i\mu_ig_i$ as their convergence measure.

### Problem L1.4: Box Constraints Produce Clipping

**Problem Statement:** Solve $\min \frac12(x-a)^2$ subject to $0 \le x \le 1$ via KKT for arbitrary
$a \in \mathbb{R}$, reporting the solution and both multipliers in each regime. Generalize to the
projection onto a box in $\mathbb{R}^n$.

*Intuition:* The answer is "clip $a$ to the interval", and the multipliers measure how hard the clipping had to push.

**Solution:**

**Step 1 (KKT system).** With $g_1 = -x \le 0$, $g_2 = x - 1 \le 0$:

$$
\mathcal{L} = \tfrac12(x-a)^2 - \mu_1x + \mu_2(x-1), \qquad \partial_x\mathcal{L} = (x-a) - \mu_1 + \mu_2 = 0
$$

plus $\mu_1,\mu_2 \ge 0$, $\mu_1x = 0$, $\mu_2(x-1) = 0$, $0\le x\le 1$.

**Step 2 (three regimes).**

- **Interior, $0 \lt x \lt 1$:** complementary slackness forces $\mu_1 = \mu_2 = 0$, so $x = a$. Consistent
  iff $0 \lt a \lt 1$.
- **Lower boundary, $x = 0$:** then $\mu_2 = 0$ (since $x - 1 = -1 \neq 0$) and stationarity gives
  $\mu_1 = -a$. Dual feasibility $\mu_1 \ge 0$ requires $a \le 0$.
- **Upper boundary, $x = 1$:** then $\mu_1 = 0$ and $\mu_2 = a - 1 \ge 0$ requires $a \ge 1$.

The three regimes are exhaustive and mutually exclusive, so

$$
x^*(a) = \begin{cases} 0, & a \le 0\\ a, & 0 \lt a \lt 1 \\ 1, & a \ge 1\end{cases} = \min\left\{\max\{a, 0\}, 1\right\} = \operatorname{clip}(a, 0, 1)
$$

**Step 3 (reading the multipliers).** $\mu_1^* = \max\{-a, 0\}$ and $\mu_2^* = \max\{a-1,0\}$: each
multiplier is the **overshoot** of $a$ past the corresponding bound, i.e. the magnitude of the restoring
force the constraint applies. By sensitivity, raising the upper bound from $1$ to $1+u$ changes $p^*$ at
rate $-\mu_2^* = 1-a$ (negative when $a \gt 1$: a looser box helps).

**Step 4 (the $n$-dimensional box).** For $\min\frac12\lVert \mathbf{x}-\mathbf{a}\rVert^2$ subject to
$\boldsymbol{\ell} \le \mathbf{x} \le \mathbf{u}$, both the objective and the constraints separate across
coordinates, so the KKT system decouples into $n$ copies of the scalar problem:

$$
x_i^* = \operatorname{clip}(a_i, \ell_i, u_i)
$$

This is the projection $\Pi_{[\boldsymbol{\ell},\mathbf{u}]}(\mathbf{a})$ used in every projected-gradient
method and in weight clipping.

$$
\boxed{x^* = \operatorname{clip}(a,0,1), \quad \mu_1^* = \max\{-a,0\}, \quad \mu_2^* = \max\{a-1,0\}}
$$

> **Key takeaway:** Separable constraints give separable KKT systems, which is why box projections, soft-thresholding, and ReLU-like clipping all emerge as one-line closed forms rather than iterative solves.

### Problem L1.5: LP Duality on a Concrete Example

**Problem Statement:** For the LP $\max\ 3x_1 + 2x_2$ subject to $x_1+x_2 \le 4$, $x_1+3x_2\le 6$,
$\mathbf{x}\ge\mathbf{0}$: write the dual, solve both, verify $p^* = d^*$, and check every complementary
slackness relation.

*Intuition:* The dual prices each resource so that no product is worth producing at a loss, and the cheapest such pricing costs exactly the maximum revenue.

**Solution:**

**Step 1 (the dual).** For $\max \mathbf{c}^T\mathbf{x}$ s.t. $A\mathbf{x}\le\mathbf{b}$, $\mathbf{x}\ge 0$
the dual is $\min \mathbf{b}^T\mathbf{y}$ s.t. $A^T\mathbf{y}\ge\mathbf{c}$, $\mathbf{y}\ge 0$:

$$
\min\ 4y_1 + 6y_2 \quad \text{s.t.} \quad y_1 + y_2 \ge 3, \quad y_1 + 3y_2 \ge 2, \quad y_1, y_2 \ge 0
$$

**Step 2 (solve the primal by vertex enumeration).** Candidate vertices of the feasible polygon:

| Vertex | Feasible? | Objective $3x_1+2x_2$ |
|---|---|---|
| $(0,0)$ | yes | $0$ |
| $(4,0)$ | yes ($4\le4$, $4\le6$) | $12$ |
| $(3,1)$ (both tight) | yes | $11$ |
| $(0,2)$ | yes | $4$ |

So $p^* = 12$ at $\mathbf{x}^* = (4,0)$.

**Step 3 (solve the dual).** Try $\mathbf{y} = (3,0)$: constraints $3 \ge 3$ $\checkmark$, $3 \ge 2$
$\checkmark$, $\mathbf{y}\ge0$ $\checkmark$; objective $4(3) + 6(0) = 12$. Weak duality says every dual
feasible point is $\ge p^* = 12$, so $\mathbf{y}^* = (3,0)$ is dual optimal and

$$
d^* = 12 = p^* \quad \checkmark
$$

**Step 4 (complementary slackness, all four relations).**

- $y_1^* = 3 \gt 0 \Rightarrow$ first primal constraint tight: $4 + 0 = 4$ $\checkmark$
- Second primal constraint slack ($4 \lt 6$) $\Rightarrow y_2^* = 0$ $\checkmark$
- $x_1^* = 4 \gt 0 \Rightarrow$ first dual constraint tight: $y_1+y_2 = 3$ $\checkmark$
- Second dual constraint slack ($3 \gt 2$) $\Rightarrow x_2^* = 0$ $\checkmark$

**Step 5 (economic reading).** $y_1^* = 3$ is the shadow price of resource 1: one extra unit of $b_1$ raises
the optimal revenue by $3$ (indeed with $b_1 = 5$ the optimum becomes $(5,0)$ with value $15$, as long as
$b_2 = 6$ still permits it). Resource 2 is not scarce at the optimum, so it is free: $y_2^* = 0$.

$$
\boxed{\mathbf{x}^* = (4,0),\ \mathbf{y}^* = (3,0),\ p^* = d^* = 12}
$$

> **Key takeaway:** For LPs, strong duality is automatic whenever either problem is feasible and bounded — no Slater check needed — and complementary slackness pairs each primal variable with a dual constraint and vice versa.

### Problem L1.6: KKT Is Sufficient for Convex Problems

**Problem Statement:** Prove that for a convex problem ($f$ and all $g_i$ convex, $h_j$ affine) with
differentiable functions, any point satisfying the KKT conditions is a **global** minimizer — with no
constraint qualification required.

*Intuition:* Convexity makes the Lagrangian convex in $\mathbf{x}$ for $\boldsymbol{\mu}\ge\mathbf{0}$, and a stationary point of a convex function is its global minimum.

**Solution:**

**Step 1 (the Lagrangian is convex in $\mathbf{x}$).** For fixed $\boldsymbol{\mu}\ge\mathbf{0}$ and any
$\boldsymbol{\lambda}$,

$$
\mathcal{L}(\mathbf{x},\boldsymbol{\mu},\boldsymbol{\lambda}) = f(\mathbf{x}) + \sum_i\mu_ig_i(\mathbf{x}) + \sum_j\lambda_jh_j(\mathbf{x})
$$

is a **nonnegative** combination of convex functions ($f$, $\mu_ig_i$) plus affine functions
($\lambda_jh_j$), hence convex in $\mathbf{x}$.

**Step 2 (stationarity gives a global minimum of $\mathcal{L}$).** KKT condition 1 says
$\nabla_{\mathbf{x}}\mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*) = \mathbf{0}$. For a
differentiable convex function, a stationary point is a global minimizer, so for every $\mathbf{x}$,

$$
\mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*) \le \mathcal{L}(\mathbf{x},\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*)
$$

**Step 3 (evaluate the left side using complementary slackness).**

$$
\mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*) = f(\mathbf{x}^*) + \underbrace{\sum_i\mu_i^*g_i(\mathbf{x}^*)}_{=0 \ \text{(comp. slack.)}} + \underbrace{\sum_j\lambda_j^*h_j(\mathbf{x}^*)}_{=0 \ \text{(feasibility)}} = f(\mathbf{x}^*)
$$

**Step 4 (bound the right side at any feasible $\mathbf{x}$).** For feasible $\mathbf{x}$,
$g_i(\mathbf{x})\le0$ and $\mu_i^*\ge0$ give $\sum_i\mu_i^*g_i(\mathbf{x}) \le 0$, and
$\sum_j\lambda_j^*h_j(\mathbf{x}) = 0$, so

$$
\mathcal{L}(\mathbf{x},\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*) \le f(\mathbf{x})
$$

**Step 5 (combine).** Chaining Steps 2–4, for every feasible $\mathbf{x}$:

$$
f(\mathbf{x}^*) = \mathcal{L}(\mathbf{x}^*,\cdot) \le \mathcal{L}(\mathbf{x},\cdot) \le f(\mathbf{x})
$$

so $\mathbf{x}^*$ is globally optimal. Moreover $g(\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*) = f(\mathbf{x}^*)$,
so the gap is zero and $(\boldsymbol{\mu}^*,\boldsymbol{\lambda}^*)$ is dual optimal: a KKT point is a
complete primal-dual optimality certificate.

$$
\boxed{\text{convex problem} + \text{KKT point} \implies \text{global optimum and zero duality gap (no CQ needed)}}
$$

> **Key takeaway:** Constraint qualifications are needed for KKT to be *necessary*, never for it to be *sufficient* — so a solver that exhibits a KKT point on a convex problem has produced a proof, not a guess.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: Water-Filling Power Allocation

**Problem Statement:** Maximize $\sum_{i=1}^n \log\left(\alpha_i + p_i\right)$ subject to
$\sum_i p_i = P$, $p_i \ge 0$. Derive the water-filling solution from KKT, then solve the instance
$\boldsymbol{\alpha} = (0.2, 0.5, 1.5)$, $P = 1$.

*Intuition:* Pour a fixed volume of water into a tank whose floor heights are $\alpha_i$; the water level equalizes and the highest floors stay dry.

**Solution:**

**Step 1 (KKT).** Minimize $-\sum_i\log(\alpha_i+p_i)$ subject to $\sum p_i = P$ and $-p_i \le 0$:

$$
\mathcal{L} = -\sum_i\log(\alpha_i+p_i) + \nu\left(\sum_ip_i - P\right) - \sum_i\mu_ip_i
$$

$$
-\frac{1}{\alpha_i+p_i} + \nu - \mu_i = 0, \qquad \mu_i \ge 0, \quad \mu_ip_i = 0
$$

**Step 2 (case analysis per coordinate).**

- If $p_i \gt 0$ then $\mu_i = 0$ and $\alpha_i + p_i = 1/\nu$, i.e. $p_i = \frac{1}{\nu} - \alpha_i$.
- If $p_i = 0$ then $\mu_i = \nu - \frac{1}{\alpha_i} \ge 0$, i.e. $\alpha_i \ge \frac{1}{\nu}$.

Both cases are summarized by the **water-filling formula**

$$
p_i^*(\nu) = \max\left\{\frac{1}{\nu} - \alpha_i,\ 0\right\}, \qquad \text{with } \nu \gt 0 \text{ chosen so that} \sum_i p_i^*(\nu) = P
$$

Calling $w = 1/\nu$ the **water level**, $p_i = (w - \alpha_i)_+$: pour water to height $w$ over floors of
height $\alpha_i$.

**Step 3 (solve the instance).** Try all three channels active:

$$
3w - (0.2+0.5+1.5) = 1 \ \Longrightarrow\ w = \frac{3.2}{3} = 1.0667
$$

but then $p_3 = 1.0667 - 1.5 \lt 0$: infeasible, so channel 3 is dry. Retry with $\{1,2\}$ active:

$$
2w - (0.2+0.5) = 1 \ \Longrightarrow\ w = 0.85
$$

Check the dry channel: $w = 0.85 \le \alpha_3 = 1.5$ $\checkmark$ (its multiplier is
$\mu_3 = \nu - 1/\alpha_3 = 1.1765 - 0.6667 = 0.5098 \ge 0$).

$$
\mathbf{p}^* = (0.65,\ 0.35,\ 0), \qquad \nu^* = \frac{1}{0.85} \approx 1.1765
$$

**Step 4 (optimal value and sensitivity).**

$$
\sum_i\log(\alpha_i+p_i^*) = \log 0.85 + \log 0.85 + \log 1.5 \approx -0.1625 - 0.1625 + 0.4055 = 0.0805
$$

By sensitivity, one extra unit of total power raises the objective at rate $\nu^* \approx 1.18$ nats per
unit — the marginal value of power, which decreases as more channels come online.

$$
\boxed{p_i^* = \left(w - \alpha_i\right)_+ \ \text{with} \ \sum_i p_i^* = P; \ \text{here } w = 0.85, \ \mathbf{p}^* = (0.65, 0.35, 0)}
$$

> **Key takeaway:** Complementary slackness is what produces the "dry channel" case; the entire algorithm is a one-dimensional root-find on the water level, which is why water-filling is $O(n\log n)$ rather than a general convex solve.

### Problem L2.2: Projection onto the Probability Simplex

**Problem Statement:** Solve $\min \frac12\lVert \mathbf{x}-\mathbf{y}\rVert^2$ subject to
$\mathbf{1}^T\mathbf{x} = 1$, $\mathbf{x} \ge \mathbf{0}$. Derive the KKT solution
$x_i = (y_i - \tau)_+$ and compute it for $\mathbf{y} = (1.0, 0.2, -0.5)$.

*Intuition:* Shift every coordinate down by a common threshold and clip at zero, choosing the threshold so the result sums to one.

**Solution:**

**Step 1 (KKT system).**

$$
\mathcal{L} = \tfrac12\lVert \mathbf{x}-\mathbf{y}\rVert^2 + \tau\left(\mathbf{1}^T\mathbf{x}-1\right) - \boldsymbol{\mu}^T\mathbf{x}
$$

$$
x_i - y_i + \tau - \mu_i = 0, \qquad \mu_i \ge 0, \quad \mu_ix_i = 0, \quad x_i \ge 0, \quad \textstyle\sum_i x_i = 1
$$

**Step 2 (per-coordinate case analysis).**

- $x_i \gt 0 \Rightarrow \mu_i = 0 \Rightarrow x_i = y_i - \tau$ (requires $y_i \gt \tau$).
- $x_i = 0 \Rightarrow \mu_i = \tau - y_i \ge 0 \Rightarrow y_i \le \tau$.

Both cases combine into

$$
x_i^* = \left(y_i - \tau\right)_+ \qquad \text{with } \tau \text{ solving } \ \Phi(\tau) := \sum_i (y_i-\tau)_+ = 1
$$

$\Phi$ is continuous, piecewise linear, strictly decreasing where positive, with $\Phi(-\infty)=+\infty$
and $\Phi(\max_iy_i) = 0$, so $\tau$ exists and is unique.

**Step 3 (algorithm).** Sort $y_{(1)} \ge y_{(2)} \ge \dots \ge y_{(n)}$ and find the largest $k$ with

$$
y_{(k)} - \frac{1}{k}\left(\sum_{r\le k}y_{(r)} - 1\right) \gt 0, \qquad \text{then} \quad \tau = \frac{1}{k}\left(\sum_{r\le k}y_{(r)} - 1\right)
$$

Cost: $O(n\log n)$ for the sort (or $O(n)$ expected with a median-of-medians pivot).

**Step 4 (the instance $\mathbf{y} = (1.0,\ 0.2,\ -0.5)$).** Sorted: $1.0, 0.2, -0.5$.

- $k = 3$: $\tau = \frac{1.0+0.2-0.5-1}{3} = \frac{-0.3}{3} = -0.1$, but then $y_{(3)} - \tau = -0.4 \le 0$
  — reject.
- $k = 2$: $\tau = \frac{1.0+0.2-1}{2} = 0.1$, and $y_{(2)} - \tau = 0.1 \gt 0$ — **accept**.

$$
\mathbf{x}^* = \left((1.0-0.1)_+,\ (0.2-0.1)_+,\ (-0.5-0.1)_+\right) = (0.9,\ 0.1,\ 0)
$$

which sums to $1$ $\checkmark$. Multipliers: $\tau^* = 0.1$, $\mu_3^* = 0.1 - (-0.5) = 0.6 \ge 0$
$\checkmark$, $\mu_1^*=\mu_2^*=0$.

**Step 5 (contrast with softmax).** Simplex projection produces **sparse** outputs (coordinates exactly
zero) because the nonnegativity constraints are *active*; softmax, the entropy-regularized analogue, is
strictly positive everywhere. This is exactly the difference between sparsemax and softmax attention.

$$
\boxed{x_i^* = \left(y_i - \tau\right)_+, \ \textstyle\sum_i x_i^* = 1; \ \text{for } \mathbf{y}=(1,0.2,-0.5): \ \tau = 0.1, \ \mathbf{x}^* = (0.9, 0.1, 0)}
$$

> **Key takeaway:** The active set of the nonnegativity constraints *is* the support of the answer, so KKT case analysis and sparsity are the same phenomenon viewed from two sides.

### Problem L2.3: The Hard-Margin SVM Dual and the Kernel Trick

**Problem Statement:** Derive the dual of
$\min \frac12\lVert \mathbf{w}\rVert^2$ subject to $y_i(\mathbf{w}^T\mathbf{x}_i + b) \ge 1$,
$i = 1,\dots,m$. Identify the support vectors and explain why the dual admits kernels.

*Intuition:* Eliminating $\mathbf{w}$ from the Lagrangian leaves an objective containing the data only through inner products.

**Solution:**

**Step 1 (Lagrangian).** With $\alpha_i \ge 0$,

$$
\mathcal{L}(\mathbf{w},b,\boldsymbol{\alpha}) = \tfrac12\lVert \mathbf{w}\rVert^2 - \sum_{i=1}^m\alpha_i\left[y_i\left(\mathbf{w}^T\mathbf{x}_i+b\right)-1\right]
$$

**Step 2 (minimize over the primal variables).**

$$
\nabla_{\mathbf{w}}\mathcal{L} = \mathbf{w} - \sum_i\alpha_iy_i\mathbf{x}_i = \mathbf{0} \ \Longrightarrow\ \mathbf{w} = \sum_i\alpha_iy_i\mathbf{x}_i
$$

$$
\frac{\partial\mathcal{L}}{\partial b} = -\sum_i\alpha_iy_i = 0
$$

The second is not a stationary point in $b$ but a *feasibility condition on the dual domain*: if
$\sum_i\alpha_iy_i \neq 0$ the Lagrangian is unbounded below in $b$ and $g = -\infty$.

**Step 3 (substitute).**

$$
g(\boldsymbol{\alpha}) = \tfrac12\left\lVert \sum_i\alpha_iy_i\mathbf{x}_i\right\rVert^2 - \sum_{i}\alpha_iy_i\mathbf{x}_i^T\!\!\left(\sum_j\alpha_jy_j\mathbf{x}_j\right) + \sum_i\alpha_i = \sum_i\alpha_i - \frac12\sum_{i,j}\alpha_i\alpha_jy_iy_j\,\mathbf{x}_i^T\mathbf{x}_j
$$

giving the **dual problem**

$$
\max_{\boldsymbol{\alpha}\ge\mathbf{0}} \ \sum_i\alpha_i - \frac12\sum_{i,j}\alpha_i\alpha_jy_iy_j\,\mathbf{x}_i^T\mathbf{x}_j \quad \text{s.t.} \quad \sum_i\alpha_iy_i = 0
$$

a concave QP in $m$ variables (independent of the feature dimension $n$).

**Step 4 (support vectors).** Complementary slackness gives

$$
\alpha_i\left[y_i\left(\mathbf{w}^T\mathbf{x}_i+b\right)-1\right] = 0
$$

so $\alpha_i \gt 0$ only for points **on** the margin, $y_i(\mathbf{w}^T\mathbf{x}_i+b) = 1$. Thus
$\mathbf{w} = \sum_{i \in SV}\alpha_iy_i\mathbf{x}_i$ depends only on those points, and $b$ is recovered
from any one of them: $b = y_i - \mathbf{w}^T\mathbf{x}_i$ (using $y_i^2 = 1$).

**Step 5 (kernels).** The data enter the dual *only* through the Gram matrix
$K_{ij} = \mathbf{x}_i^T\mathbf{x}_j$, and the classifier is

$$
f(\mathbf{x}) = \sum_{i\in SV}\alpha_iy_i\,\mathbf{x}_i^T\mathbf{x} + b
$$

which also uses only inner products. Replacing $\mathbf{x}_i^T\mathbf{x}_j$ by any positive semidefinite
kernel $k(\mathbf{x}_i,\mathbf{x}_j) = \langle\phi(\mathbf{x}_i),\phi(\mathbf{x}_j)\rangle$ trains a linear
SVM in a (possibly infinite-dimensional) feature space at a cost that depends only on $m$ — the **kernel
trick**. Strong duality holds because the problem is convex with affine constraints (Slater in the refined
form for affine inequalities).

$$
\boxed{\max_{\boldsymbol{\alpha}\ge\mathbf{0},\ \boldsymbol{\alpha}^T\mathbf{y}=0}\ \mathbf{1}^T\boldsymbol{\alpha} - \tfrac12\boldsymbol{\alpha}^T\!\left(\mathbf{y}\mathbf{y}^T\odot K\right)\boldsymbol{\alpha}, \qquad \mathbf{w} = \sum_{i\in SV}\alpha_iy_i\mathbf{x}_i}
$$

> **Key takeaway:** Duality converted a problem in $n$ features into a problem in $m$ examples whose data dependence is entirely through inner products — the single algebraic fact that made kernel methods possible.

### Problem L2.4: Soft-Margin SVM — The Support-Vector Taxonomy

**Problem Statement:** For the soft-margin SVM
$\min \frac12\lVert \mathbf{w}\rVert^2 + C\sum_i\xi_i$ s.t. $y_i(\mathbf{w}^T\mathbf{x}_i+b)\ge 1-\xi_i$,
$\xi_i \ge 0$, derive the box constraint $0 \le \alpha_i \le C$ and classify each training point by the
value of $\alpha_i$.

*Intuition:* The slack multiplier competes with the margin multiplier for the same budget $C$, and their complementarity produces a three-way classification of the data.

**Solution:**

**Step 1 (Lagrangian).** With $\alpha_i \ge 0$ for the margin constraints and $\beta_i \ge 0$ for
$\xi_i \ge 0$:

$$
\mathcal{L} = \tfrac12\lVert \mathbf{w}\rVert^2 + C\sum_i\xi_i - \sum_i\alpha_i\left[y_i(\mathbf{w}^T\mathbf{x}_i+b)-1+\xi_i\right] - \sum_i\beta_i\xi_i
$$

**Step 2 (stationarity).**

$$
\mathbf{w} = \sum_i\alpha_iy_i\mathbf{x}_i, \qquad \sum_i\alpha_iy_i = 0, \qquad \frac{\partial\mathcal{L}}{\partial\xi_i} = C - \alpha_i - \beta_i = 0
$$

**Step 3 (the box constraint).** From $\beta_i = C - \alpha_i$ and $\beta_i \ge 0$, $\alpha_i \ge 0$:

$$
0 \le \alpha_i \le C
$$

Substituting into $\mathcal{L}$ gives *exactly the same dual objective* as the hard-margin case; the entire
effect of the slack variables is to cap the multipliers:

$$
\max_{\boldsymbol{\alpha}}\ \sum_i\alpha_i - \frac12\sum_{i,j}\alpha_i\alpha_jy_iy_jK_{ij} \quad \text{s.t.}\quad \sum_i\alpha_iy_i = 0,\ \ 0 \le \alpha_i \le C
$$

**Step 4 (the taxonomy).** Complementary slackness reads
$\alpha_i\left[y_if(\mathbf{x}_i)-1+\xi_i\right] = 0$ and $\beta_i\xi_i = (C-\alpha_i)\xi_i = 0$:

| $\alpha_i$ | $\beta_i$ | forced $\xi_i$ | margin status | name |
|---|---|---|---|---|
| $\alpha_i = 0$ | $C$ | $\xi_i = 0$ | $y_if(\mathbf{x}_i) \ge 1$ | non-support vector (correct, outside margin) |
| $0 \lt \alpha_i \lt C$ | $\gt 0$ | $\xi_i = 0$ | $y_if(\mathbf{x}_i) = 1$ | **free** support vector, exactly on the margin |
| $\alpha_i = C$ | $0$ | $\xi_i \ge 0$ | $y_if(\mathbf{x}_i) \le 1$ | **bounded** support vector: inside margin or misclassified |

**Step 5 (practical consequences).** The bias $b$ is computed from any *free* support vector via
$b = y_i - \mathbf{w}^T\mathbf{x}_i$, since those are the points with $\xi_i = 0$ and an equality
constraint. The parameter $C$ is a hard cap on any single example's influence — which is exactly why the
soft-margin SVM is robust to outliers, whereas the hard-margin version ($C = \infty$) lets one mislabeled
point dominate. As $C \to \infty$ the box disappears and the hard-margin dual is recovered.

$$
\boxed{0 \le \alpha_i \le C \ \text{with} \ \beta_i = C - \alpha_i; \ \ \alpha_i = C \iff \xi_i \ \text{may be positive}}
$$

> **Key takeaway:** The regularization parameter $C$ enters the dual purely as an upper bound on the multipliers — regularization *is* a bound on shadow prices, capping how much any single constraint can be worth.

### Problem L2.5: The Ridge Dual and the Representer Theorem

**Problem Statement:** Derive the dual of
$\min_{\mathbf{w}} \frac{\lambda}{2}\lVert \mathbf{w}\rVert^2 + \frac12\lVert X\mathbf{w}-\mathbf{y}\rVert^2$
by introducing the residual as a constrained variable. Show $\mathbf{w}^* = X^T\boldsymbol{\alpha}^*$ and
compare the primal and dual solve costs.

*Intuition:* The optimal weight vector must lie in the span of the data — a fact that duality makes obvious and that kernels then exploit.

**Solution:**

**Step 1 (constrained reformulation).**

$$
\min_{\mathbf{w},\mathbf{r}}\ \frac{\lambda}{2}\lVert \mathbf{w}\rVert^2 + \frac12\lVert \mathbf{r}\rVert^2 \quad \text{s.t.}\quad \mathbf{r} = \mathbf{y} - X\mathbf{w}
$$

$$
\mathcal{L} = \frac{\lambda}{2}\lVert \mathbf{w}\rVert^2 + \frac12\lVert \mathbf{r}\rVert^2 + \boldsymbol{\alpha}^T\left(\mathbf{y}-X\mathbf{w}-\mathbf{r}\right)
$$

**Step 2 (minimize over $\mathbf{w}$ and $\mathbf{r}$).**

$$
\lambda\mathbf{w} - X^T\boldsymbol{\alpha} = \mathbf{0} \ \Longrightarrow\ \mathbf{w} = \frac{1}{\lambda}X^T\boldsymbol{\alpha}, \qquad \mathbf{r} - \boldsymbol{\alpha} = \mathbf{0} \ \Longrightarrow\ \mathbf{r} = \boldsymbol{\alpha}
$$

The first equation is already the **representer theorem**: $\mathbf{w}^* \in \operatorname{range}(X^T)$, a
linear combination of the training inputs.

**Step 3 (dual function).** Substituting,

$$
g(\boldsymbol{\alpha}) = \frac{1}{2\lambda}\lVert X^T\boldsymbol{\alpha}\rVert^2 + \frac12\lVert \boldsymbol{\alpha}\rVert^2 + \boldsymbol{\alpha}^T\mathbf{y} - \frac{1}{\lambda}\boldsymbol{\alpha}^TXX^T\boldsymbol{\alpha} - \lVert \boldsymbol{\alpha}\rVert^2
$$

$$
= -\frac{1}{2\lambda}\boldsymbol{\alpha}^TXX^T\boldsymbol{\alpha} - \frac12\lVert \boldsymbol{\alpha}\rVert^2 + \mathbf{y}^T\boldsymbol{\alpha}
$$

a concave quadratic in $\boldsymbol{\alpha} \in \mathbb{R}^m$, unconstrained.

**Step 4 (solve the dual).**

$$
\nabla g = -\frac{1}{\lambda}XX^T\boldsymbol{\alpha} - \boldsymbol{\alpha} + \mathbf{y} = \mathbf{0} \ \Longrightarrow\ \boldsymbol{\alpha}^* = \lambda\left(XX^T + \lambda I_m\right)^{-1}\mathbf{y}
$$

and therefore

$$
\mathbf{w}^* = \frac{1}{\lambda}X^T\boldsymbol{\alpha}^* = X^T\left(XX^T+\lambda I_m\right)^{-1}\mathbf{y}
$$

**Step 5 (consistency and cost).** The push-through identity
$X^T(XX^T+\lambda I_m)^{-1} = (X^TX+\lambda I_n)^{-1}X^T$ shows this equals the familiar primal ridge
solution — the two forms are algebraically identical. The costs are not:

| Route | System size | Cost |
|---|---|---|
| Primal $(X^TX+\lambda I_n)$ | $n\times n$ | $O(mn^2 + n^3)$ |
| Dual $(XX^T+\lambda I_m)$ | $m\times m$ | $O(m^2n + m^3)$ |

so the dual wins whenever $m \ll n$ (few examples, many features) — and, crucially, $XX^T$ can be replaced
by a kernel matrix $K$, giving **kernel ridge regression**
$f(\mathbf{x}) = \mathbf{k}(\mathbf{x})^T(K+\lambda I)^{-1}\mathbf{y}$, which is exactly Gaussian-process
posterior mean prediction.

$$
\boxed{\mathbf{w}^* = X^T\boldsymbol{\alpha}^*, \qquad \boldsymbol{\alpha}^* = \lambda\left(XX^T+\lambda I\right)^{-1}\mathbf{y}}
$$

> **Key takeaway:** The representer theorem is a *stationarity condition*, not a deep theorem: $\nabla_{\mathbf{w}}\mathcal{L} = 0$ literally says the weights are a combination of data vectors, and everything kernelized follows.

### Problem L2.6: Contact Mechanics as a Complementarity Problem

**Problem Statement:** A block of mass $m$ rests on a rigid table. Let $u \ge 0$ be the gap between block
and table (the non-penetration constraint is $-u \le 0$) and let $N$ be the normal contact force. Write the
static problem as a minimization of potential energy subject to non-penetration, and show that KKT produces
the physical complementarity law $N \ge 0$, $u \ge 0$, $Nu = 0$.

*Intuition:* A rigid contact either touches and pushes, or separates and exerts nothing — never both, never neither.

**Solution:**

**Step 1 (energy formulation).** Let the block's height above the table be $u$. Its potential energy is
$V(u) = mgu$, and rigidity forbids penetration:

$$
\min_{u}\ mgu \quad \text{subject to} \quad -u \le 0
$$

**Step 2 (KKT).** With $\mathcal{L} = mgu - Nu$ ($N \ge 0$ the multiplier of $-u \le 0$):

1. Stationarity: $\dfrac{d\mathcal{L}}{du} = mg - N = 0 \Rightarrow N = mg$
2. Primal feasibility: $u \ge 0$
3. Dual feasibility: $N \ge 0$
4. Complementary slackness: $N u = 0$

From 1 and 4, $N = mg \gt 0$ forces $u = 0$: the block rests **on** the table, with normal force exactly
$mg$ balancing gravity.

**Step 3 (the general law).** For a system with several potential contacts and gaps $u_i$, the same
derivation gives the **linear complementarity problem** (LCP)

$$
\mathbf{u} \ge \mathbf{0}, \qquad \mathbf{N} \ge \mathbf{0}, \qquad \mathbf{u}^T\mathbf{N} = 0
$$

usually written $0 \le \mathbf{u} \perp \mathbf{N} \ge 0$. This is exactly the Signorini contact condition
of continuum mechanics — and it *is* complementary slackness.

**Step 4 (why it must be complementary).** Physically: a contact force can only push (never pull), so
$N \ge 0$; bodies cannot interpenetrate, so $u \ge 0$; and a force can only be transmitted through actual
contact, so $N \gt 0 \Rightarrow u = 0$. Each of the three is a KKT condition, in the same order.

**Step 5 (sensitivity).** Perturb the constraint to $-u \le \epsilon$ (allow penetration depth $\epsilon$).
Then $p^*(\epsilon) = -mg\epsilon$ and

$$
\frac{dp^*}{d\epsilon} = -mg = -N \quad\checkmark
$$

matching $\partial p^*/\partial u_i = -\mu_i^*$: the multiplier is the force, and the sensitivity identity
is the statement that force times displacement is work.

$$
\boxed{0 \le u \ \perp\ N \ge 0, \quad N = mg, \quad u = 0}
$$

> **Key takeaway:** Contact dynamics simulators solve LCPs at every timestep — those solvers are KKT solvers, and "complementary slackness" and "no pulling without touching" are the same sentence in two languages.

## Level 3 — Challenge

### Problem L3.1: Slater's Condition Implies Strong Duality

**Problem Statement:** For a convex problem $\min f(\mathbf{x})$ s.t. $g_i(\mathbf{x}) \le 0$ with $f, g_i$
convex, suppose there is $\tilde{\mathbf{x}}$ with $g_i(\tilde{\mathbf{x}}) \lt 0$ for all $i$ (Slater) and
$p^*$ finite. Prove $d^* = p^*$ using a separating-hyperplane argument.

*Intuition:* The set of achievable (constraint level, objective value) pairs is convex; a supporting hyperplane at the optimum has a nonzero objective coefficient exactly because Slater provides interior room.

**Solution:**

**Step 1 (the value set).** Define

$$
\mathcal{A} = \left\{(\mathbf{u}, t) \in \mathbb{R}^m\times\mathbb{R} : \exists\,\mathbf{x} \ \text{with}\ g_i(\mathbf{x}) \le u_i \ \forall i,\ f(\mathbf{x}) \le t\right\}
$$

$\mathcal{A}$ is **convex**: if $(\mathbf{u}^1,t^1)$ and $(\mathbf{u}^2,t^2)$ are achieved by
$\mathbf{x}^1,\mathbf{x}^2$, convexity of $f$ and each $g_i$ makes
$\theta\mathbf{x}^1+(1-\theta)\mathbf{x}^2$ achieve the convex combination. By construction
$p^* = \inf\{t : (\mathbf{0},t) \in \mathcal{A}\}$.

**Step 2 (separate).** The point $(\mathbf{0}, p^*)$ is not in the interior of $\mathcal{A}$ (nothing in
$\mathcal{A}$ has $\mathbf{u} = \mathbf{0}$ and $t \lt p^*$). By the supporting hyperplane theorem for
convex sets there exists $(\boldsymbol{\mu}, \mu_0) \neq \mathbf{0}$ with

$$
\boldsymbol{\mu}^T\mathbf{u} + \mu_0 t \ \ge\ \mu_0 p^* \qquad \text{for all } (\mathbf{u},t) \in \mathcal{A}
$$

**Step 3 (signs).** $\mathcal{A}$ is upward/rightward closed (increasing any $u_i$ or $t$ stays in
$\mathcal{A}$), so letting $u_i \to \infty$ or $t\to\infty$ forces $\mu_i \ge 0$ and $\mu_0 \ge 0$.

**Step 4 (Slater rules out $\mu_0 = 0$).** Suppose $\mu_0 = 0$. Applying Step 2 at the point of
$\mathcal{A}$ generated by the Slater point $\tilde{\mathbf{x}}$, i.e.
$(\mathbf{g}(\tilde{\mathbf{x}}), f(\tilde{\mathbf{x}}))$, gives

$$
\sum_i\mu_i\,g_i(\tilde{\mathbf{x}}) \ge 0
$$

But every $g_i(\tilde{\mathbf{x}}) \lt 0$ and every $\mu_i \ge 0$, so the sum is $\le 0$, and it can equal
$0$ only if $\boldsymbol{\mu} = \mathbf{0}$. Together with $\mu_0 = 0$ this contradicts
$(\boldsymbol{\mu},\mu_0)\neq\mathbf{0}$. Hence $\mu_0 \gt 0$, and we may normalize $\mu_0 = 1$.

**Step 5 (identify the dual value).** With $\mu_0 = 1$, applying Step 2 to the pair generated by an
arbitrary $\mathbf{x}$ gives

$$
f(\mathbf{x}) + \sum_i\mu_i g_i(\mathbf{x}) \ge p^* \qquad \text{for all } \mathbf{x}
$$

Taking the infimum over $\mathbf{x}$ yields $g(\boldsymbol{\mu}) \ge p^*$. Combined with weak duality
$g(\boldsymbol{\mu}) \le d^* \le p^*$, all are equal:

$$
d^* = p^*, \quad \text{and the supremum is attained at } \boldsymbol{\mu}^* = \boldsymbol{\mu}
$$

$$
\boxed{\text{convex} + \text{Slater} \implies d^* = p^* \ \text{with the dual optimum attained}}
$$

> **Key takeaway:** Slater's role is entirely to force $\mu_0 \gt 0$ in the supporting hyperplane — i.e. to prevent a *vertical* separating hyperplane, which would carry information about the constraints only and none about the objective (the abnormal case of Fritz John).

### Problem L3.2: The Dual of the Dual of an LP Is the Primal

**Problem Statement:** For the LP $p^* = \min\{\mathbf{c}^T\mathbf{x} : A\mathbf{x}\ge\mathbf{b},\ \mathbf{x}\ge\mathbf{0}\}$,
derive the dual via the Lagrangian, then dualize the dual and show the original problem is recovered.

*Intuition:* Dualization is an involution on LPs; taking it twice returns you to the start, which is why LP duality is a symmetric statement rather than a hierarchy.

**Solution:**

**Step 1 (Lagrangian of the primal).** Dualize only the inequality $A\mathbf{x}\ge\mathbf{b}$ (i.e.
$\mathbf{b}-A\mathbf{x} \le \mathbf{0}$), keeping $\mathbf{x}\ge\mathbf{0}$ as an explicit domain:

$$
\mathcal{L}(\mathbf{x},\mathbf{y}) = \mathbf{c}^T\mathbf{x} + \mathbf{y}^T\left(\mathbf{b}-A\mathbf{x}\right) = \mathbf{b}^T\mathbf{y} + \left(\mathbf{c}-A^T\mathbf{y}\right)^T\mathbf{x}, \qquad \mathbf{y}\ge\mathbf{0}
$$

**Step 2 (dual function).** Minimizing over $\mathbf{x}\ge\mathbf{0}$, a linear function is bounded below on
the nonnegative orthant iff its coefficient vector is nonnegative, and then the infimum is $0$ at
$\mathbf{x}=\mathbf{0}$:

$$
g(\mathbf{y}) = \begin{cases} \mathbf{b}^T\mathbf{y}, & A^T\mathbf{y} \le \mathbf{c} \\ -\infty, & \text{otherwise}\end{cases}
$$

Hence the **dual LP** is

$$
d^* = \max\left\{\mathbf{b}^T\mathbf{y} \ : \ A^T\mathbf{y}\le\mathbf{c},\ \mathbf{y}\ge\mathbf{0}\right\}
$$

**Step 3 (put the dual in primal form).** Write it as a minimization:

$$
-d^* = \min\left\{(-\mathbf{b})^T\mathbf{y} \ : \ \left(-A^T\right)\mathbf{y}\ \ge\ -\mathbf{c},\ \mathbf{y}\ge\mathbf{0}\right\}
$$

This has exactly the template of Step 1 with the substitutions
$\mathbf{c}\to-\mathbf{b}$, $A \to -A^T$, $\mathbf{b}\to-\mathbf{c}$.

**Step 4 (apply the rule again).** The dual of that template is
$\max\{\mathbf{b}^{\prime T}\mathbf{z} : A^{\prime T}\mathbf{z}\le\mathbf{c}',\ \mathbf{z}\ge\mathbf{0}\}$
with the substituted data:

$$
\max\left\{(-\mathbf{c})^T\mathbf{z} \ : \ \left(-A^T\right)^T\mathbf{z} \le -\mathbf{b},\ \mathbf{z}\ge\mathbf{0}\right\} = \max\left\{-\mathbf{c}^T\mathbf{z} \ : \ A\mathbf{z}\ge\mathbf{b},\ \mathbf{z}\ge\mathbf{0}\right\}
$$

**Step 5 (recover the primal).** Negating,

$$
-\max\left\{-\mathbf{c}^T\mathbf{z}\right\} = \min\left\{\mathbf{c}^T\mathbf{z} : A\mathbf{z}\ge\mathbf{b},\ \mathbf{z}\ge\mathbf{0}\right\} = p^*
$$

the original problem, with $\mathbf{z}$ renamed $\mathbf{x}$.

$$
\boxed{\text{dual}(\text{dual}(\text{LP})) = \text{LP}: \ \min\{\mathbf{c}^T\mathbf{x}: A\mathbf{x}\ge\mathbf{b},\mathbf{x}\ge0\} \ \longleftrightarrow\ \max\{\mathbf{b}^T\mathbf{y}: A^T\mathbf{y}\le\mathbf{c},\mathbf{y}\ge0\}}
$$

> **Key takeaway:** The involution property means "primal" and "dual" are labels, not roles: a simplex code solving the dual is solving the primal, and the strong duality theorem for LPs is a statement about a symmetric pair.

### Problem L3.3: A Convex Problem With a Strictly Positive Duality Gap

**Problem Statement:** Consider $\min\ e^{-x}$ subject to $\frac{x^2}{y} \le 0$, over the domain
$\mathcal{D} = \{(x,y) : y \gt 0\}$. Show the problem is convex, compute $p^*$ and $d^*$, and exhibit the
gap. Which hypothesis of the strong-duality theorem fails?

*Intuition:* The feasible set is a boundary sliver with empty interior — Slater has nowhere to stand.

**Solution:**

**Step 1 (convexity).** $f(x,y) = e^{-x}$ is convex. The quadratic-over-linear function
$g(x,y) = x^2/y$ is convex on $y \gt 0$ (its Hessian is
$\frac{2}{y^3}\begin{bmatrix} y^2 & -xy \\ -xy & x^2\end{bmatrix} \succeq 0$, being
$\frac{2}{y^3}(y,-x)^T(y,-x)$). So this is a genuine convex program.

**Step 2 (the primal value).** On the domain $y \gt 0$, $x^2/y \le 0$ forces $x^2 \le 0$, i.e. $x = 0$.
Every $y \gt 0$ is then feasible, and

$$
p^* = e^{-0} = 1
$$

**Step 3 (the dual value).** For $\mu \ge 0$,

$$
g(\mu) = \inf_{y \gt 0,\ x}\left(e^{-x} + \mu\frac{x^2}{y}\right)
$$

Fix any $x$ and let $y \to \infty$: the second term $\to 0$ (for $\mu \ge 0$), so
$g(\mu) \le \inf_x e^{-x} = 0$. And the expression is nonnegative, so $g(\mu) = 0$ for every
$\mu \ge 0$. Hence

$$
d^* = \sup_{\mu\ge0}g(\mu) = 0
$$

**Step 4 (the gap).**

$$
p^* - d^* = 1 - 0 = 1 \gt 0
$$

a strictly positive duality gap in a convex problem.

**Step 5 (which hypothesis fails).** Slater requires a point in the domain with $g(x,y) \lt 0$, i.e.
$x^2/y \lt 0$ with $y \gt 0$ — impossible, since $x^2/y \ge 0$ always. The feasible set
$\{(0,y) : y \gt 0\}$ has **empty relative interior with respect to the constraint**: the constraint can be
satisfied only with equality. Convexity alone therefore does *not* give strong duality; the geometric
reading is that the value set $\mathcal{A}$ of Problem L3.1 admits only a *vertical* supporting hyperplane
at $(\mathbf{0}, p^*)$, i.e. $\mu_0 = 0$.

**Step 6 (the repair).** Replacing the constraint by the equivalent $x = 0$ (same feasible set, now
described by an affine equality) restores strong duality trivially: the refined Slater condition allows
affine constraints to hold with equality.

$$
\boxed{p^* = 1, \quad d^* = 0, \quad \text{gap} = 1; \ \text{Slater fails (no strictly feasible point)}}
$$

> **Key takeaway:** Strong duality is not a corollary of convexity — it needs a *qualification*, and this example is the standard reminder that a convex problem whose constraints can only be met with equality may have a gap.

### Problem L3.4: Saddle Points and the Minimax Equality

**Problem Statement:** Prove that $(\mathbf{x}^*,\boldsymbol{\mu}^*)$ with $\boldsymbol{\mu}^*\ge\mathbf{0}$
is a saddle point of the Lagrangian — meaning
$\mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}) \le \mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*) \le \mathcal{L}(\mathbf{x},\boldsymbol{\mu}^*)$
for all $\mathbf{x}$ and all $\boldsymbol{\mu}\ge\mathbf{0}$ — **if and only if** $\mathbf{x}^*$ is primal
optimal, $\boldsymbol{\mu}^*$ is dual optimal, and the duality gap is zero. Deduce the minimax equality.

*Intuition:* A saddle point is exactly a point where the order of $\min$ and $\max$ does not matter.

**Solution:**

**Step 1 (the primal is a max over multipliers).** For any $\mathbf{x}$,

$$
\sup_{\boldsymbol{\mu}\ge\mathbf{0}}\mathcal{L}(\mathbf{x},\boldsymbol{\mu}) = \sup_{\boldsymbol{\mu}\ge\mathbf{0}}\left[f(\mathbf{x}) + \sum_i\mu_ig_i(\mathbf{x})\right] = \begin{cases} f(\mathbf{x}), & \mathbf{x} \ \text{feasible} \\ +\infty, & \text{otherwise}\end{cases}
$$

(if some $g_i(\mathbf{x}) \gt 0$, drive $\mu_i \to \infty$). Hence

$$
p^* = \inf_{\mathbf{x}}\sup_{\boldsymbol{\mu}\ge\mathbf{0}}\mathcal{L}, \qquad d^* = \sup_{\boldsymbol{\mu}\ge\mathbf{0}}\inf_{\mathbf{x}}\mathcal{L}
$$

Weak duality $d^* \le p^*$ is thus the general max-min $\le$ min-max inequality.

**Step 2 ($\Rightarrow$ saddle implies optimality and zero gap).** Assume the saddle inequalities. The
right one says $\mathbf{x}^*$ minimizes $\mathcal{L}(\cdot,\boldsymbol{\mu}^*)$, so

$$
g(\boldsymbol{\mu}^*) = \mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*)
$$

The left one says $\boldsymbol{\mu}^*$ maximizes $\mathcal{L}(\mathbf{x}^*,\cdot)$ over
$\boldsymbol{\mu}\ge\mathbf{0}$; by Step 1 this supremum is $+\infty$ unless $\mathbf{x}^*$ is feasible, so
$\mathbf{x}^*$ is feasible and

$$
\mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*) = \sup_{\boldsymbol{\mu}\ge0}\mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}) = f(\mathbf{x}^*)
$$

Combining, $g(\boldsymbol{\mu}^*) = f(\mathbf{x}^*)$. With weak duality
$g(\boldsymbol{\mu}^*) \le d^* \le p^* \le f(\mathbf{x}^*)$, every inequality is an equality: $\mathbf{x}^*$
is primal optimal, $\boldsymbol{\mu}^*$ dual optimal, gap zero.

**Step 3 ($\Leftarrow$ optimality with zero gap implies saddle).** Assume $f(\mathbf{x}^*) = p^* = d^* = g(\boldsymbol{\mu}^*)$.
By Problem L1.3, $\mathbf{x}^*$ minimizes $\mathcal{L}(\cdot,\boldsymbol{\mu}^*)$ and complementary
slackness holds. Then for all $\mathbf{x}$,

$$
\mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*) \le \mathcal{L}(\mathbf{x},\boldsymbol{\mu}^*)
$$

the right saddle inequality. For the left, using feasibility of $\mathbf{x}^*$ and $\boldsymbol{\mu}\ge0$,

$$
\mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}) = f(\mathbf{x}^*) + \sum_i\mu_ig_i(\mathbf{x}^*) \le f(\mathbf{x}^*) = f(\mathbf{x}^*) + \sum_i\mu_i^*g_i(\mathbf{x}^*) = \mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*)
$$

where the last equality is complementary slackness.

**Step 4 (minimax equality).** At a saddle point,

$$
\inf_{\mathbf{x}}\sup_{\boldsymbol{\mu}\ge\mathbf{0}}\mathcal{L}(\mathbf{x},\boldsymbol{\mu}) = \mathcal{L}(\mathbf{x}^*,\boldsymbol{\mu}^*) = \sup_{\boldsymbol{\mu}\ge\mathbf{0}}\inf_{\mathbf{x}}\mathcal{L}(\mathbf{x},\boldsymbol{\mu})
$$

so the order of optimization is immaterial, and the common value is the saddle value $p^* = d^*$.

$$
\boxed{\text{saddle point of } \mathcal{L} \iff \text{primal optimal} + \text{dual optimal} + \text{zero gap} \iff \min\max = \max\min}
$$

> **Key takeaway:** Strong duality is exactly a minimax interchange theorem, which is why the same machinery underlies zero-sum game equilibria (von Neumann), GAN training objectives, and primal-dual solvers that ascend in $\boldsymbol{\mu}$ while descending in $\mathbf{x}$.
