# Topic 04: Systems of ODEs and the Matrix Exponential

## 1. First-Principles Intuition & Motivation

**The phenomenon.** Real systems almost never involve a single evolving quantity. Salt masses in two connected tanks, the two currents in an RLC circuit, predator and prey populations, the $n$ hidden units of a recurrent network — each rate of change depends on the *current values of all the other variables*. The natural object is therefore a **system of first-order ODEs**:

$$
\frac{d\mathbf{x}}{dt} = \mathbf{f}(t, \mathbf{x}), \qquad \mathbf{x}(t) \in \mathbb{R}^n
$$

**Why first-order systems are THE canonical object.** Any $n$-th order scalar ODE can be converted into a first-order system. Take

$$
y^{(n)} + a_{n-1} y^{(n-1)} + \cdots + a_1 y' + a_0 y = 0
$$

and introduce the state vector $x_1 = y,\ x_2 = y',\ \ldots,\ x_n = y^{(n-1)}$. Then $x_1' = x_2$, $x_2' = x_3$, ..., and the original equation supplies the last row:

$$
\mathbf{x}' =
\begin{pmatrix}
0 & 1 & 0 & \cdots & 0 \\
0 & 0 & 1 & \cdots & 0 \\
\vdots & & & \ddots & \vdots \\
0 & 0 & 0 & \cdots & 1 \\
-a_0 & -a_1 & -a_2 & \cdots & -a_{n-1}
\end{pmatrix}
\mathbf{x} = C\,\mathbf{x}
$$

The matrix $C$ is called the **companion matrix**; its characteristic polynomial is exactly $\lambda^n + a_{n-1}\lambda^{n-1} + \cdots + a_0$, so the classical "characteristic equation" trick for higher-order ODEs (see [`../../calculus/15_ordinary_differential_equations/`](../../calculus/15_ordinary_differential_equations/)) is secretly an eigenvalue computation. Nothing is lost and everything is unified: one theory for the constant-coefficient linear system $\mathbf{x}' = A\mathbf{x}$ covers every linear ODE of every order, and — via linearization at equilibria — governs the local behavior of every smooth nonlinear system too.

### The goal: promote $e^{at}$ to matrices

The scalar problem $x' = ax$, $x(0) = x_0$ has the unique solution $x(t) = e^{at}x_0$, where

$$
e^{at} = \sum_{k=0}^{\infty} \frac{(at)^k}{k!}
$$

The guiding first-principles question of this module: *does the same series make sense when $a$ is replaced by an $n \times n$ matrix $A$, and does it still solve the initial value problem?* Both answers are yes, and the resulting operator

$$
e^{At} = \sum_{k=0}^{\infty} \frac{(At)^k}{k!} = I + At + \frac{A^2 t^2}{2!} + \frac{A^3 t^3}{3!} + \cdots
$$

is the **matrix exponential** — the flow map that pushes any initial state forward by time $t$.

**Standing assumptions.**

1. $A \in \mathbb{R}^{n \times n}$ (or $\mathbb{C}^{n \times n}$) is a *constant* matrix; time-varying $A(t)$ breaks most formulas below unless the family commutes.
2. For forced systems $\mathbf{x}' = A\mathbf{x} + \mathbf{g}(t)$ we assume $\mathbf{g}$ is continuous, so all integrals exist.
3. We work in finite dimensions; every proof uses only completeness of $\mathbb{R}^{n\times n}$ and a submultiplicative norm, so everything transfers verbatim to bounded operators on a Banach space.

### Variables and notation

| Symbol | Meaning |
| :--- | :--- |
| $\mathbf{x}(t) \in \mathbb{R}^n$ | state vector at time $t$ |
| $A \in \mathbb{R}^{n \times n}$ | constant coefficient matrix (the generator of the flow) |
| $\lVert A \rVert$ | operator norm $\max_{\lVert v \rVert = 1} \lVert Av \rVert$; satisfies $\lVert AB \rVert \le \lVert A \rVert \lVert B \rVert$ |
| $e^{At}$ | matrix exponential, the flow map of $\mathbf{x}' = A\mathbf{x}$ |
| $\Phi(t)$ | a fundamental matrix: columns are $n$ independent solutions |
| $W(t) = \det \Phi(t)$ | Wronskian of the system |
| $\lambda_i, \mathbf{v}_i$ | eigenvalues and eigenvectors of $A$ (see [`../../linear_algebra/06_eigenvalues_eigenvectors_spectral_theory/`](../../linear_algebra/06_eigenvalues_eigenvectors_spectral_theory/)) |
| $N$ | nilpotent part in $A = \lambda I + N$ or in a Jordan block |
| $\mathbf{g}(t)$ | forcing (input) term in $\mathbf{x}' = A\mathbf{x} + \mathbf{g}(t)$ |
| $Q$ | generator of a continuous-time Markov chain, $P(t) = e^{Qt}$ |

Throughout, $[A, B] = AB - BA$ denotes the commutator, and "$A$ and $B$ commute" means $[A,B] = 0$.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition 2.1 (Matrix exponential).** For $A \in \mathbb{C}^{n \times n}$ and $t \in \mathbb{R}$,

$$
e^{At} := \sum_{k=0}^{\infty} \frac{t^k A^k}{k!}, \qquad A^0 := I
$$

The series converges absolutely in any matrix norm and uniformly for $t$ in compact intervals (Proof 3.1).

**Definition 2.2 (Submultiplicative norm).** A norm on $n \times n$ matrices is *submultiplicative* if $\lVert AB \rVert \le \lVert A \rVert\,\lVert B \rVert$. The operator norm induced by any vector norm is submultiplicative, and consequently

$$
\lVert A^k \rVert \le \lVert A \rVert^k \quad \text{for all } k \ge 0
$$

This single inequality powers every convergence argument in this module: it lets us dominate the matrix series by the scalar series $\sum \lVert A \rVert^k t^k / k! = e^{\lVert A \rVert t}$.

**Theorem 2.3 (Fundamental theorem of linear systems).** For every $A \in \mathbb{R}^{n \times n}$ and $\mathbf{x}_0 \in \mathbb{R}^n$, the initial value problem

$$
\mathbf{x}' = A\mathbf{x}, \qquad \mathbf{x}(0) = \mathbf{x}_0
$$

has the **unique** solution $\mathbf{x}(t) = e^{At}\mathbf{x}_0$, defined for all $t \in \mathbb{R}$. (Proof 3.3.)

**Theorem 2.4 (Algebraic properties).** For all $A$, $s$, $t$:

1. $e^{A \cdot 0} = I$.
2. **Semigroup law:** $e^{A(t+s)} = e^{At} e^{As}$ (Lemma 3.0); in particular $e^{At}$ is invertible with $\left(e^{At}\right)^{-1} = e^{-At}$.
3. **Derivative:** $\dfrac{d}{dt} e^{At} = A e^{At} = e^{At} A$ (Proof 3.2).
4. **Similarity invariance:** $e^{P A P^{-1} t} = P\, e^{At}\, P^{-1}$ — proved by conjugating the series term by term, since $(PAP^{-1})^k = P A^k P^{-1}$.
5. **Spectral mapping:** if $A\mathbf{v} = \lambda \mathbf{v}$ then $e^{At}\mathbf{v} = e^{\lambda t}\mathbf{v}$; the eigenvalues of $e^{A}$ are exactly $e^{\lambda_i}$.
6. **Jacobi's formula:** $\det e^{A} = e^{\operatorname{tr} A}$ (Proof 3.4).

**Definition 2.5 (Fundamental matrix).** A matrix-valued function $\Phi(t)$ is a *fundamental matrix* for $\mathbf{x}' = A\mathbf{x}$ if $\Phi'(t) = A\Phi(t)$ and $\det\Phi(t_0) \ne 0$ for some $t_0$. Its columns are then $n$ linearly independent solutions, and every solution is $\Phi(t)\mathbf{c}$ for a constant vector $\mathbf{c}$. The *principal* fundamental matrix at $0$ is $\Phi(t) = e^{At}$, singled out by $\Phi(0) = I$; any two fundamental matrices differ by a constant invertible factor on the right.

**Theorem 2.6 (Abel–Liouville).** The Wronskian $W(t) = \det \Phi(t)$ of any fundamental matrix satisfies the scalar ODE

$$
W'(t) = \operatorname{tr}(A)\, W(t), \qquad \text{hence} \qquad W(t) = W(0)\, e^{t \operatorname{tr} A}
$$

so $W$ is either never zero or identically zero (Proof 3.4). This is the volume-evolution law of the flow.

**Theorem 2.7 (Duhamel / variation of constants).** The forced problem $\mathbf{x}' = A\mathbf{x} + \mathbf{g}(t)$, $\mathbf{x}(0) = \mathbf{x}_0$, with $\mathbf{g}$ continuous, has the unique solution

$$
\mathbf{x}(t) = e^{At}\mathbf{x}_0 + \int_0^t e^{A(t-s)}\, \mathbf{g}(s)\, ds
$$

— free evolution of the initial state, plus a superposition of free evolutions of each infinitesimal "kick" $\mathbf{g}(s)\,ds$ delivered at time $s$ and propagated for the remaining time $t - s$ (Proof 3.5).

### The commutation caveat

**Theorem 2.8.** If $AB = BA$ then $e^{A+B} = e^A e^B$. The hypothesis is essential: without commutativity the identity generally fails.

*Proof of the commuting case.* When $AB = BA$ the binomial theorem holds: $(A+B)^k = \sum_{j=0}^{k} \binom{k}{j} A^j B^{k-j}$. Then, with all series absolutely convergent (so rearrangement and Cauchy products are legitimate),

$$
e^{A+B} = \sum_{k=0}^\infty \frac{1}{k!} \sum_{j=0}^k \binom{k}{j} A^j B^{k-j}
= \sum_{k=0}^\infty \sum_{j=0}^{k} \frac{A^j}{j!} \frac{B^{k-j}}{(k-j)!}
= \left( \sum_{j=0}^\infty \frac{A^j}{j!} \right)\left( \sum_{m=0}^\infty \frac{B^m}{m!} \right) = e^A e^B \qquad \blacksquare
$$

**Counterexample (non-commuting).** Take

$$
A = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}, \qquad
B = \begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix}, \qquad
AB = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix} \ne \begin{pmatrix} 0 & 0 \\ 0 & 1 \end{pmatrix} = BA
$$

Since $A^2 = B^2 = 0$, the exponentials truncate: $e^A = I + A$ and $e^B = I + B$, so

$$
e^A e^B = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}\begin{pmatrix} 1 & 0 \\ 1 & 1 \end{pmatrix} = \begin{pmatrix} 2 & 1 \\ 1 & 1 \end{pmatrix}
$$

But the sum

$$
A + B = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}, \qquad (A+B)^2 = I
$$

satisfies a two-term recursion, so splitting the series into even and odd powers gives

$$
e^{A+B} = \cosh(1)\, I + \sinh(1)\,(A+B) = \begin{pmatrix} \cosh 1 & \sinh 1 \\ \sinh 1 & \cosh 1 \end{pmatrix} \approx \begin{pmatrix} 1.543 & 1.175 \\ 1.175 & 1.543 \end{pmatrix}
$$

and $\cosh 1 \approx 1.543 \ne 2$. The discrepancy is quantified by the Baker–Campbell–Hausdorff series, whose first correction is $\tfrac{1}{2}[A,B]$ (derived in exercises L3.3).

## 3. Step-by-Step Mathematical Proofs & Derivations

This section gives six complete proofs, in logical order:

- **Lemma 3.0** — semigroup law $e^{A(t+s)} = e^{At}e^{As}$ and invertibility.
- **Proof 3.1** — the defining series converges absolutely, uniformly on compact time intervals.
- **Proof 3.2** — term-wise differentiation: $\frac{d}{dt}e^{At} = Ae^{At}$.
- **Proof 3.3** — $e^{At}\mathbf{x}_0$ solves the IVP, and no other solution exists.
- **Proof 3.4** — Abel–Liouville: $W' = \operatorname{tr}(A)W$, hence $\det e^{A} = e^{\operatorname{tr} A}$.
- **Proof 3.5** — the Duhamel / variation-of-constants formula.
- **Proof 3.6** — Putzer's algorithm for $2 \times 2$ matrices (exponentials without eigenvectors).

**Lemma 3.0 (Semigroup law).**

*Step 1.* $At$ and $As$ are scalar multiples of the same matrix, so they commute: $(At)(As) = ts A^2 = (As)(At)$.

*Step 2.* By Theorem 2.8 (whose proof used only commutativity and absolute convergence, established independently in Proof 3.1),

$$
e^{A(t+s)} = e^{At + As} = e^{At}e^{As}
$$

*Step 3.* Setting $s = -t$: $e^{At}e^{-At} = e^{A\cdot 0} = I$, and symmetrically $e^{-At}e^{At} = I$. Thus $e^{At}$ is invertible for every $t$ with

$$
\boxed{\ e^{A(t+s)} = e^{At}e^{As}, \qquad \left(e^{At}\right)^{-1} = e^{-At}\ }
$$

### Proof 3.1 — Absolute convergence of the exponential series

**Claim.** For every $A \in \mathbb{C}^{n\times n}$, the series $\sum_{k \ge 0} t^k A^k / k!$ converges absolutely in operator norm, uniformly for $t \in [-T, T]$, for every $T \gt 0$.

*Step 1 (Norm bound on terms).* Fix a submultiplicative norm. For each $k$,

$$
\left\lVert \frac{t^k A^k}{k!} \right\rVert \le \frac{\lvert t \rvert^k \lVert A \rVert^k}{k!} \le \frac{(T \lVert A \rVert)^k}{k!} =: M_k \quad \text{for all } t \in [-T, T]
$$

*Step 2 (Scalar comparison).* The scalar series $\sum_{k} M_k = e^{T\lVert A \rVert}$ converges. By the Weierstrass M-test, the matrix series converges absolutely and uniformly on $[-T, T]$: the partial sums $S_m(t) = \sum_{k=0}^m t^k A^k/k!$ satisfy, for $m \gt l$,

$$
\lVert S_m(t) - S_l(t) \rVert \le \sum_{k=l+1}^{m} \frac{(T\lVert A \rVert)^k}{k!} \xrightarrow[l,m \to \infty]{} 0
$$

so $(S_m(t))$ is uniformly Cauchy.

*Step 3 (Completeness).* $\mathbb{C}^{n \times n} \cong \mathbb{C}^{n^2}$ is complete, and all norms on it are equivalent; hence the Cauchy sequence of partial sums converges to a limit, entry by entry, and the limit function is continuous in $t$ (uniform limit of polynomials). Each entry of the series is dominated by the entrywise-absolutely-convergent scalar series, so rearrangements (used in Theorem 2.8 and Proof 3.2) are valid.

$$
\boxed{\ e^{At} \text{ exists for every square } A \text{, with } \lVert e^{At} \rVert \le e^{\lvert t \rvert \lVert A \rVert}\ }
$$

### Proof 3.2 — Differentiation of $e^{At}$ term by term

**Claim.** $t \mapsto e^{At}$ is differentiable with $\frac{d}{dt}e^{At} = Ae^{At} = e^{At}A$.

*Step 1 (Differentiate the partial sums).* Each partial sum $S_m(t) = \sum_{k=0}^{m} t^k A^k / k!$ is a matrix polynomial in $t$, and

$$
S_m'(t) = \sum_{k=1}^{m} \frac{k\, t^{k-1} A^k}{k!} = A \sum_{k=1}^{m} \frac{t^{k-1}A^{k-1}}{(k-1)!} = A\, S_{m-1}(t)
$$

*Step 2 (Uniform convergence of the differentiated series).* On $[-T,T]$,

$$
\lVert S_m'(t) - A e^{At} \rVert \le \lVert A \rVert \cdot \lVert S_{m-1}(t) - e^{At} \rVert \le \lVert A \rVert \sum_{k=m}^{\infty} \frac{(T\lVert A\rVert)^k}{k!} \to 0
$$

so $S_m' \to A e^{At}$ *uniformly* on $[-T, T]$.

*Step 3 (Swap limit and derivative).* The classical theorem on differentiation of function sequences applies: if $S_m \to S$ pointwise and $S_m' \to G$ uniformly on an interval, with each $S_m$ continuously differentiable, then $S$ is differentiable and $S' = G$. (Equivalently: integrate the uniformly convergent series $A e^{As}$ from $0$ to $t$ term by term to recover $e^{At} - I$, then invoke the fundamental theorem of calculus.) Hence $\frac{d}{dt} e^{At} = A e^{At}$.

*Step 4 (Two-sided factorization).* In Step 1 the factor $A$ can equally be pulled out on the right, $S_m'(t) = S_{m-1}(t)\, A$, since powers of $A$ commute with $A$. Therefore

$$
\boxed{\ \frac{d}{dt} e^{At} = A e^{At} = e^{At} A\ }
$$

### Proof 3.3 — $e^{At}\mathbf{x}_0$ solves the IVP, uniquely

**Claim.** $\mathbf{x}(t) = e^{At}\mathbf{x}_0$ is the one and only solution of $\mathbf{x}' = A\mathbf{x}$, $\mathbf{x}(0) = \mathbf{x}_0$.

*Step 1 (Existence).* By Proof 3.2, $\mathbf{x}'(t) = \frac{d}{dt}\left(e^{At}\right)\mathbf{x}_0 = A e^{At} \mathbf{x}_0 = A\mathbf{x}(t)$, and $\mathbf{x}(0) = e^{0}\mathbf{x}_0 = I\mathbf{x}_0 = \mathbf{x}_0$. So a solution exists — globally, for all $t \in \mathbb{R}$.

*Step 2 (Uniqueness — the derivative-of-difference trick).* Let $\mathbf{y}(t)$ be *any* differentiable solution of the IVP. Define

$$
\mathbf{z}(t) := e^{-At}\, \mathbf{y}(t)
$$

By the product rule, Proof 3.2 (applied to $-A$), and the fact that $A$ commutes with $e^{-At}$,

$$
\mathbf{z}'(t) = \left(-A e^{-At}\right)\mathbf{y}(t) + e^{-At}\, \mathbf{y}'(t) = -A e^{-At}\mathbf{y}(t) + e^{-At} A \mathbf{y}(t) = \mathbf{0}
$$

*Step 3 (Conclude).* A differentiable function with identically zero derivative on $\mathbb{R}$ is constant (apply the mean value theorem to each component). Hence $\mathbf{z}(t) = \mathbf{z}(0) = \mathbf{y}(0) = \mathbf{x}_0$ for all $t$, and multiplying by $e^{At}$ on the left (using Lemma 3.0, $e^{At}e^{-At} = I$):

$$
\mathbf{y}(t) = e^{At}\mathbf{x}_0
$$

Every solution coincides with the exponential one. (The same conclusion follows from Grönwall's inequality applied to $\lVert \mathbf{y}_1 - \mathbf{y}_2 \rVert$, which is the route that generalizes to nonlinear systems — see the Picard–Lindelöf module `../02_existence_uniqueness_picard_lindelof/`.)

$$
\boxed{\ \mathbf{x}(t) = e^{At}\mathbf{x}_0 \text{ is the unique solution of } \mathbf{x}' = A\mathbf{x},\ \mathbf{x}(0) = \mathbf{x}_0\ }
$$

### Proof 3.4 — Abel–Liouville formula and $\det e^{A} = e^{\operatorname{tr} A}$

**Claim.** If $\Phi'(t) = A\Phi(t)$, then $W(t) = \det\Phi(t)$ satisfies $W' = \operatorname{tr}(A)\,W$; consequently $\det e^{At} = e^{t\operatorname{tr} A}$.

*Step 1 (Derivative of a determinant).* Write $\Phi(t)$ in terms of its rows $\mathbf{r}_1(t), \ldots, \mathbf{r}_n(t)$. The determinant is multilinear in the rows, so its derivative is a sum over rows:

$$
W'(t) = \sum_{i=1}^{n} \det \begin{pmatrix} \mathbf{r}_1 \\ \vdots \\ \mathbf{r}_i' \\ \vdots \\ \mathbf{r}_n \end{pmatrix}
$$

*Step 2 (Use the ODE row-wise).* From $\Phi' = A\Phi$, row $i$ of $\Phi'$ is a combination of the rows of $\Phi$: $\mathbf{r}_i' = \sum_{j=1}^{n} a_{ij}\, \mathbf{r}_j$. Substitute into the $i$-th determinant and expand by multilinearity. Every term with $j \ne i$ has a repeated row ($\mathbf{r}_j$ appears in positions $i$ and $j$), so its determinant vanishes. Only the $j = i$ term survives:

$$
\det \begin{pmatrix} \mathbf{r}_1 \\ \vdots \\ \mathbf{r}_i' \\ \vdots \\ \mathbf{r}_n \end{pmatrix} = a_{ii}\, \det\Phi(t) = a_{ii}\, W(t)
$$

*Step 3 (Sum and solve the scalar ODE).* Summing over $i$ gives $W'(t) = \left(\sum_i a_{ii}\right) W(t) = \operatorname{tr}(A)\, W(t)$, a scalar linear ODE whose unique solution (by Proof 3.3 with $n = 1$) is

$$
W(t) = W(0)\, e^{t \operatorname{tr} A}
$$

*Step 4 (Specialize).* Take $\Phi(t) = e^{At}$, so $W(0) = \det I = 1$, and evaluate at $t = 1$:

$$
\boxed{\ \det e^{At} = e^{t\operatorname{tr} A} \ne 0 \quad \text{for all } t; \qquad \det e^{A} = e^{\operatorname{tr} A}\ }
$$

This re-proves invertibility of $e^{At}$ and shows the flow scales phase-space volume at the exact exponential rate $\operatorname{tr} A$ (Liouville's theorem: volume-preserving iff $\operatorname{tr} A = 0$).

### Proof 3.5 — Duhamel / variation of constants

**Claim.** The unique solution of $\mathbf{x}' = A\mathbf{x} + \mathbf{g}(t)$, $\mathbf{x}(0) = \mathbf{x}_0$, with $\mathbf{g}$ continuous, is

$$
\mathbf{x}(t) = e^{At}\mathbf{x}_0 + \int_0^t e^{A(t-s)}\,\mathbf{g}(s)\,ds
$$

*Step 1 (The integrating-factor ansatz).* Guided by the homogeneous solution, seek $\mathbf{x}(t) = e^{At}\mathbf{u}(t)$ where the "constant" $\mathbf{u}$ is allowed to vary. Substituting into the ODE:

$$
\mathbf{x}' = A e^{At}\mathbf{u} + e^{At}\mathbf{u}' \stackrel{!}{=} A e^{At}\mathbf{u} + \mathbf{g}(t)
$$

*Step 2 (Isolate the unknown).* The $A e^{At}\mathbf{u}$ terms cancel, leaving $e^{At}\mathbf{u}'(t) = \mathbf{g}(t)$. Multiplying by the inverse $e^{-At}$ (Lemma 3.0):

$$
\mathbf{u}'(t) = e^{-At}\,\mathbf{g}(t)
$$

*Step 3 (Integrate).* $\mathbf{u}$ is recovered by the fundamental theorem of calculus ($e^{-As}\mathbf{g}(s)$ is continuous):

$$
\mathbf{u}(t) = \mathbf{u}(0) + \int_0^t e^{-As}\,\mathbf{g}(s)\,ds, \qquad \mathbf{u}(0) = \mathbf{x}(0) = \mathbf{x}_0
$$

*Step 4 (Reassemble).* Multiply by $e^{At}$ and use the semigroup law $e^{At}e^{-As} = e^{A(t-s)}$ inside the integral (legitimate since $e^{At}$ is a constant matrix with respect to $s$):

$$
\mathbf{x}(t) = e^{At}\mathbf{x}_0 + \int_0^t e^{A(t-s)}\,\mathbf{g}(s)\,ds
$$

*Step 5 (Verify and confirm uniqueness).* Differentiating (Leibniz rule: the upper limit contributes $e^{A(t-t)}\mathbf{g}(t) = \mathbf{g}(t)$, the integrand contributes $A e^{A(t-s)}\mathbf{g}(s)$) gives $\mathbf{x}' = A\mathbf{x} + \mathbf{g}$, and $\mathbf{x}(0) = \mathbf{x}_0$. If $\mathbf{x}_1, \mathbf{x}_2$ both solve the problem, their difference solves the homogeneous IVP with zero initial data, hence vanishes by Proof 3.3.

$$
\boxed{\ \mathbf{x}(t) = e^{At}\mathbf{x}_0 + \int_0^t e^{A(t-s)}\mathbf{g}(s)\,ds\ }
$$

**Interpretation.** The state at time $t$ is the freely-evolved initial condition plus the superposition of every past input impulse $\mathbf{g}(s)\,ds$, each propagated forward by the flow for its remaining time $t - s$. This is exactly the "convolve the input with the impulse response" picture of linear systems theory, and the formula that state-space ML models (S4/Mamba) discretize.

### Proof 3.6 — Putzer's algorithm for $2 \times 2$ matrices

**Claim.** Let $A \in \mathbb{R}^{2\times 2}$ have eigenvalues $\lambda_1, \lambda_2$ (repetition allowed, no eigenvectors needed). Then

$$
e^{At} = r_1(t)\, I + r_2(t)\,(A - \lambda_1 I)
$$

where $r_1' = \lambda_1 r_1$, $r_1(0) = 1$ and $r_2' = \lambda_2 r_2 + r_1$, $r_2(0) = 0$.

*Step 1 (Ansatz driven by Cayley–Hamilton).* By Cayley–Hamilton, $(A - \lambda_1 I)(A - \lambda_2 I) = 0$, so $I$ and $A - \lambda_1 I$ span an algebra containing all powers of $A$; it is natural to seek $e^{At}$ in that span with time-dependent coefficients. Define $M(t) := r_1(t) I + r_2(t)(A - \lambda_1 I)$ with $r_1, r_2$ as above.

*Step 2 (Check the initial condition).* $M(0) = 1 \cdot I + 0 \cdot (A - \lambda_1 I) = I$, as required.

*Step 3 (Check the ODE).* Differentiate and substitute the scalar ODEs:

$$
M' = r_1' I + r_2' (A - \lambda_1 I) = \lambda_1 r_1 I + (\lambda_2 r_2 + r_1)(A - \lambda_1 I)
$$

Compare with $AM = r_1 A + r_2 A(A - \lambda_1 I)$. Subtracting, and noting that $r_1 A - \lambda_1 r_1 I - r_1(A - \lambda_1 I) = 0$,

$$
AM - M' = r_2\left[ A(A-\lambda_1 I) - \lambda_2 (A - \lambda_1 I)\right] = r_2\,(A - \lambda_2 I)(A - \lambda_1 I) = 0
$$

where the last equality is Cayley–Hamilton (the two factors commute, being polynomials in $A$). Hence $M' = AM$ with $M(0) = I$, so by uniqueness (Proof 3.3 applied column-wise) $M(t) = e^{At}$.

*Step 4 (Solve the scalar cascade).* $r_1(t) = e^{\lambda_1 t}$. Then $r_2$ solves a scalar Duhamel problem (Proof 3.5 with $n = 1$):

$$
r_2(t) = \int_0^t e^{\lambda_2 (t-s)} e^{\lambda_1 s}\, ds =
\begin{cases}
\dfrac{e^{\lambda_1 t} - e^{\lambda_2 t}}{\lambda_1 - \lambda_2}, & \lambda_1 \ne \lambda_2 \\[2mm]
t\, e^{\lambda_1 t}, & \lambda_1 = \lambda_2
\end{cases}
$$

$$
\boxed{\ e^{At} = e^{\lambda_1 t} I + r_2(t)\,(A - \lambda_1 I), \quad r_2 \text{ as above}\ }
$$

Putzer's algorithm extends to $n \times n$: $e^{At} = \sum_{j=1}^{n} r_j(t) P_{j-1}$ with $P_0 = I$, $P_j = (A - \lambda_j I)P_{j-1}$, and the triangular cascade $r_1' = \lambda_1 r_1$, $r_j' = \lambda_j r_j + r_{j-1}$. It requires eigenvalues only — never eigenvectors — so it handles defective matrices with no extra machinery (a $3 \times 3$ instance is worked in exercises L3.2).

## 4. Computational & Algorithmic Insights

### Closed-form strategies, by matrix type

**1. Diagonalizable $A = PDP^{-1}$.** Similarity invariance gives

$$
e^{At} = P\, e^{Dt}\, P^{-1} = P \operatorname{diag}\!\left(e^{\lambda_1 t}, \ldots, e^{\lambda_n t}\right) P^{-1}
$$

Each column of $P$ is a "mode" evolving independently at its own exponential rate — the whole point of eigendecomposition.

**2. Complex conjugate pair $\lambda = a \pm ib$.** In the real $2\times 2$ invariant plane, $A$ is similar to the rotation-scaling block $R = aI + bJ$, where $J$ is the quarter-turn matrix with $J^2 = -I$. The two terms commute, so

$$
R = \begin{pmatrix} a & -b \\ b & a \end{pmatrix} \quad\Longrightarrow\quad e^{Rt} = e^{at}\, e^{bJt} = e^{at}\begin{pmatrix} \cos bt & -\sin bt \\ \sin bt & \cos bt \end{pmatrix}
$$

— exponential growth or decay at rate $a$, rotation at angular frequency $b$. Real solutions spiral; no complex arithmetic survives to the final answer.

**3. Defective $A$: Jordan blocks and nilpotent expansion.** A Jordan block is $J = \lambda I + N$ with $N$ nilpotent ($N^m = 0$, $m \le$ block size). Since $\lambda I$ commutes with $N$,

$$
e^{Jt} = e^{\lambda t}\, e^{Nt} = e^{\lambda t}\left( I + Nt + \frac{N^2 t^2}{2!} + \cdots + \frac{N^{m-1}t^{m-1}}{(m-1)!} \right)
$$

The exponential *terminates* — a polynomial in $t$ times $e^{\lambda t}$. These $t^k e^{\lambda t}$ terms are exactly the "repeated root" solutions from the second-order theory (`../03_second_order_linear_odes/`).

**4. Putzer / Cayley–Hamilton.** Avoids the Jordan form entirely (Proof 3.6): eigenvalues plus a triangular cascade of scalar ODEs. Best hand method for defective or nearly defective small matrices.

### Numerical reality: the nineteen dubious ways

Moler and Van Loan's famous survey ("Nineteen Dubious Ways to Compute the Exponential of a Matrix", SIAM Review 1978; revisited 2003) catalogs how nearly every "obvious" algorithm fails:

- **Truncated Taylor series**: catastrophic cancellation when $\lVert A \rVert$ is large or entries have mixed signs — intermediate terms grow like $\lVert A \rVert^k / k!$ before shrinking, destroying floating-point accuracy.
- **Eigendecomposition $P e^{Dt} P^{-1}$**: fails or loses accuracy when $A$ is **defective or near-defective**, because the eigenvector matrix condition number $\kappa(P) = \lVert P \rVert \lVert P^{-1} \rVert$ blows up; the computed exponential inherits an error amplified by $\kappa(P)$. For normal matrices ($AA^{\mathsf T} = A^{\mathsf T}A$), $P$ can be chosen orthogonal, $\kappa(P) = 1$, and the method is excellent.
- **Jordan form**: numerically meaningless — the Jordan structure is discontinuous in the entries of $A$, so roundoff can change it completely.

**The production algorithm: scaling and squaring.** Used by `scipy.linalg.expm` (Al-Mohy and Higham 2009 variant):

```text
1. Choose s so that norm(A / 2^s) <= theta  (theta ~ 5.37 for the [13/13] Pade approximant)
2. Compute R = pade_13(A / 2^s)             # rational approximation, backward stable
3. Square s times: e^A ~ R^(2^s)
```

The identity behind step 3 is the semigroup law: $e^{A} = \left(e^{A/2^s}\right)^{2^s}$. Small-norm exponentials are computed accurately, and repeated squaring reassembles the answer with $O(s)$ extra multiplications.

### Cost model and the large-sparse regime

| Method | Cost | When to use |
| :--- | :--- | :--- |
| Eigendecomposition $P e^{Dt} P^{-1}$ | $O(n^3)$, one-time; then $O(n^2)$ per $t$ | normal or safely diagonalizable $A$, many values of $t$ |
| Scaling-and-squaring Pade (`expm`) | $O(n^3)$ per evaluation, roughly 20–30 matrix products | general dense $A$ up to $n$ in the low thousands |
| Putzer / Cayley–Hamilton | $O(n^4)$ naive, fine for tiny $n$ | hand computation, symbolic work, defective matrices |
| Krylov methods (`expm_multiply`) | $O(m \cdot \operatorname{nnz}(A))$ for $m$ Krylov steps | huge sparse $A$ when only $e^{At}\mathbf{v}$ is needed |

The last row is the key practical insight: for a sparse $10^6 \times 10^6$ generator (a graph Laplacian, a CTMC rate matrix, a discretized PDE), one never forms the dense matrix $e^{At}$ — one approximates the *action* $e^{At}\mathbf{v}$ in the Krylov subspace $\operatorname{span}(\mathbf{v}, A\mathbf{v}, \ldots, A^{m-1}\mathbf{v})$ with $m \ll n$, projecting the exponential onto a small $m \times m$ Hessenberg matrix. In SciPy:

```python
from scipy.linalg import expm            # dense, scaling-and-squaring
from scipy.sparse.linalg import expm_multiply  # action e^{tA} v, Krylov-based
```

**Conditioning warning.** The sensitivity of $e^A$ to perturbations in $A$ is governed by the Fréchet derivative condition number $\kappa_{\exp}(A)$, which can vastly exceed $\kappa(A)$ for non-normal matrices: two matrices agreeing to machine precision can have exponentials differing in the first digit when eigenvalues nearly collide (near-defectiveness). Always prefer `expm` over hand-rolled eigendecomposition in such regimes.

## 5. Real-World Physics & AI/ML Applications

### Linear physical networks

**Coupled tanks (compartment models).** Two well-stirred brine tanks exchanging fluid obey $\mathbf{x}' = A\mathbf{x}$ where $x_i$ is the salt mass in tank $i$ and $A$ encodes flow rates over volumes. The eigenvalue $0$ (total salt conserved in a closed loop) and negative eigenvalues (concentration differences equalizing) tell the whole story — fully worked in exercises L2.1. Pharmacokinetics uses the identical mathematics with "tanks" replaced by organs.

**Coupled oscillators and normal modes.** Two unit masses joined by three unit springs give $\mathbf{q}'' = K\mathbf{q}$ with a symmetric stiffness matrix $K$. Diagonalizing $K$ decouples the motion into **normal modes** — collective patterns that oscillate at frequencies $\omega_i = \sqrt{-\lambda_i(K)}$ — and the equivalent first-order system's $e^{At}$ is block-diagonal with rotation blocks at those frequencies (exercises L2.2). This is the mathematics of molecular vibration spectra and structural resonance analysis.

**RLC networks and heat conduction on graphs.** Kirchhoff's laws for a linear RLC network, and spatial discretization of the heat equation, both produce

$$
\mathbf{x}' = -L\mathbf{x}, \qquad L = \text{(weighted) graph Laplacian}
$$

$L$ is symmetric positive semidefinite with $L\mathbf{1} = \mathbf{0}$, so $e^{-Lt}$ drives every initial condition to the *consensus state* — the average of the initial values — at a rate set by the second-smallest eigenvalue $\lambda_2(L)$, the **algebraic connectivity** (exercises L2.4). The same operator $e^{-Lt}$ is the heat-kernel smoother of spectral graph theory and graph neural network diffusion layers.

### Stochastic and quantum evolution

**Continuous-time Markov chains.** A CTMC on $n$ states is defined by a generator $Q$ (off-diagonal rates $q_{ij} \ge 0$, rows summing to zero). The transition probabilities satisfy the Kolmogorov forward equation ("master equation") $P'(t) = P(t)Q$, whose unique solution with $P(0) = I$ is

$$
P(t) = e^{Qt}
$$

Row sums stay $1$ because $Q\mathbf{1} = \mathbf{0}$ forces $e^{Qt}\mathbf{1} = \mathbf{1}$, and entries stay nonnegative because $e^{Qt} = e^{-qt} e^{(qI + Q)t}$ with $qI + Q$ entrywise nonnegative for $q$ large enough (uniformization). Long-run behavior: $e^{Qt} \to \mathbf{1}\boldsymbol{\pi}$, the stationary distribution — computed exactly for a two-state chain in exercises L2.3.

**Quantum mechanics.** The Schrödinger equation $i\hbar\, \psi' = H\psi$ with Hermitian Hamiltonian $H$ has solution

$$
\psi(t) = e^{-iHt/\hbar}\, \psi(0)
$$

Since $(iH)^{\dagger} = -iH$ (skew-Hermitian generator), the propagator $U(t) = e^{-iHt/\hbar}$ satisfies $U^{\dagger}U = e^{iHt/\hbar}e^{-iHt/\hbar} = I$: it is **unitary**, so probability $\lVert \psi \rVert^2$ is exactly conserved — the quantum analogue of Liouville's volume conservation for $\operatorname{tr} = 0$. The commutation caveat is physical here: $e^{-i(H_1 + H_2)t} \ne e^{-iH_1 t} e^{-iH_2 t}$ for non-commuting Hamiltonians is precisely why Trotter splitting $\left(e^{-iH_1 t/m} e^{-iH_2 t/m}\right)^m$ is needed in quantum simulation.

### Machine learning: recurrences, state-space models, gradient flows

**Linear RNN dynamics and vanishing/exploding gradients.** A linear recurrence $\mathbf{h}_{k+1} = A\mathbf{h}_k$ is the discrete-time sibling of $\mathbf{x}' = A\mathbf{x}$ (indeed $\mathbf{h}_k = A^k \mathbf{h}_0$ versus $e^{At}\mathbf{x}_0$). Backpropagated gradients through $k$ steps carry a factor $\left(A^{\mathsf T}\right)^k$, and Gelfand's formula $\lVert A^k \rVert^{1/k} \to \rho(A)$ shows the gradient norm behaves like $\rho(A)^k$: it **vanishes** when $\rho(A) \lt 1$ and **explodes** when $\rho(A) \gt 1$ — the fundamental obstacle to long-range credit assignment, motivating orthogonal initialization ($\rho = 1$, $\kappa(P) = 1$) and gated architectures (exercises L2.5). Non-normal $A$ adds transient growth $\lVert A^k \rVert \gg \rho(A)^k$ before the asymptotics kick in — the same phenomenon as near-defective conditioning in Section 4.

**State-space models (S4, Mamba).** These architectures posit a continuous latent system $\mathbf{x}' = A\mathbf{x} + B u(t)$ per channel and *discretize* it with step $\Delta$ under a zero-order hold. Duhamel's formula over one step gives exactly

$$
\bar{A} = e^{A\Delta}, \qquad \bar{B} = A^{-1}\left(e^{A\Delta} - I\right)B, \qquad \mathbf{x}_{k+1} = \bar{A}\mathbf{x}_k + \bar{B}u_k
$$

so every SSM layer literally contains a matrix exponential; making $\Delta$ input-dependent (Mamba's selectivity) modulates the flow's timescale token by token (exercises L2.6). Diagonal parameterizations of $A$ with $\operatorname{Re}\lambda \lt 0$ guarantee stable long convolutions.

**Deep linear networks and gradient flow.** Training a deep linear network $W_L \cdots W_1$ by gradient descent with infinitesimal step size is the ODE $\dot{\theta} = -\nabla L(\theta)$; near a critical point it linearizes to $\dot{\boldsymbol{\delta}} = -H\boldsymbol{\delta}$ with $H$ the Hessian, so the error evolves as $e^{-Ht}\boldsymbol{\delta}_0$ — each Hessian eigendirection converges at its own rate $e^{-\lambda_i t}$, explaining why ill-conditioned Hessians (tiny $\lambda_{\min}$) train slowly. Neural ODEs (Chen et al. 2018) close the circle by making the flow itself the learned model.
