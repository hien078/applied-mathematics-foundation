### Problem L0.1: When Does $e^{A+B} = e^Ae^B$ Fail?

**Problem Statement:** Decide whether $e^{A+B} = e^Ae^B$ holds for
$A = \begin{pmatrix} 0 & 1 \\ 0 & 0\end{pmatrix}$ and $B = \begin{pmatrix} 0 & 0 \\ 1 & 0\end{pmatrix}$
by computing both sides. State the general condition under which the law of exponents survives.

**Intuition:** The scalar proof of $e^{a+b} = e^ae^b$ rearranges a double sum freely; matrices do not commute, so that rearrangement is illegal.

**Solution:**

**Step 1 (compute $e^A$ and $e^B$).** Both matrices are nilpotent: $A^2 = B^2 = 0$, so the series
truncates:

$$
e^A = I + A = \begin{pmatrix} 1 & 1 \\ 0 & 1\end{pmatrix}, \qquad e^B = I + B = \begin{pmatrix} 1 & 0 \\ 1 & 1\end{pmatrix}
$$

$$
e^Ae^B = \begin{pmatrix} 1 & 1 \\ 0 & 1\end{pmatrix}\begin{pmatrix} 1 & 0 \\ 1 & 1\end{pmatrix} = \begin{pmatrix} 2 & 1 \\ 1 & 1\end{pmatrix}
$$

**Step 2 (compute $e^{A+B}$).** Here $A + B = \begin{pmatrix} 0 & 1 \\ 1 & 0\end{pmatrix}$ satisfies
$(A+B)^2 = I$, so the series splits into even and odd parts:

$$
e^{A+B} = \sum_{k \ \text{even}}\frac{(A+B)^k}{k!} + \sum_{k \ \text{odd}}\frac{(A+B)^k}{k!} = \cosh(1)\,I + \sinh(1)\,(A+B)
$$

$$
= \begin{pmatrix} \cosh 1 & \sinh 1 \\ \sinh 1 & \cosh 1 \end{pmatrix} \approx \begin{pmatrix} 1.5431 & 1.1752 \\ 1.1752 & 1.5431\end{pmatrix}
$$

**Step 3 (compare).** The $(1,1)$ entries are $2$ versus $\cosh 1 \approx 1.5431$: the identity **fails**.
Note also that $e^Ae^B$ is not even symmetric, whereas $e^{A+B}$ must be (the exponential of a symmetric
matrix is symmetric).

**Step 4 (the general condition).** If $AB = BA$ then the binomial theorem $(A+B)^n = \sum_k\binom{n}{k}A^kB^{n-k}$
is valid, and the Cauchy product of the two series rearranges exactly as in the scalar case:

$$
AB = BA \ \Longrightarrow \ e^{A+B} = e^Ae^B
$$

Here $AB = \begin{pmatrix}1&0\\0&0\end{pmatrix}$ and $BA = \begin{pmatrix}0&0\\0&1\end{pmatrix}$, so
$[A,B] = AB - BA = \begin{pmatrix}1&0\\0&-1\end{pmatrix} \neq 0$ — and the Baker-Campbell-Hausdorff series
(Problem L3.3) quantifies exactly how the commutator measures the failure.

$$
\boxed{e^Ae^B = \begin{pmatrix}2&1\\1&1\end{pmatrix} \neq \begin{pmatrix}\cosh1 & \sinh1\\ \sinh1&\cosh1\end{pmatrix} = e^{A+B}; \quad \text{equality} \iff AB = BA}
$$

> **Key Takeaway:** The law of exponents is a statement about *commuting* families; every splitting method in numerical analysis (Strang splitting, Trotter products) exists precisely to control the error this failure creates.

### Problem L0.2: Is $e^{At}$ the Entrywise Exponential?

**Problem Statement:** For $A = \begin{pmatrix} 0 & 1 \\ 0 & 0\end{pmatrix}$, compare $e^{At}$ with the
matrix obtained by exponentiating each entry of $At$. Explain in general when (if ever) the entrywise
recipe is correct.

**Intuition:** The defining series involves matrix *powers*, which mix rows and columns; entrywise operations never do.

**Solution:**

**Step 1 (the true matrix exponential).** $A^2 = 0$, so

$$
e^{At} = I + At = \begin{pmatrix} 1 & t \\ 0 & 1\end{pmatrix}
$$

**Step 2 (the entrywise "exponential").**

$$
\left[\exp\left(a_{ij}t\right)\right] = \begin{pmatrix} e^0 & e^t \\ e^0 & e^0\end{pmatrix} = \begin{pmatrix} 1 & e^t \\ 1 & 1\end{pmatrix}
$$

These disagree in both the $(1,2)$ and the $(2,1)$ entries; worse, the entrywise version is not even the
identity at $t = 0$ in the $(2,1)$ slot, so it cannot solve any initial value problem.

**Step 3 (why the recipe is wrong).** By definition

$$
e^{At} = \sum_{k=0}^{\infty}\frac{t^k A^k}{k!}
$$

and $\left(A^k\right)_{ij} = \sum_{p_1,\dots,p_{k-1}} a_{ip_1}a_{p_1p_2}\cdots a_{p_{k-1}j}$ combines
*many* entries of $A$. Only for a **diagonal** matrix do the powers stay entrywise,
$\left(\operatorname{diag}(d_i)\right)^k = \operatorname{diag}(d_i^k)$, giving

$$
e^{\operatorname{diag}(d_1,\dots,d_n)t} = \operatorname{diag}\left(e^{d_1t},\dots,e^{d_nt}\right)
$$

**Step 4 (the correct generalization).** For diagonalizable $A = PDP^{-1}$, the exponential is
*entrywise in the eigenbasis*:

$$
e^{At} = P\,e^{Dt}\,P^{-1}
$$

which is the algorithm of Problem L1.1. Functions of matrices act on the **spectrum**, never on the
entries.

$$
\boxed{e^{At} = \begin{pmatrix}1&t\\0&1\end{pmatrix} \neq \begin{pmatrix}1&e^t\\1&1\end{pmatrix}; \ \text{entrywise is valid only for diagonal } A}
$$

> **Key Takeaway:** $e^{At}$ is a *flow operator* assembled from repeated applications of $A$, not a table of scalar exponentials — which is why polynomial factors $t^ke^{\lambda t}$ can appear at all.

### Problem L0.3: Can $e^{At}$ Be Singular?

**Problem Statement:** Determine whether there exist a matrix $A$ and a time $t$ for which $e^{At}$ is
singular. Prove your answer two ways: from the semigroup law, and from Jacobi's formula
$\det e^{At} = e^{t\operatorname{tr}A}$.

**Intuition:** A linear flow can compress volume exponentially fast, but exponential decay never reaches zero in finite time.

**Solution:**

**Answer: never.** $e^{At}$ is invertible for every square $A$ and every $t \in \mathbb{R}$.

**Proof 1 (semigroup law).** Since $At$ and $-At$ commute, Problem L0.1's criterion applies:

$$
e^{At}\,e^{-At} = e^{At - At} = e^{0} = I
$$

so $e^{At}$ has the explicit two-sided inverse $\left(e^{At}\right)^{-1} = e^{-At}$.

**Proof 2 (Jacobi's formula).** Let $\phi(t) = \det e^{At}$. Differentiating with Jacobi's identity
$\frac{d}{dt}\det M(t) = \det M(t)\,\operatorname{tr}\!\left(M^{-1}M'\right)$ and $M' = Ae^{At}$:

$$
\phi'(t) = \phi(t)\operatorname{tr}\!\left(e^{-At}Ae^{At}\right) = \phi(t)\operatorname{tr}A
$$

using the cyclic property of the trace. With $\phi(0) = \det I = 1$, this scalar ODE gives

$$
\det e^{At} = e^{t\,\operatorname{tr}A} \neq 0 \qquad \text{for all } t
$$

since a real exponential is never zero.

**Step 3 (the geometric reading).** $\operatorname{tr}A$ is the divergence of the vector field
$\mathbf{x}\mapsto A\mathbf{x}$, and $\det e^{At}$ is the factor by which the flow scales volume. Volume
shrinks like $e^{t\operatorname{tr}A}$ — arbitrarily small, but never zero.

**Step 4 (consequence: reversibility).** Every linear flow is exactly reversible: the state at time $0$ is
recovered from the state at time $t$ by applying $e^{-At}$. Contrast with the heat equation on a continuum,
where the forward map is compact and the inverse unbounded — finite-dimensional linear ODEs have no such
irreversibility.

$$
\boxed{\det e^{At} = e^{t\operatorname{tr}A} \neq 0, \qquad \left(e^{At}\right)^{-1} = e^{-At}}
$$

> **Key Takeaway:** Invertibility of $e^{At}$ is what guarantees that solution trajectories of a linear system never cross and never merge — the geometric content of the uniqueness theorem.

### Problem L0.4: Every Higher-Order ODE Is a First-Order System

**Problem Statement:** Convert $y''' + 6y'' + 11y' + 6y = 0$ into a first-order system
$\mathbf{x}' = A\mathbf{x}$, identify the companion matrix, and verify that its characteristic polynomial
reproduces the original equation's characteristic polynomial.

**Intuition:** Name the derivatives as new variables and the equation becomes a rule for shifting the list one slot at a time.

**Solution:**

**Step 1 (name the state).** Put

$$
x_1 = y, \qquad x_2 = y', \qquad x_3 = y''
$$

Then $x_1' = x_2$, $x_2' = x_3$, and the ODE supplies the last row:

$$
x_3' = y''' = -6y'' - 11y' - 6y = -6x_3 - 11x_2 - 6x_1
$$

**Step 2 (the companion matrix).**

$$
\mathbf{x}' = A\mathbf{x}, \qquad A = \begin{pmatrix} 0 & 1 & 0 \\ 0 & 0 & 1 \\ -6 & -11 & -6\end{pmatrix}
$$

The general pattern for $y^{(n)} + a_{n-1}y^{(n-1)}+\dots+a_0y = 0$ is a matrix with $1$s on the
superdiagonal and $-a_0,\dots,-a_{n-1}$ in the last row.

**Step 3 (characteristic polynomial).** Expanding $\det(A-\lambda I)$ along the first column, or quoting
the standard companion-matrix fact,

$$
\det(A - \lambda I) = -\left(\lambda^3 + 6\lambda^2 + 11\lambda + 6\right)
$$

whose roots are exactly the roots of the original characteristic equation
$\lambda^3+6\lambda^2+11\lambda+6 = 0$, namely

$$
\lambda = -1,\ -2,\ -3 \qquad \text{(since } \lambda^3+6\lambda^2+11\lambda+6 = (\lambda+1)(\lambda+2)(\lambda+3)\text{)}
$$

**Step 4 (solution and consistency).** Distinct eigenvalues give
$y(t) = c_1e^{-t}+c_2e^{-2t}+c_3e^{-3t}$, exactly the classical answer, and
$\mathbf{x}(t) = e^{At}\mathbf{x}_0$ packages all three constants into the single initial state
$\mathbf{x}_0 = (y(0), y'(0), y''(0))^T$.

**Step 5 (why this matters).** The reduction is not a special trick: it shows that the constant-coefficient
system $\mathbf{x}' = A\mathbf{x}$ is the *general* linear ODE, so a complete theory of $e^{At}$ is a
complete theory of linear ODEs of every order. It is also how every numerical ODE solver consumes a
higher-order problem.

$$
\boxed{A = \begin{pmatrix}0&1&0\\0&0&1\\-6&-11&-6\end{pmatrix}, \qquad \det(A-\lambda I) = -(\lambda+1)(\lambda+2)(\lambda+3)}
$$

> **Key Takeaway:** Order and dimension are interchangeable: one $n$-th order equation equals $n$ first-order equations, and the companion matrix is the dictionary between them.

### Problem L1.1: $e^{At}$ by Diagonalization

**Problem Statement:** Compute $e^{At}$ for $A = \begin{pmatrix} 1 & 2 \\ 3 & 2\end{pmatrix}$ by
diagonalization, and verify both $e^{A\cdot0} = I$ and $\frac{d}{dt}e^{At}\big\vert_{t=0} = A$.

**Intuition:** In the eigenbasis the system decouples into independent scalar exponentials; the change of basis puts them back together.

**Solution:**

**Step 1 (eigenvalues).** $\operatorname{tr}A = 3$, $\det A = 2 - 6 = -4$, so

$$
\lambda^2 - 3\lambda - 4 = 0 \ \Longrightarrow\ \lambda_1 = 4, \quad \lambda_2 = -1
$$

**Step 2 (eigenvectors).**

- $\lambda = 4$: $(A-4I) = \begin{pmatrix}-3&2\\3&-2\end{pmatrix}$, so $3v_1 = 2v_2$ and
  $\mathbf{v}_1 = (2,3)^T$.
- $\lambda = -1$: $(A+I) = \begin{pmatrix}2&2\\3&3\end{pmatrix}$, so $v_1 = -v_2$ and
  $\mathbf{v}_2 = (1,-1)^T$.

**Step 3 (assemble $P$ and $P^{-1}$).**

$$
P = \begin{pmatrix} 2 & 1 \\ 3 & -1\end{pmatrix}, \qquad \det P = -5, \qquad P^{-1} = \frac{1}{-5}\begin{pmatrix}-1 & -1\\ -3 & 2\end{pmatrix} = \frac{1}{5}\begin{pmatrix}1 & 1\\ 3 & -2\end{pmatrix}
$$

**Step 4 (exponentiate).**

$$
e^{At} = P\begin{pmatrix} e^{4t} & 0 \\ 0 & e^{-t}\end{pmatrix}P^{-1} = \frac{1}{5}\begin{pmatrix} 2e^{4t}+3e^{-t} & 2e^{4t}-2e^{-t} \\ 3e^{4t}-3e^{-t} & 3e^{4t}+2e^{-t}\end{pmatrix}
$$

**Step 5 (two checks).**

- At $t = 0$: $\frac15\begin{pmatrix}5&0\\0&5\end{pmatrix} = I$ $\checkmark$
- Differentiating entrywise and setting $t=0$:
  $\frac15\begin{pmatrix}8-3 & 8+2\\ 12+3 & 12-2\end{pmatrix} = \frac15\begin{pmatrix}5&10\\15&10\end{pmatrix} = \begin{pmatrix}1&2\\3&2\end{pmatrix} = A$ $\checkmark$
- Jacobi: $\det e^{At} = e^{4t}e^{-t} = e^{3t} = e^{t\operatorname{tr}A}$ $\checkmark$

$$
\boxed{e^{At} = \frac{1}{5}\begin{pmatrix} 2e^{4t}+3e^{-t} & 2e^{4t}-2e^{-t} \\ 3e^{4t}-3e^{-t} & 3e^{4t}+2e^{-t}\end{pmatrix}}
$$

The dominant eigenvalue $4 \gt 0$ makes almost every trajectory blow up along $\mathbf{v}_1 = (2,3)^T$; only
initial states on the line $\operatorname{span}\{(1,-1)\}$ decay.

> **Key Takeaway:** Diagonalization turns $e^{At}$ into $n$ scalar exponentials; the eigenvalues set the growth rates and the eigenvectors set the directions along which those rates apply.

### Problem L1.2: Complex Eigenvalues Give Rotation-Scaling

**Problem Statement:** Compute $e^{At}$ for $A = \begin{pmatrix} -1 & -2 \\ 2 & -1\end{pmatrix}$ without
using complex arithmetic, by splitting $A$ into a scalar part and a rotation generator. Describe the
trajectories.

**Intuition:** A real matrix with complex eigenvalues $a \pm ib$ is $aI$ plus something that squares to $-b^2I$ — an infinitesimal rotation.

**Solution:**

**Step 1 (eigenvalues).** $\operatorname{tr}A = -2$, $\det A = 1 + 4 = 5$, so
$\lambda^2 + 2\lambda + 5 = 0$ and

$$
\lambda = -1 \pm 2i
$$

**Step 2 (split).** Write $A = -I + J$ with

$$
J = A + I = \begin{pmatrix} 0 & -2 \\ 2 & 0\end{pmatrix}, \qquad J^2 = \begin{pmatrix}-4&0\\0&-4\end{pmatrix} = -4I
$$

so $J$ behaves like $2i$: it is $2$ times a $90^\circ$ rotation generator. Crucially $-I$ commutes with
everything, so the law of exponents is legal here:

$$
e^{At} = e^{-tI}e^{Jt} = e^{-t}\,e^{Jt}
$$

**Step 3 (exponentiate the rotation part).** Using $J^2 = -4I$, the series splits into even and odd terms
exactly like the scalar $\cos/\sin$ series:

$$
e^{Jt} = \sum_{k}\frac{(Jt)^k}{k!} = \cos(2t)\,I + \frac{\sin(2t)}{2}\,J = \begin{pmatrix}\cos 2t & -\sin 2t \\ \sin 2t & \cos 2t\end{pmatrix}
$$

a genuine rotation matrix by angle $2t$.

**Step 4 (assemble).**

$$
e^{At} = e^{-t}\begin{pmatrix}\cos 2t & -\sin 2t \\ \sin 2t & \cos 2t\end{pmatrix}
$$

**Step 5 (checks and geometry).** At $t=0$ this is $I$ $\checkmark$;
$\det e^{At} = e^{-2t}(\cos^2+\sin^2) = e^{-2t} = e^{t\operatorname{tr}A}$ $\checkmark$. Trajectories are
**logarithmic spirals** into the origin: the angular speed is $\lvert \operatorname{Im}\lambda\rvert = 2$
rad per unit time (period $\pi$), and the radius shrinks by $e^{-t}$, so
$\lVert \mathbf{x}(t)\rVert = e^{-t}\lVert \mathbf{x}_0\rVert$ exactly. The origin is a **stable spiral**.
No solution ever leaves $\mathbb{R}^2$ — the complex eigenvalues are bookkeeping for rotation, not for
complex states.

$$
\boxed{e^{At} = e^{-t}\begin{pmatrix}\cos 2t & -\sin 2t\\ \sin 2t & \cos 2t\end{pmatrix}, \qquad \lambda = -1\pm 2i}
$$

> **Key Takeaway:** $\operatorname{Re}\lambda$ is the radial growth rate and $\operatorname{Im}\lambda$ the angular frequency; every $2\times2$ real block with complex spectrum is a scaled rotation in disguise.

### Problem L1.3: A Defective Matrix and the Nilpotent Expansion

**Problem Statement:** Compute $e^{At}$ for $A = \begin{pmatrix} 3 & 1 \\ 0 & 3\end{pmatrix}$. Show that
$A$ is defective, and explain where the factor $t$ in the answer comes from.

**Intuition:** A missing eigenvector is paid for with a polynomial factor, never with a loss of solvability.

**Solution:**

**Step 1 (defectiveness).** The characteristic polynomial is $(3-\lambda)^2$, so $\lambda = 3$ has
algebraic multiplicity $2$. But

$$
A - 3I = \begin{pmatrix} 0 & 1 \\ 0 & 0\end{pmatrix}
$$

has rank $1$, so its null space is one-dimensional, spanned by $(1,0)^T$: the geometric multiplicity is
$1 \lt 2$ and $A$ is **defective** — it cannot be diagonalized.

**Step 2 (split into scalar plus nilpotent).** Write

$$
A = 3I + N, \qquad N = \begin{pmatrix}0&1\\0&0\end{pmatrix}, \qquad N^2 = 0
$$

Since $3I$ commutes with $N$, the law of exponents applies:

$$
e^{At} = e^{3tI}e^{Nt} = e^{3t}\,e^{Nt}
$$

**Step 3 (the nilpotent series truncates).**

$$
e^{Nt} = I + Nt + \frac{N^2t^2}{2} + \dots = I + Nt = \begin{pmatrix}1 & t\\ 0 & 1\end{pmatrix}
$$

**Step 4 (assemble and check).**

$$
e^{At} = e^{3t}\begin{pmatrix} 1 & t \\ 0 & 1\end{pmatrix} = \begin{pmatrix} e^{3t} & te^{3t}\\ 0 & e^{3t}\end{pmatrix}
$$

At $t=0$: $I$ $\checkmark$. Derivative at $0$: $\begin{pmatrix}3&1\\0&3\end{pmatrix} = A$ $\checkmark$.
$\det e^{At} = e^{6t} = e^{t\operatorname{tr}A}$ $\checkmark$.

**Step 5 (where the $t$ comes from).** The solution space needs two independent solutions, but only one
eigenvector exists. The second is built from a **generalized eigenvector** $\mathbf{w}$ with
$(A-3I)\mathbf{w} = \mathbf{v}$; then

$$
\mathbf{x}(t) = e^{3t}\left(\mathbf{w} + t\mathbf{v}\right)
$$

solves the system, and the $t$ is exactly the nilpotent term $Nt$. In general a Jordan block of size $m$
produces terms up to $t^{m-1}e^{\lambda t}$:

$$
e^{(\lambda I + N)t} = e^{\lambda t}\left(I + Nt + \frac{N^2t^2}{2!} + \dots + \frac{N^{m-1}t^{m-1}}{(m-1)!}\right)
$$

$$
\boxed{e^{At} = \begin{pmatrix} e^{3t} & te^{3t}\\ 0 & e^{3t}\end{pmatrix}}
$$

> **Key Takeaway:** Defectiveness costs polynomial factors $t^ke^{\lambda t}$, never solvability — and those factors are what make purely imaginary defective eigenvalues produce *unbounded* solutions (Problem L3.1).

### Problem L1.4: Solving an Initial Value Problem

**Problem Statement:** Solve $\mathbf{x}' = A\mathbf{x}$ with
$A = \begin{pmatrix} 0 & 1 \\ -2 & -3\end{pmatrix}$ and $\mathbf{x}(0) = (1, 0)^T$. Interpret the answer in
terms of the scalar ODE whose companion matrix $A$ is.

**Intuition:** Expand the initial condition in the eigenbasis; each coordinate then evolves by its own exponential.

**Solution:**

**Step 1 (spectrum).** $\operatorname{tr}A = -3$, $\det A = 2$, so $\lambda^2+3\lambda+2 = 0$ and

$$
\lambda_1 = -1, \qquad \lambda_2 = -2
$$

**Step 2 (eigenvectors).**

- $\lambda = -1$: $(A+I) = \begin{pmatrix}1&1\\-2&-2\end{pmatrix}$, so $\mathbf{v}_1 = (1,-1)^T$.
- $\lambda = -2$: $(A+2I) = \begin{pmatrix}2&1\\-2&-1\end{pmatrix}$, so $\mathbf{v}_2 = (1,-2)^T$.

**Step 3 (expand the initial condition).** Solve $c_1\mathbf{v}_1 + c_2\mathbf{v}_2 = (1,0)^T$:

$$
c_1 + c_2 = 1, \qquad -c_1 - 2c_2 = 0 \ \Longrightarrow\ c_1 = -2c_2 \ \Longrightarrow\ -c_2 = 1
$$

so $c_2 = -1$ and $c_1 = 2$.

**Step 4 (assemble the solution).**

$$
\mathbf{x}(t) = 2e^{-t}\begin{pmatrix}1\\-1\end{pmatrix} - e^{-2t}\begin{pmatrix}1\\-2\end{pmatrix} = \begin{pmatrix} 2e^{-t}-e^{-2t} \\ -2e^{-t}+2e^{-2t}\end{pmatrix}
$$

Check: $\mathbf{x}(0) = (2-1,\ -2+2)^T = (1,0)^T$ $\checkmark$.

**Step 5 (scalar interpretation).** $A$ is the companion matrix of $y'' + 3y' + 2y = 0$ with $x_1 = y$,
$x_2 = y'$. The initial data $\mathbf{x}(0) = (1,0)$ means $y(0)=1$, $y'(0)=0$: released from rest at unit
displacement. The answer

$$
y(t) = 2e^{-t}-e^{-2t}
$$

is the classical overdamped response — no oscillation (both roots real and negative), a monotone decay to
zero after $t=0$, and the slow mode $e^{-t}$ dominating for large $t$. Verify:
$y' = -2e^{-t}+2e^{-2t}$ (so $y'(0)=0$ $\checkmark$), $y'' = 2e^{-t}-4e^{-2t}$, and

$$
y''+3y'+2y = (2-6+4)e^{-t} + (-4+6-2)e^{-2t} = 0 \quad\checkmark
$$

$$
\boxed{\mathbf{x}(t) = \begin{pmatrix} 2e^{-t}-e^{-2t}\\ -2e^{-t}+2e^{-2t}\end{pmatrix}, \qquad y(t) = 2e^{-t}-e^{-2t}}
$$

> **Key Takeaway:** Solving an IVP is a change of basis: expand $\mathbf{x}_0$ in eigenvectors, multiply each coefficient by its own $e^{\lambda t}$, and reassemble — the matrix exponential is that whole procedure in one symbol.

### Problem L1.5: The Two Structural Identities

**Problem Statement:** Prove directly from the power series that (a) $\frac{d}{dt}e^{At} = Ae^{At} = e^{At}A$
and (b) $e^{A(t+s)} = e^{At}e^{As}$. Then deduce that $\mathbf{x}(t) = e^{At}\mathbf{x}_0$ is the unique
solution of the initial value problem.

**Intuition:** Term-by-term differentiation shifts the series by one power of $A$; the semigroup law is the binomial theorem for two commuting copies of the same matrix.

**Solution:**

Throughout we may differentiate and rearrange because the series
$\sum_k\frac{\lVert A\rVert^kt^k}{k!} = e^{\lVert A\rVert\lvert t\rvert}$ converges absolutely and
uniformly on compact $t$-intervals, using submultiplicativity $\lVert A^k\rVert \le \lVert A\rVert^k$.

**(a) The derivative identity.**

$$
\frac{d}{dt}e^{At} = \frac{d}{dt}\sum_{k=0}^{\infty}\frac{A^kt^k}{k!} = \sum_{k=1}^{\infty}\frac{A^kt^{k-1}}{(k-1)!} = A\sum_{j=0}^{\infty}\frac{A^jt^{j}}{j!} = Ae^{At}
$$

and since $A$ commutes with every power of itself, the factor $A$ may equally be pulled out on the right,
giving $e^{At}A$.

**(b) The semigroup law.** Because $At$ and $As$ commute, the Cauchy product of the two absolutely
convergent series may be rearranged:

$$
e^{At}e^{As} = \left(\sum_{p}\frac{A^pt^p}{p!}\right)\left(\sum_{q}\frac{A^qs^q}{q!}\right) = \sum_{k=0}^{\infty}A^k\sum_{p+q=k}\frac{t^ps^q}{p!\,q!} = \sum_{k=0}^{\infty}\frac{A^k}{k!}\sum_{p=0}^{k}\binom{k}{p}t^ps^{k-p}
$$

The inner sum is the binomial expansion of $(t+s)^k$, so

$$
e^{At}e^{As} = \sum_{k=0}^{\infty}\frac{A^k(t+s)^k}{k!} = e^{A(t+s)}
$$

**(c) Existence.** $\mathbf{x}(t) = e^{At}\mathbf{x}_0$ satisfies $\mathbf{x}(0) = I\mathbf{x}_0 = \mathbf{x}_0$
and, by (a), $\mathbf{x}'(t) = Ae^{At}\mathbf{x}_0 = A\mathbf{x}(t)$.

**(d) Uniqueness.** Let $\mathbf{y}$ be any solution and set $\mathbf{z}(t) = e^{-At}\mathbf{y}(t)$. By the
product rule and (a),

$$
\mathbf{z}'(t) = -Ae^{-At}\mathbf{y}(t) + e^{-At}\mathbf{y}'(t) = -Ae^{-At}\mathbf{y} + e^{-At}A\mathbf{y} = \mathbf{0}
$$

using that $e^{-At}$ commutes with $A$. Hence $\mathbf{z}$ is constant, $\mathbf{z}(t) = \mathbf{z}(0) = \mathbf{y}(0) = \mathbf{x}_0$,
and therefore $\mathbf{y}(t) = e^{At}\mathbf{x}_0$. $\blacksquare$

$$
\boxed{\frac{d}{dt}e^{At} = Ae^{At}, \quad e^{A(t+s)} = e^{At}e^{As}, \quad \mathbf{x}(t) = e^{At}\mathbf{x}_0 \ \text{is the unique solution}}
$$

> **Key Takeaway:** The semigroup law says the flow at time $t+s$ is the flow at time $s$ followed by the flow at time $t$ — the algebraic statement of determinism, and the reason $\{e^{At}\}$ is a one-parameter group of invertible operators.

### Problem L1.6: Fundamental Matrices and Abel-Liouville

**Problem Statement:** Define a fundamental matrix $\Phi(t)$ for $\mathbf{x}' = A(t)\mathbf{x}$ and prove
the Abel-Liouville formula

$$
\det\Phi(t) = \det\Phi(t_0)\,\exp\left(\int_{t_0}^{t}\operatorname{tr}A(s)\,ds\right)
$$

Specialize to constant $A$ and to the Wronskian of a second-order scalar ODE.

**Intuition:** The determinant measures volume, and the divergence $\operatorname{tr}A$ is the instantaneous rate at which the flow expands volume.

**Solution:**

**Step 1 (definition).** $\Phi(t) \in \mathbb{R}^{n\times n}$ is a **fundamental matrix** if its columns are
$n$ solutions of $\mathbf{x}' = A(t)\mathbf{x}$ and $\det\Phi(t_0)\neq0$ for some $t_0$. Equivalently
$\Phi' = A(t)\Phi$ with $\Phi(t_0)$ invertible. The general solution is
$\mathbf{x}(t) = \Phi(t)\mathbf{c}$, and the **state-transition matrix** is
$\Psi(t,t_0) = \Phi(t)\Phi(t_0)^{-1}$.

**Step 2 (differentiate the determinant).** Jacobi's formula gives, wherever $\Phi$ is invertible,

$$
\frac{d}{dt}\det\Phi(t) = \det\Phi(t)\,\operatorname{tr}\!\left(\Phi(t)^{-1}\Phi'(t)\right) = \det\Phi(t)\,\operatorname{tr}\!\left(\Phi^{-1}A\Phi\right) = \det\Phi(t)\,\operatorname{tr}A(t)
$$

using similarity invariance of the trace.

**Step 3 (integrate the scalar ODE).** Let $w(t) = \det\Phi(t)$. Then $w' = \operatorname{tr}A(t)\,w$ is a
separable scalar equation with solution

$$
w(t) = w(t_0)\exp\left(\int_{t_0}^t\operatorname{tr}A(s)\,ds\right)
$$

**Step 4 (the dichotomy).** Since the exponential never vanishes, $\det\Phi(t)$ is either **never** zero or
**identically** zero. So checking invertibility at a single time certifies it for all time — which
retroactively justifies Step 2 on the whole interval.

**Step 5 (specializations).**

- **Constant $A$:** taking $\Phi(t) = e^{At}$ with $\Phi(0)=I$ recovers Jacobi's formula
  $\det e^{At} = e^{t\operatorname{tr}A}$ of Problem L0.3.
- **Second-order scalar ODE $y''+p(t)y'+q(t)y = 0$:** its companion matrix is
  $A(t) = \begin{pmatrix}0&1\\ -q&-p\end{pmatrix}$ with $\operatorname{tr}A = -p(t)$, and $\det\Phi$ is
  exactly the Wronskian $W(y_1,y_2)$. Abel-Liouville becomes the classical Abel identity

$$
W(t) = W(t_0)\exp\left(-\int_{t_0}^tp(s)\,ds\right)
$$

$$
\boxed{\det\Phi(t) = \det\Phi(t_0)\exp\left(\int_{t_0}^t\operatorname{tr}A(s)\,ds\right); \ \text{constant } A: \ \det e^{At} = e^{t\operatorname{tr}A}}
$$

> **Key Takeaway:** One scalar quantity, $\operatorname{tr}A$, controls how the flow scales phase-space volume — which is why $\operatorname{tr}A = 0$ characterizes volume-preserving (Hamiltonian-like) linear flows.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: Two Mixing Tanks

**Problem Statement:** Two tanks each hold $100$ litres of brine and are connected so that liquid flows at
$10$ L/min from tank 1 to tank 2 and $10$ L/min back. Tank 1 initially contains $20$ kg of salt and tank 2
none. Set up the system, compute $e^{At}$, find the salt in each tank at time $t$, and give the half-life
of the imbalance.

**Intuition:** The total salt is conserved (a zero eigenvalue) and the difference relaxes exponentially (the negative eigenvalue).

**Solution:**

**Step 1 (model).** With $x_i(t)$ the kilograms of salt in tank $i$ and concentration $x_i/100$,

$$
x_1' = -\frac{10}{100}x_1 + \frac{10}{100}x_2, \qquad x_2' = \frac{10}{100}x_1 - \frac{10}{100}x_2
$$

$$
\mathbf{x}' = A\mathbf{x}, \qquad A = 0.1\begin{pmatrix}-1 & 1\\ 1 & -1\end{pmatrix}, \qquad \mathbf{x}(0) = \begin{pmatrix}20\\0\end{pmatrix}
$$

**Step 2 (spectrum).** $\operatorname{tr}A = -0.2$, $\det A = 0$, so

$$
\lambda_1 = 0 \ \left(\mathbf{v}_1 = (1,1)^T\right), \qquad \lambda_2 = -0.2 \ \left(\mathbf{v}_2 = (1,-1)^T\right)
$$

The zero eigenvalue is **conservation of total salt**: $\frac{d}{dt}(x_1+x_2) = 0$.

**Step 3 (matrix exponential).** In the orthogonal eigenbasis, with projectors
$P_1 = \frac12\begin{pmatrix}1&1\\1&1\end{pmatrix}$ and $P_2 = \frac12\begin{pmatrix}1&-1\\-1&1\end{pmatrix}$:

$$
e^{At} = P_1 + e^{-0.2t}P_2 = \frac12\begin{pmatrix} 1+e^{-0.2t} & 1-e^{-0.2t}\\ 1-e^{-0.2t} & 1+e^{-0.2t}\end{pmatrix}
$$

**Step 4 (the solution).**

$$
\mathbf{x}(t) = e^{At}\begin{pmatrix}20\\0\end{pmatrix} = \begin{pmatrix} 10\left(1+e^{-0.2t}\right) \\ 10\left(1-e^{-0.2t}\right)\end{pmatrix}
$$

Checks: $\mathbf{x}(0) = (20,0)^T$ $\checkmark$; $x_1+x_2 = 20$ for all $t$ $\checkmark$;
$\mathbf{x}(\infty) = (10,10)^T$ — the salt equalizes, as physical sense demands.

**Step 5 (half-life of the imbalance).** The difference $d(t) = x_1-x_2 = 20e^{-0.2t}$ obeys
$d' = -0.2d$, so

$$
t_{1/2} = \frac{\ln 2}{0.2} \approx 3.47 \ \text{minutes}
$$

and equilibration is essentially complete after about $5$ time constants,
$5/0.2 = 25$ minutes ($d \approx 0.13$ kg).

$$
\boxed{x_1(t) = 10\left(1+e^{-0.2t}\right), \quad x_2(t) = 10\left(1-e^{-0.2t}\right), \quad t_{1/2} \approx 3.47 \ \text{min}}
$$

> **Key Takeaway:** A zero eigenvalue is a conservation law and its eigenvector is the conserved direction; the remaining spectrum gives the relaxation rates toward the corresponding equilibrium manifold.

### Problem L2.2: The Damped Oscillator as a First-Order System

**Problem Statement:** Write $\ddot{y} + 2\dot{y} + 4y = 0$ as a system, compute $e^{At}$ in closed form
using the identity $(A+I)^2 = -3I$, and identify the damping ratio, damped frequency, and decay envelope.

**Intuition:** Once the "rotation part" of the matrix squares to a negative multiple of the identity, the exponential collapses to sines and cosines.

**Solution:**

**Step 1 (companion form).** With $x_1 = y$, $x_2 = \dot{y}$,

$$
A = \begin{pmatrix} 0 & 1 \\ -4 & -2\end{pmatrix}, \qquad \lambda^2 + 2\lambda + 4 = 0 \ \Longrightarrow\ \lambda = -1 \pm i\sqrt3
$$

**Step 2 (the algebraic shortcut).** Set $B = A + I = \begin{pmatrix}1&1\\-4&-1\end{pmatrix}$. Then

$$
B^2 = \begin{pmatrix}1-4 & 1-1\\ -4+4 & -4+1\end{pmatrix} = \begin{pmatrix}-3&0\\0&-3\end{pmatrix} = -3I
$$

so $B$ plays the role of $i\sqrt3$. Since $-I$ commutes with $B$,

$$
e^{At} = e^{-t}e^{Bt}
$$

**Step 3 (exponentiate $B$).** Splitting the series into even and odd powers with $B^2 = -3I$:

$$
e^{Bt} = \cos\!\left(\sqrt3\,t\right)I + \frac{\sin\!\left(\sqrt3\,t\right)}{\sqrt3}\,B
$$

**Step 4 (assemble).**

$$
e^{At} = e^{-t}\begin{pmatrix} \cos\sqrt3 t + \frac{1}{\sqrt3}\sin\sqrt3 t & \frac{1}{\sqrt3}\sin\sqrt3 t \\[2pt] -\frac{4}{\sqrt3}\sin\sqrt3 t & \cos\sqrt3 t - \frac{1}{\sqrt3}\sin\sqrt3 t\end{pmatrix}
$$

Checks: at $t=0$ this is $I$ $\checkmark$; its derivative at $t=0$ is $-I + B = A$ $\checkmark$;
$\det e^{At} = e^{-2t} = e^{t\operatorname{tr}A}$ $\checkmark$.

**Step 5 (physical parameters).** Comparing $\ddot y + 2\zeta\omega_n\dot y + \omega_n^2y = 0$:

$$
\omega_n = 2, \qquad 2\zeta\omega_n = 2 \Rightarrow \zeta = \tfrac12, \qquad \omega_d = \omega_n\sqrt{1-\zeta^2} = 2\cdot\frac{\sqrt3}{2} = \sqrt3
$$

**Underdamped** ($\zeta \lt 1$): the response rings at $\omega_d = \sqrt3 \approx 1.732$ rad/s inside the
envelope $e^{-t}$, with quality factor $Q = \frac{1}{2\zeta} = 1$. The amplitude falls by $e^{-1}$ in one
second and by $e^{-2\pi/\sqrt3} \approx 0.027$ per oscillation period $2\pi/\sqrt3 \approx 3.63$ s.

$$
\boxed{e^{At} = e^{-t}\left[\cos\!\left(\sqrt3t\right)I + \tfrac{1}{\sqrt3}\sin\!\left(\sqrt3t\right)(A+I)\right], \quad \zeta = \tfrac12,\ \omega_d = \sqrt3}
$$

> **Key Takeaway:** The eigenvalues $-\zeta\omega_n \pm i\omega_d$ carry the entire physics: real part is the decay rate, imaginary part is the ringing frequency, and the ratio is the damping ratio.

### Problem L2.3: A Continuous-Time Markov Chain

**Problem Statement:** A two-state chain has generator
$Q = \begin{pmatrix} -a & a \\ b & -b\end{pmatrix}$ with $a, b \gt 0$. Compute the transition matrix
$P(t) = e^{Qt}$ in closed form, verify that its rows are probability vectors, and identify the stationary
distribution and the relaxation time. Evaluate for $a = 2$, $b = 3$.

**Intuition:** The generator's zero eigenvalue carries the stationary distribution; the other eigenvalue sets how fast the chain forgets its start.

**Solution:**

**Step 1 (spectrum).** $Q\mathbf{1} = \mathbf{0}$ (rows sum to zero), so $\lambda_1 = 0$ with right
eigenvector $\mathbf{1} = (1,1)^T$. Since $\operatorname{tr}Q = -(a+b)$,

$$
\lambda_2 = -(a+b)
$$

with left eigenvector $\boldsymbol{\pi}$ solving $\boldsymbol{\pi}Q = \mathbf{0}$.

**Step 2 (spectral form).** Writing $e^{Qt} = R_1 + e^{-(a+b)t}R_2$ with $R_1 = \mathbf{1}\boldsymbol{\pi}$
the rank-one projector onto the stationary direction and $R_2 = I - R_1$:

$$
P(t) = \frac{1}{a+b}\begin{pmatrix} b + a e^{-(a+b)t} & a\left(1 - e^{-(a+b)t}\right) \\ b\left(1-e^{-(a+b)t}\right) & a + b e^{-(a+b)t}\end{pmatrix}
$$

**Step 3 (verification).**

- $P(0) = \frac{1}{a+b}\begin{pmatrix}b+a & 0\\ 0 & a+b\end{pmatrix} = I$ $\checkmark$
- Row sums: $\frac{b+ae^{-\cdot}+a-ae^{-\cdot}}{a+b} = \frac{a+b}{a+b} = 1$ $\checkmark$, and all entries
  are nonnegative, so each row is a probability distribution.
- $P'(0) = \frac{1}{a+b}\begin{pmatrix}-a(a+b) & a(a+b)\\ b(a+b) & -b(a+b)\end{pmatrix} = Q$ $\checkmark$

**Step 4 (stationary distribution and relaxation).** As $t\to\infty$ both rows converge to

$$
\boldsymbol{\pi} = \left(\frac{b}{a+b},\ \frac{a}{a+b}\right)
$$

independent of the starting state, and the approach is exponential with **relaxation time**

$$
\tau = \frac{1}{a+b} \qquad \text{(spectral gap } = a+b\text{)}
$$

**Step 5 (numbers for $a = 2$, $b = 3$).** $\boldsymbol{\pi} = (0.6,\ 0.4)$, $a+b = 5$, $\tau = 0.2$, and

$$
P(t) = \begin{pmatrix} 0.6 + 0.4e^{-5t} & 0.4\left(1-e^{-5t}\right) \\ 0.6\left(1-e^{-5t}\right) & 0.4 + 0.6e^{-5t}\end{pmatrix}
$$

At $t = 0.2$ ($=\tau$): $e^{-1} = 0.368$, so $P_{11} = 0.747$ — already $63\%$ of the way to stationarity.
At $t = 1$: $e^{-5} = 0.0067$ and $P_{11} = 0.6027$, essentially equilibrated.

$$
\boxed{P(t) = \frac{1}{a+b}\begin{pmatrix} b+ae^{-(a+b)t} & a\left(1-e^{-(a+b)t}\right)\\ b\left(1-e^{-(a+b)t}\right) & a+be^{-(a+b)t}\end{pmatrix}, \quad \boldsymbol{\pi} = \frac{(b,\,a)}{a+b}, \quad \tau = \frac{1}{a+b}}
$$

> **Key Takeaway:** $P(t) = e^{Qt}$ is the Kolmogorov forward/backward equation solved in one stroke; the zero eigenvalue gives the equilibrium and the spectral gap gives the mixing time — the same structure that governs MCMC convergence.

### Problem L2.4: Heat Flow on a Graph

**Problem Statement:** For the path graph on three nodes ($1 - 2 - 3$) with Laplacian
$L = \begin{pmatrix}1&-1&0\\-1&2&-1\\0&-1&1\end{pmatrix}$, solve $\mathbf{u}' = -L\mathbf{u}$ with
$\mathbf{u}(0) = (1,0,0)^T$. Identify the conserved quantity, the equilibrium, and the relaxation rate.

**Intuition:** The graph Laplacian is the discrete diffusion operator; $e^{-Lt}$ spreads mass along edges while conserving the total.

**Solution:**

**Step 1 (spectrum of $L$).** For the path $P_3$ the eigenvalues are

$$
\mu_1 = 0,\quad \mu_2 = 1,\quad \mu_3 = 3
$$

with orthogonal eigenvectors

$$
\mathbf{w}_1 = (1,1,1)^T, \qquad \mathbf{w}_2 = (1,0,-1)^T, \qquad \mathbf{w}_3 = (1,-2,1)^T
$$

(Verify $\mathbf{w}_3$: $L\mathbf{w}_3 = (1+2,\ -1-4-1,\ 2+1)^T = (3,-6,3)^T = 3\mathbf{w}_3$ $\checkmark$.)

**Step 2 (expand the initial condition).** With $\lVert \mathbf{w}_1\rVert^2 = 3$,
$\lVert \mathbf{w}_2\rVert^2 = 2$, $\lVert \mathbf{w}_3\rVert^2 = 6$ and
$\mathbf{u}_0 = (1,0,0)^T$:

$$
\mathbf{u}_0 = \frac{1}{3}\mathbf{w}_1 + \frac{1}{2}\mathbf{w}_2 + \frac{1}{6}\mathbf{w}_3
$$

(coefficients $\mathbf{w}_i^T\mathbf{u}_0/\lVert \mathbf{w}_i\rVert^2 = 1/3,\ 1/2,\ 1/6$).

**Step 3 (evolve).** Each mode decays at its own rate $e^{-\mu_it}$:

$$
\mathbf{u}(t) = \frac{1}{3}\begin{pmatrix}1\\1\\1\end{pmatrix} + \frac{e^{-t}}{2}\begin{pmatrix}1\\0\\-1\end{pmatrix} + \frac{e^{-3t}}{6}\begin{pmatrix}1\\-2\\1\end{pmatrix}
$$

componentwise

$$
u_1 = \frac13+\frac{e^{-t}}{2}+\frac{e^{-3t}}{6}, \qquad u_2 = \frac13-\frac{e^{-3t}}{3}, \qquad u_3 = \frac13-\frac{e^{-t}}{2}+\frac{e^{-3t}}{6}
$$

Check $t=0$: $(1/3+1/2+1/6,\ 0,\ 1/3-1/2+1/6) = (1,0,0)$ $\checkmark$.

**Step 4 (conservation and equilibrium).** $\mathbf{1}^TL = \mathbf{0}$, so
$\frac{d}{dt}\sum_iu_i = -\mathbf{1}^TL\mathbf{u} = 0$: total heat is conserved at $1$. As $t\to\infty$
every nonzero mode dies and

$$
\mathbf{u}(\infty) = \left(\tfrac13,\tfrac13,\tfrac13\right)^T
$$

the uniform distribution — the discrete statement that heat equalizes on a connected graph.

**Step 5 (relaxation rate).** The slowest nonzero mode decays at $\mu_2 = 1$, the **algebraic connectivity**
(Fiedler value). So $\lVert \mathbf{u}(t)-\mathbf{u}(\infty)\rVert = O(e^{-\mu_2 t}) = O(e^{-t})$; a
better-connected graph has larger $\mu_2$ and mixes faster. Note also that node $2$ (the centre) approaches
equilibrium at rate $3$ while the endpoints relax at rate $1$: the antisymmetric endpoint mode is the
bottleneck.

$$
\boxed{\mathbf{u}(t) = \tfrac13\mathbf{1} + \tfrac{e^{-t}}{2}(1,0,-1)^T + \tfrac{e^{-3t}}{6}(1,-2,1)^T, \quad \mathbf{u}(\infty) = \tfrac13\mathbf{1}, \quad \text{rate } \mu_2 = 1}
$$

> **Key Takeaway:** $e^{-Lt}$ is the graph heat kernel: its zero eigenvalue conserves mass, its spectral gap sets the mixing rate, and the same operator underlies diffusion maps, spectral clustering and graph neural network smoothing.

### Problem L2.5: Vanishing and Exploding Gradients in a Linear RNN

**Problem Statement:** A linear recurrent network has hidden state $\mathbf{h}_{k+1} = W\mathbf{h}_k$, and
its continuous-time analogue is $\mathbf{h}' = A\mathbf{h}$. Express the backpropagated gradient over $T$
steps, relate its growth to the spectrum, and state the condition on $A$ that a continuous-time model
(e.g. a state-space layer) must satisfy for stable long-range memory.

**Intuition:** Backpropagation through time multiplies the same matrix $T$ times, so the spectral radius is raised to the power $T$ — a knife-edge.

**Solution:**

**Step 1 (the gradient is a matrix power).** With $\mathbf{h}_T = W^T\mathbf{h}_0$,

$$
\frac{\partial \mathbf{h}_T}{\partial \mathbf{h}_0} = W^{T}
$$

so a loss gradient at step $T$ reaches step $0$ multiplied by $\left(W^{T}\right)^{\!\top}$.

**Step 2 (spectral growth).** If $W = P\Lambda P^{-1}$ is diagonalizable with spectral radius
$\rho(W) = \max_i\lvert \lambda_i\rvert$, then asymptotically

$$
\lVert W^T\rVert \sim C\,\rho(W)^{T}
$$

Hence

- $\rho(W) \lt 1$: **vanishing gradients**, $\rho^T \to 0$ exponentially. With $\rho = 0.9$ and $T = 100$:
  $0.9^{100} \approx 2.7\times10^{-5}$; at $T = 1000$, $\approx 1.7\times10^{-46}$ — no learning signal
  survives.
- $\rho(W) \gt 1$: **exploding gradients**. With $\rho = 1.1$ and $T = 100$: $1.1^{100} \approx 1.4\times10^{4}$.
- $\rho(W) = 1$ exactly: the borderline that orthogonal and unitary RNNs enforce by construction.

**Step 3 (continuous time).** For $\mathbf{h}' = A\mathbf{h}$ the propagator over horizon $t$ is $e^{At}$
and

$$
\lVert e^{At}\rVert \sim C\,e^{\,t\,\max_i\operatorname{Re}\lambda_i(A)}
$$

so $\operatorname{Re}\lambda$ replaces $\log\lvert \lambda\rvert$:

$$
\max_i\operatorname{Re}\lambda_i(A) \ \begin{cases} \lt 0 & \text{gradients vanish} \\ = 0 & \text{marginal, long memory} \\ \gt 0 & \text{gradients explode}\end{cases}
$$

**Step 4 (design rule for state-space models).** S4/Mamba-style layers parametrize $A$ so that
$\operatorname{Re}\lambda_i \lt 0$ but *small in magnitude*: strictly stable (bounded outputs) yet with
time constants $1/\lvert \operatorname{Re}\lambda_i\rvert$ spanning many orders of magnitude, so that some
modes retain information over thousands of steps. Typical parametrizations write
$\lambda = -\exp(\theta_{\text{re}}) + i\,\theta_{\text{im}}$, which enforces stability by construction —
the continuous analogue of constraining $\rho(W) \lt 1$ without letting it be much less than $1$.

**Step 5 (the non-normal caveat).** The spectral estimate is only *asymptotic*. If $A$ is far from normal,
$\lVert e^{At}\rVert$ can grow by orders of magnitude before decaying (Problem L3.4), so a "stable"
spectrum does not guarantee well-behaved gradients at moderate horizons — one reason orthogonal/normal
parametrizations are preferred in practice.

$$
\boxed{\left\lVert \frac{\partial\mathbf{h}_T}{\partial\mathbf{h}_0}\right\rVert \sim \rho(W)^T \ \text{(discrete)}, \qquad \lVert e^{At}\rVert \sim e^{t\max_i\operatorname{Re}\lambda_i} \ \text{(continuous)}}
$$

> **Key Takeaway:** Long-range memory is a spectral design problem: put the eigenvalues on (or just inside) the stability boundary — $\lvert \lambda\rvert = 1$ discretely, $\operatorname{Re}\lambda = 0$ continuously — and keep the eigenvector basis well conditioned.

### Problem L2.6: Zero-Order-Hold Discretization $\bar{A} = e^{A\Delta}$

**Problem Statement:** For $\mathbf{x}' = A\mathbf{x} + B u(t)$ with $u$ held constant on each interval
$[k\Delta,(k+1)\Delta)$, derive the exact discrete recursion $\mathbf{x}_{k+1} = \bar{A}\mathbf{x}_k + \bar{B}u_k$
and identify $\bar{A},\bar{B}$. Evaluate for $A = \operatorname{diag}(-0.1,-1,-10)$, $\Delta = 0.1$, and
interpret the numbers.

**Intuition:** Integrating the forced equation exactly over one sampling interval turns the ODE into a linear recursion with no discretization error at the sample points.

**Solution:**

**Step 1 (Duhamel over one interval).** The variation-of-constants formula gives, for
$t\in[k\Delta,(k+1)\Delta]$,

$$
\mathbf{x}(t) = e^{A(t-k\Delta)}\mathbf{x}(k\Delta) + \int_{k\Delta}^{t}e^{A(t-s)}Bu(s)\,ds
$$

**Step 2 (use the hold).** With $u(s) \equiv u_k$ on the interval, substitute $\tau = t-s$ at
$t = (k+1)\Delta$:

$$
\mathbf{x}_{k+1} = e^{A\Delta}\mathbf{x}_k + \left(\int_0^{\Delta}e^{A\tau}d\tau\right)Bu_k
$$

**Step 3 (identify the matrices).**

$$
\bar{A} = e^{A\Delta}, \qquad \bar{B} = \left(\int_0^{\Delta}e^{A\tau}d\tau\right)B = A^{-1}\left(e^{A\Delta}-I\right)B \quad (A \ \text{invertible})
$$

The integral formula is valid even when $A$ is singular; the $A^{-1}$ form is just a convenient closed
expression. This recursion is **exact at the sample times** — no truncation error whatsoever, which is why
zero-order hold is the standard discretization for state-space layers.

**Step 4 (the diagonal instance).** With $A = \operatorname{diag}(-0.1,-1,-10)$ and $\Delta = 0.1$:

$$
\bar{A} = \operatorname{diag}\left(e^{-0.01},\ e^{-0.1},\ e^{-1}\right) = \operatorname{diag}\left(0.99005,\ 0.90484,\ 0.36788\right)
$$

and, with $B = \mathbf{1}$, $\bar{B}_j = \frac{e^{a_j\Delta}-1}{a_j}$:

$$
\bar{B} = \left(\frac{0.99005-1}{-0.1},\ \frac{0.90484-1}{-1},\ \frac{0.36788-1}{-10}\right) = (0.09950,\ 0.09516,\ 0.06321)
$$

**Step 5 (interpretation).** The continuous time constants are $1/0.1 = 10$, $1$, $0.1$; in units of
$\Delta = 0.1$ these are $100$, $10$, $1$ steps. Accordingly:

- The slow mode retains $99\%$ of its value per step and can carry information over hundreds of steps —
  this is the mechanism by which SSMs achieve long-range memory.
- The fast mode retains only $37\%$ per step and is effectively local.
- All $\lvert \bar{\lambda}_j\rvert \lt 1$, so the discrete system is stable — guaranteed because
  $\operatorname{Re}\lambda_j \lt 0$ and $\lvert e^{\lambda\Delta}\rvert = e^{\Delta\operatorname{Re}\lambda}$.
  **ZOH maps the left half-plane into the unit disc**, so continuous stability transfers exactly, unlike
  forward Euler ($1+\lambda\Delta$), which requires $\Delta \lt 2/\lvert \lambda\rvert$ and would already be
  marginal here for the fast mode.

$$
\boxed{\bar{A} = e^{A\Delta}, \quad \bar{B} = A^{-1}\left(e^{A\Delta}-I\right)B; \quad \bar{A} = \operatorname{diag}(0.990,\ 0.905,\ 0.368)}
$$

> **Key Takeaway:** $\Delta$ is a learnable timescale: it rescales the entire spectrum through $e^{\lambda\Delta}$, letting one architecture cover memory horizons from a few steps to thousands by moving eigenvalues toward or away from the unit circle.

## Level 3 — Challenge

### Problem L3.1: The Exact Boundedness Criterion

**Problem Statement:** Prove that all solutions of $\mathbf{x}' = A\mathbf{x}$ are bounded on $[0,\infty)$
**if and only if** every eigenvalue satisfies $\operatorname{Re}\lambda \le 0$ and every eigenvalue with
$\operatorname{Re}\lambda = 0$ is semisimple (its algebraic and geometric multiplicities agree). Give a
counterexample showing that $\operatorname{Re}\lambda \le 0$ alone is not enough.

**Intuition:** A Jordan block on the imaginary axis produces polynomial factors with nothing to damp them.

**Solution:**

**Step 1 (the general form of solutions).** Over $\mathbb{C}$, put $A$ in Jordan form
$A = SJS^{-1}$ with blocks $J_i = \lambda_iI + N_i$ of size $m_i$. Then
$e^{At} = Se^{Jt}S^{-1}$ and each block contributes

$$
e^{J_it} = e^{\lambda_it}\left(I + N_it + \dots + \frac{N_i^{m_i-1}t^{m_i-1}}{(m_i-1)!}\right)
$$

so every entry of $e^{At}$ is a finite combination of terms $t^ke^{\lambda_it}$ with
$0\le k \le m_i - 1$.

**Step 2 (sufficiency).** Suppose $\operatorname{Re}\lambda_i \le 0$ for all $i$, and $m_i = 1$ whenever
$\operatorname{Re}\lambda_i = 0$.

- If $\operatorname{Re}\lambda_i \lt 0$, then $\lvert t^ke^{\lambda_it}\rvert = t^ke^{t\operatorname{Re}\lambda_i} \to 0$
  and is bounded on $[0,\infty)$ (exponentials beat polynomials).
- If $\operatorname{Re}\lambda_i = 0$, semisimplicity forces $k = 0$ and
  $\lvert e^{\lambda_it}\rvert = 1$: bounded.

Every entry of $e^{At}$ is therefore bounded, hence so is $\lVert e^{At}\rVert$, and every solution
$\mathbf{x}(t) = e^{At}\mathbf{x}_0$ is bounded.

**Step 3 (necessity, part 1).** If some $\operatorname{Re}\lambda \gt 0$ with eigenvector $\mathbf{v}$
(possibly complex), then $\mathbf{x}(t) = e^{\lambda t}\mathbf{v}$ has
$\lVert \mathbf{x}(t)\rVert = e^{t\operatorname{Re}\lambda}\lVert \mathbf{v}\rVert\to\infty$; taking real
and imaginary parts gives an unbounded real solution.

**Step 4 (necessity, part 2).** Suppose $\operatorname{Re}\lambda = 0$ but the block for $\lambda$ has size
$m \ge 2$. Then there is a generalized eigenvector $\mathbf{w}$ with $(A-\lambda I)\mathbf{w} = \mathbf{v} \neq \mathbf{0}$,
and

$$
\mathbf{x}(t) = e^{\lambda t}\left(\mathbf{w} + t\mathbf{v}\right)
$$

solves the system with $\lVert \mathbf{x}(t)\rVert \ge t\lVert \mathbf{v}\rVert - \lVert \mathbf{w}\rVert \to\infty$
(using $\lvert e^{\lambda t}\rvert = 1$). Unbounded.

**Step 5 (the counterexample).** Take

$$
A = \begin{pmatrix}0 & 1\\ 0 & 0\end{pmatrix}, \qquad e^{At} = \begin{pmatrix}1 & t\\ 0 & 1\end{pmatrix}
$$

Both eigenvalues are $0$, so $\operatorname{Re}\lambda \le 0$ holds, but $\lambda = 0$ is defective and
$\mathbf{x}(t) = (t, 1)^T$ grows linearly. Physically this is a free particle, $\ddot y = 0$: no restoring
force and no damping, so the position drifts forever. Similarly
$A = \begin{pmatrix}0&1&0&0\\-\omega^2&0&0&0\\0&0&0&1\\0&0&-\omega^2&0\end{pmatrix}$-type resonant blocks
produce $t\sin\omega t$ — secular growth.

$$
\boxed{\text{bounded} \iff \operatorname{Re}\lambda_i \le 0 \ \forall i \ \text{and every } \lambda_i \ \text{with} \ \operatorname{Re}\lambda_i = 0 \ \text{is semisimple}}
$$

Asymptotic stability ($\mathbf{x}(t)\to\mathbf{0}$ for every $\mathbf{x}_0$) is the strictly stronger
condition $\operatorname{Re}\lambda_i \lt 0$ for all $i$.

> **Key Takeaway:** Stability is a spectral condition *plus* an eigenvector-completeness condition on the boundary of the spectrum — the marginal case is exactly where the Jordan structure becomes visible in the dynamics.

### Problem L3.2: Putzer's Algorithm on a Defective $3\times3$

**Problem Statement:** Use Putzer's algorithm to compute $e^{At}$ for

$$
A = \begin{pmatrix} 1 & 1 & 0 \\ 0 & 1 & 0 \\ 1 & 0 & 2\end{pmatrix}
$$

Verify that $A$ is defective, and check the result at $t = 0$, by differentiation, and against Jacobi's
formula.

**Intuition:** Putzer replaces eigenvectors with a triangular chain of scalar ODEs, so a defective spectrum causes no extra difficulty.

**Solution:**

**Step 1 (spectrum).** Expanding $\det(A-\lambda I)$ along the third column (entries $0,0,2-\lambda$):

$$
\det(A-\lambda I) = (2-\lambda)\det\begin{pmatrix}1-\lambda & 1\\ 0 & 1-\lambda\end{pmatrix} = (2-\lambda)(1-\lambda)^2
$$

so the eigenvalues are $\lambda = 1$ (twice) and $\lambda = 2$.

**Step 2 (defectiveness).** $A - I = \begin{pmatrix}0&1&0\\0&0&0\\1&0&1\end{pmatrix}$ has rank $2$, so its
null space is one-dimensional: $\lambda = 1$ has geometric multiplicity $1 \lt 2$. **Defective**, so
diagonalization is unavailable.

**Step 3 (Putzer's recipe).** Order the eigenvalues $\lambda_1 = 1$, $\lambda_2 = 1$, $\lambda_3 = 2$ and
set $P_0 = I$, $P_1 = A-\lambda_1I$, $P_2 = (A-\lambda_2I)(A-\lambda_1I)$. Solve the triangular system

$$
r_1' = \lambda_1r_1,\ r_1(0)=1; \qquad r_j' = \lambda_jr_j + r_{j-1},\ r_j(0)=0
$$

- $r_1' = r_1 \Rightarrow r_1 = e^t$.
- $r_2' = r_2 + e^t,\ r_2(0)=0 \Rightarrow r_2 = te^t$.
- $r_3' = 2r_3 + te^t,\ r_3(0)=0$. Integrating factor $e^{-2t}$:

$$
r_3 = e^{2t}\int_0^t se^{-s}\,ds = e^{2t}\left[1-(1+t)e^{-t}\right] = e^{2t} - (1+t)e^{t}
$$

**Step 4 (the matrix powers).**

$$
P_1 = A - I = \begin{pmatrix}0&1&0\\0&0&0\\1&0&1\end{pmatrix}, \qquad P_2 = P_1^2 = \begin{pmatrix}0&0&0\\0&0&0\\1&1&1\end{pmatrix}
$$

**Step 5 (assemble).** $e^{At} = r_1P_0 + r_2P_1 + r_3P_2$:

$$
e^{At} = e^t I + te^t\begin{pmatrix}0&1&0\\0&0&0\\1&0&1\end{pmatrix} + \left(e^{2t}-(1+t)e^t\right)\begin{pmatrix}0&0&0\\0&0&0\\1&1&1\end{pmatrix}
$$

$$
= \begin{pmatrix} e^t & te^t & 0 \\ 0 & e^t & 0 \\ e^{2t}-e^t & e^{2t}-(1+t)e^t & e^{2t}\end{pmatrix}
$$

(the $(3,1)$ entry simplifies as $te^t + e^{2t}-(1+t)e^t = e^{2t}-e^t$, and the $(3,3)$ entry as
$e^t + te^t + e^{2t}-(1+t)e^t = e^{2t}$).

**Step 6 (three checks).**

- $t = 0$: $\begin{pmatrix}1&0&0\\0&1&0\\0&0&1\end{pmatrix}$ $\checkmark$
- Derivative at $t=0$: entries $1,\ 1,\ 0;\ 0,\ 1,\ 0;\ 2-1,\ 2-1-1,\ 2$, i.e.
  $\begin{pmatrix}1&1&0\\0&1&0\\1&0&2\end{pmatrix} = A$ $\checkmark$
- Jacobi: $\det e^{At} = e^t\cdot e^t\cdot e^{2t} = e^{4t}$ and $\operatorname{tr}A = 1+1+2 = 4$
  $\checkmark$

$$
\boxed{e^{At} = \begin{pmatrix} e^t & te^t & 0\\ 0 & e^t & 0\\ e^{2t}-e^t & e^{2t}-(1+t)e^t & e^{2t}\end{pmatrix}}
$$

> **Key Takeaway:** Putzer needs only the eigenvalues (with multiplicity) and the Cayley-Hamilton theorem — no eigenvectors, no Jordan basis — which makes it the method of choice for defective matrices worked by hand.

### Problem L3.3: The First BCH Commutator Term

**Problem Statement:** Expand $e^{Ah}e^{Bh}$ and $e^{(A+B)h}$ through order $h^2$ and show

$$
e^{Ah}e^{Bh} = \exp\left(\left(A+B\right)h + \frac{h^2}{2}\left[A,B\right] + O(h^3)\right)
$$

Deduce the local and global order of the Lie-Trotter splitting $\left(e^{Ah/n}e^{Bh/n}\right)^n$ and
explain why Strang splitting is better.

**Intuition:** The two products agree at first order; their difference at second order is exactly the term that measures non-commutativity.

**Solution:**

**Step 1 (expand the product).**

$$
e^{Ah}e^{Bh} = \left(I + Ah + \frac{A^2h^2}{2} + O(h^3)\right)\left(I + Bh + \frac{B^2h^2}{2}+O(h^3)\right)
$$

$$
= I + (A+B)h + \left(\frac{A^2}{2} + AB + \frac{B^2}{2}\right)h^2 + O(h^3)
$$

**Step 2 (expand the exact exponential).**

$$
e^{(A+B)h} = I + (A+B)h + \frac{(A+B)^2}{2}h^2 + O(h^3) = I + (A+B)h + \frac{A^2+AB+BA+B^2}{2}h^2 + O(h^3)
$$

**Step 3 (subtract).**

$$
e^{Ah}e^{Bh} - e^{(A+B)h} = h^2\left(AB - \frac{AB+BA}{2}\right) + O(h^3) = \frac{h^2}{2}\left(AB-BA\right) + O(h^3) = \frac{h^2}{2}[A,B] + O(h^3)
$$

**Step 4 (write it as a single exponential).** Since the discrepancy is $O(h^2)$ and
$e^{X+\epsilon} = e^X(I+\epsilon+O(\epsilon^2))$ for small $\epsilon$, absorbing the term into the exponent
gives

$$
e^{Ah}e^{Bh} = \exp\left((A+B)h + \frac{h^2}{2}[A,B] + O(h^3)\right)
$$

the first two terms of the Baker-Campbell-Hausdorff series
$\log(e^Xe^Y) = X+Y+\frac12[X,Y]+\frac{1}{12}\left([X,[X,Y]]+[Y,[Y,X]]\right)+\dots$.

**Step 5 (order of Lie-Trotter splitting).** One step of $e^{Ah}e^{Bh}$ has **local error** $O(h^2)$. Over
a fixed interval $T$ with $n = T/h$ steps, the errors accumulate linearly, giving **global error**
$O(h)$: Lie-Trotter is a **first-order** method, and

$$
\lim_{n\to\infty}\left(e^{AT/n}e^{BT/n}\right)^n = e^{(A+B)T}
$$

(the Lie product formula), valid for all $A,B$ regardless of commutation.

**Step 6 (why Strang is better).** The symmetrized (Strang) splitting

$$
S(h) = e^{Ah/2}\,e^{Bh}\,e^{Ah/2}
$$

satisfies $S(h)S(-h) = I$, i.e. it is time-symmetric, and a time-symmetric one-step map has only *even*
powers of $h$ in its error expansion. The $h^2$ commutator therefore cancels, leaving local error $O(h^3)$
and global error $O(h^2)$ — second order at essentially the same cost per step (consecutive half-steps
$e^{Ah/2}e^{Ah/2} = e^{Ah}$ can be merged). If $[A,B] = 0$ both splittings are exact.

$$
\boxed{e^{Ah}e^{Bh} = \exp\left((A+B)h + \tfrac{h^2}{2}[A,B]+O(h^3)\right); \ \text{Trotter } O(h), \ \text{Strang } O(h^2)}
$$

> **Key Takeaway:** The commutator is the exact obstruction to splitting, and its size tells you the price: operator-splitting integrators for advection-diffusion, quantum simulation, and symplectic mechanics are all engineered around cancelling BCH terms.

### Problem L3.4: Transient Growth of a Non-Normal Flow

**Problem Statement:** For $A = \begin{pmatrix} -1 & M \\ 0 & -2\end{pmatrix}$ with $M \gg 1$, compute
$e^{At}$ exactly, find the time and size of the maximum of the $(1,2)$ entry, and explain why the spectrum
alone fails to predict short-time behaviour. Take $M = 100$.

**Intuition:** Eigenvalues govern the $t \to \infty$ limit; when eigenvectors are nearly parallel, the flow can amplify enormously before that limit takes hold.

**Solution:**

**Step 1 (solve the triangular system).** The second equation decouples:
$x_2' = -2x_2 \Rightarrow x_2(t) = e^{-2t}x_2(0)$. Then

$$
x_1' = -x_1 + Mx_2 = -x_1 + Me^{-2t}x_2(0)
$$

Using the integrating factor $e^{t}$,

$$
x_1(t) = e^{-t}x_1(0) + M\int_0^te^{-(t-s)}e^{-2s}\,ds\ x_2(0) = e^{-t}x_1(0) + M\left(e^{-t}-e^{-2t}\right)x_2(0)
$$

**Step 2 (the exponential).**

$$
e^{At} = \begin{pmatrix} e^{-t} & M\left(e^{-t}-e^{-2t}\right) \\ 0 & e^{-2t}\end{pmatrix}
$$

Checks: $e^{A\cdot0} = I$ $\checkmark$; derivative at $0$ gives $\begin{pmatrix}-1 & M\\0&-2\end{pmatrix} = A$
$\checkmark$; $\det e^{At} = e^{-3t} = e^{t\operatorname{tr}A}$ $\checkmark$.

**Step 3 (maximize the coupling entry).** Let $g(t) = M\left(e^{-t}-e^{-2t}\right)$. Then

$$
g'(t) = M\left(-e^{-t}+2e^{-2t}\right) = 0 \ \Longrightarrow\ e^{-t} = 2e^{-2t} \ \Longrightarrow\ e^{t} = 2 \ \Longrightarrow\ t^\star = \ln 2
$$

$$
g(t^\star) = M\left(\tfrac12 - \tfrac14\right) = \frac{M}{4}
$$

**Step 4 (the numbers for $M = 100$).** The maximum coupling is $25$, attained at
$t^\star = \ln2 \approx 0.693$. Starting from $\mathbf{x}_0 = (0,1)^T$ (unit norm), the state at $t^\star$ is
$(25,\ 0.25)^T$ with norm $\approx 25$: the trajectory is amplified **$25$-fold** before decaying, even
though both eigenvalues are $-1$ and $-2$ and the asymptotic decay is $e^{-t}$.

**Step 5 (why the spectrum misleads).** The two eigenvectors are $(1,0)^T$ and $(M, -1)^T$, whose normalized
inner product is $\frac{M}{\sqrt{M^2+1}} \to 1$: they are nearly **parallel**, so the eigenbasis is
ill-conditioned with $\kappa(P) \approx M$. The bound

$$
\lVert e^{At}\rVert \le \kappa(P)\,e^{t\max_i\operatorname{Re}\lambda_i}
$$

carries the factor $\kappa(P) \approx 100$, and that constant is exactly what is realized transiently. For
**normal** matrices ($AA^T = A^TA$) the eigenbasis is orthogonal, $\kappa(P)=1$, and $\lVert e^{At}\rVert = e^{t\max\operatorname{Re}\lambda}$
exactly — no transient growth is possible.

**Step 6 (consequences).** Transient amplification of this kind is the linear mechanism behind subcritical
transition to turbulence in shear flows (where $M$ plays the role of the Reynolds number), the reason
pseudospectra rather than spectra are the right diagnostic for non-normal operators, and a practical
caution for recurrent and state-space models (Problem L2.5): a spectrally stable $A$ with a badly
conditioned eigenbasis can still produce exploding activations over moderate horizons.

$$
\boxed{e^{At} = \begin{pmatrix} e^{-t} & M\left(e^{-t}-e^{-2t}\right)\\ 0 & e^{-2t}\end{pmatrix}, \quad \max_t \ \text{coupling} = \frac{M}{4} = 25 \ \text{at} \ t = \ln 2}
$$

> **Key Takeaway:** Eigenvalues describe the destination, not the journey: for non-normal $A$ the transient can exceed the asymptotic behaviour by the condition number of the eigenvector basis, which is why stability analysis must look beyond the spectrum.
