### Problem L0.1: Why Transforms Help

**Problem Statement:** In one paragraph plus one displayed equation, explain *mechanically* why the Laplace transform turns a linear constant-coefficient IVP into an algebra problem. Which theorem does the work, and where do the initial conditions go?

**Intuition:** The transform is a dictionary in which "differentiate" translates to "multiply by $s$".

**Solution:**

The Laplace transform is linear, and the derivative theorem states $\mathcal{L}\{y'\} = sY(s) - y(0)$ and $\mathcal{L}\{y''\} = s^2 Y(s) - s y(0) - y'(0)$. Applying $\mathcal{L}$ to $a y'' + b y' + c y = g(t)$ therefore replaces every derivative of the *unknown function* by a *polynomial in $s$* times the single unknown quantity $Y(s)$:

$$
(a s^2 + b s + c)\, Y(s) = G(s) + (a s + b)\, y(0) + a\, y'(0).
$$

No derivatives of the unknown remain — the equation is algebraic, solved by dividing by the characteristic polynomial $a s^2 + b s + c$. The initial conditions are not imposed afterwards, as with integration constants: they enter *during* the transformation, as the boundary terms of the integration by parts inside the derivative theorem, and appear as the explicit polynomial $(as + b) y(0) + a y'(0)$ on the right-hand side. Inversion (partial fractions plus Lerch's uniqueness theorem) then returns the unique continuous solution.

$$
\boxed{\text{Derivative theorem: differentiation} \mapsto \text{multiplication by } s, \text{ with } y(0), y'(0) \text{ absorbed as algebraic data.}}
$$

> **Key Takeaway:** The transform solves the whole IVP in one algebraic pass because the derivative theorem simultaneously eliminates calculus and injects the initial state.

### Problem L0.2: Region of Convergence

**Problem Statement:** For which real $s$ does $\mathcal{L}\{e^{3t}\}(s)$ converge? Compute the transform on that set. Does the divergence for $s \le 3$ make the resulting formula $\frac{1}{s-3}$ unusable there? Explain the role of the region of convergence when we later plug in small $s$ (e.g. in the final value theorem).

**Intuition:** The kernel $e^{-st}$ must out-decay the growth of $f$.

**Solution:**

**Step 1 — direct computation.** For real $s$:

$$
\int_0^T e^{-st} e^{3t}\, dt = \int_0^T e^{-(s-3)t}\, dt = \frac{1 - e^{-(s-3)T}}{s - 3} \quad (s \neq 3).
$$

As $T \to \infty$, the term $e^{-(s-3)T}$ tends to $0$ if and only if $s - 3 \gt 0$. For $s = 3$ the integral is $\int_0^T dt = T \to \infty$. Hence the transform exists exactly on the half-line

$$
\boxed{\ \mathcal{L}\{e^{3t}\}(s) = \frac{1}{s - 3}, \qquad \text{ROC: } s \gt 3.\ }
$$

**Step 2 — the formula versus the transform.** The rational function $\frac{1}{s-3}$ is defined for all $s \neq 3$, but it *represents the integral* only on the ROC $s \gt 3$; elsewhere it is the analytic continuation of the transform, not the transform itself. This distinction is harmless for inversion (partial-fraction coefficients are algebraic identities), but it is exactly where the final value theorem can silently fail: evaluating $sF(s)$ at $s = 0$ uses a point *outside* the ROC whenever $F$ has a pole with $\operatorname{Re}(p) \ge 0$, and then the computed number need not equal $\lim_{t \to \infty} f(t)$. Indeed here $sF(s) = \frac{s}{s-3} \to 0$ as $s \to 0$, while $e^{3t} \to \infty$.

> **Key Takeaway:** A Laplace transform is a formula *plus* a half-plane; theorems that evaluate near $s = 0$ are only trustworthy when the ROC (equivalently, the pole locations) permits it.

### Problem L0.3: Linearity and Its Limits

**Problem Statement:** (a) Using linearity, compute $\mathcal{L}\{5 - 2t + 3\sin 4t\}$. (b) A student claims $\mathcal{L}\{t \sin 4t\} = \mathcal{L}\{t\} \cdot \mathcal{L}\{\sin 4t\}$. Refute the claim, state the two theorems that correctly handle products of this kind, and give the correct transform.

**Intuition:** The transform respects sums, never products — products belong to convolution.

**Solution:**

**(a)** By linearity and the table ($\mathcal{L}\{1\} = 1/s$, $\mathcal{L}\{t\} = 1/s^2$, $\mathcal{L}\{\sin 4t\} = 4/(s^2+16)$), for $s \gt 0$:

$$
\mathcal{L}\{5 - 2t + 3 \sin 4t\} = \frac{5}{s} - \frac{2}{s^2} + \frac{12}{s^2 + 16}.
$$

**(b)** The claim is false: the transform is a *linear* map, and linear maps preserve sums and scalar multiples, not products. A direct refutation: the claimed answer $\frac{1}{s^2} \cdot \frac{4}{s^2+16}$ has a double pole at $s = 0$, which would force a mode growing like $t$ *without* oscillation, whereas $t \sin 4t$ is purely oscillatory with growing amplitude — poles must sit at $s = \pm 4i$ only.

The correct tools for products are: (i) **multiplication by $t$**: $\mathcal{L}\{t f(t)\} = -F'(s)$; and (ii) the **convolution theorem**, which says a *product of transforms* corresponds to a *convolution of functions*: $F(s)G(s) = \mathcal{L}\{f * g\}$, never $\mathcal{L}\{fg\}$. Applying (i):

$$
\mathcal{L}\{t \sin 4t\} = -\frac{d}{ds} \left( \frac{4}{s^2 + 16} \right) = \frac{8s}{(s^2 + 16)^2}.
$$

$$
\boxed{\ \mathcal{L}\{t \sin 4t\} = \frac{8s}{(s^2+16)^2}, \quad \text{poles at } s = \pm 4i \text{ (double), as expected for resonant growth.}\ }
$$

> **Key Takeaway:** Sums transform term by term; products of functions need the derivative-of-transform rule, and products of transforms mean convolution in time.

### Problem L0.4: What the Delta Function Means

**Problem Statement:** Define the Dirac delta $\delta(t - c)$ as a limit of pulses, explain why it is not a function, compute $\mathcal{L}\{\delta(t-c)\}$ from the limit definition, and explain the physical meaning of forcing a mechanical system with $I_0\, \delta(t - c)$.

**Intuition:** Delta is not a graph you can draw; it is an instruction — "sample here".

**Solution:**

**Step 1 — pulses.** Let $d_\varepsilon(t - c) = \frac{1}{\varepsilon}$ for $c \le t \le c + \varepsilon$ and $0$ otherwise: a rectangle of width $\varepsilon$, height $1/\varepsilon$, area $1$. For any continuous $g$, the mean value theorem for integrals gives $\int_0^\infty d_\varepsilon(t - c) g(t)\, dt = g(\xi_\varepsilon)$ for some $\xi_\varepsilon \in [c, c + \varepsilon]$, which tends to $g(c)$ as $\varepsilon \to 0^+$.

**Step 2 — why no function can be the limit.** The pointwise limit of $d_\varepsilon(t-c)$ is $0$ for $t \neq c$ and $\infty$ at $t = c$; any genuine function equal to $0$ almost everywhere has integral $0$, not $1$. So no function reproduces the sampling behavior of Step 1. Instead $\delta(t - c)$ is *defined* by that behavior — it is a distribution (a linear functional on test functions): $\int \delta(t - c)\, g(t)\, dt = g(c)$.

**Step 3 — the transform.** Applying the sampling rule to $g(t) = e^{-st}$, or taking the limit of pulse transforms:

$$
\mathcal{L}\{d_\varepsilon(t - c)\}(s) = \frac{1}{\varepsilon} \int_c^{c+\varepsilon} e^{-st}\, dt = e^{-cs}\, \frac{1 - e^{-s\varepsilon}}{s \varepsilon} \longrightarrow e^{-cs} \cdot 1 \quad (\varepsilon \to 0^+),
$$

using $1 - e^{-x} = x - x^2/2 + \cdots$. Hence

$$
\boxed{\ \mathcal{L}\{\delta(t - c)\} = e^{-cs}, \qquad \mathcal{L}\{\delta\} = 1.\ }
$$

**Step 4 — physical meaning.** For $m x'' + kx = I_0 \delta(t - c)$, integrate the equation over $[c - \eta, c + \eta]$: the terms $\int m x''\, dt = m[x'(c+\eta) - x'(c-\eta)]$ and $\int I_0 \delta = I_0$ survive as $\eta \to 0$, while $\int kx\, dt \to 0$ since $x$ is bounded. So the velocity jumps by $I_0/m$ and the position is continuous: a delta force is an idealized hammer blow delivering momentum $I_0$ instantaneously. Its flat transform $\mathcal{L}\{\delta\} = 1$ says an impulse excites *all frequencies equally* — which is why the impulse response characterizes a linear system completely.

> **Key Takeaway:** Delta is the idealization of a unit-area pulse; it transforms to $e^{-cs}$, transfers momentum instantly, and probes every frequency of a system at once.

### Problem L1.1: Transform of Powers by Induction

**Problem Statement:** Prove by induction that $\mathcal{L}\{t^n\}(s) = \dfrac{n!}{s^{n+1}}$ for every integer $n \ge 0$ and $s \gt 0$, deriving the reduction formula you need with integration by parts.

**Intuition:** Each factor of $t$ costs one integration by parts, which trades $t^n$ for $\frac{n}{s} t^{n-1}$.

**Solution:**

**Step 1 — base case.** $\mathcal{L}\{t^0\} = \int_0^\infty e^{-st} dt = \frac{1}{s} = \frac{0!}{s^{0+1}}$ for $s \gt 0$.

**Step 2 — reduction formula.** For $n \ge 1$, integrate by parts with $u = t^n$, $dv = e^{-st} dt$, $v = -e^{-st}/s$:

$$
\int_0^T t^n e^{-st}\, dt = \left[ -\frac{t^n e^{-st}}{s} \right]_0^T + \frac{n}{s} \int_0^T t^{n-1} e^{-st}\, dt.
$$

The boundary term at $0$ vanishes since $n \ge 1$; at $T$ it is $-T^n e^{-sT}/s \to 0$ as $T \to \infty$ because exponentials beat polynomials for $s \gt 0$ (e.g. $T^n e^{-sT} \le n! (2/s)^n e^{-sT/2}$ from $e^{sT/2} \ge (sT/2)^n/n!$). Hence

$$
\mathcal{L}\{t^n\} = \frac{n}{s}\, \mathcal{L}\{t^{n-1}\}.
$$

**Step 3 — induction step.** Assume $\mathcal{L}\{t^{n-1}\} = \frac{(n-1)!}{s^n}$. Then by Step 2,

$$
\mathcal{L}\{t^n\} = \frac{n}{s} \cdot \frac{(n-1)!}{s^n} = \frac{n!}{s^{n+1}},
$$

completing the induction. $\blacksquare$

$$
\boxed{\ \mathcal{L}\{t^n\}(s) = \frac{n!}{s^{n+1}}, \qquad s \gt 0\ }
$$

> **Key Takeaway:** The factorial in the table is the accumulated cost of trading powers of $t$ for powers of $1/s$, one integration by parts at a time.

### Problem L1.2: Damped Oscillations via s-Shifting

**Problem Statement:** Starting only from $\mathcal{L}\{e^{ibt}\} = \frac{1}{s - ib}$ (for $s \gt 0$), derive $\mathcal{L}\{\cos bt\}$ and $\mathcal{L}\{\sin bt\}$, then use the s-shifting theorem to obtain $\mathcal{L}\{e^{at} \cos bt\}$ and $\mathcal{L}\{e^{at} \sin bt\}$. Verify the result for $\mathcal{L}\{e^{-t} \sin 2t\}$ and identify its poles.

**Intuition:** Damping in time is a horizontal slide in the $s$-plane.

**Solution:**

**Step 1 — split the complex exponential.** Rationalize:

$$
\mathcal{L}\{e^{ibt}\} = \frac{1}{s - ib} = \frac{s + ib}{s^2 + b^2}.
$$

Since $e^{ibt} = \cos bt + i \sin bt$ and the transform is linear (also over complex scalars), matching real and imaginary parts:

$$
\mathcal{L}\{\cos bt\} = \frac{s}{s^2 + b^2}, \qquad \mathcal{L}\{\sin bt\} = \frac{b}{s^2 + b^2}.
$$

**Step 2 — shift.** The s-shifting theorem $\mathcal{L}\{e^{at} f(t)\}(s) = F(s - a)$ replaces $s$ by $s - a$:

$$
\mathcal{L}\{e^{at} \cos bt\} = \frac{s - a}{(s-a)^2 + b^2}, \qquad \mathcal{L}\{e^{at} \sin bt\} = \frac{b}{(s-a)^2 + b^2}, \qquad s \gt a.
$$

**Step 3 — verification.** With $a = -1$, $b = 2$:

$$
\boxed{\ \mathcal{L}\{e^{-t} \sin 2t\} = \frac{2}{(s+1)^2 + 4}\ }
$$

Poles where $(s+1)^2 = -4$, i.e. $s = -1 \pm 2i$: real part $-1$ is the decay rate, imaginary part $\pm 2$ the oscillation frequency. The pole pair *is* the signal's fingerprint — damped oscillation at rate $1$, frequency $2$ — and it lies in the left half-plane, as every stable mode must.

> **Key Takeaway:** $e^{at}$ in time slides every pole by $a$ in the $s$-plane; complex pole pairs $\sigma \pm i\omega$ encode decay $\sigma$ and frequency $\omega$ at a glance.

### Problem L1.3: Inverse Transform — Distinct and Repeated Real Roots

**Problem Statement:** Invert

$$
F(s) = \frac{s + 4}{s (s + 2)^2}.
$$

Handle the distinct root at $s = 0$ and the repeated root at $s = -2$, showing how the coefficients are computed (cover-up for the easy ones, matching for the rest).

**Intuition:** Each simple pole contributes one exponential; a double pole additionally contributes $t\, e^{pt}$.

**Solution:**

**Step 1 — ansatz.** A repeated factor $(s+2)^2$ requires *both* powers:

$$
\frac{s+4}{s(s+2)^2} = \frac{A}{s} + \frac{B}{s+2} + \frac{C}{(s+2)^2}.
$$

**Step 2 — cover-up for $A$ and $C$.** Multiply by $s$ and set $s = 0$: $A = \frac{0 + 4}{(0+2)^2} = 1$. Multiply by $(s+2)^2$ and set $s = -2$: $C = \frac{-2 + 4}{-2} = -1$.

**Step 3 — match for $B$.** Multiply out: $s + 4 = A(s+2)^2 + B s (s+2) + C s$. Compare $s^2$ coefficients: $0 = A + B$, so $B = -1$. (Check with $s = 1$: right side $= 1 \cdot 9 - 1 \cdot 3 - 1 = 5 = 1 + 4$. Correct.)

**Step 4 — invert term by term.** Using $\mathcal{L}^{-1}\{1/s\} = 1$, $\mathcal{L}^{-1}\{1/(s+2)\} = e^{-2t}$, and $\mathcal{L}^{-1}\{1/(s+2)^2\} = t e^{-2t}$ (from $\mathcal{L}\{t e^{at}\} = 1/(s-a)^2$):

$$
f(t) = 1 - e^{-2t} - t e^{-2t}.
$$

**Step 5 — sanity checks.** Initial value theorem: $sF(s) = \frac{s+4}{(s+2)^2} \to 0$ as $s \to \infty$, and indeed $f(0) = 1 - 1 - 0 = 0$. Final value theorem (poles of $sF$ at $-2$, left half-plane — applicable): $sF(s) \to \frac{4}{4} = 1$ as $s \to 0$, and indeed $f(t) \to 1$.

$$
\boxed{\ f(t) = 1 - e^{-2t} - t\, e^{-2t}\ }
$$

> **Key Takeaway:** A pole of multiplicity $m$ demands all powers $1$ through $m$ in the ansatz and produces modes $t^{k} e^{pt}$ up to $k = m - 1$; the IVT/FVT give free error-detection.

### Problem L1.4: Inverse Transform — Complex Roots

**Problem Statement:** Invert

$$
F(s) = \frac{s + 3}{s^2 + 4s + 13}
$$

by completing the square (do not factor over $\mathbb{C}$), expressing the answer as a damped oscillation.

**Intuition:** An irreducible quadratic is a shifted circle: complete the square and read off $(s-a)^2 + b^2$.

**Solution:**

**Step 1 — complete the square.** $s^2 + 4s + 13 = (s + 2)^2 + 9$, so the poles are $s = -2 \pm 3i$: decay rate $2$, frequency $3$.

**Step 2 — rewrite the numerator in shifted form.** We need $s + 3$ expressed in terms of $(s + 2)$ and the constant $3$ (the $b$ of $\sin$):

$$
\frac{s + 3}{(s+2)^2 + 9} = \frac{(s + 2) + 1}{(s+2)^2 + 9} = \frac{s+2}{(s+2)^2 + 9} + \frac{1}{3} \cdot \frac{3}{(s+2)^2 + 9}.
$$

**Step 3 — invert with the shifted table.** $\mathcal{L}\{e^{-2t} \cos 3t\} = \frac{s+2}{(s+2)^2+9}$ and $\mathcal{L}\{e^{-2t} \sin 3t\} = \frac{3}{(s+2)^2+9}$, so

$$
f(t) = e^{-2t} \cos 3t + \tfrac{1}{3} e^{-2t} \sin 3t.
$$

**Step 4 — amplitude-phase form (optional but illuminating).** $f(t) = R e^{-2t} \cos(3t - \varphi)$ with $R = \sqrt{1 + \tfrac{1}{9}} = \frac{\sqrt{10}}{3}$ and $\tan \varphi = \tfrac{1}{3}$.

$$
\boxed{\ f(t) = e^{-2t} \left( \cos 3t + \tfrac{1}{3} \sin 3t \right) = \tfrac{\sqrt{10}}{3}\, e^{-2t} \cos(3t - \arctan \tfrac{1}{3})\ }
$$

> **Key Takeaway:** Never factor irreducible quadratics — complete the square, split the numerator into "shifted $s$" plus "constant", and read off cosine and sine parts directly.

### Problem L1.5: First-Order IVP

**Problem Statement:** Solve by Laplace transform:

$$
y' + 2y = 4t, \qquad y(0) = 1.
$$

Verify the answer by substitution, and identify which parts of the solution are "forced" and which are "transient".

**Intuition:** One transform, one division, one partial-fraction split — no integrating factor needed.

**Solution:**

**Step 1 — transform.** $\mathcal{L}\{y'\} = sY - y(0) = sY - 1$ and $\mathcal{L}\{4t\} = 4/s^2$:

$$
sY - 1 + 2Y = \frac{4}{s^2} \quad \Longrightarrow \quad Y(s) = \frac{1}{s + 2} + \frac{4}{s^2 (s + 2)}.
$$

**Step 2 — partial fractions.** Set $\frac{4}{s^2(s+2)} = \frac{A}{s} + \frac{B}{s^2} + \frac{C}{s+2}$. Multiply out: $4 = A s(s+2) + B(s+2) + C s^2$. At $s = 0$: $B = 2$. At $s = -2$: $C = 1$. Comparing $s^2$: $0 = A + C$, so $A = -1$. Thus

$$
Y(s) = \frac{1}{s+2} - \frac{1}{s} + \frac{2}{s^2} + \frac{1}{s+2} = -\frac{1}{s} + \frac{2}{s^2} + \frac{2}{s+2}.
$$

**Step 3 — invert.**

$$
y(t) = -1 + 2t + 2 e^{-2t}.
$$

**Step 4 — verify.** $y' = 2 - 4e^{-2t}$, so $y' + 2y = 2 - 4e^{-2t} - 2 + 4t + 4e^{-2t} = 4t$. And $y(0) = -1 + 0 + 2 = 1$. Both check.

**Step 5 — anatomy.** The pole at $s = -2$ (from the characteristic polynomial) gives the transient $2e^{-2t}$, which dies out; the double pole at $s = 0$ (inherited from the forcing $4/s^2$) gives the persistent forced response $2t - 1$, the straight line the solution rides along for large $t$.

$$
\boxed{\ y(t) = 2t - 1 + 2 e^{-2t}\ }
$$

> **Key Takeaway:** In $Y(s)$, poles from the characteristic polynomial produce transients and poles from the forcing produce the steady response — the anatomy of the solution is visible before inverting.

### Problem L1.6: Second-Order IVP with Exponential Forcing

**Problem Statement:** Solve

$$
y'' + 5y' + 6y = e^{-t}, \qquad y(0) = 0, \quad y'(0) = 1,
$$

by Laplace transform, and confirm the initial conditions from the final formula.

**Intuition:** Three simple poles ($-1$ from the forcing, $-2$ and $-3$ from the system) mean three decaying exponentials.

**Solution:**

**Step 1 — transform.** With $y(0) = 0$, $y'(0) = 1$: $\mathcal{L}\{y''\} = s^2 Y - 1$, $\mathcal{L}\{y'\} = sY$, and $\mathcal{L}\{e^{-t}\} = \frac{1}{s+1}$:

$$
(s^2 + 5s + 6) Y - 1 = \frac{1}{s+1} \quad \Longrightarrow \quad Y(s) = \frac{1}{(s+2)(s+3)} + \frac{1}{(s+1)(s+2)(s+3)},
$$

using $s^2 + 5s + 6 = (s+2)(s+3)$.

**Step 2 — combine and decompose.** $Y(s) = \frac{(s+1) + 1}{(s+1)(s+2)(s+3)} = \frac{s + 2}{(s+1)(s+2)(s+3)} = \frac{1}{(s+1)(s+3)}$ — the factor $s + 2$ cancels. Then by cover-up:

$$
\frac{1}{(s+1)(s+3)} = \frac{1/2}{s+1} - \frac{1/2}{s+3}.
$$

**Step 3 — invert.**

$$
y(t) = \tfrac{1}{2} e^{-t} - \tfrac{1}{2} e^{-3t}.
$$

**Step 4 — check.** $y(0) = \tfrac12 - \tfrac12 = 0$. $y'(t) = -\tfrac12 e^{-t} + \tfrac32 e^{-3t}$, so $y'(0) = -\tfrac12 + \tfrac32 = 1$. Substituting back: $y'' = \tfrac12 e^{-t} - \tfrac92 e^{-3t}$, and $y'' + 5y' + 6y = e^{-t} (\tfrac12 - \tfrac52 + 3) + e^{-3t}(-\tfrac92 + \tfrac{15}{2} - 3) = e^{-t}$. All conditions verified.

$$
\boxed{\ y(t) = \tfrac{1}{2} \left( e^{-t} - e^{-3t} \right)\ }
$$

The mode $e^{-2t}$ is absent: the particular initial data $y'(0) = 1$ happen to put zero weight on the $s = -2$ eigen-direction — a cancellation the transform displays as an algebraic factor cancelling, before any inversion.

> **Key Takeaway:** Pole cancellations in $Y(s)$ are mode suppressions in $y(t)$; the algebra shows which eigendirections the initial data actually excite.

### Problem L2.1: RLC Circuit with a Switch at t = 2

**Problem Statement:** A series RLC circuit with $L = 1\,\mathrm{H}$, $R = 2\,\Omega$, $C = 0.2\,\mathrm{F}$ has zero initial charge and current. At $t = 2$ seconds a $10\,\mathrm{V}$ supply is switched on: the charge $q(t)$ obeys

$$
q'' + 2 q' + 5 q = 10\, u_2(t), \qquad q(0) = 0, \quad q'(0) = 0.
$$

Solve for $q(t)$ fully, and describe the behavior before and after the switch.

**Intuition:** The circuit is an underdamped oscillator (poles $-1 \pm 2i$); after the switch it rings toward the new equilibrium charge $q = 2$.

**Solution:**

**Step 1 — transform.** Zero initial conditions kill the boundary terms; $\mathcal{L}\{u_2\} = e^{-2s}/s$:

$$
(s^2 + 2s + 5)\, Q(s) = \frac{10 e^{-2s}}{s} \quad \Longrightarrow \quad Q(s) = e^{-2s} \cdot \frac{10}{s (s^2 + 2s + 5)}.
$$

**Step 2 — partial fractions on the undelayed part.** Set $\frac{10}{s(s^2+2s+5)} = \frac{A}{s} + \frac{Bs + C}{s^2 + 2s + 5}$. Multiply out: $10 = A(s^2 + 2s + 5) + (Bs + C)s$. At $s = 0$: $A = 2$. Comparing $s^2$: $0 = A + B$, so $B = -2$; comparing $s$: $0 = 2A + C$, so $C = -4$. Completing the square, $s^2 + 2s + 5 = (s+1)^2 + 4$:

$$
\frac{10}{s(s^2+2s+5)} = \frac{2}{s} - \frac{2s + 4}{(s+1)^2 + 4} = \frac{2}{s} - \frac{2(s+1)}{(s+1)^2 + 4} - \frac{2}{(s+1)^2 + 4}.
$$

**Step 3 — invert the undelayed part.** With the shifted table (decay $1$, frequency $2$):

$$
\varphi(t) = 2 - 2 e^{-t} \cos 2t - e^{-t} \sin 2t.
$$

(Check: $\varphi(0) = 2 - 2 = 0$; $\varphi'(0) = 2 - 2 = 0$ from $\varphi' = 2e^{-t}\cos 2t + 4e^{-t}\sin 2t - \cdots$ — consistent with a step response from rest. Also $\varphi(\infty) = 2 = 10/5$, the FVT value of $s \Phi(s)$ at $s = 0$.)

**Step 4 — apply t-shifting.** $Q(s) = e^{-2s} \Phi(s)$ gives $q(t) = u_2(t)\, \varphi(t - 2)$:

$$
\boxed{\ q(t) = u_2(t) \left[ 2 - 2 e^{-(t-2)} \cos 2(t-2) - e^{-(t-2)} \sin 2(t-2) \right]\ }
$$

**Step 5 — physical reading.** For $t \lt 2$: $q \equiv 0$ — the transform enforces causality automatically. For $t \gt 2$: the charge rings at damped frequency $2\,\mathrm{rad/s}$ with envelope $e^{-(t-2)}$, converging to the equilibrium $q_\infty = CV = 0.2 \times 10 = 2$ coulombs, in agreement with the constant term. The poles $-1 \pm 2i$ of the circuit sit in the left half-plane: the ringing is transient, the step is absorbed.

> **Key Takeaway:** A switch at $t = c$ costs exactly one factor $e^{-cs}$; the underdamped pole pair dictates ring frequency and settling envelope, and the FVT reads off the final charge without inverting.

### Problem L2.2: Hammer Blow and Resonance

**Problem Statement:** An undamped mass-spring system $x'' + \omega_0^2 x = \frac{1}{m} f(t)$ (mass $m = 1\,\mathrm{kg}$, $\omega_0 = 3\,\mathrm{rad/s}$), initially at rest, is struck at $t = 1$ by a hammer delivering impulse $I_0 = 6\,\mathrm{N\,s}$: $f(t) = 6\, \delta(t - 1)$. (a) Solve for $x(t)$. (b) Show that the response to $\delta$ (the impulse response) determines the response to *any* forcing via convolution. (c) If blows of impulse $6$ are repeated every natural period $T = 2\pi/3$, show the amplitude grows without bound and connect this to the notion of resonance.

**Intuition:** Each blow adds velocity $I_0/m$; struck in phase, the additions stack coherently — the pushed swing.

**Solution:**

**(a) Single blow.** Transforming $x'' + 9x = 6\,\delta(t-1)$ with rest initial conditions, using $\mathcal{L}\{\delta(t-1)\} = e^{-s}$:

$$
(s^2 + 9) X(s) = 6 e^{-s} \quad \Longrightarrow \quad X(s) = e^{-s}\, \frac{6}{s^2 + 9} = e^{-s} \cdot 2 \cdot \frac{3}{s^2 + 9}.
$$

By the table and t-shifting:

$$
\boxed{\ x(t) = 2\, u_1(t)\, \sin 3(t - 1)\ }
$$

Before the blow, nothing; after it, free oscillation of amplitude $2 = \frac{I_0}{m \omega_0} = \frac{6}{1 \cdot 3}$. The blow creates velocity $x'(1^+) = 6 = I_0/m$ from rest, exactly the momentum-transfer interpretation of $\delta$.

**(b) Impulse response as universal kernel.** The response to a *unit* delta at $t = 0$ is $h(t) = \frac{1}{3} \sin 3t = \mathcal{L}^{-1}\{ \frac{1}{s^2 + 9} \}$. For general forcing, $X(s) = H(s) F(s)$ with $H(s) = \frac{1}{s^2+9}$, and the convolution theorem gives

$$
x(t) = (h * f)(t) = \frac{1}{3} \int_0^t \sin 3(t - \tau)\, f(\tau)\, d\tau
$$

— Duhamel's principle: knowing the answer to one idealized kick determines the answer to everything.

**(c) Periodic blows at the natural period.** Forcing $f = 6 \sum_{n=0}^{\infty} \delta(t - nT)$ with $T = 2\pi/3$ gives, by (a) and superposition, for $NT \lt t \lt (N+1)T$:

$$
x(t) = 2 \sum_{n=0}^{N} \sin 3(t - nT) = 2 \sum_{n=0}^{N} \sin(3t - 2\pi n) = 2 (N + 1) \sin 3t,
$$

since shifting the phase by $2\pi n$ changes nothing. The amplitude after $N + 1$ blows is $2(N+1) \to \infty$: every kick arrives exactly in phase with the velocity, pumping energy in each time. In the $s$-plane this is the collision of forcing frequency with the poles $\pm 3i$ *on* the imaginary axis — an undamped system driven at its natural frequency has a repeated marginal pole, hence secular growth. This is resonance: catastrophic for bridges, essential for swings, clocks, and lasers.

> **Key Takeaway:** The impulse response is the system's DNA — every response is its convolution with the input — and in-phase impulsive driving reveals resonance as coherent stacking of impulse responses.

### Problem L2.3: PID Pole Placement for a First-Order Plant

**Problem Statement:** A DC motor's speed obeys the first-order plant $P(s) = \dfrac{1}{s + 1}$ (time constant $1$ s). It is controlled with a PI controller $C(s) = K_p + \dfrac{K_i}{s}$ in unity negative feedback. (a) Derive the closed-loop transfer function $T(s)$ from reference to output. (b) Choose $K_p, K_i$ to place both closed-loop poles at $s = -3$ (critically damped, three times faster than the plant). (c) Using the final value theorem, prove the design tracks a unit step reference with zero steady-state error, and explain which term of the controller is responsible.

**Intuition:** Two knobs ($K_p$, $K_i$), two coefficients of the characteristic polynomial — full pole placement.

**Solution:**

**(a) Closed loop.** With loop transfer $C(s)P(s) = \frac{K_p s + K_i}{s(s+1)}$:

$$
T(s) = \frac{C P}{1 + C P} = \frac{K_p s + K_i}{s(s+1) + K_p s + K_i} = \frac{K_p s + K_i}{s^2 + (1 + K_p) s + K_i}.
$$

**(b) Pole placement.** Both poles at $-3$ means characteristic polynomial $(s + 3)^2 = s^2 + 6s + 9$. Matching coefficients:

$$
1 + K_p = 6 \implies K_p = 5, \qquad K_i = 9.
$$

$$
\boxed{\ K_p = 5, \quad K_i = 9 \quad \Longrightarrow \quad T(s) = \frac{5s + 9}{(s + 3)^2}\ }
$$

The closed-loop system responds with modes $e^{-3t}, t e^{-3t}$: no overshoot from the pole pattern (repeated real pole), settling roughly three times faster than the open-loop motor.

**(c) Zero steady-state error via FVT.** The error transfer function is $E(s) = \frac{1}{1 + CP} R(s)$. For a unit step $R = 1/s$:

$$
e_\infty = \lim_{s \to 0} s\, E(s) = \lim_{s \to 0} \frac{s(s+1)}{s^2 + 6s + 9} \cdot \frac{s}{s} \cdot \frac{1}{1} = \lim_{s \to 0} \frac{s (s + 1)}{s^2 + 6 s + 9} = \frac{0}{9} = 0.
$$

The FVT applies because the closed-loop poles are at $-3 \lt 0$. The factor of $s$ in the numerator that forces the limit to zero comes from the $K_i/s$ term: the integrator contributes an *infinite DC gain* ($C(s) \to \infty$ as $s \to 0$), so any persistent error would be integrated up until it is cancelled. A pure P controller ($K_i = 0$) leaves $e_\infty = \frac{1}{1 + K_p} \neq 0$: only the integral action achieves exact tracking.

> **Key Takeaway:** Feedback design is pole algebra: the controller's gains are chosen to give the closed-loop characteristic polynomial desired roots, and the integrator's pole at $s = 0$ buys zero steady-state error via the final value theorem.

### Problem L2.4: The S4 Kernel — a State-Space Model as a Convolution

**Problem Statement:** A structured state-space sequence layer (as in S4) uses the continuous-time system

$$
x'(t) = A x(t) + B u(t), \qquad y(t) = C x(t), \qquad A = \begin{pmatrix} 0 & 1 \\ -2 & -3 \end{pmatrix}, \quad B = \begin{pmatrix} 0 \\ 1 \end{pmatrix}, \quad C = \begin{pmatrix} 1 & 0 \end{pmatrix}.
$$

(a) Compute the transfer function $H(s) = C (sI - A)^{-1} B$. (b) Invert it to find the convolution kernel $K(t) = C e^{At} B$. (c) Explain, via the convolution theorem, why this one system can be *trained* as a convolution over a whole input sequence but *deployed* as a stepwise recurrence, and what the eigenvalues of $A$ say about the layer's memory.

**Intuition:** One rational function of $s$ is simultaneously a parallel convolution filter and a sequential RNN.

**Solution:**

**(a) Transfer function.** $sI - A = \begin{pmatrix} s & -1 \\ 2 & s + 3 \end{pmatrix}$, $\det(sI - A) = s(s+3) + 2 = s^2 + 3s + 2 = (s+1)(s+2)$. Then

$$
(sI - A)^{-1} = \frac{1}{(s+1)(s+2)} \begin{pmatrix} s + 3 & 1 \\ -2 & s \end{pmatrix}, \qquad (sI - A)^{-1} B = \frac{1}{(s+1)(s+2)} \begin{pmatrix} 1 \\ s \end{pmatrix},
$$

and applying $C$ picks the first entry:

$$
H(s) = \frac{1}{(s+1)(s+2)}.
$$

**(b) Kernel.** Partial fractions: $\frac{1}{(s+1)(s+2)} = \frac{1}{s+1} - \frac{1}{s+2}$, so

$$
\boxed{\ K(t) = C e^{At} B = e^{-t} - e^{-2t}, \qquad t \ge 0\ }
$$

(One can verify directly: the eigenvalues of $A$ are $-1, -2$; diagonalizing $A$ and computing $C e^{At} B$ reproduces exactly these two weights.) The layer's output is the causal convolution $y(t) = \int_0^t K(t - \tau)\, u(\tau)\, d\tau$.

**(c) Two computational faces of one object.** By the convolution theorem, $Y(s) = H(s) U(s)$ is equivalent to $y = K * u$. After discretization (step $\Delta$), this becomes a discrete convolution $y_k = \sum_j \bar{K}_j u_{k-j}$ with $\bar{K}_j \approx K(j\Delta)$:

- **Training**: the whole kernel $\bar{K}$ is materialized once and applied to the entire input sequence with FFT-based convolution — $O(L \log L)$, fully parallel across the sequence, ideal for GPUs. This is multiplication $H \cdot U$ in the frequency domain.
- **Inference**: the same system is run as the recurrence $x_{k+1} = \bar{A} x_k + \bar{B} u_k$, $y_k = C x_k$ — $O(1)$ time and memory per new token, ideal for streaming generation.

The convolution theorem is the *proof* that these two procedures compute the same function; S4's contribution was making both faces numerically efficient for large state dimension. The eigenvalues $-1, -2$ of $A$ set the memory horizon: the kernel $e^{-t} - e^{-2t}$ decays on timescales $1$ and $1/2$, so this layer forgets inputs older than a few time units. Stable long-range memory needs eigenvalues very close to (but left of) the imaginary axis — precisely what S4's HiPPO-based initialization of $A$ arranges.

> **Key Takeaway:** An SSM layer is the triple (recurrence, kernel, transfer function) — one object in three notations — and the convolution theorem is the bridge that lets deep learning train it in parallel and run it sequentially.

### Problem L2.5: Discounted Return as a Laplace Transform

**Problem Statement:** In continuous-time reinforcement learning, the value of a deterministic reward stream $r(t)$, $t \ge 0$, under discount rate $\rho \gt 0$ is $V = \int_0^\infty e^{-\rho t} r(t)\, dt$. (a) Identify $V$ as a Laplace transform evaluation. (b) Compute $V$ for the "learning curve" reward $r(t) = 1 - e^{-t}$ (performance approaching 1) with $\rho = \tfrac{1}{2}$. (c) Use the t-shifting theorem to show how delaying all rewards by $c$ changes value, and compute the value of the delayed stream $r(t) = u_c(t)\,(1 - e^{-(t - c)})$ for $c = 2$. (d) Explain what the existence condition for the transform means in RL terms.

**Intuition:** Exponential discounting *is* the Laplace kernel; the discount rate is a point on the $s$-axis.

**Solution:**

**(a)** By definition, $V = \mathcal{L}\{r\}(\rho)$: the value function of exponential discounting is the reward stream's Laplace transform evaluated at $s = \rho$. Every transform theorem becomes a statement about values.

**(b)** By linearity and the table, $\mathcal{L}\{1 - e^{-t}\}(s) = \frac{1}{s} - \frac{1}{s+1} = \frac{1}{s(s+1)}$. At $s = \rho = \tfrac12$:

$$
V = \frac{1}{\tfrac12 \cdot \tfrac32} = \frac{4}{3} \approx 1.333.
$$

For comparison, the ideal stream $r \equiv 1$ has value $1/\rho = 2$: the slow ramp-up costs $2 - \tfrac43 = \tfrac23$ units of value — the *price of learning slowly*, computed in closed form.

**(c) Delay discounts multiplicatively.** The t-shifting theorem gives $\mathcal{L}\{u_c(t) r(t-c)\}(\rho) = e^{-\rho c}\, \mathcal{L}\{r\}(\rho)$: postponing the entire stream by $c$ multiplies its value by $e^{-\rho c}$, independently of the stream's shape. For $c = 2$, $\rho = \tfrac12$:

$$
V_{\text{delayed}} = e^{-1} \cdot \frac{4}{3} \approx 0.4905.
$$

Waiting two time units costs the agent the factor $e^{-\rho c} = e^{-1} \approx 0.368$ — a 63 percent value loss, again with no integral computed by hand.

**(d) Existence = well-posed objective.** The transform (hence $V$) exists when $r$ is of exponential order $\alpha \lt \rho$: rewards may even grow exponentially, provided slower than the discounting. If reward growth matches or beats $\rho$ (e.g. $r(t) = e^{\rho t}$), the integral diverges and the discounted objective is meaningless — the RL translation of the region of convergence $s \gt \alpha$. This is why discounted control of unstable systems requires $\rho$ larger than the instability's growth rate.

$$
\boxed{\ V = \mathcal{L}\{r\}(\rho); \quad \text{here } V = \tfrac{4}{3}, \quad V_{\text{delayed by } 2} = \tfrac{4}{3} e^{-1}\ }
$$

> **Key Takeaway:** Discounting is Laplace evaluation at $s = \rho$: delays multiply value by $e^{-\rho c}$, and the ROC condition is precisely the requirement that the discounted objective converge.

### Problem L2.6: Drug Dosing with Periodic Boluses

**Problem Statement:** A drug is eliminated with first-order rate $k = \ln 2$ per hour (half-life 1 h) and administered as an intravenous bolus of $D = 100$ mg every $T = 1$ hour: the plasma amount obeys

$$
c'(t) + k\, c(t) = D \sum_{n=0}^{\infty} \delta(t - nT), \qquad c(0^-) = 0.
$$

(a) Solve for $c(t)$ by transforming and summing the resulting geometric series in $e^{-sT}$. (b) Find the steady-state peak and trough levels. (c) Show how the periodic-function formula reproduces the same transform.

**Intuition:** Each bolus adds $D$ instantly; between boluses the level decays by the factor $e^{-kT} = \tfrac12$. Peaks ratchet up toward a finite plateau.

**Solution:**

**(a) Transform and invert.** Transforming with $\mathcal{L}\{\delta(t - nT)\} = e^{-nTs}$:

$$
(s + k)\, C(s) = D \sum_{n=0}^{\infty} e^{-nTs} \quad \Longrightarrow \quad C(s) = D \sum_{n=0}^{\infty} \frac{e^{-nTs}}{s + k}.
$$

Each term inverts by t-shifting ($\mathcal{L}^{-1}\{1/(s+k)\} = e^{-kt}$):

$$
c(t) = D \sum_{n=0}^{\infty} u_{nT}(t)\, e^{-k(t - nT)} = D e^{-kt} \sum_{n=0}^{N(t)} e^{k n T}, \qquad N(t) = \lfloor t/T \rfloor.
$$

Summing the finite geometric series with ratio $e^{kT} = 2$: for $NT \le t \lt (N+1)T$,

$$
c(t) = D e^{-k(t - NT)}\, \frac{1 - e^{-k(N+1)T}}{1 - e^{-kT}}.
$$

**(b) Steady state.** As $N \to \infty$ the bracket tends to $1$, so within a late dosing interval, with $\tau = t - NT \in [0, T)$:

$$
c_{ss}(\tau) = \frac{D\, e^{-k\tau}}{1 - e^{-kT}}.
$$

With $e^{-kT} = e^{-\ln 2} = \tfrac12$: peak at $\tau = 0^+$ and trough at $\tau \to T^-$:

$$
\boxed{\ c_{\max} = \frac{D}{1 - e^{-kT}} = \frac{100}{1/2} = 200 \text{ mg}, \qquad c_{\min} = c_{\max} e^{-kT} = 100 \text{ mg}.\ }
$$

Sanity check: trough plus fresh bolus equals peak, $100 + 100 = 200$ — the plateau is self-consistent. Accumulation ratio $\frac{1}{1 - e^{-kT}} = 2$: dosing every half-life doubles the single-dose peak.

**(c) Periodic-function formula.** At steady state the input $f(t) = D\sum_n \delta(t - nT)$ is $T$-periodic, and the formula $\mathcal{L}\{f\} = \frac{1}{1 - e^{-sT}} \int_0^T e^{-st} f(t)\, dt$ with $\int_0^{T^-} e^{-st} D\,\delta(t)\, dt = D$ gives

$$
\mathcal{L}\{f\}(s) = \frac{D}{1 - e^{-sT}} = D \sum_{n=0}^{\infty} e^{-nTs},
$$

exactly the geometric series of part (a): the formula *is* the summed series, and dividing by $s + k$ returns $C(s)$. The steady-state amplitude factor $\frac{1}{1 - e^{-kT}}$ in (b) is this same geometric amplification evaluated at the elimination rate.

> **Key Takeaway:** Periodic dosing is a geometric series in $e^{-sT}$; the periodic-function formula sums it once and for all, and the accumulation factor $1/(1 - e^{-kT})$ falls out of the pole structure.

### Problem L3.1: Half-Integer Powers and the Gamma Function

**Problem Statement:** (a) Show that for $p \gt -1$, $\mathcal{L}\{t^p\}(s) = \dfrac{\Gamma(p+1)}{s^{p+1}}$ for $s \gt 0$, where $\Gamma(z) = \int_0^\infty x^{z-1} e^{-x} dx$. (b) Prove $\Gamma(\tfrac12) = \sqrt{\pi}$ using the Gaussian integral, deduce $\Gamma(\tfrac32)$, and hence compute $\mathcal{L}\{\sqrt{t}\}$ and $\mathcal{L}\{1/\sqrt{t}\}$. (c) Comment on the fact that $1/\sqrt{t}$ is unbounded at $0$ yet has a transform, while it is *not* of exponential order zero difficulty — which hypothesis of the existence theorem does it violate and why does convergence survive?

**Intuition:** The Gamma function is what the factorial becomes when integration by parts can no longer count down to zero.

**Solution:**

**(a) Substitution.** In $\int_0^\infty e^{-st} t^p\, dt$ put $x = st$ (valid for $s \gt 0$), $t = x/s$, $dt = dx/s$:

$$
\mathcal{L}\{t^p\}(s) = \int_0^\infty e^{-x} \frac{x^p}{s^p} \cdot \frac{dx}{s} = \frac{1}{s^{p+1}} \int_0^\infty x^{p} e^{-x}\, dx = \frac{\Gamma(p + 1)}{s^{p+1}},
$$

the integral converging at $0$ because $p \gt -1$ and at $\infty$ by exponential decay. For integer $p = n$, $\Gamma(n+1) = n!$ recovers L1.1.

**(b) The Gaussian evaluation.** In $\Gamma(\tfrac12) = \int_0^\infty x^{-1/2} e^{-x} dx$ substitute $x = w^2$, $dx = 2w\, dw$:

$$
\Gamma(\tfrac12) = \int_0^\infty \frac{e^{-w^2}}{w}\, 2w\, dw = 2 \int_0^\infty e^{-w^2} dw = \sqrt{\pi},
$$

by the classical polar-coordinates computation $\left( \int_{-\infty}^\infty e^{-w^2} dw \right)^2 = \int_0^{2\pi}\!\!\int_0^\infty e^{-r^2} r\, dr\, d\theta = \pi$. The functional equation $\Gamma(z + 1) = z \Gamma(z)$ (one integration by parts) gives $\Gamma(\tfrac32) = \tfrac12 \Gamma(\tfrac12) = \tfrac{\sqrt{\pi}}{2}$. Hence

$$
\boxed{\ \mathcal{L}\{\sqrt{t}\} = \frac{\sqrt{\pi}}{2\, s^{3/2}}, \qquad \mathcal{L}\{1/\sqrt{t}\} = \frac{\Gamma(\tfrac12)}{s^{1/2}} = \sqrt{\frac{\pi}{s}}, \qquad s \gt 0.\ }
$$

**(c) Beyond piecewise continuity.** $1/\sqrt{t}$ blows up at $t = 0$, so it is *not piecewise continuous on $[0, T]$* — one hypothesis of the standard existence theorem fails (growth at infinity is fine: it is of exponential order $0$). Convergence survives because the failure is an *integrable* singularity: near zero, $\int_0^1 t^{-1/2} dt = 2 \lt \infty$, and the existence proof only used piecewise continuity to guarantee integrability on finite intervals. The theorem's hypotheses are sufficient, not necessary; the honest general condition is local integrability plus exponential order. This matters in practice: $1/\sqrt{t}$ kernels are exactly what appear in heat conduction on a half-line (Topic on diffusion) and in fractional calculus, where $\mathcal{L}\{t^{-1/2}\} = \sqrt{\pi/s}$ underlies the half-derivative.

> **Key Takeaway:** The Gamma function extends the factorial in the transform table to arbitrary powers, and integrable singularities at $t = 0$ are welcome even though the textbook existence theorem does not cover them.

### Problem L3.2: Division by t and the Dirichlet Integral

**Problem Statement:** (a) Prove that if $F = \mathcal{L}\{f\}$ exists for $s \gt \alpha$ and $\lim_{t \to 0^+} \frac{f(t)}{t}$ exists, then

$$
\mathcal{L}\left\{ \frac{f(t)}{t} \right\}(s) = \int_s^\infty F(\sigma)\, d\sigma.
$$

(b) Use it to evaluate the Dirichlet integral $\displaystyle \int_0^\infty \frac{\sin t}{t}\, dt = \frac{\pi}{2}$.

**Intuition:** Multiplying by $t$ differentiates $F$; dividing by $t$ must therefore integrate $F$ — from $s$ out to infinity so that the constant of integration dies.

**Solution:**

**(a) Step 1 — define and differentiate.** Let $g(t) = f(t)/t$ (with the removable singularity at $0$ filled by the assumed limit, so $g$ satisfies the existence hypotheses) and $G = \mathcal{L}\{g\}$. Since $f(t) = t\, g(t)$, the multiplication-by-$t$ theorem gives

$$
F(s) = \mathcal{L}\{t\, g(t)\}(s) = -G'(s).
$$

**Step 2 — integrate from $s$ to $R$.** $G(s) - G(R) = \int_s^R F(\sigma)\, d\sigma$ for any $R \gt s$.

**Step 3 — dispose of $G(R)$.** By the existence theorem's decay estimate, $G(R) \to 0$ as $R \to \infty$ (every Laplace transform of an admissible function vanishes at $s$-infinity). Letting $R \to \infty$:

$$
G(s) = \int_s^\infty F(\sigma)\, d\sigma. \qquad \blacksquare
$$

**(b) Application with $f = \sin t$.** Here $\lim_{t \to 0^+} \frac{\sin t}{t} = 1$ exists, and $F(\sigma) = \frac{1}{\sigma^2 + 1}$, so for $s \gt 0$:

$$
\mathcal{L}\left\{ \frac{\sin t}{t} \right\}(s) = \int_s^\infty \frac{d\sigma}{\sigma^2 + 1} = \frac{\pi}{2} - \arctan s.
$$

Now let $s \to 0^+$. The left side tends to $\int_0^\infty \frac{\sin t}{t}\, dt$ — this step needs justification since the ROC boundary is being touched: the integral $\int_0^\infty e^{-st}\frac{\sin t}{t} dt$ converges *uniformly for $s \ge 0$* by the alternating/Dirichlet test (the partial integrals of $\sin t / t$ over successive half-periods alternate and shrink, and $e^{-st}$ is monotone in $t$), so the limit may be taken inside (Abel's theorem for Laplace integrals). The right side tends to $\frac{\pi}{2} - 0$. Therefore

$$
\boxed{\ \int_0^\infty \frac{\sin t}{t}\, dt = \frac{\pi}{2}\ }
$$

— a famous integral with *no* elementary antiderivative, evaluated by pure transform algebra plus one careful limit.

> **Key Takeaway:** Division by $t$ integrates the transform from $s$ to infinity; evaluating such identities at the ROC boundary (Abelian limits) turns transform tables into a factory for hard definite integrals.

### Problem L3.3: A Volterra Integral Equation of Convolution Type

**Problem Statement:** Solve for the continuous function $y(t)$:

$$
y(t) = t + \int_0^t \sin(t - \tau)\, y(\tau)\, d\tau.
$$

Then verify the solution by direct substitution, and explain why the Laplace transform is the natural tool for convolution-type Volterra equations while it fails for general kernels $k(t, \tau)$.

**Intuition:** The unknown appears inside a convolution — exactly the structure the convolution theorem converts to multiplication, making the equation *linear-algebraic* in $Y$.

**Solution:**

**Step 1 — recognize the convolution.** The equation is $y = t + (\sin) * y$. Transforming with the convolution theorem, with $Y = \mathcal{L}\{y\}$:

$$
Y(s) = \frac{1}{s^2} + \frac{1}{s^2 + 1}\, Y(s).
$$

**Step 2 — solve algebraically.**

$$
Y(s) \left( 1 - \frac{1}{s^2+1} \right) = \frac{1}{s^2} \quad \Longrightarrow \quad Y(s) \cdot \frac{s^2}{s^2 + 1} = \frac{1}{s^2} \quad \Longrightarrow \quad Y(s) = \frac{s^2 + 1}{s^4} = \frac{1}{s^2} + \frac{1}{s^4}.
$$

**Step 3 — invert.** From $\mathcal{L}\{t\} = 1/s^2$ and $\mathcal{L}\{t^3\} = 6/s^4$:

$$
y(t) = t + \frac{t^3}{6}.
$$

**Step 4 — verify by substitution.** Compute $\int_0^t \sin(t-\tau) (\tau + \tfrac{\tau^3}{6}) d\tau$. Using repeated integration by parts (or the known antiderivatives): $\int_0^t \sin(t - \tau)\, \tau\, d\tau = t - \sin t$ and $\int_0^t \sin(t-\tau) \frac{\tau^3}{6} d\tau = \frac{t^3}{6} - t + \sin t$. Summing: the integral equals $\frac{t^3}{6}$, and

$$
t + \frac{t^3}{6} = t + \frac{t^3}{6}. \checkmark
$$

$$
\boxed{\ y(t) = t + \frac{t^3}{6}\ }
$$

**Step 5 — scope of the method.** The transform succeeded because the kernel depends only on the *difference* $t - \tau$: the integral operator is convolution, hence multiplication after transforming, and the "inversion of an operator" became division by $1 - K(s)$ — here even converting the integral equation into effectively $y'' = t$-type behavior (note $Y = (s^2+1)/s^4$ has only poles at $0$: the sine kernel exactly cancels oscillation). For a general kernel $k(t, \tau)$ the operator is not translation-invariant, no transform diagonalizes it, and one must fall back on Picard iteration (Topic 02) or numerical quadrature. Convolution structure = the symmetry that Laplace exploits.

> **Key Takeaway:** Laplace transforms solve convolution-type Volterra equations by turning operator inversion into scalar division — the same diagonalization-by-symmetry idea as Fourier methods for translation-invariant PDEs.

### Problem L3.4: Rigorous Inversion Without Contours, and an IVT/FVT Edge Case

**Problem Statement:** (a) Let $F(s) = \dfrac{p(s)}{q(s)}$ be a strictly proper rational function ($\deg p \lt \deg q$). Prove — without any Bromwich contour integral — that $F$ is the Laplace transform of exactly one continuous function on $[0, \infty)$, namely the finite sum of terms $\frac{t^{k} e^{p_j t}}{k!}$ produced by partial fractions, and justify why "table lookup" is therefore a complete rigorous inversion algorithm for rational transforms. (b) Exhibit the failure of the final value theorem for $F(s) = \dfrac{s}{s^2 + 1}$ and the *success* of the initial value theorem for the same $F$, explaining the asymmetry. (c) What goes wrong with the whole framework if $F$ is proper but not *strictly* proper (e.g. $F(s) = \frac{s}{s+1}$)?

**Intuition:** Existence of an inverse is exhibited by construction; uniqueness is Lerch's theorem; no complex analysis required.

**Solution:**

**(a) Construction plus uniqueness.**

*Step 1 — construction.* Over $\mathbb{C}$, $q(s) = c \prod_j (s - p_j)^{m_j}$, and partial fractions give the exact algebraic identity

$$
F(s) = \sum_j \sum_{k=1}^{m_j} \frac{a_{jk}}{(s - p_j)^k},
$$

with no polynomial part because $\deg p \lt \deg q$. Define

$$
f(t) = \sum_j \sum_{k=1}^{m_j} a_{jk}\, \frac{t^{k-1} e^{p_j t}}{(k-1)!}.
$$

Each summand is continuous on $[0, \infty)$ and of exponential order $\max_j \operatorname{Re}(p_j) + \varepsilon$; by the table entry $\mathcal{L}\{t^{k-1} e^{p_j t}/(k-1)!\} = (s - p_j)^{-k}$ (an integer-power case of L1.1 plus s-shifting, valid verbatim for complex $p_j$ since the defining integral converges absolutely for $s \gt \operatorname{Re}(p_j)$) and linearity, $\mathcal{L}\{f\} = F$ on a common half-plane. Complex poles occur in conjugate pairs with conjugate coefficients when $p, q$ are real, so $f$ is real-valued.

*Step 2 — uniqueness.* Suppose $g$ is continuous with $\mathcal{L}\{g\} = F$ on some half-plane. Then $\mathcal{L}\{f - g\} = 0$ for all large $s$, and Lerch's theorem forces $f - g = 0$ at all points of continuity — everywhere, since both are continuous. Hence $f$ is *the* inverse.

*Step 3 — conclusion.* The algorithm "partial fractions, then table" thus *constructs* an inverse and Lerch *certifies* it is the only continuous one:

$$
\boxed{\ \text{For strictly proper rational } F, \ \mathcal{L}^{-1}\{F\} \text{ exists, is unique among continuous functions, and is computable by pure algebra.}\ }
$$

No contour integration was used anywhere; the Bromwich integral is needed only for non-rational transforms.

**(b) The asymmetry of IVT and FVT.** Here $f(t) = \cos t$. IVT: $sF(s) = \frac{s^2}{s^2+1} \to 1 = f(0^+)$ as $s \to \infty$ — correct, and necessarily so: $s \to +\infty$ eventually lies in *every* ROC, so the IVT limit is always an honest statement about the transform. FVT: $sF(s) \to 0$ as $s \to 0^+$, but $\cos t$ has no limit at infinity — the theorem's hypothesis fails because the poles $\pm i$ of $sF(s)$ are on the imaginary axis, not in the open left half-plane; $s = 0$ sits on the ROC boundary and the computed $0$ is merely the *Abel average* of the oscillation, not a limit. Moral: IVT is unconditional (given $f(0^+)$ exists); FVT needs the pole check every single time.

**(c) Failure of strict properness.** $F(s) = \frac{s}{s+1} = 1 - \frac{1}{s+1}$ contains the constant $1 = \mathcal{L}\{\delta\}$: no *function* has a transform that fails to vanish as $s \to \infty$ (existence theorem, decay estimate), so the inverse must contain an impulse, $f = \delta - e^{-t}$, and lives in the space of distributions. Non-proper $F$ (degree of $p$ exceeding $q$) would even require derivatives of $\delta$. Strict properness is exactly the condition separating classical function-valued inverses from distributional ones — physically, the condition that a transfer function have more poles than zeros so its impulse response is a genuine signal.

> **Key Takeaway:** For rational transforms, algebra plus Lerch's uniqueness is a complete rigorous inversion theory; the IVT is structurally safe while the FVT stands or falls with pole locations; and loss of strict properness pushes inverses out of functions into distributions.
