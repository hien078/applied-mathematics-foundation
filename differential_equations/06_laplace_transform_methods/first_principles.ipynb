# Topic 06: Laplace Transform Methods

## 1. First-Principles Intuition & Motivation

**Phenomenon.** Solving a linear initial value problem in the time domain requires calculus at every step: differentiating trial solutions, integrating factors, matching pieces across discontinuities. Yet the *structure* of the problem — a linear combination of $y, y', y''$ equated to a forcing term — looks suspiciously like a polynomial identity. The Laplace transform makes that suspicion precise: it is a linear map under which differentiation becomes multiplication by a variable $s$, so an entire ODE initial value problem becomes one algebraic equation.

**A little history.** In the 1890s Oliver Heaviside, analyzing telegraph cables, treated the differentiation operator $p = d/dt$ as if it were a number: he wrote $y = \frac{1}{p+a}\, g$ for the solution of $y' + ay = g$ and expanded such expressions in series, obtaining correct answers with no proof whatsoever. Mathematicians were scandalized; Heaviside famously replied that he did not refuse his dinner simply because he did not understand digestion. Bromwich, Carson, and above all Doetsch later showed that Heaviside's *operational calculus* is rigorous once $p$ is interpreted through the integral transform

$$
\mathcal{L}\{f\}(s) = \int_0^\infty e^{-st} f(t)\, dt,
$$

which Laplace had already studied around 1780 in probability theory. Every one of Heaviside's operator rules is a theorem about this integral.

### The three-step workflow

The method replaces one hard problem by three easy ones:

1. **Transform**: apply $\mathcal{L}$ to the ODE. Linearity and the derivative theorem convert it into an *algebraic* equation in $Y(s) = \mathcal{L}\{y\}(s)$, with the initial conditions $y(0), y'(0), \ldots$ inserted automatically.
2. **Solve algebra**: solve for $Y(s)$ by division — typically $Y(s) = \dfrac{\text{known}(s)}{\text{characteristic polynomial}(s)}$.
3. **Invert**: decompose $Y(s)$ by partial fractions and read the answer from a transform table; uniqueness (Lerch's theorem) guarantees this recovers *the* solution.

**Why bother, when Topic 03 already solves constant-coefficient equations?** Three reasons:

- **Discontinuous forcing.** A voltage switched on at $t = 2$ is $V_0 u_2(t)$ with $u_c$ the unit step. Undetermined coefficients chokes on this; the $t$-shifting theorem handles it in one line.
- **Impulsive forcing.** A hammer blow lasting a millisecond is modeled by the Dirac delta $\delta(t - c)$, which is not a function at all — yet $\mathcal{L}\{\delta(t-c)\} = e^{-cs}$ makes it algebraically trivial.
- **Systems thinking.** Dividing output by input transforms defines the *transfer function* $H(s)$, the central object of control engineering and of modern state-space sequence models: one rational function that encodes the entire input-output behavior of the system.

### Assumptions and notation

We work with functions $f : [0, \infty) \to \mathbb{R}$ (or $\mathbb{C}$). The transform never sees $t \lt 0$; when using shifting theorems we tacitly extend $f$ by zero there.

| Symbol | Meaning |
| :--- | :--- |
| $\mathcal{L}\{f\}(s)$ or $F(s)$ | Laplace transform of $f$, a function of the complex variable $s$ |
| $f(t) \leftrightarrow F(s)$ | transform pair; lowercase in time, uppercase in frequency |
| $u_c(t)$ | unit (Heaviside) step: $0$ for $t \lt c$, $1$ for $t \ge c$ |
| $\delta(t - c)$ | Dirac delta at $c$: unit impulse, $\mathcal{L}\{\delta(t-c)\} = e^{-cs}$ |
| $(f * g)(t)$ | convolution $\int_0^t f(\tau)\, g(t - \tau)\, d\tau$ |
| $H(s)$ | transfer function; $h(t) = \mathcal{L}^{-1}\{H\}$ is the impulse response |

**Standing assumptions** (justified by the existence theorem below): $f$ is *piecewise continuous* on every finite interval $[0, T]$ — finitely many jump discontinuities, finite one-sided limits — and of *exponential order* $\alpha$: there exist constants $M \gt 0$ and $T_0$ with $\lvert f(t) \rvert \le M e^{\alpha t}$ for all $t \ge T_0$. Polynomials, exponentials, sines, cosines, and their products all qualify; $e^{t^2}$ does not, and indeed has no Laplace transform for any $s$.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition 2.1 (Laplace transform).** For $f : [0, \infty) \to \mathbb{C}$ locally integrable, the Laplace transform is

$$
F(s) = \mathcal{L}\{f\}(s) = \int_0^\infty e^{-st} f(t)\, dt = \lim_{T \to \infty} \int_0^T e^{-st} f(t)\, dt,
$$

defined for every $s$ (real or complex) at which the limit exists.

**Definition 2.2 (exponential order).** $f$ is of exponential order $\alpha$ if there are constants $M \gt 0$, $T_0 \ge 0$ such that $\lvert f(t) \rvert \le M e^{\alpha t}$ for all $t \ge T_0$.

**Theorem 2.3 (Existence and region of convergence).** If $f$ is piecewise continuous on every $[0, T]$ and of exponential order $\alpha$, then $F(s)$ exists (the integral converges absolutely) for every $s$ with $\operatorname{Re}(s) \gt \alpha$, satisfies the bound $\lvert F(s) \rvert \le \frac{M}{\operatorname{Re}(s) - \alpha} + C$ for a constant $C$ depending only on $f$ on $[0, T_0]$, and $F(s) \to 0$ as $\operatorname{Re}(s) \to \infty$. The set of convergence is a right half-plane, called the **region of convergence** (ROC); $F$ is analytic there. *Proof in Section 3.*

The decay $F(s) \to 0$ is a genuine constraint: a rational function like $\frac{s^2 + 1}{s + 2}$, which grows at infinity, cannot be the transform of any function of exponential order — a useful sanity check on computations.

**Theorem 2.4 (Linearity).** For constants $a, b$ and transforms existing on a common half-plane, $\mathcal{L}\{a f + b g\} = a F + b G$. *Proof:* immediate from linearity of the integral.

**Theorem 2.5 (Lerch: injectivity).** If $f$ and $g$ are piecewise continuous, of exponential order, and $F(s) = G(s)$ for all $s \gt s_0$, then $f(t) = g(t)$ at every point where both are continuous. In particular a *continuous* function is uniquely determined by its transform. This theorem is what makes "invert by table lookup" a rigorous method: once we exhibit any continuous $f$ with transform $F$, it is the only one.

### Building the transform table from scratch, part I

Each entry below is a small computation with the definition; throughout, $s \gt \alpha$ for the relevant order $\alpha$.

**(1) Constant function.** For $s \gt 0$:

$$
\mathcal{L}\{1\}(s) = \int_0^\infty e^{-st}\, dt = \left. -\frac{e^{-st}}{s} \right\rvert_0^\infty = \frac{1}{s}.
$$

**(2) Exponential.** For $s \gt a$:

$$
\mathcal{L}\{e^{at}\}(s) = \int_0^\infty e^{-(s-a)t}\, dt = \frac{1}{s - a}.
$$

This single entry generates many others by substitution.

**(3) Sine and cosine, by complex exponentials.** Take $a = ib$ in (2), valid for $s \gt 0$:

$$
\mathcal{L}\{e^{ibt}\}(s) = \frac{1}{s - ib} = \frac{s + ib}{s^2 + b^2}.
$$

Since $e^{ibt} = \cos bt + i \sin bt$, matching real and imaginary parts gives

$$
\mathcal{L}\{\cos bt\} = \frac{s}{s^2 + b^2}, \qquad \mathcal{L}\{\sin bt\} = \frac{b}{s^2 + b^2}.
$$

**(4) Hyperbolic functions, by linearity.** From $\cosh bt = \tfrac{1}{2}(e^{bt} + e^{-bt})$ and $\sinh bt = \tfrac{1}{2}(e^{bt} - e^{-bt})$, for $s \gt \lvert b \rvert$:

$$
\mathcal{L}\{\cosh bt\} = \frac{1}{2}\left( \frac{1}{s-b} + \frac{1}{s+b} \right) = \frac{s}{s^2 - b^2}, \qquad \mathcal{L}\{\sinh bt\} = \frac{b}{s^2 - b^2}.
$$

**(5) Powers of $t$, by reduction.** Integration by parts with $u = t^n$, $dv = e^{-st} dt$ gives, for $s \gt 0$,

$$
\mathcal{L}\{t^n\} = \frac{n}{s}\, \mathcal{L}\{t^{n-1}\},
$$

and since $\mathcal{L}\{t^0\} = 1/s$, induction yields $\mathcal{L}\{t^n\} = \dfrac{n!}{s^{n+1}}$. (Non-integer powers need the Gamma function: $\mathcal{L}\{t^p\} = \Gamma(p+1)/s^{p+1}$ for $p \gt -1$; see exercise L3.1.)

### Building the transform table, part II: the s-shifting theorem

**Theorem 2.6 (s-shifting / first shifting theorem).** If $\mathcal{L}\{f\}(s) = F(s)$ for $s \gt \alpha$, then

$$
\mathcal{L}\{e^{at} f(t)\}(s) = F(s - a) \qquad \text{for } s \gt \alpha + a.
$$

*Proof:* $\int_0^\infty e^{-st} e^{at} f(t)\, dt = \int_0^\infty e^{-(s-a)t} f(t)\, dt = F(s-a)$. $\blacksquare$

Multiplying by $e^{at}$ in time simply *slides the whole frequency picture* by $a$. Two immediate consequences complete the core table:

$$
\mathcal{L}\{t^n e^{at}\} = \frac{n!}{(s-a)^{n+1}}, \qquad \mathcal{L}\{e^{at} \sin bt\} = \frac{b}{(s-a)^2 + b^2}, \qquad \mathcal{L}\{e^{at} \cos bt\} = \frac{s-a}{(s-a)^2 + b^2}.
$$

**Master table** (all derived above; ROC in parentheses):

| $f(t)$ | $F(s)$ | valid for |
| :--- | :--- | :--- |
| $1$ | $1/s$ | $s \gt 0$ |
| $t^n$ | $n!/s^{n+1}$ | $s \gt 0$ |
| $e^{at}$ | $1/(s-a)$ | $s \gt a$ |
| $\sin bt$ | $b/(s^2+b^2)$ | $s \gt 0$ |
| $\cos bt$ | $s/(s^2+b^2)$ | $s \gt 0$ |
| $\sinh bt$, $\cosh bt$ | $b/(s^2-b^2)$, $s/(s^2-b^2)$ | $s \gt \lvert b \rvert$ |
| $t^n e^{at}$ | $n!/(s-a)^{n+1}$ | $s \gt a$ |
| $e^{at}\sin bt$ | $b/((s-a)^2+b^2)$ | $s \gt a$ |
| $u_c(t)$ | $e^{-cs}/s$ | $s \gt 0$ |
| $\delta(t-c)$ | $e^{-cs}$ | all $s$ |

### Operational theorems, part I: calculus becomes algebra

Let $F = \mathcal{L}\{f\}$ exist for $s \gt \alpha$. All proofs are given in Section 3.

**Theorem 2.7 (Derivative theorem).** If $f$ is continuous, $f'$ piecewise continuous, both of exponential order $\alpha$, then for $s \gt \alpha$

$$
\mathcal{L}\{f'\}(s) = s F(s) - f(0),
$$

and by induction, $\mathcal{L}\{f^{(n)}\}(s) = s^n F(s) - s^{n-1} f(0) - s^{n-2} f'(0) - \cdots - f^{(n-1)}(0)$.

**Theorem 2.8 (t-shifting / second shifting theorem).** For $c \ge 0$,

$$
\mathcal{L}\{u_c(t)\, f(t - c)\}(s) = e^{-cs} F(s).
$$

A signal delayed by $c$ (and off before $c$) acquires the factor $e^{-cs}$ — the algebraic fingerprint of a time delay.

**Theorem 2.9 (Scaling).** For $a \gt 0$, $\mathcal{L}\{f(at)\}(s) = \frac{1}{a} F\!\left(\frac{s}{a}\right)$.

**Theorem 2.10 (Multiplication by $t$ / derivative of the transform).** $\mathcal{L}\{t f(t)\}(s) = -F'(s)$, and generally $\mathcal{L}\{t^n f(t)\} = (-1)^n F^{(n)}(s)$.

**Theorem 2.11 (Transform of an integral).**

$$
\mathcal{L}\left\{ \int_0^t f(\tau)\, d\tau \right\}(s) = \frac{F(s)}{s}.
$$

Division by $s$ integrates; multiplication by $s$ differentiates — the two halves of Heaviside's operator dictionary.

**Theorem 2.12 (Periodic functions).** If $f$ has period $T$ (and is piecewise continuous), then for $s \gt 0$

$$
\mathcal{L}\{f\}(s) = \frac{1}{1 - e^{-sT}} \int_0^T e^{-st} f(t)\, dt :
$$

one period's worth of integral, amplified by the geometric-series factor that sums all the repeats.

### Operational theorems, part II: impulses, convolution, and asymptotics

**Definition 2.13 (Dirac delta as a limit of pulses).** For $\varepsilon \gt 0$ let $d_\varepsilon(t) = \frac{1}{\varepsilon}$ on $[0, \varepsilon]$ and $0$ elsewhere — a unit-area pulse. For any continuous $g$, $\int_0^\infty d_\varepsilon(t - c)\, g(t)\, dt \to g(c)$ as $\varepsilon \to 0^+$. The **Dirac delta** $\delta(t - c)$ is the idealized limit: not a function but a *distribution* defined by its sampling action $\int \delta(t - c) g(t)\, dt = g(c)$. Consequently

$$
\mathcal{L}\{\delta(t - c)\}(s) = e^{-cs}, \qquad \mathcal{L}\{\delta\}(s) = 1,
$$

which one also obtains as $\lim_{\varepsilon \to 0^+} \mathcal{L}\{d_\varepsilon\} = \lim_{\varepsilon \to 0^+} \frac{1 - e^{-s\varepsilon}}{s\varepsilon} = 1$.

**Definition 2.14 (Convolution).** $(f * g)(t) = \int_0^t f(\tau)\, g(t - \tau)\, d\tau$. Convolution is commutative, associative, and bilinear.

**Theorem 2.15 (Convolution theorem).** Under the standing assumptions, for $s$ large enough,

$$
\mathcal{L}\{f * g\}(s) = F(s)\, G(s).
$$

**Definition 2.16 (Transfer function and impulse response).** For the system $a y'' + b y' + c y = u(t)$ with zero initial state, $Y(s) = H(s) U(s)$ where $H(s) = \frac{1}{a s^2 + b s + c}$ is the **transfer function**. Its inverse transform $h = \mathcal{L}^{-1}\{H\}$ is the **impulse response** — the output when $u = \delta$ — and the convolution theorem gives the time-domain representation $y = h * u$.

**Theorem 2.17 (Initial and final value theorems).** Suppose $f$ and $f'$ satisfy the standing assumptions.

- **IVT:** $\displaystyle \lim_{s \to \infty} s F(s) = f(0^+)$.
- **FVT:** if additionally every pole of $s F(s)$ lies in the open left half-plane $\operatorname{Re}(s) \lt 0$, then $\displaystyle \lim_{s \to 0^+} s F(s) = \lim_{t \to \infty} f(t)$.

The pole condition is essential. For $f(t) = e^{t}$, $s F(s) = \frac{s}{s-1} \to 0$ as $s \to 0$, yet $f(t) \to \infty$: the pole at $s = 1$ (right half-plane) invalidates the theorem. Similarly $f = \cos t$ gives $sF(s) = \frac{s^2}{s^2+1} \to 0$ while $f(t)$ has no limit at all (poles on the imaginary axis).

## 3. Step-by-Step Mathematical Proofs & Derivations

### Proof 1: Existence for functions of exponential order

**Claim.** If $f$ is piecewise continuous on every $[0, T]$ and $\lvert f(t) \rvert \le M e^{\alpha t}$ for $t \ge T_0$, then $\int_0^\infty e^{-st} f(t)\, dt$ converges absolutely for every real $s \gt \alpha$ (and every complex $s$ with $\operatorname{Re}(s) \gt \alpha$), and $F(s) \to 0$ as $s \to \infty$.

**Step 1 — split the integral.** Write

$$
\int_0^\infty e^{-st} \lvert f(t) \rvert\, dt = \underbrace{\int_0^{T_0} e^{-st} \lvert f(t) \rvert\, dt}_{I_1} + \underbrace{\int_{T_0}^\infty e^{-st} \lvert f(t) \rvert\, dt}_{I_2}.
$$

**Step 2 — the finite piece is harmless.** On $[0, T_0]$, $f$ is piecewise continuous, hence bounded by some $K$; therefore $I_1 \le K T_0 \lt \infty$ for every $s \ge 0$ (and $I_1 \le K T_0 e^{\lvert s \rvert T_0}$ for any real $s$, still finite).

**Step 3 — the tail converges geometrically.** For $t \ge T_0$ and $s \gt \alpha$:

$$
e^{-st} \lvert f(t) \rvert \le M e^{-st} e^{\alpha t} = M e^{-(s - \alpha) t},
$$

so

$$
I_2 \le M \int_{T_0}^\infty e^{-(s-\alpha)t}\, dt = \frac{M e^{-(s-\alpha) T_0}}{s - \alpha} \le \frac{M}{s - \alpha} \lt \infty.
$$

**Step 4 — absolute convergence and the ROC.** The comparison test gives absolute convergence of $\int_0^\infty e^{-st} f(t)\, dt$ for every $s \gt \alpha$; for complex $s$ replace $s$ by $\operatorname{Re}(s)$ in Steps 2 and 3 since $\lvert e^{-st} \rvert = e^{-\operatorname{Re}(s) t}$. The set of convergence therefore contains the half-plane $\operatorname{Re}(s) \gt \alpha$, and if the integral converges at $s_0$ it converges for all $\operatorname{Re}(s) \gt \operatorname{Re}(s_0)$ — the ROC is always a right half-plane.

**Step 5 — decay at infinity.** From Steps 2 and 3, $\lvert F(s) \rvert \le K \int_0^{T_0} e^{-st} dt + \frac{M}{s-\alpha} \le K \frac{1 - e^{-sT_0}}{s} + \frac{M}{s - \alpha}$, and both terms tend to $0$ as $s \to \infty$.

$$
\boxed{\ f \text{ piecewise continuous of exponential order } \alpha \ \Longrightarrow\ F(s) \text{ exists and is bounded by } \tfrac{M}{s - \alpha} + C \text{ for } s \gt \alpha, \ \text{with } F(s) \to 0.\ }
$$

### Proof 2: The derivative theorem, first and n-th order

**Claim.** If $f$ is continuous on $[0, \infty)$, $f'$ piecewise continuous, and both of exponential order $\alpha$, then for $s \gt \alpha$, $\mathcal{L}\{f'\} = sF(s) - f(0)$; and for $f^{(n)}$ under the analogous hypotheses,

$$
\mathcal{L}\{f^{(n)}\}(s) = s^n F(s) - \sum_{k=0}^{n-1} s^{\,n-1-k} f^{(k)}(0).
$$

**Step 1 — integrate by parts on a finite interval.** With $u = e^{-st}$, $dv = f'(t)\, dt$:

$$
\int_0^T e^{-st} f'(t)\, dt = \Big[ e^{-st} f(t) \Big]_0^T + s \int_0^T e^{-st} f(t)\, dt = e^{-sT} f(T) - f(0) + s \int_0^T e^{-st} f(t)\, dt.
$$

(If $f'$ has jumps, apply this on each subinterval of continuity; the interior boundary terms telescope because $f$ itself is continuous.)

**Step 2 — kill the boundary term using exponential order.** For $T \ge T_0$,

$$
\lvert e^{-sT} f(T) \rvert \le M e^{-sT} e^{\alpha T} = M e^{-(s - \alpha) T} \longrightarrow 0 \quad (T \to \infty),
$$

precisely because $s \gt \alpha$. This is the step where the exponential-order hypothesis earns its keep: without it the boundary term could survive and the theorem fails.

**Step 3 — take the limit.** Letting $T \to \infty$ in Step 1, both remaining terms converge (the last by Proof 1), giving

$$
\mathcal{L}\{f'\}(s) = s F(s) - f(0).
$$

**Step 4 — induction for order $n$.** The case $n = 1$ is Step 3. Assume the formula for $n - 1$ applied to $f'$, i.e. $\mathcal{L}\{f^{(n)}\} = \mathcal{L}\{(f')^{(n-1)}\} = s^{n-1} \mathcal{L}\{f'\} - \sum_{k=0}^{n-2} s^{\,n-2-k} (f')^{(k)}(0)$. Substituting $\mathcal{L}\{f'\} = sF - f(0)$ and reindexing $(f')^{(k)}(0) = f^{(k+1)}(0)$:

$$
\mathcal{L}\{f^{(n)}\} = s^n F(s) - s^{n-1} f(0) - \sum_{j=1}^{n-1} s^{\,n-1-j} f^{(j)}(0),
$$

which is the claimed formula. $\blacksquare$

$$
\boxed{\ \mathcal{L}\{f^{(n)}\}(s) = s^n F(s) - s^{n-1} f(0) - s^{n-2} f'(0) - \cdots - f^{(n-1)}(0)\ }
$$

Applied to $ay'' + by' + cy = g(t)$, this yields $(as^2 + bs + c) Y(s) = G(s) + (as + b) y(0) + a y'(0)$: the characteristic polynomial appears on the left, and the initial data enter as a polynomial on the right. Calculus has become algebra.

### Proof 3: The convolution theorem

**Claim.** If $f, g$ are piecewise continuous of exponential order $\alpha$, then for $s \gt \alpha$, $\mathcal{L}\{f * g\}(s) = F(s) G(s)$.

**Step 1 — write the product as a double integral.** For $s \gt \alpha$ both single integrals converge absolutely (Proof 1), so

$$
F(s) G(s) = \left( \int_0^\infty e^{-s\tau} f(\tau)\, d\tau \right) \left( \int_0^\infty e^{-s\sigma} g(\sigma)\, d\sigma \right) = \int_0^\infty \!\! \int_0^\infty e^{-s(\tau + \sigma)} f(\tau)\, g(\sigma)\, d\sigma\, d\tau.
$$

Absolute convergence of each factor makes the double integral absolutely convergent, which licenses every rearrangement below (Fubini–Tonelli).

**Step 2 — change variables to total time.** Fix $\tau$ and substitute $t = \tau + \sigma$ (so $\sigma = t - \tau$, $d\sigma = dt$; as $\sigma$ runs over $(0, \infty)$, $t$ runs over $(\tau, \infty)$):

$$
F(s) G(s) = \int_0^\infty \int_\tau^\infty e^{-st} f(\tau)\, g(t - \tau)\, dt\, d\tau.
$$

**Step 3 — swap the order of integration.** The region of integration is $\{ (\tau, t) : 0 \le \tau \le t \lt \infty \}$ — the wedge above the diagonal in the $(\tau, t)$-plane. Described the other way: $t$ runs over $(0, \infty)$ and, for fixed $t$, $\tau$ runs over $[0, t]$. By Fubini (justified in Step 1):

$$
F(s) G(s) = \int_0^\infty e^{-st} \left( \int_0^t f(\tau)\, g(t - \tau)\, d\tau \right) dt.
$$

**Step 4 — recognize the convolution.** The inner integral is exactly $(f * g)(t)$, so the right side is $\mathcal{L}\{f * g\}(s)$. $\blacksquare$

$$
\boxed{\ \mathcal{L}\{f * g\}(s) = F(s)\, G(s)\ }
$$

**Why this matters structurally.** The theorem says the Laplace transform is an *algebra homomorphism* from (functions under convolution) to (functions under pointwise multiplication). Every "multiply transfer functions in series", every "output equals impulse response convolved with input", and — as we will see — every "state-space sequence model trains as a convolution" is this one theorem at work.

### Proof 4: The t-shifting theorem

**Claim.** For $c \ge 0$ and $F = \mathcal{L}\{f\}$ existing for $s \gt \alpha$:

$$
\mathcal{L}\{ u_c(t)\, f(t - c) \}(s) = e^{-cs} F(s), \qquad s \gt \alpha.
$$

**Step 1 — the integrand vanishes before $c$.** The function $u_c(t) f(t-c)$ equals $0$ for $t \lt c$ and $f(t - c)$ for $t \ge c$, so

$$
\mathcal{L}\{u_c(t) f(t-c)\}(s) = \int_c^\infty e^{-st} f(t - c)\, dt.
$$

**Step 2 — substitute $\tau = t - c$.** Then $t = \tau + c$, $dt = d\tau$, and the range $t \in [c, \infty)$ becomes $\tau \in [0, \infty)$:

$$
\int_c^\infty e^{-st} f(t-c)\, dt = \int_0^\infty e^{-s(\tau + c)} f(\tau)\, d\tau = e^{-cs} \int_0^\infty e^{-s\tau} f(\tau)\, d\tau = e^{-cs} F(s). \qquad \blacksquare
$$

$$
\boxed{\ \mathcal{L}\{u_c(t) f(t-c)\} = e^{-cs} F(s)\ }
$$

**Reading it backwards (the useful direction).** Whenever an inverse-transform problem presents $e^{-cs} F(s)$, the answer is the inverse of $F$, *delayed by $c$ and switched off before $c$*: $\mathcal{L}^{-1}\{e^{-cs} F(s)\} = u_c(t) f(t - c)$. Special case $f = 1$: $\mathcal{L}\{u_c\} = e^{-cs}/s$.

**Caution.** The theorem transforms $f(t - c)$, the *shifted* function — not $f(t)$ multiplied by a step. To transform $u_c(t) g(t)$ one must first rewrite $g(t) = \tilde{g}(t - c)$ with $\tilde{g}(\tau) = g(\tau + c)$. Forgetting this re-centering is the single most common Laplace error in practice.

### Proof 5 (auxiliary): scaling, multiplication by t, integrals, periodic functions

Four short proofs complete the operational dictionary. Throughout $s \gt \alpha$.

**(a) Scaling.** For $a \gt 0$, substitute $\tau = at$:

$$
\mathcal{L}\{f(at)\}(s) = \int_0^\infty e^{-st} f(at)\, dt = \frac{1}{a} \int_0^\infty e^{-(s/a)\tau} f(\tau)\, d\tau = \frac{1}{a} F\!\left( \frac{s}{a} \right). \qquad \blacksquare
$$

**(b) Multiplication by $t$.** For $s \gt \alpha$ the integral $\int_0^\infty e^{-st} f(t)\, dt$ converges absolutely and locally uniformly in $s$ (dominate by $M e^{-(s_1 - \alpha)t}$ on any half-line $s \ge s_1 \gt \alpha$), so differentiation under the integral sign is legitimate:

$$
F'(s) = \frac{d}{ds} \int_0^\infty e^{-st} f(t)\, dt = \int_0^\infty (-t) e^{-st} f(t)\, dt = -\mathcal{L}\{t f(t)\}(s),
$$

hence $\mathcal{L}\{t f(t)\} = -F'(s)$, and iterating gives $\mathcal{L}\{t^n f(t)\} = (-1)^n F^{(n)}(s)$. $\blacksquare$ (Check: $f = 1$ gives $\mathcal{L}\{t^n\} = (-1)^n \frac{d^n}{ds^n} s^{-1} = n!/s^{n+1}$, agreeing with Section 2.)

**(c) Transform of an integral.** Let $\varphi(t) = \int_0^t f(\tau) d\tau$. Then $\varphi$ is continuous, $\varphi' = f$ at points of continuity, $\varphi(0) = 0$, and $\varphi$ is of exponential order $\max(\alpha, 0) + \varepsilon$. The derivative theorem applied to $\varphi$ gives $F(s) = \mathcal{L}\{\varphi'\} = s\, \mathcal{L}\{\varphi\} - 0$, so

$$
\mathcal{L}\left\{ \int_0^t f(\tau)\, d\tau \right\} = \frac{F(s)}{s}. \qquad \blacksquare
$$

**(d) Periodic functions.** If $f(t + T) = f(t)$, split the integral over periods and substitute $t = \tau + nT$ in the $n$-th piece:

$$
\mathcal{L}\{f\}(s) = \sum_{n=0}^\infty \int_{nT}^{(n+1)T} e^{-st} f(t)\, dt = \sum_{n=0}^\infty e^{-snT} \int_0^T e^{-s\tau} f(\tau)\, d\tau.
$$

For $s \gt 0$ the geometric series $\sum_n (e^{-sT})^n$ converges to $\frac{1}{1 - e^{-sT}}$, giving

$$
\boxed{\ \mathcal{L}\{f\}(s) = \frac{1}{1 - e^{-sT}} \int_0^T e^{-st} f(t)\, dt\ } \qquad \blacksquare
$$

### Proof 6: A discontinuous IVP solved end-to-end

**Problem.** An undamped oscillator, initially at rest, is hit at time $t = \pi$ by a constant force that stays on:

$$
y'' + y = u_\pi(t), \qquad y(0) = 0, \quad y'(0) = 0.
$$

**Step 1 — transform the equation.** Using $\mathcal{L}\{y''\} = s^2 Y - s y(0) - y'(0) = s^2 Y$ and $\mathcal{L}\{u_\pi\} = e^{-\pi s}/s$:

$$
s^2 Y(s) + Y(s) = \frac{e^{-\pi s}}{s} \quad \Longrightarrow \quad Y(s) = e^{-\pi s} \cdot \frac{1}{s (s^2 + 1)}.
$$

One line, initial conditions included: this is the entire "solving" phase.

**Step 2 — partial fractions on the non-delayed part.** Seek $\frac{1}{s(s^2+1)} = \frac{A}{s} + \frac{Bs + C}{s^2 + 1}$. Multiplying out: $1 = A(s^2 + 1) + (Bs + C)s$. At $s = 0$: $A = 1$. Matching $s^2$ coefficients: $0 = A + B$, so $B = -1$; matching $s$: $C = 0$. Hence

$$
\frac{1}{s(s^2+1)} = \frac{1}{s} - \frac{s}{s^2 + 1} \quad \leftrightarrow \quad \varphi(t) = 1 - \cos t.
$$

(Sanity check: $\varphi(0) = 0$, and $\varphi$ is the standard step response of the undamped oscillator.)

**Step 3 — undo the delay with t-shifting.** Since $Y(s) = e^{-\pi s} \Phi(s)$ with $\Phi = \mathcal{L}\{\varphi\}$, Proof 4 (read backwards) gives

$$
y(t) = u_\pi(t)\, \varphi(t - \pi) = u_\pi(t) \left[ 1 - \cos(t - \pi) \right] = u_\pi(t) \left[ 1 + \cos t \right],
$$

using $\cos(t - \pi) = -\cos t$.

**Step 4 — interpret and verify.** For $t \lt \pi$: $y \equiv 0$ (nothing has happened yet — causality is automatic). For $t \ge \pi$: $y = 1 + \cos t$ oscillates forever between $0$ and $2$ about the new equilibrium $y = 1$. At $t = \pi$: $y(\pi) = 1 + \cos\pi = 0$ and $y'(\pi) = -\sin\pi = 0$, so $y$ and $y'$ are continuous across the switch — only $y''$ jumps (by exactly $1$, matching the jump in forcing). Direct substitution confirms $y'' + y = 1$ for $t \gt \pi$.

$$
\boxed{\ y(t) = u_\pi(t)\, (1 + \cos t)\ }
$$

Compare the classical alternative: solve on $[0, \pi)$, solve again on $[\pi, \infty)$, and match $y, y'$ at $t = \pi$ by hand. The transform did the matching for us.

### Proof 7: The final value theorem

**Claim.** Suppose $f$ is continuous, $f'$ piecewise continuous, both of exponential order, and $\lim_{t \to \infty} f(t) = f_\infty$ exists (equivalently: all poles of $sF(s)$ lie in the open left half-plane). Then

$$
\lim_{s \to 0^+} s F(s) = f_\infty.
$$

**Step 1 — start from the derivative theorem.** For all $s \gt 0$ in the ROC:

$$
\int_0^\infty e^{-st} f'(t)\, dt = s F(s) - f(0).
$$

**Step 2 — compute the limit of the left side as $s \to 0^+$.** Since $f_\infty$ exists and $f'$ is integrable on $[0, \infty)$ (because $\int_0^T f' = f(T) - f(0) \to f_\infty - f(0)$ converges, and the hypotheses give absolute control of the tail), dominated convergence applies: $e^{-st} f'(t) \to f'(t)$ pointwise as $s \to 0^+$, dominated by $\lvert f'(t) \rvert$ on any interval where the improper integral converges absolutely, and by a uniform-convergence argument in general. Hence

$$
\lim_{s \to 0^+} \int_0^\infty e^{-st} f'(t)\, dt = \int_0^\infty f'(t)\, dt = \lim_{T \to \infty} \big[ f(T) - f(0) \big] = f_\infty - f(0).
$$

**Step 3 — equate.** Taking $s \to 0^+$ in Step 1: $f_\infty - f(0) = \lim_{s \to 0^+} s F(s) - f(0)$, so

$$
\boxed{\ \lim_{s \to 0^+} s F(s) = \lim_{t \to \infty} f(t)\ } \qquad \blacksquare
$$

**Step 4 — where the hypothesis is used, and the counterexample.** Step 2 required $\int_0^\infty f'\, dt$ to converge — i.e. $f$ to actually settle. Take $f(t) = e^{t}$: then $sF(s) = \frac{s}{s - 1}$, whose limit as $s \to 0^+$ is $0$, but $f(t) \to \infty$. The failure is visible in the $s$-plane: $sF(s)$ has a pole at $s = 1$ in the right half-plane, the ROC is $s \gt 1$, and the point $s = 0$ was never in the ROC — the "limit" computed there is a value of the analytic continuation, not of the transform. Likewise $f = \cos t$ (poles at $\pm i$) gives $sF(s) \to 0$ though $f$ oscillates forever. **Moral:** the FVT is a theorem about systems that already decay; check the poles of $sF(s)$ before trusting it. The initial value theorem $\lim_{s \to \infty} sF(s) = f(0^+)$ has no such trap because $s \to \infty$ stays inside every ROC.

## 4. Computational & Algorithmic Insights

### Numerical inversion is ill-posed

The forward transform is a smoothing operator: it damps high-frequency detail of $f$ exponentially. Inverting it must therefore *amplify* exponentially — the hallmark of an ill-posed problem. Concretely, the perturbation $f_\varepsilon(t) = f(t) + \varepsilon \sin(\omega t)$ changes $F$ by only $\varepsilon \omega / (s^2 + \omega^2) = O(\varepsilon/\omega)$: transforms of wildly different functions can be numerically indistinguishable, so naive inversion from sampled values of $F(s)$ on the real axis is hopeless in floating point.

Practical algorithms fall into two families:

- **Talbot's method** deforms the Bromwich contour $\operatorname{Re}(s) = \gamma$ into a cotangent-shaped path wrapping around the negative real axis, where $e^{st}$ decays; trapezoidal quadrature on that contour converges geometrically. This is the method of choice when $F$ can be evaluated at complex $s$.
- **The Gaver–Stehfest method** uses only *real* samples $F(k \ln 2 / t)$ combined with alternating binomial weights. Elegant, but the weights grow explosively: roughly one significant digit is lost per term, so it needs high-precision arithmetic beyond about 8 terms — ill-posedness made visible.

```text
Algorithm: Gaver–Stehfest inversion (fixed t > 0, even N)
  for k = 1..N:
      V_k = alternating binomial weight (grows like N^N — precision hazard)
  f(t) ~ (ln 2 / t) * sum_{k=1..N} V_k * F(k * ln 2 / t)
```

**Symbolic computation**, by contrast, is exact and robust: computer algebra systems (SymPy's `laplace_transform` / `inverse_laplace_transform`, Mathematica, Maple) implement the operational theorems and partial-fraction inversion directly. For rational $F$ the whole pipeline — transform, solve, partial fractions, invert — is exact rational-plus-table arithmetic, which is why the Laplace method automates so well.

### The algebraic workflow as an algorithm

```text
Algorithm: solve a y'' + b y' + c y = g(t), y(0) = y0, y'(0) = v0
  1. G(s) <- L{g}         (table + shifting theorems; steps/impulses give e^{-cs} factors)
  2. Y(s) <- [ G(s) + (a s + b) y0 + a v0 ] / (a s^2 + b s + c)
  3. Split off every e^{-cs} factor: Y = sum_j e^{-c_j s} R_j(s), R_j rational
  4. Partial fractions on each R_j:
       distinct real pole p:      A/(s-p)              -> A e^{pt}
       repeated pole p (mult m):  A_k/(s-p)^k, k<=m    -> A_k t^{k-1} e^{pt}/(k-1)!
       complex pair a +- i b:     (B(s-a)+Cb)/((s-a)^2+b^2) -> e^{at}(B cos bt + C sin bt)
  5. Invert term by term; apply u_{c_j}(t) shifts for each e^{-c_j s}
```

**Cost analysis.** Step 2 is $O(1)$ algebra. Step 4 requires the roots of the characteristic polynomial — for degree $n$, this is the same eigenvalue computation ($O(n^3)$ via companion-matrix QR) that the time-domain method needs; the transform does not avoid root-finding, it *organizes everything else around it*. Solving for $k$ different forcings or initial conditions reuses the same factorization: only the numerator changes. This is exactly why control engineers manipulate transfer functions symbolically all day and only invert at the very end — and often never invert at all, reading stability, steady-state gain ($H(0)$, by the FVT), and resonance peaks straight off $H(s)$.

### The s-plane: Fourier, eigenvalues, and stability

**Fourier as a boundary case.** Setting $s = i\omega$ (when the ROC allows it) turns the Laplace transform into the Fourier transform of the causal signal: $F(i\omega) = \int_0^\infty e^{-i\omega t} f(t)\, dt$. The Laplace transform is a Fourier transform with an adjustable exponential damping $e^{-\sigma t}$ built in — which is why it converges for growing signals that defeat Fourier, and why frequency-response plots (Bode, Nyquist) are evaluations of $H(s)$ on the imaginary axis.

**Poles are eigenvalues.** For the system $x' = Ax$, taking transforms gives $sX(s) - x(0) = AX(s)$, so

$$
X(s) = (sI - A)^{-1} x(0) = \frac{\operatorname{adj}(sI - A)}{\det(sI - A)}\, x(0).
$$

Every pole of every entry of $X(s)$ is a root of $\det(sI - A)$ — an eigenvalue of $A$. The partial-fraction modes $e^{pt}$ of the transform method and the eigenmodes $e^{\lambda t}$ of the spectral method (Topic 03, and `linear_algebra` Topic 06) are the *same objects seen from two sides*.

**Stability dictionary.** Because each pole $p = \sigma + i\omega$ contributes a mode $t^k e^{\sigma t} e^{i\omega t}$:

| Pole location | Mode behavior | System verdict |
| :--- | :--- | :--- |
| $\operatorname{Re}(p) \lt 0$ | decays exponentially | asymptotically stable |
| $\operatorname{Re}(p) = 0$, simple | bounded oscillation | marginally stable |
| $\operatorname{Re}(p) = 0$, repeated | grows like $t^k$ | unstable (resonance) |
| $\operatorname{Re}(p) \gt 0$ | grows exponentially | unstable |

"All poles in the open left half-plane" is simultaneously: the condition for BIBO stability, the hypothesis of the final value theorem, and the design target of every feedback controller in Section 5.

## 5. Real-World Physics & AI/ML Applications

### Physics: switched circuits, hammer blows, and diffusion

**RLC circuits with switched sources.** Kirchhoff's voltage law for a series RLC circuit gives $L q'' + R q' + q/C = v(t)$, formally identical to the damped oscillator. Real sources are switched: $v(t) = V_0 u_{t_0}(t)$ (a supply thrown on at $t_0$), or a rectangular pulse $V_0 [u_a(t) - u_b(t)]$. The transform method is the industry-standard treatment because $e^{-t_0 s}$ factors track every switch exactly, and the impedance picture emerges for free: transforming with zero initial state gives $Z(s) I(s) = V(s)$ with $Z(s) = Ls + R + \frac{1}{Cs}$ — inductors become $Ls$, capacitors $\frac{1}{Cs}$, and circuit analysis becomes complex algebra (Exercise L2.1).

**Mass-spring under a hammer blow.** A blow of impulse $I_0$ (momentum transfer) at $t = t_0$ on $m x'' + k x = f(t)$ is modeled by $f = I_0\, \delta(t - t_0)$. The solution from rest is $x(t) = \frac{I_0}{m \omega_0} u_{t_0}(t) \sin \omega_0 (t - t_0)$ with $\omega_0 = \sqrt{k/m}$: the hammer instantaneously sets velocity $I_0/m$, then free oscillation proceeds. Periodically repeated blows at the natural period drive resonance — the pushed-swing effect (Exercise L2.2).

**Diffusion and finance.** Applied to the heat equation $u_t = k u_{xx}$ in the time variable, the Laplace transform converts a PDE into an ODE in $x$ — the classical route to the heat kernel on the half-line. The Black–Scholes equation is a disguised heat equation, and Laplace transforms in the time-to-maturity variable are a standard tool for barrier-option and first-passage-time pricing, where the transform of a first-passage density is often *far* simpler than the density itself.

### Control theory: transfer functions and pole placement

Feedback control is conducted almost entirely in the $s$-domain. A plant $P(s)$ in a negative feedback loop with controller $C(s)$ has closed-loop transfer function

$$
T(s) = \frac{C(s) P(s)}{1 + C(s) P(s)},
$$

so the closed-loop poles are the roots of $1 + C(s) P(s) = 0$. **Design = moving poles.** The PID controller

$$
C(s) = K_p + \frac{K_i}{s} + K_d s
$$

is precisely the Laplace-domain form of "react to the error, its accumulated integral, and its rate of change". For a first-order plant $P(s) = \frac{1}{\tau s + 1}$, a PI controller makes the characteristic polynomial $\tau s^2 + (1 + K_p) s + K_i$: two gains, two coefficients, so the designer can place both closed-loop poles anywhere in the left half-plane (Exercise L2.3). Three standard $s$-domain readings do the rest of the analysis without any time-domain solving:

- **Steady-state accuracy** via the final value theorem: the step-response error is $\lim_{s \to 0} s \cdot \frac{1}{1 + C(s)P(s)} \cdot \frac{1}{s}$; the integrator pole $K_i/s$ forces it to zero.
- **Stability margins** from $C(i\omega) P(i\omega)$ — the Nyquist and Bode pictures, i.e. the transform on the Fourier axis.
- **Robustness**: how far the closed-loop poles sit from the imaginary axis bounds settling time and overshoot.

### Machine learning: state-space sequence models and discounting

**S4 and the transfer-function view of deep sequence models.** Structured state-space models (S4, S5, and the lineage leading to Mamba) build sequence layers from the linear time-invariant system

$$
x'(t) = A x(t) + B u(t), \qquad y(t) = C x(t),
$$

with learned matrices $A \in \mathbb{R}^{N \times N}$, $B \in \mathbb{R}^{N \times 1}$, $C \in \mathbb{R}^{1 \times N}$. Transforming with zero initial state: $Y(s) = H(s) U(s)$ with transfer function $H(s) = C (sI - A)^{-1} B$, whose impulse response is the kernel

$$
K(t) = C e^{A t} B, \qquad y = K * u.
$$

The convolution theorem is the entire reason these models are fast: *training* evaluates $y = K * u$ over a whole sequence at once with FFT-based convolution (parallel, $O(L \log L)$), while *inference* runs the same system as a step-by-step recurrence in $O(1)$ memory per step. One mathematical object, two computational modes — multiplication in the frequency domain versus recursion in the time domain. Stability of the learned layer is the left-half-plane condition on the eigenvalues of $A$; S4's HiPPO initialization chooses $A$ with exactly this structure so that $e^{At}$ implements a well-conditioned memory of the input history (Exercise L2.4).

**RL discounting is a Laplace weight.** The continuous-time discounted return of a reward stream $r(t)$ is

$$
V = \int_0^\infty e^{-\rho t} r(t)\, dt = \mathcal{L}\{r\}(\rho) :
$$

the value function is literally the Laplace transform of the reward evaluated at the discount rate $s = \rho$. Every operational theorem then has an RL reading — e.g. delaying all rewards by $c$ multiplies value by $e^{-\rho c}$ (t-shifting), and a reward stream generated by linear dynamics has value $C(\rho I - A)^{-1} B$-style closed forms (Exercise L2.5). Hyperbolic-discounting models in neuroscience arise as *mixtures* of exponential discounts, i.e. Laplace transforms integrated over $\rho$ — the brain, on this account, stores a Laplace representation of the past and future.
