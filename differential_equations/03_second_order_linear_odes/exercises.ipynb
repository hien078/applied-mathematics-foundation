### Problem L0.1: Order and Linearity Classification

**Problem Statement:** Classify each equation by order, linearity (linear or nonlinear), homogeneity (if linear), and coefficient type (constant or variable):

1. $y'' + 4y' + 4y = 0$
2. $t^2 y'' - t y' - 3y = \ln t$
3. $y'' + \sin y = 0$
4. $y\, y'' + (y')^2 = 1$

**Intuition:** Linearity is a property of how the *unknown function* enters, not of how $t$ enters — coefficients may depend on $t$ arbitrarily.

**Solution:**

The template for a second-order linear ODE is $a(t) y'' + b(t) y' + c(t) y = g(t)$: the unknown $y$ and its derivatives appear to the first power, never multiplied together, never inside a nonlinear function.

1. $y'' + 4y' + 4y = 0$: order 2. Matches the template with $a = 1, b = 4, c = 4, g = 0$. **Linear, homogeneous, constant coefficients.**

2. $t^2 y'' - t y' - 3y = \ln t$: order 2. Matches the template with $a = t^2, b = -t, c = -3, g = \ln t$. The functions of $t$ are legal coefficients and forcing. **Linear, non-homogeneous, variable coefficients** (a Cauchy–Euler equation with logarithmic forcing).

3. $y'' + \sin y = 0$: order 2. The unknown sits inside $\sin(\cdot)$, which is not of the form $c(t) y$. **Nonlinear** (the exact pendulum). Its small-angle linearization $y'' + y = 0$ is linear.

4. $y\, y'' + (y')^2 = 1$: order 2. Two violations: the product $y \cdot y''$ and the square $(y')^2$. **Nonlinear.** (Curiously, it is exactly $(y y')' = 1$, so it *can* be integrated — nonlinear does not always mean unsolvable.)

$$
\boxed{\text{1: linear homog. const.; 2: linear non-homog. variable; 3, 4: nonlinear}}
$$

> **Key Takeaway:** Linearity is checked against the template $a(t)y'' + b(t)y' + c(t)y = g(t)$: only the unknown must appear linearly — coefficients may be any functions of $t$.

### Problem L0.2: Why Exactly Two Arbitrary Constants?

**Problem Statement:** Explain precisely why the general solution of $y'' + p(t) y' + q(t) y = 0$ (with $p, q$ continuous on $I$) contains exactly two arbitrary constants — no more, no fewer. Your answer should identify the vector space involved and exhibit the isomorphism that gives its dimension.

**Intuition:** "Two integrations, two constants" is a heuristic; the real reason is that a solution is determined by its initial position *and* initial velocity.

**Solution:**

**Step 1 (the solution set is a vector space).** Let $S = \{y \in C^2(I) : L[y] = 0\}$ where $L[y] = y'' + p y' + q y$. Since $L$ is a linear operator, $S = \ker L$ is a vector subspace of $C^2(I)$: it contains $0$, and it is closed under linear combinations by superposition.

**Step 2 (the evaluation map).** Fix $t_0 \in I$ and define

$$
\Phi : S \to \mathbb{R}^2, \qquad \Phi(y) = \bigl(y(t_0),\, y'(t_0)\bigr)
$$

$\Phi$ is linear: evaluation and differentiation are both linear operations.

**Step 3 (surjective by existence).** For any target $(y_0, v_0) \in \mathbb{R}^2$, the existence theorem for linear equations produces a solution with $y(t_0) = y_0$, $y'(t_0) = v_0$ on all of $I$. Hence $\Phi$ is onto.

**Step 4 (injective by uniqueness).** If $\Phi(y) = (0, 0)$, then $y$ and the zero function solve the same initial value problem; by the uniqueness theorem $y \equiv 0$. Hence $\ker \Phi = \{0\}$ and $\Phi$ is one-to-one.

**Step 5 (conclusion).** $\Phi$ is a linear isomorphism, so

$$
\dim S = \dim \mathbb{R}^2 = 2
$$

Any basis $\{y_1, y_2\}$ (a fundamental set) expresses every solution uniquely as $y = c_1 y_1 + c_2 y_2$: exactly two free constants, in bijection with the two initial data $(y_0, v_0)$. Three solutions are always linearly dependent; one solution can never span.

$$
\boxed{\dim \ker L = 2 \ \text{ via the isomorphism } y \mapsto (y(t_0), y'(t_0))}
$$

> **Key Takeaway:** The two arbitrary constants are the coordinates of the solution in a basis of the two-dimensional null space of $L$ — and the dimension is 2 because a state is a (position, velocity) pair.

### Problem L0.3: What the Wronskian Actually Certifies

**Problem Statement:** Compute the Wronskian of $y_1 = e^{t}$ and $y_2 = e^{-t}$, both solutions of $y'' - y = 0$. State exactly what your answer certifies, and verify the result against Abel's theorem for this equation.

**Intuition:** For solutions of an ODE, one number at one time answers a question about all times.

**Solution:**

**Step 1 (compute).**

$$
W(t) = \det\begin{pmatrix} e^{t} & e^{-t} \\ e^{t} & -e^{-t} \end{pmatrix} = e^{t}(-e^{-t}) - e^{t}e^{-t} = -1 - 1 = -2
$$

**Step 2 (what it certifies).** $W(t) = -2 \neq 0$ at every $t$ (indeed at any single $t_0$, which suffices). By the Wronskian test, $\{e^{t}, e^{-t}\}$ is a **fundamental set** for $y'' - y = 0$: the pair is linearly independent, and *every* solution is $y = c_1 e^{t} + c_2 e^{-t}$ for unique constants. Equivalently, for any initial data the $2 \times 2$ system

$$
c_1 + c_2 = y_0, \qquad c_1 - c_2 = v_0
$$

has the unique solution $c_1 = \frac{y_0 + v_0}{2}$, $c_2 = \frac{y_0 - v_0}{2}$ (its matrix has determinant $W(0) = -2 \neq 0$).

**Step 3 (Abel check).** In standard form $y'' + p y' + q y = 0$ the equation has $p(t) = 0$. Abel's identity predicts

$$
W(t) = W(t_0)\, e^{-\int_{t_0}^{t} 0\, ds} = W(t_0)
$$

i.e. a *constant* Wronskian — consistent with the computed $W \equiv -2$. The dichotomy is visible: the Wronskian is nowhere zero.

$$
\boxed{W(e^{t}, e^{-t}) = -2 \ \text{ for all } t \ \Rightarrow \ \{e^{t}, e^{-t}\} \text{ is a fundamental set of } y'' - y = 0}
$$

> **Key Takeaway:** A nonzero Wronskian at a single point certifies independence and spanning on the whole interval — and equations with no $y'$ term ($p = 0$) have constant Wronskians by Abel.

### Problem L0.4: Where Superposition Is Valid — and Where It Silently Fails

**Problem Statement:** Let $y_1, y_2$ be solutions of an equation. Decide, with proof or counterexample, whether $y_1 + y_2$ is again a solution in each situation:

1. $y'' + 4y = 0$ (linear, homogeneous)
2. $y'' + 4y = 8$ (linear, non-homogeneous)
3. $y' = y^2$ (nonlinear)

**Intuition:** Superposition is a statement about the operator being linear *and* the right-hand side being zero.

**Solution:**

**Case 1 (linear homogeneous — YES).** With $L[y] = y'' + 4y$, linearity gives

$$
L[y_1 + y_2] = L[y_1] + L[y_2] = 0 + 0 = 0
$$

The sum (and any combination $c_1 y_1 + c_2 y_2$) solves. E.g. $\cos 2t + \sin 2t$ solves $y'' + 4y = 0$. ✔

**Case 2 (linear non-homogeneous — NO).** Same linearity, different bookkeeping:

$$
L[y_1 + y_2] = L[y_1] + L[y_2] = 8 + 8 = 16 \neq 8
$$

Concretely $y_1 = y_2 = 2$ are (constant) solutions of $y'' + 4y = 8$, but $y_1 + y_2 = 4$ gives $0 + 16 = 16$. The solution set is an *affine* plane $y_p + \ker L$, not a subspace — it does not contain $0$ and is not closed under addition. (What *is* true: the difference $y_1 - y_2$ of two non-homogeneous solutions is a homogeneous solution.)

**Case 3 (nonlinear — NO).** $y_1(t) = \frac{-1}{t}$ solves $y' = y^2$ (check: $y_1' = \frac{1}{t^2} = y_1^2$). Take $y_2 = y_1$, so $y_1 + y_2 = \frac{-2}{t}$. Then

$$
(y_1 + y_2)' = \frac{2}{t^2}, \qquad (y_1 + y_2)^2 = \frac{4}{t^2}
$$

and $\frac{2}{t^2} \neq \frac{4}{t^2}$: even doubling a solution fails. The map $y \mapsto y' - y^2$ is not linear, so no superposition of any kind.

$$
\boxed{\text{Superposition holds exactly for } L[y] = 0 \text{ with } L \text{ linear; it fails for } L[y] = g \neq 0 \text{ and for nonlinear equations}}
$$

> **Key Takeaway:** "Linear homogeneous" is precisely the condition making the solution set a vector space; non-homogeneous solution sets are affine translates, and nonlinear equations offer no additivity at all.

## Level 1 — Foundation

### Problem L1.1: Distinct Real Roots (Overdamped Regime)

**Problem Statement:** Solve the initial value problem

$$
y'' - 3y' + 2y = 0, \qquad y(0) = 1, \quad y'(0) = 0
$$

and determine the behavior as $t \to \infty$.

**Intuition:** Discriminant $9 - 8 = 1 \gt 0$: two real exponential modes, and the initial data decide their mixture.

**Solution:**

**Step 1 (characteristic equation).** The ansatz $y = e^{\lambda t}$ gives

$$
\lambda^2 - 3\lambda + 2 = (\lambda - 1)(\lambda - 2) = 0 \quad \Longrightarrow \quad \lambda_1 = 1, \ \lambda_2 = 2
$$

**Step 2 (general solution).** Distinct real roots give the fundamental set $\{e^{t}, e^{2t}\}$ (Wronskian $W = e^{t} \cdot 2e^{2t} - e^{t} \cdot e^{2t} = e^{3t} \neq 0$), so

$$
y = c_1 e^{t} + c_2 e^{2t}
$$

**Step 3 (fit initial conditions).** From $y(0) = 1$ and $y'(t) = c_1 e^{t} + 2 c_2 e^{2t}$ with $y'(0) = 0$:

$$
c_1 + c_2 = 1, \qquad c_1 + 2c_2 = 0
$$

Subtracting: $c_2 = -1$, hence $c_1 = 2$.

**Step 4 (behavior).** $y(t) = 2e^{t} - e^{2t} = e^{t}(2 - e^{t})$. The faster mode $e^{2t}$ dominates with a *negative* coefficient: $y$ rises from $1$? Check $y'(t) = 2e^{t} - 2e^{2t} = 2e^{t}(1 - e^{t}) \lt 0$ for $t \gt 0$ — the solution starts flat, decreases, crosses zero at $e^{t} = 2$, i.e. $t = \ln 2$, and then $y \to -\infty$. Both roots being positive, this equation is an *anti-damped* (unstable) system: generic solutions blow up at rate $e^{2t}$.

$$
\boxed{y(t) = 2e^{t} - e^{2t}, \qquad y \to -\infty \text{ at rate } e^{2t} \text{ as } t \to \infty}
$$

> **Key Takeaway:** With distinct real roots the solution is a mixture of two pure exponentials; long-time behavior is owned by the largest root whose coefficient is nonzero.

### Problem L1.2: Repeated Root (Critically Damped Regime)

**Problem Statement:** Solve

$$
y'' + 4y' + 4y = 0, \qquad y(0) = 1, \quad y'(0) = 0
$$

and show the solution never crosses zero for $t \gt 0$.

**Intuition:** Discriminant $16 - 16 = 0$: the two exponential modes have merged, and the second solution carries the factor $t$.

**Solution:**

**Step 1 (characteristic equation).**

$$
\lambda^2 + 4\lambda + 4 = (\lambda + 2)^2 = 0 \quad \Longrightarrow \quad \lambda = -2 \ \text{(double root)}
$$

**Step 2 (general solution).** By the reduction-of-order theorem for repeated roots, the fundamental set is $\{e^{-2t}, t e^{-2t}\}$:

$$
y = (c_1 + c_2 t)\, e^{-2t}
$$

**Step 3 (fit initial conditions).** $y(0) = c_1 = 1$. Differentiate:

$$
y' = c_2 e^{-2t} - 2(c_1 + c_2 t) e^{-2t} = \bigl(c_2 - 2c_1 - 2c_2 t\bigr) e^{-2t}
$$

so $y'(0) = c_2 - 2c_1 = 0$ gives $c_2 = 2$.

**Step 4 (no zero crossing).** The solution is $y(t) = (1 + 2t) e^{-2t}$. For $t \gt 0$ both factors are strictly positive, so $y(t) \gt 0$: released from rest, a critically damped system glides monotonically to equilibrium after at most one hump — here $y' = -4t e^{-2t} \le 0$, so it decays monotonically from $1$ without ever crossing zero. The polynomial factor $1 + 2t$ grows only linearly and is crushed by $e^{-2t}$; $y \to 0$ as $t \to \infty$.

$$
\boxed{y(t) = (1 + 2t)\, e^{-2t} \gt 0 \ \text{ for all } t \ge 0}
$$

> **Key Takeaway:** Critical damping produces $(c_1 + c_2 t)e^{\lambda t}$: at most one interior extremum or zero, and the fastest non-oscillatory return to equilibrium — the design point of door closers and car suspensions.

### Problem L1.3: Complex Roots (Underdamped Regime)

**Problem Statement:** Solve

$$
y'' + 2y' + 5y = 0, \qquad y(0) = 0, \quad y'(0) = 2
$$

Identify the pseudo-frequency and the decay envelope, and sketch (verbally) the phase portrait.

**Intuition:** Discriminant $4 - 20 = -16 \lt 0$: expect a decaying oscillation, built from complex roots via Euler's formula.

**Solution:**

**Step 1 (characteristic equation).**

$$
\lambda^2 + 2\lambda + 5 = 0 \quad \Longrightarrow \quad \lambda = \frac{-2 \pm \sqrt{4 - 20}}{2} = -1 \pm 2i
$$

**Step 2 (real fundamental set).** With $\alpha = -1$, $\beta = 2$, Euler's formula turns $e^{(-1 \pm 2i)t}$ into the real pair $e^{-t}\cos 2t$, $e^{-t}\sin 2t$:

$$
y = e^{-t}\bigl(c_1 \cos 2t + c_2 \sin 2t\bigr)
$$

**Step 3 (fit initial conditions).** $y(0) = c_1 = 0$, so $y = c_2 e^{-t}\sin 2t$. Then

$$
y' = c_2\bigl(-e^{-t}\sin 2t + 2e^{-t}\cos 2t\bigr), \qquad y'(0) = 2c_2 = 2 \ \Longrightarrow \ c_2 = 1
$$

**Step 4 (read off the physics).** The solution $y = e^{-t}\sin 2t$ oscillates at pseudo-frequency $\omega_d = 2$ rad per unit time (pseudo-period $\pi$), inside the exponentially shrinking envelope $\pm e^{-t}$. Successive maxima shrink by the factor $e^{-\pi} \approx 0.043$ per period (logarithmic decrement $\pi$). In the $(y, y')$ phase plane the trajectory is a clockwise spiral into the origin — a stable spiral point, the fingerprint of complex roots with negative real part.

$$
\boxed{y(t) = e^{-t}\sin 2t: \ \text{decaying oscillation, envelope } e^{-t}, \ \omega_d = 2}
$$

> **Key Takeaway:** Complex roots $\alpha \pm i\beta$ mean oscillation at frequency $\beta$ inside the envelope $e^{\alpha t}$; the real part damps, the imaginary part rotates.

### Problem L1.4: Undetermined Coefficients with Resonance

**Problem Statement:** Find the general solution of

$$
y'' + 9y = 2\cos 3t
$$

Explain why the naive trial solution fails and apply the correct modification.

**Intuition:** The forcing frequency $3$ equals the natural frequency $\sqrt{9}$ — the forcing lives inside the null space of the operator.

**Solution:**

**Step 1 (homogeneous part).** $\lambda^2 + 9 = 0$ gives $\lambda = \pm 3i$, so $y_h = c_1 \cos 3t + c_2 \sin 3t$.

**Step 2 (naive trial fails).** Try $y_p = A\cos 3t + B \sin 3t$. But then $y_p$ is itself a homogeneous solution: $y_p'' + 9y_p = 0$ identically, and $0 = 2\cos 3t$ is impossible. The trial is annihilated because $\pm 3i$ are characteristic roots — this *is* resonance.

**Step 3 (modified trial).** Multiply by $t$: $y_p = t(A\cos 3t + B\sin 3t)$. Differentiate:

$$
y_p' = (A\cos 3t + B\sin 3t) + t(-3A\sin 3t + 3B\cos 3t)
$$

$$
y_p'' = 2(-3A\sin 3t + 3B\cos 3t) + t(-9A\cos 3t - 9B\sin 3t)
$$

**Step 4 (substitute and match).** In $y_p'' + 9 y_p$ the $t$-proportional terms cancel ($-9 t (A\cos 3t + B \sin 3t) + 9t(A \cos 3t + B\sin 3t) = 0$), leaving

$$
y_p'' + 9y_p = -6A\sin 3t + 6B\cos 3t \stackrel{!}{=} 2\cos 3t
$$

Matching: $A = 0$, $6B = 2$, so $B = \frac{1}{3}$ and $y_p = \frac{t}{3}\sin 3t$. (Cross-check with the general resonance formula $\frac{F_0 t}{2\omega}\sin\omega t$: here $F_0 = 2$, $\omega = 3$, giving $\frac{2t}{6}\sin 3t$. ✔)

**Step 5 (general solution and growth).**

$$
\boxed{y = c_1 \cos 3t + c_2 \sin 3t + \frac{t}{3} \sin 3t}
$$

The particular term oscillates inside the linearly growing envelope $\pm t/3$: bounded homogeneous motion plus secular growth — the mathematical signature of driving an undamped system exactly at its natural frequency.

> **Key Takeaway:** When the forcing solves the homogeneous equation, multiply the trial by $t$; the matched solution then grows linearly — resonance converts bounded forcing into unbounded response.

### Problem L1.5: Variation of Parameters for $y'' + y = \tan t$

**Problem Statement:** Find the general solution of

$$
y'' + y = \tan t, \qquad -\frac{\pi}{2} \lt t \lt \frac{\pi}{2}
$$

Explain why undetermined coefficients cannot work here.

**Intuition:** $\tan t$ is not a polynomial-exponential-sinusoid: differentiating it never closes into a finite family, so no finite trial ansatz exists. Variation of parameters has no such restriction.

**Solution:**

**Step 1 (setup).** Homogeneous solutions of $y'' + y = 0$: $y_1 = \cos t$, $y_2 = \sin t$, with Wronskian

$$
W = \cos t \cdot \cos t - (-\sin t)\sin t = \cos^2 t + \sin^2 t = 1
$$

Undetermined coefficients fails because repeated differentiation of $\tan t$ generates $\sec^2 t$, $\sec^2 t \tan t, \ldots$ — an infinite ladder, never a finite-dimensional trial space.

**Step 2 (the variation-of-parameters system).** With $f(t) = \tan t$:

$$
u_1' = \frac{-y_2 f}{W} = -\sin t \tan t, \qquad u_2' = \frac{y_1 f}{W} = \cos t \tan t = \sin t
$$

**Step 3 (integrate $u_2$).** $u_2 = \int \sin t\, dt = -\cos t$.

**Step 4 (integrate $u_1$).** Rewrite using $\sin^2 t = 1 - \cos^2 t$:

$$
u_1 = -\int \frac{\sin^2 t}{\cos t}\, dt = -\int \frac{1 - \cos^2 t}{\cos t}\, dt = \int \bigl(\cos t - \sec t\bigr)\, dt = \sin t - \ln\lvert \sec t + \tan t \rvert
$$

**Step 5 (assemble).**

$$
y_p = u_1 y_1 + u_2 y_2 = \bigl(\sin t - \ln\lvert \sec t + \tan t\rvert\bigr)\cos t - \cos t \sin t = -\cos t \, \ln\lvert \sec t + \tan t \rvert
$$

The $\sin t \cos t$ terms cancel. On $(-\pi/2, \pi/2)$, $\sec t + \tan t \gt 0$, so the absolute value may be dropped.

$$
\boxed{y = c_1 \cos t + c_2 \sin t - \cos t \, \ln( \sec t + \tan t )}
$$

**Check (spot).** At $t = 0$: $y_p = -1 \cdot \ln 1 = 0$, and one verifies $y_p'' + y_p = \tan t$ by direct differentiation.

> **Key Takeaway:** Variation of parameters is the universal fallback: two integrals against the Wronskian solve any continuous forcing, including forcings like $\tan t$ that undetermined coefficients can never reach.

### Problem L1.6: A Cauchy–Euler Equation

**Problem Statement:** Solve for $t \gt 0$:

$$
t^2 y'' - t y' - 3y = 0
$$

both (a) by the direct power ansatz $y = t^r$ and (b) by the substitution $t = e^s$, confirming the answers agree.

**Intuition:** The coefficient of $y^{(k)}$ is $t^k$ times a constant, so each term of $t^r$ lands back on $t^r$ — powers replace exponentials.

**Solution:**

**(a) Power ansatz.** With $y = t^r$: $y' = r t^{r-1}$, $y'' = r(r-1) t^{r-2}$. Substitute:

$$
t^2 \cdot r(r-1)t^{r-2} - t \cdot r t^{r-1} - 3 t^r = \bigl(r(r-1) - r - 3\bigr) t^r = 0
$$

The **indicial equation** is

$$
r^2 - 2r - 3 = (r - 3)(r + 1) = 0 \quad \Longrightarrow \quad r = 3, \ r = -1
$$

giving the general solution $y = c_1 t^3 + c_2 t^{-1}$. Independence: $W(t^3, t^{-1}) = t^3(-t^{-2}) - 3t^2 \cdot t^{-1} = -4t \neq 0$ on $t \gt 0$.

**(b) Exponential substitution.** Let $t = e^s$, $Y(s) = y(e^s)$, and write $\dot{Y} = dY/ds$. Chain rule: $\dot{Y} = y' t$, and

$$
\ddot{Y} = \frac{d}{ds}(y' t) = y'' t \cdot t + y' t = t^2 y'' + t y' \quad \Longrightarrow \quad t^2 y'' = \ddot{Y} - \dot{Y}
$$

Substituting into the equation:

$$
(\ddot{Y} - \dot{Y}) - \dot{Y} - 3Y = 0 \quad \Longrightarrow \quad \ddot{Y} - 2\dot{Y} - 3Y = 0
$$

Constant coefficients: $\lambda^2 - 2\lambda - 3 = (\lambda - 3)(\lambda + 1) = 0$, so $Y = c_1 e^{3s} + c_2 e^{-s}$. Translating back with $e^{s} = t$: $Y = c_1 t^3 + c_2 t^{-1}$. Identical. Note the indicial and characteristic equations coincide — always, since $t^r = e^{rs}$.

$$
\boxed{y(t) = c_1 t^3 + \frac{c_2}{t}, \qquad t \gt 0}
$$

**Remark.** Repeated indicial root $r$ would give $\{t^r, t^r \ln t\}$ (the image of $\{e^{rs}, s e^{rs}\}$), and complex roots $\alpha \pm i\beta$ give $t^{\alpha}\cos(\beta \ln t)$ and $t^{\alpha}\sin(\beta \ln t)$ — oscillations in $\ln t$.

> **Key Takeaway:** Cauchy–Euler equations are constant-coefficient equations wearing a logarithmic clock: $t = e^s$ converts them exactly, and power solutions $t^r$ mirror exponential modes $e^{rs}$.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: RLC Circuit — Ringing Frequency and Quality Factor

**Problem Statement:** A series RLC circuit has $L = 10\ \text{mH}$, $C = 1\ \mu\text{F}$, $R = 4\ \Omega$. The capacitor is charged and the circuit closed at $t = 0$ with no source. (a) Write the ODE for the charge $q(t)$. (b) Compute the undamped resonant frequency $\omega_0$ and $f_0$ in Hz. (c) Compute the damping ratio $\zeta$ and quality factor $Q$; classify the response. (d) Compute the actual ringing frequency $\omega_d$ and the time for the oscillation envelope to decay to $1\%$.

**Intuition:** This is a mass–spring–damper with $L \leftrightarrow m$, $R \leftrightarrow c$, $1/C \leftrightarrow k$.

**Solution:**

**(a) The ODE.** Kirchhoff's voltage law with no source:

$$
L q'' + R q' + \frac{1}{C} q = 0 \quad \Longrightarrow \quad 0.01\, q'' + 4\, q' + 10^{6} q = 0
$$

**(b) Natural frequency.**

$$
\omega_0 = \frac{1}{\sqrt{LC}} = \frac{1}{\sqrt{10^{-2} \cdot 10^{-6}}} = \frac{1}{\sqrt{10^{-8}}} = 10^{4}\ \text{rad/s}, \qquad f_0 = \frac{\omega_0}{2\pi} \approx 1592\ \text{Hz}
$$

**(c) Damping ratio and Q.** Matching $L q'' + R q' + q/C = 0$ to $m x'' + c x' + k x = 0$:

$$
\zeta = \frac{R}{2}\sqrt{\frac{C}{L}} = \frac{4}{2}\sqrt{\frac{10^{-6}}{10^{-2}}} = 2 \times 10^{-2} = 0.02, \qquad Q = \frac{1}{2\zeta} = \frac{1}{R}\sqrt{\frac{L}{C}} = \frac{1}{4}\sqrt{10^{4}} = 25
$$

Since $\zeta = 0.02 \lt 1$, the response is strongly **underdamped**: the charge rings roughly $Q$ times ($\approx 25$ visible oscillations) before dying away.

**(d) Ringing frequency and decay.** The characteristic roots are $\lambda = -\zeta\omega_0 \pm i \omega_0\sqrt{1 - \zeta^2}$, so

$$
\omega_d = \omega_0 \sqrt{1 - \zeta^2} = 10^4 \sqrt{1 - 0.0004} \approx 0.9998 \times 10^4\ \text{rad/s}
$$

— within $0.02\%$ of $\omega_0$: for high-$Q$ circuits the ringing frequency is essentially the resonant frequency. The envelope is $e^{-\zeta \omega_0 t} = e^{-200 t}$ (decay rate $\zeta\omega_0 = R/2L = 200\ \text{s}^{-1}$). It reaches $1\%$ when

$$
e^{-200 t} = 0.01 \quad \Longrightarrow \quad t = \frac{\ln 100}{200} \approx 23\ \text{ms}
$$

$$
\boxed{\omega_0 = 10^4\ \text{rad/s} \ (f_0 \approx 1.59\ \text{kHz}), \quad \zeta = 0.02, \quad Q = 25, \quad \omega_d \approx 0.9998\,\omega_0, \quad t_{1\%} \approx 23\ \text{ms}}
$$

> **Key Takeaway:** For a series RLC, $Q = \frac{1}{R}\sqrt{L/C}$ counts the ring-down oscillations and sets the resonance sharpness; high-$Q$ circuits ring at essentially $\omega_0$ with envelope decay rate $R/2L$.

### Problem L2.2: Car Suspension — Designing for Critical Damping

**Problem Statement:** A quarter-car model has sprung mass $m = 400\ \text{kg}$ and spring stiffness $k = 2.5 \times 10^{4}\ \text{N/m}$. (a) Find the undamped natural frequency in rad/s and Hz. (b) Choose the shock-absorber coefficient $c$ that gives critical damping. (c) If worn shocks deliver only $c = 2000\ \text{N·s/m}$, find the damping ratio, the bounce frequency the passengers feel, and the number of visible oscillations after a bump (envelope to $5\%$).

**Intuition:** Critical damping is the sweet spot: return to level ride height as fast as possible without bouncing.

**Solution:**

**(a) Natural frequency.**

$$
\omega_0 = \sqrt{\frac{k}{m}} = \sqrt{\frac{2.5 \times 10^4}{400}} = \sqrt{62.5} \approx 7.91\ \text{rad/s}, \qquad f_0 = \frac{\omega_0}{2\pi} \approx 1.26\ \text{Hz}
$$

(Comfort engineering aims for ride frequencies near $1$–$1.5$ Hz — this is a realistic setup.)

**(b) Critical damping.** The regime boundary is $c^2 = 4mk$, i.e.

$$
c_{\text{crit}} = 2\sqrt{mk} = 2\sqrt{400 \times 2.5\times 10^4} = 2\sqrt{10^{7}} \approx 6325\ \text{N·s/m}
$$

At this value the response to a bump is $(c_1 + c_2 t)e^{-\omega_0 t}$: fastest settling with zero overshoot. (Real cars target $\zeta \approx 0.3$–$0.7$, trading a little overshoot for compliance over rough roads, but $c_{\text{crit}}$ is the design anchor.)

**(c) Worn shocks.** With $c = 2000$:

$$
\zeta = \frac{c}{c_{\text{crit}}} = \frac{2000}{6325} \approx 0.316
$$

Underdamped. The felt bounce frequency is

$$
\omega_d = \omega_0\sqrt{1 - \zeta^2} = 7.91\sqrt{1 - 0.1} \approx 7.50\ \text{rad/s} \approx 1.19\ \text{Hz}
$$

Envelope decay rate $\zeta\omega_0 = 2.5\ \text{s}^{-1}$: it reaches $5\%$ when $e^{-2.5 t} = 0.05$, i.e. $t = \ln 20 / 2.5 \approx 1.2\ \text{s}$. The pseudo-period is $2\pi/\omega_d \approx 0.84\ \text{s}$, so the car completes roughly

$$
\frac{1.2}{0.84} \approx 1.4 \ \text{oscillations}
$$

— one full visible bounce and a fading second, the classic feel of tired shocks.

$$
\boxed{c_{\text{crit}} = 2\sqrt{mk} \approx 6.3 \times 10^{3}\ \text{N·s/m}; \ \text{worn: } \zeta \approx 0.32, \ f_d \approx 1.19\ \text{Hz}, \ \approx 1.4 \text{ bounces}}
$$

> **Key Takeaway:** The critical coefficient $2\sqrt{mk}$ is the fastest bounce-free design; the damping ratio $c / 2\sqrt{mk}$ instantly diagnoses ride quality from three numbers.

### Problem L2.3: Forced Building Sway — Resonant Amplification

**Problem Statement:** Model a building's fundamental sway mode as $m x'' + c x' + k x = F_0 \cos\omega t$ with modal mass $m = 10^{6}\ \text{kg}$, stiffness $k = 4 \times 10^{7}\ \text{N/m}$, damping ratio $\zeta = 0.01$, and a periodic wind-vortex force of amplitude $F_0 = 10^{5}\ \text{N}$. (a) Find the natural frequency and period. (b) Find the static deflection $F_0/k$ and the steady-state amplitude when the vortex-shedding frequency locks onto $\omega = \omega_0$. (c) Explain in one paragraph why "shedding lock-in at $\omega_0$ plus low $\zeta$" is the Tacoma-Narrows-style danger recipe, and compute the amplitude if dampers raise $\zeta$ to $0.05$.

**Intuition:** Resonant amplification multiplies the static response by $Q = 1/(2\zeta)$ — small forces become large motions when damping is tiny.

**Solution:**

**(a) Natural frequency.**

$$
\omega_0 = \sqrt{\frac{k}{m}} = \sqrt{\frac{4\times 10^7}{10^6}} = \sqrt{40} \approx 6.32\ \text{rad/s}, \qquad T_0 = \frac{2\pi}{\omega_0} \approx 0.99\ \text{s}
$$

**(b) Amplitudes.** Static deflection:

$$
x_{\text{st}} = \frac{F_0}{k} = \frac{10^5}{4\times 10^7} = 2.5\ \text{mm}
$$

The steady-state amplitude of $m x'' + c x' + kx = F_0\cos\omega t$ is

$$
R(\omega) = \frac{F_0/m}{\sqrt{(\omega_0^2 - \omega^2)^2 + (2\zeta\omega_0\omega)^2}}
$$

At $\omega = \omega_0$ the first term under the root vanishes:

$$
R(\omega_0) = \frac{F_0/m}{2\zeta\omega_0^2} = \frac{F_0}{k}\cdot\frac{1}{2\zeta} = Q\, x_{\text{st}} = 50 \times 2.5\ \text{mm} = 125\ \text{mm}
$$

A $2.5$ mm static push becomes a $12.5$ cm sway — amplification by $Q = 1/(2\zeta) = 50$.

**(c) The danger recipe and the fix.** Vortex shedding behind a bluff structure produces a nearly periodic lateral force whose frequency scales with wind speed; when it approaches $\omega_0$ it can *lock in* to the structure's motion, holding $\omega \approx \omega_0$ over a range of wind speeds. The structure then sits at the top of the resonance curve, where response is $Q$ times static, energy input per cycle exceeds dissipation until amplitude is large, and (in the historical Tacoma Narrows collapse) large motion triggered self-excited aeroelastic flutter — effectively negative damping. Low structural $\zeta$ is what makes the peak tall. Raising damping to $\zeta = 0.05$ (tuned mass dampers, added dissipation) cuts $Q$ to $10$:

$$
R = 10 \times 2.5\ \text{mm} = 25\ \text{mm}
$$

— a fivefold reduction with no change to mass or stiffness.

$$
\boxed{R(\omega_0) = \frac{F_0}{k}\cdot\frac{1}{2\zeta}: \ \ 125\ \text{mm at } \zeta = 0.01, \ \ 25\ \text{mm at } \zeta = 0.05}
$$

> **Key Takeaway:** At resonance the steady amplitude is the static deflection times $Q = 1/(2\zeta)$: damping, not strength, is the lever that controls resonant response.

### Problem L2.4: Heavy-Ball Momentum — Critical Damping in Optimization

**Problem Statement:** Near a minimum, momentum gradient descent follows the heavy-ball ODE $\theta'' + \gamma\, \theta' = -\nabla\mathcal{L}(\theta)$ (unit mass). Suppose the loss is locally quadratic with Hessian eigenvalues $\mu_{\min} = 4$ and $\mu_{\max} = 100$. (a) Write the decoupled modal equations along the Hessian eigenvectors. (b) Choose the friction $\gamma$ that critically damps the *flattest* mode, and classify the resulting dynamics of the sharpest mode. (c) Explain why choosing $\gamma$ to critically damp the sharpest mode instead would be a poor idea, and connect $\gamma$ to the discrete momentum parameter $\beta \approx 1 - \gamma \sqrt{\eta}$ qualitatively.

**Intuition:** Training dynamics near a minimum is a bank of independent spring–mass–dampers, one per Hessian eigenvalue; one knob $\gamma$ must serve them all.

**Solution:**

**(a) Modal decomposition.** Write $H = U\Lambda U^{T}$ and $z = U^{T}(\theta - \theta^{*})$. Since $\nabla\mathcal{L} \approx H(\theta - \theta^{*})$, the ODE splits into independent scalar oscillators

$$
z_i'' + \gamma z_i' + \mu_i z_i = 0, \qquad \mu_i \in \{4, \ldots, 100\}
$$

Each mode has natural frequency $\sqrt{\mu_i}$ and damping ratio $\zeta_i = \dfrac{\gamma}{2\sqrt{\mu_i}}$.

**(b) Critically damp the flattest mode.** Critical damping of mode $\mu_{\min} = 4$ requires

$$
\gamma = 2\sqrt{\mu_{\min}} = 2\sqrt{4} = 4
$$

Then the flattest mode decays like $(c_1 + c_2 t)e^{-2t}$ — the fastest possible non-oscillatory convergence at rate $\sqrt{\mu_{\min}} = 2$. The sharpest mode has

$$
\zeta_{\max} = \frac{\gamma}{2\sqrt{\mu_{\max}}} = \frac{4}{2 \times 10} = 0.2 \lt 1
$$

so it is **underdamped**: it oscillates at pseudo-frequency $\sqrt{\mu_{\max}}\sqrt{1 - 0.04} \approx 9.8$ while decaying at rate $\gamma/2 = 2$ — this is exactly the ringing seen in loss curves along sharp directions, and it is harmless because its decay rate ($2$) matches the slow mode's.

**(c) Why not damp the sharp mode?** Critical damping of $\mu_{\max}$ would need $\gamma = 2\sqrt{100} = 20$, making the flattest mode's damping ratio $\zeta_{\min} = 20/(2\cdot 2) = 5$: heavily **overdamped**. Its slow root is

$$
\lambda_{\text{slow}} = \frac{-\gamma + \sqrt{\gamma^2 - 4\mu_{\min}}}{2} = \frac{-20 + \sqrt{400 - 16}}{2} \approx -0.20
$$

— convergence rate $0.20$ instead of $2$: **ten times slower**. Overall convergence is governed by the slowest mode, so the correct policy is to critically damp the flattest direction and tolerate ringing in the sharp ones. In discrete momentum GD with step $\eta$, $\beta \approx 1 - \gamma\sqrt{\eta}$: larger $\gamma$ means *smaller* momentum $\beta$; the classical optimum $\beta^{*} = \bigl(\frac{\sqrt{\kappa} - 1}{\sqrt{\kappa} + 1}\bigr)^2$ with $\kappa = \mu_{\max}/\mu_{\min} = 25$ (here $\beta^{*} = (4/6)^2 = 4/9 \approx 0.44$) is precisely the discrete shadow of critical damping of the flattest mode.

$$
\boxed{\gamma^{*} = 2\sqrt{\mu_{\min}} = 4: \ \text{flattest mode critically damped, sharp modes ring harmlessly at } \zeta = 0.2}
$$

> **Key Takeaway:** Optimal momentum is damping design: critically damp the smallest Hessian eigenvalue, since the slowest mode sets the convergence rate — over-damping to kill visible oscillations slows training dramatically.

### Problem L2.5: Seismometer vs Accelerometer — One Equation, Two Instruments

**Problem Statement:** A proof mass $m$ hangs on a spring $k$ and damper $c$ inside a case attached to the ground. If the ground moves as $u(t) = U\cos\omega t$, the mass's displacement *relative to the case*, $z$, obeys

$$
z'' + 2\zeta\omega_n z' + \omega_n^2 z = -u''(t) = \omega^2 U \cos\omega t
$$

with $\omega_n = \sqrt{k/m}$. (a) Derive the steady-state amplitude $Z(\omega)$. (b) Show that for $\omega \gg \omega_n$ the device reads ground *displacement* (seismometer) and for $\omega \ll \omega_n$ it reads ground *acceleration* (accelerometer). (c) A MEMS accelerometer has $\omega_n = 2\pi \times 5\ \text{kHz}$. What relative amplitude does it produce for a $50\ \text{Hz}$, $0.5\ \text{mm}$ vibration, and what acceleration does that correspond to?

**Intuition:** The same forced-oscillator amplitude curve, read at opposite ends, yields two different sensors.

**Solution:**

**(a) Steady-state amplitude.** This is a forced oscillator with forcing amplitude $\omega^2 U$. The standard steady-state response $z_p = Z\cos(\omega t - \phi)$ has

$$
Z(\omega) = \frac{\omega^2 U}{\sqrt{(\omega_n^2 - \omega^2)^2 + (2\zeta\omega_n\omega)^2}}, \qquad \tan\phi = \frac{2\zeta\omega_n\omega}{\omega_n^2 - \omega^2}
$$

**(b) Two asymptotic regimes.**

*Seismometer regime, $\omega \gg \omega_n$:* the $\omega^4$ term dominates the denominator, so

$$
Z \approx \frac{\omega^2 U}{\omega^2} = U
$$

The proof mass stays inertially still while the case moves around it; the relative displacement equals the ground displacement. Seismometers are built *soft and heavy* (low $\omega_n$, often below $0.1$ Hz) so that earthquake frequencies satisfy $\omega \gg \omega_n$.

*Accelerometer regime, $\omega \ll \omega_n$:* now $\omega_n^4$ dominates, so

$$
Z \approx \frac{\omega^2 U}{\omega_n^2} = \frac{a_{\max}}{\omega_n^2}
$$

where $a_{\max} = \omega^2 U$ is the peak ground acceleration: the reading is acceleration divided by the fixed constant $\omega_n^2$. Accelerometers are built *stiff and light* (high $\omega_n$) so the signal band sits far below resonance; sensitivity trades against bandwidth as $1/\omega_n^2$.

**(c) MEMS numbers.** $\omega = 2\pi \times 50 = 314.2\ \text{rad/s}$, $\omega_n = 2\pi \times 5000 = 3.14 \times 10^4\ \text{rad/s}$, ratio $\omega/\omega_n = 0.01$, so the accelerometer approximation is excellent. Peak acceleration:

$$
a_{\max} = \omega^2 U = (314.2)^2 \times 5\times 10^{-4} \approx 49.3\ \text{m/s}^2 \approx 5g
$$

Proof-mass deflection:

$$
Z = \frac{a_{\max}}{\omega_n^2} = \frac{49.3}{(3.14\times 10^4)^2} \approx 5 \times 10^{-8}\ \text{m} = 50\ \text{nm}
$$

— why MEMS devices read capacitance changes rather than visible motion.

$$
\boxed{Z(\omega) = \frac{\omega^2 U}{\sqrt{(\omega_n^2 - \omega^2)^2 + (2\zeta\omega_n\omega)^2}}: \ \ Z \to U \ (\omega \gg \omega_n), \quad Z \to \frac{\omega^2 U}{\omega_n^2} \ (\omega \ll \omega_n)}
$$

> **Key Takeaway:** One steady-state amplitude formula, evaluated far above or far below resonance, explains both seismometers (soft, reads displacement) and accelerometers (stiff, reads acceleration over $\omega_n^2$).

### Problem L2.6: Pendulum Clock — Linearization, Design, and Its Error

**Problem Statement:** (a) Linearize the pendulum equation $\theta'' + \frac{g}{\ell}\sin\theta = 0$ for small angles and find the period. (b) Compute the length $\ell$ of a "seconds pendulum" (half-period exactly $1$ s, i.e. $T = 2$ s) using $g = 9.81\ \text{m/s}^2$. (c) The next-order correction gives true period $T(\theta_0) \approx T\bigl(1 + \frac{\theta_0^2}{16}\bigr)$ for swing amplitude $\theta_0$. How many seconds per day does the clock lose if it is regulated at tiny amplitude but actually swings at $\theta_0 = 5^{\circ}$?

**Intuition:** SHM's amplitude-independent period is what makes a pendulum a clock — and the cubic term in $\sin\theta$ is the clock's error budget.

**Solution:**

**(a) Linearization.** Taylor: $\sin\theta = \theta - \frac{\theta^3}{6} + O(\theta^5)$. Dropping the cubic term for $\lvert\theta\rvert \ll 1$:

$$
\theta'' + \frac{g}{\ell}\,\theta = 0
$$

— SHM with $\omega_0 = \sqrt{g/\ell}$ and amplitude-independent period

$$
T = 2\pi\sqrt{\frac{\ell}{g}}
$$

**(b) The seconds pendulum.** Solve $T = 2$ for $\ell$:

$$
\ell = \frac{g T^2}{4\pi^2} = \frac{9.81 \times 4}{4\pi^2} = \frac{9.81}{\pi^2} \approx 0.994\ \text{m}
$$

— just under one meter (historically proposed as the definition of the meter).

**(c) Amplitude error.** $\theta_0 = 5^{\circ} = 5 \times \frac{\pi}{180} \approx 0.0873$ rad. Fractional period increase:

$$
\frac{\Delta T}{T} = \frac{\theta_0^2}{16} = \frac{(0.0873)^2}{16} \approx \frac{0.00762}{16} \approx 4.76 \times 10^{-4}
$$

Each real period is longer than the regulated one, so the clock runs *slow* by that fraction of a day:

$$
86400 \times 4.76\times 10^{-4} \approx 41\ \text{s per day}
$$

Precision clocks therefore either kept amplitudes tiny and constant (escapement design) or compensated the arc. The general lesson: linearization near equilibrium is excellent (error quadratic in amplitude appears only as $\theta_0^2/16$ in the period) but a clock integrates that tiny error 43,200 times a day.

$$
\boxed{T = 2\pi\sqrt{\ell/g}, \quad \ell_{\text{seconds}} = \frac{g}{\pi^2} \approx 0.994\ \text{m}, \quad \text{5}^{\circ}\text{ swing} \Rightarrow \approx 41\ \text{s/day slow}}
$$

> **Key Takeaway:** Linearization turns the nonlinear pendulum into exact SHM with amplitude-independent period; the neglected cubic term reappears as a quadratic-in-amplitude period error that integrated systems (clocks) cannot ignore.

## Level 3 — Challenge

### Problem L3.1: Negative Potential — At Most One Zero

**Problem Statement:** Let $q$ be continuous with $q(t) \lt 0$ on an interval $I$, and let $y$ be a nontrivial solution of

$$
y'' + q(t)\, y = 0
$$

Prove that $y$ has **at most one** zero in $I$.

**Intuition:** With $q \lt 0$ the equation reads $y'' = (-q) y$ with $-q \gt 0$: the solution curves *away* from the axis wherever it is off the axis — think $y'' = y$ with solutions $e^{\pm t}$, which cross zero at most once.

**Solution:**

**Step 1 (zeros are simple).** Suppose $y(t_1) = 0$. If also $y'(t_1) = 0$, then $y$ and the zero function solve the same IVP at $t_1$; by uniqueness $y \equiv 0$, contradicting nontriviality. Hence at every zero, $y' \neq 0$: zeros are simple crossings, and in particular zeros are isolated.

**Step 2 (setup for contradiction).** Suppose $y$ has two zeros $t_1 \lt t_2$ in $I$; take them *consecutive* (possible since zeros are isolated), so $y$ has a fixed sign on $(t_1, t_2)$. Replacing $y$ by $-y$ if necessary (still a solution), assume

$$
y(t) \gt 0 \quad \text{on } (t_1, t_2), \qquad y(t_1) = y(t_2) = 0
$$

**Step 3 (convexity).** On $(t_1, t_2)$:

$$
y'' = -q(t)\, y = \underbrace{(-q(t))}_{\gt 0} \cdot \underbrace{y}_{\gt 0} \gt 0
$$

so $y$ is strictly convex on $(t_1, t_2)$ and $y''$ is continuous up to the endpoints with $y'' \ge 0$ on $[t_1, t_2]$.

**Step 4 (convexity contradicts positivity between two zeros).** A convex function lies on or below every chord of its graph. Applying this to the chord joining $(t_1, 0)$ and $(t_2, 0)$: for $t = (1-s) t_1 + s t_2$ with $s \in (0,1)$,

$$
y(t) \le (1 - s)\, y(t_1) + s\, y(t_2) = 0
$$

This contradicts $y(t) \gt 0$ on the open interval. (Alternative endgame: $y'(t_1) \gt 0$ since $y$ enters the positive region, and $y'' \gt 0$ makes $y'$ increasing, so $y' \gt 0$ throughout — then $y(t_2) \gt y(t_1) = 0$, contradicting $y(t_2) = 0$.)

**Step 5 (conclusion).** No two zeros can exist.

$$
\boxed{q(t) \lt 0 \text{ on } I \ \Longrightarrow \ \text{every nontrivial solution of } y'' + q y = 0 \text{ has at most one zero in } I}
$$

> **Key Takeaway:** The sign of $q$ decides oscillation: $q \lt 0$ makes solutions convex away from the axis (no oscillation possible), while $q \gt 0$ bends them back toward it — the doorway to Sturm's theory in the next problem.

### Problem L3.2: A Sturm Comparison Bound on Zero Spacing

**Problem Statement:** Let $q$ be continuous with $q(t) \ge m^2 \gt 0$ for all $t$. Prove that every nontrivial solution of

$$
y'' + q(t)\, y = 0
$$

has at least one zero in every closed interval of length $\pi/m$. (Consequently, consecutive zeros are at most $\pi/m$ apart, and every solution has infinitely many zeros on $[0, \infty)$.)

**Intuition:** Compare with $u'' + m^2 u = 0$, whose solution $\sin(m(t-a))$ has zeros exactly $\pi/m$ apart; stronger restoring force ($q \ge m^2$) should oscillate *faster*.

**Solution:**

**Step 1 (setup).** Fix an interval $[a, b]$ with $b - a = \pi/m$ and suppose, for contradiction, that a nontrivial solution $y$ has **no** zero on $[a, b]$. WLOG $y \gt 0$ on $[a, b]$ (else replace $y$ by $-y$). Let

$$
u(t) = \sin\bigl(m(t - a)\bigr), \qquad u'' + m^2 u = 0, \qquad u(a) = u(b) = 0, \quad u \gt 0 \ \text{on } (a, b)
$$

**Step 2 (the comparison Wronskian).** Define $V(t) = u(t) y'(t) - u'(t) y(t)$. Differentiate and use both ODEs:

$$
V' = u y'' - u'' y = u(-q y) - (-m^2 u) y = (m^2 - q(t))\, u\, y
$$

On $(a, b)$: $u \gt 0$, $y \gt 0$, and $m^2 - q(t) \le 0$, hence

$$
V'(t) \le 0 \quad \text{on } (a, b) \qquad \text{— } V \text{ is non-increasing.}
$$

**Step 3 (evaluate $V$ at the endpoints).** Using $u(a) = u(b) = 0$, $u'(a) = m\cos 0 = m$, $u'(b) = m\cos\pi = -m$:

$$
V(a) = -u'(a)\, y(a) = -m\, y(a) \lt 0
$$

$$
V(b) = -u'(b)\, y(b) = m\, y(b) \gt 0
$$

(both use $y \gt 0$ at the endpoints, part of the no-zero assumption on the *closed* interval).

**Step 4 (contradiction).** $V$ is non-increasing on $[a, b]$ yet $V(b) \gt 0 \gt V(a)$ — impossible. Hence $y$ must vanish somewhere in $[a, b]$.

**Step 5 (consequences).** Every interval of length $\pi/m$ contains a zero, so consecutive zeros are separated by at most $\pi/m$, and on $[0, \infty)$ the zeros form an infinite sequence marching right with gaps $\le \pi/m$. (This is the Sturm comparison theorem in its simplest calibrated form; comparing two variable coefficients $q_1 \le q_2$ works the same way. Note how it complements L3.1: $q \lt 0$ forbids oscillation, $q \ge m^2 \gt 0$ forces it.)

$$
\boxed{q \ge m^2 \gt 0 \ \Longrightarrow \ \text{every solution of } y'' + qy = 0 \text{ vanishes in every interval of length } \pi/m}
$$

> **Key Takeaway:** The Wronskian-with-a-comparison-solution trick converts pointwise inequalities on $q$ into forced zero crossings: stiffer springs oscillate at least as fast as the $\sin(mt)$ yardstick.

### Problem L3.3: Increasing Stiffness Squeezes the Amplitude

**Problem Statement:** Let $q$ be continuously differentiable with $q(t) \gt 0$ and $q'(t) \ge 0$ (stiffness positive and non-decreasing), and let $y$ solve $y'' + q(t) y = 0$. Prove that the successive local maxima of $\lvert y \rvert$ are non-increasing: if $t_1 \lt t_2$ are critical points of $y$ (points with $y' = 0$), then $\lvert y(t_2)\rvert \le \lvert y(t_1) \rvert$.

**Intuition:** A pendulum whose spring stiffens over time cannot swing wider — energy per unit stiffness is the right conserved-ish quantity. This is a Tripos/Putnam-style "invent the Lyapunov function" problem.

**Solution:**

**Step 1 (choose the functional).** Define

$$
E(t) = y(t)^2 + \frac{y'(t)^2}{q(t)}
$$

This is an "amplitude-squared" surrogate: at any critical point ($y' = 0$), $E = y^2$ exactly.

**Step 2 (differentiate along solutions).** Using the quotient rule and $y'' = -q y$:

$$
E' = 2 y y' + \frac{2 y' y'' \, q - (y')^2 q'}{q^2} = 2 y y' + \frac{2 y' (-q y)}{q} - \frac{(y')^2 q'}{q^2}
$$

The first two terms cancel: $2yy' - 2y'y = 0$. Hence

$$
E'(t) = -\,\frac{q'(t)}{q(t)^2}\, y'(t)^2 \le 0
$$

since $q' \ge 0$ and $q^2 \gt 0$: **$E$ is non-increasing along every solution.**

**Step 3 (compare critical points).** Let $t_1 \lt t_2$ with $y'(t_1) = y'(t_2) = 0$. Then

$$
y(t_2)^2 = E(t_2) \le E(t_1) = y(t_1)^2
$$

so $\lvert y(t_2)\rvert \le \lvert y(t_1)\rvert$, as claimed. In particular the sequence of successive swing maxima is non-increasing; it is *strictly* decreasing whenever $q' \gt 0$ on a set where $y'$ is not identically zero between the two critical points.

**Step 4 (sanity checks).** If $q$ is constant then $q' = 0$, so $E = y^2 + (y')^2/q$ is exactly conserved (it is the mechanical energy rescaled by $2/q$) and all maxima are equal — SHM recovered. For $q(t) = t$ (Airy's equation $y'' + t y = 0$), the result predicts the visible fact that Airy oscillations shrink as $t$ grows (indeed like $t^{-1/4}$, consistent with non-increasing $E$; our elementary bound gives monotonicity though not the rate). The companion statement with $q' \le 0$ and $E$ non-decreasing shows *softening* springs swing wider.

$$
\boxed{q \gt 0, \ q' \ge 0 \ \Longrightarrow \ E = y^2 + \frac{(y')^2}{q} \text{ is non-increasing, so successive amplitude maxima cannot grow}}
$$

> **Key Takeaway:** When no exact conservation law exists, a *monotone* energy-like functional (here $y^2 + (y')^2/q$) still yields rigorous qualitative bounds — the one-line $E'$ computation is the whole proof.

### Problem L3.4: A Green's Function Preview — Boundary Value Problem for $y'' = f$

**Problem Statement:** Consider the boundary value problem

$$
y'' = f(t), \qquad y(0) = 0, \quad y(1) = 0
$$

with $f$ continuous on $[0,1]$. (a) Construct a function $G(t,s)$ such that the unique solution is

$$
y(t) = \int_0^1 G(t, s)\, f(s)\, ds
$$

(b) Verify directly that your formula satisfies the ODE and both boundary conditions. (c) Interpret $G(t,s)$ physically.

**Intuition:** Instead of solving anew for each forcing, solve once for a unit "point load" at position $s$; superposition then assembles the answer for any $f$ — the boundary-value analogue of variation of parameters.

**Solution:**

**(a) Construction.** Homogeneous solutions of $y'' = 0$ are $1$ and $t$. Choose one satisfying each boundary condition:

$$
y_1(t) = t \quad (y_1(0) = 0), \qquad y_2(t) = t - 1 \quad (y_2(1) = 0)
$$

Their Wronskian is $W = y_1 y_2' - y_1' y_2 = t \cdot 1 - 1\cdot(t - 1) = 1$ (constant, as Abel predicts with $p = 0$). The standard Green's function pattern — left solution in the left slot, right solution in the right slot, divided by $W$ — gives

$$
G(t, s) = \begin{cases} t\,(s - 1), & 0 \le t \le s \\ s\,(t - 1), & s \le t \le 1 \end{cases}
$$

Note $G$ is continuous, symmetric ($G(t,s) = G(s,t)$), negative in the interior, and its $t$-derivative jumps by $s - (s-1) = 1$ across $t = s$ — the unit impulse.

**(b) Verification.** Write the proposed solution splitting the integral at $t$:

$$
y(t) = \int_0^{t} s(t-1) f(s)\, ds + \int_{t}^{1} t(s-1) f(s)\, ds = (t - 1)\int_0^t s f(s)\, ds + t \int_t^1 (s - 1) f(s)\, ds
$$

*Boundary conditions:* at $t = 0$ both terms vanish (first integral is empty, second has factor $t = 0$): $y(0) = 0$. At $t = 1$: first term has factor $t - 1 = 0$, second integral is empty: $y(1) = 0$. ✔

*Differentiate* (Leibniz rule; the boundary terms from the moving limits cancel):

$$
y'(t) = \int_0^t s f(s)\, ds + (t-1)\, t f(t) + \int_t^1 (s-1) f(s)\, ds - t\,(t - 1) f(t) = \int_0^t s f(s)\, ds + \int_t^1 (s-1) f(s)\, ds
$$

*Differentiate again:*

$$
y''(t) = t f(t) + \bigl(-(t - 1) f(t)\bigr) = \bigl(t - t + 1\bigr) f(t) = f(t)
$$

as required. ✔

Uniqueness: the difference of two solutions solves $y'' = 0$ with zero boundary values, i.e. a line through $(0,0)$ and $(1,0)$ — identically zero.

**(c) Physical interpretation.** $G(t, s)$ is the displacement at position $t$ of a taut string (or the deflection of a beam in the membrane analogy) pinned at both ends, under a *unit point load* at position $s$: two straight segments meeting in a kink at $s$, sagging by $G(s,s) = s(s-1) \lt 0$. The integral $\int_0^1 G(t,s) f(s)\, ds$ superposes point-load responses weighted by the load density $f$ — possible only because the problem is linear. This object generalizes to $y'' + p y' + q y = f$ with any separated boundary conditions, inverts the differential operator ($G$ is the kernel of $L^{-1}$), and reappears throughout physics (potentials), PDE theory, and machine learning (kernels of Gaussian-process priors associated with differential operators).

$$
\boxed{G(t,s) = \begin{cases} t(s-1), & t \le s \\ s(t-1), & t \ge s \end{cases}, \qquad y(t) = \int_0^1 G(t,s) f(s)\, ds}
$$

> **Key Takeaway:** A Green's function solves the boundary value problem once and for all: it is the response to a unit impulse, and superposition turns it into the inverse of the differential operator — the bridge from this module to boundary value problems and PDEs.
