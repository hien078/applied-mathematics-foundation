# Topic 03: Second-Order Linear ODEs

## 1. First-Principles Intuition & Motivation

**Phenomenon.** Newton's second law is a statement about *acceleration*: $F = ma$. Whenever a system near equilibrium experiences a restoring force proportional to displacement and a friction force proportional to velocity, the law reads

$$
m x'' + c x' + k x = F_{\text{ext}}(t)
$$

This single equation describes a mass on a spring, a pendulum swinging through small angles, a car body bouncing on its suspension, and a bridge deck responding to wind. Remarkably, it also describes electricity: Kirchhoff's voltage law for a series RLC circuit gives

$$
L q'' + R q' + \frac{1}{C} q = V(t)
$$

with charge $q$ playing the role of position, inductance $L$ the role of mass, resistance $R$ the role of friction, and $1/C$ the role of stiffness. Even optimization joins the club: gradient descent with momentum is, in continuous time, a ball with mass rolling down the loss landscape with friction — a damped oscillator in each eigendirection of the Hessian.

Second-order linear ODEs are therefore not one topic among many: they are *the* universal model of anything that vibrates, rings, settles, or overshoots.

### Goal and Standing Assumptions

**Goal.** Given

$$
a(t) y'' + b(t) y' + c(t) y = g(t)
$$

we want to (i) describe the *entire* solution set, not just one solution; (ii) determine which initial data $y(t_0) = y_0$, $y'(t_0) = v_0$ pin down a unique solution; (iii) compute solutions explicitly in the constant-coefficient case; and (iv) understand forced behavior — in particular resonance, the dangerous case where forcing at the natural frequency produces unbounded response.

**Standing assumptions.** Throughout this module:

1. The coefficients $a, b, c$ and forcing $g$ are continuous on an open interval $I$.
2. $a(t) \neq 0$ on $I$, so we may divide and use the **standard form** $y'' + p(t) y' + q(t) y = f(t)$ with $p = b/a$, $q = c/a$, $f = g/a$.
3. "Solution" means a twice continuously differentiable function on $I$ satisfying the equation at every point of $I$.

The structural results (dimension of the solution space, Wronskian theory, variation of parameters) hold for *variable* continuous coefficients; explicit formulas via the characteristic equation require *constant* coefficients, with the Cauchy–Euler family as the one important variable-coefficient class that reduces to it.

### Variables and Notation

| Symbol | Meaning |
| :--- | :--- |
| $t$ | Independent variable (time) on an interval $I$ |
| $y(t)$ | Unknown function; $y'$, $y''$ its first and second derivatives |
| $D$ | Differentiation operator $Dy = y'$ |
| $L$ | Linear differential operator $L = a D^2 + b D + c$ |
| $\lambda$ | Root of the characteristic equation (a "mode" exponent) |
| $W(t)$ | Wronskian $W = y_1 y_2' - y_1' y_2$ of a solution pair |
| $\omega_0$ | Natural (undamped) angular frequency, $\omega_0 = \sqrt{k/m}$ |
| $\zeta$ | Damping ratio $\zeta = \frac{c}{2\sqrt{mk}}$; critical damping is $\zeta = 1$ |
| $Q$ | Quality factor $Q = \frac{1}{2\zeta}$, the sharpness of the resonance peak |
| $y_h, y_p$ | General homogeneous solution and one particular solution |

Two dimensionless numbers organize all oscillator behavior: the damping ratio $\zeta$ decides the *regime* (underdamped $\zeta \lt 1$, critical $\zeta = 1$, overdamped $\zeta \gt 1$), and the frequency ratio $\omega/\omega_0$ of forcing to natural frequency decides the *steady-state amplitude and phase*.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition (second-order linear ODE).** An equation of the form

$$
a(t) y'' + b(t) y' + c(t) y = g(t)
$$

with $a(t) \neq 0$ on $I$. It is **homogeneous** if $g \equiv 0$, **non-homogeneous** otherwise, and has **constant coefficients** if $a, b, c$ are constants.

**Definition (linear differential operator).** Define $L : C^2(I) \to C(I)$ by

$$
L[y] = a(t) y'' + b(t) y' + c(t) y
$$

$L$ is a *linear map* between vector spaces of functions: for all functions $u, v \in C^2(I)$ and scalars $\alpha, \beta$,

$$
L[\alpha u + \beta v] = \alpha L[u] + \beta L[v]
$$

which follows immediately from linearity of differentiation. This one-line observation has a powerful consequence.

**Theorem (superposition principle).** If $y_1$ and $y_2$ solve the homogeneous equation $L[y] = 0$, then every linear combination $c_1 y_1 + c_2 y_2$ also solves it. Equivalently: *the solution set of $L[y] = 0$ is a vector subspace of $C^2(I)$* — it is exactly $\ker L$, the null space of the operator $L$.

Note carefully what superposition does **not** say: sums of solutions of the *non-homogeneous* equation $L[y] = g$ are not solutions (they solve $L[y] = 2g$), and superposition fails entirely for nonlinear equations such as $y'' + \sin y = 0$.

### Existence, Uniqueness, and the Dimension of the Solution Space

**Theorem (existence and uniqueness for linear equations).** Suppose $p, q, f$ are continuous on the open interval $I$ and $t_0 \in I$. Then for any numbers $y_0, v_0$ the initial value problem

$$
y'' + p(t) y' + q(t) y = f(t), \qquad y(t_0) = y_0, \quad y'(t_0) = v_0
$$

has exactly one solution, and it exists on **all** of $I$ (not merely locally, as in the nonlinear Picard–Lindelöf theorem — see the survey module [`../../calculus/15_ordinary_differential_equations/`](../../calculus/15_ordinary_differential_equations/) for the general theorem).

**Theorem (the solution space is two-dimensional).** The set $S = \{y : L[y] = 0 \text{ on } I\}$ is a vector space with $\dim S = 2$.

*Why dimension two:* the evaluation map $\Phi : S \to \mathbb{R}^2$, $\Phi(y) = (y(t_0), y'(t_0))$, is linear; existence says $\Phi$ is surjective, uniqueness says $\Phi$ is injective. Hence $\Phi$ is an isomorphism and $\dim S = \dim \mathbb{R}^2 = 2$. This is the precise sense in which a second-order equation "needs two constants": a solution is determined by exactly two numbers, its initial position and initial velocity.

**Definition (fundamental set, general solution).** A pair $\{y_1, y_2\}$ of linearly independent solutions of $L[y] = 0$ is called a **fundamental set of solutions**; then every solution is $y = c_1 y_1 + c_2 y_2$ for unique constants $c_1, c_2$, and this expression is the **general solution**.

### The Wronskian and Abel's Theorem

**Definition (Wronskian).** For differentiable functions $y_1, y_2$,

$$
W(y_1, y_2)(t) = \det \begin{pmatrix} y_1(t) & y_2(t) \\ y_1'(t) & y_2'(t) \end{pmatrix} = y_1(t) y_2'(t) - y_1'(t) y_2(t)
$$

**Theorem (Abel's identity).** If $y_1, y_2$ solve $y'' + p(t) y' + q(t) y = 0$ on $I$ with $p$ continuous, then

$$
W(t) = W(t_0)\, \exp\left( -\int_{t_0}^{t} p(s)\, ds \right)
$$

Since the exponential factor is strictly positive, we get the **dichotomy**: *the Wronskian of two solutions is either identically zero on $I$ or never zero on $I$.* (Proof 3.1 below.)

**Theorem (Wronskian test).** Two solutions $y_1, y_2$ of $L[y] = 0$ form a fundamental set if and only if $W(t_0) \neq 0$ at one (equivalently every) point $t_0 \in I$. (Proof 3.2 below.)

**Warning — this is a theorem about solutions, not about functions.** Consider $y_1(t) = t^3$ and $y_2(t) = \lvert t \rvert^3$ on $(-1, 1)$: both are continuously differentiable ($y_2' = 3 t \lvert t \rvert$), and they are linearly independent — no single constant $c$ gives $y_2 = c y_1$ on both sides of $0$, since $y_2 = y_1$ for $t \gt 0$ but $y_2 = -y_1$ for $t \lt 0$. Yet on $t \gt 0$ the columns of the Wronskian matrix are equal, on $t \lt 0$ they are negatives of each other, and at $t = 0$ both columns vanish, so

$$
W(y_1, y_2)(t) \equiv 0 \quad \text{on } (-1, 1)
$$

An independent pair with identically zero Wronskian! No contradiction with the dichotomy: this pair can never be two solutions of a single equation $y'' + p y' + q y = 0$ with $p, q$ continuous on an interval containing $0$ — the Wronskian theory is *powered by* existence–uniqueness, which arbitrary functions do not enjoy.

### Constant Coefficients: the Characteristic Equation

For $a y'' + b y' + c y = 0$ with constants $a \neq 0, b, c$, try the ansatz $y = e^{\lambda t}$. Since $L[e^{\lambda t}] = (a \lambda^2 + b \lambda + c) e^{\lambda t}$ and $e^{\lambda t} \neq 0$, the ansatz solves the ODE exactly when $\lambda$ solves the **characteristic equation**

$$
a \lambda^2 + b \lambda + c = 0, \qquad \lambda = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}
$$

The discriminant $\Delta = b^2 - 4ac$ sorts all behavior into three regimes:

| Regime | Roots | Fundamental set | General solution | Oscillator name |
| :--- | :--- | :--- | :--- | :--- |
| $\Delta \gt 0$ | Real distinct $\lambda_1 \neq \lambda_2$ | $e^{\lambda_1 t}, e^{\lambda_2 t}$ | $c_1 e^{\lambda_1 t} + c_2 e^{\lambda_2 t}$ | Overdamped |
| $\Delta = 0$ | Real repeated $\lambda = -\frac{b}{2a}$ | $e^{\lambda t}, t e^{\lambda t}$ | $(c_1 + c_2 t) e^{\lambda t}$ | Critically damped |
| $\Delta \lt 0$ | Complex $\alpha \pm i \beta$ | $e^{\alpha t}\cos\beta t, e^{\alpha t}\sin\beta t$ | $e^{\alpha t}(c_1 \cos\beta t + c_2 \sin\beta t)$ | Underdamped |

The complex-root entry uses **Euler's formula** $e^{i\beta t} = \cos\beta t + i \sin\beta t$: the complex solutions $e^{(\alpha \pm i\beta)t}$ have real and imaginary parts $e^{\alpha t}\cos\beta t$ and $e^{\alpha t}\sin\beta t$, and since $L$ has real coefficients, real and imaginary parts of a complex solution are themselves solutions. One checks $W(e^{\alpha t}\cos\beta t, e^{\alpha t}\sin\beta t) = \beta e^{2\alpha t} \neq 0$ when $\beta \neq 0$, so these form a genuine fundamental set. The repeated-root solution $t e^{\lambda t}$ is *derived*, not decreed, in Proof 3.3.

### Non-Homogeneous Equations and the Cauchy–Euler Family

**Theorem (structure of the non-homogeneous solution set).** If $y_p$ is any one solution of $L[y] = g$ and $\{y_1, y_2\}$ is a fundamental set for $L[y] = 0$, then every solution of $L[y] = g$ is

$$
y = c_1 y_1 + c_2 y_2 + y_p
$$

*Proof sketch:* if $L[y] = g$ and $L[y_p] = g$ then $L[y - y_p] = 0$ by linearity, so $y - y_p \in \ker L = \operatorname{span}\{y_1, y_2\}$. Geometrically the solution set is an *affine plane*: the null space translated by $y_p$.

**Method of undetermined coefficients** (constant coefficients, special forcings): guess $y_p$ in the same family as $g$ and match coefficients.

| Forcing $g(t)$ | Trial $y_p(t)$ |
| :--- | :--- |
| $P_n(t)$ polynomial | $A_n t^n + \cdots + A_1 t + A_0$ |
| $e^{\gamma t}$ | $A e^{\gamma t}$ |
| $\cos \omega t$ or $\sin \omega t$ | $A \cos\omega t + B \sin\omega t$ |
| $e^{\gamma t} \cos\omega t$ | $e^{\gamma t}(A\cos\omega t + B\sin\omega t)$ |

**Resonance modification rule:** if the trial solution solves the homogeneous equation (i.e., $\gamma$ or $\gamma \pm i\omega$ is a characteristic root), multiply the trial by $t$; if it is a double root, by $t^2$. The reason: the naive trial lies in $\ker L$ and is annihilated, so no choice of coefficients can produce $g$; multiplying by $t$ moves it out of the null space. Proof 3.5 shows the resulting secular growth explicitly.

**Variation of parameters** (general forcing, even variable coefficients): promote constants to functions, $y_p = u_1(t) y_1 + u_2(t) y_2$; the full derivation is Proof 3.4.

**Definition (Cauchy–Euler equation).** The variable-coefficient equation

$$
t^2 y'' + a\, t y' + b\, y = 0, \qquad t \gt 0
$$

Substituting $t = e^s$ (so $s = \ln t$) converts it into a *constant-coefficient* equation. Writing $\dot{y}$ for $dy/ds$, the chain rule gives $t y' = \dot{y}$ and $t^2 y'' = \ddot{y} - \dot{y}$, hence

$$
\ddot{y} + (a - 1)\dot{y} + b y = 0
$$

whose characteristic roots $\lambda$ give solutions $y = e^{\lambda s} = t^{\lambda}$ — power functions rather than exponentials, with $t^{\lambda} \ln t$ in the repeated-root case and $t^{\alpha}\cos(\beta \ln t)$, $t^{\alpha}\sin(\beta \ln t)$ for complex roots $\alpha \pm i\beta$.

## 3. Step-by-Step Mathematical Proofs & Derivations

We now prove the six pillars of the theory:

1. **Abel's theorem** — the Wronskian obeys its own first-order ODE, forcing the all-or-nothing dichotomy.
2. **Wronskian test** — $W \neq 0$ at a point is equivalent to being a fundamental set (this is where existence–uniqueness earns its keep).
3. **Repeated-root solution** — $t e^{\lambda t}$ derived honestly by reduction of order, with the merging-roots limit as a second view.
4. **Variation of parameters** — the general particular-solution formula, with the $u_1', u_2'$ system solved via the Wronskian.
5. **Pure resonance** — the exact solution of $y'' + \omega^2 y = \cos\omega t$, exhibiting linear-in-time amplitude growth.
6. **Equivalence with a first-order system** — one second-order equation and one $2 \times 2$ system are the same object in different coordinates.

### Proof 3.1 — Abel's Theorem

**Claim.** If $y_1, y_2$ solve $y'' + p(t) y' + q(t) y = 0$ on $I$ and $t_0 \in I$, then $W(t) = W(t_0) e^{-\int_{t_0}^{t} p(s) ds}$.

**Step 1.** Differentiate the Wronskian $W = y_1 y_2' - y_1' y_2$ using the product rule:

$$
W' = y_1' y_2' + y_1 y_2'' - y_1'' y_2 - y_1' y_2' = y_1 y_2'' - y_1'' y_2
$$

The cross terms cancel — the derivative of a Wronskian only sees second derivatives.

**Step 2.** Both functions are *solutions*, so we may eliminate the second derivatives:

$$
y_1'' = -p y_1' - q y_1, \qquad y_2'' = -p y_2' - q y_2
$$

**Step 3.** Substitute into Step 1:

$$
W' = y_1(-p y_2' - q y_2) - (-p y_1' - q y_1) y_2 = -p (y_1 y_2' - y_1' y_2) - q(y_1 y_2 - y_1 y_2)
$$

The $q$ terms cancel identically, leaving the first-order linear equation

$$
W' = -p(t)\, W
$$

**Step 4.** This is a separable first-order ODE with the unique solution through $W(t_0)$:

$$
W(t) = W(t_0) \exp\left(-\int_{t_0}^{t} p(s)\, ds\right)
$$

**Step 5 (dichotomy).** The exponential factor is strictly positive for every $t \in I$. Hence if $W(t_0) = 0$ then $W \equiv 0$ on $I$; if $W(t_0) \neq 0$ then $W(t) \neq 0$ for every $t \in I$. For the original form $a y'' + b y' + c y = 0$, replace $p$ by $b/a$.

$$
\boxed{W(t) = W(t_0)\, e^{-\int_{t_0}^{t} p(s)\, ds} \quad \Longrightarrow \quad W \equiv 0 \text{ on } I \ \text{ or } \ W(t) \neq 0 \text{ for all } t \in I}
$$

### Proof 3.2 — Wronskian Nonzero iff Fundamental Set

**Claim.** Let $y_1, y_2$ solve $y'' + p y' + q y = 0$ on $I$ with $p, q$ continuous. Then $\{y_1, y_2\}$ is a fundamental set (linearly independent, hence spanning the solution space) if and only if $W(t_0) \neq 0$ for some $t_0 \in I$.

**Step 1 ($W(t_0) \neq 0 \Rightarrow$ spanning).** Let $y$ be *any* solution and set $y_0 = y(t_0)$, $v_0 = y'(t_0)$. Seek constants with $c_1 y_1 + c_2 y_2$ matching these initial data:

$$
\begin{pmatrix} y_1(t_0) & y_2(t_0) \\ y_1'(t_0) & y_2'(t_0) \end{pmatrix} \begin{pmatrix} c_1 \\ c_2 \end{pmatrix} = \begin{pmatrix} y_0 \\ v_0 \end{pmatrix}
$$

The coefficient matrix has determinant $W(t_0) \neq 0$, so unique $c_1, c_2$ exist. Now $c_1 y_1 + c_2 y_2$ and $y$ are two solutions of the same IVP; by the **uniqueness theorem** they coincide on all of $I$. Every solution is a combination, and independence follows since neither $y_1$ nor $y_2$ can be a multiple of the other without making $W(t_0) = 0$.

**Step 2 ($W(t_0) = 0 \Rightarrow$ dependent).** If $W(t_0) = 0$, the columns of the matrix above are linearly dependent: there exist $(c_1, c_2) \neq (0,0)$ with

$$
c_1 y_1(t_0) + c_2 y_2(t_0) = 0, \qquad c_1 y_1'(t_0) + c_2 y_2'(t_0) = 0
$$

Define $\phi = c_1 y_1 + c_2 y_2$. By superposition $\phi$ is a solution, and it satisfies $\phi(t_0) = 0$, $\phi'(t_0) = 0$. But the zero function solves the same IVP, so by **uniqueness** $\phi \equiv 0$ on $I$ — a nontrivial vanishing linear combination. Hence $y_1, y_2$ are linearly dependent.

**Step 3.** Combining with Abel's dichotomy, "some $t_0$" and "every $t_0$" are interchangeable: checking $W$ at one convenient point settles independence everywhere.

$$
\boxed{\{y_1, y_2\} \text{ is a fundamental set of } L[y] = 0 \iff W(y_1, y_2)(t_0) \neq 0 \text{ at a single point } t_0 \in I}
$$

**Remark.** Step 2 is exactly where the $t^3$ vs $\lvert t \rvert^3$ pathology is excluded: for mere functions, $W(t_0) = 0$ lets us build $\phi$ with zero initial data, but *nothing* forces $\phi \equiv 0$. Uniqueness of solutions is the engine of the whole equivalence.

### Proof 3.3 — The Repeated-Root Solution via Reduction of Order

**Setting.** Suppose $\Delta = b^2 - 4ac = 0$, so the characteristic equation of $a y'' + b y' + c y = 0$ has the single root $\lambda = -\frac{b}{2a}$, giving only one mode $y_1 = e^{\lambda t}$. We need a second independent solution.

**Step 1 (ansatz).** Seek $y_2 = v(t)\, y_1(t) = v(t) e^{\lambda t}$ with $v$ an unknown function (if $v$ is non-constant, $y_2$ is automatically independent of $y_1$). Compute derivatives:

$$
y_2' = (v' + \lambda v) e^{\lambda t}, \qquad y_2'' = (v'' + 2\lambda v' + \lambda^2 v) e^{\lambda t}
$$

**Step 2 (substitute).** Plug into $a y_2'' + b y_2' + c y_2$ and divide by $e^{\lambda t} \neq 0$:

$$
a v'' + (2a\lambda + b) v' + (a \lambda^2 + b \lambda + c) v = 0
$$

**Step 3 (both lower-order coefficients die).** The coefficient of $v$ vanishes because $\lambda$ is a characteristic root. The coefficient of $v'$ is $2a\lambda + b = 2a \cdot (-\tfrac{b}{2a}) + b = 0$ — and this happens **precisely because the root is repeated**: $\lambda = -b/(2a)$ is the vertex of the parabola, i.e. the double root. What survives is simply

$$
a v'' = 0 \quad \Longrightarrow \quad v'' = 0 \quad \Longrightarrow \quad v(t) = C_1 t + C_2
$$

**Step 4.** Taking $v = t$ (the non-constant choice) gives $y_2 = t e^{\lambda t}$. Independence check via the Wronskian:

$$
W(e^{\lambda t}, t e^{\lambda t}) = e^{\lambda t}(e^{\lambda t} + \lambda t e^{\lambda t}) - \lambda e^{\lambda t} \cdot t e^{\lambda t} = e^{2\lambda t} \neq 0
$$

**Step 5 (second view: merging roots).** For distinct roots $\lambda_1 \neq \lambda_2$, the combination $\dfrac{e^{\lambda_2 t} - e^{\lambda_1 t}}{\lambda_2 - \lambda_1}$ is a solution by superposition. Fix $\lambda_1 = \lambda$ and let $\lambda_2 \to \lambda$: the limit is the derivative of $\mu \mapsto e^{\mu t}$ at $\mu = \lambda$,

$$
\lim_{\lambda_2 \to \lambda} \frac{e^{\lambda_2 t} - e^{\lambda t}}{\lambda_2 - \lambda} = \frac{\partial}{\partial \mu} e^{\mu t} \bigg\vert_{\mu = \lambda} = t e^{\lambda t}
$$

The mysterious factor $t$ is a derivative in the *root variable*: when two exponential modes collide, their difference quotient survives as $t e^{\lambda t}$.

$$
\boxed{\Delta = 0 \ \Longrightarrow \ y = (c_1 + c_2 t)\, e^{-bt/(2a)} \text{ is the general solution}}
$$

### Proof 3.4 — Variation of Parameters

**Claim.** Let $\{y_1, y_2\}$ be a fundamental set for $y'' + p(t) y' + q(t) y = 0$. Then

$$
y_p(t) = -y_1(t) \int \frac{y_2(t) f(t)}{W(t)}\, dt + y_2(t) \int \frac{y_1(t) f(t)}{W(t)}\, dt
$$

solves $y'' + p y' + q y = f(t)$ for **any** continuous forcing $f$.

**Step 1 (ansatz).** Promote the constants of the general homogeneous solution to functions: $y_p = u_1(t) y_1 + u_2(t) y_2$. Then

$$
y_p' = u_1 y_1' + u_2 y_2' + (u_1' y_1 + u_2' y_2)
$$

**Step 2 (the free constraint).** Two unknown functions, one equation to satisfy — we may impose one extra condition. Choose the one that prevents second derivatives of $u_i$ from ever appearing:

$$
u_1' y_1 + u_2' y_2 = 0
$$

With this, $y_p' = u_1 y_1' + u_2 y_2'$, exactly as if $u_1, u_2$ were constants.

**Step 3 (substitute).** Differentiate again: $y_p'' = u_1 y_1'' + u_2 y_2'' + u_1' y_1' + u_2' y_2'$. Insert into the ODE and group:

$$
u_1 (y_1'' + p y_1' + q y_1) + u_2 (y_2'' + p y_2' + q y_2) + u_1' y_1' + u_2' y_2' = f
$$

Both parenthesized groups vanish because $y_1, y_2$ are homogeneous solutions, leaving $u_1' y_1' + u_2' y_2' = f$.

**Step 4 (the linear system).** Steps 2 and 3 form a $2 \times 2$ linear system for $u_1', u_2'$:

$$
\begin{pmatrix} y_1 & y_2 \\ y_1' & y_2' \end{pmatrix} \begin{pmatrix} u_1' \\ u_2' \end{pmatrix} = \begin{pmatrix} 0 \\ f \end{pmatrix}
$$

The determinant is the Wronskian $W(t)$, nonzero everywhere by Proof 3.2. Cramer's rule gives

$$
u_1' = \frac{-y_2 f}{W}, \qquad u_2' = \frac{y_1 f}{W}
$$

**Step 5 (integrate).** Antidifferentiating and assembling $y_p = u_1 y_1 + u_2 y_2$ yields the claimed formula; the arbitrary constants of integration merely add homogeneous solutions and may be dropped.

$$
\boxed{y_p = -y_1 \int \frac{y_2 f}{W}\, dt + y_2 \int \frac{y_1 f}{W}\, dt, \qquad u_1' = \frac{-y_2 f}{W}, \quad u_2' = \frac{y_1 f}{W}}
$$

**Remark.** Unlike undetermined coefficients, this works for variable coefficients and arbitrary continuous $f$ — e.g. $y'' + y = \tan t$, solved in the exercises — at the price of two integrals that may lack elementary antiderivatives.

### Proof 3.5 — Pure Resonance: Exact Secular Growth

**Claim.** The IVP $y'' + \omega^2 y = \cos \omega t$, $y(0) = y'(0) = 0$ (forcing exactly at the natural frequency) has the exact solution $y(t) = \dfrac{t}{2\omega} \sin \omega t$, whose amplitude grows linearly in $t$.

**Step 1 (naive ansatz fails).** Try $y_p = A\cos\omega t + B \sin\omega t$. Then $y_p'' = -\omega^2 y_p$, so

$$
y_p'' + \omega^2 y_p = 0 \neq \cos\omega t
$$

The trial lies in the null space of $L = D^2 + \omega^2$: no coefficients can work. This *is* resonance, algebraically.

**Step 2 (modified ansatz).** Apply the resonance rule: $y_p = t(A \cos\omega t + B\sin\omega t)$. Differentiate twice:

$$
y_p' = (A\cos\omega t + B\sin\omega t) + t(-A\omega \sin\omega t + B\omega \cos\omega t)
$$

$$
y_p'' = 2(-A\omega\sin\omega t + B\omega\cos\omega t) + t(-A\omega^2\cos\omega t - B\omega^2 \sin\omega t)
$$

**Step 3 (match).** In $y_p'' + \omega^2 y_p$ the terms proportional to $t$ cancel exactly (they reconstruct $\omega^2 y_p$ with opposite sign), leaving

$$
y_p'' + \omega^2 y_p = -2A\omega \sin\omega t + 2B\omega\cos\omega t \stackrel{!}{=} \cos\omega t
$$

Matching coefficients: $A = 0$ and $B = \dfrac{1}{2\omega}$, so $y_p = \dfrac{t}{2\omega}\sin\omega t$.

**Step 4 (initial conditions).** General solution $y = c_1 \cos\omega t + c_2 \sin\omega t + \frac{t}{2\omega}\sin\omega t$. Then $y(0) = c_1 = 0$ and $y'(0) = c_2 \omega = 0$, so $c_2 = 0$.

**Step 5 (interpretation).** The solution oscillates at frequency $\omega$ inside the *growing envelope* $\pm\, t/(2\omega)$: each forcing cycle arrives exactly in phase with the velocity and pumps energy in, never out. With any damping $\gamma \gt 0$ the growth saturates at steady-state amplitude of order $\frac{1}{\gamma \omega}$; the idealized undamped case grows without bound — the mathematical seed of the Tacoma Narrows-style failure narrative and of why soldiers break step on bridges.

$$
\boxed{y'' + \omega^2 y = \cos\omega t, \quad y(0) = y'(0) = 0 \ \Longrightarrow \ y(t) = \frac{t}{2\omega}\, \sin \omega t}
$$

### Proof 3.6 — Equivalence of a Second-Order ODE and a 2×2 First-Order System

**Claim.** The scalar equation $y'' + p(t) y' + q(t) y = f(t)$ and the planar system

$$
\mathbf{x}' = A(t)\, \mathbf{x} + \mathbf{b}(t), \qquad A(t) = \begin{pmatrix} 0 & 1 \\ -q(t) & -p(t) \end{pmatrix}, \qquad \mathbf{b}(t) = \begin{pmatrix} 0 \\ f(t) \end{pmatrix}
$$

have solution sets in natural one-to-one correspondence.

**Step 1 (scalar to system).** Given a solution $y$, define $x_1 = y$, $x_2 = y'$. Then

$$
x_1' = y' = x_2, \qquad x_2' = y'' = -q y - p y' + f = -q x_1 - p x_2 + f
$$

which is exactly $\mathbf{x}' = A\mathbf{x} + \mathbf{b}$. The map $y \mapsto (y, y')$ is well defined and injective (a function determines its derivative).

**Step 2 (system to scalar).** Conversely, let $\mathbf{x} = (x_1, x_2)$ solve the system. The first row says $x_2 = x_1'$, so $x_1$ is differentiable twice and the second row reads $x_1'' = -q x_1 - p x_1' + f$: the function $y = x_1$ solves the scalar equation, and $(x_1, x_2) = (y, y')$. The two maps are mutually inverse bijections, linear in the homogeneous case.

**Step 3 (structures match).** For constant coefficients ($f = 0$), the companion matrix $A = \begin{pmatrix} 0 & 1 \\ -q & -p \end{pmatrix}$ has characteristic polynomial

$$
\det(A - \lambda I) = \lambda(\lambda + p) + q = \lambda^2 + p\lambda + q
$$

— *identical* to the ODE's characteristic polynomial. Eigenvalues of $A$ are the characteristic roots; eigenvectors are $(1, \lambda)^{T}$, i.e. the value-and-slope profile of the mode $e^{\lambda t}$. Moreover $\operatorname{tr} A = -p$ recovers Abel's theorem: the Wronskian of two solutions is the determinant of the fundamental matrix of the system, and Liouville's formula $\det \Phi(t) = \det\Phi(t_0) e^{\int \operatorname{tr} A} = \det\Phi(t_0) e^{-\int p}$ *is* Abel's identity in system language.

$$
\boxed{y'' + p y' + q y = f \ \Longleftrightarrow \ \begin{pmatrix} x_1 \\ x_2 \end{pmatrix}' = \begin{pmatrix} 0 & 1 \\ -q & -p \end{pmatrix} \begin{pmatrix} x_1 \\ x_2 \end{pmatrix} + \begin{pmatrix} 0 \\ f \end{pmatrix}, \quad (x_1, x_2) = (y, y')}
$$

## 4. Computational & Algorithmic Insights

**Solvers see systems, not second-order equations.** Every production ODE solver (`scipy.integrate.solve_ivp`, MATLAB `ode45`, Julia `DifferentialEquations.jl`) integrates first-order systems $\mathbf{x}' = F(t, \mathbf{x})$. Proof 3.6 is therefore not a curiosity but the *interface contract*: to solve $m y'' + c y' + k y = g(t)$ numerically, you hand the solver the companion system.

```text
# Solving m*y'' + c*y' + k*y = g(t) numerically
def rhs(t, x):                 # x = [position, velocity]
    y, v = x
    return [v, (g(t) - c*v - k*y) / m]

solution = solve_ivp(rhs, t_span=(0, T), x0=[y0, v0],
                     method="RK45", rtol=1e-8)
```

Cost per step is $O(1)$ right-hand-side evaluations for the scalar problem; for an $n$-degree-of-freedom structure $M \mathbf{y}'' + C \mathbf{y}' + K \mathbf{y} = \mathbf{g}$ the same reduction gives a $2n$-dimensional system, and each step costs a few matrix–vector products, $O(n^2)$ dense or $O(\text{nnz})$ sparse.

**Two distinct numerical difficulties hide in one equation:**

1. **High frequency (oscillatory stiffness).** For $y'' + \omega^2 y = 0$ with large $\omega$, accuracy forces the step size to resolve the period: $h \ll 2\pi/\omega$. An explicit method with step $h$ commits phase error of order $h^2 \omega^3 t$ over time $t$ — long integrations of fast oscillators drift in phase even when the amplitude looks fine.
2. **Widely separated decay rates (classical stiffness).** An overdamped system with roots $\lambda_1 = -10^4$ and $\lambda_2 = -1$ is stiff: the fast mode dies instantly, yet explicit Euler is *stable* only for $h \lt 2/10^4$. The step size is dictated by a mode that is already numerically zero — the signature of stiffness. Implicit methods (BDF, implicit Euler) remove the constraint at the cost of solving a linear system per step.

### Energy Drift: Symplectic vs Non-Symplectic Integration of SHM

Write SHM $y'' = -\omega^2 y$ as the system $y' = v$, $v' = -\omega^2 y$, with conserved energy $E = \frac{1}{2} v^2 + \frac{1}{2}\omega^2 y^2$.

**Forward (explicit) Euler** updates $y_{n+1} = y_n + h v_n$, $v_{n+1} = v_n - h \omega^2 y_n$. Compute the energy after one step:

$$
E_{n+1} = \tfrac{1}{2} v_{n+1}^2 + \tfrac{1}{2}\omega^2 y_{n+1}^2 = \tfrac{1}{2}(v_n - h\omega^2 y_n)^2 + \tfrac{1}{2}\omega^2 (y_n + h v_n)^2
$$

Expanding, the cross terms cancel and

$$
E_{n+1} = (1 + h^2 \omega^2)\, E_n
$$

**Every step multiplies the energy by $1 + h^2\omega^2 \gt 1$, no matter how small $h$ is.** The numerical oscillator spirals outward exponentially: after $n$ steps the energy is inflated by $(1 + h^2\omega^2)^n \approx e^{h \omega^2 t \cdot h}$. Halving $h$ slows the drift but never eliminates it. (Backward Euler has the mirror defect: it multiplies energy by $\frac{1}{1 + h^2\omega^2}$ and spirals inward.)

**Symplectic Euler** makes one tiny change — use the *updated* position in the velocity update (or vice versa):

```text
# Symplectic (semi-implicit) Euler for y'' = -omega^2 y
y_new = y + h * v
v_new = v - h * omega**2 * y_new     # <-- y_new, not y
```

This map is *symplectic* (it preserves phase-space area exactly), and one can check it exactly conserves the *modified* energy $\tilde{E} = \frac{1}{2}v^2 + \frac{1}{2}\omega^2 y^2 + \frac{h\omega^2}{2} y v$, which stays within $O(h)$ of the true $E$ **forever**. The numerical orbit is a slightly tilted ellipse instead of a spiral: bounded energy error over arbitrarily long times. This is why molecular dynamics and orbital mechanics use symplectic integrators (leapfrog/Verlet, a second-order symplectic scheme) rather than generic Runge–Kutta: for Hamiltonian systems, *preserving structure beats reducing local error*.

| Integrator | Energy behavior on SHM | Long-time orbit |
| :--- | :--- | :--- |
| Forward Euler | Multiplied by $1 + h^2\omega^2$ each step | Outward spiral, blow-up |
| Backward Euler | Divided by $1 + h^2\omega^2$ each step | Inward spiral, false decay |
| Symplectic Euler / Verlet | Modified energy conserved exactly; true energy oscillates within a band of width $O(h)$ | Closed (tilted) ellipse |

### Algorithmic Summary

```text
SOLVE a y'' + b y' + c y = g(t):
 1. Homogeneous part:
      roots of a*L^2 + b*L + c = 0
      D > 0 : y_h = c1 exp(L1 t) + c2 exp(L2 t)
      D = 0 : y_h = (c1 + c2 t) exp(L t)
      D < 0 : y_h = exp(alpha t) (c1 cos(beta t) + c2 sin(beta t))
 2. Particular part:
      g in polynomial/exp/trig family
          -> undetermined coefficients
             (multiply trial by t for each time it hits a char. root)
      otherwise
          -> variation of parameters:
             u1' = -y2 g / (a W),  u2' = y1 g / (a W)
 3. y = y_h + y_p ; fit c1, c2 to initial conditions LAST
    (after adding y_p, never before).
 4. Numerical route: convert to x' = [[0,1],[-c/a,-b/a]] x + [0, g/a]
    and integrate; choose implicit methods if roots widely separated,
    symplectic methods if undamped and long-time.
```

Common pitfall encoded in step 3: initial conditions must be imposed on the **full** solution $y_h + y_p$. Fitting $c_1, c_2$ to the data before adding $y_p$ silently shifts the initial state by $y_p(t_0)$ and $y_p'(t_0)$.

## 5. Real-World Physics & AI/ML Applications

### The Damped Oscillator and the Energy Method

The spring–mass–damper $m x'' + c x' + k x = 0$ defines $\omega_0 = \sqrt{k/m}$ and $\zeta = \frac{c}{2\sqrt{mk}}$; the characteristic roots are $\lambda = \omega_0\left(-\zeta \pm \sqrt{\zeta^2 - 1}\right)$.

- **Underdamped** ($\zeta \lt 1$): $x = e^{-\zeta\omega_0 t} \left(c_1 \cos\omega_d t + c_2 \sin\omega_d t\right)$ with ringing frequency $\omega_d = \omega_0\sqrt{1 - \zeta^2}$.
- **Critically damped** ($\zeta = 1$): $x = (c_1 + c_2 t) e^{-\omega_0 t}$ — fastest return to equilibrium without overshoot.
- **Overdamped** ($\zeta \gt 1$): two real decaying exponentials, dominated by the *slow* root, which approaches $0$ as $\zeta \to \infty$; excessive damping makes the system sluggish.

**Energy method.** Define $E(t) = \frac{1}{2} m v^2 + \frac{1}{2} k x^2$ with $v = x'$. Differentiate along solutions:

$$
\frac{dE}{dt} = m v v' + k x x' = v\,(m x'' + k x) = v\,(-c x' ) = -c\, v^2 \le 0
$$

Energy is exactly conserved when $c = 0$ (SHM traces an ellipse $\frac{1}{2}mv^2 + \frac{1}{2}kx^2 = E_0$ in phase space) and is dissipated at the instantaneous rate $c v^2$ otherwise — dissipation happens only while the mass is *moving*, which is why underdamped decay pauses momentarily at each turning point.

**Forced steady state.** For $m x'' + c x' + k x = F_0 \cos\omega t$ the particular solution is $x_p = R\cos(\omega t - \phi)$ with

$$
R(\omega) = \frac{F_0}{\sqrt{(k - m\omega^2)^2 + (c\omega)^2}}, \qquad \tan\phi = \frac{c\omega}{k - m\omega^2}
$$

The amplitude curve $R(\omega)$ peaks near $\omega_0$ for light damping, with peak height $\approx \frac{F_0}{c\,\omega_0} = Q \cdot \frac{F_0}{k}$: the **quality factor** $Q = \frac{\sqrt{mk}}{c} = \frac{1}{2\zeta}$ measures amplification at resonance and, equivalently, the number of radians of oscillation over which free vibrations decay by $e^{-1}$. The phase $\phi$ sweeps from $0$ (below resonance, response in phase) through $\pi/2$ (at $\omega_0$) to $\pi$ (above resonance, response inverted).

### The RLC Analogy, Suspensions, and Pendula

The mechanical–electrical dictionary is *exact* — same equation, different labels:

| Mechanical $m x'' + c x' + k x = F(t)$ | Electrical (series RLC) $L q'' + R q' + q/C = V(t)$ |
| :--- | :--- |
| Position $x$ | Charge $q$ |
| Velocity $v = x'$ | Current $i = q'$ |
| Mass $m$ (inertia) | Inductance $L$ (opposes current change) |
| Damping $c$ (friction) | Resistance $R$ (dissipation) |
| Stiffness $k$ | Elastance $1/C$ |
| Natural frequency $\sqrt{k/m}$ | Resonant frequency $1/\sqrt{LC}$ |
| Kinetic energy $\frac{1}{2} m v^2$ | Magnetic energy $\frac{1}{2} L i^2$ |
| Potential energy $\frac{1}{2} k x^2$ | Capacitor energy $q^2/(2C)$ |
| Quality factor $\sqrt{mk}/c$ | Quality factor $\frac{1}{R}\sqrt{L/C}$ |

Radio tuning is literally resonance engineering: a station is selected by adjusting $C$ so that $1/\sqrt{LC}$ matches the carrier frequency, and $Q$ sets the selectivity (bandwidth $\approx \omega_0 / Q$).

**Car suspension.** A quarter-car of mass $m$ on a spring $k$ with shock absorber $c$ should return from a bump quickly *without oscillating*: engineers target (slightly below) critical damping, $c = 2\sqrt{mk}$. Underdamped means a boaty, bouncing ride; overdamped means a harsh, slow-recovering one.

**Pendulum linearization.** The exact pendulum $\theta'' + \frac{g}{\ell}\sin\theta = 0$ is *nonlinear*; for small angles $\sin\theta \approx \theta - \theta^3/6$, and dropping the cubic term gives SHM $\theta'' + \frac{g}{\ell}\theta = 0$ with period $T = 2\pi\sqrt{\ell/g}$, independent of amplitude — the isochronism that made pendulum clocks possible. The linearization is the ODE-level instance of a master idea: *near equilibrium, every smooth system is approximately linear*, with the second-order linear theory as the universal local model.

### Machine Learning: Momentum is a Damped Oscillator

**Heavy-ball ODE.** Polyak's momentum gradient descent $\theta_{k+1} = \theta_k - \eta \nabla \mathcal{L}(\theta_k) + \beta(\theta_k - \theta_{k-1})$ is the discretization of the **heavy-ball ODE**

$$
m\, \theta'' + \gamma\, \theta' = -\nabla \mathcal{L}(\theta)
$$

a ball of mass $m$ rolling on the loss surface with friction $\gamma$. Near a local minimum $\theta^*$, expand $\nabla\mathcal{L}(\theta) \approx H (\theta - \theta^*)$ with Hessian $H \succeq 0$. Diagonalizing $H = U \Lambda U^{T}$ and writing $z = U^{T}(\theta - \theta^*)$ decouples training into independent scalar oscillators — the **normal modes** of the loss landscape:

$$
m\, z_i'' + \gamma\, z_i' + \mu_i z_i = 0, \qquad \mu_i = \text{eigenvalue of } H
$$

Each Hessian eigenvalue $\mu_i$ is a spring constant; sharp directions (large $\mu_i$) ring fast, flat directions (small $\mu_i$) relax slowly. Convergence in mode $i$ is fastest at **critical damping** $\gamma = 2\sqrt{m \mu_i}$: less friction gives oscillatory over/undershoot around the minimum (the familiar zig-zag of high momentum), more friction gives overdamped crawling (the slow mode of Proof-style analysis). Since one $\gamma$ must serve all modes, the binding constraint is the *smallest* relevant eigenvalue — the optimal momentum coefficient critically damps the flattest curvature direction, which is exactly how the classical optimal $\beta = \left(\frac{\sqrt{\kappa} - 1}{\sqrt{\kappa} + 1}\right)^2$ (condition number $\kappa = \mu_{\max}/\mu_{\min}$) arises.

**Nesterov's method as an ODE.** Su, Boyd & Candès (2016) showed Nesterov's accelerated gradient is the discretization of

$$
\theta'' + \frac{3}{t}\, \theta' + \nabla \mathcal{L}(\theta) = 0
$$

— a damped oscillator whose friction coefficient **decays in time** like $3/t$: strong damping early (fast initial progress), vanishing damping late (long, nearly conservative strides). The ODE view explains acceleration's signature oscillations and yields the $O(1/t^2)$ convergence rate by an energy (Lyapunov) argument of exactly the $E$, $dE/dt$ type used above.

**Takeaway for practitioners.** Learning-rate/momentum pairs are a choice of $(m, \gamma)$; loss-curve "ringing" is literal underdamped oscillation in a sharp Hessian eigendirection; and the remedy — more momentum decay or a smaller step — is damping design, the same calculation a suspension engineer does with $c = 2\sqrt{mk}$.
