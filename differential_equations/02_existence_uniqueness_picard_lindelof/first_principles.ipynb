# Topic 02: Existence, Uniqueness, and the Picard–Lindelöf Theorem

## 1. First-Principles Intuition & Motivation

**Phenomenon.** A differential equation is a claim about determinism: the law $y' = f(t, y)$ says the instantaneous rate of change of the state is fully determined by the current time and state. Pair it with a measurement of the present, $y(t_0) = y_0$, and physics, chemistry, epidemiology, and machine learning all make the same leap of faith — that the future is now *uniquely* pinned down.

This leap is not free. Three distinct disasters can occur:

1. **No solution exists** — the model predicts nothing.
2. **Several solutions exist** — the model predicts everything, i.e., nothing useful. The same present is compatible with different futures.
3. **The solution exists but is unstable** — an unmeasurably small error in $y_0$ produces a wildly different prediction after arbitrarily short time.

Hadamard called a problem **well-posed** when all three disasters are excluded: a solution *exists*, is *unique*, and *depends continuously* on the data. This module builds the complete classical theory answering when an IVP is well-posed, with every proof from first principles.

### 1.1 What Fails Without the Theory — Two Warning Shots

**Warning shot 1 (uniqueness fails).** The innocent-looking IVP

$$
y' = 3y^{2/3}, \qquad y(0) = 0
$$

has the solution $y(t) \equiv 0$, but also $y(t) = t^3$, and in fact a *continuum* of solutions: stay at zero for any length of time you like, then depart along a shifted cubic. The right-hand side is continuous everywhere, yet determinism is destroyed. The culprit, as we will prove, is the failure of Lipschitz continuity at $y = 0$, where the slopes of secant lines $3y^{2/3}/y = 3y^{-1/3}$ are unbounded.

**Warning shot 2 (global existence fails).** The IVP

$$
y' = y^2, \qquad y(0) = 1
$$

has the perfectly smooth, even polynomial, right-hand side $f(y) = y^2$. Yet its unique solution $y(t) = 1/(1 - t)$ escapes to infinity as $t \to 1^-$. The solution simply *ceases to exist* after one unit of time — finite-time blow-up. Smoothness of $f$ guarantees local well-posedness, never global.

**Goal.** Find hypotheses on $f$ that are (i) weak enough to cover real models and (ii) strong enough to certify existence, uniqueness, and continuous dependence — and understand precisely which hypothesis blocks which disaster.

### 1.2 Assumptions, Variables, and Notation

| Symbol | Meaning |
| :--- | :--- |
| $t$ | Independent variable (time), $t \in \mathbb{R}$ |
| $y(t)$ | Unknown state; scalar for clarity, but every result holds verbatim in $\mathbb{R}^n$ with $\lvert \cdot \rvert$ replaced by a norm $\lVert \cdot \rVert$ |
| $f(t, y)$ | Right-hand side (vector field), defined on an open set $D \subseteq \mathbb{R}^2$ |
| $(t_0, y_0)$ | Initial data, a point of $D$ |
| $R$ | Closed rectangle $R = \{(t, y) : \lvert t - t_0 \rvert \le a, \ \lvert y - y_0 \rvert \le b\} \subset D$ |
| $M$ | Bound $M = \max_{(t,y) \in R} \lvert f(t, y) \rvert$, finite since $f$ is continuous on the compact set $R$ |
| $L$ | Lipschitz constant of $f$ in the $y$ variable |
| $h$ | Certified half-width of the existence interval, $h = \min(a, b/M)$ |
| $C(I)$ | Space of continuous real-valued functions on a closed interval $I$, with sup norm $\lVert y \rVert_\infty = \sup_{t \in I} \lvert y(t) \rvert$ |

**Standing assumption for the main theorem:** $f$ is continuous on $R$ and Lipschitz continuous in $y$ on $R$. Nothing more — no differentiability of $f$ is ever required.

The strategy in one line: *trade the differential equation (hard: derivatives) for an integral equation (soft: only continuity needed), view its solution as a fixed point of an operator on a function space, and invoke the contraction mapping principle.*

## 2. Rigorous Mathematical Definitions & Theorem Statements

### 2.1 Initial Value Problems and Solutions

**Definition 2.1 (IVP).** Given $f : D \to \mathbb{R}$ continuous on an open set $D \subseteq \mathbb{R}^2$ and $(t_0, y_0) \in D$, the **initial value problem** is

$$
y'(t) = f(t, y(t)), \qquad y(t_0) = y_0.
$$

**Definition 2.2 (Local solution).** A **solution on an interval** $I \ni t_0$ is a differentiable function $\varphi : I \to \mathbb{R}$ with $(t, \varphi(t)) \in D$ for all $t \in I$, satisfying $\varphi'(t) = f(t, \varphi(t))$ on $I$ and $\varphi(t_0) = y_0$. If $I$ is a proper subinterval of the $t$-projection of $D$ we call $\varphi$ a **local** solution; if $I = \mathbb{R}$ (or all of the natural time domain) it is a **global** solution.

**Definition 2.3 (Maximal interval of existence).** A solution $\varphi$ on $I_{\max} = (t_-, t_+)$ is **maximal** if it admits no extension to a solution on a strictly larger interval. Every solution extends to a unique maximal one (Theorem 2.10 below).

Note that any solution is automatically $C^1$: $\varphi'$ equals the continuous function $t \mapsto f(t, \varphi(t))$. If $f$ is $C^k$, bootstrapping shows $\varphi \in C^{k+1}$.

### 2.2 Lipschitz Continuity

**Definition 2.4 (Lipschitz condition).** $f(t, y)$ is **Lipschitz continuous in $y$** on a set $S$ if there exists $L \ge 0$ with

$$
\lvert f(t, y_1) - f(t, y_2) \rvert \le L \, \lvert y_1 - y_2 \rvert \qquad \text{for all } (t, y_1), (t, y_2) \in S.
$$

It is **locally Lipschitz** on an open set $D$ if every point of $D$ has a neighborhood on which such an $L$ exists (the constant may grow from neighborhood to neighborhood), and **globally Lipschitz** if one $L$ works on all of $D$.

**Proposition 2.5 (Bounded derivative implies Lipschitz).** If $\partial f / \partial y$ exists and satisfies $\lvert \partial f(t, y) / \partial y \rvert \le L$ on a set $S$ convex in $y$, then $f$ is Lipschitz in $y$ with constant $L$.

*Proof.* By the mean value theorem in the $y$ variable, for some $\xi$ between $y_1$ and $y_2$,

$$
\lvert f(t, y_1) - f(t, y_2) \rvert = \left\lvert \frac{\partial f}{\partial y}(t, \xi) \right\rvert \lvert y_1 - y_2 \rvert \le L \lvert y_1 - y_2 \rvert. \qquad \blacksquare
$$

The converse fails: $f(y) = \lvert y \rvert$ is globally Lipschitz ($L = 1$) but not differentiable at $0$. In particular every $C^1$ function is locally Lipschitz, so for smooth models the Picard–Lindelöf hypotheses are automatic. The hierarchy is strict:

$$
C^1 \implies \text{locally Lipschitz} \implies \text{continuous},
$$

and $f(y) = 3y^{2/3}$ sits exactly in the gap between the last two: continuous everywhere, Lipschitz on every set bounded away from $y = 0$, but not Lipschitz on any neighborhood of $0$.

### 2.3 The Integral Reformulation, Picard Iterates, and Function Spaces

**Definition 2.6 (Integral equation).** Associated with the IVP is the integral equation

$$
y(t) = y_0 + \int_{t_0}^{t} f(s, y(s)) \, ds.
$$

Its solutions need only be *continuous* a priori — the derivative has been absorbed into the integral. Proof 3.1 shows the two problems are exactly equivalent.

**Definition 2.7 (Picard operator and iterates).** For a continuous function $y$ on an interval $I \ni t_0$, define

$$
T[y]\,(t) = y_0 + \int_{t_0}^{t} f(s, y(s)) \, ds.
$$

The **Picard iterates** are $y_0(t) \equiv y_0$ and $y_{n+1} = T[y_n]$. A solution of the integral equation is precisely a **fixed point** $y = T[y]$.

**Definition 2.8 (Complete metric space; contraction).** A metric space $(X, d)$ is **complete** if every Cauchy sequence converges to a limit in $X$. A map $T : X \to X$ is a **contraction** if there exists $q \in [0, 1)$ with $d(T x, T z) \le q \, d(x, z)$ for all $x, z \in X$.

**Fact.** $C(I)$ with metric $d(y, z) = \lVert y - z \rVert_\infty$ is complete for any closed bounded interval $I$: a uniformly Cauchy sequence of continuous functions converges uniformly, and the uniform limit of continuous functions is continuous. Any closed subset of $C(I)$ — e.g., the ball $\{y : \lVert y - y_0 \rVert_\infty \le b\}$ — is again complete. This is the stage on which the whole theory plays out.

### 2.4 The Main Theorems

**Theorem 2.9 (Banach fixed-point theorem, 1922).** Let $(X, d)$ be a non-empty complete metric space and $T : X \to X$ a contraction with constant $q \lt 1$. Then $T$ has exactly one fixed point $x^*$, and for any starting point $x_0 \in X$ the iterates $x_{n+1} = T x_n$ converge to $x^*$ with the a priori error bound

$$
d(x_n, x^*) \le \frac{q^n}{1 - q} \, d(x_1, x_0).
$$

**Theorem 2.10 (Picard–Lindelöf, local version).** Let $f$ be continuous on the rectangle $R = \{(t,y) : \lvert t - t_0 \rvert \le a, \ \lvert y - y_0 \rvert \le b\}$ with $\lvert f \rvert \le M$ on $R$, and Lipschitz in $y$ on $R$ with constant $L$. Set

$$
h = \min\!\left(a, \frac{b}{M}\right).
$$

Then the IVP $y' = f(t,y)$, $y(t_0) = y_0$ has exactly one solution on $[t_0 - h, t_0 + h]$, and the Picard iterates converge uniformly to it.

The two ingredients play different roles: $M$ controls how fast the solution can *leave the rectangle vertically* (hence the $b/M$ cap on $h$), while $L$ controls how fast nearby candidate solutions can *separate* (making $T$ a contraction).

**Theorem 2.11 (Peano existence theorem, 1886).** If $f$ is merely *continuous* on $R$, the IVP still has at least one solution on $[t_0 - h, t_0 + h]$ with the same $h = \min(a, b/M)$ — but possibly many. (Proof idea: Euler polygon approximations are uniformly bounded and equicontinuous; Arzelà–Ascoli extracts a uniformly convergent subsequence whose limit solves the integral equation. Compactness replaces contraction, which is why uniqueness is lost.)

### 2.5 Grönwall, Continuation, and Continuous Dependence

**Theorem 2.12 (Grönwall's inequality, integral form).** Let $u : [t_0, t_1] \to [0, \infty)$ be continuous and suppose there are constants $A \ge 0$, $L \ge 0$ with

$$
u(t) \le A + L \int_{t_0}^{t} u(s) \, ds \qquad \text{for all } t \in [t_0, t_1].
$$

Then

$$
u(t) \le A \, e^{L (t - t_0)} \qquad \text{for all } t \in [t_0, t_1].
$$

In words: a quantity that grows at most in proportion to its own accumulated size grows at most exponentially. This single lemma delivers uniqueness ($A = 0$) and continuous dependence ($A$ equal to the initial discrepancy) in one stroke.

**Theorem 2.13 (Continuation / maximal interval).** Let $f$ be continuous and locally Lipschitz in $y$ on an open set $D$. Then every solution extends uniquely to a maximal interval $(t_-, t_+)$, and as $t \to t_+^-$ (if $t_+ \lt \infty$) the trajectory $(t, y(t))$ **leaves every compact subset** of $D$. For $D = \mathbb{R}^2$ this means: either $t_+ = \infty$, or $\lvert y(t) \rvert \to \infty$ in finite time (blow-up). A solution cannot simply "stop" in the interior of the domain.

**Theorem 2.14 (Continuous dependence on initial conditions).** Under the hypotheses of Picard–Lindelöf, if $y_1, y_2$ solve the ODE with initial values $y_1(t_0), y_2(t_0)$ then for all $t$ in the common interval of existence

$$
\lvert y_1(t) - y_2(t) \rvert \le \lvert y_1(t_0) - y_2(t_0) \rvert \, e^{L \lvert t - t_0 \rvert}.
$$

Thus on any *fixed* compact time interval the solution map $y_0 \mapsto y(t; y_0)$ is Lipschitz continuous — Hadamard's third requirement. Together, Theorems 2.10, 2.12, and 2.14 certify the IVP as **well-posed**.

## 3. Step-by-Step Mathematical Proofs & Derivations

### Proof 3.1 — Equivalence of the IVP and the Integral Equation

**Claim.** Let $f$ be continuous on $D$ and $I \ni t_0$ an interval. A function $\varphi : I \to \mathbb{R}$ is a solution of the IVP $y' = f(t, y)$, $y(t_0) = y_0$ if and only if $\varphi$ is continuous and satisfies

$$
\varphi(t) = y_0 + \int_{t_0}^{t} f(s, \varphi(s)) \, ds \qquad \text{for all } t \in I.
$$

**Step 1 (IVP $\Rightarrow$ integral equation).** Suppose $\varphi$ solves the IVP. Then $\varphi$ is differentiable, hence continuous, so $s \mapsto f(s, \varphi(s))$ is continuous on $I$ (composition of continuous maps). By the Fundamental Theorem of Calculus, Part 2, we may integrate the identity $\varphi'(s) = f(s, \varphi(s))$ from $t_0$ to $t$:

$$
\varphi(t) - \varphi(t_0) = \int_{t_0}^{t} \varphi'(s) \, ds = \int_{t_0}^{t} f(s, \varphi(s)) \, ds.
$$

Using $\varphi(t_0) = y_0$ gives the integral equation.

**Step 2 (Integral equation $\Rightarrow$ IVP).** Suppose $\varphi$ is continuous and satisfies the integral equation. The integrand $g(s) = f(s, \varphi(s))$ is continuous, so by the Fundamental Theorem of Calculus, Part 1, the function $t \mapsto \int_{t_0}^{t} g(s)\,ds$ is differentiable with derivative $g(t)$. Hence $\varphi$ is differentiable and

$$
\varphi'(t) = f(t, \varphi(t)).
$$

Evaluating the integral equation at $t = t_0$ gives $\varphi(t_0) = y_0 + 0 = y_0$.

**Step 3 (Why this matters).** The integral form requires only *continuity* of the candidate $\varphi$, so we may search for solutions in the complete metric space $C(I)$ rather than in $C^1(I)$ — and the regularity ($\varphi \in C^1$) is returned to us for free at the end.

$$
\boxed{\text{IVP} \iff \text{integral equation } \varphi = T[\varphi] \text{ on } C(I)}
$$

### Proof 3.2 — Banach Fixed-Point Theorem

**Claim.** If $(X, d)$ is a non-empty complete metric space and $T : X \to X$ satisfies $d(Tx, Tz) \le q\, d(x, z)$ with $0 \le q \lt 1$, then $T$ has a unique fixed point, and $x_{n+1} = T x_n$ converges to it from any $x_0$.

**Step 1 (Geometric decay of consecutive gaps).** Pick any $x_0 \in X$ and set $x_{n+1} = T x_n$. By induction,

$$
d(x_{n+1}, x_n) = d(T x_n, T x_{n-1}) \le q \, d(x_n, x_{n-1}) \le \cdots \le q^n \, d(x_1, x_0).
$$

**Step 2 (The iterates are Cauchy).** For $m \gt n$, the triangle inequality and the geometric series give

$$
d(x_m, x_n) \le \sum_{k=n}^{m-1} d(x_{k+1}, x_k) \le \sum_{k=n}^{m-1} q^k \, d(x_1, x_0) \le \frac{q^n}{1 - q} \, d(x_1, x_0).
$$

Since $q \lt 1$, the right side tends to $0$ as $n \to \infty$, uniformly in $m$. Hence $(x_n)$ is Cauchy.

**Step 3 (The limit exists and is a fixed point).** Completeness of $X$ yields $x^* = \lim_n x_n \in X$. A contraction is continuous (it is Lipschitz with constant $q$), so

$$
T x^* = T\left(\lim_n x_n\right) = \lim_n T x_n = \lim_n x_{n+1} = x^*.
$$

**Step 4 (Uniqueness).** If $Tx^* = x^*$ and $Tz^* = z^*$, then

$$
d(x^*, z^*) = d(T x^*, T z^*) \le q \, d(x^*, z^*),
$$

so $(1 - q)\, d(x^*, z^*) \le 0$. Since $1 - q \gt 0$ and $d \ge 0$, we get $d(x^*, z^*) = 0$, i.e., $x^* = z^*$.

**Step 5 (Error bound).** Letting $m \to \infty$ in Step 2's estimate gives the a priori bound $d(x_n, x^*) \le \frac{q^n}{1-q} d(x_1, x_0)$: convergence is at least geometric, with computable rate.

$$
\boxed{\exists! \, x^* \in X : T x^* = x^*, \quad d(x_n, x^*) \le \frac{q^n}{1 - q}\, d(x_1, x_0)}
$$

### Proof 3.3 — Picard–Lindelöf Theorem (Part I: the operator is well-defined)

**Setup.** Let $R$, $M$, $L$, and $h = \min(a, b/M)$ be as in Theorem 2.10. Write $I = [t_0 - h, t_0 + h]$ and define the closed set

$$
X = \{\, y \in C(I) : \lvert y(t) - y_0 \rvert \le b \ \text{ for all } t \in I \,\},
$$

the "tube" of continuous functions whose graphs stay inside the rectangle $R$. As a closed subset of the complete space $(C(I), \lVert \cdot \rVert_\infty)$, $X$ is itself a complete metric space, and it is non-empty (the constant function $y_0$ belongs to it).

**Step 1 ($T$ maps $X$ into $X$).** For $y \in X$ and $t \in I$, the point $(s, y(s))$ lies in $R$ for every $s$ between $t_0$ and $t$, so $\lvert f(s, y(s)) \rvert \le M$ and

$$
\lvert T[y]\,(t) - y_0 \rvert = \left\lvert \int_{t_0}^{t} f(s, y(s)) \, ds \right\rvert \le M \, \lvert t - t_0 \rvert \le M h \le M \cdot \frac{b}{M} = b.
$$

Moreover $T[y]$ is continuous (indeed $C^1$) by the Fundamental Theorem of Calculus. Hence $T[y] \in X$. This is exactly where the mysterious $h \le b/M$ earns its keep: it prevents the integral from dragging the candidate solution out of the rectangle where our bounds on $f$ hold.

**Step 2 ($T$ is Lipschitz on $X$ with constant $Lh$).** For $y, z \in X$ and $t \in I$,

$$
\lvert T[y]\,(t) - T[z]\,(t) \rvert \le \left\lvert \int_{t_0}^{t} \lvert f(s, y(s)) - f(s, z(s)) \rvert \, ds \right\rvert \le L \left\lvert \int_{t_0}^{t} \lvert y(s) - z(s) \rvert \, ds \right\rvert \le L h \, \lVert y - z \rVert_\infty.
$$

Taking the supremum over $t \in I$: $\lVert T[y] - T[z] \rVert_\infty \le L h \, \lVert y - z \rVert_\infty$.

If $Lh \lt 1$ we could stop here and invoke Banach directly — but that would force us to shrink $h$ further to below $1/L$. The next cell shows two ways to avoid this unnecessary sacrifice.

### Proof 3.3 — Picard–Lindelöf Theorem (Part II: contraction without shrinking the interval)

**Step 3 (Iterated estimate: $T^n$ is a contraction for large $n$).** Refine Step 2 by keeping the estimate *pointwise*. Claim: for all $n \ge 1$ and $t \in I$,

$$
\lvert T^n[y]\,(t) - T^n[z]\,(t) \rvert \le \frac{L^n \lvert t - t_0 \rvert^n}{n!} \, \lVert y - z \rVert_\infty.
$$

*Induction.* The case $n = 1$ is Step 2's pointwise bound $L \lvert t - t_0 \rvert \lVert y - z \rVert_\infty$. Assuming the bound for $n$, for $t \ge t_0$ (the case $t \le t_0$ is symmetric):

$$
\lvert T^{n+1}[y]\,(t) - T^{n+1}[z]\,(t) \rvert \le L \int_{t_0}^{t} \lvert T^n[y]\,(s) - T^n[z]\,(s) \rvert \, ds \le L \int_{t_0}^{t} \frac{L^n (s - t_0)^n}{n!} \, ds \, \lVert y - z \rVert_\infty = \frac{L^{n+1} (t - t_0)^{n+1}}{(n+1)!} \, \lVert y - z \rVert_\infty.
$$

Taking suprema over $t \in I$,

$$
\lVert T^n[y] - T^n[z] \rVert_\infty \le \frac{(L h)^n}{n!} \, \lVert y - z \rVert_\infty.
$$

Since $(Lh)^n / n! \to 0$ (factorials beat exponentials), some power $T^N$ is a contraction on $X$.

**Step 4 (A fixed point of $T^N$ is a fixed point of $T$).** By Banach (Proof 3.2), $T^N$ has a unique fixed point $y^* \in X$. Then $T^N (T y^*) = T (T^N y^*) = T y^*$, so $T y^*$ is *also* a fixed point of $T^N$; uniqueness forces $T y^* = y^*$. Conversely any fixed point of $T$ is a fixed point of $T^N$, so $T$ has exactly one fixed point, namely $y^*$.

**Step 5 (Conclusion).** By Proof 3.1, $y^*$ is the unique solution of the IVP on $I = [t_0 - h, t_0 + h]$, and the Picard iterates $y_{n} = T^n[y_0]$ converge uniformly to it.

*(Alternative to Steps 3–4: the **Bielecki weighted norm** $\lVert y \rVert_L = \sup_{t \in I} e^{-2L\lvert t - t_0 \rvert} \lvert y(t) \rvert$ makes $T$ itself a contraction with constant $\frac{1}{2}$ on the full interval — worked in detail as Exercise L3.4.)*

$$
\boxed{\text{The IVP has exactly one solution on } [t_0 - h, t_0 + h], \ h = \min(a, b/M)}
$$

### Proof 3.4 — Grönwall's Inequality

**Claim.** If $u \ge 0$ is continuous on $[t_0, t_1]$ and $u(t) \le A + L \int_{t_0}^{t} u(s)\,ds$ with $A, L \ge 0$, then $u(t) \le A e^{L(t - t_0)}$.

**Step 1 (Name the accumulated quantity).** Let

$$
V(t) = A + L \int_{t_0}^{t} u(s) \, ds.
$$

Then $V$ is $C^1$, $V(t_0) = A$, the hypothesis reads $u(t) \le V(t)$, and

$$
V'(t) = L\, u(t) \le L\, V(t).
$$

**Step 2 (Integrating-factor trick).** Multiply the differential inequality $V' - LV \le 0$ by the positive factor $e^{-L(t - t_0)}$:

$$
\frac{d}{dt}\left[ V(t) \, e^{-L(t - t_0)} \right] = \left( V'(t) - L V(t) \right) e^{-L(t - t_0)} \le 0.
$$

**Step 3 (Monotonicity).** The function $t \mapsto V(t) e^{-L(t - t_0)}$ is non-increasing, so for $t \in [t_0, t_1]$,

$$
V(t) \, e^{-L(t - t_0)} \le V(t_0) = A \quad \Longrightarrow \quad u(t) \le V(t) \le A \, e^{L (t - t_0)}.
$$

A mirrored argument (or the substitution $t \mapsto 2t_0 - t$) gives the two-sided form with $e^{L \lvert t - t_0 \rvert}$ for $t \le t_0$.

$$
\boxed{u(t) \le A + L \int_{t_0}^{t} u \implies u(t) \le A \, e^{L (t - t_0)}}
$$

The subtlety worth savoring: we could not simply "differentiate the hypothesis," because an inequality between functions says nothing about their derivatives. Introducing $V$ converts the integral inequality into a *differential* inequality for a genuinely differentiable object — a move used throughout analysis.

### Proof 3.5 — Uniqueness and Continuous Dependence via Grönwall

**Claim.** Let $f$ be continuous and Lipschitz in $y$ with constant $L$ on a region containing the graphs of two solutions $y_1, y_2$ of $y' = f(t, y)$ on $[t_0, t_1]$. Then

$$
\lvert y_1(t) - y_2(t) \rvert \le \lvert y_1(t_0) - y_2(t_0) \rvert \, e^{L (t - t_0)}.
$$

In particular, equal initial values force $y_1 \equiv y_2$ (uniqueness), and the solution map $y_0 \mapsto y(t; y_0)$ is Lipschitz (continuous dependence).

**Step 1 (Subtract the integral forms).** By Proof 3.1, each solution satisfies its integral equation. Subtracting,

$$
y_1(t) - y_2(t) = y_1(t_0) - y_2(t_0) + \int_{t_0}^{t} \left[ f(s, y_1(s)) - f(s, y_2(s)) \right] ds.
$$

**Step 2 (Estimate with the Lipschitz condition).** Set $u(t) = \lvert y_1(t) - y_2(t) \rvert$ and $A = u(t_0)$. The triangle inequality gives

$$
u(t) \le A + \int_{t_0}^{t} \lvert f(s, y_1(s)) - f(s, y_2(s)) \rvert \, ds \le A + L \int_{t_0}^{t} u(s) \, ds.
$$

**Step 3 (Apply Grönwall).** Proof 3.4 yields exactly $u(t) \le A e^{L(t - t_0)}$, and the mirrored version handles $t \lt t_0$.

**Step 4 (Read off the corollaries).**

- *Uniqueness:* if $y_1(t_0) = y_2(t_0)$ then $A = 0$, so $u \equiv 0$: two solutions through the same point coincide on their whole common interval. Geometrically, **solution curves never cross** — through each point of the domain passes exactly one trajectory.
- *Continuous dependence:* on a fixed interval $\lvert t - t_0 \rvert \le \tau$ the flow map is Lipschitz with constant $e^{L\tau}$. Predictions degrade at a controlled, at-worst-exponential rate as measurement error grows — Hadamard's third condition.
- *Sharpness:* for $y' = Ly$ the bound is attained with equality, so the exponential rate cannot be improved in general.

$$
\boxed{\lvert y_1(t) - y_2(t) \rvert \le \lvert y_1(t_0) - y_2(t_0) \rvert \, e^{L \lvert t - t_0 \rvert}}
$$

### Proof 3.6 — Finite-Time Blow-Up: $y' = y^2$, $y(0) = 1$

**Claim.** The unique maximal solution is $y(t) = \dfrac{1}{1 - t}$ on the maximal interval $I_{\max} = (-\infty, 1)$.

**Step 1 (Local well-posedness).** $f(y) = y^2$ is $C^1$, hence locally Lipschitz everywhere (Proposition 2.5: on $\lvert y - 1 \rvert \le b$ one may take $L = 2(1 + b)$). Picard–Lindelöf applies at every point, so through $(0, 1)$ passes exactly one maximal solution.

**Step 2 (Solve by separation, justified by uniqueness).** Since $y(0) = 1 \gt 0$ and (by uniqueness against the equilibrium $y \equiv 0$) the solution can never touch $0$, we may divide by $y^2$ on the whole interval of existence:

$$
\int_0^t \frac{y'(s)}{y(s)^2} \, ds = \int_0^t 1 \, ds \quad \Longrightarrow \quad -\frac{1}{y(t)} + \frac{1}{y(0)} = t \quad \Longrightarrow \quad y(t) = \frac{1}{1 - t}.
$$

**Step 3 (Identify the maximal interval).** The formula defines a solution on $(-\infty, 1)$ and on $(1, \infty)$, but only the branch containing $t_0 = 0$ counts: $I_{\max} = (-\infty, 1)$. As $t \to 1^-$, $y(t) \to +\infty$: the trajectory leaves every compact subset of the plane, exactly as the continuation theorem (2.13) demands — the solution does not "stop" arbitrarily, it *escapes to infinity*.

**Step 4 (Why blow-up was inevitable — a comparison argument).** For any solution with $y(0) = y_0 \gt 0$, as long as $y \ge y_0$ we have $y' = y^2 \ge y_0 y$, so $y(t) \ge y_0 e^{y_0 t}$: growth is at least exponential, which feeds back into $y' = y^2$ super-exponentially. Quantitatively, the general solution $y(t) = y_0/(1 - y_0 t)$ blows up at $t^* = 1/y_0$: *larger initial data die sooner*. The lesson: local Lipschitz gives local existence only; global existence needs either a global Lipschitz bound (note $y^2$ has unbounded slope) or an a priori bound trapping the solution in a compact set.

$$
\boxed{y(t) = \frac{1}{1 - t}, \qquad I_{\max} = (-\infty, 1), \qquad y(t) \to +\infty \text{ as } t \to 1^-}
$$

### Proof 3.7 — Non-Uniqueness: $y' = 3y^{2/3}$, $y(0) = 0$

**Claim.** The IVP has infinitely many solutions on $\mathbb{R}$; the Lipschitz hypothesis — and only that hypothesis — fails.

**Step 1 (Two obvious solutions).** The constant $y \equiv 0$ solves the IVP. So does $y(t) = t^3$: indeed $y' = 3t^2$ and $3 y^{2/3} = 3 (t^3)^{2/3} = 3t^2$ for all real $t$ (the cube root is odd, so $(t^3)^{2/3} = t^2$ with the real-root convention).

**Step 2 (A two-parameter family by waiting and gluing).** For any $c_1 \le 0 \le c_2$ define

$$
y_{c_1, c_2}(t) =
\begin{cases}
(t - c_1)^3, & t \lt c_1, \\
0, & c_1 \le t \le c_2, \\
(t - c_2)^3, & t \gt c_2.
\end{cases}
$$

On each open piece the ODE holds by Step 1 (translation invariance). At the gluing points $t = c_i$ both one-sided derivatives equal $0$ and both one-sided limits of $3y^{2/3}$ equal $0$, so $y_{c_1,c_2}$ is $C^1$ and satisfies the ODE *everywhere*, with $y(0) = 0$. Distinct pairs $(c_1, c_2)$ give distinct solutions — an uncountable family. The system can "sit" at the unstable rest state for an arbitrary time and then spontaneously depart: determinism fails.

**Step 3 (Locate the failed hypothesis).** $f(y) = 3y^{2/3}$ is continuous (so Peano correctly predicts existence, which we indeed have). But for $y \gt 0$,

$$
\frac{\lvert f(y) - f(0) \rvert}{\lvert y - 0 \rvert} = \frac{3 y^{2/3}}{y} = \frac{3}{y^{1/3}} \longrightarrow \infty \quad \text{as } y \to 0^+,
$$

so no Lipschitz constant exists on any neighborhood of $y = 0$. Picard–Lindelöf never applied at the origin; away from $y = 0$ the function is $C^1$ and uniqueness holds, which is why all solutions agree in shape and differ only in their departure times from $0$.

**Step 4 (The Picard iterates even fail to detect the interesting solutions).** Starting from $y_0(t) \equiv 0$: $y_1(t) = 0 + \int_0^t 3 \cdot 0^{2/3} ds = 0$, and inductively $y_n \equiv 0$. The iteration converges instantly — to only one of the infinitely many solutions. Contraction was the engine of both convergence *and* uniqueness; without Lipschitz, fixed points may be plentiful and the iteration blind to them.

$$
\boxed{\text{Continuity} \Rightarrow \text{existence (Peano); the failure of Lipschitz at } y = 0 \text{ destroys uniqueness}}
$$

## 4. Computational & Algorithmic Insights

### 4.1 Picard Iteration as an Algorithm

The proof of Picard–Lindelöf *is* an algorithm — each iterate is an explicit integral of the previous one:

```text
PICARD-ITERATE(f, t0, y0, n_max):
    y(t) := y0                          # constant initial guess
    for n = 1 .. n_max:
        y(t) := y0 + Integrate(f(s, y(s)), s = t0 .. t)   # symbolic or quadrature
    return y(t)
```

**Worked symbolic run for $y' = y$, $y(0) = 1$** (so $T[y]\,(t) = 1 + \int_0^t y(s)\,ds$):

$$
y_0 = 1, \quad y_1 = 1 + t, \quad y_2 = 1 + t + \frac{t^2}{2}, \quad y_3 = 1 + t + \frac{t^2}{2} + \frac{t^3}{6}, \quad \dots
$$

By induction $y_n(t) = \sum_{k=0}^{n} t^k / k!$ — the Picard iterates are exactly the Taylor partial sums of $e^t$. The contraction proof thus *re-derives the exponential series from the ODE alone*, with no prior knowledge of $e^t$. In a CAS (`sympy.integrate` in a loop) this runs verbatim; with numerical quadrature on a grid it becomes a collocation-type method (Picard iteration on Chebyshev grids underlies modern spectral integrators and the parallel-in-time "Parareal/RIDC" family).

### 4.2 Convergence Rate — Factorial Beats Geometric

From Proof 3.3, Step 3, the distance from the $n$-th iterate to the true solution obeys

$$
\lVert y_n - y^* \rVert_\infty \le \frac{(L h)^n}{n!} \, \lVert y_1 - y_0 \rVert_\infty \cdot C,
$$

with the factorial-decay factor $(Lh)^n/n!$. Two consequences:

1. **Eventual super-geometric convergence.** For any fixed $Lh$ — even $Lh = 100$ — the ratio of successive error bounds is $Lh/(n+1) \to 0$: once $n \gt Lh$ the error collapses faster than any geometric sequence. Compare Banach's generic $q^n$: the ODE structure (nested integrals shrink volumes by $1/n!$) is strictly better than the abstract theory promises.
2. **Practical trade-off.** Each iterate integrates over the whole interval, and symbolic iterates grow in complexity (for $y' = t + y$ the $n$-th iterate carries $n{+}1$ polynomial terms). Numerically, one Picard sweep costs a full-interval quadrature, whereas a Runge–Kutta step is local. This is why time-steppers won historically — but Picard's full-interval structure is *embarrassingly parallel in time*, driving its modern revival.

A useful mental table for $Lh = 2$:

| $n$ | $(Lh)^n / n!$ |
| :--- | :--- |
| 1 | $2$ |
| 3 | $1.33$ |
| 5 | $0.27$ |
| 8 | $0.0063$ |
| 12 | $8.6 \times 10^{-6}$ |

The bound first *grows* (while $n \lt Lh$), then dies factorially — exactly the transient one observes in practice.

### 4.3 Why Numerical Solvers Secretly Assume Lipschitz — and the Condition-Number Analogy

**Every convergence theorem for ODE solvers routes through $L$.** The global error bound for Euler's method with step $\Delta t$ on $[t_0, t_0 + \tau]$ is

$$
\max_k \lvert y(t_k) - y_k \rvert \le \frac{M_2 \, \Delta t}{2L} \left( e^{L \tau} - 1 \right), \qquad M_2 = \max \lvert y'' \rvert,
$$

proved by exactly the discrete analogue of Grönwall's inequality: the local truncation errors $O(\Delta t^2)$ are amplified by at most $e^{L\tau}$ as they propagate. No Lipschitz constant, no error bound — on non-Lipschitz problems (e.g., $y' = 3y^{2/3}$ near $0$) a solver silently picks *one* of the many solutions, determined by floating-point noise, and adaptive step-size controllers can chatter or stall near blow-up times.

**Condition-number analogy.** In linear algebra, $\kappa(A) = \lVert A \rVert \lVert A^{-1} \rVert$ bounds how much relative input error inflates the output of solving $Ax = b$. For an IVP on a horizon $\tau$, the Grönwall factor

$$
\kappa_{\text{IVP}}(\tau) \approx e^{L \tau}
$$

plays the same role: it bounds the output perturbation per unit input perturbation of $y_0$. A problem with $L\tau \approx 40$ has $\kappa \approx e^{40} \approx 2 \times 10^{17}$ and is numerically hopeless in double precision no matter how good the solver — the *problem*, not the algorithm, is ill-conditioned. (For contractive dynamics, one-sided Lipschitz constants $\mu \lt 0$ give $e^{\mu \tau} \lt 1$: errors *shrink*, the regime exploited by stiff solvers.) This is Hadamard's continuous-dependence condition wearing its quantitative, floating-point clothes.

## 5. Real-World Physics & AI/ML Applications

### 5.1 Determinism in Classical Mechanics — Norton's Dome

Newtonian mechanics is often described as perfectly deterministic, *because* its equations of motion are ODEs. The fine print: determinism holds only where Picard–Lindelöf applies. **Norton's dome** (Norton, 2003) is a frictionless dome shaped so that a point mass at the apex obeys

$$
\frac{d^2 r}{dt^2} = \sqrt{r},
$$

with $r$ the arc-length from the apex and initial data $r(0) = 0$, $r'(0) = 0$. The function $\sqrt{r}$ is continuous but *not Lipschitz at $r = 0$* — the same defect as $3y^{2/3}$. Direct substitution verifies two solutions:

$$
r(t) \equiv 0 \qquad \text{and} \qquad r(t) = \frac{t^4}{144} \ \ \left(\text{check: } r'' = \frac{12 t^2}{144} = \frac{t^2}{12} = \sqrt{t^4/144}\right),
$$

and by time-translation a whole family that rests at the apex for an arbitrary time $T$ and then spontaneously slides off in an arbitrary direction — with no cause, no force imbalance, no hidden perturbation. Newton's laws alone do not decide. The moral is precise, not philosophical hand-waving: *classical determinism is a theorem about Lipschitz force laws, not about mechanics per se.* Realistic interaction potentials are smooth, hence locally Lipschitz, hence deterministic; Norton's dome lives exactly on the mathematical boundary this module maps.

### 5.2 Neural ODEs — Well-Posedness, Non-Crossing Flows, and Why Augmentation Exists

A **Neural ODE** (Chen et al., 2018) replaces a residual network's discrete layers by a learned vector field:

$$
\frac{d\mathbf{z}}{dt} = f_\theta(\mathbf{z}, t), \qquad \mathbf{z}(0) = \mathbf{x}, \qquad \text{output} = \mathbf{z}(1).
$$

**Well-posedness is an architecture property.** If $f_\theta$ is built from linear layers and $1$-Lipschitz activations ($\tanh$, ReLU, sigmoid scaled), then $f_\theta$ is globally Lipschitz in $\mathbf{z}$ with

$$
L \le \prod_{i} \lVert W_i \rVert_2,
$$

the product of spectral norms. Picard–Lindelöf (vector version, verbatim) then guarantees the forward pass is the unique solution of a well-posed IVP, and Grönwall bounds its sensitivity: $\lVert \mathbf{z}_1(t) - \mathbf{z}_2(t) \rVert \le \lVert \mathbf{x}_1 - \mathbf{x}_2 \rVert e^{Lt}$ — simultaneously a robustness certificate and an explanation of training instability when spectral norms are large. **Spectral normalization** ($W \mapsto W / \sigma_{\max}(W)$) is thus not a heuristic here; it is the Picard–Lindelöf hypothesis, enforced by construction.

**Uniqueness has teeth: flows cannot cross.** By Proof 3.5, two trajectories that agree at one time agree at all times. Hence the time-$1$ flow map $\Phi : \mathbf{x} \mapsto \mathbf{z}(1)$ is injective, and (running time backwards, again well-posed) its inverse is continuous: $\Phi$ is a **homeomorphism**. Consequences:

- A $1$-D Neural ODE **cannot represent** $g(x) = -x$: a homeomorphism of $\mathbb{R}$ is monotone, and continuous non-crossing trajectories from $-1 \mapsto 1$ and $1 \mapsto -1$ would have to intersect (Exercise L2.6 works this fully).
- **Augmented Neural ODEs** (Dupont et al., 2019) fix this by lifting to $\mathbb{R}^{d + p}$: trajectories that would cross in $\mathbb{R}^d$ pass around each other in the extra dimensions — a *mathematical necessity* dictated by the uniqueness theorem, not an engineering trick.
- The same non-crossing property makes continuous normalizing flows *bijective by construction*, which is precisely why they admit exact log-density computation via the instantaneous change-of-variables formula.

### 5.3 Uniqueness in Epidemiological Forecasting

The SIR model on a population of size $N$,

$$
S' = -\beta \frac{S I}{N}, \qquad I' = \beta \frac{S I}{N} - \gamma I, \qquad R' = \gamma I,
$$

has a polynomial right-hand side — $C^\infty$, hence locally Lipschitz — so through every initial condition passes exactly one trajectory. Moreover the region $\Delta = \{S, I, R \ge 0, \ S + I + R = N\}$ is compact and invariant (on its boundary faces the vector field points inward or along the face), so by the continuation theorem solutions never blow up: **the SIR IVP is globally well-posed**. This is what licenses the daily practice of epidemic forecasting:

- *Determinism:* today's $(S_0, I_0, R_0)$ plus the parameters determine one forecast, so disagreements between model runs must come from parameters/data, never from the mathematics.
- *Continuous dependence with a quantitative horizon:* Grönwall with the Lipschitz constant of the SIR field on $\Delta$ (roughly $L \approx \beta + \gamma$ in the early phase) bounds forecast divergence by $e^{L t}$ — formalizing why two-week forecasts are informative while six-month forecasts from noisy case counts are not: an initial-data uncertainty $\epsilon$ can grow to $\epsilon e^{L\tau}$, and the horizon at which this exceeds tolerance $\eta$ is $\tau^* = L^{-1} \ln(\eta / \epsilon)$ — *logarithmic* in the data quality. Halving measurement error buys only $L^{-1}\ln 2$ extra days of trustworthy forecast.
- *Ensemble spread interpretation:* ensembles of trajectories from perturbed initial conditions cannot cross; they form nested tubes, so ensemble quantiles remain ordered — a structural fact used (often unknowingly) by forecast-aggregation pipelines.
