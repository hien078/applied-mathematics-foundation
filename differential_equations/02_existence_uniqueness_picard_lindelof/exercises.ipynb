## Level 0 — Concept Check

### Problem L0.1: Is Lipschitz Continuity Needed for Existence?

**Problem Statement:** A colleague claims: "If $f(t, y)$ is not Lipschitz in $y$, the IVP $y' = f(t,y)$, $y(t_0) = y_0$ has no solution." Is this correct? State precisely which hypothesis buys which conclusion in the classical well-posedness theorems, and give a one-line counterexample to the claim.

**Intuition:** Two different theorems are in play — Peano's and Picard–Lindelöf's — and they charge different prices for different goods.

**Solution:**

The claim is **false**. The division of labor is:

1. **Peano's existence theorem:** if $f$ is merely *continuous* on a rectangle around $(t_0, y_0)$, then at least one local solution exists. No Lipschitz condition is involved.
2. **Picard–Lindelöf:** if $f$ is continuous *and* Lipschitz in $y$ on the rectangle, then exactly one local solution exists. The Lipschitz condition is the extra hypothesis that upgrades existence to existence-plus-uniqueness.

**Counterexample to the claim:** $f(y) = 3y^{2/3}$ with $y(0) = 0$ is continuous but not Lipschitz near $y = 0$, yet solutions exist in abundance: $y \equiv 0$ and $y = t^3$ both solve the IVP. Far from having *no* solution, the problem has *infinitely many* — the failure of Lipschitz threatens uniqueness, never existence.

The correct logical ladder:

$$
\boxed{\text{continuity} \Rightarrow \text{existence (Peano)}, \qquad \text{continuity} + \text{Lipschitz} \Rightarrow \text{existence and uniqueness (Picard–Lindelöf)}}
$$

> **Key Takeaway:** Lipschitz continuity is the price of uniqueness, not of existence; dropping it can only multiply solutions, not annihilate them.

### Problem L0.2: The Integral-Equation Reformulation

**Problem Statement:** Write the IVP $y' = t^2 + \cos y$, $y(1) = 5$ as an equivalent integral equation. Explain (a) why every continuous solution of the integral equation is automatically $C^1$, and (b) why this reformulation is the key that unlocks the fixed-point proof.

**Intuition:** Integration is a smoothing operation; differentiation is a roughening one. The reformulation moves the problem to friendlier terrain.

**Solution:**

**The integral equation.** Integrating $y'(s) = s^2 + \cos y(s)$ from $1$ to $t$ and using $y(1) = 5$:

$$
y(t) = 5 + \int_{1}^{t} \left( s^2 + \cos y(s) \right) ds.
$$

**(a) Automatic regularity.** Suppose $y$ is merely continuous and satisfies this equation. Then the integrand $g(s) = s^2 + \cos y(s)$ is continuous (composition and sum of continuous functions). By the Fundamental Theorem of Calculus, Part 1, the right-hand side is differentiable in $t$ with derivative $g(t)$. Since the left side equals the right side, $y$ is differentiable and $y'(t) = t^2 + \cos y(t)$ — so $y'$ is itself continuous, i.e., $y \in C^1$. Regularity is not assumed; it is *produced* by the equation.

**(b) Why this unlocks the proof.** The differential form demands candidates from $C^1$ and involves the non-continuous-in-$\lVert \cdot \rVert_\infty$ operation of differentiation. The integral form defines an operator

$$
T[y]\,(t) = 5 + \int_1^t \left( s^2 + \cos y(s) \right) ds
$$

that maps the *complete* metric space $C(I)$ (sup norm) into itself, and integration only improves estimates: a sup-norm bound on $y - z$ yields a sup-norm bound on $T[y] - T[z]$ shrunk by the factor $L \lvert t - 1 \rvert$. Solutions of the IVP are exactly the fixed points $y = T[y]$, so the Banach fixed-point theorem applies.

$$
\boxed{y(t) = 5 + \int_1^t \left( s^2 + \cos y(s) \right) ds}
$$

> **Key Takeaway:** Recasting derivative constraints as integral constraints trades an unbounded operation for a bounded one, and turns "solve an ODE" into "find a fixed point in a complete space."

### Problem L0.3: Identifying Lipschitz Constants

**Problem Statement:** For each function, decide whether it is globally Lipschitz in $y$ on $\mathbb{R}$, and if so find the best (smallest) constant: (a) $f(t, y) = 3y + \sin t$; (b) $f(t, y) = \sin(5y) + t^2$; (c) $f(y) = y^2$.

**Intuition:** For differentiable functions, the Lipschitz constant is the largest slope in the $y$ direction; scan $\lvert \partial f / \partial y \rvert$ over the domain.

**Solution:**

**(a) $f(t,y) = 3y + \sin t$.** Directly,

$$
\lvert f(t, y_1) - f(t, y_2) \rvert = \lvert 3y_1 - 3y_2 \rvert = 3 \lvert y_1 - y_2 \rvert,
$$

so $f$ is globally Lipschitz with $L = 3$, and no smaller constant works (equality holds for every pair). The $\sin t$ term cancels — Lipschitz-in-$y$ ignores the $t$ dependence entirely.

**(b) $f(t,y) = \sin(5y) + t^2$.** Here $\partial f / \partial y = 5 \cos(5y)$, so $\lvert \partial f / \partial y \rvert \le 5$ everywhere; by the mean value theorem $f$ is globally Lipschitz with $L = 5$. The bound is attained in the limit at points where $\cos(5y) = \pm 1$ (e.g., near $y = 0$), so $L = 5$ is best possible.

**(c) $f(y) = y^2$.** For $y_2 = -y_1 = n$:

$$
\frac{\lvert f(n) - f(-n) \rvert}{\lvert n - (-n) \rvert} = \frac{0}{2n} = 0 \quad \text{— too kind; take instead } y_1 = n, \ y_2 = n + 1:
\qquad \frac{\lvert (n+1)^2 - n^2 \rvert}{1} = 2n + 1 \to \infty.
$$

The difference quotients are unbounded, so **no global Lipschitz constant exists**. However, on any strip $\lvert y \rvert \le K$ we have $\lvert y_1^2 - y_2^2 \rvert = \lvert y_1 + y_2 \rvert \lvert y_1 - y_2 \rvert \le 2K \lvert y_1 - y_2 \rvert$: $f$ is *locally* Lipschitz with $L = 2K$ on that strip.

$$
\boxed{\text{(a) } L = 3, \qquad \text{(b) } L = 5, \qquad \text{(c) only locally Lipschitz: } L = 2K \text{ on } \lvert y \rvert \le K}
$$

> **Key Takeaway:** Globally Lipschitz means globally bounded $y$-slope; polynomials of degree at least two fail globally but succeed on every bounded strip — which is exactly why they give local, not global, existence.

### Problem L0.4: Anatomy of a Lipschitz Failure

**Problem Statement:** Show directly from the definition that $f(y) = 3y^{2/3}$ is not Lipschitz on any interval $[-\delta, \delta]$ with $\delta \gt 0$, but is Lipschitz on $[\epsilon, \infty)$ for every $\epsilon \gt 0$, with constant $L = 2\epsilon^{-1/3}$.

**Intuition:** The graph of $y^{2/3}$ has a vertical-tangent cusp at the origin; secant slopes through the cusp blow up.

**Solution:**

**Failure at the origin.** Test the definition against the pair $y_1 = y \gt 0$, $y_2 = 0$:

$$
\frac{\lvert f(y) - f(0) \rvert}{\lvert y - 0 \rvert} = \frac{3 y^{2/3}}{y} = 3 y^{-1/3}.
$$

As $y \to 0^+$ this ratio tends to $+\infty$. Given any proposed constant $L$, choose $y = \min\left(\delta, (3/L)^3 / 8\right) \gt 0$; then $3y^{-1/3} \gt L$, violating the Lipschitz inequality inside $[-\delta, \delta]$. Since $\delta$ was arbitrary, no neighborhood of $0$ admits a Lipschitz constant.

**Success away from the origin.** On $[\epsilon, \infty)$, $f$ is differentiable with

$$
f'(y) = 3 \cdot \frac{2}{3} y^{-1/3} = 2 y^{-1/3}, \qquad \lvert f'(y) \rvert \le 2 \epsilon^{-1/3} \quad \text{for } y \ge \epsilon.
$$

By the mean value theorem, for $y_1, y_2 \ge \epsilon$ there is $\xi \ge \epsilon$ with

$$
\lvert f(y_1) - f(y_2) \rvert = \lvert f'(\xi) \rvert \, \lvert y_1 - y_2 \rvert \le 2\epsilon^{-1/3} \lvert y_1 - y_2 \rvert.
$$

$$
\boxed{\text{No Lipschitz constant exists near } 0; \quad L = 2\epsilon^{-1/3} \text{ works on } [\epsilon, \infty)}
$$

Consequence for the IVP $y' = 3y^{2/3}$: uniqueness holds through every initial condition $y_0 \neq 0$ (Picard–Lindelöf applies locally there), and fails exactly at $y_0 = 0$ — matching the explicit multi-solution family.

> **Key Takeaway:** Lipschitz failure is a local diagnosis: locate the points where secant slopes blow up, and expect uniqueness trouble precisely there.

## Level 1 — Foundation

### Problem L1.1: Picard Iterates for $y' = y$

**Problem Statement:** For the IVP $y' = y$, $y(0) = 1$, compute the Picard iterates $y_0, y_1, y_2, y_3$, prove by induction the general formula for $y_n$, and identify the limit.

**Intuition:** Each iterate feeds the previous one through the integral operator; polynomial in, one-degree-higher polynomial out.

**Solution:**

The Picard operator is $T[y]\,(t) = 1 + \int_0^t y(s) \, ds$.

**Step 1: First iterates.**

$$
y_0(t) = 1, \qquad y_1(t) = 1 + \int_0^t 1 \, ds = 1 + t,
$$

$$
y_2(t) = 1 + \int_0^t (1 + s) \, ds = 1 + t + \frac{t^2}{2}, \qquad y_3(t) = 1 + \int_0^t \left( 1 + s + \frac{s^2}{2} \right) ds = 1 + t + \frac{t^2}{2} + \frac{t^3}{6}.
$$

**Step 2: Induction.** Claim $y_n(t) = \sum_{k=0}^{n} \frac{t^k}{k!}$. True for $n = 0$. If it holds for $n$, then

$$
y_{n+1}(t) = 1 + \int_0^t \sum_{k=0}^{n} \frac{s^k}{k!} \, ds = 1 + \sum_{k=0}^{n} \frac{t^{k+1}}{(k+1)!} = \sum_{k=0}^{n+1} \frac{t^k}{k!}.
$$

**Step 3: Limit.** The iterates are the partial sums of the exponential series, which converges uniformly on every bounded interval. Hence

$$
y_n(t) \longrightarrow \sum_{k=0}^{\infty} \frac{t^k}{k!} = e^t,
$$

and $e^t$ indeed solves the IVP: $(e^t)' = e^t$, $e^0 = 1$.

$$
\boxed{y_n(t) = \sum_{k=0}^{n} \frac{t^k}{k!} \longrightarrow e^t}
$$

> **Key Takeaway:** Picard iteration applied to $y' = y$ manufactures the Taylor series of $e^t$ term by term — the contraction proof is a series-generating machine.

### Problem L1.2: Picard Iterates for $y' = t + y$

**Problem Statement:** For the IVP $y' = t + y$, $y(0) = 1$, compute $y_1, y_2, y_3$, conjecture the pattern, and verify the limit against the exact solution obtained by the integrating-factor method.

**Intuition:** The inhomogeneous term $t$ shifts each iterate; the pattern should converge to a combination of $e^t$ and a polynomial.

**Solution:**

**Step 1: Iterates.** Here $T[y]\,(t) = 1 + \int_0^t (s + y(s)) \, ds$ and $y_0(t) = 1$.

$$
y_1(t) = 1 + \int_0^t (s + 1) \, ds = 1 + t + \frac{t^2}{2},
$$

$$
y_2(t) = 1 + \int_0^t \left( s + 1 + s + \frac{s^2}{2} \right) ds = 1 + t + t^2 + \frac{t^3}{6},
$$

$$
y_3(t) = 1 + \int_0^t \left( s + 1 + s + s^2 + \frac{s^3}{6} \right) ds = 1 + t + t^2 + \frac{t^3}{3} + \frac{t^4}{24}.
$$

**Step 2: Pattern.** Compare with the series of $2e^t - t - 1 = 1 + t + t^2 + \frac{t^3}{3} + \frac{t^4}{12} + \cdots$ (using $2e^t = \sum 2t^k/k!$). The iterate $y_n$ matches this series through degree $n$ and carries a trailing correction term $t^{n+1}/(n+1)!$ from the last integration — visible above: $y_2$ agrees through $t^2$, $y_3$ through $t^3$.

**Step 3: Exact solution.** Rewrite as $y' - y = t$; the integrating factor is $e^{-t}$:

$$
\left( y e^{-t} \right)' = t e^{-t} \implies y e^{-t} = -t e^{-t} - e^{-t} + C \implies y(t) = -t - 1 + C e^{t}.
$$

The initial condition $y(0) = 1$ gives $C = 2$. The Picard iterates converge (uniformly on bounded intervals, by the $(Lh)^n/n!$ estimate with $L = 1$) exactly to this function.

$$
\boxed{y(t) = 2e^{t} - t - 1}
$$

> **Key Takeaway:** For linear ODEs the Picard iterates reproduce the exact solution's Taylor expansion up to the iteration order — each sweep buys one more correct derivative at $t_0$.

### Problem L1.3: Certified Existence Interval on a Rectangle

**Problem Statement:** For the IVP $y' = t y^2$, $y(0) = 1$, work on the rectangle $R = \{ \lvert t \rvert \le 1, \ \lvert y - 1 \rvert \le 1 \}$. Compute the bound $M$, a Lipschitz constant $L$ valid on $R$, and the Picard–Lindelöf half-width $h = \min(a, b/M)$. Then compare with the true maximal interval by solving the equation exactly.

**Intuition:** The rectangle argument is deliberately conservative; solving exactly shows how much lifetime it leaves on the table.

**Solution:**

**Step 1: The bound $M$.** On $R$, $\lvert t \rvert \le 1$ and $\lvert y \rvert \le 2$, so

$$
\lvert f(t,y) \rvert = \lvert t \rvert \, y^2 \le 1 \cdot 4 = 4 = M.
$$

**Step 2: Lipschitz constant on $R$.** $\partial f / \partial y = 2ty$, and $\lvert 2ty \rvert \le 2 \cdot 1 \cdot 2 = 4$ on $R$; since $R$ is convex in $y$, the mean value theorem gives $L = 4$.

**Step 3: Certified interval.** With $a = 1$, $b = 1$:

$$
h = \min\left( a, \frac{b}{M} \right) = \min\left( 1, \frac{1}{4} \right) = \frac{1}{4}.
$$

Picard–Lindelöf certifies a unique solution on $[-1/4, 1/4]$.

**Step 4: Exact solution and true lifespan.** Separate variables: $y^{-2} \, dy = t \, dt$, so $-1/y = t^2/2 + C$; from $y(0) = 1$, $C = -1$, giving

$$
y(t) = \frac{2}{2 - t^2}.
$$

This blows up when $t^2 = 2$, so the true maximal interval is $(-\sqrt{2}, \sqrt{2}) \approx (-1.414, 1.414)$ — more than five times wider than the certified $h = 0.25$. The gap is structural: the rectangle proof must budget for the worst case in which $y$ races to the rectangle's edge at maximal speed $M$ immediately.

$$
\boxed{M = 4, \quad L = 4, \quad h = \frac{1}{4}, \quad \text{true interval } (-\sqrt{2}, \sqrt{2})}
$$

> **Key Takeaway:** $h = \min(a, b/M)$ is a certificate, not a lifespan; continuation (re-running Picard–Lindelöf from later points) recovers the rest of the maximal interval.

### Problem L1.4: Maximal Interval of $y' = y^2$ as a Function of $y_0$

**Problem Statement:** Solve $y' = y^2$, $y(0) = y_0$ for every $y_0 \in \mathbb{R}$ and determine the maximal interval of existence in each case ($y_0 \gt 0$, $y_0 = 0$, $y_0 \lt 0$).

**Intuition:** Larger positive initial data should blow up sooner; negative data get pushed toward the equilibrium $0$ and should survive forever forward.

**Solution:**

**Step 1: Equilibrium case.** If $y_0 = 0$, then $y \equiv 0$ is a solution; since $f(y) = y^2$ is locally Lipschitz, it is *the* solution, with maximal interval $(-\infty, \infty)$.

**Step 2: General case by separation.** For $y_0 \neq 0$, uniqueness prevents the solution from ever touching $0$, so we may divide:

$$
\int_0^t \frac{y'(s)}{y(s)^2} \, ds = t \implies \frac{1}{y_0} - \frac{1}{y(t)} = t \implies y(t) = \frac{y_0}{1 - y_0 t}.
$$

**Step 3: Read off intervals.** The formula is singular at $t^* = 1/y_0$.

- **$y_0 \gt 0$:** the singularity sits at $t^* = 1/y_0 \gt 0$; the branch through $t = 0$ is defined for $t \lt 1/y_0$, so $I_{\max} = (-\infty, \, 1/y_0)$, with $y(t) \to +\infty$ as $t \to (1/y_0)^-$. In particular $y_0 = 1$ reproduces $I_{\max} = (-\infty, 1)$. Doubling $y_0$ halves the time to blow-up.
- **$y_0 \lt 0$:** now $t^* = 1/y_0 \lt 0$, so the branch through $0$ is $I_{\max} = (1/y_0, \, \infty)$: the solution exists for all forward time (it decays monotonically toward $0$), but blows down to $-\infty$ at the finite backward time $1/y_0$.

**Step 4: Sanity check against continuation.** In every non-global case the solution leaves each compact set as $t$ approaches the finite endpoint (it is unbounded there), consistent with Theorem 2.13 — solutions end only by escaping.

$$
\boxed{y(t) = \frac{y_0}{1 - y_0 t}; \quad I_{\max} = (-\infty, 1/y_0) \text{ if } y_0 \gt 0, \quad \mathbb{R} \text{ if } y_0 = 0, \quad (1/y_0, \infty) \text{ if } y_0 \lt 0}
$$

> **Key Takeaway:** The maximal interval depends on the initial condition, not just on the equation — a single ODE can host immortal solutions and finite-time blow-ups side by side.

### Problem L1.5: Smooth Everywhere, Global Nowhere — $y' = 1 + y^2$

**Problem Statement:** Solve $y' = 1 + y^2$, $y(0) = 0$, find the maximal interval, and explain why *every* solution of this equation — regardless of initial condition — has a maximal interval of length exactly $\pi$.

**Intuition:** The right side is at least $1$ everywhere, so solutions must keep climbing; the question is how fast the climb compounds.

**Solution:**

**Step 1: Solve.** The equation is separable, and $1 + y^2 \ge 1 \gt 0$ means no equilibria exist:

$$
\int_0^t \frac{y'(s)}{1 + y(s)^2} \, ds = t \implies \arctan y(t) - \arctan 0 = t \implies y(t) = \tan t.
$$

**Step 2: Maximal interval.** $\tan t$ solves the ODE on $(-\pi/2, \pi/2)$ and satisfies $y(0) = 0$; as $t \to (\pi/2)^-$, $y \to +\infty$ and as $t \to (-\pi/2)^+$, $y \to -\infty$. Hence

$$
I_{\max} = \left( -\frac{\pi}{2}, \frac{\pi}{2} \right),
$$

even though $f(y) = 1 + y^2$ is a polynomial — $C^\infty$ on all of $\mathbb{R}$, locally Lipschitz everywhere. Smoothness of $f$ does not buy global existence.

**Step 3: Every solution lives exactly $\pi$.** For $y(t_0) = y_0$, the same computation gives $\arctan y(t) = (t - t_0) + \arctan y_0$, so

$$
y(t) = \tan\left( t - t_0 + \arctan y_0 \right), \qquad I_{\max} = \left( t_0 - \arctan y_0 - \frac{\pi}{2}, \; t_0 - \arctan y_0 + \frac{\pi}{2} \right).
$$

Since $\arctan y_0 \in (-\pi/2, \pi/2)$, the interval always has length $\left( \frac{\pi}{2} - \arctan y_0 \right) + \left( \frac{\pi}{2} + \arctan y_0 \right) = \pi$: initial conditions shift the lifetime window but cannot stretch it.

$$
\boxed{y(t) = \tan t, \qquad I_{\max} = \left( -\frac{\pi}{2}, \frac{\pi}{2} \right), \qquad \text{every solution lives for exactly } \pi \text{ units of time}}
$$

> **Key Takeaway:** Quadratic growth in $f$ imposes a universal finite lifespan here; global existence requires taming growth (linear bounds, dissipativity, or invariant compact regions), not just smoothness.

### Problem L1.6: The Full Non-Uniqueness Family of $y' = 3y^{2/3}$

**Problem Statement:** For $y' = 3y^{2/3}$, $y(0) = 0$, verify that for every pair $c_1 \le 0 \le c_2$ the glued function

$$
y_{c_1, c_2}(t) =
\begin{cases}
(t - c_1)^3, & t \lt c_1, \\
0, & c_1 \le t \le c_2, \\
(t - c_2)^3, & t \gt c_2
\end{cases}
$$

is a $C^1$ solution on all of $\mathbb{R}$, and conclude that the IVP has uncountably many solutions.

**Intuition:** The solution can linger at the non-Lipschitz equilibrium for any amount of time before departing along a shifted cubic — in either time direction.

**Solution:**

**Step 1: Each piece solves the ODE.** On $t \gt c_2$: $y = (t - c_2)^3$ gives $y' = 3(t - c_2)^2$ and $3 y^{2/3} = 3\left( (t - c_2)^3 \right)^{2/3} = 3 (t - c_2)^2$ since $t - c_2 \gt 0$. Equal. On $t \lt c_1$: $y = (t - c_1)^3 \lt 0$ and with the real (odd) cube root, $y^{1/3} = t - c_1$, so $y^{2/3} = (t - c_1)^2$ and again $y' = 3(t-c_1)^2 = 3y^{2/3}$. On the middle piece $y' = 0 = 3 \cdot 0^{2/3}$.

**Step 2: $C^1$ matching at the seams.** At $t = c_2$: the left value and derivative are $0$ and $0$; the right piece has $y(c_2^+) = 0$ and $y'(c_2^+) = 3(t - c_2)^2$ evaluated at $t = c_2$, which is $0$. So $y$ is differentiable at $c_2$ with $y'(c_2) = 0$, and $y'$ is continuous there ($3(t - c_2)^2 \to 0$). The seam at $c_1$ is identical by symmetry. Hence $y_{c_1, c_2} \in C^1(\mathbb{R})$ and satisfies the ODE at every real $t$.

**Step 3: Initial condition and cardinality.** Since $c_1 \le 0 \le c_2$, the point $t = 0$ falls in the middle piece, so $y(0) = 0$: every member solves the IVP. Distinct parameter pairs give distinct functions (they first leave $0$ at different times), and the parameter set $\{(c_1, c_2) : c_1 \le 0 \le c_2\}$, including the limits $c_1 = -\infty$ or $c_2 = +\infty$ (pure one-sided departures and $y \equiv 0$), is uncountable.

**Step 4: Consistency check with theory.** Away from $y = 0$ the function $3y^{2/3}$ is $C^1$, so all these solutions must — and do — coincide in shape (translated cubics), differing only in the time spent at the non-Lipschitz point $y = 0$. The freedom lives exactly where Picard–Lindelöf's hypothesis dies.

$$
\boxed{\text{The IVP admits the uncountable family } \{ y_{c_1, c_2} : c_1 \le 0 \le c_2 \}}
$$

> **Key Takeaway:** Non-uniqueness is not two stray solutions but a continuum parameterized by waiting times — one degree of freedom for each visit to the Lipschitz-failing set.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: Spectral Norms Certify a Well-Posed Neural ODE

**Problem Statement:** A Neural ODE uses the two-layer vector field

$$
f_\theta(\mathbf{z}) = W_2 \tanh(W_1 \mathbf{z} + \mathbf{b}), \qquad W_1 \in \mathbb{R}^{64 \times 8}, \quad W_2 \in \mathbb{R}^{8 \times 64},
$$

with spectral norms $\lVert W_1 \rVert_2 = 2$ and $\lVert W_2 \rVert_2 = 1.5$. (a) Prove $f_\theta$ is globally Lipschitz and compute a Lipschitz constant. (b) Conclude that the flow $\dot{\mathbf{z}} = f_\theta(\mathbf{z})$, $\mathbf{z}(0) = \mathbf{x}$, exists, is unique, and is defined for *all* $t \in \mathbb{R}$.

**Intuition:** Each layer stretches distances by at most its spectral norm; $\tanh$ never stretches at all. Global Lipschitz then kills both failure modes: non-uniqueness and blow-up.

**Solution:**

**(a) Lipschitz constant by composition.** For any $\mathbf{u}, \mathbf{v} \in \mathbb{R}^{64}$, componentwise the mean value theorem gives $\lvert \tanh u_i - \tanh v_i \rvert \le \lvert u_i - v_i \rvert$ since $0 \lt \operatorname{sech}^2 \le 1$; stacking components, $\lVert \tanh \mathbf{u} - \tanh \mathbf{v} \rVert_2 \le \lVert \mathbf{u} - \mathbf{v} \rVert_2$ ($\tanh$ is $1$-Lipschitz). Then for $\mathbf{z}_1, \mathbf{z}_2 \in \mathbb{R}^8$:

$$
\lVert f_\theta(\mathbf{z}_1) - f_\theta(\mathbf{z}_2) \rVert_2 \le \lVert W_2 \rVert_2 \, \lVert \tanh(W_1 \mathbf{z}_1 + \mathbf{b}) - \tanh(W_1 \mathbf{z}_2 + \mathbf{b}) \rVert_2 \le \lVert W_2 \rVert_2 \, \lVert W_1 \rVert_2 \, \lVert \mathbf{z}_1 - \mathbf{z}_2 \rVert_2.
$$

So $L = \lVert W_2 \rVert_2 \lVert W_1 \rVert_2 = 1.5 \times 2 = 3$, valid on all of $\mathbb{R}^8$ — a *global* Lipschitz constant.

**(b) Global well-posedness.** Local existence and uniqueness follow from the vector Picard–Lindelöf theorem at every point. For globality, note the linear growth bound

$$
\lVert f_\theta(\mathbf{z}) \rVert_2 \le \lVert f_\theta(\mathbf{0}) \rVert_2 + L \lVert \mathbf{z} \rVert_2 = : c + 3 \lVert \mathbf{z} \rVert_2,
$$

so along any solution, $u(t) = \lVert \mathbf{z}(t) \rVert_2$ satisfies $u(t) \le u(0) + \int_0^t (c + 3u(s)) ds$; Grönwall gives $u(t) \le \left( u(0) + ct \right) e^{3t}$, finite for every finite $t$. The trajectory never leaves a compact set in finite time, so by the continuation theorem the maximal interval is all of $\mathbb{R}$. (Even sharper: $\lVert \tanh(\cdot) \rVert_2 \le \sqrt{64} = 8$ bounds $\lVert f_\theta \rVert_2 \le 1.5 \times 8 = 12$ globally, so trajectories move at speed at most $12$ — no blow-up possible.)

$$
\boxed{L = 3; \ \text{the Neural ODE flow exists, is unique, and is global on } \mathbb{R}}
$$

> **Key Takeaway:** Spectral-norm control is the Picard–Lindelöf hypothesis in architectural form — constrain $\prod_i \lVert W_i \rVert_2$ and well-posedness of the learned dynamics is a theorem, not a hope.

### Problem L2.2: Grönwall as a Robustness Certificate

**Problem Statement:** For the Neural ODE of Problem L2.1 ($L = 3$, integration horizon $t \in [0, 1]$), an adversary perturbs the input $\mathbf{x}$ by $\lVert \boldsymbol{\delta} \rVert_2 = 0.01$. (a) Bound the output perturbation $\lVert \mathbf{z}_1(1) - \mathbf{z}_2(1) \rVert_2$. (b) The downstream linear classifier has a decision margin of $0.5$ at this input (distance from $\mathbf{z}(1)$ to the decision boundary, with a $1$-Lipschitz classifier head). Is the prediction certifiably unchanged? (c) What is the largest input perturbation this certificate can absorb?

**Intuition:** Grönwall converts a Lipschitz constant and a time horizon into a worst-case amplification factor $e^{Lt}$ — a computable, if conservative, adversarial guarantee.

**Solution:**

**(a) Trajectory divergence bound.** Both trajectories satisfy the integral equation; subtracting and applying the Lipschitz bound,

$$
\lVert \mathbf{z}_1(t) - \mathbf{z}_2(t) \rVert_2 \le \lVert \boldsymbol{\delta} \rVert_2 + 3 \int_0^t \lVert \mathbf{z}_1(s) - \mathbf{z}_2(s) \rVert_2 \, ds,
$$

so Grönwall yields $\lVert \mathbf{z}_1(t) - \mathbf{z}_2(t) \rVert_2 \le \lVert \boldsymbol{\delta} \rVert_2 \, e^{3t}$. At $t = 1$:

$$
\lVert \mathbf{z}_1(1) - \mathbf{z}_2(1) \rVert_2 \le 0.01 \times e^{3} \approx 0.01 \times 20.086 = 0.2009.
$$

**(b) Certification check.** A $1$-Lipschitz classifier head changes its score by at most the feature perturbation, $0.2009$. Since $0.2009 \lt 0.5$ (the margin), the predicted class **cannot flip**: the prediction is certifiably robust to this perturbation.

**(c) Certified radius.** The certificate holds as long as $\lVert \boldsymbol{\delta} \rVert_2 \, e^{3} \lt 0.5$, i.e.,

$$
\lVert \boldsymbol{\delta} \rVert_2 \lt 0.5 \, e^{-3} \approx 0.0249.
$$

Two design levers are visible in the formula: halving $L$ (stronger spectral normalization) squares the amplification factor's reciprocal ($e^{-L t}$ improves from $e^{-3} \approx 0.0498$ to $e^{-1.5} \approx 0.2231$, a $4.5\times$ larger certified radius), while shortening the integration horizon acts identically through the product $Lt$. Grönwall is sharp for $\dot{u} = Lu$, so without further structure (e.g., contractive dynamics with one-sided Lipschitz $\mu \lt 0$) this exponential price is unavoidable.

$$
\boxed{\lVert \mathbf{z}_1(1) - \mathbf{z}_2(1) \rVert_2 \le 0.01 e^{3} \approx 0.201 \lt 0.5 \ \text{(certified)}; \quad \text{certified radius} \approx 0.0249}
$$

> **Key Takeaway:** $e^{Lt}$ is the condition number of a Neural ODE's forward pass — Grönwall turns architecture constants into adversarial-robustness certificates, and explains why deep (long-horizon) flows need small Lipschitz constants.

### Problem L2.3: Blow-Up Time in Autocatalytic Chemistry

**Problem Statement:** In an idealized autocatalytic reaction, the concentration $x(t)$ (mol/L) of species $X$ obeys second-order self-amplifying kinetics

$$
\frac{dx}{dt} = k x^2, \qquad k = 0.02 \ \text{L mol}^{-1}\text{s}^{-1}, \qquad x(0) = 0.5 \ \text{mol/L}.
$$

(a) Solve the IVP and find the model's blow-up time $t^*$. (b) When does the concentration double? (c) Interpret: what does the finite $t^*$ say about the model rather than about the chemistry?

**Intuition:** Each molecule of product catalyzes more production; quadratic feedback compounds faster than exponentially.

**Solution:**

**(a) Solution and blow-up time.** Separating variables (valid since $x(0) \gt 0$ and uniqueness keeps $x \gt 0$):

$$
\int_0^t \frac{x'(s)}{x(s)^2} \, ds = k t \implies \frac{1}{x_0} - \frac{1}{x(t)} = kt \implies x(t) = \frac{x_0}{1 - k x_0 t}.
$$

Blow-up occurs when the denominator vanishes:

$$
t^* = \frac{1}{k x_0} = \frac{1}{0.02 \times 0.5} = 100 \ \text{s}.
$$

**(b) Doubling time.** Solve $x(t) = 2 x_0 = 1$:

$$
\frac{0.5}{1 - 0.01 t} = 1 \implies 1 - 0.01 t = 0.5 \implies t = 50 \ \text{s}.
$$

Note the accelerating cascade: the *first* doubling takes 50 s, and solving $x = 2^n x_0$ gives $t_n = 100(1 - 2^{-n})$ s — the second doubling takes only 25 s, the third 12.5 s, all infinitely many doublings completing before $t^* = 100$ s (a geometric-series Zeno cascade).

**(c) Interpretation.** Real concentrations cannot diverge: the model's hypotheses fail before $t^*$ as reactants deplete, heat dissipates, or transport limits kick in. The correct reading of $t^*$ is a *validity horizon*: the idealized rate law predicts its own breakdown at $t = 100$ s, and any refined model (e.g., logistic-type correction $\dot{x} = kx^2(1 - x/x_{\max})$, whose right side is again locally Lipschitz but now admits an a priori bound, hence global existence) must take over well before. Blow-up analysis is how modelers locate the boundary of a model's regime — the same computation used for thermal-runaway and explosion-limit estimates.

$$
\boxed{x(t) = \frac{0.5}{1 - 0.01 t}, \qquad t^* = 100 \ \text{s}, \qquad \text{doubling at } t = 50 \ \text{s}}
$$

> **Key Takeaway:** Finite-time blow-up in a well-posed local model is information: it timestamps where the modeling assumptions — not the mathematics — give out.

### Problem L2.4: The SIR Model Is Globally Well-Posed

**Problem Statement:** For the SIR epidemic model on a population $N$,

$$
S' = -\beta \frac{SI}{N}, \qquad I' = \beta \frac{SI}{N} - \gamma I, \qquad R' = \gamma I, \qquad \beta, \gamma \gt 0,
$$

show: (a) the right-hand side is locally Lipschitz, so solutions are unique; (b) the simplex $\Delta = \{ (S, I, R) : S, I, R \ge 0, \ S + I + R = N \}$ is forward-invariant; (c) every solution starting in $\Delta$ exists for all $t \ge 0$. Conclude what uniqueness means for forecasting.

**Intuition:** Polynomial vector field plus a compact invariant region is the standard recipe for global well-posedness of population models.

**Solution:**

**(a) Local Lipschitz.** Each component of $F(S, I, R) = \left( -\beta SI/N, \ \beta SI/N - \gamma I, \ \gamma I \right)$ is a polynomial, hence $C^1$; on any ball of radius $K$ the partial derivatives are bounded (e.g., by $\beta(2K)/N + \gamma$), so $F$ is locally Lipschitz by the multivariable mean value inequality. Picard–Lindelöf (vector form) gives local existence and uniqueness through every initial state.

**(b) Forward invariance of $\Delta$.** First, $\frac{d}{dt}(S + I + R) = -\beta SI/N + \beta SI/N - \gamma I + \gamma I = 0$: the total population is conserved, so the affine plane $S + I + R = N$ is invariant. For the faces: on $\{ I = 0 \}$ the field is $(0, 0, 0)$ — the whole face consists of equilibria, and by *uniqueness* no trajectory with $I(t_1) = 0$ at some time can have had $I \neq 0$ earlier or later (it would coincide with the equilibrium through that point). Hence $I \gt 0$ initially implies $I(t) \gt 0$ for all time in the interval of existence, and $I \ge 0$ is preserved. On $\{ S = 0 \}$: $S' = 0$, and the plane $S = 0$ is invariant for the (then linear) system, so by the same uniqueness argument $S \gt 0$ is preserved. Finally $R' = \gamma I \ge 0$ keeps $R \ge 0$. Thus trajectories starting in $\Delta$ remain in $\Delta$.

**(c) Global forward existence.** $\Delta$ is compact ($0 \le S, I, R \le N$). A maximal solution confined to a compact subset of the domain can never satisfy the escape alternative of the continuation theorem in finite time, so $t_+ = +\infty$: solutions exist, uniquely, for all $t \ge 0$.

**Forecasting consequence.** Given today's state $(S_0, I_0, R_0)$ and parameters $(\beta, \gamma)$, the epidemic's entire future curve is uniquely determined — so all forecast uncertainty is attributable to *data and parameter* uncertainty, quantifiable via Grönwall/sensitivity bounds, never to ambiguity in the model's mathematics. Two correctly-implemented solvers that disagree materially are exposing numerical error, not model freedom.

$$
\boxed{F \text{ locally Lipschitz} + \Delta \text{ compact invariant} \implies \text{unique global forward solutions in } \Delta}
$$

> **Key Takeaway:** For population models, well-posedness is proved in two moves — local Lipschitz for uniqueness, invariant compact region for immortality — and uniqueness is what makes forecasting a data problem rather than a mathematics problem.

### Problem L2.5: Norton's Dome — A Lipschitz Hole in Newtonian Determinism

**Problem Statement:** On Norton's dome, a mass at the apex obeys (in arc-length coordinate $r \ge 0$, units chosen cleanly)

$$
r'' = \sqrt{r}, \qquad r(0) = 0, \quad r'(0) = 0.
$$

(a) Verify that both $r \equiv 0$ and $r(t) = t^4/144$ solve the IVP. (b) Exhibit the full family of solutions and identify exactly which hypothesis of Picard–Lindelöf fails. (c) State carefully what this does and does not say about determinism in classical mechanics.

**Intuition:** This is the second-order incarnation of $y' = 3y^{2/3}$: a rest state where the force law has a non-Lipschitz cusp.

**Solution:**

**(a) Verification.** For $r \equiv 0$: $r'' = 0 = \sqrt{0}$, and both initial conditions hold. For $r(t) = t^4/144$ (taking $t \ge 0$):

$$
r'(t) = \frac{4t^3}{144} = \frac{t^3}{36}, \qquad r''(t) = \frac{t^2}{12}, \qquad \sqrt{r(t)} = \sqrt{\frac{t^4}{144}} = \frac{t^2}{12}.
$$

So $r'' = \sqrt{r}$, with $r(0) = r'(0) = 0$. Both solve the IVP.

**(b) The family and the failed hypothesis.** For any $T \ge 0$ define

$$
r_T(t) = \begin{cases} 0, & t \le T, \\ \dfrac{(t - T)^4}{144}, & t \gt T. \end{cases}
$$

On each piece the ODE holds (translation invariance); at $t = T$ the values of $r, r', r''$ from the right are $0, 0, 0$ (powers $4, 3, 2$ of $t - T$), matching the left, so $r_T \in C^2$ and solves the IVP for every $T$ — the mass sits at the apex until the arbitrary, uncaused moment $T$, then slides off. As a first-order system $\left( r, v \right)' = \left( v, \sqrt{r} \right)$, the field $F(r, v) = (v, \sqrt{r})$ is continuous but **not Lipschitz in $r$ at $r = 0$**: $\lvert \sqrt{r} - \sqrt{0} \rvert / \lvert r - 0 \rvert = r^{-1/2} \to \infty$. Peano still grants existence — which is why solutions abound — but the uniqueness hypothesis is void precisely at the initial state $(0, 0)$.

**(c) What it means.** It does *not* show classical mechanics is generally indeterministic: for smooth ($C^1$, hence locally Lipschitz) force laws — gravity away from collisions, Hooke's law, Lennard–Jones — Picard–Lindelöf makes determinism a theorem. It *does* show determinism is a property of the force law's regularity, not of "Newton's laws" in the abstract: the dome's shape is deliberately engineered (height profile with a $3/2$-power cusp) so that the tangential force $\sqrt{r}$ has an infinite derivative at equilibrium. Newtonian determinism has fine print, and the fine print is exactly the Lipschitz condition.

$$
\boxed{r_T(t) = \frac{(t - T)_+^4}{144} \text{ solves the IVP for every } T \ge 0; \ \text{the Lipschitz hypothesis fails at } r = 0}
$$

> **Key Takeaway:** Norton's dome is $y' = 3y^{2/3}$ dressed as physics — determinism in mechanics is exactly as strong as the Lipschitz continuity of the force law.

### Problem L2.6: Why Neural ODE Flows Cannot Represent $x \mapsto -x$

**Problem Statement:** Let $f(t, z)$ be continuous and Lipschitz in $z$ on $\mathbb{R} \times \mathbb{R}$ (a $1$-D Neural ODE), with flow map $\Phi(x) = z(1)$ where $z' = f(t, z)$, $z(0) = x$. (a) Prove trajectories cannot cross: $x_1 \lt x_2$ implies $\Phi(x_1) \lt \Phi(x_2)$. (b) Conclude $\Phi$ cannot equal $g(x) = -x$, no matter how expressive $f$ is. (c) Show the map becomes representable after augmenting one dimension, and explain the general principle.

**Intuition:** Uniqueness forbids two solution curves through the same space-time point; in one dimension curves that must swap order would have to intersect.

**Solution:**

**(a) Order preservation.** Let $z_1, z_2$ solve the ODE with $z_1(0) = x_1 \lt x_2 = z_2(0)$. Suppose $z_1(t^*) \ge z_2(t^*)$ for some $t^* \in (0, 1]$. By continuity of $z_1 - z_2$ and the intermediate value theorem there is $\tau \in (0, t^*]$ with $z_1(\tau) = z_2(\tau)$. But then both functions solve the same IVP with data $z(\tau) = z_1(\tau)$, and $f$ is Lipschitz, so by the uniqueness theorem (Grönwall, run forward *and backward* from $\tau$) $z_1 \equiv z_2$ on $[0, 1]$ — contradicting $z_1(0) \neq z_2(0)$. Hence $z_1(t) \lt z_2(t)$ for all $t \in [0,1]$; in particular $\Phi(x_1) \lt \Phi(x_2)$: the flow is strictly increasing. (It is also continuous by continuous dependence, and invertible by running time backwards — a homeomorphism, in fact an increasing one.)

**(b) The target map is out of reach.** $g(x) = -x$ requires $g(-1) = 1$ and $g(1) = -1$: it maps a smaller input to a larger output and vice versa, i.e., $g$ is strictly *decreasing*. By (a), every $1$-D Neural ODE flow is strictly increasing. No choice of $f$ — however deep, wide, or nonlinear — can realize $g$. (Geometrically: the trajectory from $-1$ up to $1$ and the trajectory from $1$ down to $-1$ would have to cross somewhere in the $(t, z)$ strip.)

**(c) Augmentation fixes it.** Lift to $\mathbb{R}^2$: initialize $\mathbf{z}(0) = (x, 0)$ and let the field be the rigid rotation

$$
\frac{d}{dt}\begin{pmatrix} z_1 \\ z_2 \end{pmatrix} = \pi \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} z_1 \\ z_2 \end{pmatrix},
$$

which is linear, hence globally Lipschitz ($L = \pi$). Its time-$1$ flow is rotation by $\pi$: $(x, 0) \mapsto (-x, 0)$; projecting to the first coordinate realizes $g(x) = -x$ exactly. The trajectories never intersect in $\mathbb{R}^2$ — they pass around each other through the extra dimension. **General principle:** uniqueness makes every Neural ODE flow a homeomorphism of $\mathbb{R}^d$ isotopic to the identity, so target maps outside this class (orientation reversal, non-injective maps like $x \mapsto \lvert x \rvert$, linking rearrangements) require augmentation to $\mathbb{R}^{d+p}$, where the topological obstruction vanishes.

$$
\boxed{1\text{-D Neural ODE flows are strictly increasing homeomorphisms; } x \mapsto -x \text{ needs } \ge 1 \text{ augmented dimension}}
$$

> **Key Takeaway:** The expressivity ceiling of Neural ODEs is the uniqueness theorem wearing an architecture hat — augmentation works because extra dimensions give non-crossing trajectories room to braid.

### Problem L3.1: Osgood's Uniqueness Criterion

**Problem Statement:** Let $\omega : [0, \infty) \to [0, \infty)$ be continuous, non-decreasing, with $\omega(0) = 0$, $\omega(u) \gt 0$ for $u \gt 0$, and

$$
\int_0^{1} \frac{du}{\omega(u)} = \infty \qquad \text{(divergence at the lower limit).}
$$

Suppose $\lvert f(t, y_1) - f(t, y_2) \rvert \le \omega( \lvert y_1 - y_2 \rvert )$ for all relevant points. Prove the IVP $y' = f(t, y)$, $y(t_0) = y_0$ has at most one solution. Check that this strictly generalizes the Lipschitz criterion and that $\omega(u) = 3u^{2/3}$ fails the test (consistently with non-uniqueness).

**Intuition:** Uniqueness survives moduli slightly worse than Lipschitz — like $u \ln(1/u)$ — as long as the "time to escape from zero difference," measured by $\int du/\omega(u)$, is infinite.

**Solution:**

**Step 1: Setup.** Let $y_1, y_2$ be solutions on $[t_0, t_0 + T]$ with the same initial value, and set $\rho(t) = \lvert y_1(t) - y_2(t) \rvert \ge 0$, $\rho(t_0) = 0$. Suppose for contradiction that $\rho(t_1) \gt 0$ for some $t_1$. Let

$$
\tau = \sup \{ t \in [t_0, t_1] : \rho(t) = 0 \},
$$

which exists ($\rho(t_0) = 0$) and satisfies $\rho(\tau) = 0$ by continuity, with $\rho(t) \gt 0$ on $(\tau, t_1]$.

**Step 2: A differential inequality where $\rho \gt 0$.** From the integral equations, for $t \in (\tau, t_1]$,

$$
\rho(t) \le \int_\tau^t \lvert f(s, y_1(s)) - f(s, y_2(s)) \rvert \, ds \le \int_\tau^t \omega(\rho(s)) \, ds.
$$

Define $\sigma(t) = \int_\tau^t \omega(\rho(s)) ds$; then $\sigma$ is $C^1$, $\sigma(\tau) = 0$, $\rho \le \sigma$, and since $\omega$ is non-decreasing, $\sigma'(t) = \omega(\rho(t)) \le \omega(\sigma(t))$. Also $\sigma(t_1) \ge \rho(t_1) \gt 0$; pick $t_2 \in (\tau, t_1]$ minimal-ish with $\sigma(t) \gt 0$ on $(t_2, t_1]$ and $\sigma(t_2^+)$ arbitrarily small (possible since $\sigma(\tau) = 0$ and $\sigma$ is continuous: for any small $\epsilon \gt 0$ there is $t_\epsilon$ with $\sigma(t_\epsilon) = \epsilon$, taking the last such time before $t_1$).

**Step 3: Integrate the separable inequality.** On $[t_\epsilon, t_1]$, $\sigma \ge \epsilon \gt 0$, so we may divide by $\omega(\sigma) \gt 0$:

$$
\int_{t_\epsilon}^{t_1} \frac{\sigma'(s)}{\omega(\sigma(s))} \, ds \le t_1 - t_\epsilon \le T.
$$

The substitution $u = \sigma(s)$ turns the left side into $\displaystyle\int_{\epsilon}^{\sigma(t_1)} \frac{du}{\omega(u)}$. Hence

$$
\int_{\epsilon}^{\sigma(t_1)} \frac{du}{\omega(u)} \le T \qquad \text{for every } \epsilon \gt 0.
$$

**Step 4: Contradiction with the Osgood condition.** As $\epsilon \to 0^+$ the left side tends to $\displaystyle\int_0^{\sigma(t_1)} \frac{du}{\omega(u)} = \infty$ (the hypothesis, since $\sigma(t_1) \gt 0$), yet it is bounded by the finite $T$. Contradiction. Hence $\rho \equiv 0$: uniqueness. (The interval $[t_0 - T, t_0]$ is handled by time reversal $t \mapsto 2t_0 - t$, which preserves the hypotheses.)

**Step 5: Scope of the criterion.**

- *Generalizes Lipschitz:* $\omega(u) = Lu$ gives $\int_0^1 du/(Lu) = \infty$. Also covered: $\omega(u) = u \ln(1/u)$ near $0$, since $\int_0^{1/2} \frac{du}{u \ln(1/u)} = \left[ \ln \ln (1/u) \right]$ diverges — uniqueness holds for some non-Lipschitz fields.
- *Correctly excludes the cusp:* $\omega(u) = 3u^{2/3}$ has $\int_0^1 \frac{du}{3u^{2/3}} = \left[ u^{1/3} \right]_0^1 = 1 \lt \infty$, so Osgood is silent — as it must be, because $y' = 3y^{2/3}$ genuinely loses uniqueness.

$$
\boxed{\int_0^{\cdot} \frac{du}{\omega(u)} = \infty \implies \text{uniqueness}; \quad \text{Lipschitz is the special case } \omega(u) = Lu}
$$

> **Key Takeaway:** Uniqueness is governed by an escape-time integral: solutions can separate only if the modulus $\omega$ lets the difference climb away from zero in finite "$\omega$-time," i.e., only if $\int_0^{} du/\omega(u)$ converges.

### Problem L3.2: A Differential Comparison Theorem

**Problem Statement:** Let $f(t, y)$ be continuous and Lipschitz in $y$ with constant $L$ on a strip containing all graphs below. Suppose $u, v$ are differentiable on $[t_0, t_1]$ with

$$
u'(t) \le f(t, u(t)), \qquad v'(t) = f(t, v(t)), \qquad u(t_0) \le v(t_0).
$$

Prove $u(t) \le v(t)$ for all $t \in [t_0, t_1]$ (a *subsolution* stays below the solution). Then use the theorem to bound solutions of $y' = \sin(y) + t$ by solutions of a linear equation.

**Intuition:** If you never move faster than the flow and start below it, you can never overtake it — but the proof must handle the possibility of touching and separating, which is where Lipschitz enters.

**Solution:**

**Step 1: Set up a first-crossing argument.** Let $w = u - v$, so $w(t_0) \le 0$ and

$$
w'(t) = u'(t) - v'(t) \le f(t, u(t)) - f(t, v(t)).
$$

Suppose for contradiction $w(t_1') \gt 0$ for some $t_1' \in (t_0, t_1]$. Define

$$
\tau = \sup\{ t \in [t_0, t_1'] : w(t) \le 0 \}.
$$

The set is non-empty ($t_0$ belongs) so $\tau$ exists; by continuity $w(\tau) = 0$ and $w(t) \gt 0$ for $t \in (\tau, t_1']$.

**Step 2: Exponential estimate on the overshoot interval.** For $t \in (\tau, t_1']$, $w(t) \gt 0$, so the Lipschitz condition gives

$$
f(t, u(t)) - f(t, v(t)) \le L \, \lvert u(t) - v(t) \rvert = L \, w(t),
$$

hence $w'(t) \le L w(t)$ there. Multiply by the positive integrating factor $e^{-L(t - \tau)}$:

$$
\frac{d}{dt} \left[ w(t) e^{-L(t - \tau)} \right] = \left( w'(t) - L w(t) \right) e^{-L(t-\tau)} \le 0.
$$

**Step 3: Contradiction.** The function $w(t) e^{-L(t - \tau)}$ is non-increasing on $[\tau, t_1']$ and vanishes at $\tau$, so

$$
w(t_1') \, e^{-L(t_1' - \tau)} \le w(\tau) = 0 \implies w(t_1') \le 0,
$$

contradicting $w(t_1') \gt 0$. Therefore $u(t) \le v(t)$ on all of $[t_0, t_1]$. (Note where Lipschitz was essential: with only continuity, subsolutions can cross — e.g., $u = t^3$ is a subsolution-equality of $y' = 3y^{2/3}$ that leaves the solution $v \equiv 0$ despite equal initial data.)

**Step 4: Application.** Consider $y' = \sin(y) + t$, $y(0) = 0$, on $[0, 2]$. Since $\sin(y) \le 1$ always, any solution is a subsolution of the linear problem $v' = 1 + t$, $v(0) = 0$ — here $f(t,v) = 1 + t$ is Lipschitz in $v$ with $L = 0$, and the theorem (with the roles $u = y$, $v$ the linear solution) yields

$$
y(t) \le v(t) = t + \frac{t^2}{2} \qquad \text{on } [0, 2],
$$

so $y(2) \le 4$. Symmetrically $\sin(y) \ge -1$ gives $y(t) \ge -t + t^2/2$. Two integrations of trivial equations produce rigorous two-sided bounds on a nonlinear solution no elementary formula can express.

$$
\boxed{u' \le f(t, u), \ v' = f(t,v), \ u(t_0) \le v(t_0) \implies u \le v \text{ on } [t_0, t_1]}
$$

> **Key Takeaway:** Comparison theorems convert one-sided differential inequalities into global orderings — the workhorse for bounding nonlinear solutions by solvable ones, and yet another consequence of the Lipschitz condition via the integrating-factor/Grönwall mechanism.

### Problem L3.3: Putnam-Style — Monotonicity and Reachability for $y' = f(y)$

**Problem Statement:** Let $f : \mathbb{R} \to (0, \infty)$ be continuous (strictly positive, no Lipschitz assumption). Consider $y' = f(y)$, $y(0) = y_0$. Prove: (a) every solution is strictly increasing, and any two solutions of the IVP agree (uniqueness holds despite the absence of a Lipschitz hypothesis!); (b) the solution reaches $+\infty$ in finite forward time if and only if

$$
\int_{y_0}^{\infty} \frac{dy}{f(y)} \lt \infty;
$$

(c) evaluate the criterion for $f(y) = 1 + y^2$ and $f(y) = 1 + \lvert y \rvert$.

**Intuition:** Time elapsed is $\int dy / f(y)$ along the trajectory — the equation trades the time variable for the state variable, and everything reduces to convergence of that integral.

**Solution:**

**(a) Monotonicity and uniqueness without Lipschitz.** Any solution satisfies $y'(t) = f(y(t)) \gt 0$, so $y$ is strictly increasing on its interval of existence, hence a bijection onto its range with differentiable inverse. Define, for $y$ in the range,

$$
G(y) = \int_{y_0}^{y} \frac{d\eta}{f(\eta)}.
$$

$G$ is $C^1$ with $G' = 1/f \gt 0$, so $G$ is strictly increasing. Along any solution, by the chain rule and substitution,

$$
\frac{d}{dt} G(y(t)) = \frac{y'(t)}{f(y(t))} = 1 \implies G(y(t)) = t \quad (\text{using } G(y(0)) = G(y_0) = 0).
$$

Thus every solution satisfies the *same* implicit equation $G(y(t)) = t$, and since $G$ is strictly increasing it is injective, so $y(t) = G^{-1}(t)$ is forced: **uniqueness holds by strict positivity alone** — the trajectory has no equilibria to linger at, which was the only mechanism of non-uniqueness in the cusp examples. (Compare: $3y^{2/3}$ vanishes at $0$; positivity of $f$ excludes precisely that.)

**(b) Reachability criterion.** From $G(y(t)) = t$: the forward lifetime of the solution is

$$
t_+ = \lim_{y \to \infty} G(y) = \int_{y_0}^{\infty} \frac{dy}{f(y)} \in (0, \infty].
$$

*If the integral converges* to $t_+ \lt \infty$: as $t \to t_+^-$, $y(t) = G^{-1}(t) \to \infty$ (since $G^{-1}$ maps $[0, t_+)$ onto $[y_0, \infty)$ monotonically) — blow-up at the finite time $t_+$, and the maximal interval's right endpoint is exactly $t_+$. *If the integral diverges:* $G$ maps $[y_0, \infty)$ onto $[0, \infty)$, so $y(t) = G^{-1}(t)$ is defined for every $t \ge 0$: the solution is global forward and $y(t) \to \infty$ only as $t \to \infty$. The criterion is exact — no slack in either direction.

**(c) Two evaluations.**

$$
f(y) = 1 + y^2: \quad \int_{y_0}^{\infty} \frac{dy}{1 + y^2} = \frac{\pi}{2} - \arctan y_0 \lt \infty \implies \text{finite-time blow-up (lifetime } \frac{\pi}{2} - \arctan y_0 \text{, matching L1.5)}.
$$

$$
f(y) = 1 + \lvert y \rvert: \quad \int_{y_0}^{\infty} \frac{dy}{1 + y} = \infty \ (\text{for } y \ge 0) \implies \text{global existence.}
$$

The watershed sits at growth just beyond linear: $f(y) \sim y \ln y$ still diverges ($\int dy/(y \ln y) = \ln \ln y \to \infty$), while $f(y) \sim y (\ln y)^{1+\epsilon}$ converges — an Osgood-type dichotomy at infinity, mirroring Problem L3.1's dichotomy at zero.

$$
\boxed{\text{Blow-up in finite time} \iff \int_{y_0}^{\infty} \frac{dy}{f(y)} \lt \infty; \ \text{uniqueness needs only } f \gt 0 \text{ continuous}}
$$

> **Key Takeaway:** For autonomous scalar equations, time is the integral of $1/f$ over states — convergence of that improper integral is the exact boundary between mortal and immortal solutions, and strict positivity of $f$ buys uniqueness with no Lipschitz condition at all.

### Problem L3.4: Picard–Lindelöf via the Bielecki Weighted Norm

**Problem Statement:** Let $f$ be continuous on $[t_0 - a, t_0 + a] \times \mathbb{R}$ and globally Lipschitz in $y$ with constant $L \gt 0$. On $X = C([t_0 - a, t_0 + a])$ define the **Bielecki norm**

$$
\lVert y \rVert_{B} = \sup_{\lvert t - t_0 \rvert \le a} e^{-2L \lvert t - t_0 \rvert} \, \lvert y(t) \rvert.
$$

Prove: (a) $\lVert \cdot \rVert_B$ is a norm equivalent to $\lVert \cdot \rVert_\infty$, so $(X, \lVert \cdot \rVert_B)$ is complete; (b) the Picard operator $T$ is a contraction on $(X, \lVert \cdot \rVert_B)$ with constant $\frac{1}{2}$, *regardless of the size of $a$*; (c) conclude existence and uniqueness on the whole interval $[t_0 - a, t_0 + a]$ in one stroke, with no interval-shrinking and no iterated-power argument.

**Intuition:** The weight $e^{-2L\lvert t - t_0 \rvert}$ discounts the future at exactly the rate at which the Lipschitz coupling amplifies it — like a well-chosen discount factor that makes a divergent-looking sum contract.

**Solution:**

**(a) Equivalence and completeness.** Positivity, homogeneity, and the triangle inequality are inherited pointwise from $\lvert \cdot \rvert$; and since $0 \lt e^{-2La} \le e^{-2L\lvert t - t_0 \rvert} \le 1$,

$$
e^{-2La} \, \lVert y \rVert_\infty \le \lVert y \rVert_B \le \lVert y \rVert_\infty.
$$

The two norms are equivalent, so they generate the same topology and the same Cauchy sequences; $(X, \lVert \cdot \rVert_\infty)$ is complete, hence so is $(X, \lVert \cdot \rVert_B)$.

**(b) Contraction with constant $\frac{1}{2}$.** Take $y, z \in X$ and $t$ with $t_0 \le t \le t_0 + a$ (the branch $t \lt t_0$ is symmetric). Using the Lipschitz bound and then inserting $1 = e^{2L(s - t_0)} e^{-2L(s - t_0)}$:

$$
\lvert T[y]\,(t) - T[z]\,(t) \rvert \le L \int_{t_0}^{t} \lvert y(s) - z(s) \rvert \, ds = L \int_{t_0}^{t} e^{2L(s - t_0)} \left[ e^{-2L(s - t_0)} \lvert y(s) - z(s) \rvert \right] ds \le L \, \lVert y - z \rVert_B \int_{t_0}^{t} e^{2L (s - t_0)} \, ds.
$$

Evaluate the integral:

$$
L \int_{t_0}^{t} e^{2L(s - t_0)} \, ds = L \cdot \frac{e^{2L(t - t_0)} - 1}{2L} \le \frac{1}{2} \, e^{2L (t - t_0)}.
$$

Multiplying both sides of the resulting bound by $e^{-2L(t - t_0)}$:

$$
e^{-2L(t - t_0)} \, \lvert T[y]\,(t) - T[z]\,(t) \rvert \le \frac{1}{2} \, \lVert y - z \rVert_B.
$$

Taking the supremum over the whole interval (both branches):

$$
\lVert T[y] - T[z] \rVert_B \le \frac{1}{2} \, \lVert y - z \rVert_B.
$$

The contraction constant $\frac{1}{2}$ is *independent of $a$ and of $L$* — the weight absorbed both.

**(c) Conclusion.** $T$ maps the complete space $X$ to itself (continuity of $f$ makes $T[y]$ continuous; no rectangle confinement is needed because $f$ is globally Lipschitz in $y$ on the strip, giving the linear growth bound $\lvert f(t,y) \rvert \le \lvert f(t, y_0) \rvert + L \lvert y - y_0 \rvert$ which keeps all integrals finite). By the Banach fixed-point theorem there is exactly one fixed point in $X$, i.e., exactly one solution of the IVP on the **entire** interval $[t_0 - a, t_0 + a]$; since $a$ was arbitrary, globally Lipschitz right-hand sides yield unique solutions on all of $\mathbb{R}$. Compare the costs: the rectangle proof needed $h \le b/M$ and either $Lh \lt 1$ or the $T^n$/factorial argument; the weighted norm eliminates every restriction by re-metrizing the *space* instead of shrinking the *domain* — the discount rate $2L$ (any rate strictly greater than $L$ works, giving constant $L/\text{rate}$) outpaces the worst-case growth rate $L$ of separations, which is Grönwall's exponential in disguise.

$$
\boxed{\lVert T[y] - T[z] \rVert_B \le \frac{1}{2} \lVert y - z \rVert_B \implies \text{unique solution on } [t_0 - a, t_0 + a] \text{ for every } a}
$$

> **Key Takeaway:** When an operator fails to contract, consider changing the metric, not the problem — Bielecki's exponential weight turns Picard's operator into a $\frac{1}{2}$-contraction on any interval, unifying Picard–Lindelöf and Grönwall in a single geometric idea.
