# Topic 05: Phase Plane and Stability Analysis

## 1. First-Principles Intuition & Motivation

**Phenomenon.** The pendulum equation $\ddot{\theta} + \sin\theta = 0$ has no elementary closed-form solution. Neither does the van der Pol oscillator, the Lotka–Volterra predator–prey model, or the coupled gradient fields that drive GAN training. Faced with this, the 19th-century program of "solve for $x(t)$, then read off behavior" collapses. Poincaré's insight was to invert the question: instead of asking *what is the formula for the solution*, ask *what does the family of all solutions look like as curves in state space*.

For an **autonomous system**

$$
\dot{\mathbf{x}} = \mathbf{f}(\mathbf{x}), \qquad \mathbf{x} \in \mathbb{R}^n,
$$

the right-hand side is a **vector field**: at every point of state space, $\mathbf{f}$ pins an arrow, and solutions are the curves that flow along the arrows. When $n = 2$ the state space is the **phase plane**, and the collection of all solution curves — the **phase portrait** — can literally be drawn. Remarkably, a handful of qualitative features (equilibria and their types, nullclines, separatrices, closed orbits) determines the fate of *every* initial condition, even when no formula exists.

### 1.1 Why the phase plane is so rigid: non-crossing of trajectories

The single most consequential fact inherits directly from the Picard–Lindelöf theorem (see [`../02_existence_uniqueness_picard_lindelof/`](../02_existence_uniqueness_picard_lindelof/)): if $\mathbf{f}$ is locally Lipschitz, the solution through each point is **unique**. For an autonomous system this has a geometric translation:

> **Two distinct orbits can never cross, and a single orbit can never cross itself transversally.**

Indeed, if two solution curves passed through the same point $\mathbf{p}$, then $\mathbf{p}$ would launch two different solutions of the same initial value problem — impossible. Orbits therefore **partition** the phase plane like streamlines of a steady fluid. In two dimensions this is a straitjacket: a closed orbit is a Jordan curve that traps everything inside it forever. This topological confinement is why planar systems cannot be chaotic (Poincaré–Bendixson, Section 2.4) and why one saddle's separatrices can organize the entire global portrait.

**Goal of the module.** Build the complete toolkit for qualitative analysis:

1. **Local**: classify every equilibrium from two numbers, the trace $\tau$ and determinant $\Delta$ of the Jacobian.
2. **Certificates**: decide stability without solving anything, via Lyapunov functions and first integrals.
3. **Global**: assemble nullclines, separatrices, basins, and limit cycles into the full portrait, and know exactly when local linear information is trustworthy (Hartman–Grobman) and when it lies (centers).

### 1.2 Assumptions, variables, and notation

**Standing assumptions.** Throughout, $\mathbf{f} : \mathbb{R}^2 \to \mathbb{R}^2$ (occasionally $\mathbb{R}^n$) is continuously differentiable, so solutions exist, are unique, and depend continuously on initial data. The system is autonomous: $\mathbf{f}$ does not depend on $t$, hence time-translates of solutions are solutions and orbits are well-defined geometric objects.

| Symbol | Meaning |
| :--- | :--- |
| $\dot{\mathbf{x}} = \mathbf{f}(\mathbf{x})$ | autonomous system; in the plane $\dot{x} = f(x,y)$, $\dot{y} = g(x,y)$ |
| $\varphi_t(\mathbf{x}_0)$ | the flow: position at time $t$ of the solution starting at $\mathbf{x}_0$ |
| $\mathbf{x}^*$ | equilibrium point, $\mathbf{f}(\mathbf{x}^*) = \mathbf{0}$ |
| $J(\mathbf{x}^*)$ | Jacobian matrix $D\mathbf{f}$ evaluated at $\mathbf{x}^*$ |
| $\tau, \Delta$ | trace and determinant of $J$; discriminant $\tau^2 - 4\Delta$ |
| $V(\mathbf{x})$ | Lyapunov function candidate; $\dot{V} = \nabla V \cdot \mathbf{f}$ its orbital derivative |
| $H(\mathbf{x})$ | first integral (conserved quantity): $\dot{H} \equiv 0$ along solutions |
| $\omega(\mathbf{x}_0)$ | omega-limit set: all points the forward orbit accumulates on |
| $\lVert \cdot \rVert$ | Euclidean norm on $\mathbb{R}^n$ |

The **orbital derivative** deserves emphasis: for any scalar field $V$, the chain rule gives $\frac{d}{dt} V(\mathbf{x}(t)) = \nabla V(\mathbf{x}(t)) \cdot \mathbf{f}(\mathbf{x}(t))$, computable **without knowing the solution**. Every stability certificate in this module exploits exactly this loophole.

## 2. Rigorous Mathematical Definitions & Theorem Statements

### 2.1 Flows, orbits, invariant sets

**Definition (flow).** For each $\mathbf{x}_0$, let $\varphi_t(\mathbf{x}_0)$ denote the value at time $t$ of the maximal solution with $\varphi_0(\mathbf{x}_0) = \mathbf{x}_0$. The family $\{\varphi_t\}$ satisfies the group law $\varphi_{t+s} = \varphi_t \circ \varphi_s$ on its domain of definition — the algebraic fingerprint of autonomy.

**Definition (orbit).** The orbit through $\mathbf{x}_0$ is the curve $\gamma(\mathbf{x}_0) = \{\varphi_t(\mathbf{x}_0) : t \in I_{\max}\}$. Orbits come in exactly three flavors for planar systems: single points (equilibria), closed curves (periodic orbits), and injective curves that never close.

**Definition (invariant set).** A set $S$ is **positively invariant** if $\varphi_t(S) \subseteq S$ for all $t \ge 0$, and **invariant** if this holds for all $t \in \mathbb{R}$. Equilibria, periodic orbits, and full basins of attraction are invariant; a trapping region (Section 2.4) is positively invariant.

**Definition (equilibrium, nullclines).** $\mathbf{x}^*$ is an equilibrium iff $\mathbf{f}(\mathbf{x}^*) = \mathbf{0}$. In the plane, the $x$-**nullcline** is the set $\{f(x,y) = 0\}$ where the flow is vertical and the $y$-**nullcline** is $\{g(x,y) = 0\}$ where it is horizontal; equilibria are precisely the intersections of the two nullcline families. Nullclines carve the plane into regions of constant direction-sign, which is why a phase portrait can be sketched from sign information alone.

**Definition (omega-limit set).** $\omega(\mathbf{x}_0)$ is the set of points $\mathbf{p}$ with $\varphi_{t_k}(\mathbf{x}_0) \to \mathbf{p}$ for some sequence $t_k \to \infty$. If the forward orbit is bounded, $\omega(\mathbf{x}_0)$ is nonempty, compact, connected, and invariant — the stage on which Poincaré–Bendixson plays out.

### 2.2 Stability: the precise definitions

Let $\mathbf{x}^*$ be an equilibrium of $\dot{\mathbf{x}} = \mathbf{f}(\mathbf{x})$.

**Definition (Lyapunov stability).** $\mathbf{x}^*$ is **stable** if for every $\varepsilon > 0$ there exists $\delta > 0$ such that

$$
\lVert \mathbf{x}(0) - \mathbf{x}^* \rVert < \delta \implies \lVert \mathbf{x}(t) - \mathbf{x}^* \rVert < \varepsilon \quad \text{for all } t \ge 0.
$$

Nothing here says trajectories approach $\mathbf{x}^*$ — only that they can be kept arbitrarily close by starting close enough. A center is stable in exactly this sense.

**Definition (asymptotic stability).** $\mathbf{x}^*$ is **asymptotically stable** if it is stable *and* there exists $\delta_0 > 0$ such that $\lVert \mathbf{x}(0) - \mathbf{x}^* \rVert < \delta_0$ implies $\mathbf{x}(t) \to \mathbf{x}^*$ as $t \to \infty$. Attraction alone does **not** imply stability (there are pathological examples where all orbits return home but first make excursions of fixed size), which is why both clauses are required.

**Definition (exponential stability).** $\mathbf{x}^*$ is **exponentially stable** if there exist $M, c, \delta_0 > 0$ with

$$
\lVert \mathbf{x}(0) - \mathbf{x}^* \rVert < \delta_0 \implies \lVert \mathbf{x}(t) - \mathbf{x}^* \rVert \le M e^{-ct} \lVert \mathbf{x}(0) - \mathbf{x}^* \rVert \quad \text{for all } t \ge 0.
$$

Exponential implies asymptotic implies Lyapunov stability, and each implication is strict: the scalar system $\dot{x} = -x^3$ is asymptotically but not exponentially stable at $0$ (decay is only algebraic, $x(t) \sim t^{-1/2}$), while a center is stable but not asymptotically stable.

**Definition (instability).** $\mathbf{x}^*$ is **unstable** if it is not stable: there exists $\varepsilon_0 > 0$ such that for every $\delta > 0$, some initial condition within $\delta$ eventually leaves the $\varepsilon_0$-ball.

### 2.3 The complete classification of planar linear systems

For $\dot{\mathbf{x}} = A\mathbf{x}$ with $A \in \mathbb{R}^{2 \times 2}$, the eigenvalues are $\lambda_{1,2} = \tfrac{1}{2}\left(\tau \pm \sqrt{\tau^2 - 4\Delta}\right)$ where $\tau = \operatorname{tr} A$ and $\Delta = \det A = \lambda_1 \lambda_2$. Two numbers therefore decide everything:

| Region of the $(\tau, \Delta)$ plane | Eigenvalues | Type | Stability |
| :--- | :--- | :--- | :--- |
| $\Delta < 0$ | real, opposite signs | **saddle** | unstable |
| $\Delta > 0$, $\tau^2 - 4\Delta > 0$, $\tau < 0$ | real, both negative | **stable node** | asymptotically stable |
| $\Delta > 0$, $\tau^2 - 4\Delta > 0$, $\tau > 0$ | real, both positive | **unstable node** | unstable |
| $\Delta > 0$, $\tau^2 - 4\Delta < 0$, $\tau < 0$ | complex, negative real part | **spiral sink** | asymptotically stable |
| $\Delta > 0$, $\tau^2 - 4\Delta < 0$, $\tau > 0$ | complex, positive real part | **spiral source** | unstable |
| $\Delta > 0$, $\tau = 0$ | purely imaginary | **center** | stable, not asymptotically |
| $\tau^2 = 4\Delta \neq 0$ | repeated real | star or degenerate node | sign of $\tau$ decides |
| $\Delta = 0$ | at least one zero eigenvalue | line of equilibria / degenerate | non-isolated or borderline |

**Geometry of the diagram.** Plot $\tau$ horizontally, $\Delta$ vertically. The lower half-plane is all saddles. The upper half-plane splits along the parabola $\Delta = \tau^2/4$: below it nodes, above it spirals, with the positive $\Delta$-axis ($\tau = 0$) the razor-thin locus of centers. Crossing the $\Delta$-axis upward through $\tau = 0$ is precisely the eigenvalue-crossing that produces a **Hopf bifurcation** in nonlinear systems.

**Eigenvector geometry.** For real eigenvalues the eigenlines $\operatorname{span}(\mathbf{v}_1)$, $\operatorname{span}(\mathbf{v}_2)$ are invariant. In a stable node with $\lambda_2 < \lambda_1 < 0$, generic orbits become **tangent to the slow eigenvector** $\mathbf{v}_1$ as $t \to \infty$ (the fast component dies first) and parallel to the fast eigenvector $\mathbf{v}_2$ far away. In a saddle, the stable eigenline is the only route in, the unstable eigenline the only route out; these four half-line orbits are the **separatrices** that partition the neighboring flow.

### 2.4 Linearization, hyperbolicity, and when it fails

**Definition (Jacobian linearization).** Near an equilibrium write $\mathbf{x} = \mathbf{x}^* + \boldsymbol{\eta}$. Taylor expansion gives $\dot{\boldsymbol{\eta}} = J(\mathbf{x}^*)\, \boldsymbol{\eta} + O(\lVert \boldsymbol{\eta} \rVert^2)$; the linear system $\dot{\boldsymbol{\eta}} = J \boldsymbol{\eta}$ is the linearization.

**Definition (hyperbolic equilibrium).** $\mathbf{x}^*$ is **hyperbolic** if no eigenvalue of $J(\mathbf{x}^*)$ has zero real part.

**Theorem (Hartman–Grobman).** If $\mathbf{x}^*$ is hyperbolic, there is a homeomorphism of a neighborhood of $\mathbf{x}^*$ onto a neighborhood of $\mathbf{0}$ mapping orbits of the nonlinear system onto orbits of the linearization, preserving time direction. In particular, nodes stay nodes, saddles stay saddles, spirals stay spirals, and stability type is inherited.

**When linearization fails.** Hyperbolicity excludes exactly the borderline rows of the classification table: centers ($\tau = 0$, $\Delta > 0$) and zero eigenvalues ($\Delta = 0$). There the neglected $O(\lVert \boldsymbol{\eta} \rVert^2)$ terms are decisive. Canonical example (fully derived in Section 3.3):

$$
\dot{x} = -y + a\,x(x^2 + y^2), \qquad \dot{y} = x + a\,y(x^2 + y^2),
$$

whose linearization at the origin is the center $\dot{x} = -y$, $\dot{y} = x$ for **every** $a$, yet the origin is a spiral sink for $a < 0$ and a spiral source for $a > 0$.

### 2.5 Lyapunov's direct method and limit-cycle theorems

**Definition (Lyapunov function).** Let $\mathbf{x}^* = \mathbf{0}$ (translate if necessary) and let $V : U \to \mathbb{R}$ be $C^1$ on a neighborhood $U$ of $\mathbf{0}$. $V$ is **positive definite** if $V(\mathbf{0}) = 0$ and $V(\mathbf{x}) > 0$ for $\mathbf{x} \neq \mathbf{0}$; it is a **Lyapunov function** if additionally $\dot{V}(\mathbf{x}) = \nabla V(\mathbf{x}) \cdot \mathbf{f}(\mathbf{x}) \le 0$ on $U$, and a **strict** Lyapunov function if $\dot{V} < 0$ for $\mathbf{x} \neq \mathbf{0}$.

**Theorem (Lyapunov, 1892).** If a Lyapunov function exists, $\mathbf{x}^*$ is stable. If a strict Lyapunov function exists, $\mathbf{x}^*$ is asymptotically stable. *(Proved in Section 3.2.)*

**Theorem (LaSalle invariance principle).** Let $\Omega$ be a compact positively invariant set, $V$ a $C^1$ function with $\dot{V} \le 0$ on $\Omega$, $E = \{\mathbf{x} \in \Omega : \dot{V}(\mathbf{x}) = 0\}$, and $M$ the largest invariant subset of $E$. Then every solution starting in $\Omega$ converges to $M$ as $t \to \infty$. This upgrades non-strict Lyapunov functions (e.g. mechanical energy with damping, which vanishes on the whole axis of zero velocity) into asymptotic conclusions.

**Construction heuristics.** Quadratic forms $V = \mathbf{x}^{\mathsf{T}} P \mathbf{x}$ with $P \succ 0$ (found by solving the Lyapunov equation $A^{\mathsf{T}} P + P A = -Q$, Section 3.6); physical energies for mechanical systems; convex first integrals for conservative systems.

### 2.6 Limit cycles, their detection, and their exclusion

**Definition (limit cycle).** A limit cycle is an isolated periodic orbit: a closed orbit with an annular neighborhood containing no other closed orbit. Nearby trajectories spiral toward it (stable), away (unstable), or mixed (semi-stable). Centers produce *continua* of closed orbits, which are therefore not limit cycles.

**Theorem (Poincaré–Bendixson).** Suppose $R \subset \mathbb{R}^2$ is a compact positively invariant set for a $C^1$ planar vector field, and $R$ contains no equilibria. Then $R$ contains a periodic orbit, and every omega-limit set in $R$ is a periodic orbit. Standard use: build a **trapping region**, typically an annulus whose inner boundary repels (flow points outward from a small circle around an unstable spiral) and whose outer boundary attracts (flow points inward) — then a limit cycle lives in the annulus. Deep consequence: bounded planar dynamics is eventually stationary or periodic; **chaos is impossible in the plane**.

**Theorem (Bendixson–Dulac negative criterion).** Let $D$ be a simply connected open region and $B \in C^1(D)$ a *Dulac function* such that

$$
\frac{\partial (B f)}{\partial x} + \frac{\partial (B g)}{\partial y}
$$

has constant sign (and is not identically zero) on $D$. Then the system $\dot{x} = f$, $\dot{y} = g$ has no periodic orbit lying entirely in $D$. Taking $B \equiv 1$ gives Bendixson's original divergence test. *(Proved via Green's theorem in Section 3.4.)*

**Hopf bifurcation (preview).** If a family $\dot{\mathbf{x}} = \mathbf{f}_\mu(\mathbf{x})$ has an equilibrium whose Jacobian eigenvalues $\alpha(\mu) \pm i\beta(\mu)$ cross the imaginary axis transversally ($\alpha(\mu_0) = 0$, $\alpha'(\mu_0) \neq 0$, $\beta(\mu_0) \neq 0$), then generically a small limit cycle of amplitude $\sim \sqrt{\lvert \mu - \mu_0 \rvert}$ is born — a spiral sink losing stability spits out a stable cycle (supercritical case). This is the trace-axis crossing of the $(\tau, \Delta)$ diagram made dynamic.

**One-dimensional bifurcations (briefly).** On the line, equilibria are born, merge, and exchange stability in three normal forms: **saddle-node** $\dot{x} = \mu + x^2$ (two equilibria collide and annihilate at $\mu = 0$), **transcritical** $\dot{x} = \mu x - x^2$ (two equilibria pass through each other and swap stability), and **pitchfork** $\dot{x} = \mu x - x^3$ (a symmetric equilibrium sheds a symmetric pair). These reappear as the $\Delta = 0$ boundary crossings of planar systems.

## 3. Step-by-Step Mathematical Proofs & Derivations

### 3.1 Proof A: Classification of planar linear systems with real distinct eigenvalues

**Claim.** Let $A$ have real eigenvalues $\lambda_1 \neq \lambda_2$ with independent eigenvectors $\mathbf{v}_1, \mathbf{v}_2$. Then the origin of $\dot{\mathbf{x}} = A\mathbf{x}$ is: a stable node if $\lambda_2 < \lambda_1 < 0$; an unstable node if $0 < \lambda_1 < \lambda_2$; a saddle if $\lambda_1 < 0 < \lambda_2$ — with the geometry described below.

**Step 1 (general solution).** Since $\{\mathbf{v}_1, \mathbf{v}_2\}$ is a basis, every solution is

$$
\mathbf{x}(t) = c_1 e^{\lambda_1 t} \mathbf{v}_1 + c_2 e^{\lambda_2 t} \mathbf{v}_2 ,
$$

with $c_1, c_2$ determined by $\mathbf{x}(0)$. The eigenlines ($c_2 = 0$ or $c_1 = 0$) are invariant: solutions starting on them stay on them.

**Step 2 (stable node, $\lambda_2 < \lambda_1 < 0$).** Both exponentials decay, so $\mathbf{x}(t) \to \mathbf{0}$ for every initial condition; with $m = \max(\lVert \mathbf{v}_1 \rVert, \lVert \mathbf{v}_2 \rVert)$ one gets $\lVert \mathbf{x}(t) \rVert \le (\lvert c_1 \rvert + \lvert c_2 \rvert)\, m\, e^{\lambda_1 t}$, which is exponential stability. For the tangency: if $c_1 \neq 0$,

$$
\mathbf{x}(t) = e^{\lambda_1 t}\left( c_1 \mathbf{v}_1 + c_2 e^{(\lambda_2 - \lambda_1)t} \mathbf{v}_2 \right),
$$

and since $\lambda_2 - \lambda_1 < 0$ the bracket tends to $c_1\mathbf{v}_1$. Hence the direction of $\mathbf{x}(t)$ converges to the **slow** eigendirection $\mathbf{v}_1$: all generic orbits enter the origin tangent to the slow eigenline, and only the two orbits with $c_1 = 0$ enter along the fast one.

**Step 3 (unstable node, $0 < \lambda_1 < \lambda_2$).** Reverse time: $s = -t$ turns the system into $\frac{d\mathbf{x}}{ds} = -A\mathbf{x}$ with eigenvalues $-\lambda_2 < -\lambda_1 < 0$, a stable node. So all orbits leave the origin (tangent to the slow direction $\mathbf{v}_1$), and the origin is unstable: every neighborhood contains points whose orbits exit any fixed ball.

**Step 4 (saddle, $\lambda_1 < 0 < \lambda_2$).** If $c_2 \neq 0$ then $\lVert \mathbf{x}(t) \rVert \to \infty$ because the $e^{\lambda_2 t}$ term dominates; so the origin is unstable. Exactly the solutions with $c_2 = 0$ converge to the origin (along $\mathbf{v}_1$): the stable eigenline is the **stable separatrix**. Exactly those with $c_1 = 0$ converge to the origin in backward time: the unstable eigenline is the **unstable separatrix**. Every other orbit is asymptotic to the unstable eigenline as $t \to +\infty$ and to the stable eigenline as $t \to -\infty$, producing the hyperbola-like portrait; the four separatrix half-lines divide a neighborhood of the origin into four flow-invariant sectors.

$$
\boxed{\;\dot{\mathbf{x}} = A\mathbf{x},\ \lambda_1 \neq \lambda_2 \in \mathbb{R}:\quad \lambda_2 < \lambda_1 < 0 \Rightarrow \text{stable node (tangent to slow eigenline)},\quad \lambda_1 < 0 < \lambda_2 \Rightarrow \text{saddle with eigenline separatrices},\quad 0 < \lambda_1 < \lambda_2 \Rightarrow \text{unstable node}.\;}
$$

### 3.2 Proof B: Lyapunov's stability theorem (sublevel-set argument)

**Theorem.** Let $\mathbf{f}(\mathbf{0}) = \mathbf{0}$ and let $V$ be positive definite and $C^1$ on a neighborhood $U$ of $\mathbf{0}$ with $\dot{V} = \nabla V \cdot \mathbf{f} \le 0$ on $U$. Then $\mathbf{0}$ is stable. If moreover $\dot{V} < 0$ on $U \setminus \{\mathbf{0}\}$, then $\mathbf{0}$ is asymptotically stable.

**Step 1 (set the stage).** Fix $\varepsilon > 0$ small enough that the closed ball $\bar{B}_\varepsilon \subset U$. The sphere $S_\varepsilon = \{\lVert \mathbf{x} \rVert = \varepsilon\}$ is compact and does not contain $\mathbf{0}$, so by positive definiteness and continuity

$$
m := \min_{\mathbf{x} \in S_\varepsilon} V(\mathbf{x}) > 0 .
$$

**Step 2 (trap in a sublevel set).** Consider $\Omega = \{\mathbf{x} \in \bar{B}_\varepsilon : V(\mathbf{x}) < m\}$. By continuity of $V$ at $\mathbf{0}$ (where $V = 0$), there is $\delta \in (0, \varepsilon)$ with $V(\mathbf{x}) < m$ for all $\lVert \mathbf{x} \rVert < \delta$; hence $B_\delta \subseteq \Omega$.

**Step 3 (the trap holds).** Take $\mathbf{x}(0) \in B_\delta$. Along the solution, $\frac{d}{dt}V(\mathbf{x}(t)) = \dot{V}(\mathbf{x}(t)) \le 0$, so $V(\mathbf{x}(t)) \le V(\mathbf{x}(0)) < m$ for as long as the solution stays in $\bar{B}_\varepsilon$. Suppose the solution reached $S_\varepsilon$ at some first time $t_1$; then $V(\mathbf{x}(t_1)) \ge m$ by Step 1 — contradicting $V(\mathbf{x}(t_1)) < m$. Hence the solution never reaches $S_\varepsilon$: it remains in $B_\varepsilon$ for all $t \ge 0$ (which also guarantees global forward existence, since the orbit stays in a compact set). This is exactly the $\varepsilon$–$\delta$ statement of stability.

**Step 4 (asymptotic version).** Now assume $\dot{V} < 0$ off the origin, and take $\mathbf{x}(0) \in B_\delta$ as above. The function $t \mapsto V(\mathbf{x}(t))$ is strictly decreasing (unless the solution is the equilibrium itself) and bounded below by $0$, so it has a limit $c \ge 0$. Suppose $c > 0$. Then the orbit stays forever in the compact annular set

$$
K = \{\mathbf{x} \in \bar{B}_\varepsilon : c \le V(\mathbf{x}) \le V(\mathbf{x}(0))\},
$$

which excludes a neighborhood of the origin (near $\mathbf{0}$, $V < c$). On $K$, the continuous function $\dot{V}$ attains a maximum $-\gamma := \max_K \dot{V} < 0$. Integrating,

$$
V(\mathbf{x}(t)) \le V(\mathbf{x}(0)) - \gamma\, t \xrightarrow[t \to \infty]{} -\infty ,
$$

contradicting $V \ge 0$. Hence $c = 0$. Finally, $V(\mathbf{x}(t)) \to 0$ forces $\mathbf{x}(t) \to \mathbf{0}$: for any $\rho > 0$, the minimum $m_\rho$ of $V$ on the compact set $\{\rho \le \lVert \mathbf{x} \rVert \le \varepsilon\}$ is positive, and once $V(\mathbf{x}(t)) < m_\rho$ the orbit must satisfy $\lVert \mathbf{x}(t) \rVert < \rho$ forever after.

$$
\boxed{\;V \text{ positive definite, } \dot{V} \le 0 \Rightarrow \text{stable}; \qquad \dot{V} < 0 \text{ off the origin} \Rightarrow \text{asymptotically stable}.\;}
$$

### 3.3 Derivation C: A center that linearization cannot see

**System.** For a parameter $a \in \mathbb{R}$,

$$
\dot{x} = -y + a\,x\,(x^2 + y^2), \qquad \dot{y} = x + a\,y\,(x^2 + y^2).
$$

**Step 1 (linearize).** The Jacobian at the origin is $J = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}$ for every $a$ (the cubic terms have zero derivative at $\mathbf{0}$): eigenvalues $\pm i$, so the linearization is a **center** — a non-hyperbolic equilibrium, outside Hartman–Grobman's jurisdiction.

**Step 2 (pass to polar coordinates).** With $r^2 = x^2 + y^2$ and $\theta = \arctan(y/x)$:

$$
r\dot{r} = x\dot{x} + y\dot{y} = x(-y + a x r^2) + y(x + a y r^2) = a\,r^4 ,
$$

so $\dot{r} = a r^3$. Similarly $r^2 \dot{\theta} = x\dot{y} - y\dot{x} = x(x + a y r^2) - y(-y + a x r^2) = r^2$, so $\dot{\theta} = 1$: uniform rotation.

**Step 3 (solve the radial equation).** $\dot{r} = a r^3$ separates: $r(t)^{-2} = r_0^{-2} - 2at$. For $a < 0$, $r(t) = r_0 (1 + 2\lvert a \rvert r_0^2 t)^{-1/2} \to 0$: every orbit spirals into the origin (algebraically, not exponentially) — the origin is **asymptotically stable**, a nonlinear spiral sink. For $a > 0$, $r(t)$ blows up at finite time $t^* = 1/(2 a r_0^2)$: a spiral source, **unstable**. For $a = 0$, $r$ is constant: a true center with all orbits circles.

**Step 4 (moral).** One system family, one linearization, three incompatible fates. Non-hyperbolic equilibria transfer the verdict from the Jacobian to the higher-order terms; the correct tool is a Lyapunov function ($V = r^2$ here, with $\dot{V} = 2 a r^4$ deciding by the sign of $a$).

$$
\boxed{\;\dot{r} = a r^3,\ \dot{\theta} = 1: \quad a < 0 \text{ spiral sink},\quad a = 0 \text{ center},\quad a > 0 \text{ spiral source — linearization identical in all three cases}.\;}
$$

### 3.4 Proof D: The Lotka–Volterra first integral and closed orbits

**System.** Prey $x$, predator $y$, positive parameters $\alpha, \beta, \gamma, \delta$:

$$
\dot{x} = x(\alpha - \beta y), \qquad \dot{y} = y(\delta x - \gamma), \qquad x, y > 0 .
$$

Interior equilibrium: $\mathbf{x}^* = (\gamma/\delta,\ \alpha/\beta)$.

**Step 1 (orbit ODE).** On orbits where $\dot{x} \neq 0$, divide:

$$
\frac{dy}{dx} = \frac{y(\delta x - \gamma)}{x(\alpha - \beta y)} .
$$

**Step 2 (separate and integrate).**

$$
\frac{\alpha - \beta y}{y}\, dy = \frac{\delta x - \gamma}{x}\, dx \implies \alpha \ln y - \beta y = \delta x - \gamma \ln x + C .
$$

Rearranged, the quantity

$$
H(x, y) = \delta x - \gamma \ln x + \beta y - \alpha \ln y
$$

is constant along every orbit. Direct check without dividing: $\dot{H} = (\delta - \gamma/x)\dot{x} + (\beta - \alpha/y)\dot{y} = (\delta x - \gamma)(\alpha - \beta y) + (\beta y - \alpha)(\delta x - \gamma) = 0$ identically. So $H$ is a **first integral**.

**Step 3 ($H$ is strictly convex and proper).** The Hessian of $H$ is $\operatorname{diag}(\gamma/x^2,\ \alpha/y^2)$, positive definite on the open quadrant, so $H$ is strictly convex there. Each single-variable piece $u \mapsto \delta u - \gamma\ln u$ tends to $+\infty$ both as $u \to 0^+$ and $u \to \infty$; hence $H \to +\infty$ on the boundary of the quadrant and at infinity: $H$ is proper, with a unique minimum at the critical point $\nabla H = \mathbf{0}$, i.e. exactly at $\mathbf{x}^*$, of value $H_{\min} = H(\mathbf{x}^*)$.

**Step 4 (level sets are closed curves).** For any $h > H_{\min}$, the sublevel set $\{H \le h\}$ is compact (properness) and strictly convex (convexity of $H$), and its boundary $\{H = h\}$ is a simple closed curve encircling $\mathbf{x}^*$: a strictly convex function's level set contains no line segments and each ray from the minimum meets it exactly once.

**Step 5 (orbits are the level curves — and are periodic).** Every orbit lies in one level set $\{H = h\}$. For $h > H_{\min}$ the level curve contains no equilibrium (the only interior equilibrium is $\mathbf{x}^*$, at level $H_{\min}$), so the speed $\lVert \mathbf{f} \rVert$ has a positive minimum on this compact curve. The orbit therefore traverses the curve without stalling and, moving at speed bounded below along a curve of finite length, returns to its starting point in finite time $T$: the orbit **is** the closed curve, a periodic solution. Consequently the equilibrium $\mathbf{x}^*$ is a genuine nonlinear **center**: surrounded by a continuum of closed orbits (predator–prey cycles), Lyapunov stable (use $V = H - H_{\min}$) but not asymptotically stable.

$$
\boxed{\;H(x,y) = \delta x - \gamma \ln x + \beta y - \alpha \ln y \text{ is conserved; its level curves are simple closed curves, so every interior Lotka–Volterra orbit is periodic}.\;}
$$

### 3.5 Proof E: Bendixson–Dulac criterion via Green's theorem

**Theorem.** Let $D \subseteq \mathbb{R}^2$ be open and simply connected, $\dot{x} = f(x,y)$, $\dot{y} = g(x,y)$ a $C^1$ system, and $B \in C^1(D)$ with $\nabla \cdot (B\mathbf{f}) = \partial_x(Bf) + \partial_y(Bg)$ of one strict sign on $D$ (except possibly on a measure-zero set). Then no periodic orbit lies entirely in $D$.

**Step 1 (suppose not).** Let $\Gamma \subset D$ be a periodic orbit of period $T$, traversed by the solution $(x(t), y(t))$. Since $D$ is simply connected, the interior region $R$ enclosed by the Jordan curve $\Gamma$ also lies in $D$.

**Step 2 (the flux integral along an orbit vanishes).** Consider the line integral of the vector field $B\mathbf{f}$'s flux through $\Gamma$:

$$
\oint_\Gamma B\,( f\, dy - g\, dx ) = \int_0^T B\big( f\, \dot{y} - g\, \dot{x} \big)\, dt = \int_0^T B\,( f g - g f )\, dt = 0 ,
$$

where we substituted $\dot{x} = f$, $\dot{y} = g$: the orbit is everywhere **tangent** to the vector field, so the field has zero flux through its own orbit. This holds for any Dulac function $B$.

**Step 3 (Green's theorem).** By Green's theorem on the region $R$ with boundary $\Gamma$ (oriented positively; if the orbit runs clockwise, both sides change sign together):

$$
\oint_\Gamma B f\, dy - B g\, dx = \iint_R \left( \frac{\partial (B f)}{\partial x} + \frac{\partial (B g)}{\partial y} \right) dA .
$$

**Step 4 (contradiction).** The right-hand side integrates a function of one strict sign (nonvanishing on a set of full measure) over a region of positive area, hence is nonzero. The left-hand side is $0$ by Step 2. Contradiction — no such $\Gamma$ exists. $\blacksquare$

**Remarks.** (i) Simple connectivity is essential: on an annulus the enclosed region may leave $D$, and indeed systems with divergence of one sign on an annulus can have cycles around the hole. (ii) With $B \equiv 1$ this is Bendixson's criterion: a planar system whose divergence never changes sign has no closed orbits. (iii) Choosing $B$ cleverly (e.g. $B = 1/(xy)$ in population models, or $B = e^{\kappa y}$) is an art directly analogous to choosing a Lyapunov function.

$$
\boxed{\;\text{If } \partial_x(Bf) + \partial_y(Bg) \text{ keeps one strict sign on a simply connected } D, \text{ the system has no periodic orbit inside } D.\;}
$$

### 3.6 Proof F: Gradient systems have no periodic orbits

**Setting.** A **gradient system** is $\dot{\mathbf{x}} = -\nabla V(\mathbf{x})$ for some $C^2$ potential $V : \mathbb{R}^n \to \mathbb{R}$ (gradient descent in continuous time).

**Step 1 ($V$ decreases along orbits).** Along any solution,

$$
\frac{d}{dt} V(\mathbf{x}(t)) = \nabla V \cdot \dot{\mathbf{x}} = -\lVert \nabla V(\mathbf{x}(t)) \rVert^2 \le 0 ,
$$

with equality at time $t$ iff $\nabla V(\mathbf{x}(t)) = \mathbf{0}$, i.e. iff $\mathbf{x}(t)$ is an equilibrium.

**Step 2 (nonconstant orbits decrease strictly).** Suppose a solution satisfies $\nabla V(\mathbf{x}(t_0)) = \mathbf{0}$ at some time. Then $\mathbf{x}(t_0)$ is an equilibrium, and by uniqueness of solutions the whole solution is the constant one. Contrapositive: along a **nonconstant** solution, $\lVert \nabla V \rVert^2 > 0$ at every time, so $t \mapsto V(\mathbf{x}(t))$ is *strictly* decreasing.

**Step 3 (periodicity is impossible).** Suppose $\mathbf{x}(t)$ is a nonconstant periodic solution with period $T > 0$. Then

$$
0 = V(\mathbf{x}(T)) - V(\mathbf{x}(0)) = -\int_0^T \lVert \nabla V(\mathbf{x}(t)) \rVert^2\, dt < 0 ,
$$

where the final strict inequality holds because the integrand is continuous, nonnegative, and strictly positive at every $t$ by Step 2. Contradiction. $\blacksquare$

**Step 4 (interpretation and contrast).** Gradient flows can only descend: their omega-limit sets consist of critical points of $V$, and (with LaSalle) bounded orbits converge to the critical set. Contrast the two great classes: **gradient systems** ($\dot{\mathbf{x}} = -\nabla V$) forbid cycles because a scalar strictly decreases, while **Hamiltonian/conservative systems** ($\dot{H} = 0$, like Lotka–Volterra) favor cycles because a scalar is frozen. Most systems are neither, but recognizing a hidden gradient or conserved structure instantly settles the limit-cycle question. This is also why plain gradient descent on a fixed loss never oscillates persistently — and why GAN dynamics, which are *not* gradient systems, can (Section 5.3).

$$
\boxed{\;\dot{\mathbf{x}} = -\nabla V \text{ admits no nonconstant periodic orbit: } V \text{ strictly decreases along every nonconstant solution}.\;}
$$

### 3.7 Proof G (sketch): Linearized stability via the Lyapunov equation

**Theorem.** Let $\dot{\mathbf{x}} = A\mathbf{x} + \mathbf{h}(\mathbf{x})$ with $\mathbf{h}$ being $C^1$, $\mathbf{h}(\mathbf{0}) = \mathbf{0}$, $\lVert \mathbf{h}(\mathbf{x}) \rVert \le k \lVert \mathbf{x} \rVert^2$ near $\mathbf{0}$ (Taylor remainder of a $C^2$ field). If every eigenvalue of $A$ satisfies $\operatorname{Re}\lambda < 0$ (Hurwitz), then $\mathbf{0}$ is asymptotically — indeed exponentially — stable for the **nonlinear** system.

**Step 1 (solve the Lyapunov equation).** Fix any $Q \succ 0$ (say $Q = I$) and define

$$
P = \int_0^\infty e^{A^{\mathsf{T}} t}\, Q\, e^{A t}\, dt .
$$

The integral converges because $\lVert e^{At} \rVert \le M e^{-ct}$ for some $M, c > 0$ when $A$ is Hurwitz. $P$ is symmetric and positive definite ($\mathbf{x}^{\mathsf{T}} P \mathbf{x} = \int_0^\infty \lVert Q^{1/2} e^{At}\mathbf{x} \rVert^2 dt > 0$ for $\mathbf{x} \neq \mathbf{0}$). Moreover

$$
A^{\mathsf{T}} P + P A = \int_0^\infty \frac{d}{dt}\left( e^{A^{\mathsf{T}} t} Q e^{A t} \right) dt = \left[ e^{A^{\mathsf{T}} t} Q e^{A t} \right]_0^\infty = -Q ,
$$

using $e^{At} \to 0$. So the **Lyapunov equation** $A^{\mathsf{T}}P + PA = -Q$ is solved constructively.

**Step 2 (quadratic Lyapunov function for the nonlinear system).** Let $V(\mathbf{x}) = \mathbf{x}^{\mathsf{T}} P \mathbf{x}$. Along nonlinear solutions,

$$
\dot{V} = \dot{\mathbf{x}}^{\mathsf{T}} P \mathbf{x} + \mathbf{x}^{\mathsf{T}} P \dot{\mathbf{x}} = \mathbf{x}^{\mathsf{T}} (A^{\mathsf{T}} P + P A) \mathbf{x} + 2\, \mathbf{x}^{\mathsf{T}} P\, \mathbf{h}(\mathbf{x}) = -\mathbf{x}^{\mathsf{T}} Q \mathbf{x} + 2\, \mathbf{x}^{\mathsf{T}} P \mathbf{h}(\mathbf{x}) .
$$

**Step 3 (the quadratic term wins near the origin).** Bound each piece: $\mathbf{x}^{\mathsf{T}} Q \mathbf{x} \ge \lambda_{\min}(Q) \lVert \mathbf{x} \rVert^2$ and $2 \lvert \mathbf{x}^{\mathsf{T}} P \mathbf{h}(\mathbf{x}) \rvert \le 2 \lVert P \rVert\, k\, \lVert \mathbf{x} \rVert^3$. Hence for $\lVert \mathbf{x} \rVert < \rho := \lambda_{\min}(Q) / (4 k \lVert P \rVert)$,

$$
\dot{V} \le -\lambda_{\min}(Q) \lVert \mathbf{x} \rVert^2 + 2 k \lVert P \rVert\, \lVert \mathbf{x} \rVert^3 \le -\tfrac{1}{2}\, \lambda_{\min}(Q)\, \lVert \mathbf{x} \rVert^2 < 0 \quad (\mathbf{x} \neq \mathbf{0}).
$$

**Step 4 (conclude).** On the ball $B_\rho$, $V$ is a strict Lyapunov function, so by Theorem 3.2 the origin is asymptotically stable. Exponential rate: since $\lambda_{\min}(P)\lVert \mathbf{x} \rVert^2 \le V \le \lambda_{\max}(P) \lVert \mathbf{x} \rVert^2$, Step 3 gives $\dot{V} \le -\kappa V$ with $\kappa = \lambda_{\min}(Q)/(2\lambda_{\max}(P))$, and Grönwall yields $V(t) \le V(0)e^{-\kappa t}$, hence $\lVert \mathbf{x}(t) \rVert \le C e^{-\kappa t/2} \lVert \mathbf{x}(0) \rVert$. This proves the "stable half" of Hartman–Grobman by pure Lyapunov methods — no conjugacy needed — and is exactly how control theorists certify controllers.

$$
\boxed{\;\operatorname{Re}\lambda_i(A) < 0 \ \Rightarrow\ P = {\textstyle\int_0^\infty} e^{A^{\mathsf{T}}t} Q e^{At} dt \text{ solves } A^{\mathsf{T}}P + PA = -Q, \text{ and } V = \mathbf{x}^{\mathsf{T}}P\mathbf{x} \text{ certifies local exponential stability of the nonlinear system}.\;}
$$

## 4. Computational & Algorithmic Insights

### 4.1 Drawing phase portraits numerically

A phase portrait is computed, not divined. The standard pipeline:

```text
ALGORITHM PhasePortrait(f, window, n_grid)
  1. Sample the vector field on an n_grid x n_grid lattice:
       U[i,j], V[i,j] = f(x_i, y_j)
  2. Streamlines (matplotlib streamplot):
       - normalize speed for direction field, keep magnitude for coloring
       - seed points on a coarse sub-grid; integrate each seed forward
         and backward with adaptive RK (dopri5 / RK45)
       - terminate a streamline when it exits the window, approaches
         another streamline (density control), or stalls near an
         equilibrium (speed below tol)
  3. Overlays that carry the mathematics:
       - nullclines: contour lines f1 = 0 and f2 = 0 (marching squares)
       - equilibria: Newton iterations seeded at nullcline intersections
       - classification: eigvals of finite-difference or autodiff Jacobian,
         then tau-Delta lookup
       - separatrices: integrate from x* +- eps * v_stable BACKWARD in
         time and from x* +- eps * v_unstable FORWARD in time
```

Cost is dominated by streamline integration: $O(S \cdot L / h)$ right-hand-side evaluations for $S$ seeds, arc length $L$, mean step $h$. Two practical subtleties: (i) **saddle separatrices are unstable objects** — integrating *toward* a saddle is numerically hopeless (any rounding error is amplified by $e^{\lambda_u t}$), so always integrate *away* from the saddle along eigenvector offsets, in the appropriate time direction; (ii) near centers, non-symplectic integrators like RK4 slowly spiral (energy drift), fabricating a false spiral — for conservative systems verify with the first integral or use a symplectic method.

### 4.2 Event detection and stiffness in fast–slow systems

**Event detection.** Many phase-plane quantities are *events*: a Poincaré-section crossing ($y = 0$ with $\dot{y} > 0$), period measurement of a limit cycle, or basin escape. Modern integrators (`scipy.integrate.solve_ivp` with `events=`) detect a sign change of an event function $e(t, \mathbf{x})$ across a step, then locate the root in $t$ with the solver's **dense output** (a local interpolant), achieving event times to interpolation accuracy without shrinking steps. Period of a limit cycle: record successive upward crossings $t_k$ of a section and report $T = t_{k+1} - t_k$ after transients decay.

**Stiffness from time-scale separation.** The van der Pol oscillator with large $\mu$,

$$
\dot{x} = y, \qquad \dot{y} = \mu (1 - x^2) y - x ,
$$

is the canonical **fast–slow** system: along the relaxation cycle the state alternates between slow drift (time scale $\mu$) and near-instantaneous jumps (time scale $1/\mu$). The Jacobian eigenvalue $\approx \mu(1 - x^2)$ makes the equation **stiff** off the slow manifold: an explicit method must obey a stability restriction $h \lesssim C/\mu$ *everywhere*, even where the solution is smooth and boring, wasting $O(\mu)$ steps per cycle. Implicit stiff solvers (BDF, Radau) pay a linear-algebra cost per step ($O(n^3)$ for the Newton solves, trivial for $n = 2$) but take steps sized by *accuracy*, not stability — orders of magnitude fewer. Rule of thumb: when a phase portrait shows trajectories slamming onto an attracting slow curve and creeping along it, switch to a stiff solver.

### 4.3 Computing basins of attraction

The **basin** of an attractor $A$ is $\mathcal{B}(A) = \{\mathbf{x}_0 : \omega(\mathbf{x}_0) \subseteq A\}$ — an open invariant set whose boundaries are typically stable manifolds (separatrices) of saddles.

```text
ALGORITHM BasinMap(f, attractors, window, N)
  1. Precompute attractor list (equilibria with Re(lambda) < 0,
     limit cycles found by long integration).
  2. For each of the N x N grid points x0:
       integrate x' = f(x) with events:
         - CAPTURE: distance to attractor_k < r_capture  -> label k
         - ESCAPE:  ||x|| > R_escape                     -> label infinity
         - TIMEOUT: t > T_max                            -> label undecided
  3. Color the grid by label.
Complexity: O(N^2 * steps); embarrassingly parallel across grid points.
```

Refinements that matter in practice: (i) choose $r_{\text{capture}}$ inside a *verified* attracting neighborhood (e.g. a Lyapunov sublevel set), or fast-but-wrong labels occur near basin boundaries; (ii) near the boundary, integration time diverges — trajectories linger along the saddle's stable manifold — so `TIMEOUT` cells mark the separatrix, which is a feature, not a bug; (iii) for the damped pendulum, basins are infinitely many diagonal strips (one per rotation count before settling), and the picture sharpens as damping decreases; (iv) in dimension $\ge 3$ or for driven systems, basin boundaries can be **fractal**, making long-term prediction near the boundary practically impossible even though the dynamics is deterministic. A cheaper alternative to full integration, when a Lyapunov function $V$ is known, is to report the largest sublevel set $\{V \le c\}$ inside which $\dot{V} < 0$: a certified inner approximation of the basin — this is how regions of attraction are certified in control (sum-of-squares programming automates the search for such $V$).

## 5. Real-World Physics & AI/ML Applications

### 5.1 The damped pendulum: a complete phase-portrait case study

The pendulum with linear drag ($g/L = 1$, damping $b \ge 0$):

$$
\dot{\theta} = \omega, \qquad \dot{\omega} = -\sin\theta - b\,\omega .
$$

**Equilibria** at $(n\pi, 0)$. Jacobian $J = \begin{pmatrix} 0 & 1 \\ -\cos\theta^* & -b \end{pmatrix}$, so $\Delta = \cos\theta^*$, $\tau = -b$.

- **Bottom equilibria** ($\theta^* = 2k\pi$): $\Delta = 1 > 0$, $\tau = -b$. Undamped $b = 0$: centers (confirmed nonlinearly by energy conservation, not by linearization!). Underdamped $0 < b < 2$: spiral sinks — the familiar decaying swing. Overdamped $b \ge 2$: stable nodes — no oscillation, monotone settling.
- **Top equilibria** ($\theta^* = (2k+1)\pi$): $\Delta = -1 < 0$: **saddles** for every $b$ — the inverted pendulum, whose stable separatrices are the razor's edge between falling left and falling right.

**Global structure via energy.** $E(\theta, \omega) = \tfrac{1}{2}\omega^2 + 1 - \cos\theta$ obeys $\dot{E} = -b\,\omega^2 \le 0$. For $b = 0$, orbits are level curves of $E$: closed libration orbits for $E < 2$, running (over-the-top) orbits for $E > 2$, and the **separatrix** $E = 2$ — heteroclinic connections joining adjacent saddles. For $b > 0$, LaSalle's principle applies on each compact sublevel set: $\dot{E} = 0$ forces $\omega = 0$, and the largest invariant set inside $\{\omega = 0\}$ is the set of equilibria; hence every orbit converges to an equilibrium, and (saddles attracting only their measure-zero separatrices) almost every initial condition spirals into some bottom equilibrium. The basins are infinite spiral-armed strips indexed by how many times the pendulum swings over the top before capture — a picture worth computing with the algorithm of Section 4.3.

### 5.2 Relaxation oscillations: the van der Pol oscillator

$\ddot{x} - \mu(1 - x^2)\dot{x} + x = 0$ models vacuum-tube circuits, heartbeats (van der Pol's own application), and neuronal spiking (FitzHugh–Nagumo is its cousin). The origin is the only equilibrium; $\tau = \mu > 0$, $\Delta = 1$ make it a spiral source ($\mu < 2$) or unstable node ($\mu \ge 2$), while the nonlinear damping $-\mu(1-x^2)\dot{x}$ pumps energy in for $\lvert x \rvert < 1$ and drains it for $\lvert x \rvert > 1$. Liénard's theorem (or a Poincaré–Bendixson trapping annulus, Exercise L3.4) yields a **unique, globally attracting limit cycle**: every nonzero initial condition, however small or huge, converges to the same self-sustained oscillation — amplitude set by the system, not the initial condition, which is the defining signature of a limit cycle as against a center's initial-condition-dependent orbits. For $\mu \gg 1$ the cycle degenerates into **relaxation oscillations** — slow crawls along the cubic nullcline punctuated by instantaneous jumps — with period $T \approx (3 - 2\ln 2)\mu$, the stiff regime of Section 4.2.

### 5.3 Epidemics and ecology: thresholds and cycles

**SIR with vital dynamics.** Susceptible/infected fractions with birth-death rate $\mu$, transmission $\beta$, recovery $\gamma$:

$$
\dot{S} = \mu - \beta S I - \mu S, \qquad \dot{I} = \beta S I - (\gamma + \mu) I .
$$

The **basic reproduction number** is $R_0 = \beta / (\gamma + \mu)$. The disease-free equilibrium $(1, 0)$ has Jacobian eigenvalues $-\mu$ and $(\gamma + \mu)(R_0 - 1)$: asymptotically stable iff $R_0 < 1$ — the epidemic threshold as an eigenvalue-sign statement. For $R_0 > 1$ the disease-free state becomes a saddle and a unique **endemic equilibrium** $S^* = 1/R_0$, $I^* = \mu(R_0 - 1)/\beta$ enters the positive quadrant; its Jacobian has $\tau < 0$ and $\Delta = \mu(\gamma + \mu)(R_0 - 1) > 0$, so it is asymptotically stable (typically a spiral sink: damped epidemic waves as observed in measles data). This is a **transcritical bifurcation** at $R_0 = 1$ — the 1D normal form of Section 2.6 living inside a planar model, and the mathematical content of vaccination targets ($R_0$ effectively below 1).

**Lotka–Volterra ecology.** Section 3.4 proved every interior orbit is a closed predator–prey cycle around the center $(\gamma/\delta, \alpha/\beta)$ — matching the celebrated boom-bust cycles of lynx–hare pelt records. But the structure is fragile: the cycles are *not* limit cycles, their amplitude is set by initial conditions, and any perturbation (logistic prey growth, predator satiation) destroys the first integral and collapses the continuum of cycles into a spiral sink or, via a Hopf bifurcation, a single limit cycle (Rosenzweig–MacArthur). A model whose qualitative prediction changes under arbitrarily small perturbation is **structurally unstable** — a warning that centers in applied models should always be interrogated.

### 5.4 GAN training dynamics: a two-player vector field

A GAN is a min–max problem $\min_x \max_y V(x, y)$: the generator (parameters $x$) descends $V$ while the discriminator (parameters $y$) ascends it. **Simultaneous gradient descent-ascent** in continuous time is the vector field

$$
\dot{x} = -\nabla_x V(x, y), \qquad \dot{y} = +\nabla_y V(x, y) ,
$$

which is **not** a gradient system — no scalar potential decreases along it — so Section 3.6's no-cycles guarantee evaporates. The minimal model (Dirac-GAN): $V(x,y) = x y$, giving

$$
\dot{x} = -y, \qquad \dot{y} = x, \qquad J = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}, \qquad \lambda = \pm i .
$$

The unique equilibrium $(0,0)$ — the solution of the game — is a **center**: training orbits circle it at constant radius forever. Worse, explicit Euler discretization with step $h$ maps radii by $r \mapsto r\sqrt{1 + h^2} > r$: the discretized training **spirals outward**. This is the phase-plane explanation of the notorious oscillation and divergence of GAN training: the game's Jacobian at equilibrium has a dominant antisymmetric (rotational) part rather than a symmetric (contractive) part.

**Regularization as damping.** Adding a gradient penalty or weight decay $\varepsilon$ modifies the field to $\dot{x} = -y - \varepsilon x$, $\dot{y} = x - \varepsilon y$: now $\tau = -2\varepsilon < 0$, $\Delta = 1 + \varepsilon^2 > 0$, eigenvalues $-\varepsilon \pm i$ — a **spiral sink**. Exactly like the damped pendulum, a little dissipation converts the center into an attractor; this is the phase-plane content of gradient penalties, consensus optimization, and negative-momentum methods for games. The design problem "make GAN training converge" is literally the problem "move the equilibrium's eigenvalues off the imaginary axis into the left half-plane."

### 5.5 Optimization and attractor networks as stability theory

**Gradient flow and the PL inequality.** Training-as-flow: $\dot{\boldsymbol{\theta}} = -\nabla L(\boldsymbol{\theta})$. If $L$ satisfies the Polyak–Łojasiewicz inequality $\tfrac{1}{2}\lVert \nabla L \rVert^2 \ge \mu\, (L - L^*)$ (satisfied by strongly convex losses but also by many overparameterized networks), then

$$
\frac{d}{dt}(L - L^*) = -\lVert \nabla L \rVert^2 \le -2\mu\, (L - L^*) \implies L(\boldsymbol{\theta}(t)) - L^* \le e^{-2\mu t}\left( L(\boldsymbol{\theta}(0)) - L^* \right)
$$

by Grönwall's inequality: the loss gap is a Lyapunov function decaying exponentially, i.e. the minimum set is **exponentially stable** and linear convergence rates of gradient descent are its Euler discretization.

**Hopfield networks: computation by Lyapunov descent.** An attractor network stores memories as asymptotically stable equilibria. The continuous Hopfield model with symmetric weights admits the energy $E(\mathbf{u})$ as a strict Lyapunov function ($\dot{E} \le 0$, zero only at equilibria), so by LaSalle every trajectory converges to an equilibrium: **retrieval is convergence to an attractor, and error-correction radius is the basin of attraction**. Modern Hopfield layers and energy-based models inherit the same logic: design the energy landscape, and stability theory guarantees the computation. Neural ODEs close the circle — a learned vector field whose stability properties (certified by learned Lyapunov functions) control robustness of the model itself.
