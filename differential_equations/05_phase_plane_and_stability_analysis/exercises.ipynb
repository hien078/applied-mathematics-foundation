### Problem L0.1: Classification from Eigenvalues

**Problem Statement:** For a planar linear system $\dot{\mathbf{x}} = A\mathbf{x}$, classify the origin (type and stability) for each Jacobian spectrum:

(a) $\lambda = -2, -5$  (b) $\lambda = 3, -1$  (c) $\lambda = -1 \pm 2i$  (d) $\lambda = \pm 3i$  (e) $\lambda = 2 \pm i$

**Intuition:** Real parts control growth/decay; imaginary parts control rotation; a sign split means a saddle.

**Solution:**

The general solution is a combination of modes $e^{\lambda t}$; behavior follows from the real and imaginary parts of each $\lambda$.

(a) $\lambda = -2, -5$: both real and negative. Every mode decays monotonically; generic orbits approach tangent to the slow eigenvector (eigenvalue $-2$). **Stable node — asymptotically (indeed exponentially) stable.**

(b) $\lambda = 3, -1$: real with opposite signs. One invariant line attracts, one repels; all other orbits eventually follow the unstable eigendirection. **Saddle — unstable.**

(c) $\lambda = -1 \pm 2i$: complex with negative real part. Solutions are decaying spirals with envelope $e^{-t}$ and angular frequency $2$. **Spiral sink — asymptotically stable.**

(d) $\lambda = \pm 3i$: purely imaginary. Solutions are ellipses traversed with period $2\pi/3$; orbits neither approach nor leave the origin. **Center — Lyapunov stable but not asymptotically stable.**

(e) $\lambda = 2 \pm i$: complex with positive real part. Growing spirals with envelope $e^{2t}$. **Spiral source — unstable.**

$$
\boxed{\text{(a) stable node, (b) saddle, (c) spiral sink, (d) center, (e) spiral source}}
$$

> **Key Takeaway:** Sign pattern of real parts fixes stability; presence of imaginary parts fixes rotation — two independent dials that generate the whole classification.

### Problem L0.2: Can Trajectories Cross?

**Problem Statement:** In the phase plane of an autonomous system $\dot{\mathbf{x}} = \mathbf{f}(\mathbf{x})$ with $\mathbf{f}$ continuously differentiable, can two distinct trajectories cross? Can a single trajectory cross itself? What is the one legitimate way distinct orbits can "meet"?

**Intuition:** At a crossing point the vector field would have to point in two directions at once.

**Solution:**

**Step 1.** Since $\mathbf{f}$ is $C^1$, it is locally Lipschitz, so the Picard–Lindelöf theorem guarantees a *unique* solution through every point.

**Step 2.** Suppose two distinct orbits both passed through a point $\mathbf{p}$. Because the system is autonomous, we may time-shift each solution so both satisfy $\mathbf{x}(0) = \mathbf{p}$. Uniqueness forces them to coincide for all time — so they were the *same orbit*, contradiction. Hence distinct orbits never share a point: **no crossings**.

**Step 3.** A single trajectory crossing itself transversally would put two distinct velocity directions at one point, impossible since the velocity is the fixed vector $\mathbf{f}(\mathbf{p})$. A trajectory *can* return to a previous point tangentially along the same direction — but then, by the same uniqueness argument, it must repeat forever: it is a **closed (periodic) orbit**, not a self-crossing.

**Step 4.** The legitimate meeting: distinct orbits may share **limit points**. Trajectories can approach the same equilibrium as $t \to \infty$ (infinitely slowly, never arriving in finite time) or spiral onto the same limit cycle. Approach is asymptotic, so no point is ever shared at finite time.

$$
\boxed{\text{Uniqueness forbids all crossings; orbits partition the plane and can only share asymptotic limit sets.}}
$$

> **Key Takeaway:** The non-crossing rule is uniqueness wearing geometric clothing — it is what makes the planar phase portrait a readable, partitioned picture.

### Problem L0.3: What Is a Nullcline?

**Problem Statement:** Define the nullclines of $\dot{x} = f(x, y)$, $\dot{y} = g(x, y)$. What is the direction of the flow *on* each nullcline, where are the equilibria, and why are nullclines generally *not* trajectories? Illustrate with $\dot{x} = y - x$, $\dot{y} = 1 - xy$.

**Intuition:** Nullclines are where the flow is momentarily purely vertical or purely horizontal.

**Solution:**

**Step 1 (definitions).** The $x$-nullcline is the curve $\{f(x,y) = 0\}$: there $\dot{x} = 0$, so the flow is **vertical** (purely in $y$). The $y$-nullcline is $\{g(x,y) = 0\}$: there $\dot{y} = 0$, so the flow is **horizontal**. Equilibria are exactly the intersections of an $x$-nullcline with a $y$-nullcline, where both components vanish.

**Step 2 (why not trajectories).** A trajectory must follow the vector field at every point. On the $x$-nullcline the field is vertical, but the nullcline itself is generally not a vertical curve — so the flow *crosses* it rather than running along it. (Exception: if a nullcline happens to be an invariant curve, e.g. an axis in a population model where both conditions align.)

**Step 3 (example).** For $\dot{x} = y - x$, $\dot{y} = 1 - xy$: the $x$-nullcline is the line $y = x$ (flow vertical there), the $y$-nullcline is the hyperbola $y = 1/x$ (flow horizontal there). Intersections: $x = y$ and $xy = 1$ give $x^2 = 1$, so equilibria at $(1,1)$ and $(-1,-1)$. The two curves split the plane into regions where the signs of $(\dot{x}, \dot{y})$ are constant — e.g. at $(0, 2)$: $\dot{x} = 2 > 0$, $\dot{y} = 1 > 0$, so the flow points up-right; recording one sample point per region yields the skeleton of the entire portrait before any integration.

$$
\boxed{\text{Nullclines: curves where one velocity component vanishes; their intersections are the equilibria; the flow crosses them vertically or horizontally.}}
$$

> **Key Takeaway:** Nullclines are the free information in a phase portrait — sign analysis on the regions they carve out sketches the global flow with zero computation.

### Problem L0.4: Center versus Spiral — What Can Linearization Say?

**Problem Statement:** The Jacobian of a nonlinear system at an equilibrium has eigenvalues $\pm i\omega$ with $\omega \neq 0$. A classmate concludes: "the equilibrium is a center, so it is stable." Diagnose the error precisely. What *can* be concluded, and which tools decide the true behavior?

**Intuition:** Purely imaginary eigenvalues sit exactly on the boundary where the Hartman–Grobman theorem loses jurisdiction.

**Solution:**

**Step 1 (the error).** The Hartman–Grobman theorem guarantees the nonlinear flow is topologically conjugate to its linearization only at **hyperbolic** equilibria — all eigenvalues off the imaginary axis. With $\lambda = \pm i\omega$ the equilibrium is non-hyperbolic, so the theorem is silent, and the linear verdict "center" transfers no guarantee whatsoever.

**Step 2 (what can happen).** The truncated quadratic/cubic terms decide. The family $\dot{x} = -y + a x (x^2 + y^2)$, $\dot{y} = x + a y (x^2 + y^2)$ has linearization with $\lambda = \pm i$ for every $a$, yet in polar coordinates $\dot{r} = a r^3$: a spiral sink for $a < 0$, a true center for $a = 0$, a spiral source for $a > 0$. All three behaviors hide behind one linearization.

**Step 3 (what can be concluded).** Only this: the equilibrium is not a node or saddle, rotation is locally present, and stability is **undetermined** by first-order data.

**Step 4 (deciding tools).** (i) A **Lyapunov function**: if some positive definite $V$ has $\dot{V} < 0$ nearby, asymptotic stability; $\dot{V} \equiv 0$ gives a conserved quantity and a genuine center. (ii) A **first integral or symmetry** (Hamiltonian structure, time-reversibility) forces closed orbits. (iii) Normal-form/averaging computation of the first Lyapunov coefficient (the sign of the effective $a$ above).

$$
\boxed{\text{Purely imaginary eigenvalues: linearization is inconclusive; only nonlinear certificates decide between center, spiral sink, and spiral source.}}
$$

> **Key Takeaway:** "Linear center" is a question, not an answer — hyperbolicity is the license that lets the Jacobian speak for the nonlinear system.

### Problem L1.1: Trace–Determinant Classification Drill

**Problem Statement:** Classify the origin of $\dot{\mathbf{x}} = A\mathbf{x}$ for each matrix using only $\tau = \operatorname{tr} A$, $\Delta = \det A$, and the discriminant $\tau^2 - 4\Delta$:

$$
A_1 = \begin{pmatrix} 2 & 1 \\ 1 & 2 \end{pmatrix}, \qquad A_2 = \begin{pmatrix} 0 & 1 \\ -2 & -3 \end{pmatrix}, \qquad A_3 = \begin{pmatrix} 1 & -4 \\ 2 & -3 \end{pmatrix}, \qquad A_4 = \begin{pmatrix} 0 & 1 \\ -4 & 0 \end{pmatrix}.
$$

**Intuition:** Two invariants of the matrix suffice — no eigenvector computation needed for the type.

**Solution:**

**Matrix $A_1$:** $\tau = 4$, $\Delta = 4 - 1 = 3 > 0$, discriminant $16 - 12 = 4 > 0$. Real eigenvalues, both positive since $\tau > 0$: $\lambda = (4 \pm 2)/2 = 3, 1$. **Unstable node.**

**Matrix $A_2$:** $\tau = -3$, $\Delta = 0 - (-2) = 2 > 0$, discriminant $9 - 8 = 1 > 0$. Real, both negative: $\lambda = (-3 \pm 1)/2 = -1, -2$. **Stable node**, orbits tangent to the slow eigenvector of $\lambda = -1$.

**Matrix $A_3$:** $\tau = 1 - 3 = -2$, $\Delta = (1)(-3) - (-4)(2) = -3 + 8 = 5 > 0$, discriminant $4 - 20 = -16 < 0$. Complex pair $\lambda = -1 \pm 2i$ with negative real part. **Spiral sink.**

**Matrix $A_4$:** $\tau = 0$, $\Delta = 0 - (-4) = 4 > 0$. Purely imaginary $\lambda = \pm 2i$. **Center** (legitimately so — for a *linear* system the classification is exact, no Hartman–Grobman caveat needed): ellipses with period $\pi$.

$$
\boxed{A_1: \text{unstable node}; \quad A_2: \text{stable node}; \quad A_3: \text{spiral sink}; \quad A_4: \text{center}}
$$

> **Key Takeaway:** $(\tau, \Delta)$ plus the discriminant place any planar linear system on the trace–determinant chart in three arithmetic operations.

### Problem L1.2: Anatomy of a Saddle

**Problem Statement:** For $\dot{\mathbf{x}} = A\mathbf{x}$ with $A = \begin{pmatrix} 1 & 3 \\ 1 & -1 \end{pmatrix}$, verify the origin is a saddle, find the eigenvalues and eigenvectors, identify the stable and unstable separatrices, and describe the phase portrait including the asymptotic behavior of a generic orbit.

**Intuition:** The four eigenline half-orbits are the skeleton; every other orbit drapes over them like a hyperbola.

**Solution:**

**Step 1 (classification).** $\tau = 0$, $\Delta = (1)(-1) - (3)(1) = -4 < 0$: negative determinant means real eigenvalues of opposite sign — a **saddle**, always unstable.

**Step 2 (eigenvalues).** $\lambda = \tfrac{1}{2}(\tau \pm \sqrt{\tau^2 - 4\Delta}) = \pm \tfrac{1}{2}\sqrt{16} = \pm 2$.

**Step 3 (eigenvectors).** For $\lambda = 2$: $(A - 2I)\mathbf{v} = \mathbf{0}$ with $A - 2I = \begin{pmatrix} -1 & 3 \\ 1 & -3 \end{pmatrix}$, giving $v_1 = 3 v_2$, so $\mathbf{v}_u = (3, 1)^{\mathsf{T}}$. For $\lambda = -2$: $A + 2I = \begin{pmatrix} 3 & 3 \\ 1 & 1 \end{pmatrix}$, giving $v_1 = -v_2$, so $\mathbf{v}_s = (1, -1)^{\mathsf{T}}$.

**Step 4 (separatrices).** The general solution is $\mathbf{x}(t) = c_1 e^{2t}(3,1)^{\mathsf{T}} + c_2 e^{-2t}(1,-1)^{\mathsf{T}}$. The **stable separatrix** is the line through $(1,-1)$: initial conditions with $c_1 = 0$ decay into the origin as $e^{-2t}$. The **unstable separatrix** is the line through $(3,1)$: solutions with $c_2 = 0$ leave the origin as $e^{2t}$. These two lines are the only orbits that reach (or emanate from) the equilibrium.

**Step 5 (generic orbit).** If $c_1 \neq 0$ and $c_2 \neq 0$: as $t \to +\infty$ the $e^{2t}$ term dominates, so the orbit sweeps out to infinity **asymptotic to the unstable line** $\operatorname{span}(3,1)$; as $t \to -\infty$ it is asymptotic to the stable line $\operatorname{span}(1,-1)$. The portrait: two crossing invariant lines, four hyperbolic sectors of orbits that fly in along $(1,-1)$, turn, and fly out along $(3,1)$. The stable line divides the plane into the two "sides" that exit along opposite ends of the unstable line — the organizing role separatrices play in every nonlinear portrait containing a saddle.

$$
\boxed{\lambda = \pm 2; \quad \text{unstable separatrix } \operatorname{span}(3,1), \quad \text{stable separatrix } \operatorname{span}(1,-1)}
$$

> **Key Takeaway:** A saddle is fully described by its two eigenlines — the stable one decides *which way* orbits eventually go, making separatrices the borders of basins in nonlinear systems.

### Problem L1.3: Equilibria of a Nonlinear System (Damped Duffing)

**Problem Statement:** Find all equilibria of

$$
\dot{x} = y, \qquad \dot{y} = x - x^3 - y ,
$$

linearize at each, classify them, and describe the resulting global picture.

**Intuition:** This is a particle in the double-well potential $V(x) = -x^2/2 + x^4/4$ with unit friction.

**Solution:**

**Step 1 (equilibria).** Need $y = 0$ and $x - x^3 = x(1 - x^2) = 0$: equilibria $(0,0)$, $(1, 0)$, $(-1, 0)$.

**Step 2 (Jacobian).**

$$
J(x, y) = \begin{pmatrix} 0 & 1 \\ 1 - 3x^2 & -1 \end{pmatrix}.
$$

**Step 3 (classify $(0,0)$).** $J = \begin{pmatrix} 0 & 1 \\ 1 & -1 \end{pmatrix}$: $\tau = -1$, $\Delta = -1 < 0$. **Saddle** (hyperbolic, so Hartman–Grobman makes the nonlinear verdict rigorous). Eigenvalues $\lambda = \tfrac{1}{2}(-1 \pm \sqrt{5})$, i.e. $\lambda_u \approx 0.618$, $\lambda_s \approx -1.618$.

**Step 4 (classify $(\pm 1, 0)$).** There $1 - 3x^2 = -2$, so $J = \begin{pmatrix} 0 & 1 \\ -2 & -1 \end{pmatrix}$: $\tau = -1 < 0$, $\Delta = 2 > 0$, discriminant $1 - 8 = -7 < 0$. **Spiral sinks**, eigenvalues $\lambda = \tfrac{1}{2}(-1 \pm i\sqrt{7})$: decay rate $e^{-t/2}$ with frequency $\sqrt{7}/2 \approx 1.32$. Both are asymptotically (exponentially) stable.

**Step 5 (global picture).** Energy $E = \tfrac{1}{2}y^2 - \tfrac{1}{2}x^2 + \tfrac{1}{4}x^4$ satisfies $\dot{E} = y\dot{y} + (x^3 - x)\dot{x} = y(x - x^3 - y) + (x^3 - x)y = -y^2 \le 0$: the system dissipates. By LaSalle (on the compact sublevel sets of $E$, which are bounded since $E \to \infty$), every orbit converges to an equilibrium. The saddle at the origin attracts only its one-dimensional stable manifold; **almost every** initial condition spirals into one of the two wells $(\pm 1, 0)$. The stable manifold of the origin is the **separatrix** dividing the two basins — a bistable system whose long-run memory is one bit: which well.

$$
\boxed{(0,0) \text{ saddle}; \quad (\pm 1, 0) \text{ spiral sinks}; \quad \text{basins separated by the saddle's stable manifold}}
$$

> **Key Takeaway:** Find equilibria, linearize, classify, then let a dissipated energy plus LaSalle stitch the local pictures into a global bistable portrait.

### Problem L1.4: Lyapunov Verification for $\dot{x} = -x^3$

**Problem Statement:** Using $V(x) = \tfrac{1}{2}x^2$, prove the origin of $\dot{x} = -x^3$ is asymptotically stable. Then solve the equation exactly and show the origin is **not** exponentially stable. Why does linearization fail here?

**Intuition:** The cubic restoring force gets arbitrarily weak near the origin — the orbit keeps arriving, but ever more slowly.

**Solution:**

**Step 1 (linearization fails).** $f(x) = -x^3$ has $f'(0) = 0$: the linearization is $\dot{\eta} = 0$, eigenvalue zero — non-hyperbolic, so no conclusion transfers.

**Step 2 (Lyapunov certificate).** $V(x) = \tfrac{1}{2}x^2$ is positive definite, and

$$
\dot{V} = x \cdot \dot{x} = x(-x^3) = -x^4 < 0 \quad \text{for } x \neq 0 .
$$

$V$ is a strict Lyapunov function, so by Lyapunov's theorem the origin is **asymptotically stable**; since $V$ is radially unbounded and $\dot{V} < 0$ everywhere off the origin, the basin is all of $\mathbb{R}$ (global asymptotic stability).

**Step 3 (exact solution).** Separate variables for $x(0) = x_0 \neq 0$:

$$
\int x^{-3}\, dx = -\int dt \implies -\frac{1}{2x^2} = -t - \frac{1}{2x_0^2} \implies x(t) = \frac{x_0}{\sqrt{1 + 2 x_0^2 t}} .
$$

**Step 4 (no exponential rate).** Exponential stability would require $\lvert x(t) \rvert \le M e^{-ct} \lvert x_0 \rvert$ for fixed $M, c > 0$. But $\lvert x(t) \rvert \sim (2t)^{-1/2}$ as $t \to \infty$, and $t^{-1/2} / e^{-ct} \to \infty$, so no such $c$ exists: decay is only **algebraic**. Equivalently: $\dot{V} = -x^4 = -4V^2$ vanishes to fourth order, not second — the differential inequality $\dot{V} \le -\kappa V$ needed for Grönwall's exponential bound fails near $0$.

$$
\boxed{x(t) = x_0 (1 + 2x_0^2 t)^{-1/2}: \ \text{globally asymptotically stable, but only algebraically — not exponentially — fast}}
$$

> **Key Takeaway:** Asymptotic and exponential stability genuinely differ, and the gap opens exactly where linearization is degenerate — a Lyapunov function certifies the former even when no linear rate exists.

### Problem L1.5: Spiral Sink with Rotation Direction

**Problem Statement:** Classify the origin of $\dot{\mathbf{x}} = A\mathbf{x}$, $A = \begin{pmatrix} -1 & -2 \\ 2 & -1 \end{pmatrix}$, determine the sense of rotation (clockwise or counterclockwise), and give the exact solution in polar-like form.

**Intuition:** Eigenvalues give the type; one sample of the vector field gives the handedness that eigenvalues alone cannot.

**Solution:**

**Step 1 (classification).** $\tau = -2$, $\Delta = (-1)(-1) - (-2)(2) = 1 + 4 = 5$, discriminant $4 - 20 = -16 < 0$: complex eigenvalues $\lambda = -1 \pm 2i$, negative real part — **spiral sink**.

**Step 2 (rotation direction).** Eigenvalues cannot distinguish clockwise from counterclockwise (conjugation by a reflection flips handedness but not the spectrum). Sample the field: at $(1, 0)$, $\dot{\mathbf{x}} = (-1, 2)^{\mathsf{T}}$, whose $y$-component is positive — the flow at the positive $x$-axis moves upward: **counterclockwise** rotation.

**Step 3 (exact solution).** $A = -I + 2R$ where $R = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}$ generates rotation. Since $I$ and $R$ commute, $e^{At} = e^{-t} e^{2Rt}$, and $e^{2Rt}$ is rotation by angle $2t$:

$$
\mathbf{x}(t) = e^{-t} \begin{pmatrix} \cos 2t & -\sin 2t \\ \sin 2t & \cos 2t \end{pmatrix} \mathbf{x}(0) .
$$

In polar terms: $r(t) = r_0 e^{-t}$, $\theta(t) = \theta_0 + 2t$ — radius decays at rate $1$ while the phase advances counterclockwise at angular speed $2$, completing a turn every $\pi$ time units while shrinking by $e^{-\pi} \approx 0.043$.

$$
\boxed{\lambda = -1 \pm 2i: \ \text{counterclockwise spiral sink, } r = r_0 e^{-t}, \ \theta = \theta_0 + 2t}
$$

> **Key Takeaway:** Type from the spectrum, handedness from one arrow of the field — and matrices of the form $aI + bR$ solve in closed polar form.

### Problem L1.6: A Conserved Quantity Makes a Center Rigorous

**Problem Statement:** For the harmonic oscillator $\dot{x} = y$, $\dot{y} = -x$: (a) show $H(x,y) = \tfrac{1}{2}(x^2 + y^2)$ is conserved and deduce all orbits; (b) give an explicit $\varepsilon$–$\delta$ proof of Lyapunov stability of the origin; (c) explain why the origin is not asymptotically stable.

**Intuition:** Frozen level sets both cage the orbits (stability) and forbid them from falling in (no asymptotic stability).

**Solution:**

**Step 1 (conservation).** $\dot{H} = x\dot{x} + y\dot{y} = xy + y(-x) = 0$: $H$ is constant on every solution.

**Step 2 (orbits).** Each orbit lies on a level set $\{x^2 + y^2 = r_0^2\}$ — a circle of radius $r_0$ (or the origin for $r_0 = 0$). On such a circle the speed is $\lVert (y, -x) \rVert = r_0 > 0$, so the orbit traverses the whole circle: every nonzero orbit is periodic (indeed $x = r_0\cos t$-type solutions with period $2\pi$), traversed clockwise since at $(1,0)$ the velocity is $(0,-1)$. The origin is a **center**.

**Step 3 (stability, $\varepsilon$–$\delta$).** Given $\varepsilon > 0$, choose $\delta = \varepsilon$. If $\lVert \mathbf{x}(0) \rVert < \delta$ then for all $t$,

$$
\lVert \mathbf{x}(t) \rVert = \sqrt{2 H(\mathbf{x}(t))} = \sqrt{2 H(\mathbf{x}(0))} = \lVert \mathbf{x}(0) \rVert < \delta = \varepsilon ,
$$

using conservation of $H$. The definition of Lyapunov stability is satisfied — with the best possible $\delta$.

**Step 4 (no asymptotic stability).** Asymptotic stability requires some ball $B_{\delta_0}$ from which all solutions converge to $\mathbf{0}$. But any $\mathbf{x}(0) \neq \mathbf{0}$ in that ball keeps $\lVert \mathbf{x}(t) \rVert = \lVert \mathbf{x}(0) \rVert > 0$ forever, so $\mathbf{x}(t) \not\to \mathbf{0}$. Conservation of a positive definite quantity is *incompatible* with asymptotic stability: a first integral can certify stability but never attraction.

$$
\boxed{\text{All orbits are circles } x^2 + y^2 = r_0^2: \text{ the origin is a center — Lyapunov stable (take } \delta = \varepsilon \text{), never asymptotically stable}}
$$

> **Key Takeaway:** A conserved positive definite quantity is the perfect stability certificate and a perfect obstruction to attraction — the cleanest illustration that the two notions differ.

### Problem L2.1: Endemic Equilibrium of the SIR Model

**Problem Statement:** The SIR model with vital dynamics (fractions, birth/death rate $\mu$):

$$
\dot{S} = \mu - \beta S I - \mu S, \qquad \dot{I} = \beta S I - (\gamma + \mu) I ,
$$

with $\beta = 0.6$, $\gamma = 0.2$, $\mu = 0.1$ (per week). Compute $R_0$, decide the stability of the disease-free equilibrium $(1, 0)$, find the endemic equilibrium, classify it, and estimate the period of the damped epidemic waves.

**Intuition:** $R_0$ is an eigenvalue-sign statement in disguise; endemic disease arrives via a transcritical bifurcation at $R_0 = 1$.

**Solution:**

**Step 1 (reproduction number).** $R_0 = \dfrac{\beta}{\gamma + \mu} = \dfrac{0.6}{0.3} = 2 > 1$.

**Step 2 (disease-free equilibrium).** The Jacobian of the system is

$$
J(S, I) = \begin{pmatrix} -\beta I - \mu & -\beta S \\ \beta I & \beta S - (\gamma + \mu) \end{pmatrix} .
$$

At $(1, 0)$: $J = \begin{pmatrix} -0.1 & -0.6 \\ 0 & 0.3 \end{pmatrix}$, triangular, eigenvalues $-\mu = -0.1$ and $\beta - (\gamma+\mu) = (\gamma+\mu)(R_0 - 1) = 0.3$. One positive eigenvalue: **saddle, unstable** — the epidemic invades. (For $R_0 < 1$ this eigenvalue is negative and the disease-free state is a stable node: the threshold theorem.)

**Step 3 (endemic equilibrium).** From $\dot{I} = 0$ with $I \neq 0$: $S^* = (\gamma + \mu)/\beta = 1/R_0 = 0.5$. From $\dot{S} = 0$: $I^* = \dfrac{\mu(1 - S^*)}{\beta S^*} = \dfrac{\mu (R_0 - 1)}{\beta} = \dfrac{0.1 \times 1}{0.6} = \dfrac{1}{6} \approx 0.167$.

**Step 4 (classification).** At $(S^*, I^*)$: $\beta S^* = \gamma + \mu = 0.3$ makes the bottom-right entry zero, and $\beta I^* = 0.6/6 = 0.1$:

$$
J^* = \begin{pmatrix} -0.2 & -0.3 \\ 0.1 & 0 \end{pmatrix}, \qquad \tau = -0.2 < 0, \qquad \Delta = 0 - (-0.3)(0.1) = 0.03 > 0 .
$$

Discriminant: $\tau^2 - 4\Delta = 0.04 - 0.12 = -0.08 < 0$. **Spiral sink — asymptotically stable endemic state**, with eigenvalues $\lambda = -0.1 \pm i\sqrt{0.02}$.

**Step 5 (period of the waves).** Angular frequency $\omega = \sqrt{0.02} \approx 0.141$ per week, so the damped inter-epidemic oscillation has period

$$
T = \frac{2\pi}{\omega} \approx \frac{6.283}{0.141} \approx 44.4 \text{ weeks} ,
$$

with envelope decaying like $e^{-0.1 t}$ (about $e^{-4.4} \approx 1.2\%$ amplitude left after one period): recurrent but damping epidemic waves settling to $17\%$ prevalence — qualitatively the pattern of pre-vaccine childhood diseases.

$$
\boxed{R_0 = 2; \ (1,0) \text{ saddle}; \ (S^*, I^*) = (0.5, 1/6) \text{ spiral sink}; \ T \approx 44 \text{ weeks}}
$$

> **Key Takeaway:** Epidemic thresholds, endemic prevalence, and inter-epidemic periodicity are all read off one Jacobian — phase-plane analysis is the working language of mathematical epidemiology.

### Problem L2.2: Damped Pendulum — Classify Everything

**Problem Statement:** For the pendulum $\dot{\theta} = \omega$, $\dot{\omega} = -\sin\theta - b\,\omega$ with $b = 0.5$: find and classify all equilibria (with numerical eigenvalues), show the energy decays, and describe the basins of attraction. What changes at $b = 2$, and what is special about $b = 0$?

**Intuition:** Hanging down is a damped oscillation; balancing up is a knife-edge saddle at any damping.

**Solution:**

**Step 1 (equilibria and Jacobian).** Equilibria at $(\theta, \omega) = (n\pi, 0)$. Jacobian

$$
J = \begin{pmatrix} 0 & 1 \\ -\cos\theta^* & -b \end{pmatrix}, \qquad \tau = -b = -0.5, \qquad \Delta = \cos\theta^* .
$$

**Step 2 (bottom equilibria, $\theta^* = 2k\pi$).** $\Delta = 1 > 0$, discriminant $b^2 - 4 = 0.25 - 4 = -3.75 < 0$: **spiral sinks** with eigenvalues

$$
\lambda = \frac{-0.5 \pm i\sqrt{3.75}}{2} = -0.25 \pm 0.968\,i .
$$

Damped swinging at frequency $0.968$ (barely below the undamped $1$), envelope $e^{-t/4}$.

**Step 3 (top equilibria, $\theta^* = (2k+1)\pi$).** $\Delta = -1 < 0$: **saddles** for every $b$. Eigenvalues $\lambda = \tfrac{1}{2}(-0.5 \pm \sqrt{0.25 + 4})$, i.e. $\lambda_u \approx 0.781$ and $\lambda_s \approx -1.281$. The inverted pendulum falls — except along the measure-zero stable separatrix.

**Step 4 (energy and LaSalle).** $E = \tfrac{1}{2}\omega^2 + 1 - \cos\theta$ gives $\dot{E} = \omega\dot{\omega} + \omega\sin\theta = -b\,\omega^2 \le 0$. On each compact sublevel set, LaSalle applies: $\dot{E} = 0$ means $\omega = 0$, and the largest invariant set inside $\{\omega = 0\}$ is the equilibrium set (staying on $\omega = 0$ forces $\sin\theta = 0$). Every orbit converges to an equilibrium; since saddles capture only their stable manifolds, **almost every orbit spirals into some bottom equilibrium** $(2k\pi, 0)$.

**Step 5 (basins).** The stable separatrices of the saddles at $(\pm\pi, 0), (\pm 3\pi, 0), \dots$ are the basin boundaries. The basin of $(0,0)$ is the strip-like region between the separatrices of the two neighboring saddles; a trajectory with large initial $\omega$ whirls over the top several times, losing $\Delta E = b\int \omega^2 dt$ per revolution, until captured — so the basins wind around the cylinder as interleaved spiral bands indexed by the number of over-the-top swings.

**Step 6 (parameter regimes).** At $b = 2$ the discriminant $b^2 - 4$ hits zero: bottom equilibria become **degenerate nodes**, and for $b > 2$ **stable nodes** — overdamped, no oscillation. At $b = 0$, $\tau = 0$: linearization gives centers, and here the conclusion survives because $E$ is *exactly conserved* — closed orbits for $E < 2$, running orbits for $E > 2$, and the heteroclinic separatrix $E = 2$ connecting adjacent saddles.

$$
\boxed{\text{Bottoms: spiral sinks } (-0.25 \pm 0.968 i) \text{ for } b = 0.5, \text{ nodes for } b \ge 2, \text{ centers for } b = 0; \text{ tops: saddles always}}
$$

> **Key Takeaway:** One Jacobian formula plus one energy inequality yields the complete life of the pendulum — local types, global convergence, and basin geometry.

### Problem L2.3: GAN Training as a Phase Portrait (Dirac-GAN)

**Problem Statement:** For the bilinear min–max game $V(x, y) = xy$ (generator parameter $x$, discriminator parameter $y$), simultaneous gradient descent-ascent gives

$$
\dot{x} = -\partial_x V = -y, \qquad \dot{y} = +\partial_y V = x .
$$

(a) Classify the equilibrium and explain why GAN training oscillates. (b) Show explicit Euler with step $h$ *diverges* for every $h > 0$. (c) Add regularization (weight decay / gradient penalty) $\dot{x} = -y - \varepsilon x$, $\dot{y} = x - \varepsilon y$ with $\varepsilon = 0.1$ and reclassify. (d) For the regularized game, find the step sizes $h$ for which Euler converges.

**Intuition:** The game's vector field is a pure rotation; regularization is damping, exactly as in the pendulum.

**Solution:**

**Step 1 (a: the center).** The equilibrium (the game solution) is $(0,0)$, with

$$
J = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}, \qquad \tau = 0, \ \Delta = 1, \qquad \lambda = \pm i .
$$

A **center**: this linear system's orbits are exact circles $x^2 + y^2 = r_0^2$ (indeed $\frac{d}{dt}(x^2 + y^2) = 2x(-y) + 2y(x) = 0$). Training makes *zero* net progress — the players orbit the solution forever. The field is antisymmetric (a "curl" field), the opposite of a gradient field: Section 3.6's no-cycle theorem for gradient systems does not apply, and cycling is exactly what happens.

**Step 2 (b: Euler diverges).** Explicit Euler: $(x_{k+1}, y_{k+1}) = (x_k - h y_k,\ y_k + h x_k)$. Then

$$
x_{k+1}^2 + y_{k+1}^2 = (x_k - h y_k)^2 + (y_k + h x_k)^2 = (1 + h^2)(x_k^2 + y_k^2) .
$$

The radius grows by the factor $\sqrt{1 + h^2} > 1$ *every step, for every* $h > 0$: simultaneous-gradient GAN training on this game spirals outward and diverges — discretization converts the neutral center into an unstable spiral. (Equivalently: the eigenvalues of the update matrix are $1 \pm ih$, of modulus $\sqrt{1+h^2} > 1$.)

**Step 3 (c: regularization = damping).** With $\varepsilon = 0.1$:

$$
J_\varepsilon = \begin{pmatrix} -0.1 & -1 \\ 1 & -0.1 \end{pmatrix}, \qquad \tau = -0.2 < 0, \quad \Delta = 0.01 + 1 = 1.01 > 0, \quad \tau^2 - 4\Delta < 0 ,
$$

eigenvalues $\lambda = -0.1 \pm i$: a **spiral sink**. Continuous-time training now converges, oscillating with period $\approx 2\pi$ while contracting like $e^{-0.1 t}$. Gradient penalties move the game's eigenvalues off the imaginary axis into the left half-plane — damping for the min–max pendulum.

**Step 4 (d: stable step sizes).** Euler's update matrix is $I + hJ_\varepsilon$ with eigenvalues $1 + h(-\varepsilon \pm i)$. Convergence needs modulus below one:

$$
(1 - h\varepsilon)^2 + h^2 < 1 \iff h^2(1 + \varepsilon^2) < 2 h \varepsilon \iff 0 < h < \frac{2\varepsilon}{1 + \varepsilon^2} .
$$

For $\varepsilon = 0.1$: $h < 0.2/1.01 \approx 0.198$. Small damping buys only a small stable-step window — with the learning rate above it, even the regularized GAN diverges.

$$
\boxed{\text{Unregularized: center } (\lambda = \pm i), \text{ Euler radius factor } \sqrt{1+h^2}; \quad \varepsilon\text{-regularized: spiral sink } (\lambda = -\varepsilon \pm i), \text{ stable iff } h < 2\varepsilon/(1+\varepsilon^2)}
$$

> **Key Takeaway:** GAN oscillation is not mysterious — it is the phase portrait of a center, and every fix (penalties, damping, small learning rates) is an eigenvalue-relocation strategy readable on the trace–determinant chart.

### Problem L2.4: Hopfield Memory as Lyapunov Descent

**Problem Statement:** A two-neuron attractor network evolves by gradient descent $\dot{\mathbf{x}} = -\nabla E(\mathbf{x})$ on the energy

$$
E(x_1, x_2) = -x_1 x_2 + \tfrac{1}{4}\left( x_1^4 + x_2^4 \right).
$$

Find all equilibria, classify them, prove every trajectory converges to an equilibrium, and interpret the phase portrait as memory storage and retrieval (which patterns are stored, what is the error-correction rule?).

**Intuition:** Memories are the minima of an energy landscape; recall is rolling downhill from a corrupted pattern.

**Solution:**

**Step 1 (the system).** $\dot{x}_1 = -\partial_{x_1} E = x_2 - x_1^3$, $\dot{x}_2 = x_1 - x_2^3$.

**Step 2 (equilibria).** $x_2 = x_1^3$ and $x_1 = x_2^3$ give $x_1 = x_1^9$, so $x_1 \in \{0, 1, -1\}$ (real roots), and correspondingly the equilibria are $(0,0)$, $(1,1)$, $(-1,-1)$.

**Step 3 (classification via the Hessian).** For a gradient system the Jacobian is $J = -\nabla^2 E$, with

$$
\nabla^2 E = \begin{pmatrix} 3x_1^2 & -1 \\ -1 & 3x_2^2 \end{pmatrix} .
$$

At $(\pm 1, \pm 1)$: $\nabla^2 E = \begin{pmatrix} 3 & -1 \\ -1 & 3 \end{pmatrix}$, eigenvalues $3 \mp 1 = 2, 4$ — positive definite, so $J$ has eigenvalues $-2, -4$: **stable nodes** (local minima of $E$, exponentially stable). At $(0,0)$: $\nabla^2 E = \begin{pmatrix} 0 & -1 \\ -1 & 0 \end{pmatrix}$, eigenvalues $\pm 1$ — indefinite, so $J$ has eigenvalues $\mp 1$: a **saddle**.

**Step 4 (global convergence).** $E$ is a Lyapunov function: $\dot{E} = -\lVert \nabla E \rVert^2 \le 0$, strict off equilibria (Proof 3.6 of the module: gradient systems admit no periodic or wandering recurrent behavior). $E$ is radially unbounded (quartic dominates the bilinear term: $E \ge \tfrac{1}{4}(x_1^4 + x_2^4) - \tfrac{1}{2}(x_1^2 + x_2^2) \to \infty$), so sublevel sets are compact and LaSalle gives: **every trajectory converges to one of the three equilibria**; almost every one reaches a minimum, since the saddle attracts only its stable manifold.

**Step 5 (memory interpretation).** The stored patterns are the two minima $(1,1)$ and $(-1,-1)$ — the "all-up" and "all-down" memories written by the coupling term $-x_1x_2$ (which rewards agreement of the two neurons). Retrieval: initialize at a corrupted pattern, e.g. $(0.9, 0.2)$, and flow downhill; by the symmetry $E(x_1, x_2) = E(x_2, x_1)$ and the saddle's stable manifold lying along the anti-diagonal $x_1 = -x_2$, the basin boundary is exactly the line $x_2 = -x_1$: any initial state with $x_1 + x_2 > 0$ recalls $(1,1)$, any with $x_1 + x_2 < 0$ recalls $(-1,-1)$. The **error-correction rule is basin membership** — the network corrects any corruption that does not flip the sign of the summed activity. Capacity, spurious minima, and retrieval radius in larger Hopfield networks are precisely the count, depth, and basin size of the energy's local minima.

$$
\boxed{\text{Memories } (\pm 1, \pm 1) \text{: stable nodes } (\lambda = -2, -4); \ (0,0) \text{ saddle; basin boundary } x_1 + x_2 = 0 \text{ is the retrieval decision line}}
$$

> **Key Takeaway:** An attractor network *is* applied Lyapunov theory: storing a memory means sculpting an asymptotically stable equilibrium, and robustness of recall is the geometry of its basin.

### Problem L2.5: Van der Pol — Simulation Reasoning

**Problem Statement:** For $\dot{x} = y$, $\dot{y} = \mu(1 - x^2)y - x$: (a) classify the origin for $\mu = 0.1$ and $\mu = 10$; (b) explain why *every* nonzero initial condition ends on the same oscillation and why its amplitude is an intrinsic property; (c) explain why an explicit RK4 simulation at $\mu = 10$ needs tiny steps while an implicit BDF solver does not, and estimate the period of the relaxation oscillation.

**Intuition:** The nonlinear damping term is an energy thermostat: it pumps small oscillations up and drains large ones down.

**Solution:**

**Step 1 (a: local classification).** Jacobian at the origin: $J = \begin{pmatrix} 0 & 1 \\ -1 & \mu \end{pmatrix}$, so $\tau = \mu > 0$, $\Delta = 1$, discriminant $\mu^2 - 4$. For $\mu = 0.1$: discriminant $< 0$, **unstable spiral** (eigenvalues $0.05 \pm 0.9987\,i$). For $\mu = 10$: discriminant $= 96 > 0$, **unstable node** (eigenvalues $\approx 9.899$ and $0.101$). Either way the origin repels.

**Step 2 (b: the limit cycle thermostat).** Energy $E = \tfrac{1}{2}(x^2 + y^2)$ changes as $\dot{E} = x\dot{x} + y\dot{y} = \mu(1 - x^2)y^2$: energy is **injected** where $\lvert x \rvert < 1$ and **dissipated** where $\lvert x \rvert > 1$. Small orbits grow, large orbits shrink — by the Poincaré–Bendixson logic (trapping annulus: Problem L3.4) there is a periodic orbit in between, and Liénard's theorem (the system has the Liénard form with $F(x) = \mu(x^3/3 - x)$ odd, negative then positive, increasing to infinity) guarantees it is **unique and globally attracting**. Hence every nonzero initial condition converges to the *same* cycle: the amplitude (about $2$ in $x$, for all $\mu$) is set by the balance of injection and dissipation — a property of the system, not the initial data. This is the defining contrast with a center (Lotka–Volterra), where amplitude remembers the start.

**Step 3 (c: stiffness).** For $\mu = 10$ the cycle is a **relaxation oscillation**: slow crawls along the branches of the cubic nullcline $y = \dfrac{x}{\mu(1 - x^2)}$ (time scale $\sim \mu$) punctuated by nearly instantaneous horizontal jumps (time scale $\sim 1/\mu$). Off the slow manifold the local Jacobian entry $\partial \dot{y}/\partial y = \mu(1 - x^2)$ has magnitude up to $\sim \mu \cdot 3 = 30$ (at $x \approx 2$), so an explicit method must keep $h \lesssim 2/30 \approx 0.07$ *even during the slow phases where nothing happens*, purely for numerical stability — the signature of stiffness. An implicit BDF/Radau solver is A-stable: it damps the fast transversal mode at any step size and chooses $h$ by accuracy, taking large steps along the slow branches and refining only at the jumps — typically orders of magnitude fewer steps per period.

**Step 4 (period estimate).** The period is dominated by the two slow crawls; the classical asymptotic result is

$$
T \approx (3 - 2\ln 2)\,\mu \approx 1.614\,\mu \implies T \approx 16.1 \text{ for } \mu = 10 ,
$$

with an $O(\mu^{-1/3})$ correction — so a simulation should be run for, say, $t \in [0, 50]$ to see transient decay plus two or three full cycles.

$$
\boxed{\text{Origin unstable (spiral for } \mu = 0.1, \text{ node for } \mu = 10); \text{ unique global limit cycle; stiff for large } \mu; \ T \approx 1.614\,\mu}
$$

> **Key Takeaway:** Limit cycles are self-tuned oscillations with system-determined amplitude, and their fast–slow anatomy dictates the numerics: recognize relaxation structure, reach for a stiff solver.

### Problem L2.6: Competitive Exclusion in an Ecological Model

**Problem Statement:** Two species compete:

$$
\dot{x} = x(3 - x - 2y), \qquad \dot{y} = y(2 - x - y), \qquad x, y \ge 0 .
$$

Find all four equilibria, classify each, determine the fate of generic initial populations, and state (in terms of competition coefficients) when a two-species model instead allows *stable coexistence*.

**Intuition:** Each species obeys logistic growth alone; the cross terms decide whether the war ends in partition or annihilation.

**Solution:**

**Step 1 (equilibria).** Either $x = 0$ or $3 - x - 2y = 0$; either $y = 0$ or $2 - x - y = 0$. This yields $(0,0)$, $(3, 0)$, $(0, 2)$, and the interior solution of $x + 2y = 3$, $x + y = 2$: subtracting, $y = 1$, then $x = 1$ — coexistence candidate $(1,1)$.

**Step 2 (Jacobian).**

$$
J(x,y) = \begin{pmatrix} 3 - 2x - 2y & -2x \\ -y & 2 - x - 2y \end{pmatrix} .
$$

**Step 3 (boundary equilibria).**

At $(0,0)$: $J = \operatorname{diag}(3, 2)$ — **unstable node** (extinction of both is repelling).

At $(3,0)$: $J = \begin{pmatrix} -3 & -6 \\ 0 & -1 \end{pmatrix}$, triangular, eigenvalues $-3, -1$ — **stable node** ($x$ wins, $y$ extinct).

At $(0,2)$: $J = \begin{pmatrix} -1 & 0 \\ -2 & -2 \end{pmatrix}$, eigenvalues $-1, -2$ — **stable node** ($y$ wins, $x$ extinct).

**Step 4 (interior equilibrium).** At $(1,1)$:

$$
J = \begin{pmatrix} -1 & -2 \\ -1 & -1 \end{pmatrix}, \qquad \tau = -2, \qquad \Delta = (-1)(-1) - (-2)(-1) = 1 - 2 = -1 < 0 .
$$

Negative determinant: **saddle**. Coexistence exists but is dynamically untenable.

**Step 5 (global fate).** The axes are invariant (each is a logistic equation), the positive quadrant is positively invariant, and orbits are bounded (for $x + 2y$ large both brackets are negative). Two sinks, one interior saddle: the saddle's stable manifold is the **separatrix** splitting the quadrant into the basin of $(3,0)$ and the basin of $(0,2)$ — a *bistable* portrait. Generic outcome: **competitive exclusion** — one species dies, and *which* one depends on initial conditions (founder effect). Only initial states exactly on the separatrix approach coexistence, a measure-zero event.

**Step 6 (when coexistence is stable).** For the general Lotka–Volterra competition model $\dot{x} = x(r_1 - x - a y)$, $\dot{y} = y(r_2 - b x - y)$ (self-competition normalized to 1), the interior equilibrium is a sink precisely when $ab < 1$ — **interspecific competition weaker than intraspecific** — and a saddle when $ab > 1$. Here $a = 2, b = 1$, $ab = 2 > 1$: mutual aggression exceeds self-limitation, so the species cannot partition the niche. Weaken the cross terms (e.g. $a = b = 1/2$) and $(x^*, y^*)$ becomes a globally attracting node: peaceful coexistence.

$$
\boxed{(0,0) \text{ source}; \ (3,0), (0,2) \text{ stable nodes}; \ (1,1) \text{ saddle} \Rightarrow \text{bistable competitive exclusion; coexistence iff cross-competition } ab < 1}
$$

> **Key Takeaway:** One determinant sign at the interior equilibrium encodes an ecological law: niches partition when self-limitation beats cross-limitation, otherwise history (the basin) picks the sole survivor.

### Problem L3.1: Asymptotic Stability Where Linearization Sees a Center

**Problem Statement:** Consider

$$
\dot{x} = -y - x^3, \qquad \dot{y} = x - y^3 .
$$

Show that the linearization at the origin is a center (hence inconclusive), and prove — with a Lyapunov function and full $\varepsilon$–$\delta$ reasoning — that the origin is globally asymptotically stable. Explain why no smooth linearization argument could ever reach this conclusion.

**Intuition:** The rotation is neutral; the cubic terms bleed energy in every direction.

**Solution:**

**Step 1 (linearization).** $J(0,0) = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}$: eigenvalues $\pm i$, a linear center. Non-hyperbolic — Hartman–Grobman does not apply and the linear picture proves nothing.

**Step 2 (Lyapunov candidate).** Take $V(x,y) = \tfrac{1}{2}(x^2 + y^2)$: positive definite and radially unbounded. Its orbital derivative:

$$
\dot{V} = x(-y - x^3) + y(x - y^3) = -xy - x^4 + xy - y^4 = -(x^4 + y^4) .
$$

The rotational parts cancel *exactly* (they are the center) and the cubic terms contribute strict dissipation: $\dot{V} < 0$ for all $(x,y) \neq (0,0)$.

**Step 3 (stability, $\varepsilon$–$\delta$).** Given $\varepsilon > 0$, set $\delta = \varepsilon$. If $\lVert \mathbf{x}(0) \rVert < \delta$, then since $V$ is nonincreasing, $\tfrac{1}{2}\lVert \mathbf{x}(t) \rVert^2 = V(t) \le V(0) < \tfrac{1}{2}\delta^2$, so $\lVert \mathbf{x}(t) \rVert < \varepsilon$ for all $t \ge 0$: Lyapunov stable.

**Step 4 (attraction).** Fix any initial condition; its orbit stays in the compact ball $\bar{B}_{r_0}$, $r_0 = \lVert \mathbf{x}(0) \rVert$. $V(t)$ is decreasing and bounded below, so $V(t) \downarrow c \ge 0$. If $c > 0$, the orbit lives forever in the compact annulus $K = \{2c \le x^2 + y^2 \le r_0^2\}$, on which the continuous function $x^4 + y^4$ has a positive minimum $\gamma$ (it vanishes only at the origin, excluded from $K$). Then $\dot{V} \le -\gamma$ gives $V(t) \le V(0) - \gamma t \to -\infty$, contradicting $V \ge 0$. Hence $c = 0$, i.e. $\lVert \mathbf{x}(t) \rVert \to 0$. Since the argument works for **every** initial condition (radial unboundedness gave the compact trap), the origin is **globally asymptotically stable**.

**Step 5 (why linearization could never see this).** Any argument using only $J(0,0)$ must give the same answer for all systems sharing that Jacobian — but $\dot{x} = -y + x^3$, $\dot{y} = x + y^3$ has the *same* linearization and is unstable (its $\dot{V} = +(x^4 + y^4)$, so $V$ grows: by the same annulus argument run forward, orbits leave every ball). The Jacobian cannot separate these two systems; only nonlinear information (here, the sign of the quartic dissipation) can. Note also the convergence is slower than exponential: near the origin $\dot{V} = -(x^4 + y^4) \ge -2V^2 \cdot 2$, an algebraic-decay differential inequality, consistent with the zero spectral margin.

$$
\boxed{V = \tfrac{1}{2}(x^2 + y^2), \ \dot{V} = -(x^4 + y^4) < 0: \text{ the origin is globally asymptotically stable although its linearization is a center}}
$$

> **Key Takeaway:** Lyapunov's method strictly dominates linearization: it settles precisely the borderline cases where the spectrum abstains, by measuring what the nonlinear terms do to a single scalar.

### Problem L3.2: Killing Limit Cycles with a Dulac Function

**Problem Statement:** Show that the predator–prey-with-competition system

$$
\dot{x} = x(2 - x - y), \qquad \dot{y} = y(4x - x^2 - 3), \qquad x > 0, \ y > 0,
$$

has no periodic orbit in the open first quadrant, using the Dulac function $B(x,y) = \dfrac{1}{xy}$. Verify all hypotheses of the Bendixson–Dulac theorem, and explain why the plain Bendixson criterion ($B \equiv 1$) would have failed.

**Intuition:** Dividing out the population sizes exposes a hidden one-signed divergence.

**Solution:**

**Step 1 (hypotheses of the theorem).** The open first quadrant $D = \{x > 0, y > 0\}$ is simply connected; $f = x(2-x-y)$, $g = y(4x - x^2 - 3)$ are $C^1$ on $D$; and $B = 1/(xy)$ is $C^1$ on $D$ (this is exactly why we restrict to the *open* quadrant — $B$ blows up on the axes, which is permissible since the axes are invariant and any interior periodic orbit stays in $D$).

**Step 2 (compute the weighted divergence).**

$$
B f = \frac{2 - x - y}{y}, \qquad B g = \frac{4x - x^2 - 3}{x} .
$$

$$
\frac{\partial (Bf)}{\partial x} = \frac{\partial}{\partial x}\left( \frac{2 - x - y}{y} \right) = -\frac{1}{y}, \qquad \frac{\partial (Bg)}{\partial y} = \frac{\partial}{\partial y}\left( \frac{4x - x^2 - 3}{x} \right) = 0 .
$$

Hence

$$
\nabla \cdot (B \mathbf{f}) = -\frac{1}{y} < 0 \quad \text{everywhere on } D .
$$

**Step 3 (conclusion).** The weighted divergence has one strict sign on the simply connected region $D$; by the Bendixson–Dulac theorem (proved in the module via Green's theorem: a periodic orbit $\Gamma \subset D$ would enclose a region $R \subset D$ with $0 = \oint_\Gamma B(f\,dy - g\,dx) = \iint_R \nabla\cdot(B\mathbf{f})\, dA < 0$, a contradiction), **no periodic orbit lies in the open first quadrant**. Since populations live there, the model admits no cycles: every bounded orbit's omega-limit set contains equilibria (by Poincaré–Bendixson trichotomy, with cycles now excluded).

**Step 4 (why plain Bendixson fails).** The unweighted divergence is

$$
\nabla \cdot \mathbf{f} = (2 - 2x - y) + (4x - x^2 - 3) = -x^2 + 2x - 1 - y + \dots
$$

carefully: $\partial_x f = 2 - 2x - y$ and $\partial_y g = 4x - x^2 - 3$, so $\nabla \cdot \mathbf{f} = -x^2 + 2x - 1 - y = -(x-1)^2 - y + 0$. Wait — collect: $(2 - 2x - y) + (4x - x^2 - 3) = -x^2 + 2x - 1 - y = -(x - 1)^2 - y$. This *happens* to be negative here, but perturb the model (e.g. change prey capacity, $\dot{x} = x(2 - x/K - y)$ with large $K$) and $\nabla\cdot\mathbf{f} = 2 - 2x/K - y + 4x - x^2 - 3$ changes sign across the quadrant, while the Dulac computation still yields $\partial_x(Bf) = -1/K \cdot \dots$ — the $1/(xy)$ weight removes the sign-indefinite growth terms *structurally*: for any Lotka–Volterra-type system $\dot{x} = x\,P(x,y)$, $\dot{y} = y\,Q(x,y)$ it gives $\nabla\cdot(B\mathbf{f}) = \tfrac{1}{y}\partial_x P + \tfrac{1}{x}\partial_y Q$, depending only on the *self-interaction* slopes. Here $\partial_x P = -1$, $\partial_y Q = 0$. That robustness — insensitivity to the sign-varying cross terms — is why $1/(xy)$ is the canonical Dulac function for population models.

$$
\boxed{\nabla \cdot \left( \tfrac{1}{xy} \mathbf{f} \right) = -\tfrac{1}{y} < 0 \text{ on the open quadrant} \implies \text{no periodic orbits — cycles are impossible in this ecology}}
$$

> **Key Takeaway:** A Dulac function is a change of measure that makes hidden dissipation visible; for population models, weighting by $1/(xy)$ reduces the cycle question to the sign of self-interaction alone.

### Problem L3.3: Level-Set Anatomy of the Duffing Hamiltonian

**Problem Statement:** The undamped Duffing oscillator $\dot{x} = y$, $\dot{y} = x - x^3$ is Hamiltonian with

$$
H(x, y) = \tfrac{1}{2} y^2 - \tfrac{1}{2} x^2 + \tfrac{1}{4} x^4 .
$$

Verify conservation, classify all equilibria, and give a complete census of the level sets of $H$ (for every value $h$), identifying the separatrix and the homoclinic orbits. Deduce the period behavior near the centers and near the separatrix.

**Intuition:** The phase portrait is a topographic map of the double-well energy surface: read the orbits off the contour lines.

**Solution:**

**Step 1 (conservation and structure).** $\dot{H} = y\dot{y} + (x^3 - x)\dot{x} = y(x - x^3) + (x^3 - x)y = 0$: orbits lie on level curves of $H$. The system is $\dot{x} = \partial_y H$, $\dot{y} = -\partial_x H$ — Hamiltonian, hence area-preserving flow and no attractors of any kind.

**Step 2 (equilibria).** $y = 0$, $x - x^3 = 0$: $(0,0)$ and $(\pm 1, 0)$. Jacobian $\begin{pmatrix} 0 & 1 \\ 1 - 3x^2 & 0 \end{pmatrix}$. At $(0,0)$: $\Delta = -1$, **saddle**, eigenvalues $\pm 1$, energy $H = 0$. At $(\pm 1, 0)$: $\Delta = 2 > 0$, $\tau = 0$ — linear centers; because $H$ is conserved and $H - H_{\min}$ is positive definite near $(\pm 1, 0)$ (Hessian $\operatorname{diag}(2, 1) \succ 0$), they are genuine nonlinear **centers**, with energy $H = -\tfrac{1}{4}$.

**Step 3 (census of level sets).** Write $H = \tfrac{1}{2}y^2 + U(x)$ with double-well potential $U(x) = -\tfrac{1}{2}x^2 + \tfrac{1}{4}x^4$, $U(\pm 1) = -\tfrac{1}{4}$, $U(0) = 0$. The level set $\{H = h\}$ is $y = \pm\sqrt{2(h - U(x))}$, nonempty where $U(x) \le h$:

- $h < -\tfrac{1}{4}$: empty.
- $h = -\tfrac{1}{4}$: the two centers $(\pm 1, 0)$.
- $-\tfrac{1}{4} < h < 0$: $\{U \le h\}$ is two disjoint intervals around $\pm 1$, so the level set is **two disjoint ovals**, one around each center — small oscillations trapped in a single well.
- $h = 0$: the level curve through the saddle: $y^2 = x^2 - \tfrac{1}{2}x^4$, a **figure-eight**. Each lobe is a **homoclinic orbit** — leaving the saddle along its unstable eigenvector, swinging around one center, and returning along the stable eigenvector as $t \to +\infty$ (taking infinite time in both directions, since the saddle is an equilibrium). This figure-eight is the **separatrix** dividing single-well from double-well motion.
- $h > 0$: $\{U \le h\}$ is one interval containing both wells; the level set is a **single large closed curve** encircling all three equilibria — the particle sloshes over the central barrier.

**Step 4 (periods).** Near a center, linearization frequency is $\sqrt{2}$ (eigenvalues $\pm i\sqrt{2}$), so small in-well oscillations have period $\approx 2\pi/\sqrt{2} = \pi\sqrt{2}$. As $h \uparrow 0$ the orbit hugs the saddle, where passage time diverges logarithmically (time near a saddle scales like $\ln(1/\text{dist})$ because motion is $e^{\pm t}$ along eigenlines): the **period tends to $+\infty$ at the separatrix**, then decreases again for large $h$. Period is a non-monotonic function of energy with a logarithmic infinity at $h = 0$ — the universal signature of separatrix crossing in Hamiltonian systems (and the mechanism behind critical slowing near saddle bifurcations).

$$
\boxed{\text{Two centers at } (\pm 1, 0), \ h = -\tfrac{1}{4}; \text{ saddle at } (0,0) \text{ with figure-eight homoclinic separatrix } H = 0; \text{ ovals inside, global loops outside, period } \to \infty \text{ at } H = 0}
$$

> **Key Takeaway:** For Hamiltonian systems the entire phase portrait is the contour plot of $H$ — equilibria at critical points, orbit topology from sublevel-set topology, and dynamics (periods) from how contours crowd the saddle.

### Problem L3.4: A Poincaré–Bendixson Trapping Region

**Problem Statement:** (a) For the system

$$
\dot{x} = -y + x(1 - x^2 - y^2), \qquad \dot{y} = x + y(1 - x^2 - y^2),
$$

construct an explicit trapping annulus and use the Poincaré–Bendixson theorem to prove a periodic orbit exists; then identify it. (b) For the van der Pol oscillator $\dot{x} = y$, $\dot{y} = \mu(1 - x^2)y - x$ ($\mu > 0$), prove rigorously that the origin repels (no orbit starting away from it can converge to it, and orbits near it leave a fixed disk), and assemble the annulus argument granting the outer bound from the classical Liénard estimate.

**Intuition:** Cage the flow between a boundary that pushes out and a boundary that pushes in; with no equilibrium inside the cage, something must go in circles.

**Solution:**

**Part (a).**

**Step 1 (radial dynamics).** With $r^2 = x^2 + y^2$: $r\dot{r} = x\dot{x} + y\dot{y} = r^2(1 - r^2)$, so $\dot{r} = r(1 - r^2)$, and $\dot{\theta} = (x\dot{y} - y\dot{x})/r^2 = 1$.

**Step 2 (the annulus).** Take $A = \{\tfrac{1}{2} \le r \le 2\}$. On the inner circle $r = \tfrac{1}{2}$: $\dot{r} = \tfrac{1}{2}(1 - \tfrac{1}{4}) = \tfrac{3}{8} > 0$ — the flow crosses **into** the annulus (outward). On the outer circle $r = 2$: $\dot{r} = 2(1 - 4) = -6 < 0$ — the flow again crosses **into** the annulus (inward). Hence $A$ is compact and positively invariant: a trapping region.

**Step 3 (no equilibria inside).** Equilibria require $\dot{\theta}$ undefined or $\mathbf{f} = 0$; here $\lVert \mathbf{f} \rVert^2 = r^2 + r^2(1-r^2)^2 > 0$ for $r > 0$, so the only equilibrium is the origin, outside $A$.

**Step 4 (Poincaré–Bendixson).** A compact positively invariant region free of equilibria contains a periodic orbit. Indeed, $\dot{r} = r(1 - r^2)$ vanishes at $r = 1$: the unit circle, traversed with $\dot{\theta} = 1$ (period $2\pi$), is the limit cycle, and since $\dot{r} > 0$ for $r < 1$ and $\dot{r} < 0$ for $r > 1$, it attracts every nonzero orbit.

**Part (b).**

**Step 5 (the origin of van der Pol repels — rigorous).** Let $V = \tfrac{1}{2}(x^2 + y^2)$; along solutions $\dot{V} = xy + y\left( \mu(1 - x^2)y - x \right) = \mu(1 - x^2)y^2 \ge 0$ on the disk $D_1 = \{x^2 + y^2 < 1\}$ (there $x^2 < 1$). Consider the **time-reversed** flow ($t \mapsto -t$), for which $\dot{V}_{\mathrm{rev}} = -\mu(1 - x^2)y^2 \le 0$ on $D_1$, vanishing exactly on $E = \{y = 0\} \cap D_1$. The largest invariant set of the reversed flow inside $E$: staying on $y = 0$ forces $\dot{y}_{\mathrm{rev}} = x = 0$, i.e. only the origin. LaSalle applied to the reversed flow on a compact sublevel set $\{V \le c\} \subset D_1$ gives: every reversed orbit starting in it converges to the origin. Translating back: **every forward orbit starting in $\{0 < V \le c\}$ must leave $\{V \le c\}$** (if a forward orbit remained, its own backward orbit would stay too — but forward-invariance would then make it a full orbit inside on which $V$ is nondecreasing yet must have come from arbitrarily near the origin; concretely, since $\dot{V} \ge 0$ in $D_1$, $V$ is nondecreasing, and reversed-LaSalle shows its alpha-limit set is the origin, so the forward orbit has $V$ increasing away from $0$ and, being monotone and bounded only if it stalls — impossible off the equilibrium since the largest forward-invariant subset of $\{\dot{V} = 0\} \cap D_1$ is again the origin — must exit). Hence the circle $r = \rho$ for small $\rho$ is an inner boundary that forward orbits cross only outward-bound in net effect: the origin is a **repeller**, and $\{V \ge \rho^2/2\}$ acts as the inside wall of the cage.

**Step 6 (outer boundary via Liénard coordinates).** In the Liénard plane $u = x$, $v = y + F(u)$ with... equivalently, the classical estimate on the standard form: let $R = \tfrac{1}{2}(x^2 + z^2)$ for the Liénard system $\dot{x} = z - F(x)$, $\dot{z} = -x$, $F(x) = \mu(\tfrac{x^3}{3} - x)$ (this system is equivalent to van der Pol). Then $\dot{R} = x\dot{x} + z\dot{z} = x(z - F(x)) - zx = -x\,F(x) = \mu x^2 (1 - \tfrac{x^2}{3})$: energy is gained only in the strip $\lvert x \rvert < \sqrt{3}$, where the gain rate is bounded, $\dot{R} \le \mu \cdot \max_{x^2 \le 3} x^2(1 - x^2/3) = \tfrac{3\mu}{4}$, while outside the strip $\dot{R} < 0$, ever more strongly for large amplitude. A trajectory of large radius $\rho$ crosses the strip in time $O(1/\rho)$ (there $\lvert \dot{x} \rvert = \lvert z - F(x) \rvert \ge \rho - \max_{\lvert x \rvert \le \sqrt{3}} \lvert F \rvert$ for the portion with $\lvert z \rvert$ comparable to $\rho$), so it gains at most $O(\mu/\rho)$ of $R$ per strip crossing, but while outside the strip it loses an amount bounded below independent of $\rho$ (the loss $-xF(x)$ integrated over the long excursion grows with $\rho$). Hence for $\rho$ large enough, $R$ strictly decreases over each circuit: some large closed curve $C_\rho$ is crossed only inward. This is the classical Liénard bound (Perko, Section 3.8; Strogatz, Section 7.5).

**Step 7 (assemble).** The region $A$ between the small circle of Step 5 and the curve $C_\rho$ of Step 6 is compact and positively invariant, and contains no equilibrium (the origin — van der Pol's only equilibrium, since $\dot{x} = 0$ forces $y = 0$ and then $\dot{y} = -x = 0$ — is excluded by the inner wall). Poincaré–Bendixson delivers a periodic orbit in $A$; Liénard's uniqueness theorem upgrades it to *the* unique, globally attracting limit cycle.

$$
\boxed{\text{(a) annulus } \tfrac{1}{2} \le r \le 2 \text{ traps the flow, no interior equilibria } \Rightarrow \text{ limit cycle } r = 1; \text{ (b) reversed-time LaSalle repels the origin, Liénard energy bound caps the outside } \Rightarrow \text{ van der Pol has a periodic orbit}}
$$

> **Key Takeaway:** Poincaré–Bendixson converts two boundary sign-checks plus an equilibrium count into the existence of an oscillation — the planar mathematician's only general-purpose machine for *proving* that something cycles.
