# Topic 01: Classification and First-Order ODEs

## 1. First-Principles Intuition & Motivation

**Phenomenon.** Nature rarely hands us a formula for a quantity; it hands us a *law of change*. A cooling coffee cup does not come labeled with its temperature curve — it comes with the observation that heat flows at a rate proportional to the temperature gap. A capacitor does not announce its voltage history — it obeys a charge-balance law at every instant. Machine learning training does not reveal its loss trajectory — it follows the local rule "move against the gradient." In every case the primitive datum is a relation between a function and its derivative:

$$
\frac{dy}{dt} = f(t, y)
$$

The entire subject of first-order ODEs is the art of reconstructing the global object $y(t)$ from this local law, together with one anchoring measurement $y(t_0) = y_0$.

**Why classification comes first.** There is no universal solution algorithm for ODEs — a theorem-level fact, not a temporary gap in our knowledge (Liouville's theory of elementary integrability shows that even $y' = y^2 - t$ has no elementary closed form). What exists instead is a taxonomy: a handful of structural classes, each carrying its own complete solution method. Classification is therefore the routing step of the whole subject. Reading off whether an equation is separable, linear, exact, or substitution-reducible determines everything that follows.

### Goal and Standing Assumptions

**Goal.** Given a first-order equation, we want to:

1. **Classify** it along every structural axis: order, ODE vs PDE, linear vs nonlinear, autonomous vs non-autonomous, homogeneous vs non-homogeneous, explicit vs implicit.
2. **Route** it to the correct analytic technique and execute that technique with full justification — no unexplained symbol-pushing (in particular, we will *prove* that "multiplying both sides by $dt$" in separation of variables is a disguised chain rule, not fraction algebra).
3. **Extract structure** from the answer: the decomposition of every linear solution into homogeneous plus particular parts, the role of the integration constant as a coordinate on the solution family, and the geometric picture supplied by the direction field.

**Standing assumptions.** Throughout this notebook, unless stated otherwise:

- $f(t, y)$, $p(t)$, $q(t)$, $M(x, y)$, $N(x, y)$ are continuous on the (open) domains where we work; when we differentiate them we assume continuous differentiability there.
- Domains for exactness arguments are **simply connected** (typically open rectangles); this is exactly the hypothesis under which the Poincaré lemma converts closed forms into exact forms.
- Solutions are understood on **maximal intervals of existence**: a solution is a differentiable function on an interval, satisfying the equation pointwise, that cannot be extended further. The example $y' = 1 + y^2$, whose solutions $y = \tan(t - C)$ live only on intervals of length $\pi$, shows that a globally smooth right-hand side does not guarantee globally defined solutions.

### Variables and Notation

| Symbol | Meaning | Notes |

| $y(t)$ | unknown function (the state) | scalar-valued in this topic |
| $y'$, $\dot{y}$, $\frac{dy}{dt}$ | first derivative | interchangeable notations |
| $f(t, y)$ | right-hand side (the slope law) | defines the direction field |

A remark on the two meanings of "homogeneous": a *linear* equation is homogeneous when its forcing vanishes ($q \equiv 0$); a first-order equation $y' = f(x, y)$ is *homogeneous of degree zero* when $f(\lambda x, \lambda y) = f(x, y)$ for all $\lambda \neq 0$. The two usages are unrelated; context (linearity) disambiguates them, and we will always say "homogeneous-degree" for the second.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition 2.1 (Ordinary differential equation, order).** An *ordinary differential equation* (ODE) is a relation

$$
G\big(t,\, y,\, y',\, \dots,\, y^{(n)}\big) = 0
$$

between an unknown function $y$ of a *single* independent variable $t$ and finitely many of its derivatives. The *order* is the highest derivative order $n$ that appears. A *partial differential equation* (PDE) instead involves partial derivatives of an unknown function of several variables, e.g. the heat equation $u_t = k\,u_{xx}$; PDEs require fundamentally different techniques and are outside this topic's scope.

**Definition 2.2 (Explicit vs implicit).** A first-order ODE is in *explicit* (or normal) form when solved for the derivative, $y' = f(t, y)$; it is *implicit* when only the relation $G(t, y, y') = 0$ is available, as in $(y')^2 + y^2 = 1$. Implicit equations may bundle several explicit branches ($y' = \pm\sqrt{1 - y^2}$ here) and can possess singular solutions lying on the boundary between branches.

**Definition 2.3 (Solution, IVP).** A *solution* of $y' = f(t,y)$ on an interval $I$ is a differentiable $\varphi: I \to \mathbb{R}$ with $\varphi'(t) = f(t, \varphi(t))$ for all $t \in I$. An *initial value problem* (IVP) adjoins the condition $y(t_0) = y_0$, selecting (under mild hypotheses, uniquely) one member of the solution family.

### The Classification Axes and Why Each One Dictates Strategy

**Definition 2.4 (Linear).** A first-order ODE is *linear* if it can be written

$$
y' + p(t)\,y = q(t)
$$

with $y$ and $y'$ appearing to the first power and not multiplied together. It is *homogeneous* if $q \equiv 0$, *non-homogeneous* otherwise. Any equation not reducible to this form is *nonlinear*.

**Definition 2.5 (Autonomous).** The equation $y' = f(y)$, with no explicit $t$ on the right-hand side, is *autonomous*; $y' = f(t, y)$ with genuine $t$-dependence is *non-autonomous*.

Each axis is a lever that either grants or withholds a specific tool:

| Axis | What the "good" side buys you |
| :--- | :--- |
| **Linear** | Superposition: the solution set is an affine line, one particular solution plus the one-dimensional homogeneous family $Ce^{-\int p}$; the integrating factor solves everything in closed form. |
| **Autonomous** | Time-translation invariance: if $y(t)$ solves the equation so does $y(t - c)$; equilibria and stability are read from the sign of $f$ on the phase line, with no solving required. |
| **Separable** | The equation splits into two one-variable integrals — direct quadrature. |
| **Exact** | Solution curves are level sets of a potential $F$; solving means one partial integration, not solving a "differential" problem at all. |
| **Homogeneous-degree** | The scaling symmetry $f(\lambda x, \lambda y) = f(x,y)$ guarantees the substitution $v = y/x$ separates the equation. |
| **Explicit** | The slope is a function; the direction field is well-defined and Picard–Lindelöf applies directly. |

**Theorem 2.6 (Picard–Lindelöf, statement).** If $f$ is continuous in $(t,y)$ and Lipschitz in $y$ on a rectangle around $(t_0, y_0)$, the IVP $y' = f(t,y)$, $y(t_0) = y_0$ has exactly one solution on some interval around $t_0$. Full proof machinery (contraction mapping, Grönwall) is treated in the survey module `../../calculus/15_ordinary_differential_equations/` and revisited later in this track; in this topic we use only the statement — and we will prove uniqueness for *linear* equations by an elementary self-contained argument (Derivation 2).

### Direction Fields and Isoclines

**Definition 2.7 (Direction field).** The equation $y' = f(t, y)$ assigns to each point $(t, y)$ of its domain the slope $f(t, y)$. The resulting field of short line segments of slope $f(t,y)$ attached at each point is the *direction field* (slope field). A function is a solution precisely when its graph is everywhere tangent to the field — the ODE *is* its direction field.

**Definition 2.8 (Isocline, nullcline).** The *isocline of slope $c$* is the level set

$$
\{(t, y) : f(t, y) = c\}
$$

Along it, all field segments are parallel with slope $c$. The *nullcline* is the isocline $c = 0$; for autonomous equations $y' = f(y)$ its horizontal lines $y = y^*$ with $f(y^*) = 0$ are exactly the **equilibrium solutions**, and the sign of $f$ between nullclines determines whether solutions rise or fall there.

Isoclines convert plotting into analysis: sketching a few isoclines partitions the plane into regions of known slope sign and monotonicity, revealing equilibria, funnels, and asymptotic behavior *before* (or without) any formula. Two rigorous facts power this: (i) by uniqueness, distinct solution graphs never cross, so equilibria act as impassable barriers; (ii) a solution trapped between two isoclines inherits slope bounds that force convergence or divergence estimates. Example: for $y' = y - t$, the isoclines are the lines $y = t + c$; the isocline $c = 1$ is the line $y = t + 1$, which has slope $1$ everywhere — equal to its own isocline value — so it is itself a solution, and the general solution $y = t + 1 + Ce^t$ peels off it exponentially.

### The Two Workhorse Normal Forms

**Definition 2.9 (Separable equation).** A first-order ODE is *separable* if it can be written

$$
\frac{dy}{dt} = g(t)\,h(y)
$$

**Theorem 2.10 (Solution by quadrature).** On any interval where $h(y) \neq 0$, the solutions of a separable equation are exactly the differentiable functions satisfying

$$
H(y(t)) = G(t) + C, \qquad H(u) = \int \frac{du}{h(u)}, \quad G(t) = \int g(t)\,dt
$$

together with the constant (equilibrium) solutions $y \equiv y^*$ for each root $h(y^*) = 0$. The proof — which legitimizes the folklore "divide by $h$, multiply by $dt$, integrate" — is Derivation 0 in Section 3.

**Definition 2.11 (Linear standard form).** Every linear first-order equation can be normalized (divide by the leading coefficient on intervals where it is nonzero) to

$$
y' + p(t)\,y = q(t)
$$

**Theorem 2.12 (Complete solution of the linear equation).** Let $p, q$ be continuous on an interval $I$ and $t_0 \in I$. With $\mu(t) = e^{\int_{t_0}^{t} p(s)\,ds}$, every solution on $I$ has the form

$$
y(t) = \underbrace{\frac{1}{\mu(t)} \int_{t_0}^{t} \mu(s)\,q(s)\,ds}_{\text{particular } y_p} + \underbrace{\frac{C}{\mu(t)}}_{\text{homogeneous } y_h}
$$

and conversely every such function is a solution; the IVP $y(t_0) = y_0$ is solved uniquely by $C = y_0$. Note the structure: the solution set is the affine family $y_p + \text{span}\{e^{-\int p}\}$ — a one-dimensional affine subspace, the first-order shadow of the general "homogeneous + particular" principle of linear theory. Proof: Derivations 1 and 2.

### Exact Equations and the Substitution Classes

**Definition 2.13 (Exact equation).** Write a first-order ODE in differential form $M(x,y)\,dx + N(x,y)\,dy = 0$. The equation is *exact* on a domain $D$ if there exists a *potential* $F \in C^1(D)$ with

$$
\frac{\partial F}{\partial x} = M, \qquad \frac{\partial F}{\partial y} = N
$$

Then along any solution $y(x)$, the chain rule gives $\frac{d}{dx}F(x, y(x)) = M + N y' = 0$, so solution curves are the level sets $F(x, y) = C$.

**Theorem 2.14 (Exactness criterion).** Let $M, N \in C^1(D)$ with $D$ an open simply connected domain (e.g. a rectangle). Then the form is exact on $D$ **iff**

$$
\frac{\partial M}{\partial y} = \frac{\partial N}{\partial x} \quad \text{throughout } D
$$

Necessity is Clairaut's equality of mixed partials; sufficiency is the Poincaré lemma in dimension two, proved constructively in Derivation 3. Simple connectivity matters: on the punctured plane, $\frac{-y\,dx + x\,dy}{x^2 + y^2}$ is closed but admits no global potential.

**Definition 2.15 (Integrating factor for exactness).** A nonvanishing $\mu(x,y)$ is an *integrating factor* if $\mu M\,dx + \mu N\,dy = 0$ is exact. Derivation 3 (continued) shows: if $(M_y - N_x)/N$ depends on $x$ alone, then $\mu(x) = e^{\int (M_y - N_x)/N \,dx}$ works; if $(N_x - M_y)/M$ depends on $y$ alone, then $\mu(y) = e^{\int (N_x - M_y)/M \,dy}$ works.

**Definition 2.16 (The three substitution classes).**

- **Bernoulli**: $y' + p(t)y = q(t)y^n$ with $n \neq 0, 1$; the substitution $v = y^{1-n}$ linearizes it (Derivation 5).
- **Homogeneous-degree**: $y' = f(x, y)$ with $f(\lambda x, \lambda y) = f(x,y)$; equivalently $y' = \Phi(y/x)$, and $v = y/x$ separates it: $x v' = \Phi(v) - v$.
- **Riccati**: $y' = a(t) + b(t)y + c(t)y^2$; the substitution $y = -\dfrac{u'}{c\,u}$ converts it to a *second-order linear* equation in $u$ (Derivation 6), and one known particular solution $y_1$ reduces it to first-order linear via $y = y_1 + 1/v$.

## 3. Step-by-Step Mathematical Proofs & Derivations

This section contains the module's complete proofs. Each derivation is self-contained, numbered step by step, and ends in a boxed result.

| Derivation | Statement proved |
| :--- | :--- |
| **0** | Separation of variables is a rigorous chain-rule argument (Theorem 2.10). |
| **1** | The integrating factor $\mu = e^{\int p}$, derived from scratch, solves $y' + p(t)y = q(t)$. |
| **2** | The formula of Derivation 1 gives *all* solutions; the linear IVP has exactly one. |
| **3** | Exactness criterion $M_y = N_x$: necessity and sufficiency on a rectangle; continued with the $\mu(x)$ and $\mu(y)$ integrating-factor constructions. |
| **4** | Full closed-form solution of the logistic equation via partial fractions, with the limit $t \to \infty$. |
| **5** | Bernoulli reduction $v = y^{1-n}$ produces a linear equation. |
| **6** | The Riccati substitution $y = -u'/(cu)$ produces a second-order linear equation. |

**Derivation 0 (Separation of variables, rigorously).** Suppose $y' = g(t)h(y)$ and $y$ is a solution on an interval $I$ with $h(y(t)) \neq 0$ on $I$.

1. Divide by $h(y(t))$ (legitimate — it is nonzero): $\dfrac{y'(t)}{h(y(t))} = g(t)$.
2. Let $H$ be any antiderivative of $1/h$ on the range of $y$. By the chain rule, $\dfrac{d}{dt} H(y(t)) = H'(y(t))\,y'(t) = \dfrac{y'(t)}{h(y(t))}$.
3. Therefore $\dfrac{d}{dt}H(y(t)) = g(t)$; integrating both sides over $[t_0, t]$ (Fundamental Theorem of Calculus) gives $H(y(t)) = G(t) + C$ with $G' = g$.
4. Since $H' = 1/h$ is continuous and nonzero, $H$ is strictly monotone, hence invertible on the relevant range, so $y(t) = H^{-1}(G(t) + C)$ — an explicit solution. Conversely, differentiating $H(y) = G(t) + C$ reproduces the ODE, so nothing extraneous was introduced.
5. Finally, each root $y^*$ of $h$ furnishes the equilibrium solution $y \equiv y^*$, which the division in step 1 excluded and which must be restored by inspection.

$$
\boxed{\;y' = g(t)h(y),\ h \neq 0 \;\Longleftrightarrow\; \int \frac{dy}{h(y)} = \int g(t)\,dt + C\;}
$$

The Leibniz ritual "$\frac{dy}{h(y)} = g(t)\,dt$" is exactly steps 1–3 with the chain rule hidden in the notation — convenient, and now justified.

### Derivation 1: The Integrating Factor from Scratch

**Claim.** Every linear equation $y' + p(t)y = q(t)$ with continuous $p, q$ is solved by multiplying through by $\mu(t) = e^{\int p(t)\,dt}$.

**Derivation.** The left-hand side $y' + py$ is *almost* the derivative of a product. We look for a multiplier $\mu(t) \gt 0$ that makes it exactly one.

1. **Impose the design condition.** We want

$$
\mu(t)\,y' + \mu(t)\,p(t)\,y = \big(\mu(t)\,y\big)'
$$

2. **Expand the right side** by the product rule: $(\mu y)' = \mu y' + \mu' y$. Matching coefficients of $y$ against step 1 (the $y'$ terms already agree) forces

$$
\mu'(t) = \mu(t)\,p(t)
$$

3. **Solve the design condition.** This is itself separable (Derivation 0) and $\mu \gt 0$ removes the absolute value: $\dfrac{\mu'}{\mu} = p$ gives $\ln \mu = \int p(t)\,dt$, i.e.

$$
\mu(t) = e^{\int p(t)\,dt}
$$

Any antiderivative works; different choices rescale $\mu$ by a positive constant, which cancels later.

4. **Use it.** Multiplying the original equation by this $\mu$ (positive, so no information is lost) collapses it to

$$
\big(\mu(t)\,y\big)' = \mu(t)\,q(t)
$$

5. **Integrate** (Fundamental Theorem of Calculus) and divide by $\mu(t) \neq 0$:

$$
\mu(t)\,y(t) = \int \mu(t)\,q(t)\,dt + C \quad\Longrightarrow\quad y(t) = \frac{1}{\mu(t)}\int \mu(t)\,q(t)\,dt + \frac{C}{\mu(t)}
$$

6. **Read off the structure.** The first term is one particular solution $y_p$; the second is the general solution $Ce^{-\int p}$ of the homogeneous equation $y' + py = 0$. Every linear solution is "particular plus homogeneous."

$$
\boxed{\;y(t) = e^{-\int p\,dt}\left(\int e^{\int p\,dt}\,q(t)\,dt + C\right)\;}
$$

### Derivation 2: The Formula Captures All Solutions — Uniqueness for Linear IVPs

**Claim.** Let $p, q$ be continuous on an interval $I$, $t_0 \in I$, $y_0 \in \mathbb{R}$. The IVP $y' + p(t)y = q(t)$, $y(t_0) = y_0$ has **exactly one** solution on all of $I$, namely the formula of Derivation 1 with $C$ fixed by the initial condition. No Lipschitz/Picard machinery is needed.

**Proof.**

1. **Existence.** Take $\mu(t) = e^{\int_{t_0}^{t} p(s)\,ds}$, so $\mu(t_0) = 1$, and define

$$
y(t) = \frac{1}{\mu(t)}\left(y_0 + \int_{t_0}^{t} \mu(s)\,q(s)\,ds\right)
$$

Both $\mu$ and the integral are differentiable on $I$ (FTC, continuity of $p$ and $q$), and differentiating verifies $y' + py = q$ directly; $y(t_0) = y_0$ by construction. So a solution exists on the *entire* interval $I$ — linear equations have no finite-time blow-up inside the domain of continuity of their coefficients.

2. **Uniqueness.** Suppose $y_1, y_2$ both solve the IVP on $I$. Set $w = y_1 - y_2$. Subtracting the two equations (linearity!) gives

$$
w' + p(t)\,w = 0, \qquad w(t_0) = 0
$$

3. Multiply by $\mu(t) \gt 0$: by the design property of Derivation 1,

$$
\big(\mu(t)\,w(t)\big)' = \mu(t)\big(w' + p\,w\big) = 0 \quad \text{on } I
$$

4. A function with vanishing derivative on an interval is constant (Mean Value Theorem — this is where connectedness of $I$ enters). Hence $\mu(t)w(t) = \mu(t_0)w(t_0) = 0$ for all $t \in I$.

5. Since $\mu(t) \gt 0$ everywhere, $w \equiv 0$, i.e. $y_1 = y_2$ on $I$.

6. **Completeness of the general solution.** The same computation with an arbitrary solution $y$ shows $(\mu y)' = \mu q$, hence $\mu y$ *must* equal $y_0 + \int_{t_0}^t \mu q$ — so every solution is an instance of the boxed formula: the family in Derivation 1 is the whole solution set.

$$
\boxed{\;\text{The linear IVP has a unique solution, defined on the full interval of continuity of } p, q\;}
$$

### Derivation 3: The Exactness Criterion $M_y = N_x$ — Necessity and Sufficiency

**Setting.** $M, N \in C^1(R)$ on an open rectangle $R = (a,b) \times (c,d)$ (the argument extends to any simply connected domain — this is the two-dimensional Poincaré lemma).

**Part A: Necessity.** Suppose a potential exists: $F \in C^1$ with $F_x = M$ and $F_y = N$ on $R$.

1. Since $M, N \in C^1$, the second partials $F_{xy} = M_y$ and $F_{yx} = N_x$ exist and are continuous, so $F \in C^2(R)$.
2. By **Clairaut's theorem** (symmetry of mixed partials for $C^2$ functions), $F_{xy} = F_{yx}$.
3. Therefore $M_y = F_{xy} = F_{yx} = N_x$ throughout $R$. $\square$

**Part B: Sufficiency.** Suppose $M_y = N_x$ on $R$. Fix a base point $(x_0, y_0) \in R$ and *construct* the potential by integrating along a horizontal-then-vertical path:

$$
F(x, y) = \int_{x_0}^{x} M(s, y)\,ds + \int_{y_0}^{y} N(x_0, u)\,du
$$

1. **Check $F_x = M$**: the second integral does not involve $x$, and the Fundamental Theorem of Calculus applied to the first gives $F_x(x,y) = M(x, y)$ immediately.
2. **Check $F_y = N$**: differentiate the first integral under the integral sign (legitimate: $M_y$ is continuous on the compact segment of integration — Leibniz rule), and apply FTC to the second:

$$
F_y(x,y) = \int_{x_0}^{x} M_y(s, y)\,ds + N(x_0, y)
$$

3. **Use the hypothesis** $M_y = N_x$ inside the integral, then apply FTC in the $s$-variable:

$$
F_y(x,y) = \int_{x_0}^{x} N_x(s, y)\,ds + N(x_0, y) = \big(N(x, y) - N(x_0, y)\big) + N(x_0, y) = N(x, y)
$$

4. Both defining conditions hold, so $F$ is a potential and the equation $M\,dx + N\,dy = 0$ is exact; its solution curves are the level sets $F(x,y) = C$. $\square$

**Where simple connectivity entered.** The rectangle let the L-shaped path from $(x_0,y_0)$ to $(x,y)$ stay inside the domain. On a non-simply-connected domain the construction can fail globally: for $M = -y/(x^2+y^2)$, $N = x/(x^2+y^2)$ on the punctured plane one checks $M_y = N_x$, yet the line integral around the unit circle equals $2\pi \neq 0$, so no single-valued potential exists — the criterion is genuinely *local*, and topology governs the *global* statement.

$$
\boxed{\;\text{On a simply connected domain: } M\,dx + N\,dy \text{ exact} \iff M_y = N_x\;}
$$

### Derivation 3 (continued): Integrating Factors that Restore Exactness

**Problem.** $M\,dx + N\,dy = 0$ fails the test ($M_y \neq N_x$). Find $\mu \neq 0$ making $\mu M\,dx + \mu N\,dy = 0$ exact.

**General condition.** Exactness of the rescaled form demands $(\mu M)_y = (\mu N)_x$, i.e. by the product rule

$$
\mu_y M + \mu M_y = \mu_x N + \mu N_x
$$

a PDE for $\mu$ — generally as hard as the original problem. It becomes an ODE under structural assumptions on $\mu$:

**Case 1: $\mu = \mu(x)$ only.** Then $\mu_y = 0$ and the condition reads $\mu M_y = \mu' N + \mu N_x$, so

$$
\frac{\mu'(x)}{\mu(x)} = \frac{M_y - N_x}{N}
$$

1. The left side depends on $x$ alone; hence a *necessary and sufficient* structural condition is that $\psi(x) := (M_y - N_x)/N$ be a function of $x$ only.
2. When it is, Derivation 0 solves the separable equation $\mu' = \psi(x)\mu$:

$$
\mu(x) = e^{\int \psi(x)\,dx}, \qquad \psi = \frac{M_y - N_x}{N}
$$

3. This $\mu$ is positive, so multiplying by it neither creates nor destroys solutions, and by construction $(\mu M)_y = (\mu N)_x$, so Derivation 3 Part B now applies to the rescaled equation.

**Case 2: $\mu = \mu(y)$ only.** Symmetrically, $\mu_x = 0$ gives $\mu' M + \mu M_y = \mu N_x$, hence

$$
\frac{\mu'(y)}{\mu(y)} = \frac{N_x - M_y}{M}
$$

which is solvable as $\mu(y) = e^{\int (N_x - M_y)/M\,dy}$ exactly when $(N_x - M_y)/M$ depends on $y$ alone.

**Worked micro-example.** $(3xy + y^2)\,dx + (x^2 + xy)\,dy = 0$: here $M_y = 3x + 2y$ and $N_x = 2x + y$, so the test fails; but

$$
\frac{M_y - N_x}{N} = \frac{x + y}{x(x + y)} = \frac{1}{x}
$$

depends on $x$ alone, so $\mu = e^{\int dx/x} = x$ (taking $x \gt 0$), and $(3x^2y + xy^2)\,dx + (x^3 + x^2y)\,dy = 0$ is exact with potential $F = x^3 y + \tfrac{1}{2}x^2y^2$.

$$
\boxed{\;\mu(x) = e^{\int \frac{M_y - N_x}{N}dx} \ \text{ or } \ \mu(y) = e^{\int \frac{N_x - M_y}{M}dy}, \ \text{whenever the quotient depends on the single stated variable}\;}
$$

Deeper fact: the linear integrating factor $\mu = e^{\int p}$ of Derivation 1 is precisely Case 1 applied to $(p(t)y - q(t))\,dt + dy = 0$ — the two theories are one.

### Derivation 4: The Logistic Equation, Solved Completely

**Claim.** The logistic IVP with growth rate $r \gt 0$ and carrying capacity $K \gt 0$,

$$
\frac{dy}{dt} = r\,y\left(1 - \frac{y}{K}\right), \qquad y(0) = y_0 \in (0, K)
$$

has the unique solution $y(t) = \dfrac{K}{1 + A e^{-rt}}$ with $A = \dfrac{K - y_0}{y_0}$, and $y(t) \to K$ as $t \to \infty$.

**Derivation.**

1. **Equilibria first**: $y \equiv 0$ and $y \equiv K$ solve the equation. By uniqueness (the right side is $C^1$, Picard–Lindelöf applies), no other solution ever touches these lines; a solution starting in $(0, K)$ stays in $(0, K)$ forever. This makes the following division legal for all time.
2. **Separate** (Derivation 0):

$$
\int \frac{dy}{y\left(1 - y/K\right)} = \int r\,dt
$$

3. **Partial fractions.** Rewrite the integrand with denominator $y(K - y)$ and decompose:

$$
\frac{K}{y(K - y)} = \frac{1}{y} + \frac{1}{K - y}
$$

(check: $\frac{1}{y} + \frac{1}{K-y} = \frac{(K - y) + y}{y(K-y)} = \frac{K}{y(K-y)}$).
4. **Integrate.** For $0 \lt y \lt K$ both denominators are positive, so no absolute values are needed:

$$
\ln y - \ln(K - y) = rt + C_1 \quad\Longrightarrow\quad \ln\!\left(\frac{y}{K - y}\right) = rt + C_1
$$

5. **Exponentiate and fix the constant** with $y(0) = y_0$:

$$
\frac{y}{K - y} = \frac{y_0}{K - y_0}\,e^{rt}
$$

6. **Solve for $y$.** Write $B = \frac{y_0}{K - y_0}e^{rt}$; then $y = B(K - y)$ gives $y(1 + B) = BK$, so $y = \frac{BK}{1 + B}$; dividing top and bottom by $B$:

$$
y(t) = \frac{K}{1 + A e^{-rt}}, \qquad A = \frac{K - y_0}{y_0} \gt 0
$$

7. **Long-time limit.** As $t \to \infty$, $e^{-rt} \to 0$, so $y(t) \to K$ monotonically from below; as $t \to -\infty$, $y \to 0$: the solution is the S-shaped curve interpolating the two equilibria. Its inflection occurs where $y'' = 0$; differentiating the ODE, $y'' = r y'(1 - 2y/K)$, which vanishes (for a non-equilibrium solution) exactly at $y = K/2$ — maximal growth at half capacity, with peak rate $rK/4$.

$$
\boxed{\;y(t) = \frac{K}{1 + \frac{K - y_0}{y_0}e^{-rt}} \;\xrightarrow[t \to \infty]{}\; K\;}
$$

For $y_0 \gt K$ the same formula holds with $A \lt 0$ and $y$ decreasing to $K$: the carrying capacity is globally attracting on $(0, \infty)$.

### Derivation 5: Bernoulli Reduction to a Linear Equation

**Claim.** The Bernoulli equation, nonlinear for $n \neq 0, 1$,

$$
y' + p(t)\,y = q(t)\,y^{n}
$$

is transformed by $v = y^{1-n}$ into the linear equation $v' + (1-n)p(t)\,v = (1-n)q(t)$.

**Derivation.** Work on an interval where $y \gt 0$ (or $y \neq 0$ with appropriate powers).

1. **Normalize the nonlinearity**: divide the equation by $y^{n}$:

$$
y^{-n}\,y' + p(t)\,y^{1-n} = q(t)
$$

2. **Choose the substitution the equation asks for.** The combination $y^{1-n}$ appears; call it $v = y^{1-n}$. By the chain rule,

$$
v' = (1-n)\,y^{-n}\,y' \quad\Longrightarrow\quad y^{-n}y' = \frac{v'}{1-n}
$$

— exactly the first term of step 1. The substitution is not a guess; it is read off from the equation's own structure.
3. **Substitute**:

$$
\frac{v'}{1-n} + p(t)\,v = q(t)
$$

4. **Multiply by $(1-n)$** to reach linear standard form:

$$
v' + (1-n)\,p(t)\,v = (1-n)\,q(t)
$$

5. **Solve** with the integrating factor $\mu = e^{(1-n)\int p\,dt}$ (Derivation 1), then invert the substitution: $y = v^{1/(1-n)}$. Lost solutions: $y \equiv 0$ (an equilibrium whenever $n \gt 0$) was excluded by the division in step 1 and must be restored by inspection.

$$
\boxed{\;v = y^{1-n} \ \Longrightarrow\ v' + (1-n)p\,v = (1-n)q \ \text{(linear)}\;}
$$

Sanity check with the logistic equation: $y' - ry = -\frac{r}{K}y^2$ is Bernoulli with $n = 2$, so $v = y^{-1}$ satisfies $v' + rv = r/K$ — a constant-coefficient linear equation whose solution $v = \frac{1}{K} + Ae^{-rt}$ reproduces Derivation 4 in one line.

### Derivation 6: Riccati Equations Linearize into Second-Order Form

**Claim.** For the Riccati equation with $c(t) \neq 0$ continuously differentiable,

$$
y' = a(t) + b(t)\,y + c(t)\,y^{2}
$$

the substitution $y = -\dfrac{u'}{c\,u}$ transforms it into the second-order **linear** equation

$$
u'' - \left(b + \frac{c'}{c}\right)u' + a\,c\,u = 0
$$

**Derivation.**

1. **Differentiate the substitution.** With $y = -u'/(cu)$, the quotient rule gives

$$
y' = -\frac{u''}{c\,u} + \frac{c'\,u'}{c^{2}\,u} + \frac{(u')^{2}}{c\,u^{2}}
$$

(the three terms come from differentiating $u'$, $c$, and $u$ respectively).
2. **Compute the quadratic term.** From the substitution, $y^2 = \dfrac{(u')^2}{c^2 u^2}$, hence

$$
c\,y^{2} = \frac{(u')^{2}}{c\,u^{2}}
$$

— *identical* to the third term of step 1. This cancellation is the entire point of the choice $y = -u'/(cu)$: the substitution is rigged so that the nonlinear term is absorbed.
3. **Substitute into the ODE.** The equation $y' - b\,y - c\,y^2 = a$ becomes

$$
-\frac{u''}{c\,u} + \frac{c'\,u'}{c^{2}\,u} + \frac{b\,u'}{c\,u} = a
$$

4. **Clear denominators**: multiply through by $-c\,u$:

$$
u'' - \frac{c'}{c}\,u' - b\,u' + a\,c\,u = 0 \quad\Longrightarrow\quad u'' - \left(b + \frac{c'}{c}\right)u' + a\,c\,u = 0
$$

5. **Conversely**, any nonvanishing solution $u$ of this linear equation yields the Riccati solution $y = -u'/(cu)$; since $u$ carries two integration constants but $y$ depends only on the ratio $u'/u$, exactly one constant survives — matching the one-parameter solution family a first-order equation must have.

$$
\boxed{\;y = -\frac{u'}{c\,u}\ \Longleftrightarrow\ u'' - \Big(b + \frac{c'}{c}\Big)u' + a\,c\,u = 0\;}
$$

**Companion fact (used in the exercises).** If one particular solution $y_1$ of the Riccati equation is known, the substitution $y = y_1 + \frac{1}{v}$ yields, after expanding $y^2 = y_1^2 + \frac{2y_1}{v} + \frac{1}{v^2}$ and cancelling the $y_1$-equation, the *first-order linear* equation

$$
v' + \big(b(t) + 2c(t)\,y_1(t)\big)v = -c(t)
$$

so one lucky guess buys the entire solution family — a structure with no analogue for generic nonlinear equations, and the reason Riccati equations sit exactly on the boundary between linear and nonlinear worlds.

## 4. Computational & Algorithmic Insights

### Symbolic Solving is Classification

A computer algebra system (SymPy's `dsolve`, Maple's `dsolve`, Mathematica's `DSolve`) solves first-order ODEs by running exactly the routing logic of this module — pattern-match the class, dispatch the method:

```text
ROUTE-FIRST-ORDER-ODE(f):
    if f(t, y) == g(t) * h(y):            return separate_and_integrate()      # Derivation 0
    if f(t, y) == q(t) - p(t) * y:        return integrating_factor()          # Derivation 1
    write as M dx + N dy = 0
    if dM/dy == dN/dx:                    return potential_function()          # Derivation 3
    if (My - Nx)/N depends on x alone:    return exactify_with_mu_x()          # Derivation 3 cont.
    if (Nx - My)/M depends on y alone:    return exactify_with_mu_y()
    if f == q * y**n - p * y:             return bernoulli(v = y**(1-n))       # Derivation 5
    if f(k*x, k*y) == f(x, y):            return homogeneous(v = y/x)
    if f == a + b*y + c*y**2:             return riccati()                     # Derivation 6
    return NO_ELEMENTARY_METHOD           # fall back to series / numerics
```

Each branch reduces the problem to *quadrature* — evaluating antiderivatives — so "solvable in closed form" really means two things stacked: (i) the equation matches a class, and (ii) the resulting integrals are elementary. Both can fail independently: $y' = y^2 - t$ matches the Riccati pattern, but the resulting second-order equation is Airy's equation, whose solutions are provably non-elementary; $y' = e^{-t^2}$ is trivially "solvable" yet its solution $\int e^{-t^2}dt$ has no elementary form. Liouville's theory (differential Galois theory) makes both failures precise theorems, not open problems.

### Computing a Direction Field

The direction field is embarrassingly parallel to compute and gives qualitative answers when formulas do not exist:

```text
SLOPE-FIELD(f, t_range, y_range, n):
    for t in linspace(t_range, n):
        for y in linspace(y_range, n):
            s = f(t, y)                       # one function evaluation
            draw unit segment at (t, y) with slope s
                # normalize: direction (1, s) / sqrt(1 + s^2)
                # so steep slopes do not blow up the picture
```

- **Cost**: $O(n^2)$ evaluations of $f$ for an $n \times n$ grid — negligible; a $50 \times 50$ field is 2500 evaluations.
- **Isocline overlay**: contouring $f(t,y) = c$ (marching squares on the same grid) draws isoclines for free; the $c = 0$ contour locates equilibria.
- **Numerical caution**: normalizing segment length matters because near a vertical asymptote of $f$ raw segments of slope $10^4$ destroy the plot; the unit-vector form $(1, s)/\sqrt{1+s^2}$ degrades gracefully.
- **From field to solver**: Euler's method is literally "follow the field": $y_{k+1} = y_k + h\,f(t_k, y_k)$. Each step commits an $O(h^2)$ local error; over $T/h$ steps the global error is $O(h)$. Runge–Kutta methods sample the field several times per step to cancel error terms ($O(h^4)$ global for RK4). The numerics module of this track develops this fully; here the point is conceptual — every numerical integrator is an interrogation strategy for the direction field.

### Stiffness: A Preview

Solve the innocuous linear equation $y' = -\lambda y$ ($\lambda \gt 0$ large, e.g. an RC circuit with time constant $\tau = 1/\lambda = 10^{-6}$ s) with explicit Euler:

$$
y_{k+1} = y_k - h\lambda y_k = (1 - h\lambda)y_k \quad\Longrightarrow\quad y_k = (1 - h\lambda)^k y_0
$$

The true solution $e^{-\lambda t}$ decays monotonically; the numerical solution decays **iff** $\lvert 1 - h\lambda \rvert \lt 1$, i.e.

$$
h \lt \frac{2}{\lambda}
$$

For $\lambda = 10^6$ the step size is capped at $2$ microseconds *even after the transient has died and the solution is essentially flat* — stability, not accuracy, dictates the cost. This is **stiffness**: the presence of a fast-decaying mode forces explicit methods to take absurdly small steps. Implicit Euler, $y_{k+1} = y_k - h\lambda y_{k+1}$, gives $y_{k+1} = y_k/(1 + h\lambda)$, stable for *every* $h \gt 0$ — the trade is one (here trivial, in general nonlinear) equation solve per step.

The ML echo is exact: gradient descent on a quadratic loss with curvature $\lambda_{\max}$ is explicit Euler on the gradient flow $\dot{\theta} = -\nabla L(\theta)$, and its stability constraint on the learning rate, $\eta \lt 2/\lambda_{\max}$, is the *same inequality*. Ill-conditioned losses are stiff ODEs; momentum, preconditioning, and Adam are, in this light, cheap approximations to implicit integration.

## 5. Real-World Physics & AI/ML Applications

### Physics: One Linear Equation, Many Costumes

**Newton's cooling.** A body at temperature $T$ in ambient $T_a$ loses heat at a rate proportional to the gap:

$$
\frac{dT}{dt} = -k\,(T - T_a) \quad\Longrightarrow\quad T(t) = T_a + (T_0 - T_a)\,e^{-kt}
$$

Linear, autonomous, one stable equilibrium $T_a$. Forensic use: two temperature measurements determine $k$ and let one solve *backwards* for the time at which $T = 37$ °C (exercise L2.2).

**RC and RL circuits.** Kirchhoff's voltage law for a resistor–capacitor loop driven by $V_s(t)$ gives

$$
RC\,\frac{dV}{dt} + V = V_s(t)
$$

a linear equation with time constant $\tau = RC$; for an RL loop, $L\,\frac{dI}{dt} + RI = V_s(t)$ with $\tau = L/R$. A step input charges the capacitor along $V = V_s(1 - e^{-t/\tau})$: 63% at $\tau$, 95% at $3\tau$, 99% at $5\tau$ — the engineer's rules of thumb are corollaries of Derivation 1.

**Mixing tanks.** Salt mass $x(t)$ in a well-stirred tank of volume $V$ with inflow concentration $c_{\text{in}}$ at rate $\rho$ and equal outflow obeys

$$
\frac{dx}{dt} = \rho\,c_{\text{in}} - \frac{\rho}{V}\,x
$$

— the same linear structure; the tank "forgets" its initial contents at exponential rate $\rho/V$. This is also the exact equation of an exponential moving average, making the mixing tank the hydraulic model of momentum buffers in optimizers.

**Radioactive decay chains.** A parent–daughter chain $A \to B \to \cdot$ with rates $\lambda_A, \lambda_B$ gives a *cascade* of first-order linear equations:

$$
\frac{dN_A}{dt} = -\lambda_A N_A, \qquad \frac{dN_B}{dt} = \lambda_A N_A - \lambda_B N_B
$$

Solve the first (pure exponential), substitute into the second, and Derivation 1 yields the Bateman solution $N_B(t) = \frac{\lambda_A N_A(0)}{\lambda_B - \lambda_A}\big(e^{-\lambda_A t} - e^{-\lambda_B t}\big)$ — a first taste of solving triangular linear *systems* one integrating factor at a time.

### AI/ML: Optimization is a First-Order ODE

**Gradient flow.** Gradient descent $\theta_{k+1} = \theta_k - \eta L'(\theta_k)$ is the explicit Euler discretization (step $\eta$) of the autonomous first-order ODE

$$
\frac{d\theta}{dt} = -L'(\theta)
$$

Everything in this module transfers. The nullcline analysis of Section 2 says equilibria are the critical points $L'(\theta^*) = 0$ and the flow decreases $L$ monotonically: $\frac{d}{dt}L(\theta(t)) = L'(\theta)\,\dot{\theta} = -\big(L'(\theta)\big)^2 \leq 0$. On the quadratic loss $L(\theta) = \frac{\lambda}{2}\theta^2$ the flow is *linear*, $\dot{\theta} = -\lambda\theta$, so $\theta(t) = \theta_0 e^{-\lambda t}$ and the loss decays as $e^{-2\lambda t}$: convergence speed is curvature. In many dimensions each Hessian eigendirection decays at its own rate $e^{-\lambda_i t}$, and the slowest mode $\lambda_{\min}$ dominates — the ODE-level meaning of ill-conditioning.

**Learning-rate schedules are time-dependent coefficients.** Gradient flow with a schedule $\eta(t)$ on a quadratic is

$$
\frac{d\theta}{dt} = -\eta(t)\,\lambda\,\theta \quad\Longleftrightarrow\quad \theta' + p(t)\,\theta = 0, \qquad p(t) = \lambda\,\eta(t)
$$

a non-autonomous linear equation, solved by Derivation 1: $\theta(t) = \theta_0\,e^{-\lambda\int_0^t \eta(s)\,ds}$. The integral $\int_0^t \eta$ is the *effective training time*: a cosine or step decay schedule changes when progress happens but only $\int \eta$ decides how much. Convergence to zero requires the divergence of $\int_0^\infty \eta(s)\,ds$ — the continuous ancestor of the Robbins–Monro condition in stochastic approximation.

**Neural ODEs.** Chen et al. (2018) replace a residual network's discrete update $h_{k+1} = h_k + f(h_k, \theta_k)$ with the ODE $\frac{dh}{dt} = f(h(t), t, \theta)$, training through a black-box integrator. The forward pass of the network *is* an IVP of exactly the type classified here; the adjoint method that makes training memory-efficient is itself another first-order ODE solved backwards in time. Continuous normalizing flows and score-based diffusion models push the same idea further — modern generative modeling runs on the machinery of this module.

### The Logistic Curve: From Populations to Sigmoids

The logistic equation of Derivation 4 is the minimal model of *self-limiting growth*: exponential reproduction $ry$ throttled by the crowding factor $(1 - y/K)$. Its fingerprints:

- **Population ecology** (Verhulst, 1838): $K$ is the carrying capacity; growth is fastest at $K/2$ with peak rate $rK/4$.
- **Epidemics**: in a closed population, simple contagion models give $I' = \beta I(1 - I/N)$ for the infected count — logistic with $K = N$; the inflection at $N/2$ is the epidemic's peak-growth day, and fitting $(r, K)$ to early case counts is a (delicate) standard exercise (L2.4).
- **Machine learning**: the solution curve *is* the sigmoid. With $K = 1$, $r = 1$, $y_0 = 1/2$, Derivation 4 gives exactly

$$
y(t) = \frac{1}{1 + e^{-t}} = \sigma(t)
$$

and the logistic ODE becomes the familiar derivative identity $\sigma' = \sigma(1 - \sigma)$ used in every backpropagation through a sigmoid. Logistic regression, softmax temperature curves, and neural scaling "S-curves" all inherit the same differential structure: saturation at both ends, maximal sensitivity in the middle.

- **Technology adoption / product growth**: cumulative adoption of an innovation follows the same S-curve (Bass-type models reduce to Riccati equations, linking back to Derivation 6).

The moral of the module in one line: a tiny vocabulary of first-order structures — linear relaxation $e^{-t/\tau}$, logistic saturation $\sigma(rt)$, power-law blow-up $1/(T - t)$ — accounts for an enormous share of observed dynamics, and *classification is how you recognize which one you are looking at*.
