### Problem L0.1: Classification Drill

**Problem Statement:** Classify each equation along every applicable axis: order; ODE vs PDE; linear vs nonlinear; autonomous vs non-autonomous; homogeneous vs non-homogeneous (for linear equations); explicit vs implicit.

1. $y' + t^2 y = \sin t$
2. $y'' + \sin y = 0$
3. $u_t = k\,u_{xx}$
4. $(y')^2 + y^2 = 1$
5. $y' = y(1 - y)$

**Intuition:** Classification is the routing step: each answer below immediately names the toolbox (or rules one out).

**Solution:**

| # | Equation | Order | ODE/PDE | Linear? | Autonomous? | Homogeneous? | Explicit? |
| :--- | :--- | :--- | :--- | :--- | :--- | :--- | :--- |
| 1 | $y' + t^2 y = \sin t$ | 1 | ODE | linear | non-autonomous ($t$ appears) | non-homogeneous ($q = \sin t$) | explicit |
| 2 | $y'' + \sin y = 0$ | 2 | ODE | nonlinear ($\sin y$) | autonomous | n/a (nonlinear) | explicit |
| 3 | $u_t = k\,u_{xx}$ | 2 | PDE (two independent variables) | linear | autonomous in $t$ | homogeneous | explicit |
| 4 | $(y')^2 + y^2 = 1$ | 1 | ODE | nonlinear (squared derivative) | autonomous | n/a | implicit |
| 5 | $y' = y(1 - y)$ | 1 | ODE | nonlinear ($y^2$ term) | autonomous | n/a | explicit |

Details worth noting:

- Equation 1 is solvable outright by the integrating factor $\mu = e^{t^3/3}$ — linearity is the strongest single property on the list.
- Equation 2 (the pendulum) is nonlinear because $\sin y$ is not a first-power appearance of $y$; its small-angle *linearization* $y'' + y = 0$ is what admits closed-form solutions.
- Equation 3 involves partial derivatives of $u(t,x)$ in two variables — a PDE, outside first-order ODE methods entirely.
- Equation 4 is implicit; solving for $y'$ gives two explicit branches $y' = \pm\sqrt{1 - y^2}$, plus the singular equilibrium solutions $y \equiv \pm 1$ where the branches meet.
- Equation 5 is the logistic equation: nonlinear but both separable and Bernoulli ($n = 2$), hence fully solvable.

$$
\boxed{\text{Every downstream method choice above was decided by classification alone}}
$$

> **Key Takeaway:** Order, linearity, autonomy, and explicitness are not descriptive trivia — each axis either unlocks a specific solution technique or bars it.

### Problem L0.2: Isoclines and a Self-Solving Isocline

**Problem Statement:** For the equation $y' = y - t$: (a) describe the isoclines; (b) show that one particular isocline is itself a solution of the ODE; (c) use the direction field to predict the behavior of solutions above and below that line, and confirm with the general solution.

**Intuition:** An isocline where the prescribed slope equals the isocline's own slope is a curve the field cannot push solutions off of.

**Solution:**

**(a)** The isocline of slope $c$ is the set where $f(t,y) = y - t = c$, i.e. the straight line

$$
y = t + c
$$

All isoclines are parallel lines of slope $1$; along $y = t + c$ every field segment has slope $c$.

**(b)** A line $y = t + c$ is a solution iff its actual slope ($1$) equals the slope the field assigns on it ($c$). That happens exactly for $c = 1$:

$$
y(t) = t + 1 \quad\Longrightarrow\quad y' = 1 \quad\text{and}\quad y - t = 1
$$

Both sides agree, so $y = t + 1$ solves the ODE.

**(c)** Above the line ($y \gt t + 1$): the field slope $y - t \gt 1$ exceeds the line's slope, so solutions steepen and pull *away* upward. Below it ($y \lt t + 1$): slope is less than $1$, and solutions fall away downward. The line is a repelling solution. Confirmation: the equation is linear, $y' - y = -t$, with integrating factor $e^{-t}$:

$$
\left(e^{-t}y\right)' = -t\,e^{-t} \quad\Longrightarrow\quad e^{-t}y = (t + 1)e^{-t} + C \quad\Longrightarrow\quad y = t + 1 + C\,e^{t}
$$

The deviation from the special line is $C e^{t}$, growing exponentially — solutions diverge from $y = t + 1$ in forward time on both sides, exactly as the field predicted.

$$
\boxed{y = t + 1 + C\,e^{t}, \quad \text{with } y = t + 1 \text{ the repelling straight-line solution}}
$$

> **Key Takeaway:** Isoclines organize the direction field into level sets; when an isocline's own slope matches its assigned slope, it is a solution, and the field's behavior around it reveals stability before any formula is computed.

### Problem L0.3: Superposition Lives and Dies with Linearity

**Problem Statement:** (a) Show that for the homogeneous linear equation $y' + p(t)y = 0$, any linear combination $c_1 y_1 + c_2 y_2$ of solutions is again a solution. (b) Show by explicit counterexample that the sum of two solutions of $y' = y^2$ is generally **not** a solution.

**Intuition:** Superposition is a property of the *operator*, not of differential equations in general.

**Solution:**

**(a)** Define the operator $L[y] = y' + p(t)y$. For any differentiable $y_1, y_2$ and constants $c_1, c_2$, differentiation is linear, so

$$
L[c_1 y_1 + c_2 y_2] = (c_1 y_1 + c_2 y_2)' + p\,(c_1 y_1 + c_2 y_2) = c_1(y_1' + p y_1) + c_2(y_2' + p y_2) = c_1 L[y_1] + c_2 L[y_2]
$$

If $L[y_1] = L[y_2] = 0$, the right side vanishes, so $c_1 y_1 + c_2 y_2$ solves the homogeneous equation: the solution set is a vector space (one-dimensional, spanned by $e^{-\int p}$).

**(b)** Solutions of $y' = y^2$ (separable) are $y = \frac{1}{C - t}$ together with $y \equiv 0$. Take the two solutions on a common interval, e.g. for $t \lt 1$:

$$
y_1 = \frac{1}{1 - t}, \qquad y_2 = \frac{1}{2 - t}, \qquad s = y_1 + y_2
$$

Then

$$
s' = y_1' + y_2' = y_1^2 + y_2^2
$$

but being a solution would require $s' = s^2 = y_1^2 + 2y_1 y_2 + y_2^2$. The discrepancy is the cross term $2 y_1 y_2 = \frac{2}{(1-t)(2-t)}$, which is nonzero everywhere on the interval. So $s$ is not a solution.

$$
\boxed{L \text{ linear} \Rightarrow \text{superposition holds; the cross term } 2y_1y_2 \text{ destroys it for } y' = y^2}
$$

> **Key Takeaway:** "Homogeneous + particular" solution structure, and everything built on it, is a theorem about linear operators; the first nonlinear term (here $y^2$) generates cross terms that break it.

### Problem L0.4: Autonomy Means Time-Translation Invariance

**Problem Statement:** (a) Prove: if $y(t)$ solves the autonomous equation $y' = f(y)$ on an interval $I$, then for any constant $c$ the shifted function $z(t) = y(t - c)$ solves the same equation on the shifted interval. (b) Show this fails for the non-autonomous equation $y' = 2t$ by exhibiting a solution whose shift is not a solution.

**Intuition:** An autonomous law has no clock: only the state matters, so restarting the experiment later replays the same trajectory.

**Solution:**

**(a)** Let $z(t) = y(t - c)$. By the chain rule (inner derivative $1$):

$$
z'(t) = y'(t - c)
$$

Since $y$ solves the equation at time $t - c$: $y'(t - c) = f\big(y(t - c)\big) = f\big(z(t)\big)$. Chaining the equalities,

$$
z'(t) = f(z(t))
$$

so $z$ is a solution wherever $t - c \in I$. The whole solution family of an autonomous equation is generated from representative trajectories by horizontal translation — which is why the phase line (the $y$-axis with arrows) is a complete qualitative description.

**(b)** Take $y' = 2t$, solved by $y(t) = t^2$. The shift $z(t) = (t - c)^2$ has

$$
z'(t) = 2(t - c) \neq 2t \quad \text{whenever } c \neq 0
$$

so $z$ fails the equation: the right-hand side reads the clock $t$, not the state, and shifting the trajectory desynchronizes it from the forcing.

$$
\boxed{y' = f(y) \Rightarrow y(t - c) \text{ is a solution for every } c; \text{ false once } f \text{ depends on } t}
$$

> **Key Takeaway:** Autonomy converts a two-dimensional picture (direction field in the $(t,y)$ plane) into a one-dimensional one (phase line) — a symmetry, translation in time, is doing the compression.

### Problem L1.1: A Separable Equation with Finite-Time Blow-Up

**Problem Statement:** Solve the IVP

$$
y' = 1 + y^2, \qquad y(0) = 0
$$

justifying each manipulation, and determine the maximal interval of existence.

**Intuition:** The right-hand side is smooth everywhere, yet the quadratic growth in $y$ will be seen to destroy the solution in finite time — smoothness of $f$ buys local, not global, existence.

**Solution:**

**Step 1 — Classify.** First order, nonlinear ($y^2$), autonomous, separable with $g(t) = 1$, $h(y) = 1 + y^2$. Since $h \gt 0$ everywhere, no equilibria exist and division by $h$ is globally legal.

**Step 2 — Separate rigorously.** Dividing by $1 + y^2$:

$$
\frac{y'(t)}{1 + y(t)^2} = 1
$$

The left side is $\frac{d}{dt}\arctan(y(t))$ by the chain rule. Integrating from $0$ to $t$:

$$
\arctan(y(t)) - \arctan(y(0)) = t
$$

**Step 3 — Apply the initial condition.** $\arctan(0) = 0$, so $\arctan(y(t)) = t$, and inverting on the range of $\arctan$:

$$
y(t) = \tan t
$$

**Step 4 — Maximal interval.** The inversion is valid while $t = \arctan(y)$ stays in $\left(-\frac{\pi}{2}, \frac{\pi}{2}\right)$; as $t \to \frac{\pi}{2}^{-}$, $y(t) = \tan t \to +\infty$. The solution cannot be extended past $\pm\frac{\pi}{2}$.

$$
\boxed{y(t) = \tan t, \quad t \in \left(-\tfrac{\pi}{2},\, \tfrac{\pi}{2}\right)}
$$

> **Key Takeaway:** Separation of variables is the chain rule for $H(y(t))$ with $H' = 1/h$; and superlinear growth of $f$ in $y$ can force finite-time blow-up even when $f$ is entire.

### Problem L1.2: Linear Equation via the Integrating Factor

**Problem Statement:** Solve the IVP

$$
t\,y' + 2y = 4t^2, \qquad y(1) = 2
$$

and state the interval on which the solution is valid.

**Intuition:** Normalize to standard form first; the leading coefficient $t$ vanishing at $t = 0$ will carve the line into two candidate intervals.

**Solution:**

**Step 1 — Standard form.** For $t \neq 0$, divide by $t$:

$$
y' + \frac{2}{t}\,y = 4t
$$

so $p(t) = 2/t$, $q(t) = 4t$, both continuous on $t \gt 0$ (the component containing the initial point $t = 1$).

**Step 2 — Integrating factor.**

$$
\mu(t) = e^{\int \frac{2}{t}\,dt} = e^{2\ln t} = t^2 \qquad (t \gt 0)
$$

**Step 3 — Collapse to a total derivative.** Multiplying the standard form by $t^2$:

$$
t^2 y' + 2t\,y = 4t^3 \quad\Longleftrightarrow\quad \left(t^2 y\right)' = 4t^3
$$

**Step 4 — Integrate and solve.**

$$
t^2 y = t^4 + C \quad\Longrightarrow\quad y = t^2 + \frac{C}{t^2}
$$

**Step 5 — Initial condition.** $y(1) = 1 + C = 2$ gives $C = 1$.

**Step 6 — Interval.** The solution $y = t^2 + t^{-2}$ satisfies the original equation for all $t \neq 0$; the maximal interval through $t = 1$ is $t \gt 0$ ($y \to \infty$ as $t \to 0^{+}$, so no extension across $0$ exists). Note the structure: particular solution $t^2$ plus homogeneous family $C t^{-2} = C e^{-\int p}$.

$$
\boxed{y(t) = t^2 + \frac{1}{t^2}, \qquad t \gt 0}
$$

> **Key Takeaway:** The integrating factor turns the entire left side into $(\mu y)'$; singular points of the *normalized* coefficients (here $t = 0$) delimit the guaranteed interval of validity.

### Problem L1.3: An Exact Equation

**Problem Statement:** Verify exactness and solve

$$
(2xy + 3x^2)\,dx + (x^2 + 2y)\,dy = 0
$$

giving the general solution as an implicit family.

**Intuition:** If the form passes the curl test $M_y = N_x$, the solutions are level curves of a single potential function, found by partial integration.

**Solution:**

**Step 1 — Exactness test.** $M = 2xy + 3x^2$, $N = x^2 + 2y$:

$$
M_y = 2x, \qquad N_x = 2x
$$

Equal on all of $\mathbb{R}^2$ (simply connected), so a global potential $F$ exists.

**Step 2 — Integrate $M$ in $x$.**

$$
F(x,y) = \int (2xy + 3x^2)\,dx = x^2 y + x^3 + g(y)
$$

with the integration "constant" $g$ allowed to depend on $y$.

**Step 3 — Match $F_y = N$.**

$$
F_y = x^2 + g'(y) \overset{!}{=} x^2 + 2y \quad\Longrightarrow\quad g'(y) = 2y \quad\Longrightarrow\quad g(y) = y^2
$$

**Step 4 — Write the solution family.** $F(x,y) = x^2 y + x^3 + y^2$, and solutions are its level sets:

$$
x^2 y + x^3 + y^2 = C
$$

**Step 5 — Verify** (differentiate implicitly): $2xy\,dx + x^2\,dy + 3x^2\,dx + 2y\,dy = 0$ reproduces the original form exactly. At points with $F_y = x^2 + 2y \neq 0$, the implicit function theorem upgrades the level set to a genuine differentiable solution branch $y(x)$.

$$
\boxed{x^2 y + x^3 + y^2 = C}
$$

> **Key Takeaway:** Exact equations are not "solved" so much as *recognized*: the ODE is the statement that $F$ is constant along trajectories, and finding $F$ is two partial integrations.

### Problem L1.4: Restoring Exactness with an Integrating Factor $\mu(x)$

**Problem Statement:** The equation

$$
(3xy + y^2)\,dx + (x^2 + xy)\,dy = 0
$$

is not exact. Find an integrating factor depending on $x$ alone, make the equation exact, and solve it (work on $x \gt 0$).

**Intuition:** The defect $M_y - N_x$, measured relative to $N$, collapses to a function of $x$ alone — the signature that a $\mu(x)$ repair exists.

**Solution:**

**Step 1 — Confirm failure.** $M = 3xy + y^2$, $N = x^2 + xy$:

$$
M_y = 3x + 2y, \qquad N_x = 2x + y, \qquad M_y - N_x = x + y \neq 0
$$

**Step 2 — Test for a $\mu(x)$.**

$$
\frac{M_y - N_x}{N} = \frac{x + y}{x(x + y)} = \frac{1}{x}
$$

a function of $x$ alone — the criterion is met.

**Step 3 — Compute $\mu$.**

$$
\mu(x) = e^{\int \frac{dx}{x}} = e^{\ln x} = x \qquad (x \gt 0)
$$

**Step 4 — Rescale and re-test.** Multiply through by $x$:

$$
(3x^2 y + x y^2)\,dx + (x^3 + x^2 y)\,dy = 0
$$

Now $\tilde{M}_y = 3x^2 + 2xy$ and $\tilde{N}_x = 3x^2 + 2xy$: exact.

**Step 5 — Find the potential.**

$$
F = \int (3x^2 y + x y^2)\,dx = x^3 y + \frac{x^2 y^2}{2} + g(y)
$$

Then $F_y = x^3 + x^2 y + g'(y)$ must equal $\tilde{N} = x^3 + x^2 y$, so $g' = 0$ and $g$ is a true constant.

**Step 6 — Solution family.**

$$
\boxed{x^3 y + \tfrac{1}{2}\,x^2 y^2 = C \qquad (x \gt 0)}
$$

Since $\mu = x \gt 0$ on the working domain, multiplying by it neither created nor destroyed solutions.

> **Key Takeaway:** When $(M_y - N_x)/N$ depends only on $x$, the separable design equation $\mu' = \mu\,(M_y - N_x)/N$ hands you an integrating factor; the exactness machinery then finishes the job.

### Problem L1.5: A Bernoulli Equation

**Problem Statement:** Solve on $t \gt 0$:

$$
y' + \frac{1}{t}\,y = y^2
$$

**Intuition:** The $y^2$ right-hand side makes this Bernoulli with $n = 2$; the substitution $v = y^{1-n} = 1/y$ is dictated by the equation's own structure.

**Solution:**

**Step 1 — Identify.** Bernoulli form $y' + p\,y = q\,y^n$ with $p = 1/t$, $q = 1$, $n = 2$. Note $y \equiv 0$ is a solution (it was excluded by the division below and is restored at the end).

**Step 2 — Divide by $y^2$** (on intervals where $y \neq 0$):

$$
y^{-2}y' + \frac{1}{t}\,y^{-1} = 1
$$

**Step 3 — Substitute.** Let $v = y^{-1}$, so $v' = -y^{-2}y'$, i.e. $y^{-2}y' = -v'$:

$$
-v' + \frac{1}{t}\,v = 1 \quad\Longleftrightarrow\quad v' - \frac{1}{t}\,v = -1
$$

— linear, exactly as Derivation 5 predicts with factor $(1 - n) = -1$.

**Step 4 — Integrating factor.** $\mu = e^{-\int dt/t} = 1/t$:

$$
\left(\frac{v}{t}\right)' = -\frac{1}{t} \quad\Longrightarrow\quad \frac{v}{t} = -\ln t + C \quad\Longrightarrow\quad v = t\,(C - \ln t)
$$

**Step 5 — Undo the substitution.**

$$
y = \frac{1}{v} = \frac{1}{t\,(C - \ln t)}
$$

valid on subintervals of $t \gt 0$ where $C \neq \ln t$; the solution blows up as $t \to e^{C}$ — another finite-time singularity born from a quadratic nonlinearity.

**Step 6 — Verify** (spot check): with $w = t(C - \ln t)$, $y = 1/w$, one computes $y' = -w'/w^2 = -(C - \ln t - 1)/w^2$ and $-\frac{y}{t} + y^2 = \frac{-w/t + 1}{w^2} = \frac{-(C - \ln t) + 1}{w^2}$, which agree.

$$
\boxed{y(t) = \frac{1}{t\,(C - \ln t)} \quad \text{together with } y \equiv 0}
$$

> **Key Takeaway:** Bernoulli equations are nonlinear by classification but linear in disguise: the power substitution $v = y^{1-n}$ is read directly off the equation, and the lost equilibrium $y = 0$ must be restored by inspection.

### Problem L1.6: A Homogeneous-Degree Equation

**Problem Statement:** Solve

$$
y' = \frac{y^2 + xy}{x^2} \qquad (x \gt 0)
$$

**Intuition:** The right side is unchanged under $(x, y) \mapsto (\lambda x, \lambda y)$ — a scaling symmetry that the substitution $v = y/x$ converts into separability.

**Solution:**

**Step 1 — Verify degree-zero homogeneity.**

$$
f(\lambda x, \lambda y) = \frac{\lambda^2 y^2 + \lambda^2 xy}{\lambda^2 x^2} = \frac{y^2 + xy}{x^2} = f(x, y)
$$

so $f$ depends only on the ratio $y/x$: indeed $f = (y/x)^2 + (y/x)$.

**Step 2 — Substitute $y = v x$.** Then $y' = v + x v'$, and the equation becomes

$$
v + x\,v' = v^2 + v \quad\Longrightarrow\quad x\,\frac{dv}{dx} = v^2
$$

**Step 3 — Separate** (for $v \neq 0$; $v \equiv 0$, i.e. $y \equiv 0$, is an equilibrium solution):

$$
\int \frac{dv}{v^2} = \int \frac{dx}{x} \quad\Longrightarrow\quad -\frac{1}{v} = \ln x + C
$$

**Step 4 — Solve for $v$, then $y$.**

$$
v = -\frac{1}{\ln x + C} \quad\Longrightarrow\quad y = x\,v = -\frac{x}{\ln x + C}
$$

**Step 5 — Verify.** With $y = -x/(\ln x + C)$: differentiate, $y' = \frac{-(\ln x + C) + 1}{(\ln x + C)^2}$; and compute $\frac{y^2}{x^2} + \frac{y}{x} = \frac{1}{(\ln x + C)^2} - \frac{1}{\ln x + C} = \frac{1 - (\ln x + C)}{(\ln x + C)^2}$. The two agree.

$$
\boxed{y(x) = -\frac{x}{\ln x + C} \quad \text{together with } y \equiv 0}
$$

Each solution lives on one side of the singular abscissa $x = e^{-C}$.

> **Key Takeaway:** A scaling symmetry of $f$ is an invitation to use the scale-invariant variable $v = y/x$; symmetry reduction, here in its simplest form, is the deep principle behind most "substitution tricks."

### Problem L2.1: RC Circuit with a Step Input

**Problem Statement:** A series RC circuit has $R = 1\,\text{k}\Omega$ and $C = 100\,\mu\text{F}$. At $t = 0$ the capacitor is uncharged and a constant source $V_s = 5\,\text{V}$ is switched in, giving

$$
RC\,\frac{dV}{dt} + V = V_s, \qquad V(0) = 0
$$

Find $V(t)$, the time constant, and the time for the capacitor to reach $95\%$ of its final voltage.

**Intuition:** A step-driven linear first-order system always relaxes exponentially toward its new equilibrium with rate set by the time constant $\tau = RC$.

**Solution:**

**Step 1 — Time constant.** $\tau = RC = (10^3\,\Omega)(10^{-4}\,\text{F}) = 0.1\,\text{s}$. Standard form:

$$
V' + \frac{1}{\tau}V = \frac{V_s}{\tau}
$$

**Step 2 — Integrating factor.** $\mu = e^{t/\tau}$:

$$
\left(e^{t/\tau}V\right)' = \frac{V_s}{\tau}\,e^{t/\tau} \quad\Longrightarrow\quad e^{t/\tau}V = V_s\,e^{t/\tau} + C_1
$$

**Step 3 — Apply $V(0) = 0$.** $0 = V_s + C_1$, so $C_1 = -V_s$ and

$$
V(t) = V_s\left(1 - e^{-t/\tau}\right) = 5\left(1 - e^{-10t}\right)\ \text{V}
$$

The solution decomposes as equilibrium ($V_s$, the particular solution) plus a decaying transient ($-V_s e^{-t/\tau}$, the homogeneous part).

**Step 4 — Time to 95%.** Solve $1 - e^{-t/\tau} = 0.95$:

$$
e^{-t/\tau} = 0.05 \quad\Longrightarrow\quad t = \tau \ln 20 \approx 0.1 \times 2.996 \approx 0.30\ \text{s}
$$

confirming the engineering rule of thumb "settled after three time constants" ($\ln 20 \approx 3.0$). For reference: $63\%$ at $\tau$ ($1 - e^{-1}$), $99\%$ at $\approx 4.6\,\tau$ ($\ln 100$).

$$
\boxed{V(t) = 5\left(1 - e^{-10t}\right)\ \text{V}, \qquad \tau = 0.1\ \text{s}, \qquad t_{95\%} = \tau\ln 20 \approx 0.30\ \text{s}}
$$

> **Key Takeaway:** Every stable first-order linear system answers a step input the same way — exponential approach to the new equilibrium — and $\tau$ compresses all the physics ($R$, $C$) into one time scale.

### Problem L2.2: Newton Cooling as Forensics — Estimating Time of Death

**Problem Statement:** A body is found in a room held at $T_a = 20\,$°C. At 1:00 am its temperature is $30\,$°C; at 2:00 am it is $28\,$°C. Assuming Newton's law of cooling and normal body temperature $37\,$°C at the moment of death, estimate the time of death.

**Intuition:** Two measurements pin down the unknown rate constant $k$; then run the exponential clock backwards until it reads $37\,$°C.

**Solution:**

**Step 1 — Model and general solution.** With $u = T - T_a$ the law $T' = -k(T - T_a)$ becomes $u' = -ku$, so

$$
T(t) = 20 + u_0\,e^{-kt}
$$

Measure $t$ in hours from 1:00 am; then $u(0) = 30 - 20 = 10$.

**Step 2 — Determine $k$ from the second measurement.** $u(1) = 28 - 20 = 8$:

$$
8 = 10\,e^{-k} \quad\Longrightarrow\quad k = \ln\frac{10}{8} = \ln 1.25 \approx 0.2231\ \text{hr}^{-1}
$$

**Step 3 — Solve backwards for the death time $t_d$.** At death, $T = 37$, i.e. $u = 17$:

$$
17 = 10\,e^{-k t_d} \quad\Longrightarrow\quad e^{-k t_d} = 1.7 \quad\Longrightarrow\quad t_d = -\frac{\ln 1.7}{k}
$$

**Step 4 — Numbers.**

$$
t_d = -\frac{0.5306}{0.2231} \approx -2.38\ \text{hours}
$$

i.e. about $2$ hours $23$ minutes *before* 1:00 am.

$$
\boxed{\text{Estimated time of death} \approx 10\!:\!37\ \text{pm (about 2.4 h before the first measurement)}}
$$

**Model caveats** (what a forensic examiner would add): the ambient temperature must truly be constant, the body cools with a single effective $k$ (ignoring the initial plateau caused by internal heat conduction), and $37\,$°C at death is an average. The ODE gives a defensible point estimate; the error bars come from these assumptions.

> **Key Takeaway:** A first-order linear model turns two temperature readings into a rate constant and lets you integrate *backwards* in time — the same equation answers prediction and postdiction alike.

### Problem L2.3: Mixing Tank

**Problem Statement:** A tank holds $100$ L of pure water. Brine at concentration $2$ g/L flows in at $5$ L/min; the tank is kept perfectly mixed and drains at $5$ L/min. Let $x(t)$ be the grams of salt at time $t$ (minutes). Find $x(t)$, the limiting salt mass, and the time to reach half of it.

**Intuition:** Salt balance: inflow carries a constant salt rate; outflow carries salt at the tank's current concentration — a linear relaxation toward the inflow's equilibrium.

**Solution:**

**Step 1 — Balance law.** Inflow rate of salt: $5\ \text{L/min} \times 2\ \text{g/L} = 10$ g/min. Outflow: concentration $x/100$ g/L leaving at $5$ L/min, i.e. $x/20$ g/min. Volume stays at $100$ L (equal flows), so

$$
\frac{dx}{dt} = 10 - \frac{x}{20}, \qquad x(0) = 0
$$

**Step 2 — Solve** (linear; $\mu = e^{t/20}$):

$$
\left(e^{t/20}x\right)' = 10\,e^{t/20} \quad\Longrightarrow\quad e^{t/20}x = 200\,e^{t/20} + C \quad\Longrightarrow\quad x = 200 + C\,e^{-t/20}
$$

With $x(0) = 0$: $C = -200$, so

$$
x(t) = 200\left(1 - e^{-t/20}\right)
$$

**Step 3 — Limit.** As $t \to \infty$, $x \to 200$ g — exactly $100$ L at the inflow concentration $2$ g/L: the tank asymptotically "forgets" its initial pure water, at exponential rate $\rho/V = 1/20$ per minute.

**Step 4 — Half-way time.** $1 - e^{-t/20} = \tfrac{1}{2}$ gives

$$
t = 20\ln 2 \approx 13.9\ \text{min}
$$

$$
\boxed{x(t) = 200\left(1 - e^{-t/20}\right)\ \text{g}, \qquad x_\infty = 200\ \text{g}, \qquad t_{1/2} = 20\ln 2 \approx 13.9\ \text{min}}
$$

**ML aside:** the tank equation is formally identical to an exponential moving average $\dot{x} = \rho\,c_{\text{in}}(t) - \rho\,x/V$: momentum buffers and EMA statistics in optimizers are mixing tanks for gradients.

> **Key Takeaway:** Conservation (rate in minus rate out) plus well-mixedness yields a linear first-order ODE whose time constant $V/\rho$ is the tank's memory length.

### Problem L2.4: Logistic Epidemic — Fit and Forecast

**Problem Statement:** In a closed town of $N = 10{,}000$ people, an infection spreads according to the logistic law

$$
\frac{dI}{dt} = r\,I\left(1 - \frac{I}{N}\right), \qquad I(0) = 100
$$

with fitted growth rate $r = 0.3\ \text{day}^{-1}$. (a) Write $I(t)$ explicitly. (b) On what day does the epidemic reach its inflection point (fastest spread), and how many are infected then? (c) What is the peak daily growth rate?

**Intuition:** Early on, growth looks exponential at rate $r$; saturation bends the curve into an S whose steepest point sits exactly at half the population.

**Solution:**

**(a)** By the closed-form logistic solution (Derivation 4) with $K = N$:

$$
I(t) = \frac{N}{1 + A\,e^{-rt}}, \qquad A = \frac{N - I_0}{I_0} = \frac{9900}{100} = 99
$$

so

$$
I(t) = \frac{10000}{1 + 99\,e^{-0.3 t}}
$$

**(b)** The inflection of a logistic curve is at $I = N/2$ (from $I'' = rI'(1 - 2I/N) = 0$ on non-equilibrium solutions). Set $I(t^*) = 5000$:

$$
1 + 99\,e^{-0.3 t^*} = 2 \quad\Longrightarrow\quad e^{-0.3 t^*} = \frac{1}{99} \quad\Longrightarrow\quad t^* = \frac{\ln 99}{0.3} \approx \frac{4.595}{0.3} \approx 15.3\ \text{days}
$$

with $5000$ people infected at that moment.

**(c)** Peak daily growth is the ODE evaluated at $I = N/2$:

$$
I'_{\max} = r\,\frac{N}{2}\left(1 - \frac{1}{2}\right) = \frac{rN}{4} = \frac{0.3 \times 10000}{4} = 750\ \text{cases/day}
$$

$$
\boxed{I(t) = \frac{10000}{1 + 99\,e^{-0.3t}}, \qquad t^* = \frac{\ln 99}{0.3} \approx 15.3\ \text{days}, \qquad I'_{\max} = 750\ \text{cases/day}}
$$

**Fitting caveat:** early-phase data constrain $r$ well but $N$ poorly (exponential and logistic are nearly indistinguishable while $I \ll N$), which is why naive early-epidemic forecasts of the total size are notoriously unstable.

> **Key Takeaway:** The logistic solution formula converts three interpretable parameters ($r$, $N$, $I_0$) into concrete forecasts — inflection date, peak burden — and its early-time degeneracy explains a real forecasting pathology.

### Problem L2.5: Gradient Flow on a Quadratic Loss — Exact Convergence Rate

**Problem Statement:** Consider gradient flow $\dot{\theta} = -\nabla L(\theta)$ on the quadratic loss $L(\theta) = \frac{1}{2}\theta^{\top} A\,\theta$, where $A$ is symmetric positive definite with eigenvalues $\lambda_1 \geq \dots \geq \lambda_n \gt 0$. (a) Solve the flow exactly in the eigenbasis. (b) Show $L(\theta(t)) \leq e^{-2\lambda_n t}L(\theta(0))$ and identify the bottleneck. (c) Connect the discrete-time stability constraint of gradient *descent* to the stiffness bound $\eta \lt 2/\lambda_1$.

**Intuition:** In the eigenbasis the flow decouples into $n$ independent scalar ODEs $\dot{c}_i = -\lambda_i c_i$ — this module's simplest equation, run in parallel.

**Solution:**

**(a)** Diagonalize $A = Q\Lambda Q^{\top}$ (spectral theorem) and set $c = Q^{\top}\theta$. Since $\nabla L = A\theta$,

$$
\dot{c} = Q^{\top}\dot{\theta} = -Q^{\top}A\,\theta = -\Lambda c \quad\Longrightarrow\quad \dot{c}_i = -\lambda_i c_i \quad\Longrightarrow\quad c_i(t) = c_i(0)\,e^{-\lambda_i t}
$$

Each Hessian eigendirection relaxes independently at its own exponential rate — the linear, autonomous, homogeneous case of this module, $n$ times over.

**(b)** In eigencoordinates $L = \frac{1}{2}\sum_i \lambda_i c_i^2$, so

$$
L(\theta(t)) = \frac{1}{2}\sum_{i} \lambda_i\,c_i(0)^2\,e^{-2\lambda_i t} \leq e^{-2\lambda_n t}\,\frac{1}{2}\sum_i \lambda_i c_i(0)^2 = e^{-2\lambda_n t}\,L(\theta(0))
$$

using $e^{-2\lambda_i t} \leq e^{-2\lambda_n t}$ for every $i$. The **smallest** eigenvalue $\lambda_n$ — the flattest direction — is the bottleneck: time to shrink the loss by a factor $\varepsilon$ is $t = \frac{\ln(1/\varepsilon)}{2\lambda_n}$.

**(c)** Gradient descent $\theta_{k+1} = \theta_k - \eta A\theta_k$ is explicit Euler with step $\eta$ on this flow. In eigencoordinates $c_i^{(k)} = (1 - \eta\lambda_i)^k c_i^{(0)}$, which converges iff $\lvert 1 - \eta\lambda_i \rvert \lt 1$ for all $i$, i.e.

$$
\eta \lt \frac{2}{\lambda_1}
$$

The *largest* eigenvalue caps the step size while the *smallest* sets the speed: the number of iterations scales with the condition number $\kappa = \lambda_1/\lambda_n$. Ill-conditioned losses are stiff ODEs (fast modes constrain the step; slow modes carry the remaining work).

$$
\boxed{c_i(t) = c_i(0)e^{-\lambda_i t}, \qquad L(t) \leq e^{-2\lambda_n t}L(0), \qquad \eta_{\text{stable}} \lt 2/\lambda_1}
$$

> **Key Takeaway:** Optimization on a quadratic is exactly this module's linear ODE in each eigendirection; conditioning, learning-rate limits, and convergence rates are restatements of exponential decay and Euler stability.

### Problem L2.6: Continuous Compounding with Withdrawals

**Problem Statement:** (a) Show that discrete compounding at annual rate $r$ in $n$ installments converges to the ODE model $B' = rB$ as $n \to \infty$. (b) A retiree holds $B_0 = 500{,}000$ dollars earning $r = 5\%$ per year continuously, withdrawing $w = 30{,}000$ dollars per year continuously: $B' = rB - w$. Find $B(t)$, the equilibrium balance, and when (if ever) the account hits zero.

**Intuition:** Withdrawals add a constant forcing to exponential growth; whether the account survives depends on which side of the equilibrium $w/r$ the balance starts.

**Solution:**

**(a)** After one year of $n$ installments, $B \mapsto B\left(1 + \frac{r}{n}\right)^{n}$. Taking logarithms and using $\ln(1 + u) = u + O(u^2)$ with $u = r/n$:

$$
\left(1 + \frac{r}{n}\right)^{n} = e^{n\ln(1 + r/n)} = e^{r + O(r^2/n)} \longrightarrow e^{r}
$$

so the per-year multiplier tends to $e^r$, exactly the flow of $B' = rB$ over unit time. Continuous compounding *is* the ODE limit of discrete compounding.

**(b)** The equation $B' = rB - w$ is linear with constant coefficients. Equilibrium: $B^* = w/r = 30000/0.05 = 600{,}000$. Writing $u = B - B^*$: $u' = ru$, so

$$
B(t) = B^* + (B_0 - B^*)\,e^{rt} = 600000 - 100000\,e^{0.05 t}
$$

Since $B_0 = 500{,}000 \lt B^*$, the deviation is negative and *grows* exponentially — the equilibrium is unstable (coefficient $+r \gt 0$), so underfunded accounts drain at an accelerating pace.

**Ruin time.** Solve $B(t) = 0$:

$$
e^{0.05t} = 6 \quad\Longrightarrow\quad t = \frac{\ln 6}{0.05} \approx \frac{1.7918}{0.05} \approx 35.8\ \text{years}
$$

$$
\boxed{B(t) = 600000 - 100000\,e^{0.05t}, \qquad B^* = 600{,}000, \qquad \text{ruin at } t = \frac{\ln 6}{0.05} \approx 35.8\ \text{yr}}
$$

Had $B_0$ exceeded $600{,}000$, the same formula would show unbounded growth; the knife-edge $B_0 = w/r$ sustains withdrawals forever.

> **Key Takeaway:** The sign of the linear coefficient decides stability: $B' = rB - w$ has an *unstable* equilibrium at $w/r$, so continuous-withdrawal planning is a race between the starting balance and an exponentially amplifying gap.

### Problem L3.1: Riccati Equation with a Known Particular Solution

**Problem Statement:** Solve completely:

$$
y' = y^2 - 2ty + t^2 + 1
$$

*Hint discovered by inspection:* $y_1(t) = t$ is a solution.

**Intuition:** One particular solution of a Riccati equation linearizes the remaining unknowns: substitute $y = y_1 + 1/v$ and watch the quadratic terms cancel against the $y_1$-equation.

**Solution:**

**Step 1 — Verify the particular solution.** For $y_1 = t$: left side $y_1' = 1$; right side $t^2 - 2t^2 + t^2 + 1 = 1$. Confirmed.

**Step 2 — Recognize the structure.** The right side is a perfect square plus one:

$$
y' = (y - t)^2 + 1
$$

**Step 3 — Substitute $y = t + \dfrac{1}{v}$.** Then $y' = 1 - \dfrac{v'}{v^2}$ and $(y - t)^2 = \dfrac{1}{v^2}$, so the equation becomes

$$
1 - \frac{v'}{v^2} = \frac{1}{v^2} + 1 \quad\Longrightarrow\quad -\frac{v'}{v^2} = \frac{1}{v^2} \quad\Longrightarrow\quad v' = -1
$$

The general Riccati theory ($v' + (b + 2c\,y_1)v = -c$ with $b = -2t$, $c = 1$, $y_1 = t$) predicts exactly this: $b + 2cy_1 = -2t + 2t = 0$, leaving $v' = -1$.

**Step 4 — Solve for $v$ and back-substitute.** $v = C - t$, hence

$$
y(t) = t + \frac{1}{C - t}
$$

**Step 5 — Verify the family.** $y' = 1 + \frac{1}{(C - t)^2}$ and $(y - t)^2 + 1 = \frac{1}{(C-t)^2} + 1$. Equal.

**Step 6 — Solution structure.** As $C \to \infty$ the family degenerates to the particular solution $y = t$ itself; every other solution blows up in finite time at $t = C$ — a movable singularity whose location is set by the initial condition, the hallmark of nonlinear equations.

$$
\boxed{y(t) = t + \frac{1}{C - t}, \quad C \in \mathbb{R}, \quad \text{together with } y = t}
$$

> **Key Takeaway:** A Riccati equation is "one particular solution away" from linear: $y = y_1 + 1/v$ collapses the quadratic nonlinearity, a privilege tied to the Riccati link with second-order linear equations.

### Problem L3.2: Constructing an Integrating Factor $\mu(xy)$

**Problem Statement:** (a) Derive a criterion for the equation $M\,dx + N\,dy = 0$ to admit an integrating factor of the form $\mu = \mu(z)$ with $z = xy$. (b) Apply it to solve

$$
(xy^2 + y)\,dx + (x^2y - x)\,dy = 0 \qquad (x \gt 0,\ y \gt 0)
$$

**Intuition:** Just as $\mu(x)$ requires $(M_y - N_x)/N$ to collapse to a function of $x$, a product-form factor requires a specific quotient to collapse to a function of $xy$.

**Solution:**

**(a) Derivation of the criterion.** Demand exactness of $\mu(z)M\,dx + \mu(z)N\,dy$ with $z = xy$, so $\partial z/\partial y = x$ and $\partial z/\partial x = y$. The condition $(\mu M)_y = (\mu N)_x$ reads, by the chain rule,

$$
\mu'(z)\,x\,M + \mu\,M_y = \mu'(z)\,y\,N + \mu\,N_x
$$

Collecting terms:

$$
\frac{\mu'(z)}{\mu(z)} = \frac{N_x - M_y}{x\,M - y\,N}
$$

**Criterion:** a factor $\mu(xy)$ exists precisely when the right-hand quotient is a function of the product $z = xy$ alone; then $\mu(z) = \exp\left(\int \frac{N_x - M_y}{xM - yN}\,dz\right)$.

**(b) Application.** Here $M = xy^2 + y$, $N = x^2y - x$:

$$
M_y = 2xy + 1, \qquad N_x = 2xy - 1 \quad\Longrightarrow\quad N_x - M_y = -2
$$

and

$$
xM - yN = x^2y^2 + xy - x^2y^2 + xy = 2xy = 2z
$$

The quotient is $\dfrac{-2}{2z} = -\dfrac{1}{z}$ — a function of $z$ alone, so the criterion applies:

$$
\mu(z) = e^{\int -dz/z} = \frac{1}{z} = \frac{1}{xy}
$$

**Step 3 — Multiply and re-check.** The equation becomes

$$
\left(y + \frac{1}{x}\right)dx + \left(x - \frac{1}{y}\right)dy = 0
$$

with $\tilde{M}_y = 1 = \tilde{N}_x$: exact.

**Step 4 — Potential.** $F = \int (y + 1/x)\,dx = xy + \ln x + g(y)$; matching $F_y = x + g'(y) = x - 1/y$ gives $g = -\ln y$. Hence

$$
F(x,y) = xy + \ln\frac{x}{y}
$$

$$
\boxed{xy + \ln\frac{x}{y} = C \qquad \left(\mu = \frac{1}{xy}\right)}
$$

> **Key Takeaway:** Every ansatz $\mu(\omega(x,y))$ produces its own solvability quotient; the method is systematic, and $z = xy$ is simply the next case after $\mu(x)$ and $\mu(y)$ in a single unified scheme.

### Problem L3.3: A Putnam-Style Blow-Up Comparison

**Problem Statement:** Let $y$ solve

$$
y' = y^2 + t^2, \qquad y(0) = 1
$$

Prove that $y$ cannot exist on the whole interval $[0, 1)$. More precisely, show $y(t) \geq \dfrac{1}{1 - t}$ wherever the solution exists in $[0,1)$, and conclude the maximal time of existence $T$ satisfies $T \lt 1$.

**Intuition:** Discard the $t^2$ term to get the *smaller* right side $y^2$, whose solution $\frac{1}{1-t}$ is computable and already blows up at $t = 1$; a comparison argument transfers the blow-up.

**Solution:**

**Step 1 — The comparison candidate.** The clean equation $z' = z^2$, $z(0) = 1$ is separable: $-1/z = t + C$ with $C = -1$, so

$$
z(t) = \frac{1}{1 - t}, \qquad t \in [0, 1)
$$

**Step 2 — Set up the comparison.** Let $[0, T)$ be the maximal interval of $y$ and put $w = y - z$ on $[0, \min(T,1))$. Then $w(0) = 0$ and

$$
w' = y' - z' = y^2 - z^2 + t^2 = (y + z)\,w + t^2
$$

**Step 3 — Grönwall-style positivity.** Define the integrating-factor damped function $\phi(t) = w(t)\,e^{-\int_0^t (y + z)\,ds}$. Then

$$
\phi'(t) = \big(w' - (y + z)w\big)\,e^{-\int_0^t (y+z)\,ds} = t^2\,e^{-\int_0^t (y+z)\,ds} \geq 0
$$

Since $\phi(0) = 0$ and $\phi$ is non-decreasing, $\phi \geq 0$, and the exponential is positive, so

$$
w(t) \geq 0, \quad\text{i.e.}\quad y(t) \geq \frac{1}{1 - t} \quad \text{on } [0, \min(T, 1))
$$

**Step 4 — Conclude blow-up before $t = 1$.** By Step 3, $y(t) \geq \frac{1}{1-t}$ on $[0, \min(T,1))$; since $\frac{1}{1-t} \to \infty$ as $t \to 1^{-}$, the solution cannot remain finite up to time $1$, so the maximal existence time satisfies $T \leq 1$. The inequality is in fact strict: on $[\tfrac{1}{2}, T)$ we have $y' \geq y^2 + \tfrac{1}{4} \gt y^2$ with $y(\tfrac{1}{2}) \geq 2$, and repeating the comparison of Steps 1–3 against $\tilde{z}' = \tilde{z}^2$, $\tilde{z}(\tfrac{1}{2}) = 2$ (which blows up at $t = 1$) with the *strict* extra forcing $\tfrac{1}{4}$ pushes the singularity of $y$ strictly earlier, so $T \lt 1$.

$$
\boxed{y(t) \geq \frac{1}{1 - t} \ \text{on } [0, T), \qquad T \lt 1: \ \text{finite-time blow-up}}
$$

> **Key Takeaway:** Comparison is the analyst's substitute for a closed form: bound the right-hand side below by a solvable one, transfer the bound with a Grönwall/integrating-factor argument, and inherit qualitative conclusions (here, blow-up) for free.

### Problem L3.4: An Implicit Exact Family Through a Given Point

**Problem Statement:** Show that

$$
\left(y\cos x + 2x e^{y}\right) + \left(\sin x + x^2 e^{y} - 1\right)y' = 0
$$

is exact, find the solution passing through the point $(0, 1)$, and prove that near this point the implicit relation defines a unique differentiable function $y(x)$.

**Intuition:** Exactness gives a conserved quantity $F(x, y)$; the initial point selects the level set, and the implicit function theorem certifies it is locally a genuine solution graph.

**Solution:**

**Step 1 — Exactness.** In differential form, $M = y\cos x + 2xe^{y}$, $N = \sin x + x^2 e^{y} - 1$:

$$
M_y = \cos x + 2x e^{y}, \qquad N_x = \cos x + 2x e^{y}
$$

Equal on $\mathbb{R}^2$ (simply connected): exact.

**Step 2 — Potential.** Integrate $M$ in $x$:

$$
F(x,y) = \int \left(y\cos x + 2xe^{y}\right)dx = y\sin x + x^2 e^{y} + g(y)
$$

Match $F_y = N$:

$$
\sin x + x^2 e^{y} + g'(y) = \sin x + x^2 e^{y} - 1 \quad\Longrightarrow\quad g'(y) = -1 \quad\Longrightarrow\quad g(y) = -y
$$

So $F(x, y) = y\sin x + x^2 e^{y} - y$.

**Step 3 — Select the level set through $(0,1)$.**

$$
F(0, 1) = 1\cdot 0 + 0 - 1 = -1
$$

giving the particular implicit solution

$$
y\sin x + x^2 e^{y} - y = -1
$$

**Step 4 — Local solvability.** The implicit function theorem applies at $(0,1)$ provided $F_y(0,1) \neq 0$:

$$
F_y = \sin x + x^2 e^{y} - 1, \qquad F_y(0, 1) = 0 + 0 - 1 = -1 \neq 0
$$

Hence there is a neighborhood of $x = 0$ and a unique $C^1$ function $y(x)$ with $y(0) = 1$ satisfying the relation; differentiating $F(x, y(x)) = -1$ recovers the ODE, so this $y(x)$ is the unique local solution of the IVP. Its slope at the initial point is

$$
y'(0) = -\frac{F_x(0,1)}{F_y(0,1)} = -\frac{1\cdot\cos 0 + 2\cdot 0\cdot e}{-1} = 1
$$

$$
\boxed{y\sin x + x^2 e^{y} - y = -1, \qquad \text{a unique local solution with } y(0) = 1,\ y'(0) = 1}
$$

> **Key Takeaway:** For exact equations the honest final answer is a level set; the implicit function theorem is the bridge that converts "curve through the point with $F_y \neq 0$" into a bona fide unique solution $y(x)$ of the IVP.
