# Topic 07: Boundary Value Problems and PDE Preview

## 1. First-Principles Intuition & Motivation

**The phenomenon.** Every problem so far has been an *initial value problem* (IVP): the state and enough derivatives are given at a single instant $t_0$, and Picard–Lindelöf guarantees a unique solution marching forward. But many physical questions are not phrased that way:

- A metal rod is held at $100^\circ$C at one end and $0^\circ$C at the other. What is the steady temperature profile *between* the ends?
- A guitar string is clamped at $x = 0$ and $x = L$. Which vibration shapes are possible?
- A quantum particle is confined to a box. Which energies can it have?

In each case the data is split between **two spatial endpoints**. The unknown is a function $y(x)$ on an interval $[a, b]$, and instead of $y(a), y'(a)$ we are given one condition at $a$ and one at $b$. This is a **boundary value problem (BVP)** — and it behaves *completely differently* from an IVP.

**The shock.** Consider $y'' + y = 0$, whose general solution is $y = c_1 \cos x + c_2 \sin x$.

- With $y(0) = 0$ and $y(\pi) = 1$: the first condition forces $c_1 = 0$, but then $y(\pi) = c_2 \sin \pi = 0 \neq 1$. **No solution exists.**
- With $y(0) = 0$ and $y(\pi) = 0$: both conditions are satisfied by $y = c_2 \sin x$ for *every* $c_2$. **Infinitely many solutions exist.**

The same ODE, with equally innocent-looking data, has either zero or infinitely many solutions. Existence and uniqueness — automatic for IVPs — become the *central question* for BVPs.

### The goal and the strategy

**Goal.** Build a complete theory answering three questions:

1. **Solvability** — when does a BVP have a solution, and when is it unique?
2. **Spectrum** — for which parameter values $\lambda$ does the homogeneous problem $y'' + \lambda y = 0$ (and its generalizations) have nontrivial solutions, and what do those solutions look like?
3. **Synthesis** — how do we assemble the special solutions into a solution for *arbitrary* data, and how does this solve time-dependent PDEs?

**Strategy.** The failure mode above is not a bug; it is the fingerprint of an **eigenvalue problem**. The values of $\lambda$ for which the homogeneous BVP has nontrivial solutions are the *eigenvalues* of the differential operator, and the nontrivial solutions are its *eigenfunctions*. Sturm–Liouville theory shows these eigenfunctions form an orthogonal basis, so any reasonable function can be expanded in them — the generalized Fourier series. That expansion is exactly the tool that solves the heat and wave equations by separation of variables.

**Assumptions in force.** Throughout: coefficients are continuous on a closed bounded interval $[a, b]$; $p(x) \gt 0$ and $w(x) \gt 0$ on $[a, b]$ (regular problems); solutions are twice continuously differentiable. Singular problems (Bessel, Legendre) relax these at the endpoints and are flagged when mentioned.

### Variables and notation

| Symbol | Meaning |
| :--- | :--- |
| $y(x)$ | unknown function of the spatial variable $x \in [a, b]$ |
| $\lambda$ | spectral parameter; eigenvalue when the BVP has a nontrivial solution |
| $\varphi_n(x)$ | eigenfunction associated with the eigenvalue $\lambda_n$ |
| $p(x), q(x), w(x)$ | Sturm–Liouville coefficients: stiffness, potential, weight |
| $L[y]$ | the differential operator $(p y')' + q y$ |
| $\langle f, g \rangle_w$ | weighted inner product $\int_a^b f(x) g(x) w(x)\, dx$ |
| $G(x, \xi)$ | Green's function: kernel of the inverse operator |
| $u(x, t)$ | PDE unknown, e.g. temperature or displacement field |
| $k$ | thermal diffusivity in the heat equation $u_t = k u_{xx}$ |
| $c$ | wave speed in the wave equation $u_{tt} = c^2 u_{xx}$ |

The weighted inner product turns the function space into a geometry: eigenfunctions will be *orthogonal vectors* in this geometry, and Fourier coefficients will be *orthogonal projections* — exactly as in the spectral theorem for symmetric matrices.

## 2. Rigorous Mathematical Definitions & Theorem Statements

**Definition 2.1 (Two-point boundary value problem).** A second-order two-point BVP on $[a, b]$ consists of an ODE

$$
y'' = f(x, y, y'), \qquad a \le x \le b,
$$

together with two boundary conditions, one involving values at $a$ and one at $b$. The four classical types of (homogeneous, separated) boundary conditions at an endpoint are:

| Type | Condition | Physical meaning (heat conduction) |
| :--- | :--- | :--- |
| **Dirichlet** | $y(a) = 0$ | endpoint temperature is prescribed |
| **Neumann** | $y'(a) = 0$ | endpoint heat flux is prescribed (insulated if zero) |
| **Robin** | $\alpha y(a) + \beta y'(a) = 0$ | convective exchange: flux proportional to temperature |
| **Periodic** | $y(a) = y(b),\; y'(a) = y'(b)$ | the interval closes into a ring |

The first three are *separated* (each condition touches one endpoint); periodic conditions are *coupled*. Unlike IVPs, no general existence–uniqueness theorem holds: solvability depends on the interplay between operator, interval length, and boundary conditions.

**Definition 2.2 (Eigenvalue problem).** Given a linear operator with boundary conditions, the values $\lambda$ for which

$$
y'' + \lambda y = 0, \qquad \text{(BCs)}
$$

admits a solution $y \not\equiv 0$ are the **eigenvalues**; the nontrivial solutions are **eigenfunctions**. Eigenfunctions are determined only up to a nonzero scalar multiple.

### The three canonical spectra

Solving $y'' + \lambda y = 0$ on $[0, L]$ under each classical boundary condition gives (full derivation in Section 3):

| Boundary conditions | Eigenvalues | Eigenfunctions |
| :--- | :--- | :--- |
| Dirichlet: $y(0) = y(L) = 0$ | $\lambda_n = \left(\frac{n\pi}{L}\right)^2$, $n = 1, 2, \dots$ | $\sin\frac{n\pi x}{L}$ |
| Neumann: $y'(0) = y'(L) = 0$ | $\lambda_n = \left(\frac{n\pi}{L}\right)^2$, $n = 0, 1, 2, \dots$ | $\cos\frac{n\pi x}{L}$ (constant for $n = 0$) |
| Periodic: $y(-L) = y(L)$, $y'(-L) = y'(L)$ | $\lambda_n = \left(\frac{n\pi}{L}\right)^2$, $n = 0, 1, 2, \dots$ | $1;\ \cos\frac{n\pi x}{L},\ \sin\frac{n\pi x}{L}$ |

Three structural observations, each of which will be proved in general:

1. The spectrum is an increasing sequence of **real** numbers tending to $+\infty$ — no accumulation, no complex values.
2. In the separated cases each eigenvalue is **simple** (one eigenfunction up to scaling); in the periodic case the positive eigenvalues are **double** (both $\cos$ and $\sin$) — coupled boundary conditions are exactly what permits multiplicity 2.
3. The $n$-th eigenfunction has exactly $n - 1$ interior zeros (Sturm oscillation) — higher modes oscillate faster.

### Sturm–Liouville theory

**Definition 2.3 (Sturm–Liouville form).** A **Sturm–Liouville (SL) equation** on $[a, b]$ is

$$
\big(p(x)\, y'\big)' + q(x)\, y + \lambda\, w(x)\, y = 0,
$$

with $p \in C^1[a,b]$, $q, w \in C[a,b]$. The problem is **regular** if $p(x) \gt 0$ and $w(x) \gt 0$ on all of $[a, b]$ and the boundary conditions are separated:

$$
\alpha_1 y(a) + \alpha_2 y'(a) = 0, \qquad \beta_1 y(b) + \beta_2 y'(b) = 0,
$$

with $(\alpha_1, \alpha_2) \neq (0, 0)$ and $(\beta_1, \beta_2) \neq (0, 0)$. Any equation $a_2 y'' + a_1 y' + a_0 y + \lambda y = 0$ with $a_2 \neq 0$ can be brought to SL form by multiplying by the integrating factor $\mu(x) = \exp\!\big(\int a_1/a_2\, dx\big)/a_2$.

**Lemma 2.4 (Lagrange identity).** Write $L[y] = (p y')' + q y$. For any $u, v \in C^2[a, b]$:

$$
u\, L[v] - v\, L[u] = \frac{d}{dx}\Big[ p \big( u v' - v u' \big) \Big].
$$

Integrating gives **Green's formula** $\int_a^b (u L[v] - v L[u])\, dx = \big[ p (u v' - v u') \big]_a^b$. When $u, v$ both satisfy regular separated (or periodic) boundary conditions, the right side vanishes: $L$ is **self-adjoint** on this domain, $\langle u, L[v] \rangle = \langle L[u], v \rangle$.

**Theorem 2.5 (Regular Sturm–Liouville theorem).** A regular SL problem has:

1. infinitely many eigenvalues, all **real**, all **simple**, forming an increasing sequence $\lambda_1 \lt \lambda_2 \lt \lambda_3 \lt \cdots$ with $\lambda_n \to \infty$;
2. eigenfunctions $\varphi_n$ that are **orthogonal with weight $w$**: $\langle \varphi_m, \varphi_n \rangle_w = 0$ for $m \neq n$;
3. **completeness**: the $\varphi_n$ form a basis — every $f$ with $\int_a^b f^2 w\, dx \lt \infty$ has a generalized Fourier expansion converging to $f$ in the mean-square sense (and pointwise to $\tfrac{1}{2}[f(x^+) + f(x^-)]$ if $f$ is piecewise smooth).

This is the infinite-dimensional spectral theorem: $L$ plays the symmetric matrix, $w$ the inner product, $\varphi_n$ the orthogonal eigenvectors.

### Generalized Fourier series and Green's functions

**Definition 2.6 (Generalized Fourier coefficients).** For a regular SL problem with eigenfunctions $\{\varphi_n\}$ and a function $f$, the expansion $f(x) \sim \sum_{n} c_n \varphi_n(x)$ has coefficients obtained by orthogonal projection:

$$
c_n = \frac{\langle f, \varphi_n \rangle_w}{\lVert \varphi_n \rVert_w^2} = \frac{\int_a^b f(x)\, \varphi_n(x)\, w(x)\, dx}{\int_a^b \varphi_n(x)^2\, w(x)\, dx}.
$$

For the Dirichlet problem on $[0, L]$ this is the classical **Fourier sine series** with $c_n = \frac{2}{L} \int_0^L f(x) \sin\frac{n\pi x}{L}\, dx$, since $\lVert \sin(n\pi x/L) \rVert^2 = L/2$.

**Definition 2.7 (Green's function).** For the BVP $-y'' = f(x)$, $y(0) = y(L) = 0$, the **Green's function** is the kernel $G(x, \xi)$ such that

$$
y(x) = \int_0^L G(x, \xi)\, f(\xi)\, d\xi
$$

solves the BVP for every continuous $f$. It will be constructed explicitly in Section 3: a symmetric, piecewise-linear "tent" kernel. Conceptually, $G$ is the continuous analogue of a matrix inverse: solving the BVP is applying the integral operator with kernel $G$, i.e. $y = L^{-1} f$.

**Definition 2.8 (Second-order PDE classification).** For $A u_{xx} + 2B u_{xt} + C u_{tt} + \cdots = 0$, the sign of the discriminant $B^2 - AC$ classifies the equation: **elliptic** (negative; Laplace $u_{xx} + u_{yy} = 0$, equilibrium, pure boundary data), **parabolic** (zero; heat $u_t = k u_{xx}$, diffusion, initial + boundary data), **hyperbolic** (positive; wave $u_{tt} = c^2 u_{xx}$, propagation at finite speed, two initial conditions + boundary data). Classification dictates which data makes the problem well-posed.

## 3. Step-by-Step Mathematical Proofs & Derivations

### Proof 3.1 — Complete solution of the Dirichlet eigenproblem

**Claim.** The problem $y'' + \lambda y = 0$, $y(0) = y(L) = 0$ has nontrivial solutions exactly for $\lambda_n = (n\pi/L)^2$, $n \ge 1$, with eigenfunctions $\varphi_n(x) = \sin(n\pi x/L)$.

**Step 1 (Case $\lambda \lt 0$).** Write $\lambda = -\mu^2$ with $\mu \gt 0$. The characteristic roots are $\pm\mu$, so $y = A \cosh \mu x + B \sinh \mu x$. Then $y(0) = A = 0$, and $y(L) = B \sinh \mu L = 0$. Since $\mu L \gt 0$ and $\sinh$ vanishes only at $0$, we get $B = 0$. Only the trivial solution: **no negative eigenvalues**.

**Step 2 (Case $\lambda = 0$).** The equation is $y'' = 0$, so $y = Ax + B$. Then $y(0) = B = 0$ and $y(L) = AL = 0$ gives $A = 0$. **Zero is not an eigenvalue.**

**Step 3 (Case $\lambda \gt 0$).** Write $\lambda = \mu^2$ with $\mu \gt 0$. Then $y = A \cos \mu x + B \sin \mu x$. The condition $y(0) = 0$ forces $A = 0$, leaving $y = B \sin \mu x$.

**Step 4 (Quantization).** For a nontrivial solution we need $B \neq 0$, so $y(L) = B \sin \mu L = 0$ requires $\sin \mu L = 0$, i.e. $\mu L = n \pi$ for a positive integer $n$. Hence $\mu_n = n\pi/L$.

**Step 5 (Conclusion).** The eigenvalues and (up to scaling) eigenfunctions are

$$
\boxed{\ \lambda_n = \left( \frac{n\pi}{L} \right)^2, \qquad \varphi_n(x) = \sin \frac{n \pi x}{L}, \qquad n = 1, 2, 3, \dots\ }
$$

The exclusion of $\lambda \le 0$ in Steps 1–2 is not pedantry: it is *why* the spectrum is a discrete ladder climbing to infinity, with a smallest positive rung $\lambda_1 = (\pi/L)^2$ — the fundamental mode.

### Proof 3.2 — Orthogonality of Sturm–Liouville eigenfunctions

**Claim.** If $\varphi_m, \varphi_n$ are eigenfunctions of a regular SL problem with distinct eigenvalues $\lambda_m \neq \lambda_n$, then $\int_a^b \varphi_m \varphi_n\, w\, dx = 0$.

**Step 1 (Lagrange identity).** With $L[y] = (p y')' + q y$, expand the right-hand side derivative:

$$
\frac{d}{dx}\Big[ p ( \varphi_m \varphi_n' - \varphi_n \varphi_m' ) \Big] = \varphi_m (p \varphi_n')' - \varphi_n (p \varphi_m')' = \varphi_m L[\varphi_n] - \varphi_n L[\varphi_m],
$$

since the $q \varphi_m \varphi_n$ terms cancel and the cross terms $p \varphi_m' \varphi_n'$ cancel in the product-rule expansion.

**Step 2 (Use the eigenvalue equations).** By assumption $L[\varphi_n] = -\lambda_n w \varphi_n$ and $L[\varphi_m] = -\lambda_m w \varphi_m$. Substituting:

$$
\varphi_m L[\varphi_n] - \varphi_n L[\varphi_m] = (\lambda_m - \lambda_n)\, w\, \varphi_m \varphi_n.
$$

**Step 3 (Integrate over the interval).**

$$
(\lambda_m - \lambda_n) \int_a^b \varphi_m \varphi_n\, w\, dx = \Big[ p \big( \varphi_m \varphi_n' - \varphi_n \varphi_m' \big) \Big]_a^b.
$$

**Step 4 (Boundary terms vanish).** At $x = a$ both eigenfunctions satisfy $\alpha_1 y(a) + \alpha_2 y'(a) = 0$ with $(\alpha_1, \alpha_2) \neq (0,0)$. The vectors $(\varphi_m(a), \varphi_m'(a))$ and $(\varphi_n(a), \varphi_n'(a))$ both lie in the one-dimensional null space of the row vector $(\alpha_1, \alpha_2)$, hence are parallel, so the determinant $\varphi_m(a) \varphi_n'(a) - \varphi_n(a) \varphi_m'(a) = 0$. The same argument applies at $x = b$. (For periodic conditions the two endpoint contributions cancel instead.)

**Step 5 (Conclude).** Since $\lambda_m \neq \lambda_n$, division gives

$$
\boxed{\ \int_a^b \varphi_m(x)\, \varphi_n(x)\, w(x)\, dx = 0 \quad (m \neq n).\ }
$$

No trigonometric identities were used — orthogonality is forced by self-adjointness plus the boundary conditions alone.

### Proof 3.3 — Reality of Sturm–Liouville eigenvalues

**Claim.** Every eigenvalue of a regular SL problem (real coefficients $p, q, w$, real boundary constants) is real.

**Step 1 (Set up a complex eigenpair).** Suppose $\lambda \in \mathbb{C}$ is an eigenvalue with eigenfunction $\varphi$ (possibly complex-valued): $L[\varphi] + \lambda w \varphi = 0$ with the boundary conditions.

**Step 2 (Conjugate the problem).** Taking complex conjugates of the equation and boundary conditions — all coefficients being real — shows $\overline{\varphi}$ is an eigenfunction with eigenvalue $\overline{\lambda}$.

**Step 3 (Apply the Green's formula computation of Proof 3.2)** to the pair $u = \varphi$, $v = \overline{\varphi}$:

$$
(\lambda - \overline{\lambda}) \int_a^b \varphi\, \overline{\varphi}\, w\, dx = \Big[ p \big( \varphi\, \overline{\varphi}' - \overline{\varphi}\, \varphi' \big) \Big]_a^b = 0,
$$

the boundary terms vanishing by the same parallel-vector argument as before.

**Step 4 (The integral is positive).** Since $\varphi \overline{\varphi} = \lvert \varphi \rvert^2 \ge 0$, $w \gt 0$, and $\varphi \not\equiv 0$ is continuous, we have $\int_a^b \lvert \varphi \rvert^2 w\, dx \gt 0$.

**Step 5 (Conclude).** A nonzero factor multiplied by $(\lambda - \overline{\lambda})$ equals zero, hence $\lambda = \overline{\lambda}$:

$$
\boxed{\ \lambda \in \mathbb{R} \text{ for every eigenvalue of a regular Sturm–Liouville problem.}\ }
$$

Moreover, taking real and imaginary parts of $\varphi$ shows the eigenfunction can always be chosen real. This is the exact analogue of "symmetric matrices have real eigenvalues", with the Lagrange identity replacing $x^{\top} A y = y^{\top} A x$.

### Proof 3.4 — Green's function for the Dirichlet problem: construction

**Claim.** The BVP $-y'' = f(x)$, $y(0) = y(L) = 0$ is solved by $y(x) = \int_0^L G(x, \xi) f(\xi)\, d\xi$ with

$$
G(x, \xi) = \begin{cases} \dfrac{x (L - \xi)}{L}, & 0 \le x \le \xi, \\[2mm] \dfrac{\xi (L - x)}{L}, & \xi \le x \le L. \end{cases}
$$

**Step 1 (Ansatz from homogeneous solutions).** For fixed $\xi$, away from $x = \xi$ the kernel must satisfy $-G_{xx} = 0$, so it is linear in $x$ on each side. Let $u_1(x) = x$ (satisfies the left BC $u_1(0) = 0$) and $u_2(x) = L - x$ (satisfies the right BC $u_2(L) = 0$). Set $G(x,\xi) = C(\xi)\, u_1(x) u_2(\xi)$ for $x \le \xi$ and $G(x,\xi) = C(\xi)\, u_1(\xi) u_2(x)$ for $x \ge \xi$ — this form is automatically continuous at $x = \xi$ and satisfies both boundary conditions.

**Step 2 (Jump condition).** Integrating $-G_{xx} = \delta(x - \xi)$ across $x = \xi$ requires the derivative jump

$$
G_x(\xi^+, \xi) - G_x(\xi^-, \xi) = -1.
$$

Computing: $G_x(\xi^+, \xi) = -C u_1(\xi) = -C\xi$ and $G_x(\xi^-, \xi) = C u_2(\xi) = C(L - \xi)$, so the jump is $-C\xi - C(L-\xi) = -CL$. Setting $-CL = -1$ gives $C = 1/L$.

**Step 3 (Assemble).** Substituting $C = 1/L$ yields exactly the claimed tent-shaped kernel. Note the **symmetry** $G(x, \xi) = G(\xi, x)$ — the continuous counterpart of the fact that the inverse of a symmetric matrix is symmetric, and the reason the operator $f \mapsto \int G f$ is self-adjoint.

**Step 4 (Wronskian view).** In general, for $-(p y')' = f$ with separated BCs, $G(x,\xi) = u_1(x_{\lt}) u_2(x_{\gt}) / \big(p(\xi) W(\xi)\big)$ where $x_{\lt} = \min(x,\xi)$, $x_{\gt} = \max(x,\xi)$, and $p W$ is constant (Abel). Here $p = 1$ and $W = u_1 u_2' - u_2 u_1' = x(-1) - (L - x)(1) = -L$; the sign is absorbed in the orientation convention, reproducing $C = 1/L$.

### Proof 3.4 (continued) — Verification that the integral formula solves the BVP

**Step 5 (Split the integral).** Define

$$
y(x) = \int_0^L G(x,\xi) f(\xi)\, d\xi = \frac{L - x}{L} \int_0^x \xi\, f(\xi)\, d\xi + \frac{x}{L} \int_x^L (L - \xi) f(\xi)\, d\xi,
$$

using the $x \ge \xi$ branch on $[0, x]$ and the $x \le \xi$ branch on $[x, L]$.

**Step 6 (First derivative, Leibniz rule).** Differentiating each term (the boundary terms at $\xi = x$ cancel because the two branches agree there):

$$
y'(x) = -\frac{1}{L} \int_0^x \xi f(\xi)\, d\xi + \frac{1}{L} \int_x^L (L - \xi) f(\xi)\, d\xi.
$$

**Step 7 (Second derivative).** Differentiating again:

$$
y''(x) = -\frac{1}{L} x f(x) - \frac{1}{L} (L - x) f(x) = -f(x),
$$

so $-y'' = f$ as required.

**Step 8 (Boundary conditions).** At $x = 0$ the first term of Step 5 has factor $\int_0^0 = 0$ and the second has factor $x = 0$; hence $y(0) = 0$. Symmetrically $y(L) = 0$.

$$
\boxed{\ y(x) = \int_0^L G(x,\xi)\, f(\xi)\, d\xi \ \text{ solves } -y'' = f,\ y(0) = y(L) = 0 \ \text{ for every continuous } f.\ }
$$

**Interpretation.** The differential operator $-d^2/dx^2$ with Dirichlet conditions has been *inverted*: its inverse is the integral operator with kernel $G$. Discretize and $G$ becomes literally the inverse matrix of the tridiagonal second-difference matrix. Uniqueness follows since the homogeneous problem ($f = 0$) has only the trivial solution by Proof 3.1 (zero is not an eigenvalue).

### Proof 3.5 — Separation of variables for the heat equation

**Claim.** The initial-boundary value problem

$$
u_t = k u_{xx} \quad (0 \lt x \lt L,\ t \gt 0), \qquad u(0,t) = u(L,t) = 0, \qquad u(x,0) = f(x)
$$

is solved by $u(x,t) = \sum_{n=1}^{\infty} b_n\, e^{-k (n\pi/L)^2 t} \sin\frac{n\pi x}{L}$ with $b_n$ the Fourier sine coefficients of $f$.

**Step 1 (Product ansatz).** Seek nontrivial solutions of the PDE + boundary conditions of the form $u(x,t) = X(x) T(t)$. Substituting: $X T' = k X'' T$, and dividing by $k X T$ (nonzero where the solution is nontrivial):

$$
\frac{T'(t)}{k\, T(t)} = \frac{X''(x)}{X(x)}.
$$

**Step 2 (Separation constant).** The left side depends only on $t$, the right only on $x$; a function of $t$ alone equals a function of $x$ alone only if both are the same constant, call it $-\lambda$. This yields two ODEs:

$$
X'' + \lambda X = 0, \qquad T' = -\lambda k\, T.
$$

**Step 3 (Boundary conditions transfer to $X$).** From $u(0,t) = X(0) T(t) = 0$ for all $t$ and $T \not\equiv 0$, we get $X(0) = 0$; likewise $X(L) = 0$. So $X$ solves precisely the Dirichlet eigenproblem of Proof 3.1: $\lambda_n = (n\pi/L)^2$, $X_n = \sin(n\pi x/L)$.

**Step 4 (Time factor).** For each $n$, $T_n' = -k \lambda_n T_n$ gives $T_n(t) = e^{-k (n\pi/L)^2 t}$. Each **mode** $u_n = e^{-k(n\pi/L)^2 t} \sin(n\pi x/L)$ solves the PDE with the boundary conditions; higher modes decay *much* faster (rate proportional to $n^2$).

**Step 5 (Superposition).** The PDE and BCs are linear and homogeneous, so any (suitably convergent) series $u = \sum_n b_n u_n$ is again a solution. The exponential factors make the series and its derivatives converge uniformly for $t \ge t_0 \gt 0$ whenever the $b_n$ are bounded, justifying term-by-term differentiation.

**Step 6 (Match the initial condition).** Setting $t = 0$: we need $\sum_n b_n \sin(n\pi x/L) = f(x)$. By completeness (Theorem 2.5) and orthogonality with $\lVert \sin(n\pi x/L)\rVert^2 = L/2$:

$$
\boxed{\ u(x,t) = \sum_{n=1}^{\infty} b_n\, e^{-k \left(\frac{n\pi}{L}\right)^2 t} \sin \frac{n\pi x}{L}, \qquad b_n = \frac{2}{L} \int_0^L f(x) \sin \frac{n\pi x}{L}\, dx.\ }
$$

The recipe is fully modular: *eigenproblem in space* $\to$ *decay rates in time* $\to$ *Fourier coefficients from initial data*. Changing to Neumann conditions only swaps in the cosine eigenfunctions.

### Proof 3.6 — Uniqueness for the heat equation by the energy method

**Claim.** The initial-boundary value problem of Proof 3.5 has at most one (classical, $C^2$) solution.

**Step 1 (Difference of two solutions).** Let $u_1, u_2$ both solve the problem and set $v = u_1 - u_2$. By linearity, $v$ satisfies

$$
v_t = k v_{xx}, \qquad v(0,t) = v(L,t) = 0, \qquad v(x,0) = 0.
$$

**Step 2 (Define the energy).** Let

$$
E(t) = \int_0^L v(x,t)^2\, dx \ \ge 0, \qquad E(0) = 0.
$$

**Step 3 (Differentiate under the integral).**

$$
E'(t) = 2 \int_0^L v\, v_t\, dx = 2k \int_0^L v\, v_{xx}\, dx.
$$

**Step 4 (Integrate by parts).**

$$
\int_0^L v\, v_{xx}\, dx = \Big[ v\, v_x \Big]_0^L - \int_0^L v_x^2\, dx = -\int_0^L v_x^2\, dx,
$$

the boundary term vanishing because $v = 0$ at both ends. Hence $E'(t) = -2k \int_0^L v_x^2\, dx \le 0$.

**Step 5 (Squeeze).** $E$ is nonnegative, starts at $0$, and is non-increasing; therefore $E(t) = 0$ for all $t \ge 0$. A continuous nonnegative function with zero integral vanishes identically, so $v(x,t) = 0$ everywhere:

$$
\boxed{\ u_1 \equiv u_2 \ \text{— the heat equation with Dirichlet data has at most one classical solution.}\ }
$$

**Remark.** The same computation with Neumann or Robin conditions (with nonnegative Robin constant) also kills the boundary term, and the identical argument gives uniqueness for the wave equation using $E(t) = \int (v_t^2 + c^2 v_x^2)\, dx$. Energy methods are the workhorse of PDE uniqueness theory precisely because they need no solution formula — only integration by parts.

## 4. Computational & Algorithmic Insights

### Finite differences: the BVP as a tridiagonal linear system

Discretize $[0, L]$ with $x_i = ih$, $h = L/(n+1)$, unknowns $y_1, \dots, y_n$ at interior nodes. The second-order central difference

$$
y''(x_i) = \frac{y_{i-1} - 2 y_i + y_{i+1}}{h^2} + O(h^2)
$$

turns $-y'' = f$, $y(0) = y(L) = 0$ into the linear system $A \mathbf{y} = h^2 \mathbf{f}$ with the tridiagonal matrix $A = \mathrm{tridiag}(-1, 2, -1)$. Because $A$ is tridiagonal, symmetric, and diagonally dominant, the **Thomas algorithm** (tridiagonal Gaussian elimination) solves it stably in $O(n)$ time and memory — versus $O(n^3)$ for a dense solver:

```text
# Thomas algorithm for a x[i-1] + b x[i] + c x[i+1] = d[i]
for i = 2..n:                       # forward sweep
    m    = a[i] / b[i-1]
    b[i] = b[i] - m * c[i-1]
    d[i] = d[i] - m * d[i-1]
x[n] = d[n] / b[n]
for i = n-1..1:                     # back substitution
    x[i] = (d[i] - c[i] * x[i+1]) / b[i]
```

**Convergence.** The global error is $O(h^2)$: halving $h$ quarters the error. Neumann conditions need a ghost-node or one-sided $O(h^2)$ stencil at the boundary, otherwise the whole scheme degrades to $O(h)$ — a classic silent bug.

### The shooting method: BVPs by root-finding on the missing slope

For $y'' = g(x, y, y')$ with $y(0) = \alpha$, $y(L) = \beta$, the IVP machinery is missing only one number: the initial slope. Define

$$
F(s) = y(L; s) - \beta, \qquad \text{where } y(\cdot\,; s) \text{ solves the IVP } y(0) = \alpha,\ y'(0) = s.
$$

A root of $F$ is a solution of the BVP. Algorithm:

```text
choose bracket [s_lo, s_hi] with F(s_lo) * F(s_hi) < 0
repeat:
    s_mid = (s_lo + s_hi) / 2          # or secant / Newton step
    integrate IVP with slope s_mid      # e.g. RK4
    update bracket using sign of F(s_mid)
until |F(s_mid)| < tol
```

- **Linear BVPs**: $F$ is affine in $s$, so *two* IVP integrations plus one linear interpolation give the exact answer (up to integrator error).
- **Newton's method** on $F$ uses the variational equation: differentiate the ODE with respect to $s$ and integrate it alongside to get $F'(s)$ for free.
- **Failure modes**: for stiff or unstable ODEs, $y(L; s)$ can depend exponentially sensitively on $s$ (remedy: *multiple shooting*, subdividing the interval); nonlinear BVPs can have several roots — Bratu's problem $y'' + \lambda e^y = 0$ has **two** solutions for small $\lambda \gt 0$, and a naive bracket finds only one of them.

### Spectral view: the discrete Laplacian converges to the continuous one

The matrix $A = \mathrm{tridiag}(-1, 2, -1)$ of size $n$ has *exactly known* eigenstructure:

$$
A v^{(j)} = \mu_j v^{(j)}, \qquad \mu_j = 2 - 2\cos\frac{j\pi}{n+1}, \qquad v^{(j)}_i = \sin\frac{i j \pi}{n+1},
$$

i.e. the eigenvectors are the Dirichlet eigenfunctions *sampled on the grid*. Scaling by $1/h^2$ with $h = L/(n+1)$ and Taylor-expanding $2 - 2\cos\theta = \theta^2 - \theta^4/12 + O(\theta^6)$:

$$
\frac{\mu_j}{h^2} = \left(\frac{j\pi}{L}\right)^2 - \frac{1}{12}\, \frac{j^4 \pi^4}{L^4}\, h^2 + O(h^4) \ \longrightarrow\ \lambda_j = \left(\frac{j\pi}{L}\right)^2.
$$

Low modes (small $j$) converge at rate $O(h^2)$; high modes (with $j$ comparable to $n$) are badly wrong — a discrete grid cannot represent oscillations near or below the grid scale. Rule of thumb: trust roughly the lowest $n/2$ computed eigenvalues (and fewer for eigenfunctions).

**Spectral methods preview.** Instead of local stencils, expand the unknown in the *global* eigenbasis (sine series here, Chebyshev polynomials generally) and impose the equation on the coefficients. For smooth solutions the error decays faster than any power of $1/n$ ("spectral accuracy"), because the machinery of Section 3 diagonalizes the operator exactly — this is also the mathematical seed of the Fourier Neural Operator in Section 5.

## 5. Real-World Physics & AI/ML Applications

### Physics: one eigenproblem, four costumes

**Heated rod and heat sinks.** The steady state of $u_t = k u_{xx}$ with fixed end temperatures is the two-point BVP $u_{xx} = 0$; with internal heating $s(x)$ it becomes $-K u'' = s$, solved instantly by the Green's function of Proof 3.4. Robin conditions $-K u'(L) = h\,(u(L) - T_{\text{env}})$ model convective cooling — the design equation of every heat sink.

**Vibrating string and music.** The wave equation $u_{tt} = c^2 u_{xx}$ with clamped ends, $c = \sqrt{T/\mu}$ (tension over linear density), has standing-wave solutions $\sin(n\pi x/L)\cos(n\pi c t/L)$ with frequencies $f_n = n c / (2L)$ — an exact integer harmonic series. This is why a plucked string sounds *pitched*: its overtones $f_2 = 2f_1, f_3 = 3f_1, \dots$ reinforce a single perceived note, and why fretting (changing $L$) or tuning (changing $T$) moves the pitch. d'Alembert's formula $u = \tfrac{1}{2}[g(x - ct) + g(x + ct)] + \tfrac{1}{2c}\int_{x - ct}^{x + ct} v$ describes the same motion as counter-propagating traveling waves.

**Quantum particle in a box.** The time-independent Schrödinger equation $-\frac{\hbar^2}{2m} \psi'' = E \psi$ with $\psi(0) = \psi(L) = 0$ *is* the Dirichlet eigenproblem with $\lambda = 2mE/\hbar^2$. Proof 3.1 therefore delivers quantized energies

$$
E_n = \frac{n^2 \pi^2 \hbar^2}{2 m L^2}, \qquad n = 1, 2, 3, \dots
$$

Energy quantization is not mystical — it is Step 4 of Proof 3.1. Orthogonality of the $\psi_n$ underlies the probabilistic interpretation of expansion coefficients.

**Beam deflection.** The Euler–Bernoulli beam $E I\, y'''' = q(x)$ is a *fourth-order* BVP needing two conditions per end (clamped: $y = y' = 0$; simply supported: $y = y'' = 0$; free: $y'' = y''' = 0$). Sturm–Liouville-type theory extends, with buckling loads as eigenvalues.

### Machine learning I: eigenfunctions as features

**Fourier features and positional encodings.** The transformer positional encoding $\big(\sin(\omega_k t), \cos(\omega_k t)\big)_k$ and the random Fourier features of kernel methods are precisely eigenfunctions of the Laplacian (periodic SL problem of Section 2). They inherit the SL virtues: orthogonality (non-redundant coordinates), a frequency ladder (multi-scale resolution), and completeness (any position-dependent signal is representable). Work on neural network *spectral bias* shows models learn low-eigenvalue components first — the exact analogue of low heat-equation modes decaying slowest.

**Graph Laplacians: discrete Sturm–Liouville theory.** For a graph with adjacency $W$ and degree matrix $D$, the Laplacian $\mathcal{L} = D - W$ is symmetric positive semidefinite; on a path graph it is *exactly* the tridiagonal matrix of Section 4. Its eigenvectors are the graph's "vibration modes":

- **Laplacian eigenmaps / spectral clustering** embed data using the lowest nontrivial eigenvectors — the discrete fundamental modes, which vary slowest across edges;
- **spectral GNNs** (e.g. ChebNet, GCN) define graph convolution as a function of $\mathcal{L}$ applied in its eigenbasis, i.e. a learned multiplier on the graph's Sturm–Liouville spectrum;
- the **Fiedler value** $\lambda_2$ (smallest nontrivial eigenvalue) measures connectivity, in perfect analogy to the fundamental frequency of a string measuring its "tightness".

### Machine learning II: learning solutions and solution operators

**Physics-informed neural networks (PINNs).** To solve $-u'' = f$, $u(0) = a$, $u(1) = b$, parameterize $u_\theta$ as a neural network and minimize the residual loss

$$
\mathcal{L}(\theta) = \frac{1}{N} \sum_{i=1}^{N} \big( u_\theta''(x_i) + f(x_i) \big)^2 + \lambda_{\text{bc}} \Big[ \big(u_\theta(0) - a\big)^2 + \big(u_\theta(1) - b\big)^2 \Big],
$$

with derivatives computed by automatic differentiation. The boundary penalty is *not optional*: the interior residual alone is minimized by a whole affine family $u + c_1 x + c_2$ (the null space of the operator), exactly the non-uniqueness of Section 1. PINNs shine when data is partial and the ODE/PDE acts as a physics prior; they struggle when the loss landscape mixes stiff residual scales — an active research area.

**Neural operators: learning the Green's function.** Proof 3.4 showed the solution map $f \mapsto u$ of a linear BVP is an integral operator $u(x) = \int G(x, \xi) f(\xi)\, d\xi$. The **Fourier Neural Operator (FNO)** learns maps of exactly this form: each layer applies a learnable multiplier to the low Fourier modes of its input (plus a local term and nonlinearity). For translation-invariant kernels, integration against $G$ *is* multiplication in Fourier space — e.g. $-u'' = f$ with periodic conditions gives $\hat{u}_m = \hat{f}_m / m^2$ — so a single FNO layer can represent the exact solution operator, and stacked nonlinear layers extend the idea to nonlinear and parametric PDE families. FNO learns *the operator once* and then solves any new right-hand side with one forward pass, in contrast to a PINN, which optimizes per problem instance.
