### Problem L0.1: IVP versus BVP — Where Uniqueness Breaks

**Problem Statement:** The IVP $y'' + y = 0$, $y(0) = 0$, $y'(0) = 1$ has exactly one solution. Show by direct computation that the BVP $y'' + y = 0$ with $y(0) = 0$ and $y(\pi) = 1$ has **no** solution, while replacing the second condition by $y(\pi) = 0$ produces **infinitely many** solutions. Explain why this does not contradict Picard–Lindelöf.

**Intuition:** Picard–Lindelöf marches forward from complete data at one point; a BVP asks two half-conditions at different points to be compatible — a global question the theorem never addresses.

**Solution:**

**Step 1 — General solution.** The characteristic equation $r^2 + 1 = 0$ gives $r = \pm i$, so every solution is $y(x) = c_1 \cos x + c_2 \sin x$.

**Step 2 — Impose $y(0) = 0$.** Since $y(0) = c_1$, we need $c_1 = 0$, leaving $y(x) = c_2 \sin x$.

**Step 3 — Case $y(\pi) = 1$.** Then $y(\pi) = c_2 \sin \pi = c_2 \cdot 0 = 0$ for *every* $c_2$. The condition $0 = 1$ is impossible: **no solution exists**.

**Step 4 — Case $y(\pi) = 0$.** The condition becomes $0 = 0$, satisfied for *every* $c_2$. Thus $y(x) = c_2 \sin x$ is a solution for all $c_2 \in \mathbb{R}$: **infinitely many solutions**.

**Step 5 — Why no contradiction.** Picard–Lindelöf guarantees a unique solution of $y'' + y = 0$ given $y(x_0)$ *and* $y'(x_0)$ at a single point $x_0$. In a BVP, the slope $y'(0)$ is not prescribed; each choice of slope launches a different unique IVP trajectory, and solvability asks whether *some* trajectory hits the target at $x = \pi$. Here every trajectory through $(0,0)$ satisfies $y(\pi) = 0$, so the target $1$ is unreachable and the target $0$ is hit by all of them. The interval length $\pi$ resonates with the operator: $\lambda = 1$ is an eigenvalue of $-y''$ on $[0, \pi]$ with Dirichlet conditions.

$$
\boxed{\ y(\pi) = 1:\ \text{no solution}; \qquad y(\pi) = 0:\ y = c_2 \sin x \ \text{for every } c_2 \in \mathbb{R}.\ }
$$

> **Key Takeaway:** BVP solvability is a resonance question between the operator, the interval, and the data — existence and uniqueness must be earned case by case, not assumed.

### Problem L0.2: Classifying Boundary Conditions

**Problem Statement:** Classify each of the following boundary conditions for a function $u(x)$ on $[0, L]$ as Dirichlet, Neumann, Robin, or periodic, and give a physical interpretation of each in the context of heat conduction:

1. $u(0) = 50$
2. $u'(L) = 0$
3. $-K u'(L) = h \big( u(L) - T_{\text{env}} \big)$ with constants $K, h \gt 0$
4. $u(0) = u(L)$ and $u'(0) = u'(L)$

**Solution:**

**Step 1 — Condition 1.** The *value* of $u$ is prescribed at an endpoint: **Dirichlet**. Physically, the end of the rod is held at a fixed temperature of $50$ degrees (e.g. clamped to a large thermal reservoir).

**Step 2 — Condition 2.** The *derivative* of $u$ is prescribed: **Neumann**. By Fourier's law the heat flux is $q = -K u'$, so $u'(L) = 0$ means zero flux — a perfectly **insulated** end.

**Step 3 — Condition 3.** A *linear combination* of value and derivative is prescribed: **Robin**. It equates conductive flux out of the rod, $-K u'(L)$, to convective loss $h (u(L) - T_{\text{env}})$ into the environment — Newton's law of cooling at the boundary. As $h \to 0$ it degenerates to Neumann (insulation); as $h \to \infty$ it forces $u(L) = T_{\text{env}}$, i.e. Dirichlet.

**Step 4 — Condition 4.** Values *and* derivatives at the two ends are coupled to each other: **periodic**. The rod is bent into a ring: $x = 0$ and $x = L$ are the same physical point, so temperature and flux must match there.

$$
\boxed{\ 1:\ \text{Dirichlet}, \quad 2:\ \text{Neumann}, \quad 3:\ \text{Robin}, \quad 4:\ \text{periodic}.\ }
$$

> **Key Takeaway:** Dirichlet fixes the state, Neumann fixes the flux, Robin trades one against the other, and periodic identifies the endpoints — and Robin contains the first two as limiting cases.

### Problem L0.3: What Is an Eigenfunction?

**Problem Statement:** Verify that $\varphi(x) = \sin \frac{3\pi x}{L}$ is an eigenfunction of the operator $A[y] = -y''$ with Dirichlet conditions $y(0) = y(L) = 0$, and find its eigenvalue. Then explain: (a) why $5\varphi(x)$ is *not* counted as a different eigenfunction, and (b) why $\cos\frac{3\pi x}{L}$ is *not* an eigenfunction of this problem even though it satisfies the same ODE.

**Solution:**

**Step 1 — Check the boundary conditions.** $\varphi(0) = \sin 0 = 0$ and $\varphi(L) = \sin 3\pi = 0$. Both hold, so $\varphi$ is in the domain of the operator.

**Step 2 — Apply the operator.** Differentiating twice:

$$
\varphi''(x) = -\left( \frac{3\pi}{L} \right)^2 \sin \frac{3\pi x}{L}, \qquad \text{so} \qquad A[\varphi] = -\varphi'' = \left( \frac{3\pi}{L} \right)^2 \varphi.
$$

The operator returns a *scalar multiple* of $\varphi$, so $\varphi$ is an eigenfunction with eigenvalue $\lambda = 9\pi^2 / L^2$.

**Step 3 — Part (a): scaling.** The eigenvalue equation $A[y] = \lambda y$ is linear and homogeneous: if $A[\varphi] = \lambda \varphi$ then $A[5\varphi] = 5 A[\varphi] = \lambda (5\varphi)$. Every nonzero multiple of $\varphi$ carries the same eigenvalue, so eigenfunctions are counted as one-dimensional *directions* (rays) in function space, exactly like eigenvectors of a matrix.

**Step 4 — Part (b): boundary conditions are part of the operator.** $\cos(3\pi x/L)$ does satisfy $-y'' = (3\pi/L)^2 y$, but $\cos(0) = 1 \neq 0$, so it violates $y(0) = 0$ and lies *outside the domain* of $A$. An eigenfunction must satisfy the ODE **and** the boundary conditions; changing the boundary conditions changes the operator and its entire spectrum (Dirichlet selects sines, Neumann selects cosines).

$$
\boxed{\ A[\varphi] = \frac{9\pi^2}{L^2}\, \varphi \ \text{with } \varphi(0) = \varphi(L) = 0: \ \text{eigenfunction with } \lambda_3 = \frac{9\pi^2}{L^2}.\ }
$$

> **Key Takeaway:** An eigenfunction is a boundary-condition-respecting direction that the operator merely rescales — and the boundary conditions are as much a part of the operator as the derivatives are.

### Problem L0.4: Why Orthogonality Is the Whole Game

**Problem Statement:** Suppose $\{\varphi_n\}_{n \ge 1}$ are eigenfunctions of a regular Sturm–Liouville problem with weight $w$, and a function $f$ has the expansion $f(x) = \sum_{n=1}^{\infty} c_n \varphi_n(x)$. Using only the orthogonality relation $\int_a^b \varphi_m \varphi_n w\, dx = 0$ for $m \neq n$, derive the coefficient formula and explain why, without orthogonality, finding the $c_n$ would require solving an infinite coupled linear system.

**Solution:**

**Step 1 — Multiply and integrate.** Fix an index $m$. Multiply the expansion by $\varphi_m(x)\, w(x)$ and integrate over $[a, b]$ (assuming term-by-term integration is valid):

$$
\int_a^b f \varphi_m w\, dx = \sum_{n=1}^{\infty} c_n \int_a^b \varphi_n \varphi_m w\, dx.
$$

**Step 2 — Orthogonality collapses the sum.** Every term with $n \neq m$ vanishes. Only $n = m$ survives:

$$
\int_a^b f \varphi_m w\, dx = c_m \int_a^b \varphi_m^2 w\, dx = c_m \lVert \varphi_m \rVert_w^2.
$$

**Step 3 — Solve for the coefficient.**

$$
\boxed{\ c_m = \frac{\langle f, \varphi_m \rangle_w}{\lVert \varphi_m \rVert_w^2} = \frac{\int_a^b f(x) \varphi_m(x) w(x)\, dx}{\int_a^b \varphi_m(x)^2 w(x)\, dx}\ }
$$

Each coefficient is computed by a *single independent integral* — an orthogonal projection of $f$ onto the direction $\varphi_m$.

**Step 4 — The counterfactual.** Without orthogonality, Step 2 fails: the equation for index $m$ would read $\langle f, \varphi_m \rangle_w = \sum_n c_n \langle \varphi_n, \varphi_m \rangle_w$, a full infinite linear system with Gram matrix $M_{mn} = \langle \varphi_n, \varphi_m \rangle_w$. Every coefficient would depend on every other, and any truncation of the basis would change all previously computed coefficients. Orthogonality diagonalizes the Gram matrix, decoupling the system completely — the same reason orthonormal eigenvectors make matrix problems trivial.

> **Key Takeaway:** Orthogonality turns "solve an infinite coupled system" into "compute one integral per coefficient" — it is the computational payoff of self-adjointness.

## Level 1 — Foundation

### Problem L1.1: The Dirichlet Eigenproblem, All Cases

**Problem Statement:** Find *all* eigenvalues and eigenfunctions of

$$
y'' + \lambda y = 0, \qquad y(0) = 0, \quad y(\pi) = 0,
$$

treating the cases $\lambda \lt 0$, $\lambda = 0$, and $\lambda \gt 0$ separately and justifying why the first two cases are excluded.

**Solution:**

**Step 1 — Case $\lambda = -\mu^2 \lt 0$ ($\mu \gt 0$).** General solution $y = A \cosh \mu x + B \sinh \mu x$. From $y(0) = A = 0$ we get $y = B \sinh \mu x$. Then $y(\pi) = B \sinh \mu \pi = 0$; since $\sinh$ is strictly increasing with $\sinh 0 = 0$ and $\mu\pi \gt 0$, $\sinh \mu \pi \gt 0$, forcing $B = 0$. Only the trivial solution — no negative eigenvalues.

**Step 2 — Case $\lambda = 0$.** Then $y'' = 0$ gives $y = Ax + B$. From $y(0) = B = 0$ and $y(\pi) = A\pi = 0$ we get $A = 0$. Trivial again — zero is not an eigenvalue.

**Step 3 — Case $\lambda = \mu^2 \gt 0$ ($\mu \gt 0$).** General solution $y = A \cos \mu x + B \sin \mu x$. From $y(0) = A = 0$, $y = B \sin \mu x$. Nontriviality requires $B \neq 0$, so $y(\pi) = B \sin \mu \pi = 0$ demands

$$
\sin \mu \pi = 0 \quad \Longleftrightarrow \quad \mu \pi = n \pi, \quad n = 1, 2, 3, \dots \quad \Longleftrightarrow \quad \mu = n.
$$

**Step 4 — Collect.**

$$
\boxed{\ \lambda_n = n^2, \qquad \varphi_n(x) = \sin n x, \qquad n = 1, 2, 3, \dots\ }
$$

The spectrum is real, simple, strictly increasing, and unbounded: $1, 4, 9, 16, \dots$ — and $\varphi_n$ has exactly $n - 1$ zeros inside $(0, \pi)$, matching the Sturm oscillation theorem.

> **Key Takeaway:** The case analysis is the proof that quantization is real: hyperbolic and linear profiles cannot bend back to hit zero twice, so only oscillatory solutions — at exactly the right frequencies — survive both boundary conditions.

### Problem L1.2: The Neumann Spectrum

**Problem Statement:** Find all eigenvalues and eigenfunctions of

$$
y'' + \lambda y = 0, \qquad y'(0) = 0, \quad y'(L) = 0,
$$

and point out the structural difference from the Dirichlet spectrum of Problem L1.1.

**Solution:**

**Step 1 — Case $\lambda = -\mu^2 \lt 0$.** With $y = A \cosh \mu x + B \sinh \mu x$, $y' = \mu (A \sinh \mu x + B \cosh \mu x)$. From $y'(0) = \mu B = 0$: $B = 0$. Then $y'(L) = \mu A \sinh \mu L = 0$ forces $A = 0$. No negative eigenvalues.

**Step 2 — Case $\lambda = 0$.** $y = Ax + B$, $y' = A$. Both conditions say $A = 0$, but $B$ is **free**: $y \equiv B \neq 0$ is a nontrivial solution. So $\lambda_0 = 0$ *is* an eigenvalue with constant eigenfunction $\varphi_0(x) = 1$.

**Step 3 — Case $\lambda = \mu^2 \gt 0$.** $y = A \cos \mu x + B \sin \mu x$, $y' = \mu(-A \sin \mu x + B \cos \mu x)$. From $y'(0) = \mu B = 0$: $B = 0$, leaving $y = A \cos \mu x$. Then $y'(L) = -A \mu \sin \mu L = 0$ with $A \neq 0$ requires $\sin \mu L = 0$, i.e. $\mu = n\pi/L$.

**Step 4 — Collect.**

$$
\boxed{\ \lambda_n = \left( \frac{n\pi}{L} \right)^2, \quad \varphi_n(x) = \cos \frac{n \pi x}{L}, \quad n = 0, 1, 2, \dots \quad (\varphi_0 \equiv 1).\ }
$$

**Step 5 — Structural contrast.** Same eigenvalue formula as Dirichlet, but: (i) the count starts at $n = 0$ — insulated ends admit a constant temperature profile, since only *flux* is constrained; (ii) eigenfunctions are cosines (slope zero at the ends) instead of sines (value zero at the ends). Physically: the Dirichlet fundamental mode must bend to satisfy value constraints and thus has positive energy, while Neumann permits the zero-energy "do nothing" mode.

> **Key Takeaway:** Neumann conditions admit the constant mode with eigenvalue zero — the mathematical statement that an insulated rod conserves total heat and can rest at any uniform temperature.

### Problem L1.3: A Mixed Dirichlet–Neumann Eigenproblem

**Problem Statement:** Find all eigenvalues and eigenfunctions of

$$
y'' + \lambda y = 0, \qquad y(0) = 0, \quad y'(L) = 0,
$$

which models a string clamped at $x = 0$ with a free (frictionless-ring) end at $x = L$, or an organ pipe closed at one end.

**Solution:**

**Step 1 — Exclude $\lambda \le 0$.** For $\lambda = -\mu^2 \lt 0$: $y(0) = 0$ gives $y = B \sinh \mu x$, and $y'(L) = B\mu \cosh \mu L = 0$ forces $B = 0$ since $\cosh \gt 0$ everywhere. For $\lambda = 0$: $y = Ax$ from $y(0)=0$, and $y'(L) = A = 0$. Both cases trivial.

**Step 2 — Case $\lambda = \mu^2 \gt 0$.** From $y(0) = 0$: $y = B \sin \mu x$, so $y' = B \mu \cos \mu x$. The condition $y'(L) = 0$ with $B \neq 0$ requires

$$
\cos \mu L = 0 \quad \Longleftrightarrow \quad \mu L = \frac{\pi}{2}, \frac{3\pi}{2}, \frac{5\pi}{2}, \dots = \frac{(2n-1)\pi}{2}, \quad n = 1, 2, 3, \dots
$$

**Step 3 — Collect.**

$$
\boxed{\ \lambda_n = \left( \frac{(2n-1)\pi}{2L} \right)^2, \qquad \varphi_n(x) = \sin \frac{(2n-1)\pi x}{2L}, \qquad n = 1, 2, 3, \dots\ }
$$

**Step 4 — Interpretation.** The mode shapes fit a *quarter wavelength* (node at $x=0$, antinode at $x=L$), so only **odd multiples** of the fundamental appear: frequencies scale as $1, 3, 5, 7, \dots$ This is exactly why a closed organ pipe (or clarinet) sounds an octave lower than an open pipe of the same length and has a hollow timbre dominated by odd harmonics, while Problems L1.1–L1.2 (two like ends) produce the full harmonic series.

> **Key Takeaway:** Mixing the boundary condition types shifts the spectrum off the integer grid — the odd-harmonic series of half-open pipes is a boundary-condition fingerprint you can literally hear.

### Problem L1.4: Fourier Sine Series of $f(x) = x$

**Problem Statement:** Expand $f(x) = x$ on $(0, L)$ in the Dirichlet eigenfunctions $\varphi_n(x) = \sin(n\pi x/L)$, i.e. compute the coefficients of $x = \sum_{n \ge 1} b_n \sin(n\pi x/L)$. Then evaluate the series at $x = L/2$ to obtain a famous numerical identity, and state what the series converges to at $x = L$ and why.

**Solution:**

**Step 1 — Coefficient formula.** With $\lVert \sin(n\pi x/L) \rVert^2 = L/2$ (weight $w = 1$),

$$
b_n = \frac{2}{L} \int_0^L x \sin \frac{n\pi x}{L}\, dx.
$$

**Step 2 — Integrate by parts.** With $u = x$, $dv = \sin(n\pi x/L)\, dx$, $v = -\frac{L}{n\pi} \cos \frac{n\pi x}{L}$:

$$
\int_0^L x \sin \frac{n\pi x}{L}\, dx = \left[ -\frac{Lx}{n\pi} \cos \frac{n\pi x}{L} \right]_0^L + \frac{L}{n\pi} \int_0^L \cos \frac{n\pi x}{L}\, dx = -\frac{L^2}{n\pi} \cos n\pi + 0 = -\frac{L^2}{n\pi} (-1)^n.
$$

**Step 3 — Assemble.** $b_n = \frac{2}{L} \cdot \frac{L^2}{n\pi} (-1)^{n+1} = \frac{2L}{n\pi} (-1)^{n+1}$, so

$$
\boxed{\ x = \frac{2L}{\pi} \sum_{n=1}^{\infty} \frac{(-1)^{n+1}}{n} \sin \frac{n\pi x}{L}, \qquad 0 \lt x \lt L.\ }
$$

**Step 4 — Evaluate at $x = L/2$.** Since $\sin(n\pi/2)$ cycles through $1, 0, -1, 0, \dots$, only odd $n = 2k-1$ contribute, with sign $(-1)^{k+1}$; the coefficient sign $(-1)^{n+1}$ is $+1$ for odd $n$. Hence

$$
\frac{L}{2} = \frac{2L}{\pi} \sum_{k=1}^{\infty} \frac{(-1)^{k+1}}{2k-1} \quad \Longrightarrow \quad \frac{\pi}{4} = 1 - \frac{1}{3} + \frac{1}{5} - \frac{1}{7} + \cdots,
$$

the Leibniz–Gregory series.

**Step 5 — Behavior at $x = L$.** Every term vanishes at $x = L$, so the series converges to $0$, *not* to $f(L) = L$. The odd periodic extension of $f$ jumps from $L$ to $-L$ at $x = L$, and Fourier series converge at a jump to the midpoint $\frac{1}{2}(L + (-L)) = 0$. Nearby, partial sums overshoot by about $9\%$ of the jump — the Gibbs phenomenon.

> **Key Takeaway:** Eigenfunction expansions represent a function in the basis dictated by the boundary conditions; where the function fights those conditions (here $f(L) \neq 0$), convergence is in the mean, coefficients decay slowly (like $1/n$), and Gibbs oscillations appear.

### Problem L1.5: Solving a BVP with the Green's Function

**Problem Statement:** The Green's function of $-y'' = f$, $y(0) = y(1) = 0$ is

$$
G(x, \xi) = \begin{cases} x(1 - \xi), & 0 \le x \le \xi, \\ \xi(1 - x), & \xi \le x \le 1. \end{cases}
$$

Use $y(x) = \int_0^1 G(x, \xi) f(\xi)\, d\xi$ to solve the BVP with $f(x) = x$, and verify your answer by direct substitution.

**Solution:**

**Step 1 — Split the integral at $\xi = x$.** For $\xi \lt x$ use the branch $\xi(1-x)$; for $\xi \gt x$ use $x(1-\xi)$:

$$
y(x) = (1 - x) \int_0^x \xi \cdot \xi\, d\xi + x \int_x^1 (1 - \xi)\, \xi\, d\xi.
$$

**Step 2 — First integral.**

$$
(1 - x) \int_0^x \xi^2 d\xi = (1 - x)\, \frac{x^3}{3} = \frac{x^3}{3} - \frac{x^4}{3}.
$$

**Step 3 — Second integral.**

$$
x \int_x^1 (\xi - \xi^2)\, d\xi = x \left[ \frac{\xi^2}{2} - \frac{\xi^3}{3} \right]_x^1 = x \left( \frac{1}{6} - \frac{x^2}{2} + \frac{x^3}{3} \right) = \frac{x}{6} - \frac{x^3}{2} + \frac{x^4}{3}.
$$

**Step 4 — Add.** The quartic terms cancel:

$$
y(x) = \frac{x}{6} + \frac{x^3}{3} - \frac{x^3}{2} = \frac{x}{6} - \frac{x^3}{6}.
$$

$$
\boxed{\ y(x) = \frac{x - x^3}{6}\ }
$$

**Step 5 — Verify.** $y' = \frac{1}{6} - \frac{x^2}{2}$, $y'' = -x$, so $-y'' = x = f(x)$; and $y(0) = 0$, $y(1) = \frac{1 - 1}{6} = 0$. Both the ODE and the boundary conditions check out.

**Step 6 — Moral.** No integration constants were tuned: the boundary conditions are *baked into the kernel*. The same $G$ instantly solves the problem for any other forcing — replace $f$ and integrate again. $G$ is the inverse operator, applied one right-hand side at a time.

> **Key Takeaway:** A Green's function converts "solve a differential equation with boundary conditions" into "evaluate one definite integral" — the boundary work is done once, in the kernel, forever.

### Problem L1.6: A Solvability Dichotomy at Resonance

**Problem Statement:** For which values of $\beta$ does the BVP

$$
y'' + 4y = 0, \qquad y(0) = 1, \quad y\!\left(\frac{\pi}{2}\right) = \beta
$$

have (a) no solution, (b) exactly one solution, (c) infinitely many solutions? Prove your classification.

**Intuition:** The interval length $\pi/2$ is exactly half the period of $\cos 2x$ — the boundary conditions land on a resonance of the operator.

**Solution:**

**Step 1 — General solution.** The roots of $r^2 + 4 = 0$ are $\pm 2i$, so $y(x) = A \cos 2x + B \sin 2x$.

**Step 2 — Impose the left condition.** $y(0) = A = 1$, hence $y(x) = \cos 2x + B \sin 2x$ with $B$ still free.

**Step 3 — Evaluate at the right end.** Since $\cos(2 \cdot \pi/2) = \cos \pi = -1$ and $\sin \pi = 0$:

$$
y\!\left(\frac{\pi}{2}\right) = -1 + B \cdot 0 = -1 \qquad \text{for every } B.
$$

The free constant $B$ has *no influence whatsoever* on the value at $x = \pi/2$.

**Step 4 — Classify.** If $\beta \neq -1$, the required condition $-1 = \beta$ is impossible: **no solution**. If $\beta = -1$, the condition holds identically and $y(x) = \cos 2x + B \sin 2x$ solves the BVP for **every** $B \in \mathbb{R}$: infinitely many solutions. The case "exactly one solution" **never occurs**.

$$
\boxed{\ \beta \neq -1: \text{ no solution}; \qquad \beta = -1: \text{ the infinite family } y = \cos 2x + B \sin 2x.\ }
$$

**Step 5 — Structural explanation (Fredholm alternative).** The homogeneous problem $y'' + 4y = 0$, $y(0) = y(\pi/2) = 0$ has the nontrivial solution $\sin 2x$ — that is, $\lambda = 4$ is a Dirichlet eigenvalue of $-y''$ on $[0, \pi/2]$. Whenever the homogeneous BVP has nontrivial solutions, the inhomogeneous one has either none or infinitely many (add any multiple of $\sin 2x$), depending on a compatibility condition on the data — here $\beta = -1$. Uniqueness for all data holds exactly when the interval avoids the eigenvalues.

> **Key Takeaway:** At an eigenvalue, a BVP obeys an all-or-nothing law — zero solutions or an infinite family — and which one occurs is decided by a single compatibility condition on the boundary data.

## Level 2 — Applications in AI/ML & Physics

### Problem L2.1: Electron in a 1 nm Quantum Well

**Problem Statement:** An electron (mass $m = 9.109 \times 10^{-31}$ kg) is confined to an infinite square well of width $L = 1.00$ nm. The stationary Schrödinger equation $-\frac{\hbar^2}{2m} \psi'' = E \psi$ with $\psi(0) = \psi(L) = 0$ is the Dirichlet eigenproblem. (a) Derive the energy formula $E_n$. (b) Compute $E_1$ and $E_2$ in electron-volts. (c) Find the wavelength of the photon emitted in the $n = 2 \to 1$ transition and identify the spectral region.

Constants: $h = 6.626 \times 10^{-34}$ J·s, $c = 2.998 \times 10^8$ m/s, $1\ \text{eV} = 1.602 \times 10^{-19}$ J.

**Solution:**

**Step 1 — Reduce to the Dirichlet eigenproblem.** Writing $\lambda = 2mE/\hbar^2$, the equation becomes $\psi'' + \lambda \psi = 0$ with $\psi(0) = \psi(L) = 0$. By Problem L1.1, $\lambda_n = (n\pi/L)^2$ with $\psi_n = \sin(n\pi x/L)$. Solving for the energy, and using $\hbar = h/2\pi$:

$$
E_n = \frac{\hbar^2}{2m} \left( \frac{n\pi}{L} \right)^2 = \frac{n^2 h^2}{8 m L^2}.
$$

**Step 2 — Ground state numerically.**

$$
E_1 = \frac{(6.626 \times 10^{-34})^2}{8 \times (9.109 \times 10^{-31}) \times (1.00 \times 10^{-9})^2} = \frac{4.390 \times 10^{-67}}{7.287 \times 10^{-48}} = 6.02 \times 10^{-20} \ \text{J}.
$$

Converting: $E_1 = 6.02 \times 10^{-20} / 1.602 \times 10^{-19} \approx 0.376$ eV.

**Step 3 — First excited state.** Energies scale as $n^2$: $E_2 = 4 E_1 \approx 1.50$ eV.

**Step 4 — Transition photon.** $\Delta E = E_2 - E_1 = 3 E_1 \approx 1.13$ eV $= 1.81 \times 10^{-19}$ J. The photon wavelength is

$$
\lambda_{\text{photon}} = \frac{h c}{\Delta E} = \frac{(6.626 \times 10^{-34})(2.998 \times 10^8)}{1.81 \times 10^{-19}} \approx 1.10 \times 10^{-6} \ \text{m} = 1.10\ \mu\text{m},
$$

in the **near-infrared** — the operating regime of semiconductor quantum-well lasers and infrared photodetectors, whose emission is tuned by choosing $L$.

$$
\boxed{\ E_n = \frac{n^2 h^2}{8 m L^2}, \qquad E_1 \approx 0.376\ \text{eV}, \quad E_2 \approx 1.50\ \text{eV}, \quad \lambda_{2 \to 1} \approx 1.10\ \mu\text{m (near-IR)}.\ }
$$

> **Key Takeaway:** Quantized energy levels are Dirichlet eigenvalues wearing physical units: the $1/L^2$ scaling means engineers tune a laser's color simply by growing a well of a different width.

### Problem L2.2: Guitar String Frequencies from Tension and Density

**Problem Statement:** A guitar's high-E string has vibrating length $L = 0.650$ m and linear mass density $\mu = 4.00 \times 10^{-4}$ kg/m, tensioned at $T = 73.4$ N. Standing waves obey the wave equation $u_{tt} = c^2 u_{xx}$ with $u(0,t) = u(L,t) = 0$ and $c = \sqrt{T/\mu}$. (a) Derive the mode frequencies $f_n$ from the Dirichlet eigenproblem. (b) Compute the fundamental $f_1$ and identify the pitch. (c) By what factor must the tension change to raise the pitch a whole tone (ratio $2^{2/12} = 1.122$)?

**Solution:**

**Step 1 — Modes from separation of variables.** Substituting $u = X(x)\, \Theta(t)$ into the wave equation gives $X'' + \lambda X = 0$, $X(0) = X(L) = 0$, and $\Theta'' = -\lambda c^2 \Theta$. The spatial eigenproblem yields $\lambda_n = (n\pi/L)^2$, $X_n = \sin(n\pi x/L)$; then $\Theta_n$ oscillates at angular frequency $\omega_n = c \sqrt{\lambda_n} = n \pi c / L$, i.e.

$$
f_n = \frac{\omega_n}{2\pi} = \frac{n c}{2 L} = \frac{n}{2L} \sqrt{\frac{T}{\mu}}, \qquad n = 1, 2, 3, \dots
$$

— a perfect integer harmonic series, which is why the string sounds a single well-defined pitch.

**Step 2 — Wave speed.**

$$
c = \sqrt{\frac{T}{\mu}} = \sqrt{\frac{73.4}{4.00 \times 10^{-4}}} = \sqrt{1.835 \times 10^5} \approx 428.4 \ \text{m/s}.
$$

**Step 3 — Fundamental frequency.**

$$
f_1 = \frac{c}{2L} = \frac{428.4}{2 \times 0.650} = \frac{428.4}{1.300} \approx 329.5 \ \text{Hz},
$$

which is the note **E4** ($329.63$ Hz in equal temperament) — the guitar's high-E string, correctly in tune. Overtones: $f_2 \approx 659$ Hz (E5), $f_3 \approx 989$ Hz (B5), etc.

**Step 4 — Retuning by a whole tone.** Since $f_1 \propto \sqrt{T}$ at fixed $L, \mu$, raising the pitch by the factor $r = 1.122$ requires $T \to r^2 T = 1.260\, T \approx 92.5$ N — a $26\%$ tension increase for a $12\%$ frequency increase. (This quadratic sensitivity is why strings snap when overtuned, and why players prefer fretting, which changes $L$ linearly.)

$$
\boxed{\ f_n = \frac{n}{2L}\sqrt{\frac{T}{\mu}}, \qquad f_1 \approx 329.5\ \text{Hz (E4)}, \qquad T_{\text{new}} = 2^{4/12}\, T \approx 92.5\ \text{N}.\ }
$$

> **Key Takeaway:** A guitar is a Sturm–Liouville solver made of wood: the Dirichlet spectrum is the harmonic series you hear, and the square-root law $f \propto \sqrt{T}$ governs every tuning peg.

### Problem L2.3: Steady Heat in a Rod with a Convective (Robin) End

**Problem Statement:** A rod of length $L = 0.50$ m and thermal conductivity $K = 200$ W/(m·K) has its left end held at $u(0) = 100\,^{\circ}$C. Its right end loses heat by convection to air at $T_{\text{env}} = 20\,^{\circ}$C with coefficient $h = 40$ W/(m²·K), giving the Robin condition $-K u'(L) = h\,\big(u(L) - T_{\text{env}}\big)$. The steady temperature satisfies $u'' = 0$. Find $u(x)$, the tip temperature $u(L)$, and the heat flux through the rod, and check the flux balance at the convective end.

**Solution:**

**Step 1 — General steady profile.** $u'' = 0$ gives the linear profile $u(x) = A + Bx$.

**Step 2 — Left (Dirichlet) condition.** $u(0) = A = 100$.

**Step 3 — Right (Robin) condition.** With $u'(x) = B$ and $u(L) = 100 + BL$:

$$
-K B = h \big( 100 + BL - 20 \big) \quad \Longrightarrow \quad -200 B = 40 \big( 80 + 0.5 B \big) = 3200 + 20 B.
$$

**Step 4 — Solve for the slope.** $-220 B = 3200$, so $B = -14.55$ K/m (temperature falls toward the cooled end).

**Step 5 — Profile and tip temperature.**

$$
u(x) = 100 - 14.55\, x, \qquad u(L) = 100 - 14.55 \times 0.50 \approx 92.7\,^{\circ}\text{C}.
$$

**Step 6 — Flux and balance check.** Fourier's law: $q = -K u' = -200 \times (-14.55) \approx 2.91 \times 10^3$ W/m², flowing left to right, constant along the rod (no internal sources). Convective loss at the tip: $h (u(L) - T_{\text{env}}) = 40 \times (92.7 - 20) = 40 \times 72.7 \approx 2.91 \times 10^3$ W/m². Conduction in equals convection out — the Robin condition is exactly this balance.

$$
\boxed{\ u(x) = 100 - 14.55\,x \ \ ^{\circ}\text{C}, \qquad u(L) \approx 92.7\,^{\circ}\text{C}, \qquad q \approx 2.91\ \text{kW/m}^2.\ }
$$

**Step 7 — Limiting sanity checks.** As $h \to \infty$ (violent cooling): $B \to -160$ K/m and $u(L) \to 20$, recovering a Dirichlet end. As $h \to 0$ (insulation): $B \to 0$ and $u \equiv 100$, the Neumann limit. The computed answer sits between, as it must.

> **Key Takeaway:** Robin conditions encode a physical balance law at the boundary; solving the BVP is bookkeeping, and the two limiting cases (Dirichlet, Neumann) bracket every answer as built-in sanity checks — this is the design calculation for every heat sink and cooling fin.

### Problem L2.4: Formulating a PINN Loss for a BVP

**Problem Statement:** You wish to solve $-u'' = \sin(\pi x)$ on $(0,1)$ with $u(0) = u(1) = 0$ using a physics-informed neural network $u_\theta(x)$. (a) Write the exact solution. (b) Write the full PINN loss with collocation points $\{x_i\}_{i=1}^N \subset (0,1)$ and boundary weight $\lambda_{\text{bc}}$. (c) Prove that dropping the boundary term leaves a loss whose global minimum (value zero) is attained by an infinite family of networks, so BC terms are *mathematically necessary*, not a tuning nicety. (d) Name one alternative to penalty enforcement.

**Solution:**

**Step 1 — Exact solution.** Try $u = a \sin(\pi x)$: then $-u'' = a \pi^2 \sin(\pi x)$, so $a = 1/\pi^2$, and the boundary conditions hold automatically:

$$
u^{\ast}(x) = \frac{\sin(\pi x)}{\pi^2}.
$$

**Step 2 — PINN loss.** With automatic differentiation supplying $u_\theta''$:

$$
\mathcal{L}(\theta) = \underbrace{\frac{1}{N} \sum_{i=1}^{N} \Big( u_\theta''(x_i) + \sin(\pi x_i) \Big)^2}_{\text{PDE residual}} \; + \; \lambda_{\text{bc}} \Big[ u_\theta(0)^2 + u_\theta(1)^2 \Big].
$$

Training minimizes $\mathcal{L}$ over the network weights $\theta$ by gradient descent; no labeled solution data is needed — the equation itself is the supervision signal.

**Step 3 — Necessity of the boundary term.** Suppose the loss is only the residual term. If $v$ is any function with $v'' = 0$, i.e. $v(x) = c_1 x + c_2$, then $u^{\ast} + v$ has the *same* second derivative as $u^{\ast}$, hence exactly zero residual at every collocation point. So every member of the two-parameter family

$$
u^{\ast}(x) + c_1 x + c_2, \qquad (c_1, c_2) \in \mathbb{R}^2,
$$

achieves the global minimum $\mathcal{L}_{\text{res}} = 0$. The residual-only loss cannot distinguish them: this is precisely the null space of $d^2/dx^2$, the non-uniqueness of Section 1 of the theory notebook. Gradient descent will converge to an arbitrary member depending on initialization. Adding the boundary penalty makes $u^{\ast}$ the *unique* zero of the total loss, because $c_1 x + c_2$ vanishing at $x = 0$ and $x = 1$ forces $c_1 = c_2 = 0$.

**Step 4 — Alternative: hard constraints.** Enforce the BCs *by construction*: parameterize $u_\theta(x) = x(1 - x)\, N_\theta(x)$. Every candidate now satisfies $u_\theta(0) = u_\theta(1) = 0$ identically, the penalty term (and its weight tuning) disappears, and optimization runs only over functions in the correct affine space — usually improving conditioning.

$$
\boxed{\ \mathcal{L}(\theta) = \frac{1}{N} \sum_i \big( u_\theta''(x_i) + \sin(\pi x_i) \big)^2 + \lambda_{\text{bc}} \big[ u_\theta(0)^2 + u_\theta(1)^2 \big], \qquad u^{\ast} = \frac{\sin \pi x}{\pi^2}.\ }
$$

> **Key Takeaway:** The boundary terms in a PINN loss are the discrete shadow of BVP well-posedness — without them the loss has the operator's entire null space as spurious global minima, and no amount of training can pick the right solution.

### Problem L2.5: Fourier Neural Operators Learn Green's Functions

**Problem Statement:** For $-u'' = f$ on the periodic domain $[0, 2\pi]$ with zero-mean data, expand $f(x) = \sum_{m \neq 0} \hat{f}_m e^{imx}$. (a) Show the solution operator acts *diagonally* in Fourier space: $\hat{u}_m = \hat{f}_m / m^2$. (b) Show the solution equals convolution with a fixed kernel, i.e. a Green's-function integral. (c) Explain why a single Fourier Neural Operator (FNO) layer — which applies learned multipliers $R_m$ to the lowest $M$ Fourier modes of its input — can represent this solution operator, what its truncation error is, and why the same trained weights work at *any* grid resolution.

**Solution:**

**Step 1 — Diagonalization.** Term-by-term differentiation of $u(x) = \sum_{m \neq 0} \hat{u}_m e^{imx}$ gives $-u'' = \sum_{m \neq 0} m^2 \hat{u}_m e^{imx}$. Matching coefficients with $f$:

$$
m^2 \hat{u}_m = \hat{f}_m \quad \Longrightarrow \quad \hat{u}_m = \frac{\hat{f}_m}{m^2}.
$$

The exponentials $e^{imx}$ are the periodic Sturm–Liouville eigenfunctions, so in *its own eigenbasis* the solution operator is the diagonal multiplier $1/m^2$ — the operator analogue of inverting a diagonalized matrix.

**Step 2 — Green's function form.** By the convolution theorem, multiplication of Fourier coefficients is convolution in $x$:

$$
u(x) = \frac{1}{2\pi} \int_0^{2\pi} G(x - \xi)\, f(\xi)\, d\xi, \qquad G(z) = \sum_{m \neq 0} \frac{e^{imz}}{m^2},
$$

and this series sums to the periodic Bernoulli-type kernel $G(z) = \frac{(z - \pi)^2}{2} - \frac{\pi^2}{6}$ for $0 \le z \le 2\pi$. Translation invariance of the operator is exactly what makes the kernel a function of $x - \xi$ alone.

**Step 3 — FNO representation.** An FNO layer computes: FFT of the input, multiplication of modes $\lvert m \rvert \le M$ by learned complex weights $R_m$ (others truncated), inverse FFT, plus a local linear term and pointwise nonlinearity. Choosing $R_m = 1/m^2$ for $0 \lt \lvert m \rvert \le M$ (and identity elsewhere off) reproduces the solution operator exactly on the retained band. The training process *learns the Green's function in its diagonal representation*.

**Step 4 — Truncation error.** For input with coefficients $\hat{f}_m$, the omitted tail contributes $\sum_{\lvert m \rvert \gt M} \lvert \hat{f}_m \rvert^2 / m^4$ to the squared error. If $f$ is smooth, $\hat{f}_m$ decays faster than any polynomial, so the error is spectrally small; the $1/m^4$ factor shows the operator itself smooths, further suppressing high modes.

**Step 5 — Discretization invariance.** The learned object is the set of *continuum* multipliers $\{R_m\}$, indexed by frequency, not by grid points. Evaluating on a finer grid changes only the FFT sizes; the same $R_m$ apply. A CNN kernel, by contrast, is tied to a grid spacing. This is why FNOs train at one resolution and evaluate at another — they parameterize the operator, not a grid function.

$$
\boxed{\ \hat{u}_m = \frac{\hat{f}_m}{m^2} \ \Longleftrightarrow\ u = G * f: \ \text{the FNO's learned spectral multipliers are a Green's function in disguise.}\ }
$$

> **Key Takeaway:** Green's functions, diagonal Fourier multipliers, and FNO layers are three descriptions of one object — the inverse of a differential operator — which is why neural operators generalize across right-hand sides and grid resolutions the way a solver does.

### Problem L2.6: Discrete Laplacian Eigenvalues and Their Continuum Limit

**Problem Statement:** Let $A = \mathrm{tridiag}(-1, 2, -1)$ be the $n \times n$ finite-difference matrix for $-y''$ on $[0, L]$ with Dirichlet conditions and $h = L/(n+1)$. (a) Verify that $v^{(j)}$ with components $v^{(j)}_i = \sin\frac{ij\pi}{n+1}$ satisfies $A v^{(j)} = \mu_j v^{(j)}$ with $\mu_j = 2 - 2\cos\frac{j\pi}{n+1}$. (b) Show $\mu_j / h^2 \to (j\pi/L)^2$ as $n \to \infty$ with the leading error term. (c) For $n = 3$, $L = 1$, compare the smallest scaled eigenvalue to $\pi^2$.

**Solution:**

**Step 1 — Verify the eigenvector identity.** Write $\theta = \frac{j\pi}{n+1}$ and $v_i = \sin(i\theta)$. Row $i$ of $A v$ is $-v_{i-1} + 2v_i - v_{i+1}$. The sum formula $\sin((i \pm 1)\theta) = \sin i\theta \cos\theta \pm \cos i\theta \sin\theta$ gives

$$
v_{i-1} + v_{i+1} = 2 \cos\theta\, \sin i\theta,
$$

so $-v_{i-1} + 2 v_i - v_{i+1} = (2 - 2\cos\theta)\, \sin i\theta = \mu_j v_i$. The boundary rows work automatically because $v_0 = \sin 0 = 0$ and $v_{n+1} = \sin(j\pi) = 0$ — the discrete eigenvector is the continuous eigenfunction $\sin(j\pi x/L)$ *sampled at the grid points*, vanishing at the ghost nodes exactly as the Dirichlet conditions demand.

**Step 2 — Continuum limit.** With $\theta_j = j\pi h / L$ small for fixed $j$, Taylor expansion $\cos\theta = 1 - \frac{\theta^2}{2} + \frac{\theta^4}{24} - \cdots$ gives

$$
\frac{\mu_j}{h^2} = \frac{2 - 2\cos\theta_j}{h^2} = \frac{\theta_j^2 - \frac{\theta_j^4}{12} + O(\theta_j^6)}{h^2} = \left( \frac{j\pi}{L} \right)^2 - \frac{h^2}{12} \left( \frac{j\pi}{L} \right)^4 + O(h^4).
$$

The discrete eigenvalues converge to the Sturm–Liouville eigenvalues at rate $O(h^2)$, but the error constant grows like $j^4$: the grid resolves low modes well and high modes badly (for $j$ near $n$, $\mu_j/h^2$ saturates near $4/h^2$ instead of blowing up quadratically).

**Step 3 — Numerical check ($n = 3$, $L = 1$, $h = 1/4$).**

$$
\mu_1 = 2 - 2\cos\frac{\pi}{4} = 2 - \sqrt{2} \approx 0.5858, \qquad \frac{\mu_1}{h^2} = 16 \times 0.5858 \approx 9.37.
$$

The continuous value is $\pi^2 \approx 9.87$, so the relative error is about $5.0\%$. The predicted leading error is $\frac{h^2}{12}\pi^4 = \frac{97.41}{192} \approx 0.507$, and indeed $9.87 - 9.37 = 0.50$ — the asymptotic formula is already accurate at $n = 3$.

$$
\boxed{\ \frac{\mu_j}{h^2} = \left( \frac{j\pi}{L} \right)^2 - \frac{h^2}{12}\left( \frac{j\pi}{L} \right)^4 + O(h^4) \ \longrightarrow\ \lambda_j = \left( \frac{j\pi}{L} \right)^2.\ }
$$

**Remark (ML connection).** $A$ is the graph Laplacian of the path graph, so this problem *is* the statement that graph-Laplacian eigenvectors (used in spectral clustering, Laplacian eigenmaps, and spectral GNNs) converge to Sturm–Liouville eigenfunctions — discrete learning machinery inherits continuum spectral theory.

> **Key Takeaway:** The discrete Laplacian carries an exact sampled copy of the continuous eigensystem with $O(h^2)$ eigenvalue error growing like $j^4$ — trust the low modes, distrust the high ones, in numerics and in spectral ML alike.

### Problem L3.1: Eigenvalues of a Regular Sturm–Liouville Problem Are Simple

**Problem Statement:** Prove that every eigenvalue of a regular Sturm–Liouville problem

$$
(p y')' + q y + \lambda w y = 0, \qquad \alpha_1 y(a) + \alpha_2 y'(a) = 0, \quad \beta_1 y(b) + \beta_2 y'(b) = 0,
$$

with $p \gt 0$, $(\alpha_1, \alpha_2) \neq (0,0)$, is **simple**: its eigenspace is one-dimensional. Explain precisely where the argument fails for periodic boundary conditions (where double eigenvalues do occur).

**Solution:**

**Step 1 — Setup.** Let $\varphi$ and $\psi$ be eigenfunctions for the *same* eigenvalue $\lambda$. Both solve the same second-order linear ODE $(p y')' + (q + \lambda w) y = 0$. It suffices to show they are linearly dependent.

**Step 2 — The Wronskian vanishes at the endpoint.** Both satisfy the left boundary condition:

$$
\alpha_1 \varphi(a) + \alpha_2 \varphi'(a) = 0, \qquad \alpha_1 \psi(a) + \alpha_2 \psi'(a) = 0.
$$

This says the nonzero vector $(\alpha_1, \alpha_2)$ is orthogonal to both $(\varphi(a), \varphi'(a))$ and $(\psi(a), \psi'(a))$ in $\mathbb{R}^2$. Two vectors orthogonal to one nonzero vector in the plane are parallel, so the determinant

$$
W(a) = \varphi(a) \psi'(a) - \psi(a) \varphi'(a) = 0.
$$

**Step 3 — Abel's identity in SL form.** Compute the derivative of $p W$ where $W = \varphi \psi' - \psi \varphi'$:

$$
(pW)' = \varphi (p\psi')' - \psi (p\varphi')' = \varphi \big[ -(q + \lambda w)\psi \big] - \psi \big[ -(q + \lambda w)\varphi \big] = 0,
$$

(the terms $\varphi' p \psi' - \psi' p \varphi'$ from the product rule cancel). So $p(x) W(x)$ is **constant** on $[a, b]$.

**Step 4 — Conclude the Wronskian vanishes identically.** The constant equals its value at $a$: $p(a) W(a) = 0$. Since $p \gt 0$ everywhere, $W(x) = 0$ for all $x \in [a, b]$.

**Step 5 — Dependence.** A vanishing Wronskian of two solutions of the same second-order linear ODE (with continuous coefficients and $p \gt 0$) forces linear dependence: at any $x_0$, $(\psi(x_0), \psi'(x_0))$ is a multiple of $(\varphi(x_0), \varphi'(x_0))$ (note $(\varphi(x_0), \varphi'(x_0)) \neq (0,0)$, else $\varphi \equiv 0$ by IVP uniqueness), and IVP uniqueness propagates $\psi = c\varphi$ to the whole interval. Hence the eigenspace has dimension one:

$$
\boxed{\ \text{Every eigenvalue of a regular (separated-BC) Sturm–Liouville problem is simple.}\ }
$$

**Step 6 — Why periodic conditions escape.** The argument hinges on Step 2: a *separated* condition constrains the solution vector at a single point to a one-dimensional line, forcing $W(a) = 0$. Periodic conditions $y(a) = y(b)$, $y'(a) = y'(b)$ couple the endpoints and constrain no single-endpoint vector to a line. Indeed on $[-L, L]$ both $\cos(n\pi x/L)$ and $\sin(n\pi x/L)$ are periodic eigenfunctions for $\lambda_n = (n\pi/L)^2$ with nonvanishing Wronskian — a genuinely two-dimensional eigenspace.

> **Key Takeaway:** Simplicity of the spectrum is a boundary-condition theorem, not an operator theorem: separated conditions pin the Wronskian to zero at an endpoint, and Abel's identity does the rest.

### Problem L3.2: The Rayleigh Quotient and a Variational Bound

**Problem Statement:** For the Dirichlet problem $-y'' = \lambda y$ on $(0, 1)$, $y(0) = y(1) = 0$, define for admissible $y \not\equiv 0$ (smooth, satisfying the BCs)

$$
R[y] = \frac{\int_0^1 y'(x)^2\, dx}{\int_0^1 y(x)^2\, dx}.
$$

(a) Prove $\lambda_1 = \min_y R[y]$, with the minimum attained exactly at multiples of $\varphi_1 = \sin \pi x$. (b) Use the trial function $y(x) = x(1 - x)$ to obtain an upper bound for $\lambda_1 = \pi^2$ and compute its relative error. (State clearly where completeness of the eigenfunctions is used.)

**Solution:**

**Step 1 — Dirichlet form identity.** Integrating by parts with $y(0) = y(1) = 0$:

$$
\int_0^1 y'^2\, dx = \big[ y y' \big]_0^1 - \int_0^1 y y''\, dx = \int_0^1 y \, (-y'')\, dx.
$$

So $R[y] = \langle y, -y'' \rangle / \langle y, y \rangle$ — the quadratic form of the operator, the exact analogue of $x^{\top} A x / x^{\top} x$.

**Step 2 — Expand in the eigenbasis.** By completeness of $\{\varphi_n = \sin n\pi x\}$ (Sturm–Liouville theorem; this is where completeness is used), any admissible $y$ has $y = \sum_n c_n \varphi_n$ with $\lVert \varphi_n \rVert^2 = \tfrac{1}{2}$. Orthogonality kills cross terms in both quadratic forms; integrating $\varphi_m' \varphi_n'$ by parts likewise gives $\lambda_n \delta_{mn} \lVert \varphi_n \rVert^2$. Hence

$$
R[y] = \frac{\sum_n \lambda_n c_n^2 \lVert \varphi_n \rVert^2}{\sum_n c_n^2 \lVert \varphi_n \rVert^2} \quad \text{with } \lambda_n = n^2 \pi^2.
$$

**Step 3 — The minimum.** Since $\lambda_n \ge \lambda_1$ for all $n$, the numerator is at least $\lambda_1$ times the denominator: $R[y] \ge \lambda_1$. Equality holds iff $c_n = 0$ for every $n$ with $\lambda_n \gt \lambda_1$, i.e. iff $y$ is a multiple of $\varphi_1$. And $R[\varphi_1] = \lambda_1$ exactly. Therefore

$$
\boxed{\ \lambda_1 = \min_{y \neq 0,\ y(0)=y(1)=0} \frac{\int_0^1 y'^2\, dx}{\int_0^1 y^2\, dx}, \quad \text{attained only at } y = c \sin \pi x.\ }
$$

**Step 4 — Trial function bound.** For $y = x(1-x)$: $y' = 1 - 2x$, so

$$
\int_0^1 y'^2 dx = \int_0^1 (1 - 2x)^2 dx = \left[ -\frac{(1-2x)^3}{6} \right]_0^1 = \frac{1}{3},
$$

$$
\int_0^1 y^2 dx = \int_0^1 (x^2 - 2x^3 + x^4)\, dx = \frac{1}{3} - \frac{1}{2} + \frac{1}{5} = \frac{1}{30}.
$$

Hence $R[y] = \frac{1/3}{1/30} = 10$, giving the rigorous bound $\pi^2 \le 10$.

**Step 5 — Error.** $\pi^2 = 9.8696\dots$, so the relative error is $(10 - \pi^2)/\pi^2 \approx 1.32\%$ — a one-line computation with a crude parabola lands within $1.5\%$ of the true fundamental eigenvalue, because the Rayleigh quotient is *quadratically* insensitive to the trial function's error: writing $y = \varphi_1 + \varepsilon g$ gives $R[y] = \lambda_1 + O(\varepsilon^2)$.

> **Key Takeaway:** Eigenvalues are minima of energy ratios; any admissible guess gives a certified upper bound with error quadratic in the guess's defect — the principle behind Ritz–Galerkin methods, finite elements, and variational quantum chemistry.

### Problem L3.3: Deriving d'Alembert's Formula via Characteristics

**Problem Statement:** Derive the solution of the wave equation on the whole line,

$$
u_{tt} = c^2 u_{xx}, \qquad u(x, 0) = g(x), \quad u_t(x, 0) = v(x),
$$

by the change of variables $\xi = x - ct$, $\eta = x + ct$, obtaining d'Alembert's formula. Conclude that the solution at $(x, t)$ depends only on data in $[x - ct, x + ct]$ (finite domain of dependence).

**Solution:**

**Step 1 — Transform the operator.** With $u(x,t) = U(\xi, \eta)$, the chain rule gives $u_x = U_\xi + U_\eta$ and $u_t = -c U_\xi + c U_\eta$. Differentiating again:

$$
u_{xx} = U_{\xi\xi} + 2 U_{\xi\eta} + U_{\eta\eta}, \qquad u_{tt} = c^2 \big( U_{\xi\xi} - 2 U_{\xi\eta} + U_{\eta\eta} \big).
$$

Therefore $u_{tt} - c^2 u_{xx} = -4 c^2\, U_{\xi\eta}$, and the wave equation collapses to $U_{\xi\eta} = 0$.

**Step 2 — Integrate twice.** $\partial_\eta (U_\xi) = 0$ means $U_\xi$ is a function of $\xi$ alone; integrating in $\xi$, $U = F(\xi) + G(\eta)$ for arbitrary $C^2$ functions $F, G$. Back in physical variables:

$$
u(x, t) = F(x - ct) + G(x + ct),
$$

a right-moving and a left-moving wave — the *general* solution.

**Step 3 — Impose the initial data.** At $t = 0$:

$$
F(x) + G(x) = g(x), \qquad u_t(x,0) = -c F'(x) + c G'(x) = v(x).
$$

**Step 4 — Solve the functional system.** Integrate the second equation from $0$ to $x$: $G(x) - F(x) = \frac{1}{c} \int_0^x v(s)\, ds + C$. Adding and subtracting with $F + G = g$:

$$
F(x) = \frac{g(x)}{2} - \frac{1}{2c} \int_0^x v(s)\, ds - \frac{C}{2}, \qquad G(x) = \frac{g(x)}{2} + \frac{1}{2c} \int_0^x v(s)\, ds + \frac{C}{2}.
$$

**Step 5 — Assemble.** The constants cancel in the sum $F(x - ct) + G(x + ct)$, and the two integrals combine into one over $[x - ct, x + ct]$:

$$
\boxed{\ u(x, t) = \frac{g(x - ct) + g(x + ct)}{2} + \frac{1}{2c} \int_{x - ct}^{x + ct} v(s)\, ds.\ }
$$

**Step 6 — Domain of dependence.** The formula uses $g$ only at the two points $x \pm ct$ and $v$ only on the interval between them. Information outside $[x - ct, x + ct]$ cannot influence $u(x,t)$: disturbances propagate at speed exactly $c$ — the defining feature of hyperbolic equations, in stark contrast to the heat equation, whose solution at any $t \gt 0$ feels the entire initial datum (infinite speed of propagation). On a bounded string, reflecting the data oddly across the ends reproduces the standing-wave series of Problem L2.2 — the traveling and standing pictures are equivalent.

> **Key Takeaway:** In characteristic coordinates the wave operator factors and integrates in two lines; the resulting formula makes finite propagation speed a theorem you can read off the limits of one integral.

### Problem L3.4: Two Solutions of a Bratu-Type Nonlinear BVP via Shooting

**Problem Statement:** Consider the Bratu problem

$$
y'' + \lambda e^{y} = 0, \qquad y(0) = y(1) = 0, \qquad \lambda \gt 0.
$$

(a) Show that every solution has the form $y(x) = -2 \ln\!\left[ \cosh\!\big( \tfrac{\theta}{2}(x - \tfrac{1}{2}) \big) \big/ \cosh\tfrac{\theta}{4} \right]$ where $\theta \ge 0$ satisfies the transcendental equation $\theta = \sqrt{2\lambda}\, \cosh\frac{\theta}{4}$. (b) Analyze the number of roots $\theta$ as a function of $\lambda$ and deduce that there is a critical value $\lambda^{\ast} \approx 3.514$ with **two** solutions for $0 \lt \lambda \lt \lambda^{\ast}$, one at $\lambda^{\ast}$, and none beyond. (c) Interpret the two solutions through the shooting function $F(s) = y(1; s)$.

**Solution:**

**Step 1 — Verify the ansatz solves the ODE.** Let $z = \frac{\theta}{2}\big(x - \frac{1}{2}\big)$ and $y = -2 \ln \cosh z + 2 \ln \cosh \frac{\theta}{4}$. Then

$$
y' = -\theta \tanh z, \qquad y'' = -\frac{\theta^2}{2}\, \mathrm{sech}^2 z,
$$

while $e^{y} = \cosh^2\frac{\theta}{4}\, \mathrm{sech}^2 z$. The ODE $y'' = -\lambda e^y$ becomes

$$
-\frac{\theta^2}{2} \mathrm{sech}^2 z = -\lambda \cosh^2\frac{\theta}{4}\, \mathrm{sech}^2 z \quad \Longleftrightarrow \quad \theta^2 = 2\lambda \cosh^2 \frac{\theta}{4} \quad \Longleftrightarrow \quad \theta = \sqrt{2\lambda} \cosh\frac{\theta}{4}.
$$

Boundary conditions: at $x = 0$ and $x = 1$, $z = \mp \theta/4$ and $\cosh(\mp\theta/4) = \cosh(\theta/4)$, so the logarithm's argument is $1$ and $y = 0$. (That *every* solution is of this form follows from the standard first-integral reduction: multiplying by $y'$ and integrating gives $\tfrac{1}{2}y'^2 + \lambda e^y = \lambda e^{y_{\max}}$; solutions are positive, symmetric about $x = \tfrac{1}{2}$, and the resulting separable equation integrates to the stated $\cosh$ profile.)

**Step 2 — Count roots of the transcendental equation.** Define

$$
g(\theta) = \frac{\theta}{\cosh(\theta/4)}, \qquad \theta \ge 0; \qquad \text{roots satisfy } g(\theta) = \sqrt{2\lambda}.
$$

We have $g(0) = 0$, $g(\theta) \to 0$ as $\theta \to \infty$ (exponential beats linear), and $g \gt 0$ in between, so $g$ attains an interior maximum. Compute

$$
g'(\theta) = \frac{\cosh(\theta/4) - \frac{\theta}{4} \sinh(\theta/4)}{\cosh^2(\theta/4)} = 0 \quad \Longleftrightarrow \quad \frac{\theta}{4} \tanh\frac{\theta}{4} = 1.
$$

The function $s \tanh s$ is strictly increasing from $0$ to $\infty$, so there is a **unique** critical point: $s^{\ast} \approx 1.1997$, i.e. $\theta^{\ast} = 4 s^{\ast} \approx 4.799$; $g$ is strictly increasing before it and strictly decreasing after. The maximum value is $g(\theta^{\ast}) = 4.799 / \cosh(1.1997) \approx 4.799 / 1.810 \approx 2.651$.

**Step 3 — The fold.** The equation $g(\theta) = \sqrt{2\lambda}$ therefore has: **two** roots $\theta_1 \lt \theta^{\ast} \lt \theta_2$ when $\sqrt{2\lambda} \lt 2.651$, i.e. $\lambda \lt \lambda^{\ast} = \tfrac{1}{2}(2.651)^2 \approx 3.514$; exactly **one** root at $\lambda = \lambda^{\ast}$; and **none** for $\lambda \gt \lambda^{\ast}$ (the known Bratu constant is $\lambda^{\ast} = 3.5138\dots$). The small root $\theta_1$ gives a shallow solution (small $y_{\max} = 2\ln\cosh(\theta/4)$), the large root $\theta_2$ a tall one; as $\lambda \uparrow \lambda^{\ast}$ the two merge and annihilate — a saddle-node (fold) bifurcation.

**Step 4 — Shooting interpretation.** The shooting function integrates the IVP with $y(0) = 0$, $y'(0) = s$ and reports $F(s) = y(1; s)$. From Step 1, the solutions have initial slopes

$$
s = y'(0) = -\theta \tanh\!\big(-\tfrac{\theta}{4}\big) = \theta \tanh \frac{\theta}{4},
$$

so for $\lambda \lt \lambda^{\ast}$ the graph of $F$ crosses zero **twice** — at $s_1 = \theta_1 \tanh(\theta_1/4)$ and $s_2 = \theta_2 \tanh(\theta_2/4)$. A bisection bracket $[0, s_{\text{big}}]$ finds only whichever root its initial signs enclose; locating both requires scanning $F$ for sign changes (or continuation in $\lambda$). At $\lambda = \lambda^{\ast}$, $F$ is *tangent* to zero — Newton's method degenerates (the root is double), the numerical signature of the fold. For $\lambda \gt \lambda^{\ast}$, $F \lt 0$ everywhere: the shooting method correctly reports nonexistence.

$$
\boxed{\ \text{Bratu's BVP has two solutions for } 0 \lt \lambda \lt \lambda^{\ast} \approx 3.514, \text{ one at } \lambda^{\ast}, \text{ none for } \lambda \gt \lambda^{\ast}.\ }
$$

> **Key Takeaway:** Nonlinear BVPs replace the linear all-or-nothing law with bifurcation structure — solution branches that coexist, merge at folds, and vanish — and the shooting function $F(s)$ is the diagnostic instrument that makes that structure computable.
